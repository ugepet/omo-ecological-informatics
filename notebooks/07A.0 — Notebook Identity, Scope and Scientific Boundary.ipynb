## Objective

The objective of Notebook 07A is to recover, preserve, taxonomically harmonize,
and quality-control the historical Omo tree structural measurements without
altering the original source values or confusing realized structural attributes
with intrinsic functional traits.

The historical source contains structural summaries at the:

**Species × Management Zone × Habitat**

level for the nine combinations of:

- Core × Major River
- Core × Stream
- Core × Upland
- Buffer × Major River
- Buffer × Stream
- Buffer × Upland
- Transition × Major River
- Transition × Stream
- Transition × Upland

Reported structural variables include:

- Total DBH
- Mean DBH
- Total height
- Mean height
- Basal area
- Basal-area percentage

---

## Scientific Boundary

Notebook 07A is restricted to:

1. faithful reconstruction of the historical structural tables;
2. preservation of original species names and numerical values;
3. reconciliation with the validated Notebook 05 canonical taxonomy;
4. investigation of DBH/girth and measurement units;
5. internal mathematical consistency checks;
6. basal-area validation;
7. identification and documentation of questionable records; and
8. production of validated structural datasets for downstream analyses.

Notebook 07A will **not**:

- acquire external functional traits;
- calculate community-weighted means (CWMs);
- calculate functional-diversity indices;
- treat DBH, observed height or basal area as intrinsic species traits;
- distribute zone-habitat structural summaries artificially among individual plots;
- redesign the taxonomic decisions established in Notebooks 02–05; or
- silently correct historical numerical values.

All suspected errors or inconsistencies will be preserved and explicitly flagged.

```python
# =============================================================================
# 07A.2 — Load Canonical Taxonomic Resources
# =============================================================================

species_metadata = pd.read_csv(SPECIES_METADATA_FILE)

community_matrix = pd.read_csv(COMMUNITY_MATRIX_FILE)

taxonomic_corrections = pd.read_csv(TAXONOMIC_CORRECTIONS_FILE)

taxonomic_review = pd.read_csv(TAXONOMIC_REVIEW_FILE)

species_inventory = pd.read_csv(SPECIES_INVENTORY_FILE)


print("Taxonomic resources loaded successfully.")
print("=" * 72)

print(f"Species metadata       : {species_metadata.shape}")
print(f"Community matrix       : {community_matrix.shape}")
print(f"Taxonomic corrections  : {taxonomic_corrections.shape}")
print(f"Taxonomic review       : {taxonomic_review.shape}")
print(f"Species inventory      : {species_inventory.shape}")
```

```python
# =============================================================================
# 07A.2.1 — Validate Canonical Species Inventory
# =============================================================================

# Expected metadata columns
required_metadata_columns = {"Species"}

missing_metadata_columns = (
    required_metadata_columns - set(species_metadata.columns)
)

assert not missing_metadata_columns, (
    f"Missing metadata columns: {missing_metadata_columns}"
)


# -------------------------------------------------------------------------
# Extract canonical species lists
# -------------------------------------------------------------------------

canonical_species_metadata = (
    species_metadata["Species"]
    .dropna()
    .astype(str)
    .str.strip()
    .tolist()
)


community_metadata_columns = {
    "Plot_ID",
    "Zone",
    "Habitat",
}

canonical_species_matrix = [
    column
    for column in community_matrix.columns
    if column not in community_metadata_columns
]


# -------------------------------------------------------------------------
# Basic inventory checks
# -------------------------------------------------------------------------

n_metadata_species = len(canonical_species_metadata)

n_unique_metadata_species = len(set(canonical_species_metadata))

n_matrix_species = len(canonical_species_matrix)

n_unique_matrix_species = len(set(canonical_species_matrix))


print("Canonical species validation")
print("=" * 72)

print(
    f"Species in metadata          : {n_metadata_species}"
)

print(
    f"Unique species in metadata   : {n_unique_metadata_species}"
)

print(
    f"Species columns in matrix    : {n_matrix_species}"
)

print(
    f"Unique matrix species        : {n_unique_matrix_species}"
)


# -------------------------------------------------------------------------
# Locked Notebook 05 expectations
# -------------------------------------------------------------------------

assert n_metadata_species == 127, (
    "Canonical metadata does not contain the expected 127 species."
)

assert n_unique_metadata_species == 127, (
    "Duplicate species detected in canonical metadata."
)

assert n_matrix_species == 127, (
    "Community matrix does not contain the expected 127 species columns."
)

assert n_unique_matrix_species == 127, (
    "Duplicate species columns detected in community matrix."
)


print("\nCanonical species counts: PASSED")
```

```python
# =============================================================================
# 07A.2.2 — Confirm Metadata–Matrix Taxonomic Synchronization
# =============================================================================

metadata_species_set = set(canonical_species_metadata)

matrix_species_set = set(canonical_species_matrix)


metadata_only = sorted(
    metadata_species_set - matrix_species_set
)

matrix_only = sorted(
    matrix_species_set - metadata_species_set
)


print("Metadata–community matrix synchronization")
print("=" * 72)

print(
    f"Species present only in metadata : {len(metadata_only)}"
)

print(
    f"Species present only in matrix   : {len(matrix_only)}"
)


if metadata_only:
    print("\nMetadata-only taxa:")
    for species in metadata_only:
        print(f"  - {species}")


if matrix_only:
    print("\nMatrix-only taxa:")
    for species in matrix_only:
        print(f"  - {species}")


assert metadata_species_set == matrix_species_set, (
    "Canonical metadata and community matrix species do not match."
)


print("\nCanonical taxonomy synchronization: PASSED")
```

```python
# =============================================================================
# 07A.2.3 — Validate Pre-Harmonization Taxonomic Audit Resources
# =============================================================================

assert "Scientific_Name" in species_inventory.columns, (
    "community_species_inventory.csv must contain 'Scientific_Name'."
)

assert "Original_Name" in taxonomic_review.columns, (
    "community_taxonomic_review.csv must contain 'Original_Name'."
)

assert "Original_Name" in taxonomic_corrections.columns, (
    "community_taxonomic_corrections.csv must contain 'Original_Name'."
)


# -------------------------------------------------------------------------
# Extract original taxon names
# -------------------------------------------------------------------------

inventory_taxa = (
    species_inventory["Scientific_Name"]
    .dropna()
    .astype(str)
    .str.strip()
)

review_taxa = (
    taxonomic_review["Original_Name"]
    .dropna()
    .astype(str)
    .str.strip()
)

correction_taxa = (
    taxonomic_corrections["Original_Name"]
    .dropna()
    .astype(str)
    .str.strip()
)


# -------------------------------------------------------------------------
# Basic record-count validation
# -------------------------------------------------------------------------

print("Pre-harmonization audit resources")
print("=" * 72)

print(f"Species inventory records    : {len(inventory_taxa)}")
print(f"Taxonomic review records     : {len(review_taxa)}")
print(f"Taxonomic correction records : {len(correction_taxa)}")


assert len(inventory_taxa) == 140, (
    "Expected 140 taxa in the pre-harmonization species inventory."
)

assert len(review_taxa) == 140, (
    "Expected 140 taxa in the taxonomic review."
)

assert len(correction_taxa) == 140, (
    "Expected 140 taxa in the taxonomic corrections file."
)


# -------------------------------------------------------------------------
# Normalize names FOR COMPARISON ONLY
#
# Important:
# This does not modify any source taxonomic values.
# It only removes formatting differences caused by capitalization or
# repeated whitespace.
# -------------------------------------------------------------------------

def normalize_taxon_name(name):
    return " ".join(str(name).split()).casefold()


inventory_taxa_normalized = {
    normalize_taxon_name(name)
    for name in inventory_taxa
}

review_taxa_normalized = {
    normalize_taxon_name(name)
    for name in review_taxa
}

correction_taxa_normalized = {
    normalize_taxon_name(name)
    for name in correction_taxa
}


# -------------------------------------------------------------------------
# Compare audit resources
# -------------------------------------------------------------------------

inventory_review_diff = sorted(
    inventory_taxa_normalized - review_taxa_normalized
)

review_inventory_diff = sorted(
    review_taxa_normalized - inventory_taxa_normalized
)

inventory_correction_diff = sorted(
    inventory_taxa_normalized - correction_taxa_normalized
)

correction_inventory_diff = sorted(
    correction_taxa_normalized - inventory_taxa_normalized
)


print("\nNormalized taxonomic comparison")
print("-" * 72)

print(
    f"Inventory-only taxa vs review     : "
    f"{len(inventory_review_diff)}"
)

print(
    f"Review-only taxa vs inventory     : "
    f"{len(review_inventory_diff)}"
)

print(
    f"Inventory-only taxa vs corrections: "
    f"{len(inventory_correction_diff)}"
)

print(
    f"Corrections-only taxa vs inventory: "
    f"{len(correction_inventory_diff)}"
)


# -------------------------------------------------------------------------
# Validate normalized equivalence
# -------------------------------------------------------------------------

assert inventory_taxa_normalized == review_taxa_normalized, (
    "Species inventory and taxonomic review still differ after "
    "case/whitespace normalization."
)

assert inventory_taxa_normalized == correction_taxa_normalized, (
    "Species inventory and taxonomic corrections still differ after "
    "case/whitespace normalization."
)

assert review_taxa_normalized == correction_taxa_normalized, (
    "Taxonomic review and corrections resources differ."
)


print("\nPre-harmonization audit layer: PASSED")
print(
    "Note: Original taxon spellings/capitalization remain unchanged; "
    "normalization was used only for validation."
)
```

This section reconstructs the nine historical structural tables into a single

The reconstruction follows a strict source-preservation rule:

- numerical values are retained as reported;

- suspicious values are not corrected;

```python
# =============================================================================
# 07A.4.1 — Inspect Structural Source Document
# =============================================================================

from docx import Document


document = Document(STRUCTURE_SOURCE_FILE)


print("Structural source document inspection")
print("=" * 72)

print(f"Source file       : {STRUCTURE_SOURCE_FILE.name}")
print(f"Word tables found : {len(document.tables)}")
print(f"Paragraphs found  : {len(document.paragraphs)}")


# -------------------------------------------------------------------------
# Inspect dimensions of detected tables
# -------------------------------------------------------------------------

table_summary = []

for i, table in enumerate(document.tables, start=1):

    n_rows = len(table.rows)

    n_cols = max(
        len(row.cells)
        for row in table.rows
    )

    table_summary.append(
        {
            "Word_Table": i,
            "Rows": n_rows,
            "Columns": n_cols,
        }
    )


table_summary = pd.DataFrame(table_summary)

display(table_summary)
```

```python
# =============================================================================
# 07A.4.2 — Inspect Raw Word Table Contents
# =============================================================================

def clean_cell_text(text):
    """
    Clean Word cell text for display only.

    This does NOT modify the source document.
    """
    return " ".join(str(text).split())


print("WORD TABLE 1 — FIRST 15 ROWS")
print("=" * 100)

table_1 = document.tables[0]

for i, row in enumerate(table_1.rows[:15]):
    values = [
        clean_cell_text(cell.text)
        for cell in row.cells
    ]

    print(f"Row {i:03d}: {values}")


print("\n\nWORD TABLE 2 — FIRST 25 ROWS")
print("=" * 100)

table_2 = document.tables[1]

for i, row in enumerate(table_2.rows[:25]):
    values = [
        clean_cell_text(cell.text)
        for cell in row.cells
    ]

    print(f"Row {i:03d}: {values}")
```

## 07A.5 — Taxonomic Reconciliation of Structural Source Names

This section reconciles historical species names in the structural source with
the validated 127-species canonical inventory inherited from Notebook 05.

Taxonomic reconciliation follows a conservative hierarchy:

1. exact canonical match;
2. normalized canonical match;
3. matching supported by existing Notebook 05 taxonomic audit resources;
4. unresolved candidate identification;
5. manual review where necessary.

No ambiguous taxon will be silently assigned to a canonical species.

The historical `Source_Species` field is preserved throughout the workflow.

The principal output of this section will be:

`structure_taxonomic_matching.csv`

```python
# =============================================================================
# 07A.5.1 — Build Unique Structural Species Inventory
# =============================================================================

# One taxonomic reconciliation record per historical source-species name.
# Family information is retained as provenance but does not define the taxon.

structure_taxa = (
    structure_raw[
        [
            "Source_Species",
            "Source_Family",
        ]
    ]
    .groupby(
        "Source_Species",
        as_index=False,
        sort=True,
    )
    .agg(
        Source_Families=(
            "Source_Family",
            lambda x: "; ".join(
                sorted(
                    set(
                        str(v).strip()
                        for v in x
                        if str(v).strip()
                    )
                )
            ),
        ),
        Family_Variants=(
            "Source_Family",
            lambda x: len(
                set(
                    str(v).strip()
                    for v in x
                    if str(v).strip()
                )
            ),
        ),
    )
)


print("Historical structural species inventory")
print("=" * 72)

print(f"Structural records          : {len(structure_raw)}")
print(f"Unique source species       : {len(structure_taxa)}")

print(
    f"Species with >1 family label: "
    f"{(structure_taxa['Family_Variants'] > 1).sum()}"
)

display(structure_taxa.head(20))
```

```python
# =============================================================================
# 07A.5.1A — Taxonomic Matching Utilities
# =============================================================================

import re
import unicodedata
from difflib import SequenceMatcher


def normalize_taxon_for_matching(name):
    """
    Normalize a taxonomic name for comparison only.

    Historical source names are never modified by this function.

    Normalization:
    - handles missing values;
    - converts to string;
    - applies Unicode normalization;
    - removes leading/trailing whitespace;
    - collapses repeated internal whitespace;
    - converts to lowercase.
    """

    if pd.isna(name):
        return ""

    name = str(name)

    name = unicodedata.normalize(
        "NFKC",
        name,
    )

    name = name.strip()

    name = re.sub(
        r"\s+",
        " ",
        name,
    )

    name = name.lower()

    return name


def taxon_similarity(a, b):
    """
    Calculate similarity between two taxonomic names after
    matching-only normalization.
    """

    return SequenceMatcher(
        None,
        normalize_taxon_for_matching(a),
        normalize_taxon_for_matching(b),
    ).ratio()


# -----------------------------------------------------------------------------
# Basic utility validation
# -----------------------------------------------------------------------------

assert (
    normalize_taxon_for_matching("Alstonia  boonei")
    ==
    normalize_taxon_for_matching("Alstonia boonei")
)

assert (
    normalize_taxon_for_matching("Blighia Sapida")
    ==
    normalize_taxon_for_matching("Blighia sapida")
)

assert (
    normalize_taxon_for_matching("  Celtis zenkeri  ")
    ==
    "celtis zenkeri"
)


print("Taxonomic matching utilities initialized: PASSED")
```

```python
# =============================================================================
# 07A.5.2 — Exact Canonical Matching
# =============================================================================

canonical_species_set = set(
    canonical_species_metadata
)


structure_taxa["Canonical_Species"] = pd.NA
structure_taxa["Match_Status"] = "UNRESOLVED"
structure_taxa["Match_Method"] = pd.NA


# -----------------------------------------------------------------------------
# Exact source-name → canonical-name matching
# -----------------------------------------------------------------------------

exact_match_mask = (
    structure_taxa["Source_Species"]
    .isin(canonical_species_set)
)


structure_taxa.loc[
    exact_match_mask,
    "Canonical_Species"
] = structure_taxa.loc[
    exact_match_mask,
    "Source_Species"
]


structure_taxa.loc[
    exact_match_mask,
    "Match_Status"
] = "EXACT_CANONICAL_MATCH"


structure_taxa.loc[
    exact_match_mask,
    "Match_Method"
] = "Exact string match"


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

n_exact = int(exact_match_mask.sum())

n_unresolved = int(
    structure_taxa["Match_Status"]
    .eq("UNRESOLVED")
    .sum()
)


print("Exact canonical taxonomic matching")
print("=" * 72)

print(f"Source species examined : {len(structure_taxa)}")
print(f"Exact canonical matches : {n_exact}")
print(f"Remaining unresolved    : {n_unresolved}")


display(
    structure_taxa.loc[
        structure_taxa["Match_Status"] == "UNRESOLVED",
        [
            "Source_Species",
            "Source_Families",
        ],
    ]
)
```

```python
# =============================================================================
# 07A.5.3 — Normalized Exact Canonical Matching
# =============================================================================

# IMPORTANT:
# This section resolves ONLY harmless formatting differences:
# - capitalization
# - leading/trailing whitespace
# - repeated internal whitespace
#
# No spelling correction is performed here.


# -----------------------------------------------------------------------------
# 1. Build normalized canonical lookup
# -----------------------------------------------------------------------------

canonical_normalized_lookup = {}

for canonical_name in canonical_species_metadata:

    normalized_name = normalize_taxon_for_matching(
        canonical_name
    )

    # Protect against accidental collisions in the canonical inventory.
    if normalized_name in canonical_normalized_lookup:

        existing_name = canonical_normalized_lookup[
            normalized_name
        ]

        assert existing_name == canonical_name, (
            "Normalization collision detected in canonical taxonomy: "
            f"{existing_name!r} and {canonical_name!r}"
        )

    canonical_normalized_lookup[
        normalized_name
    ] = canonical_name


print("Normalized canonical lookup")
print("=" * 72)

print(
    f"Canonical species supplied : "
    f"{len(canonical_species_metadata)}"
)

print(
    f"Normalized lookup entries  : "
    f"{len(canonical_normalized_lookup)}"
)


assert len(canonical_normalized_lookup) == 127, (
    "Expected 127 unique normalized canonical species names."
)


# -----------------------------------------------------------------------------
# 2. Apply normalized exact matching only to unresolved source taxa
# -----------------------------------------------------------------------------

normalized_matches_applied = []


for idx in structure_taxa.index:

    if structure_taxa.at[
        idx,
        "Match_Status"
    ] != "UNRESOLVED":
        continue

    source_name = structure_taxa.at[
        idx,
        "Source_Species"
    ]

    normalized_source = normalize_taxon_for_matching(
        source_name
    )

    if normalized_source in canonical_normalized_lookup:

        canonical_name = canonical_normalized_lookup[
            normalized_source
        ]

        structure_taxa.at[
            idx,
            "Canonical_Species"
        ] = canonical_name

        structure_taxa.at[
            idx,
            "Match_Status"
        ] = "NORMALIZED_CANONICAL_MATCH"

        structure_taxa.at[
            idx,
            "Match_Method"
        ] = (
            "Case/whitespace-normalized exact match"
        )

        normalized_matches_applied.append(
            {
                "Source_Species": source_name,
                "Canonical_Species": canonical_name,
            }
        )


normalized_matches_applied = pd.DataFrame(
    normalized_matches_applied
)


# -----------------------------------------------------------------------------
# 3. Summary
# -----------------------------------------------------------------------------

n_normalized_matches = int(
    structure_taxa["Match_Status"]
    .eq("NORMALIZED_CANONICAL_MATCH")
    .sum()
)

n_unresolved = int(
    structure_taxa["Match_Status"]
    .eq("UNRESOLVED")
    .sum()
)


print("\nNormalized canonical taxonomic matching")
print("=" * 72)

print(
    f"Normalized matches applied : "
    f"{n_normalized_matches}"
)

print(
    f"Remaining unresolved       : "
    f"{n_unresolved}"
)


print("\nNormalized matches")
print("-" * 72)

display(normalized_matches_applied)


print("\nCurrent matching status")
print("-" * 72)

display(
    structure_taxa["Match_Status"]
    .value_counts()
    .rename_axis("Match_Status")
    .reset_index(name="Taxa")
)


# -----------------------------------------------------------------------------
# 4. Remaining unresolved taxa
# -----------------------------------------------------------------------------

remaining_unresolved = (
    structure_taxa.loc[
        structure_taxa["Match_Status"] == "UNRESOLVED",
        [
            "Source_Species",
            "Source_Families",
        ],
    ]
    .copy()
    .sort_values("Source_Species")
    .reset_index(drop=True)
)


print("\nRemaining unresolved taxa")
print("-" * 72)

display(remaining_unresolved)


# -----------------------------------------------------------------------------
# 5. Locked validation
#
# The validated source state established earlier showed exactly 8 harmless
# case/whitespace variants.
# -----------------------------------------------------------------------------

expected_normalized_source_taxa = {
    "Alstonia  boonei",
    "Blighia Sapida",
    "Cleistopholis Patens",
    "Homalium  aylmeri",
    "Nauclea  diderrichii",
    "Pavetta  corymbosia",
    "Strombosia  pustulata",
    "irvinga gabonensis",
}


observed_normalized_source_taxa = set(
    normalized_matches_applied[
        "Source_Species"
    ]
)


assert n_normalized_matches == 8, (
    "Expected exactly 8 normalized canonical matches."
)

assert observed_normalized_source_taxa == expected_normalized_source_taxa, (
    "Normalized matching did not resolve the expected eight "
    "case/whitespace variants."
)

assert n_unresolved == 24, (
    "Expected exactly 24 taxa to remain unresolved after "
    "normalized exact matching."
)


print("\nNormalized canonical matching validation: PASSED")
```

```python
# =============================================================================
# 07A.5.4 — Generate Controlled Candidate Matches for Unresolved Taxa
# =============================================================================

from difflib import SequenceMatcher


# -----------------------------------------------------------------------------
# Unresolved source taxa
# -----------------------------------------------------------------------------

unresolved_taxa = (
    structure_taxa.loc[
        structure_taxa["Match_Status"] == "UNRESOLVED"
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# Similarity helper
#
# Candidate generation only.
# Similarity scores MUST NOT be treated as automatic taxonomic decisions.
# -----------------------------------------------------------------------------

def taxon_similarity(a, b):
    return SequenceMatcher(
        None,
        normalize_taxon_for_matching(a),
        normalize_taxon_for_matching(b),
    ).ratio()


candidate_records = []


for _, row in unresolved_taxa.iterrows():

    source_name = row["Source_Species"]

    candidates = []

    for canonical_name in canonical_species_metadata:

        score = taxon_similarity(
            source_name,
            canonical_name,
        )

        candidates.append(
            (
                canonical_name,
                score,
            )
        )

    candidates = sorted(
        candidates,
        key=lambda x: x[1],
        reverse=True,
    )[:3]


    candidate_records.append(
        {
            "Source_Species": source_name,
            "Source_Families": row["Source_Families"],
            "Candidate_1": candidates[0][0],
            "Similarity_1": round(candidates[0][1], 3),
            "Candidate_2": candidates[1][0],
            "Similarity_2": round(candidates[1][1], 3),
            "Candidate_3": candidates[2][0],
            "Similarity_3": round(candidates[2][1], 3),
        }
    )


candidate_review = pd.DataFrame(candidate_records)


print("Candidate review for unresolved structural taxa")
print("=" * 72)

print(
    f"Unresolved source taxa : {len(candidate_review)}"
)

display(candidate_review)
```

```python
# =============================================================================
# 07A.5.5 — Apply Controlled Historical-Name Harmonizations
# =============================================================================

# These mappings are conservative reconciliations with the validated
# 127-species canonical inventory.
#
# Important:
# - Source_Species values are NEVER modified.
# - Matching uses normalized source strings only.
# - Canonical targets must already exist in the validated canonical inventory.
# - This cell is designed to be safe to rerun.

controlled_harmonizations = {
    "Albizia auxillary": "Albizia auxilary",
    "Alstonei boonei": "Alstonia boonei",
    "Anthocliesta vogelii": "Anthocleista vogelii",
    "Anthocliesta vogeli": "Anthocleista vogelii",
    "Blghia sapida": "Blighia sapida",
    "Bombax buonopozene": "Bombax buonopozense",
    "Bosquei angolensis": "Bosqueia angolensis",
    "Bridellia ferruginea": "Bridelia ferruginea",
    "Cleistoformis patent": "Cleistoformis patens",
    "Diospyros mespiliformis": "Diospyrous mespiliformis",
    "Hildergardia baarteri": "Hildegardia barteri",
    "Lonchocarpus ciliata": "Lonchocarpus cilliata",
    "Magaritaria discoidea": "Margaritaria discoidea",
    "Mesobotyra barteri": "Maesobotrya barteri",
    "Monodora lenuifolia": "Monodora tenuifolia",
    "Nesorgodonia papaverifera": "Nesogordonia papaverifera",
    "Pterocarpus santpxyloides": "Pterocarpus santoxyloides",
    "Pycnantus angolensis": "Pycnanthus angolensis",
    "Scotelia coriacea": "Scotellia coriacea",
}


# -----------------------------------------------------------------------------
# 1. Validate canonical targets
# -----------------------------------------------------------------------------

invalid_targets = sorted(
    set(controlled_harmonizations.values())
    - canonical_species_set
)

assert not invalid_targets, (
    "One or more controlled harmonization targets are not present "
    f"in the canonical 127 inventory: {invalid_targets}"
)


# -----------------------------------------------------------------------------
# 2. Normalized lookup for the 19 controlled mappings
# -----------------------------------------------------------------------------

controlled_harmonizations_normalized = {
    normalize_taxon_for_matching(source_name): canonical_name
    for source_name, canonical_name
    in controlled_harmonizations.items()
}


# -----------------------------------------------------------------------------
# 3. Apply mappings
#
# Apply only where the taxon has not already received a downstream
# individual-review classification.
# -----------------------------------------------------------------------------

protected_review_statuses = {
    "AMBIGUOUS",
    "NOT_IN_CANONICAL_127",
}


for idx in structure_taxa.index:

    source_name = structure_taxa.at[
        idx,
        "Source_Species"
    ]

    normalized_source = normalize_taxon_for_matching(
        source_name
    )

    current_status = structure_taxa.at[
        idx,
        "Match_Status"
    ]

    # Do not overwrite later individual-review decisions.
    if current_status in protected_review_statuses:
        continue

    if normalized_source in controlled_harmonizations_normalized:

        canonical_name = (
            controlled_harmonizations_normalized[
                normalized_source
            ]
        )

        structure_taxa.at[
            idx,
            "Canonical_Species"
        ] = canonical_name

        structure_taxa.at[
            idx,
            "Match_Status"
        ] = "HARMONIZED_MATCH"

        structure_taxa.at[
            idx,
            "Match_Method"
        ] = (
            "Controlled historical-name harmonization"
        )


# -----------------------------------------------------------------------------
# 4. Validate the 19 controlled harmonizations directly
# -----------------------------------------------------------------------------

controlled_source_normalized = set(
    controlled_harmonizations_normalized.keys()
)


controlled_mask = (
    structure_taxa["Source_Species"]
    .apply(normalize_taxon_for_matching)
    .isin(controlled_source_normalized)
)


controlled_results = (
    structure_taxa.loc[
        controlled_mask,
        [
            "Source_Species",
            "Canonical_Species",
            "Match_Status",
            "Match_Method",
        ],
    ]
    .copy()
    .sort_values("Source_Species")
    .reset_index(drop=True)
)


n_controlled_records = len(controlled_results)

n_harmonized = int(
    controlled_results["Match_Status"]
    .eq("HARMONIZED_MATCH")
    .sum()
)


print("Controlled historical-name harmonization")
print("=" * 72)

print(
    f"Controlled mappings defined : "
    f"{len(controlled_harmonizations)}"
)

print(
    f"Controlled source taxa found: "
    f"{n_controlled_records}"
)

print(
    f"HARMONIZED_MATCH taxa       : "
    f"{n_harmonized}"
)


display(controlled_results)


assert n_controlled_records == 19, (
    "Expected exactly 19 controlled source taxa in structure_taxa."
)

assert n_harmonized == 19, (
    "Expected exactly 19 controlled taxa to have HARMONIZED_MATCH status."
)


# -----------------------------------------------------------------------------
# 5. Identify taxa still requiring individual review
# -----------------------------------------------------------------------------

expected_review_taxa = {
    "Elaeis guinenesis",
    "Lonchocarpus cayensis",
    "Micranga barteri",
    "Pavetta crassipes",
    "Ricinidendron heudelotii",
}


review_taxa_state = (
    structure_taxa.loc[
        structure_taxa["Source_Species"].isin(
            expected_review_taxa
        ),
        [
            "Source_Species",
            "Canonical_Species",
            "Match_Status",
            "Match_Method",
        ],
    ]
    .copy()
    .sort_values("Source_Species")
    .reset_index(drop=True)
)


assert len(review_taxa_state) == 5, (
    "Expected all five individual-review taxa to exist in structure_taxa."
)


print("\nFive taxa reserved for individual review")
print("-" * 72)

display(review_taxa_state)


# -----------------------------------------------------------------------------
# 6. Current complete matching-status summary
# -----------------------------------------------------------------------------

print("\nCurrent matching status")
print("-" * 72)

status_summary = (
    structure_taxa["Match_Status"]
    .value_counts(dropna=False)
    .rename_axis("Match_Status")
    .reset_index(name="Taxa")
)

display(status_summary)


# -----------------------------------------------------------------------------
# 7. Check for unexpected unresolved taxa
#
# Only the five predefined individual-review taxa are allowed to remain
# unresolved at this stage.
# -----------------------------------------------------------------------------

remaining_unresolved = (
    structure_taxa.loc[
        structure_taxa["Match_Status"] == "UNRESOLVED",
        [
            "Source_Species",
            "Source_Families",
        ],
    ]
    .copy()
    .sort_values("Source_Species")
    .reset_index(drop=True)
)


unexpected_unresolved = set(
    remaining_unresolved["Source_Species"]
) - expected_review_taxa


assert not unexpected_unresolved, (
    "Unexpected unresolved taxa remain after controlled harmonization: "
    f"{sorted(unexpected_unresolved)}"
)


print(
    f"\nCurrently UNRESOLVED taxa : "
    f"{len(remaining_unresolved)}"
)

if len(remaining_unresolved) > 0:
    display(remaining_unresolved)


print("\nControlled harmonization validation: PASSED")
```

Five historical structural taxa remain unresolved after exact, normalized,

These taxa are reviewed individually against the validated 127-species

No taxon is forced into the canonical inventory where:

Such taxa are retained either as `AMBIGUOUS` or `NOT_IN_CANONICAL_127`.

```python
# =============================================================================
# 07A.5.6 — Individual Review of Remaining Structural Taxa
# =============================================================================

individual_taxon_decisions = {
    "Elaeis guinenesis": {
        "Canonical_Species": pd.NA,
        "Match_Status": "AMBIGUOUS",
        "Match_Method": "Individual review",
        "Taxonomic_Notes": (
            "Multiple competing Elaeis forms remain in the canonical "
            "127 inventory: Elaeis guineensis, Elaeis guinensis, and "
            "Elaeis guineense. No silent assignment made."
        ),
    },

    "Lonchocarpus cayensis": {
        "Canonical_Species": pd.NA,
        "Match_Status": "NOT_IN_CANONICAL_127",
        "Match_Method": "Individual review",
        "Taxonomic_Notes": (
            "No exact canonical equivalent. Historical source contains "
            "other Lonchocarpus taxa separately, so similarity to "
            "Lonchocarpus cyanescens is insufficient for consolidation."
        ),
    },

    "Micranga barteri": {
        "Canonical_Species": pd.NA,
        "Match_Status": "AMBIGUOUS",
        "Match_Method": "Individual review",
        "Taxonomic_Notes": (
            "Macaranga barteri exists in the canonical inventory, but "
            "Macaranga barteri and Micranga barteri occur as separate "
            "historical structural records. Automatic consolidation "
            "would therefore be unsafe."
        ),
    },

    "Pavetta crassipes": {
        "Canonical_Species": pd.NA,
        "Match_Status": "NOT_IN_CANONICAL_127",
        "Match_Method": "Individual review",
        "Taxonomic_Notes": (
            "No supported canonical equivalent identified. Similarity "
            "to Pavetta corymbosa/corymbosia is insufficient for "
            "taxonomic consolidation."
        ),
    },

    "Ricinidendron heudelotii": {
        "Canonical_Species": pd.NA,
        "Match_Status": "AMBIGUOUS",
        "Match_Method": "Individual review",
        "Taxonomic_Notes": (
            "Multiple competing Ricinodendron/Ricinidendron variants "
            "remain in the canonical 127 inventory. No silent assignment made."
        ),
    },
}


# -----------------------------------------------------------------------------
# Apply individual decisions
# -----------------------------------------------------------------------------

for source_name, decision in individual_taxon_decisions.items():

    mask = (
        structure_taxa["Source_Species"].eq(source_name)
        & structure_taxa["Match_Status"].eq("UNRESOLVED")
    )

    assert mask.sum() == 1, (
        f"Expected exactly one unresolved record for {source_name}, "
        f"found {mask.sum()}."
    )

    structure_taxa.loc[
        mask,
        "Canonical_Species"
    ] = decision["Canonical_Species"]

    structure_taxa.loc[
        mask,
        "Match_Status"
    ] = decision["Match_Status"]

    structure_taxa.loc[
        mask,
        "Match_Method"
    ] = decision["Match_Method"]


# -----------------------------------------------------------------------------
# Add taxonomic notes if not already present
# -----------------------------------------------------------------------------

if "Taxonomic_Notes" not in structure_taxa.columns:
    structure_taxa["Taxonomic_Notes"] = pd.NA


for source_name, decision in individual_taxon_decisions.items():

    mask = structure_taxa["Source_Species"].eq(
        source_name
    )

    structure_taxa.loc[
        mask,
        "Taxonomic_Notes"
    ] = decision["Taxonomic_Notes"]


# -----------------------------------------------------------------------------
# Final taxonomic status summary
# -----------------------------------------------------------------------------

status_summary = (
    structure_taxa["Match_Status"]
    .value_counts()
    .rename_axis("Match_Status")
    .reset_index(name="Taxa")
)


print("Individual taxonomic review")
print("=" * 72)

display(status_summary)


remaining_unresolved = (
    structure_taxa["Match_Status"]
    .eq("UNRESOLVED")
    .sum()
)

print(
    f"\nRemaining UNRESOLVED taxa: "
    f"{remaining_unresolved}"
)


# -----------------------------------------------------------------------------
# Display individually reviewed taxa
# -----------------------------------------------------------------------------

display(
    structure_taxa.loc[
        structure_taxa["Source_Species"].isin(
            individual_taxon_decisions.keys()
        ),
        [
            "Source_Species",
            "Source_Families",
            "Canonical_Species",
            "Match_Status",
            "Match_Method",
            "Taxonomic_Notes",
        ],
    ]
    .sort_values("Source_Species")
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# Validation
# -----------------------------------------------------------------------------

assert remaining_unresolved == 0, (
    "Unclassified structural taxa remain."
)

assert (
    structure_taxa["Match_Status"]
    .eq("AMBIGUOUS")
    .sum()
    == 3
), (
    "Expected exactly 3 ambiguous taxa."
)

assert (
    structure_taxa["Match_Status"]
    .eq("NOT_IN_CANONICAL_127")
    .sum()
    == 2
), (
    "Expected exactly 2 taxa outside the canonical 127."
)


print("\nIndividual taxonomic review: PASSED")
```

```python
# =============================================================================
# 07A.5.11 — Add Taxonomic QC Flags to Harmonized Records
# =============================================================================

def append_qc_flag(existing_flag, new_flag):
    """
    Append a QC flag without removing existing flags.
    """

    existing_flag = (
        ""
        if pd.isna(existing_flag)
        else str(existing_flag).strip()
    )

    if not existing_flag:
        return new_flag

    flags = [
        flag.strip()
        for flag in existing_flag.split(";")
        if flag.strip()
    ]

    if new_flag not in flags:
        flags.append(new_flag)

    return ";".join(flags)


# -----------------------------------------------------------------------------
# Harmonized-name flags
# -----------------------------------------------------------------------------

harmonized_mask = structure_harmonized["Match_Status"].isin(
    [
        "NORMALIZED_CANONICAL_MATCH",
        "HARMONIZED_MATCH",
    ]
)

structure_harmonized.loc[
    harmonized_mask,
    "QC_Flag"
] = structure_harmonized.loc[
    harmonized_mask,
    "QC_Flag"
].apply(
    lambda x: append_qc_flag(
        x,
        "NAME_HARMONIZED",
    )
)


# -----------------------------------------------------------------------------
# Ambiguous taxa
# -----------------------------------------------------------------------------

ambiguous_mask = (
    structure_harmonized["Match_Status"]
    .eq("AMBIGUOUS")
)

structure_harmonized.loc[
    ambiguous_mask,
    "QC_Flag"
] = structure_harmonized.loc[
    ambiguous_mask,
    "QC_Flag"
].apply(
    lambda x: append_qc_flag(
        x,
        "AMBIGUOUS_TAXON",
    )
)


# -----------------------------------------------------------------------------
# Taxa outside canonical 127
# -----------------------------------------------------------------------------

outside_mask = (
    structure_harmonized["Match_Status"]
    .eq("NOT_IN_CANONICAL_127")
)

structure_harmonized.loc[
    outside_mask,
    "QC_Flag"
] = structure_harmonized.loc[
    outside_mask,
    "QC_Flag"
].apply(
    lambda x: append_qc_flag(
        x,
        "NOT_IN_CANONICAL_127",
    )
)


print("Taxonomic QC propagation")
print("=" * 72)

print(
    f"Records with harmonized names : "
    f"{harmonized_mask.sum()}"
)

print(
    f"Records with ambiguous taxa   : "
    f"{ambiguous_mask.sum()}"
)

print(
    f"Records outside canonical 127 : "
    f"{outside_mask.sum()}"
)


print("\nCurrent record-level QC flags")
print("-" * 72)

display(
    structure_harmonized["QC_Flag"]
    .replace("", "OK")
    .value_counts()
    .rename_axis("QC_Flag")
    .reset_index(name="Records")
)
```

## 07A.6 — Structural Measurement Validation

This section evaluates the internal mathematical consistency of the historical
Omo structural measurements.

The objectives are to investigate:

- whether reported totals and means imply coherent individual counts;
- whether DBH and height imply the same number of individuals;
- whether reported basal area is mathematically consistent with mean DBH;
- whether the source variable labelled `dbh` behaves as diameter or girth;
- likely measurement units; and
- records requiring structural QC flags.

Historical source values are never overwritten.

Separate numeric working columns are created solely for validation.

```python
# =============================================================================
# 07A.6.1 — Create Numeric Working Copy
# =============================================================================

structure_validation = structure_harmonized.copy()

numeric_source_columns = [
    "Total_DBH",
    "Mean_DBH",
    "Total_Height",
    "Mean_Height",
    "Basal_Area",
    "BA_Percent",
]


print("Numeric structural validation layer")
print("=" * 72)

print(f"Records : {len(structure_validation)}")


# -----------------------------------------------------------------------------
# Helper: normalize numeric formatting for ANALYSIS ONLY
#
# This does NOT modify historical source values.
# It only removes whitespace occurring inside numeric strings.
#
# Example:
#     "805. 6" -> "805.6"
# -----------------------------------------------------------------------------

def normalize_numeric_string(value):

    if pd.isna(value):
        return value

    return "".join(
        str(value).split()
    )


# -----------------------------------------------------------------------------
# Convert historical source values to separate numeric working columns
# -----------------------------------------------------------------------------

for column in numeric_source_columns:

    numeric_column = f"{column}_Num"

    normalized_values = (
        structure_validation[column]
        .apply(normalize_numeric_string)
    )

    structure_validation[numeric_column] = pd.to_numeric(
        normalized_values,
        errors="coerce",
    )


# -----------------------------------------------------------------------------
# Check conversion failures
# -----------------------------------------------------------------------------

print("\nNumeric conversion failures")
print("-" * 72)

conversion_failure_counts = {}

for column in numeric_source_columns:

    numeric_column = f"{column}_Num"

    failures = int(
        structure_validation[numeric_column]
        .isna()
        .sum()
    )

    conversion_failure_counts[column] = failures

    print(
        f"{numeric_column:<25}: "
        f"{failures}"
    )


assert all(
    count == 0
    for count in conversion_failure_counts.values()
), (
    "One or more structural source values still cannot be "
    "converted to numeric."
)


# -----------------------------------------------------------------------------
# Explicitly document known source-format normalization
# -----------------------------------------------------------------------------

format_normalization_mask = (
    structure_validation[numeric_source_columns]
    .astype(str)
    .apply(
        lambda column:
        column.str.contains(
            r"\.\s+\d",
            regex=True,
            na=False,
        )
    )
    .any(axis=1)
)


format_normalization_records = (
    structure_validation.loc[
        format_normalization_mask,
        [
            "Raw_Record_ID",
            "Source_Species",
            "Source_Table",
            "Source_Row",
            "Total_Height",
            "Total_Height_Num",
        ],
    ]
    .copy()
)


print("\nSource-format normalization records")
print("-" * 72)

print(
    f"Records requiring numeric-format normalization : "
    f"{len(format_normalization_records)}"
)

display(format_normalization_records)


# -----------------------------------------------------------------------------
# Validation
# -----------------------------------------------------------------------------

assert len(format_normalization_records) == 1, (
    "Expected exactly one known source numeric-format artifact."
)

assert (
    format_normalization_records.iloc[0]["Raw_Record_ID"]
    == "OMO_STR_0198"
), (
    "Unexpected record identified for numeric-format normalization."
)

assert np.isclose(
    format_normalization_records.iloc[0]["Total_Height_Num"],
    805.6,
), (
    "The historical value '805. 6' was not interpreted as 805.6."
)


print("\nNumeric conversion: PASSED")
print(
    "Historical source values remain unchanged; "
    "format normalization applies only to numeric working columns."
)
```

```python
# =============================================================================
# 07A.6.2 — Infer Individual Counts from Totals and Means
# =============================================================================

structure_validation["Inferred_N_DBH"] = (
    structure_validation["Total_DBH_Num"]
    / structure_validation["Mean_DBH_Num"]
)

structure_validation["Inferred_N_Height"] = (
    structure_validation["Total_Height_Num"]
    / structure_validation["Mean_Height_Num"]
)

structure_validation["Rounded_N_DBH"] = (
    structure_validation["Inferred_N_DBH"].round()
)

structure_validation["Rounded_N_Height"] = (
    structure_validation["Inferred_N_Height"].round()
)

structure_validation["DBH_N_Integer_Error"] = (
    structure_validation["Inferred_N_DBH"]
    - structure_validation["Rounded_N_DBH"]
).abs()

structure_validation["Height_N_Integer_Error"] = (
    structure_validation["Inferred_N_Height"]
    - structure_validation["Rounded_N_Height"]
).abs()


print("Inferred individual counts")
print("=" * 72)

display(
    structure_validation[
        [
            "Raw_Record_ID",
            "Source_Species",
            "Management_Zone",
            "Habitat",
            "Total_DBH_Num",
            "Mean_DBH_Num",
            "Inferred_N_DBH",
            "Total_Height_Num",
            "Mean_Height_Num",
            "Inferred_N_Height",
        ]
    ].head(15)
)
```

```python
# =============================================================================
# 07A.6.3 — Evaluate Total/Mean Internal Consistency
# =============================================================================

COUNT_INTEGER_TOLERANCE = 0.15


structure_validation["DBH_Count_Consistent"] = (
    structure_validation["DBH_N_Integer_Error"]
    <= COUNT_INTEGER_TOLERANCE
)

structure_validation["Height_Count_Consistent"] = (
    structure_validation["Height_N_Integer_Error"]
    <= COUNT_INTEGER_TOLERANCE
)

structure_validation["DBH_Height_Count_Agree"] = (
    structure_validation["Rounded_N_DBH"]
    == structure_validation["Rounded_N_Height"]
)

structure_validation["Total_Mean_Consistent"] = (
    structure_validation["DBH_Count_Consistent"]
    & structure_validation["Height_Count_Consistent"]
    & structure_validation["DBH_Height_Count_Agree"]
)


n_records = len(structure_validation)

print("Total/mean consistency assessment")
print("=" * 72)

print(
    f"DBH internally consistent       : "
    f"{structure_validation['DBH_Count_Consistent'].sum()} / {n_records}"
)

print(
    f"Height internally consistent    : "
    f"{structure_validation['Height_Count_Consistent'].sum()} / {n_records}"
)

print(
    f"DBH–height inferred counts agree: "
    f"{structure_validation['DBH_Height_Count_Agree'].sum()} / {n_records}"
)

print(
    f"Fully total/mean consistent     : "
    f"{structure_validation['Total_Mean_Consistent'].sum()} / {n_records}"
)
```

```python
# =============================================================================
# 07A.6.5 — Classify Total/Mean Consistency Failures
# =============================================================================

def classify_count_consistency(row):
    """
    Classify the reason a structural record fails the
    total/mean consistency assessment.
    """

    failures = []

    if not row["DBH_Count_Consistent"]:
        failures.append("DBH_NONINTEGER")

    if not row["Height_Count_Consistent"]:
        failures.append("HEIGHT_NONINTEGER")

    if not row["DBH_Height_Count_Agree"]:
        failures.append("DBH_HEIGHT_COUNT_DISAGREE")

    if not failures:
        return "CONSISTENT"

    return ";".join(failures)


structure_validation["Count_Consistency_Class"] = (
    structure_validation.apply(
        classify_count_consistency,
        axis=1,
    )
)


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

consistency_class_summary = (
    structure_validation["Count_Consistency_Class"]
    .value_counts()
    .rename_axis("Consistency_Class")
    .reset_index(name="Records")
)


print("Total/mean inconsistency classification")
print("=" * 72)

display(consistency_class_summary)


# -----------------------------------------------------------------------------
# Inspect the 26 problematic records in diagnostic order
# -----------------------------------------------------------------------------

diagnostic_columns = [
    "Raw_Record_ID",
    "Source_Species",
    "Source_Table",
    "Management_Zone",
    "Habitat",
    "Total_DBH_Num",
    "Mean_DBH_Num",
    "Inferred_N_DBH",
    "DBH_N_Integer_Error",
    "Total_Height_Num",
    "Mean_Height_Num",
    "Inferred_N_Height",
    "Height_N_Integer_Error",
    "Rounded_N_DBH",
    "Rounded_N_Height",
    "Count_Consistency_Class",
]


inconsistency_diagnostics = (
    structure_validation.loc[
        ~structure_validation["Total_Mean_Consistent"],
        diagnostic_columns,
    ]
    .copy()
    .sort_values(
        [
            "Count_Consistency_Class",
            "Source_Table",
            "Source_Species",
        ]
    )
    .reset_index(drop=True)
)


print(
    f"\nRecords requiring inspection: "
    f"{len(inconsistency_diagnostics)}"
)

display(inconsistency_diagnostics)
```

### 07A.6.9 — Basal-Area Formula and DBH Interpretation

The historical tables label the structural variable as `dbh`, while the
surrounding source text refers to tree girth. The basal-area column is labelled
`BA m2`.

Rather than assuming these definitions or units, this section tests whether
the reported basal-area values are mathematically consistent with the reported
mean DBH values.

For a circular stem of diameter \(d\):

\[
BA = \frac{\pi d^2}{4}
\]

The reported basal area is compared directly with the value predicted from
`Mean_DBH`. This test is initially unit-neutral: it evaluates numerical
consistency before assigning physical units.

```python
# =============================================================================
# 07A.6.9 — Test Basal Area Against Mean DBH
# =============================================================================

import numpy as np


# -----------------------------------------------------------------------------
# Predicted basal area from reported Mean_DBH
# -----------------------------------------------------------------------------

structure_validation["BA_From_Mean_DBH"] = (
    np.pi
    * structure_validation["Mean_DBH_Num"] ** 2
    / 4
)


# -----------------------------------------------------------------------------
# Absolute and relative discrepancies
# -----------------------------------------------------------------------------

structure_validation["BA_Absolute_Error"] = (
    structure_validation["Basal_Area_Num"]
    - structure_validation["BA_From_Mean_DBH"]
).abs()


structure_validation["BA_Relative_Error"] = (
    structure_validation["BA_Absolute_Error"]
    / structure_validation["Basal_Area_Num"]
)


structure_validation["BA_Percent_Error"] = (
    structure_validation["BA_Relative_Error"]
    * 100
)


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

print("Basal-area formula validation")
print("=" * 72)

print("\nReported BA")
print("-" * 72)

display(
    structure_validation["Basal_Area_Num"]
    .describe()
    .to_frame("Reported_BA")
)


print("\nPredicted BA from Mean DBH")
print("-" * 72)

display(
    structure_validation["BA_From_Mean_DBH"]
    .describe()
    .to_frame("Predicted_BA")
)


print("\nPercentage-error distribution")
print("-" * 72)

display(
    structure_validation["BA_Percent_Error"]
    .describe()
    .to_frame("BA_Percent_Error")
)
```

```python
# =============================================================================
# 07A.6.10 — Evaluate Basal-Area Mathematical Agreement
# =============================================================================

ba_correlation = (
    structure_validation[
        [
            "Basal_Area_Num",
            "BA_From_Mean_DBH",
        ]
    ]
    .corr()
    .iloc[0, 1]
)


ba_ratio = (
    structure_validation["Basal_Area_Num"]
    / structure_validation["BA_From_Mean_DBH"]
)


structure_validation["BA_Reported_Predicted_Ratio"] = ba_ratio


print("Basal-area mathematical agreement")
print("=" * 72)

print(
    f"Correlation reported vs predicted BA : "
    f"{ba_correlation:.6f}"
)


print("\nReported / predicted BA ratio")
print("-" * 72)

display(
    ba_ratio
    .describe()
    .to_frame("Reported_Predicted_Ratio")
)
```

```python
# =============================================================================
# 07A.6.11 — Classify Basal-Area Formula Agreement
# =============================================================================

BA_RELATIVE_TOLERANCE = 0.05


structure_validation["BA_Formula_Consistent"] = (
    structure_validation["BA_Relative_Error"]
    <= BA_RELATIVE_TOLERANCE
)


n_ba_consistent = int(
    structure_validation["BA_Formula_Consistent"]
    .sum()
)

n_ba_inconsistent = int(
    (~structure_validation["BA_Formula_Consistent"])
    .sum()
)


print("Basal-area formula consistency")
print("=" * 72)

print(
    f"Within 5% of π × Mean_DBH² / 4 : "
    f"{n_ba_consistent} / {len(structure_validation)}"
)

print(
    f"Outside 5% tolerance           : "
    f"{n_ba_inconsistent} / {len(structure_validation)}"
)


# -----------------------------------------------------------------------------
# Largest discrepancies
# -----------------------------------------------------------------------------

print("\nLargest basal-area discrepancies")
print("-" * 72)

display(
    structure_validation[
        [
            "Raw_Record_ID",
            "Source_Species",
            "Source_Table",
            "Mean_DBH_Num",
            "Basal_Area_Num",
            "BA_From_Mean_DBH",
            "BA_Reported_Predicted_Ratio",
            "BA_Percent_Error",
        ]
    ]
    .sort_values(
        "BA_Percent_Error",
        ascending=False,
    )
    .head(20)
)
```

```python
# =============================================================================
# 07A.6.12 — Formalize Basal-Area and DBH Interpretation
# =============================================================================

# Very strict diagnostic threshold because the observed agreement
# is substantially stronger than the initial 5% screening tolerance.
BA_STRICT_RELATIVE_TOLERANCE = 0.001   # 0.1%


structure_validation["BA_Strictly_Consistent"] = (
    structure_validation["BA_Relative_Error"]
    <= BA_STRICT_RELATIVE_TOLERANCE
)


n_strict_consistent = int(
    structure_validation["BA_Strictly_Consistent"].sum()
)


print("Structural measurement interpretation")
print("=" * 72)

print(
    f"BA records within 0.1% of π × Mean_DBH² / 4 : "
    f"{n_strict_consistent} / {len(structure_validation)}"
)

print(
    f"Maximum observed BA percentage error          : "
    f"{structure_validation['BA_Percent_Error'].max():.4f}%"
)

print(
    f"Median reported/predicted BA ratio            : "
    f"{structure_validation['BA_Reported_Predicted_Ratio'].median():.6f}"
)


assert n_strict_consistent == len(structure_validation), (
    "Not all basal-area records satisfy the strict mathematical "
    "consistency criterion."
)


# -----------------------------------------------------------------------------
# Dataset-level interpretation
# -----------------------------------------------------------------------------

DBH_MEASUREMENT_INTERPRETATION = (
    "Diameter-based measurement"
)

DBH_UNIT_STATUS = (
    "Likely centimetres; strongly supported by magnitude and BA relationship, "
    "but not independently documented in the historical source."
)

BASAL_AREA_INTERPRETATION = (
    "Derived from Mean_DBH using pi*d^2/4; represents area associated "
    "with mean diameter rather than aggregate species-context basal area."
)

BASAL_AREA_UNIT_STATUS = (
    "Reported source heading states m2, but numerical values behave as "
    "squared DBH units. If DBH is centimetres, reported BA values are cm2."
)

BASAL_AREA_VALIDATION_STATUS = (
    "Mathematically validated for all 222 records."
)


print("\nInterpretation")
print("-" * 72)

print(f"DBH interpretation : {DBH_MEASUREMENT_INTERPRETATION}")
print(f"DBH unit status    : {DBH_UNIT_STATUS}")
print(f"BA interpretation  : {BASAL_AREA_INTERPRETATION}")
print(f"BA unit status     : {BASAL_AREA_UNIT_STATUS}")
print(f"BA validation      : {BASAL_AREA_VALIDATION_STATUS}")

print("\nBasal-area structural interpretation: PASSED")
```

### 07A.6.13 — Height Range and Unit Plausibility Assessment

The historical source reports total and mean tree height but does not provide
sufficient independent metadata to establish the measurement unit with
certainty.

This section therefore evaluates:

- the distribution of reported mean heights;
- minimum and maximum values;
- unusually low or high observations;
- consistency across management zones and habitats; and
- whether the observed magnitudes are biologically plausible if interpreted
  as metres.

```python
# =============================================================================
# 07A.6.18 — Validate BA Percentage Totals by Source Table
# =============================================================================

ba_percent_table_summary = (
    structure_validation
    .groupby(
        [
            "Source_Table",
            "Management_Zone",
            "Habitat",
        ],
        observed=True,
    )
    .agg(
        Records=("Raw_Record_ID", "count"),
        Sum_Reported_BA=("Basal_Area_Num", "sum"),
        Sum_Reported_BA_Percent=("BA_Percent_Num", "sum"),
    )
    .reset_index()
)


ba_percent_table_summary["Percent_Total_Error"] = (
    ba_percent_table_summary["Sum_Reported_BA_Percent"]
    - 100.0
).abs()


print("Basal-area percentage totals by source table")
print("=" * 72)

display(ba_percent_table_summary)


print("\nSummary")
print("-" * 72)

print(
    f"Source tables assessed : "
    f"{len(ba_percent_table_summary)}"
)

print(
    f"Minimum BA % total     : "
    f"{ba_percent_table_summary['Sum_Reported_BA_Percent'].min():.2f}"
)

print(
    f"Maximum BA % total     : "
    f"{ba_percent_table_summary['Sum_Reported_BA_Percent'].max():.2f}"
)

print(
    f"Maximum deviation from 100% : "
    f"{ba_percent_table_summary['Percent_Total_Error'].max():.2f}"
)


assert len(ba_percent_table_summary) == 9, (
    "Expected exactly nine structural source tables."
)
```
