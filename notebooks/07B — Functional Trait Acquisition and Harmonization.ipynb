# Notebook 07B — Functional Trait Acquisition and Harmonization

## Omo Forest Ecological Informatics Project

### Purpose

This notebook assembles, harmonizes and documents intrinsic functional-trait
information for the validated 127-species Omo woody-flora inventory.

The principal functional traits targeted are:

- Wood density
- Specific leaf area (SLA)
- Seed mass
- Maximum potential / maximum adult height

This notebook focuses on trait acquisition, taxonomic reconciliation,
unit harmonization, aggregation and provenance.

It does **not** perform final trait imputation or functional-diversity analysis.

Those tasks are reserved for subsequent Notebook 07 components.

### Critical analytical boundary

Observed DBH, observed tree height and basal area reconstructed in Notebook 07A
are realized stand structural attributes and are **not** treated as intrinsic
species functional traits.

The authoritative taxonomic universe for this notebook is the validated
127-species Omo canonical inventory.

## 1. Canonical Species Foundation

Notebook 07B inherits its taxonomic universe from the previously validated
Omo community data products.

No taxonomic redesign is performed here.

The canonical species table is therefore treated as an upstream dependency,
and the notebook validates that it contains exactly:

- 127 species records
- 127 unique canonical species names
- no missing canonical species names

External functional-trait names will subsequently be reconciled to these
canonical names while preserving both the Omo canonical name and the
external-source name.

```python
# =============================================================================
# 1.1 LOAD CANONICAL SPECIES METADATA
# =============================================================================

SPECIES_METADATA_FILE = PROCESSED_DIR / "omo_species_metadata.csv"

if not SPECIES_METADATA_FILE.exists():
    raise FileNotFoundError(
        "Canonical species metadata could not be found:\n"
        f"{SPECIES_METADATA_FILE}\n\n"
        "Notebook 07B requires the validated upstream "
        "'omo_species_metadata.csv' file."
    )


species_metadata = pd.read_csv(SPECIES_METADATA_FILE)


print("Canonical species metadata loaded successfully.")
print(f"File   : {SPECIES_METADATA_FILE}")
print(f"Shape  : {species_metadata.shape}")

display(species_metadata.head(10))
```

```python
# =============================================================================
# 1.2 VALIDATE CANONICAL SPECIES INVENTORY
# =============================================================================

EXPECTED_SPECIES_COUNT = 127
SPECIES_COLUMN = "Species"


required_columns = {SPECIES_COLUMN}

missing_columns = required_columns.difference(species_metadata.columns)

if missing_columns:
    raise ValueError(
        "Canonical species metadata is missing required column(s): "
        f"{sorted(missing_columns)}"
    )


n_rows = len(species_metadata)

n_unique_species = species_metadata[SPECIES_COLUMN].nunique(dropna=True)

n_missing_species = species_metadata[SPECIES_COLUMN].isna().sum()

n_duplicate_species = species_metadata[SPECIES_COLUMN].duplicated().sum()


validation_summary = pd.DataFrame(
    {
        "Check": [
            "Total metadata rows",
            "Unique canonical species",
            "Missing species names",
            "Duplicate species names",
        ],
        "Observed": [
            n_rows,
            n_unique_species,
            n_missing_species,
            n_duplicate_species,
        ],
        "Expected": [
            EXPECTED_SPECIES_COUNT,
            EXPECTED_SPECIES_COUNT,
            0,
            0,
        ],
    }
)

validation_summary["Passed"] = (
    validation_summary["Observed"]
    == validation_summary["Expected"]
)


display(validation_summary)


if not validation_summary["Passed"].all():
    raise ValueError(
        "Canonical species metadata failed Notebook 07B validation."
    )


print(
    "\nCanonical inventory validation PASSED:\n"
    f"  Species records : {n_rows}\n"
    f"  Unique species  : {n_unique_species}\n"
    f"  Missing names   : {n_missing_species}\n"
    f"  Duplicate names : {n_duplicate_species}"
)
```

## 3. Functional-Trait Source Registry and Acquisition Hierarchy

Trait acquisition follows a trait-specific source hierarchy rather than
assuming that a single database provides the best information for all traits.

Primary sources are selected according to:

- ecological appropriateness;
- taxonomic coverage;
- measurement provenance;
- accessibility and reproducibility;
- unit documentation;
- availability of observation-level or defensibly aggregated data.

The acquisition hierarchy is:

### Wood density

1. Global Wood Density Database v2 (GWDD v2)
2. TRY Plant Trait Database
3. Peer-reviewed African or tropical forest trait sources

### Specific leaf area

1. TRY Plant Trait Database
2. Peer-reviewed functional-trait datasets

### Seed mass

1. Seed Information Database (SID; Kew/SER)
2. TRY Plant Trait Database
3. Peer-reviewed seed-trait datasets

### Maximum adult height

1. TRY Plant Trait Database
2. Peer-reviewed floras or functional-trait datasets

Raw measured plant height is not automatically interpreted as species maximum
adult height. The distinction between observation-level height and derived
species-level maximum stature is retained explicitly.

No genus-level substitution or statistical imputation is performed during
initial trait acquisition.

```python
# =============================================================================
# 3.3 VALIDATE TRAIT-SOURCE REGISTRY
# =============================================================================

expected_traits = set(
    trait_registry["Trait_Code"]
)

registered_traits = set(
    trait_source_registry["Trait_Code"]
)

missing_source_traits = expected_traits.difference(
    registered_traits
)


# -----------------------------------------------------------------------------
# Check 1: every core trait has at least one registered acquisition source
# -----------------------------------------------------------------------------

if missing_source_traits:
    raise ValueError(
        "Some core traits have no registered acquisition source: "
        f"{sorted(missing_source_traits)}"
    )


# -----------------------------------------------------------------------------
# Check 2: each core trait has exactly one PRIMARY source
# -----------------------------------------------------------------------------

primary_counts = (
    trait_source_registry
    .query("Use_Role == 'PRIMARY'")
    .groupby("Trait_Code")
    .size()
)


missing_primary_traits = expected_traits.difference(
    set(primary_counts.index)
)

if missing_primary_traits:
    raise ValueError(
        "Some core traits have no PRIMARY acquisition source: "
        f"{sorted(missing_primary_traits)}"
    )


invalid_primary_counts = primary_counts[
    primary_counts != 1
]

if not invalid_primary_counts.empty:
    raise ValueError(
        "Each core trait must have exactly one PRIMARY source.\n"
        f"Invalid counts:\n{invalid_primary_counts}"
    )


# -----------------------------------------------------------------------------
# Check 3: source priorities must be unique within each trait
# -----------------------------------------------------------------------------

duplicate_priorities = (
    trait_source_registry
    .duplicated(
        subset=["Trait_Code", "Source_Priority"],
        keep=False
    )
)

if duplicate_priorities.any():
    raise ValueError(
        "Duplicate source priorities detected within one or more traits:\n"
        f"{trait_source_registry.loc[duplicate_priorities]}"
    )


# -----------------------------------------------------------------------------
# Check 4: preferred source must correspond to PRIMARY source
# -----------------------------------------------------------------------------

preferred_primary_mismatch = trait_source_registry[
    trait_source_registry["Preferred"]
    != (trait_source_registry["Use_Role"] == "PRIMARY")
]

if not preferred_primary_mismatch.empty:
    raise ValueError(
        "Preferred-source flag is inconsistent with PRIMARY designation:\n"
        f"{preferred_primary_mismatch}"
    )


# -----------------------------------------------------------------------------
# Validation summary
# -----------------------------------------------------------------------------

print("Trait-source registry validation PASSED.")
print(f"Core traits represented : {len(registered_traits)}")
print(f"Primary sources defined : {len(primary_counts)}")
print(f"Registered source entries: {len(trait_source_registry)}")
```

```python
# =============================================================================
# 4.1 CREATE CANONICAL TAXONOMIC MATCHING FOUNDATION
# =============================================================================

canonical_taxa = (
    species_metadata[[SPECIES_COLUMN]]
    .copy()
    .rename(columns={SPECIES_COLUMN: "Canonical_Species"})
)


# -----------------------------------------------------------------------------
# Basic taxonomic components
# -----------------------------------------------------------------------------

canonical_taxa["Canonical_Species"] = (
    canonical_taxa["Canonical_Species"]
    .astype("string")
    .str.strip()
)


canonical_taxa["Canonical_Genus"] = (
    canonical_taxa["Canonical_Species"]
    .str.split()
    .str[0]
)


canonical_taxa["Canonical_Specific_Epithet"] = (
    canonical_taxa["Canonical_Species"]
    .str.split()
    .str[1]
)


# -----------------------------------------------------------------------------
# Identify records that are not explicit binomial species names
#
# Example:
#     Drypetes species
#
# Such records remain part of the authoritative Omo inventory but must not
# automatically receive species-level trait matches from external databases.
# -----------------------------------------------------------------------------

canonical_taxa["Taxonomic_Resolution"] = np.where(
    canonical_taxa["Canonical_Specific_Epithet"]
    .str.lower()
    .isin(["sp", "sp.", "spp", "spp.", "species"]),
    "GENUS_LEVEL_OR_UNRESOLVED",
    "SPECIES_LEVEL"
)


# -----------------------------------------------------------------------------
# Validation
# -----------------------------------------------------------------------------

if len(canonical_taxa) != EXPECTED_SPECIES_COUNT:
    raise ValueError(
        "Canonical taxonomic matching foundation does not contain "
        f"{EXPECTED_SPECIES_COUNT} records."
    )


if canonical_taxa["Canonical_Species"].duplicated().any():
    raise ValueError(
        "Duplicate canonical species detected while constructing "
        "the taxonomic matching foundation."
    )


print("Canonical taxonomic matching foundation created successfully.")
print(f"Canonical taxa : {len(canonical_taxa)}")

print("\nTaxonomic resolution:")
print(
    canonical_taxa["Taxonomic_Resolution"]
    .value_counts(dropna=False)
)

display(canonical_taxa.head(15))
```

```python
# =============================================================================
# 4.1a INSPECT NON-SPECIES-LEVEL CANONICAL TAXA
# =============================================================================

unresolved_canonical_taxa = canonical_taxa.loc[
    canonical_taxa["Taxonomic_Resolution"]
    == "GENUS_LEVEL_OR_UNRESOLVED"
].copy()


print(
    "Canonical records requiring non-species-level handling:"
)

display(unresolved_canonical_taxa)
```

```python
# =============================================================================
# 4.2 DEFINE TAXONOMIC NORMALIZATION FUNCTION
# =============================================================================

def normalize_taxon_name(name):
    """
    Normalize a taxon name for controlled matching only.

    Important:
    - The original source name is never overwritten.
    - This function is used only to create comparison keys.
    - It does not perform taxonomic correction or synonym resolution.

    Parameters
    ----------
    name : str or missing
        Taxon name to normalize.

    Returns
    -------
    str or pd.NA
        Normalized comparison key.
    """

    if pd.isna(name):
        return pd.NA

    name = str(name).strip().lower()

    # Replace punctuation with spaces
    name = re.sub(r"[^\w\s]", " ", name)

    # Collapse repeated whitespace
    name = re.sub(r"\s+", " ", name).strip()

    return name
```

```python
# =============================================================================
# 4.2a TEST TAXONOMIC NORMALIZATION FUNCTION
# =============================================================================

normalization_tests = pd.DataFrame(
    {
        "Original_Name": [
            "Milicia excelsa",
            "  Milicia   excelsa  ",
            "Milicia excelsa.",
            "MILICIA EXCELSA",
            "Drypetes species",
            "Cola species",
            None,
        ]
    }
)

normalization_tests["Normalized_Name"] = (
    normalization_tests["Original_Name"]
    .apply(normalize_taxon_name)
)

display(normalization_tests)
```

```python
# =============================================================================
# 4.2b ADD NORMALIZED MATCHING KEY TO CANONICAL TAXA
# =============================================================================

canonical_taxa["Canonical_Normalized"] = (
    canonical_taxa["Canonical_Species"]
    .apply(normalize_taxon_name)
)


# -----------------------------------------------------------------------------
# Validate uniqueness of normalized canonical keys
# -----------------------------------------------------------------------------

duplicate_normalized = (
    canonical_taxa["Canonical_Normalized"]
    .duplicated(keep=False)
)

if duplicate_normalized.any():
    raise ValueError(
        "Normalized canonical taxonomic keys are not unique:\n"
        f"{canonical_taxa.loc[duplicate_normalized]}"
    )


print("Canonical normalized taxonomic keys created successfully.")
print(
    f"Unique normalized keys : "
    f"{canonical_taxa['Canonical_Normalized'].nunique()}"
)

display(
    canonical_taxa[
        [
            "Canonical_Species",
            "Canonical_Normalized",
            "Taxonomic_Resolution",
        ]
    ].head(15)
)
```

```python
# =============================================================================
# 4.3 DEFINE CONTROLLED TAXONOMIC MATCHING FRAMEWORK
# =============================================================================

MATCH_STATUS_VALUES = {
    "EXACT_CANONICAL_MATCH",
    "NORMALIZED_CANONICAL_MATCH",
    "HARMONIZED_MATCH",
    "GENUS_LEVEL_CANDIDATE",
    "AMBIGUOUS",
    "UNMATCHED",
}

MATCH_METHOD_VALUES = {
    "EXACT_STRING",
    "NORMALIZED_STRING",
    "DOCUMENTED_HARMONIZATION",
    "GENUS_LEVEL_ONLY",
    "MANUAL_REVIEW_REQUIRED",
    "NO_MATCH",
}


# -----------------------------------------------------------------------------
# Matching policy
# -----------------------------------------------------------------------------

TAXONOMIC_MATCHING_POLICY = {
    "exact_match_allowed": True,
    "normalized_match_allowed": True,
    "documented_harmonized_match_allowed": True,

    # Important safeguards
    "automatic_fuzzy_matching_allowed": False,
    "automatic_genus_substitution_allowed": False,
    "automatic_synonym_acceptance_allowed": False,

    # Non-species-level Omo taxa
    "non_species_level_records": [
        "Drypetes species",
        "Cola species",
    ],

    "non_species_level_species_match_allowed": False,
}


print("Controlled taxonomic matching framework initialized.")

print("\nAllowed Match_Status values:")
for value in sorted(MATCH_STATUS_VALUES):
    print(f"  - {value}")

print("\nAllowed Match_Method values:")
for value in sorted(MATCH_METHOD_VALUES):
    print(f"  - {value}")
```

```python
# =============================================================================
# 4.3a VALIDATE CONTROLLED MATCHING FRAMEWORK
# =============================================================================

required_statuses = {
    "EXACT_CANONICAL_MATCH",
    "NORMALIZED_CANONICAL_MATCH",
    "HARMONIZED_MATCH",
    "AMBIGUOUS",
    "UNMATCHED",
}

missing_statuses = required_statuses.difference(
    MATCH_STATUS_VALUES
)

if missing_statuses:
    raise ValueError(
        "Required taxonomic match statuses are missing: "
        f"{sorted(missing_statuses)}"
    )


required_methods = {
    "EXACT_STRING",
    "NORMALIZED_STRING",
    "DOCUMENTED_HARMONIZATION",
    "NO_MATCH",
}

missing_methods = required_methods.difference(
    MATCH_METHOD_VALUES
)

if missing_methods:
    raise ValueError(
        "Required taxonomic match methods are missing: "
        f"{sorted(missing_methods)}"
    )


# Verify that the special unresolved canonical taxa actually exist
canonical_species_set = set(
    canonical_taxa["Canonical_Species"]
)

missing_unresolved_taxa = set(
    TAXONOMIC_MATCHING_POLICY["non_species_level_records"]
).difference(
    canonical_species_set
)

if missing_unresolved_taxa:
    raise ValueError(
        "Configured unresolved taxa are not present in "
        "the canonical inventory: "
        f"{sorted(missing_unresolved_taxa)}"
    )


print("Taxonomic matching framework validation PASSED.")
print(
    "Automatic fuzzy matching : "
    f"{TAXONOMIC_MATCHING_POLICY['automatic_fuzzy_matching_allowed']}"
)
print(
    "Automatic genus substitution : "
    f"{TAXONOMIC_MATCHING_POLICY['automatic_genus_substitution_allowed']}"
)
print(
    "Protected unresolved taxa : "
    f"{len(TAXONOMIC_MATCHING_POLICY['non_species_level_records'])}"
)
```

```python
# =============================================================================
# 4.4 DEFINE CONTROLLED EXTERNAL TAXON MATCHING FUNCTION
# =============================================================================

def match_external_taxon(
    source_name,
    canonical_table=canonical_taxa,
):
    """
    Match an external taxon name to the authoritative Omo canonical inventory.

    Matching order
    --------------
    1. Exact string match
    2. Normalized string match
    3. Genus-level candidate detection
    4. Otherwise unmatched

    Important safeguards
    --------------------
    - No fuzzy matching
    - No synonym resolution
    - No automatic genus substitution
    - No automatic species-level assignment to unresolved Omo taxa
    """

    result = {
        "Trait_Source_Name": source_name,
        "Canonical_Species": pd.NA,
        "Match_Status": "UNMATCHED",
        "Match_Method": "NO_MATCH",
    }

    if pd.isna(source_name):
        return result

    source_name_clean = str(source_name).strip()
    source_name_normalized = normalize_taxon_name(source_name_clean)

    # -------------------------------------------------------------------------
    # 1. Exact canonical string match
    # -------------------------------------------------------------------------

    exact_matches = canonical_table.loc[
        canonical_table["Canonical_Species"] == source_name_clean
    ]

    if len(exact_matches) == 1:

        match = exact_matches.iloc[0]

        # Protect unresolved canonical records
        if match["Taxonomic_Resolution"] == "GENUS_LEVEL_OR_UNRESOLVED":
            result.update(
                {
                    "Canonical_Species": match["Canonical_Species"],
                    "Match_Status": "GENUS_LEVEL_CANDIDATE",
                    "Match_Method": "GENUS_LEVEL_ONLY",
                }
            )
            return result

        result.update(
            {
                "Canonical_Species": match["Canonical_Species"],
                "Match_Status": "EXACT_CANONICAL_MATCH",
                "Match_Method": "EXACT_STRING",
            }
        )
        return result

    # -------------------------------------------------------------------------
    # 2. Normalized canonical string match
    # -------------------------------------------------------------------------

    normalized_matches = canonical_table.loc[
        canonical_table["Canonical_Normalized"]
        == source_name_normalized
    ]

    if len(normalized_matches) == 1:

        match = normalized_matches.iloc[0]

        # Again protect unresolved canonical records
        if match["Taxonomic_Resolution"] == "GENUS_LEVEL_OR_UNRESOLVED":
            result.update(
                {
                    "Canonical_Species": match["Canonical_Species"],
                    "Match_Status": "GENUS_LEVEL_CANDIDATE",
                    "Match_Method": "GENUS_LEVEL_ONLY",
                }
            )
            return result

        result.update(
            {
                "Canonical_Species": match["Canonical_Species"],
                "Match_Status": "NORMALIZED_CANONICAL_MATCH",
                "Match_Method": "NORMALIZED_STRING",
            }
        )
        return result

    # -------------------------------------------------------------------------
    # 3. Genus-level candidate detection
    #
    # This does NOT assign a species.
    # It only records that the external taxon belongs to a genus represented
    # by an unresolved Omo canonical record such as Drypetes species.
    # -------------------------------------------------------------------------

    source_parts = source_name_clean.split()

    if len(source_parts) >= 1:

        source_genus = source_parts[0].lower()

        unresolved_matches = canonical_table.loc[
            (
                canonical_table["Taxonomic_Resolution"]
                == "GENUS_LEVEL_OR_UNRESOLVED"
            )
            &
            (
                canonical_table["Canonical_Genus"]
                .str.lower()
                == source_genus
            )
        ]

        if len(unresolved_matches) == 1:

            result.update(
                {
                    "Canonical_Species":
                        unresolved_matches.iloc[0]["Canonical_Species"],
                    "Match_Status": "GENUS_LEVEL_CANDIDATE",
                    "Match_Method": "GENUS_LEVEL_ONLY",
                }
            )
            return result

    return result
```

```python
# =============================================================================
# 4.5 VALIDATE MATCHING ENGINE AGAINST CANONICAL INVENTORY
# =============================================================================

canonical_self_match_results = (
    canonical_taxa["Canonical_Species"]
    .apply(match_external_taxon)
    .apply(pd.Series)
)


# -----------------------------------------------------------------------------
# Attach canonical taxonomic resolution for validation
# -----------------------------------------------------------------------------

canonical_self_match_validation = canonical_taxa[
    [
        "Canonical_Species",
        "Taxonomic_Resolution",
    ]
].merge(
    canonical_self_match_results[
        [
            "Trait_Source_Name",
            "Canonical_Species",
            "Match_Status",
            "Match_Method",
        ]
    ],
    left_on="Canonical_Species",
    right_on="Trait_Source_Name",
    how="left",
    suffixes=("_Expected", "_Matched"),
)


# -----------------------------------------------------------------------------
# Expected behaviour
# -----------------------------------------------------------------------------

canonical_self_match_validation["Expected_Match_Status"] = np.where(
    canonical_self_match_validation["Taxonomic_Resolution"]
    == "GENUS_LEVEL_OR_UNRESOLVED",
    "GENUS_LEVEL_CANDIDATE",
    "EXACT_CANONICAL_MATCH",
)


canonical_self_match_validation["Validation_Passed"] = (
    canonical_self_match_validation["Match_Status"]
    == canonical_self_match_validation["Expected_Match_Status"]
)


# -----------------------------------------------------------------------------
# Summary
# -----------------------------------------------------------------------------

validation_counts = (
    canonical_self_match_validation["Match_Status"]
    .value_counts(dropna=False)
)


print("Canonical self-match validation summary:")
print(validation_counts)


failed_matches = canonical_self_match_validation.loc[
    ~canonical_self_match_validation["Validation_Passed"]
]


if not failed_matches.empty:
    raise ValueError(
        "Canonical self-match validation FAILED.\n"
        f"{failed_matches}"
    )


print("\nCanonical self-match validation PASSED.")
print(
    f"Validated canonical records : "
    f"{len(canonical_self_match_validation)}"
)
```

```python
# =============================================================================
# 5.2 VALIDATE TRAIT UNIT STANDARDIZATION REGISTRY
# =============================================================================

# -----------------------------------------------------------------------------
# Check 1: every core trait must occur in the unit registry
# -----------------------------------------------------------------------------

expected_traits = set(
    trait_registry["Trait_Code"]
)

unit_registry_traits = set(
    trait_unit_registry["Trait_Code"]
)

missing_unit_traits = expected_traits.difference(
    unit_registry_traits
)

if missing_unit_traits:
    raise ValueError(
        "Some core traits have no unit-standardization rules: "
        f"{sorted(missing_unit_traits)}"
    )


# -----------------------------------------------------------------------------
# Check 2: every trait must include its preferred standard unit
# -----------------------------------------------------------------------------

preferred_units = dict(
    zip(
        trait_registry["Trait_Code"],
        trait_registry["Preferred_Unit"],
    )
)

for trait_code, preferred_unit in preferred_units.items():

    trait_rows = trait_unit_registry.loc[
        trait_unit_registry["Trait_Code"] == trait_code
    ]

    standard_units = set(
        trait_rows["Standard_Unit"]
    )

    if standard_units != {preferred_unit}:
        raise ValueError(
            f"Inconsistent standard unit for {trait_code}.\n"
            f"Expected : {preferred_unit}\n"
            f"Observed : {sorted(standard_units)}"
        )


# -----------------------------------------------------------------------------
# Check 3: no duplicate trait/source-unit rules
# -----------------------------------------------------------------------------

duplicate_rules = trait_unit_registry.duplicated(
    subset=[
        "Trait_Code",
        "Accepted_Source_Unit",
    ],
    keep=False,
)

if duplicate_rules.any():
    raise ValueError(
        "Duplicate trait/unit conversion rules detected:\n"
        f"{trait_unit_registry.loc[duplicate_rules]}"
    )


# -----------------------------------------------------------------------------
# Check 4: every trait must include an identity conversion
#
# Example:
# WD: g cm-3 -> g cm-3
# SLA: mm2 mg-1 -> mm2 mg-1
# -----------------------------------------------------------------------------

for trait_code, preferred_unit in preferred_units.items():

    identity_rule = trait_unit_registry.loc[
        (
            trait_unit_registry["Trait_Code"]
            == trait_code
        )
        &
        (
            trait_unit_registry["Accepted_Source_Unit"]
            == preferred_unit
        )
        &
        (
            trait_unit_registry["Conversion_Factor"]
            == 1.0
        )
    ]

    if len(identity_rule) != 1:
        raise ValueError(
            f"Missing or invalid identity conversion for "
            f"{trait_code} ({preferred_unit})."
        )


# -----------------------------------------------------------------------------
# Check 5: conversion factors must be positive and finite
# -----------------------------------------------------------------------------

invalid_factors = trait_unit_registry.loc[
    (
        ~np.isfinite(
            trait_unit_registry["Conversion_Factor"]
        )
    )
    |
    (
        trait_unit_registry["Conversion_Factor"]
        <= 0
    )
]

if not invalid_factors.empty:
    raise ValueError(
        "Invalid unit-conversion factors detected:\n"
        f"{invalid_factors}"
    )


# -----------------------------------------------------------------------------
# Check 6: currently supported conversion operation
# -----------------------------------------------------------------------------

invalid_operations = trait_unit_registry.loc[
    trait_unit_registry["Conversion_Operation"]
    != "MULTIPLY"
]

if not invalid_operations.empty:
    raise ValueError(
        "Unsupported conversion operations detected:\n"
        f"{invalid_operations}"
    )


# -----------------------------------------------------------------------------
# Validation summary
# -----------------------------------------------------------------------------

print("Trait unit-standardization registry validation PASSED.")
print(
    f"Core traits represented      : "
    f"{len(unit_registry_traits)}"
)
print(
    f"Unit conversion rules        : "
    f"{len(trait_unit_registry)}"
)
print(
    f"Standard units represented   : "
    f"{trait_unit_registry['Standard_Unit'].nunique()}"
)
```

```python
# =============================================================================
# 5.3 DEFINE TRAIT UNIT STANDARDIZATION FUNCTION
# =============================================================================

def standardize_trait_value(
    trait_code,
    value,
    source_unit,
    unit_registry=trait_unit_registry,
):
    """
    Standardize a functional-trait value to the preferred unit defined
    for the analytical trait.

    Rules
    -----
    - Only explicitly registered unit conversions are allowed.
    - Missing values remain missing.
    - Missing units are not guessed.
    - Unknown units are not converted.
    - Original source values must be retained separately in the raw schema.

    Returns
    -------
    dict
        Standardized value, unit, conversion flag and QC flag.
    """

    result = {
        "Standardized_Trait_Value": pd.NA,
        "Standardized_Trait_Unit": pd.NA,
        "Unit_Conversion_Applied": False,
        "Trait_Quality_Flag": "OK",
    }

    # -------------------------------------------------------------------------
    # Missing trait value
    # -------------------------------------------------------------------------

    if pd.isna(value):
        result["Trait_Quality_Flag"] = "MISSING_VALUE"
        return result

    # -------------------------------------------------------------------------
    # Missing trait code
    # -------------------------------------------------------------------------

    if pd.isna(trait_code):
        result["Trait_Quality_Flag"] = "MISSING_TRAIT_CODE"
        return result

    # -------------------------------------------------------------------------
    # Missing source unit
    # -------------------------------------------------------------------------

    if pd.isna(source_unit):
        result["Trait_Quality_Flag"] = "MISSING_UNIT"
        return result

    trait_code = str(trait_code).strip()
    source_unit = str(source_unit).strip()

    # -------------------------------------------------------------------------
    # Locate permitted conversion rule
    # -------------------------------------------------------------------------

    rules = unit_registry.loc[
        (
            unit_registry["Trait_Code"] == trait_code
        )
        &
        (
            unit_registry["Accepted_Source_Unit"] == source_unit
        )
    ]

    if len(rules) == 0:
        result["Trait_Quality_Flag"] = "UNRECOGNIZED_UNIT"
        return result

    if len(rules) > 1:
        raise ValueError(
            "Multiple unit-standardization rules found for "
            f"{trait_code} / {source_unit}."
        )

    rule = rules.iloc[0]

    # -------------------------------------------------------------------------
    # Numeric conversion
    # -------------------------------------------------------------------------

    try:
        numeric_value = float(value)
    except (TypeError, ValueError):
        result["Trait_Quality_Flag"] = "NON_NUMERIC_VALUE"
        return result

    if not np.isfinite(numeric_value):
        result["Trait_Quality_Flag"] = "NON_FINITE_VALUE"
        return result

    operation = rule["Conversion_Operation"]
    factor = float(rule["Conversion_Factor"])

    if operation == "MULTIPLY":
        standardized_value = numeric_value * factor
    else:
        raise ValueError(
            f"Unsupported conversion operation: {operation}"
        )

    result["Standardized_Trait_Value"] = standardized_value
    result["Standardized_Trait_Unit"] = rule["Standard_Unit"]

    result["Unit_Conversion_Applied"] = (
        source_unit != rule["Standard_Unit"]
        or factor != 1.0
    )

    return result
```

```python
# =============================================================================
# 5.4 VALIDATE TRAIT UNIT STANDARDIZATION FUNCTION
# =============================================================================

validation_cases = [
    {
        "Trait_Code": "WD",
        "Value": 650,
        "Source_Unit": "kg m-3",
        "Expected_Value": 0.65,
        "Expected_Unit": "g cm-3",
        "Expected_Flag": "OK",
    },
    {
        "Trait_Code": "SLA",
        "Value": 150,
        "Source_Unit": "cm2 g-1",
        "Expected_Value": 15.0,
        "Expected_Unit": "mm2 mg-1",
        "Expected_Flag": "OK",
    },
    {
        "Trait_Code": "SEED_MASS",
        "Value": 0.25,
        "Source_Unit": "g",
        "Expected_Value": 250.0,
        "Expected_Unit": "mg",
        "Expected_Flag": "OK",
    },
    {
        "Trait_Code": "MAX_HEIGHT",
        "Value": 3000,
        "Source_Unit": "cm",
        "Expected_Value": 30.0,
        "Expected_Unit": "m",
        "Expected_Flag": "OK",
    },
]


for case in validation_cases:

    result = standardize_trait_value(
        trait_code=case["Trait_Code"],
        value=case["Value"],
        source_unit=case["Source_Unit"],
    )

    if not np.isclose(
        float(result["Standardized_Trait_Value"]),
        case["Expected_Value"],
        rtol=1e-9,
        atol=1e-12,
    ):
        raise ValueError(
            f"Incorrect standardized value for "
            f"{case['Trait_Code']}."
        )

    if result["Standardized_Trait_Unit"] != case["Expected_Unit"]:
        raise ValueError(
            f"Incorrect standardized unit for "
            f"{case['Trait_Code']}."
        )

    if result["Trait_Quality_Flag"] != case["Expected_Flag"]:
        raise ValueError(
            f"Unexpected QC flag for "
            f"{case['Trait_Code']}."
        )


# -----------------------------------------------------------------------------
# Validate defensive behaviour
# -----------------------------------------------------------------------------

defensive_cases = [
    ("WD", 0.65, None, "MISSING_UNIT"),
    ("WD", 0.65, "unknown_unit", "UNRECOGNIZED_UNIT"),
    ("SLA", None, "mm2 mg-1", "MISSING_VALUE"),
    ("SLA", "not_numeric", "mm2 mg-1", "NON_NUMERIC_VALUE"),
]


for trait_code, value, unit, expected_flag in defensive_cases:

    result = standardize_trait_value(
        trait_code=trait_code,
        value=value,
        source_unit=unit,
    )

    if result["Trait_Quality_Flag"] != expected_flag:
        raise ValueError(
            f"Defensive unit-standardization test failed for "
            f"{trait_code}: expected {expected_flag}, "
            f"got {result['Trait_Quality_Flag']}."
        )


print("Trait unit-standardization function validation PASSED.")
print(f"Validated conversion cases : {len(validation_cases)}")
print(f"Validated defensive cases  : {len(defensive_cases)}")
```

```python
# =============================================================================
# 6.1 DEFINE RAW TRAIT OBSERVATION CONSTRUCTOR
# =============================================================================

def build_raw_trait_observation(
    trait_source_name,
    trait_code,
    original_trait_value,
    original_trait_unit,
    trait_database,
    trait_source,
    source_trait_name=None,
    source_trait_id=None,
    trait_record_id=None,
    reference=None,
    observation_level=None,
    country=None,
    location=None,
    aggregation_status="RAW_OBSERVATION",
    notes=None,
):
    """
    Construct one standardized raw trait-observation record.

    This function:
    - preserves the external source name;
    - reconciles the source taxon to the Omo canonical inventory;
    - standardizes trait units where permitted;
    - preserves original values and units;
    - assigns controlled matching and QC fields;
    - does not perform genus substitution;
    - does not perform statistical imputation;
    - does not aggregate multiple biological observations.

    Returns
    -------
    dict
        One observation conforming to RAW_TRAIT_COLUMNS.
    """

    # -------------------------------------------------------------------------
    # Validate requested analytical trait
    # -------------------------------------------------------------------------

    valid_trait_codes = set(
        trait_registry["Trait_Code"]
    )

    if trait_code not in valid_trait_codes:
        raise ValueError(
            f"Unrecognized Trait_Code: {trait_code}"
        )


    trait_name_lookup = dict(
        zip(
            trait_registry["Trait_Code"],
            trait_registry["Trait_Name"],
        )
    )

    trait_name = trait_name_lookup[trait_code]


    # -------------------------------------------------------------------------
    # Taxonomic matching
    # -------------------------------------------------------------------------

    taxon_match = match_external_taxon(
        trait_source_name
    )


    # -------------------------------------------------------------------------
    # Unit standardization
    # -------------------------------------------------------------------------

    standardized = standardize_trait_value(
        trait_code=trait_code,
        value=original_trait_value,
        source_unit=original_trait_unit,
    )


    # -------------------------------------------------------------------------
    # Combine taxonomic and trait QC information
    # -------------------------------------------------------------------------

    quality_flags = []

    unit_qc = standardized["Trait_Quality_Flag"]

    if unit_qc != "OK":
        quality_flags.append(unit_qc)


    match_status = taxon_match["Match_Status"]

    if match_status == "GENUS_LEVEL_CANDIDATE":
        quality_flags.append("GENUS_LEVEL_ONLY")

    elif match_status == "AMBIGUOUS":
        quality_flags.append("AMBIGUOUS_TAXON")

    elif match_status == "UNMATCHED":
        quality_flags.append("UNMATCHED_TAXON")


    if quality_flags:
        trait_quality_flag = ";".join(
            sorted(set(quality_flags))
        )
    else:
        trait_quality_flag = "OK"


    # -------------------------------------------------------------------------
    # Construct observation record
    # -------------------------------------------------------------------------

    record = {
        "Canonical_Species":
            taxon_match["Canonical_Species"],

        "Trait_Source_Name":
            trait_source_name,

        "Trait_Code":
            trait_code,

        "Trait_Name":
            trait_name,

        "Source_Trait_Name":
            source_trait_name,

        "Source_Trait_ID":
            source_trait_id,

        "Original_Trait_Value":
            original_trait_value,

        "Original_Trait_Unit":
            original_trait_unit,

        "Standardized_Trait_Value":
            standardized["Standardized_Trait_Value"],

        "Standardized_Trait_Unit":
            standardized["Standardized_Trait_Unit"],

        "Trait_Database":
            trait_database,

        "Trait_Source":
            trait_source,

        "Trait_Record_ID":
            trait_record_id,

        "Reference":
            reference,

        "Observation_Level":
            observation_level,

        "Country":
            country,

        "Location":
            location,

        "Match_Status":
            taxon_match["Match_Status"],

        "Match_Method":
            taxon_match["Match_Method"],

        "Unit_Conversion_Applied":
            standardized["Unit_Conversion_Applied"],

        "Aggregation_Status":
            aggregation_status,

        "Trait_Quality_Flag":
            trait_quality_flag,

        "Notes":
            notes,
    }


    # -------------------------------------------------------------------------
    # Enforce exact schema
    # -------------------------------------------------------------------------

    missing_fields = set(
        RAW_TRAIT_COLUMNS
    ).difference(
        record.keys()
    )

    extra_fields = set(
        record.keys()
    ).difference(
        RAW_TRAIT_COLUMNS
    )


    if missing_fields:
        raise ValueError(
            "Raw trait record is missing required fields: "
            f"{sorted(missing_fields)}"
        )

    if extra_fields:
        raise ValueError(
            "Raw trait record contains unexpected fields: "
            f"{sorted(extra_fields)}"
        )


    # Preserve exact schema order
    record = {
        column: record[column]
        for column in RAW_TRAIT_COLUMNS
    }


    return record
```

```python
# =============================================================================
# 6.2 DEFINE RAW TRAIT DATASET VALIDATION FUNCTION
# =============================================================================

def validate_raw_trait_dataset(
    data,
    expected_columns=RAW_TRAIT_COLUMNS,
    raise_on_error=True,
):
    """
    Validate a raw functional-trait dataset against the Notebook 07B
    observation schema and controlled vocabularies.

    Parameters
    ----------
    data : pandas.DataFrame
        Raw trait observation dataset.

    expected_columns : list
        Required Notebook 07B raw trait schema.

    raise_on_error : bool
        If True, raise ValueError when validation fails.

    Returns
    -------
    pandas.DataFrame
        Validation summary.
    """

    checks = []

    # -------------------------------------------------------------------------
    # Check 1: exact schema
    # -------------------------------------------------------------------------

    observed_columns = list(data.columns)

    missing_columns = [
        col for col in expected_columns
        if col not in observed_columns
    ]

    extra_columns = [
        col for col in observed_columns
        if col not in expected_columns
    ]

    checks.append(
        {
            "Check": "Required columns present",
            "Passed": len(missing_columns) == 0,
            "Details": (
                "OK"
                if not missing_columns
                else f"Missing: {missing_columns}"
            ),
        }
    )

    checks.append(
        {
            "Check": "No unexpected columns",
            "Passed": len(extra_columns) == 0,
            "Details": (
                "OK"
                if not extra_columns
                else f"Extra: {extra_columns}"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 2: column order
    # -------------------------------------------------------------------------

    checks.append(
        {
            "Check": "Column order correct",
            "Passed": observed_columns == expected_columns,
            "Details": (
                "OK"
                if observed_columns == expected_columns
                else "Column order differs from RAW_TRAIT_COLUMNS"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 3: valid Trait_Code values
    # -------------------------------------------------------------------------

    valid_trait_codes = set(
        trait_registry["Trait_Code"]
    )

    observed_trait_codes = set(
        data["Trait_Code"].dropna()
    )

    invalid_trait_codes = (
        observed_trait_codes
        .difference(valid_trait_codes)
    )

    checks.append(
        {
            "Check": "Trait codes valid",
            "Passed": len(invalid_trait_codes) == 0,
            "Details": (
                "OK"
                if not invalid_trait_codes
                else f"Invalid: {sorted(invalid_trait_codes)}"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 4: valid Match_Status values
    # -------------------------------------------------------------------------

    observed_match_status = set(
        data["Match_Status"].dropna()
    )

    invalid_match_status = (
        observed_match_status
        .difference(MATCH_STATUS_VALUES)
    )

    checks.append(
        {
            "Check": "Match statuses valid",
            "Passed": len(invalid_match_status) == 0,
            "Details": (
                "OK"
                if not invalid_match_status
                else f"Invalid: {sorted(invalid_match_status)}"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 5: valid Match_Method values
    # -------------------------------------------------------------------------

    observed_match_methods = set(
        data["Match_Method"].dropna()
    )

    invalid_match_methods = (
        observed_match_methods
        .difference(MATCH_METHOD_VALUES)
    )

    checks.append(
        {
            "Check": "Match methods valid",
            "Passed": len(invalid_match_methods) == 0,
            "Details": (
                "OK"
                if not invalid_match_methods
                else f"Invalid: {sorted(invalid_match_methods)}"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 6: matched canonical names belong to authoritative inventory
    # -------------------------------------------------------------------------

    valid_canonical_species = set(
        canonical_taxa["Canonical_Species"]
    )

    observed_canonical_species = set(
        data["Canonical_Species"].dropna()
    )

    invalid_canonical_species = (
        observed_canonical_species
        .difference(valid_canonical_species)
    )

    checks.append(
        {
            "Check": "Canonical species valid",
            "Passed": len(invalid_canonical_species) == 0,
            "Details": (
                "OK"
                if not invalid_canonical_species
                else f"Invalid: {sorted(invalid_canonical_species)}"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 7: unmatched records must not have canonical assignment
    # -------------------------------------------------------------------------

    invalid_unmatched = data.loc[
        (data["Match_Status"] == "UNMATCHED")
        &
        (data["Canonical_Species"].notna())
    ]

    checks.append(
        {
            "Check": "Unmatched taxa unassigned",
            "Passed": invalid_unmatched.empty,
            "Details": (
                "OK"
                if invalid_unmatched.empty
                else f"{len(invalid_unmatched)} invalid record(s)"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 8: standardized values require standardized units
    # -------------------------------------------------------------------------

    invalid_standardized_units = data.loc[
        data["Standardized_Trait_Value"].notna()
        &
        data["Standardized_Trait_Unit"].isna()
    ]

    checks.append(
        {
            "Check": "Standardized values have units",
            "Passed": invalid_standardized_units.empty,
            "Details": (
                "OK"
                if invalid_standardized_units.empty
                else (
                    f"{len(invalid_standardized_units)} "
                    "invalid record(s)"
                )
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Check 9: standardized units agree with trait registry
    # -------------------------------------------------------------------------

    expected_unit_lookup = dict(
        zip(
            trait_registry["Trait_Code"],
            trait_registry["Preferred_Unit"],
        )
    )

    unit_mismatch_count = 0

    for _, row in data.iterrows():

        if pd.isna(row["Standardized_Trait_Unit"]):
            continue

        expected_unit = expected_unit_lookup.get(
            row["Trait_Code"]
        )

        if row["Standardized_Trait_Unit"] != expected_unit:
            unit_mismatch_count += 1

    checks.append(
        {
            "Check": "Standardized units correct",
            "Passed": unit_mismatch_count == 0,
            "Details": (
                "OK"
                if unit_mismatch_count == 0
                else f"{unit_mismatch_count} mismatch(es)"
            ),
        }
    )

    # -------------------------------------------------------------------------
    # Create validation summary
    # -------------------------------------------------------------------------

    validation_summary = pd.DataFrame(checks)

    validation_passed = (
        validation_summary["Passed"].all()
    )

    if raise_on_error and not validation_passed:

        failed_checks = validation_summary.loc[
            ~validation_summary["Passed"]
        ]

        raise ValueError(
            "Raw trait dataset validation FAILED:\n"
            f"{failed_checks.to_string(index=False)}"
        )

    return validation_summary
```

```python
# =============================================================================
# 6.3a DEFINE SAFE RAW TRAIT APPEND FUNCTION
# =============================================================================

def append_raw_trait_records(
    master_table,
    new_records,
    validate=True,
):
    """
    Append new raw trait observations to the master table safely.

    Parameters
    ----------
    master_table : pandas.DataFrame
        Existing Notebook 07B raw trait dataset.

    new_records : pandas.DataFrame
        New records already transformed to RAW_TRAIT_COLUMNS.

    validate : bool
        If True, validate new records and the resulting combined dataset.

    Returns
    -------
    pandas.DataFrame
        Updated master raw trait dataset.
    """

    # -------------------------------------------------------------------------
    # Schema enforcement
    # -------------------------------------------------------------------------

    if list(new_records.columns) != RAW_TRAIT_COLUMNS:
        raise ValueError(
            "New raw trait records do not conform exactly to "
            "RAW_TRAIT_COLUMNS."
        )

    if list(master_table.columns) != RAW_TRAIT_COLUMNS:
        raise ValueError(
            "Master raw trait table does not conform exactly to "
            "RAW_TRAIT_COLUMNS."
        )

    # -------------------------------------------------------------------------
    # Validate incoming records
    # -------------------------------------------------------------------------

    if validate:
        validate_raw_trait_dataset(
            new_records,
            raise_on_error=True,
        )

    # -------------------------------------------------------------------------
    # Append safely
    #
    # Avoid pandas FutureWarning when the master table is completely empty.
    # -------------------------------------------------------------------------

    if master_table.empty:

        combined = new_records.copy().reset_index(drop=True)

    elif new_records.empty:

        combined = master_table.copy().reset_index(drop=True)

    else:

        combined = pd.concat(
            [
                master_table,
                new_records,
            ],
            ignore_index=True,
        )

    # -------------------------------------------------------------------------
    # Enforce canonical column order
    # -------------------------------------------------------------------------

    combined = combined[RAW_TRAIT_COLUMNS]

    # -------------------------------------------------------------------------
    # Validate combined dataset
    # -------------------------------------------------------------------------

    if validate:
        validate_raw_trait_dataset(
            combined,
            raise_on_error=True,
        )

    return combined
```

## 7. Wood Density Acquisition and Harmonization

### 7.1 Wood Density Acquisition Protocol

Wood density (WD) is the first functional trait to be systematically acquired
for the canonical Omo Forest species pool.

The acquisition and harmonization workflow follows these principles:

1. The canonical species metadata table is the authoritative taxonomic backbone.
2. Species-level wood-density observations are preferred wherever available.
3. Source taxon names are retained separately from canonical taxon names.
4. All accepted wood-density values are standardized to **g cm⁻³**.
5. Multiple source observations are retained during acquisition and quality
   control before derivation of a representative species-level value.
6. Genus-level values may be considered only when species-level information is
   unavailable and must be explicitly identified as genus-level estimates.
7. Missing values are retained as missing until a documented estimation or
   imputation procedure is applied.
8. Every accepted value must retain sufficient provenance to identify its
   source, matching pathway, standardization procedure, and quality status.

This separation between observed, taxonomically inferred, and subsequently
imputed trait information ensures that downstream functional-diversity analyses
remain reproducible and ecologically interpretable.

```python
# =============================================================================
# Section 7.1 — Wood Density Acquisition Protocol
# =============================================================================

WD_CONFIG = {
    "trait_code": "WD",
    "trait_name": "Wood density",
    "standard_unit": "g cm-3",
    "preferred_taxonomic_level": "SPECIES",
    "allow_genus_level_candidate": True,
    "allow_unflagged_imputation": False,
}

print("Wood-density acquisition configuration")
print("=" * 72)

for key, value in WD_CONFIG.items():
    print(f"{key:32s}: {value}")
```

### 7.3 Wood-Density Source Hierarchy

Wood-density information is acquired through a controlled source hierarchy to
maximize taxonomic precision while preserving provenance.

The primary source is the **Global Wood Density Database v2 (GWDD v2)**. The
database provides taxonomically standardized records and pre-aggregated
wood-density estimates at species and genus levels.

For the present workflow, the acquisition hierarchy is:

1. GWDD v2 species-level estimate;
2. Tervuren Xylarium Wood Density Database (TWDD) species-level information;
3. GWDD v2 genus-level estimate, used only when species-level information is
   unavailable and explicitly flagged as genus-derived;
4. additional documented trait sources, where necessary; and
5. unresolved missing value.

Species-level observations always take precedence over genus-level estimates.
Genus-level information is retained as an explicit fallback rather than being
treated as an observed species trait.

No wood-density value is silently imputed.

```python
# =============================================================================
# Section 7.3 — Wood-Density Source Hierarchy
# =============================================================================

WD_SOURCE_HIERARCHY = {
    1: {
        "source_code": "GWDD_V2_SPECIES",
        "source_name": "Global Wood Density Database v2",
        "taxonomic_level": "SPECIES",
        "value_type": "MODEL_DERIVED_SPECIES_ESTIMATE",
        "priority": 1,
    },
    2: {
        "source_code": "TWDD_SPECIES",
        "source_name": "Tervuren Xylarium Wood Density Database",
        "taxonomic_level": "SPECIES",
        "value_type": "SPECIES_OBSERVATION",
        "priority": 2,
    },
    3: {
        "source_code": "GWDD_V2_GENUS",
        "source_name": "Global Wood Density Database v2",
        "taxonomic_level": "GENUS",
        "value_type": "GENUS_ESTIMATE",
        "priority": 3,
    },
    4: {
        "source_code": "SUPPLEMENTARY",
        "source_name": "Supplementary documented trait source",
        "taxonomic_level": "VARIABLE",
        "value_type": "DOCUMENTED_TRAIT_VALUE",
        "priority": 4,
    },
}

print("Wood-density source hierarchy")
print("=" * 78)

for rank, source in WD_SOURCE_HIERARCHY.items():
    print(
        f"{rank}. "
        f"{source['source_code']:<20} | "
        f"{source['taxonomic_level']:<8} | "
        f"{source['value_type']}"
    )

print("\nAcquisition rule")
print("-" * 78)
print("Species-level values take precedence over genus-level estimates.")
print("Genus-level values remain explicitly flagged.")
print("Missing values are not silently imputed.")
```

### 7.3.3 Exact Species-Level Matching Against GWDD v2

The canonical Omo Forest taxa are first matched against the GWDD v2 binomial
aggregation table using exact scientific-name matching.

At this stage:

- only exact canonical binomial matches are accepted;
- the GWDD v2 `wsg_est` field is used as the preferred wood-density estimate;
- no fuzzy taxonomic matching is performed;
- no genus-level substitution is performed; and
- unmatched taxa remain explicitly unresolved.

This conservative first pass establishes the directly supported species-level
wood-density coverage before additional taxonomic reconciliation or fallback
procedures are considered.

```python
# =============================================================================
# Section 7.3.3 — Exact Species-Level Matching Against GWDD v2
# =============================================================================

# -------------------------------------------------------------------------
# Prepare a compact GWDD species lookup table
# -------------------------------------------------------------------------

gwdd_species_lookup = (
    gwdd_binomial[
        [
            "binomial",
            "genus",
            "family",
            "nb",
            "nb_sources",
            "nb_countries",
            "nb_sites",
            "wsg_est",
        ]
    ]
    .copy()
)

# Confirm uniqueness of the aggregated binomial table
duplicate_binomials = gwdd_species_lookup["binomial"].duplicated().sum()

if duplicate_binomials != 0:
    raise ValueError(
        f"GWDD aggregated binomial table contains "
        f"{duplicate_binomials} duplicated binomial names."
    )

# -------------------------------------------------------------------------
# Exact match against canonical Omo taxa
# -------------------------------------------------------------------------

wd_exact_match = wd_acquisition.merge(
    gwdd_species_lookup,
    how="left",
    left_on="Canonical_Species",
    right_on="binomial",
    validate="one_to_one",
)

# Identify exact species-level matches
exact_mask = wd_exact_match["binomial"].notna()

# -------------------------------------------------------------------------
# Populate acquisition fields for exact matches
# -------------------------------------------------------------------------

wd_exact_match.loc[
    exact_mask, "Trait_Source_Name"
] = wd_exact_match.loc[exact_mask, "binomial"]

wd_exact_match.loc[
    exact_mask, "Raw_Trait_Value"
] = wd_exact_match.loc[exact_mask, "wsg_est"]

wd_exact_match.loc[
    exact_mask, "Raw_Trait_Unit"
] = "g cm-3"

wd_exact_match.loc[
    exact_mask, "Standardized_Trait_Value"
] = wd_exact_match.loc[exact_mask, "wsg_est"]

wd_exact_match.loc[
    exact_mask, "Standardized_Trait_Unit"
] = "g cm-3"

wd_exact_match.loc[
    exact_mask, "Match_Status"
] = "EXACT_CANONICAL_MATCH"

wd_exact_match.loc[
    exact_mask, "Taxonomic_Level"
] = "SPECIES"

wd_exact_match.loc[
    exact_mask, "Trait_Source"
] = "GWDD_V2_SPECIES"

wd_exact_match.loc[
    exact_mask, "Trait_Quality_Flag"
] = "OK"

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

n_total = len(wd_exact_match)
n_exact = int(exact_mask.sum())
n_unmatched = n_total - n_exact
coverage = 100 * n_exact / n_total

print("GWDD v2 exact species-level matching")
print("=" * 78)
print(f"Canonical Omo taxa       : {n_total}")
print(f"Exact GWDD matches       : {n_exact}")
print(f"Unmatched taxa           : {n_unmatched}")
print(f"Exact-match coverage     : {coverage:.2f}%")

print("\nMatch status")
print("-" * 78)
print(wd_exact_match["Match_Status"].value_counts(dropna=False))

print("\nWood-density summary — exact matches")
print("-" * 78)
print(
    wd_exact_match.loc[
        exact_mask,
        "Standardized_Trait_Value"
    ].astype(float).describe()
)

print("\nUnmatched canonical taxa")
print("-" * 78)

unmatched_wd_taxa = (
    wd_exact_match.loc[
        ~exact_mask,
        ["Canonical_Species"]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

display(unmatched_wd_taxa)
```

### 7.3.4 Diagnostic Classification of Unmatched GWDD Taxa

Taxa failing exact binomial matching are diagnosed before any additional
wood-density values are assigned.

The first diagnostic evaluates whether the genus implied by each canonical
Omo taxon is represented in the GWDD v2 genus aggregation table.

This distinguishes:

- taxa whose genus is already represented in GWDD and may therefore require
  species-name reconciliation;
- taxa whose apparent genus itself is absent from GWDD;
- genus-level or unresolved Omo taxa; and
- taxa requiring comparison with the project's previously established
  taxonomic harmonization rules.

No genus-level wood-density values are assigned during this diagnostic step.

```python
# =============================================================================
# Section 7.3.4 — Diagnose Unmatched GWDD Taxa
# =============================================================================

# -------------------------------------------------------------------------
# Extract apparent genus from each unmatched canonical name
# -------------------------------------------------------------------------

wd_unmatched_diagnostic = unmatched_wd_taxa.copy()

wd_unmatched_diagnostic["Apparent_Genus"] = (
    wd_unmatched_diagnostic["Canonical_Species"]
    .astype(str)
    .str.strip()
    .str.split()
    .str[0]
)

# -------------------------------------------------------------------------
# Check genus presence in GWDD
# -------------------------------------------------------------------------

gwdd_genus_set = set(
    gwdd_genus["genus"]
    .dropna()
    .astype(str)
    .str.strip()
)

wd_unmatched_diagnostic["Genus_In_GWDD"] = (
    wd_unmatched_diagnostic["Apparent_Genus"]
    .isin(gwdd_genus_set)
)

wd_unmatched_diagnostic["Diagnostic_Status"] = (
    wd_unmatched_diagnostic["Genus_In_GWDD"]
    .map({
        True: "GENUS_PRESENT_SPECIES_UNMATCHED",
        False: "GENUS_NOT_FOUND"
    })
)

# Explicitly identify unresolved taxon labels
unresolved_mask = (
    wd_unmatched_diagnostic["Canonical_Species"]
    .str.lower()
    .str.endswith(" species")
    |
    ~wd_unmatched_diagnostic["Canonical_Species"]
    .str.contains(r"\s", regex=True)
)

wd_unmatched_diagnostic.loc[
    unresolved_mask,
    "Diagnostic_Status"
] = "GENUS_LEVEL_OR_UNRESOLVED"

# -------------------------------------------------------------------------
# Summary
# -------------------------------------------------------------------------

print("GWDD unmatched-taxon diagnostic")
print("=" * 82)
print(f"Unmatched taxa examined : {len(wd_unmatched_diagnostic)}")

print("\nDiagnostic status")
print("-" * 82)
print(
    wd_unmatched_diagnostic["Diagnostic_Status"]
    .value_counts(dropna=False)
)

print("\nDetailed diagnostic table")
print("-" * 82)

display(
    wd_unmatched_diagnostic[
        [
            "Canonical_Species",
            "Apparent_Genus",
            "Genus_In_GWDD",
            "Diagnostic_Status"
        ]
    ]
    .sort_values(
        ["Diagnostic_Status", "Canonical_Species"]
    )
    .reset_index(drop=True)
)
```

### 7.3.7 Candidate Taxonomic Matching Against GWDD v2

Unmatched Omo taxon names may represent historical spellings, typographic
variants, or alternative taxonomic usage that prevents exact matching against
GWDD v2.

Candidate GWDD binomials are therefore generated using controlled string
similarity. Candidate generation is diagnostic only: no wood-density value is
assigned automatically from a fuzzy match.

Where the apparent genus is represented in GWDD, candidate searching is
restricted to that genus. Where the apparent genus itself is not recognized,
the search is performed against the complete GWDD binomial set to identify
possible genus-level spelling errors.

Candidate matches are retained with similarity scores for subsequent review
and controlled approval.

```python
# =============================================================================
# Section 7.3.7 — Generate GWDD Candidate Taxonomic Matches
# =============================================================================

from difflib import SequenceMatcher

# -------------------------------------------------------------------------
# Helper
# -------------------------------------------------------------------------

def similarity(a, b):
    return SequenceMatcher(
        None,
        str(a).lower().strip(),
        str(b).lower().strip()
    ).ratio()


gwdd_names = (
    gwdd_species_lookup["binomial"]
    .dropna()
    .astype(str)
    .unique()
)

candidate_records = []

for _, row in wd_unmatched_diagnostic.iterrows():

    omo_name = row["Canonical_Species"]
    apparent_genus = row["Apparent_Genus"]
    genus_present = row["Genus_In_GWDD"]

    # Restrict search to same genus where possible
    if genus_present:
        search_names = gwdd_species_lookup.loc[
            gwdd_species_lookup["genus"] == apparent_genus,
            "binomial"
        ].dropna().astype(str).unique()
    else:
        search_names = gwdd_names

    scored = [
        (candidate, similarity(omo_name, candidate))
        for candidate in search_names
    ]

    scored = sorted(
        scored,
        key=lambda x: x[1],
        reverse=True
    )[:3]

    for rank, (candidate, score) in enumerate(scored, start=1):

        candidate_records.append({
            "Canonical_Species": omo_name,
            "Apparent_Genus": apparent_genus,
            "Genus_In_GWDD": genus_present,
            "Candidate_Rank": rank,
            "GWDD_Candidate": candidate,
            "Similarity": round(score, 3),
        })


wd_match_candidates = pd.DataFrame(candidate_records)

# -------------------------------------------------------------------------
# Display best candidate for every unmatched taxon
# -------------------------------------------------------------------------

best_candidates = (
    wd_match_candidates
    .loc[wd_match_candidates["Candidate_Rank"] == 1]
    .sort_values(
        ["Similarity", "Canonical_Species"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

print("GWDD candidate taxonomic matching")
print("=" * 92)
print(f"Unmatched Omo taxa evaluated : {len(wd_unmatched_diagnostic)}")
print(f"Candidate records generated  : {len(wd_match_candidates)}")

print("\nBest-candidate similarity distribution")
print("-" * 92)
print(best_candidates["Similarity"].describe())

print("\nBest candidate for each unmatched taxon")
print("-" * 92)

display(
    best_candidates[
        [
            "Canonical_Species",
            "GWDD_Candidate",
            "Similarity",
            "Genus_In_GWDD"
        ]
    ]
)
```

### 7.3.8 Controlled Review of GWDD Taxonomic Candidates

String similarity is used only to generate candidate taxonomic matches and is
not treated as evidence of taxonomic equivalence.

High similarity may identify typographic or orthographic variants, but it can
also produce biologically unrelated matches. Candidate matches are therefore
classified for review rather than automatically accepted.

The review framework distinguishes:

- high-similarity candidates that warrant priority taxonomic verification;
- moderate-similarity candidates requiring manual assessment;
- low-similarity candidates that should not be accepted through string
  similarity alone; and
- genus-level or unresolved Omo taxa that require a separate fallback pathway.

No wood-density values are assigned from fuzzy candidate matches at this stage.

```python
# =============================================================================
# Section 7.3.8 — Build Controlled GWDD Candidate Review Table
# =============================================================================

wd_candidate_review = best_candidates.copy()

# -------------------------------------------------------------------------
# Initial review classification
# -------------------------------------------------------------------------

def classify_candidate(row):

    name = str(row["Canonical_Species"]).strip().lower()
    score = float(row["Similarity"])

    # Explicit unresolved/genus-level Omo taxa
    if (
        name.endswith(" species")
        or " " not in name
    ):
        return "GENUS_LEVEL_OR_UNRESOLVED"

    # Similarity is used for review priority only
    if score >= 0.90:
        return "HIGH_PRIORITY_REVIEW"

    elif score >= 0.80:
        return "MANUAL_REVIEW"

    else:
        return "LOW_CONFIDENCE"


wd_candidate_review["Review_Class"] = (
    wd_candidate_review.apply(
        classify_candidate,
        axis=1
    )
)

# No fuzzy candidate is approved automatically
wd_candidate_review["Review_Decision"] = "PENDING"

# Placeholder for later validated name
wd_candidate_review["Approved_GWDD_Name"] = pd.NA

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("Controlled GWDD candidate review table")
print("=" * 92)
print(f"Taxa requiring review : {len(wd_candidate_review)}")

print("\nReview classification")
print("-" * 92)

print(
    wd_candidate_review["Review_Class"]
    .value_counts(dropna=False)
)

print("\nHigh-priority candidates")
print("-" * 92)

display(
    wd_candidate_review.loc[
        wd_candidate_review["Review_Class"]
        == "HIGH_PRIORITY_REVIEW",
        [
            "Canonical_Species",
            "GWDD_Candidate",
            "Similarity",
            "Genus_In_GWDD",
            "Review_Decision"
        ]
    ]
    .sort_values(
        "Similarity",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nCandidates requiring additional/manual review")
print("-" * 92)

display(
    wd_candidate_review.loc[
        wd_candidate_review["Review_Class"]
        != "HIGH_PRIORITY_REVIEW",
        [
            "Canonical_Species",
            "GWDD_Candidate",
            "Similarity",
            "Genus_In_GWDD",
            "Review_Class"
        ]
    ]
    .sort_values(
        ["Review_Class", "Similarity"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)
```

### 7.3.9 Export of GWDD Taxonomic Candidate Review Table

All unmatched Omo taxa and their candidate GWDD matches are exported for
controlled taxonomic review.

The review table preserves:

- the original Omo canonical taxon string;
- the apparent genus;
- whether that genus occurs in GWDD;
- the three highest-ranked GWDD candidate names;
- their string-similarity scores;
- the initial review class; and
- fields for recording the final taxonomic decision and supporting evidence.

Candidate generation does not constitute taxonomic acceptance. Only explicitly
reviewed mappings will subsequently be permitted to contribute species-level
wood-density values.

```python
# =============================================================================
# Section 7.3.9 — Export GWDD Taxonomic Candidate Review Table
# =============================================================================

# -------------------------------------------------------------------------
# Reshape top-three candidates to one row per Omo taxon
# -------------------------------------------------------------------------

candidate_wide = (
    wd_match_candidates
    .pivot(
        index=[
            "Canonical_Species",
            "Apparent_Genus",
            "Genus_In_GWDD"
        ],
        columns="Candidate_Rank",
        values=[
            "GWDD_Candidate",
            "Similarity"
        ]
    )
)

# Flatten MultiIndex column names
candidate_wide.columns = [
    f"{field}_{rank}"
    for field, rank in candidate_wide.columns
]

candidate_wide = (
    candidate_wide
    .reset_index()
)

# -------------------------------------------------------------------------
# Add existing review classification
# -------------------------------------------------------------------------

review_class_lookup = (
    wd_candidate_review[
        [
            "Canonical_Species",
            "Review_Class"
        ]
    ]
    .drop_duplicates()
)

wd_taxonomic_review = candidate_wide.merge(
    review_class_lookup,
    how="left",
    on="Canonical_Species",
    validate="one_to_one"
)

# -------------------------------------------------------------------------
# Add explicit audit fields
# -------------------------------------------------------------------------

wd_taxonomic_review["Review_Decision"] = "PENDING"
wd_taxonomic_review["Approved_GWDD_Name"] = pd.NA
wd_taxonomic_review["Decision_Basis"] = pd.NA
wd_taxonomic_review["Taxonomic_Evidence"] = pd.NA
wd_taxonomic_review["Reviewer_Notes"] = pd.NA

# Logical ordering
review_columns = [
    "Canonical_Species",
    "Apparent_Genus",
    "Genus_In_GWDD",
    "Review_Class",
    "GWDD_Candidate_1",
    "Similarity_1",
    "GWDD_Candidate_2",
    "Similarity_2",
    "GWDD_Candidate_3",
    "Similarity_3",
    "Review_Decision",
    "Approved_GWDD_Name",
    "Decision_Basis",
    "Taxonomic_Evidence",
    "Reviewer_Notes",
]

wd_taxonomic_review = (
    wd_taxonomic_review[review_columns]
    .sort_values(
        ["Review_Class", "Similarity_1"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

# -------------------------------------------------------------------------
# Save review table
# -------------------------------------------------------------------------

WD_OUTPUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

WD_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

WD_REVIEW_FILE = (
    WD_OUTPUT_DIR
    / "wd_gwdd_taxonomic_review.csv"
)

wd_taxonomic_review.to_csv(
    WD_REVIEW_FILE,
    index=False
)

# -------------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------------

print("GWDD taxonomic review table exported")
print("=" * 92)
print(f"Rows             : {len(wd_taxonomic_review)}")
print(f"Expected taxa    : {len(wd_unmatched_diagnostic)}")
print(f"Unique Omo taxa  : {wd_taxonomic_review['Canonical_Species'].nunique()}")
print(f"Pending reviews  : {(wd_taxonomic_review['Review_Decision'] == 'PENDING').sum()}")

print(f"\nOutput file:\n{WD_REVIEW_FILE}")

print("\nReview-table preview")
print("-" * 92)

display(
    wd_taxonomic_review.head(10)
)
```

### 7.3.10 High-Priority Taxonomic Verification Batch

Candidate mappings with high string similarity are isolated for formal
taxonomic verification.

Similarity alone is not sufficient for acceptance. Each candidate must be
evaluated to determine whether it represents:

- an unambiguous orthographic or typographic correction;
- an accepted taxonomic equivalent or synonym;
- a potentially ambiguous taxonomic relationship requiring further evidence;
  or
- an incorrect candidate that must be rejected.

The original Omo taxon string remains unchanged. Any approved name is used
only as a controlled trait-source matching name and is accompanied by an
explicit decision basis and evidence trail.

```python
# =============================================================================
# Section 7.3.10 — Prepare High-Priority Taxonomic Verification Batch
# =============================================================================

wd_high_priority_review = (
    wd_taxonomic_review.loc[
        wd_taxonomic_review["Review_Class"]
        == "HIGH_PRIORITY_REVIEW"
    ]
    .copy()
    .sort_values(
        ["Similarity_1", "Canonical_Species"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

# Assign stable review IDs
wd_high_priority_review.insert(
    0,
    "WD_Review_ID",
    [
        f"WD_HP_{i:03d}"
        for i in range(1, len(wd_high_priority_review) + 1)
    ]
)

print("High-priority WD taxonomic verification batch")
print("=" * 100)
print(f"Records : {len(wd_high_priority_review)}")

print("\nCandidate similarity range")
print("-" * 100)
print(
    f"{wd_high_priority_review['Similarity_1'].min():.3f}"
    f" – "
    f"{wd_high_priority_review['Similarity_1'].max():.3f}"
)

print("\nVerification batch")
print("-" * 100)

display(
    wd_high_priority_review[
        [
            "WD_Review_ID",
            "Canonical_Species",
            "GWDD_Candidate_1",
            "Similarity_1",
            "Genus_In_GWDD",
            "Review_Decision"
        ]
    ]
)
```

### 7.3.16 Application of Verified Taxonomic Mappings to GWDD v2

The 35 independently reviewed high-priority taxonomic mappings are now applied
to taxa that failed exact GWDD binomial matching.

For these records, the original Omo taxon string remains the canonical project
identifier, while the verified botanical name is used solely as the GWDD
trait-source matching name.

Wood-density values retrieved through this pathway are explicitly distinguished
from direct exact matches using the status `HARMONIZED_CANONICAL_MATCH`.

This preserves the provenance of each trait value while increasing legitimate
species-level wood-density coverage without modifying the underlying Omo
community taxonomy.

```python
# =============================================================================
# Section 7.3.16 — Apply Verified Mappings and Recompute Species-Level Coverage
# =============================================================================

# Work from the exact-match result produced in Section 7.3.3
wd_species_matched = wd_exact_match.copy()

# -------------------------------------------------------------------------
# Add controlled trait-source matching name
# -------------------------------------------------------------------------

wd_species_matched["GWDD_Match_Name"] = (
    wd_species_matched["Canonical_Species"]
)

# Apply only the explicitly verified mappings
verified_mask = (
    wd_species_matched["Canonical_Species"]
    .isin(WD_VERIFIED_HIGH_PRIORITY)
)

wd_species_matched.loc[
    verified_mask,
    "GWDD_Match_Name"
] = (
    wd_species_matched.loc[
        verified_mask,
        "Canonical_Species"
    ]
    .map(WD_VERIFIED_HIGH_PRIORITY)
)

# -------------------------------------------------------------------------
# Build WD lookup indexed by GWDD binomial
# -------------------------------------------------------------------------

gwdd_wd_lookup = (
    gwdd_species_lookup
    .set_index("binomial")
)

# -------------------------------------------------------------------------
# Retrieve GWDD information for verified mappings
# -------------------------------------------------------------------------

for idx in wd_species_matched.index[verified_mask]:

    match_name = wd_species_matched.at[
        idx,
        "GWDD_Match_Name"
    ]

    if match_name not in gwdd_wd_lookup.index:
        raise ValueError(
            f"Verified GWDD name not found: {match_name}"
        )

    gwdd_record = gwdd_wd_lookup.loc[match_name]

    wd_species_matched.at[
        idx, "Trait_Source_Name"
    ] = match_name

    wd_species_matched.at[
        idx, "Raw_Trait_Value"
    ] = gwdd_record["wsg_est"]

    wd_species_matched.at[
        idx, "Raw_Trait_Unit"
    ] = "g cm-3"

    wd_species_matched.at[
        idx, "Standardized_Trait_Value"
    ] = gwdd_record["wsg_est"]

    wd_species_matched.at[
        idx, "Standardized_Trait_Unit"
    ] = "g cm-3"

    wd_species_matched.at[
        idx, "Match_Status"
    ] = "HARMONIZED_CANONICAL_MATCH"

    wd_species_matched.at[
        idx, "Taxonomic_Level"
    ] = "SPECIES"

    wd_species_matched.at[
        idx, "Trait_Source"
    ] = "GWDD_V2_SPECIES"

    wd_species_matched.at[
        idx, "Trait_Quality_Flag"
    ] = "OK"

# -------------------------------------------------------------------------
# Recalculate coverage
# -------------------------------------------------------------------------

species_level_mask = (
    wd_species_matched["Standardized_Trait_Value"]
    .notna()
)

n_total = len(wd_species_matched)

n_exact = (
    wd_species_matched["Match_Status"]
    == "EXACT_CANONICAL_MATCH"
).sum()

n_harmonized = (
    wd_species_matched["Match_Status"]
    == "HARMONIZED_CANONICAL_MATCH"
).sum()

n_species_level = int(species_level_mask.sum())
n_remaining = n_total - n_species_level

coverage = (
    100 * n_species_level / n_total
)

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("GWDD v2 species-level coverage after verified harmonization")
print("=" * 92)
print(f"Canonical Omo taxa        : {n_total}")
print(f"Exact species matches     : {n_exact}")
print(f"Harmonized species matches: {n_harmonized}")
print(f"Total species-level WD    : {n_species_level}")
print(f"Remaining without WD      : {n_remaining}")
print(f"Species-level coverage    : {coverage:.2f}%")

print("\nMatch status")
print("-" * 92)

print(
    wd_species_matched["Match_Status"]
    .value_counts(dropna=False)
)

print("\nWood-density summary — all species-level GWDD matches")
print("-" * 92)

print(
    wd_species_matched.loc[
        species_level_mask,
        "Standardized_Trait_Value"
    ]
    .astype(float)
    .describe()
)

# -------------------------------------------------------------------------
# Remaining unresolved taxa
# -------------------------------------------------------------------------

wd_remaining_unmatched = (
    wd_species_matched.loc[
        ~species_level_mask,
        [
            "Canonical_Species",
            "GWDD_Match_Name",
            "Match_Status"
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nRemaining taxa without species-level GWDD WD")
print("-" * 92)

display(wd_remaining_unmatched)
```

### 7.3.20 Consolidation of Secondary Taxonomic Verification

The three secondary taxonomic-review batches are consolidated into a single
audit table.

Only records with an independently verified botanical name that is also present
in the GWDD v2 binomial table are used to retrieve additional species-level
wood-density values.

Taxonomically resolved taxa that remain absent from GWDD are retained as valid
but trait-missing records. Ambiguous taxa and genus-level records remain
unresolved at the species level and are carried forward to subsequent trait
sources or explicitly flagged fallback procedures.

```python
# =============================================================================
# Section 7.3.20 — Consolidate Secondary Review and Apply GWDD Matches
# =============================================================================

# -------------------------------------------------------------------------
# Consolidate the three secondary review batches
# -------------------------------------------------------------------------

wd_secondary_review = pd.concat(
    [
        wd_secondary_batch1,
        wd_secondary_batch2,
        wd_secondary_batch3,
    ],
    ignore_index=True
)

# -------------------------------------------------------------------------
# Integrity checks
# -------------------------------------------------------------------------

if len(wd_secondary_review) != 34:
    raise ValueError(
        f"Expected 34 secondary-review records, found "
        f"{len(wd_secondary_review)}."
    )

if wd_secondary_review["Canonical_Species"].duplicated().any():

    duplicates = wd_secondary_review.loc[
        wd_secondary_review["Canonical_Species"].duplicated(keep=False),
        "Canonical_Species"
    ]

    raise ValueError(
        "Duplicate canonical taxa detected in secondary review:\n"
        + duplicates.to_string(index=False)
    )

# -------------------------------------------------------------------------
# Identify verified names capable of providing species-level GWDD WD
# -------------------------------------------------------------------------

secondary_gwdd_mask = (
    wd_secondary_review["Verified_Name"].notna()
    &
    wd_secondary_review["Present_In_GWDD"]
)

wd_secondary_gwdd_matches = (
    wd_secondary_review.loc[
        secondary_gwdd_mask
    ]
    .copy()
    .reset_index(drop=True)
)

print("Secondary taxonomic review consolidation")
print("=" * 96)
print(f"Secondary-review taxa       : {len(wd_secondary_review)}")
print(
    f"Taxonomically resolved      : "
    f"{wd_secondary_review['Verified_Name'].notna().sum()}"
)
print(
    f"Additional GWDD matches     : "
    f"{len(wd_secondary_gwdd_matches)}"
)
print(
    f"Still requiring review      : "
    f"{(wd_secondary_review['Decision'] == 'REQUIRES_FURTHER_REVIEW').sum()}"
)
print(
    f"Genus-level only            : "
    f"{(wd_secondary_review['Decision'] == 'GENUS_LEVEL_ONLY').sum()}"
)

print("\nAdditional species-level GWDD matches")
print("-" * 96)

display(
    wd_secondary_gwdd_matches[
        [
            "Canonical_Species",
            "Verified_Name",
            "Decision",
            "Present_In_GWDD",
        ]
    ]
)
```

```python
# =============================================================================
# Section 7.3.20 — Apply Secondary Verified GWDD Species Matches
# =============================================================================

wd_species_final = wd_species_matched.copy()

# -------------------------------------------------------------------------
# Controlled mapping from Omo taxon -> verified GWDD binomial
# -------------------------------------------------------------------------

WD_VERIFIED_SECONDARY_GWDD = dict(
    zip(
        wd_secondary_gwdd_matches["Canonical_Species"],
        wd_secondary_gwdd_matches["Verified_Name"],
    )
)

# -------------------------------------------------------------------------
# Apply the secondary verified mappings
# -------------------------------------------------------------------------

secondary_apply_mask = (
    wd_species_final["Canonical_Species"]
    .isin(WD_VERIFIED_SECONDARY_GWDD)
)

for idx in wd_species_final.index[secondary_apply_mask]:

    omo_name = wd_species_final.at[
        idx,
        "Canonical_Species"
    ]

    match_name = WD_VERIFIED_SECONDARY_GWDD[
        omo_name
    ]

    if match_name not in gwdd_wd_lookup.index:
        raise ValueError(
            f"Verified secondary GWDD name not found: {match_name}"
        )

    gwdd_record = gwdd_wd_lookup.loc[
        match_name
    ]

    wd_species_final.at[
        idx, "GWDD_Match_Name"
    ] = match_name

    wd_species_final.at[
        idx, "Trait_Source_Name"
    ] = match_name

    wd_species_final.at[
        idx, "Raw_Trait_Value"
    ] = gwdd_record["wsg_est"]

    wd_species_final.at[
        idx, "Raw_Trait_Unit"
    ] = "g cm-3"

    wd_species_final.at[
        idx, "Standardized_Trait_Value"
    ] = gwdd_record["wsg_est"]

    wd_species_final.at[
        idx, "Standardized_Trait_Unit"
    ] = "g cm-3"

    wd_species_final.at[
        idx, "Match_Status"
    ] = "HARMONIZED_CANONICAL_MATCH"

    wd_species_final.at[
        idx, "Taxonomic_Level"
    ] = "SPECIES"

    wd_species_final.at[
        idx, "Trait_Source"
    ] = "GWDD_V2_SPECIES"

    wd_species_final.at[
        idx, "Trait_Quality_Flag"
    ] = "OK"

# -------------------------------------------------------------------------
# Final GWDD species-level diagnostics
# -------------------------------------------------------------------------

species_level_mask_final = (
    wd_species_final["Standardized_Trait_Value"]
    .notna()
)

n_total = len(wd_species_final)

n_exact = (
    wd_species_final["Match_Status"]
    == "EXACT_CANONICAL_MATCH"
).sum()

n_harmonized = (
    wd_species_final["Match_Status"]
    == "HARMONIZED_CANONICAL_MATCH"
).sum()

n_species_level = int(
    species_level_mask_final.sum()
)

n_remaining = (
    n_total - n_species_level
)

coverage = (
    100 * n_species_level / n_total
)

print("Final GWDD v2 species-level wood-density coverage")
print("=" * 96)
print(f"Canonical Omo taxa          : {n_total}")
print(f"Exact species matches       : {n_exact}")
print(f"Harmonized species matches  : {n_harmonized}")
print(f"Total species-level WD      : {n_species_level}")
print(f"Remaining without WD        : {n_remaining}")
print(f"Species-level coverage      : {coverage:.2f}%")

print("\nMatch status")
print("-" * 96)

print(
    wd_species_final["Match_Status"]
    .value_counts(dropna=False)
)

print("\nWood-density summary")
print("-" * 96)

print(
    wd_species_final.loc[
        species_level_mask_final,
        "Standardized_Trait_Value"
    ]
    .astype(float)
    .describe()
)

# -------------------------------------------------------------------------
# Remaining taxa after all species-level GWDD acquisition
# -------------------------------------------------------------------------

wd_after_gwdd_species = (
    wd_species_final.loc[
        ~species_level_mask_final,
        [
            "Canonical_Species",
            "GWDD_Match_Name",
            "Match_Status"
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nRemaining taxa after species-level GWDD acquisition")
print("-" * 96)

display(wd_after_gwdd_species)
```

### 7.4.2 TWDD Wood-Density Variable Selection and Validation

`WD_basic` is selected as the primary TWDD wood-density variable for integration
with the Omo functional-trait database.

Basic wood density expresses oven-dry mass relative to green wood volume and is
therefore the TWDD measurement most directly compatible with the wood-specific-
gravity/basic-density values represented by GWDD `wsg_est`.

The original TWDD measurement fields (`WD_green`, `WD_air`, `WD_oven`) are not
combined directly with GWDD values. Instead, TWDD's standardized `WD_basic`
field is used while retaining the original measurement and conversion metadata
for provenance.

Before taxonomic matching, the completeness, numerical range, taxonomic rank,
and duplication structure of `WD_basic` are evaluated.

```python
# =============================================================================
# Section 7.4.2 — TWDD Wood-Density Variable Validation
# =============================================================================

twdd = pd.read_excel(
    twdd_file,
    sheet_name="TWDD"
)

# -------------------------------------------------------------------------
# Basic structural validation
# -------------------------------------------------------------------------

required_twdd_columns = [
    "species",
    "genus",
    "family",
    "status_taxonomic",
    "rank_taxonomic",
    "WD_basic",
    "region",
    "country",
    "plants_sampled",
    "n_measurement",
    "source_short",
]

missing_columns = [
    col for col in required_twdd_columns
    if col not in twdd.columns
]

if missing_columns:
    raise ValueError(
        "Required TWDD columns missing:\n"
        + "\n".join(missing_columns)
    )

# -------------------------------------------------------------------------
# Convert WD_basic explicitly to numeric
# -------------------------------------------------------------------------

twdd["WD_basic_num"] = pd.to_numeric(
    twdd["WD_basic"],
    errors="coerce"
)

n_total = len(twdd)
n_wd = int(twdd["WD_basic_num"].notna().sum())
n_missing = n_total - n_wd

print("TWDD basic wood-density validation")
print("=" * 96)
print(f"Total TWDD records          : {n_total:,}")
print(f"Records with WD_basic       : {n_wd:,}")
print(f"Records missing WD_basic    : {n_missing:,}")
print(f"WD_basic completeness       : {100 * n_wd / n_total:.2f}%")

# -------------------------------------------------------------------------
# Numerical distribution
# -------------------------------------------------------------------------

print("\nWD_basic numerical summary")
print("-" * 96)

print(
    twdd["WD_basic_num"]
    .dropna()
    .describe()
)

# -------------------------------------------------------------------------
# Taxonomic status and rank among usable records
# -------------------------------------------------------------------------

usable_twdd = twdd.loc[
    twdd["WD_basic_num"].notna()
].copy()

print("\nTaxonomic status — records with WD_basic")
print("-" * 96)

print(
    usable_twdd["status_taxonomic"]
    .value_counts(dropna=False)
    .head(20)
)

print("\nTaxonomic rank — records with WD_basic")
print("-" * 96)

print(
    usable_twdd["rank_taxonomic"]
    .value_counts(dropna=False)
    .head(20)
)

# -------------------------------------------------------------------------
# Species-level coverage
# -------------------------------------------------------------------------

twdd_species_records = usable_twdd.loc[
    usable_twdd["rank_taxonomic"]
    .astype(str)
    .str.lower()
    .eq("species")
].copy()

print("\nSpecies-level TWDD coverage")
print("-" * 96)
print(
    f"Species-level records       : "
    f"{len(twdd_species_records):,}"
)
print(
    f"Unique species names        : "
    f"{twdd_species_records['species'].nunique():,}"
)

# -------------------------------------------------------------------------
# Replication structure
# -------------------------------------------------------------------------

species_record_counts = (
    twdd_species_records
    .groupby("species")
    .size()
)

print("\nSpecies replication")
print("-" * 96)

print(species_record_counts.describe())

print(
    f"\nSpecies represented once    : "
    f"{(species_record_counts == 1).sum():,}"
)

print(
    f"Species represented >1 time : "
    f"{(species_record_counts > 1).sum():,}"
)

# -------------------------------------------------------------------------
# Plausibility check
# -------------------------------------------------------------------------

implausible_twdd = usable_twdd.loc[
    (usable_twdd["WD_basic_num"] <= 0)
    |
    (usable_twdd["WD_basic_num"] > 1.5)
]

print("\nBasic-density plausibility check")
print("-" * 96)
print(
    f"Values <= 0 or > 1.5 g cm-3 : "
    f"{len(implausible_twdd)}"
)

if len(implausible_twdd) > 0:
    display(
        implausible_twdd[
            [
                "species",
                "WD_basic_num",
                "country",
                "source_short"
            ]
        ].head(20)
    )
```

### 7.4.3 Species-Level Aggregation of TWDD Wood Density

TWDD contains multiple wood-density observations for many species. To obtain a
single reproducible species-level value for functional-trait integration,
valid `WD_basic` observations are aggregated by the taxonomically resolved TWDD
species name.

The arithmetic mean is used as the species-level wood-density estimate.
Replication and dispersion are retained through the number of records,
standard deviation, minimum, maximum, and geographic/source coverage.

Only records explicitly resolved at species rank are included in this lookup.
Genus-level records are retained separately for possible later fallback and are
not mixed with species-level estimates.

```python
# =============================================================================
# Section 7.4.3 — Build Species-Level TWDD Aggregation Table
# =============================================================================

# -------------------------------------------------------------------------
# Restrict to valid species-level records
# -------------------------------------------------------------------------

twdd_species_valid = (
    usable_twdd.loc[
        usable_twdd["rank_taxonomic"]
        .astype(str)
        .str.lower()
        .eq("species")
    ]
    .copy()
)

# Clean taxonomic names
twdd_species_valid["species"] = (
    twdd_species_valid["species"]
    .astype(str)
    .str.strip()
)

# -------------------------------------------------------------------------
# Helper for counting distinct non-missing values
# -------------------------------------------------------------------------

def n_unique_nonmissing(series):
    return series.dropna().astype(str).nunique()


# -------------------------------------------------------------------------
# Species-level aggregation
# -------------------------------------------------------------------------

twdd_species_lookup = (
    twdd_species_valid
    .groupby("species", as_index=False)
    .agg(
        WD_basic_mean=("WD_basic_num", "mean"),
        WD_basic_sd=("WD_basic_num", "std"),
        WD_basic_min=("WD_basic_num", "min"),
        WD_basic_max=("WD_basic_num", "max"),
        TWDD_record_count=("WD_basic_num", "size"),
        TWDD_country_count=("country", n_unique_nonmissing),
        TWDD_source_count=("source_short", n_unique_nonmissing),
    )
)

# -------------------------------------------------------------------------
# Add useful derived diagnostics
# -------------------------------------------------------------------------

twdd_species_lookup["WD_basic_range"] = (
    twdd_species_lookup["WD_basic_max"]
    - twdd_species_lookup["WD_basic_min"]
)

# Single-record species have undefined sample SD.
# Preserve this as NaN rather than replacing it with zero.
twdd_species_lookup = (
    twdd_species_lookup
    .sort_values("species")
    .reset_index(drop=True)
)

# -------------------------------------------------------------------------
# Integrity checks
# -------------------------------------------------------------------------

if twdd_species_lookup["species"].duplicated().any():
    raise ValueError(
        "Duplicate species detected after TWDD aggregation."
    )

if len(twdd_species_lookup) != twdd_species_valid["species"].nunique():
    raise ValueError(
        "TWDD species aggregation produced an unexpected number of rows."
    )

if twdd_species_lookup["WD_basic_mean"].isna().any():
    raise ValueError(
        "Missing species-level mean WD detected after aggregation."
    )

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("TWDD species-level aggregation")
print("=" * 96)
print(
    f"Input species-level records : "
    f"{len(twdd_species_valid):,}"
)
print(
    f"Aggregated species          : "
    f"{len(twdd_species_lookup):,}"
)
print(
    f"Single-record species       : "
    f"{(twdd_species_lookup['TWDD_record_count'] == 1).sum():,}"
)
print(
    f"Multi-record species        : "
    f"{(twdd_species_lookup['TWDD_record_count'] > 1).sum():,}"
)

print("\nAggregated WD_basic summary")
print("-" * 96)

print(
    twdd_species_lookup["WD_basic_mean"]
    .describe()
)

print("\nReplication summary")
print("-" * 96)

print(
    twdd_species_lookup["TWDD_record_count"]
    .describe()
)

print("\nSpecies with greatest replication")
print("-" * 96)

display(
    twdd_species_lookup
    .sort_values(
        "TWDD_record_count",
        ascending=False
    )
    .head(10)
)

print("\nSpecies with greatest within-species WD range")
print("-" * 96)

display(
    twdd_species_lookup
    .sort_values(
        "WD_basic_range",
        ascending=False
    )
    .head(10)
)
```

### 7.4.4 Cross-Database Compatibility of GWDD and TWDD Wood Density

Before integrating TWDD values with the existing GWDD-derived Omo trait
records, species represented independently in both databases are compared.

GWDD `wsg_est` and species-level TWDD mean `WD_basic` represent closely related
wood-specific-gravity/basic-density quantities, but differences may arise from
sampling geography, individual variation, measurement protocols, aggregation,
and database-specific estimation procedures.

Compatibility is evaluated using exact species-name overlap, Pearson and
Spearman correlations, mean and median paired differences, mean absolute error
(MAE), and root mean squared error (RMSE). Large species-specific discrepancies
are retained as diagnostic information rather than removed automatically.

```python
# =============================================================================
# Section 7.4.4 — GWDD–TWDD Cross-Database Compatibility Validation
# =============================================================================

import numpy as np
from scipy.stats import pearsonr, spearmanr

# -------------------------------------------------------------------------
# Prepare GWDD species-level comparison table
# -------------------------------------------------------------------------

gwdd_compare = (
    gwdd_species_lookup[
        ["binomial", "wsg_est"]
    ]
    .copy()
)

gwdd_compare["wsg_est"] = pd.to_numeric(
    gwdd_compare["wsg_est"],
    errors="coerce"
)

gwdd_compare = (
    gwdd_compare
    .dropna(subset=["binomial", "wsg_est"])
    .drop_duplicates(subset=["binomial"])
)

# -------------------------------------------------------------------------
# Prepare TWDD species-level comparison table
# -------------------------------------------------------------------------

twdd_compare = (
    twdd_species_lookup[
        [
            "species",
            "WD_basic_mean",
            "WD_basic_sd",
            "TWDD_record_count",
            "TWDD_country_count",
            "TWDD_source_count",
        ]
    ]
    .copy()
)

# -------------------------------------------------------------------------
# Exact species-name overlap
# -------------------------------------------------------------------------

wd_crossdb = gwdd_compare.merge(
    twdd_compare,
    left_on="binomial",
    right_on="species",
    how="inner",
    validate="one_to_one"
)

wd_crossdb["Difference_TWDD_minus_GWDD"] = (
    wd_crossdb["WD_basic_mean"]
    - wd_crossdb["wsg_est"]
)

wd_crossdb["Absolute_Difference"] = (
    wd_crossdb["Difference_TWDD_minus_GWDD"]
    .abs()
)

wd_crossdb["Squared_Difference"] = (
    wd_crossdb["Difference_TWDD_minus_GWDD"] ** 2
)

# -------------------------------------------------------------------------
# Statistical compatibility metrics
# -------------------------------------------------------------------------

n_overlap = len(wd_crossdb)

if n_overlap < 3:
    raise ValueError(
        "Insufficient exact species overlap for cross-database validation."
    )

pearson_r, pearson_p = pearsonr(
    wd_crossdb["wsg_est"],
    wd_crossdb["WD_basic_mean"]
)

spearman_rho, spearman_p = spearmanr(
    wd_crossdb["wsg_est"],
    wd_crossdb["WD_basic_mean"]
)

mean_difference = (
    wd_crossdb["Difference_TWDD_minus_GWDD"]
    .mean()
)

median_difference = (
    wd_crossdb["Difference_TWDD_minus_GWDD"]
    .median()
)

mae = (
    wd_crossdb["Absolute_Difference"]
    .mean()
)

rmse = np.sqrt(
    wd_crossdb["Squared_Difference"]
    .mean()
)

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("GWDD–TWDD cross-database compatibility")
print("=" * 100)
print(f"Exact overlapping species       : {n_overlap:,}")
print(f"Pearson r                       : {pearson_r:.4f}")
print(f"Pearson p                       : {pearson_p:.3e}")
print(f"Spearman rho                    : {spearman_rho:.4f}")
print(f"Spearman p                      : {spearman_p:.3e}")
print(f"Mean TWDD - GWDD difference     : {mean_difference:.4f} g cm-3")
print(f"Median TWDD - GWDD difference   : {median_difference:.4f} g cm-3")
print(f"Mean absolute difference (MAE)  : {mae:.4f} g cm-3")
print(f"Root mean squared error (RMSE)  : {rmse:.4f} g cm-3")

print("\nPaired-value summaries")
print("-" * 100)

display(
    wd_crossdb[
        [
            "wsg_est",
            "WD_basic_mean",
            "Difference_TWDD_minus_GWDD",
            "Absolute_Difference",
        ]
    ]
    .describe()
)

# -------------------------------------------------------------------------
# Difference thresholds
# -------------------------------------------------------------------------

print("\nAbsolute-difference thresholds")
print("-" * 100)

for threshold in [0.05, 0.10, 0.15, 0.20]:
    count = (
        wd_crossdb["Absolute_Difference"]
        > threshold
    ).sum()

    pct = 100 * count / n_overlap

    print(
        f"|difference| > {threshold:.2f} g cm-3 : "
        f"{count:,} ({pct:.2f}%)"
    )

# -------------------------------------------------------------------------
# Largest discrepancies
# -------------------------------------------------------------------------

print("\nLargest GWDD–TWDD species-level discrepancies")
print("-" * 100)

display(
    wd_crossdb[
        [
            "binomial",
            "wsg_est",
            "WD_basic_mean",
            "Difference_TWDD_minus_GWDD",
            "Absolute_Difference",
            "WD_basic_sd",
            "TWDD_record_count",
            "TWDD_country_count",
            "TWDD_source_count",
        ]
    ]
    .sort_values(
        "Absolute_Difference",
        ascending=False
    )
    .head(20)
)
```

### 7.4.5 Exact and Taxonomically Verified Matching to TWDD

The 25 Omo taxa remaining after species-level GWDD acquisition are matched
against the aggregated TWDD species table.

Two controlled matching pathways are evaluated:

1. direct exact matching of the Omo canonical taxon string; and
2. exact matching through a botanically verified name established during the
   secondary taxonomic review.

Verified names are used only where an explicit taxonomic decision has already
been recorded. No new fuzzy taxonomic substitutions are introduced during this
step.

Successful TWDD matches retain the original Omo canonical name while recording
the TWDD species name and species-level wood-density estimate separately.

```python
# =============================================================================
# Section 7.4.5 — Exact TWDD Matching of Remaining Omo Taxa
# =============================================================================

# -------------------------------------------------------------------------
# Build TWDD species lookup
# -------------------------------------------------------------------------

twdd_species_set = set(
    twdd_species_lookup["species"]
    .dropna()
    .astype(str)
    .str.strip()
)

twdd_lookup_indexed = (
    twdd_species_lookup
    .set_index("species")
)

# -------------------------------------------------------------------------
# Build verified-name lookup from completed secondary taxonomic review
# -------------------------------------------------------------------------

verified_secondary_lookup = (
    wd_secondary_review.loc[
        wd_secondary_review["Verified_Name"].notna(),
        [
            "Canonical_Species",
            "Verified_Name",
            "Decision",
        ]
    ]
    .drop_duplicates(subset=["Canonical_Species"])
    .set_index("Canonical_Species")
)

# -------------------------------------------------------------------------
# Prepare matching records
# -------------------------------------------------------------------------

twdd_match_records = []

for canonical_name in wd_twdd_target["Canonical_Species"]:

    canonical_name = str(canonical_name).strip()

    # Default: try original Omo canonical name
    match_name = canonical_name
    match_pathway = "DIRECT_CANONICAL"

    # Retrieve independently verified botanical name if available
    if canonical_name in verified_secondary_lookup.index:

        verified_name = verified_secondary_lookup.at[
            canonical_name,
            "Verified_Name"
        ]

        if pd.notna(verified_name):
            verified_name = str(verified_name).strip()

            # Prefer verified taxonomic identity for source matching
            match_name = verified_name
            match_pathway = "VERIFIED_TAXONOMIC_NAME"

    # Test exact TWDD species-level presence
    present = match_name in twdd_species_set

    record = {
        "Canonical_Species": canonical_name,
        "TWDD_Match_Name": match_name,
        "TWDD_Match_Pathway": match_pathway,
        "TWDD_Exact_Match": present,
        "WD_basic_mean": np.nan,
        "WD_basic_sd": np.nan,
        "WD_basic_min": np.nan,
        "WD_basic_max": np.nan,
        "TWDD_record_count": np.nan,
        "TWDD_country_count": np.nan,
        "TWDD_source_count": np.nan,
    }

    if present:

        twdd_record = twdd_lookup_indexed.loc[
            match_name
        ]

        for field in [
            "WD_basic_mean",
            "WD_basic_sd",
            "WD_basic_min",
            "WD_basic_max",
            "TWDD_record_count",
            "TWDD_country_count",
            "TWDD_source_count",
        ]:
            record[field] = twdd_record[field]

    twdd_match_records.append(record)


wd_twdd_exact = pd.DataFrame(
    twdd_match_records
)

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

n_twdd_exact = int(
    wd_twdd_exact["TWDD_Exact_Match"].sum()
)

n_twdd_remaining = (
    len(wd_twdd_exact) - n_twdd_exact
)

print("TWDD species-level exact matching")
print("=" * 100)
print(f"Omo taxa evaluated           : {len(wd_twdd_exact)}")
print(f"Exact TWDD species matches   : {n_twdd_exact}")
print(f"Remaining unmatched          : {n_twdd_remaining}")
print(
    f"Recovery among TWDD targets  : "
    f"{100 * n_twdd_exact / len(wd_twdd_exact):.2f}%"
)

print("\nMatch pathway")
print("-" * 100)

print(
    wd_twdd_exact.loc[
        wd_twdd_exact["TWDD_Exact_Match"],
        "TWDD_Match_Pathway"
    ]
    .value_counts(dropna=False)
)

print("\nMatched taxa")
print("-" * 100)

display(
    wd_twdd_exact.loc[
        wd_twdd_exact["TWDD_Exact_Match"],
        [
            "Canonical_Species",
            "TWDD_Match_Name",
            "TWDD_Match_Pathway",
            "WD_basic_mean",
            "WD_basic_sd",
            "TWDD_record_count",
            "TWDD_country_count",
            "TWDD_source_count",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nRemaining unmatched taxa")
print("-" * 100)

display(
    wd_twdd_exact.loc[
        ~wd_twdd_exact["TWDD_Exact_Match"],
        [
            "Canonical_Species",
            "TWDD_Match_Name",
            "TWDD_Match_Pathway",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)
```

### 7.4.6 Integration of Verified TWDD Species-Level Wood Density

Exact species-level TWDD matches are incorporated only for taxa that remained
without wood-density values after GWDD acquisition.

TWDD does not overwrite existing GWDD values. The established source hierarchy
is therefore:

1. GWDD exact species-level match;
2. GWDD taxonomically harmonized species-level match;
3. TWDD verified species-level match.

For TWDD-derived records, the aggregated mean `WD_basic` value is stored in the
common wood-density unit of g cm-3, while database identity, matched taxon,
replication, dispersion, and geographic/source coverage are retained for
provenance.

Taxa still lacking species-level wood density are carried forward explicitly
for assessment of controlled genus-level fallback.

```python
# =============================================================================
# Section 7.4.6 — Integrate Verified TWDD Species-Level Matches
# =============================================================================

wd_species_twdd = wd_species_final.copy()

# -------------------------------------------------------------------------
# Extract successful TWDD matches
# -------------------------------------------------------------------------

twdd_success = (
    wd_twdd_exact.loc[
        wd_twdd_exact["TWDD_Exact_Match"]
    ]
    .copy()
    .reset_index(drop=True)
)

# Safety check: TWDD must only fill currently missing WD values
already_filled = wd_species_twdd.loc[
    wd_species_twdd["Canonical_Species"].isin(
        twdd_success["Canonical_Species"]
    )
    &
    wd_species_twdd["Standardized_Trait_Value"].notna()
]

if len(already_filled) > 0:
    raise ValueError(
        "TWDD attempted to overwrite existing species-level WD values:\n"
        + already_filled["Canonical_Species"].to_string(index=False)
    )

# -------------------------------------------------------------------------
# Add TWDD provenance columns if not already present
# -------------------------------------------------------------------------

twdd_provenance_columns = [
    "TWDD_Record_Count",
    "TWDD_WD_SD",
    "TWDD_WD_Min",
    "TWDD_WD_Max",
    "TWDD_Country_Count",
    "TWDD_Source_Count",
]

for col in twdd_provenance_columns:
    if col not in wd_species_twdd.columns:
        wd_species_twdd[col] = np.nan

# -------------------------------------------------------------------------
# Apply successful TWDD species-level records
# -------------------------------------------------------------------------

for _, row in twdd_success.iterrows():

    canonical_name = row["Canonical_Species"]

    mask = (
        wd_species_twdd["Canonical_Species"]
        == canonical_name
    )

    if mask.sum() != 1:
        raise ValueError(
            f"Expected one Omo record for {canonical_name}; "
            f"found {mask.sum()}."
        )

    wd_species_twdd.loc[
        mask, "GWDD_Match_Name"
    ] = row["TWDD_Match_Name"]

    wd_species_twdd.loc[
        mask, "Trait_Source_Name"
    ] = row["TWDD_Match_Name"]

    wd_species_twdd.loc[
        mask, "Raw_Trait_Value"
    ] = row["WD_basic_mean"]

    wd_species_twdd.loc[
        mask, "Raw_Trait_Unit"
    ] = "g cm-3"

    wd_species_twdd.loc[
        mask, "Standardized_Trait_Value"
    ] = row["WD_basic_mean"]

    wd_species_twdd.loc[
        mask, "Standardized_Trait_Unit"
    ] = "g cm-3"

    wd_species_twdd.loc[
        mask, "Match_Status"
    ] = "TWDD_SPECIES_MATCH"

    wd_species_twdd.loc[
        mask, "Taxonomic_Level"
    ] = "SPECIES"

    wd_species_twdd.loc[
        mask, "Trait_Source"
    ] = "TWDD_SPECIES"

    wd_species_twdd.loc[
        mask, "Trait_Quality_Flag"
    ] = "OK"

    # TWDD-specific provenance
    wd_species_twdd.loc[
        mask, "TWDD_Record_Count"
    ] = row["TWDD_record_count"]

    wd_species_twdd.loc[
        mask, "TWDD_WD_SD"
    ] = row["WD_basic_sd"]

    wd_species_twdd.loc[
        mask, "TWDD_WD_Min"
    ] = row["WD_basic_min"]

    wd_species_twdd.loc[
        mask, "TWDD_WD_Max"
    ] = row["WD_basic_max"]

    wd_species_twdd.loc[
        mask, "TWDD_Country_Count"
    ] = row["TWDD_country_count"]

    wd_species_twdd.loc[
        mask, "TWDD_Source_Count"
    ] = row["TWDD_source_count"]


# -------------------------------------------------------------------------
# Coverage diagnostics
# -------------------------------------------------------------------------

wd_available_mask = (
    wd_species_twdd["Standardized_Trait_Value"]
    .notna()
)

n_total = len(wd_species_twdd)
n_available = int(wd_available_mask.sum())
n_missing = n_total - n_available

print("Wood-density coverage after TWDD species-level integration")
print("=" * 100)
print(f"Canonical Omo taxa          : {n_total}")
print(f"WD available               : {n_available}")
print(f"WD still missing           : {n_missing}")
print(
    f"Species-level WD coverage  : "
    f"{100 * n_available / n_total:.2f}%"
)

print("\nMatch status")
print("-" * 100)

print(
    wd_species_twdd["Match_Status"]
    .value_counts(dropna=False)
)

print("\nTrait source")
print("-" * 100)

print(
    wd_species_twdd.loc[
        wd_available_mask,
        "Trait_Source"
    ]
    .value_counts(dropna=False)
)

print("\nTWDD additions")
print("-" * 100)

display(
    wd_species_twdd.loc[
        wd_species_twdd["Trait_Source"]
        .eq("TWDD_SPECIES"),
        [
            "Canonical_Species",
            "Trait_Source_Name",
            "Standardized_Trait_Value",
            "Standardized_Trait_Unit",
            "TWDD_Record_Count",
            "TWDD_WD_SD",
            "TWDD_Country_Count",
            "TWDD_Source_Count",
            "Trait_Quality_Flag",
        ]
    ]
)

# -------------------------------------------------------------------------
# Freeze taxa still lacking species-level WD
# -------------------------------------------------------------------------

wd_genus_target = (
    wd_species_twdd.loc[
        wd_species_twdd["Standardized_Trait_Value"]
        .isna(),
        ["Canonical_Species"]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nTaxa remaining after GWDD + TWDD species-level acquisition")
print("-" * 100)

display(wd_genus_target)
```

## 7.5 Controlled Genus-Level Wood-Density Fallback

Following species-level acquisition from GWDD v2 and TWDD, 24 Omo taxa remain
without wood-density values.

Genus-level estimates are considered only as an explicitly flagged fallback.
They do not have the same evidential status as species-level observations and
are not treated as direct trait measurements.

Before any genus-level value is assigned, each unresolved taxon is evaluated
to determine:

1. its most defensible botanical genus;
2. whether that genus is represented in GWDD;
3. whether species-level TWDD records are available for that genus; and
4. whether unresolved taxonomic identity makes genus-level estimation
   inappropriate.

Verified taxonomic decisions established earlier in Notebook 07B are reused
where available. No new species-level equivalence is inferred during this
step.

```python
# =============================================================================
# Section 7.5.1 — Construct Controlled Genus-Fallback Diagnostic
# =============================================================================

# -------------------------------------------------------------------------
# Build GWDD genus lookup
# -------------------------------------------------------------------------

gwdd_genus_lookup = (
    gwdd_genus[
        [
            "genus",
            "wsg_est",
            "nb",
            "nb_sources",
            "nb_countries",
            "nb_sites",
        ]
    ]
    .copy()
)

gwdd_genus_lookup["genus"] = (
    gwdd_genus_lookup["genus"]
    .astype(str)
    .str.strip()
)

gwdd_genus_lookup["wsg_est"] = pd.to_numeric(
    gwdd_genus_lookup["wsg_est"],
    errors="coerce"
)

gwdd_genus_lookup = (
    gwdd_genus_lookup
    .dropna(subset=["genus", "wsg_est"])
    .drop_duplicates(subset=["genus"])
    .set_index("genus")
)

# -------------------------------------------------------------------------
# Build TWDD genus summaries FROM species-level records
#
# This is deliberately independent of TWDD's genus-rank records.
# It answers: what is the distribution among identified species belonging
# to this genus?
# -------------------------------------------------------------------------

twdd_genus_from_species = (
    twdd_species_valid
    .dropna(subset=["genus", "WD_basic_num"])
    .copy()
)

twdd_genus_from_species["genus"] = (
    twdd_genus_from_species["genus"]
    .astype(str)
    .str.strip()
)

twdd_genus_summary = (
    twdd_genus_from_species
    .groupby("genus", as_index=False)
    .agg(
        TWDD_genus_WD_mean=("WD_basic_num", "mean"),
        TWDD_genus_WD_sd=("WD_basic_num", "std"),
        TWDD_genus_WD_min=("WD_basic_num", "min"),
        TWDD_genus_WD_max=("WD_basic_num", "max"),
        TWDD_genus_records=("WD_basic_num", "size"),
        TWDD_genus_species=("species", "nunique"),
        TWDD_genus_countries=("country", n_unique_nonmissing),
        TWDD_genus_sources=("source_short", n_unique_nonmissing),
    )
    .set_index("genus")
)

# -------------------------------------------------------------------------
# Resolve candidate genus for each remaining Omo taxon
# -------------------------------------------------------------------------

genus_diagnostic_records = []

for canonical_name in wd_genus_target["Canonical_Species"]:

    canonical_name = str(canonical_name).strip()

    apparent_genus = canonical_name.split()[0]

    verified_name = pd.NA
    verified_decision = pd.NA
    fallback_genus = apparent_genus
    genus_basis = "CANONICAL_NAME"

    # Use prior verified taxonomic decision where available
    if canonical_name in verified_secondary_lookup.index:

        verified_name = verified_secondary_lookup.at[
            canonical_name,
            "Verified_Name"
        ]

        verified_decision = verified_secondary_lookup.at[
            canonical_name,
            "Decision"
        ]

        if pd.notna(verified_name):

            verified_name = str(verified_name).strip()

            fallback_genus = verified_name.split()[0]
            genus_basis = "VERIFIED_TAXONOMIC_NAME"

    # Explicit genus-level source records
    if canonical_name in ["Drypetes", "Drypetes species", "Cola species"]:
        fallback_genus = apparent_genus
        genus_basis = "SOURCE_GENUS_LEVEL"

    # Database availability
    in_gwdd_genus = fallback_genus in gwdd_genus_lookup.index
    in_twdd_genus = fallback_genus in twdd_genus_summary.index

    record = {
        "Canonical_Species": canonical_name,
        "Verified_Name": verified_name,
        "Verified_Decision": verified_decision,
        "Fallback_Genus": fallback_genus,
        "Genus_Basis": genus_basis,
        "Genus_In_GWDD": in_gwdd_genus,
        "Genus_In_TWDD": in_twdd_genus,
    }

    # GWDD diagnostics
    if in_gwdd_genus:

        gw = gwdd_genus_lookup.loc[fallback_genus]

        record.update({
            "GWDD_Genus_WD": gw["wsg_est"],
            "GWDD_Genus_N": gw["nb"],
            "GWDD_Genus_Sources": gw["nb_sources"],
            "GWDD_Genus_Countries": gw["nb_countries"],
            "GWDD_Genus_Sites": gw["nb_sites"],
        })

    else:

        record.update({
            "GWDD_Genus_WD": np.nan,
            "GWDD_Genus_N": np.nan,
            "GWDD_Genus_Sources": np.nan,
            "GWDD_Genus_Countries": np.nan,
            "GWDD_Genus_Sites": np.nan,
        })

    # TWDD diagnostics
    if in_twdd_genus:

        tw = twdd_genus_summary.loc[fallback_genus]

        record.update({
            "TWDD_Genus_WD": tw["TWDD_genus_WD_mean"],
            "TWDD_Genus_SD": tw["TWDD_genus_WD_sd"],
            "TWDD_Genus_Records": tw["TWDD_genus_records"],
            "TWDD_Genus_Species": tw["TWDD_genus_species"],
            "TWDD_Genus_Countries": tw["TWDD_genus_countries"],
            "TWDD_Genus_Sources": tw["TWDD_genus_sources"],
        })

    else:

        record.update({
            "TWDD_Genus_WD": np.nan,
            "TWDD_Genus_SD": np.nan,
            "TWDD_Genus_Records": np.nan,
            "TWDD_Genus_Species": np.nan,
            "TWDD_Genus_Countries": np.nan,
            "TWDD_Genus_Sources": np.nan,
        })

    genus_diagnostic_records.append(record)


wd_genus_diagnostic = pd.DataFrame(
    genus_diagnostic_records
)

# -------------------------------------------------------------------------
# Classify database support — still NO imputation
# -------------------------------------------------------------------------

def classify_genus_support(row):

    gwdd = bool(row["Genus_In_GWDD"])
    twdd = bool(row["Genus_In_TWDD"])

    if gwdd and twdd:
        return "GWDD_AND_TWDD"

    if gwdd:
        return "GWDD_ONLY"

    if twdd:
        return "TWDD_ONLY"

    return "NO_GENUS_DATA"


wd_genus_diagnostic["Genus_Data_Support"] = (
    wd_genus_diagnostic.apply(
        classify_genus_support,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("Controlled genus-level WD fallback diagnostic")
print("=" * 110)
print(f"Taxa evaluated             : {len(wd_genus_diagnostic)}")

print("\nGenus basis")
print("-" * 110)

print(
    wd_genus_diagnostic["Genus_Basis"]
    .value_counts(dropna=False)
)

print("\nDatabase support")
print("-" * 110)

print(
    wd_genus_diagnostic["Genus_Data_Support"]
    .value_counts(dropna=False)
)

print("\nDetailed genus-fallback diagnostic")
print("-" * 110)

display(
    wd_genus_diagnostic[
        [
            "Canonical_Species",
            "Verified_Name",
            "Fallback_Genus",
            "Genus_Basis",
            "Genus_Data_Support",
            "GWDD_Genus_WD",
            "GWDD_Genus_N",
            "TWDD_Genus_WD",
            "TWDD_Genus_SD",
            "TWDD_Genus_Records",
            "TWDD_Genus_Species",
        ]
    ]
    .sort_values(
        [
            "Genus_Data_Support",
            "Canonical_Species"
        ]
    )
    .reset_index(drop=True)
)
```

### 7.5.2 Genus-Fallback Eligibility and Quality Classification

Genus-level fallback is classified according to both taxonomic confidence and
database support.

GWDD genus-level `wsg_est` is retained as the primary genus estimator where
available because GWDD is the primary wood-density source used in this
workflow and provides an explicitly aggregated genus estimate. TWDD genus
summaries are used as independent corroborating evidence rather than averaged
with GWDD.

A genus-level estimate is never treated as equivalent to a species-level
measurement. All genus-derived values will therefore receive an explicit
imputation status and quality flag.

Taxa for which the genus itself remains uncertain are withheld from imputation
until genus identity has been reviewed.

```python
# =============================================================================
# Section 7.5.2 — Genus-Fallback Eligibility and Quality Classification
# =============================================================================

wd_genus_eligibility = wd_genus_diagnostic.copy()

# -------------------------------------------------------------------------
# Controlled genus correction already defensible from prior review
# -------------------------------------------------------------------------
#
# Diospyrous melistoformis:
# Species identity remains uncertain, but "Diospyrous" is an evident
# orthographic corruption of the genus Diospyros.
#
# This correction is GENUS-LEVEL ONLY and must not be interpreted as
# resolving the species epithet.
# -------------------------------------------------------------------------

mask_diospyrous = (
    wd_genus_eligibility["Canonical_Species"]
    .eq("Diospyrous melistoformis")
)

wd_genus_eligibility.loc[
    mask_diospyrous,
    "Fallback_Genus"
] = "Diospyros"

wd_genus_eligibility.loc[
    mask_diospyrous,
    "Genus_Basis"
] = "VERIFIED_GENUS_ORTHOGRAPHY"

# -------------------------------------------------------------------------
# Refresh GWDD and TWDD genus information for corrected genus
# -------------------------------------------------------------------------

for idx in wd_genus_eligibility.index[mask_diospyrous]:

    genus = wd_genus_eligibility.at[
        idx,
        "Fallback_Genus"
    ]

    in_gwdd = genus in gwdd_genus_lookup.index
    in_twdd = genus in twdd_genus_summary.index

    wd_genus_eligibility.at[
        idx, "Genus_In_GWDD"
    ] = in_gwdd

    wd_genus_eligibility.at[
        idx, "Genus_In_TWDD"
    ] = in_twdd

    # ---------------------------------------------------------------------
    # GWDD genus information
    # ---------------------------------------------------------------------

    if in_gwdd:

        gw = gwdd_genus_lookup.loc[genus]

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_WD"
        ] = gw["wsg_est"]

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_N"
        ] = gw["nb"]

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_Sources"
        ] = gw["nb_sources"]

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_Countries"
        ] = gw["nb_countries"]

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_Sites"
        ] = gw["nb_sites"]

    else:

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_WD"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_N"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_Sources"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_Countries"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "GWDD_Genus_Sites"
        ] = np.nan

    # ---------------------------------------------------------------------
    # TWDD genus information
    # ---------------------------------------------------------------------

    if in_twdd:

        tw = twdd_genus_summary.loc[genus]

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_WD"
        ] = tw["TWDD_genus_WD_mean"]

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_SD"
        ] = tw["TWDD_genus_WD_sd"]

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Records"
        ] = tw["TWDD_genus_records"]

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Species"
        ] = tw["TWDD_genus_species"]

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Countries"
        ] = tw["TWDD_genus_countries"]

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Sources"
        ] = tw["TWDD_genus_sources"]

    else:

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_WD"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_SD"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Records"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Species"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Countries"
        ] = np.nan

        wd_genus_eligibility.at[
            idx, "TWDD_Genus_Sources"
        ] = np.nan


# -------------------------------------------------------------------------
# Recalculate database-support classification
# -------------------------------------------------------------------------

wd_genus_eligibility["Genus_Data_Support"] = (
    wd_genus_eligibility.apply(
        classify_genus_support,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Difference between independent GWDD and TWDD genus estimates
# -------------------------------------------------------------------------

wd_genus_eligibility[
    "GWDD_TWDD_Genus_Difference"
] = (
    wd_genus_eligibility["TWDD_Genus_WD"]
    - wd_genus_eligibility["GWDD_Genus_WD"]
)

wd_genus_eligibility[
    "GWDD_TWDD_Genus_Abs_Difference"
] = (
    wd_genus_eligibility[
        "GWDD_TWDD_Genus_Difference"
    ]
    .abs()
)

# -------------------------------------------------------------------------
# Taxa whose apparent genus is still unresolved
# -------------------------------------------------------------------------

GENUS_REVIEW_TAXA = {
    "Scotellia coriacea",
    "Scotellia coriaceae",
    "Scotellia coricea",
}

# -------------------------------------------------------------------------
# Eligibility classifier
# -------------------------------------------------------------------------

def classify_genus_eligibility(row):

    taxon = row["Canonical_Species"]
    support = row["Genus_Data_Support"]

    # Genus identity itself still requires review
    if taxon in GENUS_REVIEW_TAXA:
        return "GENUS_REVIEW_REQUIRED"

    # No usable genus WD information
    if support == "NO_GENUS_DATA":
        return "NO_GENUS_DATA"

    # Both independent databases support the genus
    if support == "GWDD_AND_TWDD":

        difference = row[
            "GWDD_TWDD_Genus_Abs_Difference"
        ]

        # Strong corroboration when estimates agree closely
        if (
            pd.notna(difference)
            and difference <= 0.10
        ):
            return "HIGH_CONFIDENCE_GENUS"

        return "MODERATE_CONFIDENCE_GENUS"

    # A single database can still support genus fallback,
    # but confidence is lower
    if support in {
        "GWDD_ONLY",
        "TWDD_ONLY",
    }:
        return "MODERATE_CONFIDENCE_GENUS"

    return "GENUS_REVIEW_REQUIRED"


wd_genus_eligibility["Genus_Eligibility"] = (
    wd_genus_eligibility.apply(
        classify_genus_eligibility,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Proposed fallback source
#
# GWDD remains primary where available.
# TWDD is used only if GWDD genus WD is unavailable.
#
# NO VALUES ARE APPLIED TO THE MASTER WD TABLE IN THIS SECTION.
# -------------------------------------------------------------------------

def proposed_genus_source(row):

    eligibility = row["Genus_Eligibility"]

    if eligibility not in {
        "HIGH_CONFIDENCE_GENUS",
        "MODERATE_CONFIDENCE_GENUS",
    }:
        return pd.NA

    if pd.notna(row["GWDD_Genus_WD"]):
        return "GWDD_V2_GENUS"

    if pd.notna(row["TWDD_Genus_WD"]):
        return "TWDD_GENUS"

    return pd.NA


wd_genus_eligibility[
    "Proposed_Genus_Source"
] = (
    wd_genus_eligibility.apply(
        proposed_genus_source,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Proposed fallback value — NA-safe version
# -------------------------------------------------------------------------

def proposed_genus_value(row):

    source = row["Proposed_Genus_Source"]

    # pd.NA cannot be evaluated directly in a boolean comparison
    if pd.isna(source):
        return np.nan

    if source == "GWDD_V2_GENUS":
        return row["GWDD_Genus_WD"]

    if source == "TWDD_GENUS":
        return row["TWDD_Genus_WD"]

    return np.nan


wd_genus_eligibility[
    "Proposed_Genus_WD"
] = (
    wd_genus_eligibility.apply(
        proposed_genus_value,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Proposed quality flag
# -------------------------------------------------------------------------

def proposed_genus_quality(row):

    eligibility = row["Genus_Eligibility"]

    if eligibility == "HIGH_CONFIDENCE_GENUS":
        return "GENUS_ESTIMATE_HIGH_CONFIDENCE"

    if eligibility == "MODERATE_CONFIDENCE_GENUS":
        return "GENUS_ESTIMATE_MODERATE_CONFIDENCE"

    if eligibility == "GENUS_REVIEW_REQUIRED":
        return "GENUS_REVIEW_REQUIRED"

    if eligibility == "NO_GENUS_DATA":
        return "NO_GENUS_DATA"

    return "UNCLASSIFIED"


wd_genus_eligibility[
    "Proposed_Quality_Flag"
] = (
    wd_genus_eligibility.apply(
        proposed_genus_quality,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Integrity checks
# -------------------------------------------------------------------------

if len(wd_genus_eligibility) != 24:
    raise ValueError(
        f"Expected 24 genus-fallback taxa, found "
        f"{len(wd_genus_eligibility)}."
    )

if wd_genus_eligibility[
    "Canonical_Species"
].duplicated().any():

    duplicates = wd_genus_eligibility.loc[
        wd_genus_eligibility[
            "Canonical_Species"
        ].duplicated(keep=False),
        "Canonical_Species"
    ]

    raise ValueError(
        "Duplicate taxa detected in genus eligibility table:\n"
        + duplicates.to_string(index=False)
    )

eligible_mask = (
    wd_genus_eligibility["Genus_Eligibility"]
    .isin(
        [
            "HIGH_CONFIDENCE_GENUS",
            "MODERATE_CONFIDENCE_GENUS",
        ]
    )
)

# Every eligible record must have a proposed source and value
if wd_genus_eligibility.loc[
    eligible_mask,
    "Proposed_Genus_Source"
].isna().any():

    raise ValueError(
        "At least one genus-eligible taxon has no proposed source."
    )

if wd_genus_eligibility.loc[
    eligible_mask,
    "Proposed_Genus_WD"
].isna().any():

    raise ValueError(
        "At least one genus-eligible taxon has no proposed WD value."
    )

# Non-eligible records must not silently receive a WD value
noneligible_mask = ~eligible_mask

if wd_genus_eligibility.loc[
    noneligible_mask,
    "Proposed_Genus_WD"
].notna().any():

    raise ValueError(
        "A non-eligible genus record received a proposed WD value."
    )

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("Genus-level WD eligibility classification")
print("=" * 110)
print(
    f"Taxa evaluated              : "
    f"{len(wd_genus_eligibility)}"
)

print("\nGenus data support")
print("-" * 110)

print(
    wd_genus_eligibility[
        "Genus_Data_Support"
    ]
    .value_counts(dropna=False)
)

print("\nEligibility classes")
print("-" * 110)

print(
    wd_genus_eligibility[
        "Genus_Eligibility"
    ]
    .value_counts(dropna=False)
)

print("\nProposed fallback sources")
print("-" * 110)

print(
    wd_genus_eligibility[
        "Proposed_Genus_Source"
    ]
    .value_counts(dropna=False)
)

print("\nProposed quality flags")
print("-" * 110)

print(
    wd_genus_eligibility[
        "Proposed_Quality_Flag"
    ]
    .value_counts(dropna=False)
)

print("\nDetailed eligibility table")
print("-" * 110)

display(
    wd_genus_eligibility[
        [
            "Canonical_Species",
            "Verified_Name",
            "Fallback_Genus",
            "Genus_Basis",
            "Genus_Data_Support",
            "GWDD_Genus_WD",
            "GWDD_Genus_N",
            "TWDD_Genus_WD",
            "TWDD_Genus_SD",
            "TWDD_Genus_Records",
            "TWDD_Genus_Species",
            "GWDD_TWDD_Genus_Abs_Difference",
            "Genus_Eligibility",
            "Proposed_Genus_Source",
            "Proposed_Genus_WD",
            "Proposed_Quality_Flag",
        ]
    ]
    .sort_values(
        [
            "Genus_Eligibility",
            "Canonical_Species"
        ]
    )
    .reset_index(drop=True)
)
```

### 7.5.3 Resolution of the Scottellia Genus for Genus-Level Fallback

Three unresolved Omo taxon strings contain the genus spelling `Scotellia`.

Current botanical references recognize the genus `Scottellia`. Plants of the
World Online also records `Scottellia coriacea` as a synonym of the accepted
`Scottellia klaineana`, a West and West-Central African tree occurring in
Nigeria.

Because the Omo source strings remain uncertain at species level, no species
identity is imposed here. The correction is restricted to the genus:

`Scotellia` → `Scottellia`

This permits controlled genus-level wood-density evaluation while preserving
the unresolved species-level status of the original records.

```python
# =============================================================================
# Section 7.5.3 — Resolve Scottellia Genus and Reassess Fallback Support
# =============================================================================

wd_genus_final_review = wd_genus_eligibility.copy()

# -------------------------------------------------------------------------
# Controlled genus-only correction
# -------------------------------------------------------------------------

SCOTTELLIA_TAXA = {
    "Scotellia coriacea",
    "Scotellia coriaceae",
    "Scotellia coricea",
}

scottellia_mask = (
    wd_genus_final_review["Canonical_Species"]
    .isin(SCOTTELLIA_TAXA)
)

wd_genus_final_review.loc[
    scottellia_mask,
    "Fallback_Genus"
] = "Scottellia"

wd_genus_final_review.loc[
    scottellia_mask,
    "Genus_Basis"
] = "VERIFIED_GENUS_ORTHOGRAPHY"

# -------------------------------------------------------------------------
# Refresh database support for Scottellia
# -------------------------------------------------------------------------

for idx in wd_genus_final_review.index[scottellia_mask]:

    genus = wd_genus_final_review.at[
        idx,
        "Fallback_Genus"
    ]

    in_gwdd = genus in gwdd_genus_lookup.index
    in_twdd = genus in twdd_genus_summary.index

    wd_genus_final_review.at[
        idx, "Genus_In_GWDD"
    ] = in_gwdd

    wd_genus_final_review.at[
        idx, "Genus_In_TWDD"
    ] = in_twdd

    # GWDD
    if in_gwdd:

        gw = gwdd_genus_lookup.loc[genus]

        wd_genus_final_review.at[
            idx, "GWDD_Genus_WD"
        ] = gw["wsg_est"]

        wd_genus_final_review.at[
            idx, "GWDD_Genus_N"
        ] = gw["nb"]

        wd_genus_final_review.at[
            idx, "GWDD_Genus_Sources"
        ] = gw["nb_sources"]

        wd_genus_final_review.at[
            idx, "GWDD_Genus_Countries"
        ] = gw["nb_countries"]

        wd_genus_final_review.at[
            idx, "GWDD_Genus_Sites"
        ] = gw["nb_sites"]

    else:

        wd_genus_final_review.at[
            idx, "GWDD_Genus_WD"
        ] = np.nan

    # TWDD
    if in_twdd:

        tw = twdd_genus_summary.loc[genus]

        wd_genus_final_review.at[
            idx, "TWDD_Genus_WD"
        ] = tw["TWDD_genus_WD_mean"]

        wd_genus_final_review.at[
            idx, "TWDD_Genus_SD"
        ] = tw["TWDD_genus_WD_sd"]

        wd_genus_final_review.at[
            idx, "TWDD_Genus_Records"
        ] = tw["TWDD_genus_records"]

        wd_genus_final_review.at[
            idx, "TWDD_Genus_Species"
        ] = tw["TWDD_genus_species"]

        wd_genus_final_review.at[
            idx, "TWDD_Genus_Countries"
        ] = tw["TWDD_genus_countries"]

        wd_genus_final_review.at[
            idx, "TWDD_Genus_Sources"
        ] = tw["TWDD_genus_sources"]

    else:

        wd_genus_final_review.at[
            idx, "TWDD_Genus_WD"
        ] = np.nan

# -------------------------------------------------------------------------
# Recalculate support
# -------------------------------------------------------------------------

wd_genus_final_review["Genus_Data_Support"] = (
    wd_genus_final_review.apply(
        classify_genus_support,
        axis=1
    )
)

wd_genus_final_review[
    "GWDD_TWDD_Genus_Difference"
] = (
    wd_genus_final_review["TWDD_Genus_WD"]
    - wd_genus_final_review["GWDD_Genus_WD"]
)

wd_genus_final_review[
    "GWDD_TWDD_Genus_Abs_Difference"
] = (
    wd_genus_final_review[
        "GWDD_TWDD_Genus_Difference"
    ].abs()
)

# -------------------------------------------------------------------------
# Reclassify eligibility
# -------------------------------------------------------------------------

def classify_final_genus_eligibility(row):

    support = row["Genus_Data_Support"]

    if support == "NO_GENUS_DATA":
        return "NO_GENUS_DATA"

    if support == "GWDD_AND_TWDD":

        difference = row[
            "GWDD_TWDD_Genus_Abs_Difference"
        ]

        if (
            pd.notna(difference)
            and difference <= 0.10
        ):
            return "HIGH_CONFIDENCE_GENUS"

        return "MODERATE_CONFIDENCE_GENUS"

    if support in {
        "GWDD_ONLY",
        "TWDD_ONLY",
    }:
        return "MODERATE_CONFIDENCE_GENUS"

    return "NO_GENUS_DATA"


wd_genus_final_review[
    "Genus_Eligibility"
] = (
    wd_genus_final_review.apply(
        classify_final_genus_eligibility,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Rebuild proposed source and value
# -------------------------------------------------------------------------

wd_genus_final_review[
    "Proposed_Genus_Source"
] = (
    wd_genus_final_review.apply(
        proposed_genus_source,
        axis=1
    )
)

wd_genus_final_review[
    "Proposed_Genus_WD"
] = (
    wd_genus_final_review.apply(
        proposed_genus_value,
        axis=1
    )
)

wd_genus_final_review[
    "Proposed_Quality_Flag"
] = (
    wd_genus_final_review.apply(
        proposed_genus_quality,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Diagnostics
# -------------------------------------------------------------------------

print("Final genus-level fallback review")
print("=" * 110)

print("\nDatabase support")
print("-" * 110)

print(
    wd_genus_final_review[
        "Genus_Data_Support"
    ].value_counts(dropna=False)
)

print("\nEligibility classes")
print("-" * 110)

print(
    wd_genus_final_review[
        "Genus_Eligibility"
    ].value_counts(dropna=False)
)

print("\nScottellia records")
print("-" * 110)

display(
    wd_genus_final_review.loc[
        scottellia_mask,
        [
            "Canonical_Species",
            "Fallback_Genus",
            "Genus_Data_Support",
            "GWDD_Genus_WD",
            "TWDD_Genus_WD",
            "TWDD_Genus_Species",
            "GWDD_TWDD_Genus_Abs_Difference",
            "Genus_Eligibility",
            "Proposed_Genus_Source",
            "Proposed_Genus_WD",
        ]
    ]
)
```

### 7.5.4 Application of Approved Genus-Level Wood-Density Estimates

Approved genus-level fallback values are now incorporated for taxa that remain
without species-level wood-density information after GWDD and TWDD acquisition.

GWDD genus `wsg_est` is used as the primary fallback value where available.
TWDD genus information is retained as corroborating evidence and contributes
to the confidence classification.

All genus-derived records are explicitly flagged and remain distinguishable
from species-level observations. High-confidence and moderate-confidence genus
estimates are retained separately through the trait quality flag.

Taxa for which neither GWDD nor TWDD provides defensible genus-level information
remain missing and are not silently imputed.

```python
# =============================================================================
# Section 7.5.4 — Apply Approved Genus-Level WD Estimates
# =============================================================================

wd_with_genus = wd_species_twdd.copy()

# -------------------------------------------------------------------------
# Identify genus-eligible records
# -------------------------------------------------------------------------

genus_apply_mask = (
    wd_genus_final_review["Genus_Eligibility"]
    .isin(
        [
            "HIGH_CONFIDENCE_GENUS",
            "MODERATE_CONFIDENCE_GENUS",
        ]
    )
)

wd_genus_apply = (
    wd_genus_final_review.loc[
        genus_apply_mask
    ]
    .copy()
    .reset_index(drop=True)
)

# -------------------------------------------------------------------------
# Safety checks
# -------------------------------------------------------------------------

if len(wd_genus_apply) != 22:
    raise ValueError(
        f"Expected 22 genus-level fallback records, "
        f"found {len(wd_genus_apply)}."
    )

already_filled = wd_with_genus.loc[
    wd_with_genus["Canonical_Species"].isin(
        wd_genus_apply["Canonical_Species"]
    )
    &
    wd_with_genus["Standardized_Trait_Value"].notna()
]

if len(already_filled) > 0:
    raise ValueError(
        "Genus fallback attempted to overwrite existing WD values:\n"
        + already_filled["Canonical_Species"].to_string(index=False)
    )

# -------------------------------------------------------------------------
# Add genus-provenance fields if absent
# -------------------------------------------------------------------------

genus_provenance_columns = [
    "Fallback_Genus",
    "Genus_Estimate_Source",
    "Genus_Estimate_Confidence",
    "GWDD_Genus_N",
    "GWDD_Genus_Sources",
    "GWDD_Genus_Countries",
    "GWDD_Genus_Sites",
    "TWDD_Genus_WD",
    "TWDD_Genus_SD",
    "TWDD_Genus_Records",
    "TWDD_Genus_Species",
]

for col in genus_provenance_columns:
    if col not in wd_with_genus.columns:
        wd_with_genus[col] = np.nan

# Convert text provenance columns to object dtype
for col in [
    "Fallback_Genus",
    "Genus_Estimate_Source",
    "Genus_Estimate_Confidence",
]:
    wd_with_genus[col] = wd_with_genus[col].astype("object")

# -------------------------------------------------------------------------
# Apply genus fallback
# -------------------------------------------------------------------------

for _, row in wd_genus_apply.iterrows():

    canonical_name = row["Canonical_Species"]

    mask = (
        wd_with_genus["Canonical_Species"]
        == canonical_name
    )

    if mask.sum() != 1:
        raise ValueError(
            f"Expected exactly one record for {canonical_name}; "
            f"found {mask.sum()}."
        )

    genus_name = row["Fallback_Genus"]
    genus_value = row["Proposed_Genus_WD"]
    genus_source = row["Proposed_Genus_Source"]
    genus_quality = row["Proposed_Quality_Flag"]

    # ---------------------------------------------------------------------
    # Core trait fields
    # ---------------------------------------------------------------------

    wd_with_genus.loc[
        mask, "Trait_Source_Name"
    ] = genus_name

    wd_with_genus.loc[
        mask, "Raw_Trait_Value"
    ] = genus_value

    wd_with_genus.loc[
        mask, "Raw_Trait_Unit"
    ] = "g cm-3"

    wd_with_genus.loc[
        mask, "Standardized_Trait_Value"
    ] = genus_value

    wd_with_genus.loc[
        mask, "Standardized_Trait_Unit"
    ] = "g cm-3"

    wd_with_genus.loc[
        mask, "Match_Status"
    ] = "GENUS_LEVEL_FALLBACK"

    wd_with_genus.loc[
        mask, "Taxonomic_Level"
    ] = "GENUS"

    wd_with_genus.loc[
        mask, "Trait_Source"
    ] = genus_source

    wd_with_genus.loc[
        mask, "Trait_Quality_Flag"
    ] = genus_quality

    # ---------------------------------------------------------------------
    # Provenance
    # ---------------------------------------------------------------------

    wd_with_genus.loc[
        mask, "Fallback_Genus"
    ] = genus_name

    wd_with_genus.loc[
        mask, "Genus_Estimate_Source"
    ] = genus_source

    wd_with_genus.loc[
        mask, "Genus_Estimate_Confidence"
    ] = row["Genus_Eligibility"]

    wd_with_genus.loc[
        mask, "GWDD_Genus_N"
    ] = row["GWDD_Genus_N"]

    wd_with_genus.loc[
        mask, "GWDD_Genus_Sources"
    ] = row["GWDD_Genus_Sources"]

    wd_with_genus.loc[
        mask, "GWDD_Genus_Countries"
    ] = row["GWDD_Genus_Countries"]

    wd_with_genus.loc[
        mask, "GWDD_Genus_Sites"
    ] = row["GWDD_Genus_Sites"]

    wd_with_genus.loc[
        mask, "TWDD_Genus_WD"
    ] = row["TWDD_Genus_WD"]

    wd_with_genus.loc[
        mask, "TWDD_Genus_SD"
    ] = row["TWDD_Genus_SD"]

    wd_with_genus.loc[
        mask, "TWDD_Genus_Records"
    ] = row["TWDD_Genus_Records"]

    wd_with_genus.loc[
        mask, "TWDD_Genus_Species"
    ] = row["TWDD_Genus_Species"]


# -------------------------------------------------------------------------
# Final WD coverage diagnostics
# -------------------------------------------------------------------------

wd_final_available_mask = (
    wd_with_genus["Standardized_Trait_Value"]
    .notna()
)

n_total = len(wd_with_genus)
n_available = int(wd_final_available_mask.sum())
n_missing = n_total - n_available

print("Wood-density coverage after genus-level fallback")
print("=" * 100)
print(f"Canonical Omo taxa          : {n_total}")
print(f"WD available               : {n_available}")
print(f"WD still missing           : {n_missing}")
print(
    f"Overall WD coverage        : "
    f"{100 * n_available / n_total:.2f}%"
)

print("\nMatch status")
print("-" * 100)

print(
    wd_with_genus["Match_Status"]
    .value_counts(dropna=False)
)

print("\nTrait quality flags")
print("-" * 100)

print(
    wd_with_genus["Trait_Quality_Flag"]
    .value_counts(dropna=False)
)

print("\nGenus fallback records")
print("-" * 100)

display(
    wd_with_genus.loc[
        wd_with_genus["Match_Status"]
        .eq("GENUS_LEVEL_FALLBACK"),
        [
            "Canonical_Species",
            "Trait_Source_Name",
            "Standardized_Trait_Value",
            "Trait_Source",
            "Trait_Quality_Flag",
            "Genus_Estimate_Confidence",
            "GWDD_Genus_N",
            "TWDD_Genus_WD",
            "TWDD_Genus_Species",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -------------------------------------------------------------------------
# Remaining unresolved taxa
# -------------------------------------------------------------------------

wd_still_missing = (
    wd_with_genus.loc[
        wd_with_genus["Standardized_Trait_Value"]
        .isna(),
        [
            "Canonical_Species",
            "Match_Status",
            "Trait_Quality_Flag",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nTaxa still lacking WD")
print("-" * 100)

display(wd_still_missing)
```

### 7.5.6 Supplementary Literature Wood-Density Integration

The two taxa remaining after GWDD, TWDD, and controlled genus-level fallback
were evaluated using targeted supplementary literature.

For *Bambusa vulgaris*, a species-specific mean basic density of
0.69 g cm-3 was retained from peer-reviewed literature reporting basic
density directly for the species. Independent studies demonstrate substantial
variation with culm age and longitudinal position; the selected value therefore
represents a literature-derived species estimate rather than a universal
constant.

For the Omo taxon recorded as `Musa sapiensis`, previous taxonomic review
associated the source name with the Musa sapientum / Musa × paradisiaca
complex. A published pseudostem basic-density value of 0.20 g cm-3 for
Musa sapientum was retained. Because of the taxonomic uncertainty and strong
tissue-dependent variation in Musa density, this value receives a conservative
supplementary-literature quality flag.

Both records retain explicit provenance and remain distinguishable from
GWDD/TWDD species-level observations and genus-level estimates.

```python
# =============================================================================
# Section 7.5.6 — Integrate Supplementary Literature WD Values
# =============================================================================

wd_complete = wd_with_genus.copy()

# -------------------------------------------------------------------------
# Controlled supplementary literature registry
# -------------------------------------------------------------------------

wd_supplementary_registry = pd.DataFrame(
    [
        {
            "Canonical_Species": "Bambusa vulgaris",
            "Trait_Source_Name": "Bambusa vulgaris",
            "WD_Value": 0.69,
            "WD_Unit": "g cm-3",
            "Trait_Source": "SUPPLEMENTARY_LITERATURE",
            "Match_Status": "SUPPLEMENTARY_SPECIES_MATCH",
            "Taxonomic_Level": "SPECIES",
            "Trait_Quality_Flag":
                "SUPPLEMENTARY_SPECIES_VALUE",
            "Evidence_Type":
                "Peer-reviewed species-specific basic density",
            "Evidence_Reference":
                "Yokota et al. (2022), TROPICS 30(4):83-93",
            "Evidence_DOI":
                "10.3759/tropics.MS21-06",
        },
        {
            "Canonical_Species": "Musa sapiensis",
            "Trait_Source_Name": "Musa sapientum",
            "WD_Value": 0.20,
            "WD_Unit": "g cm-3",
            "Trait_Source": "SUPPLEMENTARY_LITERATURE",
            "Match_Status": "SUPPLEMENTARY_TAXONOMIC_MATCH",
            "Taxonomic_Level": "SPECIES_COMPLEX",
            "Trait_Quality_Flag":
                "SUPPLEMENTARY_TAXONOMIC_VALUE",
            "Evidence_Type":
                "Published pseudostem basic-density value",
            "Evidence_Reference":
                "Musa sapientum pseudostem literature",
            "Evidence_DOI":
                pd.NA,
        },
    ]
)

# -------------------------------------------------------------------------
# Add supplementary provenance fields if absent
# -------------------------------------------------------------------------

supplementary_columns = [
    "Supplementary_Evidence_Type",
    "Supplementary_Reference",
    "Supplementary_DOI",
]

for col in supplementary_columns:
    if col not in wd_complete.columns:
        wd_complete[col] = pd.NA

# Ensure text columns can safely accept strings
for col in [
    "Trait_Source_Name",
    "Trait_Source",
    "Match_Status",
    "Taxonomic_Level",
    "Trait_Quality_Flag",
    "Supplementary_Evidence_Type",
    "Supplementary_Reference",
    "Supplementary_DOI",
]:
    if col in wd_complete.columns:
        wd_complete[col] = (
            wd_complete[col]
            .astype("object")
        )

# -------------------------------------------------------------------------
# Safety check — supplementary values must only fill missing records
# -------------------------------------------------------------------------

for _, row in wd_supplementary_registry.iterrows():

    taxon = row["Canonical_Species"]

    mask = (
        wd_complete["Canonical_Species"]
        .eq(taxon)
    )

    if mask.sum() != 1:
        raise ValueError(
            f"Expected exactly one record for {taxon}; "
            f"found {mask.sum()}."
        )

    existing_value = (
        wd_complete.loc[
            mask,
            "Standardized_Trait_Value"
        ]
        .iloc[0]
    )

    if pd.notna(existing_value):
        raise ValueError(
            f"Supplementary integration would overwrite "
            f"an existing WD value for {taxon}."
        )

# -------------------------------------------------------------------------
# Apply supplementary values
# -------------------------------------------------------------------------

for _, row in wd_supplementary_registry.iterrows():

    taxon = row["Canonical_Species"]

    mask = (
        wd_complete["Canonical_Species"]
        .eq(taxon)
    )

    wd_complete.loc[
        mask, "Trait_Source_Name"
    ] = row["Trait_Source_Name"]

    wd_complete.loc[
        mask, "Raw_Trait_Value"
    ] = row["WD_Value"]

    wd_complete.loc[
        mask, "Raw_Trait_Unit"
    ] = row["WD_Unit"]

    wd_complete.loc[
        mask, "Standardized_Trait_Value"
    ] = row["WD_Value"]

    wd_complete.loc[
        mask, "Standardized_Trait_Unit"
    ] = row["WD_Unit"]

    wd_complete.loc[
        mask, "Trait_Source"
    ] = row["Trait_Source"]

    wd_complete.loc[
        mask, "Match_Status"
    ] = row["Match_Status"]

    wd_complete.loc[
        mask, "Taxonomic_Level"
    ] = row["Taxonomic_Level"]

    wd_complete.loc[
        mask, "Trait_Quality_Flag"
    ] = row["Trait_Quality_Flag"]

    wd_complete.loc[
        mask, "Supplementary_Evidence_Type"
    ] = row["Evidence_Type"]

    wd_complete.loc[
        mask, "Supplementary_Reference"
    ] = row["Evidence_Reference"]

    wd_complete.loc[
        mask, "Supplementary_DOI"
    ] = row["Evidence_DOI"]


# -------------------------------------------------------------------------
# Final integrity checks
# -------------------------------------------------------------------------

if len(wd_complete) != 127:
    raise ValueError(
        f"Expected 127 canonical taxa; found "
        f"{len(wd_complete)}."
    )

if (
    wd_complete["Canonical_Species"]
    .duplicated()
    .any()
):
    raise ValueError(
        "Duplicate canonical taxa detected."
    )

n_missing = (
    wd_complete[
        "Standardized_Trait_Value"
    ]
    .isna()
    .sum()
)

if n_missing != 0:
    raise ValueError(
        f"Expected complete WD coverage; "
        f"{n_missing} taxa remain missing."
    )

# Basic plausibility
invalid_wd = wd_complete.loc[
    (
        wd_complete[
            "Standardized_Trait_Value"
        ] <= 0
    )
    |
    (
        wd_complete[
            "Standardized_Trait_Value"
        ] > 1.5
    )
]

if len(invalid_wd) > 0:
    raise ValueError(
        "Implausible WD values detected."
    )

# -------------------------------------------------------------------------
# Final diagnostics
# -------------------------------------------------------------------------

n_total = len(wd_complete)

n_available = (
    wd_complete[
        "Standardized_Trait_Value"
    ]
    .notna()
    .sum()
)

print("Final wood-density acquisition coverage")
print("=" * 100)

print(
    f"Canonical Omo taxa          : {n_total}"
)

print(
    f"WD available               : {n_available}"
)

print(
    f"WD missing                 : "
    f"{n_total - n_available}"
)

print(
    f"Final WD coverage          : "
    f"{100 * n_available / n_total:.2f}%"
)

print("\nMatch status")
print("-" * 100)

print(
    wd_complete[
        "Match_Status"
    ]
    .value_counts(dropna=False)
)

print("\nTrait source")
print("-" * 100)

print(
    wd_complete[
        "Trait_Source"
    ]
    .value_counts(dropna=False)
)

print("\nTrait quality")
print("-" * 100)

print(
    wd_complete[
        "Trait_Quality_Flag"
    ]
    .value_counts(dropna=False)
)

print("\nSupplementary literature records")
print("-" * 100)

display(
    wd_complete.loc[
        wd_complete["Trait_Source"]
        .eq("SUPPLEMENTARY_LITERATURE"),
        [
            "Canonical_Species",
            "Trait_Source_Name",
            "Standardized_Trait_Value",
            "Standardized_Trait_Unit",
            "Match_Status",
            "Taxonomic_Level",
            "Trait_Quality_Flag",
            "Supplementary_Reference",
            "Supplementary_DOI",
        ]
    ]
    .reset_index(drop=True)
)

print("\nFinal WD numerical summary")
print("-" * 100)

print(
    wd_complete[
        "Standardized_Trait_Value"
    ]
    .describe()
)
```

### 7.5.7 Final Wood-Density Type Normalization and Validation

Following completion of the hierarchical wood-density acquisition workflow,
the standardized WD field is explicitly converted to numeric form.

This step ensures that successive provenance-preserving assignments during
species-level matching, genus fallback, and supplementary integration have not
left the analytical WD field with an object data type.

The final table is then validated for completeness, uniqueness, numerical
plausibility, and provenance consistency before export.

```python
# =============================================================================
# Section 7.5.7 — Final WD Type Normalization and Validation
# =============================================================================

wd_final = wd_complete.copy()

# -------------------------------------------------------------------------
# Explicit numeric normalization
# -------------------------------------------------------------------------

wd_final["Standardized_Trait_Value"] = pd.to_numeric(
    wd_final["Standardized_Trait_Value"],
    errors="coerce"
)

wd_final["Raw_Trait_Value"] = pd.to_numeric(
    wd_final["Raw_Trait_Value"],
    errors="coerce"
)

# -------------------------------------------------------------------------
# Core integrity checks
# -------------------------------------------------------------------------

if len(wd_final) != 127:
    raise ValueError(
        f"Expected 127 canonical taxa; found {len(wd_final)}."
    )

if wd_final["Canonical_Species"].duplicated().any():

    duplicates = wd_final.loc[
        wd_final["Canonical_Species"].duplicated(keep=False),
        "Canonical_Species"
    ]

    raise ValueError(
        "Duplicate canonical taxa detected:\n"
        + duplicates.to_string(index=False)
    )

# No missing standardized WD values
n_missing = (
    wd_final["Standardized_Trait_Value"]
    .isna()
    .sum()
)

if n_missing != 0:
    raise ValueError(
        f"{n_missing} standardized WD values are missing "
        f"after numeric normalization."
    )

# -------------------------------------------------------------------------
# Unit validation
# -------------------------------------------------------------------------

unexpected_units = (
    wd_final.loc[
        wd_final["Standardized_Trait_Value"].notna(),
        "Standardized_Trait_Unit"
    ]
    .dropna()
    .unique()
)

if set(unexpected_units) != {"g cm-3"}:
    raise ValueError(
        "Unexpected standardized WD units detected: "
        f"{unexpected_units}"
    )

# -------------------------------------------------------------------------
# Numerical plausibility
# -------------------------------------------------------------------------

invalid_wd = wd_final.loc[
    (wd_final["Standardized_Trait_Value"] <= 0)
    |
    (wd_final["Standardized_Trait_Value"] > 1.5)
]

if len(invalid_wd) > 0:

    print("Implausible WD values")
    display(
        invalid_wd[
            [
                "Canonical_Species",
                "Standardized_Trait_Value",
                "Trait_Source",
            ]
        ]
    )

    raise ValueError(
        "Final WD table contains implausible values."
    )

# -------------------------------------------------------------------------
# Expected provenance counts
# -------------------------------------------------------------------------

expected_source_counts = {
    "GWDD_V2_SPECIES": 102,
    "TWDD_SPECIES": 1,
    "GWDD_V2_GENUS": 22,
    "SUPPLEMENTARY_LITERATURE": 2,
}

actual_source_counts = (
    wd_final["Trait_Source"]
    .value_counts()
    .to_dict()
)

for source, expected_n in expected_source_counts.items():

    actual_n = actual_source_counts.get(source, 0)

    if actual_n != expected_n:
        raise ValueError(
            f"Unexpected count for {source}: "
            f"expected {expected_n}, found {actual_n}."
        )

# -------------------------------------------------------------------------
# Analytical provenance class
# -------------------------------------------------------------------------

def classify_wd_provenance(row):

    source = row["Trait_Source"]

    if source == "GWDD_V2_SPECIES":
        return "SPECIES_GWDD"

    if source == "TWDD_SPECIES":
        return "SPECIES_TWDD"

    if source == "GWDD_V2_GENUS":
        return "GENUS_ESTIMATE"

    if source == "SUPPLEMENTARY_LITERATURE":

        if (
            row["Match_Status"]
            == "SUPPLEMENTARY_SPECIES_MATCH"
        ):
            return "SPECIES_SUPPLEMENTARY"

        return "TAXONOMIC_SUPPLEMENTARY"

    return "OTHER"


wd_final["WD_Provenance_Class"] = (
    wd_final.apply(
        classify_wd_provenance,
        axis=1
    )
)

# -------------------------------------------------------------------------
# Final diagnostics
# -------------------------------------------------------------------------

print("Final wood-density validation")
print("=" * 100)

print(f"Rows                       : {len(wd_final)}")
print(
    f"Unique canonical taxa      : "
    f"{wd_final['Canonical_Species'].nunique()}"
)
print(
    f"Missing standardized WD    : "
    f"{wd_final['Standardized_Trait_Value'].isna().sum()}"
)
print(
    f"WD dtype                   : "
    f"{wd_final['Standardized_Trait_Value'].dtype}"
)

print("\nWD provenance")
print("-" * 100)

print(
    wd_final["WD_Provenance_Class"]
    .value_counts(dropna=False)
)

print("\nFinal numerical WD summary")
print("-" * 100)

print(
    wd_final["Standardized_Trait_Value"]
    .describe()
)

print("\nWD values by provenance class")
print("-" * 100)

display(
    wd_final.groupby(
        "WD_Provenance_Class"
    )["Standardized_Trait_Value"]
    .agg(
        N="count",
        Mean="mean",
        SD="std",
        Minimum="min",
        Median="median",
        Maximum="max",
    )
    .round(4)
)

print("\nFinal validation status")
print("-" * 100)
print("PASS — wood-density table is complete and analytically valid.")
```

The wood-density acquisition workflow is now complete for all 127 canonical

2. species-level TWDD values;
3. controlled genus-level GWDD fallback supported, where available, by TWDD;

All records retain explicit taxonomic matching status, source provenance,

```python
# =============================================================================
# Section 7.6 — Export Final Wood-Density Dataset
# =============================================================================

from pathlib import Path

WD_OUTPUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

WD_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

WD_FINAL_FILE = (
    WD_OUTPUT_DIR
    / "omo_wood_density_final.csv"
)

WD_PROVENANCE_FILE = (
    WD_OUTPUT_DIR
    / "omo_wood_density_provenance_summary.csv"
)

# -------------------------------------------------------------------------
# Export full validated WD table
# -------------------------------------------------------------------------

wd_final.to_csv(
    WD_FINAL_FILE,
    index=False
)

# -------------------------------------------------------------------------
# Create provenance summary
# -------------------------------------------------------------------------

wd_provenance_summary = (
    wd_final.groupby(
        "WD_Provenance_Class",
        dropna=False
    )
    .agg(
        N_Taxa=("Canonical_Species", "count"),
        Mean_WD=("Standardized_Trait_Value", "mean"),
        SD_WD=("Standardized_Trait_Value", "std"),
        Min_WD=("Standardized_Trait_Value", "min"),
        Median_WD=("Standardized_Trait_Value", "median"),
        Max_WD=("Standardized_Trait_Value", "max"),
    )
    .reset_index()
)

wd_provenance_summary["Percent_of_Taxa"] = (
    100
    * wd_provenance_summary["N_Taxa"]
    / len(wd_final)
)

wd_provenance_summary.to_csv(
    WD_PROVENANCE_FILE,
    index=False
)

# -------------------------------------------------------------------------
# Post-export validation
# -------------------------------------------------------------------------

wd_export_check = pd.read_csv(
    WD_FINAL_FILE
)

if len(wd_export_check) != 127:
    raise ValueError(
        "Exported WD table does not contain 127 rows."
    )

if (
    wd_export_check["Canonical_Species"]
    .nunique()
    != 127
):
    raise ValueError(
        "Exported WD table does not contain 127 unique taxa."
    )

if (
    wd_export_check[
        "Standardized_Trait_Value"
    ]
    .isna()
    .any()
):
    raise ValueError(
        "Exported WD table contains missing standardized WD values."
    )

print("Wood-density export complete")
print("=" * 100)

print(f"Final WD table      : {WD_FINAL_FILE}")
print(f"Provenance summary : {WD_PROVENANCE_FILE}")

print("\nFinal table validation")
print("-" * 100)

print(
    f"Rows                : "
    f"{len(wd_export_check)}"
)

print(
    f"Unique taxa         : "
    f"{wd_export_check['Canonical_Species'].nunique()}"
)

print(
    f"Missing WD values   : "
    f"{wd_export_check['Standardized_Trait_Value'].isna().sum()}"
)

print("\nProvenance summary")
print("-" * 100)

display(
    wd_provenance_summary.round(4)
)

print("\nSection 7 status")
print("-" * 100)
print(
    "PASS — wood-density acquisition, harmonization, "
    "validation, provenance tracking, and export completed."
)
```

## 8. Specific Leaf Area (SLA) Acquisition and Harmonization

Specific leaf area (SLA) is a key plant functional trait describing the amount of leaf area deployed per unit of dry leaf mass. It is closely associated with plant resource-use strategy, growth rate, leaf economics, and ecological responses to environmental conditions. Species with relatively high SLA generally exhibit faster resource acquisition and shorter-lived leaves, whereas lower SLA is commonly associated with more conservative resource-use strategies and longer-lived foliage.

For the Omo Forest functional-trait dataset, SLA values will be acquired and harmonized for the 127 canonical woody taxa established during the preceding taxonomic and community-data workflows. The acquisition process will preserve the validated canonical taxonomy and will maintain explicit provenance for all trait values. Preference will be given to species-level observations from recognized plant-trait sources, while genus-level or other controlled estimates will be considered only where reliable species-level information is unavailable.

All SLA measurements will ultimately be standardized to **mm² mg⁻¹**. Original values, units, taxonomic matches, data sources, and any estimation procedures will be retained to ensure that the final functional-trait dataset remains transparent, reproducible, and suitable for subsequent community-weighted trait and functional-diversity analyses.

### 8.1 Establishment of the SLA Acquisition Target Table

Before querying external trait sources, a controlled SLA acquisition target table is constructed from the validated Omo Forest canonical species metadata. This ensures that trait acquisition is restricted to the established set of 127 canonical taxa and prevents external trait records from inadvertently altering the project's validated taxonomy.

For each canonical taxon, the target table records the canonical species name, genus, taxonomic resolution, standardized SLA unit, and fields required for subsequent source, matching, provenance, and quality-control information. Taxa resolved only to genus level are explicitly identified so that they can be handled separately during SLA acquisition rather than being treated as species-level observations.

The resulting table therefore serves as the auditable framework for the SLA acquisition and harmonization workflow.

```python
# =============================================================================
# SECTION 8.1 — ESTABLISH SLA ACQUISITION TARGET TABLE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.1 — SPECIFIC LEAF AREA (SLA) ACQUISITION TARGET TABLE")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Locate the validated 127-taxon canonical species table already in memory
# -----------------------------------------------------------------------------

def locate_canonical_species_table(namespace):
    """
    Locate the validated canonical species DataFrame from the current notebook.

    Required:
        - column 'Canonical_Species'
        - 127 unique non-null canonical taxa

    Returns
    -------
    dataframe_name : str
    dataframe      : pd.DataFrame
    """

    candidates = []

    for name, obj in namespace.items():

        if not isinstance(obj, pd.DataFrame):
            continue

        if "Canonical_Species" not in obj.columns:
            continue

        n_unique = obj["Canonical_Species"].dropna().nunique()

        if n_unique == 127:
            candidates.append((name, obj))

    if not candidates:
        raise RuntimeError(
            "Could not locate a DataFrame containing exactly 127 unique "
            "'Canonical_Species' values.\n"
            "Confirm that the validated canonical species metadata table "
            "from Sections 1–7 is still loaded."
        )

    # Prefer compact species-level metadata tables over long trait tables
    candidates.sort(key=lambda x: len(x[1]))

    return candidates[0]


canonical_df_name, canonical_source_df = locate_canonical_species_table(globals())

print(f"\nCanonical species table detected : {canonical_df_name}")
print(f"Rows in detected table           : {len(canonical_source_df):,}")
print(
    f"Unique canonical taxa            : "
    f"{canonical_source_df['Canonical_Species'].dropna().nunique():,}"
)


# -----------------------------------------------------------------------------
# 2. Extract one validated record per canonical taxon
# -----------------------------------------------------------------------------

keep_columns = [
    col for col in [
        "Canonical_Species",
        "Family",
        "Genus",
        "Taxonomic_Resolution"
    ]
    if col in canonical_source_df.columns
]

sla_targets = (
    canonical_source_df[keep_columns]
    .drop_duplicates(subset="Canonical_Species")
    .copy()
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 3. Construct genus explicitly if it is not already present
# -----------------------------------------------------------------------------

if "Genus" not in sla_targets.columns:

    sla_targets["Genus"] = (
        sla_targets["Canonical_Species"]
        .astype(str)
        .str.strip()
        .str.split()
        .str[0]
    )


# -----------------------------------------------------------------------------
# 4. Recover/derive taxonomic resolution if necessary
# -----------------------------------------------------------------------------

if "Taxonomic_Resolution" not in sla_targets.columns:

    genus_level_taxa = {
        "Drypetes species",
        "Cola species"
    }

    sla_targets["Taxonomic_Resolution"] = np.where(
        sla_targets["Canonical_Species"].isin(genus_level_taxa),
        "GENUS_LEVEL_OR_UNRESOLVED",
        "SPECIES_LEVEL"
    )


# -----------------------------------------------------------------------------
# 5. Establish SLA query fields
# -----------------------------------------------------------------------------

sla_targets["Trait_Code"] = "SLA"
sla_targets["Target_Unit"] = "mm2 mg-1"

# Species names retained exactly as validated by the Omo taxonomy workflow
sla_targets["SLA_Query_Name"] = sla_targets["Canonical_Species"]

# Flag whether strict species-level acquisition is appropriate
sla_targets["Species_Level_Query_Eligible"] = (
    sla_targets["Taxonomic_Resolution"].eq("SPECIES_LEVEL")
)

# Explicitly identify taxa that will require non-species handling
sla_targets["Requires_Genus_Level_Strategy"] = (
    ~sla_targets["Species_Level_Query_Eligible"]
)

# Acquisition fields intentionally left blank at this stage
sla_targets["SLA_Raw_Value"] = np.nan
sla_targets["SLA_Raw_Unit"] = pd.NA
sla_targets["SLA_Source_Name"] = pd.NA
sla_targets["SLA_Source_Record_ID"] = pd.NA
sla_targets["SLA_Source_DOI_or_URL"] = pd.NA
sla_targets["SLA_Match_Status"] = "NOT_YET_SEARCHED"
sla_targets["SLA_Provenance_Class"] = "NOT_YET_ASSIGNED"
sla_targets["SLA_Quality_Flag"] = "PENDING"


# -----------------------------------------------------------------------------
# 6. Reorder columns
# -----------------------------------------------------------------------------

preferred_order = [
    "Canonical_Species",
    "Family",
    "Genus",
    "Taxonomic_Resolution",
    "Trait_Code",
    "Target_Unit",
    "SLA_Query_Name",
    "Species_Level_Query_Eligible",
    "Requires_Genus_Level_Strategy",
    "SLA_Raw_Value",
    "SLA_Raw_Unit",
    "SLA_Source_Name",
    "SLA_Source_Record_ID",
    "SLA_Source_DOI_or_URL",
    "SLA_Match_Status",
    "SLA_Provenance_Class",
    "SLA_Quality_Flag"
]

sla_targets = sla_targets[
    [c for c in preferred_order if c in sla_targets.columns]
]


# -----------------------------------------------------------------------------
# 7. Validation
# -----------------------------------------------------------------------------

n_rows = len(sla_targets)
n_taxa = sla_targets["Canonical_Species"].nunique()
n_species_level = sla_targets["Species_Level_Query_Eligible"].sum()
n_genus_strategy = sla_targets["Requires_Genus_Level_Strategy"].sum()

assert n_rows == 127, (
    f"Expected 127 SLA target rows but found {n_rows}."
)

assert n_taxa == 127, (
    f"Expected 127 unique SLA target taxa but found {n_taxa}."
)

assert sla_targets["Canonical_Species"].notna().all(), \
    "Missing canonical species names detected."

assert sla_targets["Genus"].notna().all(), \
    "Missing genus values detected."

assert sla_targets["Trait_Code"].eq("SLA").all()

assert sla_targets["Target_Unit"].eq("mm2 mg-1").all()


# -----------------------------------------------------------------------------
# 8. Export acquisition target table
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"
output_dir.mkdir(parents=True, exist_ok=True)

sla_target_path = output_dir / "omo_sla_acquisition_targets.csv"

sla_targets.to_csv(
    sla_target_path,
    index=False
)


# -----------------------------------------------------------------------------
# 9. Diagnostic output
# -----------------------------------------------------------------------------

print("\nSLA acquisition target validation")
print("-" * 100)

print(f"Rows                              : {n_rows}")
print(f"Unique canonical taxa             : {n_taxa}")
print(f"Species-level query eligible      : {n_species_level}")
print(f"Require genus-level strategy      : {n_genus_strategy}")
print(f"Missing canonical species         : {sla_targets['Canonical_Species'].isna().sum()}")
print(f"Missing genus                     : {sla_targets['Genus'].isna().sum()}")

print("\nTaxonomic-resolution distribution")
print("-" * 100)
print(
    sla_targets["Taxonomic_Resolution"]
    .value_counts(dropna=False)
    .to_string()
)

print("\nTaxa requiring genus-level strategy")
print("-" * 100)

display(
    sla_targets.loc[
        sla_targets["Requires_Genus_Level_Strategy"],
        [
            "Canonical_Species",
            "Genus",
            "Taxonomic_Resolution",
            "SLA_Query_Name"
        ]
    ]
)

print("\nPreview of SLA acquisition targets")
print("-" * 100)

display(
    sla_targets.head(10)
)

print("\nExport")
print("-" * 100)
print(f"SLA acquisition targets : {sla_target_path}")

print("\nSection 8.1 completed successfully.")
```

### 8.2 SLA Source Hierarchy and Acquisition Governance

Specific leaf area measurements can vary among individuals, populations, environments, leaf ages, canopy positions, and measurement protocols. Consequently, the SLA harmonization workflow must distinguish between genuine biological variation and differences arising from taxonomy, units, measurement definitions, or data provenance.

A controlled **source hierarchy** is therefore established before SLA observations are incorporated into the Omo Forest functional-trait dataset. The hierarchy determines which records should be preferred when multiple SLA values are available for the same canonical taxon and ensures that estimates derived at broader taxonomic levels remain distinguishable from direct species-level measurements.

The principal acquisition strategy gives priority to recognized plant-trait databases containing measured species-level SLA observations. The **TRY Plant Trait Database** is treated as the primary global trait resource because it integrates numerous contributed plant-trait datasets and explicitly includes leaf area per dry mass (specific leaf area, SLA). Where appropriate and accessible, **BIEN** and other published trait datasets may provide supplementary species-level observations. Literature-derived measurements may also be used where database coverage is incomplete.

The following provenance hierarchy is adopted:

1. **SPECIES_PRIMARY** — species-level SLA measurement obtained from the primary trait source;
2. **SPECIES_SUPPLEMENTARY** — species-level measurement obtained from another recognized trait database or published dataset;
3. **SPECIES_LITERATURE** — species-level measurement obtained directly from a peer-reviewed publication or other traceable scientific source;
4. **GENUS_ESTIMATE** — estimate derived from available congeneric species where species-level information is unavailable;
5. **FAMILY_ESTIMATE** — controlled family-level estimate considered only when species- and genus-level information are unavailable;
6. **GLOBAL_FALLBACK** — broader ecological estimate used only as a final documented fallback if necessary.

Species-level observations are always preferred over taxonomic estimates. Genus-, family-, or broader estimates will never be represented as direct species measurements and will retain explicit provenance flags in the final trait table.

At this stage, no SLA values are imputed. Instead, the source hierarchy, admissible measurement definitions, unit rules, and acquisition status fields are formalized so that all subsequent SLA records can be evaluated consistently and reproducibly.

The standardized SLA unit for the project remains **mm² mg⁻¹**. Because this unit is numerically equivalent to **m² kg⁻¹**, values reported in either form can be harmonized without numerical rescaling. Other common SLA units, including **cm² g⁻¹**, require explicit conversion before integration.

### 8.3 Preparation of the Primary Species-Level SLA Acquisition Manifest

Following establishment of the SLA source hierarchy, the first acquisition phase targets direct species-level measurements from the primary trait database. Only taxa classified as species-level targets are included in this phase. The two Omo Forest taxa retained at genus-level resolution (`Cola species` and `Drypetes species`) are excluded from the primary species search and will be addressed separately through the controlled genus-level workflow.

The TRY Plant Trait Database is used as the primary source for this acquisition phase. Rather than allowing external trait records to redefine the Omo Forest species list, the validated canonical names are submitted as a fixed query set. Retrieved records will subsequently be reconciled back to these canonical taxa using explicit exact, normalized, or harmonized taxonomic matching.

For each target species, all available records corresponding to **specific leaf area (SLA; leaf area per dry leaf mass)** should initially be retained. Individual observations are not reduced to a single species value during acquisition. This separation between acquisition and aggregation is important because multiple observations may represent different populations, environments, individuals, datasets, or measurement conditions. Filtering, unit standardization, quality assessment, outlier evaluation, and species-level aggregation will therefore be performed only after the raw SLA records have been imported into the notebook.

This subsection generates a reproducible acquisition manifest containing the 125 species-level Omo Forest taxa. The manifest acts both as the query list for the primary trait search and as an audit record documenting which taxa were submitted for SLA acquisition.

```python
# =============================================================================
# SECTION 8.3A — PREPARE PRIMARY SPECIES-LEVEL SLA ACQUISITION MANIFEST
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.3A — PRIMARY SPECIES-LEVEL SLA ACQUISITION MANIFEST")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm acquisition ledger is available
# -----------------------------------------------------------------------------

if "sla_acquisition_ledger" not in globals():

    ledger_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_acquisition_ledger.csv"
    )

    if not ledger_path.exists():
        raise FileNotFoundError(
            "SLA acquisition ledger was not found. "
            "Run Section 8.2 before continuing."
        )

    sla_acquisition_ledger = pd.read_csv(ledger_path)

    print("\nSLA acquisition ledger reloaded from:")
    print(ledger_path)

else:
    print("\nUsing SLA acquisition ledger already present in notebook memory.")


# -----------------------------------------------------------------------------
# 2. Robust conversion of boolean-style columns
# -----------------------------------------------------------------------------

def as_bool(series):
    """
    Convert common Boolean representations to Python Boolean values.
    Handles actual bools and strings produced by CSV round-tripping.
    """

    if pd.api.types.is_bool_dtype(series):
        return series.fillna(False)

    return (
        series
        .astype(str)
        .str.strip()
        .str.lower()
        .map({
            "true": True,
            "false": False,
            "1": True,
            "0": False,
            "yes": True,
            "no": False,
        })
        .fillna(False)
        .astype(bool)
    )


species_eligible = as_bool(
    sla_acquisition_ledger["Species_Level_Query_Eligible"]
)


# -----------------------------------------------------------------------------
# 3. Extract species-level primary-search targets only
# -----------------------------------------------------------------------------

sla_try_manifest = (
    sla_acquisition_ledger.loc[
        species_eligible
        & sla_acquisition_ledger["Acquisition_Status"].eq(
            "PENDING_PRIMARY_SEARCH"
        )
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 4. Retain only acquisition-relevant fields
# -----------------------------------------------------------------------------

manifest_columns = [
    col for col in [
        "Canonical_Species",
        "Family",
        "Genus",
        "Taxonomic_Resolution",
        "SLA_Query_Name",
        "Trait_Code",
        "Target_Unit",
    ]
    if col in sla_try_manifest.columns
]

sla_try_manifest = sla_try_manifest[
    manifest_columns
].copy()


# -----------------------------------------------------------------------------
# 5. Add controlled primary-source metadata
# -----------------------------------------------------------------------------

sla_try_manifest["Primary_Source"] = "TRY"
sla_try_manifest["Requested_Trait"] = "Specific leaf area"
sla_try_manifest["Requested_Trait_Code"] = "SLA"

sla_try_manifest["Requested_Measurement_Definition"] = (
    "Leaf area per dry leaf mass"
)

sla_try_manifest["Acquisition_Phase"] = (
    "PRIMARY_SPECIES_LEVEL"
)

sla_try_manifest["Acquisition_Request_Status"] = (
    "READY_FOR_QUERY"
)

sla_try_manifest["Retrieved_Record_Count"] = 0
sla_try_manifest["Primary_Data_Available"] = False


# -----------------------------------------------------------------------------
# 6. Generate sequential acquisition IDs
# -----------------------------------------------------------------------------

sla_try_manifest = (
    sla_try_manifest
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

sla_try_manifest.insert(
    0,
    "SLA_Query_ID",
    [
        f"OMO_SLA_{i:04d}"
        for i in range(1, len(sla_try_manifest) + 1)
    ]
)


# -----------------------------------------------------------------------------
# 7. Construct plain species-name list for external database query
# -----------------------------------------------------------------------------

sla_try_species_names = (
    sla_try_manifest["SLA_Query_Name"]
    .dropna()
    .astype(str)
    .str.strip()
    .drop_duplicates()
    .sort_values()
    .tolist()
)


# -----------------------------------------------------------------------------
# 8. Validation
# -----------------------------------------------------------------------------

n_manifest = len(sla_try_manifest)
n_unique_queries = sla_try_manifest["SLA_Query_Name"].nunique()

assert n_manifest == 125, (
    f"Expected 125 species-level SLA targets but found {n_manifest}."
)

assert n_unique_queries == 125, (
    f"Expected 125 unique SLA query names but found {n_unique_queries}."
)

assert (
    sla_try_manifest["Canonical_Species"]
    .nunique()
    == 125
)

assert not (
    sla_try_manifest["Canonical_Species"]
    .isin(["Cola species", "Drypetes species"])
    .any()
), (
    "Genus-level taxa were incorrectly included in "
    "the primary species acquisition manifest."
)

assert (
    sla_try_manifest["Primary_Source"]
    .eq("TRY")
    .all()
)

assert (
    sla_try_manifest["Acquisition_Request_Status"]
    .eq("READY_FOR_QUERY")
    .all()
)


# -----------------------------------------------------------------------------
# 9. Export acquisition manifest
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"
output_dir.mkdir(parents=True, exist_ok=True)

manifest_path = (
    output_dir
    / "omo_sla_try_species_acquisition_manifest.csv"
)

species_list_path = (
    output_dir
    / "omo_sla_try_species_query_list.txt"
)

sla_try_manifest.to_csv(
    manifest_path,
    index=False
)

with open(
    species_list_path,
    "w",
    encoding="utf-8"
) as f:

    for species in sla_try_species_names:
        f.write(f"{species}\n")


# -----------------------------------------------------------------------------
# 10. Diagnostic output
# -----------------------------------------------------------------------------

print("\nPrimary SLA acquisition manifest validation")
print("-" * 100)

print(
    f"Manifest rows                   : "
    f"{n_manifest}"
)

print(
    f"Unique species queries          : "
    f"{n_unique_queries}"
)

print(
    f"Primary source                  : "
    f"TRY"
)

print(
    f"Requested trait                 : "
    f"Specific leaf area (SLA)"
)

print(
    f"Target standardized unit        : "
    f"mm2 mg-1"
)

print(
    f"Initial acquisition status      : "
    f"READY_FOR_QUERY"
)


print("\nFirst 15 query taxa")
print("-" * 100)

display(
    sla_try_manifest[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Primary_Source",
            "Acquisition_Request_Status",
        ]
    ].head(15)
)


print("\nLast 10 query taxa")
print("-" * 100)

display(
    sla_try_manifest[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
        ]
    ].tail(10)
)


print("\nExcluded genus-level taxa")
print("-" * 100)

excluded_taxa = (
    sla_acquisition_ledger.loc[
        ~species_eligible,
        [
            c for c in [
                "Canonical_Species",
                "Genus",
                "Taxonomic_Resolution",
                "Acquisition_Status",
            ]
            if c in sla_acquisition_ledger.columns
        ]
    ]
)

display(excluded_taxa)


print("\nExports")
print("-" * 100)

print(
    f"TRY acquisition manifest : "
    f"{manifest_path}"
)

print(
    f"TRY species query list    : "
    f"{species_list_path}"
)

print("\nSection 8.3A completed successfully.")
```

```python
# =============================================================================
# SECTION 8.3B — TRY ACCEPTED-SPECIES TAXONOMIC CROSSWALK
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import difflib

print("=" * 100)
print("SECTION 8.3B — TRY ACCEPTED-SPECIES TAXONOMIC CROSSWALK")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Locate TRY accepted-species catalogue
# -----------------------------------------------------------------------------

try_species_path = (
    Path("data")
    / "external"
    / "TRY"
    / "TryAccSpecies.txt"
)

if not try_species_path.exists():
    raise FileNotFoundError(
        f"TRY accepted-species file not found:\n{try_species_path}\n\n"
        "Place the downloaded TryAccSpecies.txt file at this location "
        "before continuing."
    )


# -----------------------------------------------------------------------------
# 2. Read TRY accepted-species catalogue
# -----------------------------------------------------------------------------

try_species = pd.read_csv(
    try_species_path,
    sep="\t",
    dtype={
        "AccSpeciesID": "Int64",
        "AccSpeciesName": "string",
    },
    low_memory=False
)

required_try_columns = {
    "AccSpeciesID",
    "AccSpeciesName",
}

missing = required_try_columns.difference(
    try_species.columns
)

if missing:
    raise ValueError(
        "TRY accepted-species file does not contain expected columns: "
        f"{sorted(missing)}"
    )


# -----------------------------------------------------------------------------
# 3. Clean matching keys without changing original names
# -----------------------------------------------------------------------------

def normalize_taxon_name(value):
    """
    Generate conservative comparison key.

    Does NOT alter the stored taxonomic name.
    """

    if pd.isna(value):
        return pd.NA

    return (
        str(value)
        .strip()
        .lower()
        .replace("_", " ")
    )


try_species["TRY_Match_Key"] = (
    try_species["AccSpeciesName"]
    .map(normalize_taxon_name)
)


# -----------------------------------------------------------------------------
# 4. Start from the validated 125-species acquisition manifest
# -----------------------------------------------------------------------------

if "sla_try_manifest" not in globals():

    manifest_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_try_species_acquisition_manifest.csv"
    )

    if not manifest_path.exists():
        raise FileNotFoundError(
            "Section 8.3A acquisition manifest not found."
        )

    sla_try_manifest = pd.read_csv(manifest_path)


sla_try_crosswalk = sla_try_manifest.copy()

sla_try_crosswalk["OMO_Match_Key"] = (
    sla_try_crosswalk["SLA_Query_Name"]
    .map(normalize_taxon_name)
)


# -----------------------------------------------------------------------------
# 5. Exact accepted-name matching
# -----------------------------------------------------------------------------

try_exact_lookup = (
    try_species[
        [
            "AccSpeciesID",
            "AccSpeciesName",
            "TRY_Match_Key",
        ]
    ]
    .drop_duplicates(subset="TRY_Match_Key")
)

sla_try_crosswalk = (
    sla_try_crosswalk
    .merge(
        try_exact_lookup,
        how="left",
        left_on="OMO_Match_Key",
        right_on="TRY_Match_Key",
    )
)


sla_try_crosswalk["TRY_Match_Status"] = np.where(
    sla_try_crosswalk["AccSpeciesID"].notna(),
    "EXACT_ACCEPTED_NAME",
    "REQUIRES_REVIEW"
)


# -----------------------------------------------------------------------------
# 6. Generate diagnostic suggestions ONLY for unresolved names
#    Suggestions are not accepted automatically.
# -----------------------------------------------------------------------------

try_names = (
    try_species["AccSpeciesName"]
    .dropna()
    .astype(str)
    .tolist()
)


def closest_try_name(name):

    matches = difflib.get_close_matches(
        str(name),
        try_names,
        n=1,
        cutoff=0.75
    )

    return matches[0] if matches else pd.NA


unresolved_mask = (
    sla_try_crosswalk["AccSpeciesID"].isna()
)

sla_try_crosswalk[
    "Suggested_TRY_Name"
] = pd.NA

sla_try_crosswalk.loc[
    unresolved_mask,
    "Suggested_TRY_Name"
] = (
    sla_try_crosswalk.loc[
        unresolved_mask,
        "SLA_Query_Name"
    ]
    .map(closest_try_name)
)


# -----------------------------------------------------------------------------
# 7. Recover suggested TRY ID for diagnostic purposes
# -----------------------------------------------------------------------------

suggestion_lookup = (
    try_species[
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    ]
    .drop_duplicates(subset="AccSpeciesName")
    .rename(
        columns={
            "AccSpeciesID": "Suggested_TRY_ID",
            "AccSpeciesName": "Suggested_TRY_Name",
        }
    )
)

sla_try_crosswalk = (
    sla_try_crosswalk
    .merge(
        suggestion_lookup,
        how="left",
        on="Suggested_TRY_Name",
    )
)


# -----------------------------------------------------------------------------
# 8. Summary diagnostics
# -----------------------------------------------------------------------------

n_total = len(sla_try_crosswalk)

n_exact = (
    sla_try_crosswalk["TRY_Match_Status"]
    .eq("EXACT_ACCEPTED_NAME")
    .sum()
)

n_review = (
    sla_try_crosswalk["TRY_Match_Status"]
    .eq("REQUIRES_REVIEW")
    .sum()
)


# -----------------------------------------------------------------------------
# 9. Export preliminary crosswalk
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

crosswalk_path = (
    output_dir
    / "omo_sla_try_taxonomic_crosswalk_preliminary.csv"
)

review_path = (
    output_dir
    / "omo_sla_try_taxa_requiring_review.csv"
)


sla_try_crosswalk.to_csv(
    crosswalk_path,
    index=False
)

sla_try_crosswalk.loc[
    unresolved_mask,
    [
        "SLA_Query_ID",
        "Canonical_Species",
        "SLA_Query_Name",
        "Suggested_TRY_Name",
        "Suggested_TRY_ID",
        "TRY_Match_Status",
    ]
].to_csv(
    review_path,
    index=False
)


# -----------------------------------------------------------------------------
# 10. Diagnostic output
# -----------------------------------------------------------------------------

print("\nTRY catalogue")
print("-" * 100)

print(
    f"TRY accepted taxa loaded        : "
    f"{len(try_species):,}"
)

print(
    f"Unique TRY accepted names       : "
    f"{try_species['AccSpeciesName'].nunique():,}"
)


print("\nOmo → TRY exact matching")
print("-" * 100)

print(
    f"Omo species-level targets       : "
    f"{n_total}"
)

print(
    f"Exact TRY accepted-name matches : "
    f"{n_exact}"
)

print(
    f"Taxa requiring review           : "
    f"{n_review}"
)

print(
    f"Exact-match coverage             : "
    f"{100 * n_exact / n_total:.2f}%"
)


print("\nTaxa requiring controlled review")
print("-" * 100)

display(
    sla_try_crosswalk.loc[
        unresolved_mask,
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Suggested_TRY_Name",
            "Suggested_TRY_ID",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Preliminary crosswalk : "
    f"{crosswalk_path}"
)

print(
    f"Review table          : "
    f"{review_path}"
)

print("\nSection 8.3B completed successfully.")
```

### 8.3C Controlled Resolution of TRY Taxonomic Matches

The preliminary Omo–TRY crosswalk identified exact accepted-name matches for approximately half of the species-level acquisition targets. The remaining records require controlled reconciliation because approximate string similarity alone cannot establish taxonomic equivalence.

Several unmatched Omo names represent straightforward spelling, transcription, or orthographic variants of accepted TRY species names. These cases can be reconciled safely where the intended taxon is unambiguous. Examples include differences in doubled consonants, omitted letters, historical misspellings, and minor genus-name transcription errors.

Other names require more careful taxonomic interpretation because the closest TRY name may represent a different species or even a different genus. These records are not resolved automatically and remain flagged for manual taxonomic review.

The reconciliation procedure therefore applies a curated mapping only to transparent orthographic variants. Exact TRY matches are retained unchanged, while ambiguous taxa remain unresolved. Genus-only names are removed from the species-level TRY request and transferred to the later genus-level SLA estimation workflow.

This conservative approach ensures that the external TRY taxonomy is used only for trait retrieval and does not overwrite the validated Omo Forest canonical taxonomy.

```python
# =============================================================================
# SECTION 8.3C — CONTROLLED RESOLUTION OF TRY TAXONOMIC MATCHES
# =============================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 100)
print("SECTION 8.3C — CONTROLLED RESOLUTION OF TRY TAXONOMIC MATCHES")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm preliminary crosswalk exists
# -----------------------------------------------------------------------------

if "sla_try_crosswalk" not in globals():

    crosswalk_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_try_taxonomic_crosswalk_preliminary.csv"
    )

    if not crosswalk_path.exists():
        raise FileNotFoundError(
            "Preliminary TRY crosswalk not found. Run Section 8.3B first."
        )

    sla_try_crosswalk = pd.read_csv(crosswalk_path)


# -----------------------------------------------------------------------------
# 2. Curated mappings for transparent orthographic / transcription variants
# -----------------------------------------------------------------------------

controlled_try_name_map = {

    "Annonidium manni": "Anonidium mannii",
    "Blighia unijigata": "Blighia unijugata",

    "Chrysophylum albidum": "Chrysophyllum albidum",
    "Citrus sinenses": "Citrus sinensis",
    "Cleistoformis patens": "Cleistopholis patens",
    "Corynante pachycercus": "Corynanthe pachyceras",

    "Dialium guinensis": "Dialium guineense",
    "Diallium guineense": "Dialium guineense",

    "Diospyrous dendo": "Diospyros dendo",
    "Diospyrous melistoformis": "Diospyros mespiliformis",
    "Diospyrous mespiliformis": "Diospyros mespiliformis",

    "Dracena manni": "Dracaena mannii",

    "Elaeis guineense": "Elaeis guineensis",
    "Elaeis guinensis": "Elaeis guineensis",

    "Ficus mucosa": "Ficus mucuso",

    "Irvinga gabonensis": "Irvingia gabonensis",

    "Lecaniodiscus cupaniodes": "Lecaniodiscus cupanioides",

    "Macranga bateri": "Macaranga barteri",

    "Margartaria discoidea": "Margaritaria discoidea",

    "Millettia thoningii": "Millettia thonningii",

    "Mitragyna cilliata": "Mitragyna ciliata",

    "Musanga cecropiodes": "Musanga cecropioides",
    "Musanga cecropoides": "Musanga cecropioides",
    "Musanga cecropoidies": "Musanga cecropioides",

    "Nauclea diderrichi": "Nauclea diderrichii",
    "Nuclea diderrichi": "Nauclea diderrichii",

    "Nesodogornia papaverifera": "Nesogordonia papaverifera",
    "Nesodogornia papaverivera": "Nesogordonia papaverifera",

    "Nothospondia staudtii": "Nothospondias staudtii",

    "Pavetta corymbosia": "Pavetta corymbosa",

    "Pterocarpus santoxyloides": "Pterocarpus santalinoides",

    "Pycanthus angolensis": "Pycnanthus angolensis",

    "Raulvolfia vomitoria": "Rauvolfia vomitoria",
    "Rauvolvia vomitoria": "Rauvolfia vomitoria",

    "Ricinidendron heudeloti": "Ricinodendron heudelotii",
    "Ricinodendron heudeloti": "Ricinodendron heudelotii",
    "Ricinodendron heudeotii": "Ricinodendron heudelotii",

    "Rothmania hispida": "Rothmannia hispida",

    "Scotellia coriacea": "Scottellia coriacea",
    "Scotellia coriaceae": "Scottellia coriacea",
    "Scotellia coricea": "Scottellia coriacea",

    "Spondia mombin": "Spondias mombin",

    "Strombosia postulata": "Strombosia pustulata",
    "Strombosia putulata": "Strombosia pustulata",

    "Termialia superba": "Terminalia superba",

    "Uapaca heudeloti": "Uapaca heudelotii",

    "Zanthoxylum xanthoxyloides": "Zanthoxylum zanthoxyloides",
}


# -----------------------------------------------------------------------------
# 3. Identify genus-only targets that must leave species acquisition
# -----------------------------------------------------------------------------

genus_only_names = {
    "Drypetes",
}


# -----------------------------------------------------------------------------
# 4. Build final controlled matching fields
# -----------------------------------------------------------------------------

sla_try_resolved = sla_try_crosswalk.copy()

sla_try_resolved["Resolved_TRY_Name"] = pd.NA
sla_try_resolved["Resolved_TRY_ID"] = pd.Series(
    pd.NA,
    index=sla_try_resolved.index,
    dtype="Int64"
)

sla_try_resolved["Resolution_Status"] = "UNRESOLVED"
sla_try_resolved["Resolution_Method"] = pd.NA


# -----------------------------------------------------------------------------
# 5. Preserve exact matches
# -----------------------------------------------------------------------------

exact_mask = (
    sla_try_resolved["TRY_Match_Status"]
    .eq("EXACT_ACCEPTED_NAME")
)

sla_try_resolved.loc[
    exact_mask,
    "Resolved_TRY_Name"
] = sla_try_resolved.loc[
    exact_mask,
    "AccSpeciesName"
]

sla_try_resolved.loc[
    exact_mask,
    "Resolved_TRY_ID"
] = sla_try_resolved.loc[
    exact_mask,
    "AccSpeciesID"
].astype("Int64")

sla_try_resolved.loc[
    exact_mask,
    "Resolution_Status"
] = "RESOLVED"

sla_try_resolved.loc[
    exact_mask,
    "Resolution_Method"
] = "EXACT_ACCEPTED_NAME"


# -----------------------------------------------------------------------------
# 6. Apply controlled orthographic mappings
# -----------------------------------------------------------------------------

try_name_to_id = (
    try_species[
        ["AccSpeciesName", "AccSpeciesID"]
    ]
    .drop_duplicates("AccSpeciesName")
    .set_index("AccSpeciesName")["AccSpeciesID"]
    .to_dict()
)


for omo_name, try_name in controlled_try_name_map.items():

    mask = (
        sla_try_resolved["SLA_Query_Name"]
        .eq(omo_name)
    )

    if not mask.any():
        continue

    try_id = try_name_to_id.get(try_name)

    if pd.isna(try_id):
        # Mapping remains unresolved if the target accepted name
        # is absent from the current TRY catalogue.
        continue

    sla_try_resolved.loc[
        mask,
        "Resolved_TRY_Name"
    ] = try_name

    sla_try_resolved.loc[
        mask,
        "Resolved_TRY_ID"
    ] = int(try_id)

    sla_try_resolved.loc[
        mask,
        "Resolution_Status"
    ] = "RESOLVED"

    sla_try_resolved.loc[
        mask,
        "Resolution_Method"
    ] = "CONTROLLED_ORTHOGRAPHIC_MATCH"


# -----------------------------------------------------------------------------
# 7. Reclassify genus-only targets
# -----------------------------------------------------------------------------

genus_mask = (
    sla_try_resolved["SLA_Query_Name"]
    .isin(genus_only_names)
)

sla_try_resolved.loc[
    genus_mask,
    "Resolved_TRY_Name"
] = pd.NA

sla_try_resolved.loc[
    genus_mask,
    "Resolved_TRY_ID"
] = pd.NA

sla_try_resolved.loc[
    genus_mask,
    "Resolution_Status"
] = "GENUS_LEVEL"

sla_try_resolved.loc[
    genus_mask,
    "Resolution_Method"
] = "REMOVED_FROM_SPECIES_QUERY"


# -----------------------------------------------------------------------------
# 8. Identify unresolved taxa
# -----------------------------------------------------------------------------

unresolved_mask = (
    sla_try_resolved["Resolution_Status"]
    .eq("UNRESOLVED")
)

resolved_mask = (
    sla_try_resolved["Resolution_Status"]
    .eq("RESOLVED")
)


# -----------------------------------------------------------------------------
# 9. Build unique TRY ID request list
# -----------------------------------------------------------------------------

resolved_ids = (
    sla_try_resolved.loc[
        resolved_mask,
        "Resolved_TRY_ID"
    ]
    .dropna()
    .astype(int)
)

unique_try_ids = (
    sorted(resolved_ids.unique())
)

try_id_string = ",".join(
    str(x) for x in unique_try_ids
)


# -----------------------------------------------------------------------------
# 10. Diagnostics
# -----------------------------------------------------------------------------

n_exact = (
    sla_try_resolved["Resolution_Method"]
    .eq("EXACT_ACCEPTED_NAME")
    .sum()
)

n_controlled = (
    sla_try_resolved["Resolution_Method"]
    .eq("CONTROLLED_ORTHOGRAPHIC_MATCH")
    .sum()
)

n_genus = (
    sla_try_resolved["Resolution_Status"]
    .eq("GENUS_LEVEL")
    .sum()
)

n_unresolved = unresolved_mask.sum()

n_resolved_rows = resolved_mask.sum()
n_unique_ids = len(unique_try_ids)


# -----------------------------------------------------------------------------
# 11. Export
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

resolved_crosswalk_path = (
    output_dir
    / "omo_sla_try_taxonomic_crosswalk_resolved.csv"
)

unresolved_path = (
    output_dir
    / "omo_sla_try_taxa_unresolved_after_controlled_matching.csv"
)

try_id_path = (
    output_dir
    / "omo_sla_try_accspecies_ids.txt"
)


sla_try_resolved.to_csv(
    resolved_crosswalk_path,
    index=False
)

sla_try_resolved.loc[
    unresolved_mask,
    [
        "SLA_Query_ID",
        "Canonical_Species",
        "SLA_Query_Name",
        "Suggested_TRY_Name",
        "Suggested_TRY_ID",
    ]
].to_csv(
    unresolved_path,
    index=False
)

with open(
    try_id_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(try_id_string)


# -----------------------------------------------------------------------------
# 12. Display diagnostics
# -----------------------------------------------------------------------------

print("\nControlled TRY taxonomic resolution")
print("-" * 100)

print(f"Exact accepted-name matches      : {n_exact}")
print(f"Controlled orthographic matches  : {n_controlled}")
print(f"Genus-level targets removed      : {n_genus}")
print(f"Still unresolved                 : {n_unresolved}")
print(f"Resolved Omo query rows          : {n_resolved_rows}")
print(f"Unique resolved TRY IDs          : {n_unique_ids}")


print("\nGenus-level taxa removed from species query")
print("-" * 100)

display(
    sla_try_resolved.loc[
        genus_mask,
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Resolution_Status",
        ]
    ]
)


print("\nTaxa still requiring taxonomic review")
print("-" * 100)

display(
    sla_try_resolved.loc[
        unresolved_mask,
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Suggested_TRY_Name",
            "Suggested_TRY_ID",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Resolved crosswalk : "
    f"{resolved_crosswalk_path}"
)

print(
    f"Unresolved taxa    : "
    f"{unresolved_path}"
)

print(
    f"TRY AccSpecies IDs : "
    f"{try_id_path}"
)

print("\nSection 8.3C completed successfully.")
```

### 8.3E Finalization of the Primary TRY Species Request

Following exact, controlled orthographic, and verified taxonomic reconciliation, the primary TRY acquisition set is finalized using only species for which a defensible correspondence to a current TRY accepted-species identifier has been established.

Taxa that remain unresolved are not forced into the TRY request through approximate string similarity or uncertain synonymy. Instead, they are retained as explicitly unresolved acquisition targets and will be revisited during the supplementary-source phase using additional trait databases, literature searches, or controlled genus-level estimation where necessary.

This conservative treatment prevents erroneous external taxonomic assignments from propagating into the functional-trait dataset. It also maintains the distinction between the validated Omo Forest canonical taxonomy and the taxonomic representations used solely for retrieving trait observations.

The finalized TRY request therefore consists of the unique accepted-species identifiers derived from resolved Omo query records. Because several Omo canonical records correspond to the same accepted TRY taxon, the number of unique TRY species identifiers is lower than the number of resolved Omo query rows.

```python
# =============================================================================
# SECTION 8.3E — FINALIZE PRIMARY TRY SLA REQUEST
# =============================================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("SECTION 8.3E — FINALIZE PRIMARY TRY SLA REQUEST")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Work from verified crosswalk
# -----------------------------------------------------------------------------

if "sla_try_resolved" not in globals():

    verified_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_try_taxonomic_crosswalk_verified.csv"
    )

    if not verified_path.exists():
        raise FileNotFoundError(
            "Verified TRY taxonomic crosswalk not found. "
            "Run Section 8.3D first."
        )

    sla_try_resolved = pd.read_csv(verified_path)


# -----------------------------------------------------------------------------
# 2. Define primary TRY request eligibility
# -----------------------------------------------------------------------------

resolved_mask = (
    sla_try_resolved["Resolution_Status"]
    .eq("RESOLVED")
)

unresolved_mask = (
    sla_try_resolved["Resolution_Status"]
    .eq("UNRESOLVED")
)

genus_mask = (
    sla_try_resolved["Resolution_Status"]
    .eq("GENUS_LEVEL")
)


# -----------------------------------------------------------------------------
# 3. Build final unique TRY request table
# -----------------------------------------------------------------------------

try_request_table = (
    sla_try_resolved.loc[
        resolved_mask,
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Resolved_TRY_Name",
            "Resolved_TRY_ID",
            "Resolution_Method",
        ]
    ]
    .copy()
)

try_request_unique = (
    try_request_table[
        [
            "Resolved_TRY_ID",
            "Resolved_TRY_Name",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "Resolved_TRY_Name",
            "Resolved_TRY_ID",
        ]
    )
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 4. Generate final AccSpeciesID string
# -----------------------------------------------------------------------------

final_try_ids = (
    try_request_unique["Resolved_TRY_ID"]
    .dropna()
    .astype(int)
    .tolist()
)

final_try_id_string = ",".join(
    str(x) for x in final_try_ids
)


# -----------------------------------------------------------------------------
# 5. Prepare deferred taxa table
# -----------------------------------------------------------------------------

sla_try_deferred = (
    sla_try_resolved.loc[
        unresolved_mask | genus_mask,
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Resolution_Status",
            "Suggested_TRY_Name",
            "Suggested_TRY_ID",
        ]
    ]
    .copy()
)

sla_try_deferred["Next_Acquisition_Strategy"] = (
    "SUPPLEMENTARY_SOURCE_REVIEW"
)

sla_try_deferred.loc[
    genus_mask,
    "Next_Acquisition_Strategy"
] = "GENUS_LEVEL_STRATEGY"


# -----------------------------------------------------------------------------
# 6. Validation
# -----------------------------------------------------------------------------

n_resolved_rows = resolved_mask.sum()
n_unique_try_ids = len(final_try_ids)
n_unresolved = unresolved_mask.sum()
n_genus = genus_mask.sum()
n_deferred = len(sla_try_deferred)

assert len(set(final_try_ids)) == n_unique_try_ids

assert try_request_unique["Resolved_TRY_ID"].notna().all()

assert (
    n_resolved_rows
    + n_unresolved
    + n_genus
    == len(sla_try_resolved)
)


# -----------------------------------------------------------------------------
# 7. Export final TRY request files
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

final_request_path = (
    output_dir
    / "omo_sla_try_primary_request_species.csv"
)

final_id_path = (
    output_dir
    / "omo_sla_try_primary_request_accspecies_ids.txt"
)

deferred_path = (
    output_dir
    / "omo_sla_try_deferred_taxa.csv"
)


try_request_unique.to_csv(
    final_request_path,
    index=False
)

with open(
    final_id_path,
    "w",
    encoding="utf-8"
) as f:
    f.write(final_try_id_string)

sla_try_deferred.to_csv(
    deferred_path,
    index=False
)


# -----------------------------------------------------------------------------
# 8. Diagnostics
# -----------------------------------------------------------------------------

print("\nPrimary TRY request finalization")
print("-" * 100)

print(
    f"Resolved Omo query rows          : "
    f"{n_resolved_rows}"
)

print(
    f"Unique TRY species IDs           : "
    f"{n_unique_try_ids}"
)

print(
    f"Unresolved taxa deferred         : "
    f"{n_unresolved}"
)

print(
    f"Genus-level taxa deferred        : "
    f"{n_genus}"
)

print(
    f"Total deferred taxa              : "
    f"{n_deferred}"
)


print("\nDeferred taxa")
print("-" * 100)

display(
    sla_try_deferred[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Resolution_Status",
            "Next_Acquisition_Strategy",
        ]
    ]
)


print("\nFinal TRY request preview")
print("-" * 100)

display(
    try_request_unique.head(20)
)


print("\nFinal AccSpeciesID string")
print("-" * 100)

print(final_try_id_string)


print("\nExports")
print("-" * 100)

print(
    f"TRY request table : "
    f"{final_request_path}"
)

print(
    f"TRY ID list       : "
    f"{final_id_path}"
)

print(
    f"Deferred taxa     : "
    f"{deferred_path}"
)

print("\nSection 8.3E completed successfully.")
```

### 8.4 Supplementary SLA Acquisition for TRY-Deferred Taxa

The primary TRY acquisition workflow successfully established defensible accepted-species identifiers for most of the species-level SLA query records. However, a small subset of taxa could not be incorporated into the primary request without introducing uncertain taxonomic substitutions. These records are therefore retained as deferred acquisition targets rather than being matched automatically to superficially similar TRY names.

The deferred set contains two conceptually different groups. The first comprises species-level Omo taxa whose source names could not yet be reconciled confidently with the current TRY accepted-species catalogue. These taxa require additional taxonomic verification followed, where possible, by acquisition from supplementary trait databases or traceable published sources. The second comprises genus-level targets for which direct species-level SLA assignment would be inappropriate. Such records require a separately documented genus-level estimation strategy.

Supplementary acquisition follows the same provenance principles established for the overall functional-trait workflow. Direct species-level measurements remain preferable to estimates, and no genus- or family-derived value will be represented as an observed species-level SLA measurement. Candidate values will retain their original taxonomic name, source, unit, measurement definition, and acquisition method before any harmonization or aggregation is performed.

This subsection therefore establishes a supplementary acquisition register for all TRY-deferred taxa. The register separates taxonomic resolution from trait acquisition and provides explicit fields for recording subsequent database, literature, genus-level, or broader fallback searches.

The TRY primary request remains independently documented as **TRY Data Request 51853**, covering Trait IDs **3115, 3116, and 3117** and the verified set of 82 unique TRY accepted-species identifiers. Results from that request will later be integrated with supplementary SLA records only after both acquisition streams have undergone independent quality control.

```python
# =============================================================================
# SECTION 8.4A — CONSTRUCT SUPPLEMENTARY SLA ACQUISITION REGISTER
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.4A — SUPPLEMENTARY SLA ACQUISITION REGISTER")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Load the deferred TRY taxa
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

if "sla_try_deferred" not in globals():

    deferred_path = (
        output_dir
        / "omo_sla_try_deferred_taxa.csv"
    )

    if not deferred_path.exists():
        raise FileNotFoundError(
            "Deferred SLA taxa table was not found. "
            "Run Section 8.3E before continuing."
        )

    sla_try_deferred = pd.read_csv(deferred_path)

    print("\nDeferred taxa table reloaded from:")
    print(deferred_path)

else:
    print("\nUsing deferred taxa table already present in notebook memory.")


# -----------------------------------------------------------------------------
# 2. Validate deferred acquisition foundation
# -----------------------------------------------------------------------------

required_columns = {
    "SLA_Query_ID",
    "Canonical_Species",
    "SLA_Query_Name",
    "Resolution_Status",
    "Next_Acquisition_Strategy",
}

missing_columns = required_columns.difference(
    sla_try_deferred.columns
)

if missing_columns:
    raise ValueError(
        "Deferred taxa table is missing required columns: "
        f"{sorted(missing_columns)}"
    )


assert len(sla_try_deferred) == 11, (
    f"Expected 11 deferred SLA records, "
    f"found {len(sla_try_deferred)}."
)

assert (
    sla_try_deferred["SLA_Query_ID"]
    .nunique()
    == 11
), "Duplicate SLA query IDs detected."


# -----------------------------------------------------------------------------
# 3. Construct supplementary acquisition register
# -----------------------------------------------------------------------------

sla_supplementary_register = (
    sla_try_deferred.copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 4. Classify acquisition pathway
# -----------------------------------------------------------------------------

sla_supplementary_register[
    "Supplementary_Target_Class"
] = np.where(
    sla_supplementary_register[
        "Resolution_Status"
    ].eq("GENUS_LEVEL"),
    "GENUS_LEVEL_TARGET",
    "SPECIES_LEVEL_UNRESOLVED"
)


# -----------------------------------------------------------------------------
# 5. Define ordered search strategy
# -----------------------------------------------------------------------------

sla_supplementary_register[
    "Search_Priority_1"
] = np.where(
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ].eq("SPECIES_LEVEL_UNRESOLVED"),
    "TAXONOMIC_VERIFICATION",
    "CONGENERIC_SLA_RECORDS"
)

sla_supplementary_register[
    "Search_Priority_2"
] = np.where(
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ].eq("SPECIES_LEVEL_UNRESOLVED"),
    "SUPPLEMENTARY_TRAIT_DATABASE",
    "GENUS_LEVEL_AGGREGATION"
)

sla_supplementary_register[
    "Search_Priority_3"
] = np.where(
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ].eq("SPECIES_LEVEL_UNRESOLVED"),
    "PEER_REVIEWED_LITERATURE",
    "FAMILY_LEVEL_REVIEW"
)

sla_supplementary_register[
    "Search_Priority_4"
] = np.where(
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ].eq("SPECIES_LEVEL_UNRESOLVED"),
    "GENUS_LEVEL_ESTIMATE_IF_NEEDED",
    "CONTROLLED_FALLBACK_IF_NEEDED"
)


# -----------------------------------------------------------------------------
# 6. Add acquisition and provenance fields
# -----------------------------------------------------------------------------

new_fields = {
    "Verified_Taxon_Name": pd.NA,
    "Taxonomic_Verification_Source": pd.NA,
    "Taxonomic_Verification_Status": "PENDING",

    "Trait_Search_Name": pd.NA,

    "Supplementary_Source_Type": pd.NA,
    "Supplementary_Source_Name": pd.NA,
    "Supplementary_Source_Record_ID": pd.NA,
    "Supplementary_Source_Citation": pd.NA,
    "Supplementary_Source_DOI_or_URL": pd.NA,

    "Raw_SLA_Value": np.nan,
    "Raw_SLA_Unit": pd.NA,
    "Measurement_Definition": pd.NA,

    "Observation_Count": pd.NA,
    "Source_Min_SLA": np.nan,
    "Source_Median_SLA": np.nan,
    "Source_Mean_SLA": np.nan,
    "Source_Max_SLA": np.nan,

    "Selected_SLA_Value": np.nan,
    "Selection_Method": pd.NA,

    "Proposed_Provenance_Class": pd.NA,
    "Supplementary_Match_Status": "NOT_YET_SEARCHED",
    "Supplementary_Quality_Flag": "PENDING",

    "Acquisition_Notes": pd.NA,
}


for column, default in new_fields.items():

    if column not in sla_supplementary_register.columns:
        sla_supplementary_register[column] = default


# -----------------------------------------------------------------------------
# 7. Establish initial trait-search names
# -----------------------------------------------------------------------------

species_mask = (
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ].eq("SPECIES_LEVEL_UNRESOLVED")
)

genus_mask = (
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ].eq("GENUS_LEVEL_TARGET")
)


# Species-level unresolved records initially retain their Omo query name.
sla_supplementary_register.loc[
    species_mask,
    "Trait_Search_Name"
] = sla_supplementary_register.loc[
    species_mask,
    "SLA_Query_Name"
]


# Genus-level records use the genus only.
sla_supplementary_register.loc[
    genus_mask,
    "Trait_Search_Name"
] = (
    sla_supplementary_register.loc[
        genus_mask,
        "SLA_Query_Name"
    ]
    .astype(str)
    .str.strip()
    .str.split()
    .str[0]
)


# -----------------------------------------------------------------------------
# 8. Record TRY request provenance separately
# -----------------------------------------------------------------------------

try_request_provenance = pd.DataFrame(
    [
        {
            "Request_ID": 51853,
            "Database": "TRY",
            "Trait_IDs": "3115, 3116, 3117",
            "Requested_Data_Class": "PUBLIC_ONLY",
            "Unique_TRY_Species_IDs": 82,
            "Total_Trait_Measurements_Available": 647,
            "Public_Trait_Measurements_Requested": 535,
            "Restricted_Measurements_Not_Requested": 112,
            "Request_Status": "SUBMITTED_AWAITING_DOWNLOAD",
            "Project_Title": (
                "Omo Forest Ecological Informatics: "
                "Functional Trait Harmonization and Biodiversity Analysis"
            ),
        }
    ]
)


# -----------------------------------------------------------------------------
# 9. Validation
# -----------------------------------------------------------------------------

n_total = len(sla_supplementary_register)

n_species_unresolved = (
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ]
    .eq("SPECIES_LEVEL_UNRESOLVED")
    .sum()
)

n_genus_targets = (
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ]
    .eq("GENUS_LEVEL_TARGET")
    .sum()
)


assert n_total == 11

assert n_species_unresolved == 10, (
    f"Expected 10 unresolved species-level targets, "
    f"found {n_species_unresolved}."
)

assert n_genus_targets == 1, (
    f"Expected 1 genus-level deferred target, "
    f"found {n_genus_targets}."
)

assert (
    sla_supplementary_register[
        "Trait_Search_Name"
    ]
    .notna()
    .all()
)

assert (
    try_request_provenance.loc[
        0,
        "Public_Trait_Measurements_Requested"
    ]
    == 535
)

assert (
    try_request_provenance.loc[
        0,
        "Total_Trait_Measurements_Available"
    ]
    == 647
)


# -----------------------------------------------------------------------------
# 10. Export supplementary register and TRY request provenance
# -----------------------------------------------------------------------------

supplementary_register_path = (
    output_dir
    / "omo_sla_supplementary_acquisition_register.csv"
)

try_request_provenance_path = (
    output_dir
    / "omo_sla_try_request_51853_provenance.csv"
)


sla_supplementary_register.to_csv(
    supplementary_register_path,
    index=False
)

try_request_provenance.to_csv(
    try_request_provenance_path,
    index=False
)


# -----------------------------------------------------------------------------
# 11. Diagnostics
# -----------------------------------------------------------------------------

print("\nSupplementary SLA acquisition register")
print("-" * 100)

print(
    f"Total deferred records          : "
    f"{n_total}"
)

print(
    f"Species-level unresolved        : "
    f"{n_species_unresolved}"
)

print(
    f"Genus-level targets             : "
    f"{n_genus_targets}"
)


print("\nAcquisition pathway distribution")
print("-" * 100)

print(
    sla_supplementary_register[
        "Supplementary_Target_Class"
    ]
    .value_counts()
    .to_string()
)


print("\nSupplementary acquisition targets")
print("-" * 100)

display(
    sla_supplementary_register[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Supplementary_Target_Class",
            "Trait_Search_Name",
            "Search_Priority_1",
            "Search_Priority_2",
        ]
    ]
)


print("\nTRY request provenance")
print("-" * 100)

display(try_request_provenance)


print("\nExports")
print("-" * 100)

print(
    f"Supplementary register : "
    f"{supplementary_register_path}"
)

print(
    f"TRY request provenance : "
    f"{try_request_provenance_path}"
)

print("\nSection 8.4A completed successfully.")
```

### 8.4B Controlled Taxonomic Verification of Supplementary SLA Targets

Before supplementary SLA values are searched or assigned, the unresolved species-level targets are subjected to an independent taxonomic verification step. This step is restricted to determining defensible trait-search aliases and does not alter the validated Omo Forest canonical taxonomy.

Accepted-name resources are used to distinguish three situations:

1. **VERIFIED_ACCEPTED_OR_SYNONYM** — the Omo query name can be linked confidently to a currently accepted taxon or documented synonym;
2. **SEARCH_ALIAS_WITH_NOMENCLATURAL_NOTE** — a practical trait-search name can be used, but the taxonomic relationship is more complex than a simple accepted-name substitution;
3. **UNRESOLVED_REQUIRES_SOURCE_REVIEW** — the available evidence is insufficient for a reliable external taxonomic assignment.

Only the first two categories are allowed to proceed directly to supplementary species-level SLA searches. Taxa remaining unresolved will retain their original Omo canonical identity and will later be evaluated through source-document review, genus-level trait information, or other controlled fallback procedures.

The genus-level `Drypetes` target remains outside this species-level verification process and continues through the separate congeneric SLA pathway.

```python
# =============================================================================
# SECTION 8.4B — CONTROLLED TAXONOMIC VERIFICATION OF SUPPLEMENTARY SLA TARGETS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.4B — CONTROLLED TAXONOMIC VERIFICATION OF SUPPLEMENTARY SLA TARGETS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm supplementary register is available
# -----------------------------------------------------------------------------

if "sla_supplementary_register" not in globals():

    supplementary_register_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_supplementary_acquisition_register.csv"
    )

    if not supplementary_register_path.exists():
        raise FileNotFoundError(
            "Supplementary SLA acquisition register not found. "
            "Run Section 8.4A first."
        )

    sla_supplementary_register = pd.read_csv(
        supplementary_register_path
    )


# -----------------------------------------------------------------------------
# 2. Controlled verification registry
#    This is an acquisition alias registry only.
#    It does NOT alter Canonical_Species.
# -----------------------------------------------------------------------------

taxonomic_verification_registry = pd.DataFrame(
    [
        {
            "SLA_Query_Name": "Aningeria robusta",
            "Verified_Taxon_Name": "Aningeria pierrei",
            "Verification_Status": "VERIFIED_ACCEPTED_OR_SYNONYM",
            "Verification_Source": "POWO",
            "Verification_Note": (
                "Aningeria robusta is treated as a synonym "
                "of Aningeria pierrei."
            ),
        },

        {
            "SLA_Query_Name": "Citrus sinenses",
            "Verified_Taxon_Name": "Citrus sinensis",
            "Verification_Status": "SEARCH_ALIAS_WITH_NOMENCLATURAL_NOTE",
            "Verification_Source": "POWO",
            "Verification_Note": (
                "Source spelling normalized to Citrus sinensis. "
                "Current POWO places Citrus × sinensis within "
                "the Citrus × aurantium complex."
            ),
        },

        {
            "SLA_Query_Name": "Melacantha alnifolia",
            "Verified_Taxon_Name": "Malacantha alnifolia",
            "Verification_Status": "VERIFIED_ACCEPTED_OR_SYNONYM",
            "Verification_Source": "POWO",
            "Verification_Note": (
                "Orthographic correction; Malacantha alnifolia "
                "is an accepted tropical African species."
            ),
        },

        {
            "SLA_Query_Name": "Mitragyna cilliata",
            "Verified_Taxon_Name": "Mitragyna ciliata",
            "Verification_Status": "VERIFIED_ACCEPTED_OR_SYNONYM",
            "Verification_Source": "POWO",
            "Verification_Note": (
                "Orthographic correction; Mitragyna ciliata "
                "is an accepted West and Central African species."
            ),
        },

        {
            "SLA_Query_Name": "Albizia auxilary",
            "Verified_Taxon_Name": pd.NA,
            "Verification_Status": "UNRESOLVED_REQUIRES_SOURCE_REVIEW",
            "Verification_Source": "PENDING",
            "Verification_Note": (
                "No defensible current accepted-name mapping "
                "established."
            ),
        },

        {
            "SLA_Query_Name": "Amphimas tetracoides",
            "Verified_Taxon_Name": pd.NA,
            "Verification_Status": "UNRESOLVED_REQUIRES_SOURCE_REVIEW",
            "Verification_Source": "PENDING",
            "Verification_Note": (
                "No defensible current accepted-name mapping "
                "established."
            ),
        },

        {
            "SLA_Query_Name": "Diospyros undebunda",
            "Verified_Taxon_Name": pd.NA,
            "Verification_Status": "UNRESOLVED_REQUIRES_SOURCE_REVIEW",
            "Verification_Source": "PENDING",
            "Verification_Note": (
                "No defensible current accepted-name mapping "
                "established."
            ),
        },

        {
            "SLA_Query_Name": "Homalium alymeri",
            "Verified_Taxon_Name": pd.NA,
            "Verification_Status": "UNRESOLVED_REQUIRES_SOURCE_REVIEW",
            "Verification_Source": "PENDING",
            "Verification_Note": (
                "Name occurs in regional forest literature, "
                "but no authoritative current accepted-name "
                "resolution has yet been established."
            ),
        },

        {
            "SLA_Query_Name": "Lonchocarpus cilliata",
            "Verified_Taxon_Name": pd.NA,
            "Verification_Status": "UNRESOLVED_REQUIRES_SOURCE_REVIEW",
            "Verification_Source": "PENDING",
            "Verification_Note": (
                "Likely requires source-document or historical "
                "taxonomic review before external trait matching."
            ),
        },

        {
            "SLA_Query_Name": "Musa sapiensis",
            "Verified_Taxon_Name": pd.NA,
            "Verification_Status": "UNRESOLVED_REQUIRES_SOURCE_REVIEW",
            "Verification_Source": "PENDING",
            "Verification_Note": (
                "Name is widely encountered as a non-standard "
                "banana name but is not assigned automatically "
                "to a modern accepted Musa taxon."
            ),
        },
    ]
)


# -----------------------------------------------------------------------------
# 3. Merge verification results into supplementary register
# -----------------------------------------------------------------------------

verification_columns = [
    "SLA_Query_Name",
    "Verified_Taxon_Name",
    "Verification_Status",
    "Verification_Source",
    "Verification_Note",
]

sla_supplementary_verified = (
    sla_supplementary_register
    .drop(
        columns=[
            c for c in [
                "Verified_Taxon_Name",
                "Taxonomic_Verification_Source",
                "Taxonomic_Verification_Status",
            ]
            if c in sla_supplementary_register.columns
        ]
    )
    .merge(
        taxonomic_verification_registry[
            verification_columns
        ],
        how="left",
        on="SLA_Query_Name",
    )
)


# -----------------------------------------------------------------------------
# 4. Restore controlled names for genus-level target
# -----------------------------------------------------------------------------

genus_mask = (
    sla_supplementary_verified[
        "Supplementary_Target_Class"
    ].eq("GENUS_LEVEL_TARGET")
)

sla_supplementary_verified.loc[
    genus_mask,
    "Verification_Status"
] = "GENUS_LEVEL_TARGET"

sla_supplementary_verified.loc[
    genus_mask,
    "Verification_Source"
] = "OMO_CANONICAL_WORKFLOW"

sla_supplementary_verified.loc[
    genus_mask,
    "Verification_Note"
] = (
    "Genus-level acquisition target; species-level "
    "taxonomic resolution is not attempted."
)

sla_supplementary_verified.loc[
    genus_mask,
    "Verified_Taxon_Name"
] = "Drypetes"


# -----------------------------------------------------------------------------
# 5. Define trait-search name after verification
# -----------------------------------------------------------------------------

searchable_mask = (
    sla_supplementary_verified[
        "Verification_Status"
    ].isin(
        [
            "VERIFIED_ACCEPTED_OR_SYNONYM",
            "SEARCH_ALIAS_WITH_NOMENCLATURAL_NOTE",
        ]
    )
)

unresolved_mask = (
    sla_supplementary_verified[
        "Verification_Status"
    ].eq("UNRESOLVED_REQUIRES_SOURCE_REVIEW")
)


sla_supplementary_verified.loc[
    searchable_mask,
    "Trait_Search_Name"
] = (
    sla_supplementary_verified.loc[
        searchable_mask,
        "Verified_Taxon_Name"
    ]
)

sla_supplementary_verified.loc[
    unresolved_mask,
    "Trait_Search_Name"
] = (
    sla_supplementary_verified.loc[
        unresolved_mask,
        "SLA_Query_Name"
    ]
)


# -----------------------------------------------------------------------------
# 6. Define next acquisition action
# -----------------------------------------------------------------------------

sla_supplementary_verified[
    "Verification_Next_Action"
] = pd.NA

sla_supplementary_verified.loc[
    searchable_mask,
    "Verification_Next_Action"
] = "SEARCH_SUPPLEMENTARY_SPECIES_SLA"

sla_supplementary_verified.loc[
    unresolved_mask,
    "Verification_Next_Action"
] = "SOURCE_DOCUMENT_TAXONOMIC_REVIEW"

sla_supplementary_verified.loc[
    genus_mask,
    "Verification_Next_Action"
] = "SEARCH_CONGENERIC_SLA"


# -----------------------------------------------------------------------------
# 7. Validation
# -----------------------------------------------------------------------------

n_verified = (
    sla_supplementary_verified[
        "Verification_Status"
    ].eq("VERIFIED_ACCEPTED_OR_SYNONYM")
    .sum()
)

n_alias = (
    sla_supplementary_verified[
        "Verification_Status"
    ].eq("SEARCH_ALIAS_WITH_NOMENCLATURAL_NOTE")
    .sum()
)

n_unresolved = (
    sla_supplementary_verified[
        "Verification_Status"
    ].eq("UNRESOLVED_REQUIRES_SOURCE_REVIEW")
    .sum()
)

n_genus = (
    sla_supplementary_verified[
        "Verification_Status"
    ].eq("GENUS_LEVEL_TARGET")
    .sum()
)


assert len(sla_supplementary_verified) == 11

assert n_verified == 3
assert n_alias == 1
assert n_unresolved == 6
assert n_genus == 1

assert (
    n_verified
    + n_alias
    + n_unresolved
    + n_genus
    == 11
)


# -----------------------------------------------------------------------------
# 8. Export verified supplementary register
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

verified_register_path = (
    output_dir
    / "omo_sla_supplementary_taxonomic_verification.csv"
)

unresolved_review_path = (
    output_dir
    / "omo_sla_supplementary_taxa_requiring_source_review.csv"
)


sla_supplementary_verified.to_csv(
    verified_register_path,
    index=False
)

sla_supplementary_verified.loc[
    unresolved_mask
].to_csv(
    unresolved_review_path,
    index=False
)


# -----------------------------------------------------------------------------
# 9. Diagnostics
# -----------------------------------------------------------------------------

print("\nSupplementary taxonomic verification")
print("-" * 100)

print(
    f"Verified accepted/synonym taxa     : "
    f"{n_verified}"
)

print(
    f"Search aliases with nomenclatural note : "
    f"{n_alias}"
)

print(
    f"Still requiring source review      : "
    f"{n_unresolved}"
)

print(
    f"Genus-level targets                : "
    f"{n_genus}"
)


print("\nVerification status distribution")
print("-" * 100)

print(
    sla_supplementary_verified[
        "Verification_Status"
    ]
    .value_counts()
    .to_string()
)


print("\nSupplementary verification table")
print("-" * 100)

display(
    sla_supplementary_verified[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Verified_Taxon_Name",
            "Verification_Status",
            "Trait_Search_Name",
            "Verification_Next_Action",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Verified supplementary register : "
    f"{verified_register_path}"
)

print(
    f"Source-review taxa              : "
    f"{unresolved_review_path}"
)

print("\nSection 8.4B completed successfully.")
```

### 8.4C Internal Source Review of Persistently Unresolved SLA Taxa

Six species-level SLA targets remain unresolved after controlled comparison with TRY and external accepted-name resources. Because these names originate from the Omo Forest inventory and earlier taxonomic harmonization stages, further resolution should first consult the project's own validated taxonomic evidence rather than relying on approximate external name matching.

This subsection therefore revisits the taxonomic tables and harmonization outputs generated during the earlier Omo Forest workflow. The objective is not to redesign or overwrite the validated taxonomy, but to determine whether the unresolved SLA query names correspond to documented spelling corrections, historical source names, accepted canonical taxa, or previously recorded harmonization decisions.

For each unresolved name, all matching records found in available project taxonomic tables are collected using both exact and normalized text comparison. Relevant fields such as source name, canonical name, harmonized name, accepted name, family, match status, and taxonomic notes are retained where available.

No taxonomic decision is made automatically in this subsection. Instead, the resulting evidence table provides an auditable basis for deciding whether each unresolved SLA target can be assigned a defensible trait-search alias or should remain unresolved and proceed to genus-level or broader trait estimation.

```python
# =============================================================================
# SECTION 8.4C — INTERNAL SOURCE REVIEW OF PERSISTENTLY UNRESOLVED SLA TAXA
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print("SECTION 8.4C — INTERNAL SOURCE REVIEW OF PERSISTENTLY UNRESOLVED SLA TAXA")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Define the six unresolved species-level targets
# -----------------------------------------------------------------------------

persistent_unresolved = [
    "Albizia auxilary",
    "Amphimas tetracoides",
    "Diospyros undebunda",
    "Homalium alymeri",
    "Lonchocarpus cilliata",
    "Musa sapiensis",
]

print("\nPersistently unresolved SLA taxa")
print("-" * 100)

for name in persistent_unresolved:
    print(f"- {name}")


# -----------------------------------------------------------------------------
# 2. Conservative normalization function
# -----------------------------------------------------------------------------

def normalize_name(value):
    """
    Normalize taxonomic text for comparison only.
    Original values remain unchanged.
    """

    if pd.isna(value):
        return ""

    value = str(value).strip().lower()
    value = re.sub(r"\s+", " ", value)

    return value


target_keys = {
    normalize_name(name): name
    for name in persistent_unresolved
}


# -----------------------------------------------------------------------------
# 3. Candidate taxonomic column names
# -----------------------------------------------------------------------------

taxonomic_column_keywords = [
    "species",
    "taxon",
    "canonical",
    "accepted",
    "source",
    "harmon",
    "resolved",
    "original",
    "scientific",
    "name",
]


# -----------------------------------------------------------------------------
# 4. Search all DataFrames currently in notebook memory
# -----------------------------------------------------------------------------

evidence_records = []

for df_name, obj in list(globals().items()):

    if not isinstance(obj, pd.DataFrame):
        continue

    if obj.empty:
        continue

    candidate_columns = [
        col for col in obj.columns
        if any(
            keyword in str(col).lower()
            for keyword in taxonomic_column_keywords
        )
    ]

    if not candidate_columns:
        continue

    for col in candidate_columns:

        series = obj[col]

        # Skip obviously numeric columns
        if pd.api.types.is_numeric_dtype(series):
            continue

        normalized_series = (
            series
            .astype("string")
            .map(normalize_name)
        )

        for target_key, target_name in target_keys.items():

            exact_mask = normalized_series.eq(target_key)

            if exact_mask.any():

                matching_rows = obj.loc[
                    exact_mask
                ].copy()

                for row_index, row in matching_rows.iterrows():

                    evidence_records.append(
                        {
                            "Target_SLA_Name": target_name,
                            "Evidence_DataFrame": df_name,
                            "Evidence_Row_Index": row_index,
                            "Matched_Column": col,
                            "Matched_Value": row[col],
                            "Match_Type": "EXACT_NORMALIZED_TEXT",
                            "Full_Row": row.to_dict(),
                        }
                    )


# -----------------------------------------------------------------------------
# 5. Convert evidence into structured long-form table
# -----------------------------------------------------------------------------

evidence_summary_rows = []

preferred_evidence_fields = [
    "Canonical_Species",
    "Source_Species",
    "Source_Name",
    "Scientific_Name",
    "Accepted_Name",
    "Accepted_Species",
    "Harmonized_Name",
    "Resolved_Name",
    "Family",
    "Genus",
    "Match_Status",
    "Taxonomic_Status",
    "Taxonomic_Resolution",
    "Harmonization_Note",
    "Taxonomic_Note",
]


for record in evidence_records:

    row_dict = record["Full_Row"]

    summary = {
        "Target_SLA_Name": record["Target_SLA_Name"],
        "Evidence_DataFrame": record["Evidence_DataFrame"],
        "Evidence_Row_Index": record["Evidence_Row_Index"],
        "Matched_Column": record["Matched_Column"],
        "Matched_Value": record["Matched_Value"],
        "Match_Type": record["Match_Type"],
    }

    for field in preferred_evidence_fields:

        if field in row_dict:
            summary[field] = row_dict[field]

    evidence_summary_rows.append(summary)


internal_taxonomic_evidence = pd.DataFrame(
    evidence_summary_rows
)


# -----------------------------------------------------------------------------
# 6. Build per-target evidence counts
# -----------------------------------------------------------------------------

if internal_taxonomic_evidence.empty:

    evidence_counts = pd.DataFrame(
        {
            "Target_SLA_Name": persistent_unresolved,
            "Evidence_Record_Count": 0,
        }
    )

else:

    evidence_counts = (
        internal_taxonomic_evidence
        .groupby(
            "Target_SLA_Name",
            as_index=False
        )
        .size()
        .rename(
            columns={
                "size": "Evidence_Record_Count"
            }
        )
    )

    evidence_counts = (
        pd.DataFrame(
            {
                "Target_SLA_Name": persistent_unresolved
            }
        )
        .merge(
            evidence_counts,
            how="left",
            on="Target_SLA_Name"
        )
    )

    evidence_counts[
        "Evidence_Record_Count"
    ] = (
        evidence_counts[
            "Evidence_Record_Count"
        ]
        .fillna(0)
        .astype(int)
    )


# -----------------------------------------------------------------------------
# 7. Search likely exported taxonomic CSV files if available
# -----------------------------------------------------------------------------

project_search_roots = [
    Path("outputs"),
    Path("data"),
]

csv_evidence_records = []

for root in project_search_roots:

    if not root.exists():
        continue

    for csv_path in root.rglob("*.csv"):

        try:
            df = pd.read_csv(
                csv_path,
                low_memory=False
            )
        except Exception:
            continue

        candidate_columns = [
            col for col in df.columns
            if any(
                keyword in str(col).lower()
                for keyword in taxonomic_column_keywords
            )
        ]

        if not candidate_columns:
            continue

        for col in candidate_columns:

            if pd.api.types.is_numeric_dtype(df[col]):
                continue

            norm_series = (
                df[col]
                .astype("string")
                .map(normalize_name)
            )

            for target_key, target_name in target_keys.items():

                mask = norm_series.eq(target_key)

                if not mask.any():
                    continue

                matches = df.loc[mask]

                for row_index, row in matches.iterrows():

                    record = {
                        "Target_SLA_Name": target_name,
                        "Evidence_File": str(csv_path),
                        "Evidence_Row_Index": row_index,
                        "Matched_Column": col,
                        "Matched_Value": row[col],
                        "Match_Type": "EXACT_NORMALIZED_TEXT",
                    }

                    for field in preferred_evidence_fields:

                        if field in row.index:
                            record[field] = row[field]

                    csv_evidence_records.append(
                        record
                    )


file_taxonomic_evidence = pd.DataFrame(
    csv_evidence_records
)


# -----------------------------------------------------------------------------
# 8. Export evidence tables
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"
output_dir.mkdir(parents=True, exist_ok=True)

memory_evidence_path = (
    output_dir
    / "omo_sla_internal_taxonomic_evidence_memory.csv"
)

file_evidence_path = (
    output_dir
    / "omo_sla_internal_taxonomic_evidence_files.csv"
)

evidence_counts_path = (
    output_dir
    / "omo_sla_internal_taxonomic_evidence_counts.csv"
)


internal_taxonomic_evidence.to_csv(
    memory_evidence_path,
    index=False
)

file_taxonomic_evidence.to_csv(
    file_evidence_path,
    index=False
)

evidence_counts.to_csv(
    evidence_counts_path,
    index=False
)


# -----------------------------------------------------------------------------
# 9. Diagnostics
# -----------------------------------------------------------------------------

print("\nInternal taxonomic evidence counts")
print("-" * 100)

display(evidence_counts)


print("\nEvidence found in notebook DataFrames")
print("-" * 100)

if internal_taxonomic_evidence.empty:

    print(
        "No exact normalized matches were found "
        "in currently loaded DataFrames."
    )

else:

    display(
        internal_taxonomic_evidence
        .sort_values(
            [
                "Target_SLA_Name",
                "Evidence_DataFrame",
            ]
        )
    )


print("\nEvidence found in exported project CSV files")
print("-" * 100)

if file_taxonomic_evidence.empty:

    print(
        "No exact normalized matches were found "
        "in searched project CSV files."
    )

else:

    display(
        file_taxonomic_evidence
        .sort_values(
            [
                "Target_SLA_Name",
                "Evidence_File",
            ]
        )
    )


print("\nExports")
print("-" * 100)

print(
    f"Notebook-memory evidence : "
    f"{memory_evidence_path}"
)

print(
    f"Project-file evidence    : "
    f"{file_evidence_path}"
)

print(
    f"Evidence counts          : "
    f"{evidence_counts_path}"
)

print("\nSection 8.4C completed successfully.")
```

### 8.4D Targeted Review of Pre-Functional-Trait Taxonomic Evidence

The broad internal evidence search identified numerous occurrences of the six persistently unresolved SLA query names. However, most of these occurrences originated from functional-trait tables created during the current notebook and therefore represent repeated propagation of the same canonical names rather than independent evidence of taxonomic identity.

To avoid circular taxonomic reasoning, the review is refined to consider only project records established **before functional-trait acquisition**. Outputs generated during the SLA and wood-density workflows are excluded. Particular attention is given to taxonomic standardization, harmonization, community-data restructuring, species metadata, and structural-data reconciliation products originating from earlier Omo Forest notebooks.

The objective is to determine whether any of the six unresolved names previously had a documented relationship between a source taxon and a canonical, accepted, corrected, or harmonized taxon. Only such pre-existing mappings can legitimately inform the supplementary SLA acquisition strategy without retrospectively modifying the validated project taxonomy.

This subsection remains diagnostic. No taxonomic substitutions or SLA values are assigned automatically.

```python
# =============================================================================
# SECTION 8.4D — TARGETED REVIEW OF PRE-FUNCTIONAL-TRAIT TAXONOMIC EVIDENCE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print("SECTION 8.4D — TARGETED REVIEW OF PRE-FUNCTIONAL-TRAIT TAXONOMIC EVIDENCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Persistently unresolved SLA targets
# -----------------------------------------------------------------------------

persistent_unresolved = [
    "Albizia auxilary",
    "Amphimas tetracoides",
    "Diospyros undebunda",
    "Homalium alymeri",
    "Lonchocarpus cilliata",
    "Musa sapiensis",
]


def normalize_taxon(value):

    if pd.isna(value):
        return ""

    value = str(value).strip().lower()
    value = re.sub(r"\s+", " ", value)

    return value


target_keys = {
    normalize_taxon(name): name
    for name in persistent_unresolved
}


# -----------------------------------------------------------------------------
# 2. Columns likely to contain useful taxonomic mapping evidence
# -----------------------------------------------------------------------------

mapping_keywords = [
    "source_species",
    "source_name",
    "original",
    "raw_species",
    "scientific",
    "accepted",
    "canonical",
    "harmon",
    "resolved",
    "standard",
    "match",
    "taxonomic",
    "species",
]


# -----------------------------------------------------------------------------
# 3. Files/directories explicitly excluded to avoid circular evidence
# -----------------------------------------------------------------------------

excluded_path_fragments = [
    "functional_traits",
    "omo_sla_",
    "wood_density",
    "wd_",
]

search_roots = [
    Path("outputs"),
    Path("data"),
]


# -----------------------------------------------------------------------------
# 4. Search only pre-functional-trait CSV evidence
# -----------------------------------------------------------------------------

records = []

for root in search_roots:

    if not root.exists():
        continue

    for csv_path in root.rglob("*.csv"):

        path_text = str(csv_path).lower()

        # Exclude current functional-trait outputs
        if any(
            fragment.lower() in path_text
            for fragment in excluded_path_fragments
        ):
            continue

        try:
            df = pd.read_csv(
                csv_path,
                low_memory=False
            )
        except Exception:
            continue

        if df.empty:
            continue

        candidate_columns = [
            col for col in df.columns
            if any(
                keyword in str(col).lower()
                for keyword in mapping_keywords
            )
        ]

        if not candidate_columns:
            continue

        for col in candidate_columns:

            if pd.api.types.is_numeric_dtype(df[col]):
                continue

            normalized = (
                df[col]
                .astype("string")
                .map(normalize_taxon)
            )

            for target_key, target_name in target_keys.items():

                mask = normalized.eq(target_key)

                if not mask.any():
                    continue

                for row_index, row in df.loc[mask].iterrows():

                    record = {
                        "Target_SLA_Name": target_name,
                        "Evidence_File": str(csv_path),
                        "Evidence_Row_Index": row_index,
                        "Matched_Column": col,
                        "Matched_Value": row[col],
                    }

                    # Retain potentially informative columns
                    for field in df.columns:

                        field_lower = str(field).lower()

                        if any(
                            keyword in field_lower
                            for keyword in [
                                "species",
                                "name",
                                "canonical",
                                "accepted",
                                "harmon",
                                "resolved",
                                "status",
                                "match",
                                "family",
                                "genus",
                                "note",
                            ]
                        ):
                            record[field] = row[field]

                    records.append(record)


pretrait_taxonomic_evidence = pd.DataFrame(records)


# -----------------------------------------------------------------------------
# 5. Remove duplicate evidence rows
# -----------------------------------------------------------------------------

if not pretrait_taxonomic_evidence.empty:

    identifying_columns = [
        c for c in [
            "Target_SLA_Name",
            "Evidence_File",
            "Evidence_Row_Index",
            "Matched_Column",
        ]
        if c in pretrait_taxonomic_evidence.columns
    ]

    pretrait_taxonomic_evidence = (
        pretrait_taxonomic_evidence
        .drop_duplicates(
            subset=identifying_columns
        )
        .reset_index(drop=True)
    )


# -----------------------------------------------------------------------------
# 6. Build evidence summary by target
# -----------------------------------------------------------------------------

summary_rows = []

for target in persistent_unresolved:

    if pretrait_taxonomic_evidence.empty:

        subset = pd.DataFrame()

    else:

        subset = (
            pretrait_taxonomic_evidence.loc[
                pretrait_taxonomic_evidence[
                    "Target_SLA_Name"
                ].eq(target)
            ]
        )

    summary_rows.append(
        {
            "Target_SLA_Name": target,
            "PreTrait_Evidence_Rows": len(subset),
            "Distinct_Evidence_Files": (
                subset["Evidence_File"].nunique()
                if not subset.empty
                else 0
            ),
            "Evidence_Available": (
                "YES"
                if len(subset) > 0
                else "NO"
            ),
        }
    )


pretrait_evidence_summary = pd.DataFrame(
    summary_rows
)


# -----------------------------------------------------------------------------
# 7. Identify rows containing an alternative taxonomic name
# -----------------------------------------------------------------------------

alternative_columns = [
    col
    for col in pretrait_taxonomic_evidence.columns
    if any(
        keyword in str(col).lower()
        for keyword in [
            "canonical",
            "accepted",
            "harmonized",
            "harmonised",
            "resolved",
            "standardized",
            "standardised",
        ]
    )
] if not pretrait_taxonomic_evidence.empty else []


mapping_candidates = []

if not pretrait_taxonomic_evidence.empty:

    for _, row in pretrait_taxonomic_evidence.iterrows():

        target = row["Target_SLA_Name"]
        target_key = normalize_taxon(target)

        for col in alternative_columns:

            value = row.get(col)

            if pd.isna(value):
                continue

            candidate_key = normalize_taxon(value)

            if (
                candidate_key
                and candidate_key != target_key
            ):

                mapping_candidates.append(
                    {
                        "Target_SLA_Name": target,
                        "Evidence_File": row[
                            "Evidence_File"
                        ],
                        "Evidence_Row_Index": row[
                            "Evidence_Row_Index"
                        ],
                        "Mapping_Column": col,
                        "Candidate_Taxon": value,
                    }
                )


pretrait_mapping_candidates = pd.DataFrame(
    mapping_candidates
)

if not pretrait_mapping_candidates.empty:

    pretrait_mapping_candidates = (
        pretrait_mapping_candidates
        .drop_duplicates()
        .sort_values(
            [
                "Target_SLA_Name",
                "Candidate_Taxon",
            ]
        )
        .reset_index(drop=True)
    )


# -----------------------------------------------------------------------------
# 8. Export
# -----------------------------------------------------------------------------

output_dir = (
    Path("outputs")
    / "functional_traits"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

evidence_path = (
    output_dir
    / "omo_sla_pretrait_taxonomic_evidence.csv"
)

summary_path = (
    output_dir
    / "omo_sla_pretrait_taxonomic_evidence_summary.csv"
)

mapping_path = (
    output_dir
    / "omo_sla_pretrait_taxonomic_mapping_candidates.csv"
)


pretrait_taxonomic_evidence.to_csv(
    evidence_path,
    index=False
)

pretrait_evidence_summary.to_csv(
    summary_path,
    index=False
)

pretrait_mapping_candidates.to_csv(
    mapping_path,
    index=False
)


# -----------------------------------------------------------------------------
# 9. Diagnostics
# -----------------------------------------------------------------------------

print("\nPre-functional-trait evidence summary")
print("-" * 100)

display(
    pretrait_evidence_summary
)


print("\nPotential historical taxonomic mappings")
print("-" * 100)

if pretrait_mapping_candidates.empty:

    print(
        "No alternative canonical/accepted/harmonized "
        "names were found in pre-functional-trait CSV outputs."
    )

else:

    display(
        pretrait_mapping_candidates
    )


print("\nDetailed pre-functional-trait evidence")
print("-" * 100)

if pretrait_taxonomic_evidence.empty:

    print(
        "No pre-functional-trait evidence was located "
        "for the six target names."
    )

else:

    display(
        pretrait_taxonomic_evidence
        .sort_values(
            [
                "Target_SLA_Name",
                "Evidence_File",
            ]
        )
    )


print("\nExports")
print("-" * 100)

print(
    f"Detailed evidence : "
    f"{evidence_path}"
)

print(
    f"Evidence summary  : "
    f"{summary_path}"
)

print(
    f"Mapping candidates: "
    f"{mapping_path}"
)

print(
    "\nSection 8.4D completed successfully."
)
```

### 8.4E Final Supplementary SLA Acquisition Pathways

The targeted review of pre-functional-trait project outputs identified no independent historical mapping evidence for the six persistently unresolved species names. Their repeated occurrence in the current functional-trait workflow therefore reflects preservation of the validated Omo canonical taxonomy rather than evidence for an alternative accepted species identity.

Accordingly, no additional species-level taxonomic substitutions are imposed on these records. The six taxa remain unchanged in the canonical Omo Forest dataset, but their SLA acquisition pathway is transferred to controlled genus-level estimation. This distinction allows trait coverage to be developed without representing an uncertain species identity as taxonomically verified.

The supplementary acquisition strategy consequently contains two pathways. Four taxa with independently supported species-level search aliases proceed to supplementary species-level SLA acquisition. The original genus-level `Drypetes` target and the six persistently unresolved species-level taxa proceed to congeneric SLA acquisition using their validated genus identities.

Genus-derived values will be explicitly labelled as estimates and will not be represented as direct species observations. Where multiple suitable congeneric SLA observations are available, a robust genus-level summary will be calculated following the same provenance and quality-control principles used elsewhere in the functional-trait workflow.

This approach preserves the validated Omo Forest taxonomy while separating taxonomic certainty from the level at which functional-trait information is available.

```python
# =============================================================================
# SECTION 8.4E — FINAL SUPPLEMENTARY SLA ACQUISITION PATHWAYS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.4E — FINAL SUPPLEMENTARY SLA ACQUISITION PATHWAYS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm verified supplementary register is available
# -----------------------------------------------------------------------------

if "sla_supplementary_verified" not in globals():

    verified_register_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_supplementary_taxonomic_verification.csv"
    )

    if not verified_register_path.exists():
        raise FileNotFoundError(
            "Supplementary taxonomic verification table not found. "
            "Run Section 8.4B first."
        )

    sla_supplementary_verified = pd.read_csv(
        verified_register_path
    )


# -----------------------------------------------------------------------------
# 2. Persistently unresolved taxa established by Section 8.4D
# -----------------------------------------------------------------------------

persistent_unresolved = {
    "Albizia auxilary",
    "Amphimas tetracoides",
    "Diospyros undebunda",
    "Homalium alymeri",
    "Lonchocarpus cilliata",
    "Musa sapiensis",
}


# -----------------------------------------------------------------------------
# 3. Derive genus conservatively from the canonical/query name
# -----------------------------------------------------------------------------

def extract_genus(name):

    if pd.isna(name):
        return pd.NA

    parts = str(name).strip().split()

    if not parts:
        return pd.NA

    return parts[0]


sla_supplementary_final = (
    sla_supplementary_verified.copy()
)

sla_supplementary_final[
    "Acquisition_Genus"
] = (
    sla_supplementary_final[
        "Canonical_Species"
    ]
    .map(extract_genus)
)


# -----------------------------------------------------------------------------
# 4. Define final acquisition pathway
# -----------------------------------------------------------------------------

sla_supplementary_final[
    "Final_Acquisition_Pathway"
] = pd.NA

sla_supplementary_final[
    "Final_Search_Name"
] = pd.NA

sla_supplementary_final[
    "Planned_Provenance_Class"
] = pd.NA


# Species-level aliases that passed verification
species_search_mask = (
    sla_supplementary_final[
        "Verification_Status"
    ].isin(
        [
            "VERIFIED_ACCEPTED_OR_SYNONYM",
            "SEARCH_ALIAS_WITH_NOMENCLATURAL_NOTE",
        ]
    )
)

sla_supplementary_final.loc[
    species_search_mask,
    "Final_Acquisition_Pathway"
] = "SUPPLEMENTARY_SPECIES_SEARCH"

sla_supplementary_final.loc[
    species_search_mask,
    "Final_Search_Name"
] = (
    sla_supplementary_final.loc[
        species_search_mask,
        "Trait_Search_Name"
    ]
)

sla_supplementary_final.loc[
    species_search_mask,
    "Planned_Provenance_Class"
] = "SPECIES_SUPPLEMENTARY"


# Original genus-level target
original_genus_mask = (
    sla_supplementary_final[
        "Verification_Status"
    ].eq("GENUS_LEVEL_TARGET")
)

sla_supplementary_final.loc[
    original_genus_mask,
    "Final_Acquisition_Pathway"
] = "CONGENERIC_SLA_SEARCH"

sla_supplementary_final.loc[
    original_genus_mask,
    "Final_Search_Name"
] = (
    sla_supplementary_final.loc[
        original_genus_mask,
        "Acquisition_Genus"
    ]
)

sla_supplementary_final.loc[
    original_genus_mask,
    "Planned_Provenance_Class"
] = "GENUS_ESTIMATE"


# Persistently unresolved species-level taxa
persistent_mask = (
    sla_supplementary_final[
        "SLA_Query_Name"
    ].isin(persistent_unresolved)
)

sla_supplementary_final.loc[
    persistent_mask,
    "Final_Acquisition_Pathway"
] = "CONGENERIC_SLA_SEARCH"

sla_supplementary_final.loc[
    persistent_mask,
    "Final_Search_Name"
] = (
    sla_supplementary_final.loc[
        persistent_mask,
        "Acquisition_Genus"
    ]
)

sla_supplementary_final.loc[
    persistent_mask,
    "Planned_Provenance_Class"
] = "GENUS_ESTIMATE"


# -----------------------------------------------------------------------------
# 5. Validate that every deferred record now has an acquisition pathway
# -----------------------------------------------------------------------------

assert len(sla_supplementary_final) == 11

assert (
    sla_supplementary_final[
        "Final_Acquisition_Pathway"
    ]
    .notna()
    .all()
), "One or more deferred taxa lack a final acquisition pathway."

assert (
    sla_supplementary_final[
        "Final_Search_Name"
    ]
    .notna()
    .all()
), "One or more deferred taxa lack a final search name."

assert (
    sla_supplementary_final[
        "Planned_Provenance_Class"
    ]
    .notna()
    .all()
)


# -----------------------------------------------------------------------------
# 6. Build species-level supplementary search manifest
# -----------------------------------------------------------------------------

sla_supp_species_manifest = (
    sla_supplementary_final.loc[
        sla_supplementary_final[
            "Final_Acquisition_Pathway"
        ].eq("SUPPLEMENTARY_SPECIES_SEARCH"),
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Verified_Taxon_Name",
            "Final_Search_Name",
            "Verification_Status",
            "Planned_Provenance_Class",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 7. Build genus-level search manifest
# -----------------------------------------------------------------------------

sla_genus_manifest = (
    sla_supplementary_final.loc[
        sla_supplementary_final[
            "Final_Acquisition_Pathway"
        ].eq("CONGENERIC_SLA_SEARCH"),
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Acquisition_Genus",
            "Final_Search_Name",
            "Planned_Provenance_Class",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 8. Validation counts
# -----------------------------------------------------------------------------

n_species_search = len(
    sla_supp_species_manifest
)

n_genus_search = len(
    sla_genus_manifest
)

n_total = len(
    sla_supplementary_final
)

assert n_species_search == 4, (
    f"Expected 4 supplementary species searches, "
    f"found {n_species_search}."
)

assert n_genus_search == 7, (
    f"Expected 7 genus-level searches, "
    f"found {n_genus_search}."
)

assert (
    n_species_search
    + n_genus_search
    == n_total
)


# -----------------------------------------------------------------------------
# 9. Export final supplementary acquisition manifests
# -----------------------------------------------------------------------------

output_dir = (
    Path("outputs")
    / "functional_traits"
)

final_path = (
    output_dir
    / "omo_sla_supplementary_final_pathways.csv"
)

species_manifest_path = (
    output_dir
    / "omo_sla_supplementary_species_search_manifest.csv"
)

genus_manifest_path = (
    output_dir
    / "omo_sla_genus_search_manifest.csv"
)


sla_supplementary_final.to_csv(
    final_path,
    index=False
)

sla_supp_species_manifest.to_csv(
    species_manifest_path,
    index=False
)

sla_genus_manifest.to_csv(
    genus_manifest_path,
    index=False
)


# -----------------------------------------------------------------------------
# 10. Diagnostics
# -----------------------------------------------------------------------------

print("\nFinal supplementary SLA pathways")
print("-" * 100)

print(
    f"Supplementary species searches : "
    f"{n_species_search}"
)

print(
    f"Congeneric SLA searches        : "
    f"{n_genus_search}"
)

print(
    f"Total deferred targets         : "
    f"{n_total}"
)


print("\nSpecies-level supplementary search manifest")
print("-" * 100)

display(
    sla_supp_species_manifest
)


print("\nGenus-level SLA search manifest")
print("-" * 100)

display(
    sla_genus_manifest
)


print("\nFinal pathway distribution")
print("-" * 100)

print(
    sla_supplementary_final[
        "Final_Acquisition_Pathway"
    ]
    .value_counts()
    .to_string()
)


print("\nExports")
print("-" * 100)

print(
    f"Final pathways   : "
    f"{final_path}"
)

print(
    f"Species manifest : "
    f"{species_manifest_path}"
)

print(
    f"Genus manifest   : "
    f"{genus_manifest_path}"
)

print(
    "\nSection 8.4E completed successfully."
)
```

## 8.5 Supplementary Species-Level SLA Acquisition

Four deferred Omo Forest taxa have defensible external species-level search aliases but were not incorporated into the primary TRY request because their accepted or normalized names were not represented directly in the current TRY accepted-species catalogue used during acquisition.

These taxa are therefore subjected to a supplementary species-level SLA search before any genus-level estimation is considered. The supplementary search preserves the same evidence hierarchy established for the primary SLA workflow: measured species-level observations are preferred, source provenance must be traceable, measurement definitions must correspond to leaf area per dry leaf mass, and original units must be retained before standardization.

The four supplementary search targets are:

- *Aningeria pierrei* for the Omo canonical taxon *Aningeria robusta*;
- *Citrus sinensis* for *Citrus sinenses*;
- *Malacantha alnifolia* for *Melacantha alnifolia*; and
- *Mitragyna ciliata* for *Mitragyna cilliata*.

No SLA value is accepted solely because a numerical value is associated with a similarly named taxon. Each candidate record must retain the searched taxon, reported taxon, source, measurement definition, original value, original unit, and bibliographic or database provenance.

### 8.5A Preparation of the Supplementary Species-Level SLA Search Register

Before external supplementary searches are conducted, a dedicated acquisition register is generated for the four verified species-level aliases. This register separates candidate acquisition from subsequent acceptance, unit harmonization, quality control, and species-level aggregation.

```python
# =============================================================================
# SECTION 8.5A — PREPARE SUPPLEMENTARY SPECIES-LEVEL SLA SEARCH REGISTER
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.5A — SUPPLEMENTARY SPECIES-LEVEL SLA SEARCH REGISTER")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Load species-level supplementary manifest if necessary
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

if "sla_supp_species_manifest" not in globals():

    species_manifest_path = (
        output_dir
        / "omo_sla_supplementary_species_search_manifest.csv"
    )

    if not species_manifest_path.exists():
        raise FileNotFoundError(
            "Supplementary species search manifest not found. "
            "Run Section 8.4E first."
        )

    sla_supp_species_manifest = pd.read_csv(
        species_manifest_path
    )

    print("\nSupplementary species manifest reloaded from:")
    print(species_manifest_path)

else:
    print("\nUsing supplementary species manifest already in notebook memory.")


# -----------------------------------------------------------------------------
# 2. Validate the four species-level targets
# -----------------------------------------------------------------------------

assert len(sla_supp_species_manifest) == 4, (
    f"Expected 4 supplementary species targets, "
    f"found {len(sla_supp_species_manifest)}."
)

assert (
    sla_supp_species_manifest["Final_Search_Name"]
    .nunique()
    == 4
)

expected_search_names = {
    "Aningeria pierrei",
    "Citrus sinensis",
    "Malacantha alnifolia",
    "Mitragyna ciliata",
}

actual_search_names = set(
    sla_supp_species_manifest[
        "Final_Search_Name"
    ].dropna()
)

assert actual_search_names == expected_search_names, (
    "Supplementary search-name set differs from "
    "the validated Section 8.4E targets."
)


# -----------------------------------------------------------------------------
# 3. Create one controlled acquisition row per target
# -----------------------------------------------------------------------------

sla_supp_species_register = (
    sla_supp_species_manifest.copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 4. Establish controlled source-search hierarchy
# -----------------------------------------------------------------------------

sla_supp_species_register[
    "Search_Source_1"
] = "RECOGNIZED_TRAIT_DATABASE"

sla_supp_species_register[
    "Search_Source_2"
] = "PEER_REVIEWED_LITERATURE"

sla_supp_species_register[
    "Search_Source_3"
] = "TRACEABLE_SCIENTIFIC_DATASET"

sla_supp_species_register[
    "Fallback_Strategy"
] = "CONGENERIC_SLA_ESTIMATE"


# -----------------------------------------------------------------------------
# 5. Add acquisition fields
# -----------------------------------------------------------------------------

acquisition_fields = {
    "Candidate_Record_Found": False,

    "Candidate_Source_Type": pd.NA,
    "Candidate_Source_Name": pd.NA,
    "Candidate_Dataset_Name": pd.NA,

    "Candidate_Reported_Taxon": pd.NA,
    "Candidate_Match_Status": "NOT_YET_SEARCHED",

    "Candidate_SLA_Value": np.nan,
    "Candidate_SLA_Unit": pd.NA,

    "Candidate_Measurement_Definition": pd.NA,
    "Candidate_Observation_Count": pd.NA,

    "Candidate_Min_SLA": np.nan,
    "Candidate_Median_SLA": np.nan,
    "Candidate_Mean_SLA": np.nan,
    "Candidate_Max_SLA": np.nan,

    "Candidate_Source_Record_ID": pd.NA,
    "Candidate_DOI_or_URL": pd.NA,
    "Candidate_Citation": pd.NA,

    "Candidate_Access_Date": pd.NA,

    "Candidate_Quality_Flag": "PENDING",

    "Candidate_Accepted": False,

    "Acceptance_Decision": "PENDING_SEARCH",

    "Standardized_SLA_Value": np.nan,
    "Standardized_SLA_Unit": "mm2 mg-1",

    "Final_Provenance_Class": pd.NA,

    "Acquisition_Notes": pd.NA,
}


for column, default in acquisition_fields.items():

    if column not in sla_supp_species_register.columns:
        sla_supp_species_register[column] = default


# -----------------------------------------------------------------------------
# 6. Add search identifiers
# -----------------------------------------------------------------------------

sla_supp_species_register.insert(
    0,
    "Supplementary_Search_ID",
    [
        f"OMO_SLA_SUPP_{i:03d}"
        for i in range(
            1,
            len(sla_supp_species_register) + 1
        )
    ]
)


# -----------------------------------------------------------------------------
# 7. Establish current acquisition state
# -----------------------------------------------------------------------------

sla_supp_species_register[
    "Supplementary_Search_Status"
] = "READY_FOR_EXTERNAL_SEARCH"

sla_supp_species_register[
    "Final_Provenance_Class"
] = "SPECIES_SUPPLEMENTARY"


# -----------------------------------------------------------------------------
# 8. Validation
# -----------------------------------------------------------------------------

assert len(sla_supp_species_register) == 4

assert (
    sla_supp_species_register[
        "Supplementary_Search_ID"
    ].nunique()
    == 4
)

assert (
    sla_supp_species_register[
        "Supplementary_Search_Status"
    ]
    .eq("READY_FOR_EXTERNAL_SEARCH")
    .all()
)

assert (
    sla_supp_species_register[
        "Standardized_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)


# -----------------------------------------------------------------------------
# 9. Export search register
# -----------------------------------------------------------------------------

register_path = (
    output_dir
    / "omo_sla_supplementary_species_acquisition_register.csv"
)

search_names_path = (
    output_dir
    / "omo_sla_supplementary_species_search_names.txt"
)


sla_supp_species_register.to_csv(
    register_path,
    index=False
)

with open(
    search_names_path,
    "w",
    encoding="utf-8"
) as f:

    for name in (
        sla_supp_species_register[
            "Final_Search_Name"
        ]
        .dropna()
        .tolist()
    ):
        f.write(f"{name}\n")


# -----------------------------------------------------------------------------
# 10. Diagnostics
# -----------------------------------------------------------------------------

print("\nSupplementary species-level SLA search register")
print("-" * 100)

print(
    f"Species-level search targets   : "
    f"{len(sla_supp_species_register)}"
)

print(
    f"Unique search names            : "
    f"{sla_supp_species_register['Final_Search_Name'].nunique()}"
)

print(
    f"Search status                  : "
    f"READY_FOR_EXTERNAL_SEARCH"
)


print("\nSearch targets")
print("-" * 100)

display(
    sla_supp_species_register[
        [
            "Supplementary_Search_ID",
            "Canonical_Species",
            "Final_Search_Name",
            "Verification_Status",
            "Search_Source_1",
            "Fallback_Strategy",
            "Supplementary_Search_Status",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Acquisition register : "
    f"{register_path}"
)

print(
    f"Species search names : "
    f"{search_names_path}"
)

print("\nSection 8.5A completed successfully.")
```

### 8.5B First-Pass Supplementary Species-Level SLA Evidence Search

The four supplementary species-level SLA targets were subjected to a targeted search of traceable scientific sources. Candidate records were evaluated for taxonomic correspondence, measurement definition, numerical SLA value, unit, experimental context, and suitability for representing the species in the Omo Forest functional-trait dataset.

The first-pass search produced different levels of evidence among the four taxa. A numerical SLA measurement was identified for *Citrus sinensis*, but the values originate from a controlled citrus interstock and salinity experiment. These measurements are therefore retained as candidate literature evidence rather than being accepted immediately as a generalized species-level SLA value.

For *Aningeria pierrei*, *Malacantha alnifolia*, and *Mitragyna ciliata*, the search confirmed the taxonomic identity and ecological occurrence of the species but did not yield a sufficiently explicit, directly extractable species-level SLA measurement suitable for immediate integration. These taxa therefore remain open for additional supplementary database or literature searching.

This subsection records the search outcome without assigning final SLA values. Candidate measurements remain separate from accepted trait values until measurement comparability and representativeness have been reviewed.

```python
# =============================================================================
# SECTION 8.5B — RECORD FIRST-PASS SUPPLEMENTARY SLA EVIDENCE SEARCH
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.5B — FIRST-PASS SUPPLEMENTARY SLA EVIDENCE SEARCH")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm supplementary species register is available
# -----------------------------------------------------------------------------

if "sla_supp_species_register" not in globals():

    register_path = (
        Path("outputs")
        / "functional_traits"
        / "omo_sla_supplementary_species_acquisition_register.csv"
    )

    if not register_path.exists():
        raise FileNotFoundError(
            "Supplementary species acquisition register not found. "
            "Run Section 8.5A first."
        )

    sla_supp_species_register = pd.read_csv(register_path)


# -----------------------------------------------------------------------------
# 2. Define first-pass evidence outcomes
# -----------------------------------------------------------------------------

first_pass_evidence = pd.DataFrame(
    [
        {
            "Final_Search_Name": "Aningeria pierrei",
            "Candidate_Record_Found": False,
            "Candidate_Source_Type": "LITERATURE_AND_TRAIT_DATABASE_SEARCH",
            "Candidate_Source_Name": "First-pass external search",
            "Candidate_Dataset_Name": pd.NA,
            "Candidate_Reported_Taxon": "Aningeria pierrei",
            "Candidate_Match_Status": "VERIFIED_TAXON_NO_USABLE_SLA_FOUND",
            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,
            "Candidate_Measurement_Definition": pd.NA,
            "Candidate_Observation_Count": pd.NA,
            "Candidate_Source_Record_ID": pd.NA,
            "Candidate_DOI_or_URL": pd.NA,
            "Candidate_Citation": (
                "Taxonomic identity confirmed, but no directly extractable "
                "species-level SLA value located in first-pass search."
            ),
            "Candidate_Quality_Flag": "SEARCH_INCOMPLETE",
            "Candidate_Accepted": False,
            "Acceptance_Decision": "CONTINUE_SUPPLEMENTARY_SEARCH",
            "Acquisition_Notes": (
                "Do not estimate from unrelated Aningeria species yet."
            ),
        },

        {
            "Final_Search_Name": "Citrus sinensis",
            "Candidate_Record_Found": True,
            "Candidate_Source_Type": "PEER_REVIEWED_LITERATURE",
            "Candidate_Source_Name": "Zapata et al. (2004)",
            "Candidate_Dataset_Name": (
                "Interstock-induced mechanism of increased growth "
                "and salt resistance of orange trees"
            ),
            "Candidate_Reported_Taxon": "Citrus sinensis",
            "Candidate_Match_Status": "SPECIES_LEVEL_EXPERIMENTAL_MATCH",

            # Mean of reported control and salt-treatment SLA values:
            # (9.4 + 9.1) / 2 = 9.25 m2 kg-1
            "Candidate_SLA_Value": 9.25,
            "Candidate_SLA_Unit": "m2 kg-1",

            "Candidate_Measurement_Definition": (
                "Leaf area per unit leaf dry mass"
            ),

            "Candidate_Observation_Count": 2,

            "Candidate_Min_SLA": 9.1,
            "Candidate_Median_SLA": 9.25,
            "Candidate_Mean_SLA": 9.25,
            "Candidate_Max_SLA": 9.4,

            "Candidate_Source_Record_ID": "PMID:15294757",
            "Candidate_DOI_or_URL": "PMID:15294757",

            "Candidate_Citation": (
                "Zapata et al. 2004. Interstock-induced mechanism "
                "of increased growth and salt resistance of orange "
                "(Citrus sinensis) trees."
            ),

            "Candidate_Quality_Flag": "CONTEXT_SPECIFIC",

            "Candidate_Accepted": False,

            "Acceptance_Decision": (
                "RETAIN_AS_CANDIDATE_PENDING_REPRESENTATIVENESS_REVIEW"
            ),

            "Acquisition_Notes": (
                "Reported SLA values 9.4 and 9.1 m2 kg-1 derive from "
                "a specific cultivar/rootstock/interstock experiment "
                "under control and salt treatments; not yet accepted "
                "as a general species-level value."
            ),
        },

        {
            "Final_Search_Name": "Malacantha alnifolia",
            "Candidate_Record_Found": False,
            "Candidate_Source_Type": "LITERATURE_AND_TRAIT_DATABASE_SEARCH",
            "Candidate_Source_Name": "First-pass external search",
            "Candidate_Dataset_Name": pd.NA,
            "Candidate_Reported_Taxon": "Malacantha alnifolia",
            "Candidate_Match_Status": "VERIFIED_TAXON_NO_USABLE_SLA_FOUND",
            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,
            "Candidate_Measurement_Definition": pd.NA,
            "Candidate_Observation_Count": pd.NA,
            "Candidate_Source_Record_ID": pd.NA,
            "Candidate_DOI_or_URL": pd.NA,
            "Candidate_Citation": (
                "Taxonomic identity confirmed, but no directly extractable "
                "species-level SLA value located in first-pass search."
            ),
            "Candidate_Quality_Flag": "SEARCH_INCOMPLETE",
            "Candidate_Accepted": False,
            "Acceptance_Decision": "CONTINUE_SUPPLEMENTARY_SEARCH",
            "Acquisition_Notes": (
                "Accepted species confirmed; further trait-specific "
                "search required."
            ),
        },

        {
            "Final_Search_Name": "Mitragyna ciliata",
            "Candidate_Record_Found": False,
            "Candidate_Source_Type": "LITERATURE_AND_TRAIT_DATABASE_SEARCH",
            "Candidate_Source_Name": "First-pass external search",
            "Candidate_Dataset_Name": pd.NA,
            "Candidate_Reported_Taxon": "Mitragyna ciliata",
            "Candidate_Match_Status": "VERIFIED_TAXON_NO_USABLE_SLA_FOUND",
            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,
            "Candidate_Measurement_Definition": pd.NA,
            "Candidate_Observation_Count": pd.NA,
            "Candidate_Source_Record_ID": pd.NA,
            "Candidate_DOI_or_URL": pd.NA,
            "Candidate_Citation": (
                "Species appears in West African forest literature, "
                "but no directly extractable species-level SLA value "
                "was confirmed in the first-pass search."
            ),
            "Candidate_Quality_Flag": "SEARCH_INCOMPLETE",
            "Candidate_Accepted": False,
            "Acceptance_Decision": "CONTINUE_SUPPLEMENTARY_SEARCH",
            "Acquisition_Notes": (
                "A search result associated the species with SLA terminology, "
                "but the underlying accessible study did not provide a "
                "verified Mitragyna ciliata SLA value."
            ),
        },
    ]
)


# -----------------------------------------------------------------------------
# 3. Merge evidence into acquisition register
# -----------------------------------------------------------------------------

fields_to_replace = [
    col
    for col in first_pass_evidence.columns
    if col != "Final_Search_Name"
]

sla_supp_species_evidence = (
    sla_supp_species_register
    .drop(
        columns=[
            col for col in fields_to_replace
            if col in sla_supp_species_register.columns
        ]
    )
    .merge(
        first_pass_evidence,
        how="left",
        on="Final_Search_Name"
    )
)


# -----------------------------------------------------------------------------
# 4. Standardize numerical candidate SLA where applicable
# -----------------------------------------------------------------------------

sla_supp_species_evidence[
    "Standardized_SLA_Value"
] = np.nan

sla_supp_species_evidence[
    "Standardized_SLA_Unit"
] = "mm2 mg-1"


# m2 kg-1 is numerically equivalent to mm2 mg-1
m2kg_mask = (
    sla_supp_species_evidence[
        "Candidate_SLA_Unit"
    ].eq("m2 kg-1")
    & sla_supp_species_evidence[
        "Candidate_SLA_Value"
    ].notna()
)

sla_supp_species_evidence.loc[
    m2kg_mask,
    "Standardized_SLA_Value"
] = (
    sla_supp_species_evidence.loc[
        m2kg_mask,
        "Candidate_SLA_Value"
    ]
)


# -----------------------------------------------------------------------------
# 5. Define first-pass search status
# -----------------------------------------------------------------------------

sla_supp_species_evidence[
    "Supplementary_Search_Status"
] = np.where(
    sla_supp_species_evidence[
        "Candidate_Record_Found"
    ].fillna(False),
    "CANDIDATE_FOUND_REVIEW_REQUIRED",
    "NO_USABLE_VALUE_FIRST_PASS"
)


# -----------------------------------------------------------------------------
# 6. Validation
# -----------------------------------------------------------------------------

n_candidates = (
    sla_supp_species_evidence[
        "Candidate_Record_Found"
    ]
    .fillna(False)
    .sum()
)

n_no_value = (
    sla_supp_species_evidence[
        "Supplementary_Search_Status"
    ]
    .eq("NO_USABLE_VALUE_FIRST_PASS")
    .sum()
)

n_accepted = (
    sla_supp_species_evidence[
        "Candidate_Accepted"
    ]
    .fillna(False)
    .sum()
)


assert len(sla_supp_species_evidence) == 4

assert n_candidates == 1
assert n_no_value == 3
assert n_accepted == 0


# -----------------------------------------------------------------------------
# 7. Export
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

evidence_path = (
    output_dir
    / "omo_sla_supplementary_species_first_pass_evidence.csv"
)

sla_supp_species_evidence.to_csv(
    evidence_path,
    index=False
)


# -----------------------------------------------------------------------------
# 8. Diagnostics
# -----------------------------------------------------------------------------

print("\nFirst-pass supplementary SLA evidence")
print("-" * 100)

print(
    f"Search targets                 : "
    f"{len(sla_supp_species_evidence)}"
)

print(
    f"Numerical candidates located  : "
    f"{n_candidates}"
)

print(
    f"No usable value in first pass : "
    f"{n_no_value}"
)

print(
    f"Candidates accepted           : "
    f"{n_accepted}"
)


print("\nEvidence summary")
print("-" * 100)

display(
    sla_supp_species_evidence[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Candidate_Record_Found",
            "Candidate_SLA_Value",
            "Candidate_SLA_Unit",
            "Standardized_SLA_Value",
            "Candidate_Quality_Flag",
            "Acceptance_Decision",
            "Supplementary_Search_Status",
        ]
    ]
)


print("\nExport")
print("-" * 100)

print(
    f"First-pass evidence : "
    f"{evidence_path}"
)

print("\nSection 8.5B completed successfully.")
```

### 8.5C Closure of the Supplementary Species-Level SLA Search

A second targeted review of the supplementary species-level targets did not identify sufficiently general, directly extractable SLA measurements for *Aningeria pierrei*, *Malacantha alnifolia*, or *Mitragyna ciliata*. Although their taxonomic identities are well established, available accessible sources did not provide numerical species-level SLA measurements suitable for direct incorporation into the Omo Forest functional-trait dataset.

For *Citrus sinensis*, published experimental studies confirm that SLA has been measured, including the candidate values retained during the first-pass review. However, these measurements derive from cultivar-, rootstock-, treatment-, and experimental-context combinations and are therefore not considered sufficiently representative for immediate adoption as a generalized species-level SLA value.

The supplementary species-level search is therefore closed without accepting any of the four candidate taxa as final species-level SLA observations. All four records are retained with their search provenance and are transferred to the controlled genus-level acquisition stage. The existing *Citrus sinensis* literature value remains preserved as contextual candidate evidence but is not used to calculate the final functional-trait value unless later genus-level evidence proves less defensible.

This decision prioritizes comparability and provenance over nominal species-level completeness and prevents context-specific experimental measurements from being treated as general species traits.

```python
# =============================================================================
# SECTION 8.5C — CLOSE SUPPLEMENTARY SPECIES-LEVEL SLA SEARCH
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.5C — CLOSE SUPPLEMENTARY SPECIES-LEVEL SLA SEARCH")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Load first-pass evidence if necessary
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

if "sla_supp_species_evidence" not in globals():

    evidence_path = (
        output_dir
        / "omo_sla_supplementary_species_first_pass_evidence.csv"
    )

    if not evidence_path.exists():
        raise FileNotFoundError(
            "First-pass supplementary SLA evidence not found. "
            "Run Section 8.5B first."
        )

    sla_supp_species_evidence = pd.read_csv(
        evidence_path
    )

    print("\nFirst-pass supplementary evidence reloaded from:")
    print(evidence_path)

else:
    print("\nUsing supplementary SLA evidence already in notebook memory.")


# -----------------------------------------------------------------------------
# 2. Define final species-level search decisions
# -----------------------------------------------------------------------------

sla_supp_species_closed = (
    sla_supp_species_evidence.copy()
)

sla_supp_species_closed[
    "Species_Level_Search_Outcome"
] = "NO_FINAL_SPECIES_VALUE"

sla_supp_species_closed[
    "Final_Species_SLA_Accepted"
] = False

sla_supp_species_closed[
    "Transfer_To_Genus_Strategy"
] = True

sla_supp_species_closed[
    "Closure_Reason"
] = pd.NA


# -----------------------------------------------------------------------------
# 3. Assign explicit closure reasons
# -----------------------------------------------------------------------------

no_value_mask = (
    sla_supp_species_closed[
        "Candidate_Record_Found"
    ]
    .fillna(False)
    .eq(False)
)

sla_supp_species_closed.loc[
    no_value_mask,
    "Closure_Reason"
] = (
    "No directly extractable, sufficiently documented "
    "species-level SLA value identified."
)


candidate_mask = (
    sla_supp_species_closed[
        "Candidate_Record_Found"
    ]
    .fillna(False)
    .eq(True)
)

sla_supp_species_closed.loc[
    candidate_mask,
    "Closure_Reason"
] = (
    "Numerical species-level candidate exists but is "
    "experimental/context-specific and not adopted as a "
    "general species-level SLA value."
)


# -----------------------------------------------------------------------------
# 4. Preserve Citrus candidate explicitly
# -----------------------------------------------------------------------------

citrus_mask = (
    sla_supp_species_closed[
        "Final_Search_Name"
    ].eq("Citrus sinensis")
)

assert citrus_mask.sum() == 1

sla_supp_species_closed.loc[
    citrus_mask,
    "Species_Level_Search_Outcome"
] = "CONTEXT_SPECIFIC_CANDIDATE_RETAINED"

sla_supp_species_closed.loc[
    citrus_mask,
    "Final_Species_SLA_Accepted"
] = False

sla_supp_species_closed.loc[
    citrus_mask,
    "Transfer_To_Genus_Strategy"
] = True


# -----------------------------------------------------------------------------
# 5. Establish genus-search names
# -----------------------------------------------------------------------------

def extract_genus(name):

    if pd.isna(name):
        return pd.NA

    parts = str(name).strip().split()

    if len(parts) == 0:
        return pd.NA

    return parts[0]


sla_supp_species_closed[
    "Genus_Search_Name"
] = (
    sla_supp_species_closed[
        "Final_Search_Name"
    ]
    .map(extract_genus)
)


# -----------------------------------------------------------------------------
# 6. Create genus-transfer manifest for these four taxa
# -----------------------------------------------------------------------------

sla_species_to_genus_manifest = (
    sla_supp_species_closed.loc[
        sla_supp_species_closed[
            "Transfer_To_Genus_Strategy"
        ].fillna(False),
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Final_Search_Name",
            "Genus_Search_Name",
            "Species_Level_Search_Outcome",
            "Candidate_Record_Found",
            "Candidate_SLA_Value",
            "Candidate_SLA_Unit",
            "Candidate_Quality_Flag",
            "Closure_Reason",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

sla_species_to_genus_manifest[
    "Planned_Provenance_Class"
] = "GENUS_ESTIMATE"


# -----------------------------------------------------------------------------
# 7. Validation
# -----------------------------------------------------------------------------

n_targets = len(
    sla_supp_species_closed
)

n_species_accepted = (
    sla_supp_species_closed[
        "Final_Species_SLA_Accepted"
    ]
    .fillna(False)
    .sum()
)

n_transferred = (
    sla_supp_species_closed[
        "Transfer_To_Genus_Strategy"
    ]
    .fillna(False)
    .sum()
)

n_context_candidates = (
    sla_supp_species_closed[
        "Species_Level_Search_Outcome"
    ]
    .eq("CONTEXT_SPECIFIC_CANDIDATE_RETAINED")
    .sum()
)


assert n_targets == 4

assert n_species_accepted == 0

assert n_transferred == 4

assert n_context_candidates == 1

assert (
    sla_supp_species_closed[
        "Genus_Search_Name"
    ]
    .notna()
    .all()
)


# -----------------------------------------------------------------------------
# 8. Export
# -----------------------------------------------------------------------------

closure_path = (
    output_dir
    / "omo_sla_supplementary_species_search_closure.csv"
)

transfer_path = (
    output_dir
    / "omo_sla_species_to_genus_transfer_manifest.csv"
)


sla_supp_species_closed.to_csv(
    closure_path,
    index=False
)

sla_species_to_genus_manifest.to_csv(
    transfer_path,
    index=False
)


# -----------------------------------------------------------------------------
# 9. Diagnostics
# -----------------------------------------------------------------------------

print("\nSupplementary species-level search closure")
print("-" * 100)

print(
    f"Species-level targets reviewed : "
    f"{n_targets}"
)

print(
    f"Final species SLA accepted     : "
    f"{n_species_accepted}"
)

print(
    f"Transferred to genus strategy  : "
    f"{n_transferred}"
)

print(
    f"Context-specific candidates retained : "
    f"{n_context_candidates}"
)


print("\nSpecies-to-genus transfer manifest")
print("-" * 100)

display(
    sla_species_to_genus_manifest[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Genus_Search_Name",
            "Species_Level_Search_Outcome",
            "Candidate_SLA_Value",
            "Candidate_SLA_Unit",
            "Planned_Provenance_Class",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Species search closure : "
    f"{closure_path}"
)

print(
    f"Genus transfer manifest: "
    f"{transfer_path}"
)

print("\nSection 8.5C completed successfully.")
```

## 8.6 Controlled Genus-Level SLA Acquisition

The remaining deferred SLA targets could not be assigned defensible species-level values from either the primary TRY request pathway or supplementary species-level evidence. These taxa therefore proceed to a controlled genus-level acquisition stage.

Genus-level estimates are treated strictly as inferred trait values rather than direct species observations. The canonical Omo Forest taxon is preserved unchanged, while the acquisition workflow searches for measured SLA observations among congeneric species. Where multiple suitable congeneric records are available, the genus estimate will be derived using a robust summary statistic after unit standardization and quality filtering.

Before any genus-level SLA values are collected, the target list is consolidated so that each genus is searched only once. The resulting genus acquisition manifest records which Omo taxa depend on each genus estimate and preserves the provenance class `GENUS_ESTIMATE`.

This stage therefore separates three components:

1. the Omo canonical taxon requiring an estimate;
2. the genus used for congeneric SLA acquisition; and
3. the eventual genus-level summary value derived from measured congeneric records.

No genus-level SLA values are assigned in this subsection. The purpose is to establish the unique acquisition targets and audit structure for the subsequent search and estimation steps.

### 8.6A Consolidation of Genus-Level SLA Targets

```python
# =============================================================================
# SECTION 8.6A — CONSOLIDATE GENUS-LEVEL SLA TARGETS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.6A — CONSOLIDATE GENUS-LEVEL SLA TARGETS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Load original genus manifest
# -----------------------------------------------------------------------------

output_dir = Path("outputs") / "functional_traits"

if "sla_genus_manifest" not in globals():

    genus_manifest_path = (
        output_dir
        / "omo_sla_genus_search_manifest.csv"
    )

    if not genus_manifest_path.exists():
        raise FileNotFoundError(
            "Original genus-level SLA manifest not found. "
            "Run Section 8.4E first."
        )

    sla_genus_manifest = pd.read_csv(
        genus_manifest_path
    )

    print("\nOriginal genus manifest reloaded from:")
    print(genus_manifest_path)

else:
    print("\nUsing original genus manifest already in notebook memory.")


# -----------------------------------------------------------------------------
# 2. Load species-to-genus transfer manifest
# -----------------------------------------------------------------------------

if "sla_species_to_genus_manifest" not in globals():

    transfer_path = (
        output_dir
        / "omo_sla_species_to_genus_transfer_manifest.csv"
    )

    if not transfer_path.exists():
        raise FileNotFoundError(
            "Species-to-genus transfer manifest not found. "
            "Run Section 8.5C first."
        )

    sla_species_to_genus_manifest = pd.read_csv(
        transfer_path
    )

    print("\nSpecies-to-genus transfer manifest reloaded from:")
    print(transfer_path)

else:
    print("\nUsing species-to-genus transfer manifest already in notebook memory.")


# -----------------------------------------------------------------------------
# 3. Standardize the two manifest structures
# -----------------------------------------------------------------------------

original_genus_targets = (
    sla_genus_manifest[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Acquisition_Genus",
            "Final_Search_Name",
            "Planned_Provenance_Class",
        ]
    ]
    .copy()
)

original_genus_targets = original_genus_targets.rename(
    columns={
        "Final_Search_Name": "Genus_Search_Name"
    }
)

original_genus_targets[
    "Genus_Target_Origin"
] = "ORIGINAL_GENUS_PATHWAY"


transferred_genus_targets = (
    sla_species_to_genus_manifest[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Genus_Search_Name",
            "Planned_Provenance_Class",
        ]
    ]
    .copy()
)

transferred_genus_targets[
    "Acquisition_Genus"
] = (
    transferred_genus_targets[
        "Genus_Search_Name"
    ]
)

transferred_genus_targets[
    "Genus_Target_Origin"
] = "TRANSFERRED_FROM_SPECIES_SEARCH"


# -----------------------------------------------------------------------------
# 4. Combine all genus-level target records
# -----------------------------------------------------------------------------

sla_genus_targets_all = pd.concat(
    [
        original_genus_targets,
        transferred_genus_targets,
    ],
    ignore_index=True,
    sort=False
)


# -----------------------------------------------------------------------------
# 5. Reorder fields
# -----------------------------------------------------------------------------

preferred_columns = [
    "SLA_Query_ID",
    "Canonical_Species",
    "SLA_Query_Name",
    "Acquisition_Genus",
    "Genus_Search_Name",
    "Genus_Target_Origin",
    "Planned_Provenance_Class",
]

sla_genus_targets_all = (
    sla_genus_targets_all[
        preferred_columns
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 6. Validate target-level queue
# -----------------------------------------------------------------------------

assert len(sla_genus_targets_all) == 11, (
    f"Expected 11 genus-level target records, "
    f"found {len(sla_genus_targets_all)}."
)

assert (
    sla_genus_targets_all[
        "Planned_Provenance_Class"
    ]
    .eq("GENUS_ESTIMATE")
    .all()
)

assert (
    sla_genus_targets_all[
        "Genus_Search_Name"
    ]
    .notna()
    .all()
)


# -----------------------------------------------------------------------------
# 7. Build unique genus acquisition manifest
# -----------------------------------------------------------------------------

genus_grouped = (
    sla_genus_targets_all
    .groupby(
        "Genus_Search_Name",
        as_index=False
    )
    .agg(
        Target_Record_Count=(
            "SLA_Query_ID",
            "count"
        ),
        Target_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
        Target_Query_IDs=(
            "SLA_Query_ID",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
        Target_Origins=(
            "Genus_Target_Origin",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
)


# -----------------------------------------------------------------------------
# 8. Add genus acquisition fields
# -----------------------------------------------------------------------------

genus_grouped = (
    genus_grouped
    .sort_values("Genus_Search_Name")
    .reset_index(drop=True)
)

genus_grouped.insert(
    0,
    "Genus_Search_ID",
    [
        f"OMO_SLA_GENUS_{i:03d}"
        for i in range(
            1,
            len(genus_grouped) + 1
        )
    ]
)


genus_grouped[
    "Trait_Code"
] = "SLA"

genus_grouped[
    "Target_Unit"
] = "mm2 mg-1"

genus_grouped[
    "Preferred_Source"
] = "TRY_OR_SUPPLEMENTARY_TRAIT_DATABASE"

genus_grouped[
    "Acquisition_Status"
] = "READY_FOR_GENUS_SEARCH"

genus_grouped[
    "Candidate_Congeneric_Record_Count"
] = 0

genus_grouped[
    "Candidate_Congeneric_Species_Count"
] = 0

genus_grouped[
    "Selected_Genus_SLA"
] = np.nan

genus_grouped[
    "Selected_Genus_SLA_Unit"
] = "mm2 mg-1"

genus_grouped[
    "Genus_Estimation_Method"
] = pd.NA

genus_grouped[
    "Quality_Flag"
] = "PENDING"

genus_grouped[
    "Provenance_Class"
] = "GENUS_ESTIMATE"


sla_unique_genus_manifest = genus_grouped.copy()


# -----------------------------------------------------------------------------
# 9. Validation
# -----------------------------------------------------------------------------

n_target_records = len(
    sla_genus_targets_all
)

n_unique_genera = len(
    sla_unique_genus_manifest
)

assert n_target_records == 11

assert (
    sla_unique_genus_manifest[
        "Genus_Search_Name"
    ]
    .nunique()
    == n_unique_genera
)

assert (
    sla_unique_genus_manifest[
        "Acquisition_Status"
    ]
    .eq("READY_FOR_GENUS_SEARCH")
    .all()
)


# -----------------------------------------------------------------------------
# 10. Export
# -----------------------------------------------------------------------------

all_targets_path = (
    output_dir
    / "omo_sla_all_genus_level_targets.csv"
)

unique_genus_path = (
    output_dir
    / "omo_sla_unique_genus_acquisition_manifest.csv"
)

genus_names_path = (
    output_dir
    / "omo_sla_unique_genus_search_names.txt"
)


sla_genus_targets_all.to_csv(
    all_targets_path,
    index=False
)

sla_unique_genus_manifest.to_csv(
    unique_genus_path,
    index=False
)

with open(
    genus_names_path,
    "w",
    encoding="utf-8"
) as f:

    for genus in (
        sla_unique_genus_manifest[
            "Genus_Search_Name"
        ]
        .dropna()
        .tolist()
    ):
        f.write(f"{genus}\n")


# -----------------------------------------------------------------------------
# 11. Diagnostics
# -----------------------------------------------------------------------------

print("\nGenus-level SLA target consolidation")
print("-" * 100)

print(
    f"Genus-level target records      : "
    f"{n_target_records}"
)

print(
    f"Unique genera requiring search  : "
    f"{n_unique_genera}"
)


print("\nAll genus-level target records")
print("-" * 100)

display(
    sla_genus_targets_all
)


print("\nUnique genus acquisition manifest")
print("-" * 100)

display(
    sla_unique_genus_manifest[
        [
            "Genus_Search_ID",
            "Genus_Search_Name",
            "Target_Record_Count",
            "Target_Canonical_Taxa",
            "Acquisition_Status",
            "Provenance_Class",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"All genus targets : "
    f"{all_targets_path}"
)

print(
    f"Unique genus manifest : "
    f"{unique_genus_path}"
)

print(
    f"Genus search names : "
    f"{genus_names_path}"
)

print("\nSection 8.6A completed successfully.")
```

### 8.6C Raw BIEN SLA Import, Coverage Assessment and Structural Inspection

The BIEN genus-level query returned 181 raw specific-leaf-area observations. Before calculating any genus-level estimates, the raw extraction is imported unchanged into the Python workflow and evaluated for structural integrity, taxonomic coverage, units, missing values, source provenance, and numerical characteristics.

The BIEN extraction provided observations for four of the eleven target genera: *Albizia*, *Diospyros*, *Drypetes*, and *Lonchocarpus*. The remaining seven genera returned no SLA observations and are retained explicitly as uncovered acquisition targets rather than being silently removed from the workflow.

Coverage strength differs substantially among the represented genera. Consequently, genus estimates will not be treated as equivalent merely because BIEN contains records. Subsequent quality control will consider both the number of observations and the number of represented congeneric species before genus-level values are accepted.

No observations are removed and no genus estimates are calculated in this subsection. The purpose is to preserve the raw BIEN extraction and establish its empirical coverage and structural characteristics prior to harmonization.

```python
# =============================================================================
# SECTION 8.6C — RAW BIEN SLA IMPORT, COVERAGE AND STRUCTURAL INSPECTION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.6C — RAW BIEN SLA IMPORT, COVERAGE AND STRUCTURAL INSPECTION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Define project-root-aware paths
# -----------------------------------------------------------------------------

from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

bien_raw_path = (
    PROJECT_ROOT
    / "data"
    / "external"
    / "BIEN"
    / "omo_sla_bien_genus_raw.csv"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

print("PROJECT_ROOT :", PROJECT_ROOT)
print("BIEN raw file:", bien_raw_path)
print("File exists  :", bien_raw_path.exists())

# -----------------------------------------------------------------------------
# 2. Import raw BIEN data
# -----------------------------------------------------------------------------

bien_sla_raw = pd.read_csv(
    bien_raw_path,
    low_memory=False
)

print("\nRaw BIEN import")
print("-" * 100)

print(f"Rows imported    : {len(bien_sla_raw):,}")
print(f"Columns imported : {bien_sla_raw.shape[1]:,}")


# -----------------------------------------------------------------------------
# 3. Validate expected schema
# -----------------------------------------------------------------------------

expected_columns = [
    "scrubbed_genus",
    "scrubbed_species_binomial",
    "trait_name",
    "trait_value",
    "unit",
    "method",
    "latitude",
    "longitude",
    "elevation_m",
    "url_source",
    "project_pi",
    "project_pi_contact",
    "access",
    "id",
]

missing_columns = [
    col
    for col in expected_columns
    if col not in bien_sla_raw.columns
]

assert not missing_columns, (
    "Expected BIEN columns are missing: "
    + ", ".join(missing_columns)
)

assert len(bien_sla_raw) == 181, (
    f"Expected 181 BIEN records, found {len(bien_sla_raw)}."
)

assert bien_sla_raw.shape[1] == 14, (
    f"Expected 14 BIEN columns, found {bien_sla_raw.shape[1]}."
)


# -----------------------------------------------------------------------------
# 4. Validate trait identity
# -----------------------------------------------------------------------------

trait_distribution = (
    bien_sla_raw["trait_name"]
    .value_counts(dropna=False)
)

assert (
    bien_sla_raw["trait_name"]
    .eq("leaf area per leaf dry mass")
    .all()
), "Unexpected trait names detected in BIEN extraction."


# -----------------------------------------------------------------------------
# 5. Convert trait value to numeric for diagnostics only
# -----------------------------------------------------------------------------

bien_sla_raw["trait_value_numeric"] = pd.to_numeric(
    bien_sla_raw["trait_value"],
    errors="coerce"
)

n_numeric = (
    bien_sla_raw["trait_value_numeric"]
    .notna()
    .sum()
)

n_non_numeric = (
    bien_sla_raw["trait_value_numeric"]
    .isna()
    .sum()
)


# -----------------------------------------------------------------------------
# 6. Load the 11-genus acquisition manifest if necessary
# -----------------------------------------------------------------------------

if "sla_unique_genus_manifest" not in globals():

    unique_genus_path = (
        output_dir
        / "omo_sla_unique_genus_acquisition_manifest.csv"
    )

    if not unique_genus_path.exists():
        raise FileNotFoundError(
            "Unique genus acquisition manifest not found. "
            "Run Section 8.6A first."
        )

    sla_unique_genus_manifest = pd.read_csv(
        unique_genus_path
    )


target_genera = set(
    sla_unique_genus_manifest[
        "Genus_Search_Name"
    ]
    .dropna()
    .astype(str)
)


# -----------------------------------------------------------------------------
# 7. Check returned genera
# -----------------------------------------------------------------------------

returned_genera = set(
    bien_sla_raw[
        "scrubbed_genus"
    ]
    .dropna()
    .astype(str)
)

unexpected_genera = (
    returned_genera - target_genera
)

assert not unexpected_genera, (
    "Unexpected genera returned by BIEN: "
    + ", ".join(sorted(unexpected_genera))
)


# -----------------------------------------------------------------------------
# 8. Calculate genus-level coverage
# -----------------------------------------------------------------------------

bien_genus_coverage = (
    bien_sla_raw
    .groupby("scrubbed_genus")
    .agg(
        BIEN_Record_Count=(
            "id",
            "count"
        ),
        BIEN_Unique_Species=(
            "scrubbed_species_binomial",
            "nunique"
        ),
        BIEN_Numeric_SLA_Count=(
            "trait_value_numeric",
            "count"
        ),
        BIEN_Unique_Sources=(
            "url_source",
            "nunique"
        ),
    )
    .reset_index()
    .rename(
        columns={
            "scrubbed_genus":
                "Genus_Search_Name"
        }
    )
)


# -----------------------------------------------------------------------------
# 9. Merge against all 11 target genera
# -----------------------------------------------------------------------------

bien_genus_coverage_complete = (
    sla_unique_genus_manifest[
        [
            "Genus_Search_ID",
            "Genus_Search_Name",
            "Target_Canonical_Taxa",
        ]
    ]
    .merge(
        bien_genus_coverage,
        how="left",
        on="Genus_Search_Name"
    )
)


count_columns = [
    "BIEN_Record_Count",
    "BIEN_Unique_Species",
    "BIEN_Numeric_SLA_Count",
    "BIEN_Unique_Sources",
]

bien_genus_coverage_complete[
    count_columns
] = (
    bien_genus_coverage_complete[
        count_columns
    ]
    .fillna(0)
    .astype(int)
)


bien_genus_coverage_complete[
    "BIEN_Coverage_Status"
] = np.where(
    bien_genus_coverage_complete[
        "BIEN_Record_Count"
    ] > 0,
    "BIEN_RECORDS_AVAILABLE",
    "NO_BIEN_SLA_RECORDS"
)


# -----------------------------------------------------------------------------
# 10. Unit distribution
# -----------------------------------------------------------------------------

unit_distribution = (
    bien_sla_raw["unit"]
    .value_counts(dropna=False)
    .rename_axis("BIEN_Unit")
    .reset_index(name="Record_Count")
)


# -----------------------------------------------------------------------------
# 11. Missingness inspection
# -----------------------------------------------------------------------------

missingness = pd.DataFrame(
    {
        "Column": bien_sla_raw.columns,
        "Missing_Count": [
            bien_sla_raw[col].isna().sum()
            for col in bien_sla_raw.columns
        ],
    }
)

missingness["Missing_Percent"] = (
    100
    * missingness["Missing_Count"]
    / len(bien_sla_raw)
)


# -----------------------------------------------------------------------------
# 12. Raw numerical diagnostics — NO filtering yet
# -----------------------------------------------------------------------------

raw_numeric_summary = (
    bien_sla_raw[
        "trait_value_numeric"
    ]
    .describe()
    .to_frame(
        name="Raw_BIEN_SLA"
    )
)


genus_numeric_summary = (
    bien_sla_raw
    .groupby("scrubbed_genus")[
        "trait_value_numeric"
    ]
    .agg(
        [
            "count",
            "min",
            "median",
            "mean",
            "max",
        ]
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 13. Validation of known R results
# -----------------------------------------------------------------------------

assert len(returned_genera) == 4, (
    f"Expected BIEN coverage for 4 genera, "
    f"found {len(returned_genera)}."
)

assert (
    bien_genus_coverage_complete[
        "BIEN_Coverage_Status"
    ]
    .eq("BIEN_RECORDS_AVAILABLE")
    .sum()
    == 4
)

assert (
    bien_genus_coverage_complete[
        "BIEN_Coverage_Status"
    ]
    .eq("NO_BIEN_SLA_RECORDS")
    .sum()
    == 7
)


# -----------------------------------------------------------------------------
# 14. Export inspection tables
# -----------------------------------------------------------------------------

coverage_path = (
    output_dir
    / "omo_sla_bien_genus_coverage.csv"
)

unit_path = (
    output_dir
    / "omo_sla_bien_unit_distribution.csv"
)

missingness_path = (
    output_dir
    / "omo_sla_bien_raw_missingness.csv"
)

numeric_path = (
    output_dir
    / "omo_sla_bien_raw_numeric_summary.csv"
)

genus_numeric_path = (
    output_dir
    / "omo_sla_bien_raw_genus_numeric_summary.csv"
)


bien_genus_coverage_complete.to_csv(
    coverage_path,
    index=False
)

unit_distribution.to_csv(
    unit_path,
    index=False
)

missingness.to_csv(
    missingness_path,
    index=False
)

raw_numeric_summary.to_csv(
    numeric_path
)

genus_numeric_summary.to_csv(
    genus_numeric_path,
    index=False
)


# -----------------------------------------------------------------------------
# 15. Diagnostics
# -----------------------------------------------------------------------------

print("\nTrait validation")
print("-" * 100)
print(trait_distribution.to_string())


print("\nNumeric SLA validation")
print("-" * 100)
print(f"Numeric SLA records     : {n_numeric:,}")
print(f"Non-numeric SLA records : {n_non_numeric:,}")


print("\nBIEN coverage across all target genera")
print("-" * 100)

display(
    bien_genus_coverage_complete
)


print("\nCoverage summary")
print("-" * 100)

print(
    bien_genus_coverage_complete[
        "BIEN_Coverage_Status"
    ]
    .value_counts()
    .to_string()
)


print("\nBIEN unit distribution")
print("-" * 100)

display(unit_distribution)


print("\nRaw SLA numerical summary")
print("-" * 100)

display(raw_numeric_summary)


print("\nRaw SLA summary by genus")
print("-" * 100)

display(genus_numeric_summary)


print("\nMissingness")
print("-" * 100)

display(missingness)


print("\nExports")
print("-" * 100)

print(f"Genus coverage   : {coverage_path}")
print(f"Unit distribution: {unit_path}")
print(f"Missingness      : {missingness_path}")
print(f"Numeric summary  : {numeric_path}")
print(f"Genus summary    : {genus_numeric_path}")

print("\nSection 8.6C completed successfully.")
```

### 8.6D Quality Control and Robust Filtering of BIEN Genus-Level SLA Records

The raw BIEN SLA extraction contains complete numerical values for all 181 records and uses a single unit (`m² kg⁻¹`), which is numerically equivalent to the project standard `mm² mg⁻¹`. However, inspection of the raw distributions identified several extreme observations, including values close to zero and unusually high values relative to the central tendency of their genera.

Because genus-level estimates are intended to represent robust central trait tendencies rather than individual anomalous measurements, the BIEN records are subjected to a conservative quality-control procedure prior to aggregation.

Quality control is performed separately within each genus using the interquartile range (IQR) method. Records falling below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR` are flagged as statistical outliers. These records are not deleted from the raw dataset; instead, they are retained with explicit QC flags and excluded only from the robust genus-summary calculation.

The procedure also retains the number of observations and number of congeneric species contributing to each filtered genus estimate. This allows the final genus-level SLA values to be accompanied by an evidence-strength classification reflecting both sample size and taxonomic breadth.

No genus estimate is transferred to the Omo taxa until the filtered distributions and retained-record counts have been inspected.

```python
# =============================================================================
# SECTION 8.6D — QUALITY CONTROL AND ROBUST FILTERING OF BIEN SLA RECORDS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.6D — QUALITY CONTROL AND ROBUST FILTERING OF BIEN SLA RECORDS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm BIEN raw table is available
# -----------------------------------------------------------------------------

if "bien_sla_raw" not in globals():
    raise RuntimeError(
        "bien_sla_raw is not available in notebook memory. "
        "Run Section 8.6C first."
    )

bien_sla_qc = bien_sla_raw.copy()


# -----------------------------------------------------------------------------
# 2. Standardize BIEN unit numerically
# -----------------------------------------------------------------------------

assert (
    bien_sla_qc["unit"]
    .dropna()
    .eq("m2.kg-1")
    .all()
), "Unexpected BIEN units detected."

bien_sla_qc["SLA_Standardized"] = (
    bien_sla_qc["trait_value_numeric"]
)

bien_sla_qc["SLA_Standardized_Unit"] = "mm2 mg-1"


# -----------------------------------------------------------------------------
# 3. Calculate genus-specific IQR thresholds
# -----------------------------------------------------------------------------

def genus_iqr_bounds(group):
    q1 = group["SLA_Standardized"].quantile(0.25)
    q3 = group["SLA_Standardized"].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    return pd.Series(
        {
            "Q1": q1,
            "Q3": q3,
            "IQR": iqr,
            "Lower_IQR_Bound": lower,
            "Upper_IQR_Bound": upper,
        }
    )


bien_iqr_thresholds = (
    bien_sla_qc
    .groupby("scrubbed_genus")
    .apply(genus_iqr_bounds)
    .reset_index()
)


# -----------------------------------------------------------------------------
# 4. Attach QC thresholds to each record
# -----------------------------------------------------------------------------

bien_sla_qc = (
    bien_sla_qc
    .merge(
        bien_iqr_thresholds,
        how="left",
        on="scrubbed_genus"
    )
)


# -----------------------------------------------------------------------------
# 5. Flag statistical outliers
# -----------------------------------------------------------------------------

bien_sla_qc["IQR_Outlier"] = (
    (bien_sla_qc["SLA_Standardized"] < bien_sla_qc["Lower_IQR_Bound"])
    |
    (bien_sla_qc["SLA_Standardized"] > bien_sla_qc["Upper_IQR_Bound"])
)

bien_sla_qc["QC_Status"] = np.where(
    bien_sla_qc["IQR_Outlier"],
    "EXCLUDE_IQR_OUTLIER",
    "RETAIN"
)


# -----------------------------------------------------------------------------
# 6. Summarize QC by genus
# -----------------------------------------------------------------------------

bien_qc_summary = (
    bien_sla_qc
    .groupby("scrubbed_genus")
    .agg(
        Raw_Record_Count=("id", "count"),
        Raw_Species_Count=("scrubbed_species_binomial", "nunique"),

        Outlier_Record_Count=("IQR_Outlier", "sum"),

        Retained_Record_Count=(
            "QC_Status",
            lambda x: (x == "RETAIN").sum()
        ),

        Raw_Min=("SLA_Standardized", "min"),
        Raw_Median=("SLA_Standardized", "median"),
        Raw_Mean=("SLA_Standardized", "mean"),
        Raw_Max=("SLA_Standardized", "max"),
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 7. Summarize retained values only
# -----------------------------------------------------------------------------

bien_retained = (
    bien_sla_qc.loc[
        bien_sla_qc["QC_Status"].eq("RETAIN")
    ]
    .copy()
)

bien_retained_summary = (
    bien_retained
    .groupby("scrubbed_genus")
    .agg(
        Retained_Record_Count=("id", "count"),
        Retained_Species_Count=("scrubbed_species_binomial", "nunique"),
        Retained_Min=("SLA_Standardized", "min"),
        Retained_Median=("SLA_Standardized", "median"),
        Retained_Mean=("SLA_Standardized", "mean"),
        Retained_Max=("SLA_Standardized", "max"),
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 8. Merge QC and retained summaries
# -----------------------------------------------------------------------------

bien_genus_qc_summary = (
    bien_qc_summary
    .drop(columns=["Retained_Record_Count"])
    .merge(
        bien_retained_summary,
        how="left",
        on="scrubbed_genus"
    )
    .merge(
        bien_iqr_thresholds,
        how="left",
        on="scrubbed_genus"
    )
)


# -----------------------------------------------------------------------------
# 9. Evidence-strength classification
# -----------------------------------------------------------------------------

def evidence_strength(row):

    n_records = row["Retained_Record_Count"]
    n_species = row["Retained_Species_Count"]

    if n_records >= 20 and n_species >= 5:
        return "STRONG"

    if n_records >= 5 and n_species >= 3:
        return "MODERATE"

    if n_records >= 2 and n_species >= 2:
        return "LIMITED"

    return "WEAK"


bien_genus_qc_summary[
    "Evidence_Strength"
] = bien_genus_qc_summary.apply(
    evidence_strength,
    axis=1
)


# -----------------------------------------------------------------------------
# 10. Candidate genus estimate
#     Median chosen for robustness
# -----------------------------------------------------------------------------

bien_genus_qc_summary[
    "Candidate_Genus_SLA"
] = (
    bien_genus_qc_summary[
        "Retained_Median"
    ]
)

bien_genus_qc_summary[
    "Candidate_Genus_SLA_Unit"
] = "mm2 mg-1"

bien_genus_qc_summary[
    "Estimation_Method"
] = "MEDIAN_AFTER_GENUS_SPECIFIC_IQR_FILTER"


# -----------------------------------------------------------------------------
# 11. Validate
# -----------------------------------------------------------------------------

assert (
    bien_genus_qc_summary["scrubbed_genus"].nunique()
    == 4
)

assert (
    bien_retained["IQR_Outlier"]
    .eq(False)
    .all()
)

assert (
    bien_genus_qc_summary[
        "Candidate_Genus_SLA"
    ]
    .notna()
    .all()
)


# -----------------------------------------------------------------------------
# 12. Export
# -----------------------------------------------------------------------------

output_dir = PROJECT_ROOT / "outputs" / "functional_traits"

qc_records_path = (
    output_dir
    / "omo_sla_bien_record_level_qc.csv"
)

qc_summary_path = (
    output_dir
    / "omo_sla_bien_genus_qc_summary.csv"
)

retained_path = (
    output_dir
    / "omo_sla_bien_retained_records.csv"
)


bien_sla_qc.to_csv(
    qc_records_path,
    index=False
)

bien_genus_qc_summary.to_csv(
    qc_summary_path,
    index=False
)

bien_retained.to_csv(
    retained_path,
    index=False
)


# -----------------------------------------------------------------------------
# 13. Diagnostics
# -----------------------------------------------------------------------------

print("\nBIEN genus-level QC summary")
print("-" * 100)

display(
    bien_genus_qc_summary[
        [
            "scrubbed_genus",
            "Raw_Record_Count",
            "Raw_Species_Count",
            "Outlier_Record_Count",
            "Retained_Record_Count",
            "Retained_Species_Count",
            "Raw_Median",
            "Retained_Median",
            "Candidate_Genus_SLA",
            "Evidence_Strength",
        ]
    ]
)


print("\nOutlier records")
print("-" * 100)

outlier_records = (
    bien_sla_qc.loc[
        bien_sla_qc["IQR_Outlier"]
    ]
    [
        [
            "scrubbed_genus",
            "scrubbed_species_binomial",
            "trait_value_numeric",
            "unit",
            "Lower_IQR_Bound",
            "Upper_IQR_Bound",
            "url_source",
            "id",
        ]
    ]
    .sort_values(
        [
            "scrubbed_genus",
            "trait_value_numeric",
        ]
    )
)

display(outlier_records)


print("\nQC status distribution")
print("-" * 100)

print(
    bien_sla_qc["QC_Status"]
    .value_counts()
    .to_string()
)


print("\nEvidence-strength distribution")
print("-" * 100)

print(
    bien_genus_qc_summary[
        "Evidence_Strength"
    ]
    .value_counts()
    .to_string()
)


print("\nExports")
print("-" * 100)

print(f"Record-level QC : {qc_records_path}")
print(f"Genus QC summary: {qc_summary_path}")
print(f"Retained records: {retained_path}")

print("\nSection 8.6D completed successfully.")
```

### 8.6E Acceptance of BIEN Genus-Level SLA Estimates and Identification of Residual Gaps

Following genus-specific quality control, BIEN provided usable congeneric SLA evidence for four of the eleven genus-level acquisition targets: *Albizia*, *Diospyros*, *Drypetes*, and *Lonchocarpus*.

A total of 177 of the 181 retrieved BIEN observations were retained after genus-specific IQR filtering. Four statistical outliers were excluded from genus-level aggregation but remain preserved in the record-level quality-control table. The median of the retained observations is used as the genus-level SLA estimator because it provides a robust measure of central tendency while reducing sensitivity to extreme observations and unequal sampling among records.

The accepted BIEN genus estimates are 15.1200 mm² mg⁻¹ for *Albizia*, 18.5284 mm² mg⁻¹ for *Diospyros*, 15.1485 mm² mg⁻¹ for *Drypetes*, and 20.7700 mm² mg⁻¹ for *Lonchocarpus*. Evidence strength ranges from LIMITED for *Drypetes*, which is represented by only two observations from two species, to STRONG for *Diospyros* and *Lonchocarpus*.

These values are retained explicitly as `GENUS_ESTIMATE` rather than species-level observations. The canonical Omo Forest taxonomic names remain unchanged.

Seven genus-level targets remain without BIEN SLA observations: *Amphimas*, *Aningeria*, *Citrus*, *Homalium*, *Malacantha*, *Mitragyna*, and *Musa*. These residual gaps proceed to the next controlled supplementary acquisition stage rather than receiving arbitrary global estimates.

### 8.7B Targeted Literature Evidence Review for Residual SLA Targets

The seven residual SLA targets were subjected to a targeted supplementary evidence review using species-level and congeneric scientific literature and trait-data sources.

The evidence obtained was heterogeneous. For some taxa, published studies confirmed that specific leaf area had been measured but did not provide a directly extractable numerical species-level value suitable for incorporation into the Omo Forest trait table. For others, evidence was available only at genus, cultivar, community, or experimental-treatment level.

A previously identified *Citrus sinensis* literature record provides numerical SLA values of 9.4 and 9.1 m² kg⁻¹ under contrasting experimental conditions. Their midpoint, 9.25 m² kg⁻¹, is retained as a documented species-level literature candidate. Because the measurement derives from a specific experimental context, it is carried forward for explicit acceptance review rather than being incorporated automatically.

For *Aningeria*, *Amphimas*, *Malacantha*, *Mitragyna*, *Homalium*, and *Musa*, the targeted review did not identify a sufficiently general and directly extractable numerical species-level SLA value suitable for immediate acceptance. These taxa therefore remain candidates for congeneric or broader controlled estimation.

This subsection records the outcome of the targeted search and preserves the distinction between:

- numerical literature evidence suitable for further acceptance review;
- qualitative evidence that SLA has been measured but without an extractable value;
- genus- or cultivar-level evidence that may inform fallback estimation; and
- absence of sufficiently defensible numerical evidence.

No additional SLA value is incorporated into the final trait table at this stage.

```python
# =============================================================================
# SECTION 8.7B — TARGETED LITERATURE EVIDENCE REVIEW
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.7B — TARGETED LITERATURE EVIDENCE REVIEW")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm residual search register is available
# -----------------------------------------------------------------------------

if "sla_residual_search_register" not in globals():
    raise RuntimeError(
        "sla_residual_search_register is not available. "
        "Run Section 8.7A first."
    )


# -----------------------------------------------------------------------------
# 2. Build controlled evidence-review registry
# -----------------------------------------------------------------------------

residual_evidence_review = pd.DataFrame(
    [
        {
            "Genus_Search_Name": "Amphimas",
            "Preferred_Species_Search_Name": "Amphimas tetracoides",

            "Evidence_Level": "CONGENERIC_OR_COMMUNITY_LEVEL",
            "Numeric_Species_SLA_Found": False,

            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,

            "Evidence_Status": (
                "SLA_RELEVANT_EVIDENCE_NO_EXTRACTABLE_SPECIES_VALUE"
            ),

            "Evidence_Interpretation": (
                "SLA-related evidence exists for Amphimas or associated "
                "forest-tree communities, but no directly extractable "
                "species-level value suitable for this target was established."
            ),

            "Recommended_Next_Action": "GENUS_OR_CONTROLLED_FALLBACK",
        },

        {
            "Genus_Search_Name": "Aningeria",
            "Preferred_Species_Search_Name": "Aningeria pierrei",

            "Evidence_Level": "SPECIES_LEVEL_MEASUREMENT_CONFIRMED",
            "Numeric_Species_SLA_Found": False,

            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,

            "Evidence_Status": (
                "SPECIES_SLA_MEASURED_NUMERIC_VALUE_NOT_EXTRACTED"
            ),

            "Evidence_Interpretation": (
                "Published experimental evidence confirms that SLA was "
                "measured for the taxon, but no sufficiently defensible "
                "numerical species-level value was extracted for integration."
            ),

            "Recommended_Next_Action": "GENUS_OR_CONTROLLED_FALLBACK",
        },

        {
            "Genus_Search_Name": "Citrus",
            "Preferred_Species_Search_Name": "Citrus sinensis",

            "Evidence_Level": "SPECIES_LEVEL_EXPERIMENTAL",

            "Numeric_Species_SLA_Found": True,

            "Candidate_SLA_Value": 9.25,
            "Candidate_SLA_Unit": "mm2 mg-1",

            "Evidence_Status": (
                "NUMERICAL_SPECIES_CANDIDATE_CONTEXT_SPECIFIC"
            ),

            "Evidence_Interpretation": (
                "Published experimental SLA values of 9.4 and "
                "9.1 m2 kg-1 yield a midpoint of 9.25 mm2 mg-1. "
                "The value remains context-specific and requires "
                "an explicit acceptance decision."
            ),

            "Recommended_Next_Action": "SPECIES_CANDIDATE_ACCEPTANCE_REVIEW",
        },

        {
            "Genus_Search_Name": "Homalium",
            "Preferred_Species_Search_Name": "Homalium alymeri",

            "Evidence_Level": "INSUFFICIENT",

            "Numeric_Species_SLA_Found": False,

            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,

            "Evidence_Status": "NO_DEFENSIBLE_NUMERICAL_SLA_FOUND",

            "Evidence_Interpretation": (
                "No sufficiently traceable numerical species-level "
                "SLA evidence was established during targeted review."
            ),

            "Recommended_Next_Action": "GENUS_OR_CONTROLLED_FALLBACK",
        },

        {
            "Genus_Search_Name": "Malacantha",
            "Preferred_Species_Search_Name": "Malacantha alnifolia",

            "Evidence_Level": "TAXON_CONFIRMED_NO_NUMERIC_SLA",

            "Numeric_Species_SLA_Found": False,

            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,

            "Evidence_Status": "NO_DEFENSIBLE_NUMERICAL_SLA_FOUND",

            "Evidence_Interpretation": (
                "The taxonomic search name is defensible, but no "
                "directly extractable species-level SLA value suitable "
                "for integration was established."
            ),

            "Recommended_Next_Action": "GENUS_OR_CONTROLLED_FALLBACK",
        },

        {
            "Genus_Search_Name": "Mitragyna",
            "Preferred_Species_Search_Name": "Mitragyna ciliata",

            "Evidence_Level": "TAXON_CONFIRMED_NO_NUMERIC_SLA",

            "Numeric_Species_SLA_Found": False,

            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,

            "Evidence_Status": "NO_DEFENSIBLE_NUMERICAL_SLA_FOUND",

            "Evidence_Interpretation": (
                "The species identity is sufficiently defined for "
                "searching, but no directly extractable numerical SLA "
                "value was established during targeted review."
            ),

            "Recommended_Next_Action": "GENUS_OR_CONTROLLED_FALLBACK",
        },

        {
            "Genus_Search_Name": "Musa",
            "Preferred_Species_Search_Name": "Musa sapiensis",

            "Evidence_Level": "GENUS_OR_CULTIVAR_LEVEL",

            "Numeric_Species_SLA_Found": False,

            "Candidate_SLA_Value": np.nan,
            "Candidate_SLA_Unit": pd.NA,

            "Evidence_Status": (
                "GENUS_SLA_EVIDENCE_NOT_SPECIES_REPRESENTATIVE"
            ),

            "Evidence_Interpretation": (
                "SLA measurements are available for Musa cultivars and "
                "other Musa taxa, but substantial cultivar and treatment "
                "variation prevents assignment of a single species-level "
                "value to the unresolved Omo taxon."
            ),

            "Recommended_Next_Action": "GENUS_OR_CONTROLLED_FALLBACK",
        },
    ]
)


# -----------------------------------------------------------------------------
# 3. Merge evidence review with acquisition register
# -----------------------------------------------------------------------------

sla_residual_evidence = (
    sla_residual_search_register
    .merge(
        residual_evidence_review,
        how="left",
        on=[
            "Genus_Search_Name",
            "Preferred_Species_Search_Name",
        ]
    )
)


# -----------------------------------------------------------------------------
# 4. Preserve prior Citrus candidate and verify consistency
# -----------------------------------------------------------------------------

citrus_mask = (
    sla_residual_evidence[
        "Genus_Search_Name"
    ].eq("Citrus")
)

assert citrus_mask.sum() == 1

existing_citrus = (
    sla_residual_evidence.loc[
        citrus_mask,
        "Existing_Candidate_SLA"
    ]
    .iloc[0]
)

review_citrus = (
    sla_residual_evidence.loc[
        citrus_mask,
        "Candidate_SLA_Value"
    ]
    .iloc[0]
)

assert np.isclose(
    existing_citrus,
    review_citrus,
), (
    "The previously retained Citrus candidate does not match "
    "the targeted evidence-review candidate."
)


# -----------------------------------------------------------------------------
# 5. Classify numerical-evidence availability
# -----------------------------------------------------------------------------

sla_residual_evidence[
    "Numerical_Evidence_Class"
] = np.where(
    sla_residual_evidence[
        "Numeric_Species_SLA_Found"
    ].fillna(False),
    "NUMERICAL_SPECIES_CANDIDATE",
    "NO_ACCEPTABLE_SPECIES_NUMERIC_VALUE"
)


# -----------------------------------------------------------------------------
# 6. No value is accepted in Section 8.7B
# -----------------------------------------------------------------------------

sla_residual_evidence[
    "Final_SLA_Accepted_8_7B"
] = False

sla_residual_evidence[
    "Acceptance_Status_8_7B"
] = "PENDING_FINAL_DECISION"


# -----------------------------------------------------------------------------
# 7. Validation
# -----------------------------------------------------------------------------

n_targets = len(
    sla_residual_evidence
)

n_numeric_candidates = (
    sla_residual_evidence[
        "Numeric_Species_SLA_Found"
    ]
    .fillna(False)
    .sum()
)

n_no_numeric = (
    sla_residual_evidence[
        "Numerical_Evidence_Class"
    ]
    .eq("NO_ACCEPTABLE_SPECIES_NUMERIC_VALUE")
    .sum()
)

n_accepted = (
    sla_residual_evidence[
        "Final_SLA_Accepted_8_7B"
    ]
    .fillna(False)
    .sum()
)


assert n_targets == 7
assert n_numeric_candidates == 1
assert n_no_numeric == 6
assert n_accepted == 0


# -----------------------------------------------------------------------------
# 8. Export evidence-review table
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

evidence_path = (
    output_dir
    / "omo_sla_residual_targeted_evidence_review.csv"
)

sla_residual_evidence.to_csv(
    evidence_path,
    index=False
)


# -----------------------------------------------------------------------------
# 9. Diagnostics
# -----------------------------------------------------------------------------

print("\nResidual SLA evidence-review summary")
print("-" * 100)

print(
    f"Residual targets reviewed       : "
    f"{n_targets}"
)

print(
    f"Numerical species candidates    : "
    f"{n_numeric_candidates}"
)

print(
    f"No acceptable species value     : "
    f"{n_no_numeric}"
)

print(
    f"Values accepted in Section 8.7B : "
    f"{n_accepted}"
)


print("\nEvidence-review table")
print("-" * 100)

display(
    sla_residual_evidence[
        [
            "Canonical_Species",
            "Genus_Search_Name",
            "Preferred_Species_Search_Name",
            "Evidence_Level",
            "Numeric_Species_SLA_Found",
            "Candidate_SLA_Value",
            "Candidate_SLA_Unit",
            "Evidence_Status",
            "Recommended_Next_Action",
            "Acceptance_Status_8_7B",
        ]
    ]
)


print("\nEvidence-status distribution")
print("-" * 100)

print(
    sla_residual_evidence[
        "Evidence_Status"
    ]
    .value_counts()
    .to_string()
)


print("\nExport")
print("-" * 100)

print(
    f"Residual evidence review : "
    f"{evidence_path}"
)

print("\nSection 8.7B completed successfully.")
```

### 8.7C Acceptance of Residual Species-Level SLA Evidence and Final Fallback Routing

The targeted literature review identified one residual taxon for which a defensible numerical species-level SLA estimate can be retained: *Citrus sinenses*, searched under the corrected name *Citrus sinensis*.

Published experimental evidence reported SLA values of 9.4 and 9.1 m² kg⁻¹ under contrasting conditions. Their midpoint, 9.25 m² kg⁻¹, is numerically equivalent to 9.25 mm² mg⁻¹ and agrees exactly with the candidate value retained during the earlier supplementary search. The convergence of these evidence streams supports acceptance of 9.25 mm² mg⁻¹ as a literature-derived species-level estimate.

The value is assigned the provenance class `SPECIES_LITERATURE` and retains an explicit quality flag indicating that it represents an experimental literature estimate rather than a multi-study species mean.

No sufficiently defensible numerical species-level SLA value was obtained for the remaining six targets. These taxa therefore proceed to the final controlled fallback stage:

- *Amphimas tetracoides*
- *Aningeria robusta*
- *Homalium alymeri*
- *Melacantha alnifolia*
- *Mitragyna cilliata*
- *Musa sapiensis*

This preserves the project acquisition hierarchy by exhausting defensible species-level evidence before broader estimates are introduced.

```python
# =============================================================================
# SECTION 8.7C — ACCEPT RESIDUAL SPECIES SLA AND ROUTE FINAL FALLBACKS
# =============================================================================

import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.7C — ACCEPT RESIDUAL SPECIES SLA AND ROUTE FINAL FALLBACKS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm evidence-review table
# -----------------------------------------------------------------------------

if "sla_residual_evidence" not in globals():
    raise RuntimeError(
        "sla_residual_evidence is not available. "
        "Run Section 8.7B first."
    )


# -----------------------------------------------------------------------------
# 2. Initialize final decision fields
# -----------------------------------------------------------------------------

sla_residual_decisions = (
    sla_residual_evidence.copy()
)

sla_residual_decisions[
    "Accepted_SLA_Value"
] = np.nan

sla_residual_decisions[
    "Accepted_SLA_Unit"
] = pd.NA

sla_residual_decisions[
    "Accepted_Provenance_Class"
] = pd.NA

sla_residual_decisions[
    "Accepted_Source_Class"
] = pd.NA

sla_residual_decisions[
    "Accepted_Quality_Flag"
] = pd.NA

sla_residual_decisions[
    "Final_Acquisition_Decision"
] = "FINAL_FALLBACK_REQUIRED"


# -----------------------------------------------------------------------------
# 3. Accept Citrus literature estimate
# -----------------------------------------------------------------------------

citrus_mask = (
    sla_residual_decisions[
        "Genus_Search_Name"
    ].eq("Citrus")
)

assert citrus_mask.sum() == 1

assert np.isclose(
    sla_residual_decisions.loc[
        citrus_mask,
        "Candidate_SLA_Value"
    ].iloc[0],
    9.25
)


sla_residual_decisions.loc[
    citrus_mask,
    "Accepted_SLA_Value"
] = 9.25

sla_residual_decisions.loc[
    citrus_mask,
    "Accepted_SLA_Unit"
] = "mm2 mg-1"

sla_residual_decisions.loc[
    citrus_mask,
    "Accepted_Provenance_Class"
] = "SPECIES_LITERATURE"

sla_residual_decisions.loc[
    citrus_mask,
    "Accepted_Source_Class"
] = "PEER_REVIEWED_LITERATURE"

sla_residual_decisions.loc[
    citrus_mask,
    "Accepted_Quality_Flag"
] = "EXPERIMENTAL_LITERATURE_ESTIMATE"

sla_residual_decisions.loc[
    citrus_mask,
    "Final_Acquisition_Decision"
] = "ACCEPT_SPECIES_LITERATURE"


# -----------------------------------------------------------------------------
# 4. Create accepted residual species table
# -----------------------------------------------------------------------------

sla_residual_species_accepted = (
    sla_residual_decisions.loc[
        sla_residual_decisions[
            "Final_Acquisition_Decision"
        ].eq("ACCEPT_SPECIES_LITERATURE")
    ]
    [
        [
            "Canonical_Species",
            "Preferred_Species_Search_Name",
            "Accepted_SLA_Value",
            "Accepted_SLA_Unit",
            "Accepted_Provenance_Class",
            "Accepted_Source_Class",
            "Accepted_Quality_Flag",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 5. Create final fallback manifest
# -----------------------------------------------------------------------------

sla_final_fallback_targets = (
    sla_residual_decisions.loc[
        sla_residual_decisions[
            "Final_Acquisition_Decision"
        ].eq("FINAL_FALLBACK_REQUIRED")
    ]
    [
        [
            "Genus_Search_ID",
            "Canonical_Species",
            "Genus_Search_Name",
            "Preferred_Species_Search_Name",
            "Alternative_Search_Name",
            "Taxonomic_Search_Status",
            "Evidence_Status",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


sla_final_fallback_targets[
    "Fallback_Priority_1"
] = "CONGENERIC_LITERATURE_ESTIMATE"

sla_final_fallback_targets[
    "Fallback_Priority_2"
] = "FAMILY_ESTIMATE"

sla_final_fallback_targets[
    "Fallback_Priority_3"
] = "CONTROLLED_GLOBAL_ESTIMATE"

sla_final_fallback_targets[
    "Fallback_Status"
] = "READY_FOR_FINAL_FALLBACK"


# -----------------------------------------------------------------------------
# 6. Validation
# -----------------------------------------------------------------------------

n_species_accepted = len(
    sla_residual_species_accepted
)

n_final_fallback = len(
    sla_final_fallback_targets
)

assert n_species_accepted == 1
assert n_final_fallback == 6

assert (
    sla_residual_species_accepted[
        "Canonical_Species"
    ].iloc[0]
    == "Citrus sinenses"
)

assert np.isclose(
    sla_residual_species_accepted[
        "Accepted_SLA_Value"
    ].iloc[0],
    9.25
)

expected_fallback_taxa = {
    "Amphimas tetracoides",
    "Aningeria robusta",
    "Homalium alymeri",
    "Melacantha alnifolia",
    "Mitragyna cilliata",
    "Musa sapiensis",
}

assert set(
    sla_final_fallback_targets[
        "Canonical_Species"
    ]
) == expected_fallback_taxa


# -----------------------------------------------------------------------------
# 7. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

decision_path = (
    output_dir
    / "omo_sla_residual_acquisition_decisions.csv"
)

accepted_path = (
    output_dir
    / "omo_sla_residual_species_literature_accepted.csv"
)

fallback_path = (
    output_dir
    / "omo_sla_final_fallback_targets.csv"
)


sla_residual_decisions.to_csv(
    decision_path,
    index=False
)

sla_residual_species_accepted.to_csv(
    accepted_path,
    index=False
)

sla_final_fallback_targets.to_csv(
    fallback_path,
    index=False
)


# -----------------------------------------------------------------------------
# 8. Diagnostics
# -----------------------------------------------------------------------------

print("\nResidual acquisition decisions")
print("-" * 100)

display(
    sla_residual_decisions[
        [
            "Canonical_Species",
            "Preferred_Species_Search_Name",
            "Candidate_SLA_Value",
            "Accepted_SLA_Value",
            "Accepted_Provenance_Class",
            "Accepted_Quality_Flag",
            "Final_Acquisition_Decision",
        ]
    ]
)


print("\nAccepted species-level literature SLA")
print("-" * 100)

display(
    sla_residual_species_accepted
)


print("\nFinal fallback targets")
print("-" * 100)

display(
    sla_final_fallback_targets[
        [
            "Canonical_Species",
            "Genus_Search_Name",
            "Preferred_Species_Search_Name",
            "Fallback_Priority_1",
            "Fallback_Priority_2",
            "Fallback_Priority_3",
            "Fallback_Status",
        ]
    ]
)


print("\nDecision summary")
print("-" * 100)

print(
    f"Species literature values accepted : "
    f"{n_species_accepted}"
)

print(
    f"Final fallback targets             : "
    f"{n_final_fallback}"
)


print("\nExports")
print("-" * 100)

print(f"Decision register : {decision_path}")
print(f"Accepted species  : {accepted_path}")
print(f"Fallback targets  : {fallback_path}")

print("\nSection 8.7C completed successfully.")
```

## 8.8 Primary TRY SLA Data Integration

The completed TRY data request (Request 51853) is now available and is integrated before proceeding to lower-priority taxonomic fallback estimates.

The TRY release represents the highest-priority external SLA evidence in the acquisition hierarchy because the request was constructed specifically for the taxonomically resolved Omo Forest species and the selected specific-leaf-area trait definitions.

Accordingly, any valid species-level TRY SLA measurements identified during this stage supersede genus-, family-, or global-level estimates developed during supplementary acquisition.

The integration workflow preserves the raw TRY release unchanged and proceeds through explicit schema inspection, trait validation, taxonomic crosswalking, unit harmonization, measurement-level quality control, species-level aggregation, and provenance tracking.

### 8.8A TRY Download Inventory and Raw Schema Inspection

TRY Request 51853 was submitted for the public records associated with SLA Trait IDs 3115, 3116, and 3117 and 82 unique resolved TRY species identifiers. The completed release contains the primary data file `51853.txt` together with the TRY data-release documentation and intellectual-property guidelines.

This subsection inventories the release and inspects the raw data structure without modifying, filtering, aggregating, or otherwise transforming the TRY measurements.

```python
# =============================================================================
# SECTION 8.8A — TRY DOWNLOAD INVENTORY AND RAW SCHEMA INSPECTION
# =============================================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("SECTION 8.8A — TRY DOWNLOAD INVENTORY AND RAW SCHEMA INSPECTION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Define project and TRY directories
# -----------------------------------------------------------------------------

if "PROJECT_ROOT" not in globals():
    PROJECT_ROOT = Path(
        r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
    )

try_dir = (
    PROJECT_ROOT
    / "notebooks"
    / "data"
    / "external"
    / "TRY"
)

try_data_path = (
    try_dir
    / "51853.txt"
)

TRY_ENCODING = "latin-1"


print(f"\nPROJECT_ROOT : {PROJECT_ROOT}")
print(f"TRY directory: {try_dir}")
print(f"TRY data file: {try_data_path}")
print(f"File exists  : {try_data_path.exists()}")


if not try_data_path.exists():
    raise FileNotFoundError(
        f"\nTRY release file not found:\n{try_data_path}\n\n"
        "Confirm that 51853.txt is present in the expected TRY directory."
    )


# -----------------------------------------------------------------------------
# 2. Inventory TRY release directory
# -----------------------------------------------------------------------------

release_inventory = []

for path in sorted(try_dir.iterdir()):

    if path.is_file():

        release_inventory.append(
            {
                "File_Name": path.name,
                "File_Extension": path.suffix.lower(),
                "File_Size_Bytes": path.stat().st_size,
                "File_Size_MB": round(
                    path.stat().st_size / (1024 ** 2),
                    3
                ),
            }
        )


try_release_inventory = pd.DataFrame(
    release_inventory
)


print("\nTRY release directory inventory")
print("-" * 100)

display(
    try_release_inventory
)


# -----------------------------------------------------------------------------
# 3. Inspect first physical lines of the raw TRY file
# -----------------------------------------------------------------------------

print("\nRaw file preview — first 5 physical lines")
print("-" * 100)


with open(
    try_data_path,
    "r",
    encoding=TRY_ENCODING,
    errors="strict"
) as f:

    raw_preview_lines = [
        f.readline().rstrip("\n\r")
        for _ in range(5)
    ]


for i, line in enumerate(
    raw_preview_lines,
    start=1
):

    print(f"\nLINE {i}")
    print(line[:1500])


# -----------------------------------------------------------------------------
# 4. Detect delimiter from header line
# -----------------------------------------------------------------------------

header_line = raw_preview_lines[0]

delimiter_candidates = {
    "TAB": "\t",
    "SEMICOLON": ";",
    "COMMA": ",",
}

delimiter_counts = {
    name: header_line.count(delimiter)
    for name, delimiter in delimiter_candidates.items()
}


detected_delimiter_name = max(
    delimiter_counts,
    key=delimiter_counts.get
)

detected_delimiter = (
    delimiter_candidates[
        detected_delimiter_name
    ]
)


print("\nDelimiter inspection")
print("-" * 100)

for name, count in delimiter_counts.items():
    print(f"{name:<12}: {count}")


print(
    f"\nDetected delimiter : "
    f"{detected_delimiter_name}"
)


if delimiter_counts[
    detected_delimiter_name
] == 0:

    raise RuntimeError(
        "Could not determine the delimiter of 51853.txt "
        "from the header line."
    )


# -----------------------------------------------------------------------------
# 5. Import raw TRY release
#    No biological filtering or transformation is performed here
# -----------------------------------------------------------------------------

try_sla_raw = pd.read_csv(
    try_data_path,
    sep=detected_delimiter,
    encoding=TRY_ENCODING,
    low_memory=False
)


print("\nRaw TRY import")
print("-" * 100)

print(
    f"Rows imported    : "
    f"{len(try_sla_raw):,}"
)

print(
    f"Columns imported : "
    f"{len(try_sla_raw.columns):,}"
)


# -----------------------------------------------------------------------------
# 6. Display exact raw schema
# -----------------------------------------------------------------------------

try_schema = pd.DataFrame(
    {
        "Column_Number": range(
            1,
            len(try_sla_raw.columns) + 1
        ),

        "Column_Name": (
            try_sla_raw.columns
            .astype(str)
            .tolist()
        ),

        "Pandas_Dtype": [
            str(dtype)
            for dtype in try_sla_raw.dtypes
        ],

        "Missing_Count": [
            try_sla_raw[col].isna().sum()
            for col in try_sla_raw.columns
        ],

        "Nonmissing_Count": [
            try_sla_raw[col].notna().sum()
            for col in try_sla_raw.columns
        ],
    }
)


print("\nTRY raw column schema")
print("-" * 100)

display(
    try_schema
)


# -----------------------------------------------------------------------------
# 7. Display first 10 records exactly as imported
# -----------------------------------------------------------------------------

print("\nFirst 10 TRY records")
print("-" * 100)

display(
    try_sla_raw.head(10)
)


# -----------------------------------------------------------------------------
# 8. Identify likely taxonomic, trait, value, unit and provenance columns
# -----------------------------------------------------------------------------

column_keywords = [
    "trait",
    "species",
    "accspecies",
    "taxon",
    "name",
    "value",
    "unit",
    "dataset",
    "dataid",
    "observation",
    "source",
    "reference",
]

candidate_columns = []

for col in try_sla_raw.columns:

    normalized_col = str(col).lower()

    matched_keywords = [
        keyword
        for keyword in column_keywords
        if keyword in normalized_col
    ]

    if matched_keywords:

        candidate_columns.append(
            {
                "Column_Name": col,
                "Matched_Keywords": ", ".join(
                    matched_keywords
                ),
            }
        )


try_candidate_columns = pd.DataFrame(
    candidate_columns
)


print("\nPotential trait/taxonomic/provenance columns")
print("-" * 100)

if not try_candidate_columns.empty:

    display(
        try_candidate_columns
    )

else:

    print(
        "No candidate columns were detected automatically. "
        "The raw schema above should be reviewed manually."
    )


# -----------------------------------------------------------------------------
# 9. Basic structural diagnostics only
# -----------------------------------------------------------------------------

print("\nBasic structural diagnostics")
print("-" * 100)

print(
    f"Duplicate complete rows : "
    f"{try_sla_raw.duplicated().sum():,}"
)

print(
    f"Completely empty rows   : "
    f"{try_sla_raw.isna().all(axis=1).sum():,}"
)


# -----------------------------------------------------------------------------
# 10. Export release inventory and raw schema
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


inventory_path = (
    output_dir
    / "omo_sla_try_request_51853_release_inventory.csv"
)

schema_path = (
    output_dir
    / "omo_sla_try_request_51853_raw_schema.csv"
)

candidate_columns_path = (
    output_dir
    / "omo_sla_try_request_51853_candidate_columns.csv"
)


try_release_inventory.to_csv(
    inventory_path,
    index=False
)

try_schema.to_csv(
    schema_path,
    index=False
)

try_candidate_columns.to_csv(
    candidate_columns_path,
    index=False
)


# -----------------------------------------------------------------------------
# 11. Final diagnostics
# -----------------------------------------------------------------------------

print("\nRelease validation")
print("-" * 100)

print(
    f"Primary TRY data file : "
    f"{try_data_path.name}"
)

print(
    f"Encoding              : "
    f"{TRY_ENCODING}"
)

print(
    f"Delimiter             : "
    f"{detected_delimiter_name}"
)

print(
    f"Raw rows              : "
    f"{len(try_sla_raw):,}"
)

print(
    f"Raw columns           : "
    f"{len(try_sla_raw.columns):,}"
)


print("\nExports")
print("-" * 100)

print(
    f"Release inventory : "
    f"{inventory_path}"
)

print(
    f"Raw schema        : "
    f"{schema_path}"
)

print(
    f"Candidate columns : "
    f"{candidate_columns_path}"
)


print(
    "\nNo TRY measurements were filtered, "
    "aggregated, standardized or otherwise modified "
    "in Section 8.8A."
)

print("\nSection 8.8A completed successfully.")
```

### 8.8B Extraction and Validation of Requested TRY SLA Measurements

The raw TRY release contains both requested SLA measurements and ancillary observation-level information such as treatment conditions, sampling dates, growth media, and other contextual variables. Only rows corresponding to the three requested SLA Trait IDs are retained for trait-value analysis.

The requested SLA definitions are:

- Trait ID 3115 — leaf area per leaf dry mass, petiole excluded;
- Trait ID 3116 — leaf area per leaf dry mass, petiole included; and
- Trait ID 3117 — leaf area per leaf dry mass, petiole inclusion or exclusion unspecified.

The TRY release already provides standardized SLA values in the project target unit of **mm² mg⁻¹**. Therefore, `StdValue` and `UnitName` are used as the primary standardized measurement fields, while original values and units remain preserved for provenance.

This subsection validates the expected 535 public SLA measurements, assesses their distribution across trait definitions, species, datasets, and units, and establishes the measurement-level table that will be used for subsequent quality control.

No species-level aggregation is performed at this stage.

```python
# =============================================================================
# SECTION 8.8B — EXTRACT AND VALIDATE REQUESTED TRY SLA MEASUREMENTS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8B — EXTRACT AND VALIDATE REQUESTED TRY SLA MEASUREMENTS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm raw TRY table is available
# -----------------------------------------------------------------------------

if "try_sla_raw" not in globals():
    raise RuntimeError(
        "try_sla_raw is not available in notebook memory. "
        "Run Section 8.8A first."
    )


# -----------------------------------------------------------------------------
# 2. Define requested TRY SLA Trait IDs
# -----------------------------------------------------------------------------

REQUESTED_TRY_SLA_TRAITS = {
    3115: "SLA_PETIOLE_EXCLUDED",
    3116: "SLA_PETIOLE_INCLUDED",
    3117: "SLA_PETIOLE_UNSPECIFIED",
}

requested_trait_ids = set(
    REQUESTED_TRY_SLA_TRAITS.keys()
)


# -----------------------------------------------------------------------------
# 3. Extract only requested SLA measurement rows
# -----------------------------------------------------------------------------

try_sla_measurements = (
    try_sla_raw.loc[
        try_sla_raw["TraitID"]
        .isin(requested_trait_ids)
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 4. Add controlled SLA definition labels
# -----------------------------------------------------------------------------

try_sla_measurements[
    "SLA_Definition_Class"
] = (
    try_sla_measurements["TraitID"]
    .astype(int)
    .map(REQUESTED_TRY_SLA_TRAITS)
)


# -----------------------------------------------------------------------------
# 5. Validate expected measurement count
# -----------------------------------------------------------------------------

n_measurements = len(
    try_sla_measurements
)

assert n_measurements == 535, (
    f"Expected 535 public TRY SLA measurements, "
    f"found {n_measurements}."
)


# -----------------------------------------------------------------------------
# 6. Validate Trait IDs
# -----------------------------------------------------------------------------

returned_trait_ids = set(
    try_sla_measurements[
        "TraitID"
    ]
    .dropna()
    .astype(int)
    .unique()
)

assert returned_trait_ids.issubset(
    requested_trait_ids
), (
    "Unexpected TRY Trait IDs detected: "
    f"{sorted(returned_trait_ids - requested_trait_ids)}"
)


# -----------------------------------------------------------------------------
# 7. Validate standardized values
# -----------------------------------------------------------------------------

try_sla_measurements[
    "StdValue_numeric"
] = pd.to_numeric(
    try_sla_measurements["StdValue"],
    errors="coerce"
)

n_numeric_stdvalue = (
    try_sla_measurements[
        "StdValue_numeric"
    ]
    .notna()
    .sum()
)

n_missing_stdvalue = (
    try_sla_measurements[
        "StdValue_numeric"
    ]
    .isna()
    .sum()
)


# -----------------------------------------------------------------------------
# 8. Validate standardized units
# -----------------------------------------------------------------------------

unit_distribution = (
    try_sla_measurements[
        "UnitName"
    ]
    .value_counts(dropna=False)
    .rename_axis("UnitName")
    .reset_index(name="Record_Count")
)


# -----------------------------------------------------------------------------
# 9. Trait-definition distribution
# -----------------------------------------------------------------------------

trait_definition_summary = (
    try_sla_measurements
    .groupby(
        [
            "TraitID",
            "TraitName",
            "SLA_Definition_Class",
        ],
        dropna=False
    )
    .agg(
        Record_Count=(
            "ObsDataID",
            "count"
        ),
        Unique_TRY_Species=(
            "AccSpeciesID",
            "nunique"
        ),
        Unique_Datasets=(
            "DatasetID",
            "nunique"
        ),
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 10. Species-level measurement coverage
# -----------------------------------------------------------------------------

try_species_coverage = (
    try_sla_measurements
    .groupby(
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
    .agg(
        TRY_SLA_Record_Count=(
            "ObsDataID",
            "count"
        ),
        TRY_Unique_Trait_Definitions=(
            "TraitID",
            "nunique"
        ),
        TRY_Unique_Datasets=(
            "DatasetID",
            "nunique"
        ),
        TRY_Unique_Observations=(
            "ObservationID",
            "nunique"
        ),
        TRY_Min_SLA=(
            "StdValue_numeric",
            "min"
        ),
        TRY_Median_SLA=(
            "StdValue_numeric",
            "median"
        ),
        TRY_Mean_SLA=(
            "StdValue_numeric",
            "mean"
        ),
        TRY_Max_SLA=(
            "StdValue_numeric",
            "max"
        ),
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 11. Dataset-level provenance summary
# -----------------------------------------------------------------------------

try_dataset_summary = (
    try_sla_measurements
    .groupby(
        [
            "DatasetID",
            "Dataset",
        ]
    )
    .agg(
        Record_Count=(
            "ObsDataID",
            "count"
        ),
        Unique_TRY_Species=(
            "AccSpeciesID",
            "nunique"
        ),
        Unique_Observations=(
            "ObservationID",
            "nunique"
        ),
    )
    .reset_index()
    .sort_values(
        "Record_Count",
        ascending=False
    )
)


# -----------------------------------------------------------------------------
# 12. Missingness summary for core fields
# -----------------------------------------------------------------------------

core_fields = [
    "AccSpeciesID",
    "AccSpeciesName",
    "ObservationID",
    "ObsDataID",
    "TraitID",
    "TraitName",
    "DatasetID",
    "Dataset",
    "StdValue",
    "UnitName",
    "OrigValueStr",
    "OrigUnitStr",
    "ValueKindName",
    "ErrorRisk",
    "Reference",
]

core_missingness = pd.DataFrame(
    {
        "Field": core_fields,
        "Missing_Count": [
            try_sla_measurements[
                field
            ].isna().sum()
            for field in core_fields
        ],
    }
)

core_missingness[
    "Missing_Percent"
] = (
    100
    * core_missingness[
        "Missing_Count"
    ]
    / len(try_sla_measurements)
)


# -----------------------------------------------------------------------------
# 13. Check complete-row duplication within extracted SLA rows
# -----------------------------------------------------------------------------

n_duplicate_rows = (
    try_sla_measurements
    .duplicated()
    .sum()
)


# -----------------------------------------------------------------------------
# 14. Build measurement-level working table
# -----------------------------------------------------------------------------

working_columns = [
    "LastName",
    "FirstName",
    "DatasetID",
    "Dataset",
    "SpeciesName",
    "AccSpeciesID",
    "AccSpeciesName",
    "ObservationID",
    "ObsDataID",
    "TraitID",
    "TraitName",
    "SLA_Definition_Class",
    "DataID",
    "DataName",
    "OriglName",
    "OrigValueStr",
    "OrigUnitStr",
    "ValueKindName",
    "OrigUncertaintyStr",
    "UncertaintyName",
    "Replicates",
    "StdValue",
    "StdValue_numeric",
    "UnitName",
    "ErrorRisk",
    "Reference",
    "Comment",
]

try_sla_measurement_working = (
    try_sla_measurements[
        working_columns
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 15. Validation
# -----------------------------------------------------------------------------

assert (
    try_sla_measurement_working[
        "AccSpeciesID"
    ]
    .notna()
    .all()
)

assert (
    try_sla_measurement_working[
        "AccSpeciesName"
    ]
    .notna()
    .all()
)

assert (
    try_sla_measurement_working[
        "TraitID"
    ]
    .notna()
    .all()
)

assert (
    try_sla_measurement_working[
        "StdValue_numeric"
    ]
    .notna()
    .all()
), (
    "One or more requested TRY SLA rows lack "
    "a numeric standardized SLA value."
)

assert (
    try_sla_measurement_working[
        "UnitName"
    ]
    .eq("mm2 mg-1")
    .all()
), (
    "Unexpected standardized TRY SLA unit detected."
)


# -----------------------------------------------------------------------------
# 16. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

measurement_path = (
    output_dir
    / "omo_sla_try_request_51853_measurements_raw.csv"
)

trait_summary_path = (
    output_dir
    / "omo_sla_try_trait_definition_summary.csv"
)

species_coverage_path = (
    output_dir
    / "omo_sla_try_species_measurement_coverage.csv"
)

dataset_summary_path = (
    output_dir
    / "omo_sla_try_dataset_summary.csv"
)

missingness_path = (
    output_dir
    / "omo_sla_try_measurement_missingness.csv"
)


try_sla_measurement_working.to_csv(
    measurement_path,
    index=False
)

trait_definition_summary.to_csv(
    trait_summary_path,
    index=False
)

try_species_coverage.to_csv(
    species_coverage_path,
    index=False
)

try_dataset_summary.to_csv(
    dataset_summary_path,
    index=False
)

core_missingness.to_csv(
    missingness_path,
    index=False
)


# -----------------------------------------------------------------------------
# 17. Diagnostics
# -----------------------------------------------------------------------------

print("\nTRY SLA measurement extraction")
print("-" * 100)

print(
    f"Requested SLA measurements : "
    f"{n_measurements}"
)

print(
    f"Numeric StdValue records    : "
    f"{n_numeric_stdvalue}"
)

print(
    f"Missing StdValue records    : "
    f"{n_missing_stdvalue}"
)

print(
    f"Duplicate complete rows     : "
    f"{n_duplicate_rows}"
)


print("\nTrait-definition summary")
print("-" * 100)

display(
    trait_definition_summary
)


print("\nStandardized unit distribution")
print("-" * 100)

display(
    unit_distribution
)


print("\nTRY species-level coverage")
print("-" * 100)

print(
    f"TRY species with SLA records : "
    f"{try_species_coverage['AccSpeciesID'].nunique()}"
)

display(
    try_species_coverage
    .sort_values(
        [
            "TRY_SLA_Record_Count",
            "AccSpeciesName",
        ],
        ascending=[
            False,
            True,
        ]
    )
)


print("\nDataset-level provenance summary")
print("-" * 100)

display(
    try_dataset_summary
)


print("\nCore-field missingness")
print("-" * 100)

display(
    core_missingness
)


print("\nExports")
print("-" * 100)

print(
    f"Measurement-level TRY SLA : "
    f"{measurement_path}"
)

print(
    f"Trait summary             : "
    f"{trait_summary_path}"
)

print(
    f"Species coverage          : "
    f"{species_coverage_path}"
)

print(
    f"Dataset summary           : "
    f"{dataset_summary_path}"
)

print(
    f"Missingness               : "
    f"{missingness_path}"
)

print(
    "\nSection 8.8B completed successfully."
)
```

### 8.8C Crosswalking TRY SLA Measurements to Omo Forest Canonical Taxa

The 535 public TRY SLA measurements represent 36 accepted TRY species. These external taxonomic identities must now be reconciled back to the validated Omo Forest canonical taxa before species-level SLA aggregation can be performed.

The reconciliation uses the taxonomic crosswalk established during Sections 8.3B–8.3E. TRY accepted-species identifiers are treated solely as external trait-retrieval keys; they do not replace or modify the canonical Omo Forest taxonomy.

Because several Omo canonical names or historical spelling variants may resolve to the same TRY accepted species, a single TRY species can provide SLA evidence for more than one Omo acquisition record. Such relationships are retained explicitly rather than collapsed silently.

This subsection therefore determines:

- which Omo canonical taxa have direct species-level TRY SLA measurements;
- which resolved Omo taxa were included in the TRY request but returned no public SLA records;
- how many Omo records share the same accepted TRY taxon; and
- which taxa remain dependent on supplementary, genus-level, or broader fallback estimates.

No TRY SLA values are aggregated or accepted as final species values in this subsection.

```python
# =============================================================================
# SECTION 8.8C — CROSSWALK TRY SLA MEASUREMENTS TO OMO CANONICAL TAXA
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8C — CROSSWALK TRY SLA MEASUREMENTS TO OMO CANONICAL TAXA")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm TRY measurement table is available
# -----------------------------------------------------------------------------

if "try_sla_measurement_working" not in globals():
    raise RuntimeError(
        "try_sla_measurement_working is not available. "
        "Run Section 8.8B first."
    )


# -----------------------------------------------------------------------------
# 2. Load the verified Omo → TRY taxonomic crosswalk if necessary
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

if "sla_try_resolved" not in globals():

    verified_crosswalk_path = (
        output_dir
        / "omo_sla_try_taxonomic_crosswalk_verified.csv"
    )

    if not verified_crosswalk_path.exists():
        raise FileNotFoundError(
            "Verified Omo → TRY taxonomic crosswalk was not found. "
            "Run Sections 8.3C–8.3D first."
        )

    sla_try_resolved = pd.read_csv(
        verified_crosswalk_path
    )


# -----------------------------------------------------------------------------
# 3. Ensure resolved TRY IDs are numeric
# -----------------------------------------------------------------------------

sla_try_resolved[
    "Resolved_TRY_ID_numeric"
] = pd.to_numeric(
    sla_try_resolved["Resolved_TRY_ID"],
    errors="coerce"
).astype("Int64")


# -----------------------------------------------------------------------------
# 4. Restrict crosswalk to defensibly resolved species-level records
# -----------------------------------------------------------------------------

resolved_crosswalk = (
    sla_try_resolved.loc[
        sla_try_resolved[
            "Resolution_Status"
        ].eq("RESOLVED")
        &
        sla_try_resolved[
            "Resolved_TRY_ID_numeric"
        ].notna()
    ]
    [
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",
            "Resolved_TRY_Name",
            "Resolved_TRY_ID_numeric",
            "Resolution_Method",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Resolved_TRY_ID_numeric": "AccSpeciesID"
        }
    )
)


# -----------------------------------------------------------------------------
# 5. Identify TRY species that actually returned SLA measurements
# -----------------------------------------------------------------------------

measured_try_species = (
    try_sla_measurement_working[
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    ]
    .drop_duplicates()
    .copy()
)

measured_try_species[
    "AccSpeciesID"
] = (
    pd.to_numeric(
        measured_try_species[
            "AccSpeciesID"
        ],
        errors="coerce"
    )
    .astype("Int64")
)


# -----------------------------------------------------------------------------
# 6. Attach TRY measurement availability to resolved Omo records
# -----------------------------------------------------------------------------

omo_try_coverage = (
    resolved_crosswalk
    .merge(
        measured_try_species,
        how="left",
        on="AccSpeciesID"
    )
)


omo_try_coverage[
    "TRY_SLA_Available"
] = (
    omo_try_coverage[
        "AccSpeciesName"
    ].notna()
)


omo_try_coverage[
    "TRY_Coverage_Status"
] = np.where(
    omo_try_coverage[
        "TRY_SLA_Available"
    ],
    "SPECIES_TRY_SLA_AVAILABLE",
    "REQUESTED_BUT_NO_PUBLIC_SLA"
)


# -----------------------------------------------------------------------------
# 7. Add measurement counts for covered taxa
# -----------------------------------------------------------------------------

try_measurement_counts = (
    try_sla_measurement_working
    .groupby(
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
    .agg(
        TRY_SLA_Record_Count=(
            "ObsDataID",
            "count"
        ),
        TRY_Unique_Trait_Definitions=(
            "TraitID",
            "nunique"
        ),
        TRY_Unique_Datasets=(
            "DatasetID",
            "nunique"
        ),
        TRY_Unique_Observations=(
            "ObservationID",
            "nunique"
        ),
    )
    .reset_index()
)


omo_try_coverage = (
    omo_try_coverage
    .drop(
        columns=[
            "AccSpeciesName"
        ]
    )
    .merge(
        try_measurement_counts,
        how="left",
        on="AccSpeciesID"
    )
)


count_columns = [
    "TRY_SLA_Record_Count",
    "TRY_Unique_Trait_Definitions",
    "TRY_Unique_Datasets",
    "TRY_Unique_Observations",
]

omo_try_coverage[
    count_columns
] = (
    omo_try_coverage[
        count_columns
    ]
    .fillna(0)
    .astype(int)
)


# -----------------------------------------------------------------------------
# 8. Determine multiplicity of Omo records per accepted TRY species
# -----------------------------------------------------------------------------

omo_per_try_species = (
    resolved_crosswalk
    .groupby(
        [
            "AccSpeciesID",
            "Resolved_TRY_Name",
        ],
        as_index=False
    )
    .agg(
        Omo_Query_Record_Count=(
            "SLA_Query_ID",
            "count"
        ),
        Omo_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
)


omo_try_coverage = (
    omo_try_coverage
    .merge(
        omo_per_try_species[
            [
                "AccSpeciesID",
                "Omo_Query_Record_Count",
                "Omo_Canonical_Taxa",
            ]
        ],
        how="left",
        on="AccSpeciesID"
    )
)


# -----------------------------------------------------------------------------
# 9. Summary counts
# -----------------------------------------------------------------------------

n_resolved_omo_rows = len(
    omo_try_coverage
)

n_covered_omo_rows = (
    omo_try_coverage[
        "TRY_Coverage_Status"
    ]
    .eq("SPECIES_TRY_SLA_AVAILABLE")
    .sum()
)

n_uncovered_omo_rows = (
    omo_try_coverage[
        "TRY_Coverage_Status"
    ]
    .eq("REQUESTED_BUT_NO_PUBLIC_SLA")
    .sum()
)

n_unique_try_requested = (
    resolved_crosswalk[
        "AccSpeciesID"
    ].nunique()
)

n_unique_try_measured = (
    omo_try_coverage.loc[
        omo_try_coverage[
            "TRY_SLA_Available"
        ],
        "AccSpeciesID"
    ]
    .nunique()
)

n_unique_omo_taxa_covered = (
    omo_try_coverage.loc[
        omo_try_coverage[
            "TRY_SLA_Available"
        ],
        "Canonical_Species"
    ]
    .nunique()
)


# -----------------------------------------------------------------------------
# 10. Identify TRY species shared by multiple Omo records
# -----------------------------------------------------------------------------

shared_try_taxa = (
    omo_per_try_species.loc[
        omo_per_try_species[
            "Omo_Query_Record_Count"
        ] > 1
    ]
    .copy()
    .sort_values(
        [
            "Omo_Query_Record_Count",
            "Resolved_TRY_Name",
        ],
        ascending=[
            False,
            True,
        ]
    )
)


# -----------------------------------------------------------------------------
# 11. Build covered and uncovered Omo tables
# -----------------------------------------------------------------------------

omo_try_covered = (
    omo_try_coverage.loc[
        omo_try_coverage[
            "TRY_SLA_Available"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

omo_try_uncovered = (
    omo_try_coverage.loc[
        ~omo_try_coverage[
            "TRY_SLA_Available"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 12. Validation
# -----------------------------------------------------------------------------

assert n_resolved_omo_rows == 114, (
    f"Expected 114 resolved Omo query rows, "
    f"found {n_resolved_omo_rows}."
)

assert n_unique_try_requested == 82, (
    f"Expected 82 unique resolved TRY IDs, "
    f"found {n_unique_try_requested}."
)

assert n_unique_try_measured == 36, (
    f"Expected 36 TRY species with SLA measurements, "
    f"found {n_unique_try_measured}."
)

assert (
    n_covered_omo_rows
    + n_uncovered_omo_rows
    == n_resolved_omo_rows
)


# -----------------------------------------------------------------------------
# 13. Export
# -----------------------------------------------------------------------------

coverage_path = (
    output_dir
    / "omo_sla_omo_try_species_coverage_crosswalk.csv"
)

covered_path = (
    output_dir
    / "omo_sla_omo_taxa_with_try_sla.csv"
)

uncovered_path = (
    output_dir
    / "omo_sla_omo_taxa_without_try_sla.csv"
)

shared_path = (
    output_dir
    / "omo_sla_try_taxa_shared_by_multiple_omo_records.csv"
)


omo_try_coverage.to_csv(
    coverage_path,
    index=False
)

omo_try_covered.to_csv(
    covered_path,
    index=False
)

omo_try_uncovered.to_csv(
    uncovered_path,
    index=False
)

shared_try_taxa.to_csv(
    shared_path,
    index=False
)


# -----------------------------------------------------------------------------
# 14. Diagnostics
# -----------------------------------------------------------------------------

print("\nOmo → TRY species-level SLA coverage")
print("-" * 100)

print(
    f"Resolved Omo query rows          : "
    f"{n_resolved_omo_rows}"
)

print(
    f"Unique TRY species requested     : "
    f"{n_unique_try_requested}"
)

print(
    f"TRY species with SLA records     : "
    f"{n_unique_try_measured}"
)

print(
    f"Omo query rows with TRY SLA      : "
    f"{n_covered_omo_rows}"
)

print(
    f"Unique Omo taxa with TRY SLA     : "
    f"{n_unique_omo_taxa_covered}"
)

print(
    f"Resolved Omo rows without TRY SLA: "
    f"{n_uncovered_omo_rows}"
)


print("\nCoverage-status distribution")
print("-" * 100)

print(
    omo_try_coverage[
        "TRY_Coverage_Status"
    ]
    .value_counts()
    .to_string()
)


print("\nOmo taxa with direct TRY SLA coverage")
print("-" * 100)

display(
    omo_try_covered[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "Resolved_TRY_Name",
            "AccSpeciesID",
            "TRY_SLA_Record_Count",
            "TRY_Unique_Trait_Definitions",
            "TRY_Unique_Datasets",
            "Resolution_Method",
        ]
    ]
    .sort_values(
        [
            "Canonical_Species"
        ]
    )
)


print("\nResolved Omo taxa without public TRY SLA measurements")
print("-" * 100)

display(
    omo_try_uncovered[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "Resolved_TRY_Name",
            "AccSpeciesID",
            "Resolution_Method",
            "TRY_Coverage_Status",
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
)


print("\nTRY accepted species shared by multiple Omo records")
print("-" * 100)

if shared_try_taxa.empty:
    print(
        "No accepted TRY species are shared by multiple Omo query records."
    )
else:
    display(shared_try_taxa)


print("\nExports")
print("-" * 100)

print(f"Complete crosswalk : {coverage_path}")
print(f"Covered Omo taxa   : {covered_path}")
print(f"Uncovered Omo taxa : {uncovered_path}")
print(f"Shared TRY taxa    : {shared_path}")

print("\nSection 8.8C completed successfully.")
```

### 8.8D Quality Control of TRY SLA Measurements

The 535 extracted TRY SLA measurements provide direct species-level evidence for 36 accepted TRY taxa and, through the established crosswalk, 50 Omo Forest canonical taxa. Before species-level SLA values are calculated, the measurement records are subjected to explicit quality control.

Several species display highly asymmetric or implausibly low standardized SLA values in the raw summaries. Such records may reflect unit conversion problems, data-entry errors, unusual experimental contexts, or valid but extreme biological observations. Consequently, raw measurements are not filtered solely on numerical magnitude.

Quality control therefore considers:

- TRY `ErrorRisk`;
- standardized SLA value;
- original reported value and unit;
- dataset provenance;
- within-species distribution;
- and statistical outlier status.

TRY records are retained in the raw archive regardless of QC outcome. Records judged unsuitable for species-level aggregation are only excluded from the working summary table and remain fully traceable through their original identifiers and provenance.

Species-level filtering uses a conservative combination of TRY error-risk screening and within-species interquartile-range diagnostics. The resulting retained measurements will subsequently be summarized using robust species-level medians.

```python
# =============================================================================
# SECTION 8.8D — QUALITY CONTROL OF TRY SLA MEASUREMENTS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8D — QUALITY CONTROL OF TRY SLA MEASUREMENTS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm TRY measurement table is available
# -----------------------------------------------------------------------------

if "try_sla_measurement_working" not in globals():
    raise RuntimeError(
        "try_sla_measurement_working is not available. "
        "Run Section 8.8B first."
    )

try_sla_qc = (
    try_sla_measurement_working.copy()
)


# -----------------------------------------------------------------------------
# 2. Validate standardized SLA units
# -----------------------------------------------------------------------------

assert (
    try_sla_qc["UnitName"]
    .eq("mm2 mg-1")
    .all()
), "Unexpected standardized units detected."


# -----------------------------------------------------------------------------
# 3. Ensure core numeric fields
# -----------------------------------------------------------------------------

try_sla_qc[
    "SLA_Std"
] = pd.to_numeric(
    try_sla_qc["StdValue_numeric"],
    errors="coerce"
)

try_sla_qc[
    "ErrorRisk_numeric"
] = pd.to_numeric(
    try_sla_qc["ErrorRisk"],
    errors="coerce"
)


assert (
    try_sla_qc["SLA_Std"]
    .notna()
    .all()
)


# -----------------------------------------------------------------------------
# 4. Inspect ErrorRisk distribution
# -----------------------------------------------------------------------------

errorrisk_distribution = (
    try_sla_qc[
        "ErrorRisk_numeric"
    ]
    .value_counts(
        dropna=False
    )
    .sort_index()
    .rename_axis(
        "ErrorRisk"
    )
    .reset_index(
        name="Record_Count"
    )
)


# -----------------------------------------------------------------------------
# 5. Define conservative TRY error-risk rule
#    ErrorRisk >= 4 is treated as elevated risk.
# -----------------------------------------------------------------------------

try_sla_qc[
    "ErrorRisk_Flag"
] = np.where(
    try_sla_qc[
        "ErrorRisk_numeric"
    ] >= 4,
    "ELEVATED_ERROR_RISK",
    "ACCEPTABLE_ERROR_RISK"
)


# -----------------------------------------------------------------------------
# 6. Calculate within-species IQR thresholds
# -----------------------------------------------------------------------------

species_iqr = (
    try_sla_qc
    .groupby(
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )["SLA_Std"]
    .agg(
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
    )
    .reset_index()
)

species_iqr[
    "IQR"
] = (
    species_iqr["Q3"]
    - species_iqr["Q1"]
)

species_iqr[
    "Lower_IQR_Bound"
] = (
    species_iqr["Q1"]
    - 1.5 * species_iqr["IQR"]
)

species_iqr[
    "Upper_IQR_Bound"
] = (
    species_iqr["Q3"]
    + 1.5 * species_iqr["IQR"]
)


# -----------------------------------------------------------------------------
# 7. Attach species thresholds
# -----------------------------------------------------------------------------

try_sla_qc = (
    try_sla_qc
    .merge(
        species_iqr,
        how="left",
        on=[
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
)


# -----------------------------------------------------------------------------
# 8. Statistical outlier flag
# -----------------------------------------------------------------------------

try_sla_qc[
    "IQR_Outlier"
] = (
    (
        try_sla_qc["SLA_Std"]
        < try_sla_qc["Lower_IQR_Bound"]
    )
    |
    (
        try_sla_qc["SLA_Std"]
        > try_sla_qc["Upper_IQR_Bound"]
    )
)


# -----------------------------------------------------------------------------
# 9. Flag extremely low SLA records for diagnostic review
#    This is diagnostic only, not an automatic exclusion rule.
# -----------------------------------------------------------------------------

try_sla_qc[
    "Extreme_Low_Diagnostic"
] = (
    try_sla_qc[
        "SLA_Std"
    ] < 1.0
)


# -----------------------------------------------------------------------------
# 10. Combine QC decision
# -----------------------------------------------------------------------------

try_sla_qc[
    "QC_Status"
] = "RETAIN"

try_sla_qc.loc[
    try_sla_qc[
        "ErrorRisk_Flag"
    ].eq(
        "ELEVATED_ERROR_RISK"
    ),
    "QC_Status"
] = "EXCLUDE_ERROR_RISK"

try_sla_qc.loc[
    try_sla_qc[
        "IQR_Outlier"
    ]
    &
    try_sla_qc[
        "QC_Status"
    ].eq("RETAIN"),
    "QC_Status"
] = "EXCLUDE_IQR_OUTLIER"


# -----------------------------------------------------------------------------
# 11. Build retained measurement table
# -----------------------------------------------------------------------------

try_sla_retained = (
    try_sla_qc.loc[
        try_sla_qc[
            "QC_Status"
        ].eq("RETAIN")
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 12. Species-level QC summary
# -----------------------------------------------------------------------------

try_species_qc_summary = (
    try_sla_qc
    .groupby(
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
    .agg(
        Raw_Record_Count=(
            "ObsDataID",
            "count"
        ),
        ErrorRisk_Excluded_Count=(
            "QC_Status",
            lambda x: (
                x == "EXCLUDE_ERROR_RISK"
            ).sum()
        ),
        IQR_Excluded_Count=(
            "QC_Status",
            lambda x: (
                x == "EXCLUDE_IQR_OUTLIER"
            ).sum()
        ),
        Extreme_Low_Count=(
            "Extreme_Low_Diagnostic",
            "sum"
        ),
        Raw_Min_SLA=(
            "SLA_Std",
            "min"
        ),
        Raw_Median_SLA=(
            "SLA_Std",
            "median"
        ),
        Raw_Max_SLA=(
            "SLA_Std",
            "max"
        ),
    )
    .reset_index()
)


retained_species_summary = (
    try_sla_retained
    .groupby(
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
    .agg(
        Retained_Record_Count=(
            "ObsDataID",
            "count"
        ),
        Retained_Dataset_Count=(
            "DatasetID",
            "nunique"
        ),
        Retained_Trait_Definition_Count=(
            "TraitID",
            "nunique"
        ),
        Retained_Min_SLA=(
            "SLA_Std",
            "min"
        ),
        Retained_Median_SLA=(
            "SLA_Std",
            "median"
        ),
        Retained_Mean_SLA=(
            "SLA_Std",
            "mean"
        ),
        Retained_Max_SLA=(
            "SLA_Std",
            "max"
        ),
    )
    .reset_index()
)


try_species_qc_summary = (
    try_species_qc_summary
    .merge(
        retained_species_summary,
        how="left",
        on=[
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
)


# -----------------------------------------------------------------------------
# 13. Evidence-strength classification
# -----------------------------------------------------------------------------

def classify_try_strength(row):

    n_records = row["Retained_Record_Count"]
    n_datasets = row["Retained_Dataset_Count"]

    if pd.isna(n_records):
        return "NO_RETAINED_RECORDS"

    if n_records >= 10 and n_datasets >= 3:
        return "STRONG"

    if n_records >= 3 and n_datasets >= 2:
        return "MODERATE"

    if n_records >= 1:
        return "LIMITED"

    return "NO_RETAINED_RECORDS"


try_species_qc_summary[
    "Evidence_Strength"
] = (
    try_species_qc_summary.apply(
        classify_try_strength,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 14. Diagnostic subsets
# -----------------------------------------------------------------------------

extreme_low_records = (
    try_sla_qc.loc[
        try_sla_qc[
            "Extreme_Low_Diagnostic"
        ]
    ]
    [
        [
            "AccSpeciesName",
            "SLA_Std",
            "OrigValueStr",
            "OrigUnitStr",
            "StdValue",
            "UnitName",
            "ErrorRisk_numeric",
            "Dataset",
            "ObservationID",
            "ObsDataID",
            "Reference",
            "QC_Status",
        ]
    ]
    .sort_values(
        [
            "AccSpeciesName",
            "SLA_Std",
        ]
    )
)


excluded_records = (
    try_sla_qc.loc[
        ~try_sla_qc[
            "QC_Status"
        ].eq("RETAIN")
    ]
    [
        [
            "AccSpeciesName",
            "SLA_Std",
            "OrigValueStr",
            "OrigUnitStr",
            "ErrorRisk_numeric",
            "IQR_Outlier",
            "Dataset",
            "ObservationID",
            "ObsDataID",
            "QC_Status",
        ]
    ]
    .sort_values(
        [
            "AccSpeciesName",
            "SLA_Std",
        ]
    )
)


# -----------------------------------------------------------------------------
# 15. Validation
# -----------------------------------------------------------------------------

assert len(
    try_sla_qc
) == 535

assert (
    len(
        try_sla_retained
    )
    <= 535
)

assert (
    try_species_qc_summary[
        "AccSpeciesID"
    ].nunique()
    == 36
)


# -----------------------------------------------------------------------------
# 16. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

qc_path = (
    output_dir
    / "omo_sla_try_measurement_level_qc.csv"
)

retained_path = (
    output_dir
    / "omo_sla_try_measurements_retained.csv"
)

species_qc_path = (
    output_dir
    / "omo_sla_try_species_qc_summary.csv"
)

extreme_low_path = (
    output_dir
    / "omo_sla_try_extreme_low_diagnostics.csv"
)

excluded_path = (
    output_dir
    / "omo_sla_try_excluded_measurements.csv"
)


try_sla_qc.to_csv(
    qc_path,
    index=False
)

try_sla_retained.to_csv(
    retained_path,
    index=False
)

try_species_qc_summary.to_csv(
    species_qc_path,
    index=False
)

extreme_low_records.to_csv(
    extreme_low_path,
    index=False
)

excluded_records.to_csv(
    excluded_path,
    index=False
)


# -----------------------------------------------------------------------------
# 17. Diagnostics
# -----------------------------------------------------------------------------

print("\nTRY ErrorRisk distribution")
print("-" * 100)

display(
    errorrisk_distribution
)


print("\nTRY measurement QC status")
print("-" * 100)

print(
    try_sla_qc[
        "QC_Status"
    ]
    .value_counts()
    .to_string()
)


print("\nMeasurement counts")
print("-" * 100)

print(
    f"Raw TRY SLA measurements : "
    f"{len(try_sla_qc)}"
)

print(
    f"Retained measurements    : "
    f"{len(try_sla_retained)}"
)

print(
    f"Excluded measurements    : "
    f"{len(try_sla_qc) - len(try_sla_retained)}"
)

print(
    f"Extreme-low diagnostics  : "
    f"{len(extreme_low_records)}"
)


print("\nSpecies-level QC summary")
print("-" * 100)

display(
    try_species_qc_summary[
        [
            "AccSpeciesName",
            "Raw_Record_Count",
            "ErrorRisk_Excluded_Count",
            "IQR_Excluded_Count",
            "Extreme_Low_Count",
            "Retained_Record_Count",
            "Retained_Dataset_Count",
            "Raw_Median_SLA",
            "Retained_Median_SLA",
            "Evidence_Strength",
        ]
    ]
    .sort_values(
        "AccSpeciesName"
    )
)


print("\nExtreme-low SLA diagnostic records")
print("-" * 100)

if extreme_low_records.empty:
    print(
        "No SLA records below 1.0 mm2 mg-1."
    )
else:
    display(
        extreme_low_records
    )


print("\nExcluded TRY SLA measurements")
print("-" * 100)

if excluded_records.empty:
    print(
        "No TRY SLA measurements were excluded."
    )
else:
    display(
        excluded_records
    )


print("\nExports")
print("-" * 100)

print(f"Measurement QC      : {qc_path}")
print(f"Retained records    : {retained_path}")
print(f"Species QC summary  : {species_qc_path}")
print(f"Extreme-low records : {extreme_low_path}")
print(f"Excluded records    : {excluded_path}")

print("\nSection 8.8D completed successfully.")
```

```python
# =============================================================================
# SECTION 8.8E1 — DIAGNOSTIC REVIEW OF ZERO-RETENTION TRY SLA SPECIES
# =============================================================================

import numpy as np
import pandas as pd
from pathlib import Path

print("=" * 100)
print("SECTION 8.8E1 — DIAGNOSTIC REVIEW OF ZERO-RETENTION TRY SLA SPECIES")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. Confirm required notebook objects
# -----------------------------------------------------------------------------

required_objects = [
    "try_sla_qc",
    "try_species_qc_summary",
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(f"  - {obj}" for obj in missing_objects)
        + "\n\nRun Section 8.8D before Section 8.8E1."
    )


# -----------------------------------------------------------------------------
# 2. Identify species for which automatic QC retained no measurements
# -----------------------------------------------------------------------------

zero_retention_species = (
    try_species_qc_summary.loc[
        try_species_qc_summary["Evidence_Strength"]
        .eq("NO_RETAINED_RECORDS"),
        "AccSpeciesName"
    ]
    .dropna()
    .astype(str)
    .drop_duplicates()
    .sort_values()
    .tolist()
)

print("\nZero-retention species")
print("-" * 100)
print(f"Species requiring diagnostic review : {len(zero_retention_species)}")

for species in zero_retention_species:
    print(f"  - {species}")


# -----------------------------------------------------------------------------
# 3. Extract all original measurements for these species
# -----------------------------------------------------------------------------

zero_retention_records = (
    try_sla_qc.loc[
        try_sla_qc["AccSpeciesName"].isin(zero_retention_species)
    ]
    .copy()
)

if zero_retention_records.empty:
    raise RuntimeError(
        "No measurement records were recovered for the zero-retention species."
    )


# -----------------------------------------------------------------------------
# 4. Create broad numerical diagnostic classes
#
# IMPORTANT:
# These are diagnostic categories only.
# They do NOT determine acceptance or rejection.
# -----------------------------------------------------------------------------

zero_retention_records["SLA_Diagnostic_Class"] = np.select(
    [
        zero_retention_records["SLA_Std"] < 1,
        zero_retention_records["SLA_Std"].between(1, 5, inclusive="left"),
        zero_retention_records["SLA_Std"].between(5, 10, inclusive="left"),
        zero_retention_records["SLA_Std"].between(10, 40, inclusive="both"),
        zero_retention_records["SLA_Std"] > 40,
    ],
    [
        "EXTREME_LOW_LT_1",
        "VERY_LOW_1_TO_LT_5",
        "LOW_5_TO_LT_10",
        "CENTRAL_DIAGNOSTIC_RANGE_10_TO_40",
        "HIGH_GT_40",
    ],
    default="UNCLASSIFIED"
)


# -----------------------------------------------------------------------------
# 5. Dataset-level summaries
# -----------------------------------------------------------------------------

dataset_diagnostic = (
    zero_retention_records
    .groupby(
        [
            "AccSpeciesName",
            "DatasetID",
            "Dataset",
            "OrigUnitStr",
        ],
        dropna=False
    )
    .agg(
        Record_Count=("SLA_Std", "size"),
        Min_SLA=("SLA_Std", "min"),
        Median_SLA=("SLA_Std", "median"),
        Mean_SLA=("SLA_Std", "mean"),
        Max_SLA=("SLA_Std", "max"),
        Min_ErrorRisk=("ErrorRisk_numeric", "min"),
        Median_ErrorRisk=("ErrorRisk_numeric", "median"),
        Max_ErrorRisk=("ErrorRisk_numeric", "max"),
        Unique_Observations=("ObservationID", "nunique"),
    )
    .reset_index()
    .sort_values(
        ["AccSpeciesName", "Median_SLA", "Dataset"],
        na_position="last"
    )
)


# -----------------------------------------------------------------------------
# 6. Species-level diagnostic summary
# -----------------------------------------------------------------------------

species_diagnostic = (
    zero_retention_records
    .groupby("AccSpeciesName", dropna=False)
    .agg(
        Raw_Record_Count=("SLA_Std", "size"),
        Dataset_Count=("DatasetID", "nunique"),
        Observation_Count=("ObservationID", "nunique"),
        Min_SLA=("SLA_Std", "min"),
        Median_SLA=("SLA_Std", "median"),
        Mean_SLA=("SLA_Std", "mean"),
        Max_SLA=("SLA_Std", "max"),
        Extreme_Low_Count=(
            "SLA_Diagnostic_Class",
            lambda x: (x == "EXTREME_LOW_LT_1").sum()
        ),
        Central_Range_Count=(
            "SLA_Diagnostic_Class",
            lambda x: (
                x == "CENTRAL_DIAGNOSTIC_RANGE_10_TO_40"
            ).sum()
        ),
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 7. Identify independent support within the central diagnostic range
#
# This remains diagnostic evidence only.
# -----------------------------------------------------------------------------

central_candidates = (
    zero_retention_records.loc[
        zero_retention_records["SLA_Diagnostic_Class"]
        .eq("CENTRAL_DIAGNOSTIC_RANGE_10_TO_40")
    ]
    .copy()
)

if not central_candidates.empty:

    central_support = (
        central_candidates
        .groupby("AccSpeciesName", dropna=False)
        .agg(
            Candidate_Record_Count=("SLA_Std", "size"),
            Candidate_Dataset_Count=("DatasetID", "nunique"),
            Candidate_Observation_Count=("ObservationID", "nunique"),
            Candidate_Min_SLA=("SLA_Std", "min"),
            Candidate_Median_SLA=("SLA_Std", "median"),
            Candidate_Max_SLA=("SLA_Std", "max"),
        )
        .reset_index()
    )

else:

    central_support = pd.DataFrame(
        columns=[
            "AccSpeciesName",
            "Candidate_Record_Count",
            "Candidate_Dataset_Count",
            "Candidate_Observation_Count",
            "Candidate_Min_SLA",
            "Candidate_Median_SLA",
            "Candidate_Max_SLA",
        ]
    )


species_diagnostic = species_diagnostic.merge(
    central_support,
    on="AccSpeciesName",
    how="left"
)

count_columns = [
    "Candidate_Record_Count",
    "Candidate_Dataset_Count",
    "Candidate_Observation_Count",
]

for col in count_columns:
    species_diagnostic[col] = (
        species_diagnostic[col]
        .fillna(0)
        .astype(int)
    )


# -----------------------------------------------------------------------------
# 8. Assign diagnostic review priority
#
# Again: this is NOT an acceptance decision.
# -----------------------------------------------------------------------------

def assign_review_priority(row):

    if row["Candidate_Dataset_Count"] >= 2:
        return "HIGH_PRIORITY_FOR_RESCUE_REVIEW"

    elif row["Candidate_Record_Count"] >= 2:
        return "REVIEW_SINGLE_DATASET_SUPPORT"

    elif row["Candidate_Record_Count"] == 1:
        return "REVIEW_SINGLE_RECORD"

    else:
        return "NO_CENTRAL_RANGE_RESCUE_CANDIDATE"


species_diagnostic["Rescue_Review_Priority"] = (
    species_diagnostic.apply(
        assign_review_priority,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 9. Create record-level review register
# -----------------------------------------------------------------------------

review_columns = [
    "AccSpeciesName",
    "AccSpeciesID",
    "SLA_Std",
    "OrigValueStr",
    "OrigUnitStr",
    "StdValue",
    "UnitName",
    "ErrorRisk_numeric",
    "SLA_Diagnostic_Class",
    "DatasetID",
    "Dataset",
    "ObservationID",
    "ObsDataID",
    "TraitID",
    "TraitName",
    "Reference",
    "QC_Status",
]

available_review_columns = [
    col for col in review_columns
    if col in zero_retention_records.columns
]

try_rescue_review_register = (
    zero_retention_records[
        available_review_columns
    ]
    .copy()
    .sort_values(
        ["AccSpeciesName", "SLA_Std", "Dataset"]
    )
)

# Explicitly leave decisions unresolved at this stage.
try_rescue_review_register["Rescue_Decision"] = "PENDING_REVIEW"
try_rescue_review_register["Rescue_Reason"] = pd.NA


# -----------------------------------------------------------------------------
# 10. Display diagnostics
# -----------------------------------------------------------------------------

print("\nSpecies-level diagnostic summary")
print("-" * 100)

display(
    species_diagnostic.sort_values(
        [
            "Rescue_Review_Priority",
            "AccSpeciesName"
        ]
    )
)


print("\nDataset-level diagnostic summary")
print("-" * 100)

display(dataset_diagnostic)


print("\nCentral-range rescue candidates")
print("-" * 100)

if central_candidates.empty:
    print("No central-range candidates identified.")
else:
    display(
        central_candidates[
            [
                col for col in [
                    "AccSpeciesName",
                    "SLA_Std",
                    "OrigValueStr",
                    "OrigUnitStr",
                    "ErrorRisk_numeric",
                    "Dataset",
                    "ObservationID",
                    "Reference",
                    "QC_Status",
                ]
                if col in central_candidates.columns
            ]
        ].sort_values(
            ["AccSpeciesName", "SLA_Std"]
        )
    )


# -----------------------------------------------------------------------------
# 11. Export diagnostic evidence
# -----------------------------------------------------------------------------

output_dir = PROJECT_ROOT / "outputs" / "functional_traits"
output_dir.mkdir(parents=True, exist_ok=True)

species_path = (
    output_dir /
    "omo_sla_try_zero_retention_species_diagnostic.csv"
)

dataset_path = (
    output_dir /
    "omo_sla_try_zero_retention_dataset_diagnostic.csv"
)

candidate_path = (
    output_dir /
    "omo_sla_try_zero_retention_central_candidates.csv"
)

review_path = (
    output_dir /
    "omo_sla_try_rescue_review_register.csv"
)

species_diagnostic.to_csv(
    species_path,
    index=False
)

dataset_diagnostic.to_csv(
    dataset_path,
    index=False
)

central_candidates.to_csv(
    candidate_path,
    index=False
)

try_rescue_review_register.to_csv(
    review_path,
    index=False
)


# -----------------------------------------------------------------------------
# -----------------------------------------------------------------------------
# 12. Final report
# -----------------------------------------------------------------------------

n_multi_dataset_support = int(
    (
        species_diagnostic["Candidate_Dataset_Count"] >= 2
    ).sum()
)

print("\nDiagnostic summary")
print("-" * 100)

print(
    f"Zero-retention species reviewed : "
    f"{len(zero_retention_species)}"
)

print(
    f"Measurements examined           : "
    f"{len(zero_retention_records)}"
)

print(
    f"Central-range candidates        : "
    f"{len(central_candidates)}"
)

print(
    f"Species with >=2 supporting datasets : "
    f"{n_multi_dataset_support}"
)


print("\nExports")
print("-" * 100)

print(f"Species diagnostics : {species_path}")
print(f"Dataset diagnostics : {dataset_path}")
print(f"Central candidates  : {candidate_path}")
print(f"Review register     : {review_path}")

print(
    "\nSection 8.8E1 completed successfully."
)
```

### 8.8E2 Controlled Rescue of Coherent TRY SLA Measurements

The diagnostic review of species for which the automatic TRY quality-control procedure retained no measurements identified two distinct situations.

For *Blighia sapida* and *Nesogordonia papaverifera*, multiple independent datasets provide mutually consistent SLA measurements within a biologically coherent numerical range. These measurements are retained despite elevated TRY `ErrorRisk`, because cross-dataset agreement provides stronger direct evidence for their suitability than the generic ErrorRisk threshold alone.

For *Albizia zygia* and *Sterculia rhinopetala*, a smaller number of internally coherent measurements were identified within individual datasets. These records are rescued conservatively and assigned a limited-confidence rescue status because independent multi-dataset confirmation is absent.

No measurements are rescued for *Carica papaya*, *Sterculia oblonga*, or *Terminalia superba*. Their available TRY records consist exclusively of extremely low SLA values from a single source and lack independent corroborating measurements.

Rescue decisions are applied at the individual TRY record level using unique `ObsDataID` identifiers. All original measurements and automatic QC classifications remain preserved. A separate final QC status records whether each measurement was retained automatically, rescued following manual evidence review, or excluded.

This controlled intervention avoids weakening the general quality-control rule while preventing coherent, independently supported species-level SLA evidence from being discarded solely because of elevated TRY ErrorRisk.

```python
# =============================================================================
# SECTION 8.8E2 — CONTROLLED RESCUE OF COHERENT TRY SLA MEASUREMENTS
# =============================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 100)
print("SECTION 8.8E2 — CONTROLLED RESCUE OF COHERENT TRY SLA MEASUREMENTS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "try_sla_qc",
    "try_rescue_review_register",
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\nRun Sections 8.8D and 8.8E1 first."
    )


# -----------------------------------------------------------------------------
# 2. Define explicit record-level rescue decisions
#
# Records are identified by ObsDataID so that decisions are reproducible
# and independent of DataFrame row order.
# -----------------------------------------------------------------------------

rescue_registry = pd.DataFrame(
    [
        # ---------------------------------------------------------------------
        # Albizia zygia — limited-confidence rescue
        # ---------------------------------------------------------------------
        {
            "ObsDataID": 25404399,
            "Expected_Species": "Albizia zygia",
            "Rescue_Class": "RESCUE_LIMITED_SINGLE_DATASET",
            "Rescue_Reason": (
                "Plausible SLA value within same dataset; "
                "extreme-low companion records remain excluded."
            ),
        },
        {
            "ObsDataID": 25404406,
            "Expected_Species": "Albizia zygia",
            "Rescue_Class": "RESCUE_LIMITED_SINGLE_DATASET",
            "Rescue_Reason": (
                "Plausible SLA value within same dataset; "
                "extreme-low companion records remain excluded."
            ),
        },

        # ---------------------------------------------------------------------
        # Blighia sapida — strong rescue due to independent dataset agreement
        # ---------------------------------------------------------------------
        {
            "ObsDataID": 13257578,
            "Expected_Species": "Blighia sapida",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "Coherent SLA measurement supported by agreement "
                "across independent datasets."
            ),
        },
        {
            "ObsDataID": 20438544,
            "Expected_Species": "Blighia sapida",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "Independent dataset agrees closely with TROBIT "
                "measurements."
            ),
        },
        {
            "ObsDataID": 13257661,
            "Expected_Species": "Blighia sapida",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "Coherent SLA measurement supported by agreement "
                "across independent datasets."
            ),
        },
        {
            "ObsDataID": 13257745,
            "Expected_Species": "Blighia sapida",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "Coherent SLA measurement supported by agreement "
                "across independent datasets."
            ),
        },

        # ---------------------------------------------------------------------
        # Nesogordonia papaverifera — strong rescue
        # ---------------------------------------------------------------------
        {
            "ObsDataID": 20438979,
            "Expected_Species": "Nesogordonia papaverifera",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "17–18 mm2 mg-1 measurement cluster independently "
                "supported across two datasets; Tene low-value cluster "
                "remains excluded."
            ),
        },
        {
            "ObsDataID": 13238227,
            "Expected_Species": "Nesogordonia papaverifera",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "17–18 mm2 mg-1 measurement cluster independently "
                "supported across two datasets."
            ),
        },
        {
            "ObsDataID": 13238291,
            "Expected_Species": "Nesogordonia papaverifera",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "17–18 mm2 mg-1 measurement cluster independently "
                "supported across two datasets."
            ),
        },
        {
            "ObsDataID": 13238204,
            "Expected_Species": "Nesogordonia papaverifera",
            "Rescue_Class": "RESCUE_MULTI_DATASET_SUPPORT",
            "Rescue_Reason": (
                "17–18 mm2 mg-1 measurement cluster independently "
                "supported across two datasets."
            ),
        },

        # ---------------------------------------------------------------------
        # Sterculia rhinopetala — limited-confidence rescue
        # ---------------------------------------------------------------------
        {
            "ObsDataID": 13238355,
            "Expected_Species": "Sterculia rhinopetala",
            "Rescue_Class": "RESCUE_LIMITED_SINGLE_DATASET",
            "Rescue_Reason": (
                "Two coherent TROBIT measurements around "
                "12 mm2 mg-1; extreme-low Tene records remain excluded."
            ),
        },
        {
            "ObsDataID": 13238397,
            "Expected_Species": "Sterculia rhinopetala",
            "Rescue_Class": "RESCUE_LIMITED_SINGLE_DATASET",
            "Rescue_Reason": (
                "Two coherent TROBIT measurements around "
                "12 mm2 mg-1; extreme-low Tene records remain excluded."
            ),
        },
    ]
)


# -----------------------------------------------------------------------------
# 3. Validate rescue IDs exist exactly once
# -----------------------------------------------------------------------------

source_ids = set(
    pd.to_numeric(
        try_sla_qc["ObsDataID"],
        errors="coerce"
    )
    .dropna()
    .astype(int)
)

rescue_ids = set(
    rescue_registry["ObsDataID"]
    .astype(int)
)

missing_rescue_ids = (
    rescue_ids - source_ids
)

assert not missing_rescue_ids, (
    "Rescue registry contains ObsDataID values "
    "not found in TRY QC data: "
    f"{sorted(missing_rescue_ids)}"
)


assert (
    rescue_registry["ObsDataID"].nunique()
    == len(rescue_registry)
), "Duplicate ObsDataID values found in rescue registry."


# -----------------------------------------------------------------------------
# 4. Validate taxon identity for every rescue record
# -----------------------------------------------------------------------------

rescue_validation = (
    try_sla_qc[
        [
            "ObsDataID",
            "AccSpeciesName",
            "SLA_Std",
            "Dataset",
            "QC_Status",
        ]
    ]
    .merge(
        rescue_registry,
        how="inner",
        on="ObsDataID"
    )
)


assert len(rescue_validation) == 12, (
    f"Expected 12 rescue records, "
    f"found {len(rescue_validation)}."
)


identity_mismatch = (
    rescue_validation[
        "AccSpeciesName"
    ]
    != rescue_validation[
        "Expected_Species"
    ]
)

assert not identity_mismatch.any(), (
    "One or more rescue ObsDataID values do not match "
    "the expected species."
)


# -----------------------------------------------------------------------------
# 5. Create final TRY QC table
# -----------------------------------------------------------------------------

try_sla_final_qc = (
    try_sla_qc.copy()
)

try_sla_final_qc[
    "Final_QC_Status"
] = (
    try_sla_final_qc[
        "QC_Status"
    ]
)

try_sla_final_qc[
    "Rescue_Class"
] = pd.NA

try_sla_final_qc[
    "Rescue_Reason"
] = pd.NA


# -----------------------------------------------------------------------------
# 6. Apply controlled rescue
# -----------------------------------------------------------------------------

try_sla_final_qc = (
    try_sla_final_qc
    .merge(
        rescue_registry[
            [
                "ObsDataID",
                "Rescue_Class",
                "Rescue_Reason",
            ]
        ],
        how="left",
        on="ObsDataID",
        suffixes=("", "_registry")
    )
)


rescue_mask = (
    try_sla_final_qc[
        "Rescue_Class_registry"
    ].notna()
)


try_sla_final_qc.loc[
    rescue_mask,
    "Final_QC_Status"
] = "RETAIN_RESCUED"


try_sla_final_qc.loc[
    rescue_mask,
    "Rescue_Class"
] = (
    try_sla_final_qc.loc[
        rescue_mask,
        "Rescue_Class_registry"
    ]
)


try_sla_final_qc.loc[
    rescue_mask,
    "Rescue_Reason"
] = (
    try_sla_final_qc.loc[
        rescue_mask,
        "Rescue_Reason_registry"
    ]
)


try_sla_final_qc = (
    try_sla_final_qc.drop(
        columns=[
            "Rescue_Class_registry",
            "Rescue_Reason_registry",
        ]
    )
)


# -----------------------------------------------------------------------------
# 7. Create definitive retained TRY table
# -----------------------------------------------------------------------------

final_retention_mask = (
    try_sla_final_qc[
        "Final_QC_Status"
    ].isin(
        [
            "RETAIN",
            "RETAIN_RESCUED",
        ]
    )
)


try_sla_final_retained = (
    try_sla_final_qc.loc[
        final_retention_mask
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 8. Final species-level summary after rescue
# -----------------------------------------------------------------------------

try_final_species_summary = (
    try_sla_final_retained
    .groupby(
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    )
    .agg(
        Final_Record_Count=(
            "ObsDataID",
            "count"
        ),
        Final_Dataset_Count=(
            "DatasetID",
            "nunique"
        ),
        Final_Observation_Count=(
            "ObservationID",
            "nunique"
        ),
        Final_Trait_Definition_Count=(
            "TraitID",
            "nunique"
        ),
        Final_Min_SLA=(
            "SLA_Std",
            "min"
        ),
        Final_Median_SLA=(
            "SLA_Std",
            "median"
        ),
        Final_Mean_SLA=(
            "SLA_Std",
            "mean"
        ),
        Final_Max_SLA=(
            "SLA_Std",
            "max"
        ),
        Rescued_Record_Count=(
            "Final_QC_Status",
            lambda x: (
                x == "RETAIN_RESCUED"
            ).sum()
        ),
    )
    .reset_index()
)


# -----------------------------------------------------------------------------
# 9. Assign final evidence-quality class
# -----------------------------------------------------------------------------

def final_try_evidence_strength(row):

    n_records = row["Final_Record_Count"]
    n_datasets = row["Final_Dataset_Count"]
    n_rescued = row["Rescued_Record_Count"]

    if n_records >= 10 and n_datasets >= 3:
        return "STRONG"

    if n_records >= 3 and n_datasets >= 2:
        return "MODERATE"

    if n_rescued > 0 and n_datasets >= 2:
        return "MODERATE_RESCUED"

    if n_rescued > 0:
        return "LIMITED_RESCUED"

    if n_records >= 1:
        return "LIMITED"

    return "NO_DATA"


try_final_species_summary[
    "Final_Evidence_Strength"
] = (
    try_final_species_summary.apply(
        final_try_evidence_strength,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 10. Determine remaining TRY species without usable records
# -----------------------------------------------------------------------------

all_measured_species = set(
    try_sla_qc[
        "AccSpeciesName"
    ].unique()
)

final_retained_species = set(
    try_sla_final_retained[
        "AccSpeciesName"
    ].unique()
)

remaining_no_try_value = sorted(
    all_measured_species
    - final_retained_species
)


# -----------------------------------------------------------------------------
# 11. Validation
# -----------------------------------------------------------------------------

n_raw = len(
    try_sla_final_qc
)

n_auto_retained = (
    try_sla_final_qc[
        "Final_QC_Status"
    ].eq("RETAIN")
    .sum()
)

n_rescued = (
    try_sla_final_qc[
        "Final_QC_Status"
    ].eq("RETAIN_RESCUED")
    .sum()
)

n_final_retained = len(
    try_sla_final_retained
)

n_final_species = (
    try_sla_final_retained[
        "AccSpeciesID"
    ].nunique()
)


assert n_raw == 535
assert n_rescued == 12
assert n_final_retained == 430

assert n_final_species == 33, (
    f"Expected 33 TRY species with final usable SLA records, "
    f"found {n_final_species}."
)

assert len(
    remaining_no_try_value
) == 3


expected_remaining = {
    "Carica papaya",
    "Sterculia oblonga",
    "Terminalia superba",
}

assert set(
    remaining_no_try_value
) == expected_remaining


# -----------------------------------------------------------------------------
# 12. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

rescue_registry_path = (
    output_dir
    / "omo_sla_try_controlled_rescue_registry.csv"
)

final_qc_path = (
    output_dir
    / "omo_sla_try_final_measurement_qc.csv"
)

final_retained_path = (
    output_dir
    / "omo_sla_try_final_retained_measurements.csv"
)

final_species_path = (
    output_dir
    / "omo_sla_try_final_species_summary.csv"
)


rescue_registry.to_csv(
    rescue_registry_path,
    index=False
)

try_sla_final_qc.to_csv(
    final_qc_path,
    index=False
)

try_sla_final_retained.to_csv(
    final_retained_path,
    index=False
)

try_final_species_summary.to_csv(
    final_species_path,
    index=False
)


# -----------------------------------------------------------------------------
# 13. Diagnostics
# -----------------------------------------------------------------------------

print("\nControlled rescue validation")
print("-" * 100)

display(
    rescue_validation[
        [
            "AccSpeciesName",
            "SLA_Std",
            "Dataset",
            "QC_Status",
            "Rescue_Class",
        ]
    ]
    .sort_values(
        [
            "AccSpeciesName",
            "SLA_Std",
        ]
    )
)


print("\nFinal TRY QC status distribution")
print("-" * 100)

print(
    try_sla_final_qc[
        "Final_QC_Status"
    ]
    .value_counts()
    .to_string()
)


print("\nFinal TRY measurement counts")
print("-" * 100)

print(
    f"Raw SLA measurements       : "
    f"{n_raw}"
)

print(
    f"Automatically retained     : "
    f"{n_auto_retained}"
)

print(
    f"Controlled rescue records  : "
    f"{n_rescued}"
)

print(
    f"Final retained measurements: "
    f"{n_final_retained}"
)

print(
    f"Final TRY species coverage : "
    f"{n_final_species}"
)


print("\nSpecies summaries after controlled rescue")
print("-" * 100)

display(
    try_final_species_summary[
        [
            "AccSpeciesName",
            "Final_Record_Count",
            "Final_Dataset_Count",
            "Rescued_Record_Count",
            "Final_Min_SLA",
            "Final_Median_SLA",
            "Final_Max_SLA",
            "Final_Evidence_Strength",
        ]
    ]
    .sort_values(
        "AccSpeciesName"
    )
)


print("\nTRY species remaining without usable SLA")
print("-" * 100)

for species in remaining_no_try_value:
    print(f"- {species}")


print("\nExports")
print("-" * 100)

print(f"Rescue registry : {rescue_registry_path}")
print(f"Final QC table  : {final_qc_path}")
print(f"Final retained  : {final_retained_path}")
print(f"Species summary : {final_species_path}")

print(
    "\nSection 8.8E2 completed successfully."
)
```

### 8.8F Final TRY Species-Level SLA Aggregation and Omo Forest Crosswalk

Following measurement-level quality control and controlled rescue review, 430 TRY SLA measurements are retained as suitable for species-level aggregation. These records represent 33 accepted TRY species.

The final TRY species-level SLA estimate is calculated as the median of retained standardized measurements for each accepted species. The median is used because TRY observations originate from multiple datasets, environments, experimental settings, and measurement contexts, and therefore a robust central statistic is preferable to an arithmetic mean for harmonized trait integration.

Each species-level estimate retains the number of contributing measurements, datasets, observations, trait-definition classes, rescued records, and the final evidence-strength classification established during quality control.

The resulting TRY species estimates are then mapped back to the validated Omo Forest canonical taxa using the controlled Omo-to-TRY taxonomic crosswalk developed earlier in the acquisition workflow. TRY accepted names remain external trait-retrieval identities only and do not replace Omo canonical taxonomy.

Species-level TRY estimates have priority over supplementary literature, genus-, family-, and global-level estimates. Omo taxa associated with TRY species for which all measurements failed final quality control remain unresolved and continue through the fallback hierarchy.

No lower-priority trait source is merged into the final SLA table in this subsection.

```python
# =============================================================================
# SECTION 8.8F — FINAL TRY SPECIES-LEVEL SLA AGGREGATION AND OMO CROSSWALK
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8F — FINAL TRY SPECIES-LEVEL SLA AGGREGATION AND OMO CROSSWALK")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "try_sla_final_retained",
    "try_final_species_summary",
    "resolved_crosswalk",
]

missing_objects = [
    obj
    for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Sections 8.8C and 8.8E2 first."
    )


# -----------------------------------------------------------------------------
# 2. Build final TRY species-level SLA table
# -----------------------------------------------------------------------------

try_species_final = (
    try_final_species_summary
    .copy()
    .rename(
        columns={
            "Final_Median_SLA": "TRY_Species_SLA",
        }
    )
)


try_species_final[
    "TRY_Species_SLA_Unit"
] = "mm2 mg-1"

try_species_final[
    "Trait_Code"
] = "SLA"

try_species_final[
    "Trait_Source"
] = "TRY"

try_species_final[
    "Provenance_Class"
] = "SPECIES_TRY"

try_species_final[
    "Aggregation_Method"
] = "MEDIAN_OF_FINAL_RETAINED_TRY_MEASUREMENTS"


# -----------------------------------------------------------------------------
# 3. Assign final TRY quality flag
# -----------------------------------------------------------------------------

quality_map = {
    "STRONG": "TRY_STRONG",
    "MODERATE": "TRY_MODERATE",
    "LIMITED": "TRY_LIMITED",
    "MODERATE_RESCUED": "TRY_MODERATE_RESCUED",
    "LIMITED_RESCUED": "TRY_LIMITED_RESCUED",
}

try_species_final[
    "TRY_Quality_Flag"
] = (
    try_species_final[
        "Final_Evidence_Strength"
    ]
    .map(quality_map)
    .fillna("TRY_REVIEW")
)


# -----------------------------------------------------------------------------
# 4. Select final species fields
# -----------------------------------------------------------------------------

try_species_final = (
    try_species_final[
        [
            "AccSpeciesID",
            "AccSpeciesName",
            "TRY_Species_SLA",
            "TRY_Species_SLA_Unit",

            "Final_Record_Count",
            "Final_Dataset_Count",
            "Final_Observation_Count",
            "Final_Trait_Definition_Count",
            "Rescued_Record_Count",

            "Final_Min_SLA",
            "Final_Mean_SLA",
            "Final_Max_SLA",

            "Final_Evidence_Strength",
            "TRY_Quality_Flag",

            "Trait_Code",
            "Trait_Source",
            "Provenance_Class",
            "Aggregation_Method",
        ]
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 5. Ensure TRY IDs have compatible type
# -----------------------------------------------------------------------------

try_species_final[
    "AccSpeciesID"
] = (
    pd.to_numeric(
        try_species_final[
            "AccSpeciesID"
        ],
        errors="coerce"
    )
    .astype("Int64")
)

resolved_crosswalk[
    "AccSpeciesID"
] = (
    pd.to_numeric(
        resolved_crosswalk[
            "AccSpeciesID"
        ],
        errors="coerce"
    )
    .astype("Int64")
)


# -----------------------------------------------------------------------------
# 6. Crosswalk final TRY estimates to Omo canonical taxa
# -----------------------------------------------------------------------------

omo_try_final = (
    resolved_crosswalk
    .merge(
        try_species_final,
        how="left",
        on="AccSpeciesID"
    )
)


# -----------------------------------------------------------------------------
# 7. Classify final TRY acquisition status
# -----------------------------------------------------------------------------

omo_try_final[
    "Final_TRY_SLA_Status"
] = np.where(
    omo_try_final[
        "TRY_Species_SLA"
    ].notna(),
    "FINAL_SPECIES_TRY_SLA_AVAILABLE",
    "NO_USABLE_FINAL_TRY_SLA"
)


# -----------------------------------------------------------------------------
# 8. Split covered and uncovered Omo records
# -----------------------------------------------------------------------------

omo_try_final_covered = (
    omo_try_final.loc[
        omo_try_final[
            "TRY_Species_SLA"
        ].notna()
    ]
    .copy()
    .reset_index(drop=True)
)


omo_try_final_uncovered = (
    omo_try_final.loc[
        omo_try_final[
            "TRY_Species_SLA"
        ].isna()
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 9. Identify taxa that had raw TRY measurements but no final usable TRY value
# -----------------------------------------------------------------------------

measured_try_ids = set(
    pd.to_numeric(
        try_sla_qc[
            "AccSpeciesID"
        ],
        errors="coerce"
    )
    .dropna()
    .astype(int)
)

final_try_ids = set(
    try_species_final[
        "AccSpeciesID"
    ]
    .dropna()
    .astype(int)
)

qc_failed_try_ids = (
    measured_try_ids
    - final_try_ids
)


omo_try_qc_failed = (
    omo_try_final.loc[
        omo_try_final[
            "AccSpeciesID"
        ]
        .astype("Int64")
        .isin(qc_failed_try_ids)
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 10. Summary counts
# -----------------------------------------------------------------------------

n_final_try_species = (
    try_species_final[
        "AccSpeciesID"
    ].nunique()
)

n_omo_try_covered = len(
    omo_try_final_covered
)

n_unique_omo_try_covered = (
    omo_try_final_covered[
        "Canonical_Species"
    ].nunique()
)

n_omo_try_uncovered = len(
    omo_try_final_uncovered
)

n_qc_failed_try_species = len(
    qc_failed_try_ids
)

n_omo_qc_failed = len(
    omo_try_qc_failed
)


# -----------------------------------------------------------------------------
# 11. Validation
# -----------------------------------------------------------------------------

assert n_final_try_species == 33, (
    f"Expected 33 accepted TRY species, "
    f"found {n_final_try_species}."
)

assert (
    try_species_final[
        "TRY_Species_SLA"
    ]
    .notna()
    .all()
)

assert (
    try_species_final[
        "TRY_Species_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)

assert (
    try_species_final[
        "Provenance_Class"
    ]
    .eq("SPECIES_TRY")
    .all()
)

assert (
    n_omo_try_covered
    + n_omo_try_uncovered
    == 114
)

assert n_qc_failed_try_species == 3, (
    f"Expected 3 TRY species to fail final QC, "
    f"found {n_qc_failed_try_species}."
)


# Based on the controlled taxonomic crosswalk:
# Carica papaya = 1 Omo record
# Sterculia oblonga = 1 Omo record
# Terminalia superba = 2 Omo records

assert n_omo_qc_failed == 4, (
    f"Expected 4 Omo query rows associated with "
    f"QC-failed TRY species, found {n_omo_qc_failed}."
)


# -----------------------------------------------------------------------------
# 12. Build Omo-ready TRY SLA table
# -----------------------------------------------------------------------------

omo_try_sla_final = (
    omo_try_final_covered[
        [
            "SLA_Query_ID",
            "Canonical_Species",
            "SLA_Query_Name",

            "Resolved_TRY_Name",
            "AccSpeciesID",
            "Resolution_Method",

            "TRY_Species_SLA",
            "TRY_Species_SLA_Unit",

            "Final_Record_Count",
            "Final_Dataset_Count",
            "Final_Observation_Count",
            "Final_Trait_Definition_Count",
            "Rescued_Record_Count",

            "Final_Evidence_Strength",
            "TRY_Quality_Flag",

            "Trait_Source",
            "Provenance_Class",
            "Aggregation_Method",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


omo_try_sla_final = (
    omo_try_sla_final
    .rename(
        columns={
            "TRY_Species_SLA":
                "Standardized_SLA_Value",

            "TRY_Species_SLA_Unit":
                "Standardized_SLA_Unit",
        }
    )
)


# -----------------------------------------------------------------------------
# 13. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

try_species_path = (
    output_dir
    / "omo_sla_try_final_species_estimates.csv"
)

omo_try_path = (
    output_dir
    / "omo_sla_final_species_try_values.csv"
)

uncovered_path = (
    output_dir
    / "omo_sla_resolved_taxa_without_final_try_value.csv"
)

qc_failed_path = (
    output_dir
    / "omo_sla_try_qc_failed_omo_taxa.csv"
)


try_species_final.to_csv(
    try_species_path,
    index=False
)

omo_try_sla_final.to_csv(
    omo_try_path,
    index=False
)

omo_try_final_uncovered.to_csv(
    uncovered_path,
    index=False
)

omo_try_qc_failed.to_csv(
    qc_failed_path,
    index=False
)


# -----------------------------------------------------------------------------
# 14. Diagnostics
# -----------------------------------------------------------------------------

print("\nFinal TRY species-level SLA estimates")
print("-" * 100)

print(
    f"Accepted TRY species estimates : "
    f"{n_final_try_species}"
)

display(
    try_species_final[
        [
            "AccSpeciesName",
            "TRY_Species_SLA",
            "Final_Record_Count",
            "Final_Dataset_Count",
            "Rescued_Record_Count",
            "Final_Evidence_Strength",
            "TRY_Quality_Flag",
        ]
    ]
    .sort_values(
        "AccSpeciesName"
    )
)


print("\nFinal Omo species-level TRY coverage")
print("-" * 100)

print(
    f"Omo query rows with final TRY SLA : "
    f"{n_omo_try_covered}"
)

print(
    f"Unique Omo taxa with final TRY SLA: "
    f"{n_unique_omo_try_covered}"
)

print(
    f"Resolved Omo rows without final TRY SLA : "
    f"{n_omo_try_uncovered}"
)


print("\nOmo-ready final TRY SLA table")
print("-" * 100)

display(
    omo_try_sla_final[
        [
            "Canonical_Species",
            "Resolved_TRY_Name",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "Final_Record_Count",
            "Final_Dataset_Count",
            "Final_Evidence_Strength",
            "TRY_Quality_Flag",
            "Provenance_Class",
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
)


print("\nTRY species failing final measurement QC")
print("-" * 100)

print(
    f"TRY species failing QC : "
    f"{n_qc_failed_try_species}"
)

print(
    f"Affected Omo query rows: "
    f"{n_omo_qc_failed}"
)

if omo_try_qc_failed.empty:

    print(
        "No Omo taxa were associated with "
        "QC-failed TRY species."
    )

else:

    display(
        omo_try_qc_failed[
            [
                "SLA_Query_ID",
                "Canonical_Species",
                "Resolved_TRY_Name",
                "AccSpeciesID",
                "Resolution_Method",
                "Final_TRY_SLA_Status",
            ]
        ]
        .sort_values(
            "Canonical_Species"
        )
    )


print("\nTRY provenance distribution")
print("-" * 100)

print(
    omo_try_sla_final[
        "TRY_Quality_Flag"
    ]
    .value_counts()
    .to_string()
)


print("\nExports")
print("-" * 100)

print(
    f"TRY species estimates : "
    f"{try_species_path}"
)

print(
    f"Omo TRY SLA values    : "
    f"{omo_try_path}"
)

print(
    f"Without final TRY     : "
    f"{uncovered_path}"
)

print(
    f"TRY QC-failed taxa    : "
    f"{qc_failed_path}"
)

print(
    "\nSection 8.8F completed successfully."
)
```

### 8.8G Integrated SLA Coverage Reassessment After TRY Acquisition

Completion of the TRY species-level acquisition substantially changes the functional-trait coverage state established during the earlier supplementary and fallback stages.

The final TRY quality-control workflow produced accepted species-level SLA estimates for 33 TRY taxa, corresponding to 46 Omo Forest canonical taxa after controlled taxonomic crosswalking. These species-level TRY values have the highest acquisition priority and therefore supersede any lower-level estimate that might otherwise be available for the same canonical taxon.

The current SLA evidence hierarchy is applied in the following order:

1. `SPECIES_TRY`
2. `SPECIES_LITERATURE`
3. `GENUS_ESTIMATE`
4. unresolved / further fallback required

The literature-derived *Citrus sinensis* estimate and previously accepted BIEN genus-level estimates remain available only where no valid species-level TRY value exists.

This subsection reconstructs SLA coverage from the complete 127-taxon acquisition foundation, assigns the highest-priority currently available value to each canonical taxon, and identifies the true residual gaps that remain after integration of TRY, literature, and genus-level evidence.

No family-level or global fallback values are introduced in this subsection.

```python
# =============================================================================
# SECTION 8.8G — INTEGRATED SLA COVERAGE REASSESSMENT AFTER TRY ACQUISITION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8G — INTEGRATED SLA COVERAGE REASSESSMENT AFTER TRY ACQUISITION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Define current and legacy output directories
# -----------------------------------------------------------------------------

current_output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

legacy_output_dir = (
    PROJECT_ROOT
    / "notebooks"
    / "outputs"
    / "functional_traits"
)

current_output_dir.mkdir(
    parents=True,
    exist_ok=True
)

print("\nOutput directories")
print("-" * 100)

print(
    f"Current output directory : "
    f"{current_output_dir}"
)

print(
    f"Legacy output directory  : "
    f"{legacy_output_dir}"
)


# Use current directory for all NEW exports from this point onward.
output_dir = current_output_dir


# -----------------------------------------------------------------------------
# 2. Locate complete 127-taxon SLA acquisition foundation
# -----------------------------------------------------------------------------

candidate_acquisition_paths = [
    current_output_dir
    / "omo_sla_acquisition_targets.csv",

    legacy_output_dir
    / "omo_sla_acquisition_targets.csv",
]


acquisition_targets_path = None

for candidate_path in candidate_acquisition_paths:

    if candidate_path.exists():

        acquisition_targets_path = (
            candidate_path
        )

        break


if acquisition_targets_path is None:

    raise FileNotFoundError(
        "Could not locate omo_sla_acquisition_targets.csv "
        "in either the current or legacy output directory.\n\n"
        "Checked:\n"
        + "\n".join(
            str(path)
            for path in candidate_acquisition_paths
        )
    )


print(
    f"\nSLA acquisition foundation located at:\n"
    f"{acquisition_targets_path}"
)


sla_all_targets = pd.read_csv(
    acquisition_targets_path
)


# -----------------------------------------------------------------------------
# 3. Validate complete canonical foundation
# -----------------------------------------------------------------------------

assert len(sla_all_targets) == 127, (
    f"Expected 127 SLA acquisition targets, "
    f"found {len(sla_all_targets)}."
)

assert (
    sla_all_targets["Canonical_Species"].nunique()
    == 127
), "Canonical SLA target table contains duplicate taxa."


# -----------------------------------------------------------------------------
# 4. Load final species-level TRY values
# -----------------------------------------------------------------------------

try_final_path = (
    output_dir
    / "omo_sla_final_species_try_values.csv"
)

if not try_final_path.exists():
    raise FileNotFoundError(
        "Final species-level TRY SLA table not found. "
        "Run Section 8.8F first."
    )

sla_try_final_values = pd.read_csv(
    try_final_path
)

assert (
    sla_try_final_values[
        "Canonical_Species"
    ].nunique()
    == len(sla_try_final_values)
)


# -----------------------------------------------------------------------------
# 5. Load accepted species-level literature values
# -----------------------------------------------------------------------------

literature_path = (
    output_dir
    / "omo_sla_residual_species_literature_accepted.csv"
)

if literature_path.exists():

    sla_literature_values = pd.read_csv(
        literature_path
    )

else:

    sla_literature_values = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Accepted_SLA_Value",
            "Accepted_SLA_Unit",
            "Accepted_Provenance_Class",
            "Accepted_Source_Class",
            "Accepted_Quality_Flag",
        ]
    )


# -----------------------------------------------------------------------------
# 6. Load accepted BIEN genus-level estimates attached to Omo targets
# -----------------------------------------------------------------------------

genus_status_path = (
    output_dir
    / "omo_sla_genus_target_acquisition_status.csv"
)

if genus_status_path.exists():

    sla_genus_status = pd.read_csv(
        genus_status_path
    )

    sla_genus_values = (
        sla_genus_status.loc[
            sla_genus_status[
                "Genus_Acquisition_Status"
            ].eq("GENUS_ESTIMATE_ACQUIRED")
        ]
        .copy()
    )

else:

    sla_genus_values = pd.DataFrame()


# -----------------------------------------------------------------------------
# 7. Prepare compact TRY evidence table
# -----------------------------------------------------------------------------

try_compact = (
    sla_try_final_values[
        [
            "Canonical_Species",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "Resolved_TRY_Name",
            "AccSpeciesID",
            "Final_Record_Count",
            "Final_Dataset_Count",
            "Final_Evidence_Strength",
            "TRY_Quality_Flag",
            "Provenance_Class",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Standardized_SLA_Value":
                "TRY_SLA_Value",

            "Standardized_SLA_Unit":
                "TRY_SLA_Unit",

            "Provenance_Class":
                "TRY_Provenance_Class",
        }
    )
)


# -----------------------------------------------------------------------------
# 8. Prepare compact literature evidence table
# -----------------------------------------------------------------------------

if not sla_literature_values.empty:

    literature_compact = (
        sla_literature_values[
            [
                "Canonical_Species",
                "Accepted_SLA_Value",
                "Accepted_SLA_Unit",
                "Accepted_Provenance_Class",
                "Accepted_Source_Class",
                "Accepted_Quality_Flag",
            ]
        ]
        .copy()
        .rename(
            columns={
                "Accepted_SLA_Value":
                    "Literature_SLA_Value",

                "Accepted_SLA_Unit":
                    "Literature_SLA_Unit",

                "Accepted_Provenance_Class":
                    "Literature_Provenance_Class",

                "Accepted_Source_Class":
                    "Literature_Source_Class",

                "Accepted_Quality_Flag":
                    "Literature_Quality_Flag",
            }
        )
    )

else:

    literature_compact = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Literature_SLA_Value",
            "Literature_SLA_Unit",
            "Literature_Provenance_Class",
            "Literature_Source_Class",
            "Literature_Quality_Flag",
        ]
    )


# -----------------------------------------------------------------------------
# 9. Prepare compact genus-level evidence table
# -----------------------------------------------------------------------------

if not sla_genus_values.empty:

    genus_compact = (
        sla_genus_values[
            [
                "Canonical_Species",
                "Genus_Search_Name",
                "Accepted_Genus_SLA",
                "Accepted_Genus_SLA_Unit",
                "Retained_Record_Count",
                "Retained_Species_Count",
                "Evidence_Strength",
                "Estimation_Method",
                "Trait_Source",
            ]
        ]
        .copy()
        .rename(
            columns={
                "Accepted_Genus_SLA":
                    "Genus_SLA_Value",

                "Accepted_Genus_SLA_Unit":
                    "Genus_SLA_Unit",

                "Evidence_Strength":
                    "Genus_Evidence_Strength",

                "Trait_Source":
                    "Genus_Trait_Source",
            }
        )
    )

else:

    genus_compact = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Genus_Search_Name",
            "Genus_SLA_Value",
            "Genus_SLA_Unit",
            "Retained_Record_Count",
            "Retained_Species_Count",
            "Genus_Evidence_Strength",
            "Estimation_Method",
            "Genus_Trait_Source",
        ]
    )


# -----------------------------------------------------------------------------
# 10. Merge all current evidence streams onto the 127-taxon foundation
# -----------------------------------------------------------------------------

sla_integrated = (
    sla_all_targets
    .merge(
        try_compact,
        how="left",
        on="Canonical_Species"
    )
    .merge(
        literature_compact,
        how="left",
        on="Canonical_Species"
    )
    .merge(
        genus_compact,
        how="left",
        on="Canonical_Species"
    )
)


assert len(sla_integrated) == 127


# -----------------------------------------------------------------------------
# 11. Apply source hierarchy
#
# Priority:
#   1. Species-level TRY
#   2. Species-level literature
#   3. Genus estimate
# -----------------------------------------------------------------------------

sla_integrated[
    "Current_SLA_Value"
] = np.nan

sla_integrated[
    "Current_SLA_Unit"
] = pd.NA

sla_integrated[
    "Current_Provenance_Class"
] = pd.NA

sla_integrated[
    "Current_Source"
] = pd.NA

sla_integrated[
    "Current_Quality_Flag"
] = pd.NA


# -----------------------------------------------------------------------------
# 12. Priority 3 — genus estimate
# -----------------------------------------------------------------------------

genus_mask = (
    sla_integrated[
        "Genus_SLA_Value"
    ].notna()
)

sla_integrated.loc[
    genus_mask,
    "Current_SLA_Value"
] = (
    sla_integrated.loc[
        genus_mask,
        "Genus_SLA_Value"
    ]
)

sla_integrated.loc[
    genus_mask,
    "Current_SLA_Unit"
] = (
    sla_integrated.loc[
        genus_mask,
        "Genus_SLA_Unit"
    ]
)

sla_integrated.loc[
    genus_mask,
    "Current_Provenance_Class"
] = "GENUS_ESTIMATE"

sla_integrated.loc[
    genus_mask,
    "Current_Source"
] = "BIEN"

sla_integrated.loc[
    genus_mask,
    "Current_Quality_Flag"
] = (
    "GENUS_"
    + sla_integrated.loc[
        genus_mask,
        "Genus_Evidence_Strength"
    ].fillna("UNKNOWN")
)


# -----------------------------------------------------------------------------
# 13. Priority 2 — species literature overrides genus estimate
# -----------------------------------------------------------------------------

literature_mask = (
    sla_integrated[
        "Literature_SLA_Value"
    ].notna()
)

sla_integrated.loc[
    literature_mask,
    "Current_SLA_Value"
] = (
    sla_integrated.loc[
        literature_mask,
        "Literature_SLA_Value"
    ]
)

sla_integrated.loc[
    literature_mask,
    "Current_SLA_Unit"
] = (
    sla_integrated.loc[
        literature_mask,
        "Literature_SLA_Unit"
    ]
)

sla_integrated.loc[
    literature_mask,
    "Current_Provenance_Class"
] = (
    sla_integrated.loc[
        literature_mask,
        "Literature_Provenance_Class"
    ]
)

sla_integrated.loc[
    literature_mask,
    "Current_Source"
] = (
    sla_integrated.loc[
        literature_mask,
        "Literature_Source_Class"
    ]
)

sla_integrated.loc[
    literature_mask,
    "Current_Quality_Flag"
] = (
    sla_integrated.loc[
        literature_mask,
        "Literature_Quality_Flag"
    ]
)


# -----------------------------------------------------------------------------
# 14. Priority 1 — species TRY overrides all lower-level evidence
# -----------------------------------------------------------------------------

try_mask = (
    sla_integrated[
        "TRY_SLA_Value"
    ].notna()
)

sla_integrated.loc[
    try_mask,
    "Current_SLA_Value"
] = (
    sla_integrated.loc[
        try_mask,
        "TRY_SLA_Value"
    ]
)

sla_integrated.loc[
    try_mask,
    "Current_SLA_Unit"
] = (
    sla_integrated.loc[
        try_mask,
        "TRY_SLA_Unit"
    ]
)

sla_integrated.loc[
    try_mask,
    "Current_Provenance_Class"
] = "SPECIES_TRY"

sla_integrated.loc[
    try_mask,
    "Current_Source"
] = "TRY"

sla_integrated.loc[
    try_mask,
    "Current_Quality_Flag"
] = (
    sla_integrated.loc[
        try_mask,
        "TRY_Quality_Flag"
    ]
)


# -----------------------------------------------------------------------------
# 15. Classify current acquisition state
# -----------------------------------------------------------------------------

sla_integrated[
    "Current_Acquisition_Status"
] = np.where(
    sla_integrated[
        "Current_SLA_Value"
    ].notna(),
    "SLA_VALUE_AVAILABLE",
    "SLA_VALUE_STILL_MISSING"
)


# -----------------------------------------------------------------------------
# 16. Validate standardized units of currently available values
# -----------------------------------------------------------------------------

available_mask = (
    sla_integrated[
        "Current_SLA_Value"
    ].notna()
)

assert (
    sla_integrated.loc[
        available_mask,
        "Current_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
), (
    "One or more currently available SLA values "
    "are not in mm2 mg-1."
)


# -----------------------------------------------------------------------------
# 17. Build residual gap register
# -----------------------------------------------------------------------------

sla_remaining_gaps = (
    sla_integrated.loc[
        sla_integrated[
            "Current_Acquisition_Status"
        ].eq("SLA_VALUE_STILL_MISSING")
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 18. Add next acquisition strategy
# -----------------------------------------------------------------------------

sla_remaining_gaps[
    "Next_Acquisition_Strategy"
] = "GENUS_LEVEL_SEARCH"

# Genus-level/unresolved canonical taxa also proceed through controlled
# genus-level evidence acquisition.

if "Taxonomic_Resolution" in sla_remaining_gaps.columns:

    genus_level_mask = (
        sla_remaining_gaps[
            "Taxonomic_Resolution"
        ].eq("GENUS_LEVEL_OR_UNRESOLVED")
    )

    sla_remaining_gaps.loc[
        genus_level_mask,
        "Next_Acquisition_Strategy"
    ] = "CONTROLLED_GENUS_LEVEL_SEARCH"


# -----------------------------------------------------------------------------
# 19. Summary counts
# -----------------------------------------------------------------------------

n_total = len(
    sla_integrated
)

n_try = (
    sla_integrated[
        "Current_Provenance_Class"
    ]
    .eq("SPECIES_TRY")
    .sum()
)

n_literature = (
    sla_integrated[
        "Current_Provenance_Class"
    ]
    .eq("SPECIES_LITERATURE")
    .sum()
)

n_genus = (
    sla_integrated[
        "Current_Provenance_Class"
    ]
    .eq("GENUS_ESTIMATE")
    .sum()
)

n_available = (
    sla_integrated[
        "Current_SLA_Value"
    ]
    .notna()
    .sum()
)

n_missing = (
    sla_integrated[
        "Current_SLA_Value"
    ]
    .isna()
    .sum()
)


# -----------------------------------------------------------------------------
# 20. Validation
# -----------------------------------------------------------------------------

assert n_total == 127

assert (
    n_available
    + n_missing
    == 127
)

assert (
    sla_integrated[
        "Canonical_Species"
    ].nunique()
    == 127
)


# -----------------------------------------------------------------------------
# 21. Export
# -----------------------------------------------------------------------------

integrated_path = (
    output_dir
    / "omo_sla_current_integrated_coverage.csv"
)

remaining_path = (
    output_dir
    / "omo_sla_remaining_gaps_after_try.csv"
)

provenance_summary_path = (
    output_dir
    / "omo_sla_current_provenance_summary.csv"
)


sla_integrated.to_csv(
    integrated_path,
    index=False
)

sla_remaining_gaps.to_csv(
    remaining_path,
    index=False
)


provenance_summary = (
    sla_integrated[
        "Current_Provenance_Class"
    ]
    .fillna("MISSING")
    .value_counts()
    .rename_axis(
        "Provenance_Class"
    )
    .reset_index(
        name="Taxon_Count"
    )
)

provenance_summary[
    "Percent_of_127"
] = (
    100
    * provenance_summary[
        "Taxon_Count"
    ]
    / 127
)

provenance_summary.to_csv(
    provenance_summary_path,
    index=False
)


# -----------------------------------------------------------------------------
# 22. Diagnostics
# -----------------------------------------------------------------------------

print("\nCurrent integrated SLA coverage")
print("-" * 100)

print(
    f"Canonical taxa                  : "
    f"{n_total}"
)

print(
    f"Species-level TRY values        : "
    f"{n_try}"
)

print(
    f"Species-level literature values : "
    f"{n_literature}"
)

print(
    f"Genus-level estimates           : "
    f"{n_genus}"
)

print(
    f"Current SLA values available    : "
    f"{n_available}"
)

print(
    f"Remaining SLA gaps              : "
    f"{n_missing}"
)


print("\nCurrent provenance distribution")
print("-" * 100)

display(
    provenance_summary
)


print("\nRemaining SLA acquisition gaps")
print("-" * 100)

display(
    sla_remaining_gaps[
        [
            col
            for col in [
                "Canonical_Species",
                "Genus",
                "Taxonomic_Resolution",
                "SLA_Query_Name",
                "Next_Acquisition_Strategy",
            ]
            if col in sla_remaining_gaps.columns
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Integrated coverage : "
    f"{integrated_path}"
)

print(
    f"Remaining gaps      : "
    f"{remaining_path}"
)

print(
    f"Provenance summary  : "
    f"{provenance_summary_path}"
)

print(
    "\nSection 8.8G completed successfully."
)
```

### 8.8H Residual Taxonomic Normalization and TRY-Congeneric Opportunity Assessment

Following integration of species-level TRY data, literature evidence, and previously acquired BIEN genus estimates, 76 canonical Omo Forest taxa remain without SLA values.

Before initiating additional external acquisition, the residual taxa are normalized to defensible acquisition genera. Historical spelling variants and orthographic forms in the canonical Omo taxonomy are not used directly as external search keys where an accepted TRY taxonomic resolution is already available.

For taxa previously resolved against TRY, the acquisition genus is derived from the accepted TRY species name. For unresolved or genus-level taxa, the controlled supplementary search names established earlier are used where available. The canonical Omo taxon remains unchanged throughout this process.

The normalized residual genera are then compared against the 33 TRY species for which final species-level SLA estimates were accepted. Where one or more accepted TRY species belong to the same genus as a residual target, those measurements constitute higher-priority congeneric evidence that can potentially support a `GENUS_ESTIMATE` without requiring a new external database query.

No new genus estimates are assigned in this subsection. The purpose is to establish the corrected residual genus manifest and identify which missing taxa can potentially be addressed using already acquired TRY evidence.

```python
# =============================================================================
# SECTION 8.8H — RESIDUAL TAXONOMIC NORMALIZATION AND TRY-CONGENERIC
#                OPPORTUNITY ASSESSMENT
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8H — RESIDUAL TAXONOMIC NORMALIZATION AND TRY-CONGENERIC OPPORTUNITY ASSESSMENT")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "sla_remaining_gaps",
    "resolved_crosswalk",
    "try_species_final",
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Sections 8.8F and 8.8G first."
    )


output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# 2. Start from the 76 residual taxa
# -----------------------------------------------------------------------------

sla_residual_normalized = (
    sla_remaining_gaps
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 3. Recover authoritative TRY taxonomic resolution
#
# Important:
# sla_remaining_gaps may already contain Resolved_TRY_Name fields originating
# from the final TRY-value merge. For taxa without final SLA values those
# fields are normally empty. Therefore we explicitly remove those incomplete
# fields and restore the authoritative taxonomic resolution from the verified
# Omo → TRY crosswalk.
# -----------------------------------------------------------------------------

try_resolution_columns = [
    "Resolved_TRY_Name",
    "AccSpeciesID",
    "Resolution_Method",
]

existing_resolution_columns = [
    col
    for col in try_resolution_columns
    if col in sla_residual_normalized.columns
]

if existing_resolution_columns:

    sla_residual_normalized = (
        sla_residual_normalized
        .drop(
            columns=existing_resolution_columns
        )
    )


resolved_name_lookup = (
    resolved_crosswalk[
        [
            "Canonical_Species",
            "Resolved_TRY_Name",
            "AccSpeciesID",
            "Resolution_Method",
        ]
    ]
    .drop_duplicates(
        subset="Canonical_Species"
    )
    .copy()
)


sla_residual_normalized = (
    sla_residual_normalized
    .merge(
        resolved_name_lookup,
        how="left",
        on="Canonical_Species",
        validate="one_to_one"
    )
)


# -----------------------------------------------------------------------------
# 4. Validate restored TRY-resolution structure
# -----------------------------------------------------------------------------

n_with_try_resolution = (
    sla_residual_normalized[
        "Resolved_TRY_Name"
    ]
    .notna()
    .sum()
)

n_without_try_resolution = (
    sla_residual_normalized[
        "Resolved_TRY_Name"
    ]
    .isna()
    .sum()
)


print("\nResidual TRY-resolution fields")
print("-" * 100)

print(
    f"Residual taxa with accepted TRY names    : "
    f"{n_with_try_resolution}"
)

print(
    f"Residual taxa without accepted TRY names : "
    f"{n_without_try_resolution}"
)


# There were 68 resolved Omo query rows without a final usable TRY SLA value
# following Section 8.8F.

assert n_with_try_resolution == 68, (
    f"Expected 68 residual taxa with verified TRY taxonomic "
    f"resolution, found {n_with_try_resolution}."
)

assert n_without_try_resolution == 8, (
    f"Expected 8 residual taxa without verified TRY taxonomic "
    f"resolution, found {n_without_try_resolution}."
)


# -----------------------------------------------------------------------------
# 5. Define controlled search-name overrides for taxa not resolved through TRY
# -----------------------------------------------------------------------------

controlled_search_overrides = {
    "Amphimas tetracoides": "Amphimas tetracoides",
    "Aningeria robusta": "Aningeria pierrei",
    "Homalium alymeri": "Homalium alymeri",
    "Melacantha alnifolia": "Malacantha alnifolia",
    "Mitragyna cilliata": "Mitragyna ciliata",
    "Musa sapiensis": "Musa sapiensis",
    "Cola species": "Cola",
    "Drypetes species": "Drypetes",
}


# -----------------------------------------------------------------------------
# 6. Establish normalized acquisition search name
#
# Priority:
#   1. Accepted TRY name
#   2. Controlled supplementary search name
#   3. Original SLA query name
# -----------------------------------------------------------------------------

sla_residual_normalized[
    "Normalized_Search_Name"
] = (
    sla_residual_normalized[
        "Resolved_TRY_Name"
    ]
)


for canonical_name, search_name in controlled_search_overrides.items():

    mask = (
        sla_residual_normalized[
            "Canonical_Species"
        ].eq(canonical_name)
        &
        sla_residual_normalized[
            "Normalized_Search_Name"
        ].isna()
    )

    sla_residual_normalized.loc[
        mask,
        "Normalized_Search_Name"
    ] = search_name


remaining_name_mask = (
    sla_residual_normalized[
        "Normalized_Search_Name"
    ].isna()
)

sla_residual_normalized.loc[
    remaining_name_mask,
    "Normalized_Search_Name"
] = (
    sla_residual_normalized.loc[
        remaining_name_mask,
        "SLA_Query_Name"
    ]
)


# -----------------------------------------------------------------------------
# 7. Derive normalized acquisition genus
# -----------------------------------------------------------------------------

def extract_acquisition_genus(name):

    if pd.isna(name):
        return pd.NA

    text = str(name).strip()

    if not text:
        return pd.NA

    return text.split()[0]


sla_residual_normalized[
    "Normalized_Acquisition_Genus"
] = (
    sla_residual_normalized[
        "Normalized_Search_Name"
    ]
    .map(extract_acquisition_genus)
)


# -----------------------------------------------------------------------------
# 8. Record normalization source
# -----------------------------------------------------------------------------

sla_residual_normalized[
    "Genus_Normalization_Source"
] = "ORIGINAL_QUERY_NAME"


try_resolved_mask = (
    sla_residual_normalized[
        "Resolved_TRY_Name"
    ].notna()
)

sla_residual_normalized.loc[
    try_resolved_mask,
    "Genus_Normalization_Source"
] = "TRY_ACCEPTED_NAME"


for canonical_name in controlled_search_overrides:

    override_mask = (
        sla_residual_normalized[
            "Canonical_Species"
        ].eq(canonical_name)
        &
        sla_residual_normalized[
            "Resolved_TRY_Name"
        ].isna()
    )

    sla_residual_normalized.loc[
        override_mask,
        "Genus_Normalization_Source"
    ] = "CONTROLLED_SUPPLEMENTARY_NAME"


# -----------------------------------------------------------------------------
# 9. Validate all residual taxa now have a normalized genus
# -----------------------------------------------------------------------------

assert len(
    sla_residual_normalized
) == 76

assert (
    sla_residual_normalized[
        "Normalized_Acquisition_Genus"
    ]
    .notna()
    .all()
)


# -----------------------------------------------------------------------------
# 10. Derive genus from final accepted TRY species
# -----------------------------------------------------------------------------

try_species_for_genus = (
    try_species_final[
        [
            "AccSpeciesID",
            "AccSpeciesName",
            "TRY_Species_SLA",
            "Final_Record_Count",
            "Final_Dataset_Count",
            "Final_Evidence_Strength",
            "TRY_Quality_Flag",
        ]
    ]
    .copy()
)


try_species_for_genus[
    "TRY_Genus"
] = (
    try_species_for_genus[
        "AccSpeciesName"
    ]
    .map(extract_acquisition_genus)
)


# -----------------------------------------------------------------------------
# 11. Summarize existing TRY species evidence by genus
# -----------------------------------------------------------------------------

try_genus_available = (
    try_species_for_genus
    .groupby(
        "TRY_Genus",
        as_index=False
    )
    .agg(
        TRY_Congeneric_Species_Count=(
            "AccSpeciesName",
            "nunique"
        ),

        TRY_Congeneric_Species=(
            "AccSpeciesName",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        TRY_Congeneric_Species_Median_SLA=(
            "TRY_Species_SLA",
            "median"
        ),

        TRY_Congeneric_Min_SLA=(
            "TRY_Species_SLA",
            "min"
        ),

        TRY_Congeneric_Max_SLA=(
            "TRY_Species_SLA",
            "max"
        ),
    )
    .rename(
        columns={
            "TRY_Genus":
                "Normalized_Acquisition_Genus"
        }
    )
)


# -----------------------------------------------------------------------------
# 12. Attach congeneric TRY opportunities to residual taxa
# -----------------------------------------------------------------------------

sla_residual_normalized = (
    sla_residual_normalized
    .merge(
        try_genus_available,
        how="left",
        on="Normalized_Acquisition_Genus"
    )
)


sla_residual_normalized[
    "TRY_Congeneric_Available"
] = (
    sla_residual_normalized[
        "TRY_Congeneric_Species_Count"
    ]
    .fillna(0)
    .gt(0)
)


count_col = (
    "TRY_Congeneric_Species_Count"
)

sla_residual_normalized[
    count_col
] = (
    sla_residual_normalized[
        count_col
    ]
    .fillna(0)
    .astype(int)
)


# -----------------------------------------------------------------------------
# 13. Define next acquisition pathway
# -----------------------------------------------------------------------------

sla_residual_normalized[
    "Next_Residual_Pathway"
] = np.where(
    sla_residual_normalized[
        "TRY_Congeneric_Available"
    ],
    "USE_EXISTING_TRY_CONGENERIC_EVIDENCE",
    "QUERY_SUPPLEMENTARY_GENUS_SOURCE"
)


# -----------------------------------------------------------------------------
# 14. Create unique residual genus manifest
# -----------------------------------------------------------------------------

residual_genus_manifest = (
    sla_residual_normalized
    .groupby(
        "Normalized_Acquisition_Genus",
        as_index=False
    )
    .agg(
        Residual_Taxon_Count=(
            "Canonical_Species",
            "count"
        ),

        Residual_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        TRY_Congeneric_Available=(
            "TRY_Congeneric_Available",
            "max"
        ),

        TRY_Congeneric_Species_Count=(
            "TRY_Congeneric_Species_Count",
            "max"
        ),

        TRY_Congeneric_Species=(
            "TRY_Congeneric_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        TRY_Congeneric_Species_Median_SLA=(
            "TRY_Congeneric_Species_Median_SLA",
            "first"
        ),
    )
    .sort_values(
        "Normalized_Acquisition_Genus"
    )
    .reset_index(drop=True)
)


residual_genus_manifest.insert(
    0,
    "Residual_Genus_ID",
    [
        f"OMO_SLA_RESID_GENUS_{i:03d}"
        for i in range(
            1,
            len(residual_genus_manifest) + 1
        )
    ]
)


residual_genus_manifest[
    "Acquisition_Pathway"
] = np.where(
    residual_genus_manifest[
        "TRY_Congeneric_Available"
    ],
    "TRY_CONGENERIC_REVIEW",
    "SUPPLEMENTARY_GENUS_QUERY"
)


# -----------------------------------------------------------------------------
# 15. Summary counts
# -----------------------------------------------------------------------------

n_residual_taxa = len(
    sla_residual_normalized
)

n_unique_residual_genera = (
    residual_genus_manifest[
        "Normalized_Acquisition_Genus"
    ].nunique()
)

n_try_congeneric_taxa = (
    sla_residual_normalized[
        "TRY_Congeneric_Available"
    ].sum()
)

n_external_genus_taxa = (
    n_residual_taxa
    - n_try_congeneric_taxa
)

n_try_congeneric_genera = (
    residual_genus_manifest[
        "TRY_Congeneric_Available"
    ].sum()
)

n_external_query_genera = (
    n_unique_residual_genera
    - n_try_congeneric_genera
)


# -----------------------------------------------------------------------------
# 16. Validation
# -----------------------------------------------------------------------------

assert n_residual_taxa == 76

assert (
    n_try_congeneric_taxa
    + n_external_genus_taxa
    == 76
)


# -----------------------------------------------------------------------------
# 17. Export
# -----------------------------------------------------------------------------

normalized_path = (
    output_dir
    / "omo_sla_residual_taxonomic_normalization.csv"
)

genus_manifest_path = (
    output_dir
    / "omo_sla_residual_unique_genus_manifest.csv"
)

try_congeneric_path = (
    output_dir
    / "omo_sla_residual_try_congeneric_opportunities.csv"
)

external_genus_path = (
    output_dir
    / "omo_sla_residual_external_genus_queries.csv"
)


sla_residual_normalized.to_csv(
    normalized_path,
    index=False
)

residual_genus_manifest.to_csv(
    genus_manifest_path,
    index=False
)

sla_residual_normalized.loc[
    sla_residual_normalized[
        "TRY_Congeneric_Available"
    ]
].to_csv(
    try_congeneric_path,
    index=False
)

residual_genus_manifest.loc[
    ~residual_genus_manifest[
        "TRY_Congeneric_Available"
    ]
].to_csv(
    external_genus_path,
    index=False
)


# -----------------------------------------------------------------------------
# 18. Diagnostics
# -----------------------------------------------------------------------------

print("\nResidual taxonomic normalization")
print("-" * 100)

print(
    f"Residual canonical taxa            : "
    f"{n_residual_taxa}"
)

print(
    f"Unique normalized residual genera  : "
    f"{n_unique_residual_genera}"
)

print(
    f"Taxa with existing TRY congeneric evidence : "
    f"{n_try_congeneric_taxa}"
)

print(
    f"Taxa requiring supplementary genus search : "
    f"{n_external_genus_taxa}"
)


print("\nNormalization-source distribution")
print("-" * 100)

print(
    sla_residual_normalized[
        "Genus_Normalization_Source"
    ]
    .value_counts()
    .to_string()
)


print("\nResidual taxa with existing TRY congeneric evidence")
print("-" * 100)

try_congeneric_display = (
    sla_residual_normalized.loc[
        sla_residual_normalized[
            "TRY_Congeneric_Available"
        ],
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "TRY_Congeneric_Species_Count",
            "TRY_Congeneric_Species",
            "TRY_Congeneric_Species_Median_SLA",
        ]
    ]
    .sort_values(
        [
            "Normalized_Acquisition_Genus",
            "Canonical_Species",
        ]
    )
)

if try_congeneric_display.empty:

    print(
        "No residual taxa have existing TRY congeneric evidence."
    )

else:

    display(
        try_congeneric_display
    )


print("\nUnique residual genus manifest")
print("-" * 100)

display(
    residual_genus_manifest[
        [
            "Residual_Genus_ID",
            "Normalized_Acquisition_Genus",
            "Residual_Taxon_Count",
            "Residual_Canonical_Taxa",
            "TRY_Congeneric_Available",
            "TRY_Congeneric_Species_Count",
            "TRY_Congeneric_Species",
            "Acquisition_Pathway",
        ]
    ]
)


print("\nGenus acquisition summary")
print("-" * 100)

print(
    f"Residual genera with TRY evidence : "
    f"{n_try_congeneric_genera}"
)

print(
    f"Residual genera needing external query : "
    f"{n_external_query_genera}"
)


print("\nExports")
print("-" * 100)

print(
    f"Normalized residual taxa : "
    f"{normalized_path}"
)

print(
    f"Residual genus manifest  : "
    f"{genus_manifest_path}"
)

print(
    f"TRY congeneric options   : "
    f"{try_congeneric_path}"
)

print(
    f"External genus queries   : "
    f"{external_genus_path}"
)

print(
    "\nSection 8.8H completed successfully."
)
```

### 8.8I Controlled TRY-Congeneric SLA Estimation

The residual-gap assessment identified four genera for which accepted TRY species-level SLA evidence is already available from a congeneric taxon: *Blighia*, *Diospyros*, *Homalium*, and *Sterculia*.

For each of these genera, the currently available TRY evidence consists of a single accepted species. Consequently, the resulting values are not treated as direct species-level observations or as strongly supported multi-species genus means. Instead, they are retained as controlled `GENUS_ESTIMATE` values derived from the best available congeneric evidence.

The source species and its accepted TRY SLA value remain explicitly recorded for provenance. The resulting estimates are assigned only to residual Omo taxa belonging to the same normalized acquisition genus and are not allowed to override any existing species-level TRY or literature value.

Because the genus evidence is based on one congeneric species per genus, all estimates are assigned a conservative limited-evidence quality classification.

```python
# =============================================================================
# SECTION 8.8I — CONTROLLED TRY-CONGENERIC SLA ESTIMATION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8I — CONTROLLED TRY-CONGENERIC SLA ESTIMATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "sla_residual_normalized",
    "try_species_final",
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Section 8.8H first."
    )


# -----------------------------------------------------------------------------
# 2. Restrict to residual taxa with TRY congeneric evidence
# -----------------------------------------------------------------------------

try_congeneric_targets = (
    sla_residual_normalized.loc[
        sla_residual_normalized[
            "TRY_Congeneric_Available"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(try_congeneric_targets) == 7, (
    f"Expected 7 residual taxa with TRY congeneric evidence, "
    f"found {len(try_congeneric_targets)}."
)


# -----------------------------------------------------------------------------
# 3. Derive genus for accepted TRY species
# -----------------------------------------------------------------------------

try_species_genus = (
    try_species_final.copy()
)

try_species_genus[
    "TRY_Genus"
] = (
    try_species_genus[
        "AccSpeciesName"
    ]
    .astype(str)
    .str.split()
    .str[0]
)


# -----------------------------------------------------------------------------
# 4. Build one-source-species genus evidence table
# -----------------------------------------------------------------------------

try_genus_evidence = (
    try_species_genus.loc[
        try_species_genus[
            "TRY_Genus"
        ].isin(
            try_congeneric_targets[
                "Normalized_Acquisition_Genus"
            ].unique()
        )
    ]
    [
        [
            "TRY_Genus",
            "AccSpeciesID",
            "AccSpeciesName",
            "TRY_Species_SLA",
            "TRY_Species_SLA_Unit",
            "Final_Record_Count",
            "Final_Dataset_Count",
            "Final_Evidence_Strength",
            "TRY_Quality_Flag",
        ]
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 5. Validate one accepted TRY species per target genus
# -----------------------------------------------------------------------------

species_per_genus = (
    try_genus_evidence
    .groupby("TRY_Genus")
    ["AccSpeciesName"]
    .nunique()
)

assert (
    species_per_genus.eq(1).all()
), (
    "One or more target genera contain multiple accepted TRY species. "
    "Aggregation logic should be revised before proceeding."
)


# -----------------------------------------------------------------------------
# 6. Construct controlled genus estimates
# -----------------------------------------------------------------------------

try_congeneric_estimates = (
    try_genus_evidence
    .rename(
        columns={
            "TRY_Genus":
                "Normalized_Acquisition_Genus",

            "AccSpeciesName":
                "Congeneric_Source_Species",

            "AccSpeciesID":
                "Congeneric_Source_TRY_ID",

            "TRY_Species_SLA":
                "Congeneric_Genus_SLA",

            "TRY_Species_SLA_Unit":
                "Congeneric_Genus_SLA_Unit",

            "Final_Record_Count":
                "Source_Record_Count",

            "Final_Dataset_Count":
                "Source_Dataset_Count",

            "Final_Evidence_Strength":
                "Source_Evidence_Strength",

            "TRY_Quality_Flag":
                "Source_TRY_Quality_Flag",
        }
    )
    .copy()
)


try_congeneric_estimates[
    "Congeneric_Provenance_Class"
] = "GENUS_ESTIMATE"

try_congeneric_estimates[
    "Congeneric_Estimation_Method"
] = "SINGLE_CONGENERIC_TRY_SPECIES_PROXY"

try_congeneric_estimates[
    "Congeneric_Evidence_Strength"
] = "LIMITED"

try_congeneric_estimates[
    "Congeneric_Quality_Flag"
] = "GENUS_SINGLE_TRY_SPECIES_PROXY"


# -----------------------------------------------------------------------------
# 7. Attach genus estimates to residual Omo taxa
# -----------------------------------------------------------------------------

omo_try_congeneric_values = (
    try_congeneric_targets
    .merge(
        try_congeneric_estimates,
        how="left",
        on="Normalized_Acquisition_Genus"
    )
)


# -----------------------------------------------------------------------------
# 8. Validate all 7 target taxa received values
# -----------------------------------------------------------------------------

assert (
    omo_try_congeneric_values[
        "Congeneric_Genus_SLA"
    ]
    .notna()
    .all()
)

assert (
    omo_try_congeneric_values[
        "Congeneric_Genus_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)


# -----------------------------------------------------------------------------
# 9. Build compact accepted genus table
# -----------------------------------------------------------------------------

omo_try_congeneric_accepted = (
    omo_try_congeneric_values[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",

            "Congeneric_Source_Species",
            "Congeneric_Source_TRY_ID",

            "Congeneric_Genus_SLA",
            "Congeneric_Genus_SLA_Unit",

            "Source_Record_Count",
            "Source_Dataset_Count",
            "Source_Evidence_Strength",
            "Source_TRY_Quality_Flag",

            "Congeneric_Provenance_Class",
            "Congeneric_Estimation_Method",
            "Congeneric_Evidence_Strength",
            "Congeneric_Quality_Flag",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Congeneric_Genus_SLA":
                "Standardized_SLA_Value",

            "Congeneric_Genus_SLA_Unit":
                "Standardized_SLA_Unit",

            "Congeneric_Provenance_Class":
                "Provenance_Class",

            "Congeneric_Quality_Flag":
                "Quality_Flag",
        }
    )
)


# -----------------------------------------------------------------------------
# 10. Summary
# -----------------------------------------------------------------------------

n_taxa = len(
    omo_try_congeneric_accepted
)

n_genera = (
    omo_try_congeneric_accepted[
        "Normalized_Acquisition_Genus"
    ].nunique()
)


# -----------------------------------------------------------------------------
# 11. Validation
# -----------------------------------------------------------------------------

assert n_taxa == 7
assert n_genera == 4

assert (
    omo_try_congeneric_accepted[
        "Provenance_Class"
    ]
    .eq("GENUS_ESTIMATE")
    .all()
)


# -----------------------------------------------------------------------------
# 12. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

genus_evidence_path = (
    output_dir
    / "omo_sla_try_congeneric_genus_evidence.csv"
)

accepted_path = (
    output_dir
    / "omo_sla_try_congeneric_genus_estimates_accepted.csv"
)


try_congeneric_estimates.to_csv(
    genus_evidence_path,
    index=False
)

omo_try_congeneric_accepted.to_csv(
    accepted_path,
    index=False
)


# -----------------------------------------------------------------------------
# 13. Diagnostics
# -----------------------------------------------------------------------------

print("\nAccepted TRY-congeneric SLA estimates")
print("-" * 100)

display(
    omo_try_congeneric_accepted[
        [
            "Canonical_Species",
            "Normalized_Acquisition_Genus",
            "Congeneric_Source_Species",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "Source_Record_Count",
            "Source_Dataset_Count",
            "Source_TRY_Quality_Flag",
            "Provenance_Class",
            "Quality_Flag",
        ]
    ]
    .sort_values(
        [
            "Normalized_Acquisition_Genus",
            "Canonical_Species",
        ]
    )
)


print("\nTRY-congeneric acquisition summary")
print("-" * 100)

print(
    f"Residual taxa assigned genus estimates : "
    f"{n_taxa}"
)

print(
    f"Genera represented                    : "
    f"{n_genera}"
)


print("\nExports")
print("-" * 100)

print(
    f"Genus evidence : "
    f"{genus_evidence_path}"
)

print(
    f"Accepted values: "
    f"{accepted_path}"
)

print(
    "\nSection 8.8I completed successfully."
)
```

### 8.8J BIEN Query Manifest for Remaining Residual Genera

After incorporation of accepted species-level TRY values, the literature-derived *Citrus* estimate, previously accepted BIEN genus estimates, and controlled TRY-congeneric estimates, 69 Omo Forest taxa remain without SLA values.

These taxa belong to 46 normalized acquisition genera that currently lack usable genus-level evidence within the assembled Omo Forest functional-trait workflow.

The next acquisition stage therefore prepares a single BIEN genus-query manifest for these 46 genera. Taxonomic normalization established in Section 8.8H is retained so that historical spelling variants and orthographic forms are not queried as separate genera.

The requested BIEN trait remains `leaf area per leaf dry mass`, with all returned measurements subsequently standardized to the project target unit of `mm2 mg-1`.

No BIEN records are queried or incorporated in this subsection. The purpose is to generate the definitive external genus acquisition manifest and genus-name list for use in R.

```python
# =============================================================================
# SECTION 8.8J — BIEN QUERY MANIFEST FOR REMAINING RESIDUAL GENERA
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8J — BIEN QUERY MANIFEST FOR REMAINING RESIDUAL GENERA")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "residual_genus_manifest",
    "omo_try_congeneric_accepted",
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Sections 8.8H and 8.8I first."
    )


# -----------------------------------------------------------------------------
# 2. Restrict to genera still requiring supplementary acquisition
# -----------------------------------------------------------------------------

bien_residual_genus_manifest = (
    residual_genus_manifest.loc[
        residual_genus_manifest[
            "Acquisition_Pathway"
        ].eq("SUPPLEMENTARY_GENUS_QUERY")
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 3. Validate expected genus count
# -----------------------------------------------------------------------------

n_bien_genera = len(
    bien_residual_genus_manifest
)

assert n_bien_genera == 46, (
    f"Expected 46 residual genera for BIEN query, "
    f"found {n_bien_genera}."
)

assert (
    bien_residual_genus_manifest[
        "Normalized_Acquisition_Genus"
    ].nunique()
    == 46
)


# -----------------------------------------------------------------------------
# 4. Add BIEN query governance
# -----------------------------------------------------------------------------

bien_residual_genus_manifest[
    "Requested_Trait"
] = "leaf area per leaf dry mass"

bien_residual_genus_manifest[
    "Target_Unit"
] = "mm2 mg-1"

bien_residual_genus_manifest[
    "Source"
] = "BIEN"

bien_residual_genus_manifest[
    "Query_Status"
] = "READY_FOR_BIEN_QUERY"

bien_residual_genus_manifest[
    "Planned_Provenance_Class"
] = "GENUS_ESTIMATE"


# -----------------------------------------------------------------------------
# 5. Add clean BIEN query IDs
# -----------------------------------------------------------------------------

bien_residual_genus_manifest[
    "BIEN_Genus_Query_ID"
] = [
    f"OMO_SLA_BIEN_RESID_{i:03d}"
    for i in range(
        1,
        len(bien_residual_genus_manifest) + 1
    )
]


# -----------------------------------------------------------------------------
# 6. Reorder columns
# -----------------------------------------------------------------------------

preferred_columns = [
    "BIEN_Genus_Query_ID",
    "Normalized_Acquisition_Genus",
    "Residual_Taxon_Count",
    "Residual_Canonical_Taxa",
    "Requested_Trait",
    "Target_Unit",
    "Source",
    "Query_Status",
    "Planned_Provenance_Class",
]

remaining_columns = [
    col
    for col in bien_residual_genus_manifest.columns
    if col not in preferred_columns
]

bien_residual_genus_manifest = (
    bien_residual_genus_manifest[
        preferred_columns
        + remaining_columns
    ]
)


# -----------------------------------------------------------------------------
# 7. Prepare genus-name vector
# -----------------------------------------------------------------------------

bien_genus_names = (
    bien_residual_genus_manifest[
        "Normalized_Acquisition_Genus"
    ]
    .dropna()
    .astype(str)
    .drop_duplicates()
    .sort_values()
    .tolist()
)


# -----------------------------------------------------------------------------
# 8. Validate taxonomic cleanliness
# -----------------------------------------------------------------------------

problematic_legacy_genera = {
    "Chrysophylum",
    "Cleistoformis",
    "Diallium",
    "Diospyrous",
    "Dracena",
    "Macranga",
    "Nuclea",
    "Raulvolfia",
    "Rauvolvia",
    "Rothmania",
    "Scotellia",
    "Termialia",
}

detected_problematic = (
    set(bien_genus_names)
    & problematic_legacy_genera
)

assert not detected_problematic, (
    "Legacy/unharmonized genera detected in BIEN manifest: "
    + ", ".join(
        sorted(detected_problematic)
    )
)


# -----------------------------------------------------------------------------
# 9. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

manifest_path = (
    output_dir
    / "omo_sla_bien_residual_genus_query_manifest.csv"
)

genus_names_path = (
    output_dir
    / "omo_sla_bien_residual_genus_names.txt"
)


bien_residual_genus_manifest.to_csv(
    manifest_path,
    index=False
)


with open(
    genus_names_path,
    "w",
    encoding="utf-8"
) as f:

    for genus in bien_genus_names:
        f.write(f"{genus}\n")


# -----------------------------------------------------------------------------
# 10. Diagnostics
# -----------------------------------------------------------------------------

print("\nResidual BIEN genus query manifest")
print("-" * 100)

print(
    f"Residual genera prepared : "
    f"{n_bien_genera}"
)

display(
    bien_residual_genus_manifest[
        [
            "BIEN_Genus_Query_ID",
            "Normalized_Acquisition_Genus",
            "Residual_Taxon_Count",
            "Residual_Canonical_Taxa",
            "Requested_Trait",
            "Query_Status",
        ]
    ]
)


print("\nBIEN genus names")
print("-" * 100)

for genus in bien_genus_names:
    print(genus)


print("\nExports")
print("-" * 100)

print(
    f"BIEN query manifest : "
    f"{manifest_path}"
)

print(
    f"BIEN genus names    : "
    f"{genus_names_path}"
)

print(
    "\nSection 8.8J completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8K2 — IMPORT, COVERAGE AND QC INSPECTION OF RESIDUAL BIEN SLA DATA
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd

print("=" * 100)
print("SECTION 8.8K2 — IMPORT, COVERAGE AND QC INSPECTION OF RESIDUAL BIEN SLA DATA")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Project paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

bien_raw_path = (
    PROJECT_ROOT
    / "data"
    / "external"
    / "BIEN"
    / "omo_sla_bien_residual_genus_raw.csv"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

output_dir.mkdir(parents=True, exist_ok=True)

manifest_path = (
    output_dir
    / "omo_sla_bien_residual_genus_query_manifest.csv"
)

print("\nInput files")
print("-" * 100)
print(f"BIEN raw file : {bien_raw_path}")
print(f"Exists        : {bien_raw_path.exists()}")
print(f"Query manifest: {manifest_path}")
print(f"Exists        : {manifest_path.exists()}")

if not bien_raw_path.exists():
    raise FileNotFoundError(
        "Residual BIEN SLA file not found:\n"
        f"{bien_raw_path}"
    )

if not manifest_path.exists():
    raise FileNotFoundError(
        "Residual BIEN query manifest not found:\n"
        f"{manifest_path}"
    )


# -----------------------------------------------------------------------------
# 2. Import raw BIEN data and query manifest
# -----------------------------------------------------------------------------

bien_residual_raw = pd.read_csv(
    bien_raw_path,
    low_memory=False
)

bien_residual_manifest = pd.read_csv(
    manifest_path,
    low_memory=False
)

print("\nRaw BIEN residual import")
print("-" * 100)
print(f"Rows imported       : {len(bien_residual_raw)}")
print(f"Columns imported    : {bien_residual_raw.shape[1]}")
print(f"Requested genera    : {len(bien_residual_manifest)}")


# -----------------------------------------------------------------------------
# 3. Structural validation
# -----------------------------------------------------------------------------

required_bien_columns = [
    "scrubbed_genus",
    "scrubbed_species_binomial",
    "trait_name",
    "trait_value",
    "unit",
    "url_source",
    "id",
]

missing_columns = [
    col
    for col in required_bien_columns
    if col not in bien_residual_raw.columns
]

if missing_columns:
    raise KeyError(
        "Required BIEN columns missing:\n"
        + "\n".join(missing_columns)
    )

required_manifest_columns = [
    "BIEN_Genus_Query_ID",
    "Normalized_Acquisition_Genus",
    "Residual_Taxon_Count",
    "Residual_Canonical_Taxa",
]

missing_manifest_columns = [
    col
    for col in required_manifest_columns
    if col not in bien_residual_manifest.columns
]

if missing_manifest_columns:
    raise KeyError(
        "Required manifest columns missing:\n"
        + "\n".join(missing_manifest_columns)
    )


# -----------------------------------------------------------------------------
# 4. Validate requested trait
# -----------------------------------------------------------------------------

trait_distribution = (
    bien_residual_raw["trait_name"]
    .fillna("<MISSING>")
    .value_counts(dropna=False)
    .rename_axis("Trait_Name")
    .reset_index(name="Record_Count")
)

print("\nTrait validation")
print("-" * 100)
print(trait_distribution.to_string(index=False))


# -----------------------------------------------------------------------------
# 5. Convert SLA to numeric
# -----------------------------------------------------------------------------

bien_residual_raw["trait_value_numeric"] = pd.to_numeric(
    bien_residual_raw["trait_value"],
    errors="coerce"
)

numeric_count = (
    bien_residual_raw["trait_value_numeric"]
    .notna()
    .sum()
)

non_numeric_count = (
    bien_residual_raw["trait_value_numeric"]
    .isna()
    .sum()
)

print("\nNumeric SLA validation")
print("-" * 100)
print(f"Numeric SLA records     : {numeric_count}")
print(f"Non-numeric SLA records : {non_numeric_count}")


# -----------------------------------------------------------------------------
# 6. Unit inspection
#
# BIEN reports the requested SLA trait in m2 kg-1.
#
# Numerically:
#
#     1 m2 kg-1 = 1 mm2 mg-1
#
# Therefore the numerical values do NOT require a scale conversion.
# We retain the raw unit here and assign the standardized unit separately.
# -----------------------------------------------------------------------------

unit_distribution = (
    bien_residual_raw["unit"]
    .fillna("<MISSING>")
    .value_counts(dropna=False)
    .rename_axis("BIEN_Unit")
    .reset_index(name="Record_Count")
)

bien_residual_raw["Standardized_SLA_Value"] = (
    bien_residual_raw["trait_value_numeric"]
)

bien_residual_raw["Standardized_SLA_Unit"] = "mm2 mg-1"

print("\nBIEN unit distribution")
print("-" * 100)
print(unit_distribution.to_string(index=False))


# -----------------------------------------------------------------------------
# 7. Coverage by requested genus
# -----------------------------------------------------------------------------

genus_record_counts = (
    bien_residual_raw
    .groupby("scrubbed_genus", dropna=False)
    .size()
    .rename("BIEN_Record_Count")
    .reset_index()
)

genus_species_counts = (
    bien_residual_raw
    .dropna(subset=["scrubbed_genus"])
    .groupby("scrubbed_genus")["scrubbed_species_binomial"]
    .nunique(dropna=True)
    .rename("BIEN_Unique_Species")
    .reset_index()
)

genus_numeric_counts = (
    bien_residual_raw
    .dropna(subset=["scrubbed_genus"])
    .groupby("scrubbed_genus")["Standardized_SLA_Value"]
    .apply(lambda x: x.notna().sum())
    .rename("BIEN_Numeric_SLA_Count")
    .reset_index()
)

genus_source_counts = (
    bien_residual_raw
    .dropna(subset=["scrubbed_genus"])
    .groupby("scrubbed_genus")["url_source"]
    .nunique(dropna=True)
    .rename("BIEN_Unique_Sources")
    .reset_index()
)

bien_genus_coverage = (
    bien_residual_manifest[
        [
            "BIEN_Genus_Query_ID",
            "Normalized_Acquisition_Genus",
            "Residual_Taxon_Count",
            "Residual_Canonical_Taxa",
        ]
    ]
    .merge(
        genus_record_counts,
        how="left",
        left_on="Normalized_Acquisition_Genus",
        right_on="scrubbed_genus",
    )
    .drop(columns=["scrubbed_genus"], errors="ignore")
    .merge(
        genus_species_counts,
        how="left",
        left_on="Normalized_Acquisition_Genus",
        right_on="scrubbed_genus",
    )
    .drop(columns=["scrubbed_genus"], errors="ignore")
    .merge(
        genus_numeric_counts,
        how="left",
        left_on="Normalized_Acquisition_Genus",
        right_on="scrubbed_genus",
    )
    .drop(columns=["scrubbed_genus"], errors="ignore")
    .merge(
        genus_source_counts,
        how="left",
        left_on="Normalized_Acquisition_Genus",
        right_on="scrubbed_genus",
    )
    .drop(columns=["scrubbed_genus"], errors="ignore")
)

count_columns = [
    "BIEN_Record_Count",
    "BIEN_Unique_Species",
    "BIEN_Numeric_SLA_Count",
    "BIEN_Unique_Sources",
]

bien_genus_coverage[count_columns] = (
    bien_genus_coverage[count_columns]
    .fillna(0)
    .astype(int)
)

bien_genus_coverage["BIEN_Coverage_Status"] = np.where(
    bien_genus_coverage["BIEN_Numeric_SLA_Count"] > 0,
    "BIEN_RECORDS_AVAILABLE",
    "NO_BIEN_SLA_RECORDS",
)

print("\nBIEN coverage across requested residual genera")
print("-" * 100)
print(
    bien_genus_coverage.to_string(
        index=False
    )
)


# -----------------------------------------------------------------------------
# 8. Coverage summary
# -----------------------------------------------------------------------------

coverage_summary = (
    bien_genus_coverage["BIEN_Coverage_Status"]
    .value_counts()
)

print("\nCoverage summary")
print("-" * 100)
print(coverage_summary.to_string())


# -----------------------------------------------------------------------------
# 9. Raw numerical summary
# -----------------------------------------------------------------------------

raw_numeric_summary = (
    bien_residual_raw["Standardized_SLA_Value"]
    .describe()
    .to_frame("Raw_BIEN_SLA")
)

print("\nRaw residual BIEN SLA numerical summary")
print("-" * 100)
print(raw_numeric_summary.to_string())


# -----------------------------------------------------------------------------
# 10. Numerical summary by genus
# -----------------------------------------------------------------------------

numeric_valid = bien_residual_raw.dropna(
    subset=[
        "scrubbed_genus",
        "Standardized_SLA_Value",
    ]
).copy()

genus_numeric_summary = (
    numeric_valid
    .groupby("scrubbed_genus")["Standardized_SLA_Value"]
    .agg(
        count="count",
        min="min",
        median="median",
        mean="mean",
        max="max",
    )
    .reset_index()
    .sort_values(
        "count",
        ascending=False
    )
)

print("\nRaw SLA summary by returned genus")
print("-" * 100)
print(
    genus_numeric_summary.to_string(
        index=False
    )
)


# -----------------------------------------------------------------------------
# 11. Identify biologically suspicious raw measurements
#
# These are DIAGNOSTIC flags only.
#
# They are NOT yet exclusions.
#
# A broad diagnostic range is used so that suspicious orders-of-magnitude
# values can be inspected before formal genus-specific QC.
# -----------------------------------------------------------------------------

diagnostic_lower_bound = 1.0
diagnostic_upper_bound = 100.0

bien_residual_raw["Diagnostic_Value_Flag"] = np.select(
    [
        bien_residual_raw["Standardized_SLA_Value"].isna(),
        bien_residual_raw["Standardized_SLA_Value"]
        < diagnostic_lower_bound,
        bien_residual_raw["Standardized_SLA_Value"]
        > diagnostic_upper_bound,
    ],
    [
        "NON_NUMERIC",
        "SUSPICIOUSLY_LOW",
        "SUSPICIOUSLY_HIGH",
    ],
    default="CENTRAL_DIAGNOSTIC_RANGE",
)

diagnostic_flag_summary = (
    bien_residual_raw["Diagnostic_Value_Flag"]
    .value_counts()
)

print("\nDiagnostic value-flag distribution")
print("-" * 100)
print(diagnostic_flag_summary.to_string())


# -----------------------------------------------------------------------------
# 12. Suspicious-value records
# -----------------------------------------------------------------------------

suspicious_records = (
    bien_residual_raw[
        bien_residual_raw[
            "Diagnostic_Value_Flag"
        ].isin(
            [
                "SUSPICIOUSLY_LOW",
                "SUSPICIOUSLY_HIGH",
            ]
        )
    ]
    [
        [
            "scrubbed_genus",
            "scrubbed_species_binomial",
            "Standardized_SLA_Value",
            "unit",
            "method",
            "url_source",
            "project_pi",
            "id",
            "Diagnostic_Value_Flag",
        ]
    ]
    .sort_values(
        [
            "scrubbed_genus",
            "Standardized_SLA_Value",
        ]
    )
)

print("\nSuspicious raw SLA measurements")
print("-" * 100)

if suspicious_records.empty:
    print("No suspicious measurements identified.")
else:
    print(
        suspicious_records.to_string(
            index=False
        )
    )


# -----------------------------------------------------------------------------
# 13. Suspicious-value summary by genus
# -----------------------------------------------------------------------------

suspicious_genus_summary = (
    bien_residual_raw
    .assign(
        Suspicious_Record=lambda df:
        df["Diagnostic_Value_Flag"].isin(
            [
                "SUSPICIOUSLY_LOW",
                "SUSPICIOUSLY_HIGH",
            ]
        )
    )
    .groupby("scrubbed_genus", dropna=False)
    .agg(
        Raw_Record_Count=("id", "size"),
        Suspicious_Record_Count=(
            "Suspicious_Record",
            "sum",
        ),
    )
    .reset_index()
)

suspicious_genus_summary[
    "Suspicious_Percent"
] = (
    suspicious_genus_summary[
        "Suspicious_Record_Count"
    ]
    / suspicious_genus_summary[
        "Raw_Record_Count"
    ]
    * 100
)

suspicious_genus_summary = (
    suspicious_genus_summary
    .sort_values(
        [
            "Suspicious_Record_Count",
            "Raw_Record_Count",
        ],
        ascending=[False, False],
    )
)

print("\nSuspicious-value summary by genus")
print("-" * 100)
print(
    suspicious_genus_summary.to_string(
        index=False
    )
)


# -----------------------------------------------------------------------------
# 14. Species-level structural summary
# -----------------------------------------------------------------------------

species_summary = (
    numeric_valid
    .groupby(
        [
            "scrubbed_genus",
            "scrubbed_species_binomial",
        ],
        dropna=False,
    )
    .agg(
        Record_Count=(
            "Standardized_SLA_Value",
            "count",
        ),
        Min_SLA=(
            "Standardized_SLA_Value",
            "min",
        ),
        Median_SLA=(
            "Standardized_SLA_Value",
            "median",
        ),
        Mean_SLA=(
            "Standardized_SLA_Value",
            "mean",
        ),
        Max_SLA=(
            "Standardized_SLA_Value",
            "max",
        ),
        Unique_Sources=(
            "url_source",
            "nunique",
        ),
    )
    .reset_index()
    .sort_values(
        [
            "scrubbed_genus",
            "Record_Count",
        ],
        ascending=[True, False],
    )
)

print("\nSpecies-level raw SLA summary")
print("-" * 100)
print(
    species_summary.to_string(
        index=False
    )
)


# -----------------------------------------------------------------------------
# 15. Missingness inspection
# -----------------------------------------------------------------------------

missingness = pd.DataFrame(
    {
        "Column": bien_residual_raw.columns,
        "Missing_Count": [
            bien_residual_raw[col].isna().sum()
            for col in bien_residual_raw.columns
        ],
    }
)

missingness["Missing_Percent"] = (
    missingness["Missing_Count"]
    / len(bien_residual_raw)
    * 100
)

print("\nMissingness")
print("-" * 100)
print(
    missingness.to_string(
        index=False
    )
)


# -----------------------------------------------------------------------------
# 16. Export diagnostic products
# -----------------------------------------------------------------------------

coverage_export = (
    output_dir
    / "omo_sla_bien_residual_genus_coverage.csv"
)

unit_export = (
    output_dir
    / "omo_sla_bien_residual_unit_distribution.csv"
)

numeric_export = (
    output_dir
    / "omo_sla_bien_residual_raw_numeric_summary.csv"
)

genus_summary_export = (
    output_dir
    / "omo_sla_bien_residual_raw_genus_numeric_summary.csv"
)

species_summary_export = (
    output_dir
    / "omo_sla_bien_residual_raw_species_numeric_summary.csv"
)

suspicious_export = (
    output_dir
    / "omo_sla_bien_residual_suspicious_records.csv"
)

suspicious_genus_export = (
    output_dir
    / "omo_sla_bien_residual_suspicious_genus_summary.csv"
)

missingness_export = (
    output_dir
    / "omo_sla_bien_residual_raw_missingness.csv"
)

bien_genus_coverage.to_csv(
    coverage_export,
    index=False
)

unit_distribution.to_csv(
    unit_export,
    index=False
)

raw_numeric_summary.to_csv(
    numeric_export
)

genus_numeric_summary.to_csv(
    genus_summary_export,
    index=False
)

species_summary.to_csv(
    species_summary_export,
    index=False
)

suspicious_records.to_csv(
    suspicious_export,
    index=False
)

suspicious_genus_summary.to_csv(
    suspicious_genus_export,
    index=False
)

missingness.to_csv(
    missingness_export,
    index=False
)


# -----------------------------------------------------------------------------
# 17. Final checkpoint
# -----------------------------------------------------------------------------

returned_genera = (
    bien_genus_coverage[
        bien_genus_coverage[
            "BIEN_Coverage_Status"
        ]
        == "BIEN_RECORDS_AVAILABLE"
    ]
)

missing_genera = (
    bien_genus_coverage[
        bien_genus_coverage[
            "BIEN_Coverage_Status"
        ]
        == "NO_BIEN_SLA_RECORDS"
    ]
)

print("\nSection 8.8K2 checkpoint")
print("-" * 100)
print(
    f"Requested residual genera       : "
    f"{len(bien_residual_manifest)}"
)
print(
    f"Genera with BIEN SLA records    : "
    f"{len(returned_genera)}"
)
print(
    f"Genera without BIEN SLA records : "
    f"{len(missing_genera)}"
)
print(
    f"Raw BIEN SLA records            : "
    f"{len(bien_residual_raw)}"
)
print(
    f"Numeric SLA records             : "
    f"{numeric_count}"
)
print(
    f"Suspicious diagnostic records   : "
    f"{len(suspicious_records)}"
)

print("\nExports")
print("-" * 100)
print(f"Genus coverage     : {coverage_export}")
print(f"Unit distribution : {unit_export}")
print(f"Numeric summary    : {numeric_export}")
print(f"Genus summary      : {genus_summary_export}")
print(f"Species summary    : {species_summary_export}")
print(f"Suspicious records : {suspicious_export}")
print(f"Suspicious genera  : {suspicious_genus_export}")
print(f"Missingness        : {missingness_export}")

print(
    "\nSection 8.8K2 completed successfully."
)
```

### 8.8K3 Robust QC, Species-Level Aggregation and BIEN Genus SLA Estimation

The residual BIEN acquisition returned 695 SLA measurements across 18 of the 46 requested genera. Inspection identified seven records with standardized values below 1 mm² mg⁻¹ and on the order of 10⁻⁵. These measurements are inconsistent in scale with the surrounding observations and are treated as clear numerical anomalies.

To prevent unequal sampling intensity from biasing genus-level estimates, BIEN measurements are not aggregated directly at the genus level. Instead, quality-controlled observations are first summarized to species-level medians. Each identified species therefore contributes one equally weighted value to its genus.

The final candidate genus SLA is then calculated as the median of the retained species-level medians. Genus-only records lacking a species binomial are preserved diagnostically but are not included in the primary species-balanced genus estimate.

Evidence strength is classified using taxonomic breadth and supporting data volume. Multi-species genera receive stronger confidence than single-species proxies, while genera represented by only one identifiable species remain explicitly flagged as limited-evidence genus estimates.

No candidate BIEN genus value is assigned to Omo Forest taxa in this subsection.

```python
# =============================================================================
# SECTION 8.8K3 — ROBUST QC, SPECIES-LEVEL AGGREGATION AND
#                 BIEN GENUS SLA ESTIMATION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8K3 — ROBUST QC, SPECIES-LEVEL AGGREGATION AND BIEN GENUS SLA ESTIMATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "bien_residual_raw",
    "bien_genus_coverage",
]

missing_objects = [
    obj for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Section 8.8K2 first."
    )


# -----------------------------------------------------------------------------
# 2. Create working QC table
# -----------------------------------------------------------------------------

bien_residual_qc = (
    bien_residual_raw
    .copy()
)


# -----------------------------------------------------------------------------
# 3. Define clear low-scale anomaly rule
#
# Section 8.8K2 identified seven values below 1 mm2 mg-1.
# These are orders of magnitude below the remaining SLA observations and
# originate from the same anomalous scale pattern.
#
# The raw records remain preserved; only the aggregation working table
# excludes them.
# -----------------------------------------------------------------------------

bien_residual_qc[
    "Low_Scale_Anomaly"
] = (
    bien_residual_qc[
        "Standardized_SLA_Value"
    ] < 1.0
)


bien_residual_qc[
    "QC_Status"
] = np.where(
    bien_residual_qc[
        "Low_Scale_Anomaly"
    ],
    "EXCLUDE_CLEAR_LOW_SCALE_ANOMALY",
    "RETAIN"
)


# -----------------------------------------------------------------------------
# 4. Validate expected anomaly count
# -----------------------------------------------------------------------------

n_clear_anomalies = (
    bien_residual_qc[
        "Low_Scale_Anomaly"
    ].sum()
)

assert n_clear_anomalies == 7, (
    f"Expected 7 low-scale anomalies, "
    f"found {n_clear_anomalies}."
)


# -----------------------------------------------------------------------------
# 5. Create retained BIEN measurement table
# -----------------------------------------------------------------------------

bien_residual_retained = (
    bien_residual_qc.loc[
        bien_residual_qc[
            "QC_Status"
        ].eq("RETAIN")
    ]
    .copy()
    .reset_index(drop=True)
)


n_retained = len(
    bien_residual_retained
)

assert n_retained == 688, (
    f"Expected 688 retained BIEN records, "
    f"found {n_retained}."
)


# -----------------------------------------------------------------------------
# 6. Separate identified-species and genus-only records
# -----------------------------------------------------------------------------

bien_identified_species_records = (
    bien_residual_retained.loc[
        bien_residual_retained[
            "scrubbed_species_binomial"
        ].notna()
    ]
    .copy()
)

bien_genus_only_records = (
    bien_residual_retained.loc[
        bien_residual_retained[
            "scrubbed_species_binomial"
        ].isna()
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 7. Species-level aggregation
#
# Each species contributes one median SLA value to its genus.
# -----------------------------------------------------------------------------

bien_species_medians = (
    bien_identified_species_records
    .groupby(
        [
            "scrubbed_genus",
            "scrubbed_species_binomial",
        ],
        as_index=False
    )
    .agg(
        Species_Record_Count=(
            "id",
            "count"
        ),

        Species_Source_Count=(
            "url_source",
            "nunique"
        ),

        Species_Min_SLA=(
            "Standardized_SLA_Value",
            "min"
        ),

        Species_Median_SLA=(
            "Standardized_SLA_Value",
            "median"
        ),

        Species_Mean_SLA=(
            "Standardized_SLA_Value",
            "mean"
        ),

        Species_Max_SLA=(
            "Standardized_SLA_Value",
            "max"
        ),
    )
)


# -----------------------------------------------------------------------------
# 8. Genus-level aggregation from species medians
# -----------------------------------------------------------------------------

bien_candidate_genus_estimates = (
    bien_species_medians
    .groupby(
        "scrubbed_genus",
        as_index=False
    )
    .agg(
        Retained_Species_Count=(
            "scrubbed_species_binomial",
            "nunique"
        ),

        Species_Median_Min=(
            "Species_Median_SLA",
            "min"
        ),

        Candidate_Genus_SLA=(
            "Species_Median_SLA",
            "median"
        ),

        Species_Median_Mean=(
            "Species_Median_SLA",
            "mean"
        ),

        Species_Median_Max=(
            "Species_Median_SLA",
            "max"
        ),

        Total_Identified_Record_Count=(
            "Species_Record_Count",
            "sum"
        ),

        Total_Species_Source_Count=(
            "Species_Source_Count",
            "sum"
        ),
    )
)


# -----------------------------------------------------------------------------
# 9. Attach raw and retained record counts by genus
# -----------------------------------------------------------------------------

raw_genus_counts = (
    bien_residual_qc
    .groupby(
        "scrubbed_genus",
        as_index=False
    )
    .agg(
        Raw_Record_Count=(
            "id",
            "count"
        ),

        Excluded_Anomaly_Count=(
            "Low_Scale_Anomaly",
            "sum"
        ),
    )
)


retained_genus_counts = (
    bien_residual_retained
    .groupby(
        "scrubbed_genus",
        as_index=False
    )
    .agg(
        Retained_Record_Count=(
            "id",
            "count"
        ),

        Retained_Source_Count=(
            "url_source",
            "nunique"
        ),
    )
)


# -----------------------------------------------------------------------------
# 10. Summarize genus-only records
# -----------------------------------------------------------------------------

genus_only_summary = (
    bien_genus_only_records
    .groupby(
        "scrubbed_genus",
        as_index=False
    )
    .agg(
        Genus_Only_Record_Count=(
            "id",
            "count"
        ),

        Genus_Only_Median_SLA=(
            "Standardized_SLA_Value",
            "median"
        ),

        Genus_Only_Min_SLA=(
            "Standardized_SLA_Value",
            "min"
        ),

        Genus_Only_Max_SLA=(
            "Standardized_SLA_Value",
            "max"
        ),
    )
)


# -----------------------------------------------------------------------------
# 11. Merge genus evidence diagnostics
# -----------------------------------------------------------------------------

bien_candidate_genus_estimates = (
    bien_candidate_genus_estimates
    .merge(
        raw_genus_counts,
        how="left",
        on="scrubbed_genus"
    )
    .merge(
        retained_genus_counts,
        how="left",
        on="scrubbed_genus"
    )
    .merge(
        genus_only_summary,
        how="left",
        on="scrubbed_genus"
    )
)


genus_only_columns = [
    "Genus_Only_Record_Count",
]

for col in genus_only_columns:
    bien_candidate_genus_estimates[col] = (
        bien_candidate_genus_estimates[col]
        .fillna(0)
        .astype(int)
    )


# -----------------------------------------------------------------------------
# 12. Evidence-strength classification
#
# Strong:
#   >= 5 identified species and >= 20 retained records
#
# Moderate:
#   >= 3 identified species and >= 5 retained records
#
# Limited:
#   >= 2 identified species
#
# Single-species proxy:
#   1 identified species
# -----------------------------------------------------------------------------

def classify_bien_genus_strength(row):

    n_species = row[
        "Retained_Species_Count"
    ]

    n_records = row[
        "Retained_Record_Count"
    ]

    if (
        n_species >= 5
        and n_records >= 20
    ):
        return "STRONG"

    if (
        n_species >= 3
        and n_records >= 5
    ):
        return "MODERATE"

    if n_species >= 2:
        return "LIMITED"

    if n_species == 1:
        return "SINGLE_SPECIES_PROXY"

    return "NO_IDENTIFIED_SPECIES"


bien_candidate_genus_estimates[
    "Evidence_Strength"
] = (
    bien_candidate_genus_estimates.apply(
        classify_bien_genus_strength,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 13. Candidate-quality flag
# -----------------------------------------------------------------------------

quality_map = {
    "STRONG":
        "BIEN_GENUS_STRONG",

    "MODERATE":
        "BIEN_GENUS_MODERATE",

    "LIMITED":
        "BIEN_GENUS_LIMITED",

    "SINGLE_SPECIES_PROXY":
        "BIEN_GENUS_SINGLE_SPECIES_PROXY",

    "NO_IDENTIFIED_SPECIES":
        "BIEN_GENUS_NO_IDENTIFIED_SPECIES",
}


bien_candidate_genus_estimates[
    "Candidate_Quality_Flag"
] = (
    bien_candidate_genus_estimates[
        "Evidence_Strength"
    ]
    .map(quality_map)
)


# -----------------------------------------------------------------------------
# 14. Add provenance and estimation metadata
# -----------------------------------------------------------------------------

bien_candidate_genus_estimates[
    "Candidate_Genus_SLA_Unit"
] = "mm2 mg-1"

bien_candidate_genus_estimates[
    "Provenance_Class"
] = "GENUS_ESTIMATE"

bien_candidate_genus_estimates[
    "Trait_Source"
] = "BIEN"

bien_candidate_genus_estimates[
    "Aggregation_Method"
] = (
    "MEDIAN_OF_IDENTIFIED_SPECIES_MEDIANS_AFTER_CLEAR_ANOMALY_FILTER"
)


# -----------------------------------------------------------------------------
# 15. Handle Carica separately
#
# Carica has a BIEN record but no scrubbed_species_binomial.
# Therefore it cannot enter the species-balanced genus estimator.
# It remains a separate genus-only candidate requiring review.
# -----------------------------------------------------------------------------

carica_genus_only = (
    bien_genus_only_records.loc[
        bien_genus_only_records[
            "scrubbed_genus"
        ].eq("Carica")
    ]
    .copy()
)


if not carica_genus_only.empty:

    carica_candidate = pd.DataFrame(
        [
            {
                "scrubbed_genus":
                    "Carica",

                "Retained_Species_Count":
                    0,

                "Species_Median_Min":
                    np.nan,

                "Candidate_Genus_SLA":
                    carica_genus_only[
                        "Standardized_SLA_Value"
                    ].median(),

                "Species_Median_Mean":
                    np.nan,

                "Species_Median_Max":
                    np.nan,

                "Total_Identified_Record_Count":
                    0,

                "Total_Species_Source_Count":
                    0,

                "Raw_Record_Count":
                    len(
                        bien_residual_qc.loc[
                            bien_residual_qc[
                                "scrubbed_genus"
                            ].eq("Carica")
                        ]
                    ),

                "Excluded_Anomaly_Count":
                    int(
                        bien_residual_qc.loc[
                            bien_residual_qc[
                                "scrubbed_genus"
                            ].eq("Carica"),
                            "Low_Scale_Anomaly"
                        ].sum()
                    ),

                "Retained_Record_Count":
                    len(carica_genus_only),

                "Retained_Source_Count":
                    carica_genus_only[
                        "url_source"
                    ].nunique(),

                "Genus_Only_Record_Count":
                    len(carica_genus_only),

                "Genus_Only_Median_SLA":
                    carica_genus_only[
                        "Standardized_SLA_Value"
                    ].median(),

                "Genus_Only_Min_SLA":
                    carica_genus_only[
                        "Standardized_SLA_Value"
                    ].min(),

                "Genus_Only_Max_SLA":
                    carica_genus_only[
                        "Standardized_SLA_Value"
                    ].max(),

                "Evidence_Strength":
                    "GENUS_ONLY_SINGLE_RECORD",

                "Candidate_Quality_Flag":
                    "BIEN_GENUS_ONLY_SINGLE_RECORD_REVIEW",

                "Candidate_Genus_SLA_Unit":
                    "mm2 mg-1",

                "Provenance_Class":
                    "GENUS_ESTIMATE",

                "Trait_Source":
                    "BIEN",

                "Aggregation_Method":
                    "GENUS_ONLY_RECORD_PENDING_REVIEW",
            }
        ]
    )

    bien_candidate_genus_estimates = (
        pd.concat(
            [
                bien_candidate_genus_estimates,
                carica_candidate,
            ],
            ignore_index=True
        )
    )


# -----------------------------------------------------------------------------
# 16. Sort candidate table
# -----------------------------------------------------------------------------

bien_candidate_genus_estimates = (
    bien_candidate_genus_estimates
    .sort_values(
        "scrubbed_genus"
    )
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 17. Validation
# -----------------------------------------------------------------------------

n_candidate_genera = (
    bien_candidate_genus_estimates[
        "scrubbed_genus"
    ].nunique()
)

assert n_candidate_genera == 18, (
    f"Expected candidate summaries for 18 BIEN genera, "
    f"found {n_candidate_genera}."
)

assert (
    bien_candidate_genus_estimates[
        "Candidate_Genus_SLA"
    ]
    .notna()
    .all()
)

assert (
    bien_candidate_genus_estimates[
        "Candidate_Genus_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)


# -----------------------------------------------------------------------------
# 18. Build excluded anomaly table
# -----------------------------------------------------------------------------

bien_clear_anomalies = (
    bien_residual_qc.loc[
        bien_residual_qc[
            "Low_Scale_Anomaly"
        ]
    ]
    [
        [
            "scrubbed_genus",
            "scrubbed_species_binomial",
            "Standardized_SLA_Value",
            "unit",
            "url_source",
            "project_pi",
            "id",
            "QC_Status",
        ]
    ]
    .copy()
    .sort_values(
        [
            "scrubbed_genus",
            "Standardized_SLA_Value",
        ]
    )
)


# -----------------------------------------------------------------------------
# 19. Export
# -----------------------------------------------------------------------------

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

qc_path = (
    output_dir
    / "omo_sla_bien_residual_measurement_qc.csv"
)

retained_path = (
    output_dir
    / "omo_sla_bien_residual_measurements_retained.csv"
)

species_median_path = (
    output_dir
    / "omo_sla_bien_residual_species_medians.csv"
)

genus_candidate_path = (
    output_dir
    / "omo_sla_bien_residual_candidate_genus_estimates.csv"
)

genus_only_path = (
    output_dir
    / "omo_sla_bien_residual_genus_only_records.csv"
)

anomaly_path = (
    output_dir
    / "omo_sla_bien_residual_clear_low_scale_anomalies.csv"
)


bien_residual_qc.to_csv(
    qc_path,
    index=False
)

bien_residual_retained.to_csv(
    retained_path,
    index=False
)

bien_species_medians.to_csv(
    species_median_path,
    index=False
)

bien_candidate_genus_estimates.to_csv(
    genus_candidate_path,
    index=False
)

bien_genus_only_records.to_csv(
    genus_only_path,
    index=False
)

bien_clear_anomalies.to_csv(
    anomaly_path,
    index=False
)


# -----------------------------------------------------------------------------
# 20. Diagnostics
# -----------------------------------------------------------------------------

print("\nResidual BIEN measurement QC")
print("-" * 100)

print(
    f"Raw measurements              : "
    f"{len(bien_residual_qc)}"
)

print(
    f"Clear low-scale anomalies     : "
    f"{n_clear_anomalies}"
)

print(
    f"Retained measurements         : "
    f"{n_retained}"
)

print(
    f"Identified-species records    : "
    f"{len(bien_identified_species_records)}"
)

print(
    f"Genus-only records            : "
    f"{len(bien_genus_only_records)}"
)


print("\nCandidate BIEN genus SLA estimates")
print("-" * 100)

display(
    bien_candidate_genus_estimates[
        [
            "scrubbed_genus",
            "Candidate_Genus_SLA",
            "Candidate_Genus_SLA_Unit",
            "Retained_Species_Count",
            "Retained_Record_Count",
            "Retained_Source_Count",
            "Genus_Only_Record_Count",
            "Evidence_Strength",
            "Candidate_Quality_Flag",
            "Aggregation_Method",
        ]
    ]
)


print("\nEvidence-strength distribution")
print("-" * 100)

print(
    bien_candidate_genus_estimates[
        "Evidence_Strength"
    ]
    .value_counts()
    .to_string()
)


print("\nExcluded clear low-scale anomalies")
print("-" * 100)

display(
    bien_clear_anomalies
)


print("\nGenus-only retained records")
print("-" * 100)

if bien_genus_only_records.empty:

    print(
        "No genus-only BIEN records."
    )

else:

    display(
        bien_genus_only_records[
            [
                "scrubbed_genus",
                "scrubbed_species_binomial",
                "Standardized_SLA_Value",
                "url_source",
                "project_pi",
                "id",
            ]
        ]
    )


print("\nExports")
print("-" * 100)

print(
    f"Measurement QC      : "
    f"{qc_path}"
)

print(
    f"Retained records    : "
    f"{retained_path}"
)

print(
    f"Species medians     : "
    f"{species_median_path}"
)

print(
    f"Genus candidates    : "
    f"{genus_candidate_path}"
)

print(
    f"Genus-only records  : "
    f"{genus_only_path}"
)

print(
    f"Excluded anomalies  : "
    f"{anomaly_path}"
)

print(
    "\nSection 8.8K3 completed successfully."
)
```

### 8.8K4 Controlled Acceptance and Omo Crosswalk of BIEN Genus-Level SLA Estimates

The residual BIEN acquisition produced candidate SLA estimates for 18 genera after removal of seven clear low-scale anomalies and species-balanced aggregation.

Seventeen genera contain identifiable species-level BIEN records and are accepted as genus-level SLA estimates. Their evidence strength remains explicitly classified as `STRONG`, `MODERATE`, `LIMITED`, or `SINGLE_SPECIES_PROXY` according to the number of contributing species and measurements.

The genus *Carica* is treated separately. Although BIEN returned a standardized SLA value of 52.144 mm² mg⁻¹, the record lacks a scrubbed species binomial and is represented by only one genus-only observation. Consequently, this value is retained as a candidate requiring further review and is not incorporated into the accepted SLA dataset at this stage.

Accepted BIEN genus estimates are mapped only to residual Omo Forest taxa sharing the corresponding normalized acquisition genus. They retain the provenance class `GENUS_ESTIMATE` and do not supersede species-level TRY, species-level literature, or previously accepted higher-priority evidence.

Following this integration, the remaining SLA gaps are recalculated before family-level or broader fallback strategies are considered.

```python
# =============================================================================
# SECTION 8.8K4 — CONTROLLED ACCEPTANCE AND OMO CROSSWALK OF
#                 BIEN GENUS-LEVEL SLA ESTIMATES
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8K4 — CONTROLLED ACCEPTANCE AND OMO CROSSWALK OF BIEN GENUS SLA ESTIMATES")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required notebook objects
# -----------------------------------------------------------------------------

required_objects = [
    "bien_candidate_genus_estimates",
    "sla_residual_normalized",
    "omo_try_congeneric_accepted",
]

missing_objects = [
    obj
    for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Sections 8.8I and 8.8K3 first."
    )


output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# 2. Separate accepted BIEN genus candidates from review-required candidates
#
# Carica is deliberately withheld because its only retained record lacks an
# identified species binomial.
# -----------------------------------------------------------------------------

bien_genus_accepted = (
    bien_candidate_genus_estimates.loc[
        ~bien_candidate_genus_estimates[
            "Evidence_Strength"
        ].eq("GENUS_ONLY_SINGLE_RECORD")
    ]
    .copy()
    .reset_index(drop=True)
)


bien_genus_pending_review = (
    bien_candidate_genus_estimates.loc[
        bien_candidate_genus_estimates[
            "Evidence_Strength"
        ].eq("GENUS_ONLY_SINGLE_RECORD")
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 3. Validate acceptance partition
# -----------------------------------------------------------------------------

assert len(
    bien_candidate_genus_estimates
) == 18

assert len(
    bien_genus_accepted
) == 17, (
    f"Expected 17 accepted BIEN genus estimates, "
    f"found {len(bien_genus_accepted)}."
)

assert len(
    bien_genus_pending_review
) == 1, (
    f"Expected 1 review-required BIEN genus estimate, "
    f"found {len(bien_genus_pending_review)}."
)

assert (
    bien_genus_pending_review[
        "scrubbed_genus"
    ].iloc[0]
    == "Carica"
)


# -----------------------------------------------------------------------------
# 4. Add final acceptance metadata
# -----------------------------------------------------------------------------

bien_genus_accepted[
    "Acceptance_Status"
] = "ACCEPTED_GENUS_ESTIMATE"

bien_genus_accepted[
    "Accepted_Provenance_Class"
] = "GENUS_ESTIMATE"

bien_genus_accepted[
    "Accepted_Source"
] = "BIEN"

bien_genus_accepted[
    "Accepted_SLA_Value"
] = (
    bien_genus_accepted[
        "Candidate_Genus_SLA"
    ]
)

bien_genus_accepted[
    "Accepted_SLA_Unit"
] = "mm2 mg-1"

bien_genus_accepted[
    "Accepted_Quality_Flag"
] = (
    bien_genus_accepted[
        "Candidate_Quality_Flag"
    ]
)


# -----------------------------------------------------------------------------
# 5. Prepare accepted genus lookup
# -----------------------------------------------------------------------------

accepted_genus_lookup = (
    bien_genus_accepted[
        [
            "scrubbed_genus",
            "Accepted_SLA_Value",
            "Accepted_SLA_Unit",
            "Evidence_Strength",
            "Accepted_Quality_Flag",
            "Retained_Species_Count",
            "Retained_Record_Count",
            "Retained_Source_Count",
            "Excluded_Anomaly_Count",
            "Aggregation_Method",
            "Accepted_Provenance_Class",
            "Accepted_Source",
            "Acceptance_Status",
        ]
    ]
    .copy()
    .rename(
        columns={
            "scrubbed_genus":
                "Normalized_Acquisition_Genus",

            "Evidence_Strength":
                "BIEN_Genus_Evidence_Strength",

            "Retained_Species_Count":
                "BIEN_Retained_Species_Count",

            "Retained_Record_Count":
                "BIEN_Retained_Record_Count",

            "Retained_Source_Count":
                "BIEN_Retained_Source_Count",

            "Excluded_Anomaly_Count":
                "BIEN_Excluded_Anomaly_Count",

            "Aggregation_Method":
                "BIEN_Aggregation_Method",
        }
    )
)


# -----------------------------------------------------------------------------
# 6. Attach accepted BIEN genus estimates to residual Omo taxa
# -----------------------------------------------------------------------------

sla_residual_with_bien = (
    sla_residual_normalized
    .merge(
        accepted_genus_lookup,
        how="left",
        on="Normalized_Acquisition_Genus",
        validate="many_to_one"
    )
)


# -----------------------------------------------------------------------------
# 7. Identify Omo taxa newly covered by BIEN genus estimates
#
# Exclude the 7 taxa already assigned TRY-congeneric estimates in 8.8I.
# -----------------------------------------------------------------------------

new_bien_mask = (
    sla_residual_with_bien[
        "Accepted_SLA_Value"
    ].notna()
    &
    ~sla_residual_with_bien[
        "TRY_Congeneric_Available"
    ]
)


omo_bien_genus_accepted = (
    sla_residual_with_bien.loc[
        new_bien_mask
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 8. Build compact Omo-ready BIEN genus table
# -----------------------------------------------------------------------------

omo_bien_genus_values = (
    omo_bien_genus_accepted[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",

            "Accepted_SLA_Value",
            "Accepted_SLA_Unit",

            "BIEN_Genus_Evidence_Strength",
            "Accepted_Quality_Flag",

            "BIEN_Retained_Species_Count",
            "BIEN_Retained_Record_Count",
            "BIEN_Retained_Source_Count",
            "BIEN_Excluded_Anomaly_Count",

            "BIEN_Aggregation_Method",
            "Accepted_Provenance_Class",
            "Accepted_Source",
            "Acceptance_Status",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Accepted_SLA_Value":
                "Standardized_SLA_Value",

            "Accepted_SLA_Unit":
                "Standardized_SLA_Unit",

            "Accepted_Provenance_Class":
                "Provenance_Class",

            "Accepted_Source":
                "Trait_Source",

            "Accepted_Quality_Flag":
                "Quality_Flag",
        }
    )
)


# -----------------------------------------------------------------------------
# 9. Validate expected new coverage
# -----------------------------------------------------------------------------

n_new_bien_taxa = len(
    omo_bien_genus_values
)

n_new_bien_genera = (
    omo_bien_genus_values[
        "Normalized_Acquisition_Genus"
    ].nunique()
)


assert n_new_bien_genera == 17, (
    f"Expected 17 accepted BIEN genera, "
    f"found {n_new_bien_genera}."
)

assert n_new_bien_taxa == 27, (
    f"Expected 27 Omo taxa to receive new BIEN genus estimates, "
    f"found {n_new_bien_taxa}."
)


# -----------------------------------------------------------------------------
# 10. Confirm all accepted values are standardized correctly
# -----------------------------------------------------------------------------

assert (
    omo_bien_genus_values[
        "Standardized_SLA_Value"
    ]
    .notna()
    .all()
)

assert (
    omo_bien_genus_values[
        "Standardized_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)

assert (
    omo_bien_genus_values[
        "Provenance_Class"
    ]
    .eq("GENUS_ESTIMATE")
    .all()
)


# -----------------------------------------------------------------------------
# 11. Identify remaining residual taxa after:
#
#     7 TRY-congeneric estimates
#    27 accepted BIEN genus estimates
#
# from the original 76 residual taxa.
# -----------------------------------------------------------------------------

covered_by_try_congeneric = set(
    omo_try_congeneric_accepted[
        "Canonical_Species"
    ]
)

covered_by_bien_genus = set(
    omo_bien_genus_values[
        "Canonical_Species"
    ]
)


assert covered_by_try_congeneric.isdisjoint(
    covered_by_bien_genus
), (
    "Overlap detected between TRY-congeneric and newly accepted "
    "BIEN genus estimates."
)


covered_residual_taxa = (
    covered_by_try_congeneric
    | covered_by_bien_genus
)


sla_post_bien_remaining_gaps = (
    sla_residual_normalized.loc[
        ~sla_residual_normalized[
            "Canonical_Species"
        ].isin(
            covered_residual_taxa
        )
    ]
    .copy()
    .reset_index(drop=True)
)


n_remaining = len(
    sla_post_bien_remaining_gaps
)

assert n_remaining == 42, (
    f"Expected 42 SLA gaps after BIEN genus integration, "
    f"found {n_remaining}."
)


# -----------------------------------------------------------------------------
# 12. Flag Carica explicitly in remaining-gap table
# -----------------------------------------------------------------------------

sla_post_bien_remaining_gaps[
    "Post_BIEN_Status"
] = "NO_ACCEPTED_BIEN_GENUS_ESTIMATE"


carica_mask = (
    sla_post_bien_remaining_gaps[
        "Normalized_Acquisition_Genus"
    ].eq("Carica")
)

sla_post_bien_remaining_gaps.loc[
    carica_mask,
    "Post_BIEN_Status"
] = "BIEN_GENUS_ONLY_CANDIDATE_REVIEW_REQUIRED"


# -----------------------------------------------------------------------------
# 13. Assign next acquisition strategy
# -----------------------------------------------------------------------------

sla_post_bien_remaining_gaps[
    "Next_Acquisition_Strategy"
] = "FAMILY_LEVEL_EVIDENCE"


sla_post_bien_remaining_gaps.loc[
    carica_mask,
    "Next_Acquisition_Strategy"
] = (
    "REVIEW_CARICA_BEFORE_FAMILY_FALLBACK"
)


# -----------------------------------------------------------------------------
# 14. Summarize post-BIEN remaining genera
# -----------------------------------------------------------------------------

post_bien_genus_summary = (
    sla_post_bien_remaining_gaps
    .groupby(
        "Normalized_Acquisition_Genus",
        as_index=False
    )
    .agg(
        Remaining_Taxon_Count=(
            "Canonical_Species",
            "count"
        ),

        Remaining_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Post_BIEN_Status=(
            "Post_BIEN_Status",
            "first"
        ),

        Next_Acquisition_Strategy=(
            "Next_Acquisition_Strategy",
            "first"
        ),
    )
    .sort_values(
        "Normalized_Acquisition_Genus"
    )
    .reset_index(drop=True)
)


n_remaining_genera = (
    post_bien_genus_summary[
        "Normalized_Acquisition_Genus"
    ].nunique()
)


# -----------------------------------------------------------------------------
# 15. Overall SLA coverage checkpoint
#
# Previous coverage after 8.8I:
#   51 established pre-8.8I values
#    7 TRY-congeneric values
#   --------------------------------
#   58
#
# New BIEN genus values:
#   27
#
# Total:
#   85 / 127
# -----------------------------------------------------------------------------

previous_coverage = 58

current_total_coverage = (
    previous_coverage
    + n_new_bien_taxa
)

assert current_total_coverage == 85

assert (
    current_total_coverage
    + n_remaining
    == 127
)


# -----------------------------------------------------------------------------
# 16. Export
# -----------------------------------------------------------------------------

accepted_genus_path = (
    output_dir
    / "omo_sla_bien_residual_accepted_genus_estimates.csv"
)

omo_values_path = (
    output_dir
    / "omo_sla_bien_residual_omo_genus_values.csv"
)

pending_review_path = (
    output_dir
    / "omo_sla_bien_residual_genus_candidates_pending_review.csv"
)

remaining_path = (
    output_dir
    / "omo_sla_remaining_gaps_after_bien_genus_integration.csv"
)

remaining_genus_path = (
    output_dir
    / "omo_sla_remaining_genera_after_bien_integration.csv"
)


bien_genus_accepted.to_csv(
    accepted_genus_path,
    index=False
)

omo_bien_genus_values.to_csv(
    omo_values_path,
    index=False
)

bien_genus_pending_review.to_csv(
    pending_review_path,
    index=False
)

sla_post_bien_remaining_gaps.to_csv(
    remaining_path,
    index=False
)

post_bien_genus_summary.to_csv(
    remaining_genus_path,
    index=False
)


# -----------------------------------------------------------------------------
# 17. Diagnostics
# -----------------------------------------------------------------------------

print("\nBIEN genus acceptance summary")
print("-" * 100)

print(
    f"Candidate BIEN genera             : "
    f"{len(bien_candidate_genus_estimates)}"
)

print(
    f"Accepted BIEN genera              : "
    f"{len(bien_genus_accepted)}"
)

print(
    f"Candidates retained for review    : "
    f"{len(bien_genus_pending_review)}"
)

print(
    f"Omo taxa newly covered by BIEN    : "
    f"{n_new_bien_taxa}"
)


print("\nAccepted BIEN genus estimates")
print("-" * 100)

display(
    bien_genus_accepted[
        [
            "scrubbed_genus",
            "Accepted_SLA_Value",
            "Accepted_SLA_Unit",
            "Retained_Species_Count",
            "Retained_Record_Count",
            "Evidence_Strength",
            "Accepted_Quality_Flag",
        ]
    ]
)


print("\nOmo taxa receiving BIEN genus estimates")
print("-" * 100)

display(
    omo_bien_genus_values[
        [
            "Canonical_Species",
            "Normalized_Acquisition_Genus",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "BIEN_Genus_Evidence_Strength",
            "Quality_Flag",
            "Provenance_Class",
        ]
    ]
    .sort_values(
        [
            "Normalized_Acquisition_Genus",
            "Canonical_Species",
        ]
    )
)


print("\nBIEN candidate retained for review")
print("-" * 100)

display(
    bien_genus_pending_review[
        [
            "scrubbed_genus",
            "Candidate_Genus_SLA",
            "Candidate_Genus_SLA_Unit",
            "Retained_Species_Count",
            "Retained_Record_Count",
            "Genus_Only_Record_Count",
            "Evidence_Strength",
            "Candidate_Quality_Flag",
        ]
    ]
)


print("\nPost-BIEN SLA coverage")
print("-" * 100)

print(
    f"Coverage before this subsection : "
    f"{previous_coverage}"
)

print(
    f"New BIEN genus coverage         : "
    f"{n_new_bien_taxa}"
)

print(
    f"Current SLA coverage            : "
    f"{current_total_coverage} / 127"
)

print(
    f"Current coverage percent        : "
    f"{100 * current_total_coverage / 127:.2f}%"
)

print(
    f"Remaining SLA gaps              : "
    f"{n_remaining}"
)

print(
    f"Remaining normalized genera     : "
    f"{n_remaining_genera}"
)


print("\nRemaining SLA gaps")
print("-" * 100)

display(
    sla_post_bien_remaining_gaps[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Post_BIEN_Status",
            "Next_Acquisition_Strategy",
        ]
    ]
)


print("\nRemaining genus summary")
print("-" * 100)

display(
    post_bien_genus_summary
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted genus estimates : "
    f"{accepted_genus_path}"
)

print(
    f"Omo BIEN genus values    : "
    f"{omo_values_path}"
)

print(
    f"Pending review           : "
    f"{pending_review_path}"
)

print(
    f"Remaining SLA gaps       : "
    f"{remaining_path}"
)

print(
    f"Remaining genera         : "
    f"{remaining_genus_path}"
)

print(
    "\nSection 8.8K4 completed successfully."
)
```

## Section 8.8L — Targeted Species-Level Rescue Planning for Remaining SLA Gaps

Following integration of accepted BIEN genus-level SLA estimates, SLA coverage
increased to 85 of 127 canonical Omo taxa (66.93%), leaving 42 taxa unresolved.

Before applying broader family-level estimates, the remaining taxa are subjected
to a targeted species-level rescue stage. This stage prioritizes empirical
species-level evidence wherever possible and distinguishes:

1. taxa with normalized species names suitable for targeted literature search;
2. taxa represented by multiple historical or orthographic variants;
3. taxa requiring special review because potentially usable evidence already exists;
4. genuinely genus-level or unresolved taxa; and
5. taxa that may ultimately require family-level estimation.

No new SLA values are assigned in this subsection. The purpose is to establish
a controlled, auditable acquisition manifest for the remaining taxa.

```python
# =============================================================================
# SECTION 8.8L
# TARGETED SPECIES-LEVEL RESCUE PLANNING FOR REMAINING SLA GAPS
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8L — TARGETED SPECIES-LEVEL RESCUE PLANNING "
    "FOR REMAINING SLA GAPS"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Establish output directory
# -----------------------------------------------------------------------------

project_root = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    project_root
    / "outputs"
    / "functional_traits"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


# -----------------------------------------------------------------------------
# 2. Load remaining SLA gaps from Section 8.8K4
# -----------------------------------------------------------------------------

remaining_gap_path = (
    output_dir
    / "omo_sla_remaining_gaps_after_bien_genus_integration.csv"
)

if not remaining_gap_path.exists():
    raise FileNotFoundError(
        "Remaining SLA gap table from Section 8.8K4 not found:\n"
        f"{remaining_gap_path}"
    )

sla_remaining = pd.read_csv(
    remaining_gap_path
)

print("\nRemaining-gap foundation")
print("-" * 100)
print(f"Remaining taxa loaded : {len(sla_remaining)}")


# -----------------------------------------------------------------------------
# 3. Validate required fields
# -----------------------------------------------------------------------------

required_columns = [
    "Canonical_Species",
    "Normalized_Search_Name",
    "Normalized_Acquisition_Genus",
    "Post_BIEN_Status",
    "Next_Acquisition_Strategy",
]

missing_columns = [
    col
    for col in required_columns
    if col not in sla_remaining.columns
]

if missing_columns:
    raise KeyError(
        "Required columns missing from remaining SLA gap table:\n"
        + "\n".join(missing_columns)
    )


# -----------------------------------------------------------------------------
# 4. Basic text cleaning
# -----------------------------------------------------------------------------

for col in [
    "Canonical_Species",
    "Normalized_Search_Name",
    "Normalized_Acquisition_Genus",
]:
    sla_remaining[col] = (
        sla_remaining[col]
        .astype("string")
        .str.strip()
    )


# -----------------------------------------------------------------------------
# 5. Determine whether normalized search name is species-level
#
# A simple binomial test is appropriate here because these names have already
# passed through the controlled taxonomic normalization stages.
# -----------------------------------------------------------------------------

def is_species_level_name(name):

    if pd.isna(name):
        return False

    name = str(name).strip()

    if name == "":
        return False

    parts = name.split()

    if len(parts) < 2:
        return False

    non_species_terms = {
        "species",
        "sp",
        "sp.",
        "spp",
        "spp.",
    }

    if parts[1].lower() in non_species_terms:
        return False

    return True


sla_remaining[
    "Species_Level_Search_Available"
] = (
    sla_remaining[
        "Normalized_Search_Name"
    ]
    .apply(is_species_level_name)
)


# -----------------------------------------------------------------------------
# 6. Identify canonical-name variants sharing one normalized search name
#
# This prevents duplicated acquisition effort for historical spellings and
# orthographic variants.
# -----------------------------------------------------------------------------

normalized_name_counts = (
    sla_remaining
    .groupby(
        "Normalized_Search_Name",
        dropna=False
    )
    .size()
    .rename(
        "Canonical_Variant_Count"
    )
    .reset_index()
)

sla_rescue_manifest = (
    sla_remaining
    .merge(
        normalized_name_counts,
        on="Normalized_Search_Name",
        how="left"
    )
)


variant_groups = (
    sla_rescue_manifest
    .groupby(
        "Normalized_Search_Name",
        dropna=False
    )["Canonical_Species"]
    .agg(
        lambda x: " | ".join(
            sorted(
                set(
                    x.dropna().astype(str)
                )
            )
        )
    )
    .rename(
        "Canonical_Variants"
    )
    .reset_index()
)

sla_rescue_manifest = (
    sla_rescue_manifest
    .merge(
        variant_groups,
        on="Normalized_Search_Name",
        how="left"
    )
)


# -----------------------------------------------------------------------------
# 7. Flag special cases
# -----------------------------------------------------------------------------

sla_rescue_manifest[
    "Existing_Special_Evidence"
] = "NONE"


# Carica has a BIEN genus-only single record requiring explicit review.
carica_mask = (
    sla_rescue_manifest[
        "Normalized_Acquisition_Genus"
    ]
    .eq("Carica")
)

sla_rescue_manifest.loc[
    carica_mask,
    "Existing_Special_Evidence"
] = (
    "BIEN_GENUS_ONLY_SINGLE_RECORD"
)


# Genus-level/unresolved taxa should not be treated as normal species searches.
unresolved_mask = ~sla_rescue_manifest[
    "Species_Level_Search_Available"
]

sla_rescue_manifest.loc[
    unresolved_mask,
    "Existing_Special_Evidence"
] = (
    "GENUS_LEVEL_OR_UNRESOLVED"
)


# -----------------------------------------------------------------------------
# 8. Assign rescue pathway
# -----------------------------------------------------------------------------

def assign_rescue_pathway(row):

    if (
        row["Existing_Special_Evidence"]
        == "BIEN_GENUS_ONLY_SINGLE_RECORD"
    ):
        return "SPECIAL_EVIDENCE_REVIEW"

    if not row["Species_Level_Search_Available"]:
        return "GENUS_OR_FAMILY_FALLBACK"

    if row["Canonical_Variant_Count"] > 1:
        return "NORMALIZED_SPECIES_TARGETED_SEARCH"

    return "SPECIES_TARGETED_SEARCH"


sla_rescue_manifest[
    "Rescue_Pathway"
] = (
    sla_rescue_manifest
    .apply(
        assign_rescue_pathway,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 9. Assign rescue priority
#
# Priority order:
#   1. Existing special evidence
#   2. Species names representing multiple canonical variants
#   3. Other normalized species-level searches
#   4. Genus/unresolved taxa
# -----------------------------------------------------------------------------

def assign_rescue_priority(row):

    pathway = row["Rescue_Pathway"]

    if pathway == "SPECIAL_EVIDENCE_REVIEW":
        return 1

    if pathway == "NORMALIZED_SPECIES_TARGETED_SEARCH":
        return 2

    if pathway == "SPECIES_TARGETED_SEARCH":
        return 3

    return 4


sla_rescue_manifest[
    "Rescue_Priority"
] = (
    sla_rescue_manifest
    .apply(
        assign_rescue_priority,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 10. Define recommended search terms
# -----------------------------------------------------------------------------

sla_rescue_manifest[
    "Primary_Search_Term"
] = (
    sla_rescue_manifest[
        "Normalized_Search_Name"
    ]
)


sla_rescue_manifest[
    "Recommended_Search_String"
] = (
    '"'
    + sla_rescue_manifest[
        "Primary_Search_Term"
    ].fillna("")
    + '" '
    + '"specific leaf area" OR SLA OR '
    + '"leaf mass per area"'
)


# -----------------------------------------------------------------------------
# 11. Create unique normalized species search manifest
#
# Several Omo canonical strings refer to the same normalized species.
# Search each normalized species only once.
# -----------------------------------------------------------------------------

species_search_manifest = (
    sla_rescue_manifest[
        sla_rescue_manifest[
            "Species_Level_Search_Available"
        ]
    ]
    .sort_values(
        [
            "Rescue_Priority",
            "Normalized_Search_Name",
        ]
    )
    .drop_duplicates(
        subset=[
            "Normalized_Search_Name"
        ]
    )
    [
        [
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Canonical_Variant_Count",
            "Canonical_Variants",
            "Existing_Special_Evidence",
            "Rescue_Pathway",
            "Rescue_Priority",
            "Recommended_Search_String",
        ]
    ]
    .reset_index(drop=True)
)


species_search_manifest.insert(
    0,
    "SLA_Rescue_Query_ID",
    [
        f"OMO_SLA_RESCUE_{i:03d}"
        for i in range(
            1,
            len(species_search_manifest) + 1
        )
    ]
)


# -----------------------------------------------------------------------------
# 12. Identify taxa not eligible for normal species-level rescue
# -----------------------------------------------------------------------------

fallback_manifest = (
    sla_rescue_manifest[
        ~sla_rescue_manifest[
            "Species_Level_Search_Available"
        ]
    ]
    [
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Existing_Special_Evidence",
            "Rescue_Pathway",
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 13. Summary statistics
# -----------------------------------------------------------------------------

n_remaining = len(
    sla_rescue_manifest
)

n_species_searchable = int(
    sla_rescue_manifest[
        "Species_Level_Search_Available"
    ].sum()
)

n_unique_species_searches = len(
    species_search_manifest
)

n_variant_taxa = int(
    (
        sla_rescue_manifest[
            "Canonical_Variant_Count"
        ] > 1
    ).sum()
)

n_special_review = int(
    (
        sla_rescue_manifest[
            "Rescue_Pathway"
        ]
        == "SPECIAL_EVIDENCE_REVIEW"
    ).sum()
)

n_fallback = int(
    (
        sla_rescue_manifest[
            "Rescue_Pathway"
        ]
        == "GENUS_OR_FAMILY_FALLBACK"
    ).sum()
)


# -----------------------------------------------------------------------------
# 14. Export
# -----------------------------------------------------------------------------

rescue_manifest_path = (
    output_dir
    / "omo_sla_targeted_species_rescue_manifest.csv"
)

species_query_path = (
    output_dir
    / "omo_sla_unique_species_rescue_queries.csv"
)

fallback_path = (
    output_dir
    / "omo_sla_non_species_rescue_fallback_manifest.csv"
)


sla_rescue_manifest.to_csv(
    rescue_manifest_path,
    index=False
)

species_search_manifest.to_csv(
    species_query_path,
    index=False
)

fallback_manifest.to_csv(
    fallback_path,
    index=False
)


# -----------------------------------------------------------------------------
# 15. Reporting
# -----------------------------------------------------------------------------

print("\nTargeted rescue planning summary")
print("-" * 100)
print(
    f"Remaining canonical taxa              : "
    f"{n_remaining}"
)
print(
    f"Taxa eligible for species search      : "
    f"{n_species_searchable}"
)
print(
    f"Unique normalized species searches    : "
    f"{n_unique_species_searches}"
)
print(
    f"Taxa belonging to variant groups      : "
    f"{n_variant_taxa}"
)
print(
    f"Taxa requiring special evidence review: "
    f"{n_special_review}"
)
print(
    f"Genus/unresolved fallback taxa         : "
    f"{n_fallback}"
)


print("\nRescue pathway distribution")
print("-" * 100)

print(
    sla_rescue_manifest[
        "Rescue_Pathway"
    ]
    .value_counts()
    .to_string()
)


print("\nUnique targeted species search manifest")
print("-" * 100)

display(
    species_search_manifest
)


if len(fallback_manifest) > 0:

    print(
        "\nTaxa not eligible for ordinary "
        "species-level rescue"
    )
    print("-" * 100)

    display(
        fallback_manifest
    )


print("\nExports")
print("-" * 100)
print(
    f"Complete rescue manifest : "
    f"{rescue_manifest_path}"
)
print(
    f"Unique species queries   : "
    f"{species_query_path}"
)
print(
    f"Fallback manifest        : "
    f"{fallback_path}"
)


print(
    "\nSection 8.8L completed successfully."
)
```

## Section 8.8M — Targeted Literature SLA Rescue Acquisition Register

Following taxonomic normalization and consolidation of the remaining SLA gaps,
31 unique species-level search targets remain eligible for targeted empirical
rescue.

This stage searches peer-reviewed literature, recognized trait repositories,
institutional datasets, and other traceable scientific sources for direct
species-level measurements of specific leaf area (SLA). Search names follow the
normalized taxonomic identities established during the preceding workflow.

Each candidate measurement must retain:

- canonical Omo taxon or taxa represented;
- normalized taxonomic search name;
- reported SLA value;
- reported unit;
- standardized SLA value in mm² mg⁻¹;
- bibliographic or database source;
- study context where available;
- evidence type;
- and an explicit acceptance decision.

Species-level empirical evidence is preferred over genus-, family-, or global
fallback estimates. No numerical value is accepted solely because it falls
within an expected SLA range.

The 31 search targets are divided into manageable review batches so that each
candidate can be evaluated individually and transparently.

```python
# =============================================================================
# SECTION 8.8M1 — TARGETED LITERATURE SLA RESCUE ACQUISITION REGISTER
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import math

print("=" * 100)
print("SECTION 8.8M1 — TARGETED LITERATURE SLA RESCUE ACQUISITION REGISTER")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

query_path = (
    output_dir
    / "omo_sla_unique_species_rescue_queries.csv"
)


if not query_path.exists():
    raise FileNotFoundError(
        "Species rescue-query table not found:\n"
        f"{query_path}"
    )


# -----------------------------------------------------------------------------
# 2. Load validated 8.8L search targets
# -----------------------------------------------------------------------------

sla_rescue_queries = pd.read_csv(
    query_path,
    low_memory=False
)


assert len(sla_rescue_queries) == 31, (
    f"Expected 31 unique rescue searches, "
    f"found {len(sla_rescue_queries)}."
)


# -----------------------------------------------------------------------------
# 3. Create literature acquisition register
# -----------------------------------------------------------------------------

sla_literature_register = (
    sla_rescue_queries.copy()
)


sla_literature_register[
    "Preferred_Source_1"
] = "PEER_REVIEWED_LITERATURE"

sla_literature_register[
    "Preferred_Source_2"
] = "RECOGNIZED_TRAIT_DATABASE"

sla_literature_register[
    "Preferred_Source_3"
] = "SCIENTIFIC_DATA_REPOSITORY"

sla_literature_register[
    "Target_Trait"
] = "specific leaf area"

sla_literature_register[
    "Target_Unit"
] = "mm2 mg-1"

sla_literature_register[
    "Search_Status"
] = "READY_FOR_TARGETED_LITERATURE_SEARCH"

sla_literature_register[
    "Candidate_Record_Found"
] = False

sla_literature_register[
    "Candidate_SLA_Value"
] = np.nan

sla_literature_register[
    "Candidate_SLA_Unit"
] = pd.NA

sla_literature_register[
    "Standardized_SLA_Value"
] = np.nan

sla_literature_register[
    "Evidence_Source"
] = pd.NA

sla_literature_register[
    "Evidence_Reference"
] = pd.NA

sla_literature_register[
    "Evidence_Context"
] = pd.NA

sla_literature_register[
    "Evidence_Quality_Flag"
] = pd.NA

sla_literature_register[
    "Acceptance_Decision"
] = "PENDING_SEARCH"


# -----------------------------------------------------------------------------
# 4. Assign controlled search batches
#
# Six taxa per batch gives six manageable batches:
# 6 + 6 + 6 + 6 + 6 + 1 = 31
# -----------------------------------------------------------------------------

BATCH_SIZE = 6

sla_literature_register[
    "Search_Batch"
] = [
    math.floor(i / BATCH_SIZE) + 1
    for i in range(
        len(sla_literature_register)
    )
]


n_batches = (
    sla_literature_register[
        "Search_Batch"
    ].nunique()
)


# -----------------------------------------------------------------------------
# 5. Create compact batch manifest
# -----------------------------------------------------------------------------

batch_manifest = (
    sla_literature_register
    .groupby(
        "Search_Batch",
        as_index=False
    )
    .agg(
        Search_Target_Count=(
            "Normalized_Search_Name",
            "count"
        ),

        Search_Targets=(
            "Normalized_Search_Name",
            lambda x: " | ".join(
                x.astype(str)
            )
        ),
    )
)


# -----------------------------------------------------------------------------
# 6. Validation
# -----------------------------------------------------------------------------

assert (
    sla_literature_register[
        "Normalized_Search_Name"
    ].nunique()
    == 31
)

assert (
    batch_manifest[
        "Search_Target_Count"
    ].sum()
    == 31
)


# -----------------------------------------------------------------------------
# 7. Export
# -----------------------------------------------------------------------------

register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register.csv"
)

batch_path = (
    output_dir
    / "omo_sla_targeted_literature_search_batches.csv"
)


sla_literature_register.to_csv(
    register_path,
    index=False
)

batch_manifest.to_csv(
    batch_path,
    index=False
)


# -----------------------------------------------------------------------------
# 8. Diagnostics
# -----------------------------------------------------------------------------

print("\nTargeted literature acquisition register")
print("-" * 100)

print(
    f"Unique species-level targets : "
    f"{len(sla_literature_register)}"
)

print(
    f"Search batches              : "
    f"{n_batches}"
)


print("\nSearch-batch manifest")
print("-" * 100)

display(
    batch_manifest
)


print("\nBatch 1 targets")
print("-" * 100)

display(
    sla_literature_register.loc[
        sla_literature_register[
            "Search_Batch"
        ].eq(1),
        [
            "SLA_Rescue_Query_ID",
            "Normalized_Search_Name",
            "Canonical_Variants",
            "Rescue_Pathway",
            "Recommended_Search_String",
            "Search_Status",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Literature acquisition register : "
    f"{register_path}"
)

print(
    f"Search-batch manifest           : "
    f"{batch_path}"
)

print(
    "\nSection 8.8M1 completed successfully."
)
```

## Section 8.8M2 — Record and Accept Batch 1 Targeted Literature Evidence

Batch 1 of the targeted literature rescue search yielded usable species-level SLA
evidence for three normalized taxa:

- *Carica papaya*
- *Elaeis guineensis*
- *Nauclea diderrichii*

The accepted values are retained as species-level literature evidence and remain
explicitly linked to their bibliographic provenance and contextual interpretation.
Where several historical Omo canonical names resolve to the same normalized
species, the same accepted species-level SLA value is propagated to each mapped
canonical taxon.

No value is assigned to *Cleistopholis patens*, *Lecaniodiscus cupanioides*, or
*Pavetta corymbosa* in this subsection because the first-pass search did not yield
a sufficiently traceable numerical species-level SLA measurement.

Species-level literature values remain higher priority than genus-, family-, or
global fallback estimates.

```python
# =============================================================================
# SECTION 8.8M2 — RECORD AND ACCEPT BATCH 1 TARGETED LITERATURE EVIDENCE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8M2 — RECORD AND ACCEPT BATCH 1 TARGETED LITERATURE EVIDENCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Confirm required objects
# -----------------------------------------------------------------------------

required_objects = [
    "sla_literature_register",
    "sla_rescue_manifest",
]

missing_objects = [
    obj
    for obj in required_objects
    if obj not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Required notebook objects are missing:\n"
        + "\n".join(
            f"  - {obj}"
            for obj in missing_objects
        )
        + "\n\nRun Sections 8.8L and 8.8M1 first."
    )


output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# 2. Define accepted Batch 1 literature evidence
#
# Values already standardized to mm2 mg-1.
# -----------------------------------------------------------------------------

batch1_accepted = pd.DataFrame(
    [
        {
            "Normalized_Search_Name": "Carica papaya",
            "Accepted_SLA_Value": 29.755,
            "Accepted_SLA_Unit": "mm2 mg-1",
            "Accepted_Provenance_Class": "SPECIES_LITERATURE",
            "Accepted_Source_Class": "PEER_REVIEWED_LITERATURE",
            "Accepted_Quality_Flag": "CONTEXT_QUALIFIED_SPECIES_LITERATURE",
            "Evidence_Context": (
                "Median of four reported treatment-specific SLA values "
                "for Carica papaya."
            ),
            "Acceptance_Decision": "ACCEPT_SPECIES_LITERATURE",
        },
        {
            "Normalized_Search_Name": "Elaeis guineensis",
            "Accepted_SLA_Value": 7.375,
            "Accepted_SLA_Unit": "mm2 mg-1",
            "Accepted_Provenance_Class": "SPECIES_LITERATURE",
            "Accepted_Source_Class": "PEER_REVIEWED_LITERATURE",
            "Accepted_Quality_Flag": "SPECIES_LITERATURE_MULTI_MEAN_SUPPORT",
            "Evidence_Context": (
                "Median of four directly reported site/genotype SLA means."
            ),
            "Acceptance_Decision": "ACCEPT_SPECIES_LITERATURE",
        },
        {
            "Normalized_Search_Name": "Nauclea diderrichii",
            "Accepted_SLA_Value": 26.000,
            "Accepted_SLA_Unit": "mm2 mg-1",
            "Accepted_Provenance_Class": "SPECIES_LITERATURE",
            "Accepted_Source_Class": "PEER_REVIEWED_LITERATURE",
            "Accepted_Quality_Flag": "EXPERIMENTAL_CONTEXT_SPECIES_LITERATURE",
            "Evidence_Context": (
                "Median of four treatment-specific SLA values reported "
                "for Nauclea diderrichii."
            ),
            "Acceptance_Decision": "ACCEPT_SPECIES_LITERATURE",
        },
    ]
)


# -----------------------------------------------------------------------------
# 3. Define unresolved Batch 1 searches
# -----------------------------------------------------------------------------

batch1_unresolved_names = [
    "Cleistopholis patens",
    "Lecaniodiscus cupanioides",
    "Pavetta corymbosa",
]


batch1_unresolved = pd.DataFrame(
    {
        "Normalized_Search_Name": batch1_unresolved_names,
        "Batch1_Search_Outcome": "NO_USABLE_VALUE_FIRST_PASS",
        "Next_Action": "CONTINUE_TO_LATER_FALLBACK",
    }
)


# -----------------------------------------------------------------------------
# 4. Validate Batch 1 partition
# -----------------------------------------------------------------------------

batch1_manifest = (
    sla_literature_register.loc[
        sla_literature_register[
            "Search_Batch"
        ].eq(1)
    ]
    .copy()
)


assert len(batch1_manifest) == 6, (
    f"Expected 6 Batch 1 search targets, found {len(batch1_manifest)}."
)


accepted_names = set(
    batch1_accepted[
        "Normalized_Search_Name"
    ]
)

unresolved_names = set(
    batch1_unresolved[
        "Normalized_Search_Name"
    ]
)

assert accepted_names.isdisjoint(
    unresolved_names
)

assert (
    accepted_names
    | unresolved_names
) == set(
    batch1_manifest[
        "Normalized_Search_Name"
    ]
)


# -----------------------------------------------------------------------------
# 5. Propagate accepted normalized species values to Omo canonical taxa
# -----------------------------------------------------------------------------

batch1_omo_values = (
    sla_rescue_manifest.loc[
        sla_rescue_manifest[
            "Normalized_Search_Name"
        ].isin(
            accepted_names
        )
    ]
    [
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Canonical_Variant_Count",
            "Canonical_Variants",
        ]
    ]
    .merge(
        batch1_accepted,
        how="left",
        on="Normalized_Search_Name",
        validate="many_to_one"
    )
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 6. Build Omo-ready accepted table
# -----------------------------------------------------------------------------

batch1_omo_values = (
    batch1_omo_values
    .rename(
        columns={
            "Accepted_SLA_Value": "Standardized_SLA_Value",
            "Accepted_SLA_Unit": "Standardized_SLA_Unit",
            "Accepted_Provenance_Class": "Provenance_Class",
            "Accepted_Source_Class": "Trait_Source",
            "Accepted_Quality_Flag": "Quality_Flag",
        }
    )
)


# -----------------------------------------------------------------------------
# 7. Expected new canonical-taxon coverage
#
# Carica papaya          -> 1
# Elaeis guineensis      -> 3 canonical variants
# Nauclea diderrichii    -> 3 canonical variants
# Total                  -> 7
# -----------------------------------------------------------------------------

n_new_batch1_taxa = len(
    batch1_omo_values
)

assert n_new_batch1_taxa == 7, (
    f"Expected 7 Omo canonical taxa from Batch 1 accepted values, "
    f"found {n_new_batch1_taxa}."
)


assert (
    batch1_omo_values[
        "Standardized_SLA_Value"
    ]
    .notna()
    .all()
)

assert (
    batch1_omo_values[
        "Standardized_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)

assert (
    batch1_omo_values[
        "Provenance_Class"
    ]
    .eq("SPECIES_LITERATURE")
    .all()
)


# -----------------------------------------------------------------------------
# 8. Update Batch 1 search-status register
# -----------------------------------------------------------------------------

sla_literature_register_updated = (
    sla_literature_register.copy()
)


for _, row in batch1_accepted.iterrows():

    mask = (
        sla_literature_register_updated[
            "Normalized_Search_Name"
        ].eq(
            row["Normalized_Search_Name"]
        )
    )

    sla_literature_register_updated.loc[
        mask,
        "Candidate_Record_Found"
    ] = True

    sla_literature_register_updated.loc[
        mask,
        "Standardized_SLA_Value"
    ] = row[
        "Accepted_SLA_Value"
    ]

    sla_literature_register_updated.loc[
        mask,
        "Candidate_SLA_Unit"
    ] = row[
        "Accepted_SLA_Unit"
    ]

    sla_literature_register_updated.loc[
        mask,
        "Evidence_Source"
    ] = row[
        "Accepted_Source_Class"
    ]

    sla_literature_register_updated.loc[
        mask,
        "Evidence_Context"
    ] = row[
        "Evidence_Context"
    ]

    sla_literature_register_updated.loc[
        mask,
        "Evidence_Quality_Flag"
    ] = row[
        "Accepted_Quality_Flag"
    ]

    sla_literature_register_updated.loc[
        mask,
        "Acceptance_Decision"
    ] = row[
        "Acceptance_Decision"
    ]

    sla_literature_register_updated.loc[
        mask,
        "Search_Status"
    ] = "COMPLETED_ACCEPTED"


for unresolved_name in batch1_unresolved_names:

    mask = (
        sla_literature_register_updated[
            "Normalized_Search_Name"
        ].eq(
            unresolved_name
        )
    )

    sla_literature_register_updated.loc[
        mask,
        "Candidate_Record_Found"
    ] = False

    sla_literature_register_updated.loc[
        mask,
        "Acceptance_Decision"
    ] = "NO_USABLE_VALUE_FIRST_PASS"

    sla_literature_register_updated.loc[
        mask,
        "Search_Status"
    ] = "COMPLETED_NO_USABLE_VALUE"


# -----------------------------------------------------------------------------
# 9. Recalculate current SLA coverage
#
# Coverage after 8.8K4 = 85
# Batch 1 adds 7 canonical taxa.
# -----------------------------------------------------------------------------

coverage_before_batch1 = 85

coverage_after_batch1 = (
    coverage_before_batch1
    + n_new_batch1_taxa
)

remaining_after_batch1 = (
    127
    - coverage_after_batch1
)


assert coverage_after_batch1 == 92
assert remaining_after_batch1 == 35


# -----------------------------------------------------------------------------
# 10. Export
# -----------------------------------------------------------------------------

accepted_evidence_path = (
    output_dir
    / "omo_sla_literature_batch1_accepted_species_evidence.csv"
)

accepted_omo_path = (
    output_dir
    / "omo_sla_literature_batch1_accepted_omo_values.csv"
)

unresolved_path = (
    output_dir
    / "omo_sla_literature_batch1_unresolved.csv"
)

updated_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch1.csv"
)


batch1_accepted.to_csv(
    accepted_evidence_path,
    index=False
)

batch1_omo_values.to_csv(
    accepted_omo_path,
    index=False
)

batch1_unresolved.to_csv(
    unresolved_path,
    index=False
)

sla_literature_register_updated.to_csv(
    updated_register_path,
    index=False
)


# -----------------------------------------------------------------------------
# 11. Diagnostics
# -----------------------------------------------------------------------------

print("\nBatch 1 literature acceptance summary")
print("-" * 100)

print(
    f"Batch 1 unique species searches : "
    f"{len(batch1_manifest)}"
)

print(
    f"Accepted species-level values   : "
    f"{len(batch1_accepted)}"
)

print(
    f"No usable first-pass values     : "
    f"{len(batch1_unresolved)}"
)

print(
    f"Omo canonical taxa newly covered: "
    f"{n_new_batch1_taxa}"
)


print("\nAccepted species-level literature evidence")
print("-" * 100)

display(
    batch1_accepted
)


print("\nOmo taxa receiving Batch 1 species-level literature SLA")
print("-" * 100)

display(
    batch1_omo_values[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "Quality_Flag",
            "Provenance_Class",
        ]
    ]
    .sort_values(
        [
            "Normalized_Search_Name",
            "Canonical_Species",
        ]
    )
)


print("\nBatch 1 unresolved targets")
print("-" * 100)

display(
    batch1_unresolved
)


print("\nCoverage after Batch 1 literature rescue")
print("-" * 100)

print(
    f"Coverage before Batch 1 : "
    f"{coverage_before_batch1} / 127"
)

print(
    f"New Batch 1 coverage    : "
    f"{n_new_batch1_taxa}"
)

print(
    f"Current SLA coverage    : "
    f"{coverage_after_batch1} / 127"
)

print(
    f"Coverage percent        : "
    f"{100 * coverage_after_batch1 / 127:.2f}%"
)

print(
    f"Remaining SLA gaps      : "
    f"{remaining_after_batch1}"
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted species evidence : "
    f"{accepted_evidence_path}"
)

print(
    f"Accepted Omo values       : "
    f"{accepted_omo_path}"
)

print(
    f"Unresolved Batch 1        : "
    f"{unresolved_path}"
)

print(
    f"Updated acquisition register : "
    f"{updated_register_path}"
)

print(
    "\nSection 8.8M2 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8M3 — RECORD AND ACCEPT BATCH 2 TARGETED LITERATURE EVIDENCE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8M3 — RECORD AND ACCEPT BATCH 2 TARGETED LITERATURE EVIDENCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Project paths
# -----------------------------------------------------------------------------

project_root = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = project_root / "outputs" / "functional_traits"

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


# -----------------------------------------------------------------------------
# 2. Load the literature acquisition register updated after Batch 1
# -----------------------------------------------------------------------------

batch1_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch1.csv"
)

if not batch1_register_path.exists():
    raise FileNotFoundError(
        "Updated Batch 1 literature acquisition register not found:\n"
        f"{batch1_register_path}"
    )

literature_register = pd.read_csv(
    batch1_register_path
)

required_columns = {
    "Normalized_Search_Name",
    "Canonical_Variants",
}

missing_columns = (
    required_columns
    - set(literature_register.columns)
)

if missing_columns:
    raise KeyError(
        "Required columns missing from literature acquisition register: "
        f"{sorted(missing_columns)}"
    )


# -----------------------------------------------------------------------------
# 3. Define Batch 2 targets
# -----------------------------------------------------------------------------

batch2_targets = [
    "Scottellia coriacea",
    "Uapaca heudelotii",
    "Afzelia africana",
    "Amphimas tetracoides",
    "Aningeria pierrei",
    "Anthocleista djalonensis",
]

batch2_register = literature_register[
    literature_register[
        "Normalized_Search_Name"
    ].isin(batch2_targets)
].copy()

found_targets = set(
    batch2_register[
        "Normalized_Search_Name"
    ].dropna()
)

missing_targets = (
    set(batch2_targets)
    - found_targets
)

if missing_targets:
    raise ValueError(
        "Some expected Batch 2 targets were not found in the "
        "literature acquisition register:\n"
        f"{sorted(missing_targets)}"
    )

if len(batch2_register) != 6:
    raise ValueError(
        "Expected exactly 6 Batch 2 species-level search targets, "
        f"but found {len(batch2_register)}."
    )


# -----------------------------------------------------------------------------
# 4. Record defensible Batch 2 literature evidence
#
# Afzelia africana:
#   Published species-level SLA = 193.4 cm2 g-1
#
# Unit equivalence:
#   1 cm2 g-1 = 0.1 mm2 mg-1
#
# Therefore:
#   193.4 cm2 g-1 = 19.34 mm2 mg-1
#
# The remaining five targets did not yield a sufficiently traceable numerical
# species-level SLA value during the controlled literature search.
# -----------------------------------------------------------------------------

batch2_evidence_records = [
    {
        "Normalized_Search_Name": "Afzelia africana",
        "Accepted_SLA_Value": 19.34,
        "Accepted_SLA_Unit": "mm2 mg-1",
        "Accepted_Provenance_Class": "SPECIES_LITERATURE",
        "Accepted_Source_Class": "PEER_REVIEWED_LITERATURE",
        "Accepted_Quality_Flag": "DIRECT_SPECIES_LITERATURE_VALUE",
        "Evidence_Context": (
            "Published mean species-level SLA reported as "
            "193.4 cm2 g-1; converted to 19.34 mm2 mg-1."
        ),
        "Acceptance_Decision": "ACCEPT_SPECIES_LITERATURE",
    }
]

batch2_accepted_species = pd.DataFrame(
    batch2_evidence_records
)


# -----------------------------------------------------------------------------
# 5. Define unresolved Batch 2 targets
# -----------------------------------------------------------------------------

batch2_unresolved_names = [
    "Scottellia coriacea",
    "Uapaca heudelotii",
    "Amphimas tetracoides",
    "Aningeria pierrei",
    "Anthocleista djalonensis",
]

batch2_unresolved = batch2_register[
    batch2_register[
        "Normalized_Search_Name"
    ].isin(batch2_unresolved_names)
][
    [
        "Normalized_Search_Name",
        "Canonical_Variants",
    ]
].copy()

batch2_unresolved[
    "Batch2_Search_Outcome"
] = "NO_USABLE_VALUE_FIRST_PASS"

batch2_unresolved[
    "Next_Action"
] = "CONTINUE_TO_LATER_FALLBACK"


# -----------------------------------------------------------------------------
# 6. Map accepted species evidence back to Omo canonical taxa
# -----------------------------------------------------------------------------

accepted_lookup = (
    batch2_accepted_species
    .set_index("Normalized_Search_Name")
    .to_dict("index")
)

omo_batch2_rows = []

for _, row in batch2_register.iterrows():

    search_name = row["Normalized_Search_Name"]

    if search_name not in accepted_lookup:
        continue

    evidence = accepted_lookup[
        search_name
    ]

    canonical_variants = str(
        row["Canonical_Variants"]
    ).split("|")

    canonical_variants = [
        x.strip()
        for x in canonical_variants
        if x.strip()
    ]

    for canonical_species in canonical_variants:

        omo_batch2_rows.append(
            {
                "Canonical_Species":
                    canonical_species,

                "Normalized_Search_Name":
                    search_name,

                "Standardized_SLA_Value":
                    evidence[
                        "Accepted_SLA_Value"
                    ],

                "Standardized_SLA_Unit":
                    evidence[
                        "Accepted_SLA_Unit"
                    ],

                "Quality_Flag":
                    evidence[
                        "Accepted_Quality_Flag"
                    ],

                "Provenance_Class":
                    evidence[
                        "Accepted_Provenance_Class"
                    ],
            }
        )

omo_batch2_values = pd.DataFrame(
    omo_batch2_rows
)

if omo_batch2_values.empty:
    raise RuntimeError(
        "No Omo canonical taxa were mapped from the accepted "
        "Batch 2 literature evidence."
    )


# -----------------------------------------------------------------------------
# 7. Validate accepted Batch 2 evidence
# -----------------------------------------------------------------------------

if batch2_accepted_species[
    "Normalized_Search_Name"
].duplicated().any():

    raise ValueError(
        "Duplicate normalized species names detected in "
        "Batch 2 accepted evidence."
    )

if omo_batch2_values[
    "Canonical_Species"
].duplicated().any():

    duplicate_taxa = (
        omo_batch2_values.loc[
            omo_batch2_values[
                "Canonical_Species"
            ].duplicated(keep=False),
            "Canonical_Species"
        ]
        .unique()
        .tolist()
    )

    raise ValueError(
        "Duplicate Omo canonical taxa generated from Batch 2: "
        f"{duplicate_taxa}"
    )


# -----------------------------------------------------------------------------
# 8. Update acquisition register
# -----------------------------------------------------------------------------

updated_register = (
    literature_register.copy()
)

if "Search_Status" not in updated_register.columns:
    updated_register[
        "Search_Status"
    ] = pd.NA

if "Accepted_SLA_Value" not in updated_register.columns:
    updated_register[
        "Accepted_SLA_Value"
    ] = np.nan

if "Accepted_SLA_Unit" not in updated_register.columns:
    updated_register[
        "Accepted_SLA_Unit"
    ] = pd.NA

if "Literature_Acquisition_Decision" not in updated_register.columns:
    updated_register[
        "Literature_Acquisition_Decision"
    ] = pd.NA


# Accepted species
for _, evidence in batch2_accepted_species.iterrows():

    mask = (
        updated_register[
            "Normalized_Search_Name"
        ]
        == evidence[
            "Normalized_Search_Name"
        ]
    )

    updated_register.loc[
        mask,
        "Search_Status"
    ] = "SPECIES_LITERATURE_ACCEPTED"

    updated_register.loc[
        mask,
        "Accepted_SLA_Value"
    ] = evidence[
        "Accepted_SLA_Value"
    ]

    updated_register.loc[
        mask,
        "Accepted_SLA_Unit"
    ] = evidence[
        "Accepted_SLA_Unit"
    ]

    updated_register.loc[
        mask,
        "Literature_Acquisition_Decision"
    ] = "ACCEPT_SPECIES_LITERATURE"


# Unresolved species
for species_name in batch2_unresolved_names:

    mask = (
        updated_register[
            "Normalized_Search_Name"
        ]
        == species_name
    )

    updated_register.loc[
        mask,
        "Search_Status"
    ] = "NO_USABLE_VALUE_FIRST_PASS"

    updated_register.loc[
        mask,
        "Literature_Acquisition_Decision"
    ] = "CONTINUE_TO_LATER_FALLBACK"


# -----------------------------------------------------------------------------
# 9. Coverage accounting
#
# Locked state after Batch 1:
#     92 / 127
#
# Afzelia africana represents one canonical Omo taxon.
# -----------------------------------------------------------------------------

coverage_before_batch2 = 92

new_batch2_taxa = (
    omo_batch2_values[
        "Canonical_Species"
    ].nunique()
)

current_coverage = (
    coverage_before_batch2
    + new_batch2_taxa
)

total_taxa = 127

remaining_gaps = (
    total_taxa
    - current_coverage
)

coverage_percent = (
    current_coverage
    / total_taxa
    * 100
)

if current_coverage != 93:
    raise ValueError(
        "Unexpected Batch 2 coverage result. "
        f"Expected 93/127 but obtained "
        f"{current_coverage}/127."
    )


# -----------------------------------------------------------------------------
# 10. Export
# -----------------------------------------------------------------------------

accepted_species_path = (
    output_dir
    / "omo_sla_literature_batch2_accepted_species_evidence.csv"
)

accepted_omo_path = (
    output_dir
    / "omo_sla_literature_batch2_accepted_omo_values.csv"
)

unresolved_path = (
    output_dir
    / "omo_sla_literature_batch2_unresolved.csv"
)

updated_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch2.csv"
)

batch2_accepted_species.to_csv(
    accepted_species_path,
    index=False
)

omo_batch2_values.to_csv(
    accepted_omo_path,
    index=False
)

batch2_unresolved.to_csv(
    unresolved_path,
    index=False
)

updated_register.to_csv(
    updated_register_path,
    index=False
)


# -----------------------------------------------------------------------------
# 11. Report
# -----------------------------------------------------------------------------

print("\nBatch 2 literature acceptance summary")
print("-" * 100)

print(
    f"Batch 2 unique species searches : "
    f"{len(batch2_register)}"
)

print(
    f"Accepted species-level values   : "
    f"{len(batch2_accepted_species)}"
)

print(
    f"No usable first-pass values     : "
    f"{len(batch2_unresolved)}"
)

print(
    f"Omo canonical taxa newly covered: "
    f"{new_batch2_taxa}"
)


print("\nAccepted species-level literature evidence")
print("-" * 100)

display(
    batch2_accepted_species
)


print("\nOmo taxa receiving Batch 2 species-level literature SLA")
print("-" * 100)

display(
    omo_batch2_values
)


print("\nBatch 2 unresolved targets")
print("-" * 100)

display(
    batch2_unresolved[
        [
            "Normalized_Search_Name",
            "Batch2_Search_Outcome",
            "Next_Action",
        ]
    ]
)


print("\nCoverage after Batch 2 literature rescue")
print("-" * 100)

print(
    f"Coverage before Batch 2 : "
    f"{coverage_before_batch2} / {total_taxa}"
)

print(
    f"New Batch 2 coverage    : "
    f"{new_batch2_taxa}"
)

print(
    f"Current SLA coverage    : "
    f"{current_coverage} / {total_taxa}"
)

print(
    f"Coverage percent        : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps      : "
    f"{remaining_gaps}"
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted species evidence : "
    f"{accepted_species_path}"
)

print(
    f"Accepted Omo values       : "
    f"{accepted_omo_path}"
)

print(
    f"Unresolved Batch 2        : "
    f"{unresolved_path}"
)

print(
    f"Updated acquisition register : "
    f"{updated_register_path}"
)


print(
    "\nSection 8.8M3 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8M4 — RECORD AND ACCEPT BATCH 3 TARGETED LITERATURE EVIDENCE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8M4 — RECORD AND ACCEPT BATCH 3 TARGETED LITERATURE EVIDENCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Project paths
# -----------------------------------------------------------------------------

project_root = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    project_root
    / "outputs"
    / "functional_traits"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


# -----------------------------------------------------------------------------
# 2. Load acquisition register updated after Batch 2
# -----------------------------------------------------------------------------

batch2_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch2.csv"
)

if not batch2_register_path.exists():
    raise FileNotFoundError(
        "Updated Batch 2 literature acquisition register not found:\n"
        f"{batch2_register_path}"
    )

literature_register = pd.read_csv(
    batch2_register_path
)


# -----------------------------------------------------------------------------
# 3. Define Batch 3 targets
# -----------------------------------------------------------------------------

batch3_targets = [
    "Anthocleista vogelii",
    "Anthonotha macrophylla",
    "Baphia nitida",
    "Bombax buonopozense",
    "Bridelia ferruginea",
    "Cola nitida",
]

batch3_register = (
    literature_register.loc[
        literature_register[
            "Normalized_Search_Name"
        ].isin(batch3_targets)
    ]
    .copy()
)


assert len(batch3_register) == 6, (
    f"Expected 6 Batch 3 targets, "
    f"found {len(batch3_register)}."
)


found_targets = set(
    batch3_register[
        "Normalized_Search_Name"
    ]
)

missing_targets = (
    set(batch3_targets)
    - found_targets
)

assert not missing_targets, (
    f"Missing Batch 3 targets: "
    f"{sorted(missing_targets)}"
)


# -----------------------------------------------------------------------------
# 4. Accepted Batch 3 literature evidence
#
# Bombax buonopozense:
#   Published median SLA = 255.46 cm2 g-1
#
# Unit conversion:
#   255.46 cm2 g-1 = 25.546 mm2 mg-1
# -----------------------------------------------------------------------------

batch3_accepted_species = pd.DataFrame(
    [
        {
            "Normalized_Search_Name":
                "Bombax buonopozense",

            "Accepted_SLA_Value":
                25.546,

            "Accepted_SLA_Unit":
                "mm2 mg-1",

            "Accepted_Provenance_Class":
                "SPECIES_LITERATURE",

            "Accepted_Source_Class":
                "PEER_REVIEWED_LITERATURE",

            "Accepted_Quality_Flag":
                "DIRECT_SPECIES_LITERATURE_MEDIAN",

            "Evidence_Context":
                (
                    "Published species-level median SLA reported as "
                    "255.46 cm2 g-1; converted to 25.546 mm2 mg-1."
                ),

            "Acceptance_Decision":
                "ACCEPT_SPECIES_LITERATURE",
        }
    ]
)


# -----------------------------------------------------------------------------
# 5. Unresolved Batch 3 targets
# -----------------------------------------------------------------------------

batch3_unresolved_names = [
    "Anthocleista vogelii",
    "Anthonotha macrophylla",
    "Baphia nitida",
    "Bridelia ferruginea",
    "Cola nitida",
]


batch3_unresolved = (
    batch3_register.loc[
        batch3_register[
            "Normalized_Search_Name"
        ].isin(batch3_unresolved_names),
        [
            "Normalized_Search_Name",
            "Canonical_Variants",
        ]
    ]
    .copy()
)


batch3_unresolved[
    "Batch3_Search_Outcome"
] = "NO_USABLE_VALUE_FIRST_PASS"

batch3_unresolved[
    "Next_Action"
] = "CONTINUE_TO_LATER_FALLBACK"


# -----------------------------------------------------------------------------
# 6. Validate Batch 3 partition
# -----------------------------------------------------------------------------

accepted_names = set(
    batch3_accepted_species[
        "Normalized_Search_Name"
    ]
)

unresolved_names = set(
    batch3_unresolved[
        "Normalized_Search_Name"
    ]
)

assert accepted_names.isdisjoint(
    unresolved_names
)

assert (
    accepted_names
    | unresolved_names
) == set(batch3_targets)


# -----------------------------------------------------------------------------
# 7. Propagate accepted normalized species value to canonical Omo taxa
# -----------------------------------------------------------------------------

batch3_omo_rows = []

for _, evidence in batch3_accepted_species.iterrows():

    search_name = (
        evidence[
            "Normalized_Search_Name"
        ]
    )

    matching_rows = (
        batch3_register.loc[
            batch3_register[
                "Normalized_Search_Name"
            ].eq(search_name)
        ]
    )

    if matching_rows.empty:
        raise RuntimeError(
            f"No acquisition-register row found for {search_name}."
        )

    for _, row in matching_rows.iterrows():

        variants = str(
            row["Canonical_Variants"]
        ).split("|")

        variants = [
            v.strip()
            for v in variants
            if v.strip()
        ]

        for canonical_species in variants:

            batch3_omo_rows.append(
                {
                    "Canonical_Species":
                        canonical_species,

                    "Normalized_Search_Name":
                        search_name,

                    "Standardized_SLA_Value":
                        evidence[
                            "Accepted_SLA_Value"
                        ],

                    "Standardized_SLA_Unit":
                        evidence[
                            "Accepted_SLA_Unit"
                        ],

                    "Quality_Flag":
                        evidence[
                            "Accepted_Quality_Flag"
                        ],

                    "Provenance_Class":
                        evidence[
                            "Accepted_Provenance_Class"
                        ],
                }
            )


batch3_omo_values = pd.DataFrame(
    batch3_omo_rows
)


# -----------------------------------------------------------------------------
# 8. Validate new canonical-taxon coverage
# -----------------------------------------------------------------------------

n_new_batch3_taxa = (
    batch3_omo_values[
        "Canonical_Species"
    ]
    .nunique()
)

assert n_new_batch3_taxa == 1, (
    f"Expected 1 newly covered Omo taxon, "
    f"found {n_new_batch3_taxa}."
)


assert (
    batch3_omo_values[
        "Standardized_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)

assert (
    batch3_omo_values[
        "Provenance_Class"
    ]
    .eq("SPECIES_LITERATURE")
    .all()
)


# -----------------------------------------------------------------------------
# 9. Update literature acquisition register
# -----------------------------------------------------------------------------

updated_register = (
    literature_register.copy()
)


required_update_columns = [
    "Search_Status",
    "Accepted_SLA_Value",
    "Accepted_SLA_Unit",
    "Literature_Acquisition_Decision",
]

for col in required_update_columns:

    if col not in updated_register.columns:

        if col == "Accepted_SLA_Value":
            updated_register[col] = np.nan
        else:
            updated_register[col] = pd.NA


# Accepted species
for _, evidence in batch3_accepted_species.iterrows():

    mask = (
        updated_register[
            "Normalized_Search_Name"
        ]
        .eq(
            evidence[
                "Normalized_Search_Name"
            ]
        )
    )

    updated_register.loc[
        mask,
        "Search_Status"
    ] = "SPECIES_LITERATURE_ACCEPTED"

    updated_register.loc[
        mask,
        "Accepted_SLA_Value"
    ] = evidence[
        "Accepted_SLA_Value"
    ]

    updated_register.loc[
        mask,
        "Accepted_SLA_Unit"
    ] = evidence[
        "Accepted_SLA_Unit"
    ]

    updated_register.loc[
        mask,
        "Literature_Acquisition_Decision"
    ] = "ACCEPT_SPECIES_LITERATURE"


# Unresolved species
for species_name in batch3_unresolved_names:

    mask = (
        updated_register[
            "Normalized_Search_Name"
        ].eq(species_name)
    )

    updated_register.loc[
        mask,
        "Search_Status"
    ] = "NO_USABLE_VALUE_FIRST_PASS"

    updated_register.loc[
        mask,
        "Literature_Acquisition_Decision"
    ] = "CONTINUE_TO_LATER_FALLBACK"


# -----------------------------------------------------------------------------
# 10. Coverage accounting
#
# Locked state after Batch 2:
#     93 / 127
#
# Batch 3 adds Bombax buonopozense:
#      1 canonical taxon
# -----------------------------------------------------------------------------

coverage_before_batch3 = 93

coverage_after_batch3 = (
    coverage_before_batch3
    + n_new_batch3_taxa
)

remaining_after_batch3 = (
    127
    - coverage_after_batch3
)

coverage_percent = (
    100
    * coverage_after_batch3
    / 127
)


assert coverage_after_batch3 == 94
assert remaining_after_batch3 == 33


# -----------------------------------------------------------------------------
# 11. Export
# -----------------------------------------------------------------------------

accepted_species_path = (
    output_dir
    / "omo_sla_literature_batch3_accepted_species_evidence.csv"
)

accepted_omo_path = (
    output_dir
    / "omo_sla_literature_batch3_accepted_omo_values.csv"
)

unresolved_path = (
    output_dir
    / "omo_sla_literature_batch3_unresolved.csv"
)

updated_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch3.csv"
)


batch3_accepted_species.to_csv(
    accepted_species_path,
    index=False
)

batch3_omo_values.to_csv(
    accepted_omo_path,
    index=False
)

batch3_unresolved.to_csv(
    unresolved_path,
    index=False
)

updated_register.to_csv(
    updated_register_path,
    index=False
)


# -----------------------------------------------------------------------------
# 12. Diagnostics
# -----------------------------------------------------------------------------

print("\nBatch 3 literature acceptance summary")
print("-" * 100)

print(
    f"Batch 3 unique species searches : "
    f"{len(batch3_register)}"
)

print(
    f"Accepted species-level values   : "
    f"{len(batch3_accepted_species)}"
)

print(
    f"No usable first-pass values     : "
    f"{len(batch3_unresolved)}"
)

print(
    f"Omo canonical taxa newly covered: "
    f"{n_new_batch3_taxa}"
)


print("\nAccepted species-level literature evidence")
print("-" * 100)

display(
    batch3_accepted_species
)


print("\nOmo taxa receiving Batch 3 species-level literature SLA")
print("-" * 100)

display(
    batch3_omo_values
)


print("\nBatch 3 unresolved targets")
print("-" * 100)

display(
    batch3_unresolved[
        [
            "Normalized_Search_Name",
            "Batch3_Search_Outcome",
            "Next_Action",
        ]
    ]
)


print("\nCoverage after Batch 3 literature rescue")
print("-" * 100)

print(
    f"Coverage before Batch 3 : "
    f"{coverage_before_batch3} / 127"
)

print(
    f"New Batch 3 coverage    : "
    f"{n_new_batch3_taxa}"
)

print(
    f"Current SLA coverage    : "
    f"{coverage_after_batch3} / 127"
)

print(
    f"Coverage percent        : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps      : "
    f"{remaining_after_batch3}"
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted species evidence : "
    f"{accepted_species_path}"
)

print(
    f"Accepted Omo values       : "
    f"{accepted_omo_path}"
)

print(
    f"Unresolved Batch 3        : "
    f"{unresolved_path}"
)

print(
    f"Updated acquisition register : "
    f"{updated_register_path}"
)


print(
    "\nSection 8.8M4 completed successfully."
)
```

## Section 8.8M5 — Record Batch 4 Literature Search Outcomes

Batch 4 of the targeted literature rescue included six normalized species:

- *Funtumia africana*
- *Funtumia elastica*
- *Hexalobus crispiflorus*
- *Hildegardia barteri*
- *Maesobotrya barteri*
- *Malacantha alnifolia*

Relevant SLA or functional-trait evidence was identified for *Funtumia africana*
and *Funtumia elastica*, but the accessible sources did not provide a sufficiently
complete and traceable numerical species-level SLA value suitable for acceptance.

For the remaining four species, no defensible numerical SLA value was recovered
during the first-pass literature search.

No new SLA values are therefore incorporated in this subsection. All six taxa
remain unresolved and continue to later fallback acquisition stages, while the
presence of partial evidence is retained explicitly for provenance.

```python
# =============================================================================
# SECTION 8.8M5 — RECORD BATCH 4 LITERATURE SEARCH OUTCOMES
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8M5 — RECORD BATCH 4 LITERATURE SEARCH OUTCOMES")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Project paths
# -----------------------------------------------------------------------------

project_root = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    project_root
    / "outputs"
    / "functional_traits"
)

output_dir.mkdir(
    parents=True,
    exist_ok=True
)


# -----------------------------------------------------------------------------
# 2. Load acquisition register updated after Batch 3
# -----------------------------------------------------------------------------

batch3_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch3.csv"
)

if not batch3_register_path.exists():
    raise FileNotFoundError(
        "Updated Batch 3 literature acquisition register not found:\n"
        f"{batch3_register_path}"
    )

literature_register = pd.read_csv(
    batch3_register_path
)


# -----------------------------------------------------------------------------
# 3. Define Batch 4 targets
# -----------------------------------------------------------------------------

batch4_targets = [
    "Funtumia africana",
    "Funtumia elastica",
    "Hexalobus crispiflorus",
    "Hildegardia barteri",
    "Maesobotrya barteri",
    "Malacantha alnifolia",
]

batch4_register = (
    literature_register.loc[
        literature_register[
            "Normalized_Search_Name"
        ].isin(batch4_targets)
    ]
    .copy()
)


assert len(batch4_register) == 6, (
    f"Expected 6 Batch 4 targets, "
    f"found {len(batch4_register)}."
)


found_targets = set(
    batch4_register[
        "Normalized_Search_Name"
    ]
)

missing_targets = (
    set(batch4_targets)
    - found_targets
)

assert not missing_targets, (
    f"Missing Batch 4 targets: "
    f"{sorted(missing_targets)}"
)


# -----------------------------------------------------------------------------
# 4. Define literature outcomes
# -----------------------------------------------------------------------------

batch4_outcomes = pd.DataFrame(
    [
        {
            "Normalized_Search_Name":
                "Funtumia africana",

            "Batch4_Search_Outcome":
                "EVIDENCE_FOUND_NUMERIC_VALUE_NOT_ACCEPTED",

            "Evidence_Status":
                "DIRECT_SLA_STUDY_FOUND",

            "Evidence_Context":
                (
                    "A controlled study measured SLA directly across "
                    "soil-moisture treatments, but the accessible source "
                    "did not expose a complete numerical set suitable for "
                    "a defensible central estimate."
                ),

            "Acceptance_Decision":
                "NO_VALUE_ACCEPTED",

            "Next_Action":
                "CONTINUE_TO_LATER_FALLBACK",
        },

        {
            "Normalized_Search_Name":
                "Funtumia elastica",

            "Batch4_Search_Outcome":
                "EVIDENCE_FOUND_NUMERIC_VALUE_NOT_ACCEPTED",

            "Evidence_Status":
                "FUNCTIONAL_TRAIT_STUDY_FOUND",

            "Evidence_Context":
                (
                    "Functional-trait studies including the species were "
                    "located, but no clean traceable numerical species-level "
                    "SLA value was recoverable from the accessible material."
                ),

            "Acceptance_Decision":
                "NO_VALUE_ACCEPTED",

            "Next_Action":
                "CONTINUE_TO_LATER_FALLBACK",
        },

        {
            "Normalized_Search_Name":
                "Hexalobus crispiflorus",

            "Batch4_Search_Outcome":
                "NO_USABLE_VALUE_FIRST_PASS",

            "Evidence_Status":
                "NO_DEFENSIBLE_NUMERICAL_SLA",

            "Evidence_Context":
                (
                    "No traceable numerical species-level SLA value "
                    "recovered in the first-pass search."
                ),

            "Acceptance_Decision":
                "NO_VALUE_ACCEPTED",

            "Next_Action":
                "CONTINUE_TO_LATER_FALLBACK",
        },

        {
            "Normalized_Search_Name":
                "Hildegardia barteri",

            "Batch4_Search_Outcome":
                "NO_USABLE_VALUE_FIRST_PASS",

            "Evidence_Status":
                "NO_DEFENSIBLE_NUMERICAL_SLA",

            "Evidence_Context":
                (
                    "No traceable numerical species-level SLA value "
                    "recovered in the first-pass search."
                ),

            "Acceptance_Decision":
                "NO_VALUE_ACCEPTED",

            "Next_Action":
                "CONTINUE_TO_LATER_FALLBACK",
        },

        {
            "Normalized_Search_Name":
                "Maesobotrya barteri",

            "Batch4_Search_Outcome":
                "NO_USABLE_VALUE_FIRST_PASS",

            "Evidence_Status":
                "NO_DEFENSIBLE_NUMERICAL_SLA",

            "Evidence_Context":
                (
                    "No traceable numerical species-level SLA value "
                    "recovered in the first-pass search."
                ),

            "Acceptance_Decision":
                "NO_VALUE_ACCEPTED",

            "Next_Action":
                "CONTINUE_TO_LATER_FALLBACK",
        },

        {
            "Normalized_Search_Name":
                "Malacantha alnifolia",

            "Batch4_Search_Outcome":
                "NO_USABLE_VALUE_FIRST_PASS",

            "Evidence_Status":
                "NO_DEFENSIBLE_NUMERICAL_SLA",

            "Evidence_Context":
                (
                    "No traceable numerical species-level SLA value "
                    "recovered in the first-pass search."
                ),

            "Acceptance_Decision":
                "NO_VALUE_ACCEPTED",

            "Next_Action":
                "CONTINUE_TO_LATER_FALLBACK",
        },
    ]
)


# -----------------------------------------------------------------------------
# 5. Validate Batch 4 outcome table
# -----------------------------------------------------------------------------

assert len(batch4_outcomes) == 6

assert (
    set(
        batch4_outcomes[
            "Normalized_Search_Name"
        ]
    )
    == set(batch4_targets)
)


# -----------------------------------------------------------------------------
# 6. Attach canonical variants for traceability
# -----------------------------------------------------------------------------

batch4_outcomes = (
    batch4_outcomes
    .merge(
        batch4_register[
            [
                "Normalized_Search_Name",
                "Canonical_Variants",
                "Normalized_Acquisition_Genus",
            ]
        ],
        how="left",
        on="Normalized_Search_Name",
        validate="one_to_one"
    )
)


# -----------------------------------------------------------------------------
# 7. Update acquisition register
# -----------------------------------------------------------------------------

updated_register = (
    literature_register.copy()
)


required_update_columns = [
    "Search_Status",
    "Literature_Acquisition_Decision",
    "Evidence_Context",
    "Evidence_Quality_Flag",
]

for col in required_update_columns:

    if col not in updated_register.columns:
        updated_register[col] = pd.NA


for _, outcome in batch4_outcomes.iterrows():

    mask = (
        updated_register[
            "Normalized_Search_Name"
        ].eq(
            outcome[
                "Normalized_Search_Name"
            ]
        )
    )

    updated_register.loc[
        mask,
        "Search_Status"
    ] = outcome[
        "Batch4_Search_Outcome"
    ]

    updated_register.loc[
        mask,
        "Literature_Acquisition_Decision"
    ] = outcome[
        "Acceptance_Decision"
    ]

    updated_register.loc[
        mask,
        "Evidence_Context"
    ] = outcome[
        "Evidence_Context"
    ]

    updated_register.loc[
        mask,
        "Evidence_Quality_Flag"
    ] = outcome[
        "Evidence_Status"
    ]


# -----------------------------------------------------------------------------
# 8. Coverage accounting
#
# No values accepted in Batch 4.
# Coverage remains 94 / 127.
# -----------------------------------------------------------------------------

coverage_before_batch4 = 94

new_batch4_taxa = 0

coverage_after_batch4 = (
    coverage_before_batch4
    + new_batch4_taxa
)

remaining_after_batch4 = (
    127
    - coverage_after_batch4
)

coverage_percent = (
    100
    * coverage_after_batch4
    / 127
)


assert coverage_after_batch4 == 94
assert remaining_after_batch4 == 33


# -----------------------------------------------------------------------------
# 9. Summary counts
# -----------------------------------------------------------------------------

n_partial_evidence = (
    batch4_outcomes[
        "Batch4_Search_Outcome"
    ]
    .eq(
        "EVIDENCE_FOUND_NUMERIC_VALUE_NOT_ACCEPTED"
    )
    .sum()
)

n_no_value = (
    batch4_outcomes[
        "Batch4_Search_Outcome"
    ]
    .eq(
        "NO_USABLE_VALUE_FIRST_PASS"
    )
    .sum()
)


# -----------------------------------------------------------------------------
# 10. Export
# -----------------------------------------------------------------------------

outcomes_path = (
    output_dir
    / "omo_sla_literature_batch4_search_outcomes.csv"
)

updated_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch4.csv"
)


batch4_outcomes.to_csv(
    outcomes_path,
    index=False
)

updated_register.to_csv(
    updated_register_path,
    index=False
)


# -----------------------------------------------------------------------------
# 11. Diagnostics
# -----------------------------------------------------------------------------

print("\nBatch 4 literature search summary")
print("-" * 100)

print(
    f"Batch 4 unique species searches          : "
    f"{len(batch4_register)}"
)

print(
    f"Accepted species-level values            : "
    f"{new_batch4_taxa}"
)

print(
    f"Evidence found but value not accepted    : "
    f"{n_partial_evidence}"
)

print(
    f"No usable numerical value first pass     : "
    f"{n_no_value}"
)


print("\nBatch 4 search outcomes")
print("-" * 100)

display(
    batch4_outcomes[
        [
            "Normalized_Search_Name",
            "Canonical_Variants",
            "Batch4_Search_Outcome",
            "Evidence_Status",
            "Acceptance_Decision",
            "Next_Action",
        ]
    ]
)


print("\nCoverage after Batch 4 literature rescue")
print("-" * 100)

print(
    f"Coverage before Batch 4 : "
    f"{coverage_before_batch4} / 127"
)

print(
    f"New Batch 4 coverage    : "
    f"{new_batch4_taxa}"
)

print(
    f"Current SLA coverage    : "
    f"{coverage_after_batch4} / 127"
)

print(
    f"Coverage percent        : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps      : "
    f"{remaining_after_batch4}"
)


print("\nExports")
print("-" * 100)

print(
    f"Batch 4 search outcomes : "
    f"{outcomes_path}"
)

print(
    f"Updated acquisition register : "
    f"{updated_register_path}"
)


print(
    "\nSection 8.8M5 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8M6 — RECORD AND ACCEPT BATCH 5 TARGETED LITERATURE EVIDENCE
# =============================================================================
#
# Purpose:
#   Record the results of the fifth targeted species-level literature search
#   for residual SLA gaps.
#
#   Batch 5 targets:
#       - Millettia thonningii
#       - Mitragyna ciliata
#       - Monodora tenuifolia
#       - Musa sapiensis
#       - Myrianthus arboreus
#       - Nothospondias staudtii
#
#   One defensible species-level SLA value was recovered:
#
#       Myrianthus arboreus = 25.1 mm2 mg-1
#
#   The remaining five species are retained as unresolved and will proceed
#   to later fallback acquisition.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# -----------------------------------------------------------------------------
# 1. Section header
# -----------------------------------------------------------------------------

print("=" * 100)
print(
    "SECTION 8.8M6 — RECORD AND ACCEPT BATCH 5 TARGETED "
    "LITERATURE EVIDENCE"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 2. Confirm output directory
# -----------------------------------------------------------------------------

project_root = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = project_root / "outputs" / "functional_traits"
output_dir.mkdir(parents=True, exist_ok=True)


# -----------------------------------------------------------------------------
# 3. Locate latest literature acquisition register
# -----------------------------------------------------------------------------

register_candidates = [
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch4.csv",

    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch3.csv",

    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch2.csv",

    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch1.csv",

    output_dir
    / "omo_sla_targeted_literature_acquisition_register.csv",
]

register_path = next(
    (p for p in register_candidates if p.exists()),
    None
)

if register_path is None:
    raise FileNotFoundError(
        "No targeted literature acquisition register could be located."
    )

literature_register = pd.read_csv(register_path)

print("\nLiterature acquisition register")
print("-" * 100)
print(f"Loaded from : {register_path}")
print(f"Rows        : {len(literature_register)}")


# -----------------------------------------------------------------------------
# 4. Define Batch 5 targets
# -----------------------------------------------------------------------------

batch5_targets = [
    "Millettia thonningii",
    "Mitragyna ciliata",
    "Monodora tenuifolia",
    "Musa sapiensis",
    "Myrianthus arboreus",
    "Nothospondias staudtii",
]

batch5_register = literature_register[
    literature_register["Normalized_Search_Name"].isin(
        batch5_targets
    )
].copy()

if len(batch5_register) != 6:
    print(
        "\nWARNING: Expected 6 Batch 5 targets but located "
        f"{len(batch5_register)} in the acquisition register."
    )


# -----------------------------------------------------------------------------
# 5. Record accepted literature evidence
#
# Source-derived value:
#
#   Myrianthus arboreus
#   SLA = 25.1 m2 kg-1
#
# Unit equivalence:
#
#   1 m2 kg-1 = 1 mm2 mg-1
#
# Therefore:
#
#   25.1 m2 kg-1 = 25.1 mm2 mg-1
#
# -----------------------------------------------------------------------------

batch5_accepted_species = pd.DataFrame(
    [
        {
            "Normalized_Search_Name": "Myrianthus arboreus",
            "Accepted_SLA_Value": 25.1,
            "Accepted_SLA_Unit": "mm2 mg-1",
            "Accepted_Provenance_Class": "SPECIES_LITERATURE",
            "Accepted_Source_Class": "PEER_REVIEWED_LITERATURE",
            "Accepted_Quality_Flag":
                "DIRECT_SPECIES_LITERATURE_VALUE",
            "Evidence_Context":
                "Direct species-level SLA value reported as "
                "25.1 m2 kg-1; numerically equivalent to "
                "25.1 mm2 mg-1.",
            "Acceptance_Decision":
                "ACCEPT_SPECIES_LITERATURE",
        }
    ]
)


# -----------------------------------------------------------------------------
# 6. Record unresolved Batch 5 targets
# -----------------------------------------------------------------------------

batch5_unresolved_names = [
    "Millettia thonningii",
    "Mitragyna ciliata",
    "Monodora tenuifolia",
    "Musa sapiensis",
    "Nothospondias staudtii",
]

batch5_unresolved = pd.DataFrame(
    {
        "Normalized_Search_Name": batch5_unresolved_names,
        "Batch5_Search_Outcome":
            "NO_USABLE_VALUE_FIRST_PASS",
        "Next_Action":
            "CONTINUE_TO_LATER_FALLBACK",
    }
)


# -----------------------------------------------------------------------------
# 7. Map accepted evidence back to Omo canonical taxa
#
# Use the complete rescue manifest because it preserves the relationship
# between normalized search names and Omo canonical names.
# -----------------------------------------------------------------------------

rescue_manifest_path = (
    output_dir
    / "omo_sla_targeted_species_rescue_manifest.csv"
)

if not rescue_manifest_path.exists():
    raise FileNotFoundError(
        "Targeted species rescue manifest not found:\n"
        f"{rescue_manifest_path}"
    )

rescue_manifest = pd.read_csv(rescue_manifest_path)

myrianthus_mapping = rescue_manifest[
    rescue_manifest["Normalized_Search_Name"].eq(
        "Myrianthus arboreus"
    )
].copy()

if myrianthus_mapping.empty:
    raise ValueError(
        "Myrianthus arboreus was not found in the targeted "
        "species rescue manifest."
    )

batch5_omo_values = (
    myrianthus_mapping[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
        ]
    ]
    .drop_duplicates()
    .copy()
)

batch5_omo_values["Standardized_SLA_Value"] = 25.1
batch5_omo_values["Standardized_SLA_Unit"] = "mm2 mg-1"
batch5_omo_values["Quality_Flag"] = (
    "DIRECT_SPECIES_LITERATURE_VALUE"
)
batch5_omo_values["Provenance_Class"] = (
    "SPECIES_LITERATURE"
)


# -----------------------------------------------------------------------------
# 8. Update acquisition register
# -----------------------------------------------------------------------------

updated_register = literature_register.copy()

# Ensure result fields exist.
result_columns = {
    "Literature_Search_Outcome": pd.NA,
    "Accepted_SLA_Value": np.nan,
    "Accepted_SLA_Unit": pd.NA,
    "Accepted_Provenance_Class": pd.NA,
    "Accepted_Source_Class": pd.NA,
    "Accepted_Quality_Flag": pd.NA,
    "Acceptance_Decision": pd.NA,
    "Next_Action": pd.NA,
}

for col, default_value in result_columns.items():
    if col not in updated_register.columns:
        updated_register[col] = default_value


# Accepted Myrianthus record
accepted_mask = (
    updated_register["Normalized_Search_Name"]
    .eq("Myrianthus arboreus")
)

updated_register.loc[
    accepted_mask,
    "Literature_Search_Outcome"
] = "USABLE_SPECIES_LEVEL_VALUE_FOUND"

updated_register.loc[
    accepted_mask,
    "Accepted_SLA_Value"
] = 25.1

updated_register.loc[
    accepted_mask,
    "Accepted_SLA_Unit"
] = "mm2 mg-1"

updated_register.loc[
    accepted_mask,
    "Accepted_Provenance_Class"
] = "SPECIES_LITERATURE"

updated_register.loc[
    accepted_mask,
    "Accepted_Source_Class"
] = "PEER_REVIEWED_LITERATURE"

updated_register.loc[
    accepted_mask,
    "Accepted_Quality_Flag"
] = "DIRECT_SPECIES_LITERATURE_VALUE"

updated_register.loc[
    accepted_mask,
    "Acceptance_Decision"
] = "ACCEPT_SPECIES_LITERATURE"

updated_register.loc[
    accepted_mask,
    "Next_Action"
] = "ACQUISITION_COMPLETE"


# Unresolved Batch 5 records
unresolved_mask = (
    updated_register["Normalized_Search_Name"]
    .isin(batch5_unresolved_names)
)

updated_register.loc[
    unresolved_mask,
    "Literature_Search_Outcome"
] = "NO_USABLE_VALUE_FIRST_PASS"

updated_register.loc[
    unresolved_mask,
    "Acceptance_Decision"
] = "NO_VALUE_ACCEPTED"

updated_register.loc[
    unresolved_mask,
    "Next_Action"
] = "CONTINUE_TO_LATER_FALLBACK"


# -----------------------------------------------------------------------------
# 9. Coverage accounting
#
# Previous validated checkpoint:
#
#   94 / 127 after Batch 4
#
# -----------------------------------------------------------------------------

coverage_before = 94
total_taxa = 127

new_coverage = (
    batch5_omo_values["Canonical_Species"]
    .nunique()
)

coverage_after = coverage_before + new_coverage
remaining_gaps = total_taxa - coverage_after
coverage_percent = coverage_after / total_taxa * 100


# -----------------------------------------------------------------------------
# 10. Validation
# -----------------------------------------------------------------------------

assert len(batch5_accepted_species) == 1

assert (
    batch5_accepted_species[
        "Normalized_Search_Name"
    ].iloc[0]
    == "Myrianthus arboreus"
)

assert np.isclose(
    batch5_accepted_species[
        "Accepted_SLA_Value"
    ].iloc[0],
    25.1
)

assert (
    batch5_accepted_species[
        "Accepted_SLA_Unit"
    ].iloc[0]
    == "mm2 mg-1"
)

assert len(batch5_unresolved) == 5

assert new_coverage == 1

assert coverage_after == 95

assert remaining_gaps == 32


# -----------------------------------------------------------------------------
# 11. Export
# -----------------------------------------------------------------------------

accepted_species_path = (
    output_dir
    / "omo_sla_literature_batch5_accepted_species_evidence.csv"
)

accepted_omo_path = (
    output_dir
    / "omo_sla_literature_batch5_accepted_omo_values.csv"
)

unresolved_path = (
    output_dir
    / "omo_sla_literature_batch5_unresolved.csv"
)

updated_register_path = (
    output_dir
    / "omo_sla_targeted_literature_acquisition_register_updated_batch5.csv"
)

batch5_accepted_species.to_csv(
    accepted_species_path,
    index=False
)

batch5_omo_values.to_csv(
    accepted_omo_path,
    index=False
)

batch5_unresolved.to_csv(
    unresolved_path,
    index=False
)

updated_register.to_csv(
    updated_register_path,
    index=False
)


# -----------------------------------------------------------------------------
# 12. Reporting
# -----------------------------------------------------------------------------

print("\nBatch 5 literature acceptance summary")
print("-" * 100)

print(
    f"Batch 5 unique species searches : "
    f"{len(batch5_targets)}"
)

print(
    f"Accepted species-level values   : "
    f"{len(batch5_accepted_species)}"
)

print(
    f"No usable first-pass values     : "
    f"{len(batch5_unresolved)}"
)

print(
    f"Omo canonical taxa newly covered: "
    f"{new_coverage}"
)


print("\nAccepted species-level literature evidence")
print("-" * 100)

display(
    batch5_accepted_species
)


print("\nOmo taxa receiving Batch 5 species-level literature SLA")
print("-" * 100)

display(
    batch5_omo_values
)


print("\nBatch 5 unresolved targets")
print("-" * 100)

display(
    batch5_unresolved
)


print("\nCoverage after Batch 5 literature rescue")
print("-" * 100)

print(
    f"Coverage before Batch 5 : "
    f"{coverage_before} / {total_taxa}"
)

print(
    f"New Batch 5 coverage    : "
    f"{new_coverage}"
)

print(
    f"Current SLA coverage    : "
    f"{coverage_after} / {total_taxa}"
)

print(
    f"Coverage percent        : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps      : "
    f"{remaining_gaps}"
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted species evidence : "
    f"{accepted_species_path}"
)

print(
    f"Accepted Omo values       : "
    f"{accepted_omo_path}"
)

print(
    f"Unresolved Batch 5        : "
    f"{unresolved_path}"
)

print(
    f"Updated acquisition register : "
    f"{updated_register_path}"
)


print(
    "\nSection 8.8M6 completed successfully."
)
```

## Section 8.8N — Post-Literature SLA Evidence Reconciliation

Following completion of all targeted species-level literature rescue batches,
the SLA acquisition workflow is reconciled across the complete set of 127
canonical Omo Forest taxa.

All accepted evidence streams are combined into a single provenance-aware
ledger. Where more than one value is available for a canonical taxon, the
highest-priority evidence is retained according to the established hierarchy:

1. `SPECIES_TRY`
2. `SPECIES_LITERATURE`
3. `GENUS_ESTIMATE`

Genus-level estimates include both BIEN-derived genus values and controlled
TRY-congeneric proxies. Lower-priority evidence is never allowed to replace a
valid species-level value.

This reconciliation provides a definitive post-literature SLA coverage state,
identifies any overlapping evidence sources, and produces the residual taxon
set requiring broader taxonomic fallback.

No family-level values are introduced in this subsection.

```python
# =============================================================================
# SECTION 8.8N — POST-LITERATURE SLA EVIDENCE RECONCILIATION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8N — POST-LITERATURE SLA EVIDENCE RECONCILIATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

legacy_output_dir = (
    PROJECT_ROOT
    / "notebooks"
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# 2. Locate complete 127-taxon foundation
# -----------------------------------------------------------------------------

foundation_candidates = [
    output_dir
    / "omo_sla_acquisition_targets.csv",

    legacy_output_dir
    / "omo_sla_acquisition_targets.csv",
]

foundation_path = next(
    (
        p for p in foundation_candidates
        if p.exists()
    ),
    None
)

if foundation_path is None:

    raise FileNotFoundError(
        "Could not locate omo_sla_acquisition_targets.csv."
    )


sla_foundation = pd.read_csv(
    foundation_path
)


assert len(sla_foundation) == 127

assert (
    sla_foundation[
        "Canonical_Species"
    ].nunique()
    == 127
)


print("\nFoundation")
print("-" * 100)
print(f"Loaded from     : {foundation_path}")
print(f"Canonical taxa  : {len(sla_foundation)}")


# -----------------------------------------------------------------------------
# 3. Helper for standardized evidence ledger
# -----------------------------------------------------------------------------

ledger_parts = []


def add_evidence(
    df,
    source_label,
    value_col,
    unit_col,
    provenance_col=None,
    provenance_value=None,
    quality_col=None,
    quality_value=None,
    trait_source_col=None,
    trait_source_value=None,
):

    if df is None or df.empty:
        return

    required = {
        "Canonical_Species",
        value_col,
    }

    missing = required - set(df.columns)

    if missing:
        raise KeyError(
            f"{source_label}: missing required columns "
            f"{sorted(missing)}"
        )

    temp = pd.DataFrame()

    temp[
        "Canonical_Species"
    ] = df[
        "Canonical_Species"
    ].astype(str).str.strip()

    temp[
        "Standardized_SLA_Value"
    ] = pd.to_numeric(
        df[value_col],
        errors="coerce"
    )

    if unit_col in df.columns:

        temp[
            "Standardized_SLA_Unit"
        ] = df[unit_col]

    else:

        temp[
            "Standardized_SLA_Unit"
        ] = "mm2 mg-1"


    if (
        provenance_col is not None
        and provenance_col in df.columns
    ):

        temp[
            "Provenance_Class"
        ] = df[
            provenance_col
        ]

    else:

        temp[
            "Provenance_Class"
        ] = provenance_value


    if (
        quality_col is not None
        and quality_col in df.columns
    ):

        temp[
            "Quality_Flag"
        ] = df[
            quality_col
        ]

    else:

        temp[
            "Quality_Flag"
        ] = quality_value


    if (
        trait_source_col is not None
        and trait_source_col in df.columns
    ):

        temp[
            "Trait_Source"
        ] = df[
            trait_source_col
        ]

    else:

        temp[
            "Trait_Source"
        ] = trait_source_value


    temp[
        "Evidence_Stream"
    ] = source_label


    temp = temp.loc[
        temp[
            "Standardized_SLA_Value"
        ].notna()
    ].copy()

    ledger_parts.append(
        temp
    )


# -----------------------------------------------------------------------------
# 4. Species-level TRY evidence
# -----------------------------------------------------------------------------

try_path = (
    output_dir
    / "omo_sla_final_species_try_values.csv"
)

try_values = pd.read_csv(
    try_path
)

add_evidence(
    df=try_values,
    source_label="FINAL_SPECIES_TRY",
    value_col="Standardized_SLA_Value",
    unit_col="Standardized_SLA_Unit",
    provenance_col="Provenance_Class",
    quality_col="TRY_Quality_Flag",
    trait_source_col="Trait_Source",
)


# -----------------------------------------------------------------------------
# 5. Earlier accepted species-literature evidence
#
# Citrus sinenses / Citrus sinensis
# -----------------------------------------------------------------------------

early_literature_path = (
    output_dir
    / "omo_sla_residual_species_literature_accepted.csv"
)

if early_literature_path.exists():

    early_literature = pd.read_csv(
        early_literature_path
    )

    add_evidence(
        df=early_literature,
        source_label="EARLY_SPECIES_LITERATURE",
        value_col="Accepted_SLA_Value",
        unit_col="Accepted_SLA_Unit",
        provenance_col="Accepted_Provenance_Class",
        quality_col="Accepted_Quality_Flag",
        trait_source_col="Accepted_Source_Class",
    )


# -----------------------------------------------------------------------------
# 6. Earlier accepted BIEN genus estimates
# -----------------------------------------------------------------------------

early_genus_path = (
    output_dir
    / "omo_sla_genus_target_acquisition_status.csv"
)

if early_genus_path.exists():

    early_genus = pd.read_csv(
        early_genus_path
    )

    early_genus = early_genus.loc[
        early_genus[
            "Genus_Acquisition_Status"
        ].eq(
            "GENUS_ESTIMATE_ACQUIRED"
        )
    ].copy()

    add_evidence(
        df=early_genus,
        source_label="EARLY_BIEN_GENUS",
        value_col="Accepted_Genus_SLA",
        unit_col="Accepted_Genus_SLA_Unit",
        provenance_value="GENUS_ESTIMATE",
        quality_col="Evidence_Strength",
        trait_source_value="BIEN",
    )


# -----------------------------------------------------------------------------
# 7. TRY-congeneric accepted estimates
# -----------------------------------------------------------------------------

try_congeneric_path = (
    output_dir
    / "omo_sla_try_congeneric_genus_estimates_accepted.csv"
)

if try_congeneric_path.exists():

    try_congeneric = pd.read_csv(
        try_congeneric_path
    )

    add_evidence(
        df=try_congeneric,
        source_label="TRY_CONGENERIC_GENUS",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_value="TRY_CONGENERIC",
    )


# -----------------------------------------------------------------------------
# 8. Residual BIEN genus values
# -----------------------------------------------------------------------------

bien_genus_path = (
    output_dir
    / "omo_sla_bien_residual_omo_genus_values.csv"
)

if bien_genus_path.exists():

    bien_genus_values = pd.read_csv(
        bien_genus_path
    )

    add_evidence(
        df=bien_genus_values,
        source_label="RESIDUAL_BIEN_GENUS",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
    )


# -----------------------------------------------------------------------------
# 9. Targeted literature rescue batches
# -----------------------------------------------------------------------------

literature_batch_files = [
    (
        "LITERATURE_BATCH1",
        "omo_sla_literature_batch1_accepted_omo_values.csv"
    ),
    (
        "LITERATURE_BATCH2",
        "omo_sla_literature_batch2_accepted_omo_values.csv"
    ),
    (
        "LITERATURE_BATCH3",
        "omo_sla_literature_batch3_accepted_omo_values.csv"
    ),
    (
        "LITERATURE_BATCH5",
        "omo_sla_literature_batch5_accepted_omo_values.csv"
    ),
    (
        "LITERATURE_BATCH6",
        "omo_sla_literature_batch6_accepted_omo_values.csv"
    ),
]


for source_label, filename in literature_batch_files:

    path = (
        output_dir
        / filename
    )

    if not path.exists():
        continue

    df = pd.read_csv(
        path
    )

    add_evidence(
        df=df,
        source_label=source_label,
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
        trait_source_value="PEER_REVIEWED_LITERATURE",
    )


# -----------------------------------------------------------------------------
# 10. Combine evidence ledger
# -----------------------------------------------------------------------------

sla_evidence_ledger = pd.concat(
    ledger_parts,
    ignore_index=True
)


# -----------------------------------------------------------------------------
# 11. Validate units
# -----------------------------------------------------------------------------

unit_values = (
    sla_evidence_ledger[
        "Standardized_SLA_Unit"
    ]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

assert set(unit_values) == {
    "mm2 mg-1"
}, (
    f"Unexpected SLA units detected: "
    f"{sorted(unit_values)}"
)


# -----------------------------------------------------------------------------
# 12. Assign evidence priority
# -----------------------------------------------------------------------------

priority_map = {
    "SPECIES_TRY": 1,
    "SPECIES_LITERATURE": 2,
    "GENUS_ESTIMATE": 3,
}


sla_evidence_ledger[
    "Evidence_Priority"
] = (
    sla_evidence_ledger[
        "Provenance_Class"
    ].map(
        priority_map
    )
)


if (
    sla_evidence_ledger[
        "Evidence_Priority"
    ].isna().any()
):

    unknown_classes = (
        sla_evidence_ledger.loc[
            sla_evidence_ledger[
                "Evidence_Priority"
            ].isna(),
            "Provenance_Class"
        ]
        .unique()
    )

    raise ValueError(
        "Unexpected provenance classes in SLA ledger: "
        f"{unknown_classes}"
    )


# -----------------------------------------------------------------------------
# 13. Detect taxa with multiple evidence streams
# -----------------------------------------------------------------------------

evidence_counts = (
    sla_evidence_ledger
    .groupby(
        "Canonical_Species"
    )
    .agg(
        Evidence_Record_Count=(
            "Evidence_Stream",
            "size"
        ),

        Evidence_Stream_Count=(
            "Evidence_Stream",
            "nunique"
        ),

        Evidence_Streams=(
            "Evidence_Stream",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),

        Provenance_Classes=(
            "Provenance_Class",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
    .reset_index()
)


overlapping_evidence = (
    evidence_counts.loc[
        evidence_counts[
            "Evidence_Record_Count"
        ] > 1
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 14. Select highest-priority evidence for each canonical taxon
# -----------------------------------------------------------------------------

sla_evidence_ranked = (
    sla_evidence_ledger
    .sort_values(
        [
            "Canonical_Species",
            "Evidence_Priority",
            "Evidence_Stream",
        ]
    )
    .reset_index(drop=True)
)


sla_selected = (
    sla_evidence_ranked
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ],
        keep="first"
    )
    .copy()
)


# -----------------------------------------------------------------------------
# 15. Merge selected evidence to complete 127-taxon foundation
# -----------------------------------------------------------------------------

sla_post_literature = (
    sla_foundation
    .merge(
        sla_selected[
            [
                "Canonical_Species",
                "Standardized_SLA_Value",
                "Standardized_SLA_Unit",
                "Provenance_Class",
                "Quality_Flag",
                "Trait_Source",
                "Evidence_Stream",
                "Evidence_Priority",
            ]
        ],
        how="left",
        on="Canonical_Species",
        validate="one_to_one"
    )
)


# -----------------------------------------------------------------------------
# 16. Current acquisition status
# -----------------------------------------------------------------------------

sla_post_literature[
    "SLA_Acquisition_Status"
] = np.where(
    sla_post_literature[
        "Standardized_SLA_Value"
    ].notna(),
    "SLA_VALUE_AVAILABLE",
    "SLA_VALUE_STILL_MISSING"
)


# -----------------------------------------------------------------------------
# 17. Identify definitive residual gaps
# -----------------------------------------------------------------------------

sla_post_literature_gaps = (
    sla_post_literature.loc[
        sla_post_literature[
            "Standardized_SLA_Value"
        ].isna()
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 18. Recover normalized taxonomic information for remaining gaps
# -----------------------------------------------------------------------------

normalization_path = (
    output_dir
    / "omo_sla_residual_taxonomic_normalization.csv"
)

if normalization_path.exists():

    normalization = pd.read_csv(
        normalization_path
    )

    normalization_keep = [
        col
        for col in [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Genus_Normalization_Source",
        ]
        if col in normalization.columns
    ]

    sla_post_literature_gaps = (
        sla_post_literature_gaps
        .merge(
            normalization[
                normalization_keep
            ]
            .drop_duplicates(
                "Canonical_Species"
            ),
            how="left",
            on="Canonical_Species",
            suffixes=(
                "",
                "_normalized"
            )
        )
    )


# -----------------------------------------------------------------------------
# 19. Assign next fallback class
# -----------------------------------------------------------------------------

sla_post_literature_gaps[
    "Next_Fallback_Strategy"
] = (
    "FAMILY_LEVEL_ESTIMATE"
)


# Genuine genus-level taxon remains distinct.
cola_species_mask = (
    sla_post_literature_gaps[
        "Canonical_Species"
    ].eq(
        "Cola species"
    )
)

sla_post_literature_gaps.loc[
    cola_species_mask,
    "Next_Fallback_Strategy"
] = (
    "GENUS_LEVEL_OR_FAMILY_ESTIMATE"
)


# -----------------------------------------------------------------------------
# 20. Coverage statistics
# -----------------------------------------------------------------------------

n_total = len(
    sla_post_literature
)

n_available = int(
    sla_post_literature[
        "Standardized_SLA_Value"
    ].notna().sum()
)

n_missing = int(
    sla_post_literature[
        "Standardized_SLA_Value"
    ].isna().sum()
)


provenance_summary = (
    sla_post_literature[
        "Provenance_Class"
    ]
    .fillna(
        "MISSING"
    )
    .value_counts()
    .rename_axis(
        "Provenance_Class"
    )
    .reset_index(
        name="Taxon_Count"
    )
)


provenance_summary[
    "Percent_of_127"
] = (
    provenance_summary[
        "Taxon_Count"
    ]
    / 127
    * 100
)


# -----------------------------------------------------------------------------
# 21. Validation
# -----------------------------------------------------------------------------

assert n_total == 127

assert n_available == 96, (
    f"Expected 96 taxa with SLA values, "
    f"found {n_available}."
)

assert n_missing == 31, (
    f"Expected 31 remaining SLA gaps, "
    f"found {n_missing}."
)

assert (
    n_available
    + n_missing
    == 127
)

assert (
    sla_post_literature[
        "Canonical_Species"
    ].nunique()
    == 127
)


# -----------------------------------------------------------------------------
# 22. Export
# -----------------------------------------------------------------------------

ledger_path = (
    output_dir
    / "omo_sla_post_literature_evidence_ledger.csv"
)

selected_path = (
    output_dir
    / "omo_sla_post_literature_selected_evidence.csv"
)

integrated_path = (
    output_dir
    / "omo_sla_post_literature_integrated_coverage.csv"
)

gaps_path = (
    output_dir
    / "omo_sla_post_literature_remaining_gaps.csv"
)

overlap_path = (
    output_dir
    / "omo_sla_post_literature_overlapping_evidence.csv"
)

provenance_path = (
    output_dir
    / "omo_sla_post_literature_provenance_summary.csv"
)


sla_evidence_ledger.to_csv(
    ledger_path,
    index=False
)

sla_selected.to_csv(
    selected_path,
    index=False
)

sla_post_literature.to_csv(
    integrated_path,
    index=False
)

sla_post_literature_gaps.to_csv(
    gaps_path,
    index=False
)

overlapping_evidence.to_csv(
    overlap_path,
    index=False
)

provenance_summary.to_csv(
    provenance_path,
    index=False
)


# -----------------------------------------------------------------------------
# 23. Diagnostics
# -----------------------------------------------------------------------------

print("\nPost-literature SLA reconciliation")
print("-" * 100)

print(
    f"Canonical taxa               : "
    f"{n_total}"
)

print(
    f"Taxa with accepted SLA       : "
    f"{n_available}"
)

print(
    f"Remaining SLA gaps           : "
    f"{n_missing}"
)

print(
    f"Current SLA coverage         : "
    f"{100 * n_available / n_total:.2f}%"
)


print("\nFinal evidence provenance distribution")
print("-" * 100)

display(
    provenance_summary
)


print("\nTaxa with overlapping accepted evidence")
print("-" * 100)

if overlapping_evidence.empty:

    print(
        "No overlapping evidence streams detected."
    )

else:

    display(
        overlapping_evidence
    )


print("\nDefinitive remaining SLA gaps")
print("-" * 100)

gap_display_columns = [
    col
    for col in [
        "Canonical_Species",
        "Normalized_Search_Name",
        "Normalized_Acquisition_Genus",
        "Taxonomic_Resolution",
        "Next_Fallback_Strategy",
    ]
    if col in sla_post_literature_gaps.columns
]

display(
    sla_post_literature_gaps[
        gap_display_columns
    ]
)


print("\nExports")
print("-" * 100)

print(f"Evidence ledger      : {ledger_path}")
print(f"Selected evidence    : {selected_path}")
print(f"Integrated coverage  : {integrated_path}")
print(f"Remaining gaps       : {gaps_path}")
print(f"Evidence overlaps    : {overlap_path}")
print(f"Provenance summary   : {provenance_path}")

print(
    "\nSection 8.8N completed successfully."
)
```

## Section 8.8O — Residual Family Assignment and Family-Level Fallback Manifest

Following reconciliation of all species-level TRY, species-level literature,
and genus-level SLA evidence, 31 canonical Omo Forest taxa remain unresolved.
These correspond to 25 distinct normalized taxonomic targets after accounting
for historical spellings and orthographic variants.

Before numerical family-level estimates are introduced, each residual taxon is
assigned to a controlled botanical family. Canonical Omo names remain unchanged;
family assignments are attached to the normalized acquisition identities used
throughout the trait-acquisition workflow.

The family-level fallback stage is intentionally separated into two steps:

1. taxonomic assignment and construction of the family acquisition manifest;
2. acquisition and quality assessment of family-level SLA evidence.

No numerical SLA value is imputed in this subsection.

```python
# =============================================================================
# SECTION 8.8O — RESIDUAL FAMILY ASSIGNMENT AND FAMILY-LEVEL FALLBACK MANIFEST
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8O — RESIDUAL FAMILY ASSIGNMENT AND FAMILY-LEVEL FALLBACK MANIFEST")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

gap_path = (
    output_dir
    / "omo_sla_post_literature_remaining_gaps.csv"
)

if not gap_path.exists():
    raise FileNotFoundError(
        "Post-literature residual gap table not found:\n"
        f"{gap_path}"
    )


# -----------------------------------------------------------------------------
# 2. Load definitive 31-taxon residual set
# -----------------------------------------------------------------------------

sla_family_targets = pd.read_csv(
    gap_path
)

assert len(sla_family_targets) == 31, (
    f"Expected 31 residual canonical taxa, "
    f"found {len(sla_family_targets)}."
)


# -----------------------------------------------------------------------------
# 3. Controlled family assignment
#
# Assign by normalized acquisition genus rather than historical canonical
# spelling.
# -----------------------------------------------------------------------------

family_lookup = {
    "Amphimas": "Fabaceae",
    "Aningeria": "Sapotaceae",
    "Anthocleista": "Gentianaceae",
    "Anthonotha": "Fabaceae",
    "Baphia": "Fabaceae",
    "Bridelia": "Phyllanthaceae",
    "Cleistopholis": "Annonaceae",
    "Cola": "Malvaceae",
    "Funtumia": "Apocynaceae",
    "Hexalobus": "Annonaceae",
    "Hildegardia": "Malvaceae",
    "Lecaniodiscus": "Sapindaceae",
    "Maesobotrya": "Phyllanthaceae",
    "Malacantha": "Sapotaceae",
    "Millettia": "Fabaceae",
    "Mitragyna": "Rubiaceae",
    "Monodora": "Annonaceae",
    "Musa": "Musaceae",
    "Nothospondias": "Anacardiaceae",
    "Pavetta": "Rubiaceae",
    "Scottellia": "Salicaceae",
    "Uapaca": "Phyllanthaceae",
}


# -----------------------------------------------------------------------------
# 4. Apply family assignments
# -----------------------------------------------------------------------------

sla_family_targets[
    "Assigned_Family"
] = (
    sla_family_targets[
        "Normalized_Acquisition_Genus"
    ]
    .map(family_lookup)
)


# -----------------------------------------------------------------------------
# 5. Validate complete family coverage
# -----------------------------------------------------------------------------

unassigned = (
    sla_family_targets.loc[
        sla_family_targets[
            "Assigned_Family"
        ].isna(),
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
        ]
    ]
)


if not unassigned.empty:

    print("\nUnassigned residual taxa")
    print("-" * 100)
    display(unassigned)

    raise RuntimeError(
        "One or more residual taxa lack a controlled family assignment."
    )


# -----------------------------------------------------------------------------
# 6. Construct unique normalized-target table
#
# Several canonical strings refer to the same normalized taxon.
# -----------------------------------------------------------------------------

normalized_family_targets = (
    sla_family_targets
    .groupby(
        [
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Assigned_Family",
        ],
        as_index=False,
        dropna=False
    )
    .agg(
        Canonical_Taxon_Count=(
            "Canonical_Species",
            "count"
        ),

        Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Taxonomic_Resolution=(
            "Taxonomic_Resolution",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
    .sort_values(
        [
            "Assigned_Family",
            "Normalized_Search_Name",
        ]
    )
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 7. Validate normalized target count
# -----------------------------------------------------------------------------

n_normalized_targets = len(
    normalized_family_targets
)

assert n_normalized_targets == 25, (
    f"Expected 25 distinct normalized residual targets, "
    f"found {n_normalized_targets}."
)


# -----------------------------------------------------------------------------
# 8. Construct unique family acquisition manifest
# -----------------------------------------------------------------------------

family_manifest = (
    normalized_family_targets
    .groupby(
        "Assigned_Family",
        as_index=False
    )
    .agg(
        Residual_Normalized_Target_Count=(
            "Normalized_Search_Name",
            "count"
        ),

        Residual_Canonical_Taxon_Count=(
            "Canonical_Taxon_Count",
            "sum"
        ),

        Residual_Genera=(
            "Normalized_Acquisition_Genus",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Residual_Normalized_Targets=(
            "Normalized_Search_Name",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
    .sort_values(
        "Assigned_Family"
    )
    .reset_index(drop=True)
)


family_manifest.insert(
    0,
    "SLA_Family_Query_ID",
    [
        f"OMO_SLA_FAMILY_{i:03d}"
        for i in range(
            1,
            len(family_manifest) + 1
        )
    ]
)


# -----------------------------------------------------------------------------
# 9. Define family acquisition governance
# -----------------------------------------------------------------------------

family_manifest[
    "Target_Trait"
] = "specific leaf area"

family_manifest[
    "Target_Unit"
] = "mm2 mg-1"

family_manifest[
    "Preferred_Evidence_1"
] = "MULTI_SPECIES_FAMILY_TRAIT_DATA"

family_manifest[
    "Preferred_Evidence_2"
] = "PEER_REVIEWED_FAMILY_LEVEL_TRAIT_SYNTHESIS"

family_manifest[
    "Fallback_Evidence"
] = "CONTROLLED_GLOBAL_SLA_ESTIMATE"

family_manifest[
    "Planned_Provenance_Class"
] = "FAMILY_ESTIMATE"

family_manifest[
    "Acquisition_Status"
] = "READY_FOR_FAMILY_EVIDENCE_ASSESSMENT"


# -----------------------------------------------------------------------------
# 10. Add family-level fallback class to canonical residual table
# -----------------------------------------------------------------------------

sla_family_targets[
    "Planned_Provenance_Class"
] = "FAMILY_ESTIMATE"

sla_family_targets[
    "Family_Acquisition_Status"
] = (
    "READY_FOR_FAMILY_EVIDENCE_ASSESSMENT"
)


# Cola species remains explicitly genus-level taxonomically, even if its
# eventual numerical fallback comes from the family.
cola_species_mask = (
    sla_family_targets[
        "Canonical_Species"
    ].eq("Cola species")
)

sla_family_targets.loc[
    cola_species_mask,
    "Family_Acquisition_Status"
] = (
    "GENUS_LEVEL_TARGET_REQUIRING_FAMILY_EVIDENCE"
)


# -----------------------------------------------------------------------------
# 11. Summary counts
# -----------------------------------------------------------------------------

n_canonical = len(
    sla_family_targets
)

n_normalized = len(
    normalized_family_targets
)

n_families = (
    family_manifest[
        "Assigned_Family"
    ].nunique()
)

n_genera = (
    normalized_family_targets[
        "Normalized_Acquisition_Genus"
    ].nunique()
)


# -----------------------------------------------------------------------------
# 12. Validation
# -----------------------------------------------------------------------------

assert n_canonical == 31
assert n_normalized == 25

assert (
    family_manifest[
        "Residual_Canonical_Taxon_Count"
    ].sum()
    == 31
)

assert (
    family_manifest[
        "Residual_Normalized_Target_Count"
    ].sum()
    == 25
)

assert (
    sla_family_targets[
        "Assigned_Family"
    ].notna().all()
)


# -----------------------------------------------------------------------------
# 13. Export
# -----------------------------------------------------------------------------

canonical_path = (
    output_dir
    / "omo_sla_family_fallback_canonical_targets.csv"
)

normalized_path = (
    output_dir
    / "omo_sla_family_fallback_normalized_targets.csv"
)

family_manifest_path = (
    output_dir
    / "omo_sla_family_acquisition_manifest.csv"
)


sla_family_targets.to_csv(
    canonical_path,
    index=False
)

normalized_family_targets.to_csv(
    normalized_path,
    index=False
)

family_manifest.to_csv(
    family_manifest_path,
    index=False
)


# -----------------------------------------------------------------------------
# 14. Diagnostics
# -----------------------------------------------------------------------------

print("\nResidual family fallback structure")
print("-" * 100)

print(
    f"Residual canonical taxa          : "
    f"{n_canonical}"
)

print(
    f"Distinct normalized targets      : "
    f"{n_normalized}"
)

print(
    f"Distinct residual genera         : "
    f"{n_genera}"
)

print(
    f"Families requiring assessment    : "
    f"{n_families}"
)


print("\nFamily distribution of residual canonical taxa")
print("-" * 100)

family_distribution = (
    sla_family_targets[
        "Assigned_Family"
    ]
    .value_counts()
    .rename_axis(
        "Family"
    )
    .reset_index(
        name="Canonical_Taxon_Count"
    )
)

display(
    family_distribution
)


print("\nNormalized residual targets")
print("-" * 100)

display(
    normalized_family_targets
)


print("\nFamily acquisition manifest")
print("-" * 100)

display(
    family_manifest[
        [
            "SLA_Family_Query_ID",
            "Assigned_Family",
            "Residual_Normalized_Target_Count",
            "Residual_Canonical_Taxon_Count",
            "Residual_Genera",
            "Acquisition_Status",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Canonical family targets : "
    f"{canonical_path}"
)

print(
    f"Normalized targets       : "
    f"{normalized_path}"
)

print(
    f"Family manifest          : "
    f"{family_manifest_path}"
)


print(
    "\nSection 8.8O completed successfully."
)
```

## Section 8.8P — Internal Family-Level SLA Evidence Assessment

The 31 unresolved canonical taxa belong to 12 botanical families. Before external
family-level or global SLA estimates are introduced, the existing accepted Omo
Forest SLA dataset is evaluated for within-family empirical support.

The assessment uses the 96 taxa for which SLA values have already been accepted
from species-level TRY data, species-level literature, or controlled genus-level
estimates. Accepted values are grouped by botanical family and summarized using
the median as a robust measure of central tendency.

Family estimates are not yet assigned to unresolved taxa. Instead, each residual
family is classified according to the amount and taxonomic breadth of internal
evidence available.

Evidence strength considers:

- number of accepted canonical taxa;
- number of distinct genera represented;
- provenance composition of the supporting SLA values;
- and numerical spread within each family.

Families with adequate multi-genus support may subsequently provide controlled
`FAMILY_ESTIMATE` values. Families with weak or absent internal support will
require external family-level evidence or a final controlled global fallback.

```python
# =============================================================================
# SECTION 8.8P0 — DISCOVER AUTHORITATIVE FAMILY METADATA
# =============================================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("SECTION 8.8P0 — DISCOVER AUTHORITATIVE FAMILY METADATA")
print("=" * 100)

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

foundation_candidates = [
    PROJECT_ROOT / "outputs" / "functional_traits"
    / "omo_sla_post_literature_integrated_coverage.csv",

    PROJECT_ROOT / "notebooks" / "outputs" / "functional_traits"
    / "omo_sla_acquisition_targets.csv",
]

foundation_path = next(
    (p for p in foundation_candidates if p.exists()),
    None
)

if foundation_path is None:
    raise FileNotFoundError(
        "Could not locate the 127-taxon Omo foundation."
    )

foundation = pd.read_csv(foundation_path)

canonical_species = set(
    foundation["Canonical_Species"]
    .dropna()
    .astype(str)
    .str.strip()
)

print(f"\nCanonical species expected : {len(canonical_species)}")


# -----------------------------------------------------------------------------
# Search all CSV files for plausible species + family columns
# -----------------------------------------------------------------------------

family_keywords = [
    "family",
    "taxonomic_family",
    "accepted_family",
    "canonical_family",
]

species_keywords = [
    "canonical_species",
    "species",
    "scientific_name",
    "canonical_name",
    "accepted_name",
]

candidate_results = []

csv_files = list(PROJECT_ROOT.rglob("*.csv"))

print(f"CSV files scanned         : {len(csv_files)}")


for path in csv_files:

    try:
        df = pd.read_csv(
            path,
            low_memory=False,
            nrows=5000
        )
    except Exception:
        continue

    lower_lookup = {
        str(col).strip().lower(): col
        for col in df.columns
    }

    family_col = next(
        (
            lower_lookup[key]
            for key in family_keywords
            if key in lower_lookup
        ),
        None
    )

    if family_col is None:
        continue

    species_col = next(
        (
            lower_lookup[key]
            for key in species_keywords
            if key in lower_lookup
        ),
        None
    )

    if species_col is None:
        continue

    temp = df[
        [
            species_col,
            family_col,
        ]
    ].copy()

    temp.columns = [
        "Candidate_Species",
        "Candidate_Family",
    ]

    temp["Candidate_Species"] = (
        temp["Candidate_Species"]
        .astype("string")
        .str.strip()
    )

    temp["Candidate_Family"] = (
        temp["Candidate_Family"]
        .astype("string")
        .str.strip()
    )

    temp = temp.dropna(
        subset=[
            "Candidate_Species",
            "Candidate_Family",
        ]
    )

    overlap = (
        set(temp["Candidate_Species"])
        & canonical_species
    )

    candidate_results.append(
        {
            "File_Path": str(path),
            "Species_Column": species_col,
            "Family_Column": family_col,
            "Rows_Read": len(df),
            "Unique_Species": temp[
                "Candidate_Species"
            ].nunique(),
            "Canonical_Overlap": len(overlap),
            "Coverage_Percent": (
                100 * len(overlap) / 127
            ),
        }
    )


family_metadata_candidates = pd.DataFrame(
    candidate_results
)


if family_metadata_candidates.empty:

    print(
        "\nNo CSV containing both a recognizable "
        "species column and family column was found."
    )

else:

    family_metadata_candidates = (
        family_metadata_candidates
        .sort_values(
            [
                "Canonical_Overlap",
                "Unique_Species",
            ],
            ascending=False
        )
        .reset_index(drop=True)
    )

    print("\nCandidate family metadata files")
    print("-" * 100)

    display(
        family_metadata_candidates
    )


# -----------------------------------------------------------------------------
# Best candidate
# -----------------------------------------------------------------------------

if not family_metadata_candidates.empty:

    best = family_metadata_candidates.iloc[0]

    print("\nBest candidate")
    print("-" * 100)

    print(
        f"File              : "
        f"{best['File_Path']}"
    )

    print(
        f"Species column    : "
        f"{best['Species_Column']}"
    )

    print(
        f"Family column     : "
        f"{best['Family_Column']}"
    )

    print(
        f"Canonical overlap : "
        f"{int(best['Canonical_Overlap'])} / 127"
    )

    print(
        f"Coverage          : "
        f"{best['Coverage_Percent']:.2f}%"
    )


print(
    "\nSection 8.8P0 completed successfully."
)
```

### 8.8P1 Construct Consolidated Genus–Family Authority

No single project CSV provides complete family assignments for all 127 current
canonical Omo taxa. The strongest directly overlapping source covers 70 taxa,
while additional taxonomic reference files contain valid family information
under alternative or accepted scientific names.

To support the family-level SLA assessment without manually assigning families
to already resolved taxa, family evidence is consolidated at the genus level
from multiple validated project sources.

Genus–family relationships are extracted from:

1. the processed tree analysis dataset;
2. the wood-density harmonization output containing family information; and
3. the project species dictionary.

Family assignments are accepted only where a genus has a unique family
relationship across the contributing sources. Any genus associated with
conflicting families is explicitly flagged for review rather than resolved
automatically.

The resulting genus–family authority is then used to assign families to the
96 taxa with accepted SLA values.

```python
# =============================================================================
# SECTION 8.8P1 — CONSTRUCT CONSOLIDATED GENUS–FAMILY AUTHORITY
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8P1 — CONSTRUCT CONSOLIDATED GENUS–FAMILY AUTHORITY")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

tree_path = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "tree_analysis_ready.csv"
)

dictionary_path = (
    PROJECT_ROOT
    / "data"
    / "reference"
    / "species_dictionary.csv"
)


# -----------------------------------------------------------------------------
# 2. Locate wood-density file containing family information
# -----------------------------------------------------------------------------

wd_candidates = []

for path in output_dir.glob("*.csv"):

    if "wood" not in path.name.lower():
        continue

    try:
        tmp = pd.read_csv(
            path,
            nrows=5,
            low_memory=False
        )
    except Exception:
        continue

    lower_cols = {
        str(c).lower(): c
        for c in tmp.columns
    }

    if (
        "canonical_species" in lower_cols
        and "family" in lower_cols
    ):
        wd_candidates.append(path)


print("\nInput sources")
print("-" * 100)

print(f"Tree dataset exists       : {tree_path.exists()}")
print(f"Species dictionary exists : {dictionary_path.exists()}")

print(
    f"Wood-density candidates   : "
    f"{len(wd_candidates)}"
)

for path in wd_candidates:
    print(f"  - {path}")


# -----------------------------------------------------------------------------
# 3. Helper: extract clean genus-family relationships
# -----------------------------------------------------------------------------

def genus_from_name(name):

    if pd.isna(name):
        return pd.NA

    text = str(name).strip()

    if not text:
        return pd.NA

    return text.split()[0]


def prepare_genus_family(
    df,
    species_col,
    family_col,
    source_name
):

    temp = df[
        [
            species_col,
            family_col,
        ]
    ].copy()

    temp.columns = [
        "Scientific_Name",
        "Family",
    ]

    temp[
        "Scientific_Name"
    ] = (
        temp[
            "Scientific_Name"
        ]
        .astype("string")
        .str.strip()
    )

    temp[
        "Family"
    ] = (
        temp[
            "Family"
        ]
        .astype("string")
        .str.strip()
    )

    temp = temp.dropna(
        subset=[
            "Scientific_Name",
            "Family",
        ]
    )

    temp = temp.loc[
        ~temp[
            "Family"
        ].isin(
            [
                "",
                "nan",
                "<NA>",
                "Unknown",
                "UNKNOWN",
            ]
        )
    ].copy()

    temp[
        "Genus"
    ] = (
        temp[
            "Scientific_Name"
        ]
        .map(genus_from_name)
    )

    temp[
        "Family_Source"
    ] = source_name

    return (
        temp[
            [
                "Scientific_Name",
                "Genus",
                "Family",
                "Family_Source",
            ]
        ]
        .drop_duplicates()
    )


# -----------------------------------------------------------------------------
# 4. Extract genus-family evidence from tree dataset
# -----------------------------------------------------------------------------

genus_family_parts = []


if tree_path.exists():

    tree_df = pd.read_csv(
        tree_path,
        low_memory=False
    )

    tree_family = prepare_genus_family(
        df=tree_df,
        species_col="Species",
        family_col="Family",
        source_name="TREE_ANALYSIS_READY"
    )

    genus_family_parts.append(
        tree_family
    )


# -----------------------------------------------------------------------------
# 5. Extract genus-family evidence from species dictionary
# -----------------------------------------------------------------------------

if dictionary_path.exists():

    dictionary_df = pd.read_csv(
        dictionary_path,
        low_memory=False
    )

    dictionary_family = prepare_genus_family(
        df=dictionary_df,
        species_col="Scientific_Name",
        family_col="Family",
        source_name="SPECIES_DICTIONARY"
    )

    genus_family_parts.append(
        dictionary_family
    )


# -----------------------------------------------------------------------------
# 6. Extract genus-family evidence from all suitable WD outputs
# -----------------------------------------------------------------------------

for wd_path in wd_candidates:

    wd_df = pd.read_csv(
        wd_path,
        low_memory=False
    )

    family_col = next(
        col
        for col in wd_df.columns
        if str(col).lower() == "family"
    )

    species_col = next(
        col
        for col in wd_df.columns
        if str(col).lower() == "canonical_species"
    )

    wd_family = prepare_genus_family(
        df=wd_df,
        species_col=species_col,
        family_col=family_col,
        source_name=f"WOOD_DENSITY:{wd_path.name}"
    )

    genus_family_parts.append(
        wd_family
    )


# -----------------------------------------------------------------------------
# 7. Combine all genus-family evidence
# -----------------------------------------------------------------------------

if not genus_family_parts:

    raise RuntimeError(
        "No genus-family evidence could be extracted."
    )


genus_family_evidence = (
    pd.concat(
        genus_family_parts,
        ignore_index=True
    )
    .drop_duplicates()
)


print("\nCombined genus-family evidence")
print("-" * 100)

print(
    f"Evidence records : "
    f"{len(genus_family_evidence)}"
)

print(
    f"Unique genera    : "
    f"{genus_family_evidence['Genus'].nunique()}"
)

print(
    f"Unique families  : "
    f"{genus_family_evidence['Family'].nunique()}"
)


# -----------------------------------------------------------------------------
# 8. Detect genus-family conflicts
# -----------------------------------------------------------------------------

genus_family_counts = (
    genus_family_evidence
    .groupby("Genus")
    ["Family"]
    .nunique()
    .rename(
        "Distinct_Family_Count"
    )
    .reset_index()
)


conflicting_genera = (
    genus_family_counts.loc[
        genus_family_counts[
            "Distinct_Family_Count"
        ] > 1
    ]
    .copy()
)


if conflicting_genera.empty:

    conflict_detail = pd.DataFrame()

else:

    conflict_detail = (
        genus_family_evidence.loc[
            genus_family_evidence[
                "Genus"
            ].isin(
                conflicting_genera[
                    "Genus"
                ]
            )
        ]
        .sort_values(
            [
                "Genus",
                "Family",
                "Family_Source",
            ]
        )
        .copy()
    )


print("\nGenus-family conflicts")
print("-" * 100)

print(
    f"Conflicting genera : "
    f"{len(conflicting_genera)}"
)

if not conflict_detail.empty:
    display(conflict_detail)


# -----------------------------------------------------------------------------
# 9. Build authority only from non-conflicting genera
# -----------------------------------------------------------------------------

non_conflicting_genera = (
    genus_family_counts.loc[
        genus_family_counts[
            "Distinct_Family_Count"
        ].eq(1),
        "Genus"
    ]
)


genus_family_authority = (
    genus_family_evidence.loc[
        genus_family_evidence[
            "Genus"
        ].isin(
            non_conflicting_genera
        )
    ]
    .groupby(
        "Genus",
        as_index=False
    )
    .agg(
        Assigned_Family=(
            "Family",
            "first"
        ),

        Supporting_Source_Count=(
            "Family_Source",
            "nunique"
        ),

        Supporting_Sources=(
            "Family_Source",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),

        Supporting_Name_Count=(
            "Scientific_Name",
            "nunique"
        ),
    )
)


# -----------------------------------------------------------------------------
# 10. Load post-literature accepted SLA taxa
# -----------------------------------------------------------------------------

integrated_path = (
    output_dir
    / "omo_sla_post_literature_integrated_coverage.csv"
)

sla_integrated = pd.read_csv(
    integrated_path
)


accepted_sla = (
    sla_integrated.loc[
        sla_integrated[
            "Standardized_SLA_Value"
        ].notna()
    ]
    .copy()
)


assert len(
    accepted_sla
) == 96


# -----------------------------------------------------------------------------
# 11. Recover best available genus for each accepted taxon
#
# First preference:
#   Resolved TRY name, where present.
#
# Otherwise:
#   canonical first token.
# -----------------------------------------------------------------------------

if (
    "Resolved_TRY_Name"
    in accepted_sla.columns
):

    accepted_sla[
        "Family_Match_Name"
    ] = (
        accepted_sla[
            "Resolved_TRY_Name"
        ]
        .fillna(
            accepted_sla[
                "Canonical_Species"
            ]
        )
    )

else:

    accepted_sla[
        "Family_Match_Name"
    ] = (
        accepted_sla[
            "Canonical_Species"
        ]
    )


accepted_sla[
    "Family_Match_Genus"
] = (
    accepted_sla[
        "Family_Match_Name"
    ]
    .map(genus_from_name)
)


# -----------------------------------------------------------------------------
# 12. Attach consolidated family authority
# -----------------------------------------------------------------------------

accepted_sla_with_family = (
    accepted_sla
    .merge(
        genus_family_authority,
        how="left",
        left_on="Family_Match_Genus",
        right_on="Genus",
        validate="many_to_one"
    )
)


# -----------------------------------------------------------------------------
# 13. Assess coverage specifically for the 12 residual families
# -----------------------------------------------------------------------------

family_manifest_path = (
    output_dir
    / "omo_sla_family_acquisition_manifest.csv"
)

family_manifest = pd.read_csv(
    family_manifest_path
)


target_families = set(
    family_manifest[
        "Assigned_Family"
    ]
)


family_assignment_summary = pd.DataFrame(
    {
        "Accepted_SLA_Taxa": [
            len(
                accepted_sla_with_family
            )
        ],

        "Accepted_Taxa_With_Family": [
            accepted_sla_with_family[
                "Assigned_Family"
            ].notna().sum()
        ],

        "Accepted_Taxa_Without_Family": [
            accepted_sla_with_family[
                "Assigned_Family"
            ].isna().sum()
        ],

        "Accepted_Taxa_In_12_Target_Families": [
            accepted_sla_with_family[
                "Assigned_Family"
            ].isin(
                target_families
            ).sum()
        ],
    }
)


# -----------------------------------------------------------------------------
# 14. Missing-family diagnostics
# -----------------------------------------------------------------------------

accepted_family_missing = (
    accepted_sla_with_family.loc[
        accepted_sla_with_family[
            "Assigned_Family"
        ].isna(),
        [
            "Canonical_Species",
            "Family_Match_Name",
            "Family_Match_Genus",
            "Provenance_Class",
        ]
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 15. Export
# -----------------------------------------------------------------------------

evidence_path = (
    output_dir
    / "omo_sla_genus_family_evidence_combined.csv"
)

authority_path = (
    output_dir
    / "omo_sla_genus_family_authority.csv"
)

conflict_path = (
    output_dir
    / "omo_sla_genus_family_conflicts.csv"
)

accepted_family_path = (
    output_dir
    / "omo_sla_accepted_values_with_family_assignment.csv"
)

missing_path = (
    output_dir
    / "omo_sla_accepted_values_missing_family_assignment.csv"
)


genus_family_evidence.to_csv(
    evidence_path,
    index=False
)

genus_family_authority.to_csv(
    authority_path,
    index=False
)

conflict_detail.to_csv(
    conflict_path,
    index=False
)

accepted_sla_with_family.to_csv(
    accepted_family_path,
    index=False
)

accepted_family_missing.to_csv(
    missing_path,
    index=False
)


# -----------------------------------------------------------------------------
# 16. Diagnostics
# -----------------------------------------------------------------------------

print("\nConsolidated genus-family authority")
print("-" * 100)

print(
    f"Authoritative genera : "
    f"{len(genus_family_authority)}"
)


print("\nAccepted SLA family-assignment coverage")
print("-" * 100)

display(
    family_assignment_summary
)


print("\nAccepted taxa without family assignment")
print("-" * 100)

if accepted_family_missing.empty:

    print(
        "All 96 accepted SLA taxa received family assignments."
    )

else:

    display(
        accepted_family_missing
    )


print("\nCoverage within the 12 residual families")
print("-" * 100)

target_family_support = (
    accepted_sla_with_family.loc[
        accepted_sla_with_family[
            "Assigned_Family"
        ].isin(
            target_families
        )
    ]
    .groupby(
        "Assigned_Family"
    )
    .agg(
        Accepted_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Accepted_Genera=(
            "Family_Match_Genus",
            "nunique"
        ),
    )
    .reindex(
        sorted(
            target_families
        ),
        fill_value=0
    )
    .reset_index()
)

display(
    target_family_support
)


print("\nExports")
print("-" * 100)

print(
    f"Combined evidence : "
    f"{evidence_path}"
)

print(
    f"Genus authority   : "
    f"{authority_path}"
)

print(
    f"Conflicts         : "
    f"{conflict_path}"
)

print(
    f"Accepted + family : "
    f"{accepted_family_path}"
)

print(
    f"Missing family    : "
    f"{missing_path}"
)


print(
    "\nSection 8.8P1 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8P2 — HARMONIZE ACCEPTED SLA TAXA TO NORMALIZED GENERA
#                  AND RECALCULATE FAMILY SUPPORT
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8P2 — HARMONIZE ACCEPTED SLA TAXA TO NORMALIZED GENERA "
    "AND RECALCULATE FAMILY SUPPORT"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

accepted_family_path = (
    output_dir
    / "omo_sla_accepted_values_with_family_assignment.csv"
)

authority_path = (
    output_dir
    / "omo_sla_genus_family_authority.csv"
)

normalization_path = (
    output_dir
    / "omo_sla_residual_taxonomic_normalization.csv"
)

verified_crosswalk_path = (
    output_dir
    / "omo_sla_try_taxonomic_crosswalk_verified.csv"
)

family_manifest_path = (
    output_dir
    / "omo_sla_family_acquisition_manifest.csv"
)


# -----------------------------------------------------------------------------
# 2. Validate required inputs
# -----------------------------------------------------------------------------

required_paths = [
    accepted_family_path,
    authority_path,
    family_manifest_path,
]

missing_paths = [
    p for p in required_paths
    if not p.exists()
]

if missing_paths:
    raise FileNotFoundError(
        "Required files are missing:\n"
        + "\n".join(str(p) for p in missing_paths)
    )


accepted_sla = pd.read_csv(
    accepted_family_path,
    low_memory=False
)

genus_family_authority = pd.read_csv(
    authority_path,
    low_memory=False
)

family_manifest = pd.read_csv(
    family_manifest_path,
    low_memory=False
)


assert len(accepted_sla) == 96


# -----------------------------------------------------------------------------
# 3. Begin normalized-name recovery table
# -----------------------------------------------------------------------------

accepted_harmonized = (
    accepted_sla.copy()
)

accepted_harmonized[
    "Recovered_Normalized_Name"
] = pd.NA

accepted_harmonized[
    "Recovered_Normalized_Genus"
] = pd.NA

accepted_harmonized[
    "Genus_Normalization_Method"
] = pd.NA


# -----------------------------------------------------------------------------
# 4. Use residual-taxonomic-normalization output where available
# -----------------------------------------------------------------------------

if normalization_path.exists():

    residual_normalization = pd.read_csv(
        normalization_path,
        low_memory=False
    )

    normalization_lookup = (
        residual_normalization[
            [
                "Canonical_Species",
                "Normalized_Search_Name",
                "Normalized_Acquisition_Genus",
            ]
        ]
        .drop_duplicates(
            subset="Canonical_Species"
        )
        .copy()
    )

    accepted_harmonized = (
        accepted_harmonized
        .merge(
            normalization_lookup,
            how="left",
            on="Canonical_Species",
            validate="one_to_one"
        )
    )

    has_residual_normalization = (
        accepted_harmonized[
            "Normalized_Acquisition_Genus"
        ].notna()
    )

    accepted_harmonized.loc[
        has_residual_normalization,
        "Recovered_Normalized_Name"
    ] = accepted_harmonized.loc[
        has_residual_normalization,
        "Normalized_Search_Name"
    ]

    accepted_harmonized.loc[
        has_residual_normalization,
        "Recovered_Normalized_Genus"
    ] = accepted_harmonized.loc[
        has_residual_normalization,
        "Normalized_Acquisition_Genus"
    ]

    accepted_harmonized.loc[
        has_residual_normalization,
        "Genus_Normalization_Method"
    ] = "RESIDUAL_TAXONOMIC_NORMALIZATION"


# -----------------------------------------------------------------------------
# 5. Recover TRY accepted names where residual normalization is unavailable
# -----------------------------------------------------------------------------

if verified_crosswalk_path.exists():

    verified_crosswalk = pd.read_csv(
        verified_crosswalk_path,
        low_memory=False
    )

    try_lookup = (
        verified_crosswalk[
            [
                "Canonical_Species",
                "Resolved_TRY_Name",
            ]
        ]
        .dropna(
            subset=["Resolved_TRY_Name"]
        )
        .drop_duplicates(
            subset="Canonical_Species"
        )
        .copy()
    )

    try_lookup[
        "TRY_Resolved_Genus"
    ] = (
        try_lookup[
            "Resolved_TRY_Name"
        ]
        .astype(str)
        .str.strip()
        .str.split()
        .str[0]
    )

    accepted_harmonized = (
        accepted_harmonized
        .merge(
            try_lookup,
            how="left",
            on="Canonical_Species",
            validate="one_to_one"
        )
    )

    use_try = (
        accepted_harmonized[
            "Recovered_Normalized_Genus"
        ].isna()
        &
        accepted_harmonized[
            "TRY_Resolved_Genus"
        ].notna()
    )

    accepted_harmonized.loc[
        use_try,
        "Recovered_Normalized_Name"
    ] = accepted_harmonized.loc[
        use_try,
        "Resolved_TRY_Name"
    ]

    accepted_harmonized.loc[
        use_try,
        "Recovered_Normalized_Genus"
    ] = accepted_harmonized.loc[
        use_try,
        "TRY_Resolved_Genus"
    ]

    accepted_harmonized.loc[
        use_try,
        "Genus_Normalization_Method"
    ] = "VERIFIED_TRY_ACCEPTED_NAME"


# -----------------------------------------------------------------------------
# 6. Controlled genus corrections for known historical/orthographic variants
#
# These are taxa already resolved during earlier sections of Notebook 07B.
# They are included here only as a final normalization safeguard.
# -----------------------------------------------------------------------------

controlled_genus_overrides = {

    "Annonidium manni": "Anonidium",
    "Chrysophylum albidum": "Chrysophyllum",
    "Corynante pachycercus": "Corynanthe",
    "Diallium guineense": "Dialium",

    "Diospyrous dendo": "Diospyros",
    "Diospyrous melistoformis": "Diospyros",
    "Diospyrous mespiliformis": "Diospyros",

    "Dracena manni": "Dracaena",

    "Irvinga gabonensis": "Irvingia",

    "Macranga bateri": "Macaranga",

    "Margartaria discoidea": "Margaritaria",

    "Nesodogornia papaverifera": "Nesogordonia",
    "Nesodogornia papaverivera": "Nesogordonia",

    "Pycanthus angolensis": "Pycnanthus",

    "Raulvolfia vomitoria": "Rauvolfia",
    "Rauvolvia vomitoria": "Rauvolfia",

    "Ricinidendron heudeloti": "Ricinodendron",

    "Rothmania hispida": "Rothmannia",

    "Spondia mombin": "Spondias",

    "Strombosia postulata": "Strombosia",
    "Strombosia putulata": "Strombosia",

    "Termialia superba": "Terminalia",
}


for canonical_name, corrected_genus in controlled_genus_overrides.items():

    mask = (
        accepted_harmonized[
            "Canonical_Species"
        ].eq(canonical_name)
    )

    if mask.any():

        accepted_harmonized.loc[
            mask,
            "Recovered_Normalized_Genus"
        ] = corrected_genus

        accepted_harmonized.loc[
            mask,
            "Genus_Normalization_Method"
        ] = (
            "CONTROLLED_PROJECT_TAXONOMIC_OVERRIDE"
        )


# -----------------------------------------------------------------------------
# 7. Final fallback to canonical first-token genus
# -----------------------------------------------------------------------------

fallback_mask = (
    accepted_harmonized[
        "Recovered_Normalized_Genus"
    ].isna()
)

accepted_harmonized.loc[
    fallback_mask,
    "Recovered_Normalized_Genus"
] = (
    accepted_harmonized.loc[
        fallback_mask,
        "Canonical_Species"
    ]
    .astype(str)
    .str.strip()
    .str.split()
    .str[0]
)

accepted_harmonized.loc[
    fallback_mask,
    "Genus_Normalization_Method"
] = "CANONICAL_FIRST_TOKEN"


# -----------------------------------------------------------------------------
# 8. Remove old incomplete family assignment
# -----------------------------------------------------------------------------

columns_to_remove = [
    "Genus",
    "Assigned_Family",
    "Supporting_Source_Count",
    "Supporting_Sources",
    "Supporting_Name_Count",
]

accepted_harmonized = (
    accepted_harmonized.drop(
        columns=[
            c for c in columns_to_remove
            if c in accepted_harmonized.columns
        ],
        errors="ignore"
    )
)


# -----------------------------------------------------------------------------
# 9. Reattach family using normalized genus authority
# -----------------------------------------------------------------------------

family_lookup = (
    genus_family_authority[
        [
            "Genus",
            "Assigned_Family",
            "Supporting_Source_Count",
            "Supporting_Sources",
            "Supporting_Name_Count",
        ]
    ]
    .drop_duplicates(
        subset="Genus"
    )
    .copy()
)


accepted_harmonized = (
    accepted_harmonized
    .merge(
        family_lookup,
        how="left",
        left_on="Recovered_Normalized_Genus",
        right_on="Genus",
        validate="many_to_one"
    )
)


# -----------------------------------------------------------------------------
# 10. Diagnose remaining unassigned accepted taxa
# -----------------------------------------------------------------------------

accepted_family_missing_after_normalization = (
    accepted_harmonized.loc[
        accepted_harmonized[
            "Assigned_Family"
        ].isna()
    ]
    [
        [
            "Canonical_Species",
            "Recovered_Normalized_Name",
            "Recovered_Normalized_Genus",
            "Genus_Normalization_Method",
            "Provenance_Class",
        ]
    ]
    .copy()
)


n_with_family = (
    accepted_harmonized[
        "Assigned_Family"
    ].notna().sum()
)

n_without_family = (
    accepted_harmonized[
        "Assigned_Family"
    ].isna().sum()
)


# -----------------------------------------------------------------------------
# 11. Restrict to the 12 residual families
# -----------------------------------------------------------------------------

target_families = set(
    family_manifest[
        "Assigned_Family"
    ]
)


target_family_records = (
    accepted_harmonized.loc[
        accepted_harmonized[
            "Assigned_Family"
        ].isin(
            target_families
        )
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 12. Recalculate family support
#
# Important:
# Historical orthographic variants must not inflate taxonomic breadth.
#
# We therefore count unique normalized genera, while canonical taxa are retained
# only as supporting evidence records.
# -----------------------------------------------------------------------------

family_support_summary = (
    target_family_records
    .groupby(
        "Assigned_Family",
        as_index=False
    )
    .agg(
        Accepted_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Accepted_Normalized_Genera=(
            "Recovered_Normalized_Genus",
            "nunique"
        ),

        Internal_Min_SLA=(
            "Standardized_SLA_Value",
            "min"
        ),

        Internal_Median_SLA=(
            "Standardized_SLA_Value",
            "median"
        ),

        Internal_Mean_SLA=(
            "Standardized_SLA_Value",
            "mean"
        ),

        Internal_Max_SLA=(
            "Standardized_SLA_Value",
            "max"
        ),
    )
)


# -----------------------------------------------------------------------------
# 13. Provenance composition
# -----------------------------------------------------------------------------

family_provenance = (
    target_family_records
    .groupby(
        [
            "Assigned_Family",
            "Provenance_Class",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
    .reset_index()
)


for col in [
    "SPECIES_TRY",
    "SPECIES_LITERATURE",
    "GENUS_ESTIMATE",
]:

    if col not in family_provenance.columns:
        family_provenance[col] = 0


# -----------------------------------------------------------------------------
# 14. Merge all 12 families, including those with no internal support
# -----------------------------------------------------------------------------

family_evidence_assessment = (
    family_manifest
    .merge(
        family_support_summary,
        how="left",
        on="Assigned_Family"
    )
    .merge(
        family_provenance[
            [
                "Assigned_Family",
                "SPECIES_TRY",
                "SPECIES_LITERATURE",
                "GENUS_ESTIMATE",
            ]
        ],
        how="left",
        on="Assigned_Family"
    )
)


count_cols = [
    "Accepted_Canonical_Taxa",
    "Accepted_Normalized_Genera",
    "SPECIES_TRY",
    "SPECIES_LITERATURE",
    "GENUS_ESTIMATE",
]


family_evidence_assessment[
    count_cols
] = (
    family_evidence_assessment[
        count_cols
    ]
    .fillna(0)
    .astype(int)
)


# -----------------------------------------------------------------------------
# 15. Classify evidence strength
# -----------------------------------------------------------------------------

def classify_family_support(row):

    n_taxa = row[
        "Accepted_Canonical_Taxa"
    ]

    n_genera = row[
        "Accepted_Normalized_Genera"
    ]

    if (
        n_taxa >= 5
        and n_genera >= 3
    ):
        return "STRONG"

    if (
        n_taxa >= 3
        and n_genera >= 2
    ):
        return "MODERATE"

    if n_taxa >= 2:
        return "LIMITED"

    if n_taxa == 1:
        return "SINGLE_TAXON"

    return "NONE"


family_evidence_assessment[
    "Internal_Family_Evidence_Strength"
] = (
    family_evidence_assessment.apply(
        classify_family_support,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 16. Recommended family pathway
# -----------------------------------------------------------------------------

family_evidence_assessment[
    "Recommended_Family_Pathway"
] = np.select(
    [
        family_evidence_assessment[
            "Internal_Family_Evidence_Strength"
        ].isin(
            [
                "STRONG",
                "MODERATE",
            ]
        ),

        family_evidence_assessment[
            "Internal_Family_Evidence_Strength"
        ].isin(
            [
                "LIMITED",
                "SINGLE_TAXON",
            ]
        ),
    ],
    [
        "INTERNAL_FAMILY_ESTIMATE_CANDIDATE",
        "INTERNAL_SUPPORT_TOO_LIMITED_REQUIRES_EXTERNAL_REVIEW",
    ],
    default=(
        "EXTERNAL_FAMILY_OR_GLOBAL_EVIDENCE_REQUIRED"
    )
)


# -----------------------------------------------------------------------------
# 17. Validation
# -----------------------------------------------------------------------------

assert len(
    accepted_harmonized
) == 96

assert len(
    family_evidence_assessment
) == 12

assert (
    family_evidence_assessment[
        "Assigned_Family"
    ].nunique()
    == 12
)


# -----------------------------------------------------------------------------
# 18. Export
# -----------------------------------------------------------------------------

harmonized_path = (
    output_dir
    / "omo_sla_accepted_values_family_harmonized.csv"
)

missing_path = (
    output_dir
    / "omo_sla_accepted_values_still_missing_family.csv"
)

assessment_path = (
    output_dir
    / "omo_sla_internal_family_evidence_assessment_harmonized.csv"
)


accepted_harmonized.to_csv(
    harmonized_path,
    index=False
)

accepted_family_missing_after_normalization.to_csv(
    missing_path,
    index=False
)

family_evidence_assessment.to_csv(
    assessment_path,
    index=False
)


# -----------------------------------------------------------------------------
# 19. Diagnostics
# -----------------------------------------------------------------------------

print("\nAccepted SLA family assignment after taxonomic harmonization")
print("-" * 100)

print(
    f"Accepted SLA taxa         : "
    f"{len(accepted_harmonized)}"
)

print(
    f"With family assignment    : "
    f"{n_with_family}"
)

print(
    f"Still without family      : "
    f"{n_without_family}"
)


print("\nAccepted taxa still lacking family")
print("-" * 100)

if accepted_family_missing_after_normalization.empty:

    print(
        "All accepted taxa now have family assignments."
    )

else:

    display(
        accepted_family_missing_after_normalization
    )


print("\nHarmonized internal family evidence assessment")
print("-" * 100)

display(
    family_evidence_assessment[
        [
            "Assigned_Family",
            "Residual_Canonical_Taxon_Count",
            "Accepted_Canonical_Taxa",
            "Accepted_Normalized_Genera",
            "Internal_Median_SLA",
            "Internal_Min_SLA",
            "Internal_Max_SLA",
            "SPECIES_TRY",
            "SPECIES_LITERATURE",
            "GENUS_ESTIMATE",
            "Internal_Family_Evidence_Strength",
            "Recommended_Family_Pathway",
        ]
    ]
)


print("\nEvidence-strength distribution")
print("-" * 100)

print(
    family_evidence_assessment[
        "Internal_Family_Evidence_Strength"
    ]
    .value_counts()
    .to_string()
)


print("\nRecommended pathways")
print("-" * 100)

print(
    family_evidence_assessment[
        "Recommended_Family_Pathway"
    ]
    .value_counts()
    .to_string()
)


print("\nExports")
print("-" * 100)

print(
    f"Harmonized accepted SLA : "
    f"{harmonized_path}"
)

print(
    f"Still missing family    : "
    f"{missing_path}"
)

print(
    f"Family assessment       : "
    f"{assessment_path}"
)


print(
    "\nSection 8.8P2 completed successfully."
)
```

### 8.8P3 Species-Evidence-Only Hierarchical Family SLA Assessment

The preliminary internal family assessment demonstrated that several residual
families are represented within the accepted Omo SLA dataset. However, family
fallback estimates should not be derived directly from all accepted values.

Two safeguards are introduced before family-level estimation.

First, historical spelling variants representing the same biological taxon are
collapsed to a single normalized evidence unit so that duplicated canonical
strings do not artificially increase taxonomic support.

Second, existing `GENUS_ESTIMATE` values are excluded from construction of
family-level estimates. This avoids nested imputation in which a taxonomic proxy
is subsequently used to generate a broader taxonomic proxy.

Internal family evidence is therefore restricted to empirical species-level
values from:

- `SPECIES_TRY`
- `SPECIES_LITERATURE`

A hierarchical estimator is then used:

1. one robust SLA value per normalized biological taxon;
2. median SLA per genus;
3. median of genus medians per family.

This gives each represented genus equal influence on the family estimate and
provides a conservative basis for determining which residual families can
support an internal `FAMILY_ESTIMATE`.

```python
# =============================================================================
# SECTION 8.8P3 — SPECIES-EVIDENCE-ONLY HIERARCHICAL FAMILY SLA ASSESSMENT
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8P3 — SPECIES-EVIDENCE-ONLY HIERARCHICAL "
    "FAMILY SLA ASSESSMENT"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

harmonized_path = (
    output_dir
    / "omo_sla_accepted_values_family_harmonized.csv"
)

family_manifest_path = (
    output_dir
    / "omo_sla_family_acquisition_manifest.csv"
)


if not harmonized_path.exists():
    raise FileNotFoundError(
        f"Harmonized accepted SLA table not found:\n{harmonized_path}"
    )

if not family_manifest_path.exists():
    raise FileNotFoundError(
        f"Family acquisition manifest not found:\n{family_manifest_path}"
    )


accepted = pd.read_csv(
    harmonized_path,
    low_memory=False
)

family_manifest = pd.read_csv(
    family_manifest_path,
    low_memory=False
)


assert len(accepted) == 96
assert len(family_manifest) == 12


# -----------------------------------------------------------------------------
# 2. Restrict family support to species-level empirical evidence
#
# GENUS_ESTIMATE is deliberately excluded to avoid nested imputation.
# -----------------------------------------------------------------------------

empirical_species = (
    accepted.loc[
        accepted[
            "Provenance_Class"
        ].isin(
            [
                "SPECIES_TRY",
                "SPECIES_LITERATURE",
            ]
        )
    ]
    .copy()
    .reset_index(drop=True)
)


print("\nEmpirical species-level evidence")
print("-" * 100)

print(
    f"All accepted SLA taxa       : "
    f"{len(accepted)}"
)

print(
    f"Species-level evidence rows : "
    f"{len(empirical_species)}"
)

print(
    "Excluded genus-estimate rows: "
    f"{(accepted['Provenance_Class'] == 'GENUS_ESTIMATE').sum()}"
)


# -----------------------------------------------------------------------------
# 3. Build normalized biological taxon key
#
# Prefer recovered normalized name.
# If absent, use canonical species.
#
# Some controlled overrides are supplied for known orthographic variants that
# already passed taxonomic review earlier in the SLA workflow.
# -----------------------------------------------------------------------------

empirical_species[
    "Evidence_Taxon_Name"
] = (
    empirical_species[
        "Recovered_Normalized_Name"
    ]
    .fillna(
        empirical_species[
            "Canonical_Species"
        ]
    )
    .astype(str)
    .str.strip()
)


controlled_taxon_name_overrides = {

    "Annonidium manni":
        "Anonidium mannii",

    "Corynante pachycercus":
        "Corynanthe pachyceras",

    "Diospyrous melistoformis":
        "Diospyros mespiliformis",

    "Diospyrous mespiliformis":
        "Diospyros mespiliformis",

    "Irvinga gabonensis":
        "Irvingia gabonensis",

    "Margartaria discoidea":
        "Margaritaria discoidea",

    "Nesodogornia papaverifera":
        "Nesogordonia papaverifera",

    "Nesodogornia papaverivera":
        "Nesogordonia papaverifera",

    "Pycanthus angolensis":
        "Pycnanthus angolensis",

    "Ricinidendron heudeloti":
        "Ricinodendron heudelotii",

    "Spondia mombin":
        "Spondias mombin",

    "Strombosia postulata":
        "Strombosia pustulata",

    "Strombosia putulata":
        "Strombosia pustulata",
}


for canonical_name, normalized_name in (
    controlled_taxon_name_overrides.items()
):

    mask = (
        empirical_species[
            "Canonical_Species"
        ].eq(canonical_name)
    )

    empirical_species.loc[
        mask,
        "Evidence_Taxon_Name"
    ] = normalized_name


# -----------------------------------------------------------------------------
# 4. Derive normalized evidence genus
# -----------------------------------------------------------------------------

empirical_species[
    "Evidence_Genus"
] = (
    empirical_species[
        "Evidence_Taxon_Name"
    ]
    .astype(str)
    .str.split()
    .str[0]
)


# -----------------------------------------------------------------------------
# 5. Preserve only records with a usable family assignment
# -----------------------------------------------------------------------------

empirical_with_family = (
    empirical_species.loc[
        empirical_species[
            "Assigned_Family"
        ].notna()
    ]
    .copy()
)


target_families = set(
    family_manifest[
        "Assigned_Family"
    ]
)


empirical_target_family = (
    empirical_with_family.loc[
        empirical_with_family[
            "Assigned_Family"
        ].isin(
            target_families
        )
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 6. Collapse canonical spelling variants to one biological taxon
#
# If several canonical strings map to the same biological taxon and contain
# the same accepted value, they contribute only once.
#
# Median is used defensively in case multiple legitimate empirical estimates
# exist for the same normalized taxon.
# -----------------------------------------------------------------------------

taxon_level_evidence = (
    empirical_target_family
    .groupby(
        [
            "Assigned_Family",
            "Evidence_Genus",
            "Evidence_Taxon_Name",
        ],
        as_index=False
    )
    .agg(
        Taxon_SLA=(
            "Standardized_SLA_Value",
            "median"
        ),

        Canonical_Record_Count=(
            "Canonical_Species",
            "nunique"
        ),

        Canonical_Names=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),

        Evidence_Provenance=(
            "Provenance_Class",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
)


# -----------------------------------------------------------------------------
# 7. Genus-level aggregation
#
# Each empirical species contributes to a genus median.
# -----------------------------------------------------------------------------

genus_level_evidence = (
    taxon_level_evidence
    .groupby(
        [
            "Assigned_Family",
            "Evidence_Genus",
        ],
        as_index=False
    )
    .agg(
        Empirical_Taxon_Count=(
            "Evidence_Taxon_Name",
            "nunique"
        ),

        Genus_Median_SLA=(
            "Taxon_SLA",
            "median"
        ),

        Genus_Min_SLA=(
            "Taxon_SLA",
            "min"
        ),

        Genus_Max_SLA=(
            "Taxon_SLA",
            "max"
        ),

        Supporting_Taxa=(
            "Evidence_Taxon_Name",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
)


# -----------------------------------------------------------------------------
# 8. Family-level candidate = median of genus medians
# -----------------------------------------------------------------------------

family_hierarchical_summary = (
    genus_level_evidence
    .groupby(
        "Assigned_Family",
        as_index=False
    )
    .agg(
        Empirical_Genus_Count=(
            "Evidence_Genus",
            "nunique"
        ),

        Empirical_Taxon_Count=(
            "Empirical_Taxon_Count",
            "sum"
        ),

        Family_GenusMedian_Min=(
            "Genus_Median_SLA",
            "min"
        ),

        Candidate_Family_SLA=(
            "Genus_Median_SLA",
            "median"
        ),

        Family_GenusMedian_Mean=(
            "Genus_Median_SLA",
            "mean"
        ),

        Family_GenusMedian_Max=(
            "Genus_Median_SLA",
            "max"
        ),

        Supporting_Genera=(
            "Evidence_Genus",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
)


# -----------------------------------------------------------------------------
# 9. Add species-level provenance composition
# -----------------------------------------------------------------------------

provenance_summary = (
    empirical_target_family
    .groupby(
        [
            "Assigned_Family",
            "Provenance_Class",
        ]
    )
    .size()
    .unstack(
        fill_value=0
    )
    .reset_index()
)


for col in [
    "SPECIES_TRY",
    "SPECIES_LITERATURE",
]:

    if col not in provenance_summary.columns:
        provenance_summary[col] = 0


# -----------------------------------------------------------------------------
# 10. Merge with all 12 residual families
# -----------------------------------------------------------------------------

family_hierarchical_assessment = (
    family_manifest
    .merge(
        family_hierarchical_summary,
        how="left",
        on="Assigned_Family"
    )
    .merge(
        provenance_summary[
            [
                "Assigned_Family",
                "SPECIES_TRY",
                "SPECIES_LITERATURE",
            ]
        ],
        how="left",
        on="Assigned_Family"
    )
)


count_cols = [
    "Empirical_Genus_Count",
    "Empirical_Taxon_Count",
    "SPECIES_TRY",
    "SPECIES_LITERATURE",
]


family_hierarchical_assessment[
    count_cols
] = (
    family_hierarchical_assessment[
        count_cols
    ]
    .fillna(0)
    .astype(int)
)


# -----------------------------------------------------------------------------
# 11. Evidence-strength rules
#
# VERY_STRONG:
#     >=5 empirical genera
#
# STRONG:
#     >=3 empirical genera
#
# MODERATE:
#     2 empirical genera
#
# SINGLE_GENUS:
#     1 empirical genus
#
# NONE:
#     0 empirical genera
#
# Genus breadth is emphasized because the family estimate is the median of
# independent genus medians.
# -----------------------------------------------------------------------------

def classify_hierarchical_family(row):

    n_genera = row[
        "Empirical_Genus_Count"
    ]

    if n_genera >= 5:
        return "VERY_STRONG"

    if n_genera >= 3:
        return "STRONG"

    if n_genera == 2:
        return "MODERATE"

    if n_genera == 1:
        return "SINGLE_GENUS"

    return "NONE"


family_hierarchical_assessment[
    "Family_Evidence_Strength"
] = (
    family_hierarchical_assessment.apply(
        classify_hierarchical_family,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 12. Controlled acceptance recommendation
#
# At this stage:
#
# VERY_STRONG / STRONG:
#     suitable internal family estimate
#
# MODERATE:
#     retain as candidate but require review
#
# SINGLE_GENUS / NONE:
#     external evidence required
# -----------------------------------------------------------------------------

family_hierarchical_assessment[
    "Family_Estimate_Recommendation"
] = np.select(
    [
        family_hierarchical_assessment[
            "Family_Evidence_Strength"
        ].isin(
            [
                "VERY_STRONG",
                "STRONG",
            ]
        ),

        family_hierarchical_assessment[
            "Family_Evidence_Strength"
        ].eq(
            "MODERATE"
        ),
    ],
    [
        "ACCEPT_INTERNAL_FAMILY_CANDIDATE",
        "REVIEW_INTERNAL_FAMILY_CANDIDATE",
    ],
    default=(
        "EXTERNAL_FAMILY_EVIDENCE_REQUIRED"
    )
)


# -----------------------------------------------------------------------------
# 13. Candidate unit and provenance
# -----------------------------------------------------------------------------

family_hierarchical_assessment[
    "Candidate_Family_SLA_Unit"
] = "mm2 mg-1"

family_hierarchical_assessment[
    "Planned_Provenance_Class"
] = "FAMILY_ESTIMATE"

family_hierarchical_assessment[
    "Aggregation_Method"
] = (
    "MEDIAN_OF_EMPIRICAL_GENUS_MEDIANS"
)


# -----------------------------------------------------------------------------
# 14. Validation
# -----------------------------------------------------------------------------

assert len(
    family_hierarchical_assessment
) == 12

assert (
    family_hierarchical_assessment[
        "Assigned_Family"
    ].nunique()
    == 12
)


# -----------------------------------------------------------------------------
# 15. Export
# -----------------------------------------------------------------------------

taxon_path = (
    output_dir
    / "omo_sla_family_empirical_taxon_evidence.csv"
)

genus_path = (
    output_dir
    / "omo_sla_family_empirical_genus_evidence.csv"
)

assessment_path = (
    output_dir
    / "omo_sla_family_hierarchical_evidence_assessment.csv"
)


taxon_level_evidence.to_csv(
    taxon_path,
    index=False
)

genus_level_evidence.to_csv(
    genus_path,
    index=False
)

family_hierarchical_assessment.to_csv(
    assessment_path,
    index=False
)


# -----------------------------------------------------------------------------
# 16. Diagnostics
# -----------------------------------------------------------------------------

print("\nSpecies-evidence-only family assessment")
print("-" * 100)

display(
    family_hierarchical_assessment[
        [
            "Assigned_Family",
            "Residual_Canonical_Taxon_Count",
            "Empirical_Taxon_Count",
            "Empirical_Genus_Count",
            "Candidate_Family_SLA",
            "Family_GenusMedian_Min",
            "Family_GenusMedian_Max",
            "SPECIES_TRY",
            "SPECIES_LITERATURE",
            "Family_Evidence_Strength",
            "Family_Estimate_Recommendation",
        ]
    ]
)


print("\nFamily evidence-strength distribution")
print("-" * 100)

print(
    family_hierarchical_assessment[
        "Family_Evidence_Strength"
    ]
    .value_counts()
    .to_string()
)


print("\nFamily recommendation distribution")
print("-" * 100)

print(
    family_hierarchical_assessment[
        "Family_Estimate_Recommendation"
    ]
    .value_counts()
    .to_string()
)


print("\nEmpirical genus-level support")
print("-" * 100)

display(
    genus_level_evidence[
        [
            "Assigned_Family",
            "Evidence_Genus",
            "Empirical_Taxon_Count",
            "Genus_Median_SLA",
            "Supporting_Taxa",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Taxon evidence  : "
    f"{taxon_path}"
)

print(
    f"Genus evidence  : "
    f"{genus_path}"
)

print(
    f"Family assessment: "
    f"{assessment_path}"
)


print(
    "\nSection 8.8P3 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8Q — ACCEPT STRONG INTERNAL FAMILY SLA ESTIMATES
#                AND IDENTIFY REMAINING EXTERNAL FALLBACK FAMILIES
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8Q — ACCEPT STRONG INTERNAL FAMILY SLA ESTIMATES "
    "AND IDENTIFY REMAINING EXTERNAL FALLBACK FAMILIES"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

family_assessment_path = (
    output_dir
    / "omo_sla_family_hierarchical_evidence_assessment.csv"
)

canonical_targets_path = (
    output_dir
    / "omo_sla_family_fallback_canonical_targets.csv"
)

if not family_assessment_path.exists():
    raise FileNotFoundError(
        f"Family assessment not found:\n{family_assessment_path}"
    )

if not canonical_targets_path.exists():
    raise FileNotFoundError(
        f"Family fallback target table not found:\n{canonical_targets_path}"
    )


family_assessment = pd.read_csv(
    family_assessment_path
)

family_targets = pd.read_csv(
    canonical_targets_path
)


# -----------------------------------------------------------------------------
# 2. Select only strong internal family candidates
# -----------------------------------------------------------------------------

accepted_family_estimates = (
    family_assessment.loc[
        family_assessment[
            "Family_Estimate_Recommendation"
        ].eq(
            "ACCEPT_INTERNAL_FAMILY_CANDIDATE"
        )
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(accepted_family_estimates) == 3, (
    f"Expected 3 accepted internal family candidates, "
    f"found {len(accepted_family_estimates)}."
)


# -----------------------------------------------------------------------------
# 3. Add acceptance metadata
# -----------------------------------------------------------------------------

accepted_family_estimates[
    "Accepted_Family_SLA"
] = (
    accepted_family_estimates[
        "Candidate_Family_SLA"
    ]
)

accepted_family_estimates[
    "Accepted_Family_SLA_Unit"
] = "mm2 mg-1"

accepted_family_estimates[
    "Accepted_Provenance_Class"
] = "FAMILY_ESTIMATE"

accepted_family_estimates[
    "Accepted_Source_Class"
] = (
    "INTERNAL_EMPIRICAL_SPECIES_EVIDENCE"
)

accepted_family_estimates[
    "Accepted_Quality_Flag"
] = (
    "HIERARCHICAL_MEDIAN_OF_EMPIRICAL_GENUS_MEDIANS"
)

accepted_family_estimates[
    "Acceptance_Status"
] = "ACCEPTED_INTERNAL_FAMILY_ESTIMATE"


# -----------------------------------------------------------------------------
# 4. Attach accepted family estimates to unresolved canonical taxa
# -----------------------------------------------------------------------------

family_lookup = (
    accepted_family_estimates[
        [
            "Assigned_Family",
            "Accepted_Family_SLA",
            "Accepted_Family_SLA_Unit",
            "Family_Evidence_Strength",
            "Empirical_Taxon_Count",
            "Empirical_Genus_Count",
            "Supporting_Genera",
            "Accepted_Provenance_Class",
            "Accepted_Source_Class",
            "Accepted_Quality_Flag",
            "Acceptance_Status",
        ]
    ]
    .copy()
)


targets_with_family_estimate = (
    family_targets
    .merge(
        family_lookup,
        how="left",
        on="Assigned_Family",
        validate="many_to_one"
    )
)


# -----------------------------------------------------------------------------
# 5. Identify newly covered taxa
# -----------------------------------------------------------------------------

new_family_values = (
    targets_with_family_estimate.loc[
        targets_with_family_estimate[
            "Accepted_Family_SLA"
        ].notna()
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 6. Build Omo-ready family SLA table
# -----------------------------------------------------------------------------

omo_family_values = (
    new_family_values[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Assigned_Family",
            "Accepted_Family_SLA",
            "Accepted_Family_SLA_Unit",
            "Family_Evidence_Strength",
            "Empirical_Taxon_Count",
            "Empirical_Genus_Count",
            "Supporting_Genera",
            "Accepted_Provenance_Class",
            "Accepted_Source_Class",
            "Accepted_Quality_Flag",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Accepted_Family_SLA":
                "Standardized_SLA_Value",

            "Accepted_Family_SLA_Unit":
                "Standardized_SLA_Unit",

            "Accepted_Provenance_Class":
                "Provenance_Class",

            "Accepted_Source_Class":
                "Trait_Source",

            "Accepted_Quality_Flag":
                "Quality_Flag",
        }
    )
)


# -----------------------------------------------------------------------------
# 7. Coverage accounting
#
# Current coverage before family estimates:
#     96 / 127
#
# Count how many canonical residual taxa fall into the three accepted families.
# -----------------------------------------------------------------------------

new_family_taxa = (
    omo_family_values[
        "Canonical_Species"
    ].nunique()
)

coverage_before_family = 96

coverage_after_family = (
    coverage_before_family
    + new_family_taxa
)

remaining_after_family = (
    127
    - coverage_after_family
)

coverage_percent = (
    100
    * coverage_after_family
    / 127
)


# -----------------------------------------------------------------------------
# 8. Remaining unresolved family targets
# -----------------------------------------------------------------------------

covered_taxa = set(
    omo_family_values[
        "Canonical_Species"
    ]
)

remaining_family_targets = (
    family_targets.loc[
        ~family_targets[
            "Canonical_Species"
        ].isin(
            covered_taxa
        )
    ]
    .copy()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 9. Remaining family summary
# -----------------------------------------------------------------------------

remaining_family_summary = (
    remaining_family_targets
    .groupby(
        "Assigned_Family",
        as_index=False
    )
    .agg(
        Remaining_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Remaining_Normalized_Targets=(
            "Normalized_Search_Name",
            "nunique"
        ),

        Remaining_Genera=(
            "Normalized_Acquisition_Genus",
            "nunique"
        ),
    )
)


# Attach family evidence status
remaining_family_summary = (
    remaining_family_summary
    .merge(
        family_assessment[
            [
                "Assigned_Family",
                "Family_Evidence_Strength",
                "Candidate_Family_SLA",
                "Family_Estimate_Recommendation",
            ]
        ],
        how="left",
        on="Assigned_Family",
        validate="one_to_one"
    )
)


# -----------------------------------------------------------------------------
# 10. Validation
# -----------------------------------------------------------------------------

assert (
    accepted_family_estimates[
        "Assigned_Family"
    ].nunique()
    == 3
)

assert (
    omo_family_values[
        "Standardized_SLA_Value"
    ].notna().all()
)

assert (
    omo_family_values[
        "Standardized_SLA_Unit"
    ].eq("mm2 mg-1").all()
)

assert (
    omo_family_values[
        "Provenance_Class"
    ].eq("FAMILY_ESTIMATE").all()
)

assert (
    new_family_taxa
    + len(remaining_family_targets)
    == 31
)


# -----------------------------------------------------------------------------
# 11. Export
# -----------------------------------------------------------------------------

accepted_family_path = (
    output_dir
    / "omo_sla_internal_family_estimates_accepted.csv"
)

omo_family_values_path = (
    output_dir
    / "omo_sla_internal_family_omo_values.csv"
)

remaining_targets_path = (
    output_dir
    / "omo_sla_remaining_targets_after_internal_family_estimates.csv"
)

remaining_family_summary_path = (
    output_dir
    / "omo_sla_remaining_external_family_requirements.csv"
)


accepted_family_estimates.to_csv(
    accepted_family_path,
    index=False
)

omo_family_values.to_csv(
    omo_family_values_path,
    index=False
)

remaining_family_targets.to_csv(
    remaining_targets_path,
    index=False
)

remaining_family_summary.to_csv(
    remaining_family_summary_path,
    index=False
)


# -----------------------------------------------------------------------------
# 12. Diagnostics
# -----------------------------------------------------------------------------

print("\nAccepted internal family SLA estimates")
print("-" * 100)

display(
    accepted_family_estimates[
        [
            "Assigned_Family",
            "Accepted_Family_SLA",
            "Accepted_Family_SLA_Unit",
            "Empirical_Taxon_Count",
            "Empirical_Genus_Count",
            "Supporting_Genera",
            "Family_Evidence_Strength",
            "Accepted_Quality_Flag",
        ]
    ]
)


print("\nOmo taxa receiving internal family SLA estimates")
print("-" * 100)

display(
    omo_family_values[
        [
            "Canonical_Species",
            "Assigned_Family",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "Family_Evidence_Strength",
            "Quality_Flag",
            "Provenance_Class",
        ]
    ]
)


print("\nCoverage after internal family estimates")
print("-" * 100)

print(
    f"Coverage before family estimates : "
    f"{coverage_before_family} / 127"
)

print(
    f"New family-level coverage        : "
    f"{new_family_taxa}"
)

print(
    f"Current SLA coverage             : "
    f"{coverage_after_family} / 127"
)

print(
    f"Coverage percent                 : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps               : "
    f"{remaining_after_family}"
)


print("\nFamilies still requiring external evidence")
print("-" * 100)

display(
    remaining_family_summary
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted family estimates : "
    f"{accepted_family_path}"
)

print(
    f"Omo family values         : "
    f"{omo_family_values_path}"
)

print(
    f"Remaining targets          : "
    f"{remaining_targets_path}"
)

print(
    f"External family needs      : "
    f"{remaining_family_summary_path}"
)


print(
    "\nSection 8.8Q completed successfully."
)
```

## Section 8.8R1 — External Family SLA Evidence Acquisition Manifest

Following acceptance of strong internally derived family estimates for
Apocynaceae, Fabaceae and Malvaceae, 22 canonical Omo taxa remain unresolved
across nine families.

The remaining families differ substantially in their existing empirical support.
Accordingly, external evidence acquisition is stratified into three pathways:

1. **External corroboration**
   - families already possessing a moderate internal family candidate;
2. **External augmentation**
   - families represented internally by only one empirical genus;
3. **External primary acquisition**
   - families with no usable internal empirical family support.

External family estimates will preferentially be based on multi-species,
multi-genus SLA evidence from recognized trait databases or peer-reviewed
trait syntheses. Family estimates will not be constructed from a single
congeneric species where broader family evidence is required.

No new numerical SLA values are assigned in this subsection.

```python
# =============================================================================
# SECTION 8.8R1 — EXTERNAL FAMILY SLA EVIDENCE ACQUISITION MANIFEST
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8R1 — EXTERNAL FAMILY SLA EVIDENCE ACQUISITION MANIFEST")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

family_requirements_path = (
    output_dir
    / "omo_sla_remaining_external_family_requirements.csv"
)

remaining_targets_path = (
    output_dir
    / "omo_sla_remaining_targets_after_internal_family_estimates.csv"
)


for path in [
    family_requirements_path,
    remaining_targets_path,
]:
    if not path.exists():
        raise FileNotFoundError(
            f"Required Section 8.8Q output not found:\n{path}"
        )


family_requirements = pd.read_csv(
    family_requirements_path
)

remaining_targets = pd.read_csv(
    remaining_targets_path
)


# -----------------------------------------------------------------------------
# 2. Validate locked 8.8Q state
# -----------------------------------------------------------------------------

assert len(remaining_targets) == 22, (
    f"Expected 22 residual canonical taxa, "
    f"found {len(remaining_targets)}."
)

assert (
    family_requirements[
        "Assigned_Family"
    ].nunique()
    == 9
), (
    "Expected 9 residual families."
)


# -----------------------------------------------------------------------------
# 3. Assign controlled external-evidence pathway
# -----------------------------------------------------------------------------

def external_pathway(row):

    strength = row[
        "Family_Evidence_Strength"
    ]

    if strength == "MODERATE":
        return "EXTERNAL_CORROBORATION"

    if strength == "SINGLE_GENUS":
        return "EXTERNAL_AUGMENTATION"

    if strength == "NONE":
        return "EXTERNAL_PRIMARY_ACQUISITION"

    return "EXTERNAL_REVIEW"


family_external_manifest = (
    family_requirements.copy()
)

family_external_manifest[
    "External_Acquisition_Pathway"
] = (
    family_external_manifest.apply(
        external_pathway,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 4. Assign search priority
#
# Priority 1:
#   no internal empirical family evidence
#
# Priority 2:
#   moderate candidate needing corroboration
#
# Priority 3:
#   single-genus evidence requiring augmentation
# -----------------------------------------------------------------------------

priority_map = {
    "EXTERNAL_PRIMARY_ACQUISITION": 1,
    "EXTERNAL_CORROBORATION": 2,
    "EXTERNAL_AUGMENTATION": 3,
    "EXTERNAL_REVIEW": 4,
}

family_external_manifest[
    "External_Search_Priority"
] = (
    family_external_manifest[
        "External_Acquisition_Pathway"
    ].map(priority_map)
)


# -----------------------------------------------------------------------------
# 5. Evidence requirements
# -----------------------------------------------------------------------------

def minimum_evidence_requirement(row):

    pathway = row[
        "External_Acquisition_Pathway"
    ]

    if pathway == "EXTERNAL_CORROBORATION":
        return (
            "MULTI_GENUS_EXTERNAL_EVIDENCE_TO_VALIDATE_INTERNAL_CANDIDATE"
        )

    if pathway == "EXTERNAL_AUGMENTATION":
        return (
            "AT_LEAST_2_ADDITIONAL_EXTERNAL_GENERA_PREFERRED"
        )

    if pathway == "EXTERNAL_PRIMARY_ACQUISITION":
        return (
            "MULTI_SPECIES_MULTI_GENUS_EXTERNAL_FAMILY_EVIDENCE"
        )

    return "MANUAL_REVIEW"


family_external_manifest[
    "Minimum_Evidence_Requirement"
] = (
    family_external_manifest.apply(
        minimum_evidence_requirement,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 6. Preferred source hierarchy
# -----------------------------------------------------------------------------

family_external_manifest[
    "Preferred_Source_1"
] = "RECOGNIZED_TRAIT_DATABASE"

family_external_manifest[
    "Preferred_Source_2"
] = "PEER_REVIEWED_TRAIT_SYNTHESIS"

family_external_manifest[
    "Preferred_Source_3"
] = "SCIENTIFIC_DATA_REPOSITORY"

family_external_manifest[
    "Final_Fallback_Source"
] = "CONTROLLED_GLOBAL_SLA_DISTRIBUTION"


# -----------------------------------------------------------------------------
# 7. Query metadata
# -----------------------------------------------------------------------------

family_external_manifest[
    "Requested_Trait"
] = "specific leaf area"

family_external_manifest[
    "Target_Unit"
] = "mm2 mg-1"

family_external_manifest[
    "Planned_Provenance_Class"
] = "FAMILY_ESTIMATE"

family_external_manifest[
    "External_Search_Status"
] = "READY_FOR_EXTERNAL_FAMILY_QUERY"


# -----------------------------------------------------------------------------
# 8. Generate family query IDs
# -----------------------------------------------------------------------------

family_external_manifest = (
    family_external_manifest
    .sort_values(
        [
            "External_Search_Priority",
            "Assigned_Family",
        ]
    )
    .reset_index(drop=True)
)


family_external_manifest.insert(
    0,
    "External_Family_Query_ID",
    [
        f"OMO_SLA_EXT_FAMILY_{i:03d}"
        for i in range(
            1,
            len(family_external_manifest) + 1
        )
    ]
)


# -----------------------------------------------------------------------------
# 9. Attach residual taxa to family search manifest
# -----------------------------------------------------------------------------

family_target_detail = (
    remaining_targets
    .groupby(
        "Assigned_Family",
        as_index=False
    )
    .agg(
        Residual_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Residual_Normalized_Targets=(
            "Normalized_Search_Name",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Residual_Genera=(
            "Normalized_Acquisition_Genus",
            lambda x: " | ".join(
                sorted(
                    pd.Series(x)
                    .dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
)


family_external_manifest = (
    family_external_manifest
    .merge(
        family_target_detail,
        how="left",
        on="Assigned_Family",
        validate="one_to_one"
    )
)


# -----------------------------------------------------------------------------
# 10. Build compact family-name file
# -----------------------------------------------------------------------------

external_family_names = (
    family_external_manifest[
        "Assigned_Family"
    ]
    .drop_duplicates()
    .tolist()
)


# -----------------------------------------------------------------------------
# 11. Validation
# -----------------------------------------------------------------------------

assert len(
    family_external_manifest
) == 9

assert (
    family_external_manifest[
        "Remaining_Canonical_Taxa"
    ].sum()
    == 22
)

assert (
    family_external_manifest[
        "Assigned_Family"
    ].nunique()
    == 9
)


# -----------------------------------------------------------------------------
# 12. Export
# -----------------------------------------------------------------------------

manifest_path = (
    output_dir
    / "omo_sla_external_family_acquisition_manifest.csv"
)

target_detail_path = (
    output_dir
    / "omo_sla_external_family_target_detail.csv"
)

family_names_path = (
    output_dir
    / "omo_sla_external_family_names.txt"
)


family_external_manifest.to_csv(
    manifest_path,
    index=False
)

remaining_targets.to_csv(
    target_detail_path,
    index=False
)

with open(
    family_names_path,
    "w",
    encoding="utf-8"
) as f:

    for family in external_family_names:
        f.write(
            f"{family}\n"
        )


# -----------------------------------------------------------------------------
# 13. Diagnostics
# -----------------------------------------------------------------------------

print("\nExternal family acquisition summary")
print("-" * 100)

print(
    f"Residual canonical taxa      : "
    f"{len(remaining_targets)}"
)

print(
    f"Families requiring evidence  : "
    f"{len(family_external_manifest)}"
)


print("\nExternal acquisition pathway distribution")
print("-" * 100)

print(
    family_external_manifest[
        "External_Acquisition_Pathway"
    ]
    .value_counts()
    .to_string()
)


print("\nExternal family acquisition manifest")
print("-" * 100)

display(
    family_external_manifest[
        [
            "External_Family_Query_ID",
            "Assigned_Family",
            "Remaining_Canonical_Taxa",
            "Remaining_Normalized_Targets",
            "Remaining_Genera",
            "Family_Evidence_Strength",
            "Candidate_Family_SLA",
            "External_Acquisition_Pathway",
            "External_Search_Priority",
            "Minimum_Evidence_Requirement",
        ]
    ]
)


print("\nFamilies requiring external acquisition")
print("-" * 100)

for family in external_family_names:
    print(family)


print("\nExports")
print("-" * 100)

print(
    f"External family manifest : "
    f"{manifest_path}"
)

print(
    f"Residual target detail   : "
    f"{target_detail_path}"
)

print(
    f"Family-name list         : "
    f"{family_names_path}"
)


print(
    "\nSection 8.8R1 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8R2 — IMPORT, QC AND HIERARCHICAL AGGREGATION OF
#                  EXTERNAL BIEN FAMILY SLA EVIDENCE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8R2 — IMPORT, QC AND HIERARCHICAL AGGREGATION "
    "OF EXTERNAL BIEN FAMILY SLA EVIDENCE"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

bien_path = (
    PROJECT_ROOT
    / "data"
    / "external"
    / "BIEN"
    / "omo_sla_bien_external_family_raw.csv"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

if not bien_path.exists():
    raise FileNotFoundError(
        f"External-family BIEN file not found:\n{bien_path}"
    )


# -----------------------------------------------------------------------------
# 2. Import raw external-family BIEN data
# -----------------------------------------------------------------------------

bien_family_raw = pd.read_csv(
    bien_path,
    low_memory=False
)

print("\nRaw BIEN external-family import")
print("-" * 100)

print(
    f"Rows imported    : "
    f"{len(bien_family_raw)}"
)

print(
    f"Columns imported : "
    f"{len(bien_family_raw.columns)}"
)


# -----------------------------------------------------------------------------
# 3. Validate expected columns
# -----------------------------------------------------------------------------

required_columns = [
    "scrubbed_family",
    "scrubbed_genus",
    "scrubbed_species_binomial",
    "trait_name",
    "trait_value",
    "unit",
]

missing_columns = [
    col
    for col in required_columns
    if col not in bien_family_raw.columns
]

if missing_columns:
    raise KeyError(
        "Missing required BIEN columns:\n"
        + "\n".join(missing_columns)
    )


# -----------------------------------------------------------------------------
# 4. Restrict to intended SLA trait
# -----------------------------------------------------------------------------

bien_family = (
    bien_family_raw.loc[
        bien_family_raw[
            "trait_name"
        ].eq(
            "leaf area per leaf dry mass"
        )
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 5. Standardize numeric SLA
#
# BIEN unit:
#     m2 kg-1
#
# Numerically equivalent to:
#     mm2 mg-1
# -----------------------------------------------------------------------------

bien_family[
    "SLA_Std"
] = pd.to_numeric(
    bien_family[
        "trait_value"
    ],
    errors="coerce"
)

bien_family[
    "SLA_Std_Unit"
] = "mm2 mg-1"


# -----------------------------------------------------------------------------
# 6. Basic validity filter
# -----------------------------------------------------------------------------

bien_family[
    "QC_Status"
] = "RETAIN"


bien_family.loc[
    bien_family[
        "SLA_Std"
    ].isna(),
    "QC_Status"
] = "EXCLUDE_NON_NUMERIC"


bien_family.loc[
    bien_family[
        "SLA_Std"
    ] <= 0,
    "QC_Status"
] = "EXCLUDE_NON_POSITIVE"


# -----------------------------------------------------------------------------
# 7. Remove clear low-scale anomalies
#
# Previous BIEN extractions showed some values around 0.00001 that are
# almost certainly unit/scale conversion artifacts.
#
# Use a conservative threshold:
#     SLA < 0.5 mm2 mg-1
#
# This is only an anomaly screen, not a biological upper-bound filter.
# -----------------------------------------------------------------------------

bien_family.loc[
    (
        bien_family[
            "QC_Status"
        ].eq("RETAIN")
        &
        (
            bien_family[
                "SLA_Std"
            ] < 0.5
        )
    ),
    "QC_Status"
] = "EXCLUDE_LOW_SCALE_ANOMALY"


# -----------------------------------------------------------------------------
# 8. Family-specific IQR screening
#
# Apply only after removing obvious scale errors.
# -----------------------------------------------------------------------------

pre_iqr = (
    bien_family.loc[
        bien_family[
            "QC_Status"
        ].eq("RETAIN")
    ]
    .copy()
)


family_bounds = (
    pre_iqr
    .groupby(
        "scrubbed_family"
    )[
        "SLA_Std"
    ]
    .quantile(
        [0.25, 0.75]
    )
    .unstack()
    .reset_index()
    .rename(
        columns={
            0.25: "Q1",
            0.75: "Q3",
        }
    )
)

family_bounds[
    "IQR"
] = (
    family_bounds[
        "Q3"
    ]
    -
    family_bounds[
        "Q1"
    ]
)

family_bounds[
    "Lower_IQR_Bound"
] = (
    family_bounds[
        "Q1"
    ]
    -
    1.5
    * family_bounds[
        "IQR"
    ]
)

family_bounds[
    "Upper_IQR_Bound"
] = (
    family_bounds[
        "Q3"
    ]
    +
    1.5
    * family_bounds[
        "IQR"
    ]
)


bien_family = (
    bien_family
    .merge(
        family_bounds[
            [
                "scrubbed_family",
                "Lower_IQR_Bound",
                "Upper_IQR_Bound",
            ]
        ],
        how="left",
        on="scrubbed_family"
    )
)


iqr_outlier_mask = (
    bien_family[
        "QC_Status"
    ].eq("RETAIN")
    &
    (
        (
            bien_family[
                "SLA_Std"
            ]
            <
            bien_family[
                "Lower_IQR_Bound"
            ]
        )
        |
        (
            bien_family[
                "SLA_Std"
            ]
            >
            bien_family[
                "Upper_IQR_Bound"
            ]
        )
    )
)


bien_family.loc[
    iqr_outlier_mask,
    "QC_Status"
] = "EXCLUDE_IQR_OUTLIER"


# -----------------------------------------------------------------------------
# 9. Retained measurements
# -----------------------------------------------------------------------------

bien_family_retained = (
    bien_family.loc[
        bien_family[
            "QC_Status"
        ].eq("RETAIN")
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 10. Species-level aggregation
#
# Species median prevents repeated measurements from dominating.
# -----------------------------------------------------------------------------

species_level = (
    bien_family_retained
    .groupby(
        [
            "scrubbed_family",
            "scrubbed_genus",
            "scrubbed_species_binomial",
        ],
        as_index=False
    )
    .agg(
        Species_Record_Count=(
            "SLA_Std",
            "size"
        ),

        Species_Median_SLA=(
            "SLA_Std",
            "median"
        ),

        Species_Min_SLA=(
            "SLA_Std",
            "min"
        ),

        Species_Max_SLA=(
            "SLA_Std",
            "max"
        ),
    )
)


# -----------------------------------------------------------------------------
# 11. Genus-level aggregation
#
# Each species receives equal weight within genus.
# -----------------------------------------------------------------------------

genus_level = (
    species_level
    .groupby(
        [
            "scrubbed_family",
            "scrubbed_genus",
        ],
        as_index=False
    )
    .agg(
        Genus_Species_Count=(
            "scrubbed_species_binomial",
            "nunique"
        ),

        Genus_Median_SLA=(
            "Species_Median_SLA",
            "median"
        ),

        Genus_Min_Species_Median=(
            "Species_Median_SLA",
            "min"
        ),

        Genus_Max_Species_Median=(
            "Species_Median_SLA",
            "max"
        ),
    )
)


# -----------------------------------------------------------------------------
# 12. Family-level hierarchical aggregation
#
# Family candidate = median of genus medians.
# -----------------------------------------------------------------------------

family_level = (
    genus_level
    .groupby(
        "scrubbed_family",
        as_index=False
    )
    .agg(
        External_Genus_Count=(
            "scrubbed_genus",
            "nunique"
        ),

        External_Species_Count=(
            "Genus_Species_Count",
            "sum"
        ),

        External_Family_SLA=(
            "Genus_Median_SLA",
            "median"
        ),

        External_GenusMedian_Min=(
            "Genus_Median_SLA",
            "min"
        ),

        External_GenusMedian_Max=(
            "Genus_Median_SLA",
            "max"
        ),

        Supporting_Genera=(
            "scrubbed_genus",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
)


family_level[
    "Standardized_SLA_Unit"
] = "mm2 mg-1"

family_level[
    "Aggregation_Method"
] = (
    "MEDIAN_OF_GENUS_MEDIANS_FROM_SPECIES_MEDIANS"
)


# -----------------------------------------------------------------------------
# 13. External evidence strength
# -----------------------------------------------------------------------------

def classify_external_strength(row):

    n_genera = row[
        "External_Genus_Count"
    ]

    n_species = row[
        "External_Species_Count"
    ]

    if (
        n_genera >= 10
        and n_species >= 20
    ):
        return "VERY_STRONG"

    if (
        n_genera >= 5
        and n_species >= 10
    ):
        return "STRONG"

    if n_genera >= 3:
        return "MODERATE"

    if n_genera >= 2:
        return "LIMITED"

    if n_genera == 1:
        return "SINGLE_GENUS"

    return "NONE"


family_level[
    "External_Evidence_Strength"
] = (
    family_level.apply(
        classify_external_strength,
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 14. Attach all 9 expected families, including missing Musaceae
# -----------------------------------------------------------------------------

expected_families = [
    "Gentianaceae",
    "Musaceae",
    "Sapotaceae",
    "Rubiaceae",
    "Anacardiaceae",
    "Annonaceae",
    "Phyllanthaceae",
    "Salicaceae",
    "Sapindaceae",
]


family_coverage = pd.DataFrame(
    {
        "Assigned_Family":
            expected_families
    }
)


family_coverage = (
    family_coverage
    .merge(
        family_level.rename(
            columns={
                "scrubbed_family":
                    "Assigned_Family"
            }
        ),
        how="left",
        on="Assigned_Family",
        validate="one_to_one"
    )
)


family_coverage[
    "External_Family_Status"
] = np.where(
    family_coverage[
        "External_Family_SLA"
    ].notna(),
    "EXTERNAL_BIEN_FAMILY_EVIDENCE_AVAILABLE",
    "NO_EXTERNAL_BIEN_FAMILY_EVIDENCE"
)


# -----------------------------------------------------------------------------
# 15. Validation
# -----------------------------------------------------------------------------

assert len(
    bien_family_raw
) == 4363, (
    f"Expected 4363 raw records, "
    f"found {len(bien_family_raw)}."
)

assert (
    bien_family_raw[
        "scrubbed_family"
    ].nunique()
    == 8
)

assert (
    family_coverage[
        "Assigned_Family"
    ].nunique()
    == 9
)


# -----------------------------------------------------------------------------
# 16. Summaries
# -----------------------------------------------------------------------------

qc_summary = (
    bien_family[
        "QC_Status"
    ]
    .value_counts()
    .rename_axis(
        "QC_Status"
    )
    .reset_index(
        name="Record_Count"
    )
)


raw_family_summary = (
    bien_family
    .groupby(
        "scrubbed_family",
        as_index=False
    )
    .agg(
        Raw_Record_Count=(
            "SLA_Std",
            "size"
        ),

        Raw_Species_Count=(
            "scrubbed_species_binomial",
            "nunique"
        ),

        Raw_Genus_Count=(
            "scrubbed_genus",
            "nunique"
        ),

        Raw_Min_SLA=(
            "SLA_Std",
            "min"
        ),

        Raw_Median_SLA=(
            "SLA_Std",
            "median"
        ),

        Raw_Max_SLA=(
            "SLA_Std",
            "max"
        ),
    )
)


# -----------------------------------------------------------------------------
# 17. Export
# -----------------------------------------------------------------------------

qc_path = (
    output_dir
    / "omo_sla_bien_external_family_record_qc.csv"
)

retained_path = (
    output_dir
    / "omo_sla_bien_external_family_retained_records.csv"
)

species_path = (
    output_dir
    / "omo_sla_bien_external_family_species_summary.csv"
)

genus_path = (
    output_dir
    / "omo_sla_bien_external_family_genus_summary.csv"
)

family_path = (
    output_dir
    / "omo_sla_bien_external_family_hierarchical_estimates.csv"
)

coverage_path = (
    output_dir
    / "omo_sla_bien_external_family_coverage.csv"
)


bien_family.to_csv(
    qc_path,
    index=False
)

bien_family_retained.to_csv(
    retained_path,
    index=False
)

species_level.to_csv(
    species_path,
    index=False
)

genus_level.to_csv(
    genus_path,
    index=False
)

family_level.to_csv(
    family_path,
    index=False
)

family_coverage.to_csv(
    coverage_path,
    index=False
)


# -----------------------------------------------------------------------------
# 18. Diagnostics
# -----------------------------------------------------------------------------

print("\nExternal BIEN family QC summary")
print("-" * 100)

display(
    qc_summary
)


print("\nRaw family summary")
print("-" * 100)

display(
    raw_family_summary
)


print("\nHierarchical external family SLA estimates")
print("-" * 100)

display(
    family_level[
        [
            "scrubbed_family",
            "External_Family_SLA",
            "Standardized_SLA_Unit",
            "External_Genus_Count",
            "External_Species_Count",
            "External_GenusMedian_Min",
            "External_GenusMedian_Max",
            "External_Evidence_Strength",
        ]
    ]
)


print("\nCoverage across all 9 required families")
print("-" * 100)

display(
    family_coverage[
        [
            "Assigned_Family",
            "External_Family_SLA",
            "External_Genus_Count",
            "External_Species_Count",
            "External_Evidence_Strength",
            "External_Family_Status",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Record-level QC      : "
    f"{qc_path}"
)

print(
    f"Retained records     : "
    f"{retained_path}"
)

print(
    f"Species summary      : "
    f"{species_path}"
)

print(
    f"Genus summary        : "
    f"{genus_path}"
)

print(
    f"Family estimates     : "
    f"{family_path}"
)

print(
    f"Family coverage      : "
    f"{coverage_path}"
)


print(
    "\nSection 8.8R2 completed successfully."
)
```

## Section 8.8R3 — Accept External BIEN Family SLA Estimates and Isolate Final Residual Gap

External BIEN family-level SLA evidence was successfully recovered for eight of
the nine residual families.

Following quality control, external family estimates were derived hierarchically:

1. median SLA within species;
2. median of species medians within genus;
3. median of genus medians within family.

This procedure prevents heavily sampled species or genera from dominating the
family-level estimate.

All eight recovered families have strong multi-genus support and are accepted
as external `FAMILY_ESTIMATE` values. The BIEN-derived estimate is used directly
rather than averaged with earlier internal candidates, thereby avoiding
unnecessary blending of evidence sources.

Musaceae was not represented in the BIEN extraction and therefore remains
unresolved for a final controlled fallback stage.

```python
# =============================================================================
# SECTION 8.8R3 — ACCEPT EXTERNAL BIEN FAMILY SLA ESTIMATES
#                  AND ISOLATE FINAL RESIDUAL GAP
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8R3 — ACCEPT EXTERNAL BIEN FAMILY SLA ESTIMATES "
    "AND ISOLATE FINAL RESIDUAL GAP"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

family_coverage_path = (
    output_dir
    / "omo_sla_bien_external_family_coverage.csv"
)

remaining_targets_path = (
    output_dir
    / "omo_sla_remaining_targets_after_internal_family_estimates.csv"
)

family_requirements_path = (
    output_dir
    / "omo_sla_remaining_external_family_requirements.csv"
)


for path in [
    family_coverage_path,
    remaining_targets_path,
    family_requirements_path,
]:
    if not path.exists():
        raise FileNotFoundError(
            f"Required input not found:\n{path}"
        )


family_coverage = pd.read_csv(
    family_coverage_path
)

remaining_targets = pd.read_csv(
    remaining_targets_path
)

family_requirements = pd.read_csv(
    family_requirements_path
)


# -----------------------------------------------------------------------------
# 2. Validate locked state
# -----------------------------------------------------------------------------

assert len(remaining_targets) == 22, (
    f"Expected 22 residual canonical taxa, "
    f"found {len(remaining_targets)}."
)

assert (
    remaining_targets[
        "Assigned_Family"
    ].nunique()
    == 9
)

assert len(family_coverage) == 9


# -----------------------------------------------------------------------------
# 3. Select externally supported BIEN families
# -----------------------------------------------------------------------------

accepted_external_families = (
    family_coverage.loc[
        family_coverage[
            "External_Family_Status"
        ].eq(
            "EXTERNAL_BIEN_FAMILY_EVIDENCE_AVAILABLE"
        )
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(accepted_external_families) == 8, (
    f"Expected 8 BIEN-covered families, "
    f"found {len(accepted_external_families)}."
)


# -----------------------------------------------------------------------------
# 4. Validate external evidence strength
# -----------------------------------------------------------------------------

acceptable_strengths = {
    "STRONG",
    "VERY_STRONG",
}

unexpected_strength = (
    accepted_external_families.loc[
        ~accepted_external_families[
            "External_Evidence_Strength"
        ].isin(
            acceptable_strengths
        )
    ]
)


if not unexpected_strength.empty:

    print("\nExternal estimates failing acceptance threshold")
    print("-" * 100)

    display(
        unexpected_strength[
            [
                "Assigned_Family",
                "External_Family_SLA",
                "External_Genus_Count",
                "External_Species_Count",
                "External_Evidence_Strength",
            ]
        ]
    )

    raise RuntimeError(
        "One or more external family estimates lack adequate "
        "multi-genus support."
    )


# -----------------------------------------------------------------------------
# 5. Build accepted external family table
# -----------------------------------------------------------------------------

accepted_external_family_estimates = (
    accepted_external_families[
        [
            "Assigned_Family",
            "External_Family_SLA",
            "Standardized_SLA_Unit",
            "External_Genus_Count",
            "External_Species_Count",
            "Supporting_Genera",
            "External_Evidence_Strength",
            "Aggregation_Method",
        ]
    ]
    .copy()
    .rename(
        columns={
            "External_Family_SLA":
                "Accepted_Family_SLA",

            "Standardized_SLA_Unit":
                "Accepted_Family_SLA_Unit",
        }
    )
)


accepted_external_family_estimates[
    "Accepted_Provenance_Class"
] = "FAMILY_ESTIMATE"

accepted_external_family_estimates[
    "Accepted_Source_Class"
] = "BIEN_EXTERNAL_FAMILY"

accepted_external_family_estimates[
    "Accepted_Quality_Flag"
] = (
    "EXTERNAL_BIEN_HIERARCHICAL_FAMILY_ESTIMATE"
)

accepted_external_family_estimates[
    "Acceptance_Status"
] = "ACCEPTED_EXTERNAL_FAMILY_ESTIMATE"


# -----------------------------------------------------------------------------
# 6. Internal-vs-external corroboration diagnostic
# -----------------------------------------------------------------------------

internal_candidates = (
    family_requirements[
        [
            "Assigned_Family",
            "Candidate_Family_SLA",
            "Family_Evidence_Strength",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Candidate_Family_SLA":
                "Internal_Candidate_Family_SLA",

            "Family_Evidence_Strength":
                "Internal_Evidence_Strength",
        }
    )
)


accepted_external_family_estimates = (
    accepted_external_family_estimates
    .merge(
        internal_candidates,
        how="left",
        on="Assigned_Family",
        validate="one_to_one"
    )
)


accepted_external_family_estimates[
    "Internal_External_Absolute_Difference"
] = (
    (
        accepted_external_family_estimates[
            "Accepted_Family_SLA"
        ]
        -
        accepted_external_family_estimates[
            "Internal_Candidate_Family_SLA"
        ]
    )
    .abs()
)


accepted_external_family_estimates[
    "Internal_External_Percent_Difference"
] = np.where(
    accepted_external_family_estimates[
        "Internal_Candidate_Family_SLA"
    ].notna()
    &
    accepted_external_family_estimates[
        "Internal_Candidate_Family_SLA"
    ].ne(0),

    100
    * accepted_external_family_estimates[
        "Internal_External_Absolute_Difference"
    ]
    /
    accepted_external_family_estimates[
        "Internal_Candidate_Family_SLA"
    ],

    np.nan
)


# -----------------------------------------------------------------------------
# 7. Construct clean residual-target foundation
#
# IMPORTANT:
# Remove any pre-existing SLA result columns before attaching the newly
# accepted external-family values. This prevents duplicate column names.
# -----------------------------------------------------------------------------

result_columns_to_drop = [
    "Standardized_SLA_Value",
    "Standardized_SLA_Unit",
    "Provenance_Class",
    "Trait_Source",
    "Quality_Flag",

    "Accepted_Family_SLA",
    "Accepted_Family_SLA_Unit",
    "Accepted_Provenance_Class",
    "Accepted_Source_Class",
    "Accepted_Quality_Flag",

    "External_Genus_Count",
    "External_Species_Count",
    "External_Evidence_Strength",
]


remaining_targets_clean = (
    remaining_targets.drop(
        columns=[
            col
            for col in result_columns_to_drop
            if col in remaining_targets.columns
        ],
        errors="ignore"
    )
    .copy()
)


# -----------------------------------------------------------------------------
# 8. Attach accepted external family evidence
# -----------------------------------------------------------------------------

external_lookup = (
    accepted_external_family_estimates[
        [
            "Assigned_Family",
            "Accepted_Family_SLA",
            "Accepted_Family_SLA_Unit",
            "External_Genus_Count",
            "External_Species_Count",
            "External_Evidence_Strength",
            "Accepted_Provenance_Class",
            "Accepted_Source_Class",
            "Accepted_Quality_Flag",
        ]
    ]
    .copy()
)


targets_after_external_merge = (
    remaining_targets_clean
    .merge(
        external_lookup,
        how="left",
        on="Assigned_Family",
        validate="many_to_one"
    )
)


# -----------------------------------------------------------------------------
# 9. Build newly covered Omo values
#
# Create the standardized fields explicitly rather than renaming columns.
# -----------------------------------------------------------------------------

external_omo_values = (
    targets_after_external_merge.loc[
        targets_after_external_merge[
            "Accepted_Family_SLA"
        ].notna()
    ]
    .copy()
    .reset_index(drop=True)
)


external_omo_values[
    "Standardized_SLA_Value"
] = external_omo_values[
    "Accepted_Family_SLA"
]

external_omo_values[
    "Standardized_SLA_Unit"
] = external_omo_values[
    "Accepted_Family_SLA_Unit"
]

external_omo_values[
    "Provenance_Class"
] = external_omo_values[
    "Accepted_Provenance_Class"
]

external_omo_values[
    "Trait_Source"
] = external_omo_values[
    "Accepted_Source_Class"
]

external_omo_values[
    "Quality_Flag"
] = external_omo_values[
    "Accepted_Quality_Flag"
]


# -----------------------------------------------------------------------------
# 10. Defensive duplicate-column validation
# -----------------------------------------------------------------------------

duplicate_columns = (
    external_omo_values.columns[
        external_omo_values.columns.duplicated()
    ]
    .tolist()
)

assert len(duplicate_columns) == 0, (
    f"Duplicate columns detected: {duplicate_columns}"
)


# -----------------------------------------------------------------------------
# 11. Determine final unresolved taxa
# -----------------------------------------------------------------------------

covered_external_taxa = set(
    external_omo_values[
        "Canonical_Species"
    ]
)


final_residual = (
    remaining_targets_clean.loc[
        ~remaining_targets_clean[
            "Canonical_Species"
        ].isin(
            covered_external_taxa
        )
    ]
    .copy()
    .reset_index(drop=True)
)


final_residual[
    "Final_Fallback_Strategy"
] = (
    "TARGETED_EXTERNAL_OR_CONTROLLED_GLOBAL_ESTIMATE"
)

final_residual[
    "Final_Fallback_Status"
] = (
    "READY_FOR_FINAL_FALLBACK"
)


# -----------------------------------------------------------------------------
# 12. Coverage accounting
# -----------------------------------------------------------------------------

coverage_before_external_family = 105

new_external_family_taxa = (
    external_omo_values[
        "Canonical_Species"
    ].nunique()
)

coverage_after_external_family = (
    coverage_before_external_family
    + new_external_family_taxa
)

remaining_after_external_family = (
    127
    - coverage_after_external_family
)

coverage_percent = (
    100
    * coverage_after_external_family
    / 127
)


# -----------------------------------------------------------------------------
# 13. Validation
# -----------------------------------------------------------------------------

assert new_external_family_taxa == 21, (
    f"Expected 21 newly covered taxa, "
    f"found {new_external_family_taxa}."
)

assert coverage_after_external_family == 126

assert remaining_after_external_family == 1

assert len(final_residual) == 1

assert (
    final_residual[
        "Assigned_Family"
    ].iloc[0]
    == "Musaceae"
), (
    "Expected Musaceae to be the sole residual family."
)


# Now this returns ONE Series, so .all() returns one Boolean.
assert external_omo_values[
    "Standardized_SLA_Value"
].notna().all()


assert external_omo_values[
    "Standardized_SLA_Unit"
].eq(
    "mm2 mg-1"
).all()


assert external_omo_values[
    "Provenance_Class"
].eq(
    "FAMILY_ESTIMATE"
).all()


assert external_omo_values[
    "Trait_Source"
].eq(
    "BIEN_EXTERNAL_FAMILY"
).all()


# -----------------------------------------------------------------------------
# 14. Family-level Omo coverage summary
# -----------------------------------------------------------------------------

external_family_taxon_counts = (
    external_omo_values
    .groupby(
        "Assigned_Family",
        as_index=False
    )
    .agg(
        Omo_Taxa_Assigned=(
            "Canonical_Species",
            "nunique"
        ),

        External_Family_SLA=(
            "Standardized_SLA_Value",
            "first"
        ),

        External_Genus_Count=(
            "External_Genus_Count",
            "first"
        ),

        External_Species_Count=(
            "External_Species_Count",
            "first"
        ),

        External_Evidence_Strength=(
            "External_Evidence_Strength",
            "first"
        ),
    )
)


# -----------------------------------------------------------------------------
# 15. Export
# -----------------------------------------------------------------------------

accepted_family_path = (
    output_dir
    / "omo_sla_external_bien_family_estimates_accepted.csv"
)

omo_values_path = (
    output_dir
    / "omo_sla_external_bien_family_omo_values.csv"
)

final_residual_path = (
    output_dir
    / "omo_sla_final_residual_after_external_family_estimates.csv"
)

family_taxon_summary_path = (
    output_dir
    / "omo_sla_external_bien_family_taxon_coverage.csv"
)


accepted_external_family_estimates.to_csv(
    accepted_family_path,
    index=False
)

external_omo_values.to_csv(
    omo_values_path,
    index=False
)

final_residual.to_csv(
    final_residual_path,
    index=False
)

external_family_taxon_counts.to_csv(
    family_taxon_summary_path,
    index=False
)


# -----------------------------------------------------------------------------
# 16. Diagnostics
# -----------------------------------------------------------------------------

print("\nAccepted external BIEN family SLA estimates")
print("-" * 100)

display(
    accepted_external_family_estimates[
        [
            "Assigned_Family",
            "Accepted_Family_SLA",
            "Accepted_Family_SLA_Unit",
            "External_Genus_Count",
            "External_Species_Count",
            "External_Evidence_Strength",
            "Internal_Candidate_Family_SLA",
            "Internal_External_Percent_Difference",
            "Accepted_Quality_Flag",
        ]
    ]
)


print("\nOmo taxon coverage from external BIEN family estimates")
print("-" * 100)

display(
    external_family_taxon_counts
)


print("\nOmo taxa receiving external family estimates")
print("-" * 100)

display(
    external_omo_values[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Assigned_Family",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "External_Evidence_Strength",
            "Quality_Flag",
            "Provenance_Class",
        ]
    ]
)


print("\nCoverage after external BIEN family acquisition")
print("-" * 100)

print(
    f"Coverage before external families : "
    f"{coverage_before_external_family} / 127"
)

print(
    f"New external family coverage      : "
    f"{new_external_family_taxa}"
)

print(
    f"Current SLA coverage              : "
    f"{coverage_after_external_family} / 127"
)

print(
    f"Coverage percent                  : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps                : "
    f"{remaining_after_external_family}"
)


print("\nFinal residual SLA target")
print("-" * 100)

display(
    final_residual[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Assigned_Family",
            "Final_Fallback_Strategy",
            "Final_Fallback_Status",
        ]
    ]
)


print("\nExports")
print("-" * 100)

print(
    f"Accepted family estimates : "
    f"{accepted_family_path}"
)

print(
    f"Omo family values         : "
    f"{omo_values_path}"
)

print(
    f"Final residual target      : "
    f"{final_residual_path}"
)

print(
    f"Family taxon coverage      : "
    f"{family_taxon_summary_path}"
)


print(
    "\nSection 8.8R3 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8S1 — FINAL MUSACEAE SLA RESCUE REGISTER
# =============================================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("SECTION 8.8S1 — FINAL MUSACEAE SLA RESCUE REGISTER")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

residual_path = (
    output_dir
    / "omo_sla_final_residual_after_external_family_estimates.csv"
)

if not residual_path.exists():
    raise FileNotFoundError(
        f"Final residual table not found:\n{residual_path}"
    )


final_residual = pd.read_csv(
    residual_path
)


# -----------------------------------------------------------------------------
# 2. Validate sole residual taxon
# -----------------------------------------------------------------------------

assert len(final_residual) == 1

assert (
    final_residual[
        "Canonical_Species"
    ].iloc[0]
    == "Musa sapiensis"
)

assert (
    final_residual[
        "Assigned_Family"
    ].iloc[0]
    == "Musaceae"
)


# -----------------------------------------------------------------------------
# 3. Define controlled taxonomic search identities
#
# Earlier taxonomic review indicated that the historical Omo name
# Musa sapiensis should be searched using cultivated Musa aliases.
# -----------------------------------------------------------------------------

musa_rescue_register = pd.DataFrame(
    [
        {
            "Canonical_Species":
                "Musa sapiensis",

            "Normalized_Acquisition_Genus":
                "Musa",

            "Assigned_Family":
                "Musaceae",

            "Primary_Search_Name":
                "Musa paradisiaca",

            "Alternative_Search_Name_1":
                "Musa x paradisiaca",

            "Alternative_Search_Name_2":
                "Musa acuminata",

            "Trait":
                "specific leaf area",

            "Target_Unit":
                "mm2 mg-1",

            "Preferred_Evidence_1":
                "DIRECT_MUSA_SPECIES_SLA",

            "Preferred_Evidence_2":
                "MULTI_SPECIES_MUSA_GENUS_SLA",

            "Preferred_Evidence_3":
                "MUSACEAE_FAMILY_TRAIT_DATA",

            "Final_Fallback":
                "CONTROLLED_GLOBAL_SLA_ESTIMATE",

            "Acquisition_Status":
                "READY_FOR_FINAL_TARGETED_SEARCH",
        }
    ]
)


# -----------------------------------------------------------------------------
# 4. Generate search strings
# -----------------------------------------------------------------------------

search_strings = pd.DataFrame(
    {
        "Search_Query": [
            '"Musa paradisiaca" "specific leaf area"',
            '"Musa x paradisiaca" "specific leaf area"',
            '"Musa acuminata" "specific leaf area"',
            '"Musa" "specific leaf area" SLA',
            '"Musaceae" "specific leaf area"',
        ]
    }
)


# -----------------------------------------------------------------------------
# 5. Export
# -----------------------------------------------------------------------------

register_path = (
    output_dir
    / "omo_sla_final_musaceae_rescue_register.csv"
)

search_path = (
    output_dir
    / "omo_sla_final_musaceae_search_strings.txt"
)


musa_rescue_register.to_csv(
    register_path,
    index=False
)

with open(
    search_path,
    "w",
    encoding="utf-8"
) as f:

    for query in search_strings[
        "Search_Query"
    ]:
        f.write(
            f"{query}\n"
        )


# -----------------------------------------------------------------------------
# 6. Diagnostics
# -----------------------------------------------------------------------------

print("\nFinal SLA residual")
print("-" * 100)

display(
    final_residual[
        [
            "Canonical_Species",
            "Normalized_Search_Name",
            "Normalized_Acquisition_Genus",
            "Assigned_Family",
        ]
    ]
)


print("\nMusaceae rescue register")
print("-" * 100)

display(
    musa_rescue_register
)


print("\nTargeted search strings")
print("-" * 100)

display(
    search_strings
)


print("\nExports")
print("-" * 100)

print(
    f"Musaceae rescue register : "
    f"{register_path}"
)

print(
    f"Search strings           : "
    f"{search_path}"
)


print(
    "\nSection 8.8S1 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8S2 — ACCEPT FINAL MUSA SLA ESTIMATE AND COMPLETE SLA COVERAGE
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 8.8S2 — ACCEPT FINAL MUSA SLA ESTIMATE "
    "AND COMPLETE SLA COVERAGE"
)
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

residual_path = (
    output_dir
    / "omo_sla_final_residual_after_external_family_estimates.csv"
)

if not residual_path.exists():
    raise FileNotFoundError(
        f"Final residual table not found:\n{residual_path}"
    )


final_residual = pd.read_csv(
    residual_path
)


# -----------------------------------------------------------------------------
# 2. Validate single residual taxon
# -----------------------------------------------------------------------------

assert len(final_residual) == 1

assert (
    final_residual[
        "Canonical_Species"
    ].iloc[0]
    == "Musa sapiensis"
)

assert (
    final_residual[
        "Normalized_Acquisition_Genus"
    ].iloc[0]
    == "Musa"
)

assert (
    final_residual[
        "Assigned_Family"
    ].iloc[0]
    == "Musaceae"
)


# -----------------------------------------------------------------------------
# 3. Final Musa evidence decision
#
# Literature evidence:
#
# East African highland banana:
#     Average SLA = 12 m2 kg-1
#
# Additional published Musa AAB ranges:
#     8.7–13.6 m2 kg-1
#     10.0–14.9 m2 kg-1
#
# Independent banana crop modelling parameter:
#     SLA = 13 m2 kg-1
#
# Since:
#     1 m2 kg-1 = 1 mm2 mg-1
#
# Accepted representative Musa value:
#     12.0 mm2 mg-1
# -----------------------------------------------------------------------------

accepted_musa_sla = 12.0


musa_evidence = pd.DataFrame(
    [
        {
            "Canonical_Species":
                "Musa sapiensis",

            "Normalized_Search_Name":
                "Musa x paradisiaca",

            "Normalized_Acquisition_Genus":
                "Musa",

            "Assigned_Family":
                "Musaceae",

            "Standardized_SLA_Value":
                accepted_musa_sla,

            "Standardized_SLA_Unit":
                "mm2 mg-1",

            "Provenance_Class":
                "GENUS_LITERATURE_ESTIMATE",

            "Trait_Source":
                "PEER_REVIEWED_AND_TECHNICAL_LITERATURE",

            "Quality_Flag":
                "MULTI_STUDY_MUSA_GENUS_ESTIMATE",

            "Primary_Evidence_Value":
                12.0,

            "Primary_Evidence_Unit":
                "m2 kg-1",

            "Supporting_Evidence_Value":
                13.0,

            "Supporting_Evidence_Unit":
                "m2 kg-1",

            "Published_Range_1":
                "8.7-13.6 m2 kg-1",

            "Published_Range_2":
                "10.0-14.9 m2 kg-1",

            "Acceptance_Decision":
                "ACCEPT_CONTROLLED_MUSA_GENUS_ESTIMATE",
        }
    ]
)


# -----------------------------------------------------------------------------
# 4. Validate value and unit
# -----------------------------------------------------------------------------

assert np.isclose(
    musa_evidence[
        "Standardized_SLA_Value"
    ].iloc[0],
    12.0
)

assert (
    musa_evidence[
        "Standardized_SLA_Unit"
    ].iloc[0]
    == "mm2 mg-1"
)


# -----------------------------------------------------------------------------
# 5. Coverage accounting
#
# Locked state after 8.8R3:
#     126 / 127
# -----------------------------------------------------------------------------

coverage_before_musa = 126
new_musa_coverage = 1

coverage_after_musa = (
    coverage_before_musa
    + new_musa_coverage
)

remaining_gaps = (
    127
    - coverage_after_musa
)

coverage_percent = (
    100
    * coverage_after_musa
    / 127
)


assert coverage_after_musa == 127
assert remaining_gaps == 0


# -----------------------------------------------------------------------------
# 6. Final residual closure record
# -----------------------------------------------------------------------------

musa_closure = (
    final_residual.copy()
)

musa_closure[
    "Standardized_SLA_Value"
] = accepted_musa_sla

musa_closure[
    "Standardized_SLA_Unit"
] = "mm2 mg-1"

musa_closure[
    "Provenance_Class"
] = "GENUS_LITERATURE_ESTIMATE"

musa_closure[
    "Trait_Source"
] = (
    "PEER_REVIEWED_AND_TECHNICAL_LITERATURE"
)

musa_closure[
    "Quality_Flag"
] = (
    "MULTI_STUDY_MUSA_GENUS_ESTIMATE"
)

musa_closure[
    "Final_Fallback_Status"
] = (
    "FINAL_SLA_VALUE_ACQUIRED"
)


# -----------------------------------------------------------------------------
# 7. Export
# -----------------------------------------------------------------------------

evidence_path = (
    output_dir
    / "omo_sla_final_musa_evidence.csv"
)

closure_path = (
    output_dir
    / "omo_sla_final_musa_closure.csv"
)


musa_evidence.to_csv(
    evidence_path,
    index=False
)

musa_closure.to_csv(
    closure_path,
    index=False
)


# -----------------------------------------------------------------------------
# 8. Diagnostics
# -----------------------------------------------------------------------------

print("\nFinal Musa SLA evidence")
print("-" * 100)

display(
    musa_evidence
)


print("\nFinal residual closure")
print("-" * 100)

display(
    musa_closure[
        [
            "Canonical_Species",
            "Normalized_Acquisition_Genus",
            "Assigned_Family",
            "Standardized_SLA_Value",
            "Standardized_SLA_Unit",
            "Provenance_Class",
            "Quality_Flag",
            "Final_Fallback_Status",
        ]
    ]
)


print("\nFinal SLA coverage")
print("-" * 100)

print(
    f"Coverage before Musa rescue : "
    f"{coverage_before_musa} / 127"
)

print(
    f"New Musa coverage           : "
    f"{new_musa_coverage}"
)

print(
    f"Final SLA coverage          : "
    f"{coverage_after_musa} / 127"
)

print(
    f"Coverage percent            : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Remaining SLA gaps          : "
    f"{remaining_gaps}"
)


print("\nExports")
print("-" * 100)

print(
    f"Musa evidence : "
    f"{evidence_path}"
)

print(
    f"Musa closure  : "
    f"{closure_path}"
)


print(
    "\nSection 8.8S2 completed successfully."
)
```

```python
# =============================================================================
# SECTION 8.8T — FINAL SLA CONSOLIDATION, PROVENANCE SUMMARY AND VALIDATION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 8.8T — FINAL SLA CONSOLIDATION, PROVENANCE SUMMARY AND VALIDATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 1. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

output_dir = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

legacy_output_dir = (
    PROJECT_ROOT
    / "notebooks"
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# 2. Locate 127-taxon foundation
# -----------------------------------------------------------------------------

foundation_candidates = [
    output_dir / "omo_sla_acquisition_targets.csv",
    legacy_output_dir / "omo_sla_acquisition_targets.csv",
]

foundation_path = next(
    (p for p in foundation_candidates if p.exists()),
    None
)

if foundation_path is None:
    raise FileNotFoundError(
        "Could not locate the 127-taxon SLA acquisition foundation."
    )

foundation = pd.read_csv(
    foundation_path,
    low_memory=False
)

assert len(foundation) == 127
assert foundation["Canonical_Species"].nunique() == 127


# -----------------------------------------------------------------------------
# 3. Evidence ledger helper
# -----------------------------------------------------------------------------

ledger_parts = []


def add_final_evidence(
    df,
    source_stream,
    value_col,
    unit_col,
    provenance_col=None,
    provenance_value=None,
    quality_col=None,
    quality_value=None,
    trait_source_col=None,
    trait_source_value=None,
):
    if df is None or df.empty:
        return

    required = {
        "Canonical_Species",
        value_col,
    }

    missing = required - set(df.columns)

    if missing:
        raise KeyError(
            f"{source_stream}: missing columns {sorted(missing)}"
        )

    temp = pd.DataFrame()

    temp["Canonical_Species"] = (
        df["Canonical_Species"]
        .astype(str)
        .str.strip()
    )

    temp["Standardized_SLA_Value"] = pd.to_numeric(
        df[value_col],
        errors="coerce"
    )

    if unit_col in df.columns:
        temp["Standardized_SLA_Unit"] = df[unit_col]
    else:
        temp["Standardized_SLA_Unit"] = "mm2 mg-1"

    if provenance_col and provenance_col in df.columns:
        temp["Provenance_Class"] = df[provenance_col]
    else:
        temp["Provenance_Class"] = provenance_value

    if quality_col and quality_col in df.columns:
        temp["Quality_Flag"] = df[quality_col]
    else:
        temp["Quality_Flag"] = quality_value

    if trait_source_col and trait_source_col in df.columns:
        temp["Trait_Source"] = df[trait_source_col]
    else:
        temp["Trait_Source"] = trait_source_value

    temp["Evidence_Stream"] = source_stream

    temp = temp.loc[
        temp["Standardized_SLA_Value"].notna()
    ].copy()

    ledger_parts.append(temp)


# -----------------------------------------------------------------------------
# 4. Species-level TRY
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_final_species_try_values.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="SPECIES_TRY_FINAL",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="TRY_Quality_Flag",
        trait_source_value="TRY",
    )


# -----------------------------------------------------------------------------
# 5. Early species literature
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_residual_species_literature_accepted.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="EARLY_SPECIES_LITERATURE",
        value_col="Accepted_SLA_Value",
        unit_col="Accepted_SLA_Unit",
        provenance_col="Accepted_Provenance_Class",
        quality_col="Accepted_Quality_Flag",
        trait_source_col="Accepted_Source_Class",
    )


# -----------------------------------------------------------------------------
# 6. Targeted literature batches
# -----------------------------------------------------------------------------

literature_files = [
    "omo_sla_literature_batch1_accepted_omo_values.csv",
    "omo_sla_literature_batch2_accepted_omo_values.csv",
    "omo_sla_literature_batch3_accepted_omo_values.csv",
    "omo_sla_literature_batch5_accepted_omo_values.csv",
    "omo_sla_literature_batch6_accepted_omo_values.csv",
]

for filename in literature_files:

    path = output_dir / filename

    if not path.exists():
        continue

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream=f"LITERATURE:{filename}",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
        trait_source_value="PEER_REVIEWED_LITERATURE",
    )


# -----------------------------------------------------------------------------
# 7. Earlier BIEN genus estimates
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_genus_target_acquisition_status.csv"
)

if path.exists():

    df = pd.read_csv(path)

    df = df.loc[
        df["Genus_Acquisition_Status"]
        .eq("GENUS_ESTIMATE_ACQUIRED")
    ].copy()

    add_final_evidence(
        df=df,
        source_stream="EARLY_BIEN_GENUS",
        value_col="Accepted_Genus_SLA",
        unit_col="Accepted_Genus_SLA_Unit",
        provenance_value="GENUS_ESTIMATE",
        quality_col="Evidence_Strength",
        trait_source_value="BIEN",
    )


# -----------------------------------------------------------------------------
# 8. TRY congeneric estimates
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_try_congeneric_genus_estimates_accepted.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="TRY_CONGENERIC_GENUS",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_value="TRY_CONGENERIC",
    )


# -----------------------------------------------------------------------------
# 9. Residual BIEN genus estimates
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_bien_residual_omo_genus_values.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="RESIDUAL_BIEN_GENUS",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
    )


# -----------------------------------------------------------------------------
# 10. Accepted internal family estimates
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_internal_family_omo_values.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="INTERNAL_FAMILY_ESTIMATE",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
    )


# -----------------------------------------------------------------------------
# 11. Accepted external BIEN family estimates
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_external_bien_family_omo_values.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="EXTERNAL_BIEN_FAMILY_ESTIMATE",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
    )


# -----------------------------------------------------------------------------
# 12. Final Musa estimate
# -----------------------------------------------------------------------------

path = (
    output_dir
    / "omo_sla_final_musa_closure.csv"
)

if path.exists():

    df = pd.read_csv(path)

    add_final_evidence(
        df=df,
        source_stream="FINAL_MUSA_GENUS_LITERATURE",
        value_col="Standardized_SLA_Value",
        unit_col="Standardized_SLA_Unit",
        provenance_col="Provenance_Class",
        quality_col="Quality_Flag",
        trait_source_col="Trait_Source",
    )


# -----------------------------------------------------------------------------
# 13. Combine all evidence
# -----------------------------------------------------------------------------

sla_final_ledger = pd.concat(
    ledger_parts,
    ignore_index=True
)


# -----------------------------------------------------------------------------
# 14. Validate standardized units
# -----------------------------------------------------------------------------

units = (
    sla_final_ledger[
        "Standardized_SLA_Unit"
    ]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)

assert set(units) == {"mm2 mg-1"}, (
    f"Unexpected SLA units detected: {units}"
)


# -----------------------------------------------------------------------------
# 15. Final evidence precedence
# -----------------------------------------------------------------------------

priority_map = {
    "SPECIES_TRY": 1,
    "SPECIES_LITERATURE": 2,
    "GENUS_ESTIMATE": 3,
    "FAMILY_ESTIMATE": 4,
    "GENUS_LITERATURE_ESTIMATE": 5,
}


sla_final_ledger[
    "Evidence_Priority"
] = (
    sla_final_ledger[
        "Provenance_Class"
    ].map(priority_map)
)


if sla_final_ledger[
    "Evidence_Priority"
].isna().any():

    unknown = (
        sla_final_ledger.loc[
            sla_final_ledger[
                "Evidence_Priority"
            ].isna(),
            "Provenance_Class"
        ]
        .dropna()
        .unique()
    )

    raise ValueError(
        f"Unknown provenance classes detected: {unknown}"
    )


# -----------------------------------------------------------------------------
# 16. Detect evidence overlaps
# -----------------------------------------------------------------------------

overlap_summary = (
    sla_final_ledger
    .groupby(
        "Canonical_Species"
    )
    .agg(
        Evidence_Record_Count=(
            "Evidence_Stream",
            "size"
        ),

        Evidence_Stream_Count=(
            "Evidence_Stream",
            "nunique"
        ),

        Evidence_Streams=(
            "Evidence_Stream",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),

        Provenance_Classes=(
            "Provenance_Class",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
    .reset_index()
)


overlap_taxa = (
    overlap_summary.loc[
        overlap_summary[
            "Evidence_Record_Count"
        ] > 1
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# 17. Select final evidence
# -----------------------------------------------------------------------------

sla_final_ranked = (
    sla_final_ledger
    .sort_values(
        [
            "Canonical_Species",
            "Evidence_Priority",
            "Evidence_Stream",
        ]
    )
    .reset_index(drop=True)
)


sla_final_selected = (
    sla_final_ranked
    .drop_duplicates(
        subset="Canonical_Species",
        keep="first"
    )
    .copy()
)


# -----------------------------------------------------------------------------
# 18. Merge with complete 127-taxon foundation
# -----------------------------------------------------------------------------

sla_final = (
    foundation
    .merge(
        sla_final_selected[
            [
                "Canonical_Species",
                "Standardized_SLA_Value",
                "Standardized_SLA_Unit",
                "Provenance_Class",
                "Quality_Flag",
                "Trait_Source",
                "Evidence_Stream",
                "Evidence_Priority",
            ]
        ],
        how="left",
        on="Canonical_Species",
        validate="one_to_one"
    )
)


# -----------------------------------------------------------------------------
# 19. Final validation
# -----------------------------------------------------------------------------

assert len(sla_final) == 127
assert sla_final["Canonical_Species"].nunique() == 127

n_missing = (
    sla_final[
        "Standardized_SLA_Value"
    ].isna().sum()
)

assert n_missing == 0, (
    f"Final SLA table still contains {n_missing} missing values."
)

assert (
    sla_final[
        "Standardized_SLA_Unit"
    ]
    .eq("mm2 mg-1")
    .all()
)

assert (
    sla_final[
        "Standardized_SLA_Value"
    ]
    .gt(0)
    .all()
)


# -----------------------------------------------------------------------------
# 20. Final numerical summary
# -----------------------------------------------------------------------------

sla_numeric_summary = (
    sla_final[
        "Standardized_SLA_Value"
    ]
    .describe()
    .to_frame(
        name="SLA_mm2_mg"
    )
)


# -----------------------------------------------------------------------------
# 21. Provenance summary
# -----------------------------------------------------------------------------

sla_provenance_summary = (
    sla_final
    .groupby(
        "Provenance_Class",
        as_index=False
    )
    .agg(
        N_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Mean_SLA=(
            "Standardized_SLA_Value",
            "mean"
        ),

        SD_SLA=(
            "Standardized_SLA_Value",
            "std"
        ),

        Min_SLA=(
            "Standardized_SLA_Value",
            "min"
        ),

        Median_SLA=(
            "Standardized_SLA_Value",
            "median"
        ),

        Max_SLA=(
            "Standardized_SLA_Value",
            "max"
        ),
    )
)


sla_provenance_summary[
    "Percent_of_Taxa"
] = (
    sla_provenance_summary[
        "N_Taxa"
    ]
    / 127
    * 100
)


# -----------------------------------------------------------------------------
# 22. Quality-flag summary
# -----------------------------------------------------------------------------

quality_summary = (
    sla_final[
        "Quality_Flag"
    ]
    .fillna("UNSPECIFIED")
    .value_counts()
    .rename_axis(
        "Quality_Flag"
    )
    .reset_index(
        name="N_Taxa"
    )
)


# -----------------------------------------------------------------------------
# 23. Export definitive SLA products
# -----------------------------------------------------------------------------

final_path = (
    output_dir
    / "omo_sla_final.csv"
)

provenance_path = (
    output_dir
    / "omo_sla_final_provenance_summary.csv"
)

quality_path = (
    output_dir
    / "omo_sla_final_quality_summary.csv"
)

numeric_path = (
    output_dir
    / "omo_sla_final_numeric_summary.csv"
)

ledger_path = (
    output_dir
    / "omo_sla_final_evidence_ledger.csv"
)

selected_path = (
    output_dir
    / "omo_sla_final_selected_evidence.csv"
)

overlap_path = (
    output_dir
    / "omo_sla_final_evidence_overlaps.csv"
)


sla_final.to_csv(
    final_path,
    index=False
)

sla_provenance_summary.to_csv(
    provenance_path,
    index=False
)

quality_summary.to_csv(
    quality_path,
    index=False
)

sla_numeric_summary.to_csv(
    numeric_path
)

sla_final_ledger.to_csv(
    ledger_path,
    index=False
)

sla_final_selected.to_csv(
    selected_path,
    index=False
)

overlap_taxa.to_csv(
    overlap_path,
    index=False
)


# -----------------------------------------------------------------------------
# 24. Diagnostics
# -----------------------------------------------------------------------------

print("\nFinal SLA table validation")
print("-" * 100)

print(
    f"Rows                : "
    f"{len(sla_final)}"
)

print(
    f"Unique taxa         : "
    f"{sla_final['Canonical_Species'].nunique()}"
)

print(
    f"Missing SLA values  : "
    f"{n_missing}"
)

print(
    f"Minimum SLA         : "
    f"{sla_final['Standardized_SLA_Value'].min():.4f}"
)

print(
    f"Median SLA          : "
    f"{sla_final['Standardized_SLA_Value'].median():.4f}"
)

print(
    f"Mean SLA            : "
    f"{sla_final['Standardized_SLA_Value'].mean():.4f}"
)

print(
    f"Maximum SLA         : "
    f"{sla_final['Standardized_SLA_Value'].max():.4f}"
)


print("\nFinal provenance summary")
print("-" * 100)

display(
    sla_provenance_summary
)


print("\nFinal quality-flag summary")
print("-" * 100)

display(
    quality_summary
)


print("\nEvidence overlap diagnostic")
print("-" * 100)

if overlap_taxa.empty:

    print(
        "No taxa contain multiple accepted evidence streams."
    )

else:

    print(
        f"Taxa with multiple evidence records : "
        f"{len(overlap_taxa)}"
    )

    display(
        overlap_taxa
    )


print("\nExports")
print("-" * 100)

print(
    f"Final SLA table      : "
    f"{final_path}"
)

print(
    f"Provenance summary   : "
    f"{provenance_path}"
)

print(
    f"Quality summary      : "
    f"{quality_path}"
)

print(
    f"Numeric summary      : "
    f"{numeric_path}"
)

print(
    f"Evidence ledger      : "
    f"{ledger_path}"
)

print(
    f"Selected evidence    : "
    f"{selected_path}"
)

print(
    f"Evidence overlaps    : "
    f"{overlap_path}"
)


print(
    "\nSection 8.8T completed successfully."
)
```

```python
# =============================================================================
# SECTION 9.0A — SEED MASS EXISTING-EVIDENCE AND PROJECT-FILE AUDIT
# =============================================================================
#
# Purpose
# -------
# Establish the starting evidence base for Seed Mass before any new acquisition.
#
# Standardized trait definition
# -----------------------------
# Trait code   : SEED_MASS
# Standard unit: mg
# Interpretation:
#     Individual seed dry mass where possible.
#
# This cell:
#   1. Defines the Seed Mass trait standard.
#   2. Searches relevant project directories for existing Seed Mass / TRY /
#      functional-trait files.
#   3. Inspects CSV/TSV files for columns that may contain Seed Mass evidence.
#   4. Does NOT modify, estimate, harmonize, or overwrite any trait values.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import re

# -----------------------------------------------------------------------------
# 1. PROJECT PATHS
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

DATA_DIR = PROJECT_ROOT / "data"
OUTPUT_DIR = PROJECT_ROOT / "outputs"
FUNCTIONAL_OUTPUT_DIR = OUTPUT_DIR / "functional_traits"

assert PROJECT_ROOT.exists(), (
    f"Project root not found:\n{PROJECT_ROOT}"
)

assert FUNCTIONAL_OUTPUT_DIR.exists(), (
    f"Functional-trait output directory not found:\n"
    f"{FUNCTIONAL_OUTPUT_DIR}"
)

# -----------------------------------------------------------------------------
# 2. STANDARDIZED SEED MASS DEFINITION
# -----------------------------------------------------------------------------

SEED_MASS_CONFIG = {
    "Trait_Code": "SEED_MASS",
    "Trait_Name": "Seed Mass",
    "Standardized_Unit": "mg",
    "Preferred_Interpretation": "Individual seed dry mass",
    "Preferred_Taxonomic_Level": "SPECIES",
    "Allow_Genus_Fallback": True,
    "Allow_Family_Fallback": True,
    "Silent_Imputation_Allowed": False,
}

print("=" * 100)
print("SECTION 9.0A — SEED MASS EXISTING-EVIDENCE AND PROJECT-FILE AUDIT")
print("=" * 100)

print("\nStandardized Seed Mass definition")
print("-" * 100)
for key, value in SEED_MASS_CONFIG.items():
    print(f"{key:<30}: {value}")

# -----------------------------------------------------------------------------
# 3. SEARCH TERMS
# -----------------------------------------------------------------------------

# Terms used to identify potentially relevant file names.
filename_terms = [
    "seed",
    "seedmass",
    "seed_mass",
    "seed mass",
    "try",
    "trait",
    "functional",
]

# Terms used to identify potentially relevant columns.
column_patterns = [
    r"\bseed\b",
    r"seed.?mass",
    r"seed.?weight",
    r"seed.?dry.?mass",
    r"seed.?dry.?weight",
    r"thousand.?seed",
    r"1000.?seed",
    r"100.?seed",
    r"\btry\b",
    r"trait",
]

column_regex = re.compile(
    "|".join(column_patterns),
    flags=re.IGNORECASE
)

# -----------------------------------------------------------------------------
# 4. FIND POTENTIALLY RELEVANT FILES
# -----------------------------------------------------------------------------

search_roots = [
    DATA_DIR,
    FUNCTIONAL_OUTPUT_DIR,
]

all_files = []

for root in search_roots:
    if not root.exists():
        continue

    for path in root.rglob("*"):
        if path.is_file():
            all_files.append(path)

relevant_name_files = []

for path in all_files:
    name_lower = path.name.lower()

    if any(term in name_lower for term in filename_terms):
        relevant_name_files.append(path)

relevant_name_files = sorted(set(relevant_name_files))

print("\nPotentially relevant files based on filename")
print("-" * 100)

if relevant_name_files:
    for i, path in enumerate(relevant_name_files, start=1):
        try:
            rel = path.relative_to(PROJECT_ROOT)
        except ValueError:
            rel = path

        print(f"{i:>3}. {rel}")
else:
    print("No filename-based Seed Mass / TRY / trait candidates found.")

# -----------------------------------------------------------------------------
# 5. INSPECT TABULAR FILES FOR SEED-MASS-RELATED COLUMNS
# -----------------------------------------------------------------------------

tabular_extensions = {".csv", ".tsv", ".txt"}

column_audit_records = []

for path in all_files:

    if path.suffix.lower() not in tabular_extensions:
        continue

    # Avoid reading extremely large files unnecessarily.
    # Only the header / first few rows are required for this audit.
    try:

        if path.suffix.lower() == ".tsv":
            df_preview = pd.read_csv(
                path,
                sep="\t",
                nrows=5,
                low_memory=False
            )

        else:
            # sep=None allows Python to infer common delimiters for CSV/TXT.
            df_preview = pd.read_csv(
                path,
                sep=None,
                engine="python",
                nrows=5
            )

    except Exception:
        # File may not actually be a delimited text table.
        continue

    matching_columns = [
        col for col in df_preview.columns
        if column_regex.search(str(col))
    ]

    if matching_columns:

        try:
            rel = path.relative_to(PROJECT_ROOT)
        except ValueError:
            rel = path

        column_audit_records.append({
            "File": str(rel),
            "Matching_Columns": " | ".join(map(str, matching_columns)),
            "N_Columns": len(df_preview.columns),
        })

column_audit = pd.DataFrame(column_audit_records)

print("\nTabular files containing Seed Mass / TRY / trait-related columns")
print("-" * 100)

if not column_audit.empty:

    column_audit = (
        column_audit
        .sort_values("File")
        .reset_index(drop=True)
    )

    display(column_audit)

else:
    print("No relevant columns detected in inspected tabular files.")

# -----------------------------------------------------------------------------
# 6. EXPORT THE INITIAL FILE/COLUMN AUDIT
# -----------------------------------------------------------------------------

audit_export = (
    FUNCTIONAL_OUTPUT_DIR
    / "seed_mass_existing_evidence_file_audit.csv"
)

if not column_audit.empty:

    column_audit.to_csv(
        audit_export,
        index=False
    )

    print("\nAudit table exported")
    print("-" * 100)
    print(audit_export)

else:
    print("\nNo column-audit CSV exported because no candidate columns were found.")

# -----------------------------------------------------------------------------
# 7. BASIC VALIDATION
# -----------------------------------------------------------------------------

assert SEED_MASS_CONFIG["Trait_Code"] == "SEED_MASS"
assert SEED_MASS_CONFIG["Standardized_Unit"] == "mg"
assert SEED_MASS_CONFIG["Silent_Imputation_Allowed"] is False

print("\nInitial Seed Mass infrastructure audit completed successfully.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.0B — AUDIT EXISTING TRY RELEASE FOR SEED MASS EVIDENCE
# =============================================================================
#
# Purpose
# -------
# Determine whether the TRY release already present in the project contains
# Seed Mass measurements before requesting or downloading any new data.
#
# No measurements are accepted, converted, harmonized, or estimated here.
# =============================================================================

from pathlib import Path
import pandas as pd
import re

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FUNCTIONAL_OUTPUT_DIR = (
    PROJECT_ROOT / "outputs" / "functional_traits"
)

TRY_RAW_FILE = (
    FUNCTIONAL_OUTPUT_DIR
    / "omo_sla_try_request_51853_measurements_raw.csv"
)

# -----------------------------------------------------------------------------
# 1. VALIDATE EXISTING TRY FILE
# -----------------------------------------------------------------------------

assert TRY_RAW_FILE.exists(), (
    "Expected TRY raw measurement file was not found:\n"
    f"{TRY_RAW_FILE}"
)

try_raw = pd.read_csv(
    TRY_RAW_FILE,
    low_memory=False
)

assert len(try_raw) > 0, "TRY raw measurement table is empty."

print("=" * 100)
print("SECTION 9.0B — AUDIT EXISTING TRY RELEASE FOR SEED MASS EVIDENCE")
print("=" * 100)

print("\nExisting TRY release")
print("-" * 100)
print(f"File    : {TRY_RAW_FILE}")
print(f"Rows    : {len(try_raw):,}")
print(f"Columns : {len(try_raw.columns):,}")

# -----------------------------------------------------------------------------
# 2. DISPLAY RAW SCHEMA
# -----------------------------------------------------------------------------

print("\nTRY raw schema")
print("-" * 100)

for i, col in enumerate(try_raw.columns, start=1):
    print(f"{i:>3}. {col}")

# -----------------------------------------------------------------------------
# 3. IDENTIFY TRAIT-ID / TRAIT-NAME COLUMNS
# -----------------------------------------------------------------------------

trait_id_candidates = [
    c for c in try_raw.columns
    if str(c).strip().lower() in {
        "traitid",
        "trait_id",
        "trait id"
    }
]

trait_name_candidates = [
    c for c in try_raw.columns
    if str(c).strip().lower() in {
        "traitname",
        "trait_name",
        "trait name"
    }
]

assert trait_name_candidates, (
    "No recognizable TraitName column was found in the TRY raw table."
)

TRAIT_NAME_COL = trait_name_candidates[0]
TRAIT_ID_COL = trait_id_candidates[0] if trait_id_candidates else None

print("\nDetected trait-definition columns")
print("-" * 100)
print(f"Trait name column : {TRAIT_NAME_COL}")
print(f"Trait ID column   : {TRAIT_ID_COL}")

# -----------------------------------------------------------------------------
# 4. ENUMERATE UNIQUE TRAIT DEFINITIONS
# -----------------------------------------------------------------------------

trait_cols = [TRAIT_NAME_COL]

if TRAIT_ID_COL is not None:
    trait_cols.insert(0, TRAIT_ID_COL)

trait_inventory = (
    try_raw[trait_cols]
    .drop_duplicates()
    .sort_values(
        by=trait_cols,
        key=lambda x: x.astype(str)
    )
    .reset_index(drop=True)
)

print("\nUnique TRY trait definitions present in existing release")
print("-" * 100)
print(f"Unique definitions : {len(trait_inventory):,}")

display(trait_inventory)

# -----------------------------------------------------------------------------
# 5. SEARCH TRAIT DEFINITIONS FOR SEED-MASS TERMINOLOGY
# -----------------------------------------------------------------------------

seed_pattern = re.compile(
    r"seed|diaspore|propagule|mass|weight",
    flags=re.IGNORECASE
)

seed_candidate_mask = (
    trait_inventory[TRAIT_NAME_COL]
    .astype(str)
    .str.contains(seed_pattern, na=False)
)

seed_trait_candidates = (
    trait_inventory.loc[seed_candidate_mask]
    .copy()
    .reset_index(drop=True)
)

print("\nPotential Seed Mass-related TRY trait definitions")
print("-" * 100)

if seed_trait_candidates.empty:
    print(
        "NONE FOUND — the existing TRY release does not contain an "
        "obvious Seed Mass-related trait definition."
    )
else:
    print(
        f"Candidate trait definitions found : "
        f"{len(seed_trait_candidates):,}"
    )
    display(seed_trait_candidates)

# -----------------------------------------------------------------------------
# 6. COUNT RAW RECORDS ASSOCIATED WITH EACH CANDIDATE
# -----------------------------------------------------------------------------

if not seed_trait_candidates.empty:

    candidate_names = set(
        seed_trait_candidates[TRAIT_NAME_COL]
        .dropna()
        .astype(str)
    )

    candidate_records = try_raw[
        try_raw[TRAIT_NAME_COL]
        .astype(str)
        .isin(candidate_names)
    ].copy()

    candidate_record_counts = (
        candidate_records
        .groupby(trait_cols, dropna=False)
        .size()
        .reset_index(name="N_Raw_Records")
        .sort_values(
            "N_Raw_Records",
            ascending=False
        )
        .reset_index(drop=True)
    )

    print("\nRaw-record counts for candidate definitions")
    print("-" * 100)

    display(candidate_record_counts)

else:

    candidate_record_counts = pd.DataFrame()

# -----------------------------------------------------------------------------
# 7. EXPORT AUDIT TABLES
# -----------------------------------------------------------------------------

inventory_export = (
    FUNCTIONAL_OUTPUT_DIR
    / "omo_seed_mass_try_existing_trait_inventory.csv"
)

candidate_export = (
    FUNCTIONAL_OUTPUT_DIR
    / "omo_seed_mass_try_existing_candidate_traits.csv"
)

trait_inventory.to_csv(
    inventory_export,
    index=False
)

seed_trait_candidates.to_csv(
    candidate_export,
    index=False
)

print("\nAudit exports")
print("-" * 100)
print(f"Trait inventory   : {inventory_export}")
print(f"Seed candidates   : {candidate_export}")

# -----------------------------------------------------------------------------
# 8. VALIDATION
# -----------------------------------------------------------------------------

assert trait_inventory[TRAIT_NAME_COL].notna().any(), (
    "TRY trait inventory contains no usable trait names."
)

print("\nExisting TRY Seed Mass evidence audit completed successfully.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.0C — AUDIT EXISTING STRUCTURED TRAIT TABLES FOR SEED MASS RECORDS
# =============================================================================
#
# Purpose
# -------
# Determine whether any existing Notebook 07B/project tables already contain
# actual SEED_MASS records.
#
# Important
# ---------
# This searches ROW VALUES for explicit Seed Mass trait codes/names.
# It does not infer Seed Mass from generic words such as "mass".
# No acquisition, harmonization, conversion, or fallback estimation occurs.
# =============================================================================

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FUNCTIONAL_OUTPUT_DIR = (
    PROJECT_ROOT / "outputs" / "functional_traits"
)

DATA_DIR = PROJECT_ROOT / "data"

assert PROJECT_ROOT.exists()
assert FUNCTIONAL_OUTPUT_DIR.exists()

print("=" * 100)
print("SECTION 9.0C — EXISTING STRUCTURED SEED MASS RECORD AUDIT")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. SEARCH ALL PROJECT TABULAR FILES LIKELY TO CONTAIN TRAIT RECORDS
# -----------------------------------------------------------------------------

search_roots = [
    DATA_DIR,
    FUNCTIONAL_OUTPUT_DIR,
]

tabular_extensions = {".csv", ".tsv", ".txt"}

candidate_files = []

for root in search_roots:
    if not root.exists():
        continue

    for path in root.rglob("*"):
        if (
            path.is_file()
            and path.suffix.lower() in tabular_extensions
        ):
            candidate_files.append(path)

candidate_files = sorted(set(candidate_files))

print("\nTabular files considered")
print("-" * 100)
print(f"N files : {len(candidate_files):,}")

# -----------------------------------------------------------------------------
# 2. TRAIT-FIELD NAMES THAT MAY HOLD AN EXPLICIT SEED MASS IDENTIFIER
# -----------------------------------------------------------------------------

trait_field_names = {
    "trait_code",
    "traitcode",
    "trait",
    "trait_name",
    "traitname",
    "target_trait",
    "requested_trait",
}

# Strict accepted Seed Mass identifiers.
seed_mass_identifiers = {
    "SEED_MASS",
    "SEED MASS",
    "SEED MASS (MG)",
    "SEED DRY MASS",
    "SEED DRY WEIGHT",
}

# -----------------------------------------------------------------------------
# 3. INSPECT TABLE VALUES
# -----------------------------------------------------------------------------

hits = []
files_with_trait_fields = []

for path in candidate_files:

    try:
        if path.suffix.lower() == ".tsv":
            df = pd.read_csv(
                path,
                sep="\t",
                low_memory=False
            )
        else:
            df = pd.read_csv(
                path,
                sep=None,
                engine="python"
            )

    except Exception:
        continue

    if df.empty:
        continue

    trait_cols = [
        col for col in df.columns
        if str(col).strip().lower() in trait_field_names
    ]

    if not trait_cols:
        continue

    try:
        rel_path = str(path.relative_to(PROJECT_ROOT))
    except ValueError:
        rel_path = str(path)

    files_with_trait_fields.append({
        "File": rel_path,
        "N_Rows": len(df),
        "Trait_Fields": " | ".join(map(str, trait_cols)),
    })

    for trait_col in trait_cols:

        normalized = (
            df[trait_col]
            .astype("string")
            .str.strip()
            .str.upper()
        )

        mask = normalized.isin(seed_mass_identifiers)

        if not mask.any():
            continue

        matched = df.loc[mask].copy()

        # Retain scientifically useful fields where available.
        preferred_columns = [
            trait_col,
            "Canonical_Species",
            "Trait_Source_Name",
            "SpeciesName",
            "AccSpeciesName",
            "Standardized_Trait_Value",
            "Standardized_Trait_Unit",
            "Raw_Trait_Value",
            "Raw_Trait_Unit",
            "Trait_Source",
            "Source",
            "Match_Status",
            "Trait_Quality_Flag",
            "Quality_Flag",
        ]

        retained_columns = []

        for col in preferred_columns:
            if col in matched.columns and col not in retained_columns:
                retained_columns.append(col)

        for row_index, row in matched.iterrows():

            record = {
                "File": rel_path,
                "Source_Row_Index": row_index,
                "Trait_Field": trait_col,
                "Trait_Field_Value": row[trait_col],
            }

            for col in retained_columns:
                record[col] = row[col]

            hits.append(record)

# -----------------------------------------------------------------------------
# 4. BUILD AUDIT TABLES
# -----------------------------------------------------------------------------

trait_field_audit = pd.DataFrame(files_with_trait_fields)
seed_mass_existing = pd.DataFrame(hits)

print("\nFiles containing explicit trait-identifier fields")
print("-" * 100)

if trait_field_audit.empty:
    print("No structured trait-identifier fields were found.")
else:
    print(f"Files with trait fields : {len(trait_field_audit):,}")
    display(
        trait_field_audit
        .sort_values("File")
        .reset_index(drop=True)
    )

print("\nExisting explicit SEED_MASS records")
print("-" * 100)

if seed_mass_existing.empty:

    print("NONE FOUND")
    print(
        "No existing structured project row is explicitly identified "
        "as SEED_MASS."
    )

else:

    print(f"Seed Mass records found : {len(seed_mass_existing):,}")

    display(
        seed_mass_existing
        .sort_values(
            ["File", "Source_Row_Index"]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 5. EXPORT AUDIT RESULTS
# -----------------------------------------------------------------------------

trait_field_export = (
    FUNCTIONAL_OUTPUT_DIR
    / "omo_seed_mass_existing_trait_field_audit.csv"
)

seed_mass_export = (
    FUNCTIONAL_OUTPUT_DIR
    / "omo_seed_mass_existing_records.csv"
)

trait_field_audit.to_csv(
    trait_field_export,
    index=False
)

seed_mass_existing.to_csv(
    seed_mass_export,
    index=False
)

print("\nAudit exports")
print("-" * 100)
print(f"Trait-field audit : {trait_field_export}")
print(f"Seed Mass records : {seed_mass_export}")

# -----------------------------------------------------------------------------
# 6. VALIDATION / INTERPRETATION
# -----------------------------------------------------------------------------

print("\nExisting-evidence interpretation")
print("-" * 100)

if seed_mass_existing.empty:
    print("Existing structured Seed Mass evidence : 0 records")
    print("Status                                : NEW ACQUISITION REQUIRED")
else:
    print(
        f"Existing structured Seed Mass evidence : "
        f"{len(seed_mass_existing):,} records"
    )
    print("Status                                : EXISTING EVIDENCE REQUIRES REVIEW")

print("\nStructured Seed Mass record audit completed successfully.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.0D — BUILD CANONICAL SEED MASS ACQUISITION MANIFEST
# =============================================================================
#
# Purpose
# -------
# Create the authoritative 127-taxon target manifest for Seed Mass acquisition
# using the taxonomic foundation already established in Notebook 07B.
#
# This cell:
#   1. Searches existing project outputs/data for the canonical taxon table.
#   2. Identifies the table containing the 127 canonical taxa.
#   3. Validates species-level vs genus/unresolved status.
#   4. Creates a Seed Mass acquisition manifest.
#
# No external trait acquisition occurs here.
# =============================================================================

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

DATA_DIR = PROJECT_ROOT / "data"
OUTPUT_DIR = PROJECT_ROOT / "outputs"
FUNCTIONAL_OUTPUT_DIR = OUTPUT_DIR / "functional_traits"

assert PROJECT_ROOT.exists()
assert FUNCTIONAL_OUTPUT_DIR.exists()

print("=" * 100)
print("SECTION 9.0D — CANONICAL SEED MASS ACQUISITION MANIFEST")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. SEARCH FOR EXISTING TABLES CONTAINING THE CANONICAL TAXON FIELD
# -----------------------------------------------------------------------------

search_roots = [
    DATA_DIR,
    OUTPUT_DIR,
]

tabular_extensions = {".csv", ".tsv"}

canonical_field_candidates = {
    "canonical_species",
    "canonical_taxon",
    "canonical_name",
    "accepted_species",
}

candidate_tables = []

for root in search_roots:

    if not root.exists():
        continue

    for path in root.rglob("*"):

        if (
            not path.is_file()
            or path.suffix.lower() not in tabular_extensions
        ):
            continue

        try:

            if path.suffix.lower() == ".tsv":
                preview = pd.read_csv(
                    path,
                    sep="\t",
                    nrows=5,
                    low_memory=False
                )
            else:
                preview = pd.read_csv(
                    path,
                    nrows=5,
                    low_memory=False
                )

        except Exception:
            continue

        normalized_columns = {
            str(col).strip().lower(): col
            for col in preview.columns
        }

        matched_fields = [
            original
            for normalized, original in normalized_columns.items()
            if normalized in canonical_field_candidates
        ]

        if not matched_fields:
            continue

        try:
            rel = str(path.relative_to(PROJECT_ROOT))
        except ValueError:
            rel = str(path)

        candidate_tables.append({
            "Path": path,
            "File": rel,
            "Canonical_Fields": matched_fields
        })

print("\nCandidate canonical-taxon tables")
print("-" * 100)

if not candidate_tables:
    raise RuntimeError(
        "No table containing a recognizable canonical taxon field was found."
    )

candidate_summary = pd.DataFrame([
    {
        "File": x["File"],
        "Canonical_Fields": " | ".join(x["Canonical_Fields"])
    }
    for x in candidate_tables
])

display(candidate_summary)

# -----------------------------------------------------------------------------
# 2. FIND TABLES HAVING EXACTLY 127 UNIQUE CANONICAL TAXA
# -----------------------------------------------------------------------------

valid_taxon_tables = []

for item in candidate_tables:

    path = item["Path"]

    try:

        if path.suffix.lower() == ".tsv":
            df = pd.read_csv(
                path,
                sep="\t",
                low_memory=False
            )
        else:
            df = pd.read_csv(
                path,
                low_memory=False
            )

    except Exception:
        continue

    for canonical_col in item["Canonical_Fields"]:

        taxa = (
            df[canonical_col]
            .dropna()
            .astype(str)
            .str.strip()
        )

        unique_taxa = taxa[taxa.ne("")].drop_duplicates()

        if len(unique_taxa) == 127:

            valid_taxon_tables.append({
                "Path": path,
                "File": item["File"],
                "Canonical_Column": canonical_col,
                "DataFrame": df,
                "Taxa": unique_taxa
            })

print("\nTables containing exactly 127 unique canonical taxa")
print("-" * 100)

if not valid_taxon_tables:
    raise RuntimeError(
        "No project table was found containing exactly "
        "127 unique canonical taxa."
    )

for i, item in enumerate(valid_taxon_tables, start=1):
    print(
        f"{i:>2}. {item['File']} "
        f"[column: {item['Canonical_Column']}]"
    )

# -----------------------------------------------------------------------------
# 3. SELECT PREFERRED TAXON SOURCE
# -----------------------------------------------------------------------------
#
# Preference:
#   1. Files explicitly associated with species metadata
#   2. Functional-trait files
#   3. Other validated project outputs
#
# This avoids depending unnecessarily on the completed SLA values themselves.
# -----------------------------------------------------------------------------

def source_priority(item):

    name = item["File"].lower()

    if "species_metadata" in name:
        return 0
    if "functional_traits" in name:
        return 1
    return 2


valid_taxon_tables = sorted(
    valid_taxon_tables,
    key=source_priority
)

selected = valid_taxon_tables[0]

SOURCE_FILE = selected["Path"]
CANONICAL_COL = selected["Canonical_Column"]

taxon_df = selected["DataFrame"].copy()

print("\nSelected authoritative taxon source")
print("-" * 100)
print(f"File             : {selected['File']}")
print(f"Canonical column : {CANONICAL_COL}")

# -----------------------------------------------------------------------------
# 4. EXTRACT AND VALIDATE THE 127 CANONICAL TAXA
# -----------------------------------------------------------------------------

canonical_taxa = (
    taxon_df[[CANONICAL_COL]]
    .dropna()
    .copy()
)

canonical_taxa[CANONICAL_COL] = (
    canonical_taxa[CANONICAL_COL]
    .astype(str)
    .str.strip()
)

canonical_taxa = (
    canonical_taxa[
        canonical_taxa[CANONICAL_COL].ne("")
    ]
    .drop_duplicates()
    .sort_values(CANONICAL_COL)
    .reset_index(drop=True)
)

assert len(canonical_taxa) == 127, (
    f"Expected 127 canonical taxa, found {len(canonical_taxa)}."
)

assert canonical_taxa[CANONICAL_COL].is_unique

# -----------------------------------------------------------------------------
# 5. ASSIGN BASIC TAXONOMIC ACQUISITION STATUS
# -----------------------------------------------------------------------------

known_genus_or_unresolved = {
    "Drypetes species",
    "Cola species",
}

canonical_taxa["Taxonomic_Resolution"] = canonical_taxa[
    CANONICAL_COL
].apply(
    lambda x:
        "GENUS_LEVEL_OR_UNRESOLVED"
        if x in known_genus_or_unresolved
        else "SPECIES_LEVEL"
)

canonical_taxa["Trait_Code"] = "SEED_MASS"
canonical_taxa["Standardized_Unit"] = "mg"

canonical_taxa["Existing_Seed_Mass_Evidence"] = False

canonical_taxa["Acquisition_Status"] = "NOT_ACQUIRED"

canonical_taxa["Preferred_Evidence_Level"] = (
    canonical_taxa["Taxonomic_Resolution"]
    .map({
        "SPECIES_LEVEL": "SPECIES",
        "GENUS_LEVEL_OR_UNRESOLVED": "GENUS"
    })
)

# -----------------------------------------------------------------------------
# 6. VALIDATE EXPECTED TAXONOMIC FOUNDATION
# -----------------------------------------------------------------------------

resolution_summary = (
    canonical_taxa["Taxonomic_Resolution"]
    .value_counts()
)

print("\nTaxonomic resolution summary")
print("-" * 100)
print(resolution_summary.to_string())

assert resolution_summary.get("SPECIES_LEVEL", 0) == 125, (
    "Expected 125 species-level taxa."
)

assert resolution_summary.get(
    "GENUS_LEVEL_OR_UNRESOLVED", 0
) == 2, (
    "Expected 2 genus-level/unresolved taxa."
)

actual_unresolved = set(
    canonical_taxa.loc[
        canonical_taxa["Taxonomic_Resolution"]
        == "GENUS_LEVEL_OR_UNRESOLVED",
        CANONICAL_COL
    ]
)

assert actual_unresolved == known_genus_or_unresolved, (
    "Genus/unresolved taxa differ from the validated Notebook 07B foundation."
)

print("\nValidated genus/unresolved taxa")
print("-" * 100)

for taxon in sorted(actual_unresolved):
    print(f" - {taxon}")

# -----------------------------------------------------------------------------
# 7. CREATE SEED MASS ACQUISITION MANIFEST
# -----------------------------------------------------------------------------

seed_mass_manifest = canonical_taxa.rename(
    columns={
        CANONICAL_COL: "Canonical_Species"
    }
)

manifest_columns = [
    "Canonical_Species",
    "Taxonomic_Resolution",
    "Trait_Code",
    "Standardized_Unit",
    "Preferred_Evidence_Level",
    "Existing_Seed_Mass_Evidence",
    "Acquisition_Status",
]

seed_mass_manifest = seed_mass_manifest[
    manifest_columns
]

print("\nInitial Seed Mass acquisition manifest")
print("-" * 100)

display(seed_mass_manifest)

# -----------------------------------------------------------------------------
# 8. EXPORT
# -----------------------------------------------------------------------------

manifest_export = (
    FUNCTIONAL_OUTPUT_DIR
    / "omo_seed_mass_acquisition_manifest_initial.csv"
)

seed_mass_manifest.to_csv(
    manifest_export,
    index=False
)

print("\nManifest export")
print("-" * 100)
print(manifest_export)

# -----------------------------------------------------------------------------
# 9. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(seed_mass_manifest) == 127
assert seed_mass_manifest["Canonical_Species"].is_unique
assert (
    seed_mass_manifest["Trait_Code"] == "SEED_MASS"
).all()
assert (
    seed_mass_manifest["Standardized_Unit"] == "mg"
).all()
assert (
    seed_mass_manifest["Existing_Seed_Mass_Evidence"] == False
).all()
assert (
    seed_mass_manifest["Acquisition_Status"] == "NOT_ACQUIRED"
).all()

print("\nSeed Mass canonical acquisition manifest created successfully.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.0E — CROSS-TRAIT CANONICAL TAXON BACKBONE VALIDATION
# =============================================================================
#
# Purpose
# -------
# Confirm that the 127 taxon labels placed in the Seed Mass acquisition
# manifest are exactly the same taxonomic backbone used by the completed
# Wood Density and SLA final tables.
#
# This is a validation-only step.
# No taxonomy is changed and no Seed Mass evidence is acquired.
# =============================================================================

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

SEED_MANIFEST_FILE = (
    FT_DIR / "omo_seed_mass_acquisition_manifest_initial.csv"
)

WD_FINAL_FILE = (
    FT_DIR / "omo_wood_density_final.csv"
)

SLA_FINAL_FILE = (
    FT_DIR / "omo_sla_final.csv"
)

# -----------------------------------------------------------------------------
# 1. VALIDATE FILES
# -----------------------------------------------------------------------------

for path in [
    SEED_MANIFEST_FILE,
    WD_FINAL_FILE,
    SLA_FINAL_FILE,
]:
    assert path.exists(), f"Required file not found:\n{path}"

seed = pd.read_csv(SEED_MANIFEST_FILE, low_memory=False)
wd = pd.read_csv(WD_FINAL_FILE, low_memory=False)
sla = pd.read_csv(SLA_FINAL_FILE, low_memory=False)

for name, df in [
    ("Seed Mass manifest", seed),
    ("WD final", wd),
    ("SLA final", sla),
]:
    assert "Canonical_Species" in df.columns, (
        f"{name} has no Canonical_Species column."
    )

print("=" * 100)
print("SECTION 9.0E — CROSS-TRAIT CANONICAL TAXON BACKBONE VALIDATION")
print("=" * 100)

# -----------------------------------------------------------------------------
# 2. NORMALIZE ONLY WHITESPACE
# -----------------------------------------------------------------------------
#
# IMPORTANT:
# We deliberately do NOT correct spelling or taxonomy here.
# This test asks whether the established project identifiers themselves
# are consistent across trait products.
# -----------------------------------------------------------------------------

def get_taxa(df):
    return (
        df["Canonical_Species"]
        .dropna()
        .astype(str)
        .str.strip()
    )

seed_taxa = get_taxa(seed)
wd_taxa = get_taxa(wd)
sla_taxa = get_taxa(sla)

print("\nBasic backbone counts")
print("-" * 100)

summary = pd.DataFrame({
    "Dataset": [
        "Seed Mass initial manifest",
        "Wood Density final",
        "SLA final",
    ],
    "Rows": [
        len(seed),
        len(wd),
        len(sla),
    ],
    "Unique_Canonical_Taxa": [
        seed_taxa.nunique(),
        wd_taxa.nunique(),
        sla_taxa.nunique(),
    ],
    "Missing_Canonical_Name": [
        seed["Canonical_Species"].isna().sum(),
        wd["Canonical_Species"].isna().sum(),
        sla["Canonical_Species"].isna().sum(),
    ],
})

display(summary)

# -----------------------------------------------------------------------------
# 3. CONVERT TO EXACT TAXON SETS
# -----------------------------------------------------------------------------

seed_set = set(seed_taxa)
wd_set = set(wd_taxa)
sla_set = set(sla_taxa)

# -----------------------------------------------------------------------------
# 4. PAIRWISE DIFFERENCES
# -----------------------------------------------------------------------------

comparison_records = []

comparisons = [
    ("SEED_MANIFEST", seed_set, "WD_FINAL", wd_set),
    ("SEED_MANIFEST", seed_set, "SLA_FINAL", sla_set),
    ("WD_FINAL", wd_set, "SLA_FINAL", sla_set),
]

for left_name, left_set, right_name, right_set in comparisons:

    left_only = sorted(left_set - right_set)
    right_only = sorted(right_set - left_set)

    comparison_records.append({
        "Comparison": f"{left_name} vs {right_name}",
        "Left_N": len(left_set),
        "Right_N": len(right_set),
        "Shared_N": len(left_set & right_set),
        "Left_Only_N": len(left_only),
        "Right_Only_N": len(right_only),
        "Exact_Set_Match": left_set == right_set,
    })

comparison_summary = pd.DataFrame(comparison_records)

print("\nPairwise canonical-taxon comparisons")
print("-" * 100)
display(comparison_summary)

# -----------------------------------------------------------------------------
# 5. SHOW ANY DISCREPANT TAXA
# -----------------------------------------------------------------------------

discrepancy_records = []

for left_name, left_set, right_name, right_set in comparisons:

    for taxon in sorted(left_set - right_set):
        discrepancy_records.append({
            "Comparison": f"{left_name} vs {right_name}",
            "Taxon": taxon,
            "Status": f"ONLY_IN_{left_name}",
        })

    for taxon in sorted(right_set - left_set):
        discrepancy_records.append({
            "Comparison": f"{left_name} vs {right_name}",
            "Taxon": taxon,
            "Status": f"ONLY_IN_{right_name}",
        })

discrepancy_table = pd.DataFrame(discrepancy_records)

print("\nCanonical-taxon discrepancies")
print("-" * 100)

if discrepancy_table.empty:
    print("NONE — all three datasets contain exactly the same canonical taxon set.")
else:
    display(discrepancy_table)

# -----------------------------------------------------------------------------
# 6. CHECK ROW-ORDER CONSISTENCY
# -----------------------------------------------------------------------------

seed_order = seed_taxa.tolist()
wd_order = wd_taxa.tolist()
sla_order = sla_taxa.tolist()

order_summary = pd.DataFrame({
    "Comparison": [
        "Seed manifest vs WD final",
        "Seed manifest vs SLA final",
        "WD final vs SLA final",
    ],
    "Same_Row_Order": [
        seed_order == wd_order,
        seed_order == sla_order,
        wd_order == sla_order,
    ],
})

print("\nRow-order comparison")
print("-" * 100)
display(order_summary)

# -----------------------------------------------------------------------------
# 7. EXPORT VALIDATION RESULTS
# -----------------------------------------------------------------------------

summary_file = (
    FT_DIR / "omo_seed_mass_taxon_backbone_validation_summary.csv"
)

discrepancy_file = (
    FT_DIR / "omo_seed_mass_taxon_backbone_discrepancies.csv"
)

comparison_summary.to_csv(
    summary_file,
    index=False
)

discrepancy_table.to_csv(
    discrepancy_file,
    index=False
)

print("\nValidation exports")
print("-" * 100)
print(f"Summary       : {summary_file}")
print(f"Discrepancies : {discrepancy_file}")

# -----------------------------------------------------------------------------
# 8. STRONG VALIDATION ASSERTIONS
# -----------------------------------------------------------------------------

assert len(seed_set) == 127, (
    f"Seed Mass manifest has {len(seed_set)} unique taxa instead of 127."
)

assert len(wd_set) == 127, (
    f"WD final has {len(wd_set)} unique taxa instead of 127."
)

assert len(sla_set) == 127, (
    f"SLA final has {len(sla_set)} unique taxa instead of 127."
)

assert seed_set == wd_set, (
    "Seed Mass manifest and WD final do not use the same canonical taxon set."
)

assert seed_set == sla_set, (
    "Seed Mass manifest and SLA final do not use the same canonical taxon set."
)

assert wd_set == sla_set, (
    "WD and SLA final tables do not use the same canonical taxon set."
)

print("\nCanonical taxon backbone validation PASSED.")
print("Seed Mass, WD and SLA all use the same 127 project taxon identifiers.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1A — PREPARE BIEN SEED MASS ACQUISITION TARGET MANIFEST
# =============================================================================
#
# Purpose
# -------
# Prepare the validated Omo taxon list for first-pass Seed Mass acquisition
# from BIEN.
#
# This cell does NOT query BIEN yet.
# It creates a clean acquisition manifest that:
#   1. preserves the 127 canonical project identifiers;
#   2. separates species-level and genus/unresolved targets;
#   3. derives query names without changing the project backbone;
#   4. exports the manifest for the next acquisition step.
# =============================================================================

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

SEED_MANIFEST_FILE = (
    FT_DIR / "omo_seed_mass_acquisition_manifest_initial.csv"
)

assert SEED_MANIFEST_FILE.exists(), (
    f"Seed Mass manifest not found:\n{SEED_MANIFEST_FILE}"
)

seed_manifest = pd.read_csv(
    SEED_MANIFEST_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1A — PREPARE BIEN SEED MASS ACQUISITION TARGET MANIFEST")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT MANIFEST
# -----------------------------------------------------------------------------

required_columns = {
    "Canonical_Species",
    "Taxonomic_Resolution",
    "Trait_Code",
    "Standardized_Unit",
    "Preferred_Evidence_Level",
    "Existing_Seed_Mass_Evidence",
    "Acquisition_Status",
}

missing_columns = required_columns - set(seed_manifest.columns)

assert not missing_columns, (
    f"Missing required columns: {sorted(missing_columns)}"
)

assert len(seed_manifest) == 127
assert seed_manifest["Canonical_Species"].is_unique
assert (seed_manifest["Trait_Code"] == "SEED_MASS").all()
assert (seed_manifest["Standardized_Unit"] == "mg").all()

# -----------------------------------------------------------------------------
# 2. DERIVE QUERY FIELDS
# -----------------------------------------------------------------------------
#
# IMPORTANT:
# Canonical_Species remains unchanged.
# Query_Name is only an acquisition field.
# -----------------------------------------------------------------------------

bien_manifest = seed_manifest.copy()

bien_manifest["Query_Name"] = (
    bien_manifest["Canonical_Species"]
    .astype(str)
    .str.strip()
)

# Extract genus from the project identifier.
bien_manifest["Query_Genus"] = (
    bien_manifest["Canonical_Species"]
    .astype(str)
    .str.strip()
    .str.split()
    .str[0]
)

# Species-level records are queried first as full binomials/project names.
# Genus/unresolved taxa are explicitly marked for genus-level handling.
bien_manifest["BIEN_Query_Level"] = (
    bien_manifest["Taxonomic_Resolution"]
    .map({
        "SPECIES_LEVEL": "SPECIES",
        "GENUS_LEVEL_OR_UNRESOLVED": "GENUS",
    })
)

bien_manifest["Requested_Trait"] = "seed mass"
bien_manifest["Requested_Standard_Unit"] = "mg"

bien_manifest["BIEN_Query_Status"] = "PENDING"
bien_manifest["BIEN_Raw_Record_Count"] = 0

# -----------------------------------------------------------------------------
# 3. VALIDATE TARGET COUNTS
# -----------------------------------------------------------------------------

query_level_summary = (
    bien_manifest["BIEN_Query_Level"]
    .value_counts()
    .rename_axis("BIEN_Query_Level")
    .reset_index(name="N_Taxa")
)

print("\nBIEN acquisition target summary")
print("-" * 100)
display(query_level_summary)

assert (
    query_level_summary
    .set_index("BIEN_Query_Level")
    .loc["SPECIES", "N_Taxa"]
    == 125
)

assert (
    query_level_summary
    .set_index("BIEN_Query_Level")
    .loc["GENUS", "N_Taxa"]
    == 2
)

# -----------------------------------------------------------------------------
# 4. DISPLAY GENUS/UNRESOLVED TARGETS
# -----------------------------------------------------------------------------

genus_targets = bien_manifest.loc[
    bien_manifest["BIEN_Query_Level"] == "GENUS",
    [
        "Canonical_Species",
        "Query_Genus",
        "BIEN_Query_Level",
        "Requested_Trait",
    ]
].copy()

print("\nGenus/unresolved BIEN targets")
print("-" * 100)
display(genus_targets)

# -----------------------------------------------------------------------------
# 5. DISPLAY SAMPLE OF SPECIES-LEVEL TARGETS
# -----------------------------------------------------------------------------

species_targets = bien_manifest.loc[
    bien_manifest["BIEN_Query_Level"] == "SPECIES"
].copy()

print("\nSpecies-level BIEN targets")
print("-" * 100)
print(f"N species-level targets : {len(species_targets)}")

display(
    species_targets[
        [
            "Canonical_Species",
            "Query_Name",
            "Query_Genus",
            "Requested_Trait",
            "BIEN_Query_Status",
        ]
    ].head(20)
)

# -----------------------------------------------------------------------------
# 6. EXPORT BIEN ACQUISITION MANIFEST
# -----------------------------------------------------------------------------

export_file = (
    FT_DIR / "omo_seed_mass_bien_acquisition_manifest.csv"
)

bien_manifest.to_csv(
    export_file,
    index=False
)

print("\nBIEN acquisition manifest export")
print("-" * 100)
print(export_file)

# -----------------------------------------------------------------------------
# 7. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(bien_manifest) == 127
assert bien_manifest["Canonical_Species"].is_unique
assert bien_manifest["Query_Genus"].notna().all()
assert bien_manifest["BIEN_Query_Level"].notna().all()
assert (bien_manifest["Requested_Trait"] == "seed mass").all()
assert (bien_manifest["BIEN_Query_Status"] == "PENDING").all()

print("\nBIEN Seed Mass acquisition target manifest prepared successfully.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1E — BIEN RAW-DATA INTEGRITY AND TAXONOMIC CROSSWALK AUDIT
# =============================================================================
#
# Purpose
# -------
# Bring the full BIEN Seed Mass acquisition back into Python and determine
# how the returned BIEN taxonomic names relate to the validated Omo
# 127-taxon backbone.
#
# This cell:
#   1. validates the raw BIEN file;
#   2. validates trait labels, units and numeric values;
#   3. audits BIEN taxonomic fields;
#   4. tests exact matching against Omo canonical names using several
#      returned BIEN name fields;
#   5. identifies returned records requiring taxonomic reconciliation.
#
# IMPORTANT:
#   - No records are removed.
#   - No duplicate filtering occurs.
#   - No aggregation occurs.
#   - No genus fallback occurs.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

BIEN_DIR = (
    PROJECT_ROOT / "data" / "external" / "BIEN"
)

MANIFEST_FILE = (
    FT_DIR / "omo_seed_mass_bien_acquisition_manifest.csv"
)

RAW_FILE = (
    BIEN_DIR / "omo_seed_mass_bien_species_raw.csv"
)

# -----------------------------------------------------------------------------
# 1. VALIDATE FILES
# -----------------------------------------------------------------------------

assert MANIFEST_FILE.exists(), (
    f"Manifest not found:\n{MANIFEST_FILE}"
)

assert RAW_FILE.exists(), (
    f"Raw BIEN Seed Mass file not found:\n{RAW_FILE}"
)

manifest = pd.read_csv(
    MANIFEST_FILE,
    low_memory=False
)

raw = pd.read_csv(
    RAW_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1E — BIEN RAW-DATA INTEGRITY AND TAXONOMIC CROSSWALK AUDIT")
print("=" * 100)

print("\nInput tables")
print("-" * 100)

print(
    f"Manifest : {len(manifest):,} rows | "
    f"{manifest['Canonical_Species'].nunique():,} canonical taxa"
)

print(
    f"BIEN raw : {len(raw):,} rows | "
    f"{len(raw.columns):,} columns"
)

# -----------------------------------------------------------------------------
# 2. BASIC RAW-DATA VALIDATION
# -----------------------------------------------------------------------------

required_raw_cols = {
    "scrubbed_species_binomial",
    "trait_name",
    "trait_value",
    "unit",
    "method",
    "url_source",
    "source_citation",
    "verbatim_scientific_name",
    "name_submitted",
    "name_matched",
    "scrubbed_taxon_canonical",
    "scrubbed_taxonomic_status",
    "access",
    "id",
}

missing = required_raw_cols - set(raw.columns)

assert not missing, (
    f"Raw BIEN file is missing expected columns: {sorted(missing)}"
)

assert len(raw) == 77, (
    f"Expected 77 raw BIEN records, found {len(raw)}."
)

# -----------------------------------------------------------------------------
# 3. TRAIT AND UNIT INTEGRITY
# -----------------------------------------------------------------------------

trait_clean = (
    raw["trait_name"]
    .astype("string")
    .str.strip()
    .str.lower()
)

unit_clean = (
    raw["unit"]
    .astype("string")
    .str.strip()
    .str.lower()
)

print("\nTrait / unit integrity")
print("-" * 100)

print("Trait labels:")
print(trait_clean.value_counts(dropna=False).to_string())

print("\nUnits:")
print(unit_clean.value_counts(dropna=False).to_string())

assert trait_clean.eq("seed mass").all(), (
    "Non-Seed-Mass BIEN records detected."
)

assert unit_clean.eq("mg").all(), (
    "Unexpected Seed Mass units detected."
)

# -----------------------------------------------------------------------------
# 4. NUMERIC VALUE AUDIT
# -----------------------------------------------------------------------------

raw["Trait_Value_mg"] = pd.to_numeric(
    raw["trait_value"],
    errors="coerce"
)

numeric_summary = raw["Trait_Value_mg"].describe()

print("\nRaw numeric Seed Mass summary")
print("-" * 100)
print(numeric_summary.to_string())

n_missing_numeric = raw["Trait_Value_mg"].isna().sum()
n_nonpositive = raw["Trait_Value_mg"].le(0).sum()

print(f"\nNon-numeric/missing values : {n_missing_numeric}")
print(f"Non-positive values        : {n_nonpositive}")

# Do not remove anything yet.
assert n_missing_numeric == 0, (
    "Non-numeric BIEN Seed Mass values require review."
)

assert n_nonpositive == 0, (
    "Zero/negative BIEN Seed Mass values require review."
)

# -----------------------------------------------------------------------------
# 5. BIEN TAXONOMIC FIELD SUMMARY
# -----------------------------------------------------------------------------

taxonomic_fields = [
    "name_submitted",
    "verbatim_scientific_name",
    "name_matched",
    "scrubbed_species_binomial",
    "scrubbed_taxon_canonical",
]

print("\nBIEN taxonomic-field uniqueness")
print("-" * 100)

taxonomic_summary = pd.DataFrame({
    "Field": taxonomic_fields,
    "N_Unique_Nonmissing": [
        raw[c].dropna().astype(str).str.strip().nunique()
        for c in taxonomic_fields
    ],
    "N_Missing": [
        raw[c].isna().sum()
        for c in taxonomic_fields
    ],
})

display(taxonomic_summary)

# -----------------------------------------------------------------------------
# 6. PROJECT CANONICAL TAXON SET
# -----------------------------------------------------------------------------

omo_taxa = (
    manifest["Canonical_Species"]
    .dropna()
    .astype(str)
    .str.strip()
)

assert omo_taxa.nunique() == 127
assert len(omo_taxa) == 127

omo_set = set(omo_taxa)

# -----------------------------------------------------------------------------
# 7. EXACT MATCH AUDIT ACROSS BIEN NAME FIELDS
# -----------------------------------------------------------------------------

match_records = []

for field in taxonomic_fields:

    cleaned = (
        raw[field]
        .astype("string")
        .str.strip()
    )

    exact_mask = cleaned.isin(omo_set)

    match_records.append({
        "BIEN_Name_Field": field,
        "N_Raw_Records_Exact_Match": int(exact_mask.sum()),
        "N_Unique_Omo_Taxa_Matched": int(
            cleaned[exact_mask].nunique()
        ),
        "Percent_Raw_Records_Matched": round(
            100 * exact_mask.mean(),
            2
        ),
    })

match_summary = pd.DataFrame(match_records)

print("\nExact-match performance against Omo canonical backbone")
print("-" * 100)
display(match_summary)

# -----------------------------------------------------------------------------
# 8. RECORD-LEVEL MULTI-FIELD EXACT MATCH
# -----------------------------------------------------------------------------
#
# A raw BIEN record is considered directly crosswalkable at this stage if
# ANY recognized BIEN taxonomic name field exactly equals an Omo canonical
# project identifier.
# -----------------------------------------------------------------------------

def exact_omo_matches(row):

    matches = []

    for field in taxonomic_fields:

        value = row.get(field)

        if pd.isna(value):
            continue

        value = str(value).strip()

        if value in omo_set:
            matches.append(value)

    return sorted(set(matches))


raw["Exact_Omo_Matches"] = raw.apply(
    exact_omo_matches,
    axis=1
)

raw["N_Exact_Omo_Matches"] = (
    raw["Exact_Omo_Matches"]
    .apply(len)
)

raw["Exact_Omo_Match"] = raw[
    "Exact_Omo_Matches"
].apply(
    lambda x: x[0] if len(x) == 1 else pd.NA
)

raw["Crosswalk_Status"] = np.select(
    [
        raw["N_Exact_Omo_Matches"].eq(1),
        raw["N_Exact_Omo_Matches"].gt(1),
    ],
    [
        "EXACT_CROSSWALK",
        "AMBIGUOUS_MULTIPLE_EXACT",
    ],
    default="NO_EXACT_CROSSWALK"
)

crosswalk_summary = (
    raw["Crosswalk_Status"]
    .value_counts(dropna=False)
    .rename_axis("Crosswalk_Status")
    .reset_index(name="N_Raw_Records")
)

print("\nRecord-level exact crosswalk summary")
print("-" * 100)
display(crosswalk_summary)

# -----------------------------------------------------------------------------
# 9. UNIQUE TAXA SUPPORTED BY EXACT CROSSWALK
# -----------------------------------------------------------------------------

exact_supported_taxa = sorted(
    raw.loc[
        raw["Crosswalk_Status"].eq("EXACT_CROSSWALK"),
        "Exact_Omo_Match"
    ]
    .dropna()
    .unique()
)

print("\nOmo taxa with at least one exact-crosswalk BIEN record")
print("-" * 100)

print(f"N taxa : {len(exact_supported_taxa)}")

for taxon in exact_supported_taxa:
    print(f" - {taxon}")

# -----------------------------------------------------------------------------
# 10. RECORDS REQUIRING TAXONOMIC RECONCILIATION
# -----------------------------------------------------------------------------

review_cols = [
    "id",
    "name_submitted",
    "verbatim_scientific_name",
    "name_matched",
    "scrubbed_species_binomial",
    "scrubbed_taxon_canonical",
    "trait_value",
    "unit",
    "source_citation",
    "Crosswalk_Status",
]

taxonomic_review = raw.loc[
    ~raw["Crosswalk_Status"].eq("EXACT_CROSSWALK"),
    review_cols
].copy()

print("\nBIEN records requiring taxonomic reconciliation")
print("-" * 100)

print(f"N raw records : {len(taxonomic_review)}")

if taxonomic_review.empty:
    print("NONE")
else:
    display(taxonomic_review)

# -----------------------------------------------------------------------------
# 11. SOURCE / METHOD DIAGNOSTIC
# -----------------------------------------------------------------------------

source_summary = (
    raw.groupby(
        ["url_source", "source_citation"],
        dropna=False
    )
    .size()
    .reset_index(name="N_Raw_Records")
    .sort_values(
        "N_Raw_Records",
        ascending=False
    )
    .reset_index(drop=True)
)

print("\nBIEN Seed Mass source summary")
print("-" * 100)
display(source_summary)

method_summary = (
    raw["method"]
    .fillna("<MISSING>")
    .astype(str)
    .value_counts()
    .rename_axis("Method")
    .reset_index(name="N_Raw_Records")
)

print("\nBIEN Seed Mass method summary")
print("-" * 100)
display(method_summary)

# -----------------------------------------------------------------------------
# 12. EXPORT AUDIT TABLES
# -----------------------------------------------------------------------------

record_audit_export = (
    FT_DIR / "omo_seed_mass_bien_raw_crosswalk_audit.csv"
)

review_export = (
    FT_DIR / "omo_seed_mass_bien_taxonomic_review.csv"
)

source_export = (
    FT_DIR / "omo_seed_mass_bien_source_summary.csv"
)

method_export = (
    FT_DIR / "omo_seed_mass_bien_method_summary.csv"
)

raw_export = raw.copy()

# Convert list field to readable text for CSV export.
raw_export["Exact_Omo_Matches"] = (
    raw_export["Exact_Omo_Matches"]
    .apply(lambda x: " | ".join(x))
)

raw_export.to_csv(
    record_audit_export,
    index=False
)

taxonomic_review.to_csv(
    review_export,
    index=False
)

source_summary.to_csv(
    source_export,
    index=False
)

method_summary.to_csv(
    method_export,
    index=False
)

print("\nAudit exports")
print("-" * 100)
print(f"Record crosswalk : {record_audit_export}")
print(f"Taxonomic review : {review_export}")
print(f"Source summary   : {source_export}")
print(f"Method summary   : {method_export}")

# -----------------------------------------------------------------------------
# 13. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert (
    raw["N_Exact_Omo_Matches"] <= 1
).all(), (
    "At least one BIEN record maps exactly to multiple Omo canonical taxa."
)

print("\nBIEN raw-data integrity and taxonomic crosswalk audit completed successfully.")
print("No measurements have yet been accepted, deduplicated or aggregated.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1F — BIEN DUPLICATE, SOURCE-DIVERSITY AND WITHIN-SPECIES VARIABILITY AUDIT
# =============================================================================
#
# Purpose
# -------
# Diagnose redundancy and biological variation in the 77 direct BIEN
# species-level Seed Mass records before deciding what constitutes an
# accepted measurement set.
#
# This cell:
#   1. identifies exact duplicated rows;
#   2. identifies repeated species-value-source combinations;
#   3. summarizes unique values and source diversity per species;
#   4. quantifies within-species variation;
#   5. flags large within-species ranges for review.
#
# IMPORTANT:
#   - No records are removed.
#   - No values are aggregated into final species estimates.
#   - No fallback estimation occurs.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

CROSSWALK_FILE = (
    FT_DIR / "omo_seed_mass_bien_raw_crosswalk_audit.csv"
)

assert CROSSWALK_FILE.exists(), (
    f"Required crosswalk audit file not found:\n{CROSSWALK_FILE}"
)

df = pd.read_csv(
    CROSSWALK_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1F — BIEN DUPLICATE, SOURCE-DIVERSITY AND WITHIN-SPECIES VARIABILITY AUDIT")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. REQUIRED FIELDS
# -----------------------------------------------------------------------------

required_cols = {
    "id",
    "Exact_Omo_Match",
    "trait_value",
    "Trait_Value_mg",
    "unit",
    "method",
    "url_source",
    "source_citation",
    "scrubbed_species_binomial",
}

missing = required_cols - set(df.columns)

assert not missing, (
    f"Missing expected columns: {sorted(missing)}"
)

assert len(df) == 77
assert df["Exact_Omo_Match"].notna().all()

df["Trait_Value_mg"] = pd.to_numeric(
    df["Trait_Value_mg"],
    errors="coerce"
)

assert df["Trait_Value_mg"].notna().all()
assert df["Trait_Value_mg"].gt(0).all()

# -----------------------------------------------------------------------------
# 2. COMPLETE ROW DUPLICATES
# -----------------------------------------------------------------------------
#
# BIEN 'id' is deliberately excluded from the scientific duplicate signature,
# because otherwise independently stored copies of the same measurement could
# never appear duplicated.
# -----------------------------------------------------------------------------

scientific_signature_cols = [
    "Exact_Omo_Match",
    "Trait_Value_mg",
    "method",
    "url_source",
    "source_citation",
]

complete_duplicate_mask = df.duplicated(
    subset=scientific_signature_cols,
    keep=False
)

complete_duplicate_records = (
    df.loc[
        complete_duplicate_mask,
        [
            "id",
            "Exact_Omo_Match",
            "Trait_Value_mg",
            "method",
            "url_source",
            "source_citation",
        ]
    ]
    .sort_values(
        ["Exact_Omo_Match", "Trait_Value_mg"]
    )
    .reset_index(drop=True)
)

print("\nRepeated scientific-signature records")
print("-" * 100)

print(
    f"Raw records belonging to repeated signatures : "
    f"{complete_duplicate_mask.sum()}"
)

if complete_duplicate_records.empty:
    print("NONE")
else:
    display(complete_duplicate_records)

# -----------------------------------------------------------------------------
# 3. UNIQUE SCIENTIFIC SIGNATURES
# -----------------------------------------------------------------------------

signature_counts = (
    df.groupby(
        scientific_signature_cols,
        dropna=False
    )
    .size()
    .reset_index(name="N_Raw_Records")
    .sort_values(
        ["N_Raw_Records", "Exact_Omo_Match"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

repeated_signatures = (
    signature_counts[
        signature_counts["N_Raw_Records"] > 1
    ]
    .copy()
)

print("\nRepeated species-value-source signatures")
print("-" * 100)

print(
    f"Unique repeated signatures : "
    f"{len(repeated_signatures)}"
)

if repeated_signatures.empty:
    print("NONE")
else:
    display(repeated_signatures)

# -----------------------------------------------------------------------------
# 4. SOURCE NORMALIZATION FOR DIAGNOSTIC COUNTING
# -----------------------------------------------------------------------------

df["Source_Key"] = (
    df["source_citation"]
    .fillna(df["url_source"])
    .fillna("<UNKNOWN_SOURCE>")
    .astype(str)
    .str.strip()
)

# -----------------------------------------------------------------------------
# 5. PER-SPECIES SUMMARY
# -----------------------------------------------------------------------------

species_summary = (
    df.groupby(
        "Exact_Omo_Match",
        dropna=False
    )
    .agg(
        N_Raw_Records=("Trait_Value_mg", "size"),
        N_Unique_Values=("Trait_Value_mg", "nunique"),
        N_Unique_Sources=("Source_Key", "nunique"),
        Min_mg=("Trait_Value_mg", "min"),
        Median_mg=("Trait_Value_mg", "median"),
        Mean_mg=("Trait_Value_mg", "mean"),
        Max_mg=("Trait_Value_mg", "max"),
    )
    .reset_index()
    .rename(
        columns={
            "Exact_Omo_Match": "Canonical_Species"
        }
    )
)

species_summary["Fold_Range"] = (
    species_summary["Max_mg"]
    / species_summary["Min_mg"]
)

species_summary["Log10_Range"] = (
    np.log10(species_summary["Max_mg"])
    - np.log10(species_summary["Min_mg"])
)

species_summary = species_summary.sort_values(
    "Fold_Range",
    ascending=False
).reset_index(drop=True)

print("\nPer-species raw Seed Mass summary")
print("-" * 100)

display(species_summary)

# -----------------------------------------------------------------------------
# 6. FLAG LARGE WITHIN-SPECIES VARIATION
# -----------------------------------------------------------------------------
#
# This is a REVIEW flag, not an exclusion rule.
#
# Fold range >= 5 is deliberately conservative and merely identifies taxa
# whose reported maximum is at least five times their minimum.
# -----------------------------------------------------------------------------

species_summary["Variability_Flag"] = np.where(
    species_summary["Fold_Range"] >= 5,
    "REVIEW_LARGE_WITHIN_SPECIES_RANGE",
    "OK_INITIAL"
)

variability_review = (
    species_summary[
        species_summary["Variability_Flag"]
        == "REVIEW_LARGE_WITHIN_SPECIES_RANGE"
    ]
    .copy()
)

print("\nSpecies flagged for large within-species variation")
print("-" * 100)

if variability_review.empty:
    print("NONE")
else:
    display(variability_review)

# -----------------------------------------------------------------------------
# 7. RECORD-LEVEL DETAIL FOR FLAGGED SPECIES
# -----------------------------------------------------------------------------

flagged_species = set(
    variability_review["Canonical_Species"]
)

flagged_records = (
    df[
        df["Exact_Omo_Match"].isin(flagged_species)
    ][
        [
            "id",
            "Exact_Omo_Match",
            "Trait_Value_mg",
            "method",
            "url_source",
            "source_citation",
        ]
    ]
    .sort_values(
        ["Exact_Omo_Match", "Trait_Value_mg"]
    )
    .reset_index(drop=True)
)

print("\nRecord-level detail for flagged species")
print("-" * 100)

if flagged_records.empty:
    print("NONE")
else:
    display(flagged_records)

# -----------------------------------------------------------------------------
# 8. SOURCE DIVERSITY BY SPECIES
# -----------------------------------------------------------------------------

source_diversity = (
    df.groupby(
        ["Exact_Omo_Match", "Source_Key"],
        dropna=False
    )
    .size()
    .reset_index(name="N_Raw_Records")
    .sort_values(
        ["Exact_Omo_Match", "N_Raw_Records"],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

print("\nSpecies × source evidence structure")
print("-" * 100)

display(source_diversity)

# -----------------------------------------------------------------------------
# 9. GLOBAL DUPLICATION SUMMARY
# -----------------------------------------------------------------------------

n_raw = len(df)

n_unique_signatures = (
    df[scientific_signature_cols]
    .drop_duplicates()
    .shape[0]
)

n_redundant_records = (
    n_raw - n_unique_signatures
)

duplication_summary = pd.DataFrame({
    "Metric": [
        "Raw BIEN records",
        "Unique scientific signatures",
        "Redundant records beyond first signature occurrence",
        "Species represented",
        "Species with >=2 unique sources",
        "Species flagged for large within-species range",
    ],
    "Value": [
        n_raw,
        n_unique_signatures,
        n_redundant_records,
        df["Exact_Omo_Match"].nunique(),
        int(
            (
                species_summary["N_Unique_Sources"] >= 2
            ).sum()
        ),
        len(variability_review),
    ],
})

print("\nGlobal duplicate / evidence summary")
print("-" * 100)

display(duplication_summary)

# -----------------------------------------------------------------------------
# 10. EXPORT DIAGNOSTICS
# -----------------------------------------------------------------------------

signature_export = (
    FT_DIR / "omo_seed_mass_bien_repeated_signatures.csv"
)

species_summary_export = (
    FT_DIR / "omo_seed_mass_bien_species_raw_summary.csv"
)

variability_export = (
    FT_DIR / "omo_seed_mass_bien_variability_review.csv"
)

flagged_records_export = (
    FT_DIR / "omo_seed_mass_bien_variability_review_records.csv"
)

source_diversity_export = (
    FT_DIR / "omo_seed_mass_bien_species_source_structure.csv"
)

repeated_signatures.to_csv(
    signature_export,
    index=False
)

species_summary.to_csv(
    species_summary_export,
    index=False
)

variability_review.to_csv(
    variability_export,
    index=False
)

flagged_records.to_csv(
    flagged_records_export,
    index=False
)

source_diversity.to_csv(
    source_diversity_export,
    index=False
)

print("\nDiagnostic exports")
print("-" * 100)
print(f"Repeated signatures : {signature_export}")
print(f"Species summary      : {species_summary_export}")
print(f"Variability review   : {variability_export}")
print(f"Flagged records      : {flagged_records_export}")
print(f"Source structure     : {source_diversity_export}")

# -----------------------------------------------------------------------------
# 11. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert species_summary["Canonical_Species"].nunique() == 15
assert species_summary["Min_mg"].gt(0).all()
assert species_summary["Max_mg"].ge(
    species_summary["Min_mg"]
).all()

print("\nBIEN duplicate and within-species variability audit completed successfully.")
print("No records have been removed or aggregated.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1G — BIEN DEDUPLICATION AND PROVISIONAL MEASUREMENT-LEVEL QC
# =============================================================================
#
# Purpose
# -------
# Convert the 77 raw BIEN Seed Mass records into a defensible provisional
# measurement-level evidence table by:
#
#   1. removing repeated scientific-signature redundancy;
#   2. preserving one representative BIEN record for each unique signature;
#   3. flagging high-variability taxa for manual/review-stage treatment;
#   4. retaining all other positive, numeric, mg Seed Mass measurements.
#
# IMPORTANT
# ---------
# This cell DOES NOT calculate final species-level Seed Mass estimates.
# Nauclea diderrichii and Elaeis guineensis are retained in the evidence
# ledger but flagged for review.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

CROSSWALK_FILE = (
    FT_DIR / "omo_seed_mass_bien_raw_crosswalk_audit.csv"
)

assert CROSSWALK_FILE.exists(), (
    f"Required file not found:\n{CROSSWALK_FILE}"
)

raw = pd.read_csv(
    CROSSWALK_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1G — BIEN DEDUPLICATION AND PROVISIONAL MEASUREMENT-LEVEL QC")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT
# -----------------------------------------------------------------------------

required_cols = {
    "id",
    "Exact_Omo_Match",
    "Trait_Value_mg",
    "trait_name",
    "unit",
    "method",
    "url_source",
    "source_citation",
}

missing = required_cols - set(raw.columns)

assert not missing, (
    f"Missing required columns: {sorted(missing)}"
)

raw["Trait_Value_mg"] = pd.to_numeric(
    raw["Trait_Value_mg"],
    errors="coerce"
)

assert len(raw) == 77
assert raw["Trait_Value_mg"].notna().all()
assert raw["Trait_Value_mg"].gt(0).all()
assert raw["Exact_Omo_Match"].notna().all()

# -----------------------------------------------------------------------------
# 2. DEFINE SCIENTIFIC DUPLICATE SIGNATURE
# -----------------------------------------------------------------------------

signature_cols = [
    "Exact_Omo_Match",
    "Trait_Value_mg",
    "method",
    "url_source",
    "source_citation",
]

# Preserve original ordering for reproducibility.
raw = raw.reset_index(drop=True)

raw["Original_Record_Order"] = np.arange(
    1,
    len(raw) + 1
)

# Count how many raw BIEN rows share each scientific signature.
raw["Signature_Record_Count"] = (
    raw.groupby(
        signature_cols,
        dropna=False
    )["id"]
    .transform("size")
)

# -----------------------------------------------------------------------------
# 3. DEDUPLICATE SCIENTIFIC SIGNATURES
# -----------------------------------------------------------------------------
#
# Keep one representative row per species-value-method-source combination.
#
# We are NOT claiming that equal values can never represent independent
# observations. Here duplication is removed only when value + method +
# source citation + source URL are simultaneously identical.
# -----------------------------------------------------------------------------

deduplicated = (
    raw.sort_values(
        "Original_Record_Order"
    )
    .drop_duplicates(
        subset=signature_cols,
        keep="first"
    )
    .copy()
    .reset_index(drop=True)
)

deduplicated["Deduplication_Status"] = np.where(
    deduplicated["Signature_Record_Count"] > 1,
    "RETAINED_REPRESENTATIVE_OF_REPEATED_SIGNATURE",
    "UNIQUE_SIGNATURE"
)

print("\nDeduplication summary")
print("-" * 100)

print(f"Raw BIEN records                 : {len(raw)}")
print(f"Unique scientific signatures     : {len(deduplicated)}")
print(
    f"Redundant rows removed            : "
    f"{len(raw) - len(deduplicated)}"
)

assert len(deduplicated) == 58, (
    f"Expected 58 unique scientific signatures, "
    f"found {len(deduplicated)}."
)

# -----------------------------------------------------------------------------
# 4. DEFINE HIGH-VARIABILITY REVIEW TAXA
# -----------------------------------------------------------------------------

high_variability_taxa = {
    "Nauclea diderrichii",
    "Elaeis guineensis",
}

deduplicated["Variability_Status"] = np.where(
    deduplicated["Exact_Omo_Match"].isin(
        high_variability_taxa
    ),
    "REVIEW_HIGH_WITHIN_SPECIES_VARIATION",
    "INITIAL_VARIABILITY_OK"
)

# -----------------------------------------------------------------------------
# 5. ASSIGN PROVISIONAL MEASUREMENT QC STATUS
# -----------------------------------------------------------------------------

deduplicated["Measurement_QC_Status"] = np.select(
    [
        deduplicated["Trait_Value_mg"].isna(),
        deduplicated["Trait_Value_mg"].le(0),
        deduplicated["Exact_Omo_Match"].isin(
            high_variability_taxa
        ),
    ],
    [
        "EXCLUDE_NON_NUMERIC",
        "EXCLUDE_NONPOSITIVE",
        "HOLD_FOR_VARIABILITY_REVIEW",
    ],
    default="PROVISIONALLY_RETAIN"
)

print("\nMeasurement-level QC status")
print("-" * 100)

qc_summary = (
    deduplicated["Measurement_QC_Status"]
    .value_counts()
    .rename_axis("Measurement_QC_Status")
    .reset_index(name="N_Records")
)

display(qc_summary)

# -----------------------------------------------------------------------------
# 6. PROVISIONALLY RETAINED MEASUREMENTS
# -----------------------------------------------------------------------------

provisionally_retained = (
    deduplicated[
        deduplicated["Measurement_QC_Status"]
        == "PROVISIONALLY_RETAIN"
    ]
    .copy()
    .reset_index(drop=True)
)

held_for_review = (
    deduplicated[
        deduplicated["Measurement_QC_Status"]
        == "HOLD_FOR_VARIABILITY_REVIEW"
    ]
    .copy()
    .reset_index(drop=True)
)

print("\nProvisional retained evidence")
print("-" * 100)

print(
    f"Retained measurement signatures : "
    f"{len(provisionally_retained)}"
)

print(
    f"Species represented              : "
    f"{provisionally_retained['Exact_Omo_Match'].nunique()}"
)

print("\nHeld for variability review")
print("-" * 100)

print(
    f"Held measurement signatures : "
    f"{len(held_for_review)}"
)

print(
    f"Held species                : "
    f"{held_for_review['Exact_Omo_Match'].nunique()}"
)

if not held_for_review.empty:
    display(
        held_for_review[
            [
                "id",
                "Exact_Omo_Match",
                "Trait_Value_mg",
                "method",
                "url_source",
                "source_citation",
                "Measurement_QC_Status",
            ]
        ]
        .sort_values(
            ["Exact_Omo_Match", "Trait_Value_mg"]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 7. SPECIES-LEVEL PROVISIONAL COVERAGE
# -----------------------------------------------------------------------------

all_species = sorted(
    deduplicated["Exact_Omo_Match"]
    .dropna()
    .unique()
)

retained_species = set(
    provisionally_retained["Exact_Omo_Match"]
)

held_species = set(
    held_for_review["Exact_Omo_Match"]
)

coverage_records = []

for species in all_species:

    if species in retained_species:
        status = "PROVISIONAL_BIEN_EVIDENCE"
    elif species in held_species:
        status = "BIEN_EVIDENCE_HELD_FOR_REVIEW"
    else:
        status = "NO_USABLE_BIEN_EVIDENCE"

    coverage_records.append({
        "Canonical_Species": species,
        "BIEN_Provisional_Status": status
    })

coverage = pd.DataFrame(coverage_records)

print("\nSpecies-level provisional BIEN status")
print("-" * 100)

display(
    coverage[
        "BIEN_Provisional_Status"
    ]
    .value_counts()
    .rename_axis("BIEN_Provisional_Status")
    .reset_index(name="N_Species")
)

# -----------------------------------------------------------------------------
# 8. SOURCE DIVERSITY AFTER DEDUPLICATION
# -----------------------------------------------------------------------------

deduplicated["Source_Key"] = (
    deduplicated["source_citation"]
    .fillna(deduplicated["url_source"])
    .fillna("<UNKNOWN_SOURCE>")
    .astype(str)
    .str.strip()
)

post_dedup_species_summary = (
    deduplicated.groupby(
        "Exact_Omo_Match"
    )
    .agg(
        N_Unique_Measurement_Signatures=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        N_Sources=(
            "Source_Key",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Median_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
    .rename(
        columns={
            "Exact_Omo_Match":
                "Canonical_Species"
        }
    )
)

post_dedup_species_summary["Fold_Range"] = (
    post_dedup_species_summary["Max_mg"]
    / post_dedup_species_summary["Min_mg"]
)

post_dedup_species_summary["Species_QC_Status"] = np.where(
    post_dedup_species_summary[
        "Canonical_Species"
    ].isin(high_variability_taxa),
    "REVIEW_HIGH_WITHIN_SPECIES_VARIATION",
    "PROVISIONAL_OK"
)

print("\nPost-deduplication species evidence summary")
print("-" * 100)

display(
    post_dedup_species_summary
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 9. EXPORT QC TABLES
# -----------------------------------------------------------------------------

dedup_export = (
    FT_DIR /
    "omo_seed_mass_bien_deduplicated_measurements.csv"
)

retained_export = (
    FT_DIR /
    "omo_seed_mass_bien_provisionally_retained_measurements.csv"
)

held_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_hold_measurements.csv"
)

summary_export = (
    FT_DIR /
    "omo_seed_mass_bien_post_dedup_species_summary.csv"
)

coverage_export = (
    FT_DIR /
    "omo_seed_mass_bien_provisional_species_status.csv"
)

deduplicated.to_csv(
    dedup_export,
    index=False
)

provisionally_retained.to_csv(
    retained_export,
    index=False
)

held_for_review.to_csv(
    held_export,
    index=False
)

post_dedup_species_summary.to_csv(
    summary_export,
    index=False
)

coverage.to_csv(
    coverage_export,
    index=False
)

print("\nQC exports")
print("-" * 100)
print(f"Deduplicated measurements : {dedup_export}")
print(f"Provisionally retained     : {retained_export}")
print(f"Variability hold           : {held_export}")
print(f"Species summary            : {summary_export}")
print(f"Species status             : {coverage_export}")

# -----------------------------------------------------------------------------
# 10. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(deduplicated) == 58

assert (
    len(deduplicated)
    + (len(raw) - len(deduplicated))
    == len(raw)
)

assert set(
    held_for_review["Exact_Omo_Match"]
) == high_variability_taxa

assert (
    provisionally_retained[
        "Trait_Value_mg"
    ].gt(0).all()
)

print("\nBIEN Seed Mass deduplication and provisional QC completed successfully.")
print("No final species-level Seed Mass estimates have yet been calculated.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1H — TARGETED REVIEW OF HELD BIEN SEED MASS RECORDS
# =============================================================================
#
# Purpose
# -------
# Investigate the 7 BIEN Seed Mass measurements held because of extreme
# within-species variability for:
#
#   - Elaeis guineensis
#   - Nauclea diderrichii
#
# This cell examines:
#   1. full taxonomic metadata;
#   2. submitted/verbatim/scrubbed names;
#   3. BIEN record identifiers;
#   4. methods and sources;
#   5. value structure on linear and log10 scales;
#   6. whether suspicious values are associated with distinct metadata.
#
# IMPORTANT:
#   - No values are accepted or rejected here.
#   - No aggregation occurs.
#   - This is a forensic evidence-review step only.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

HOLD_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_variability_hold_measurements.csv"
)

assert HOLD_FILE.exists(), (
    f"Variability-hold file not found:\n{HOLD_FILE}"
)

held = pd.read_csv(
    HOLD_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1H — TARGETED REVIEW OF HELD BIEN SEED MASS RECORDS")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE HELD EVIDENCE
# -----------------------------------------------------------------------------

held_taxa_expected = {
    "Elaeis guineensis",
    "Nauclea diderrichii",
}

assert len(held) == 7, (
    f"Expected 7 held records, found {len(held)}."
)

assert set(
    held["Exact_Omo_Match"]
) == held_taxa_expected, (
    "Held taxa differ from the validated Section 9.1G result."
)

held["Trait_Value_mg"] = pd.to_numeric(
    held["Trait_Value_mg"],
    errors="coerce"
)

assert held["Trait_Value_mg"].notna().all()
assert held["Trait_Value_mg"].gt(0).all()

# -----------------------------------------------------------------------------
# 2. FULL TAXONOMIC AND SOURCE METADATA
# -----------------------------------------------------------------------------

review_fields = [
    "id",
    "Exact_Omo_Match",
    "Trait_Value_mg",
    "name_submitted",
    "verbatim_scientific_name",
    "name_matched",
    "scrubbed_species_binomial",
    "scrubbed_taxon_canonical",
    "family_matched",
    "scrubbed_family",
    "matched_taxonomic_status",
    "scrubbed_taxonomic_status",
    "tnrs_warning",
    "method",
    "url_source",
    "source_citation",
    "project_pi",
    "access",
]

available_review_fields = [
    c for c in review_fields
    if c in held.columns
]

held_review = (
    held[available_review_fields]
    .sort_values(
        ["Exact_Omo_Match", "Trait_Value_mg"]
    )
    .reset_index(drop=True)
)

print("\nFull held-record metadata")
print("-" * 100)

display(held_review)

# -----------------------------------------------------------------------------
# 3. TAXONOMIC CONSISTENCY CHECK
# -----------------------------------------------------------------------------

taxonomic_check_fields = [
    c for c in [
        "name_submitted",
        "verbatim_scientific_name",
        "name_matched",
        "scrubbed_species_binomial",
        "scrubbed_taxon_canonical",
        "matched_taxonomic_status",
        "scrubbed_taxonomic_status",
        "tnrs_warning",
    ]
    if c in held.columns
]

taxonomic_diagnostics = []

for species, group in held.groupby("Exact_Omo_Match"):

    record = {
        "Canonical_Species": species,
        "N_Records": len(group),
    }

    for col in taxonomic_check_fields:

        values = (
            group[col]
            .dropna()
            .astype(str)
            .str.strip()
        )

        record[f"{col}_N_Unique"] = values.nunique()

        record[f"{col}_Values"] = (
            " | ".join(sorted(values.unique()))
            if len(values) > 0
            else "<MISSING>"
        )

    taxonomic_diagnostics.append(record)

taxonomic_diagnostics = pd.DataFrame(
    taxonomic_diagnostics
)

print("\nTaxonomic consistency diagnostics")
print("-" * 100)

display(taxonomic_diagnostics)

# -----------------------------------------------------------------------------
# 4. SOURCE / METHOD CONSISTENCY
# -----------------------------------------------------------------------------

source_method_summary = (
    held.groupby(
        [
            "Exact_Omo_Match",
            "method",
            "url_source",
            "source_citation",
        ],
        dropna=False
    )
    .agg(
        N_Records=("Trait_Value_mg", "size"),
        Min_mg=("Trait_Value_mg", "min"),
        Median_mg=("Trait_Value_mg", "median"),
        Max_mg=("Trait_Value_mg", "max"),
    )
    .reset_index()
)

print("\nHeld-record source × method structure")
print("-" * 100)

display(source_method_summary)

# -----------------------------------------------------------------------------
# 5. VALUE STRUCTURE
# -----------------------------------------------------------------------------

held["Log10_Seed_Mass"] = np.log10(
    held["Trait_Value_mg"]
)

value_diagnostics = (
    held.groupby(
        "Exact_Omo_Match"
    )
    .agg(
        N_Records=("Trait_Value_mg", "size"),
        Min_mg=("Trait_Value_mg", "min"),
        Median_mg=("Trait_Value_mg", "median"),
        Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: 10 ** np.mean(np.log10(x))
        ),
        Mean_mg=("Trait_Value_mg", "mean"),
        Max_mg=("Trait_Value_mg", "max"),
        Min_Log10=("Log10_Seed_Mass", "min"),
        Median_Log10=("Log10_Seed_Mass", "median"),
        Max_Log10=("Log10_Seed_Mass", "max"),
    )
    .reset_index()
    .rename(
        columns={
            "Exact_Omo_Match":
                "Canonical_Species"
        }
    )
)

value_diagnostics["Fold_Range"] = (
    value_diagnostics["Max_mg"]
    / value_diagnostics["Min_mg"]
)

value_diagnostics["Orders_of_Magnitude"] = (
    value_diagnostics["Max_Log10"]
    - value_diagnostics["Min_Log10"]
)

print("\nValue-distribution diagnostics")
print("-" * 100)

display(value_diagnostics)

# -----------------------------------------------------------------------------
# 6. NEIGHBOURING VALUE RATIOS
# -----------------------------------------------------------------------------
#
# Large jumps between adjacent sorted measurements can help identify
# potentially heterogeneous groups without imposing an arbitrary cutoff.
# -----------------------------------------------------------------------------

jump_records = []

for species, group in held.groupby("Exact_Omo_Match"):

    values = sorted(
        group["Trait_Value_mg"]
        .astype(float)
        .tolist()
    )

    for i in range(1, len(values)):

        jump_records.append({
            "Canonical_Species": species,
            "Lower_Value_mg": values[i - 1],
            "Upper_Value_mg": values[i],
            "Adjacent_Fold_Jump": (
                values[i] / values[i - 1]
            ),
            "Adjacent_Log10_Jump": (
                np.log10(values[i])
                - np.log10(values[i - 1])
            ),
        })

jump_table = pd.DataFrame(jump_records)

print("\nAdjacent-value jump diagnostics")
print("-" * 100)

display(
    jump_table
    .sort_values(
        "Adjacent_Fold_Jump",
        ascending=False
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 7. RECORD-LEVEL REVIEW REGISTER
# -----------------------------------------------------------------------------

review_register = held_review.copy()

review_register["Review_Decision"] = "PENDING"
review_register["Review_Rationale"] = ""
review_register["Evidence_Action"] = "HOLD"

print("\nTargeted review register")
print("-" * 100)

display(review_register)

# -----------------------------------------------------------------------------
# 8. EXPORT TARGETED REVIEW TABLES
# -----------------------------------------------------------------------------

metadata_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_metadata_review.csv"
)

taxonomic_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_taxonomic_diagnostics.csv"
)

value_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_value_diagnostics.csv"
)

jump_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_adjacent_jumps.csv"
)

register_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_review_register.csv"
)

held_review.to_csv(
    metadata_export,
    index=False
)

taxonomic_diagnostics.to_csv(
    taxonomic_export,
    index=False
)

value_diagnostics.to_csv(
    value_export,
    index=False
)

jump_table.to_csv(
    jump_export,
    index=False
)

review_register.to_csv(
    register_export,
    index=False
)

print("\nReview exports")
print("-" * 100)

print(f"Metadata review       : {metadata_export}")
print(f"Taxonomic diagnostics : {taxonomic_export}")
print(f"Value diagnostics     : {value_export}")
print(f"Adjacent jumps        : {jump_export}")
print(f"Review register       : {register_export}")

# -----------------------------------------------------------------------------
# 9. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(review_register) == 7
assert (
    review_register["Review_Decision"]
    == "PENDING"
).all()

assert (
    review_register["Evidence_Action"]
    == "HOLD"
).all()

print("\nTargeted BIEN variability review prepared successfully.")
print("All seven measurements remain on HOLD pending evidence interpretation.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1I — RESOLVE HELD BIEN SEED MASS MEASUREMENTS
# =============================================================================
#
# Purpose
# -------
# Resolve the seven BIEN measurements held in Section 9.1G using the
# targeted evidence review completed in Section 9.1H.
#
# Evidence decisions
# ------------------
# Elaeis guineensis:
#   Retain all four BIEN measurements.
#   Large variation is biologically plausible and is preserved explicitly
#   with a high-variability quality flag.
#
# Nauclea diderrichii:
#   Retain 0.30 mg and 2.83 mg.
#   Exclude 317.46 mg as inconsistent with independent species-specific
#   seed-weight evidence.
#
# IMPORTANT
# ---------
# No final species-level aggregation occurs in this cell.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

RETAINED_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_provisionally_retained_measurements.csv"
)

HOLD_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_variability_hold_measurements.csv"
)

assert RETAINED_FILE.exists()
assert HOLD_FILE.exists()

retained = pd.read_csv(
    RETAINED_FILE,
    low_memory=False
)

held = pd.read_csv(
    HOLD_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1I — RESOLVE HELD BIEN SEED MASS MEASUREMENTS")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT STATE
# -----------------------------------------------------------------------------

assert len(retained) == 51
assert len(held) == 7

held["Trait_Value_mg"] = pd.to_numeric(
    held["Trait_Value_mg"],
    errors="coerce"
)

assert held["Trait_Value_mg"].notna().all()

assert set(
    held["Exact_Omo_Match"]
) == {
    "Elaeis guineensis",
    "Nauclea diderrichii",
}

# -----------------------------------------------------------------------------
# 2. INITIALIZE REVIEW DECISION FIELDS
# -----------------------------------------------------------------------------

held["Review_Decision"] = pd.NA
held["Evidence_Action"] = pd.NA
held["Review_Rationale"] = pd.NA
held["External_Evidence_Class"] = pd.NA

# -----------------------------------------------------------------------------
# 3. ELAEIS GUINEENSIS
# -----------------------------------------------------------------------------
#
# Retain all four measurements.
#
# External seed literature demonstrates substantial seed-mass variation
# associated with oil-palm fruit/seed types. Therefore the BIEN range is
# preserved rather than statistically trimmed.
# -----------------------------------------------------------------------------

elaeis_mask = (
    held["Exact_Omo_Match"]
    .eq("Elaeis guineensis")
)

held.loc[
    elaeis_mask,
    "Review_Decision"
] = "ACCEPT"

held.loc[
    elaeis_mask,
    "Evidence_Action"
] = "RETAIN_MEASUREMENT"

held.loc[
    elaeis_mask,
    "External_Evidence_Class"
] = "BIOLOGICALLY_PLAUSIBLE_HIGH_VARIABILITY"

held.loc[
    elaeis_mask,
    "Review_Rationale"
] = (
    "Retained after targeted external review. "
    "Published oil-palm seed information supports substantial "
    "within-species seed-mass variation; no individual BIEN value "
    "is rejected solely on statistical extremeness."
)

# -----------------------------------------------------------------------------
# 4. NAUCLEA DIDERRICHII
# -----------------------------------------------------------------------------
#
# Independent species-specific evidence reports approximately 1–18 g
# per 1000 seeds, corresponding to about 1–18 mg per seed.
#
# 0.30 and 2.83 mg are retained as small-seed measurements.
# 317.46 mg is rejected as biologically inconsistent with the external
# species-specific seed evidence and morphology.
# -----------------------------------------------------------------------------

nauclea_accept_mask = (
    held["Exact_Omo_Match"]
    .eq("Nauclea diderrichii")
    &
    held["Trait_Value_mg"]
    .isin([0.30, 2.83])
)

nauclea_reject_mask = (
    held["Exact_Omo_Match"]
    .eq("Nauclea diderrichii")
    &
    np.isclose(
        held["Trait_Value_mg"],
        317.46
    )
)

held.loc[
    nauclea_accept_mask,
    "Review_Decision"
] = "ACCEPT"

held.loc[
    nauclea_accept_mask,
    "Evidence_Action"
] = "RETAIN_MEASUREMENT"

held.loc[
    nauclea_accept_mask,
    "External_Evidence_Class"
] = "CONSISTENT_WITH_SPECIES_SEED_EVIDENCE"

held.loc[
    nauclea_accept_mask,
    "Review_Rationale"
] = (
    "Retained after targeted external review. "
    "Value is compatible with the very small seeds and published "
    "species-specific 1000-seed-weight evidence."
)

held.loc[
    nauclea_reject_mask,
    "Review_Decision"
] = "REJECT"

held.loc[
    nauclea_reject_mask,
    "Evidence_Action"
] = "EXCLUDE_MEASUREMENT"

held.loc[
    nauclea_reject_mask,
    "External_Evidence_Class"
] = "BIOLOGICALLY_INCONSISTENT_OUTLIER"

held.loc[
    nauclea_reject_mask,
    "Review_Rationale"
] = (
    "Excluded after targeted external review. "
    "317.46 mg is inconsistent with independent species-specific "
    "1000-seed-weight evidence and the documented very small seed size."
)

# -----------------------------------------------------------------------------
# 5. VALIDATE ALL SEVEN DECISIONS
# -----------------------------------------------------------------------------

assert held["Review_Decision"].notna().all()
assert held["Evidence_Action"].notna().all()
assert held["Review_Rationale"].notna().all()

decision_summary = (
    held["Review_Decision"]
    .value_counts()
    .rename_axis("Review_Decision")
    .reset_index(name="N_Measurements")
)

print("\nHeld-measurement decisions")
print("-" * 100)

display(decision_summary)

assert (
    (held["Review_Decision"] == "ACCEPT").sum()
    == 6
)

assert (
    (held["Review_Decision"] == "REJECT").sum()
    == 1
)

# -----------------------------------------------------------------------------
# 6. DISPLAY COMPLETE REVIEW REGISTER
# -----------------------------------------------------------------------------

display_cols = [
    "id",
    "Exact_Omo_Match",
    "Trait_Value_mg",
    "Review_Decision",
    "Evidence_Action",
    "External_Evidence_Class",
    "Review_Rationale",
]

print("\nResolved held-measurement register")
print("-" * 100)

display(
    held[display_cols]
    .sort_values(
        ["Exact_Omo_Match", "Trait_Value_mg"]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 7. CREATE FINAL ACCEPTED BIEN MEASUREMENT SET
# -----------------------------------------------------------------------------

held_accepted = (
    held[
        held["Review_Decision"]
        .eq("ACCEPT")
    ]
    .copy()
)

held_rejected = (
    held[
        held["Review_Decision"]
        .eq("REJECT")
    ]
    .copy()
)

# Add equivalent evidence fields to the original 51 retained records.
retained["Review_Decision"] = "ACCEPT"
retained["Evidence_Action"] = "RETAIN_MEASUREMENT"

retained["External_Evidence_Class"] = (
    "NO_TARGETED_EXTERNAL_REVIEW_REQUIRED"
)

retained["Review_Rationale"] = (
    "Passed BIEN measurement-level QC and did not require "
    "targeted variability review."
)

accepted_measurements = pd.concat(
    [
        retained,
        held_accepted,
    ],
    ignore_index=True,
    sort=False
)

# -----------------------------------------------------------------------------
# 8. VALIDATE FINAL MEASUREMENT SET
# -----------------------------------------------------------------------------

print("\nAccepted BIEN measurement evidence")
print("-" * 100)

print(
    f"Previously provisionally retained : {len(retained)}"
)

print(
    f"Accepted from held review         : {len(held_accepted)}"
)

print(
    f"Rejected from held review         : {len(held_rejected)}"
)

print(
    f"Final accepted measurements       : {len(accepted_measurements)}"
)

print(
    f"Species represented               : "
    f"{accepted_measurements['Exact_Omo_Match'].nunique()}"
)

assert len(accepted_measurements) == 57
assert len(held_rejected) == 1

assert (
    accepted_measurements[
        "Exact_Omo_Match"
    ].nunique()
    == 15
)

assert (
    accepted_measurements[
        "Trait_Value_mg"
    ].gt(0).all()
)

# -----------------------------------------------------------------------------
# 9. POST-REVIEW SPECIES SUMMARY
# -----------------------------------------------------------------------------

post_review_summary = (
    accepted_measurements.groupby(
        "Exact_Omo_Match"
    )
    .agg(
        N_Accepted_Measurements=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Median_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
    .rename(
        columns={
            "Exact_Omo_Match":
                "Canonical_Species"
        }
    )
)

post_review_summary["Fold_Range"] = (
    post_review_summary["Max_mg"]
    /
    post_review_summary["Min_mg"]
)

post_review_summary["Evidence_Quality_Flag"] = np.where(
    post_review_summary[
        "Canonical_Species"
    ].eq("Elaeis guineensis"),
    "HIGH_INTRASPECIFIC_VARIABILITY_RETAINED",
    np.where(
        post_review_summary[
            "Canonical_Species"
        ].eq("Nauclea diderrichii"),
        "TARGETED_OUTLIER_REVIEW_COMPLETED",
        "BIEN_QC_PASSED"
    )
)

print("\nPost-review species evidence summary")
print("-" * 100)

display(
    post_review_summary
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 10. EXPORT RESOLVED EVIDENCE
# -----------------------------------------------------------------------------

register_export = (
    FT_DIR /
    "omo_seed_mass_bien_variability_review_register_resolved.csv"
)

accepted_export = (
    FT_DIR /
    "omo_seed_mass_bien_accepted_measurements.csv"
)

rejected_export = (
    FT_DIR /
    "omo_seed_mass_bien_rejected_measurements.csv"
)

summary_export = (
    FT_DIR /
    "omo_seed_mass_bien_post_review_species_summary.csv"
)

held.to_csv(
    register_export,
    index=False
)

accepted_measurements.to_csv(
    accepted_export,
    index=False
)

held_rejected.to_csv(
    rejected_export,
    index=False
)

post_review_summary.to_csv(
    summary_export,
    index=False
)

print("\nResolved evidence exports")
print("-" * 100)

print(f"Resolved register      : {register_export}")
print(f"Accepted measurements  : {accepted_export}")
print(f"Rejected measurements  : {rejected_export}")
print(f"Post-review summary    : {summary_export}")

print(
    "\nHeld BIEN Seed Mass evidence resolved successfully."
)

print(
    "57 measurements across 15 species are now accepted for "
    "subsequent species-level aggregation."
)

print(
    "One Nauclea diderrichii measurement (317.46 mg) was "
    "excluded with explicit external-evidence justification."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1J — CANDIDATE SPECIES-LEVEL SEED MASS AGGREGATION DIAGNOSTIC
# =============================================================================
#
# Purpose
# -------
# Compare scientifically defensible species-level aggregation strategies
# for the 57 accepted BIEN Seed Mass measurements.
#
# Candidate summaries:
#
#   1. Arithmetic mean
#   2. Median
#   3. Geometric mean
#   4. Median of source-level medians
#
# The fourth approach reduces over-weighting of sources contributing many
# repeated/related observations.
#
# IMPORTANT
# ---------
# This cell does NOT yet select the final species-level Seed Mass estimate.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

ACCEPTED_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_accepted_measurements.csv"
)

assert ACCEPTED_FILE.exists(), (
    f"Accepted BIEN measurement file not found:\n{ACCEPTED_FILE}"
)

df = pd.read_csv(
    ACCEPTED_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1J — CANDIDATE SPECIES-LEVEL SEED MASS AGGREGATION DIAGNOSTIC")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT
# -----------------------------------------------------------------------------

required_cols = {
    "Exact_Omo_Match",
    "Trait_Value_mg",
    "source_citation",
    "url_source",
}

missing = required_cols - set(df.columns)

assert not missing, (
    f"Missing required columns: {sorted(missing)}"
)

df["Trait_Value_mg"] = pd.to_numeric(
    df["Trait_Value_mg"],
    errors="coerce"
)

assert len(df) == 57
assert df["Trait_Value_mg"].notna().all()
assert df["Trait_Value_mg"].gt(0).all()
assert df["Exact_Omo_Match"].nunique() == 15

# -----------------------------------------------------------------------------
# 2. CREATE SOURCE KEY
# -----------------------------------------------------------------------------

df["Source_Key"] = (
    df["source_citation"]
    .fillna(df["url_source"])
    .fillna("<UNKNOWN_SOURCE>")
    .astype(str)
    .str.strip()
)

# -----------------------------------------------------------------------------
# 3. SOURCE-LEVEL MEDIANS
# -----------------------------------------------------------------------------

source_level = (
    df.groupby(
        ["Exact_Omo_Match", "Source_Key"],
        dropna=False
    )
    .agg(
        N_Measurements=("Trait_Value_mg", "size"),
        Source_Median_mg=("Trait_Value_mg", "median"),
        Source_Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: float(
                np.exp(
                    np.mean(
                        np.log(
                            x.astype(float)
                        )
                    )
                )
            )
        ),
        Source_Min_mg=("Trait_Value_mg", "min"),
        Source_Max_mg=("Trait_Value_mg", "max"),
    )
    .reset_index()
)

# -----------------------------------------------------------------------------
# 4. SPECIES-LEVEL CANDIDATE SUMMARIES
# -----------------------------------------------------------------------------

species_base = (
    df.groupby(
        "Exact_Omo_Match"
    )
    .agg(
        N_Accepted_Measurements=("Trait_Value_mg", "size"),
        N_Unique_Values=("Trait_Value_mg", "nunique"),
        N_Sources=("Source_Key", "nunique"),
        Min_mg=("Trait_Value_mg", "min"),
        Arithmetic_Mean_mg=("Trait_Value_mg", "mean"),
        Median_mg=("Trait_Value_mg", "median"),
        Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: float(
                np.exp(
                    np.mean(
                        np.log(
                            x.astype(float)
                        )
                    )
                )
            )
        ),
        Max_mg=("Trait_Value_mg", "max"),
    )
    .reset_index()
    .rename(
        columns={
            "Exact_Omo_Match": "Canonical_Species"
        }
    )
)

source_balanced = (
    source_level.groupby(
        "Exact_Omo_Match"
    )
    .agg(
        Median_of_Source_Medians_mg=(
            "Source_Median_mg",
            "median"
        ),
        Mean_of_Source_Medians_mg=(
            "Source_Median_mg",
            "mean"
        ),
    )
    .reset_index()
    .rename(
        columns={
            "Exact_Omo_Match": "Canonical_Species"
        }
    )
)

comparison = species_base.merge(
    source_balanced,
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)

# -----------------------------------------------------------------------------
# 5. QUANTIFY DIFFERENCES AMONG CANDIDATE SUMMARIES
# -----------------------------------------------------------------------------

comparison["Mean_vs_Median_Ratio"] = (
    comparison["Arithmetic_Mean_mg"]
    /
    comparison["Median_mg"]
)

comparison["Geom_vs_Median_Ratio"] = (
    comparison["Geometric_Mean_mg"]
    /
    comparison["Median_mg"]
)

comparison["SourceMedian_vs_PooledMedian_Ratio"] = (
    comparison["Median_of_Source_Medians_mg"]
    /
    comparison["Median_mg"]
)

comparison["Fold_Range"] = (
    comparison["Max_mg"]
    /
    comparison["Min_mg"]
)

# -----------------------------------------------------------------------------
# 6. IDENTIFY SPECIES WHERE AGGREGATION METHOD MATTERS
# -----------------------------------------------------------------------------
#
# Review flag if source-balanced median differs from pooled median by >25%
# in either direction.
#
# This is a diagnostic threshold only, not a rejection criterion.
# -----------------------------------------------------------------------------

comparison["Aggregation_Sensitivity_Flag"] = np.where(
    (
        comparison[
            "SourceMedian_vs_PooledMedian_Ratio"
        ] > 1.25
    )
    |
    (
        comparison[
            "SourceMedian_vs_PooledMedian_Ratio"
        ] < 0.80
    ),
    "REVIEW_AGGREGATION_SENSITIVITY",
    "LOW_AGGREGATION_SENSITIVITY"
)

print("\nCandidate species-level aggregation summaries")
print("-" * 100)

display(
    comparison
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 7. AGGREGATION-SENSITIVE SPECIES
# -----------------------------------------------------------------------------

sensitive = (
    comparison[
        comparison["Aggregation_Sensitivity_Flag"]
        == "REVIEW_AGGREGATION_SENSITIVITY"
    ]
    .copy()
)

print("\nSpecies sensitive to aggregation method")
print("-" * 100)

if sensitive.empty:
    print("NONE")
else:
    display(
        sensitive[
            [
                "Canonical_Species",
                "N_Accepted_Measurements",
                "N_Sources",
                "Median_mg",
                "Geometric_Mean_mg",
                "Median_of_Source_Medians_mg",
                "SourceMedian_vs_PooledMedian_Ratio",
                "Aggregation_Sensitivity_Flag",
            ]
        ]
        .sort_values(
            "SourceMedian_vs_PooledMedian_Ratio"
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 8. SOURCE-LEVEL DETAIL
# -----------------------------------------------------------------------------

print("\nSource-level median structure")
print("-" * 100)

display(
    source_level
    .sort_values(
        ["Exact_Omo_Match", "Source_Key"]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 9. GLOBAL METHOD COMPARISON
# -----------------------------------------------------------------------------

method_differences = pd.DataFrame({
    "Metric": [
        "Species represented",
        "Species with one accepted measurement",
        "Species with multiple accepted measurements",
        "Species with multiple sources",
        "Species sensitive to aggregation method",
    ],
    "Value": [
        len(comparison),
        int(
            (
                comparison["N_Accepted_Measurements"] == 1
            ).sum()
        ),
        int(
            (
                comparison["N_Accepted_Measurements"] > 1
            ).sum()
        ),
        int(
            (
                comparison["N_Sources"] > 1
            ).sum()
        ),
        len(sensitive),
    ],
})

print("\nGlobal aggregation diagnostic")
print("-" * 100)

display(method_differences)

# -----------------------------------------------------------------------------
# 10. EXPORT DIAGNOSTICS
# -----------------------------------------------------------------------------

comparison_export = (
    FT_DIR /
    "omo_seed_mass_bien_species_aggregation_comparison.csv"
)

source_level_export = (
    FT_DIR /
    "omo_seed_mass_bien_source_level_summaries.csv"
)

sensitive_export = (
    FT_DIR /
    "omo_seed_mass_bien_aggregation_sensitivity_review.csv"
)

comparison.to_csv(
    comparison_export,
    index=False
)

source_level.to_csv(
    source_level_export,
    index=False
)

sensitive.to_csv(
    sensitive_export,
    index=False
)

print("\nDiagnostic exports")
print("-" * 100)

print(f"Aggregation comparison : {comparison_export}")
print(f"Source-level summaries : {source_level_export}")
print(f"Sensitivity review     : {sensitive_export}")

# -----------------------------------------------------------------------------
# 11. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(comparison) == 15
assert comparison["Canonical_Species"].is_unique

assert comparison[
    [
        "Arithmetic_Mean_mg",
        "Median_mg",
        "Geometric_Mean_mg",
        "Median_of_Source_Medians_mg",
    ]
].gt(0).all().all()

print("\nSeed Mass aggregation-method diagnostic completed successfully.")
print("No final species-level aggregation method has yet been selected.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.1K — FINAL BIEN SPECIES-LEVEL SEED MASS ESTIMATES
# =============================================================================
#
# Purpose
# -------
# Convert the 57 accepted BIEN Seed Mass measurements into finalized
# species-level direct-evidence estimates for the 15 Omo taxa represented.
#
# Final aggregation rule
# ----------------------
# Species Seed Mass estimate = median of accepted deduplicated measurements.
#
# Rationale
# ---------
# - Seed mass is often strongly right-skewed.
# - Median is robust to extreme values.
# - Scientific-signature duplication has already been removed.
# - Section 9.1J showed no species with strong aggregation sensitivity.
# - Source-balanced summaries are retained as diagnostics, not discarded.
#
# IMPORTANT
# ---------
# These are DIRECT SPECIES-LEVEL BIEN estimates only.
# No genus or family fallback occurs here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

ACCEPTED_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_accepted_measurements.csv"
)

AGGREGATION_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_species_aggregation_comparison.csv"
)

assert ACCEPTED_FILE.exists(), (
    f"Accepted measurement file not found:\n{ACCEPTED_FILE}"
)

assert AGGREGATION_FILE.exists(), (
    f"Aggregation diagnostic file not found:\n{AGGREGATION_FILE}"
)

accepted = pd.read_csv(
    ACCEPTED_FILE,
    low_memory=False
)

agg = pd.read_csv(
    AGGREGATION_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.1K — FINAL BIEN SPECIES-LEVEL SEED MASS ESTIMATES")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT STATE
# -----------------------------------------------------------------------------

assert len(accepted) == 57
assert accepted["Exact_Omo_Match"].nunique() == 15

accepted["Trait_Value_mg"] = pd.to_numeric(
    accepted["Trait_Value_mg"],
    errors="coerce"
)

assert accepted["Trait_Value_mg"].notna().all()
assert accepted["Trait_Value_mg"].gt(0).all()

assert len(agg) == 15
assert agg["Canonical_Species"].is_unique

# -----------------------------------------------------------------------------
# 2. BUILD FINAL BIEN SPECIES ESTIMATES
# -----------------------------------------------------------------------------

bien_species_final = agg.copy()

bien_species_final["Seed_Mass_mg"] = (
    bien_species_final["Median_mg"]
)

bien_species_final["Trait_Code"] = "SEED_MASS"
bien_species_final["Standardized_Trait_Unit"] = "mg"

bien_species_final["Trait_Source"] = "SPECIES_BIEN"

bien_species_final["Aggregation_Method"] = (
    "MEDIAN_OF_ACCEPTED_DEDUPLICATED_MEASUREMENTS"
)

# -----------------------------------------------------------------------------
# 3. ADD SPECIES-LEVEL QUALITY FLAGS
# -----------------------------------------------------------------------------

def assign_quality_flag(row):

    species = row["Canonical_Species"]

    if species == "Elaeis guineensis":
        return "HIGH_INTRASPECIFIC_VARIABILITY_RETAINED"

    if species == "Nauclea diderrichii":
        return "TARGETED_OUTLIER_REVIEW_COMPLETED"

    if row["N_Accepted_Measurements"] == 1:
        return "SINGLE_ACCEPTED_SPECIES_MEASUREMENT"

    if row["N_Sources"] >= 2:
        return "MULTI_SOURCE_SPECIES_EVIDENCE"

    return "MULTIPLE_ACCEPTED_SPECIES_MEASUREMENTS"


bien_species_final["Trait_Quality_Flag"] = (
    bien_species_final.apply(
        assign_quality_flag,
        axis=1
    )
)

# -----------------------------------------------------------------------------
# 4. EVIDENCE-STRENGTH CLASS
# -----------------------------------------------------------------------------

def evidence_strength(row):

    n = row["N_Accepted_Measurements"]
    s = row["N_Sources"]

    if s >= 2 and n >= 3:
        return "HIGH"

    if n >= 2:
        return "MODERATE"

    return "LIMITED"


bien_species_final["Evidence_Strength"] = (
    bien_species_final.apply(
        evidence_strength,
        axis=1
    )
)

# -----------------------------------------------------------------------------
# 5. SELECT FINAL OUTPUT COLUMNS
# -----------------------------------------------------------------------------

final_columns = [
    "Canonical_Species",
    "Trait_Code",
    "Seed_Mass_mg",
    "Standardized_Trait_Unit",
    "Trait_Source",
    "Aggregation_Method",
    "N_Accepted_Measurements",
    "N_Unique_Values",
    "N_Sources",
    "Min_mg",
    "Median_mg",
    "Geometric_Mean_mg",
    "Max_mg",
    "Median_of_Source_Medians_mg",
    "Fold_Range",
    "Trait_Quality_Flag",
    "Evidence_Strength",
]

bien_species_final = bien_species_final[
    final_columns
].copy()

bien_species_final = (
    bien_species_final
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 6. VALIDATION
# -----------------------------------------------------------------------------

assert len(bien_species_final) == 15
assert bien_species_final["Canonical_Species"].is_unique

assert (
    bien_species_final["Trait_Code"]
    == "SEED_MASS"
).all()

assert (
    bien_species_final[
        "Standardized_Trait_Unit"
    ] == "mg"
).all()

assert (
    bien_species_final["Trait_Source"]
    == "SPECIES_BIEN"
).all()

assert bien_species_final["Seed_Mass_mg"].gt(0).all()

assert np.allclose(
    bien_species_final["Seed_Mass_mg"],
    bien_species_final["Median_mg"],
)

# -----------------------------------------------------------------------------
# 7. DISPLAY FINAL BIEN SPECIES ESTIMATES
# -----------------------------------------------------------------------------

print("\nFinal BIEN species-level Seed Mass estimates")
print("-" * 100)

display(bien_species_final)

# -----------------------------------------------------------------------------
# 8. QUALITY / EVIDENCE SUMMARY
# -----------------------------------------------------------------------------

print("\nTrait quality summary")
print("-" * 100)

quality_summary = (
    bien_species_final["Trait_Quality_Flag"]
    .value_counts()
    .rename_axis("Trait_Quality_Flag")
    .reset_index(name="N_Species")
)

display(quality_summary)

print("\nEvidence-strength summary")
print("-" * 100)

strength_summary = (
    bien_species_final["Evidence_Strength"]
    .value_counts()
    .rename_axis("Evidence_Strength")
    .reset_index(name="N_Species")
)

display(strength_summary)

# -----------------------------------------------------------------------------
# 9. NUMERIC SUMMARY
# -----------------------------------------------------------------------------

print("\nFinal BIEN Seed Mass numeric summary")
print("-" * 100)

print(
    bien_species_final[
        "Seed_Mass_mg"
    ].describe().to_string()
)

# -----------------------------------------------------------------------------
# 10. EXPORT FINAL DIRECT BIEN SPECIES EVIDENCE
# -----------------------------------------------------------------------------

final_export = (
    FT_DIR /
    "omo_seed_mass_bien_final_species_values.csv"
)

quality_export = (
    FT_DIR /
    "omo_seed_mass_bien_final_quality_summary.csv"
)

bien_species_final.to_csv(
    final_export,
    index=False
)

quality_summary.to_csv(
    quality_export,
    index=False
)

print("\nFinal BIEN exports")
print("-" * 100)

print(f"Species values  : {final_export}")
print(f"Quality summary : {quality_export}")

# -----------------------------------------------------------------------------
# 11. FINAL STATUS
# -----------------------------------------------------------------------------

print("\nBIEN species-level Seed Mass consolidation completed successfully.")
print(f"Direct BIEN species coverage : {len(bien_species_final)} / 127")
print(
    f"Coverage (%)                 : "
    f"{100 * len(bien_species_final) / 127:.2f}"
)

print(
    "Remaining taxa require additional species evidence or "
    "explicit fallback acquisition."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2A — RESIDUAL SEED MASS TARGETS AND VALIDATED TAXONOMIC QUERY NAMES
# =============================================================================
#
# Purpose
# -------
# Build the residual Seed Mass acquisition manifest after direct BIEN
# species-level consolidation and recover validated alternative/normalized
# taxonomic names already established during Notebook 07B WD/SLA work.
#
# IMPORTANT
# ---------
# Canonical_Species remains the project identifier.
# Alternative names are acquisition/query names only.
#
# No external database query occurs in this cell.
# No genus/family fallback occurs.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

SEED_MANIFEST_FILE = (
    FT_DIR / "omo_seed_mass_acquisition_manifest_initial.csv"
)

BIEN_FINAL_FILE = (
    FT_DIR / "omo_seed_mass_bien_final_species_values.csv"
)

assert SEED_MANIFEST_FILE.exists()
assert BIEN_FINAL_FILE.exists()

manifest = pd.read_csv(
    SEED_MANIFEST_FILE,
    low_memory=False
)

bien_final = pd.read_csv(
    BIEN_FINAL_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2A — RESIDUAL SEED MASS TARGETS AND VALIDATED TAXONOMIC QUERY NAMES")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE CURRENT COVERAGE
# -----------------------------------------------------------------------------

assert len(manifest) == 127
assert manifest["Canonical_Species"].is_unique

assert len(bien_final) == 15
assert bien_final["Canonical_Species"].is_unique

covered_species = set(
    bien_final["Canonical_Species"]
    .astype(str)
    .str.strip()
)

residual = manifest[
    ~manifest["Canonical_Species"]
    .astype(str)
    .str.strip()
    .isin(covered_species)
].copy()

residual = residual.reset_index(drop=True)

print("\nCurrent Seed Mass coverage")
print("-" * 100)

print(f"Canonical taxa                : {len(manifest)}")
print(f"Direct BIEN species values    : {len(bien_final)}")
print(f"Residual taxa                 : {len(residual)}")

assert len(residual) == 112

# -----------------------------------------------------------------------------
# 2. LOCATE EXISTING TAXONOMIC-HARMONIZATION TABLES
# -----------------------------------------------------------------------------
#
# We search existing WD/SLA outputs for tables containing Canonical_Species
# plus likely accepted/normalized/evidence-name fields.
# -----------------------------------------------------------------------------

candidate_files = []

for path in FT_DIR.glob("*.csv"):

    name = path.name.lower()

    if not (
        name.startswith("omo_sla")
        or name.startswith("omo_wood")
        or name.startswith("wd_")
    ):
        continue

    try:
        preview = pd.read_csv(
            path,
            nrows=5,
            low_memory=False
        )
    except Exception:
        continue

    if "Canonical_Species" not in preview.columns:
        continue

    candidate_files.append(path)

print("\nExisting trait tables considered for taxonomic aliases")
print("-" * 100)

print(f"N candidate files : {len(candidate_files)}")

# -----------------------------------------------------------------------------
# 3. DEFINE POSSIBLE TAXONOMIC ALIAS FIELDS
# -----------------------------------------------------------------------------

alias_keywords = [
    "matched",
    "normalized",
    "accepted",
    "resolved",
    "harmonized",
    "source_species",
    "trait_source_name",
    "evidence_target",
    "query_name",
    "replacement",
    "corrected",
    "taxon",
]

excluded_alias_fields = {
    "Canonical_Species",
    "Taxonomic_Resolution",
}

# -----------------------------------------------------------------------------
# 4. HARVEST ALTERNATIVE NAMES
# -----------------------------------------------------------------------------

alias_records = []

residual_set = set(
    residual["Canonical_Species"]
    .astype(str)
    .str.strip()
)

for path in candidate_files:

    try:
        df = pd.read_csv(
            path,
            low_memory=False
        )
    except Exception:
        continue

    if "Canonical_Species" not in df.columns:
        continue

    df["Canonical_Species"] = (
        df["Canonical_Species"]
        .astype("string")
        .str.strip()
    )

    df = df[
        df["Canonical_Species"].isin(residual_set)
    ].copy()

    if df.empty:
        continue

    possible_alias_cols = []

    for col in df.columns:

        if col in excluded_alias_fields:
            continue

        col_lower = str(col).lower()

        if any(
            keyword in col_lower
            for keyword in alias_keywords
        ):
            possible_alias_cols.append(col)

    for _, row in df.iterrows():

        canonical = row["Canonical_Species"]

        for col in possible_alias_cols:

            value = row.get(col)

            if pd.isna(value):
                continue

            value = str(value).strip()

            if not value:
                continue

            # Basic scientific-name-like filter:
            # retain only short textual strings, not status codes,
            # provenance labels, URLs or numeric values.
            parts = value.split()

            if not (1 <= len(parts) <= 4):
                continue

            if value.upper() == value and "_" in value:
                continue

            if value.startswith("http"):
                continue

            if any(char.isdigit() for char in value):
                continue

            # Require first token to look like a genus name.
            first = parts[0]

            if not (
                first
                and first[0].isupper()
                and first.replace("-", "").isalpha()
            ):
                continue

            alias_records.append({
                "Canonical_Species": canonical,
                "Alternative_Query_Name": value,
                "Alias_Field": col,
                "Alias_Source_File": path.name,
            })

aliases = pd.DataFrame(alias_records)

# -----------------------------------------------------------------------------
# 5. CLEAN / DEDUPLICATE ALIASES
# -----------------------------------------------------------------------------

if not aliases.empty:

    aliases["Canonical_Species"] = (
        aliases["Canonical_Species"]
        .astype(str)
        .str.strip()
    )

    aliases["Alternative_Query_Name"] = (
        aliases["Alternative_Query_Name"]
        .astype(str)
        .str.strip()
    )

    # Remove trivial self-aliases.
    aliases = aliases[
        aliases["Alternative_Query_Name"]
        != aliases["Canonical_Species"]
    ].copy()

    aliases = (
        aliases
        .drop_duplicates(
            subset=[
                "Canonical_Species",
                "Alternative_Query_Name",
            ]
        )
        .sort_values(
            [
                "Canonical_Species",
                "Alternative_Query_Name",
            ]
        )
        .reset_index(drop=True)
    )

print("\nRecovered validated alternative taxonomic names")
print("-" * 100)

if aliases.empty:

    print("No alternative names recovered.")

else:

    print(f"Alias relationships : {len(aliases)}")
    print(
        f"Residual taxa with >=1 alternative name : "
        f"{aliases['Canonical_Species'].nunique()}"
    )

    display(aliases)

# -----------------------------------------------------------------------------
# 6. BUILD RESIDUAL ACQUISITION MANIFEST
# -----------------------------------------------------------------------------

residual_manifest = residual[
    [
        "Canonical_Species",
        "Taxonomic_Resolution",
        "Preferred_Evidence_Level",
    ]
].copy()

residual_manifest["Trait_Code"] = "SEED_MASS"
residual_manifest["Standardized_Unit"] = "mg"

residual_manifest["Primary_Query_Name"] = (
    residual_manifest["Canonical_Species"]
)

# Collapse alternative names into a pipe-separated field.
if not aliases.empty:

    collapsed_aliases = (
        aliases.groupby("Canonical_Species")[
            "Alternative_Query_Name"
        ]
        .apply(
            lambda x: " | ".join(
                sorted(set(x))
            )
        )
        .reset_index(
            name="Validated_Alternative_Query_Names"
        )
    )

    residual_manifest = residual_manifest.merge(
        collapsed_aliases,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )

else:

    residual_manifest[
        "Validated_Alternative_Query_Names"
    ] = pd.NA

residual_manifest["Has_Validated_Alternative_Name"] = (
    residual_manifest[
        "Validated_Alternative_Query_Names"
    ].notna()
)

residual_manifest["Next_Acquisition_Status"] = np.where(
    residual_manifest["Taxonomic_Resolution"]
    .eq("GENUS_LEVEL_OR_UNRESOLVED"),
    "DEFER_TO_GENUS_STAGE",
    np.where(
        residual_manifest[
            "Has_Validated_Alternative_Name"
        ],
        "REQUERY_SPECIES_WITH_VALIDATED_ALIAS",
        "SPECIES_RESCUE_REQUIRED"
    )
)

# -----------------------------------------------------------------------------
# 7. SUMMARIZE RESIDUAL TARGETS
# -----------------------------------------------------------------------------

print("\nResidual Seed Mass acquisition status")
print("-" * 100)

status_summary = (
    residual_manifest[
        "Next_Acquisition_Status"
    ]
    .value_counts()
    .rename_axis(
        "Next_Acquisition_Status"
    )
    .reset_index(name="N_Taxa")
)

display(status_summary)

print("\nResidual taxa with validated alternative query names")
print("-" * 100)

alias_targets = residual_manifest[
    residual_manifest[
        "Has_Validated_Alternative_Name"
    ]
].copy()

if alias_targets.empty:
    print("NONE")
else:
    display(
        alias_targets[
            [
                "Canonical_Species",
                "Validated_Alternative_Query_Names",
                "Next_Acquisition_Status",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 8. VALIDATE GENUS/UNRESOLVED TAXA REMAIN SEPARATE
# -----------------------------------------------------------------------------

unresolved_targets = residual_manifest[
    residual_manifest["Taxonomic_Resolution"]
    .eq("GENUS_LEVEL_OR_UNRESOLVED")
]

print("\nGenus/unresolved residual taxa")
print("-" * 100)

display(
    unresolved_targets[
        [
            "Canonical_Species",
            "Taxonomic_Resolution",
            "Next_Acquisition_Status",
        ]
    ]
)

assert set(
    unresolved_targets["Canonical_Species"]
) == {
    "Cola species",
    "Drypetes species",
}

# -----------------------------------------------------------------------------
# 9. EXPORT
# -----------------------------------------------------------------------------

alias_export = (
    FT_DIR /
    "omo_seed_mass_validated_taxonomic_aliases.csv"
)

residual_export = (
    FT_DIR /
    "omo_seed_mass_residual_species_acquisition_manifest.csv"
)

aliases.to_csv(
    alias_export,
    index=False
)

residual_manifest.to_csv(
    residual_export,
    index=False
)

print("\nResidual acquisition exports")
print("-" * 100)

print(f"Validated aliases : {alias_export}")
print(f"Residual manifest : {residual_export}")

# -----------------------------------------------------------------------------
# 10. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(residual_manifest) == 112
assert residual_manifest["Canonical_Species"].is_unique

assert (
    set(residual_manifest["Canonical_Species"])
    .isdisjoint(covered_species)
)

print("\nResidual Seed Mass taxonomic rescue manifest created successfully.")
print("No new external evidence has yet been acquired.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2B — SANITIZE AND PRIORITIZE VALIDATED SPECIES-LEVEL RESCUE ALIASES
# =============================================================================
#
# Purpose
# -------
# Clean the broad alias harvest from Section 9.2A so that only scientifically
# plausible species-level alternative names are retained for Seed Mass rescue.
#
# Specifically exclude:
#   - status labels such as SPECIES / GENUS;
#   - Boolean strings such as True / False;
#   - genus-only names;
#   - URLs / provenance labels;
#   - trivial aliases identical to Canonical_Species.
#
# Retain:
#   - binomials;
#   - hybrid expressions such as "Musa x paradisiaca";
#   - validated accepted/replacement species names from prior WD/SLA work.
#
# No external database query occurs here.
# =============================================================================

from pathlib import Path
import pandas as pd
import re

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

ALIAS_FILE = (
    FT_DIR / "omo_seed_mass_validated_taxonomic_aliases.csv"
)

RESIDUAL_FILE = (
    FT_DIR / "omo_seed_mass_residual_species_acquisition_manifest.csv"
)

assert ALIAS_FILE.exists()
assert RESIDUAL_FILE.exists()

aliases = pd.read_csv(
    ALIAS_FILE,
    low_memory=False
)

residual = pd.read_csv(
    RESIDUAL_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2B — SANITIZE AND PRIORITIZE VALIDATED SPECIES-LEVEL RESCUE ALIASES")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT
# -----------------------------------------------------------------------------

required_alias_cols = {
    "Canonical_Species",
    "Alternative_Query_Name",
    "Alias_Field",
    "Alias_Source_File",
}

missing = required_alias_cols - set(aliases.columns)

assert not missing, (
    f"Alias table missing required fields: {sorted(missing)}"
)

assert len(residual) == 112
assert residual["Canonical_Species"].is_unique

# -----------------------------------------------------------------------------
# 2. NORMALIZE TEXT
# -----------------------------------------------------------------------------

for col in [
    "Canonical_Species",
    "Alternative_Query_Name",
    "Alias_Field",
    "Alias_Source_File",
]:
    aliases[col] = (
        aliases[col]
        .astype("string")
        .str.strip()
    )

# -----------------------------------------------------------------------------
# 3. DEFINE NON-TAXONOMIC TOKENS
# -----------------------------------------------------------------------------

invalid_exact_tokens = {
    "",
    "SPECIES",
    "GENUS",
    "FAMILY",
    "TRUE",
    "FALSE",
    "NONE",
    "NAN",
    "NA",
    "UNKNOWN",
    "UNRESOLVED",
}

# -----------------------------------------------------------------------------
# 4. SCIENTIFIC-NAME FILTER
# -----------------------------------------------------------------------------

def looks_like_species_name(value):
    """
    Accept ordinary binomials and simple hybrid forms such as:
        Musa x paradisiaca
    Reject genus-only strings and status/provenance tokens.
    """

    if pd.isna(value):
        return False

    value = str(value).strip()

    if not value:
        return False

    if value.upper() in invalid_exact_tokens:
        return False

    if value.lower().startswith("http"):
        return False

    if any(ch.isdigit() for ch in value):
        return False

    parts = value.split()

    # Ordinary binomial:
    # Genus species
    if len(parts) == 2:
        genus, epithet = parts

        if not (
            genus
            and genus[0].isupper()
            and genus.replace("-", "").isalpha()
        ):
            return False

        if not (
            epithet
            and epithet[0].islower()
            and epithet.replace("-", "").isalpha()
        ):
            return False

        return True

    # Simple hybrid:
    # Genus x species
    if len(parts) == 3:
        genus, hybrid_marker, epithet = parts

        if hybrid_marker.lower() not in {"x", "×"}:
            return False

        if not (
            genus
            and genus[0].isupper()
            and genus.replace("-", "").isalpha()
        ):
            return False

        if not (
            epithet
            and epithet[0].islower()
            and epithet.replace("-", "").isalpha()
        ):
            return False

        return True

    return False


aliases["Looks_Like_Species_Name"] = (
    aliases["Alternative_Query_Name"]
    .apply(looks_like_species_name)
)

# -----------------------------------------------------------------------------
# 5. REMOVE SELF-ALIASES
# -----------------------------------------------------------------------------

aliases["Is_Self_Alias"] = (
    aliases["Alternative_Query_Name"]
    .eq(aliases["Canonical_Species"])
)

clean_aliases = aliases[
    aliases["Looks_Like_Species_Name"]
    &
    ~aliases["Is_Self_Alias"]
].copy()

# -----------------------------------------------------------------------------
# 6. PRIORITIZE ALIAS FIELD TYPES
# -----------------------------------------------------------------------------
#
# Strongest fields are those explicitly indicating normalized / accepted /
# corrected / resolved taxonomic names.
# -----------------------------------------------------------------------------

def alias_priority(field):

    field = str(field).lower()

    if "recovered_normalized_name" in field:
        return 1

    if "accepted" in field:
        return 2

    if "resolved" in field:
        return 3

    if "normalized" in field:
        return 4

    if "corrected" in field:
        return 5

    if "matched" in field:
        return 6

    if "trait_source_name" in field:
        return 7

    return 9


clean_aliases["Alias_Priority"] = (
    clean_aliases["Alias_Field"]
    .apply(alias_priority)
)

# -----------------------------------------------------------------------------
# 7. DEDUPLICATE SAME CANONICAL → ALIAS RELATIONSHIP
# -----------------------------------------------------------------------------

clean_aliases = (
    clean_aliases
    .sort_values(
        [
            "Canonical_Species",
            "Alternative_Query_Name",
            "Alias_Priority",
        ]
    )
    .drop_duplicates(
        subset=[
            "Canonical_Species",
            "Alternative_Query_Name",
        ],
        keep="first"
    )
    .reset_index(drop=True)
)

print("\nAlias sanitation summary")
print("-" * 100)

print(f"Raw harvested alias relationships      : {len(aliases)}")
print(f"Valid species-level alias relationships : {len(clean_aliases)}")
print(
    f"Residual taxa with >=1 valid alias      : "
    f"{clean_aliases['Canonical_Species'].nunique()}"
)

# -----------------------------------------------------------------------------
# 8. BUILD ONE PRIORITIZED QUERY NAME PER CANONICAL TAXON
# -----------------------------------------------------------------------------

best_alias = (
    clean_aliases
    .sort_values(
        [
            "Canonical_Species",
            "Alias_Priority",
            "Alternative_Query_Name",
        ]
    )
    .groupby(
        "Canonical_Species",
        as_index=False
    )
    .first()
)

best_alias = best_alias[
    [
        "Canonical_Species",
        "Alternative_Query_Name",
        "Alias_Field",
        "Alias_Source_File",
        "Alias_Priority",
    ]
].rename(
    columns={
        "Alternative_Query_Name":
            "Preferred_Rescue_Query_Name",
        "Alias_Field":
            "Preferred_Alias_Field",
        "Alias_Source_File":
            "Preferred_Alias_Source_File",
    }
)

# -----------------------------------------------------------------------------
# 9. MERGE WITH RESIDUAL MANIFEST
# -----------------------------------------------------------------------------

rescue_manifest = residual.merge(
    best_alias,
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)

rescue_manifest["Has_Usable_Species_Alias"] = (
    rescue_manifest[
        "Preferred_Rescue_Query_Name"
    ].notna()
)

# Genus/unresolved taxa remain excluded from species rescue.
rescue_manifest["Species_Rescue_Status"] = "NO_VALIDATED_SPECIES_ALIAS"

rescue_manifest.loc[
    rescue_manifest[
        "Taxonomic_Resolution"
    ].eq("GENUS_LEVEL_OR_UNRESOLVED"),
    "Species_Rescue_Status"
] = "DEFER_TO_GENUS_STAGE"

rescue_manifest.loc[
    rescue_manifest[
        "Taxonomic_Resolution"
    ].eq("SPECIES_LEVEL")
    &
    rescue_manifest[
        "Has_Usable_Species_Alias"
    ],
    "Species_Rescue_Status"
] = "READY_FOR_BIEN_ALIAS_REQUERY"

# -----------------------------------------------------------------------------
# 10. SUMMARY
# -----------------------------------------------------------------------------

status_summary = (
    rescue_manifest[
        "Species_Rescue_Status"
    ]
    .value_counts()
    .rename_axis("Species_Rescue_Status")
    .reset_index(name="N_Taxa")
)

print("\nSanitized species-rescue status")
print("-" * 100)

display(status_summary)

# -----------------------------------------------------------------------------
# 11. DISPLAY VALIDATED RESCUE NAMES
# -----------------------------------------------------------------------------

ready = rescue_manifest[
    rescue_manifest[
        "Species_Rescue_Status"
    ].eq("READY_FOR_BIEN_ALIAS_REQUERY")
].copy()

print("\nValidated species aliases ready for BIEN re-query")
print("-" * 100)

print(f"N taxa ready : {len(ready)}")

if ready.empty:
    print("NONE")
else:
    display(
        ready[
            [
                "Canonical_Species",
                "Preferred_Rescue_Query_Name",
                "Preferred_Alias_Field",
                "Preferred_Alias_Source_File",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 12. TAXA STILL WITHOUT A VALID SPECIES ALIAS
# -----------------------------------------------------------------------------

still_unresolved = rescue_manifest[
    rescue_manifest[
        "Species_Rescue_Status"
    ].eq("NO_VALIDATED_SPECIES_ALIAS")
].copy()

print("\nSpecies-level residual taxa without validated rescue alias")
print("-" * 100)

print(f"N taxa : {len(still_unresolved)}")

if still_unresolved.empty:
    print("NONE")
else:
    display(
        still_unresolved[
            [
                "Canonical_Species",
                "Taxonomic_Resolution",
                "Species_Rescue_Status",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 13. EXPORT
# -----------------------------------------------------------------------------

clean_alias_export = (
    FT_DIR /
    "omo_seed_mass_validated_species_aliases_clean.csv"
)

rescue_manifest_export = (
    FT_DIR /
    "omo_seed_mass_species_alias_rescue_manifest.csv"
)

ready_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_requery_targets.csv"
)

clean_aliases.to_csv(
    clean_alias_export,
    index=False
)

rescue_manifest.to_csv(
    rescue_manifest_export,
    index=False
)

ready.to_csv(
    ready_export,
    index=False
)

print("\nSanitized alias exports")
print("-" * 100)

print(f"Clean aliases        : {clean_alias_export}")
print(f"Rescue manifest      : {rescue_manifest_export}")
print(f"BIEN re-query targets: {ready_export}")

# -----------------------------------------------------------------------------
# 14. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(rescue_manifest) == 112
assert rescue_manifest["Canonical_Species"].is_unique

assert set(
    rescue_manifest.loc[
        rescue_manifest[
            "Species_Rescue_Status"
        ].eq("DEFER_TO_GENUS_STAGE"),
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Drypetes species",
}

assert (
    ready["Preferred_Rescue_Query_Name"]
    .apply(looks_like_species_name)
    .all()
)

print("\nValidated species-level taxonomic rescue aliases sanitized successfully.")
print("No external Seed Mass query has yet been performed.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2C — EXCLUDE CONGENERIC PSEUDO-ALIASES AND FINALIZE
#                BIEN SPECIES RE-QUERY TARGETS
# =============================================================================
#
# Purpose
# -------
# Refine Section 9.2B by removing alternative names derived from congeneric
# fallback evidence rather than genuine taxonomic normalization.
#
# Scientific rule
# ---------------
# A species may be re-queried under an alternative name ONLY when that name
# represents a validated taxonomic normalization, spelling correction,
# synonym, or accepted-name mapping.
#
# Congeneric source species used previously for SLA estimation are NOT
# species aliases and must not be used as direct Seed Mass species evidence.
#
# No external query occurs in this cell.
# =============================================================================

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

CLEAN_ALIAS_FILE = (
    FT_DIR / "omo_seed_mass_validated_species_aliases_clean.csv"
)

RESCUE_MANIFEST_FILE = (
    FT_DIR / "omo_seed_mass_species_alias_rescue_manifest.csv"
)

assert CLEAN_ALIAS_FILE.exists()
assert RESCUE_MANIFEST_FILE.exists()

aliases = pd.read_csv(CLEAN_ALIAS_FILE, low_memory=False)
residual = pd.read_csv(RESCUE_MANIFEST_FILE, low_memory=False)

print("=" * 100)
print("SECTION 9.2C — FINALIZE TRUE TAXONOMIC ALIASES FOR BIEN SPECIES RE-QUERY")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT
# -----------------------------------------------------------------------------

assert len(residual) == 112
assert residual["Canonical_Species"].is_unique

required = {
    "Canonical_Species",
    "Alternative_Query_Name",
    "Alias_Field",
    "Alias_Source_File",
}

missing = required - set(aliases.columns)

assert not missing, (
    f"Clean alias table missing fields: {sorted(missing)}"
)

# -----------------------------------------------------------------------------
# 2. IDENTIFY CONGENERIC PSEUDO-ALIASES
# -----------------------------------------------------------------------------

aliases["Alias_Field"] = (
    aliases["Alias_Field"]
    .astype("string")
    .str.strip()
)

aliases["Alias_Source_File"] = (
    aliases["Alias_Source_File"]
    .astype("string")
    .str.strip()
)

congeneric_mask = (
    aliases["Alias_Field"]
    .str.contains(
        "congeneric",
        case=False,
        na=False
    )
    |
    aliases["Alias_Source_File"]
    .str.contains(
        "congeneric",
        case=False,
        na=False
    )
)

congeneric_excluded = aliases[
    congeneric_mask
].copy()

true_aliases = aliases[
    ~congeneric_mask
].copy()

print("\nCongeneric pseudo-aliases excluded")
print("-" * 100)

print(f"N relationships excluded : {len(congeneric_excluded)}")
print(
    f"N canonical taxa affected: "
    f"{congeneric_excluded['Canonical_Species'].nunique()}"
)

if congeneric_excluded.empty:
    print("NONE")
else:
    display(
        congeneric_excluded[
            [
                "Canonical_Species",
                "Alternative_Query_Name",
                "Alias_Field",
                "Alias_Source_File",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 3. REMOVE ANY REMAINING SELF-ALIASES
# -----------------------------------------------------------------------------

true_aliases = true_aliases[
    true_aliases["Alternative_Query_Name"]
    != true_aliases["Canonical_Species"]
].copy()

true_aliases = (
    true_aliases
    .drop_duplicates(
        subset=[
            "Canonical_Species",
            "Alternative_Query_Name",
        ]
    )
    .sort_values(
        [
            "Canonical_Species",
            "Alias_Priority",
            "Alternative_Query_Name",
        ]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 4. SELECT BEST TRUE TAXONOMIC ALIAS PER CANONICAL TAXON
# -----------------------------------------------------------------------------

best_true_alias = (
    true_aliases
    .sort_values(
        [
            "Canonical_Species",
            "Alias_Priority",
            "Alternative_Query_Name",
        ]
    )
    .groupby(
        "Canonical_Species",
        as_index=False
    )
    .first()
)

best_true_alias = best_true_alias[
    [
        "Canonical_Species",
        "Alternative_Query_Name",
        "Alias_Field",
        "Alias_Source_File",
        "Alias_Priority",
    ]
].rename(
    columns={
        "Alternative_Query_Name":
            "Validated_BIEN_Query_Name",
        "Alias_Field":
            "Validated_Alias_Field",
        "Alias_Source_File":
            "Validated_Alias_Source_File",
    }
)

# -----------------------------------------------------------------------------
# 5. REBUILD RESIDUAL MANIFEST FROM CLEAN STATE
# -----------------------------------------------------------------------------

base_cols = [
    "Canonical_Species",
    "Taxonomic_Resolution",
]

for optional_col in [
    "Trait_Code",
    "Standardized_Unit",
]:
    if optional_col in residual.columns:
        base_cols.append(optional_col)

final_manifest = residual[
    base_cols
].copy()

final_manifest = final_manifest.merge(
    best_true_alias,
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)

final_manifest["Has_True_Taxonomic_Alias"] = (
    final_manifest[
        "Validated_BIEN_Query_Name"
    ].notna()
)

# -----------------------------------------------------------------------------
# 6. ASSIGN FINAL RE-QUERY STATUS
# -----------------------------------------------------------------------------

final_manifest["BIEN_Requery_Status"] = (
    "NO_ALTERNATIVE_SPECIES_NAME"
)

final_manifest.loc[
    final_manifest[
        "Taxonomic_Resolution"
    ].eq("GENUS_LEVEL_OR_UNRESOLVED"),
    "BIEN_Requery_Status"
] = "DEFER_TO_GENUS_STAGE"

final_manifest.loc[
    final_manifest[
        "Taxonomic_Resolution"
    ].eq("SPECIES_LEVEL")
    &
    final_manifest[
        "Has_True_Taxonomic_Alias"
    ],
    "BIEN_Requery_Status"
] = "READY_FOR_BIEN_SPECIES_REQUERY"

# -----------------------------------------------------------------------------
# 7. SUMMARY
# -----------------------------------------------------------------------------

print("\nFinal BIEN species re-query status")
print("-" * 100)

status_summary = (
    final_manifest[
        "BIEN_Requery_Status"
    ]
    .value_counts()
    .rename_axis("BIEN_Requery_Status")
    .reset_index(name="N_Taxa")
)

display(status_summary)

# -----------------------------------------------------------------------------
# 8. FINAL RE-QUERY TARGETS
# -----------------------------------------------------------------------------

requery_targets = final_manifest[
    final_manifest[
        "BIEN_Requery_Status"
    ].eq("READY_FOR_BIEN_SPECIES_REQUERY")
].copy()

print("\nTrue taxonomic aliases ready for BIEN Seed Mass re-query")
print("-" * 100)

print(f"N taxa ready : {len(requery_targets)}")

if requery_targets.empty:
    print("NONE")
else:
    display(
        requery_targets[
            [
                "Canonical_Species",
                "Validated_BIEN_Query_Name",
                "Validated_Alias_Field",
                "Validated_Alias_Source_File",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 9. TAXA WITHOUT ALTERNATIVE SPECIES NAME
# -----------------------------------------------------------------------------

no_alias = final_manifest[
    final_manifest[
        "BIEN_Requery_Status"
    ].eq("NO_ALTERNATIVE_SPECIES_NAME")
].copy()

print("\nSpecies taxa without an alternative taxonomic query name")
print("-" * 100)

print(f"N taxa : {len(no_alias)}")

# These are NOT failures.
# Their canonical names were already queried in Section 9.1.
# They simply have no additional validated synonym/correction to try.

if no_alias.empty:
    print("NONE")
else:
    display(
        no_alias[
            [
                "Canonical_Species",
                "Taxonomic_Resolution",
                "BIEN_Requery_Status",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 10. SCIENTIFIC VALIDATION
# -----------------------------------------------------------------------------

# Congeneric evidence must not survive as a species re-query alias.
assert not (
    requery_targets[
        "Validated_Alias_Field"
    ]
    .astype(str)
    .str.contains(
        "congeneric",
        case=False,
        na=False
    )
).any()

assert not (
    requery_targets[
        "Validated_Alias_Source_File"
    ]
    .astype(str)
    .str.contains(
        "congeneric",
        case=False,
        na=False
    )
).any()

assert set(
    final_manifest.loc[
        final_manifest[
            "BIEN_Requery_Status"
        ].eq("DEFER_TO_GENUS_STAGE"),
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Drypetes species",
}

assert (
    len(requery_targets)
    + len(no_alias)
    + 2
    == 112
)

# -----------------------------------------------------------------------------
# 11. EXPORT
# -----------------------------------------------------------------------------

true_alias_export = (
    FT_DIR /
    "omo_seed_mass_true_taxonomic_species_aliases.csv"
)

excluded_export = (
    FT_DIR /
    "omo_seed_mass_congeneric_pseudo_aliases_excluded.csv"
)

final_manifest_export = (
    FT_DIR /
    "omo_seed_mass_final_species_requery_manifest.csv"
)

requery_export = (
    FT_DIR /
    "omo_seed_mass_bien_species_requery_targets_final.csv"
)

true_aliases.to_csv(
    true_alias_export,
    index=False
)

congeneric_excluded.to_csv(
    excluded_export,
    index=False
)

final_manifest.to_csv(
    final_manifest_export,
    index=False
)

requery_targets.to_csv(
    requery_export,
    index=False
)

print("\nFinal taxonomic rescue exports")
print("-" * 100)

print(f"True species aliases     : {true_alias_export}")
print(f"Excluded pseudo-aliases  : {excluded_export}")
print(f"Final residual manifest  : {final_manifest_export}")
print(f"BIEN re-query targets    : {requery_export}")

print("\nSpecies-level BIEN re-query manifest finalized successfully.")
print("Congeneric fallback evidence has been explicitly excluded.")
print("No external Seed Mass query has yet been performed.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2E — BIEN ALIAS-RESCUE CROSSWALK AND EXISTING-EVIDENCE OVERLAP AUDIT
# =============================================================================
#
# Purpose
# -------
# Audit the raw BIEN Seed Mass evidence recovered through validated taxonomic
# aliases in Section 9.2D.
#
# This cell determines:
#   1. whether returned BIEN taxonomy is consistent with the alias queried;
#   2. which Omo canonical taxa already possess accepted BIEN evidence;
#   3. which taxa represent genuine new species-level coverage;
#   4. whether alias-requery rows duplicate measurements already accepted
#      during Section 9.1;
#   5. which records should proceed to measurement-level QC.
#
# IMPORTANT:
# No new Seed Mass measurement is accepted in this cell.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

ALIAS_RAW_FILE = (
    FT_DIR / "omo_seed_mass_bien_alias_requery_raw.csv"
)

ALIAS_STATUS_FILE = (
    FT_DIR / "omo_seed_mass_bien_alias_requery_status.csv"
)

EXISTING_ACCEPTED_FILE = (
    FT_DIR / "omo_seed_mass_bien_accepted_measurements.csv"
)

EXISTING_FINAL_FILE = (
    FT_DIR / "omo_seed_mass_bien_final_species_values.csv"
)

assert ALIAS_RAW_FILE.exists()
assert ALIAS_STATUS_FILE.exists()
assert EXISTING_ACCEPTED_FILE.exists()
assert EXISTING_FINAL_FILE.exists()

alias_raw = pd.read_csv(
    ALIAS_RAW_FILE,
    low_memory=False
)

alias_status = pd.read_csv(
    ALIAS_STATUS_FILE,
    low_memory=False
)

existing = pd.read_csv(
    EXISTING_ACCEPTED_FILE,
    low_memory=False
)

existing_final = pd.read_csv(
    EXISTING_FINAL_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2E — BIEN ALIAS-RESCUE CROSSWALK AND EXISTING-EVIDENCE OVERLAP AUDIT")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. BASIC INPUT VALIDATION
# -----------------------------------------------------------------------------

print("\nInput tables")
print("-" * 100)

print(f"Alias raw evidence        : {len(alias_raw)} rows")
print(f"Alias canonical status    : {len(alias_status)} taxa")
print(f"Existing accepted BIEN    : {len(existing)} measurements")
print(f"Existing final BIEN values: {len(existing_final)} species")

assert len(alias_raw) == 61
assert len(alias_status) == 54
assert len(existing_final) == 15

required_alias_cols = {
    "Omo_Canonical_Species",
    "Validated_BIEN_Query_Name",
    "trait_name",
    "trait_value",
    "unit",
}

missing = required_alias_cols - set(alias_raw.columns)

assert not missing, (
    f"Alias raw table missing required fields: {sorted(missing)}"
)

# -----------------------------------------------------------------------------
# 2. NORMALIZE IMPORTANT TEXT FIELDS
# -----------------------------------------------------------------------------

text_cols = [
    "Omo_Canonical_Species",
    "Validated_BIEN_Query_Name",
    "trait_name",
    "unit",
    "name_submitted",
    "verbatim_scientific_name",
    "name_matched",
    "scrubbed_species_binomial",
    "scrubbed_taxon_canonical",
]

for col in text_cols:
    if col in alias_raw.columns:
        alias_raw[col] = (
            alias_raw[col]
            .astype("string")
            .str.strip()
        )

# Numeric Seed Mass
alias_raw["Trait_Value_mg"] = pd.to_numeric(
    alias_raw["trait_value"],
    errors="coerce"
)

assert alias_raw["Trait_Value_mg"].notna().all()
assert (alias_raw["Trait_Value_mg"] > 0).all()

assert (
    alias_raw["trait_name"]
    .str.lower()
    .eq("seed mass")
    .all()
)

assert (
    alias_raw["unit"]
    .str.lower()
    .eq("mg")
    .all()
)

# -----------------------------------------------------------------------------
# 3. TAXONOMIC CONSISTENCY OF RETURNED BIEN RECORDS
# -----------------------------------------------------------------------------

taxonomic_fields = [
    c for c in [
        "name_submitted",
        "verbatim_scientific_name",
        "name_matched",
        "scrubbed_species_binomial",
        "scrubbed_taxon_canonical",
    ]
    if c in alias_raw.columns
]

def normalize_taxon(x):
    if pd.isna(x):
        return pd.NA
    return " ".join(str(x).strip().lower().split())

query_norm = (
    alias_raw["Validated_BIEN_Query_Name"]
    .map(normalize_taxon)
)

tax_match_matrix = pd.DataFrame(
    {
        field: (
            alias_raw[field]
            .map(normalize_taxon)
            .eq(query_norm)
        )
        for field in taxonomic_fields
    }
)

if taxonomic_fields:
    alias_raw["Query_Name_Matched_By_BIEN"] = (
        tax_match_matrix.any(axis=1)
    )
else:
    alias_raw["Query_Name_Matched_By_BIEN"] = False

print("\nReturned-taxonomy consistency")
print("-" * 100)

print(
    alias_raw["Query_Name_Matched_By_BIEN"]
    .value_counts(dropna=False)
    .rename_axis("Query_Name_Matched_By_BIEN")
    .reset_index(name="N_Raw_Records")
)

# Any mismatch is retained for explicit review, not silently discarded.
taxonomy_review = alias_raw[
    ~alias_raw["Query_Name_Matched_By_BIEN"]
].copy()

print("\nRecords requiring returned-taxonomy review")
print("-" * 100)

print(f"N records : {len(taxonomy_review)}")

if not taxonomy_review.empty:
    display(
        taxonomy_review[
            [
                "Omo_Canonical_Species",
                "Validated_BIEN_Query_Name",
                *taxonomic_fields,
                "Trait_Value_mg",
            ]
        ].reset_index(drop=True)
    )
else:
    print("NONE")

# -----------------------------------------------------------------------------
# 4. IDENTIFY TAXA ALREADY COVERED BY SECTION 9.1
# -----------------------------------------------------------------------------

existing_species = set(
    existing_final["Canonical_Species"]
    .astype(str)
    .str.strip()
)

alias_raw["Existing_BIEN_Species_Coverage"] = (
    alias_raw["Omo_Canonical_Species"]
    .isin(existing_species)
)

coverage_class = (
    alias_raw[
        [
            "Omo_Canonical_Species",
            "Existing_BIEN_Species_Coverage",
        ]
    ]
    .drop_duplicates()
    .copy()
)

coverage_class["Alias_Rescue_Coverage_Class"] = np.where(
    coverage_class["Existing_BIEN_Species_Coverage"],
    "ALREADY_COVERED_BY_SECTION_9_1",
    "NEW_CANONICAL_TAXON_COVERAGE"
)

print("\nAlias-requery canonical coverage classification")
print("-" * 100)

coverage_summary = (
    coverage_class[
        "Alias_Rescue_Coverage_Class"
    ]
    .value_counts()
    .rename_axis("Alias_Rescue_Coverage_Class")
    .reset_index(name="N_Taxa")
)

display(coverage_summary)

print("\nCanonical taxa with genuinely new BIEN coverage")
print("-" * 100)

new_taxa = (
    coverage_class[
        coverage_class[
            "Alias_Rescue_Coverage_Class"
        ].eq("NEW_CANONICAL_TAXON_COVERAGE")
    ]
    ["Omo_Canonical_Species"]
    .sort_values()
    .tolist()
)

print(f"N taxa : {len(new_taxa)}")

for taxon in new_taxa:
    print(f" - {taxon}")

# -----------------------------------------------------------------------------
# 5. CONSTRUCT DUPLICATE SIGNATURES
# -----------------------------------------------------------------------------
#
# We compare alias evidence against previously accepted BIEN evidence.
#
# BIEN record ID is preferred where available. Otherwise we compare a
# conservative scientific signature using taxon/value/source/method fields.
# -----------------------------------------------------------------------------

def clean_string_series(series):
    return (
        series
        .astype("string")
        .fillna("")
        .str.strip()
        .str.lower()
        .str.replace(r"\s+", " ", regex=True)
    )

# Determine BIEN ID field.
candidate_id_fields = [
    "id",
    "trait_id",
    "observation_id",
]

id_field = next(
    (
        c for c in candidate_id_fields
        if c in alias_raw.columns and c in existing.columns
    ),
    None
)

print("\nDuplicate-signature strategy")
print("-" * 100)

print(
    "Shared BIEN identifier field :",
    id_field if id_field is not None else "NONE"
)

# Fields used for fallback scientific signature.
signature_fields = [
    c for c in [
        "trait_value",
        "unit",
        "method",
        "url_source",
        "source_citation",
    ]
    if c in alias_raw.columns and c in existing.columns
]

print(
    "Shared scientific fields     :",
    ", ".join(signature_fields)
)

# -----------------------------------------------------------------------------
# 6. RECORD-ID OVERLAP
# -----------------------------------------------------------------------------

if id_field is not None:

    existing_ids = set(
        existing[id_field]
        .dropna()
        .astype(str)
        .str.strip()
    )

    alias_raw["Duplicate_By_BIEN_ID"] = (
        alias_raw[id_field]
        .astype("string")
        .str.strip()
        .isin(existing_ids)
    )

else:

    alias_raw["Duplicate_By_BIEN_ID"] = False

# -----------------------------------------------------------------------------
# 7. FALLBACK SCIENTIFIC-SIGNATURE OVERLAP
# -----------------------------------------------------------------------------

def make_signature(df, fields):

    parts = []

    for field in fields:

        if field == "trait_value":
            numeric = pd.to_numeric(
                df[field],
                errors="coerce"
            )

            part = numeric.map(
                lambda x: (
                    f"{x:.12g}"
                    if pd.notna(x)
                    else ""
                )
            )

        else:
            part = clean_string_series(df[field])

        parts.append(part)

    if not parts:
        return pd.Series(
            "",
            index=df.index,
            dtype="string"
        )

    signature = parts[0].astype("string")

    for part in parts[1:]:
        signature = (
            signature
            + " || "
            + part.astype("string")
        )

    return signature


if signature_fields:

    alias_raw["Scientific_Signature"] = make_signature(
        alias_raw,
        signature_fields
    )

    existing_signature = make_signature(
        existing,
        signature_fields
    )

    existing_signature_set = set(
        existing_signature.tolist()
    )

    alias_raw["Duplicate_By_Scientific_Signature"] = (
        alias_raw["Scientific_Signature"]
        .isin(existing_signature_set)
    )

else:

    alias_raw["Scientific_Signature"] = ""
    alias_raw["Duplicate_By_Scientific_Signature"] = False

# Combined overlap status.
alias_raw["Existing_Evidence_Overlap"] = (
    alias_raw["Duplicate_By_BIEN_ID"]
    |
    alias_raw["Duplicate_By_Scientific_Signature"]
)

# -----------------------------------------------------------------------------
# 8. ASSIGN AUDIT DISPOSITION
# -----------------------------------------------------------------------------

alias_raw["Alias_Evidence_Audit_Status"] = (
    "PROCEED_TO_MEASUREMENT_QC"
)

alias_raw.loc[
    ~alias_raw["Query_Name_Matched_By_BIEN"],
    "Alias_Evidence_Audit_Status"
] = "HOLD_FOR_TAXONOMIC_REVIEW"

alias_raw.loc[
    alias_raw["Existing_Evidence_Overlap"]
    &
    alias_raw["Query_Name_Matched_By_BIEN"],
    "Alias_Evidence_Audit_Status"
] = "EXISTING_BIEN_EVIDENCE_DUPLICATE"

print("\nRecord-level alias evidence audit")
print("-" * 100)

audit_summary = (
    alias_raw[
        "Alias_Evidence_Audit_Status"
    ]
    .value_counts()
    .rename_axis("Alias_Evidence_Audit_Status")
    .reset_index(name="N_Raw_Records")
)

display(audit_summary)

# -----------------------------------------------------------------------------
# 9. SPECIES-LEVEL AUDIT SUMMARY
# -----------------------------------------------------------------------------

species_audit = (
    alias_raw
    .groupby(
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
        ],
        as_index=False
    )
    .agg(
        N_Raw_Records=(
            "Trait_Value_mg",
            "size"
        ),
        N_Taxonomically_Consistent=(
            "Query_Name_Matched_By_BIEN",
            "sum"
        ),
        N_Existing_Evidence_Duplicates=(
            "Existing_Evidence_Overlap",
            "sum"
        ),
        N_Proceed_To_QC=(
            "Alias_Evidence_Audit_Status",
            lambda x: (
                x == "PROCEED_TO_MEASUREMENT_QC"
            ).sum()
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Median_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
)

species_audit["Existing_BIEN_Species_Coverage"] = (
    species_audit["Omo_Canonical_Species"]
    .isin(existing_species)
)

species_audit["Species_Alias_Audit_Status"] = np.select(
    [
        species_audit[
            "N_Taxonomically_Consistent"
        ].eq(0),

        species_audit[
            "N_Proceed_To_QC"
        ].eq(0),

        species_audit[
            "N_Proceed_To_QC"
        ].gt(0),
    ],
    [
        "TAXONOMIC_REVIEW_REQUIRED",
        "NO_NEW_MEASUREMENTS_AFTER_OVERLAP",
        "NEW_MEASUREMENTS_AVAILABLE_FOR_QC",
    ],
    default="REVIEW"
)

print("\nSpecies-level alias evidence audit")
print("-" * 100)

display(
    species_audit
    .sort_values("Omo_Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 10. EXTRACT RECORDS ELIGIBLE FOR NEXT QC STAGE
# -----------------------------------------------------------------------------

qc_candidates = alias_raw[
    alias_raw[
        "Alias_Evidence_Audit_Status"
    ].eq("PROCEED_TO_MEASUREMENT_QC")
].copy()

print("\nAlias-rescue measurements eligible for next QC stage")
print("-" * 100)

print(f"N records : {len(qc_candidates)}")
print(
    "N canonical taxa :",
    qc_candidates[
        "Omo_Canonical_Species"
    ].nunique()
)

if not qc_candidates.empty:

    display(
        qc_candidates[
            [
                "Omo_Canonical_Species",
                "Validated_BIEN_Query_Name",
                "Trait_Value_mg",
                "Query_Name_Matched_By_BIEN",
                "Existing_Evidence_Overlap",
            ]
        ]
        .sort_values(
            [
                "Omo_Canonical_Species",
                "Trait_Value_mg",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 11. EXPORT
# -----------------------------------------------------------------------------

audit_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_requery_crosswalk_overlap_audit.csv"
)

species_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_requery_species_audit.csv"
)

qc_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_requery_qc_candidates.csv"
)

taxonomy_review_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_requery_taxonomic_review.csv"
)

alias_raw.to_csv(
    audit_export,
    index=False
)

species_audit.to_csv(
    species_export,
    index=False
)

qc_candidates.to_csv(
    qc_export,
    index=False
)

taxonomy_review.to_csv(
    taxonomy_review_export,
    index=False
)

print("\nAudit exports")
print("-" * 100)

print(f"Crosswalk / overlap audit : {audit_export}")
print(f"Species audit             : {species_export}")
print(f"QC candidates             : {qc_export}")
print(f"Taxonomic review          : {taxonomy_review_export}")

# -----------------------------------------------------------------------------
# 12. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(alias_raw) == 61

assert (
    alias_raw[
        "Alias_Evidence_Audit_Status"
    ]
    .notna()
    .all()
)

assert (
    len(qc_candidates)
    +
    (
        alias_raw[
            "Alias_Evidence_Audit_Status"
        ].ne("PROCEED_TO_MEASUREMENT_QC")
    ).sum()
    == 61
)

print("\nBIEN alias-rescue crosswalk and overlap audit completed successfully.")
print("No alias-rescue measurement has yet been accepted.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2F — CARRY FORWARD PRIOR BIEN QC INTO TAXONOMIC-ALIAS RESCUE
# =============================================================================
#
# Purpose
# -------
# Correctly handle BIEN records recovered through validated taxonomic aliases.
#
# Scientific principle
# --------------------
# If an Omo canonical taxon is a validated spelling variant / synonym of a
# species whose BIEN measurements were already reviewed in Section 9.1,
# those reviewed measurements remain legitimate species-level evidence for
# the rescued canonical taxon.
#
# Therefore:
#
#   PRIOR ACCEPTED BIEN RECORD
#       -> reuse for rescued canonical taxon
#
#   PRIOR REJECTED BIEN RECORD
#       -> preserve rejection
#
#   PREVIOUSLY UNSEEN BIEN RECORD
#       -> send to new measurement-level QC
#
# Duplication is evaluated WITHIN each rescued Omo canonical taxon, not across
# different canonical taxa that legitimately share the same accepted species.
#
# No species-level aggregation occurs here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

ALIAS_RAW_FILE = (
    FT_DIR / "omo_seed_mass_bien_alias_requery_raw.csv"
)

PRIOR_ACCEPTED_FILE = (
    FT_DIR / "omo_seed_mass_bien_accepted_measurements.csv"
)

PRIOR_REJECTED_FILE = (
    FT_DIR / "omo_seed_mass_bien_rejected_measurements.csv"
)

assert ALIAS_RAW_FILE.exists()
assert PRIOR_ACCEPTED_FILE.exists()
assert PRIOR_REJECTED_FILE.exists()

alias_raw = pd.read_csv(
    ALIAS_RAW_FILE,
    low_memory=False
)

prior_accepted = pd.read_csv(
    PRIOR_ACCEPTED_FILE,
    low_memory=False
)

prior_rejected = pd.read_csv(
    PRIOR_REJECTED_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2F — CARRY FORWARD PRIOR BIEN QC INTO TAXONOMIC-ALIAS RESCUE")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT STATE
# -----------------------------------------------------------------------------

print("\nInput evidence")
print("-" * 100)

print(f"Alias-rescue raw mapped records : {len(alias_raw)}")
print(f"Prior accepted measurements     : {len(prior_accepted)}")
print(f"Prior rejected measurements     : {len(prior_rejected)}")

assert len(alias_raw) == 61
assert len(prior_accepted) == 57
assert len(prior_rejected) == 1

required_alias_cols = {
    "Omo_Canonical_Species",
    "Validated_BIEN_Query_Name",
    "trait_value",
    "unit",
    "id",
}

missing = required_alias_cols - set(alias_raw.columns)

assert not missing, (
    f"Alias evidence missing required fields: {sorted(missing)}"
)

# -----------------------------------------------------------------------------
# 2. NORMALIZE IDENTIFIERS / VALUES
# -----------------------------------------------------------------------------

for df in [
    alias_raw,
    prior_accepted,
    prior_rejected,
]:

    if "id" in df.columns:
        df["BIEN_Record_ID"] = (
            df["id"]
            .astype("string")
            .str.strip()
        )

alias_raw["Trait_Value_mg"] = pd.to_numeric(
    alias_raw["trait_value"],
    errors="coerce"
)

assert alias_raw["Trait_Value_mg"].notna().all()
assert alias_raw["Trait_Value_mg"].gt(0).all()

# -----------------------------------------------------------------------------
# 3. BUILD PRIOR QC ID SETS
# -----------------------------------------------------------------------------

accepted_ids = set(
    prior_accepted["BIEN_Record_ID"]
    .dropna()
    .astype(str)
)

rejected_ids = set(
    prior_rejected["BIEN_Record_ID"]
    .dropna()
    .astype(str)
)

assert accepted_ids.isdisjoint(rejected_ids), (
    "A BIEN record appears in both prior accepted and rejected evidence."
)

print("\nPrior reviewed BIEN IDs")
print("-" * 100)

print(f"Accepted IDs : {len(accepted_ids)}")
print(f"Rejected IDs : {len(rejected_ids)}")

# -----------------------------------------------------------------------------
# 4. INHERIT PRIOR QC DECISIONS
# -----------------------------------------------------------------------------

alias_raw["Prior_QC_Status"] = np.select(
    [
        alias_raw["BIEN_Record_ID"].isin(accepted_ids),
        alias_raw["BIEN_Record_ID"].isin(rejected_ids),
    ],
    [
        "PRIOR_ACCEPTED_EVIDENCE_REUSED",
        "PRIOR_REJECTED_EVIDENCE_EXCLUDED",
    ],
    default="NEW_EVIDENCE_REQUIRES_QC"
)

prior_qc_summary = (
    alias_raw["Prior_QC_Status"]
    .value_counts()
    .rename_axis("Prior_QC_Status")
    .reset_index(name="N_Raw_Records")
)

print("\nInherited prior-QC classification")
print("-" * 100)

display(prior_qc_summary)

# -----------------------------------------------------------------------------
# 5. VERIFY THE NAUCLEA DECISION IS CARRIED FORWARD
# -----------------------------------------------------------------------------

nauclea_rescue = alias_raw[
    alias_raw["Validated_BIEN_Query_Name"]
    .eq("Nauclea diderrichii")
].copy()

print("\nNauclea alias-rescue evidence after prior-QC inheritance")
print("-" * 100)

display(
    nauclea_rescue[
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
            "BIEN_Record_ID",
            "Trait_Value_mg",
            "Prior_QC_Status",
        ]
    ]
    .sort_values(
        [
            "Omo_Canonical_Species",
            "Trait_Value_mg",
        ]
    )
    .reset_index(drop=True)
)

# Expected:
#   0.30   -> previously accepted
#   2.83   -> previously accepted
#   317.46 -> previously rejected

for canonical in [
    "Nauclea diderrichi",
    "Nuclea diderrichi",
]:

    sub = nauclea_rescue[
        nauclea_rescue[
            "Omo_Canonical_Species"
        ].eq(canonical)
    ]

    assert (
        sub.loc[
            np.isclose(
                sub["Trait_Value_mg"],
                0.30
            ),
            "Prior_QC_Status"
        ]
        .eq("PRIOR_ACCEPTED_EVIDENCE_REUSED")
        .all()
    )

    assert (
        sub.loc[
            np.isclose(
                sub["Trait_Value_mg"],
                2.83
            ),
            "Prior_QC_Status"
        ]
        .eq("PRIOR_ACCEPTED_EVIDENCE_REUSED")
        .all()
    )

    assert (
        sub.loc[
            np.isclose(
                sub["Trait_Value_mg"],
                317.46
            ),
            "Prior_QC_Status"
        ]
        .eq("PRIOR_REJECTED_EVIDENCE_EXCLUDED")
        .all()
    )

# -----------------------------------------------------------------------------
# 6. SCIENTIFIC SIGNATURE FOR WITHIN-CANONICAL DEDUPLICATION
# -----------------------------------------------------------------------------

signature_cols = [
    "Omo_Canonical_Species",
    "Trait_Value_mg",
]

for optional in [
    "method",
    "url_source",
    "source_citation",
]:
    if optional in alias_raw.columns:
        signature_cols.append(optional)

alias_raw["Within_Canonical_Signature_Count"] = (
    alias_raw.groupby(
        signature_cols,
        dropna=False
    )["BIEN_Record_ID"]
    .transform("size")
)

# Keep one representative of each identical scientific signature
# WITHIN the rescued Omo canonical taxon.

alias_dedup = (
    alias_raw
    .drop_duplicates(
        subset=signature_cols,
        keep="first"
    )
    .copy()
    .reset_index(drop=True)
)

print("\nWithin-canonical deduplication")
print("-" * 100)

print(f"Mapped raw evidence rows        : {len(alias_raw)}")
print(f"Unique within-taxon signatures : {len(alias_dedup)}")
print(
    f"Redundant rows removed          : "
    f"{len(alias_raw) - len(alias_dedup)}"
)

# -----------------------------------------------------------------------------
# 7. ASSIGN RESCUE MEASUREMENT DISPOSITION
# -----------------------------------------------------------------------------

alias_dedup["Rescue_Measurement_Status"] = np.select(
    [
        alias_dedup[
            "Prior_QC_Status"
        ].eq("PRIOR_REJECTED_EVIDENCE_EXCLUDED"),

        alias_dedup[
            "Prior_QC_Status"
        ].eq("PRIOR_ACCEPTED_EVIDENCE_REUSED"),

        alias_dedup[
            "Prior_QC_Status"
        ].eq("NEW_EVIDENCE_REQUIRES_QC"),
    ],
    [
        "EXCLUDE_PRIOR_REJECTED_RECORD",
        "ACCEPT_PRIOR_REVIEWED_RECORD",
        "SEND_TO_NEW_MEASUREMENT_QC",
    ],
    default="REVIEW"
)

disposition_summary = (
    alias_dedup[
        "Rescue_Measurement_Status"
    ]
    .value_counts()
    .rename_axis(
        "Rescue_Measurement_Status"
    )
    .reset_index(name="N_Measurements")
)

print("\nPost-dedup rescue measurement disposition")
print("-" * 100)

display(disposition_summary)

# -----------------------------------------------------------------------------
# 8. PRIOR-ACCEPTED EVIDENCE READY FOR DIRECT REUSE
# -----------------------------------------------------------------------------

reused_accepted = alias_dedup[
    alias_dedup[
        "Rescue_Measurement_Status"
    ].eq("ACCEPT_PRIOR_REVIEWED_RECORD")
].copy()

new_qc = alias_dedup[
    alias_dedup[
        "Rescue_Measurement_Status"
    ].eq("SEND_TO_NEW_MEASUREMENT_QC")
].copy()

reused_rejected = alias_dedup[
    alias_dedup[
        "Rescue_Measurement_Status"
    ].eq("EXCLUDE_PRIOR_REJECTED_RECORD")
].copy()

print("\nPrior-reviewed measurements reused for rescued canonical taxa")
print("-" * 100)

print(f"N measurements     : {len(reused_accepted)}")
print(
    f"N canonical taxa   : "
    f"{reused_accepted['Omo_Canonical_Species'].nunique()}"
)

if not reused_accepted.empty:
    display(
        reused_accepted[
            [
                "Omo_Canonical_Species",
                "Validated_BIEN_Query_Name",
                "BIEN_Record_ID",
                "Trait_Value_mg",
                "Prior_QC_Status",
            ]
        ]
        .sort_values(
            [
                "Omo_Canonical_Species",
                "Trait_Value_mg",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 9. NEW EVIDENCE REQUIRING FRESH QC
# -----------------------------------------------------------------------------

print("\nPreviously unseen alias-rescue evidence requiring QC")
print("-" * 100)

print(f"N measurements   : {len(new_qc)}")
print(
    f"N canonical taxa : "
    f"{new_qc['Omo_Canonical_Species'].nunique()}"
)

if not new_qc.empty:

    new_qc_species = (
        new_qc.groupby(
            [
                "Omo_Canonical_Species",
                "Validated_BIEN_Query_Name",
            ]
        )
        .agg(
            N_Measurements=(
                "Trait_Value_mg",
                "size"
            ),
            N_Unique_Values=(
                "Trait_Value_mg",
                "nunique"
            ),
            Min_mg=(
                "Trait_Value_mg",
                "min"
            ),
            Median_mg=(
                "Trait_Value_mg",
                "median"
            ),
            Max_mg=(
                "Trait_Value_mg",
                "max"
            ),
        )
        .reset_index()
    )

    new_qc_species["Fold_Range"] = (
        new_qc_species["Max_mg"]
        /
        new_qc_species["Min_mg"]
    )

    display(
        new_qc_species
        .sort_values(
            "Omo_Canonical_Species"
        )
        .reset_index(drop=True)
    )

else:

    new_qc_species = pd.DataFrame()
    print("NONE")

# -----------------------------------------------------------------------------
# 10. PRIOR-REJECTED RECORDS REMAIN EXCLUDED
# -----------------------------------------------------------------------------

print("\nPrior-rejected records preserved as rejected")
print("-" * 100)

print(f"N records : {len(reused_rejected)}")

if not reused_rejected.empty:

    display(
        reused_rejected[
            [
                "Omo_Canonical_Species",
                "Validated_BIEN_Query_Name",
                "BIEN_Record_ID",
                "Trait_Value_mg",
                "Prior_QC_Status",
            ]
        ]
        .sort_values(
            [
                "Omo_Canonical_Species",
                "Trait_Value_mg",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 11. CANONICAL-TAXON RESCUE STATUS
# -----------------------------------------------------------------------------

taxon_status_records = []

for taxon, group in alias_dedup.groupby(
    "Omo_Canonical_Species"
):

    n_reused = (
        group[
            "Rescue_Measurement_Status"
        ]
        .eq("ACCEPT_PRIOR_REVIEWED_RECORD")
        .sum()
    )

    n_new = (
        group[
            "Rescue_Measurement_Status"
        ]
        .eq("SEND_TO_NEW_MEASUREMENT_QC")
        .sum()
    )

    n_rejected = (
        group[
            "Rescue_Measurement_Status"
        ]
        .eq("EXCLUDE_PRIOR_REJECTED_RECORD")
        .sum()
    )

    if n_new > 0:
        status = "NEW_QC_REQUIRED"
    elif n_reused > 0:
        status = "PRIOR_REVIEWED_EVIDENCE_AVAILABLE"
    else:
        status = "NO_ACCEPTABLE_EVIDENCE_YET"

    taxon_status_records.append({
        "Canonical_Species": taxon,
        "N_Prior_Accepted_Reused": int(n_reused),
        "N_New_QC_Required": int(n_new),
        "N_Prior_Rejected": int(n_rejected),
        "Alias_Rescue_Status": status,
    })

taxon_status = pd.DataFrame(
    taxon_status_records
)

print("\nCanonical-taxon alias rescue status")
print("-" * 100)

display(
    taxon_status
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nRescue-status summary")
print("-" * 100)

display(
    taxon_status[
        "Alias_Rescue_Status"
    ]
    .value_counts()
    .rename_axis(
        "Alias_Rescue_Status"
    )
    .reset_index(name="N_Taxa")
)

# -----------------------------------------------------------------------------
# 12. EXPORT
# -----------------------------------------------------------------------------

dedup_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_rescue_deduplicated.csv"
)

reuse_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_prior_accepted_reused.csv"
)

new_qc_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_qc_required.csv"
)

rejected_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_prior_rejected.csv"
)

taxon_status_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_rescue_taxon_status.csv"
)

alias_dedup.to_csv(
    dedup_export,
    index=False
)

reused_accepted.to_csv(
    reuse_export,
    index=False
)

new_qc.to_csv(
    new_qc_export,
    index=False
)

reused_rejected.to_csv(
    rejected_export,
    index=False
)

taxon_status.to_csv(
    taxon_status_export,
    index=False
)

print("\nRescue-QC exports")
print("-" * 100)

print(f"Deduplicated rescue evidence : {dedup_export}")
print(f"Prior accepted reused        : {reuse_export}")
print(f"New evidence for QC          : {new_qc_export}")
print(f"Prior rejected carried over  : {rejected_export}")
print(f"Taxon rescue status          : {taxon_status_export}")

# -----------------------------------------------------------------------------
# 13. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert alias_dedup[
    "Rescue_Measurement_Status"
].notna().all()

assert set(
    reused_rejected[
        "Validated_BIEN_Query_Name"
    ].unique()
) <= {
    "Nauclea diderrichii"
}

assert (
    len(reused_accepted)
    + len(new_qc)
    + len(reused_rejected)
    == len(alias_dedup)
)

print("\nPrior BIEN QC decisions successfully propagated into alias rescue evidence.")
print("Previously accepted records are reusable for validated taxonomic variants.")
print("Previously rejected records remain rejected.")
print("Only genuinely unseen BIEN evidence proceeds to fresh measurement QC.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2G — FRESH QC DIAGNOSTIC FOR NEW BIEN ALIAS-RESCUE EVIDENCE
# =============================================================================
#
# Purpose
# -------
# Examine the 27 genuinely new BIEN Seed Mass measurements identified in
# Section 9.2F before accepting any of them.
#
# Target canonical taxa:
#   - Artocarpus communis
#   - Cassia saimen
#   - Diospyrous melistoformis
#   - Diospyrous mespiliformis
#   - Spondia mombin
#
# This cell examines:
#   1. taxonomic consistency;
#   2. source and method diversity;
#   3. value redundancy after prior deduplication;
#   4. within-species ranges on raw and log10 scales;
#   5. adjacent value jumps;
#   6. source-specific value ranges;
#   7. taxa requiring targeted external review.
#
# IMPORTANT:
# No new measurement is accepted or rejected here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

NEW_QC_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_qc_required.csv"
)

assert NEW_QC_FILE.exists(), (
    f"New-QC evidence file not found:\n{NEW_QC_FILE}"
)

df = pd.read_csv(
    NEW_QC_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2G — FRESH QC DIAGNOSTIC FOR NEW BIEN ALIAS-RESCUE EVIDENCE")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE EXPECTED INPUT
# -----------------------------------------------------------------------------

expected_taxa = {
    "Artocarpus communis",
    "Cassia saimen",
    "Diospyrous melistoformis",
    "Diospyrous mespiliformis",
    "Spondia mombin",
}

assert len(df) == 27, (
    f"Expected 27 new measurements, found {len(df)}."
)

assert set(
    df["Omo_Canonical_Species"]
) == expected_taxa

df["Trait_Value_mg"] = pd.to_numeric(
    df["Trait_Value_mg"],
    errors="coerce"
)

assert df["Trait_Value_mg"].notna().all()
assert df["Trait_Value_mg"].gt(0).all()

print("\nInput evidence")
print("-" * 100)

print(f"Measurements     : {len(df)}")
print(
    f"Canonical taxa   : "
    f"{df['Omo_Canonical_Species'].nunique()}"
)

# -----------------------------------------------------------------------------
# 2. AVAILABLE METADATA FIELDS
# -----------------------------------------------------------------------------

metadata_fields = [
    c for c in [
        "BIEN_Record_ID",
        "id",
        "Omo_Canonical_Species",
        "Validated_BIEN_Query_Name",
        "name_submitted",
        "verbatim_scientific_name",
        "name_matched",
        "scrubbed_species_binomial",
        "scrubbed_taxon_canonical",
        "scrubbed_taxonomic_status",
        "tnrs_warning",
        "method",
        "url_source",
        "source_citation",
        "project_pi",
        "access",
    ]
    if c in df.columns
]

print("\nAvailable review metadata fields")
print("-" * 100)

for col in metadata_fields:
    print(f" - {col}")

# -----------------------------------------------------------------------------
# 3. TAXONOMIC CONSISTENCY
# -----------------------------------------------------------------------------

tax_fields = [
    c for c in [
        "name_submitted",
        "verbatim_scientific_name",
        "name_matched",
        "scrubbed_species_binomial",
        "scrubbed_taxon_canonical",
        "scrubbed_taxonomic_status",
        "tnrs_warning",
    ]
    if c in df.columns
]

taxonomic_records = []

for taxon, group in df.groupby(
    "Omo_Canonical_Species"
):

    record = {
        "Canonical_Species": taxon,
        "Validated_Query_Name": (
            group["Validated_BIEN_Query_Name"]
            .iloc[0]
        ),
        "N_Measurements": len(group),
    }

    for col in tax_fields:

        vals = (
            group[col]
            .dropna()
            .astype(str)
            .str.strip()
        )

        record[f"{col}_N_Unique"] = (
            vals.nunique()
        )

        record[f"{col}_Values"] = (
            " | ".join(
                sorted(vals.unique())
            )
            if len(vals)
            else "<MISSING>"
        )

    taxonomic_records.append(record)

taxonomic_diagnostics = pd.DataFrame(
    taxonomic_records
)

print("\nTaxonomic consistency diagnostics")
print("-" * 100)

display(taxonomic_diagnostics)

# -----------------------------------------------------------------------------
# 4. SOURCE KEY
# -----------------------------------------------------------------------------

df["Source_Key"] = (
    df["source_citation"]
    .fillna(df["url_source"])
    .fillna("<UNKNOWN_SOURCE>")
    .astype(str)
    .str.strip()
)

# -----------------------------------------------------------------------------
# 5. SOURCE / METHOD STRUCTURE
# -----------------------------------------------------------------------------

source_method = (
    df.groupby(
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
            "Source_Key",
            "method",
        ],
        dropna=False
    )
    .agg(
        N_Measurements=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Median_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
)

source_method["Fold_Range"] = (
    source_method["Max_mg"]
    /
    source_method["Min_mg"]
)

print("\nSpecies × source × method structure")
print("-" * 100)

display(
    source_method
    .sort_values(
        [
            "Omo_Canonical_Species",
            "Source_Key",
        ]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 6. SPECIES-LEVEL VALUE DIAGNOSTICS
# -----------------------------------------------------------------------------

species_summary = (
    df.groupby(
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
        ]
    )
    .agg(
        N_Measurements=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        N_Sources=(
            "Source_Key",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Arithmetic_Mean_mg=(
            "Trait_Value_mg",
            "mean"
        ),
        Median_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: float(
                np.exp(
                    np.mean(
                        np.log(
                            x.astype(float)
                        )
                    )
                )
            )
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
)

species_summary["Fold_Range"] = (
    species_summary["Max_mg"]
    /
    species_summary["Min_mg"]
)

species_summary["Orders_of_Magnitude"] = (
    np.log10(
        species_summary["Max_mg"]
    )
    -
    np.log10(
        species_summary["Min_mg"]
    )
)

print("\nSpecies-level fresh-evidence summary")
print("-" * 100)

display(
    species_summary
    .sort_values(
        "Omo_Canonical_Species"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 7. ADJACENT-VALUE JUMP ANALYSIS
# -----------------------------------------------------------------------------

jump_records = []

for taxon, group in df.groupby(
    "Omo_Canonical_Species"
):

    values = sorted(
        group["Trait_Value_mg"]
        .astype(float)
        .unique()
    )

    for i in range(
        1,
        len(values)
    ):

        lower = values[i - 1]
        upper = values[i]

        jump_records.append({
            "Canonical_Species": taxon,
            "Lower_Value_mg": lower,
            "Upper_Value_mg": upper,
            "Adjacent_Fold_Jump": (
                upper / lower
            ),
            "Adjacent_Log10_Jump": (
                np.log10(upper)
                - np.log10(lower)
            ),
        })

jump_table = pd.DataFrame(
    jump_records
)

print("\nLargest adjacent Seed Mass jumps")
print("-" * 100)

if jump_table.empty:
    print("No species contains multiple unique measurements.")
else:
    display(
        jump_table
        .sort_values(
            "Adjacent_Fold_Jump",
            ascending=False
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 8. IDENTIFY HIGH-VARIABILITY TAXA
# -----------------------------------------------------------------------------
#
# As before, >=5-fold is a REVIEW trigger only.
# It is NOT an automatic rejection rule.
# -----------------------------------------------------------------------------

species_summary["Variability_Flag"] = np.where(
    species_summary["Fold_Range"] >= 5,
    "TARGETED_REVIEW_REQUIRED",
    "INITIAL_VARIABILITY_OK"
)

review_taxa = species_summary[
    species_summary[
        "Variability_Flag"
    ].eq("TARGETED_REVIEW_REQUIRED")
].copy()

print("\nTaxa requiring targeted variability review")
print("-" * 100)

if review_taxa.empty:
    print("NONE")
else:
    display(
        review_taxa
        .sort_values(
            "Fold_Range",
            ascending=False
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 9. SINGLE-MEASUREMENT TAXA
# -----------------------------------------------------------------------------

single_measurement_taxa = (
    species_summary[
        species_summary[
            "N_Measurements"
        ].eq(1)
    ]
    .copy()
)

print("\nSingle-measurement rescued taxa")
print("-" * 100)

if single_measurement_taxa.empty:
    print("NONE")
else:
    display(
        single_measurement_taxa[
            [
                "Omo_Canonical_Species",
                "Validated_BIEN_Query_Name",
                "N_Measurements",
                "Min_mg",
                "Median_mg",
                "Max_mg",
            ]
        ]
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 10. FULL RECORD DETAIL FOR REVIEW TAXA
# -----------------------------------------------------------------------------

review_species_set = set(
    review_taxa[
        "Omo_Canonical_Species"
    ]
)

review_records = (
    df[
        df[
            "Omo_Canonical_Species"
        ].isin(review_species_set)
    ]
    .copy()
)

detail_cols = [
    c for c in [
        "BIEN_Record_ID",
        "id",
        "Omo_Canonical_Species",
        "Validated_BIEN_Query_Name",
        "Trait_Value_mg",
        "method",
        "url_source",
        "source_citation",
        "name_submitted",
        "scrubbed_species_binomial",
        "scrubbed_taxonomic_status",
        "tnrs_warning",
    ]
    if c in review_records.columns
]

print("\nRecord-level evidence for targeted-review taxa")
print("-" * 100)

if review_records.empty:
    print("NONE")
else:
    display(
        review_records[
            detail_cols
        ]
        .sort_values(
            [
                "Omo_Canonical_Species",
                "Trait_Value_mg",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 11. PROVISIONAL QC ROUTING
# -----------------------------------------------------------------------------

routing = species_summary[
    [
        "Omo_Canonical_Species",
        "Validated_BIEN_Query_Name",
        "N_Measurements",
        "N_Unique_Values",
        "N_Sources",
        "Fold_Range",
        "Variability_Flag",
    ]
].copy()

routing["Fresh_QC_Route"] = np.select(
    [
        routing[
            "Variability_Flag"
        ].eq("TARGETED_REVIEW_REQUIRED"),

        routing[
            "N_Measurements"
        ].eq(1),
    ],
    [
        "TARGETED_EXTERNAL_REVIEW",
        "SINGLE_MEASUREMENT_SOURCE_REVIEW",
    ],
    default="PROVISIONAL_QC_PASS"
)

print("\nFresh-evidence QC routing")
print("-" * 100)

display(
    routing
    .sort_values(
        "Omo_Canonical_Species"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 12. EXPORT
# -----------------------------------------------------------------------------

taxonomic_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_taxonomic_diagnostics.csv"
)

source_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_source_method_summary.csv"
)

species_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_species_summary.csv"
)

jump_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_adjacent_jumps.csv"
)

review_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_targeted_review_records.csv"
)

routing_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_qc_routing.csv"
)

taxonomic_diagnostics.to_csv(
    taxonomic_export,
    index=False
)

source_method.to_csv(
    source_export,
    index=False
)

species_summary.to_csv(
    species_export,
    index=False
)

jump_table.to_csv(
    jump_export,
    index=False
)

review_records.to_csv(
    review_export,
    index=False
)

routing.to_csv(
    routing_export,
    index=False
)

print("\nFresh-QC diagnostic exports")
print("-" * 100)

print(f"Taxonomic diagnostics : {taxonomic_export}")
print(f"Source/method summary : {source_export}")
print(f"Species summary       : {species_export}")
print(f"Adjacent jumps        : {jump_export}")
print(f"Review records        : {review_export}")
print(f"QC routing            : {routing_export}")

# -----------------------------------------------------------------------------
# 13. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(species_summary) == 5

assert set(
    species_summary[
        "Omo_Canonical_Species"
    ]
) == expected_taxa

assert (
    species_summary["Min_mg"] > 0
).all()

assert (
    species_summary["Max_mg"]
    >= species_summary["Min_mg"]
).all()

print("\nFresh alias-rescue BIEN evidence diagnostic completed successfully.")
print("No new measurements have yet been accepted or rejected.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2H — RESOLVE CLEAN FRESH ALIAS-RESCUE EVIDENCE AND
#                ISOLATE SPONDIAS DEFINITION AMBIGUITY
# =============================================================================
#
# Purpose
# -------
# Resolve the 27 genuinely new BIEN alias-rescue Seed Mass measurements
# after Section 9.2G diagnostic review.
#
# Decisions
# ---------
#
# ACCEPT:
#   Artocarpus communis -> Artocarpus altilis
#       5898 mg independently corroborated by Kew 1000-seed weight.
#
#   Cassia saimen -> Senna siamea
#       tight multi-source range; passes measurement QC.
#
#   Diospyrous melistoformis -> Diospyros mespiliformis
#       two consistent BIEN measurements.
#
#   Diospyrous mespiliformis -> Diospyros mespiliformis
#       same validated species evidence.
#
# HOLD:
#   Spondia mombin -> Spondias mombin
#       substantial evidence-definition heterogeneity:
#       "seed", true seed, stone/endocarp and dispersal unit may not be
#       equivalent in this species.
#
# IMPORTANT:
#   No Spondias measurement is rejected here.
#   No species-level aggregation occurs yet.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

NEW_QC_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_qc_required.csv"
)

assert NEW_QC_FILE.exists()

df = pd.read_csv(
    NEW_QC_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2H — RESOLVE CLEAN FRESH ALIAS-RESCUE EVIDENCE")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT
# -----------------------------------------------------------------------------

expected_taxa = {
    "Artocarpus communis",
    "Cassia saimen",
    "Diospyrous melistoformis",
    "Diospyrous mespiliformis",
    "Spondia mombin",
}

assert len(df) == 27

assert set(
    df["Omo_Canonical_Species"]
) == expected_taxa

df["Trait_Value_mg"] = pd.to_numeric(
    df["Trait_Value_mg"],
    errors="coerce"
)

assert df["Trait_Value_mg"].notna().all()
assert df["Trait_Value_mg"].gt(0).all()

# -----------------------------------------------------------------------------
# 2. INITIALIZE REVIEW FIELDS
# -----------------------------------------------------------------------------

df["Fresh_QC_Decision"] = pd.NA
df["Evidence_Action"] = pd.NA
df["Evidence_Interpretation"] = pd.NA
df["Trait_Definition_Flag"] = pd.NA
df["Review_Rationale"] = pd.NA

# -----------------------------------------------------------------------------
# 3. ARTОCARPUS COMMUNIS / ARTОCARPUS ALTILIS
# -----------------------------------------------------------------------------

artocarpus_mask = (
    df["Omo_Canonical_Species"]
    .eq("Artocarpus communis")
)

assert artocarpus_mask.sum() == 1

assert np.isclose(
    df.loc[
        artocarpus_mask,
        "Trait_Value_mg"
    ].iloc[0],
    5898.0
)

df.loc[
    artocarpus_mask,
    "Fresh_QC_Decision"
] = "ACCEPT"

df.loc[
    artocarpus_mask,
    "Evidence_Action"
] = "RETAIN_MEASUREMENT"

df.loc[
    artocarpus_mask,
    "Evidence_Interpretation"
] = "SPECIES_LEVEL_VALIDATED_ALIAS"

df.loc[
    artocarpus_mask,
    "Trait_Definition_Flag"
] = "SEED_MASS_EXTERNALLY_CORROBORATED"

df.loc[
    artocarpus_mask,
    "Review_Rationale"
] = (
    "Accepted. BIEN value of 5898 mg is independently corroborated "
    "by Kew reporting an average 1000-seed weight of 5898 g for "
    "Artocarpus altilis."
)

# -----------------------------------------------------------------------------
# 4. CASSIA SAIMEN / SENNA SIAMEA
# -----------------------------------------------------------------------------

cassia_mask = (
    df["Omo_Canonical_Species"]
    .eq("Cassia saimen")
)

assert cassia_mask.sum() == 9

df.loc[
    cassia_mask,
    "Fresh_QC_Decision"
] = "ACCEPT"

df.loc[
    cassia_mask,
    "Evidence_Action"
] = "RETAIN_MEASUREMENT"

df.loc[
    cassia_mask,
    "Evidence_Interpretation"
] = "SPECIES_LEVEL_VALIDATED_ALIAS"

df.loc[
    cassia_mask,
    "Trait_Definition_Flag"
] = "SEED_MASS_QC_PASSED"

df.loc[
    cassia_mask,
    "Review_Rationale"
] = (
    "Accepted after BIEN QC. Nine deduplicated measurements from "
    "two sources show a narrow 22.7-28.99 mg range."
)

# -----------------------------------------------------------------------------
# 5. DIOSPYROS VALIDATED-ALIAS TAXA
# -----------------------------------------------------------------------------

diospyros_taxa = {
    "Diospyrous melistoformis",
    "Diospyrous mespiliformis",
}

diospyros_mask = (
    df["Omo_Canonical_Species"]
    .isin(diospyros_taxa)
)

assert diospyros_mask.sum() == 4

df.loc[
    diospyros_mask,
    "Fresh_QC_Decision"
] = "ACCEPT"

df.loc[
    diospyros_mask,
    "Evidence_Action"
] = "RETAIN_MEASUREMENT"

df.loc[
    diospyros_mask,
    "Evidence_Interpretation"
] = "SPECIES_LEVEL_VALIDATED_ALIAS"

df.loc[
    diospyros_mask,
    "Trait_Definition_Flag"
] = "SEED_MASS_QC_PASSED"

df.loc[
    diospyros_mask,
    "Review_Rationale"
] = (
    "Accepted after BIEN QC. Validated mapping to Diospyros "
    "mespiliformis returned two consistent measurements "
    "(365 and 500 mg)."
)

# -----------------------------------------------------------------------------
# 6. SPONDIA MOMBIN / SPONDIAS MOMBIN
# -----------------------------------------------------------------------------
#
# DO NOT reject values solely because they are statistically extreme.
#
# External evidence indicates that Spondias literature and seed databases
# may variously treat:
#
#   - true individual seed,
#   - lignified endocarp / stone,
#   - dispersal unit
#
# as the measured entity.
#
# Because our standardized trait definition is individual seed dry mass
# where possible, all 13 values remain on HOLD until definition-specific
# evidence is resolved.
# -----------------------------------------------------------------------------

spondias_mask = (
    df["Omo_Canonical_Species"]
    .eq("Spondia mombin")
)

assert spondias_mask.sum() == 13

df.loc[
    spondias_mask,
    "Fresh_QC_Decision"
] = "HOLD"

df.loc[
    spondias_mask,
    "Evidence_Action"
] = "DEFER_DEFINITION_REVIEW"

df.loc[
    spondias_mask,
    "Evidence_Interpretation"
] = "SPECIES_LEVEL_BUT_MEASUREMENT_ENTITY_AMBIGUOUS"

df.loc[
    spondias_mask,
    "Trait_Definition_Flag"
] = "SEED_VS_ENDOCARP_DISPERSAL_UNIT_AMBIGUITY"

df.loc[
    spondias_mask,
    "Review_Rationale"
] = (
    "Held for definition-specific review. External botanical evidence "
    "shows that the lignified endocarp/stone is the Spondias dispersal "
    "unit and may contain multiple true seeds. BIEN values therefore "
    "may mix true seed mass with stone/endocarp mass."
)

# -----------------------------------------------------------------------------
# 7. VALIDATE ALL DECISIONS
# -----------------------------------------------------------------------------

assert df["Fresh_QC_Decision"].notna().all()
assert df["Evidence_Action"].notna().all()
assert df["Review_Rationale"].notna().all()

decision_summary = (
    df["Fresh_QC_Decision"]
    .value_counts()
    .rename_axis("Fresh_QC_Decision")
    .reset_index(name="N_Measurements")
)

print("\nFresh-QC decisions")
print("-" * 100)

display(decision_summary)

# Expected:
#   ACCEPT = 14
#   HOLD   = 13

assert (
    (df["Fresh_QC_Decision"] == "ACCEPT").sum()
    == 14
)

assert (
    (df["Fresh_QC_Decision"] == "HOLD").sum()
    == 13
)

# -----------------------------------------------------------------------------
# 8. ACCEPTED NEW ALIAS-RESCUE MEASUREMENTS
# -----------------------------------------------------------------------------

accepted_new = (
    df[
        df["Fresh_QC_Decision"]
        .eq("ACCEPT")
    ]
    .copy()
    .reset_index(drop=True)
)

held_definition = (
    df[
        df["Fresh_QC_Decision"]
        .eq("HOLD")
    ]
    .copy()
    .reset_index(drop=True)
)

print("\nAccepted new alias-rescue evidence")
print("-" * 100)

print(
    f"Accepted measurements : {len(accepted_new)}"
)

print(
    f"Canonical taxa        : "
    f"{accepted_new['Omo_Canonical_Species'].nunique()}"
)

display(
    accepted_new[
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
            "Trait_Value_mg",
            "Fresh_QC_Decision",
            "Trait_Definition_Flag",
        ]
    ]
    .sort_values(
        [
            "Omo_Canonical_Species",
            "Trait_Value_mg",
        ]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 9. SPONDIAS HOLD REGISTER
# -----------------------------------------------------------------------------

print("\nSpondias definition-review hold")
print("-" * 100)

print(
    f"Held measurements : {len(held_definition)}"
)

display(
    held_definition[
        [
            "BIEN_Record_ID",
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
            "Trait_Value_mg",
            "method",
            "source_citation",
            "Fresh_QC_Decision",
            "Trait_Definition_Flag",
        ]
    ]
    .sort_values(
        "Trait_Value_mg"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 10. SPECIES-LEVEL ACCEPTED-EVIDENCE SUMMARY
# -----------------------------------------------------------------------------

accepted_species_summary = (
    accepted_new.groupby(
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
        ]
    )
    .agg(
        N_Accepted_Measurements=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Median_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: float(
                np.exp(
                    np.mean(
                        np.log(
                            x.astype(float)
                        )
                    )
                )
            )
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
)

accepted_species_summary["Fold_Range"] = (
    accepted_species_summary["Max_mg"]
    /
    accepted_species_summary["Min_mg"]
)

print("\nAccepted fresh alias-rescue species summary")
print("-" * 100)

display(
    accepted_species_summary
    .sort_values(
        "Omo_Canonical_Species"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 11. EXPORT
# -----------------------------------------------------------------------------

review_register_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_fresh_qc_resolved_register.csv"
)

accepted_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_accepted_measurements.csv"
)

spondias_hold_export = (
    FT_DIR /
    "omo_seed_mass_spondias_definition_review_hold.csv"
)

accepted_summary_export = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_accepted_species_summary.csv"
)

df.to_csv(
    review_register_export,
    index=False
)

accepted_new.to_csv(
    accepted_export,
    index=False
)

held_definition.to_csv(
    spondias_hold_export,
    index=False
)

accepted_species_summary.to_csv(
    accepted_summary_export,
    index=False
)

print("\nResolved fresh-QC exports")
print("-" * 100)

print(f"Resolved register    : {review_register_export}")
print(f"Accepted measurements: {accepted_export}")
print(f"Spondias hold        : {spondias_hold_export}")
print(f"Accepted summary     : {accepted_summary_export}")

# -----------------------------------------------------------------------------
# 12. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(accepted_new) == 14

assert set(
    accepted_new["Omo_Canonical_Species"]
) == {
    "Artocarpus communis",
    "Cassia saimen",
    "Diospyrous melistoformis",
    "Diospyrous mespiliformis",
}

assert set(
    held_definition["Omo_Canonical_Species"]
) == {
    "Spondia mombin"
}

print("\nClean fresh alias-rescue evidence resolved successfully.")
print("Four canonical taxa now have newly accepted species-level BIEN evidence.")
print("Spondia mombin remains explicitly held for seed-vs-endocarp definition review.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2I — RESOLVE SPONDIAS SEED-vs-ENDOCARP DEFINITION AMBIGUITY
# =============================================================================
#
# Purpose
# -------
# Resolve the 13 held BIEN measurements for:
#
#     Omo canonical taxon : Spondia mombin
#     Validated name      : Spondias mombin
#
# Standardized trait definition
# -----------------------------
# Individual seed dry mass (mg), where possible.
#
# Evidence interpretation
# -----------------------
# Retain:
#     24.264 mg
#     26.000 mg
#
# These two independent trait-study observations form a coherent low-mass
# cluster compatible with true seed dry mass.
#
# Definition-exclude:
#     remaining 11 measurements
#
# These are NOT declared biologically erroneous. They are excluded from the
# standardized individual-seed estimate because external botanical evidence
# indicates that Spondias datasets may report the lignified endocarp/stone or
# dispersal unit as "seed", which is not equivalent to embryo + endosperm.
#
# No broader Seed Mass integration occurs in this cell.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

SPONDIAS_HOLD_FILE = (
    FT_DIR /
    "omo_seed_mass_spondias_definition_review_hold.csv"
)

assert SPONDIAS_HOLD_FILE.exists(), (
    f"Spondias hold file not found:\n{SPONDIAS_HOLD_FILE}"
)

df = pd.read_csv(
    SPONDIAS_HOLD_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2I — RESOLVE SPONDIAS SEED-vs-ENDOCARP DEFINITION AMBIGUITY")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE INPUT
# -----------------------------------------------------------------------------

assert len(df) == 13, (
    f"Expected 13 held Spondias records, found {len(df)}."
)

assert set(
    df["Omo_Canonical_Species"]
) == {"Spondia mombin"}

assert set(
    df["Validated_BIEN_Query_Name"]
) == {"Spondias mombin"}

df["Trait_Value_mg"] = pd.to_numeric(
    df["Trait_Value_mg"],
    errors="coerce"
)

assert df["Trait_Value_mg"].notna().all()
assert df["Trait_Value_mg"].gt(0).all()

print("\nHeld Spondias evidence")
print("-" * 100)

print(f"Measurements : {len(df)}")
print(
    f"Range        : "
    f"{df['Trait_Value_mg'].min():.3f} - "
    f"{df['Trait_Value_mg'].max():.3f} mg"
)

# -----------------------------------------------------------------------------
# 2. INITIALIZE DEFINITION-REVIEW FIELDS
# -----------------------------------------------------------------------------

df["Definition_Review_Decision"] = pd.NA
df["Evidence_Action"] = pd.NA
df["Measurement_Entity_Class"] = pd.NA
df["Definition_Review_Rationale"] = pd.NA

# -----------------------------------------------------------------------------
# 3. RETAIN TRUE-SEED-COMPATIBLE LOW-MASS EVIDENCE
# -----------------------------------------------------------------------------

true_seed_values = [24.264, 26.000]

true_seed_mask = np.isclose(
    df["Trait_Value_mg"].to_numpy()[:, None],
    np.array(true_seed_values)[None, :],
    rtol=0,
    atol=1e-9
).any(axis=1)

assert true_seed_mask.sum() == 2, (
    "Expected exactly two true-seed-compatible Spondias records."
)

df.loc[
    true_seed_mask,
    "Definition_Review_Decision"
] = "ACCEPT"

df.loc[
    true_seed_mask,
    "Evidence_Action"
] = "RETAIN_FOR_INDIVIDUAL_SEED_MASS"

df.loc[
    true_seed_mask,
    "Measurement_Entity_Class"
] = "TRUE_SEED_COMPATIBLE"

df.loc[
    true_seed_mask,
    "Definition_Review_Rationale"
] = (
    "Retained for standardized individual seed dry mass. "
    "Low-mass independent trait-study evidence is mutually consistent "
    "and compatible with measurements of seed tissue rather than the "
    "multi-seeded lignified endocarp/dispersal unit."
)

# -----------------------------------------------------------------------------
# 4. DEFINITION-EXCLUDE THE REMAINING RECORDS
# -----------------------------------------------------------------------------
#
# IMPORTANT:
# These records are not labelled erroneous.
#
# They are excluded because the measured entity cannot be demonstrated to
# correspond to an individual true seed under the Notebook 07B definition.
# -----------------------------------------------------------------------------

ambiguous_mask = ~true_seed_mask

assert ambiguous_mask.sum() == 11

df.loc[
    ambiguous_mask,
    "Definition_Review_Decision"
] = "EXCLUDE_FROM_STANDARDIZED_SEED_MASS"

df.loc[
    ambiguous_mask,
    "Evidence_Action"
] = "RETAIN_IN_LEDGER_ONLY"

df.loc[
    ambiguous_mask,
    "Measurement_Entity_Class"
] = "SEED_ENDOCARP_DISPERSAL_UNIT_AMBIGUOUS"

df.loc[
    ambiguous_mask,
    "Definition_Review_Rationale"
] = (
    "Not used for standardized individual seed dry mass. "
    "External botanical evidence shows that Spondias endocarps/stones "
    "are multi-seeded dispersal units and are frequently treated as "
    "'seed' in propagation and seed-weight sources. The measurement "
    "entity therefore cannot be assumed to equal one true seed."
)

# -----------------------------------------------------------------------------
# 5. VALIDATE DECISIONS
# -----------------------------------------------------------------------------

assert df["Definition_Review_Decision"].notna().all()
assert df["Evidence_Action"].notna().all()
assert df["Measurement_Entity_Class"].notna().all()

decision_summary = (
    df["Definition_Review_Decision"]
    .value_counts()
    .rename_axis("Definition_Review_Decision")
    .reset_index(name="N_Measurements")
)

print("\nDefinition-review decisions")
print("-" * 100)

display(decision_summary)

assert (
    (df["Definition_Review_Decision"] == "ACCEPT").sum()
    == 2
)

assert (
    (
        df["Definition_Review_Decision"]
        == "EXCLUDE_FROM_STANDARDIZED_SEED_MASS"
    ).sum()
    == 11
)

# -----------------------------------------------------------------------------
# 6. DISPLAY COMPLETE RESOLUTION REGISTER
# -----------------------------------------------------------------------------

display_cols = [
    c for c in [
        "BIEN_Record_ID",
        "Omo_Canonical_Species",
        "Validated_BIEN_Query_Name",
        "Trait_Value_mg",
        "method",
        "source_citation",
        "Definition_Review_Decision",
        "Evidence_Action",
        "Measurement_Entity_Class",
    ]
    if c in df.columns
]

print("\nResolved Spondias evidence register")
print("-" * 100)

display(
    df[display_cols]
    .sort_values("Trait_Value_mg")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 7. EXTRACT ACCEPTED INDIVIDUAL-SEED EVIDENCE
# -----------------------------------------------------------------------------

accepted = (
    df[
        df["Definition_Review_Decision"]
        .eq("ACCEPT")
    ]
    .copy()
    .reset_index(drop=True)
)

definition_excluded = (
    df[
        df["Definition_Review_Decision"]
        .eq("EXCLUDE_FROM_STANDARDIZED_SEED_MASS")
    ]
    .copy()
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 8. CALCULATE PROVISIONAL SPONDIAS SPECIES ESTIMATE
# -----------------------------------------------------------------------------
#
# Same robust aggregation principle used elsewhere:
# median of accepted compatible measurements.
# -----------------------------------------------------------------------------

spondias_estimate = float(
    accepted["Trait_Value_mg"].median()
)

assert np.isclose(
    spondias_estimate,
    25.132
)

species_summary = pd.DataFrame({
    "Canonical_Species": [
        "Spondia mombin"
    ],
    "Validated_Taxonomic_Name": [
        "Spondias mombin"
    ],
    "Trait_Code": [
        "SEED_MASS"
    ],
    "Seed_Mass_mg": [
        spondias_estimate
    ],
    "Standardized_Trait_Unit": [
        "mg"
    ],
    "N_Accepted_Measurements": [
        len(accepted)
    ],
    "N_Definition_Excluded_Measurements": [
        len(definition_excluded)
    ],
    "Min_Accepted_mg": [
        accepted["Trait_Value_mg"].min()
    ],
    "Median_Accepted_mg": [
        accepted["Trait_Value_mg"].median()
    ],
    "Max_Accepted_mg": [
        accepted["Trait_Value_mg"].max()
    ],
    "Aggregation_Method": [
        "MEDIAN_OF_DEFINITION_COMPATIBLE_MEASUREMENTS"
    ],
    "Trait_Source": [
        "SPECIES_BIEN_VALIDATED_ALIAS"
    ],
    "Trait_Quality_Flag": [
        "SEED_ENDOCARP_DEFINITION_REVIEW_COMPLETED"
    ],
    "Evidence_Strength": [
        "MODERATE"
    ],
})

print("\nResolved Spondias individual Seed Mass estimate")
print("-" * 100)

display(species_summary)

# -----------------------------------------------------------------------------
# 9. EXPORT
# -----------------------------------------------------------------------------

register_export = (
    FT_DIR /
    "omo_seed_mass_spondias_definition_review_resolved.csv"
)

accepted_export = (
    FT_DIR /
    "omo_seed_mass_spondias_accepted_true_seed_measurements.csv"
)

excluded_export = (
    FT_DIR /
    "omo_seed_mass_spondias_definition_excluded_measurements.csv"
)

summary_export = (
    FT_DIR /
    "omo_seed_mass_spondias_species_value.csv"
)

df.to_csv(
    register_export,
    index=False
)

accepted.to_csv(
    accepted_export,
    index=False
)

definition_excluded.to_csv(
    excluded_export,
    index=False
)

species_summary.to_csv(
    summary_export,
    index=False
)

print("\nSpondias resolution exports")
print("-" * 100)

print(f"Resolved register : {register_export}")
print(f"Accepted evidence : {accepted_export}")
print(f"Definition-excluded evidence : {excluded_export}")
print(f"Species value     : {summary_export}")

# -----------------------------------------------------------------------------
# 10. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(accepted) == 2
assert len(definition_excluded) == 11

assert accepted["Trait_Value_mg"].isin(
    [24.264, 26.000]
).all()

assert species_summary.loc[
    0,
    "Seed_Mass_mg"
] > 0

print("\nSpondias definition ambiguity resolved successfully.")
print(
    "Final provisional individual Seed Mass estimate: "
    f"{spondias_estimate:.3f} mg"
)
print(
    "Eleven non-comparable/ambiguous dispersal-unit records remain "
    "preserved in the evidence ledger but are not used in the estimate."
)
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.2J — CONSOLIDATE DIRECT + TAXONOMIC-RESCUE BIEN SPECIES VALUES
# =============================================================================
#
# Purpose
# -------
# Consolidate all accepted species-level BIEN Seed Mass evidence obtained
# through:
#
#   A. Direct canonical-name BIEN acquisition        : 15 taxa
#   B. Validated taxonomic-alias rescue              : 13 taxa
#
# Expected combined direct species-level BIEN coverage:
#
#       28 / 127 taxa
#
# Provenance remains explicit:
#
#   SPECIES_BIEN
#   SPECIES_BIEN_VALIDATED_ALIAS
#
# No genus/family estimation occurs.
# No additional external acquisition occurs.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

DIRECT_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_final_species_values.csv"
)

REUSED_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_alias_prior_accepted_reused.csv"
)

NEW_ACCEPTED_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_alias_new_accepted_measurements.csv"
)

SPONDIAS_FILE = (
    FT_DIR /
    "omo_seed_mass_spondias_species_value.csv"
)

# -----------------------------------------------------------------------------
# 1. VALIDATE FILES
# -----------------------------------------------------------------------------

for path in [
    DIRECT_FILE,
    REUSED_FILE,
    NEW_ACCEPTED_FILE,
    SPONDIAS_FILE,
]:
    assert path.exists(), f"Required file not found:\n{path}"

direct = pd.read_csv(
    DIRECT_FILE,
    low_memory=False
)

reused = pd.read_csv(
    REUSED_FILE,
    low_memory=False
)

new_accepted = pd.read_csv(
    NEW_ACCEPTED_FILE,
    low_memory=False
)

spondias = pd.read_csv(
    SPONDIAS_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.2J — CONSOLIDATE DIRECT + TAXONOMIC-RESCUE BIEN SPECIES VALUES")
print("=" * 100)

print("\nInput evidence")
print("-" * 100)

print(f"Direct BIEN species values       : {len(direct)} taxa")
print(f"Prior-reviewed rescue records    : {len(reused)} measurements")
print(f"Fresh accepted rescue records    : {len(new_accepted)} measurements")
print(f"Spondias resolved species values : {len(spondias)} taxa")

assert len(direct) == 15
assert len(reused) == 19
assert len(new_accepted) == 14
assert len(spondias) == 1

# -----------------------------------------------------------------------------
# 2. CONSOLIDATE PRIOR-REVIEWED ALIAS-RESCUE MEASUREMENTS
# -----------------------------------------------------------------------------

reused["Trait_Value_mg"] = pd.to_numeric(
    reused["Trait_Value_mg"],
    errors="coerce"
)

assert reused["Trait_Value_mg"].notna().all()
assert reused["Trait_Value_mg"].gt(0).all()

# Source diversity where available.
if "source_citation" in reused.columns:

    reused["Source_Key"] = (
        reused["source_citation"]
        .fillna(
            reused.get(
                "url_source",
                pd.Series(
                    index=reused.index,
                    dtype="object"
                )
            )
        )
        .fillna("<UNKNOWN_SOURCE>")
        .astype(str)
        .str.strip()
    )

else:

    reused["Source_Key"] = "<UNKNOWN_SOURCE>"

reused_species = (
    reused.groupby(
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
        ],
        dropna=False
    )
    .agg(
        N_Accepted_Measurements=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        N_Sources=(
            "Source_Key",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Seed_Mass_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: float(
                np.exp(
                    np.mean(
                        np.log(
                            x.astype(float)
                        )
                    )
                )
            )
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
    .rename(
        columns={
            "Omo_Canonical_Species":
                "Canonical_Species",
            "Validated_BIEN_Query_Name":
                "Validated_Taxonomic_Name",
        }
    )
)

reused_species["Trait_Code"] = "SEED_MASS"
reused_species["Standardized_Trait_Unit"] = "mg"
reused_species["Trait_Source"] = "SPECIES_BIEN_VALIDATED_ALIAS"

reused_species["Aggregation_Method"] = (
    "MEDIAN_OF_PRIOR_REVIEWED_BIEN_MEASUREMENTS"
)

reused_species["Trait_Quality_Flag"] = (
    "VALIDATED_ALIAS_PRIOR_BIEN_QC_REUSED"
)

reused_species["Evidence_Strength"] = np.select(
    [
        (
            reused_species["N_Accepted_Measurements"] >= 3
        )
        &
        (
            reused_species["N_Sources"] >= 2
        ),

        reused_species[
            "N_Accepted_Measurements"
        ] >= 2,
    ],
    [
        "HIGH",
        "MODERATE",
    ],
    default="LIMITED"
)

print("\nPrior-reviewed alias-rescue species values")
print("-" * 100)

print(
    f"N canonical taxa : "
    f"{reused_species['Canonical_Species'].nunique()}"
)

display(
    reused_species
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

assert (
    reused_species[
        "Canonical_Species"
    ].nunique()
    == 8
)

# -----------------------------------------------------------------------------
# 3. CONSOLIDATE FRESH ACCEPTED ALIAS-RESCUE MEASUREMENTS
# -----------------------------------------------------------------------------

new_accepted["Trait_Value_mg"] = pd.to_numeric(
    new_accepted["Trait_Value_mg"],
    errors="coerce"
)

assert new_accepted["Trait_Value_mg"].notna().all()
assert new_accepted["Trait_Value_mg"].gt(0).all()

if "source_citation" in new_accepted.columns:

    new_accepted["Source_Key"] = (
        new_accepted["source_citation"]
        .fillna(
            new_accepted.get(
                "url_source",
                pd.Series(
                    index=new_accepted.index,
                    dtype="object"
                )
            )
        )
        .fillna("<UNKNOWN_SOURCE>")
        .astype(str)
        .str.strip()
    )

else:

    new_accepted["Source_Key"] = "<UNKNOWN_SOURCE>"

fresh_species = (
    new_accepted.groupby(
        [
            "Omo_Canonical_Species",
            "Validated_BIEN_Query_Name",
        ],
        dropna=False
    )
    .agg(
        N_Accepted_Measurements=(
            "Trait_Value_mg",
            "size"
        ),
        N_Unique_Values=(
            "Trait_Value_mg",
            "nunique"
        ),
        N_Sources=(
            "Source_Key",
            "nunique"
        ),
        Min_mg=(
            "Trait_Value_mg",
            "min"
        ),
        Seed_Mass_mg=(
            "Trait_Value_mg",
            "median"
        ),
        Geometric_Mean_mg=(
            "Trait_Value_mg",
            lambda x: float(
                np.exp(
                    np.mean(
                        np.log(
                            x.astype(float)
                        )
                    )
                )
            )
        ),
        Max_mg=(
            "Trait_Value_mg",
            "max"
        ),
    )
    .reset_index()
    .rename(
        columns={
            "Omo_Canonical_Species":
                "Canonical_Species",
            "Validated_BIEN_Query_Name":
                "Validated_Taxonomic_Name",
        }
    )
)

fresh_species["Trait_Code"] = "SEED_MASS"
fresh_species["Standardized_Trait_Unit"] = "mg"
fresh_species["Trait_Source"] = "SPECIES_BIEN_VALIDATED_ALIAS"

fresh_species["Aggregation_Method"] = (
    "MEDIAN_OF_ACCEPTED_DEDUPLICATED_MEASUREMENTS"
)

fresh_species["Trait_Quality_Flag"] = np.where(
    fresh_species[
        "Canonical_Species"
    ].eq("Artocarpus communis"),
    "VALIDATED_ALIAS_EXTERNALLY_CORROBORATED",
    "VALIDATED_ALIAS_BIEN_QC_PASSED"
)

fresh_species["Evidence_Strength"] = np.select(
    [
        (
            fresh_species["N_Accepted_Measurements"] >= 3
        )
        &
        (
            fresh_species["N_Sources"] >= 2
        ),

        fresh_species[
            "N_Accepted_Measurements"
        ] >= 2,
    ],
    [
        "HIGH",
        "MODERATE",
    ],
    default="LIMITED"
)

print("\nFresh accepted alias-rescue species values")
print("-" * 100)

display(
    fresh_species
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

assert (
    fresh_species[
        "Canonical_Species"
    ].nunique()
    == 4
)

# -----------------------------------------------------------------------------
# 4. STANDARDIZE SPONDIAS RESOLVED VALUE
# -----------------------------------------------------------------------------

assert len(spondias) == 1
assert spondias.loc[0, "Canonical_Species"] == "Spondia mombin"

spondias_species = pd.DataFrame({
    "Canonical_Species": [
        spondias.loc[
            0,
            "Canonical_Species"
        ]
    ],
    "Validated_Taxonomic_Name": [
        spondias.loc[
            0,
            "Validated_Taxonomic_Name"
        ]
    ],
    "Trait_Code": [
        "SEED_MASS"
    ],
    "Seed_Mass_mg": [
        float(
            spondias.loc[
                0,
                "Seed_Mass_mg"
            ]
        )
    ],
    "Standardized_Trait_Unit": [
        "mg"
    ],
    "Trait_Source": [
        "SPECIES_BIEN_VALIDATED_ALIAS"
    ],
    "Aggregation_Method": [
        spondias.loc[
            0,
            "Aggregation_Method"
        ]
    ],
    "N_Accepted_Measurements": [
        int(
            spondias.loc[
                0,
                "N_Accepted_Measurements"
            ]
        )
    ],
    "N_Unique_Values": [
        int(
            spondias.loc[
                0,
                "N_Accepted_Measurements"
            ]
        )
    ],
    "N_Sources": [
        2
    ],
    "Min_mg": [
        float(
            spondias.loc[
                0,
                "Min_Accepted_mg"
            ]
        )
    ],
    "Geometric_Mean_mg": [
        float(
            np.sqrt(
                spondias.loc[
                    0,
                    "Min_Accepted_mg"
                ]
                *
                spondias.loc[
                    0,
                    "Max_Accepted_mg"
                ]
            )
        )
    ],
    "Max_mg": [
        float(
            spondias.loc[
                0,
                "Max_Accepted_mg"
            ]
        )
    ],
    "Trait_Quality_Flag": [
        spondias.loc[
            0,
            "Trait_Quality_Flag"
        ]
    ],
    "Evidence_Strength": [
        spondias.loc[
            0,
            "Evidence_Strength"
        ]
    ],
})

# -----------------------------------------------------------------------------
# 5. COMBINE THE 13 RESCUED CANONICAL TAXA
# -----------------------------------------------------------------------------

rescue_columns = [
    "Canonical_Species",
    "Validated_Taxonomic_Name",
    "Trait_Code",
    "Seed_Mass_mg",
    "Standardized_Trait_Unit",
    "Trait_Source",
    "Aggregation_Method",
    "N_Accepted_Measurements",
    "N_Unique_Values",
    "N_Sources",
    "Min_mg",
    "Geometric_Mean_mg",
    "Max_mg",
    "Trait_Quality_Flag",
    "Evidence_Strength",
]

rescue_species = pd.concat(
    [
        reused_species[rescue_columns],
        fresh_species[rescue_columns],
        spondias_species[rescue_columns],
    ],
    ignore_index=True
)

rescue_species = (
    rescue_species
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nFinal BIEN taxonomic-rescue species table")
print("-" * 100)

print(f"Rows        : {len(rescue_species)}")
print(
    f"Unique taxa : "
    f"{rescue_species['Canonical_Species'].nunique()}"
)

display(rescue_species)

assert len(rescue_species) == 13
assert rescue_species["Canonical_Species"].is_unique

# -----------------------------------------------------------------------------
# 6. STANDARDIZE DIRECT BIEN TABLE FOR COMBINATION
# -----------------------------------------------------------------------------

direct_species = pd.DataFrame({
    "Canonical_Species":
        direct["Canonical_Species"],

    "Validated_Taxonomic_Name":
        direct["Canonical_Species"],

    "Trait_Code":
        direct["Trait_Code"],

    "Seed_Mass_mg":
        pd.to_numeric(
            direct["Seed_Mass_mg"],
            errors="coerce"
        ),

    "Standardized_Trait_Unit":
        direct["Standardized_Trait_Unit"],

    "Trait_Source":
        direct["Trait_Source"],

    "Aggregation_Method":
        direct["Aggregation_Method"],

    "N_Accepted_Measurements":
        direct["N_Accepted_Measurements"],

    "N_Unique_Values":
        direct["N_Unique_Values"],

    "N_Sources":
        direct["N_Sources"],

    "Min_mg":
        direct["Min_mg"],

    "Geometric_Mean_mg":
        direct["Geometric_Mean_mg"],

    "Max_mg":
        direct["Max_mg"],

    "Trait_Quality_Flag":
        direct["Trait_Quality_Flag"],

    "Evidence_Strength":
        direct["Evidence_Strength"],
})

# -----------------------------------------------------------------------------
# 7. CHECK DIRECT-vs-RESCUE CANONICAL OVERLAP
# -----------------------------------------------------------------------------

direct_set = set(
    direct_species[
        "Canonical_Species"
    ]
)

rescue_set = set(
    rescue_species[
        "Canonical_Species"
    ]
)

overlap = sorted(
    direct_set & rescue_set
)

print("\nDirect-vs-rescue canonical overlap")
print("-" * 100)

if overlap:
    print(overlap)
else:
    print("NONE")

assert not overlap, (
    "A canonical taxon appears in both direct and rescue BIEN species tables."
)

# -----------------------------------------------------------------------------
# 8. COMBINE ALL CURRENT BIEN SPECIES-LEVEL VALUES
# -----------------------------------------------------------------------------

bien_species_all = pd.concat(
    [
        direct_species,
        rescue_species,
    ],
    ignore_index=True
)

bien_species_all = (
    bien_species_all
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nCurrent accepted BIEN species-level Seed Mass evidence")
print("-" * 100)

print(f"Rows        : {len(bien_species_all)}")
print(
    f"Unique taxa : "
    f"{bien_species_all['Canonical_Species'].nunique()}"
)

print(
    f"Coverage    : "
    f"{100 * len(bien_species_all) / 127:.2f}%"
)

display(bien_species_all)

# -----------------------------------------------------------------------------
# 9. PROVENANCE SUMMARY
# -----------------------------------------------------------------------------

provenance_summary = (
    bien_species_all[
        "Trait_Source"
    ]
    .value_counts()
    .rename_axis("Trait_Source")
    .reset_index(name="N_Taxa")
)

provenance_summary["Percent_of_127_Taxa"] = (
    100
    *
    provenance_summary["N_Taxa"]
    / 127
)

print("\nCurrent BIEN provenance summary")
print("-" * 100)

display(provenance_summary)

# -----------------------------------------------------------------------------
# 10. EVIDENCE-STRENGTH SUMMARY
# -----------------------------------------------------------------------------

strength_summary = (
    bien_species_all[
        "Evidence_Strength"
    ]
    .value_counts()
    .rename_axis("Evidence_Strength")
    .reset_index(name="N_Taxa")
)

print("\nCurrent BIEN evidence-strength summary")
print("-" * 100)

display(strength_summary)

# -----------------------------------------------------------------------------
# 11. NUMERIC SUMMARY
# -----------------------------------------------------------------------------

print("\nCurrent BIEN species-level Seed Mass numeric summary")
print("-" * 100)

print(
    bien_species_all[
        "Seed_Mass_mg"
    ]
    .describe()
    .to_string()
)

# -----------------------------------------------------------------------------
# 12. EXPORT
# -----------------------------------------------------------------------------

rescue_export = (
    FT_DIR /
    "omo_seed_mass_bien_final_taxonomic_rescue_values.csv"
)

combined_export = (
    FT_DIR /
    "omo_seed_mass_bien_all_species_values_current.csv"
)

provenance_export = (
    FT_DIR /
    "omo_seed_mass_bien_current_provenance_summary.csv"
)

rescue_species.to_csv(
    rescue_export,
    index=False
)

bien_species_all.to_csv(
    combined_export,
    index=False
)

provenance_summary.to_csv(
    provenance_export,
    index=False
)

print("\nConsolidation exports")
print("-" * 100)

print(f"Taxonomic rescue values : {rescue_export}")
print(f"All BIEN species values : {combined_export}")
print(f"Provenance summary      : {provenance_export}")

# -----------------------------------------------------------------------------
# 13. STRONG FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(direct_species) == 15
assert len(rescue_species) == 13

assert len(bien_species_all) == 28

assert (
    bien_species_all[
        "Canonical_Species"
    ].nunique()
    == 28
)

assert (
    bien_species_all[
        "Seed_Mass_mg"
    ].notna()
    .all()
)

assert (
    bien_species_all[
        "Seed_Mass_mg"
    ].gt(0)
    .all()
)

assert (
    bien_species_all[
        "Trait_Code"
    ].eq("SEED_MASS")
    .all()
)

assert (
    bien_species_all[
        "Standardized_Trait_Unit"
    ].eq("mg")
    .all()
)

print("\nBIEN species-level Seed Mass consolidation PASSED.")
print("Direct canonical BIEN taxa       : 15")
print("Validated taxonomic rescue taxa  : 13")
print("Total species-level BIEN coverage: 28 / 127")
print(f"Coverage percentage              : {100 * 28 / 127:.2f}%")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3A — REMAINING SEED MASS TARGETS vs TRY ACCEPTED TAXONOMY
# =============================================================================
#
# Purpose
# -------
# Prepare the next species-level Seed Mass acquisition stage using TRY.
#
# Starting state:
#     Canonical Omo taxa               : 127
#     Species-level BIEN coverage       : 28
#     Remaining taxa                    : 99
#
# Expected residual structure:
#     Species-level                     : 97
#     Genus/unresolved                  : 2
#
# This cell:
#   1. identifies the remaining 99 taxa;
#   2. locates the existing TRY accepted-species catalogue;
#   3. tests canonical and validated alternative names against TRY;
#   4. selects a preferred TRY query name where possible;
#   5. creates a preliminary Seed Mass TRY request manifest.
#
# IMPORTANT:
# No TRY request is submitted here.
# No trait measurements are downloaded.
# No genus/family fallback is performed.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

BIEN_CURRENT_FILE = (
    FT_DIR / "omo_seed_mass_bien_all_species_values_current.csv"
)

SEED_MANIFEST_FILE = (
    FT_DIR / "omo_seed_mass_acquisition_manifest_initial.csv"
)

TRUE_ALIAS_FILE = (
    FT_DIR / "omo_seed_mass_true_taxonomic_species_aliases.csv"
)

assert BIEN_CURRENT_FILE.exists(), (
    f"Current BIEN Seed Mass table not found:\n{BIEN_CURRENT_FILE}"
)

assert SEED_MANIFEST_FILE.exists(), (
    f"Seed Mass acquisition manifest not found:\n{SEED_MANIFEST_FILE}"
)

bien = pd.read_csv(
    BIEN_CURRENT_FILE,
    low_memory=False
)

manifest = pd.read_csv(
    SEED_MANIFEST_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.3A — REMAINING SEED MASS TARGETS vs TRY ACCEPTED TAXONOMY")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE CURRENT BIEN-COVERED STATE
# -----------------------------------------------------------------------------

assert len(manifest) == 127, (
    f"Expected 127 canonical taxa, found {len(manifest)}."
)

assert manifest["Canonical_Species"].is_unique, (
    "Canonical_Species is not unique in the Seed Mass manifest."
)

assert len(bien) == 28, (
    f"Expected 28 current BIEN-covered taxa, found {len(bien)}."
)

assert bien["Canonical_Species"].is_unique, (
    "Canonical_Species is not unique in current BIEN coverage."
)

covered = set(
    bien["Canonical_Species"]
    .dropna()
    .astype(str)
    .str.strip()
)

residual = (
    manifest[
        ~manifest["Canonical_Species"]
        .astype(str)
        .str.strip()
        .isin(covered)
    ]
    .copy()
    .reset_index(drop=True)
)

print("\nCurrent residual Seed Mass targets")
print("-" * 100)

print(f"Canonical taxa             : {len(manifest)}")
print(f"BIEN-covered taxa          : {len(bien)}")
print(f"Remaining taxa             : {len(residual)}")

assert len(residual) == 99, (
    f"Expected 99 remaining taxa, found {len(residual)}."
)

# -----------------------------------------------------------------------------
# 2. VALIDATE TAXONOMIC RESOLUTION
# -----------------------------------------------------------------------------

required_manifest_cols = {
    "Canonical_Species",
    "Taxonomic_Resolution",
}

missing_manifest_cols = (
    required_manifest_cols
    - set(residual.columns)
)

assert not missing_manifest_cols, (
    "Residual manifest missing required fields: "
    f"{sorted(missing_manifest_cols)}"
)

resolution_summary = (
    residual["Taxonomic_Resolution"]
    .value_counts(dropna=False)
    .rename_axis("Taxonomic_Resolution")
    .reset_index(name="N_Taxa")
)

print("\nResidual taxonomic resolution")
print("-" * 100)

display(resolution_summary)

n_species_level = (
    residual["Taxonomic_Resolution"]
    .eq("SPECIES_LEVEL")
    .sum()
)

n_genus_unresolved = (
    residual["Taxonomic_Resolution"]
    .eq("GENUS_LEVEL_OR_UNRESOLVED")
    .sum()
)

assert n_species_level == 97, (
    f"Expected 97 species-level residual taxa, found {n_species_level}."
)

assert n_genus_unresolved == 2, (
    "Expected 2 genus-level/unresolved residual taxa, "
    f"found {n_genus_unresolved}."
)

unresolved = residual[
    residual["Taxonomic_Resolution"]
    .eq("GENUS_LEVEL_OR_UNRESOLVED")
].copy()

assert set(
    unresolved["Canonical_Species"]
) == {
    "Cola species",
    "Drypetes species",
}, (
    "Genus/unresolved residual taxa differ from expected "
    "Cola species and Drypetes species."
)

print("\nGenus/unresolved taxa deferred from TRY species request")
print("-" * 100)

display(
    unresolved[
        [
            "Canonical_Species",
            "Taxonomic_Resolution",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 3. LOCATE EXISTING TRY ACCEPTED-SPECIES CATALOGUE
# -----------------------------------------------------------------------------

possible_try_files = []

for root in [
    PROJECT_ROOT / "data",
    FT_DIR,
    PROJECT_ROOT,
]:

    if not root.exists():
        continue

    for path in root.rglob("TryAccSpecies.txt"):

        if path.is_file():
            possible_try_files.append(path)

possible_try_files = sorted(
    set(possible_try_files)
)

print("\nTRY accepted-species catalogue search")
print("-" * 100)

if not possible_try_files:
    raise FileNotFoundError(
        "TryAccSpecies.txt was not found anywhere under the project root."
    )

for i, path in enumerate(
    possible_try_files,
    start=1
):
    print(f"{i:>2}. {path}")

TRY_SPECIES_FILE = possible_try_files[0]

print("\nSelected TRY species catalogue")
print("-" * 100)
print(TRY_SPECIES_FILE)

# -----------------------------------------------------------------------------
# 4. LOAD TRY ACCEPTED-SPECIES CATALOGUE
# -----------------------------------------------------------------------------
#
# TRY accepted-species catalogues are commonly tab-delimited.
#
# Strategy:
#   1. Try tab-delimited parsing first.
#   2. If this produces only one column, retry using separator inference
#      with the Python engine.
#
# IMPORTANT:
# low_memory=False is NOT used with engine="python".
# -----------------------------------------------------------------------------

try:

    try_species = pd.read_csv(
        TRY_SPECIES_FILE,
        sep="\t",
        low_memory=False
    )

    # If tab parsing failed to separate fields,
    # retry with automatic delimiter inference.
    if len(try_species.columns) == 1:

        try_species = pd.read_csv(
            TRY_SPECIES_FILE,
            sep=None,
            engine="python"
        )

except Exception as e:

    raise RuntimeError(
        "Unable to read TRY accepted-species catalogue.\n\n"
        f"File:\n{TRY_SPECIES_FILE}\n\n"
        f"Original error:\n{e}"
    )

assert not try_species.empty, (
    "TRY accepted-species catalogue loaded but contains no rows."
)

assert len(try_species.columns) > 1, (
    "TRY accepted-species catalogue appears not to have been parsed "
    "into multiple columns."
)

print("\nTRY accepted-species catalogue")
print("-" * 100)

print(f"Rows    : {len(try_species):,}")
print(f"Columns : {len(try_species.columns):,}")

print("\nColumns")

for i, col in enumerate(
    try_species.columns,
    start=1
):
    print(f"{i:>3}. {col}")

# -----------------------------------------------------------------------------
# 5. IDENTIFY TRY ACCEPTED-NAME / ID COLUMNS
# -----------------------------------------------------------------------------

normalized_cols = {
    str(c).strip().lower(): c
    for c in try_species.columns
}

name_candidates = [
    "accspeciesname",
    "acc_species_name",
    "accepted_species_name",
    "speciesname",
    "species_name",
]

id_candidates = [
    "accspeciesid",
    "acc_species_id",
    "accepted_species_id",
    "speciesid",
    "species_id",
]

TRY_NAME_COL = next(
    (
        normalized_cols[name]
        for name in name_candidates
        if name in normalized_cols
    ),
    None
)

TRY_ID_COL = next(
    (
        normalized_cols[name]
        for name in id_candidates
        if name in normalized_cols
    ),
    None
)

assert TRY_NAME_COL is not None, (
    "Could not identify the TRY accepted species-name column "
    "in TryAccSpecies.txt."
)

print("\nDetected TRY taxonomy fields")
print("-" * 100)

print(f"Accepted-name field : {TRY_NAME_COL}")
print(
    f"Accepted-ID field   : "
    f"{TRY_ID_COL if TRY_ID_COL is not None else '<NOT FOUND>'}"
)

# -----------------------------------------------------------------------------
# 6. CLEAN TRY ACCEPTED SPECIES NAMES
# -----------------------------------------------------------------------------

try_species["_TRY_Name_Clean"] = (
    try_species[TRY_NAME_COL]
    .astype("string")
    .str.strip()
)

try_species = try_species[
    try_species["_TRY_Name_Clean"]
    .notna()
    &
    try_species["_TRY_Name_Clean"]
    .ne("")
].copy()

try_name_set = set(
    try_species["_TRY_Name_Clean"]
)

print(
    f"\nUnique TRY accepted names : "
    f"{len(try_name_set):,}"
)

assert len(try_name_set) > 0, (
    "No usable accepted species names were recovered from TryAccSpecies.txt."
)

# -----------------------------------------------------------------------------
# 7. LOAD PREVIOUSLY VALIDATED TRUE TAXONOMIC ALIASES
# -----------------------------------------------------------------------------

if TRUE_ALIAS_FILE.exists():

    true_aliases = pd.read_csv(
        TRUE_ALIAS_FILE,
        low_memory=False
    )

    required_alias_cols = {
        "Canonical_Species",
        "Alternative_Query_Name",
    }

    missing_alias_cols = (
        required_alias_cols
        - set(true_aliases.columns)
    )

    assert not missing_alias_cols, (
        "True taxonomic alias table missing fields: "
        f"{sorted(missing_alias_cols)}"
    )

    true_aliases["Canonical_Species"] = (
        true_aliases["Canonical_Species"]
        .astype("string")
        .str.strip()
    )

    true_aliases["Alternative_Query_Name"] = (
        true_aliases["Alternative_Query_Name"]
        .astype("string")
        .str.strip()
    )

else:

    true_aliases = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Alternative_Query_Name",
        ]
    )

# -----------------------------------------------------------------------------
# 8. RESTRICT TO 97 RESIDUAL SPECIES-LEVEL TAXA
# -----------------------------------------------------------------------------

species_residual = residual[
    residual["Taxonomic_Resolution"]
    .eq("SPECIES_LEVEL")
].copy()

species_residual = (
    species_residual
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

assert len(species_residual) == 97
assert species_residual["Canonical_Species"].is_unique

species_residual_set = set(
    species_residual["Canonical_Species"]
)

true_aliases = true_aliases[
    true_aliases["Canonical_Species"]
    .isin(species_residual_set)
].copy()

print("\nPreviously validated taxonomic aliases available for residual taxa")
print("-" * 100)

print(
    f"Residual species with >=1 validated alias : "
    f"{true_aliases['Canonical_Species'].nunique()}"
)

# -----------------------------------------------------------------------------
# 9. BUILD ALL CANDIDATE TRY QUERY NAMES
# -----------------------------------------------------------------------------

candidate_records = []

for canonical in sorted(
    species_residual_set
):

    # Canonical project name itself.
    candidate_records.append({
        "Canonical_Species": canonical,
        "Candidate_TRY_Name": canonical,
        "Candidate_Name_Type": "CANONICAL",
    })

    # Previously validated taxonomic alternatives.
    alias_values = (
        true_aliases.loc[
            true_aliases["Canonical_Species"]
            .eq(canonical),
            "Alternative_Query_Name"
        ]
        .dropna()
        .astype(str)
        .str.strip()
        .unique()
    )

    for alias in sorted(alias_values):

        if not alias:
            continue

        if alias == canonical:
            continue

        candidate_records.append({
            "Canonical_Species": canonical,
            "Candidate_TRY_Name": alias,
            "Candidate_Name_Type": "VALIDATED_ALIAS",
        })

candidate_table = pd.DataFrame(
    candidate_records
)

assert not candidate_table.empty

candidate_table["Exact_TRY_Match"] = (
    candidate_table["Candidate_TRY_Name"]
    .isin(try_name_set)
)

# -----------------------------------------------------------------------------
# 10. ATTACH TRY ACCEPTED SPECIES IDS
# -----------------------------------------------------------------------------

try_lookup_cols = [
    "_TRY_Name_Clean"
]

if TRY_ID_COL is not None:
    try_lookup_cols.append(
        TRY_ID_COL
    )

try_lookup = (
    try_species[
        try_lookup_cols
    ]
    .drop_duplicates(
        subset=["_TRY_Name_Clean"]
    )
    .copy()
)

candidate_table = candidate_table.merge(
    try_lookup,
    left_on="Candidate_TRY_Name",
    right_on="_TRY_Name_Clean",
    how="left",
    validate="many_to_one"
)

candidate_table = candidate_table.drop(
    columns=["_TRY_Name_Clean"]
)

# -----------------------------------------------------------------------------
# 11. ASSIGN CANDIDATE PRIORITY
# -----------------------------------------------------------------------------
#
# Preference:
#   1. canonical exact match
#   2. validated taxonomic alias exact match
# -----------------------------------------------------------------------------

candidate_table["Candidate_Priority"] = (
    candidate_table[
        "Candidate_Name_Type"
    ]
    .map({
        "CANONICAL": 1,
        "VALIDATED_ALIAS": 2,
    })
)

assert candidate_table[
    "Candidate_Priority"
].notna().all()

# -----------------------------------------------------------------------------
# 12. SELECT PREFERRED EXACT TRY MATCH PER CANONICAL TAXON
# -----------------------------------------------------------------------------

exact_candidates = (
    candidate_table[
        candidate_table["Exact_TRY_Match"]
    ]
    .sort_values(
        [
            "Canonical_Species",
            "Candidate_Priority",
            "Candidate_TRY_Name",
        ]
    )
    .copy()
)

if not exact_candidates.empty:

    best_try_match = (
        exact_candidates
        .groupby(
            "Canonical_Species",
            as_index=False
        )
        .first()
    )

    best_cols = [
        "Canonical_Species",
        "Candidate_TRY_Name",
        "Candidate_Name_Type",
    ]

    if TRY_ID_COL is not None:
        best_cols.append(
            TRY_ID_COL
        )

    best_try_match = best_try_match[
        best_cols
    ].rename(
        columns={
            "Candidate_TRY_Name":
                "Preferred_TRY_Query_Name",

            "Candidate_Name_Type":
                "TRY_Name_Match_Type",

            **(
                {
                    TRY_ID_COL:
                        "TRY_AccSpeciesID"
                }
                if TRY_ID_COL is not None
                else {}
            )
        }
    )

else:

    best_try_match = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Preferred_TRY_Query_Name",
            "TRY_Name_Match_Type",
            "TRY_AccSpeciesID",
        ]
    )

# -----------------------------------------------------------------------------
# 13. BUILD PRELIMINARY TRY SEED MASS REQUEST MANIFEST
# -----------------------------------------------------------------------------

try_manifest = species_residual[
    [
        "Canonical_Species",
        "Taxonomic_Resolution",
    ]
].copy()

try_manifest = try_manifest.merge(
    best_try_match,
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)

# Guarantee ID field exists even if TRY catalogue did not expose one.
if "TRY_AccSpeciesID" not in try_manifest.columns:
    try_manifest["TRY_AccSpeciesID"] = pd.NA

try_manifest["Trait_Code"] = "SEED_MASS"
try_manifest["Standardized_Unit"] = "mg"

try_manifest["TRY_Taxonomic_Status"] = np.where(
    try_manifest[
        "Preferred_TRY_Query_Name"
    ].notna(),
    "EXACT_TRY_ACCEPTED_NAME_AVAILABLE",
    "NO_EXACT_TRY_NAME_MATCH"
)

try_manifest["TRY_Request_Status"] = np.where(
    try_manifest[
        "Preferred_TRY_Query_Name"
    ].notna(),
    "READY_FOR_TRY_SEED_MASS_REQUEST",
    "TRY_TAXONOMIC_REVIEW_REQUIRED"
)

# -----------------------------------------------------------------------------
# 14. SUMMARY
# -----------------------------------------------------------------------------

print("\nTRY taxonomic match summary for 97 residual species-level taxa")
print("-" * 100)

status_summary = (
    try_manifest[
        "TRY_Request_Status"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "TRY_Request_Status"
    )
    .reset_index(name="N_Taxa")
)

display(status_summary)

match_type_summary = (
    try_manifest[
        "TRY_Name_Match_Type"
    ]
    .fillna("<NO_MATCH>")
    .value_counts(dropna=False)
    .rename_axis(
        "TRY_Name_Match_Type"
    )
    .reset_index(name="N_Taxa")
)

print("\nTRY preferred-name match type")
print("-" * 100)

display(match_type_summary)

# -----------------------------------------------------------------------------
# 15. DISPLAY READY TARGETS
# -----------------------------------------------------------------------------

ready = try_manifest[
    try_manifest[
        "TRY_Request_Status"
    ].eq(
        "READY_FOR_TRY_SEED_MASS_REQUEST"
    )
].copy()

print("\nResidual species ready for TRY Seed Mass request")
print("-" * 100)

print(f"N taxa : {len(ready)}")

if ready.empty:

    print("NONE")

else:

    display(
        ready[
            [
                "Canonical_Species",
                "Preferred_TRY_Query_Name",
                "TRY_Name_Match_Type",
                "TRY_AccSpeciesID",
                "TRY_Request_Status",
            ]
        ]
        .sort_values(
            "Canonical_Species"
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 16. DISPLAY TRY TAXONOMIC GAPS
# -----------------------------------------------------------------------------

try_gaps = try_manifest[
    try_manifest[
        "TRY_Request_Status"
    ].eq(
        "TRY_TAXONOMIC_REVIEW_REQUIRED"
    )
].copy()

print("\nResidual species without an exact TRY accepted-name match")
print("-" * 100)

print(f"N taxa : {len(try_gaps)}")

if try_gaps.empty:

    print("NONE")

else:

    display(
        try_gaps[
            [
                "Canonical_Species",
                "TRY_Request_Status",
            ]
        ]
        .sort_values(
            "Canonical_Species"
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 17. ADD TWO GENUS/UNRESOLVED TAXA TO OVERALL STATUS
# -----------------------------------------------------------------------------

genus_deferred = unresolved[
    [
        "Canonical_Species",
        "Taxonomic_Resolution",
    ]
].copy()

genus_deferred["Trait_Code"] = "SEED_MASS"
genus_deferred["Standardized_Unit"] = "mg"

genus_deferred["TRY_Request_Status"] = (
    "DEFER_TO_GENUS_LEVEL_STAGE"
)

print("\nDeferred genus/unresolved targets")
print("-" * 100)

display(
    genus_deferred
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 18. OVERALL RESIDUAL TRY PREPARATION SUMMARY
# -----------------------------------------------------------------------------

overall_summary = pd.DataFrame({
    "Category": [
        "Species-level residual taxa",
        "Ready for TRY Seed Mass request",
        "TRY taxonomic review required",
        "Genus/unresolved deferred",
        "Total residual taxa",
    ],
    "N_Taxa": [
        len(try_manifest),
        len(ready),
        len(try_gaps),
        len(genus_deferred),
        len(try_manifest) + len(genus_deferred),
    ],
})

print("\nOverall TRY Seed Mass preparation summary")
print("-" * 100)

display(overall_summary)

# -----------------------------------------------------------------------------
# 19. EXPORT
# -----------------------------------------------------------------------------

candidate_export = (
    FT_DIR /
    "omo_seed_mass_try_residual_taxonomic_candidates.csv"
)

manifest_export = (
    FT_DIR /
    "omo_seed_mass_try_species_request_manifest_preliminary.csv"
)

ready_export = (
    FT_DIR /
    "omo_seed_mass_try_species_request_ready.csv"
)

gap_export = (
    FT_DIR /
    "omo_seed_mass_try_taxonomic_review_required.csv"
)

deferred_export = (
    FT_DIR /
    "omo_seed_mass_try_genus_deferred.csv"
)

candidate_table.to_csv(
    candidate_export,
    index=False
)

try_manifest.to_csv(
    manifest_export,
    index=False
)

ready.to_csv(
    ready_export,
    index=False
)

try_gaps.to_csv(
    gap_export,
    index=False
)

genus_deferred.to_csv(
    deferred_export,
    index=False
)

print("\nTRY preparation exports")
print("-" * 100)

print(f"Candidate names    : {candidate_export}")
print(f"Request manifest   : {manifest_export}")
print(f"Ready request taxa : {ready_export}")
print(f"Taxonomic gaps     : {gap_export}")
print(f"Genus deferred     : {deferred_export}")

# -----------------------------------------------------------------------------
# 20. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(try_manifest) == 97

assert try_manifest[
    "Canonical_Species"
].is_unique

assert (
    len(ready)
    + len(try_gaps)
    == 97
)

assert (
    len(try_manifest)
    + len(genus_deferred)
    == 99
)

assert set(
    genus_deferred[
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Drypetes species",
}

assert (
    try_manifest[
        "Trait_Code"
    ].eq("SEED_MASS")
    .all()
)

assert (
    try_manifest[
        "Standardized_Unit"
    ].eq("mg")
    .all()
)

print("\nTRY Seed Mass residual taxonomic audit completed successfully.")
print("No TRY Seed Mass request has yet been submitted.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3B — CONTROLLED RESOLUTION OF RESIDUAL TRY TAXONOMIC GAPS
# =============================================================================
#
# Purpose
# -------
# Resolve the 10 species-level taxa that remained unmatched against the
# TRY accepted-species catalogue in Section 9.3A.
#
# Strategy
# --------
# 1. Start from the 10 validated TRY taxonomic gaps.
# 2. Search existing WD/SLA functional-trait outputs for previously
#    established taxonomic normalizations / accepted names.
# 3. Retain only scientifically plausible species-level alternative names.
# 4. Test those names against TryAccSpecies.txt.
# 5. Separate:
#
#       RESOLVED_FROM_EXISTING_VALIDATED_EVIDENCE
#       STILL_REQUIRES_TAXONOMIC_REVIEW
#
# IMPORTANT
# ---------
# - Canonical_Species remains unchanged.
# - No new synonymy is invented.
# - Congeneric fallback species are NOT accepted as taxonomic aliases.
# - No TRY request is submitted in this cell.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

TRY_GAP_FILE = (
    FT_DIR /
    "omo_seed_mass_try_taxonomic_review_required.csv"
)

TRY_PRELIM_MANIFEST_FILE = (
    FT_DIR /
    "omo_seed_mass_try_species_request_manifest_preliminary.csv"
)

TRY_SPECIES_FILE = (
    PROJECT_ROOT /
    "data" /
    "external" /
    "TRY" /
    "TryAccSpecies.txt"
)

assert TRY_GAP_FILE.exists(), (
    f"TRY gap file not found:\n{TRY_GAP_FILE}"
)

assert TRY_PRELIM_MANIFEST_FILE.exists(), (
    f"TRY preliminary manifest not found:\n{TRY_PRELIM_MANIFEST_FILE}"
)

assert TRY_SPECIES_FILE.exists(), (
    f"TRY accepted-species catalogue not found:\n{TRY_SPECIES_FILE}"
)

gaps = pd.read_csv(
    TRY_GAP_FILE,
    low_memory=False
)

prelim = pd.read_csv(
    TRY_PRELIM_MANIFEST_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.3B — CONTROLLED RESOLUTION OF RESIDUAL TRY TAXONOMIC GAPS")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE THE 10 EXPECTED GAP TAXA
# -----------------------------------------------------------------------------

expected_gaps = {
    "Albizia auxilary",
    "Amphimas tetracoides",
    "Aningeria robusta",
    "Citrus sinenses",
    "Diospyros undebunda",
    "Drypetes",
    "Homalium alymeri",
    "Lonchocarpus cilliata",
    "Mitragyna cilliata",
    "Musa sapiensis",
}

assert len(gaps) == 10, (
    f"Expected 10 TRY taxonomic gaps, found {len(gaps)}."
)

assert set(
    gaps["Canonical_Species"]
) == expected_gaps, (
    "TRY gap taxa differ from the validated Section 9.3A result."
)

print("\nTRY taxonomic gaps entering controlled review")
print("-" * 100)

display(
    gaps[
        [
            "Canonical_Species",
            "TRY_Request_Status",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 2. LOAD TRY ACCEPTED-SPECIES CATALOGUE
# -----------------------------------------------------------------------------

try:
    try_species = pd.read_csv(
        TRY_SPECIES_FILE,
        sep="\t",
        low_memory=False
    )

    if len(try_species.columns) == 1:
        try_species = pd.read_csv(
            TRY_SPECIES_FILE,
            sep=None,
            engine="python"
        )

except Exception as e:
    raise RuntimeError(
        f"Unable to read TRY catalogue:\n{TRY_SPECIES_FILE}\n\n{e}"
    )

assert {
    "AccSpeciesID",
    "AccSpeciesName"
}.issubset(try_species.columns)

try_species["AccSpeciesName"] = (
    try_species["AccSpeciesName"]
    .astype("string")
    .str.strip()
)

try_lookup = (
    try_species[
        [
            "AccSpeciesID",
            "AccSpeciesName",
        ]
    ]
    .dropna(subset=["AccSpeciesName"])
    .drop_duplicates(
        subset=["AccSpeciesName"]
    )
    .copy()
)

try_name_set = set(
    try_lookup["AccSpeciesName"]
)

# -----------------------------------------------------------------------------
# 3. IDENTIFY EXISTING WD/SLA TABLES THAT MAY CONTAIN VALIDATED NAMES
# -----------------------------------------------------------------------------

candidate_files = []

for path in FT_DIR.glob("*.csv"):

    lower_name = path.name.lower()

    if not (
        lower_name.startswith("omo_sla")
        or lower_name.startswith("omo_wood")
        or lower_name.startswith("omo_seed_mass_true_taxonomic")
    ):
        continue

    # Explicitly avoid congeneric fallback tables.
    if "congeneric" in lower_name:
        continue

    try:
        preview = pd.read_csv(
            path,
            nrows=5,
            low_memory=False
        )
    except Exception:
        continue

    if "Canonical_Species" in preview.columns:
        candidate_files.append(path)

candidate_files = sorted(
    set(candidate_files)
)

print("\nExisting validated taxonomic evidence tables considered")
print("-" * 100)

print(f"N files : {len(candidate_files)}")

# -----------------------------------------------------------------------------
# 4. FIELDS THAT MAY REPRESENT TRUE TAXONOMIC NORMALIZATION
# -----------------------------------------------------------------------------

positive_field_keywords = [
    "accepted",
    "normalized",
    "resolved",
    "corrected",
    "replacement",
    "harmonized",
    "recovered_normalized",
    "trait_source_name",
    "alternative_query_name",
]

negative_field_keywords = [
    "genus",
    "family",
    "congeneric",
    "fallback",
    "estimate",
    "provenance",
    "status",
    "flag",
    "level",
]

# -----------------------------------------------------------------------------
# 5. SCIENTIFIC-NAME VALIDATOR
# -----------------------------------------------------------------------------

def looks_like_species_name(value):
    """
    Accept:
        Genus species
        Genus x species
        Genus × species

    Reject:
        genus-only strings
        status labels
        numeric/provenance text
    """

    if pd.isna(value):
        return False

    value = str(value).strip()

    if not value:
        return False

    if value.lower().startswith("http"):
        return False

    if any(ch.isdigit() for ch in value):
        return False

    parts = value.split()

    if len(parts) == 2:

        genus, epithet = parts

        return (
            len(genus) > 1
            and genus[0].isupper()
            and genus.replace("-", "").isalpha()
            and len(epithet) > 1
            and epithet[0].islower()
            and epithet.replace("-", "").isalpha()
        )

    if len(parts) == 3:

        genus, marker, epithet = parts

        return (
            marker.lower() in {"x", "×"}
            and len(genus) > 1
            and genus[0].isupper()
            and genus.replace("-", "").isalpha()
            and len(epithet) > 1
            and epithet[0].islower()
            and epithet.replace("-", "").isalpha()
        )

    return False

# -----------------------------------------------------------------------------
# 6. HARVEST PREVIOUSLY VALIDATED TAXONOMIC ALTERNATIVES
# -----------------------------------------------------------------------------

gap_set = set(
    gaps["Canonical_Species"]
    .astype(str)
    .str.strip()
)

evidence_records = []

for path in candidate_files:

    try:
        df = pd.read_csv(
            path,
            low_memory=False
        )
    except Exception:
        continue

    if "Canonical_Species" not in df.columns:
        continue

    df["Canonical_Species"] = (
        df["Canonical_Species"]
        .astype("string")
        .str.strip()
    )

    df = df[
        df["Canonical_Species"]
        .isin(gap_set)
    ].copy()

    if df.empty:
        continue

    for col in df.columns:

        if col == "Canonical_Species":
            continue

        col_lower = str(col).lower()

        # Require positive taxonomic signal.
        if not any(
            k in col_lower
            for k in positive_field_keywords
        ):
            continue

        # Reject fallback / genus / estimate fields.
        if any(
            k in col_lower
            for k in negative_field_keywords
        ):
            continue

        for _, row in df.iterrows():

            canonical = row["Canonical_Species"]
            value = row.get(col)

            if not looks_like_species_name(value):
                continue

            value = str(value).strip()

            if value == canonical:
                continue

            evidence_records.append({
                "Canonical_Species": canonical,
                "Candidate_Validated_Name": value,
                "Evidence_Field": col,
                "Evidence_Source_File": path.name,
            })

existing_candidates = pd.DataFrame(
    evidence_records
)

if existing_candidates.empty:

    existing_candidates = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Candidate_Validated_Name",
            "Evidence_Field",
            "Evidence_Source_File",
        ]
    )

else:

    existing_candidates = (
        existing_candidates
        .drop_duplicates(
            subset=[
                "Canonical_Species",
                "Candidate_Validated_Name",
            ]
        )
        .sort_values(
            [
                "Canonical_Species",
                "Candidate_Validated_Name",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 7. ADD SPECIFIC VALIDATED DECISIONS ALREADY ESTABLISHED IN NOTEBOOK 07B
# -----------------------------------------------------------------------------
#
# These mappings were previously established during the WD/SLA taxonomic
# harmonization workflow.
#
# They are NOT newly inferred here.
# -----------------------------------------------------------------------------

previously_validated = pd.DataFrame(
    [
        {
            "Canonical_Species": "Aningeria robusta",
            "Candidate_Validated_Name": "Aningeria pierrei",
            "Evidence_Field": "PRIOR_NOTEBOOK_07B_VALIDATION",
            "Evidence_Source_File": "WD_SLA_VALIDATED_TAXONOMY",
        },
        {
            "Canonical_Species": "Citrus sinenses",
            "Candidate_Validated_Name": "Citrus sinensis",
            "Evidence_Field": "PRIOR_NOTEBOOK_07B_VALIDATION",
            "Evidence_Source_File": "WD_SLA_VALIDATED_TAXONOMY",
        },
        {
            "Canonical_Species": "Mitragyna cilliata",
            "Candidate_Validated_Name": "Mitragyna ciliata",
            "Evidence_Field": "PRIOR_NOTEBOOK_07B_VALIDATION",
            "Evidence_Source_File": "WD_SLA_VALIDATED_TAXONOMY",
        },
        {
            "Canonical_Species": "Musa sapiensis",
            "Candidate_Validated_Name": "Musa x paradisiaca",
            "Evidence_Field": "PRIOR_NOTEBOOK_07B_VALIDATION",
            "Evidence_Source_File": "SLA_VALIDATED_TAXONOMY",
        },
    ]
)

all_candidates = pd.concat(
    [
        existing_candidates,
        previously_validated,
    ],
    ignore_index=True
)

all_candidates = (
    all_candidates
    .drop_duplicates(
        subset=[
            "Canonical_Species",
            "Candidate_Validated_Name",
        ]
    )
    .sort_values(
        [
            "Canonical_Species",
            "Candidate_Validated_Name",
        ]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 8. TEST EACH VALIDATED CANDIDATE AGAINST TRY
# -----------------------------------------------------------------------------

all_candidates["Exact_TRY_Match"] = (
    all_candidates[
        "Candidate_Validated_Name"
    ].isin(try_name_set)
)

all_candidates = all_candidates.merge(
    try_lookup,
    left_on="Candidate_Validated_Name",
    right_on="AccSpeciesName",
    how="left",
    validate="many_to_one"
)

all_candidates = all_candidates.drop(
    columns=["AccSpeciesName"]
)

print("\nPreviously validated alternative names tested against TRY")
print("-" * 100)

if all_candidates.empty:
    print("NONE")
else:
    display(
        all_candidates
        .sort_values(
            [
                "Canonical_Species",
                "Candidate_Validated_Name",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 9. SELECT EXACT TRY MATCHES
# -----------------------------------------------------------------------------

resolved_candidates = (
    all_candidates[
        all_candidates["Exact_TRY_Match"]
    ]
    .copy()
)

# Prefer an explicitly prior-validated Notebook 07B decision when multiple
# candidates happen to survive.
resolved_candidates["Evidence_Priority"] = np.where(
    resolved_candidates[
        "Evidence_Field"
    ].eq("PRIOR_NOTEBOOK_07B_VALIDATION"),
    1,
    2
)

resolved_best = (
    resolved_candidates
    .sort_values(
        [
            "Canonical_Species",
            "Evidence_Priority",
            "Candidate_Validated_Name",
        ]
    )
    .groupby(
        "Canonical_Species",
        as_index=False
    )
    .first()
)

resolved_best = resolved_best[
    [
        "Canonical_Species",
        "Candidate_Validated_Name",
        "AccSpeciesID",
        "Evidence_Field",
        "Evidence_Source_File",
    ]
].rename(
    columns={
        "Candidate_Validated_Name":
            "Resolved_TRY_Query_Name",
        "AccSpeciesID":
            "TRY_AccSpeciesID",
        "Evidence_Field":
            "Resolution_Evidence_Field",
        "Evidence_Source_File":
            "Resolution_Evidence_Source",
    }
)

# -----------------------------------------------------------------------------
# 10. BUILD THE 10-TAXON RESOLUTION REGISTER
# -----------------------------------------------------------------------------

review_register = gaps[
    [
        "Canonical_Species",
    ]
].copy()

review_register = review_register.merge(
    resolved_best,
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)

review_register["Resolution_Status"] = np.where(
    review_register[
        "Resolved_TRY_Query_Name"
    ].notna(),
    "RESOLVED_FROM_EXISTING_VALIDATED_EVIDENCE",
    "STILL_REQUIRES_TAXONOMIC_REVIEW"
)

review_register["TRY_Request_Status"] = np.where(
    review_register[
        "Resolved_TRY_Query_Name"
    ].notna(),
    "READY_FOR_TRY_SEED_MASS_REQUEST",
    "TRY_TAXONOMIC_REVIEW_REQUIRED"
)

print("\nControlled TRY gap-resolution register")
print("-" * 100)

display(
    review_register
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 11. SUMMARY
# -----------------------------------------------------------------------------

resolution_summary = (
    review_register[
        "Resolution_Status"
    ]
    .value_counts()
    .rename_axis("Resolution_Status")
    .reset_index(name="N_Taxa")
)

print("\nResolution summary")
print("-" * 100)

display(resolution_summary)

resolved_taxa = review_register[
    review_register[
        "Resolution_Status"
    ].eq(
        "RESOLVED_FROM_EXISTING_VALIDATED_EVIDENCE"
    )
].copy()

still_unresolved = review_register[
    review_register[
        "Resolution_Status"
    ].eq(
        "STILL_REQUIRES_TAXONOMIC_REVIEW"
    )
].copy()

print("\nResolved TRY Seed Mass taxa")
print("-" * 100)

print(f"N taxa : {len(resolved_taxa)}")

if resolved_taxa.empty:
    print("NONE")
else:
    display(
        resolved_taxa[
            [
                "Canonical_Species",
                "Resolved_TRY_Query_Name",
                "TRY_AccSpeciesID",
                "Resolution_Evidence_Source",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

print("\nTaxa still requiring taxonomic review")
print("-" * 100)

print(f"N taxa : {len(still_unresolved)}")

if still_unresolved.empty:
    print("NONE")
else:
    display(
        still_unresolved[
            [
                "Canonical_Species",
                "Resolution_Status",
                "TRY_Request_Status",
            ]
        ]
        .sort_values("Canonical_Species")
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 12. UPDATE THE PRELIMINARY 97-TAXON TRY MANIFEST
# -----------------------------------------------------------------------------

updated_manifest = prelim.copy()

assert len(updated_manifest) == 97
assert updated_manifest["Canonical_Species"].is_unique

for _, row in resolved_taxa.iterrows():

    mask = (
        updated_manifest[
            "Canonical_Species"
        ].eq(
            row["Canonical_Species"]
        )
    )

    assert mask.sum() == 1

    updated_manifest.loc[
        mask,
        "Preferred_TRY_Query_Name"
    ] = row["Resolved_TRY_Query_Name"]

    updated_manifest.loc[
        mask,
        "TRY_Name_Match_Type"
    ] = "VALIDATED_ALIAS"

    updated_manifest.loc[
        mask,
        "TRY_AccSpeciesID"
    ] = row["TRY_AccSpeciesID"]

    updated_manifest.loc[
        mask,
        "TRY_Taxonomic_Status"
    ] = "EXACT_TRY_ACCEPTED_NAME_AVAILABLE"

    updated_manifest.loc[
        mask,
        "TRY_Request_Status"
    ] = "READY_FOR_TRY_SEED_MASS_REQUEST"

# -----------------------------------------------------------------------------
# 13. UPDATED OVERALL TRY-READY COUNTS
# -----------------------------------------------------------------------------

updated_summary = (
    updated_manifest[
        "TRY_Request_Status"
    ]
    .value_counts()
    .rename_axis("TRY_Request_Status")
    .reset_index(name="N_Taxa")
)

print("\nUpdated TRY species request status")
print("-" * 100)

display(updated_summary)

updated_ready = updated_manifest[
    updated_manifest[
        "TRY_Request_Status"
    ].eq(
        "READY_FOR_TRY_SEED_MASS_REQUEST"
    )
].copy()

updated_gaps = updated_manifest[
    updated_manifest[
        "TRY_Request_Status"
    ].eq(
        "TRY_TAXONOMIC_REVIEW_REQUIRED"
    )
].copy()

print(
    f"\nTRY-ready species-level taxa : "
    f"{len(updated_ready)} / 97"
)

print(
    f"Still requiring review       : "
    f"{len(updated_gaps)} / 97"
)

# -----------------------------------------------------------------------------
# 14. EXPORT
# -----------------------------------------------------------------------------

candidate_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_validated_name_candidates.csv"
)

review_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_resolution_register.csv"
)

resolved_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_resolved.csv"
)

still_unresolved_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_still_unresolved.csv"
)

updated_manifest_export = (
    FT_DIR /
    "omo_seed_mass_try_species_request_manifest_after_gap_resolution.csv"
)

updated_ready_export = (
    FT_DIR /
    "omo_seed_mass_try_species_request_ready_after_gap_resolution.csv"
)

all_candidates.to_csv(
    candidate_export,
    index=False
)

review_register.to_csv(
    review_export,
    index=False
)

resolved_taxa.to_csv(
    resolved_export,
    index=False
)

still_unresolved.to_csv(
    still_unresolved_export,
    index=False
)

updated_manifest.to_csv(
    updated_manifest_export,
    index=False
)

updated_ready.to_csv(
    updated_ready_export,
    index=False
)

print("\nTRY taxonomic-resolution exports")
print("-" * 100)

print(f"Validated candidates : {candidate_export}")
print(f"Resolution register  : {review_export}")
print(f"Resolved taxa        : {resolved_export}")
print(f"Still unresolved     : {still_unresolved_export}")
print(f"Updated manifest     : {updated_manifest_export}")
print(f"Updated ready list   : {updated_ready_export}")

# -----------------------------------------------------------------------------
# 15. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(review_register) == 10

assert (
    len(resolved_taxa)
    + len(still_unresolved)
    == 10
)

assert len(updated_manifest) == 97

assert (
    len(updated_ready)
    + len(updated_gaps)
    == 97
)

assert updated_ready[
    "Preferred_TRY_Query_Name"
].notna().all()

assert updated_ready[
    "TRY_AccSpeciesID"
].notna().all()

print("\nControlled TRY taxonomic gap resolution completed successfully.")
print("No new taxonomic synonymy was invented.")
print("No TRY Seed Mass request has yet been submitted.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3C — GENUS-CONSTRAINED TRY CANDIDATE DIAGNOSIS
# =============================================================================
#
# Purpose
# -------
# Diagnose the 10 residual TRY taxonomic gaps by examining the nearest
# accepted TRY names without automatically accepting any mapping.
#
# Strategy
# --------
# For each unresolved Omo taxon:
#
#   1. identify its stated genus;
#   2. retrieve all TRY accepted species within that genus;
#   3. rank candidates by string similarity;
#   4. report the top genus-constrained candidates;
#   5. also report the closest global TRY name for diagnostic purposes;
#   6. preserve external-taxonomy candidate names already established
#      during Notebook 07B, but DO NOT automatically accept them.
#
# IMPORTANT
# ---------
# This is a diagnostic/review step only.
#
# No new taxonomic mapping is accepted.
# No TRY request is submitted.
# No trait measurements are downloaded.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
from difflib import SequenceMatcher

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

TRY_SPECIES_FILE = (
    PROJECT_ROOT /
    "data" /
    "external" /
    "TRY" /
    "TryAccSpecies.txt"
)

UNRESOLVED_FILE = (
    FT_DIR /
    "omo_seed_mass_try_gap_still_unresolved.csv"
)

assert TRY_SPECIES_FILE.exists(), (
    f"TRY catalogue not found:\n{TRY_SPECIES_FILE}"
)

assert UNRESOLVED_FILE.exists(), (
    f"Unresolved TRY gap file not found:\n{UNRESOLVED_FILE}"
)

print("=" * 100)
print("SECTION 9.3C — GENUS-CONSTRAINED TRY CANDIDATE DIAGNOSIS")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. LOAD THE 10 UNRESOLVED TAXA
# -----------------------------------------------------------------------------

unresolved = pd.read_csv(
    UNRESOLVED_FILE,
    low_memory=False
)

expected_taxa = {
    "Albizia auxilary",
    "Amphimas tetracoides",
    "Aningeria robusta",
    "Citrus sinenses",
    "Diospyros undebunda",
    "Drypetes",
    "Homalium alymeri",
    "Lonchocarpus cilliata",
    "Mitragyna cilliata",
    "Musa sapiensis",
}

assert len(unresolved) == 10

assert set(
    unresolved["Canonical_Species"]
) == expected_taxa

print("\nResidual TRY taxonomic gaps")
print("-" * 100)

display(
    unresolved[
        [
            "Canonical_Species",
            "Resolution_Status",
            "TRY_Request_Status",
        ]
    ]
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 2. LOAD TRY ACCEPTED-SPECIES CATALOGUE
# -----------------------------------------------------------------------------

try:
    try_species = pd.read_csv(
        TRY_SPECIES_FILE,
        sep="\t",
        low_memory=False
    )

    if len(try_species.columns) == 1:
        try_species = pd.read_csv(
            TRY_SPECIES_FILE,
            sep=None,
            engine="python"
        )

except Exception as e:

    raise RuntimeError(
        f"Unable to load TRY catalogue:\n{e}"
    )

required_try_cols = {
    "AccSpeciesID",
    "AccSpeciesName",
}

assert required_try_cols.issubset(
    try_species.columns
)

try_species = try_species[
    [
        "AccSpeciesID",
        "AccSpeciesName",
    ]
].copy()

try_species["AccSpeciesName"] = (
    try_species["AccSpeciesName"]
    .astype("string")
    .str.strip()
)

try_species = try_species[
    try_species["AccSpeciesName"]
    .notna()
    &
    try_species["AccSpeciesName"]
    .ne("")
].copy()

try_species = (
    try_species
    .drop_duplicates(
        subset=["AccSpeciesName"]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 3. EXTRACT GENUS FROM TRY ACCEPTED NAMES
# -----------------------------------------------------------------------------

try_species["TRY_Genus"] = (
    try_species["AccSpeciesName"]
    .str.split()
    .str[0]
)

# -----------------------------------------------------------------------------
# 4. STRING-SIMILARITY FUNCTION
# -----------------------------------------------------------------------------

def similarity(a, b):
    """
    SequenceMatcher ratio on lowercase normalized text.
    """

    a = " ".join(
        str(a).lower().strip().split()
    )

    b = " ".join(
        str(b).lower().strip().split()
    )

    return SequenceMatcher(
        None,
        a,
        b
    ).ratio()


# -----------------------------------------------------------------------------
# 5. EXTERNAL / PRIOR NOTEBOOK TAXONOMIC CANDIDATES
# -----------------------------------------------------------------------------
#
# These are review candidates, NOT automatic resolutions.
#
# Values below are based on previously validated Notebook 07B decisions
# and/or authoritative taxonomic review already undertaken.
# -----------------------------------------------------------------------------

external_review_candidates = {
    "Albizia auxilary": [],
    "Amphimas tetracoides": [],
    "Aningeria robusta": [
        "Aningeria pierrei",
    ],
    "Citrus sinenses": [
        "Citrus sinensis",
        "Citrus x sinensis",
        "Citrus × sinensis",
        "Citrus aurantium",
    ],
    "Diospyros undebunda": [],
    "Drypetes": [],
    "Homalium alymeri": [],
    "Lonchocarpus cilliata": [],
    "Mitragyna cilliata": [
        "Mitragyna ciliata",
        "Hallea ciliata",
    ],
    "Musa sapiensis": [
        "Musa sapientum",
        "Musa x sapientum",
        "Musa × sapientum",
        "Musa x paradisiaca",
        "Musa × paradisiaca",
    ],
}

# -----------------------------------------------------------------------------
# 6. GENERATE GENUS-CONSTRAINED CANDIDATES
# -----------------------------------------------------------------------------

candidate_rows = []

TOP_N = 10

for canonical in sorted(
    expected_taxa
):

    tokens = canonical.split()

    stated_genus = (
        tokens[0]
        if len(tokens) >= 1
        else ""
    )

    # -------------------------------------------------------------------------
    # SAME-GENUS TRY CANDIDATES
    # -------------------------------------------------------------------------

    genus_pool = try_species[
        try_species["TRY_Genus"]
        .eq(stated_genus)
    ].copy()

    if not genus_pool.empty:

        genus_pool["Similarity"] = (
            genus_pool[
                "AccSpeciesName"
            ]
            .apply(
                lambda x: similarity(
                    canonical,
                    x
                )
            )
        )

        genus_pool = (
            genus_pool
            .sort_values(
                [
                    "Similarity",
                    "AccSpeciesName",
                ],
                ascending=[
                    False,
                    True,
                ]
            )
            .head(TOP_N)
        )

        for rank, (_, row) in enumerate(
            genus_pool.iterrows(),
            start=1
        ):

            candidate_rows.append({
                "Canonical_Species":
                    canonical,

                "Stated_Genus":
                    stated_genus,

                "Candidate_Source":
                    "TRY_SAME_GENUS",

                "Candidate_Rank":
                    rank,

                "Candidate_Name":
                    row["AccSpeciesName"],

                "TRY_AccSpeciesID":
                    row["AccSpeciesID"],

                "Similarity":
                    row["Similarity"],

                "Exact_TRY_Name":
                    True,
            })

    # -------------------------------------------------------------------------
    # EXTERNAL / NOTEBOOK REVIEW CANDIDATES
    # -------------------------------------------------------------------------

    for candidate in external_review_candidates.get(
        canonical,
        []
    ):

        exact_match = try_species[
            try_species[
                "AccSpeciesName"
            ].eq(candidate)
        ]

        if len(exact_match) == 1:

            try_id = (
                exact_match[
                    "AccSpeciesID"
                ].iloc[0]
            )

            exact_try = True

        else:

            try_id = pd.NA
            exact_try = False

        candidate_rows.append({
            "Canonical_Species":
                canonical,

            "Stated_Genus":
                stated_genus,

            "Candidate_Source":
                "EXTERNAL_OR_PRIOR_VALIDATED",

            "Candidate_Rank":
                pd.NA,

            "Candidate_Name":
                candidate,

            "TRY_AccSpeciesID":
                try_id,

            "Similarity":
                similarity(
                    canonical,
                    candidate
                ),

            "Exact_TRY_Name":
                exact_try,
        })

candidate_table = pd.DataFrame(
    candidate_rows
)

# -----------------------------------------------------------------------------
# 7. GLOBAL CLOSEST TRY NAME
# -----------------------------------------------------------------------------
#
# Used only when the same-genus search may fail because the genus itself
# could be misspelled or obsolete.
# -----------------------------------------------------------------------------

global_rows = []

for canonical in sorted(
    expected_taxa
):

    scores = (
        try_species[
            "AccSpeciesName"
        ]
        .apply(
            lambda x: similarity(
                canonical,
                x
            )
        )
    )

    best_indices = (
        scores
        .nlargest(5)
        .index
    )

    for rank, idx in enumerate(
        best_indices,
        start=1
    ):

        row = try_species.loc[idx]

        global_rows.append({
            "Canonical_Species":
                canonical,

            "Candidate_Source":
                "TRY_GLOBAL",

            "Candidate_Rank":
                rank,

            "Candidate_Name":
                row["AccSpeciesName"],

            "TRY_AccSpeciesID":
                row["AccSpeciesID"],

            "Similarity":
                scores.loc[idx],
        })

global_candidates = pd.DataFrame(
    global_rows
)

# -----------------------------------------------------------------------------
# 8. DIAGNOSTIC SUMMARY PER TAXON
# -----------------------------------------------------------------------------

diagnostic_rows = []

for canonical in sorted(
    expected_taxa
):

    tokens = canonical.split()
    genus = tokens[0]

    genus_n = int(
        try_species[
            "TRY_Genus"
        ]
        .eq(genus)
        .sum()
    )

    same_genus = candidate_table[
        (
            candidate_table[
                "Canonical_Species"
            ].eq(canonical)
        )
        &
        (
            candidate_table[
                "Candidate_Source"
            ].eq("TRY_SAME_GENUS")
        )
    ]

    if not same_genus.empty:

        top_genus_name = (
            same_genus
            .sort_values(
                "Candidate_Rank"
            )
            .iloc[0][
                "Candidate_Name"
            ]
        )

        top_genus_score = (
            same_genus
            .sort_values(
                "Candidate_Rank"
            )
            .iloc[0][
                "Similarity"
            ]
        )

    else:

        top_genus_name = pd.NA
        top_genus_score = np.nan

    ext = candidate_table[
        (
            candidate_table[
                "Canonical_Species"
            ].eq(canonical)
        )
        &
        (
            candidate_table[
                "Candidate_Source"
            ].eq(
                "EXTERNAL_OR_PRIOR_VALIDATED"
            )
        )
    ]

    external_exact_try = (
        ext[
            ext[
                "Exact_TRY_Name"
            ].eq(True)
        ][
            "Candidate_Name"
        ]
        .tolist()
    )

    diagnostic_rows.append({
        "Canonical_Species":
            canonical,

        "Stated_Genus":
            genus,

        "N_TRY_Species_In_Stated_Genus":
            genus_n,

        "Top_Same_Genus_TRY_Candidate":
            top_genus_name,

        "Top_Same_Genus_Similarity":
            top_genus_score,

        "External_Candidates_Exact_In_TRY":
            (
                " | ".join(
                    external_exact_try
                )
                if external_exact_try
                else ""
            ),
    })

diagnostic_summary = pd.DataFrame(
    diagnostic_rows
)

# -----------------------------------------------------------------------------
# 9. DISPLAY SUMMARY
# -----------------------------------------------------------------------------

print("\nTRY candidate diagnostic summary")
print("-" * 100)

display(
    diagnostic_summary
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 10. DISPLAY SAME-GENUS TOP CANDIDATES
# -----------------------------------------------------------------------------

print("\nTop TRY same-genus candidates")
print("-" * 100)

same_genus_display = candidate_table[
    candidate_table[
        "Candidate_Source"
    ].eq("TRY_SAME_GENUS")
].copy()

if same_genus_display.empty:

    print("NONE")

else:

    display(
        same_genus_display[
            [
                "Canonical_Species",
                "Candidate_Rank",
                "Candidate_Name",
                "TRY_AccSpeciesID",
                "Similarity",
            ]
        ]
        .sort_values(
            [
                "Canonical_Species",
                "Candidate_Rank",
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 11. DISPLAY EXTERNAL / PRIOR VALIDATED CANDIDATES
# -----------------------------------------------------------------------------

print("\nExternal / prior-validated candidates tested against TRY")
print("-" * 100)

external_display = candidate_table[
    candidate_table[
        "Candidate_Source"
    ].eq(
        "EXTERNAL_OR_PRIOR_VALIDATED"
    )
].copy()

if external_display.empty:

    print("NONE")

else:

    display(
        external_display[
            [
                "Canonical_Species",
                "Candidate_Name",
                "TRY_AccSpeciesID",
                "Similarity",
                "Exact_TRY_Name",
            ]
        ]
        .sort_values(
            [
                "Canonical_Species",
                "Exact_TRY_Name",
                "Similarity",
            ],
            ascending=[
                True,
                False,
                False,
            ]
        )
        .reset_index(drop=True)
    )

# -----------------------------------------------------------------------------
# 12. DISPLAY GLOBAL TOP-5 MATCHES
# -----------------------------------------------------------------------------

print("\nTop 5 global TRY candidates for each unresolved taxon")
print("-" * 100)

display(
    global_candidates
    .sort_values(
        [
            "Canonical_Species",
            "Candidate_Rank",
        ]
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 13. INITIAL REVIEW ROUTING
# -----------------------------------------------------------------------------
#
# This does NOT resolve names.
#
# It simply identifies which cases have an externally supported candidate
# that also exists exactly in TRY.
# -----------------------------------------------------------------------------

review_routing = diagnostic_summary.copy()

review_routing["Review_Route"] = np.where(
    review_routing[
        "External_Candidates_Exact_In_TRY"
    ].ne(""),
    "AUTHORITATIVE_CANDIDATE_AVAILABLE_IN_TRY",
    "MANUAL_TAXONOMIC_REVIEW_REQUIRED"
)

# A genus-only source string must not be forced into a species-level request.
review_routing.loc[
    review_routing[
        "Canonical_Species"
    ].eq("Drypetes"),
    "Review_Route"
] = "GENUS_ONLY_STRING_REQUIRES_SPECIAL_REVIEW"

print("\nTaxonomic review routing")
print("-" * 100)

display(
    review_routing
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)

print("\nReview-route summary")
print("-" * 100)

display(
    review_routing[
        "Review_Route"
    ]
    .value_counts()
    .rename_axis(
        "Review_Route"
    )
    .reset_index(
        name="N_Taxa"
    )
)

# -----------------------------------------------------------------------------
# 14. EXPORT DIAGNOSTICS
# -----------------------------------------------------------------------------

candidate_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_genus_candidate_diagnostics.csv"
)

global_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_global_candidate_diagnostics.csv"
)

summary_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_candidate_summary.csv"
)

routing_export = (
    FT_DIR /
    "omo_seed_mass_try_gap_review_routing.csv"
)

candidate_table.to_csv(
    candidate_export,
    index=False
)

global_candidates.to_csv(
    global_export,
    index=False
)

diagnostic_summary.to_csv(
    summary_export,
    index=False
)

review_routing.to_csv(
    routing_export,
    index=False
)

print("\nDiagnostic exports")
print("-" * 100)

print(f"Genus candidates : {candidate_export}")
print(f"Global candidates: {global_export}")
print(f"Summary          : {summary_export}")
print(f"Review routing   : {routing_export}")

# -----------------------------------------------------------------------------
# 15. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(diagnostic_summary) == 10
assert len(review_routing) == 10

assert set(
    diagnostic_summary[
        "Canonical_Species"
    ]
) == expected_taxa

assert (
    review_routing.loc[
        review_routing[
            "Canonical_Species"
        ].eq("Drypetes"),
        "Review_Route"
    ]
    .eq(
        "GENUS_ONLY_STRING_REQUIRES_SPECIAL_REVIEW"
    )
    .all()
)

print("\nTRY residual taxonomic candidate diagnosis completed successfully.")
print("No new taxonomic mapping has been accepted.")
print("No TRY Seed Mass request has yet been submitted.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3D — FREEZE FINAL TRY SEED MASS REQUEST AND RESIDUAL ROUTING
# =============================================================================
#
# Purpose
# -------
# Close the TRY taxonomic-preparation phase and freeze a defensible
# species-level Seed Mass request manifest.
#
# Current state
# -------------
# Total canonical taxa                         : 127
# Already covered by BIEN                     : 28
# Residual taxa                               : 99
#
# TRY routing:
#   Clean species-level TRY request            : 87
#   Species-level non-TRY rescue               : 9
#   Genus-only acquisition stage               : 3
#
# The three genus-stage targets are:
#   - Cola species
#   - Drypetes species
#   - Drypetes
#
# NOTE:
# "Drypetes" remains unchanged as a canonical project identifier.
# We are only overriding its ACQUISITION QUERY LEVEL because the string
# itself is genus-only.
#
# No TRY request is submitted in this cell.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

BIEN_FILE = (
    FT_DIR /
    "omo_seed_mass_bien_all_species_values_current.csv"
)

TRY_MANIFEST_FILE = (
    FT_DIR /
    "omo_seed_mass_try_species_request_manifest_after_gap_resolution.csv"
)

TRY_GAP_FILE = (
    FT_DIR /
    "omo_seed_mass_try_gap_still_unresolved.csv"
)

GENUS_DEFERRED_FILE = (
    FT_DIR /
    "omo_seed_mass_try_genus_deferred.csv"
)

for path in [
    BIEN_FILE,
    TRY_MANIFEST_FILE,
    TRY_GAP_FILE,
    GENUS_DEFERRED_FILE,
]:
    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )

bien = pd.read_csv(
    BIEN_FILE,
    low_memory=False
)

try_manifest = pd.read_csv(
    TRY_MANIFEST_FILE,
    low_memory=False
)

try_gaps = pd.read_csv(
    TRY_GAP_FILE,
    low_memory=False
)

genus_deferred_original = pd.read_csv(
    GENUS_DEFERRED_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.3D — FREEZE FINAL TRY SEED MASS REQUEST AND RESIDUAL ROUTING")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE STARTING STATE
# -----------------------------------------------------------------------------

assert len(bien) == 28
assert bien["Canonical_Species"].is_unique

assert len(try_manifest) == 97
assert try_manifest["Canonical_Species"].is_unique

assert len(try_gaps) == 10

assert set(
    genus_deferred_original["Canonical_Species"]
) == {
    "Cola species",
    "Drypetes species",
}

print("\nStarting coverage")
print("-" * 100)

print(f"BIEN species-level coverage : {len(bien)}")
print(f"Residual species manifest   : {len(try_manifest)}")
print(f"Unresolved TRY gaps         : {len(try_gaps)}")
print(f"Original genus deferred     : {len(genus_deferred_original)}")

# -----------------------------------------------------------------------------
# 2. DEFINE ACQUISITION-LEVEL OVERRIDE FOR GENUS-ONLY 'Drypetes'
# -----------------------------------------------------------------------------
#
# This does NOT change Canonical_Species or the validated project backbone.
#
# It only prevents a genus-only string from entering a species-level TRY
# request as though it were a binomial.
# -----------------------------------------------------------------------------

genus_only_override = {
    "Drypetes"
}

assert genus_only_override.issubset(
    set(
        try_manifest["Canonical_Species"]
    )
)

# -----------------------------------------------------------------------------
# 3. BUILD FINAL CLEAN TRY SPECIES REQUEST
# -----------------------------------------------------------------------------

final_try_request = try_manifest[
    try_manifest[
        "TRY_Request_Status"
    ].eq(
        "READY_FOR_TRY_SEED_MASS_REQUEST"
    )
].copy()

# Explicitly remove any acquisition-level genus-only override,
# should one ever enter the ready table.
final_try_request = final_try_request[
    ~final_try_request[
        "Canonical_Species"
    ].isin(genus_only_override)
].copy()

final_try_request = (
    final_try_request
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nFinal clean TRY species-level request")
print("-" * 100)

print(f"N canonical taxa : {len(final_try_request)}")

assert len(final_try_request) == 87

assert final_try_request[
    "Preferred_TRY_Query_Name"
].notna().all()

assert final_try_request[
    "TRY_AccSpeciesID"
].notna().all()

assert final_try_request[
    "Canonical_Species"
].is_unique

# -----------------------------------------------------------------------------
# 4. VALIDATE QUERY-NAME STRUCTURE
# -----------------------------------------------------------------------------

def looks_species_level(name):

    if pd.isna(name):
        return False

    parts = str(name).strip().split()

    # Ordinary binomial.
    if len(parts) == 2:
        return True

    # Simple hybrid name:
    # Genus x epithet / Genus × epithet
    if (
        len(parts) == 3
        and parts[1].lower() in {"x", "×"}
    ):
        return True

    return False


final_try_request[
    "Species_Level_Query_Name_Valid"
] = (
    final_try_request[
        "Preferred_TRY_Query_Name"
    ]
    .apply(looks_species_level)
)

invalid_query_names = final_try_request[
    ~final_try_request[
        "Species_Level_Query_Name_Valid"
    ]
].copy()

print("\nSpecies-level TRY query-name validation")
print("-" * 100)

if invalid_query_names.empty:
    print("All 87 TRY query names have species-level structure.")
else:
    display(
        invalid_query_names[
            [
                "Canonical_Species",
                "Preferred_TRY_Query_Name",
            ]
        ]
    )

assert invalid_query_names.empty

# -----------------------------------------------------------------------------
# 5. SPECIES-LEVEL TAXA DEFERRED TO NON-TRY RESCUE
# -----------------------------------------------------------------------------
#
# Remove "Drypetes" from the ten TRY gaps because its canonical string itself
# is genus-only and belongs in the genus acquisition stage.
# -----------------------------------------------------------------------------

species_non_try_rescue = try_gaps[
    ~try_gaps[
        "Canonical_Species"
    ].isin(genus_only_override)
].copy()

species_non_try_rescue[
    "Seed_Mass_Acquisition_Route"
] = "TARGETED_SPECIES_NON_TRY_RESCUE"

species_non_try_rescue[
    "Route_Rationale"
] = (
    "No defensible exact TRY accepted-name mapping available; "
    "retain as species-level target for authoritative database "
    "or literature rescue before any genus/family fallback."
)

species_non_try_rescue = (
    species_non_try_rescue
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nSpecies-level targets deferred to non-TRY rescue")
print("-" * 100)

print(f"N taxa : {len(species_non_try_rescue)}")

display(
    species_non_try_rescue[
        [
            "Canonical_Species",
            "Seed_Mass_Acquisition_Route",
        ]
    ]
)

assert len(species_non_try_rescue) == 9

# -----------------------------------------------------------------------------
# 6. BUILD FINAL GENUS-LEVEL ACQUISITION ROUTE
# -----------------------------------------------------------------------------

genus_stage_records = []

# Existing two validated genus/unresolved project taxa.
for _, row in genus_deferred_original.iterrows():

    genus_stage_records.append({
        "Canonical_Species":
            row["Canonical_Species"],

        "Original_Taxonomic_Resolution":
            row.get(
                "Taxonomic_Resolution",
                pd.NA
            ),

        "Seed_Mass_Acquisition_Level":
            "GENUS",

        "Seed_Mass_Acquisition_Route":
            "DEFER_TO_GENUS_LEVEL_STAGE",

        "Route_Rationale":
            "Canonical project taxon is genus-level/unresolved.",
    })

# Acquisition override for canonical string "Drypetes".
genus_stage_records.append({
    "Canonical_Species":
        "Drypetes",

    "Original_Taxonomic_Resolution":
        (
            try_manifest.loc[
                try_manifest[
                    "Canonical_Species"
                ].eq("Drypetes"),
                "Taxonomic_Resolution"
            ]
            .iloc[0]
        ),

    "Seed_Mass_Acquisition_Level":
        "GENUS",

    "Seed_Mass_Acquisition_Route":
        "DEFER_TO_GENUS_LEVEL_STAGE",

    "Route_Rationale":
        (
            "Canonical identifier is a genus-only string; "
            "acquisition-level override prevents treatment as "
            "direct species evidence."
        ),
})

genus_stage = pd.DataFrame(
    genus_stage_records
)

genus_stage = (
    genus_stage
    .drop_duplicates(
        subset=["Canonical_Species"]
    )
    .sort_values("Canonical_Species")
    .reset_index(drop=True)
)

print("\nFinal genus-level Seed Mass acquisition targets")
print("-" * 100)

display(genus_stage)

assert len(genus_stage) == 3

assert set(
    genus_stage["Canonical_Species"]
) == {
    "Cola species",
    "Drypetes",
    "Drypetes species",
}

# -----------------------------------------------------------------------------
# 7. FINAL TRY REQUEST ID LIST
# -----------------------------------------------------------------------------

try_request_ids = (
    final_try_request[
        "TRY_AccSpeciesID"
    ]
    .dropna()
    .astype(int)
    .astype(str)
    .tolist()
)

n_unique_try_ids = len(
    set(try_request_ids)
)

print("\nTRY request identifier summary")
print("-" * 100)

print(f"Canonical Omo taxa in request : {len(final_try_request)}")
print(f"Unique TRY AccSpeciesIDs      : {n_unique_try_ids}")

# Several Omo spelling variants can legitimately map to the same
# accepted TRY species, so unique IDs may be fewer than 87.

# -----------------------------------------------------------------------------
# 8. BUILD UNIQUE TRY QUERY TABLE
# -----------------------------------------------------------------------------

unique_try_queries = (
    final_try_request[
        [
            "Preferred_TRY_Query_Name",
            "TRY_AccSpeciesID",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "Preferred_TRY_Query_Name"
    )
    .reset_index(drop=True)
)

unique_try_queries[
    "TRY_AccSpeciesID"
] = (
    unique_try_queries[
        "TRY_AccSpeciesID"
    ]
    .astype(int)
)

print("\nUnique TRY accepted species to request")
print("-" * 100)

print(
    f"N unique accepted TRY taxa : "
    f"{len(unique_try_queries)}"
)

display(unique_try_queries)

assert len(
    unique_try_queries
) == n_unique_try_ids

# -----------------------------------------------------------------------------
# 9. CREATE REQUEST-ID STRING
# -----------------------------------------------------------------------------
#
# This is convenient for copying into the TRY request interface if it accepts
# AccSpeciesID values.
# -----------------------------------------------------------------------------

try_id_string = ",".join(
    unique_try_queries[
        "TRY_AccSpeciesID"
    ]
    .astype(str)
    .tolist()
)

id_text_export = (
    FT_DIR /
    "omo_seed_mass_try_final_accspecies_ids.txt"
)

with open(
    id_text_export,
    "w",
    encoding="utf-8"
) as f:

    f.write(try_id_string)

# -----------------------------------------------------------------------------
# 10. ACCOUNT FOR ALL 127 CANONICAL TAXA
# -----------------------------------------------------------------------------

bien_set = set(
    bien["Canonical_Species"]
)

try_set = set(
    final_try_request["Canonical_Species"]
)

non_try_set = set(
    species_non_try_rescue["Canonical_Species"]
)

genus_set = set(
    genus_stage["Canonical_Species"]
)

# All four acquisition categories must be mutually exclusive.
assert not (bien_set & try_set)
assert not (bien_set & non_try_set)
assert not (bien_set & genus_set)
assert not (try_set & non_try_set)
assert not (try_set & genus_set)
assert not (non_try_set & genus_set)

all_accounted = (
    bien_set
    | try_set
    | non_try_set
    | genus_set
)

print("\nFinal Seed Mass acquisition routing")
print("-" * 100)

routing_summary = pd.DataFrame({
    "Acquisition_Category": [
        "SPECIES_BIEN_RESOLVED",
        "SPECIES_TRY_REQUEST",
        "SPECIES_NON_TRY_RESCUE",
        "GENUS_LEVEL_STAGE",
    ],
    "N_Taxa": [
        len(bien_set),
        len(try_set),
        len(non_try_set),
        len(genus_set),
    ],
})

routing_summary[
    "Percent_of_127"
] = (
    100
    *
    routing_summary["N_Taxa"]
    / 127
)

display(routing_summary)

print(
    f"\nTotal canonical taxa accounted for : "
    f"{len(all_accounted)}"
)

assert len(all_accounted) == 127

assert (
    len(bien_set)
    + len(try_set)
    + len(non_try_set)
    + len(genus_set)
    == 127
)

# -----------------------------------------------------------------------------
# 11. EXPORT FINAL TRY PREPARATION TABLES
# -----------------------------------------------------------------------------

request_export = (
    FT_DIR /
    "omo_seed_mass_try_final_species_request_manifest.csv"
)

unique_query_export = (
    FT_DIR /
    "omo_seed_mass_try_final_unique_species_queries.csv"
)

non_try_export = (
    FT_DIR /
    "omo_seed_mass_species_non_try_rescue_targets.csv"
)

genus_export = (
    FT_DIR /
    "omo_seed_mass_genus_stage_targets.csv"
)

routing_export = (
    FT_DIR /
    "omo_seed_mass_acquisition_routing_after_try_taxonomy.csv"
)

final_try_request.to_csv(
    request_export,
    index=False
)

unique_try_queries.to_csv(
    unique_query_export,
    index=False
)

species_non_try_rescue.to_csv(
    non_try_export,
    index=False
)

genus_stage.to_csv(
    genus_export,
    index=False
)

routing_summary.to_csv(
    routing_export,
    index=False
)

print("\nFinal TRY preparation exports")
print("-" * 100)

print(f"Final species request : {request_export}")
print(f"Unique TRY queries    : {unique_query_export}")
print(f"TRY AccSpeciesIDs     : {id_text_export}")
print(f"Non-TRY rescue targets: {non_try_export}")
print(f"Genus-stage targets   : {genus_export}")
print(f"Acquisition routing   : {routing_export}")

# -----------------------------------------------------------------------------
# 12. FINAL VALIDATION
# -----------------------------------------------------------------------------

assert len(final_try_request) == 87
assert len(species_non_try_rescue) == 9
assert len(genus_stage) == 3

assert (
    final_try_request[
        "Preferred_TRY_Query_Name"
    ]
    .notna()
    .all()
)

assert (
    final_try_request[
        "TRY_AccSpeciesID"
    ]
    .notna()
    .all()
)

print("\nFinal TRY Seed Mass request manifest frozen successfully.")
print("Species-level TRY request taxa : 87")
print("Species-level non-TRY rescue   : 9")
print("Genus-stage acquisition taxa   : 3")
print("All 127 canonical Omo taxa are accounted for.")
print("No TRY request has yet been submitted.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3E — PREPARE FINAL TRY SEED DRY MASS REQUEST PACKAGE
# =============================================================================
#
# Purpose
# -------
# Create the final submission-ready package for a NEW TRY request for:
#
#       TraitID 26 — Seed dry mass
#
# Current request scope:
#       Omo canonical taxa represented : 87
#       Unique TRY accepted species     : 64
#
# The request will use the unique TRY AccSpeciesIDs so that spelling variants
# in the Omo canonical backbone do not generate redundant TRY species queries.
#
# IMPORTANT
# ---------
# This cell DOES NOT submit the TRY request.
# Submission will be performed manually through the TRY web interface.
# =============================================================================

from pathlib import Path
import pandas as pd
import textwrap

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

FT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

FINAL_REQUEST_FILE = (
    FT_DIR /
    "omo_seed_mass_try_final_species_request_manifest.csv"
)

UNIQUE_QUERY_FILE = (
    FT_DIR /
    "omo_seed_mass_try_final_unique_species_queries.csv"
)

assert FINAL_REQUEST_FILE.exists(), (
    f"Final TRY request manifest not found:\n{FINAL_REQUEST_FILE}"
)

assert UNIQUE_QUERY_FILE.exists(), (
    f"Unique TRY query table not found:\n{UNIQUE_QUERY_FILE}"
)

request_manifest = pd.read_csv(
    FINAL_REQUEST_FILE,
    low_memory=False
)

unique_queries = pd.read_csv(
    UNIQUE_QUERY_FILE,
    low_memory=False
)

print("=" * 100)
print("SECTION 9.3E — PREPARE FINAL TRY SEED DRY MASS REQUEST PACKAGE")
print("=" * 100)

# -----------------------------------------------------------------------------
# 1. VALIDATE FROZEN REQUEST STATE
# -----------------------------------------------------------------------------

assert len(request_manifest) == 87
assert request_manifest["Canonical_Species"].is_unique

assert len(unique_queries) == 64
assert unique_queries["TRY_AccSpeciesID"].notna().all()
assert unique_queries["Preferred_TRY_Query_Name"].notna().all()

unique_queries["TRY_AccSpeciesID"] = (
    pd.to_numeric(
        unique_queries["TRY_AccSpeciesID"],
        errors="raise"
    )
    .astype(int)
)

assert unique_queries["TRY_AccSpeciesID"].is_unique

print("\nFrozen TRY request state")
print("-" * 100)

print(
    f"Omo canonical taxa represented : "
    f"{len(request_manifest)}"
)

print(
    f"Unique TRY accepted species     : "
    f"{len(unique_queries)}"
)

# -----------------------------------------------------------------------------
# 2. DEFINE TRY TRAIT
# -----------------------------------------------------------------------------

TRY_TRAIT_ID = 26
TRY_TRAIT_NAME = "Seed dry mass"

print("\nTRY trait requested")
print("-" * 100)

print(f"Trait ID   : {TRY_TRAIT_ID}")
print(f"Trait name : {TRY_TRAIT_NAME}")

# -----------------------------------------------------------------------------
# 3. PREPARE UNIQUE SPECIES ID LIST
# -----------------------------------------------------------------------------

species_ids = (
    unique_queries[
        "TRY_AccSpeciesID"
    ]
    .astype(str)
    .tolist()
)

# TRY interface may accept pasted IDs as line-separated or comma-separated.
species_ids_lines = "\n".join(
    species_ids
)

species_ids_commas = ",".join(
    species_ids
)

assert len(species_ids) == 64

# -----------------------------------------------------------------------------
# 4. PREPARE TRAIT ID TEXT
# -----------------------------------------------------------------------------

trait_id_text = str(
    TRY_TRAIT_ID
)

# -----------------------------------------------------------------------------
# 5. PREPARE REQUEST TITLE
# -----------------------------------------------------------------------------

request_title = (
    "Omo Forest Ecological Informatics: "
    "Seed Dry Mass Functional Trait Harmonization"
)

# -----------------------------------------------------------------------------
# 6. PREPARE PROJECT DESCRIPTION
# -----------------------------------------------------------------------------

project_description = textwrap.dedent(
    """
    This request supports the Omo Forest Ecological Informatics Project,
    which is developing a harmonized functional-trait dataset for 127
    canonical woody taxa recorded in the Omo Forest/Biosphere Reserve,
    Nigeria.

    The present request targets Seed dry mass (TRY TraitID 26) for the
    remaining species-level taxa not already resolved from other accepted
    trait sources. Taxonomic names have been harmonized against the TRY
    accepted-species catalogue before submission.

    Seed mass will be standardized to milligrams per individual dry seed
    where compatible with the TRY trait definition. Raw records will undergo
    taxonomic validation, unit checking, duplicate detection, measurement-
    level quality control, provenance tracking and robust species-level
    aggregation before use.

    The resulting trait dataset will support functional-diversity,
    community-ecology and biodiversity-informatics analyses of the Omo
    Forest woody flora.
    """
).strip()

# -----------------------------------------------------------------------------
# 7. REQUEST DATA-ACCESS MODE
# -----------------------------------------------------------------------------

data_access_mode = "PUBLIC DATA ONLY"

# -----------------------------------------------------------------------------
# 8. PREPARE REQUEST SUMMARY TABLE
# -----------------------------------------------------------------------------

request_summary = pd.DataFrame({
    "Request_Field": [
        "Trait_ID",
        "Trait_Name",
        "Unique_TRY_Species",
        "Omo_Canonical_Taxa_Represented",
        "Requested_Access",
        "Request_Title",
    ],
    "Value": [
        TRY_TRAIT_ID,
        TRY_TRAIT_NAME,
        len(unique_queries),
        len(request_manifest),
        data_access_mode,
        request_title,
    ],
})

print("\nTRY Seed Mass request summary")
print("-" * 100)

display(request_summary)

# -----------------------------------------------------------------------------
# 9. PREPARE SPECIES CROSSWALK FOR REQUEST PROVENANCE
# -----------------------------------------------------------------------------

crosswalk = request_manifest[
    [
        "Canonical_Species",
        "Preferred_TRY_Query_Name",
        "TRY_Name_Match_Type",
        "TRY_AccSpeciesID",
    ]
].copy()

crosswalk["TRY_AccSpeciesID"] = (
    pd.to_numeric(
        crosswalk["TRY_AccSpeciesID"],
        errors="raise"
    )
    .astype(int)
)

crosswalk = (
    crosswalk
    .sort_values(
        [
            "Preferred_TRY_Query_Name",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)

print("\nOmo → TRY request crosswalk")
print("-" * 100)

display(crosswalk)

# -----------------------------------------------------------------------------
# 10. DISPLAY UNIQUE TRY SPECIES REQUEST LIST
# -----------------------------------------------------------------------------

print("\nUnique TRY accepted species included in request")
print("-" * 100)

display(
    unique_queries
    .sort_values(
        "Preferred_TRY_Query_Name"
    )
    .reset_index(drop=True)
)

# -----------------------------------------------------------------------------
# 11. EXPORT SUBMISSION MATERIALS
# -----------------------------------------------------------------------------

trait_id_export = (
    FT_DIR /
    "omo_seed_mass_try_request_trait_id.txt"
)

species_ids_lines_export = (
    FT_DIR /
    "omo_seed_mass_try_request_species_ids_lines.txt"
)

species_ids_commas_export = (
    FT_DIR /
    "omo_seed_mass_try_request_species_ids_commas.txt"
)

title_export = (
    FT_DIR /
    "omo_seed_mass_try_request_title.txt"
)

description_export = (
    FT_DIR /
    "omo_seed_mass_try_request_project_description.txt"
)

summary_export = (
    FT_DIR /
    "omo_seed_mass_try_request_submission_summary.csv"
)

crosswalk_export = (
    FT_DIR /
    "omo_seed_mass_try_request_species_crosswalk.csv"
)

with open(
    trait_id_export,
    "w",
    encoding="utf-8"
) as f:
    f.write(trait_id_text)

with open(
    species_ids_lines_export,
    "w",
    encoding="utf-8"
) as f:
    f.write(species_ids_lines)

with open(
    species_ids_commas_export,
    "w",
    encoding="utf-8"
) as f:
    f.write(species_ids_commas)

with open(
    title_export,
    "w",
    encoding="utf-8"
) as f:
    f.write(request_title)

with open(
    description_export,
    "w",
    encoding="utf-8"
) as f:
    f.write(project_description)

request_summary.to_csv(
    summary_export,
    index=False
)

crosswalk.to_csv(
    crosswalk_export,
    index=False
)

print("\nSubmission-package exports")
print("-" * 100)

print(f"Trait ID              : {trait_id_export}")
print(f"Species IDs — lines   : {species_ids_lines_export}")
print(f"Species IDs — commas  : {species_ids_commas_export}")
print(f"Request title         : {title_export}")
print(f"Project description   : {description_export}")
print(f"Submission summary    : {summary_export}")
print(f"Species crosswalk     : {crosswalk_export}")

# -----------------------------------------------------------------------------
# 12. DISPLAY COPY-READY REQUEST INFORMATION
# -----------------------------------------------------------------------------

print("\n" + "=" * 100)
print("COPY-READY TRY REQUEST INFORMATION")
print("=" * 100)

print("\nTRAIT")
print("-" * 100)

print(
    f"{TRY_TRAIT_ID} — {TRY_TRAIT_NAME}"
)

print("\nREQUEST TITLE")
print("-" * 100)

print(request_title)

print("\nDATA ACCESS")
print("-" * 100)

print(data_access_mode)

print("\nPROJECT DESCRIPTION")
print("-" * 100)

print(project_description)

print("\nTRY ACCEPTED SPECIES IDs — COMMA-SEPARATED")
print("-" * 100)

print(species_ids_commas)

# -----------------------------------------------------------------------------
# 13. STRONG FINAL VALIDATION
# -----------------------------------------------------------------------------

assert TRY_TRAIT_ID == 26
assert TRY_TRAIT_NAME == "Seed dry mass"

assert len(species_ids) == 64

assert len(
    set(species_ids)
) == 64

assert len(crosswalk) == 87

assert crosswalk[
    "TRY_AccSpeciesID"
].notna().all()

print("\nFinal TRY Seed Dry Mass submission package prepared successfully.")
print("Trait requested                : 26 — Seed dry mass")
print("Unique TRY species IDs         : 64")
print("Omo canonical taxa represented : 87")
print("Requested access               : PUBLIC DATA ONLY")
print("No TRY request has yet been submitted.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3F — TRY REQUEST 51911 RAW-DATA INTEGRITY,
#                SEED MASS EXTRACTION AND ASSOCIATED-METADATA AUDIT
# =============================================================================
#
# Purpose
# -------
# 1. Load the downloaded TRY Request 51911 file.
# 2. Preserve the complete raw TRY return.
# 3. Distinguish actual Seed Mass measurements from associated metadata rows.
# 4. Confirm that TraitID 26 represents the returned Seed Mass measurements.
# 5. Audit numeric values, units and taxonomic identifiers.
# 6. Crosswalk returned TRY taxa against the frozen Section 9.3D request.
# 7. Associate metadata records with Seed Mass observations through ObservationID.
# 8. Export auditable measurement and metadata evidence tables.
#
# IMPORTANT
# ---------
# This section performs DATA INTEGRITY AND CROSSWALK AUDITING ONLY.
#
# No TRY measurement is accepted or rejected here.
# No deduplication is performed here.
# No species-level Seed Mass estimate is calculated here.
# No fallback acquisition is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re


# =============================================================================
# 1. PROJECT PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

# Frozen request outputs from Section 9.3D
TRY_REQUEST_MANIFEST_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_final_species_request_manifest.csv"
)

TRY_UNIQUE_QUERY_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_final_unique_species_queries.csv"
)

# -------------------------------------------------------------------------
# TRY Request 51911 download
#
# Search likely project locations first.
# If you have stored 51911.txt somewhere else, add that folder below.
# -------------------------------------------------------------------------

TRY_FILE_CANDIDATES = [
    PROJECT_ROOT / "51911.txt",
    OUTPUT_DIR / "51911.txt",
    PROJECT_ROOT / "data" / "51911.txt",
    PROJECT_ROOT / "data" / "raw" / "51911.txt",
    PROJECT_ROOT / "data" / "raw" / "functional_traits" / "51911.txt",
    PROJECT_ROOT / "inputs" / "51911.txt",
    PROJECT_ROOT / "inputs" / "functional_traits" / "51911.txt",
    Path.home() / "Downloads" / "51911.txt",
]

TRY_RETURN_FILE = next(
    (p for p in TRY_FILE_CANDIDATES if p.exists()),
    None
)

if TRY_RETURN_FILE is None:
    raise FileNotFoundError(
        "Could not locate TRY Request 51911 file.\n\n"
        "Expected filename: 51911.txt\n\n"
        "Checked:\n" +
        "\n".join(f" - {p}" for p in TRY_FILE_CANDIDATES)
    )


# =============================================================================
# 2. VALIDATE REQUIRED INPUT FILES
# =============================================================================

required_files = [
    TRY_REQUEST_MANIFEST_FILE,
    TRY_UNIQUE_QUERY_FILE,
]

missing_files = [
    p for p in required_files
    if not p.exists()
]

if missing_files:
    raise FileNotFoundError(
        "Required Section 9.3D input file(s) missing:\n" +
        "\n".join(f" - {p}" for p in missing_files)
    )


print("=" * 100)
print(
    "SECTION 9.3F — TRY REQUEST 51911 RAW-DATA INTEGRITY, "
    "SEED MASS EXTRACTION AND ASSOCIATED-METADATA AUDIT"
)
print("=" * 100)

print("\nInput files")
print("-" * 100)
print(f"TRY return              : {TRY_RETURN_FILE}")
print(f"Frozen request manifest : {TRY_REQUEST_MANIFEST_FILE}")
print(f"Unique TRY queries      : {TRY_UNIQUE_QUERY_FILE}")


# =============================================================================
# 3. HELPER FUNCTIONS
# =============================================================================

def clean_text(series):
    """
    Normalize whitespace while preserving missing values.
    """
    s = series.astype("string")
    s = s.str.strip()
    s = s.str.replace(r"\s+", " ", regex=True)
    return s


def normalize_taxon(series):
    """
    Conservative normalization for comparison only.
    Does not alter biological taxonomy.
    """
    s = clean_text(series)
    s = s.str.lower()
    s = s.str.replace(r"\s+", " ", regex=True)
    return s


def find_column(df, candidates, required=True):
    """
    Locate the first matching column, case-insensitively.
    """
    lookup = {
        str(col).strip().lower(): col
        for col in df.columns
    }

    for candidate in candidates:
        key = candidate.strip().lower()

        if key in lookup:
            return lookup[key]

    if required:
        raise KeyError(
            "Could not locate required column. "
            f"Candidates tested: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


def first_nonmissing(series):
    """
    Return first non-missing value in a series.
    """
    s = series.dropna()

    if len(s) == 0:
        return pd.NA

    return s.iloc[0]


# =============================================================================
# 4. LOAD FROZEN SECTION 9.3D REQUEST TABLES
# =============================================================================

request_manifest = pd.read_csv(
    TRY_REQUEST_MANIFEST_FILE,
    low_memory=False
)

unique_queries = pd.read_csv(
    TRY_UNIQUE_QUERY_FILE,
    low_memory=False
)

print("\nFrozen TRY request")
print("-" * 100)
print(
    f"Canonical Omo taxa requested : "
    f"{request_manifest['Canonical_Species'].nunique()}"
)

print(
    f"Unique TRY query rows         : "
    f"{len(unique_queries)}"
)


# =============================================================================
# 5. LOAD TRY REQUEST 51911 — ROBUST ENCODING HANDLING
# =============================================================================
#
# TRY exports may contain contributor names/references encoded with
# Windows-1252 / Latin-1 characters rather than strict UTF-8.
#
# Try common encodings in a controlled order.
# =============================================================================

raw = None
encoding_used = None
load_errors = []

encoding_candidates = [
    "utf-8-sig",
    "cp1252",
    "latin1",
]

for enc in encoding_candidates:

    try:

        raw = pd.read_csv(
            TRY_RETURN_FILE,
            sep="\t",
            dtype="string",
            low_memory=False,
            encoding=enc
        )

        encoding_used = enc
        break

    except UnicodeDecodeError as e:

        load_errors.append(
            f"{enc}: {e}"
        )

if raw is None:

    raise RuntimeError(
        "Unable to decode TRY Request 51911 using expected encodings.\n\n"
        + "\n".join(load_errors)
    )

assert not raw.empty, (
    "TRY Request 51911 loaded but contains no rows."
)

assert len(raw.columns) > 1, (
    "TRY Request 51911 does not appear to have parsed as a tab-delimited table."
)

# Remove completely empty trailing columns only.
raw = raw.dropna(
    axis=1,
    how="all"
).copy()

# Remove unnamed columns only when they contain no information.
unnamed_empty = [
    col
    for col in raw.columns
    if str(col).startswith("Unnamed:")
    and raw[col].isna().all()
]

if unnamed_empty:
    raw = raw.drop(
        columns=unnamed_empty
    )

print("\nTRY Request 51911 raw return")
print("-" * 100)

print(f"File          : {TRY_RETURN_FILE}")
print(f"Encoding used : {encoding_used}")
print(f"Rows          : {len(raw)}")
print(f"Columns       : {len(raw.columns)}")

print("\nReturned columns")
print("-" * 100)

for i, col in enumerate(
    raw.columns,
    start=1
):
    print(f"{i:02d}. {col}")


# =============================================================================
# 6. IDENTIFY CORE TRY COLUMNS
# =============================================================================

TRAIT_ID_COL = find_column(
    raw,
    ["TraitID"]
)

TRAIT_NAME_COL = find_column(
    raw,
    ["TraitName"],
    required=False
)

OBSERVATION_ID_COL = find_column(
    raw,
    ["ObservationID"]
)

OBS_DATA_ID_COL = find_column(
    raw,
    ["ObsDataID"],
    required=False
)

ACC_SPECIES_ID_COL = find_column(
    raw,
    ["AccSpeciesID"]
)

ACC_SPECIES_NAME_COL = find_column(
    raw,
    ["AccSpeciesName"]
)

DATA_ID_COL = find_column(
    raw,
    ["DataID"],
    required=False
)

DATA_NAME_COL = find_column(
    raw,
    ["DataName"],
    required=False
)

ORIG_NAME_COL = find_column(
    raw,
    ["OriglName", "OrigName"],
    required=False
)

ORIG_VALUE_COL = find_column(
    raw,
    ["OrigValueStr", "OrigValue"],
    required=False
)

ORIG_UNIT_COL = find_column(
    raw,
    ["OrigUnitStr", "OrigUnit"],
    required=False
)

STD_VALUE_COL = find_column(
    raw,
    ["StdValue", "StdValueStr"]
)

UNIT_NAME_COL = find_column(
    raw,
    ["UnitName"]
)

REFERENCE_COL = find_column(
    raw,
    ["Reference"],
    required=False
)

COMMENT_COL = find_column(
    raw,
    ["Comment"],
    required=False
)

DATASET_ID_COL = find_column(
    raw,
    ["DatasetID"],
    required=False
)

DATASET_COL = find_column(
    raw,
    ["Dataset"],
    required=False
)

DATASET_SPECIES_COL = find_column(
    raw,
    ["SpeciesName"],
    required=False
)


print("\nDetected core TRY fields")
print("-" * 100)

detected_fields = {
    "Trait ID": TRAIT_ID_COL,
    "Trait name": TRAIT_NAME_COL,
    "Observation ID": OBSERVATION_ID_COL,
    "ObsData ID": OBS_DATA_ID_COL,
    "Accepted species ID": ACC_SPECIES_ID_COL,
    "Accepted species name": ACC_SPECIES_NAME_COL,
    "Data ID": DATA_ID_COL,
    "Data name": DATA_NAME_COL,
    "Original value": ORIG_VALUE_COL,
    "Original unit": ORIG_UNIT_COL,
    "Standardized value": STD_VALUE_COL,
    "Standardized unit": UNIT_NAME_COL,
    "Reference": REFERENCE_COL,
    "Comment": COMMENT_COL,
}

for label, col in detected_fields.items():
    print(f"{label:<24}: {col}")


# =============================================================================
# 7. TRAIT-ID STRUCTURE AUDIT
# =============================================================================

trait_id_numeric = pd.to_numeric(
    raw[TRAIT_ID_COL],
    errors="coerce"
)

raw["TraitID_Numeric"] = trait_id_numeric

print("\nTrait ID structure")
print("-" * 100)

trait_id_display = (
    raw[TRAIT_ID_COL]
    .fillna("<MISSING>")
    .value_counts(dropna=False)
    .rename_axis("Raw_TraitID")
    .reset_index(name="N_Rows")
)

display(trait_id_display)


# =============================================================================
# 8. ISOLATE ACTUAL SEED MASS MEASUREMENTS
# =============================================================================
#
# Diagnostic established:
#
#     TraitID = 26
#
# identifies the returned Seed Mass measurement records.
#
# Rows with missing TraitID contain associated metadata and must be retained
# separately rather than treated as malformed measurements.
# =============================================================================

SEED_MASS_TRAIT_ID = 26

seed_mass = raw[
    raw["TraitID_Numeric"].eq(SEED_MASS_TRAIT_ID)
].copy()

associated_metadata = raw[
    raw["TraitID_Numeric"].isna()
].copy()

other_trait_rows = raw[
    raw["TraitID_Numeric"].notna()
    &
    ~raw["TraitID_Numeric"].eq(SEED_MASS_TRAIT_ID)
].copy()


print("\nTRY return row classification")
print("-" * 100)

row_class_summary = pd.DataFrame(
    {
        "TRY_Row_Class": [
            "SEED_MASS_MEASUREMENT",
            "ASSOCIATED_METADATA",
            "OTHER_NUMERIC_TRAIT_ID",
        ],
        "N_Rows": [
            len(seed_mass),
            len(associated_metadata),
            len(other_trait_rows),
        ],
    }
)

display(row_class_summary)

assert (
    len(seed_mass)
    + len(associated_metadata)
    + len(other_trait_rows)
    == len(raw)
), "TRY row classification does not account for all returned rows."


# =============================================================================
# 9. CONFIRM TRAIT NAME
# =============================================================================

print("\nSeed Mass trait confirmation")
print("-" * 100)
print(f"TraitID : {SEED_MASS_TRAIT_ID}")

if TRAIT_NAME_COL is not None:

    trait_names = (
        seed_mass[TRAIT_NAME_COL]
        .dropna()
        .astype(str)
        .str.strip()
        .value_counts()
    )

    print("\nTrait names associated with TraitID 26:")
    print(trait_names)

else:
    print(
        "TraitName column unavailable; "
        "TraitID 26 classification retained from request structure."
    )


# =============================================================================
# 10. NUMERIC VALUE INTEGRITY
# =============================================================================

seed_mass["StdValue_Numeric"] = pd.to_numeric(
    seed_mass[STD_VALUE_COL],
    errors="coerce"
)

n_missing_numeric = (
    seed_mass["StdValue_Numeric"]
    .isna()
    .sum()
)

n_nonpositive = (
    seed_mass["StdValue_Numeric"]
    .le(0)
    .fillna(False)
    .sum()
)

print("\nSeed Mass numeric integrity")
print("-" * 100)

print(
    f"Seed Mass measurement rows : {len(seed_mass)}"
)

print(
    f"Non-numeric/missing values : {n_missing_numeric}"
)

print(
    f"Non-positive values        : {n_nonpositive}"
)

print("\nReturned standardized-value summary")
print("-" * 100)

display(
    seed_mass["StdValue_Numeric"]
    .describe()
    .to_frame("StdValue")
)


# =============================================================================
# 11. UNIT INTEGRITY
# =============================================================================

seed_mass["Unit_Clean"] = clean_text(
    seed_mass[UNIT_NAME_COL]
)

print("\nReturned standardized units")
print("-" * 100)

unit_summary = (
    seed_mass["Unit_Clean"]
    .fillna("<MISSING>")
    .value_counts(dropna=False)
    .rename_axis("UnitName")
    .reset_index(name="N_Measurements")
)

display(unit_summary)


# =============================================================================
# 12. RETURNED TAXONOMIC STRUCTURE
# =============================================================================

seed_mass["AccSpeciesID_Numeric"] = pd.to_numeric(
    seed_mass[ACC_SPECIES_ID_COL],
    errors="coerce"
)

seed_mass["AccSpeciesName_Clean"] = clean_text(
    seed_mass[ACC_SPECIES_NAME_COL]
)

print("\nReturned TRY Seed Mass taxonomic coverage")
print("-" * 100)

print(
    "Unique accepted species IDs   : "
    f"{seed_mass['AccSpeciesID_Numeric'].nunique(dropna=True)}"
)

print(
    "Unique accepted species names : "
    f"{seed_mass['AccSpeciesName_Clean'].nunique(dropna=True)}"
)


# =============================================================================
# 13. PREPARE FROZEN TRY REQUEST CROSSWALK
# =============================================================================

REQUEST_ID_COL = find_column(
    unique_queries,
    ["TRY_AccSpeciesID", "AccSpeciesID"]
)

REQUEST_NAME_COL = find_column(
    unique_queries,
    [
        "Preferred_TRY_Query_Name",
        "TRY_Query_Name",
        "AccSpeciesName"
    ]
)

unique_queries["TRY_AccSpeciesID_Numeric"] = pd.to_numeric(
    unique_queries[REQUEST_ID_COL],
    errors="coerce"
)

unique_queries["Preferred_TRY_Query_Name_Clean"] = clean_text(
    unique_queries[REQUEST_NAME_COL]
)

assert (
    unique_queries["TRY_AccSpeciesID_Numeric"]
    .notna()
    .all()
), (
    "Frozen TRY request contains non-numeric "
    "TRY AccSpeciesID values."
)


# =============================================================================
# 14. AUDIT RETURNED ACCSPECIES IDs AGAINST FROZEN REQUEST
# =============================================================================

requested_ids = set(
    unique_queries["TRY_AccSpeciesID_Numeric"]
    .dropna()
    .astype(int)
)

returned_ids = set(
    seed_mass["AccSpeciesID_Numeric"]
    .dropna()
    .astype(int)
)

unexpected_ids = sorted(
    returned_ids - requested_ids
)

requested_ids_with_seed_mass = sorted(
    requested_ids & returned_ids
)

requested_ids_without_seed_mass = sorted(
    requested_ids - returned_ids
)


print("\nTRY request-versus-return ID audit")
print("-" * 100)

print(
    f"Unique TRY IDs requested             : {len(requested_ids)}"
)

print(
    f"Requested IDs returning Seed Mass    : "
    f"{len(requested_ids_with_seed_mass)}"
)

print(
    f"Requested IDs without Seed Mass      : "
    f"{len(requested_ids_without_seed_mass)}"
)

print(
    f"Unexpected returned TRY IDs          : "
    f"{len(unexpected_ids)}"
)

if unexpected_ids:
    print("\nUnexpected IDs:")
    print(unexpected_ids)
else:
    print("\nUnexpected IDs: NONE")


# =============================================================================
# 15. BUILD REQUEST-RETURN CROSSWALK
# =============================================================================

request_return_crosswalk = unique_queries.copy()

request_return_crosswalk[
    "Returned_Seed_Mass"
] = request_return_crosswalk[
    "TRY_AccSpeciesID_Numeric"
].isin(returned_ids)

measurement_counts = (
    seed_mass
    .groupby(
        "AccSpeciesID_Numeric",
        dropna=False
    )
    .size()
    .rename("N_Returned_Seed_Mass_Measurements")
    .reset_index()
)

request_return_crosswalk = (
    request_return_crosswalk
    .merge(
        measurement_counts,
        left_on="TRY_AccSpeciesID_Numeric",
        right_on="AccSpeciesID_Numeric",
        how="left"
    )
)

request_return_crosswalk[
    "N_Returned_Seed_Mass_Measurements"
] = (
    request_return_crosswalk[
        "N_Returned_Seed_Mass_Measurements"
    ]
    .fillna(0)
    .astype(int)
)

request_return_crosswalk[
    "TRY_Return_Status"
] = np.where(
    request_return_crosswalk["Returned_Seed_Mass"],
    "SEED_MASS_RETURNED",
    "NO_SEED_MASS_RETURNED"
)


print("\nFrozen TRY query return status")
print("-" * 100)

display(
    request_return_crosswalk[
        "TRY_Return_Status"
    ]
    .value_counts()
    .rename_axis("TRY_Return_Status")
    .reset_index(name="N_TRY_Queries")
)


# =============================================================================
# 16. MAP RETURNED TRY IDs BACK TO CANONICAL OMO TAXA
# =============================================================================

MANIFEST_ID_COL = find_column(
    request_manifest,
    ["TRY_AccSpeciesID", "AccSpeciesID"]
)

request_manifest[
    "TRY_AccSpeciesID_Numeric"
] = pd.to_numeric(
    request_manifest[MANIFEST_ID_COL],
    errors="coerce"
)

canonical_lookup = (
    request_manifest[
        [
            "Canonical_Species",
            "TRY_AccSpeciesID_Numeric"
        ]
    ]
    .dropna(
        subset=["TRY_AccSpeciesID_Numeric"]
    )
    .drop_duplicates()
)

seed_mass = seed_mass.merge(
    canonical_lookup,
    left_on="AccSpeciesID_Numeric",
    right_on="TRY_AccSpeciesID_Numeric",
    how="left"
)


print("\nCanonical Omo coverage in returned TRY Seed Mass")
print("-" * 100)

print(
    "Canonical taxa represented : "
    f"{seed_mass['Canonical_Species'].nunique(dropna=True)}"
)

print(
    "Unmapped measurement rows  : "
    f"{seed_mass['Canonical_Species'].isna().sum()}"
)


# =============================================================================
# 17. AUDIT POSSIBLE ONE-TO-MANY CANONICAL MAPPINGS
# =============================================================================
#
# Section 9.3D showed 87 Omo taxa mapping to 64 unique TRY IDs.
# Therefore a TRY accepted taxon may legitimately represent more than one
# Omo canonical string. Preserve this explicitly.
# =============================================================================

id_to_canonical = (
    canonical_lookup
    .groupby("TRY_AccSpeciesID_Numeric")
    ["Canonical_Species"]
    .agg(
        lambda x: " | ".join(
            sorted(set(x.dropna().astype(str)))
        )
    )
    .rename("Mapped_Omo_Canonical_Taxa")
    .reset_index()
)

id_to_canonical_counts = (
    canonical_lookup
    .groupby("TRY_AccSpeciesID_Numeric")
    ["Canonical_Species"]
    .nunique()
    .rename("N_Mapped_Omo_Canonical_Taxa")
    .reset_index()
)

id_to_canonical = id_to_canonical.merge(
    id_to_canonical_counts,
    on="TRY_AccSpeciesID_Numeric",
    how="left"
)

print("\nTRY-ID to Omo-canonical mapping structure")
print("-" * 100)

display(
    id_to_canonical[
        "N_Mapped_Omo_Canonical_Taxa"
    ]
    .value_counts()
    .sort_index()
    .rename_axis("N_Omo_Taxa_Per_TRY_ID")
    .reset_index(name="N_TRY_IDs")
)


# =============================================================================
# 18. ASSOCIATED METADATA AUDIT
# =============================================================================
#
# Missing TraitID rows are not treated as malformed.
# They are preserved as associated observation-level metadata.
# =============================================================================

associated_metadata["ObservationID_Clean"] = clean_text(
    associated_metadata[OBSERVATION_ID_COL]
)

seed_mass["ObservationID_Clean"] = clean_text(
    seed_mass[OBSERVATION_ID_COL]
)

seed_observation_ids = set(
    seed_mass["ObservationID_Clean"]
    .dropna()
)

associated_metadata[
    "Linked_To_Seed_Mass_Observation"
] = associated_metadata[
    "ObservationID_Clean"
].isin(seed_observation_ids)


print("\nAssociated metadata linkage")
print("-" * 100)

print(
    f"Metadata rows                     : "
    f"{len(associated_metadata)}"
)

print(
    f"Linked to Seed Mass ObservationID : "
    f"{associated_metadata['Linked_To_Seed_Mass_Observation'].sum()}"
)

print(
    f"Unlinked metadata rows            : "
    f"{(~associated_metadata['Linked_To_Seed_Mass_Observation']).sum()}"
)


# =============================================================================
# 19. METADATA TYPE SUMMARY
# =============================================================================

if DATA_NAME_COL is not None:

    metadata_type_summary = (
        associated_metadata[
            DATA_NAME_COL
        ]
        .fillna("<MISSING>")
        .value_counts(dropna=False)
        .rename_axis("Metadata_DataName")
        .reset_index(name="N_Rows")
    )

    print("\nAssociated metadata types")
    print("-" * 100)

    display(metadata_type_summary)

else:

    metadata_type_summary = pd.DataFrame(
        columns=[
            "Metadata_DataName",
            "N_Rows"
        ]
    )

    print(
        "\nDataName field unavailable; "
        "metadata-type summary not produced."
    )


# =============================================================================
# 20. MATERIAL-WEIGHED / SEED-DEFINITION EVIDENCE
# =============================================================================
#
# Metadata DataName such as "Seed mass comment" may contain evidence about
# what biological entity was weighed. Preserve this for later QC.
# =============================================================================

if DATA_NAME_COL is not None and ORIG_VALUE_COL is not None:

    material_mask = (
        associated_metadata[DATA_NAME_COL]
        .astype("string")
        .str.contains(
            "seed mass comment|material",
            case=False,
            na=False,
            regex=True
        )
    )

    material_evidence = (
        associated_metadata[
            material_mask
        ]
        .copy()
    )

else:

    material_evidence = associated_metadata.iloc[0:0].copy()


print("\nSeed-definition/material-weighed metadata")
print("-" * 100)

print(
    f"Potential definition-evidence rows : "
    f"{len(material_evidence)}"
)

if len(material_evidence) > 0:

    cols_to_show = [
        OBSERVATION_ID_COL,
        ACC_SPECIES_ID_COL,
        ACC_SPECIES_NAME_COL,
    ]

    if DATA_NAME_COL is not None:
        cols_to_show.append(DATA_NAME_COL)

    if ORIG_VALUE_COL is not None:
        cols_to_show.append(ORIG_VALUE_COL)

    display(
        material_evidence[
            list(dict.fromkeys(cols_to_show))
        ]
        .head(30)
    )


# =============================================================================
# 21. OBSERVATION-LEVEL METADATA COUNTS
# =============================================================================

metadata_counts_by_observation = (
    associated_metadata[
        associated_metadata[
            "Linked_To_Seed_Mass_Observation"
        ]
    ]
    .groupby(
        "ObservationID_Clean"
    )
    .size()
    .rename(
        "N_Associated_Metadata_Rows"
    )
    .reset_index()
)

seed_mass = seed_mass.merge(
    metadata_counts_by_observation,
    on="ObservationID_Clean",
    how="left"
)

seed_mass[
    "N_Associated_Metadata_Rows"
] = (
    seed_mass[
        "N_Associated_Metadata_Rows"
    ]
    .fillna(0)
    .astype(int)
)


print("\nSeed Mass observation metadata coverage")
print("-" * 100)

metadata_coverage_summary = pd.DataFrame(
    {
        "Metric": [
            "Seed Mass measurement rows",
            "Unique Seed Mass ObservationIDs",
            "Measurements with associated metadata",
            "Measurements without associated metadata",
        ],
        "Value": [
            len(seed_mass),
            seed_mass[
                "ObservationID_Clean"
            ].nunique(dropna=True),
            (
                seed_mass[
                    "N_Associated_Metadata_Rows"
                ] > 0
            ).sum(),
            (
                seed_mass[
                    "N_Associated_Metadata_Rows"
                ] == 0
            ).sum(),
        ]
    }
)

display(metadata_coverage_summary)


# =============================================================================
# 22. BUILD SPECIES-LEVEL RAW RETURN SUMMARY
# =============================================================================

species_raw_summary = (
    seed_mass
    .groupby(
        [
            "Canonical_Species",
            "AccSpeciesID_Numeric",
            "AccSpeciesName_Clean",
        ],
        dropna=False
    )
    .agg(
        N_Raw_Measurements=(
            "StdValue_Numeric",
            "size"
        ),
        N_Numeric_Measurements=(
            "StdValue_Numeric",
            lambda x: x.notna().sum()
        ),
        N_Unique_Values=(
            "StdValue_Numeric",
            "nunique"
        ),
        Min_StdValue=(
            "StdValue_Numeric",
            "min"
        ),
        Median_StdValue=(
            "StdValue_Numeric",
            "median"
        ),
        Max_StdValue=(
            "StdValue_Numeric",
            "max"
        ),
        N_Unique_ObservationIDs=(
            "ObservationID_Clean",
            "nunique"
        ),
    )
    .reset_index()
)

species_raw_summary[
    "Fold_Range"
] = np.where(
    (
        species_raw_summary["Min_StdValue"].notna()
        &
        species_raw_summary["Max_StdValue"].notna()
        &
        species_raw_summary["Min_StdValue"].gt(0)
    ),
    (
        species_raw_summary["Max_StdValue"]
        /
        species_raw_summary["Min_StdValue"]
    ),
    np.nan
)


print("\nReturned Seed Mass species-level raw evidence summary")
print("-" * 100)

display(species_raw_summary)


# =============================================================================
# 23. BUILD REQUESTED TAXA WITH NO RETURNED SEED MASS
# =============================================================================

no_seed_mass_return = request_return_crosswalk[
    request_return_crosswalk[
        "TRY_Return_Status"
    ].eq("NO_SEED_MASS_RETURNED")
].copy()


print("\nRequested TRY accepted taxa with no returned Seed Mass")
print("-" * 100)

print(
    f"N TRY accepted taxa : "
    f"{len(no_seed_mass_return)}"
)

if len(no_seed_mass_return) > 0:

    display(
        no_seed_mass_return[
            [
                REQUEST_NAME_COL,
                REQUEST_ID_COL,
                "TRY_Return_Status"
            ]
        ]
    )

else:
    print("NONE")


# =============================================================================
# 24. INTEGRITY FLAGS
# =============================================================================

seed_mass["TRY_Raw_Integrity_Status"] = "RAW_SEED_MASS_EVIDENCE"

seed_mass.loc[
    seed_mass[
        "StdValue_Numeric"
    ].isna(),
    "TRY_Raw_Integrity_Status"
] = "REVIEW_NON_NUMERIC_VALUE"

seed_mass.loc[
    seed_mass[
        "StdValue_Numeric"
    ].le(0).fillna(False),
    "TRY_Raw_Integrity_Status"
] = "REVIEW_NON_POSITIVE_VALUE"

seed_mass.loc[
    seed_mass[
        "AccSpeciesID_Numeric"
    ].isna(),
    "TRY_Raw_Integrity_Status"
] = "REVIEW_MISSING_ACC_SPECIES_ID"

seed_mass.loc[
    ~seed_mass[
        "AccSpeciesID_Numeric"
    ].fillna(-1).astype(int).isin(requested_ids),
    "TRY_Raw_Integrity_Status"
] = "REVIEW_UNEXPECTED_TRY_TAXON"


print("\nRaw Seed Mass integrity status")
print("-" * 100)

display(
    seed_mass[
        "TRY_Raw_Integrity_Status"
    ]
    .value_counts()
    .rename_axis(
        "TRY_Raw_Integrity_Status"
    )
    .reset_index(
        name="N_Measurements"
    )
)


# =============================================================================
# 25. FINAL AUDIT SUMMARY
# =============================================================================

summary_table = pd.DataFrame(
    {
        "Metric": [
            "Total TRY return rows",
            "Seed Mass measurement rows",
            "Associated metadata rows",
            "Other numeric TraitID rows",
            "Unique returned TRY species IDs",
            "Canonical Omo taxa represented",
            "Unique TRY IDs requested",
            "Requested TRY IDs with Seed Mass",
            "Requested TRY IDs without Seed Mass",
            "Unexpected returned TRY IDs",
            "Non-numeric/missing Seed Mass values",
            "Non-positive Seed Mass values",
        ],
        "Value": [
            len(raw),
            len(seed_mass),
            len(associated_metadata),
            len(other_trait_rows),
            seed_mass[
                "AccSpeciesID_Numeric"
            ].nunique(dropna=True),
            seed_mass[
                "Canonical_Species"
            ].nunique(dropna=True),
            len(requested_ids),
            len(requested_ids_with_seed_mass),
            len(requested_ids_without_seed_mass),
            len(unexpected_ids),
            int(n_missing_numeric),
            int(n_nonpositive),
        ],
    }
)

print("\nTRY Request 51911 integrity summary")
print("-" * 100)

display(summary_table)


# =============================================================================
# 26. EXPORT AUDIT PRODUCTS
# =============================================================================

RAW_PRESERVED_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_raw_preserved.csv"
)

SEED_MASS_RAW_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_seed_mass_raw_measurements.csv"
)

METADATA_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_associated_metadata.csv"
)

MATERIAL_EVIDENCE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_material_definition_metadata.csv"
)

REQUEST_RETURN_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_request_return_crosswalk.csv"
)

SPECIES_SUMMARY_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_raw_species_summary.csv"
)

NO_RETURN_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_no_seed_mass_return.csv"
)

ID_CANONICAL_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_try_id_canonical_mapping.csv"
)

METADATA_TYPE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_metadata_type_summary.csv"
)

INTEGRITY_SUMMARY_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_integrity_summary.csv"
)


raw.to_csv(
    RAW_PRESERVED_FILE,
    index=False
)

seed_mass.to_csv(
    SEED_MASS_RAW_FILE,
    index=False
)

associated_metadata.to_csv(
    METADATA_FILE,
    index=False
)

material_evidence.to_csv(
    MATERIAL_EVIDENCE_FILE,
    index=False
)

request_return_crosswalk.to_csv(
    REQUEST_RETURN_FILE,
    index=False
)

species_raw_summary.to_csv(
    SPECIES_SUMMARY_FILE,
    index=False
)

no_seed_mass_return.to_csv(
    NO_RETURN_FILE,
    index=False
)

id_to_canonical.to_csv(
    ID_CANONICAL_FILE,
    index=False
)

metadata_type_summary.to_csv(
    METADATA_TYPE_FILE,
    index=False
)

summary_table.to_csv(
    INTEGRITY_SUMMARY_FILE,
    index=False
)


# =============================================================================
# 27. REPORT EXPORTS
# =============================================================================

print("\nAudit exports")
print("-" * 100)

print(
    f"Preserved raw return     : {RAW_PRESERVED_FILE}"
)

print(
    f"Seed Mass measurements   : {SEED_MASS_RAW_FILE}"
)

print(
    f"Associated metadata      : {METADATA_FILE}"
)

print(
    f"Material-definition data : {MATERIAL_EVIDENCE_FILE}"
)

print(
    f"Request-return crosswalk : {REQUEST_RETURN_FILE}"
)

print(
    f"Species raw summary      : {SPECIES_SUMMARY_FILE}"
)

print(
    f"No-return TRY taxa       : {NO_RETURN_FILE}"
)

print(
    f"TRY-ID canonical mapping : {ID_CANONICAL_FILE}"
)

print(
    f"Metadata type summary    : {METADATA_TYPE_FILE}"
)

print(
    f"Integrity summary        : {INTEGRITY_SUMMARY_FILE}"
)


# =============================================================================
# 28. COMPLETION MESSAGE
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Request 51911 raw-data integrity and "
    "Seed Mass extraction audit completed successfully."
)

print(
    f"Seed Mass measurement rows preserved : "
    f"{len(seed_mass)}"
)

print(
    f"Associated metadata rows preserved    : "
    f"{len(associated_metadata)}"
)

print(
    "No TRY Seed Mass measurement has yet been "
    "accepted, rejected, deduplicated or aggregated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3F-R — RECONCILE RAW TRY MEASUREMENTS AND CANONICAL-MAPPING EXPANSION
# =============================================================================
#
# Purpose
# -------
# Reconcile the two legitimate but conceptually different row counts produced
# during Section 9.3F:
#
#   1. ORIGINAL TRY EVIDENCE LAYER
#      One row per Seed Mass record returned by TRY.
#
#   2. CANONICAL-MAPPING LAYER
#      The original TRY records expanded across Omo canonical taxa where more
#      than one project taxon maps to the same TRY accepted-species ID.
#
# This section prevents canonical mapping expansion from being interpreted as
# independent biological measurements.
#
# IMPORTANT
# ---------
# No measurement is accepted, rejected, biologically deduplicated, or
# aggregated in this section.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3F-R — RECONCILE RAW TRY MEASUREMENTS "
    "AND CANONICAL-MAPPING EXPANSION"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

RAW_MEASUREMENT_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_seed_mass_raw_measurements.csv"
)

REQUEST_MANIFEST_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_final_species_request_manifest.csv"
)

OUT_ORIGINAL_EVIDENCE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_original_measurement_evidence.csv"
)

OUT_CANONICAL_EXPANSION = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_canonical_mapping_expansion.csv"
)

OUT_NON_NUMERIC_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_non_numeric_measurement_review.csv"
)

OUT_CLEAN_QC_INPUT = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurements_for_qc.csv"
)

OUT_RECONCILIATION_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_reconciliation_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

assert RAW_MEASUREMENT_FILE.exists(), (
    f"Missing Section 9.3F measurement export:\n{RAW_MEASUREMENT_FILE}"
)

assert REQUEST_MANIFEST_FILE.exists(), (
    f"Missing frozen TRY request manifest:\n{REQUEST_MANIFEST_FILE}"
)


# =============================================================================
# 3. LOAD SECTION 9.3F SEED MASS EXPORT
# =============================================================================

raw = pd.read_csv(
    RAW_MEASUREMENT_FILE,
    dtype="string",
    low_memory=False
)

request_manifest = pd.read_csv(
    REQUEST_MANIFEST_FILE,
    dtype="string",
    low_memory=False
)

print("\nInput tables")
print("-" * 100)

print(
    f"Section 9.3F Seed Mass export : "
    f"{len(raw)} rows"
)

print(
    f"Frozen TRY request manifest   : "
    f"{len(request_manifest)} rows"
)


# =============================================================================
# 4. IDENTIFY REQUIRED TRY FIELDS
# =============================================================================

required_fields = [
    "ObservationID",
    "AccSpeciesID",
    "AccSpeciesName",
    "TraitID",
    "TraitName",
    "StdValue",
    "UnitName",
]

missing_fields = [
    col
    for col in required_fields
    if col not in raw.columns
]

assert not missing_fields, (
    "Required TRY measurement fields are missing:\n"
    + "\n".join(missing_fields)
)

print("\nRequired TRY fields confirmed")
print("-" * 100)

for col in required_fields:
    print(f" - {col}")


# =============================================================================
# 5. RECOVER ORIGINAL TRY MEASUREMENT RECORD IDENTIFIERS
# =============================================================================
#
# IMPORTANT:
# In TRY Request 51911:
#
#   ObservationID = observation-level grouping identifier
#   ObsDataID      = individual returned data-record identifier
#   DataID         = data-variable identifier and is NOT suitable as the
#                    primary unique measurement-record key
#
# Therefore ObsDataID is used as the primary evidence identifier.
# =============================================================================

obs_data_id = (
    raw["ObsDataID"]
    .astype("string")
    .str.strip()
)

obs_id = (
    raw["ObservationID"]
    .astype("string")
    .str.strip()
)

acc_id = (
    raw["AccSpeciesID"]
    .astype("string")
    .str.strip()
)

trait_id = (
    raw["TraitID"]
    .astype("string")
    .str.strip()
)

std_value = (
    raw["StdValue"]
    .astype("string")
    .str.strip()
)

orig_value = (
    raw["OrigValueStr"]
    .astype("string")
    .str.strip()
    if "OrigValueStr" in raw.columns
    else pd.Series(
        pd.NA,
        index=raw.index,
        dtype="string"
    )
)

# Deterministic fallback if ObsDataID were ever missing.
fallback_key = (
    "OBS_"
    + obs_id.fillna("MISSING")
    + "_ACC_"
    + acc_id.fillna("MISSING")
    + "_TRAIT_"
    + trait_id.fillna("MISSING")
    + "_STD_"
    + std_value.fillna("MISSING")
    + "_ORIG_"
    + orig_value.fillna("MISSING")
)

raw["TRY_Seed_Mass_Record_Key"] = np.where(
    obs_data_id.notna()
    &
    obs_data_id.ne(""),
    "OBSDATA_" + obs_data_id,
    fallback_key
)

assert (
    raw["TRY_Seed_Mass_Record_Key"]
    .notna()
    .all()
)


# =============================================================================
# 6. CONFIRM TRAIT STRUCTURE
# =============================================================================

trait_id_numeric = pd.to_numeric(
    raw["TraitID"],
    errors="coerce"
)

print("\nTrait structure")
print("-" * 100)

trait_summary = (
    raw["TraitID"]
    .fillna("<MISSING>")
    .value_counts(dropna=False)
    .rename_axis("TraitID")
    .reset_index(name="N_Rows")
)

display(trait_summary)

assert (
    trait_id_numeric
    .dropna()
    .eq(26)
    .all()
), (
    "Measurement export contains a numeric TraitID other than 26."
)


# =============================================================================
# 7. RECONSTRUCT UNIQUE ORIGINAL TRY RECORDS
# =============================================================================
#
# Section 9.3F's mapped export contains repeated copies of the same TRY
# measurement whenever one accepted TRY taxon maps to multiple Omo canonical
# strings. ObsDataID allows those replicated mapping rows to collapse back to
# the original TRY evidence layer.
# =============================================================================

key_frequency = (
    raw[
        "TRY_Seed_Mass_Record_Key"
    ]
    .value_counts()
)

print("\nTRY evidence-key frequency before reconciliation")
print("-" * 100)

print(
    f"Mapped rows                    : {len(raw)}"
)

print(
    f"Unique ObsData-based keys      : "
    f"{raw['TRY_Seed_Mass_Record_Key'].nunique()}"
)

print(
    f"Keys represented more than once: "
    f"{(key_frequency > 1).sum()}"
)

original_evidence = (
    raw
    .sort_values(
        [
            "TRY_Seed_Mass_Record_Key",
            "Canonical_Species",
        ],
        na_position="last"
    )
    .drop_duplicates(
        subset=[
            "TRY_Seed_Mass_Record_Key"
        ],
        keep="first"
    )
    .copy()
    .reset_index(drop=True)
)

print("\nOriginal TRY evidence reconstruction")
print("-" * 100)

print(
    f"Rows in Section 9.3F mapped export : "
    f"{len(raw)}"
)

print(
    f"Unique original TRY record keys    : "
    f"{len(original_evidence)}"
)

print(
    f"Unique ObservationIDs              : "
    f"{original_evidence['ObservationID'].nunique()}"
)

print(
    f"Unique ObsDataIDs                  : "
    f"{original_evidence['ObsDataID'].nunique()}"
)

assert len(original_evidence) == 178, (
    "Expected 178 original TRY Seed Mass records after collapsing "
    "canonical mapping expansion, "
    f"found {len(original_evidence)}."
)

assert (
    original_evidence[
        "TRY_Seed_Mass_Record_Key"
    ].nunique()
    == 178
), (
    "Original TRY evidence keys are not unique after reconciliation."
)

# =============================================================================
# 8. NUMERIC INTEGRITY — PANDAS-SAFE VERSION
# =============================================================================

original_evidence[
    "StdValue_Numeric"
] = pd.to_numeric(
    original_evidence["StdValue"],
    errors="coerce"
)

original_evidence[
    "TRY_Raw_Value_Status"
] = "NUMERIC_POSITIVE_VALUE"


missing_mask = (
    original_evidence[
        "StdValue_Numeric"
    ]
    .isna()
)

original_evidence.loc[
    missing_mask,
    "TRY_Raw_Value_Status"
] = "REVIEW_NON_NUMERIC_OR_MISSING"


nonpositive_mask = (
    original_evidence[
        "StdValue_Numeric"
    ]
    .le(0)
    .fillna(False)
    .astype(bool)
)

original_evidence.loc[
    nonpositive_mask,
    "TRY_Raw_Value_Status"
] = "REVIEW_NON_POSITIVE_VALUE"


print("\nOriginal TRY Seed Mass numeric integrity")
print("-" * 100)

value_status_summary = (
    original_evidence[
        "TRY_Raw_Value_Status"
    ]
    .value_counts(dropna=False)
    .rename_axis(
        "TRY_Raw_Value_Status"
    )
    .reset_index(
        name="N_Measurements"
    )
)

display(value_status_summary)


non_numeric_review = original_evidence[
    original_evidence[
        "TRY_Raw_Value_Status"
    ]
    != "NUMERIC_POSITIVE_VALUE"
].copy()


print("\nMeasurements requiring value review")
print("-" * 100)

print(
    f"N measurements : "
    f"{len(non_numeric_review)}"
)

if len(non_numeric_review) > 0:

    review_cols = [
        col
        for col in [
            "TRY_Seed_Mass_Record_Key",
            "ObservationID",
            "ObsDataID",
            "DataID",
            "AccSpeciesID",
            "AccSpeciesName",
            "OriglName",
            "OrigValueStr",
            "OrigUnitStr",
            "StdValue",
            "UnitName",
            "Reference",
            "Comment",
            "TRY_Raw_Value_Status",
        ]
        if col in non_numeric_review.columns
    ]

    display(
        non_numeric_review[
            review_cols
        ]
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 9. STANDARDIZED UNIT AUDIT
# =============================================================================

print("\nOriginal TRY standardized units")
print("-" * 100)

unit_summary = (
    original_evidence[
        "UnitName"
    ]
    .fillna("<MISSING>")
    .astype("string")
    .str.strip()
    .value_counts(dropna=False)
    .rename_axis("UnitName")
    .reset_index(name="N_Measurements")
)

display(unit_summary)


positive_numeric = original_evidence[
    original_evidence[
        "TRY_Raw_Value_Status"
    ]
    .eq("NUMERIC_POSITIVE_VALUE")
].copy()


unexpected_units = positive_numeric[
    ~positive_numeric[
        "UnitName"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
].copy()


assert len(unexpected_units) == 0, (
    "Positive numeric TRY Seed Mass records contain "
    "standardized units other than mg."
)


# =============================================================================
# 10. BUILD CLEAN ORIGINAL QC INPUT
# =============================================================================

qc_input = positive_numeric.copy()

qc_input[
    "Evidence_Layer"
] = "ORIGINAL_TRY_SEED_MASS_RECORD"

qc_input[
    "Independent_Measurement_Flag"
] = True

qc_input[
    "Section_9_3G_Input_Status"
] = "READY_FOR_TRY_MEASUREMENT_QC"


assert len(qc_input) == 177, (
    "Expected 177 positive numeric TRY Seed Mass records "
    f"for Section 9.3G, found {len(qc_input)}."
)


assert (
    qc_input[
        "TRY_Seed_Mass_Record_Key"
    ]
    .nunique()
    == len(qc_input)
), (
    "QC input contains duplicated original TRY record keys."
)


print("\nClean independent TRY measurement input")
print("-" * 100)

print(
    f"Original TRY records       : "
    f"{len(original_evidence)}"
)

print(
    f"Ready for measurement QC   : "
    f"{len(qc_input)}"
)

print(
    f"Held for value review      : "
    f"{len(non_numeric_review)}"
)

print(
    f"Unique returned TRY taxa   : "
    f"{qc_input['AccSpeciesID'].nunique()}"
)


# =============================================================================
# 11. BUILD CANONICAL MAPPING TABLE FROM THE FROZEN REQUEST MANIFEST
# =============================================================================

assert {
    "Canonical_Species",
    "TRY_AccSpeciesID"
}.issubset(
    request_manifest.columns
), (
    "Frozen TRY request manifest lacks required "
    "Canonical_Species / TRY_AccSpeciesID fields."
)


canonical_mapping = request_manifest[
    [
        "Canonical_Species",
        "TRY_AccSpeciesID",
    ]
].copy()


canonical_mapping[
    "TRY_AccSpeciesID_Merge"
] = (
    canonical_mapping[
        "TRY_AccSpeciesID"
    ]
    .astype("string")
    .str.strip()
    .str.replace(
        r"\.0$",
        "",
        regex=True
    )
)


canonical_mapping = (
    canonical_mapping[
        [
            "Canonical_Species",
            "TRY_AccSpeciesID_Merge",
        ]
    ]
    .dropna()
    .drop_duplicates()
    .reset_index(drop=True)
)


print("\nCanonical mapping structure")
print("-" * 100)

print(
    f"Unique Omo canonical taxa       : "
    f"{canonical_mapping['Canonical_Species'].nunique()}"
)

print(
    f"Unique TRY accepted species IDs : "
    f"{canonical_mapping['TRY_AccSpeciesID_Merge'].nunique()}"
)


# =============================================================================
# 12. PREPARE ORIGINAL TRY IDS FOR MAPPING
# =============================================================================

original_evidence[
    "TRY_AccSpeciesID_Merge"
] = (
    original_evidence[
        "AccSpeciesID"
    ]
    .astype("string")
    .str.strip()
    .str.replace(
        r"\.0$",
        "",
        regex=True
    )
)




# =============================================================================
# 13. CREATE CANONICAL-MAPPING EXPANSION
# =============================================================================
#
# IMPORTANT:
#
# original_evidence was reconstructed from the Section 9.3F mapped export.
# That earlier export may already contain Canonical_Species because TRY
# records had previously been expanded across Omo project taxa.
#
# For the independent TRY evidence layer, that inherited canonical assignment
# must NOT be carried forward. We therefore remove any pre-existing
# Canonical_Species field before rebuilding the mapping cleanly from the
# frozen request manifest.
#
# Each original TRY record may then legitimately map to one or more Omo
# canonical taxa in this separate provenance layer.
# =============================================================================

# -------------------------------------------------------------------------
# Remove inherited canonical mapping fields from the independent evidence
# before performing the authoritative mapping.
# -------------------------------------------------------------------------

mapping_columns_to_remove = [
    col
    for col in [
        "Canonical_Species",
        "Canonical_Species_x",
        "Canonical_Species_y",
        "TRY_AccSpeciesID_Merge_x",
        "TRY_AccSpeciesID_Merge_y",
    ]
    if col in original_evidence.columns
]

original_for_mapping = original_evidence.drop(
    columns=mapping_columns_to_remove,
    errors="ignore"
).copy()


# -------------------------------------------------------------------------
# Ensure one clean TRY ID merge field is present.
# -------------------------------------------------------------------------

original_for_mapping[
    "TRY_AccSpeciesID_Merge"
] = (
    original_for_mapping[
        "AccSpeciesID"
    ]
    .astype("string")
    .str.strip()
    .str.replace(
        r"\.0$",
        "",
        regex=True
    )
)


# -------------------------------------------------------------------------
# Validate mapping table before merge.
# -------------------------------------------------------------------------

assert {
    "TRY_AccSpeciesID_Merge",
    "Canonical_Species",
}.issubset(
    canonical_mapping.columns
), (
    "Canonical mapping table lacks required merge fields."
)

assert (
    canonical_mapping[
        "TRY_AccSpeciesID_Merge"
    ]
    .notna()
    .all()
)

assert (
    canonical_mapping[
        "Canonical_Species"
    ]
    .notna()
    .all()
)


# -------------------------------------------------------------------------
# Build canonical expansion.
# -------------------------------------------------------------------------

canonical_expansion = original_for_mapping.merge(
    canonical_mapping,
    on="TRY_AccSpeciesID_Merge",
    how="left",
    validate="many_to_many"
)


canonical_expansion[
    "Evidence_Layer"
] = "CANONICAL_MAPPING_EXPANSION"

canonical_expansion[
    "Independent_Measurement_Flag"
] = False

canonical_expansion[
    "Mapping_Interpretation"
] = (
    "ONE_ORIGINAL_TRY_RECORD_MAY_MAP_TO_MULTIPLE_OMO_CANONICAL_TAXA"
)


# -------------------------------------------------------------------------
# Strong structural validation.
# -------------------------------------------------------------------------

assert "Canonical_Species" in canonical_expansion.columns, (
    "Canonical_Species was not created correctly during mapping."
)

assert not any(
    col in canonical_expansion.columns
    for col in [
        "Canonical_Species_x",
        "Canonical_Species_y",
    ]
), (
    "Duplicate canonical-name columns remain after reconciliation."
)


print("\nCanonical mapping expansion")
print("-" * 100)

print(
    f"Original TRY evidence rows : "
    f"{len(original_evidence)}"
)

print(
    f"Canonical-expanded rows     : "
    f"{len(canonical_expansion)}"
)

print(
    f"Unique original TRY records : "
    f"{canonical_expansion['TRY_Seed_Mass_Record_Key'].nunique()}"
)

print(
    f"Canonical taxa represented  : "
    f"{canonical_expansion['Canonical_Species'].nunique()}"
)

print(
    f"Unmapped expansion rows     : "
    f"{canonical_expansion['Canonical_Species'].isna().sum()}"
)


# =============================================================================
# 14. CANONICAL EXPANSION MULTIPLICITY
# =============================================================================

expansion_multiplicity = (
    canonical_expansion
    .groupby(
        "TRY_Seed_Mass_Record_Key",
        dropna=False
    )
    .agg(
        N_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        )
    )
    .reset_index()
)


print("\nCanonical expansion multiplicity")
print("-" * 100)

multiplicity_summary = (
    expansion_multiplicity[
        "N_Canonical_Taxa"
    ]
    .value_counts()
    .sort_index()
    .rename_axis(
        "N_Omo_Canonical_Taxa"
    )
    .reset_index(
        name="N_Original_TRY_Records"
    )
)

display(multiplicity_summary)


# =============================================================================
# 15. CONFIRM THAT THE 177 QC ROWS REMAIN INDEPENDENT
# =============================================================================

qc_input[
    "TRY_AccSpeciesID_Merge"
] = (
    qc_input[
        "AccSpeciesID"
    ]
    .astype("string")
    .str.strip()
    .str.replace(
        r"\.0$",
        "",
        regex=True
    )
)


assert (
    qc_input[
        "TRY_Seed_Mass_Record_Key"
    ].nunique()
    == 177
)

assert (
    qc_input[
        "Independent_Measurement_Flag"
    ].all()
)


# =============================================================================
# 16. RECONCILIATION SUMMARY
# =============================================================================

reconciliation_summary = pd.DataFrame(
    {
        "Metric": [
            "Rows in Section 9.3F mapped Seed Mass export",
            "Original independent TRY Seed Mass records",
            "Positive numeric original records",
            "Records requiring value review",
            "Unique returned TRY accepted species IDs",
            "Canonical mapping expansion rows",
            "Canonical taxa represented in expansion",
            "Unique original TRY records in expansion",
            "Records entering Section 9.3G QC",
        ],
        "Value": [
            len(raw),
            len(original_evidence),
            len(qc_input),
            len(non_numeric_review),
            original_evidence[
                "TRY_AccSpeciesID_Merge"
            ].nunique(),
            len(canonical_expansion),
            canonical_expansion[
                "Canonical_Species"
            ].nunique(),
            canonical_expansion[
                "TRY_Seed_Mass_Record_Key"
            ].nunique(),
            len(qc_input),
        ],
    }
)


print("\nMeasurement reconciliation summary")
print("-" * 100)

display(reconciliation_summary)


# =============================================================================
# 17. STRONG VALIDATION
# =============================================================================

assert len(original_evidence) == 178

assert len(qc_input) == 177

assert len(non_numeric_review) == 1

assert (
    original_evidence[
        "TRY_Seed_Mass_Record_Key"
    ].nunique()
    == 178
)

assert (
    qc_input[
        "TRY_Seed_Mass_Record_Key"
    ].nunique()
    == 177
)

assert (
    canonical_expansion[
        "TRY_Seed_Mass_Record_Key"
    ].nunique()
    == 178
)

assert (
    canonical_expansion[
        "Canonical_Species"
    ]
    .notna()
    .all()
), (
    "At least one original TRY record could not be mapped "
    "back to an Omo canonical taxon."
)


print("\nReconciliation assertions")
print("-" * 100)

print(
    "Original TRY Seed Mass rows confirmed : 178"
)

print(
    "Positive numeric QC input confirmed   : 177"
)

print(
    "Value-review hold confirmed           : 1"
)

print(
    "Standardized unit confirmed           : mg"
)

print(
    "Canonical expansion excluded from independent QC count."
)


# =============================================================================
# 18. EXPORT
# =============================================================================

original_evidence.to_csv(
    OUT_ORIGINAL_EVIDENCE,
    index=False
)

canonical_expansion.to_csv(
    OUT_CANONICAL_EXPANSION,
    index=False
)

non_numeric_review.to_csv(
    OUT_NON_NUMERIC_REVIEW,
    index=False
)

qc_input.to_csv(
    OUT_CLEAN_QC_INPUT,
    index=False
)

reconciliation_summary.to_csv(
    OUT_RECONCILIATION_SUMMARY,
    index=False
)


# =============================================================================
# 19. REPORT EXPORTS
# =============================================================================

print("\nReconciliation exports")
print("-" * 100)

print(
    f"Original evidence     : "
    f"{OUT_ORIGINAL_EVIDENCE}"
)

print(
    f"Canonical expansion   : "
    f"{OUT_CANONICAL_EXPANSION}"
)

print(
    f"Value-review register : "
    f"{OUT_NON_NUMERIC_REVIEW}"
)

print(
    f"Section 9.3G QC input : "
    f"{OUT_CLEAN_QC_INPUT}"
)

print(
    f"Summary               : "
    f"{OUT_RECONCILIATION_SUMMARY}"
)


# =============================================================================
# 20. COMPLETION MESSAGE
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Request 51911 measurement/canonical-mapping "
    "reconciliation completed successfully."
)

print(
    f"Independent TRY Seed Mass records preserved : "
    f"{len(original_evidence)}"
)

print(
    f"Positive numeric records entering QC        : "
    f"{len(qc_input)}"
)

print(
    f"Records held for value review               : "
    f"{len(non_numeric_review)}"
)

print(
    "Canonical mapping expansion is retained strictly "
    "as a taxonomic/provenance layer."
)

print(
    "No measurement has been accepted, rejected, "
    "biologically deduplicated or aggregated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3G — TRY SEED MASS MEASUREMENT-LEVEL QC
#                AND EVIDENCE CLASSIFICATION
# =============================================================================
#
# Purpose
# -------
# Perform measurement-level quality assessment of the 177 positive numeric
# Seed dry mass records returned by TRY Request 51911.
#
# Evidence base
# -------------
# Independent TRY Seed Mass records ready for QC : 177
# Associated TRY metadata rows                   : 285
# Non-numeric/value-review records               : 1
#
# QC principles
# -------------
# - Preserve TRY TraitID 26 as the primary trait-definition evidence.
# - Use associated observation metadata to identify explicit seed evidence.
# - Flag possible seed-vs-diaspore/fruit/endocarp ambiguity for review.
# - Do NOT reject measurements solely because they are numerically extreme.
# - Do NOT perform species-level aggregation yet.
# - Do NOT treat Omo canonical aliases as independent measurements.
#
# Outputs
# -------
# Measurement QC register
# Provisionally retained measurements
# Definition-review holds
# Duplicate-signature diagnostic
# Species-level pre-aggregation evidence summary
# Metadata linkage summary
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print(
    "SECTION 9.3G — TRY SEED MASS MEASUREMENT-LEVEL QC "
    "AND EVIDENCE CLASSIFICATION"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

QC_INPUT_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurements_for_qc.csv"
)

METADATA_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_associated_metadata.csv"
)

VALUE_REVIEW_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_non_numeric_measurement_review.csv"
)

REQUEST_MANIFEST_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_final_species_request_manifest.csv"
)


OUT_QC_REGISTER = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_qc_register.csv"
)

OUT_PROVISIONAL_RETAINED = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_provisionally_retained_measurements.csv"
)

OUT_DEFINITION_HOLD = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_definition_review_hold.csv"
)

OUT_DUPLICATE_DIAGNOSTIC = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_duplicate_signature_diagnostic.csv"
)

OUT_SPECIES_PREAGG = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_species_preaggregation_summary.csv"
)

OUT_METADATA_LINKAGE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_metadata_linkage_summary.csv"
)

OUT_QC_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_qc_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

for path in [
    QC_INPUT_FILE,
    METADATA_FILE,
    VALUE_REVIEW_FILE,
    REQUEST_MANIFEST_FILE,
]:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

qc = pd.read_csv(
    QC_INPUT_FILE,
    dtype="string",
    low_memory=False
)

metadata = pd.read_csv(
    METADATA_FILE,
    dtype="string",
    low_memory=False
)

value_review = pd.read_csv(
    VALUE_REVIEW_FILE,
    dtype="string",
    low_memory=False
)

request_manifest = pd.read_csv(
    REQUEST_MANIFEST_FILE,
    dtype="string",
    low_memory=False
)


print("\nInput evidence")
print("-" * 100)

print(f"Positive numeric TRY measurements : {len(qc)}")
print(f"Associated metadata rows          : {len(metadata)}")
print(f"Value-review hold rows            : {len(value_review)}")
print(f"Frozen request manifest rows      : {len(request_manifest)}")


assert len(qc) == 177, (
    f"Expected 177 measurements entering QC, found {len(qc)}."
)

assert len(value_review) == 1, (
    f"Expected one value-review hold, found {len(value_review)}."
)


# =============================================================================
# 4. CORE STRUCTURAL VALIDATION
# =============================================================================

required_qc_fields = {
    "TRY_Seed_Mass_Record_Key",
    "ObservationID",
    "ObsDataID",
    "AccSpeciesID",
    "AccSpeciesName",
    "TraitID",
    "TraitName",
    "StdValue",
    "UnitName",
}

missing_qc_fields = (
    required_qc_fields
    - set(qc.columns)
)

assert not missing_qc_fields, (
    "QC input missing required fields:\n"
    f"{sorted(missing_qc_fields)}"
)


qc["StdValue_Numeric"] = pd.to_numeric(
    qc["StdValue"],
    errors="coerce"
)

qc["TraitID_Numeric"] = pd.to_numeric(
    qc["TraitID"],
    errors="coerce"
)

qc["AccSpeciesID_Numeric"] = pd.to_numeric(
    qc["AccSpeciesID"],
    errors="coerce"
)


assert qc["StdValue_Numeric"].notna().all()
assert qc["StdValue_Numeric"].gt(0).all()

assert qc["TraitID_Numeric"].eq(26).all()

assert (
    qc["UnitName"]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
)

assert (
    qc["TRY_Seed_Mass_Record_Key"]
    .is_unique
), (
    "Independent QC input unexpectedly contains duplicate TRY record keys."
)


print("\nStructural integrity")
print("-" * 100)

print("TraitID                 : 26 only")
print("Trait                    : Seed dry mass")
print("Standardized unit        : mg only")
print("Positive numeric values  : 177 / 177")
print("Unique TRY record keys   : 177 / 177")


# =============================================================================
# 5. NORMALIZE OBSERVATION IDS FOR METADATA LINKAGE
# =============================================================================

qc["ObservationID_Link"] = (
    qc["ObservationID"]
    .astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)

metadata["ObservationID_Link"] = (
    metadata["ObservationID"]
    .astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)


# =============================================================================
# 6. PREPARE METADATA VALUE TEXT
# =============================================================================
#
# Important:
#
# We search actual metadata VALUES for biological-definition evidence.
#
# We deliberately do NOT use DataName itself when identifying words such as
# "diaspore", because some TRY field labels contain explanatory wording like
# "Seed mass reference (sometimes diaspore ...)" even when the actual record
# is not a diaspore measurement.
# =============================================================================

metadata_value_columns = [
    col
    for col in [
        "OrigValueStr",
        "Comment",
    ]
    if col in metadata.columns
]


if metadata_value_columns:

    metadata["Metadata_Value_Text"] = (
        metadata[
            metadata_value_columns
        ]
        .fillna("")
        .astype(str)
        .agg(" | ".join, axis=1)
        .str.strip(" |")
    )

else:

    metadata["Metadata_Value_Text"] = ""


# =============================================================================
# 7. IDENTIFY DEFINITION-RELEVANT METADATA ROWS
# =============================================================================

if "DataName" in metadata.columns:

    definition_metadata_mask = (
        metadata["DataName"]
        .astype("string")
        .str.contains(
            (
                "seed material description"
                "|seed mass comment"
                "|comments, notes, methods"
            ),
            case=False,
            na=False,
            regex=True
        )
    )

else:

    definition_metadata_mask = pd.Series(
        False,
        index=metadata.index
    )


definition_metadata = metadata[
    definition_metadata_mask
].copy()


print("\nDefinition-relevant associated metadata")
print("-" * 100)

print(
    f"Rows considered definition-relevant : "
    f"{len(definition_metadata)}"
)


# =============================================================================
# 8. COLLAPSE DEFINITION METADATA BY OBSERVATION
# =============================================================================

if not definition_metadata.empty:

    definition_by_observation = (
        definition_metadata
        .groupby(
            "ObservationID_Link",
            dropna=False
        )
        .agg(
            N_Definition_Metadata_Rows=(
                "ObservationID_Link",
                "size"
            ),

            Definition_Metadata_Text=(
                "Metadata_Value_Text",
                lambda x: " || ".join(
                    [
                        str(v).strip()
                        for v in x
                        if pd.notna(v)
                        and str(v).strip()
                    ]
                )
            ),
        )
        .reset_index()
    )

else:

    definition_by_observation = pd.DataFrame(
        columns=[
            "ObservationID_Link",
            "N_Definition_Metadata_Rows",
            "Definition_Metadata_Text",
        ]
    )


# =============================================================================
# 9. TOTAL METADATA COUNT PER SEED MASS OBSERVATION
# =============================================================================

metadata_counts = (
    metadata
    .groupby(
        "ObservationID_Link",
        dropna=False
    )
    .size()
    .rename(
        "N_Associated_Metadata_Rows"
    )
    .reset_index()
)


# =============================================================================
# 10. ATTACH METADATA TO THE 177 INDEPENDENT MEASUREMENTS
# =============================================================================
#
# The Section 9.3F-R QC input may already contain metadata-derived columns
# inherited from the earlier integrity audit.
#
# To avoid merge suffixes such as:
#
#     N_Associated_Metadata_Rows_x
#     N_Associated_Metadata_Rows_y
#
# remove those derived fields first and recompute them cleanly from the
# preserved 285-row associated-metadata table.
# =============================================================================

preexisting_metadata_fields = [
    "N_Associated_Metadata_Rows",
    "N_Definition_Metadata_Rows",
    "Definition_Metadata_Text",
]

fields_present = [
    col
    for col in preexisting_metadata_fields
    if col in qc.columns
]

if fields_present:

    print("\nRemoving inherited metadata-derived fields before clean relink")
    print("-" * 100)

    for col in fields_present:
        print(f" - {col}")

    qc = qc.drop(
        columns=fields_present
    )

else:

    print("\nNo inherited metadata-derived fields required removal.")


# -----------------------------------------------------------------------------
# Merge total metadata counts.
# -----------------------------------------------------------------------------

qc = qc.merge(
    metadata_counts,
    on="ObservationID_Link",
    how="left",
    validate="many_to_one"
)


# -----------------------------------------------------------------------------
# Merge definition-relevant metadata.
# -----------------------------------------------------------------------------

qc = qc.merge(
    definition_by_observation,
    on="ObservationID_Link",
    how="left",
    validate="many_to_one"
)


# -----------------------------------------------------------------------------
# Standardize resulting metadata fields.
# -----------------------------------------------------------------------------

qc["N_Associated_Metadata_Rows"] = (
    pd.to_numeric(
        qc["N_Associated_Metadata_Rows"],
        errors="coerce"
    )
    .fillna(0)
    .astype(int)
)

qc["N_Definition_Metadata_Rows"] = (
    pd.to_numeric(
        qc["N_Definition_Metadata_Rows"],
        errors="coerce"
    )
    .fillna(0)
    .astype(int)
)

qc["Definition_Metadata_Text"] = (
    qc["Definition_Metadata_Text"]
    .fillna("")
    .astype(str)
)


# -----------------------------------------------------------------------------
# Validate that the clean merge did not generate suffix columns.
# -----------------------------------------------------------------------------

unexpected_suffix_columns = [
    col
    for col in qc.columns
    if (
        col.startswith("N_Associated_Metadata_Rows_")
        or
        col.startswith("N_Definition_Metadata_Rows_")
        or
        col.startswith("Definition_Metadata_Text_")
    )
]

assert not unexpected_suffix_columns, (
    "Unexpected metadata merge-suffix columns remain:\n"
    f"{unexpected_suffix_columns}"
)


print("\nClean metadata relink completed")
print("-" * 100)

print(
    f"Measurements in QC table               : "
    f"{len(qc)}"
)

print(
    f"Measurements with associated metadata  : "
    f"{qc['N_Associated_Metadata_Rows'].gt(0).sum()}"
)

print(
    f"Measurements without associated metadata: "
    f"{qc['N_Associated_Metadata_Rows'].eq(0).sum()}"
)

print(
    f"Measurements with definition metadata  : "
    f"{qc['N_Definition_Metadata_Rows'].gt(0).sum()}"
)


# =============================================================================
# 11. DEFINITION-EVIDENCE CLASSIFICATION
# =============================================================================
#
# Explicit true-seed evidence:
#    metadata values explicitly describing the weighed entity as "Seed".
#
# Ambiguous/non-standard dispersal-unit evidence:
#    fruit, diaspore, endocarp, stone, pyrene etc.
#
# These patterns create REVIEW HOLDS rather than automatic rejections.
# =============================================================================

definition_text_lower = (
    qc["Definition_Metadata_Text"]
    .str.lower()
)


explicit_seed_mask = (
    definition_text_lower
    .str.contains(
        r"(^|\|\||\|)\s*seed\s*($|\|\||\|)",
        regex=True,
        na=False
    )
    |
    definition_text_lower
    .str.fullmatch(
        r"\s*seed\s*",
        na=False
    )
)


ambiguity_pattern = (
    r"\bdiaspore\b"
    r"|\bendocarp\b"
    r"|\bstone\b"
    r"|\bpyrene\b"
    r"|\bwhole fruit\b"
    r"|\bfruit mass\b"
    r"|\bfruit\b"
)


definition_ambiguity_mask = (
    definition_text_lower
    .str.contains(
        ambiguity_pattern,
        regex=True,
        na=False
    )
)


qc["Definition_Evidence_Class"] = (
    "TRY_TRAIT26_NO_CONTRADICTORY_DEFINITION_METADATA"
)

qc.loc[
    explicit_seed_mask,
    "Definition_Evidence_Class"
] = "EXPLICIT_SEED_MATERIAL_EVIDENCE"

qc.loc[
    definition_ambiguity_mask,
    "Definition_Evidence_Class"
] = "POTENTIAL_DISPERSAL_UNIT_AMBIGUITY"


# =============================================================================
# 12. SOURCE / DATASET AUDIT
# =============================================================================

source_fields = [
    c
    for c in [
        "DatasetID",
        "Dataset",
        "Reference",
    ]
    if c in qc.columns
]


if "DatasetID" in qc.columns:

    qc["Source_Key"] = (
        qc["DatasetID"]
        .fillna("<MISSING_DATASET>")
        .astype(str)
        .str.strip()
    )

elif "Reference" in qc.columns:

    qc["Source_Key"] = (
        qc["Reference"]
        .fillna("<MISSING_REFERENCE>")
        .astype(str)
        .str.strip()
    )

else:

    qc["Source_Key"] = "<UNKNOWN_SOURCE>"


# =============================================================================
# 13. ERROR-RISK INVENTORY
# =============================================================================
#
# We inventory ErrorRisk but do not automatically reject any record based on
# the field because TRY's coded semantics must be interpreted explicitly.
# =============================================================================

if "ErrorRisk" in qc.columns:

    error_risk_summary = (
        qc["ErrorRisk"]
        .fillna("<MISSING>")
        .value_counts(dropna=False)
        .rename_axis("ErrorRisk")
        .reset_index(name="N_Measurements")
    )

else:

    error_risk_summary = pd.DataFrame(
        {
            "ErrorRisk": ["<FIELD_NOT_AVAILABLE>"],
            "N_Measurements": [len(qc)],
        }
    )


print("\nTRY ErrorRisk inventory")
print("-" * 100)

display(error_risk_summary)


# =============================================================================
# 14. SCIENTIFIC DUPLICATE-SIGNATURE DIAGNOSTIC
# =============================================================================
#
# ObsDataID / TRY record keys are already unique.
#
# Here we ask a different question:
# Do scientifically identical measurement signatures recur?
#
# We include ObservationID so genuinely separate observations are retained.
# No rows are removed in this section.
# =============================================================================

duplicate_signature_fields = [
    col
    for col in [
        "AccSpeciesID",
        "ObservationID",
        "DatasetID",
        "StdValue",
        "UnitName",
        "Reference",
    ]
    if col in qc.columns
]


qc["Scientific_Duplicate_Signature"] = (
    qc[
        duplicate_signature_fields
    ]
    .fillna("<MISSING>")
    .astype(str)
    .agg(" || ".join, axis=1)
)


signature_counts = (
    qc[
        "Scientific_Duplicate_Signature"
    ]
    .value_counts()
)


duplicate_signature_values = set(
    signature_counts[
        signature_counts > 1
    ].index
)


qc["Duplicate_Signature_Status"] = np.where(
    qc[
        "Scientific_Duplicate_Signature"
    ]
    .isin(
        duplicate_signature_values
    ),
    "REPEATED_SCIENTIFIC_SIGNATURE",
    "UNIQUE_SCIENTIFIC_SIGNATURE"
)


duplicate_diagnostic = qc[
    qc[
        "Duplicate_Signature_Status"
    ].eq(
        "REPEATED_SCIENTIFIC_SIGNATURE"
    )
].copy()


print("\nScientific duplicate-signature diagnostic")
print("-" * 100)

print(
    f"Duplicate signatures      : "
    f"{len(duplicate_signature_values)}"
)

print(
    f"Rows in duplicate groups  : "
    f"{len(duplicate_diagnostic)}"
)


# =============================================================================
# 15. MEASUREMENT-LEVEL QC CLASSIFICATION
# =============================================================================
#
# Definition ambiguity receives a HOLD.
#
# Duplicate scientific signatures are not automatically rejected because
# repeated measurements may be legitimate; they remain provisionally retained
# but explicitly flagged for the later deduplication stage.
# =============================================================================

qc["Measurement_QC_Status"] = (
    "PROVISIONALLY_RETAIN"
)


qc.loc[
    definition_ambiguity_mask,
    "Measurement_QC_Status"
] = "HOLD_FOR_DEFINITION_REVIEW"


qc["Trait_Quality_Flag"] = np.select(
    [
        definition_ambiguity_mask.to_numpy(
            dtype=bool
        ),

        explicit_seed_mask.to_numpy(
            dtype=bool
        ),

        qc[
            "N_Associated_Metadata_Rows"
        ].gt(0).to_numpy(
            dtype=bool
        ),
    ],
    [
        "POTENTIAL_SEED_DISPERSAL_UNIT_AMBIGUITY",
        "EXPLICIT_SEED_MATERIAL_CONFIRMED",
        "TRY_METADATA_AVAILABLE_NO_DEFINITION_CONFLICT",
    ],
    default="TRY_TRAIT26_NO_ASSOCIATED_METADATA"
)


# Add duplicate status as a supplementary flag rather than overriding
# biological-definition status.

qc["Duplicate_Review_Flag"] = np.where(
    qc[
        "Duplicate_Signature_Status"
    ].eq(
        "REPEATED_SCIENTIFIC_SIGNATURE"
    ),
    "REVIEW_DURING_DEDUPLICATION",
    "NO_DUPLICATE_SIGNATURE_FLAG"
)


# =============================================================================
# 16. QC STATUS SUMMARY
# =============================================================================

qc_status_summary = (
    qc[
        "Measurement_QC_Status"
    ]
    .value_counts()
    .rename_axis(
        "Measurement_QC_Status"
    )
    .reset_index(
        name="N_Measurements"
    )
)


print("\nMeasurement-level QC status")
print("-" * 100)

display(qc_status_summary)


quality_summary = (
    qc[
        "Trait_Quality_Flag"
    ]
    .value_counts()
    .rename_axis(
        "Trait_Quality_Flag"
    )
    .reset_index(
        name="N_Measurements"
    )
)


print("\nTrait-quality classification")
print("-" * 100)

display(quality_summary)


# =============================================================================
# 17. DEFINITION-REVIEW HOLD
# =============================================================================

definition_hold = qc[
    qc[
        "Measurement_QC_Status"
    ].eq(
        "HOLD_FOR_DEFINITION_REVIEW"
    )
].copy()


print("\nMeasurements held for definition review")
print("-" * 100)

print(
    f"N measurements : "
    f"{len(definition_hold)}"
)


if len(definition_hold) > 0:

    hold_columns = [
        col
        for col in [
            "TRY_Seed_Mass_Record_Key",
            "ObservationID",
            "AccSpeciesID",
            "AccSpeciesName",
            "StdValue_Numeric",
            "UnitName",
            "Dataset",
            "Reference",
            "Definition_Metadata_Text",
            "Trait_Quality_Flag",
        ]
        if col in definition_hold.columns
    ]

    display(
        definition_hold[
            hold_columns
        ]
        .sort_values(
            [
                "AccSpeciesName",
                "StdValue_Numeric",
            ]
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 18. PROVISIONALLY RETAINED MEASUREMENTS
# =============================================================================

provisional = qc[
    qc[
        "Measurement_QC_Status"
    ].eq(
        "PROVISIONALLY_RETAIN"
    )
].copy()


print("\nProvisionally retained TRY Seed Mass evidence")
print("-" * 100)

print(
    f"Measurements : "
    f"{len(provisional)}"
)

print(
    f"TRY taxa     : "
    f"{provisional['AccSpeciesID_Numeric'].nunique()}"
)


# =============================================================================
# 19. SPECIES-LEVEL PRE-AGGREGATION EVIDENCE SUMMARY
# =============================================================================
#
# This is diagnostic only.
# No species-level Seed Mass estimate is selected here.
# =============================================================================

species_preagg = (
    qc
    .groupby(
        [
            "AccSpeciesID_Numeric",
            "AccSpeciesName",
        ],
        dropna=False
    )
    .agg(
        N_QC_Input_Measurements=(
            "StdValue_Numeric",
            "size"
        ),

        N_Provisionally_Retained=(
            "Measurement_QC_Status",
            lambda x: (
                x
                == "PROVISIONALLY_RETAIN"
            ).sum()
        ),

        N_Definition_Holds=(
            "Measurement_QC_Status",
            lambda x: (
                x
                == "HOLD_FOR_DEFINITION_REVIEW"
            ).sum()
        ),

        N_Unique_Values=(
            "StdValue_Numeric",
            "nunique"
        ),

        N_Sources=(
            "Source_Key",
            "nunique"
        ),

        Min_mg=(
            "StdValue_Numeric",
            "min"
        ),

        Median_mg=(
            "StdValue_Numeric",
            "median"
        ),

        Max_mg=(
            "StdValue_Numeric",
            "max"
        ),
    )
    .reset_index()
)


species_preagg[
    "Fold_Range"
] = np.where(
    species_preagg[
        "Min_mg"
    ].gt(0),
    (
        species_preagg[
            "Max_mg"
        ]
        /
        species_preagg[
            "Min_mg"
        ]
    ),
    np.nan
)


species_preagg[
    "Species_Preaggregation_Status"
] = np.where(
    species_preagg[
        "N_Definition_Holds"
    ].gt(0),
    "HAS_DEFINITION_REVIEW_HOLD",
    "NO_DEFINITION_HOLD"
)


print("\nTRY species-level pre-aggregation evidence summary")
print("-" * 100)

display(
    species_preagg
    .sort_values(
        "AccSpeciesName"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 20. HIGH WITHIN-SPECIES VARIABILITY DIAGNOSTIC
# =============================================================================
#
# Large fold ranges are NOT automatically erroneous.
# We simply identify taxa that may deserve targeted review later.
# =============================================================================

variability_review = species_preagg[
    species_preagg[
        "Fold_Range"
    ].ge(10)
].copy()


print("\nHigh within-species variability diagnostic (>=10-fold)")
print("-" * 100)

print(
    f"N TRY taxa : "
    f"{len(variability_review)}"
)

if len(variability_review) > 0:

    display(
        variability_review[
            [
                "AccSpeciesID_Numeric",
                "AccSpeciesName",
                "N_QC_Input_Measurements",
                "Min_mg",
                "Median_mg",
                "Max_mg",
                "Fold_Range",
            ]
        ]
        .sort_values(
            "Fold_Range",
            ascending=False
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 21. METADATA-LINKAGE SUMMARY
# =============================================================================

metadata_linkage_summary = pd.DataFrame(
    {
        "Metric": [
            "Measurements entering QC",
            "Measurements with any associated metadata",
            "Measurements without associated metadata",
            "Measurements with definition metadata",
            "Measurements with explicit seed-material evidence",
            "Measurements with definition ambiguity",
        ],
        "Value": [
            len(qc),

            int(
                qc[
                    "N_Associated_Metadata_Rows"
                ]
                .gt(0)
                .sum()
            ),

            int(
                qc[
                    "N_Associated_Metadata_Rows"
                ]
                .eq(0)
                .sum()
            ),

            int(
                qc[
                    "N_Definition_Metadata_Rows"
                ]
                .gt(0)
                .sum()
            ),

            int(
                explicit_seed_mask.sum()
            ),

            int(
                definition_ambiguity_mask.sum()
            ),
        ],
    }
)


print("\nMetadata linkage summary")
print("-" * 100)

display(metadata_linkage_summary)


# =============================================================================
# 22. OVERALL QC SUMMARY
# =============================================================================

qc_summary = pd.DataFrame(
    {
        "Metric": [
            "Independent positive numeric TRY records",
            "TRY accepted species represented",
            "Provisionally retained measurements",
            "Definition-review holds",
            "Repeated scientific-signature rows",
            "Species with >=10-fold raw variability",
            "Separate non-numeric/value-review records",
        ],
        "Value": [
            len(qc),

            qc[
                "AccSpeciesID_Numeric"
            ].nunique(),

            len(provisional),

            len(definition_hold),

            len(duplicate_diagnostic),

            len(variability_review),

            len(value_review),
        ],
    }
)


print("\nTRY Seed Mass measurement-QC summary")
print("-" * 100)

display(qc_summary)


# =============================================================================
# 23. STRONG VALIDATION
# =============================================================================

assert len(qc) == 177

assert (
    len(provisional)
    + len(definition_hold)
    == 177
)

assert qc[
    "TRY_Seed_Mass_Record_Key"
].is_unique

assert qc[
    "StdValue_Numeric"
].notna().all()

assert qc[
    "StdValue_Numeric"
].gt(0).all()

assert qc[
    "TraitID_Numeric"
].eq(26).all()

assert (
    qc[
        "UnitName"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
)


# =============================================================================
# 24. EXPORT
# =============================================================================

qc.to_csv(
    OUT_QC_REGISTER,
    index=False
)

provisional.to_csv(
    OUT_PROVISIONAL_RETAINED,
    index=False
)

definition_hold.to_csv(
    OUT_DEFINITION_HOLD,
    index=False
)

duplicate_diagnostic.to_csv(
    OUT_DUPLICATE_DIAGNOSTIC,
    index=False
)

species_preagg.to_csv(
    OUT_SPECIES_PREAGG,
    index=False
)

metadata_linkage_summary.to_csv(
    OUT_METADATA_LINKAGE,
    index=False
)

qc_summary.to_csv(
    OUT_QC_SUMMARY,
    index=False
)


# =============================================================================
# 25. REPORT EXPORTS
# =============================================================================

print("\nMeasurement-QC exports")
print("-" * 100)

print(
    f"QC register             : "
    f"{OUT_QC_REGISTER}"
)

print(
    f"Provisionally retained  : "
    f"{OUT_PROVISIONAL_RETAINED}"
)

print(
    f"Definition-review hold  : "
    f"{OUT_DEFINITION_HOLD}"
)

print(
    f"Duplicate diagnostic    : "
    f"{OUT_DUPLICATE_DIAGNOSTIC}"
)

print(
    f"Species pre-aggregation : "
    f"{OUT_SPECIES_PREAGG}"
)

print(
    f"Metadata linkage        : "
    f"{OUT_METADATA_LINKAGE}"
)

print(
    f"QC summary              : "
    f"{OUT_QC_SUMMARY}"
)


# =============================================================================
# 26. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Request 51911 Seed Mass measurement-level QC "
    "completed successfully."
)

print(
    f"Measurements assessed            : "
    f"{len(qc)}"
)

print(
    f"Provisionally retained           : "
    f"{len(provisional)}"
)

print(
    f"Held for definition review       : "
    f"{len(definition_hold)}"
)

print(
    "No final TRY measurement has yet been accepted, "
    "deduplicated or aggregated to species level."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3H — EXCEPTIONAL-RECORD REVIEW:
#                DRACAENA MANNII VARIABILITY + CELTIS ZENKERI VALUE HOLD
# =============================================================================
#
# Purpose
# -------
# Investigate the only two exceptional TRY issues remaining after Section 9.3G:
#
#   1. Dracaena mannii
#      - 4 positive numeric measurements
#      - ~24-fold range
#
#   2. Celtis zenkeri
#      - 1 separate non-numeric/missing standardized-value record
#
# This section is diagnostic only.
#
# No measurement is accepted, rejected, or aggregated here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3H — EXCEPTIONAL-RECORD REVIEW: "
    "DRACAENA MANNII + CELTIS ZENKERI"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

QC_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_qc_register.csv"
)

VALUE_REVIEW_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_non_numeric_measurement_review.csv"
)

METADATA_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_associated_metadata.csv"
)

OUT_DRACAENA_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_dracaena_mannii_variability_review.csv"
)

OUT_DRACAENA_METADATA = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_dracaena_mannii_metadata_review.csv"
)

OUT_CELTIS_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_celtis_zenkeri_value_review.csv"
)

OUT_EXCEPTION_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_exceptional_record_review_summary.csv"
)


# =============================================================================
# 2. LOAD INPUTS
# =============================================================================

for path in [
    QC_REGISTER_FILE,
    VALUE_REVIEW_FILE,
    METADATA_FILE,
]:
    assert path.exists(), (
        f"Required file not found:\n{path}"
    )

qc = pd.read_csv(
    QC_REGISTER_FILE,
    dtype="string",
    low_memory=False
)

value_review = pd.read_csv(
    VALUE_REVIEW_FILE,
    dtype="string",
    low_memory=False
)

metadata = pd.read_csv(
    METADATA_FILE,
    dtype="string",
    low_memory=False
)

print("\nInput tables")
print("-" * 100)

print(f"QC register rows        : {len(qc)}")
print(f"Value-review rows       : {len(value_review)}")
print(f"Metadata rows           : {len(metadata)}")


# =============================================================================
# 3. NUMERIC NORMALIZATION
# =============================================================================

qc["StdValue_Numeric"] = pd.to_numeric(
    qc["StdValue"],
    errors="coerce"
)

qc["AccSpeciesID_Numeric"] = pd.to_numeric(
    qc["AccSpeciesID"],
    errors="coerce"
)

qc["ObservationID_Link"] = (
    qc["ObservationID"]
    .astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)

metadata["ObservationID_Link"] = (
    metadata["ObservationID"]
    .astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)


# =============================================================================
# 4. EXTRACT DRACAENA MANNII
# =============================================================================

dracaena = qc[
    qc["AccSpeciesName"]
    .astype("string")
    .str.strip()
    .eq("Dracaena mannii")
].copy()

assert len(dracaena) == 4, (
    f"Expected 4 Dracaena mannii records, found {len(dracaena)}."
)

dracaena = (
    dracaena
    .sort_values("StdValue_Numeric")
    .reset_index(drop=True)
)

print("\nDracaena mannii TRY Seed Mass measurements")
print("-" * 100)

dracaena_cols = [
    col for col in [
        "TRY_Seed_Mass_Record_Key",
        "ObservationID",
        "ObsDataID",
        "DatasetID",
        "Dataset",
        "SpeciesName",
        "AccSpeciesID",
        "AccSpeciesName",
        "OriglName",
        "OrigValueStr",
        "OrigUnitStr",
        "StdValue",
        "UnitName",
        "ValueKindName",
        "Reference",
        "Comment",
        "ErrorRisk",
        "Definition_Evidence_Class",
        "Trait_Quality_Flag",
    ]
    if col in dracaena.columns
]

display(
    dracaena[
        dracaena_cols
    ]
)


# =============================================================================
# 5. DRACAENA NUMERIC STRUCTURE
# =============================================================================

dracaena_summary = pd.DataFrame({
    "Metric": [
        "N_Measurements",
        "N_Unique_Values",
        "N_Datasets",
        "N_References",
        "Min_mg",
        "Median_mg",
        "Mean_mg",
        "Max_mg",
        "Fold_Range",
    ],
    "Value": [
        len(dracaena),
        dracaena["StdValue_Numeric"].nunique(),
        dracaena["DatasetID"].nunique()
            if "DatasetID" in dracaena.columns else pd.NA,
        dracaena["Reference"].nunique()
            if "Reference" in dracaena.columns else pd.NA,
        dracaena["StdValue_Numeric"].min(),
        dracaena["StdValue_Numeric"].median(),
        dracaena["StdValue_Numeric"].mean(),
        dracaena["StdValue_Numeric"].max(),
        (
            dracaena["StdValue_Numeric"].max()
            /
            dracaena["StdValue_Numeric"].min()
        ),
    ],
})

print("\nDracaena mannii variability summary")
print("-" * 100)

display(dracaena_summary)


# =============================================================================
# 6. DRACAENA ASSOCIATED METADATA
# =============================================================================

dracaena_obs = set(
    dracaena["ObservationID_Link"]
    .dropna()
)

dracaena_metadata = metadata[
    metadata["ObservationID_Link"]
    .isin(dracaena_obs)
].copy()

print("\nDracaena mannii associated metadata")
print("-" * 100)

print(
    f"Metadata rows : {len(dracaena_metadata)}"
)

if len(dracaena_metadata) > 0:

    dracaena_meta_cols = [
        col for col in [
            "ObservationID",
            "AccSpeciesID",
            "AccSpeciesName",
            "DataID",
            "DataName",
            "OriglName",
            "OrigValueStr",
            "OrigUnitStr",
            "Reference",
            "Comment",
        ]
        if col in dracaena_metadata.columns
    ]

    display(
        dracaena_metadata[
            dracaena_meta_cols
        ]
        .sort_values(
            [
                "ObservationID",
                "DataName",
            ]
        )
        .reset_index(drop=True)
    )


# =============================================================================
# 7. SOURCE-LEVEL DRACAENA SUMMARY
# =============================================================================

source_group_cols = [
    col for col in [
        "DatasetID",
        "Dataset",
        "Reference",
    ]
    if col in dracaena.columns
]

if source_group_cols:

    dracaena_source_summary = (
        dracaena
        .groupby(
            source_group_cols,
            dropna=False
        )
        .agg(
            N_Measurements=(
                "StdValue_Numeric",
                "size"
            ),
            Min_mg=(
                "StdValue_Numeric",
                "min"
            ),
            Median_mg=(
                "StdValue_Numeric",
                "median"
            ),
            Max_mg=(
                "StdValue_Numeric",
                "max"
            ),
        )
        .reset_index()
    )

else:

    dracaena_source_summary = pd.DataFrame()

print("\nDracaena mannii source-level structure")
print("-" * 100)

if len(dracaena_source_summary) > 0:
    display(dracaena_source_summary)
else:
    print("No source grouping fields available.")


# =============================================================================
# 8. EXTRACT HELD CELTIS ZENKERI RECORD
# =============================================================================

celtis_hold = value_review[
    value_review["AccSpeciesName"]
    .astype("string")
    .str.strip()
    .eq("Celtis zenkeri")
].copy()

assert len(celtis_hold) == 1, (
    f"Expected one Celtis zenkeri held record, found {len(celtis_hold)}."
)

print("\nHeld Celtis zenkeri TRY record")
print("-" * 100)

celtis_cols = [
    col for col in [
        "TRY_Seed_Mass_Record_Key",
        "ObservationID",
        "ObsDataID",
        "DataID",
        "DatasetID",
        "Dataset",
        "SpeciesName",
        "AccSpeciesID",
        "AccSpeciesName",
        "OriglName",
        "OrigValueStr",
        "OrigUnitStr",
        "StdValue",
        "UnitName",
        "ValueKindName",
        "Reference",
        "Comment",
        "ErrorRisk",
        "TRY_Raw_Value_Status",
    ]
    if col in celtis_hold.columns
]

display(
    celtis_hold[
        celtis_cols
    ]
)


# =============================================================================
# 9. CHECK CELTIS OBSERVATION METADATA
# =============================================================================

celtis_hold["ObservationID_Link"] = (
    celtis_hold["ObservationID"]
    .astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
)

celtis_obs = set(
    celtis_hold["ObservationID_Link"]
    .dropna()
)

celtis_metadata = metadata[
    metadata["ObservationID_Link"]
    .isin(celtis_obs)
].copy()

print("\nCeltis zenkeri held-record metadata")
print("-" * 100)

print(
    f"Metadata rows : {len(celtis_metadata)}"
)

if len(celtis_metadata) > 0:

    celtis_meta_cols = [
        col for col in [
            "ObservationID",
            "AccSpeciesID",
            "AccSpeciesName",
            "DataID",
            "DataName",
            "OriglName",
            "OrigValueStr",
            "OrigUnitStr",
            "Reference",
            "Comment",
        ]
        if col in celtis_metadata.columns
    ]

    display(
        celtis_metadata[
            celtis_meta_cols
        ]
        .sort_values("DataName")
        .reset_index(drop=True)
    )


# =============================================================================
# 10. CELTIS EXISTING NUMERIC EVIDENCE
# =============================================================================
#
# Compare the held record with the three usable Celtis zenkeri records already
# present in the 177-record QC set.
# =============================================================================

celtis_numeric = qc[
    qc["AccSpeciesName"]
    .astype("string")
    .str.strip()
    .eq("Celtis zenkeri")
].copy()

assert len(celtis_numeric) == 3, (
    f"Expected 3 numeric Celtis zenkeri records, found {len(celtis_numeric)}."
)

print("\nExisting numeric Celtis zenkeri measurements")
print("-" * 100)

display(
    celtis_numeric[
        [
            col for col in [
                "TRY_Seed_Mass_Record_Key",
                "ObservationID",
                "ObsDataID",
                "StdValue_Numeric",
                "UnitName",
                "Dataset",
                "Reference",
            ]
            if col in celtis_numeric.columns
        ]
    ]
    .sort_values("StdValue_Numeric")
    .reset_index(drop=True)
)


# =============================================================================
# 11. DIAGNOSE WHETHER HELD CELTIS VALUE IS RECOVERABLE
# =============================================================================

held_orig_numeric = pd.to_numeric(
    celtis_hold["OrigValueStr"],
    errors="coerce"
)

held_std_numeric = pd.to_numeric(
    celtis_hold["StdValue"],
    errors="coerce"
)

celtis_hold["Original_Value_Numeric"] = held_orig_numeric

celtis_hold["Held_Value_Diagnostic"] = np.where(
    held_std_numeric.notna(),
    "STANDARDIZED_VALUE_ALREADY_NUMERIC",
    np.where(
        held_orig_numeric.notna(),
        "ORIGINAL_VALUE_NUMERIC_STDVALUE_MISSING",
        "NO_NUMERIC_VALUE_AVAILABLE"
    )
)

print("\nCeltis held-value diagnostic")
print("-" * 100)

display(
    celtis_hold[
        [
            col for col in [
                "ObservationID",
                "ObsDataID",
                "OriglName",
                "OrigValueStr",
                "OrigUnitStr",
                "StdValue",
                "UnitName",
                "Original_Value_Numeric",
                "Held_Value_Diagnostic",
            ]
            if col in celtis_hold.columns
        ]
    ]
)


# =============================================================================
# 12. EXCEPTIONAL-RECORD REVIEW ROUTING
# =============================================================================
#
# Still no accept/reject decisions.
# We simply route the exceptional cases to the next decision step.
# =============================================================================

dracaena_fold = (
    dracaena["StdValue_Numeric"].max()
    /
    dracaena["StdValue_Numeric"].min()
)

exception_summary = pd.DataFrame({
    "Taxon": [
        "Dracaena mannii",
        "Celtis zenkeri",
    ],
    "Issue": [
        "HIGH_WITHIN_SPECIES_VARIABILITY",
        "MISSING_STANDARDIZED_VALUE",
    ],
    "N_Affected_Records": [
        len(dracaena),
        len(celtis_hold),
    ],
    "Diagnostic_Value": [
        dracaena_fold,
        (
            celtis_hold[
                "Held_Value_Diagnostic"
            ].iloc[0]
        ),
    ],
    "Next_Action": [
        (
            "REVIEW_SOURCE_AND_MEASUREMENT_CONTEXT_"
            "BEFORE_AGGREGATION"
        ),
        (
            "DETERMINE_IF_ORIGINAL_VALUE_IS_SAFELY_"
            "RECOVERABLE_OR_KEEP_EXCLUDED"
        ),
    ],
})

print("\nExceptional-record review routing")
print("-" * 100)

display(exception_summary)


# =============================================================================
# 13. EXPORT
# =============================================================================

dracaena.to_csv(
    OUT_DRACAENA_REVIEW,
    index=False
)

dracaena_metadata.to_csv(
    OUT_DRACAENA_METADATA,
    index=False
)

celtis_hold.to_csv(
    OUT_CELTIS_REVIEW,
    index=False
)

exception_summary.to_csv(
    OUT_EXCEPTION_SUMMARY,
    index=False
)


print("\nExceptional-review exports")
print("-" * 100)

print(
    f"Dracaena review    : "
    f"{OUT_DRACAENA_REVIEW}"
)

print(
    f"Dracaena metadata  : "
    f"{OUT_DRACAENA_METADATA}"
)

print(
    f"Celtis value review: "
    f"{OUT_CELTIS_REVIEW}"
)

print(
    f"Review summary     : "
    f"{OUT_EXCEPTION_SUMMARY}"
)


# =============================================================================
# 14. FINAL VALIDATION
# =============================================================================

assert len(dracaena) == 4
assert len(celtis_hold) == 1
assert len(celtis_numeric) == 3

assert dracaena_fold >= 10

print("\nExceptional TRY Seed Mass record review completed successfully.")
print("No exceptional record has yet been accepted or rejected.")
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3I — CROSS-DATASET SCIENTIFIC DEDUPLICATION
#                AND EXCEPTIONAL-RECORD RESOLUTION
# =============================================================================
#
# Purpose
# -------
# Resolve the remaining exceptional-record issues and prepare a defensible
# TRY measurement pool for species-level aggregation.
#
# Key principles
# --------------
# 1. The non-numeric Celtis zenkeri record remains preserved but cannot enter
#    numerical aggregation.
#
# 2. Distinct TRY rows are NOT automatically independent scientific evidence.
#    The same underlying biological value may be represented in multiple TRY
#    dataset versions.
#
# 3. Cross-dataset duplicate candidates are detected conservatively using:
#
#       accepted TRY species
#       standardized value
#       standardized unit
#       original-value context
#       source/reference lineage
#
# 4. Exact repeated values are NOT removed merely because values are equal.
#    Deduplication requires supporting source/context similarity.
#
# 5. No species-level Seed Mass estimate is calculated in this section.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print(
    "SECTION 9.3I — CROSS-DATASET SCIENTIFIC DEDUPLICATION "
    "AND EXCEPTIONAL-RECORD RESOLUTION"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

QC_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_qc_register.csv"
)

CELTIS_REVIEW_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_celtis_zenkeri_value_review.csv"
)

DRACAENA_REVIEW_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_dracaena_mannii_variability_review.csv"
)

OUT_DEDUP_AUDIT = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_cross_dataset_deduplication_audit.csv"
)

OUT_DUPLICATE_GROUPS = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_cross_dataset_duplicate_groups.csv"
)

OUT_FINAL_MEASUREMENTS = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurements_ready_for_species_aggregation.csv"
)

OUT_EXCLUDED_DUPLICATES = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_duplicate_measurements_excluded.csv"
)

OUT_CELTIS_RESOLUTION = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_celtis_value_hold_resolved.csv"
)

OUT_SPECIES_POSTDEDUP = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_post_dedup_species_summary.csv"
)

OUT_DEDUP_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_cross_dataset_dedup_summary.csv"
)


# =============================================================================
# 2. LOAD INPUTS
# =============================================================================

for path in [
    QC_REGISTER_FILE,
    CELTIS_REVIEW_FILE,
    DRACAENA_REVIEW_FILE,
]:
    assert path.exists(), (
        f"Required file not found:\n{path}"
    )

qc = pd.read_csv(
    QC_REGISTER_FILE,
    dtype="string",
    low_memory=False
)

celtis_hold = pd.read_csv(
    CELTIS_REVIEW_FILE,
    dtype="string",
    low_memory=False
)

dracaena_review = pd.read_csv(
    DRACAENA_REVIEW_FILE,
    dtype="string",
    low_memory=False
)


print("\nInput evidence")
print("-" * 100)

print(f"Positive numeric TRY measurements : {len(qc)}")
print(f"Celtis value-review records        : {len(celtis_hold)}")
print(f"Dracaena review records            : {len(dracaena_review)}")

assert len(qc) == 177
assert len(celtis_hold) == 1
assert len(dracaena_review) == 4


# =============================================================================
# 3. NUMERIC NORMALIZATION
# =============================================================================

qc["StdValue_Numeric"] = pd.to_numeric(
    qc["StdValue"],
    errors="coerce"
)

qc["AccSpeciesID_Numeric"] = pd.to_numeric(
    qc["AccSpeciesID"],
    errors="coerce"
)

assert qc["StdValue_Numeric"].notna().all()
assert qc["StdValue_Numeric"].gt(0).all()


# =============================================================================
# 4. FORMALLY RESOLVE THE CELTIS NON-NUMERIC HOLD
# =============================================================================
#
# Section 9.3H confirmed:
#
#     OrigValueStr : missing
#     StdValue     : missing
#
# Therefore no numerical value can be reconstructed without invention.
# =============================================================================

celtis_hold["Exceptional_Record_Decision"] = (
    "EXCLUDE_FROM_NUMERIC_AGGREGATION"
)

celtis_hold["Evidence_Action"] = (
    "RETAIN_IN_EVIDENCE_LEDGER_ONLY"
)

celtis_hold["Resolution_Rationale"] = (
    "TRY Seed dry mass record contains neither a numeric original value "
    "nor a numeric standardized value; no defensible numerical recovery "
    "is possible."
)

print("\nCeltis zenkeri value-hold resolution")
print("-" * 100)

display(
    celtis_hold[
        [
            col for col in [
                "TRY_Seed_Mass_Record_Key",
                "ObservationID",
                "ObsDataID",
                "AccSpeciesName",
                "OrigValueStr",
                "StdValue",
                "UnitName",
                "Exceptional_Record_Decision",
                "Evidence_Action",
                "Resolution_Rationale",
            ]
            if col in celtis_hold.columns
        ]
    ]
)


# =============================================================================
# 5. NORMALIZE SOURCE TEXT FOR CROSS-DATASET DUPLICATE DIAGNOSIS
# =============================================================================

def normalize_text(value):

    if pd.isna(value):
        return ""

    value = str(value).lower().strip()

    value = re.sub(
        r"https?://\S+",
        "",
        value
    )

    value = re.sub(
        r"\b20\d{2}\b|\b19\d{2}\b",
        "",
        value
    )

    value = re.sub(
        r"[^a-z0-9]+",
        " ",
        value
    )

    value = re.sub(
        r"\s+",
        " ",
        value
    ).strip()

    return value


for col in [
    "Dataset",
    "Reference",
    "OriglName",
    "OrigValueStr",
    "OrigUnitStr",
    "Comment",
]:

    if col not in qc.columns:
        qc[col] = pd.NA


qc["Reference_Normalized"] = (
    qc["Reference"]
    .apply(normalize_text)
)

qc["Original_Name_Normalized"] = (
    qc["OriglName"]
    .apply(normalize_text)
)

qc["Original_Unit_Normalized"] = (
    qc["OrigUnitStr"]
    .apply(normalize_text)
)


# =============================================================================
# 6. CREATE CONSERVATIVE SCIENTIFIC DUPLICATE SIGNATURE
# =============================================================================
#
# ObservationID and DatasetID are intentionally excluded here.
#
# The purpose is to detect cases where the SAME measurement is reproduced
# through multiple TRY dataset versions.
#
# We use rounded standardized value only to avoid string-format differences.
# =============================================================================

qc["StdValue_Signature"] = (
    qc["StdValue_Numeric"]
    .round(9)
    .astype(str)
)


qc["Cross_Dataset_Scientific_Signature"] = (
    qc["AccSpeciesID"]
    .fillna("<MISSING>")
    .astype(str)
    .str.strip()
    + " || "
    + qc["StdValue_Signature"]
    + " || "
    + qc["UnitName"]
    .fillna("<MISSING>")
    .astype(str)
    .str.strip()
    .str.lower()
    + " || "
    + qc["Original_Name_Normalized"]
    + " || "
    + qc["Original_Unit_Normalized"]
)


# =============================================================================
# 7. IDENTIFY REPEATED CROSS-DATASET SIGNATURES
# =============================================================================

signature_counts = (
    qc[
        "Cross_Dataset_Scientific_Signature"
    ]
    .value_counts()
)

repeated_signatures = set(
    signature_counts[
        signature_counts > 1
    ].index
)


qc["Cross_Dataset_Duplicate_Candidate"] = (
    qc[
        "Cross_Dataset_Scientific_Signature"
    ]
    .isin(
        repeated_signatures
    )
)


duplicate_candidates = qc[
    qc[
        "Cross_Dataset_Duplicate_Candidate"
    ]
].copy()


print("\nCross-dataset duplicate-candidate diagnostic")
print("-" * 100)

print(
    f"Repeated scientific signatures : "
    f"{len(repeated_signatures)}"
)

print(
    f"Rows in candidate groups        : "
    f"{len(duplicate_candidates)}"
)


if len(duplicate_candidates) > 0:

    candidate_cols = [
        col for col in [
            "TRY_Seed_Mass_Record_Key",
            "ObservationID",
            "ObsDataID",
            "DatasetID",
            "Dataset",
            "AccSpeciesID",
            "AccSpeciesName",
            "OriglName",
            "OrigValueStr",
            "OrigUnitStr",
            "StdValue_Numeric",
            "UnitName",
            "Reference",
            "Cross_Dataset_Scientific_Signature",
        ]
        if col in duplicate_candidates.columns
    ]

    display(
        duplicate_candidates[
            candidate_cols
        ]
        .sort_values(
            [
                "AccSpeciesName",
                "StdValue_Numeric",
                "DatasetID",
            ]
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 8. COUNT DISTINCT DATASETS / REFERENCES WITHIN CANDIDATE GROUPS
# =============================================================================

if len(duplicate_candidates) > 0:

    duplicate_group_summary = (
        duplicate_candidates
        .groupby(
            "Cross_Dataset_Scientific_Signature",
            dropna=False
        )
        .agg(
            AccSpeciesID=(
                "AccSpeciesID",
                "first"
            ),

            AccSpeciesName=(
                "AccSpeciesName",
                "first"
            ),

            StdValue_mg=(
                "StdValue_Numeric",
                "first"
            ),

            N_Rows=(
                "TRY_Seed_Mass_Record_Key",
                "size"
            ),

            N_Datasets=(
                "DatasetID",
                "nunique"
            ),

            N_ObservationIDs=(
                "ObservationID",
                "nunique"
            ),

            N_References=(
                "Reference",
                "nunique"
            ),
        )
        .reset_index()
    )

else:

    duplicate_group_summary = pd.DataFrame()


print("\nCross-dataset duplicate-group summary")
print("-" * 100)

if len(duplicate_group_summary) > 0:
    display(
        duplicate_group_summary
        .sort_values(
            [
                "AccSpeciesName",
                "StdValue_mg",
            ]
        )
        .reset_index(drop=True)
    )
else:
    print("NONE")


# =============================================================================
# 9. INITIAL DEDUPLICATION DECISION
# =============================================================================
#
# Conservative rule:
#
# A repeated signature is collapsed only when:
#
#   - same accepted TRY species
#   - same standardized value
#   - same standardized unit
#   - same original measurement concept/unit
#
# This is interpreted as duplicated representation of the same measurement
# across TRY dataset versions.
#
# One record is retained deterministically; all others remain in the ledger
# with explicit duplicate status.
# =============================================================================

qc["Deduplication_Decision"] = "RETAIN"

qc["Deduplication_Rationale"] = (
    "UNIQUE_OR_DISTINCT_TRY_MEASUREMENT"
)


if repeated_signatures:

    repeated_mask = (
        qc[
            "Cross_Dataset_Scientific_Signature"
        ]
        .isin(
            repeated_signatures
        )
    )

    repeated = qc[
        repeated_mask
    ].copy()

    # Prefer the most recent/highest DatasetID numerically where possible,
    # only as a deterministic representation of an otherwise duplicated
    # scientific value.
    repeated["_DatasetID_Numeric"] = pd.to_numeric(
        repeated["DatasetID"],
        errors="coerce"
    )

    repeated = repeated.sort_values(
        [
            "Cross_Dataset_Scientific_Signature",
            "_DatasetID_Numeric",
            "TRY_Seed_Mass_Record_Key",
        ],
        ascending=[
            True,
            False,
            True,
        ]
    )

    keep_keys = (
        repeated
        .groupby(
            "Cross_Dataset_Scientific_Signature",
            as_index=False
        )
        .first()[
            "TRY_Seed_Mass_Record_Key"
        ]
    )

    duplicate_exclude_mask = (
        repeated_mask
        &
        ~qc[
            "TRY_Seed_Mass_Record_Key"
        ].isin(
            set(keep_keys)
        )
    )

    duplicate_keep_mask = (
        repeated_mask
        &
        qc[
            "TRY_Seed_Mass_Record_Key"
        ].isin(
            set(keep_keys)
        )
    )

    qc.loc[
        duplicate_keep_mask,
        "Deduplication_Decision"
    ] = "RETAIN_REPRESENTATIVE"

    qc.loc[
        duplicate_keep_mask,
        "Deduplication_Rationale"
    ] = (
        "REPRESENTATIVE_OF_REPEATED_CROSS_DATASET_SCIENTIFIC_SIGNATURE"
    )

    qc.loc[
        duplicate_exclude_mask,
        "Deduplication_Decision"
    ] = "EXCLUDE_DUPLICATE_REPRESENTATION"

    qc.loc[
        duplicate_exclude_mask,
        "Deduplication_Rationale"
    ] = (
        "SAME_SPECIES_VALUE_UNIT_AND_MEASUREMENT_CONTEXT_REPRESENTED_"
        "IN_ANOTHER_TRY_DATASET_RECORD"
    )


# =============================================================================
# 10. INSPECT DRACAENA MANNII AFTER DEDUPLICATION
# =============================================================================

dracaena_post = qc[
    qc["AccSpeciesName"]
    .astype("string")
    .str.strip()
    .eq("Dracaena mannii")
].copy()


print("\nDracaena mannii deduplication decisions")
print("-" * 100)

display(
    dracaena_post[
        [
            col for col in [
                "TRY_Seed_Mass_Record_Key",
                "DatasetID",
                "Dataset",
                "ObservationID",
                "StdValue_Numeric",
                "OriglName",
                "OrigUnitStr",
                "Reference",
                "Deduplication_Decision",
                "Deduplication_Rationale",
            ]
            if col in dracaena_post.columns
        ]
    ]
    .sort_values(
        "StdValue_Numeric"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 11. FINAL MEASUREMENT POOL
# =============================================================================

final_measurements = qc[
    ~qc[
        "Deduplication_Decision"
    ].eq(
        "EXCLUDE_DUPLICATE_REPRESENTATION"
    )
].copy()


excluded_duplicates = qc[
    qc[
        "Deduplication_Decision"
    ].eq(
        "EXCLUDE_DUPLICATE_REPRESENTATION"
    )
].copy()


print("\nCross-dataset deduplication outcome")
print("-" * 100)

decision_summary = (
    qc[
        "Deduplication_Decision"
    ]
    .value_counts()
    .rename_axis(
        "Deduplication_Decision"
    )
    .reset_index(
        name="N_Measurements"
    )
)

display(decision_summary)

print(
    f"\nMeasurements before deduplication : "
    f"{len(qc)}"
)

print(
    f"Duplicate representations removed : "
    f"{len(excluded_duplicates)}"
)

print(
    f"Measurements retained             : "
    f"{len(final_measurements)}"
)


# =============================================================================
# 12. POST-DEDUP SPECIES SUMMARY
# =============================================================================

species_postdedup = (
    final_measurements
    .groupby(
        [
            "AccSpeciesID_Numeric",
            "AccSpeciesName",
        ],
        dropna=False
    )
    .agg(
        N_Retained_Measurements=(
            "StdValue_Numeric",
            "size"
        ),

        N_Unique_Values=(
            "StdValue_Numeric",
            "nunique"
        ),

        N_Datasets=(
            "DatasetID",
            "nunique"
        ),

        Min_mg=(
            "StdValue_Numeric",
            "min"
        ),

        Median_mg=(
            "StdValue_Numeric",
            "median"
        ),

        Mean_mg=(
            "StdValue_Numeric",
            "mean"
        ),

        Max_mg=(
            "StdValue_Numeric",
            "max"
        ),
    )
    .reset_index()
)


species_postdedup[
    "Fold_Range"
] = np.where(
    species_postdedup[
        "Min_mg"
    ].gt(0),
    (
        species_postdedup[
            "Max_mg"
        ]
        /
        species_postdedup[
            "Min_mg"
        ]
    ),
    np.nan
)


print("\nPost-deduplication species evidence summary")
print("-" * 100)

display(
    species_postdedup
    .sort_values(
        "AccSpeciesName"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 13. DRACAENA POST-DEDUP SUMMARY
# =============================================================================

dracaena_final = final_measurements[
    final_measurements[
        "AccSpeciesName"
    ]
    .astype("string")
    .str.strip()
    .eq(
        "Dracaena mannii"
    )
].copy()


print("\nDracaena mannii evidence after cross-dataset deduplication")
print("-" * 100)

print(
    f"Measurements retained : "
    f"{len(dracaena_final)}"
)

print(
    f"Unique values         : "
    f"{dracaena_final['StdValue_Numeric'].nunique()}"
)

print(
    f"Values (mg)           : "
    f"{sorted(dracaena_final['StdValue_Numeric'].tolist())}"
)

print(
    f"Median (mg)           : "
    f"{dracaena_final['StdValue_Numeric'].median():.6f}"
)

print(
    f"Fold range            : "
    f"{dracaena_final['StdValue_Numeric'].max() / dracaena_final['StdValue_Numeric'].min():.6f}"
)


# =============================================================================
# 14. BUILD DEDUPLICATION SUMMARY
# =============================================================================

dedup_summary = pd.DataFrame({
    "Metric": [
        "Positive numeric measurements entering Section 9.3I",
        "Cross-dataset repeated scientific signatures",
        "Rows in repeated-signature groups",
        "Duplicate representations excluded",
        "Measurements retained for species aggregation",
        "TRY accepted species represented",
        "Celtis non-numeric records retained in ledger only",
    ],
    "Value": [
        len(qc),
        len(repeated_signatures),
        len(duplicate_candidates),
        len(excluded_duplicates),
        len(final_measurements),
        final_measurements[
            "AccSpeciesID_Numeric"
        ].nunique(),
        len(celtis_hold),
    ],
})


print("\nCross-dataset deduplication summary")
print("-" * 100)

display(dedup_summary)


# =============================================================================
# 15. VALIDATION
# =============================================================================

assert len(qc) == 177

assert len(celtis_hold) == 1

assert (
    len(final_measurements)
    + len(excluded_duplicates)
    == 177
)

assert final_measurements[
    "TRY_Seed_Mass_Record_Key"
].is_unique

assert final_measurements[
    "StdValue_Numeric"
].notna().all()

assert final_measurements[
    "StdValue_Numeric"
].gt(0).all()

assert (
    final_measurements[
        "UnitName"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
)


# =============================================================================
# 16. EXPORT
# =============================================================================

qc.to_csv(
    OUT_DEDUP_AUDIT,
    index=False
)

duplicate_group_summary.to_csv(
    OUT_DUPLICATE_GROUPS,
    index=False
)

final_measurements.to_csv(
    OUT_FINAL_MEASUREMENTS,
    index=False
)

excluded_duplicates.to_csv(
    OUT_EXCLUDED_DUPLICATES,
    index=False
)

celtis_hold.to_csv(
    OUT_CELTIS_RESOLUTION,
    index=False
)

species_postdedup.to_csv(
    OUT_SPECIES_POSTDEDUP,
    index=False
)

dedup_summary.to_csv(
    OUT_DEDUP_SUMMARY,
    index=False
)


# =============================================================================
# 17. EXPORT REPORT
# =============================================================================

print("\nSection 9.3I exports")
print("-" * 100)

print(
    f"Deduplication audit    : "
    f"{OUT_DEDUP_AUDIT}"
)

print(
    f"Duplicate groups       : "
    f"{OUT_DUPLICATE_GROUPS}"
)

print(
    f"Final measurement pool : "
    f"{OUT_FINAL_MEASUREMENTS}"
)

print(
    f"Excluded duplicates    : "
    f"{OUT_EXCLUDED_DUPLICATES}"
)

print(
    f"Celtis resolution      : "
    f"{OUT_CELTIS_RESOLUTION}"
)

print(
    f"Species summary        : "
    f"{OUT_SPECIES_POSTDEDUP}"
)

print(
    f"Dedup summary          : "
    f"{OUT_DEDUP_SUMMARY}"
)


# =============================================================================
# 18. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Seed Mass exceptional-record resolution and "
    "cross-dataset scientific deduplication completed successfully."
)

print(
    f"Positive numeric TRY records assessed : "
    f"{len(qc)}"
)

print(
    f"Duplicate representations excluded    : "
    f"{len(excluded_duplicates)}"
)

print(
    f"Measurements ready for aggregation    : "
    f"{len(final_measurements)}"
)

print(
    "The non-numeric Celtis zenkeri record remains preserved "
    "in the evidence ledger but excluded from numerical aggregation."
)

print(
    "No species-level Seed Mass estimate has yet been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3I-R — SOURCE-LINEAGE-AWARE DUPLICATE RECONCILIATION
# =============================================================================
#
# Purpose
# -------
# Correct the preliminary Section 9.3I deduplication strategy.
#
# Section 9.3I showed that TRY contains multiple representations of Kew
# Seed Information Database (SID) evidence across dataset versions.
#
# The earlier signature included OriglName, which incorrectly separated
# equivalent Kew records labelled:
#
#     "Seed Mass"
#     "Mean 1000 Seed Weight"
#     "1000 Seed Weight (g)"
#
# This section therefore performs a SOURCE-LINEAGE-aware reconciliation.
#
# Scientific rule
# ---------------
# KEW SID:
#     same accepted TRY species
#     + same standardized Seed Mass value
#     + same standardized unit
#     = same database evidence signature across SID versions
#
# NON-KEW:
#     retain independent records unless the SAME explicit source/reference
#     and same standardized value indicate an exact repeated representation.
#
# IMPORTANT
# ---------
# - Start again from the 177 positive numeric QC records.
# - Do not use the preliminary 111-row Section 9.3I pool as input.
# - No species-level aggregation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print(
    "SECTION 9.3I-R — SOURCE-LINEAGE-AWARE DUPLICATE RECONCILIATION"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

QC_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurement_qc_register.csv"
)

CELTIS_RESOLUTION_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_celtis_value_hold_resolved.csv"
)

OUT_LINEAGE_AUDIT = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_source_lineage_dedup_audit.csv"
)

OUT_DUPLICATE_GROUPS = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_source_lineage_duplicate_groups.csv"
)

OUT_RETAINED = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurements_ready_for_species_aggregation.csv"
)

OUT_EXCLUDED = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_source_lineage_duplicates_excluded.csv"
)

OUT_SPECIES_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_source_lineage_post_dedup_species_summary.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_source_lineage_dedup_summary.csv"
)


# =============================================================================
# 2. LOAD ORIGINAL 177-RECORD QC EVIDENCE
# =============================================================================

assert QC_REGISTER_FILE.exists(), (
    f"QC register not found:\n{QC_REGISTER_FILE}"
)

qc = pd.read_csv(
    QC_REGISTER_FILE,
    dtype="string",
    low_memory=False
)

assert len(qc) == 177, (
    f"Expected 177 positive numeric TRY records, found {len(qc)}."
)

print("\nInput evidence")
print("-" * 100)

print(
    f"Independent positive numeric TRY records : "
    f"{len(qc)}"
)


# =============================================================================
# 3. NUMERIC NORMALIZATION
# =============================================================================

qc["StdValue_Numeric"] = pd.to_numeric(
    qc["StdValue"],
    errors="coerce"
)

qc["AccSpeciesID_Numeric"] = pd.to_numeric(
    qc["AccSpeciesID"],
    errors="coerce"
)

assert qc["StdValue_Numeric"].notna().all()
assert qc["StdValue_Numeric"].gt(0).all()

assert (
    qc["TRY_Seed_Mass_Record_Key"]
    .is_unique
)


# =============================================================================
# 4. NORMALIZE SOURCE TEXT
# =============================================================================

for col in [
    "DatasetID",
    "Dataset",
    "Reference",
    "OriglName",
    "OrigUnitStr",
]:
    if col not in qc.columns:
        qc[col] = pd.NA


def normalize_source_text(x):

    if pd.isna(x):
        return ""

    x = str(x).lower().strip()

    x = re.sub(
        r"https?://\S+",
        " ",
        x
    )

    x = re.sub(
        r"[^a-z0-9]+",
        " ",
        x
    )

    x = re.sub(
        r"\s+",
        " ",
        x
    ).strip()

    return x


qc["Dataset_Normalized"] = (
    qc["Dataset"]
    .apply(normalize_source_text)
)

qc["Reference_Normalized"] = (
    qc["Reference"]
    .apply(normalize_source_text)
)


# =============================================================================
# 5. IDENTIFY KEW SID SOURCE LINEAGE
# =============================================================================
#
# Dataset IDs observed in this TRY return include several historical/current
# Kew SID representations, including:
#
#     21
#     119
#     565
#
# We classify using both Dataset and Reference text rather than hard-coding
# only those IDs.
# =============================================================================

kew_text = (
    qc["Dataset_Normalized"]
    + " "
    + qc["Reference_Normalized"]
)

qc["Source_Lineage"] = np.where(
    (
        kew_text
        .str.contains(
            r"\bkew\b|seed information database|\bsid\b",
            regex=True,
            na=False
        )
    ),
    "KEW_SEED_INFORMATION_DATABASE",
    "OTHER_SOURCE"
)


print("\nSource-lineage classification")
print("-" * 100)

display(
    qc[
        "Source_Lineage"
    ]
    .value_counts()
    .rename_axis(
        "Source_Lineage"
    )
    .reset_index(
        name="N_Measurements"
    )
)


# =============================================================================
# 6. STANDARDIZE NUMERICAL VALUE SIGNATURE
# =============================================================================

qc["Value_Signature"] = (
    qc[
        "StdValue_Numeric"
    ]
    .round(9)
    .map(
        lambda x: f"{x:.9f}"
    )
)


qc["Unit_Signature"] = (
    qc[
        "UnitName"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
)


# =============================================================================
# 7. BUILD SOURCE-LINEAGE-AWARE SCIENTIFIC SIGNATURE
# =============================================================================
#
# KEW SID:
#
#     Species + standardized value + unit + Kew lineage
#
# This intentionally ignores OriglName and DatasetID.
#
# OTHER SOURCES:
#
#     Species + value + unit + normalized reference + DatasetID
#
# This avoids collapsing coincidentally equal values from genuinely distinct
# studies.
# =============================================================================

kew_mask = (
    qc["Source_Lineage"]
    .eq(
        "KEW_SEED_INFORMATION_DATABASE"
    )
)


qc["Scientific_Evidence_Signature"] = ""


qc.loc[
    kew_mask,
    "Scientific_Evidence_Signature"
] = (
    "KEW_SID || "
    + qc.loc[
        kew_mask,
        "AccSpeciesID"
    ].astype(str)
    + " || "
    + qc.loc[
        kew_mask,
        "Value_Signature"
    ]
    + " || "
    + qc.loc[
        kew_mask,
        "Unit_Signature"
    ]
)


non_kew_mask = ~kew_mask


qc.loc[
    non_kew_mask,
    "Scientific_Evidence_Signature"
] = (
    "OTHER || "
    + qc.loc[
        non_kew_mask,
        "AccSpeciesID"
    ].astype(str)
    + " || "
    + qc.loc[
        non_kew_mask,
        "Value_Signature"
    ]
    + " || "
    + qc.loc[
        non_kew_mask,
        "Unit_Signature"
    ]
    + " || "
    + qc.loc[
        non_kew_mask,
        "DatasetID"
    ].fillna("<MISSING>").astype(str)
    + " || "
    + qc.loc[
        non_kew_mask,
        "Reference_Normalized"
    ]
)


# =============================================================================
# 8. IDENTIFY REPEATED SCIENTIFIC EVIDENCE SIGNATURES
# =============================================================================

signature_counts = (
    qc[
        "Scientific_Evidence_Signature"
    ]
    .value_counts()
)

repeated_signatures = set(
    signature_counts[
        signature_counts > 1
    ].index
)


qc[
    "Repeated_Evidence_Signature"
] = (
    qc[
        "Scientific_Evidence_Signature"
    ]
    .isin(
        repeated_signatures
    )
)


duplicate_candidates = qc[
    qc[
        "Repeated_Evidence_Signature"
    ]
].copy()


print("\nSource-lineage repeated evidence diagnostic")
print("-" * 100)

print(
    f"Repeated evidence signatures : "
    f"{len(repeated_signatures)}"
)

print(
    f"Rows in repeated groups      : "
    f"{len(duplicate_candidates)}"
)


# =============================================================================
# 9. BUILD DUPLICATE-GROUP SUMMARY
# =============================================================================

if len(duplicate_candidates) > 0:

    duplicate_groups = (
        duplicate_candidates
        .groupby(
            "Scientific_Evidence_Signature",
            dropna=False
        )
        .agg(
            AccSpeciesID=(
                "AccSpeciesID",
                "first"
            ),

            AccSpeciesName=(
                "AccSpeciesName",
                "first"
            ),

            Source_Lineage=(
                "Source_Lineage",
                "first"
            ),

            StdValue_mg=(
                "StdValue_Numeric",
                "first"
            ),

            N_Rows=(
                "TRY_Seed_Mass_Record_Key",
                "size"
            ),

            N_Datasets=(
                "DatasetID",
                "nunique"
            ),

            N_ObservationIDs=(
                "ObservationID",
                "nunique"
            ),

            Dataset_IDs=(
                "DatasetID",
                lambda x: " | ".join(
                    sorted(
                        set(
                            x.dropna().astype(str)
                        )
                    )
                )
            ),

            Original_Field_Names=(
                "OriglName",
                lambda x: " | ".join(
                    sorted(
                        set(
                            x.dropna().astype(str)
                        )
                    )
                )
            ),
        )
        .reset_index()
    )

else:

    duplicate_groups = pd.DataFrame()


print("\nSource-lineage duplicate groups")
print("-" * 100)

if len(duplicate_groups) > 0:

    display(
        duplicate_groups
        .sort_values(
            [
                "AccSpeciesName",
                "StdValue_mg",
            ]
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 10. DEDUPLICATION DECISION
# =============================================================================
#
# One representative is retained per repeated scientific evidence signature.
#
# Representative preference for Kew SID:
#
#     Dataset 565 > Dataset 119 > Dataset 21
#
# because 565 is the more recent Kew SID representation visible in the
# returned TRY data.
#
# This preference affects only which duplicate row represents the evidence;
# it does NOT alter the biological value.
# =============================================================================

dataset_priority = {
    "565": 3,
    "119": 2,
    "21": 1,
}


qc[
    "Dataset_Representation_Priority"
] = (
    qc[
        "DatasetID"
    ]
    .astype("string")
    .map(
        dataset_priority
    )
    .fillna(0)
    .astype(int)
)


qc[
    "Deduplication_Decision"
] = "RETAIN_DISTINCT_EVIDENCE"

qc[
    "Deduplication_Rationale"
] = (
    "NO_REPEATED_SOURCE_LINEAGE_EVIDENCE_SIGNATURE"
)


for signature in repeated_signatures:

    group = qc[
        qc[
            "Scientific_Evidence_Signature"
        ].eq(
            signature
        )
    ].copy()

    group = group.sort_values(
        [
            "Dataset_Representation_Priority",
            "TRY_Seed_Mass_Record_Key",
        ],
        ascending=[
            False,
            True,
        ]
    )

    representative_key = (
        group.iloc[0][
            "TRY_Seed_Mass_Record_Key"
        ]
    )

    group_keys = set(
        group[
            "TRY_Seed_Mass_Record_Key"
        ]
    )

    qc.loc[
        qc[
            "TRY_Seed_Mass_Record_Key"
        ].eq(
            representative_key
        ),
        "Deduplication_Decision"
    ] = "RETAIN_REPRESENTATIVE"

    qc.loc[
        qc[
            "TRY_Seed_Mass_Record_Key"
        ].eq(
            representative_key
        ),
        "Deduplication_Rationale"
    ] = (
        "REPRESENTATIVE_OF_REPEATED_SOURCE_LINEAGE_EVIDENCE"
    )

    exclude_keys = (
        group_keys
        - {representative_key}
    )

    qc.loc[
        qc[
            "TRY_Seed_Mass_Record_Key"
        ].isin(
            exclude_keys
        ),
        "Deduplication_Decision"
    ] = "EXCLUDE_DUPLICATE_REPRESENTATION"

    qc.loc[
        qc[
            "TRY_Seed_Mass_Record_Key"
        ].isin(
            exclude_keys
        ),
        "Deduplication_Rationale"
    ] = (
        "REPEATED_REPRESENTATION_OF_SAME_SOURCE_LINEAGE_"
        "SPECIES_AND_STANDARDIZED_VALUE"
    )


# =============================================================================
# 11. CREATE FINAL RETAINED MEASUREMENT POOL
# =============================================================================

retained = qc[
    ~qc[
        "Deduplication_Decision"
    ]
    .eq(
        "EXCLUDE_DUPLICATE_REPRESENTATION"
    )
].copy()


excluded = qc[
    qc[
        "Deduplication_Decision"
    ]
    .eq(
        "EXCLUDE_DUPLICATE_REPRESENTATION"
    )
].copy()


print("\nSource-lineage deduplication outcome")
print("-" * 100)

display(
    qc[
        "Deduplication_Decision"
    ]
    .value_counts()
    .rename_axis(
        "Deduplication_Decision"
    )
    .reset_index(
        name="N_Measurements"
    )
)

print(
    f"\nMeasurements before reconciliation : "
    f"{len(qc)}"
)

print(
    f"Duplicate representations excluded : "
    f"{len(excluded)}"
)

print(
    f"Measurements retained              : "
    f"{len(retained)}"
)


# =============================================================================
# 12. SPECIES-LEVEL POST-DEDUP DIAGNOSTIC
# =============================================================================

species_summary = (
    retained
    .groupby(
        [
            "AccSpeciesID_Numeric",
            "AccSpeciesName",
        ],
        dropna=False
    )
    .agg(
        N_Retained_Measurements=(
            "StdValue_Numeric",
            "size"
        ),

        N_Unique_Values=(
            "StdValue_Numeric",
            "nunique"
        ),

        N_Source_Lineages=(
            "Source_Lineage",
            "nunique"
        ),

        Min_mg=(
            "StdValue_Numeric",
            "min"
        ),

        Median_mg=(
            "StdValue_Numeric",
            "median"
        ),

        Mean_mg=(
            "StdValue_Numeric",
            "mean"
        ),

        Max_mg=(
            "StdValue_Numeric",
            "max"
        ),
    )
    .reset_index()
)


species_summary[
    "Fold_Range"
] = np.where(
    species_summary[
        "Min_mg"
    ].gt(0),

    species_summary[
        "Max_mg"
    ]
    /
    species_summary[
        "Min_mg"
    ],

    np.nan
)


print("\nPost-reconciliation species evidence summary")
print("-" * 100)

display(
    species_summary
    .sort_values(
        "AccSpeciesName"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 13. DRACAENA MANNII CHECK
# =============================================================================

dracaena = retained[
    retained[
        "AccSpeciesName"
    ]
    .astype("string")
    .str.strip()
    .eq(
        "Dracaena mannii"
    )
].copy()


print("\nDracaena mannii after source-lineage reconciliation")
print("-" * 100)

display(
    dracaena[
        [
            col
            for col in [
                "TRY_Seed_Mass_Record_Key",
                "DatasetID",
                "Dataset",
                "ObservationID",
                "StdValue_Numeric",
                "OriglName",
                "OrigUnitStr",
                "Source_Lineage",
                "Deduplication_Decision",
            ]
            if col in dracaena.columns
        ]
    ]
    .sort_values(
        "StdValue_Numeric"
    )
    .reset_index(drop=True)
)


print(
    f"\nDracaena retained measurements : "
    f"{len(dracaena)}"
)

print(
    f"Dracaena unique values         : "
    f"{dracaena['StdValue_Numeric'].nunique()}"
)

print(
    f"Dracaena values (mg)           : "
    f"{sorted(dracaena['StdValue_Numeric'].tolist())}"
)

print(
    f"Dracaena median (mg)           : "
    f"{dracaena['StdValue_Numeric'].median():.6f}"
)


# =============================================================================
# 14. SUMMARY
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Positive numeric TRY records entering reconciliation",
        "Kew SID records",
        "Other-source records",
        "Repeated scientific evidence signatures",
        "Rows in repeated evidence groups",
        "Duplicate representations excluded",
        "Measurements retained for species aggregation",
        "TRY accepted species represented",
        "Dracaena mannii retained measurements",
    ],
    "Value": [
        len(qc),

        int(
            qc[
                "Source_Lineage"
            ]
            .eq(
                "KEW_SEED_INFORMATION_DATABASE"
            )
            .sum()
        ),

        int(
            qc[
                "Source_Lineage"
            ]
            .eq(
                "OTHER_SOURCE"
            )
            .sum()
        ),

        len(repeated_signatures),

        len(duplicate_candidates),

        len(excluded),

        len(retained),

        retained[
            "AccSpeciesID_Numeric"
        ].nunique(),

        len(dracaena),
    ],
})


print("\nSource-lineage reconciliation summary")
print("-" * 100)

display(summary)


# =============================================================================
# 15. STRONG VALIDATION
# =============================================================================

assert len(qc) == 177

assert (
    len(retained)
    + len(excluded)
    == 177
)

assert retained[
    "TRY_Seed_Mass_Record_Key"
].is_unique

assert retained[
    "StdValue_Numeric"
].notna().all()

assert retained[
    "StdValue_Numeric"
].gt(0).all()

assert (
    retained[
        "UnitName"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
)

assert (
    retained[
        "AccSpeciesID_Numeric"
    ]
    .nunique()
    == 33
), (
    "Source-lineage deduplication unexpectedly removed "
    "all evidence for one or more TRY taxa."
)

# Crucial Dracaena check:
assert (
    dracaena[
        "StdValue_Numeric"
    ]
    .value_counts()
    .max()
    == 1
), (
    "Dracaena mannii still contains repeated identical "
    "Kew SID value representations."
)


# =============================================================================
# 16. EXPORT
# =============================================================================

qc.to_csv(
    OUT_LINEAGE_AUDIT,
    index=False
)

duplicate_groups.to_csv(
    OUT_DUPLICATE_GROUPS,
    index=False
)

retained.to_csv(
    OUT_RETAINED,
    index=False
)

excluded.to_csv(
    OUT_EXCLUDED,
    index=False
)

species_summary.to_csv(
    OUT_SPECIES_SUMMARY,
    index=False
)

summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 17. REPORT EXPORTS
# =============================================================================

print("\nSection 9.3I-R exports")
print("-" * 100)

print(
    f"Source-lineage audit   : "
    f"{OUT_LINEAGE_AUDIT}"
)

print(
    f"Duplicate groups       : "
    f"{OUT_DUPLICATE_GROUPS}"
)

print(
    f"Retained measurements  : "
    f"{OUT_RETAINED}"
)

print(
    f"Excluded duplicates    : "
    f"{OUT_EXCLUDED}"
)

print(
    f"Species summary        : "
    f"{OUT_SPECIES_SUMMARY}"
)

print(
    f"Reconciliation summary : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 18. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Seed Mass source-lineage-aware duplicate reconciliation "
    "completed successfully."
)

print(
    f"Original positive numeric evidence : "
    f"{len(qc)}"
)

print(
    f"Duplicate representations excluded : "
    f"{len(excluded)}"
)

print(
    f"Measurements ready for aggregation : "
    f"{len(retained)}"
)

print(
    f"TRY accepted species retained       : "
    f"{retained['AccSpeciesID_Numeric'].nunique()}"
)

print(
    "No species-level Seed Mass values have yet been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3J — TRY CANDIDATE SPECIES-LEVEL SEED MASS
#                AGGREGATION DIAGNOSTIC
# =============================================================================
#
# Purpose
# -------
# Evaluate candidate species-level aggregation approaches for the
# source-lineage-reconciled TRY Seed Mass evidence.
#
# Input evidence
# --------------
# Reconciled positive numeric measurements : 74
# TRY accepted species represented          : 33
#
# Candidate estimators
# --------------------
# - Arithmetic mean
# - Median
# - Geometric mean
#
# Diagnostics
# -----------
# - Number of retained measurements
# - Number of unique values
# - Number of datasets represented
# - Minimum / maximum
# - Fold range
# - Mean-to-median ratio
# - Geometric-mean-to-median ratio
# - Aggregation sensitivity
#
# IMPORTANT
# ---------
# This section does NOT select the final TRY species-level estimator.
# No species-level values are finalized here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3J — TRY CANDIDATE SPECIES-LEVEL "
    "SEED MASS AGGREGATION DIAGNOSTIC"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

MEASUREMENT_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurements_ready_for_species_aggregation.csv"
)

OUT_AGGREGATION_COMPARISON = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_species_aggregation_comparison.csv"
)

OUT_SENSITIVITY_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_aggregation_sensitivity_review.csv"
)

OUT_SPECIES_STRUCTURE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_species_evidence_structure.csv"
)

OUT_DIAGNOSTIC_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_aggregation_diagnostic_summary.csv"
)


# =============================================================================
# 2. LOAD RECONCILED TRY EVIDENCE
# =============================================================================

assert MEASUREMENT_FILE.exists(), (
    f"Reconciled TRY measurement file not found:\n{MEASUREMENT_FILE}"
)

measurements = pd.read_csv(
    MEASUREMENT_FILE,
    low_memory=False
)

print("\nInput evidence")
print("-" * 100)

print(
    f"Retained measurements : "
    f"{len(measurements)}"
)

print(
    f"TRY accepted species  : "
    f"{measurements['AccSpeciesID'].nunique()}"
)


assert len(measurements) == 74, (
    "Expected 74 source-lineage-reconciled TRY measurements, "
    f"found {len(measurements)}."
)

assert (
    measurements[
        "AccSpeciesID"
    ].nunique()
    == 33
), (
    "Expected 33 TRY accepted species."
)


# =============================================================================
# 3. NUMERIC VALIDATION
# =============================================================================

measurements[
    "Seed_Mass_mg"
] = pd.to_numeric(
    measurements["StdValue"],
    errors="coerce"
)

assert (
    measurements[
        "Seed_Mass_mg"
    ]
    .notna()
    .all()
), (
    "Reconciled TRY measurement pool contains non-numeric Seed Mass values."
)

assert (
    measurements[
        "Seed_Mass_mg"
    ]
    .gt(0)
    .all()
), (
    "Reconciled TRY measurement pool contains non-positive Seed Mass values."
)

assert (
    measurements[
        "UnitName"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
), (
    "Reconciled TRY measurements contain units other than mg."
)


# =============================================================================
# 4. GEOMETRIC-MEAN HELPER
# =============================================================================

def geometric_mean(values):

    values = pd.to_numeric(
        values,
        errors="coerce"
    )

    values = values[
        values.notna()
        &
        values.gt(0)
    ]

    if len(values) == 0:
        return np.nan

    return float(
        np.exp(
            np.mean(
                np.log(values)
            )
        )
    )


# =============================================================================
# 5. SPECIES-LEVEL AGGREGATION COMPARISON
# =============================================================================

species_rows = []

group_fields = [
    "AccSpeciesID",
    "AccSpeciesName",
]

for (
    acc_species_id,
    acc_species_name
), group in measurements.groupby(
    group_fields,
    dropna=False
):

    values = (
        group[
            "Seed_Mass_mg"
        ]
        .dropna()
        .astype(float)
    )

    n_measurements = len(values)

    n_unique_values = (
        values.nunique()
    )

    n_datasets = (
        group[
            "DatasetID"
        ].nunique()
        if "DatasetID" in group.columns
        else np.nan
    )

    n_source_lineages = (
        group[
            "Source_Lineage"
        ].nunique()
        if "Source_Lineage" in group.columns
        else np.nan
    )

    min_value = values.min()

    max_value = values.max()

    median_value = values.median()

    mean_value = values.mean()

    geom_value = geometric_mean(
        values
    )

    fold_range = (
        max_value / min_value
        if min_value > 0
        else np.nan
    )

    mean_median_ratio = (
        mean_value / median_value
        if median_value > 0
        else np.nan
    )

    geom_median_ratio = (
        geom_value / median_value
        if median_value > 0
        else np.nan
    )

    species_rows.append(
        {
            "AccSpeciesID":
                acc_species_id,

            "AccSpeciesName":
                acc_species_name,

            "N_Retained_Measurements":
                n_measurements,

            "N_Unique_Values":
                n_unique_values,

            "N_Datasets":
                n_datasets,

            "N_Source_Lineages":
                n_source_lineages,

            "Min_mg":
                min_value,

            "Arithmetic_Mean_mg":
                mean_value,

            "Median_mg":
                median_value,

            "Geometric_Mean_mg":
                geom_value,

            "Max_mg":
                max_value,

            "Fold_Range":
                fold_range,

            "Mean_vs_Median_Ratio":
                mean_median_ratio,

            "Geom_vs_Median_Ratio":
                geom_median_ratio,
        }
    )


aggregation_comparison = pd.DataFrame(
    species_rows
)


# =============================================================================
# 6. AGGREGATION-SENSITIVITY CLASSIFICATION
# =============================================================================
#
# Sensitivity is flagged when:
#
#   - arithmetic mean differs from median by >50%, OR
#   - geometric mean differs from median by >50%, OR
#   - within-species fold range >=10.
#
# These are diagnostic thresholds only.
# =============================================================================

aggregation_comparison[
    "Aggregation_Sensitivity_Flag"
] = "LOW_AGGREGATION_SENSITIVITY"


mean_sensitive = (
    aggregation_comparison[
        "Mean_vs_Median_Ratio"
    ].gt(1.5)
    |
    aggregation_comparison[
        "Mean_vs_Median_Ratio"
    ].lt(1 / 1.5)
)


geom_sensitive = (
    aggregation_comparison[
        "Geom_vs_Median_Ratio"
    ].gt(1.5)
    |
    aggregation_comparison[
        "Geom_vs_Median_Ratio"
    ].lt(1 / 1.5)
)


range_sensitive = (
    aggregation_comparison[
        "Fold_Range"
    ].ge(10)
)


aggregation_comparison.loc[
    (
        mean_sensitive
        |
        geom_sensitive
        |
        range_sensitive
    ),
    "Aggregation_Sensitivity_Flag"
] = "REVIEW_AGGREGATION_SENSITIVITY"


# =============================================================================
# 7. EVIDENCE-STRENGTH DIAGNOSTIC
# =============================================================================

def classify_evidence(row):

    n = row[
        "N_Retained_Measurements"
    ]

    unique_values = row[
        "N_Unique_Values"
    ]

    if n == 1:
        return "SINGLE_MEASUREMENT"

    if unique_values == 1:
        return "REPEATED_SOURCE_VALUE_COLLAPSED"

    if n >= 3:
        return "MULTIPLE_DISTINCT_VALUES"

    return "TWO_DISTINCT_VALUES"


aggregation_comparison[
    "Evidence_Structure"
] = aggregation_comparison.apply(
    classify_evidence,
    axis=1
)


# =============================================================================
# 8. SORT TABLE
# =============================================================================

aggregation_comparison = (
    aggregation_comparison
    .sort_values(
        "AccSpeciesName"
    )
    .reset_index(drop=True)
)


print("\nCandidate TRY species-level aggregation summaries")
print("-" * 100)

display(
    aggregation_comparison
)


# =============================================================================
# 9. SENSITIVE TAXA
# =============================================================================

sensitivity_review = aggregation_comparison[
    aggregation_comparison[
        "Aggregation_Sensitivity_Flag"
    ].eq(
        "REVIEW_AGGREGATION_SENSITIVITY"
    )
].copy()


print("\nSpecies sensitive to aggregation method or extreme within-species range")
print("-" * 100)

print(
    f"N species : "
    f"{len(sensitivity_review)}"
)

if len(sensitivity_review) > 0:

    display(
        sensitivity_review[
            [
                "AccSpeciesID",
                "AccSpeciesName",
                "N_Retained_Measurements",
                "N_Unique_Values",
                "Min_mg",
                "Arithmetic_Mean_mg",
                "Median_mg",
                "Geometric_Mean_mg",
                "Max_mg",
                "Fold_Range",
                "Mean_vs_Median_Ratio",
                "Geom_vs_Median_Ratio",
                "Aggregation_Sensitivity_Flag",
            ]
        ]
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 10. EVIDENCE-STRUCTURE SUMMARY
# =============================================================================

evidence_structure = (
    aggregation_comparison[
        "Evidence_Structure"
    ]
    .value_counts()
    .rename_axis(
        "Evidence_Structure"
    )
    .reset_index(
        name="N_Species"
    )
)


print("\nTRY species evidence structure")
print("-" * 100)

display(
    evidence_structure
)


# =============================================================================
# 11. SINGLE-MEASUREMENT SPECIES
# =============================================================================

single_measurement_species = (
    aggregation_comparison[
        aggregation_comparison[
            "N_Retained_Measurements"
        ].eq(1)
    ]
    .copy()
)


print("\nTRY species represented by one retained measurement")
print("-" * 100)

print(
    f"N species : "
    f"{len(single_measurement_species)}"
)

if len(single_measurement_species) > 0:

    display(
        single_measurement_species[
            [
                "AccSpeciesID",
                "AccSpeciesName",
                "Median_mg",
                "Evidence_Structure",
            ]
        ]
        .reset_index(drop=True)
    )


# =============================================================================
# 12. MULTIPLE-VALUE SPECIES
# =============================================================================

multiple_value_species = (
    aggregation_comparison[
        aggregation_comparison[
            "N_Unique_Values"
        ].gt(1)
    ]
    .copy()
)


print("\nTRY species with multiple distinct retained values")
print("-" * 100)

print(
    f"N species : "
    f"{len(multiple_value_species)}"
)


# =============================================================================
# 13. GLOBAL DIAGNOSTIC SUMMARY
# =============================================================================

diagnostic_summary = pd.DataFrame({
    "Metric": [
        "Species represented",
        "Retained reconciled measurements",
        "Species with one retained measurement",
        "Species with multiple retained measurements",
        "Species with multiple distinct values",
        "Species with >=10-fold range",
        "Species flagged for aggregation sensitivity",
    ],
    "Value": [
        len(
            aggregation_comparison
        ),

        len(
            measurements
        ),

        int(
            aggregation_comparison[
                "N_Retained_Measurements"
            ]
            .eq(1)
            .sum()
        ),

        int(
            aggregation_comparison[
                "N_Retained_Measurements"
            ]
            .gt(1)
            .sum()
        ),

        int(
            aggregation_comparison[
                "N_Unique_Values"
            ]
            .gt(1)
            .sum()
        ),

        int(
            aggregation_comparison[
                "Fold_Range"
            ]
            .ge(10)
            .sum()
        ),

        len(
            sensitivity_review
        ),
    ],
})


print("\nGlobal TRY aggregation diagnostic")
print("-" * 100)

display(
    diagnostic_summary
)


# =============================================================================
# 14. DRACAENA MANNII EXPLICIT CHECK
# =============================================================================

dracaena_diag = aggregation_comparison[
    aggregation_comparison[
        "AccSpeciesName"
    ]
    .astype(str)
    .str.strip()
    .eq(
        "Dracaena mannii"
    )
].copy()


assert len(dracaena_diag) == 1


print("\nDracaena mannii aggregation diagnostic")
print("-" * 100)

display(
    dracaena_diag
)


# =============================================================================
# 15. VALIDATION
# =============================================================================

assert len(
    aggregation_comparison
) == 33

assert (
    aggregation_comparison[
        "N_Retained_Measurements"
    ].sum()
    == 74
)

assert (
    aggregation_comparison[
        "Median_mg"
    ].notna().all()
)

assert (
    aggregation_comparison[
        "Median_mg"
    ].gt(0).all()
)

assert (
    aggregation_comparison[
        "Geometric_Mean_mg"
    ].notna().all()
)


# =============================================================================
# 16. EXPORT
# =============================================================================

aggregation_comparison.to_csv(
    OUT_AGGREGATION_COMPARISON,
    index=False
)

sensitivity_review.to_csv(
    OUT_SENSITIVITY_REVIEW,
    index=False
)

evidence_structure.to_csv(
    OUT_SPECIES_STRUCTURE,
    index=False
)

diagnostic_summary.to_csv(
    OUT_DIAGNOSTIC_SUMMARY,
    index=False
)


# =============================================================================
# 17. REPORT EXPORTS
# =============================================================================

print("\nAggregation-diagnostic exports")
print("-" * 100)

print(
    f"Aggregation comparison : "
    f"{OUT_AGGREGATION_COMPARISON}"
)

print(
    f"Sensitivity review     : "
    f"{OUT_SENSITIVITY_REVIEW}"
)

print(
    f"Evidence structure     : "
    f"{OUT_SPECIES_STRUCTURE}"
)

print(
    f"Diagnostic summary     : "
    f"{OUT_DIAGNOSTIC_SUMMARY}"
)


# =============================================================================
# 18. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Seed Mass aggregation-method diagnostic completed successfully."
)

print(
    "No final species-level TRY Seed Mass aggregation method "
    "has yet been selected."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3K — FINALIZE TRY SPECIES-LEVEL SEED MASS
#                AND MAP TO OMO CANONICAL TAXA
# =============================================================================
#
# Purpose
# -------
# 1. Finalize species-level Seed Mass values for TRY Request 51911.
# 2. Use the median of source-lineage-reconciled positive measurements.
# 3. Preserve measurement counts, ranges, evidence structure and sensitivity.
# 4. Map final TRY accepted-species values back to the frozen Omo request
#    manifest.
# 5. Identify Omo canonical taxa resolved and unresolved by TRY.
#
# Final estimator
# ---------------
#     MEDIAN_OF_SOURCE_LINEAGE_RECONCILED_MEASUREMENTS
#
# IMPORTANT
# ---------
# - No BIEN evidence is merged here.
# - No non-TRY rescue is performed here.
# - No genus/family fallback is performed here.
# - This section closes the TRY species-level acquisition stage only.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3K — FINALIZE TRY SPECIES-LEVEL SEED MASS "
    "AND MAP TO OMO CANONICAL TAXA"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

MEASUREMENT_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_measurements_ready_for_species_aggregation.csv"
)

AGGREGATION_DIAGNOSTIC_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_species_aggregation_comparison.csv"
)

REQUEST_MANIFEST_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_final_species_request_manifest.csv"
)

VALUE_HOLD_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_celtis_value_hold_resolved.csv"
)


# Outputs
OUT_TRY_SPECIES_VALUES = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_species_values.csv"
)

OUT_TRY_CANONICAL_VALUES = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_omo_canonical_values.csv"
)

OUT_TRY_CANONICAL_GAPS = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_omo_canonical_gaps.csv"
)

OUT_TRY_SPECIES_QUALITY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_species_quality_summary.csv"
)

OUT_TRY_CANONICAL_STATUS = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_canonical_coverage_status.csv"
)

OUT_TRY_FINAL_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

for path in [
    MEASUREMENT_FILE,
    AGGREGATION_DIAGNOSTIC_FILE,
    REQUEST_MANIFEST_FILE,
]:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

measurements = pd.read_csv(
    MEASUREMENT_FILE,
    low_memory=False
)

aggregation = pd.read_csv(
    AGGREGATION_DIAGNOSTIC_FILE,
    low_memory=False
)

request_manifest = pd.read_csv(
    REQUEST_MANIFEST_FILE,
    low_memory=False
)


print("\nInput evidence")
print("-" * 100)

print(
    f"Reconciled TRY measurements : "
    f"{len(measurements)}"
)

print(
    f"TRY aggregation rows        : "
    f"{len(aggregation)}"
)

print(
    f"Frozen Omo request taxa     : "
    f"{len(request_manifest)}"
)


assert len(measurements) == 74, (
    f"Expected 74 reconciled TRY measurements, found {len(measurements)}."
)

assert len(aggregation) == 33, (
    f"Expected 33 TRY accepted-species aggregation rows, "
    f"found {len(aggregation)}."
)

assert len(request_manifest) == 87, (
    f"Expected 87 Omo canonical taxa in frozen TRY request manifest, "
    f"found {len(request_manifest)}."
)


# =============================================================================
# 4. IDENTIFY REQUIRED FIELDS
# =============================================================================

required_aggregation_fields = {
    "AccSpeciesID",
    "AccSpeciesName",
    "N_Retained_Measurements",
    "N_Unique_Values",
    "N_Datasets",
    "N_Source_Lineages",
    "Min_mg",
    "Arithmetic_Mean_mg",
    "Median_mg",
    "Geometric_Mean_mg",
    "Max_mg",
    "Fold_Range",
    "Aggregation_Sensitivity_Flag",
    "Evidence_Structure",
}

missing_aggregation_fields = (
    required_aggregation_fields
    - set(aggregation.columns)
)

assert not missing_aggregation_fields, (
    "Aggregation diagnostic missing required columns:\n"
    f"{sorted(missing_aggregation_fields)}"
)


required_manifest_fields = {
    "Canonical_Species",
    "TRY_AccSpeciesID",
}

missing_manifest_fields = (
    required_manifest_fields
    - set(request_manifest.columns)
)

assert not missing_manifest_fields, (
    "Frozen request manifest missing required columns:\n"
    f"{sorted(missing_manifest_fields)}"
)


# =============================================================================
# 5. NUMERIC NORMALIZATION
# =============================================================================

aggregation["TRY_AccSpeciesID"] = pd.to_numeric(
    aggregation["AccSpeciesID"],
    errors="coerce"
).astype("Int64")


for col in [
    "N_Retained_Measurements",
    "N_Unique_Values",
    "N_Datasets",
    "N_Source_Lineages",
    "Min_mg",
    "Arithmetic_Mean_mg",
    "Median_mg",
    "Geometric_Mean_mg",
    "Max_mg",
    "Fold_Range",
]:

    aggregation[col] = pd.to_numeric(
        aggregation[col],
        errors="coerce"
    )


request_manifest["TRY_AccSpeciesID"] = pd.to_numeric(
    request_manifest["TRY_AccSpeciesID"],
    errors="coerce"
).astype("Int64")


assert aggregation["TRY_AccSpeciesID"].notna().all()

assert request_manifest["TRY_AccSpeciesID"].notna().all()

assert aggregation["Median_mg"].notna().all()

assert aggregation["Median_mg"].gt(0).all()


# =============================================================================
# 6. FORMALLY DEFINE FINAL TRY AGGREGATION RULE
# =============================================================================

FINAL_AGGREGATION_METHOD = (
    "MEDIAN_OF_SOURCE_LINEAGE_RECONCILED_MEASUREMENTS"
)

TRAIT_CODE = "SEED_MASS"

STANDARDIZED_UNIT = "mg"

TRAIT_SOURCE = "SPECIES_TRY"


# =============================================================================
# 7. BUILD FINAL TRY ACCEPTED-SPECIES VALUE TABLE
# =============================================================================

final_try_species = aggregation[
    [
        "TRY_AccSpeciesID",
        "AccSpeciesName",
        "N_Retained_Measurements",
        "N_Unique_Values",
        "N_Datasets",
        "N_Source_Lineages",
        "Min_mg",
        "Arithmetic_Mean_mg",
        "Median_mg",
        "Geometric_Mean_mg",
        "Max_mg",
        "Fold_Range",
        "Aggregation_Sensitivity_Flag",
        "Evidence_Structure",
    ]
].copy()


final_try_species = final_try_species.rename(
    columns={
        "AccSpeciesName":
            "TRY_Accepted_Species",

        "Median_mg":
            "Seed_Mass_mg",
    }
)


final_try_species["Trait_Code"] = TRAIT_CODE

final_try_species[
    "Standardized_Trait_Unit"
] = STANDARDIZED_UNIT

final_try_species[
    "Trait_Source"
] = TRAIT_SOURCE

final_try_species[
    "Aggregation_Method"
] = FINAL_AGGREGATION_METHOD


# =============================================================================
# 8. QUALITY-FLAG CLASSIFICATION
# =============================================================================

def classify_quality(row):

    n = int(
        row["N_Retained_Measurements"]
    )

    sensitivity = (
        row["Aggregation_Sensitivity_Flag"]
    )

    if sensitivity == "REVIEW_AGGREGATION_SENSITIVITY":

        if row["TRY_Accepted_Species"] == "Dracaena mannii":
            return (
                "HIGH_WITHIN_SPECIES_VARIABILITY_"
                "ROBUST_MEDIAN_RETAINED"
            )

        return (
            "AGGREGATION_SENSITIVITY_"
            "ROBUST_MEDIAN_RETAINED"
        )

    if n == 1:
        return "SINGLE_RECONCILED_TRY_MEASUREMENT"

    if n == 2:
        return "TWO_DISTINCT_RECONCILED_TRY_MEASUREMENTS"

    return "MULTIPLE_RECONCILED_TRY_MEASUREMENTS"


final_try_species[
    "Trait_Quality_Flag"
] = final_try_species.apply(
    classify_quality,
    axis=1
)


# =============================================================================
# 9. EVIDENCE-STRENGTH CLASSIFICATION
# =============================================================================

def classify_strength(row):

    n = int(
        row["N_Retained_Measurements"]
    )

    if n == 1:
        return "LIMITED"

    if n == 2:
        return "MODERATE"

    if (
        row["Aggregation_Sensitivity_Flag"]
        == "REVIEW_AGGREGATION_SENSITIVITY"
    ):
        return "MODERATE"

    return "HIGH"


final_try_species[
    "Evidence_Strength"
] = final_try_species.apply(
    classify_strength,
    axis=1
)


# =============================================================================
# 10. REORDER FINAL TRY SPECIES TABLE
# =============================================================================

final_try_species = final_try_species[
    [
        "TRY_AccSpeciesID",
        "TRY_Accepted_Species",
        "Trait_Code",
        "Seed_Mass_mg",
        "Standardized_Trait_Unit",
        "Trait_Source",
        "Aggregation_Method",
        "N_Retained_Measurements",
        "N_Unique_Values",
        "N_Datasets",
        "N_Source_Lineages",
        "Min_mg",
        "Arithmetic_Mean_mg",
        "Geometric_Mean_mg",
        "Max_mg",
        "Fold_Range",
        "Evidence_Structure",
        "Aggregation_Sensitivity_Flag",
        "Trait_Quality_Flag",
        "Evidence_Strength",
    ]
]


final_try_species = (
    final_try_species
    .sort_values(
        "TRY_Accepted_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 11. VALIDATE FINAL TRY ACCEPTED-SPECIES VALUES
# =============================================================================

assert len(final_try_species) == 33

assert final_try_species[
    "TRY_AccSpeciesID"
].is_unique

assert final_try_species[
    "Seed_Mass_mg"
].notna().all()

assert final_try_species[
    "Seed_Mass_mg"
].gt(0).all()


print("\nFinal TRY accepted-species Seed Mass values")
print("-" * 100)

display(
    final_try_species
)


# =============================================================================
# 12. QUALITY SUMMARY
# =============================================================================

quality_summary = (
    final_try_species[
        "Trait_Quality_Flag"
    ]
    .value_counts()
    .rename_axis(
        "Trait_Quality_Flag"
    )
    .reset_index(
        name="N_TRY_Species"
    )
)


print("\nTRY species quality summary")
print("-" * 100)

display(
    quality_summary
)


strength_summary = (
    final_try_species[
        "Evidence_Strength"
    ]
    .value_counts()
    .rename_axis(
        "Evidence_Strength"
    )
    .reset_index(
        name="N_TRY_Species"
    )
)


print("\nTRY evidence-strength summary")
print("-" * 100)

display(
    strength_summary
)


# =============================================================================
# 13. EXPLICITLY CHECK THE TWO AGGREGATION-SENSITIVE TAXA
# =============================================================================

sensitive_species = final_try_species[
    final_try_species[
        "Aggregation_Sensitivity_Flag"
    ].eq(
        "REVIEW_AGGREGATION_SENSITIVITY"
    )
].copy()


print("\nFinal treatment of aggregation-sensitive TRY taxa")
print("-" * 100)

display(
    sensitive_species[
        [
            "TRY_Accepted_Species",
            "N_Retained_Measurements",
            "Min_mg",
            "Seed_Mass_mg",
            "Arithmetic_Mean_mg",
            "Geometric_Mean_mg",
            "Max_mg",
            "Fold_Range",
            "Trait_Quality_Flag",
            "Evidence_Strength",
        ]
    ]
)


assert set(
    sensitive_species[
        "TRY_Accepted_Species"
    ]
) == {
    "Dracaena mannii",
    "Macaranga alnifolia",
}, (
    "Unexpected aggregation-sensitive TRY taxa."
)


# =============================================================================
# 14. PREPARE FROZEN OMO REQUEST MANIFEST FOR MAPPING
# =============================================================================

manifest_mapping = request_manifest.copy()


# Preserve useful taxonomic provenance fields if present.
preferred_manifest_columns = [
    col
    for col in [
        "Canonical_Species",
        "Preferred_TRY_Query_Name",
        "TRY_Name_Match_Type",
        "TRY_AccSpeciesID",
    ]
    if col in manifest_mapping.columns
]


manifest_mapping = (
    manifest_mapping[
        preferred_manifest_columns
    ]
    .copy()
)


assert manifest_mapping[
    "Canonical_Species"
].is_unique, (
    "Frozen TRY request manifest does not contain unique Omo canonical taxa."
)


# =============================================================================
# 15. MAP FINAL TRY VALUES BACK TO THE 87 OMO CANONICAL TAXA
# =============================================================================

canonical_try = manifest_mapping.merge(
    final_try_species,
    on="TRY_AccSpeciesID",
    how="left",
    validate="many_to_one"
)


canonical_try[
    "TRY_Seed_Mass_Status"
] = np.where(
    canonical_try[
        "Seed_Mass_mg"
    ].notna(),
    "TRY_SEED_MASS_RESOLVED",
    "NO_USABLE_TRY_SEED_MASS_RETURNED"
)


resolved_canonical = canonical_try[
    canonical_try[
        "TRY_Seed_Mass_Status"
    ].eq(
        "TRY_SEED_MASS_RESOLVED"
    )
].copy()


unresolved_canonical = canonical_try[
    canonical_try[
        "TRY_Seed_Mass_Status"
    ].eq(
        "NO_USABLE_TRY_SEED_MASS_RETURNED"
    )
].copy()


print("\nOmo canonical mapping of final TRY Seed Mass evidence")
print("-" * 100)

print(
    f"Omo taxa submitted through TRY : "
    f"{len(canonical_try)}"
)

print(
    f"Omo taxa resolved by TRY       : "
    f"{len(resolved_canonical)}"
)

print(
    f"Omo taxa unresolved by TRY     : "
    f"{len(unresolved_canonical)}"
)


# =============================================================================
# 16. CANONICAL COVERAGE STATUS
# =============================================================================

canonical_status_summary = (
    canonical_try[
        "TRY_Seed_Mass_Status"
    ]
    .value_counts()
    .rename_axis(
        "TRY_Seed_Mass_Status"
    )
    .reset_index(
        name="N_Omo_Canonical_Taxa"
    )
)


canonical_status_summary[
    "Percent_of_87_TRY_Targets"
] = (
    canonical_status_summary[
        "N_Omo_Canonical_Taxa"
    ]
    / 87
    * 100
)


print("\nOmo canonical TRY coverage summary")
print("-" * 100)

display(
    canonical_status_summary
)


# =============================================================================
# 17. DISPLAY RESOLVED OMO CANONICAL TAXA
# =============================================================================

print("\nOmo canonical taxa receiving final TRY Seed Mass values")
print("-" * 100)

display(
    resolved_canonical[
        [
            col
            for col in [
                "Canonical_Species",
                "Preferred_TRY_Query_Name",
                "TRY_Name_Match_Type",
                "TRY_AccSpeciesID",
                "TRY_Accepted_Species",
                "Seed_Mass_mg",
                "Trait_Source",
                "Trait_Quality_Flag",
                "Evidence_Strength",
            ]
            if col in resolved_canonical.columns
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 18. DISPLAY UNRESOLVED TRY TARGETS
# =============================================================================

print("\nOmo canonical taxa submitted to TRY but still unresolved")
print("-" * 100)

print(
    f"N taxa : "
    f"{len(unresolved_canonical)}"
)

if len(unresolved_canonical) > 0:

    display(
        unresolved_canonical[
            [
                col
                for col in [
                    "Canonical_Species",
                    "Preferred_TRY_Query_Name",
                    "TRY_Name_Match_Type",
                    "TRY_AccSpeciesID",
                    "TRY_Seed_Mass_Status",
                ]
                if col in unresolved_canonical.columns
            ]
        ]
        .sort_values(
            "Canonical_Species"
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 19. FINAL NUMERIC SUMMARY OF 33 TRY SPECIES VALUES
# =============================================================================

numeric_summary = (
    final_try_species[
        "Seed_Mass_mg"
    ]
    .describe()
)


print("\nFinal TRY species-level Seed Mass numeric summary")
print("-" * 100)

print(
    numeric_summary.to_string()
)


# =============================================================================
# 20. FINAL TRY SUMMARY TABLE
# =============================================================================

final_summary = pd.DataFrame({
    "Metric": [
        "TRY Request ID",
        "TRY Trait ID",
        "TRY Trait",
        "Original positive numeric TRY measurements",
        "Source-lineage-reconciled measurements",
        "TRY accepted species with final values",
        "Omo canonical TRY targets",
        "Omo canonical taxa resolved by TRY",
        "Omo canonical taxa unresolved by TRY",
        "Aggregation-sensitive TRY species",
        "Final aggregation method",
        "Standardized unit",
    ],

    "Value": [
        51911,
        26,
        "Seed dry mass",
        177,
        len(measurements),
        len(final_try_species),
        len(canonical_try),
        len(resolved_canonical),
        len(unresolved_canonical),
        len(sensitive_species),
        FINAL_AGGREGATION_METHOD,
        STANDARDIZED_UNIT,
    ],
})


print("\nFinal TRY Seed Mass acquisition summary")
print("-" * 100)

display(
    final_summary
)


# =============================================================================
# 21. STRONG VALIDATION
# =============================================================================

assert len(final_try_species) == 33

assert len(canonical_try) == 87

assert (
    len(resolved_canonical)
    + len(unresolved_canonical)
    == 87
)

# Previous canonical-expansion audit established that returned TRY evidence
# maps to 46 Omo canonical taxa.
assert len(resolved_canonical) == 46, (
    "Expected 46 Omo canonical taxa to receive TRY Seed Mass values "
    f"based on the validated request-return crosswalk, "
    f"found {len(resolved_canonical)}."
)

assert len(unresolved_canonical) == 41, (
    "Expected 41 submitted Omo canonical taxa to remain unresolved "
    f"after TRY, found {len(unresolved_canonical)}."
)

assert resolved_canonical[
    "Seed_Mass_mg"
].notna().all()

assert resolved_canonical[
    "Seed_Mass_mg"
].gt(0).all()

assert (
    resolved_canonical[
        "Standardized_Trait_Unit"
    ]
    .eq("mg")
    .all()
)

assert (
    resolved_canonical[
        "Trait_Source"
    ]
    .eq("SPECIES_TRY")
    .all()
)


print("\nFinal TRY validation")
print("-" * 100)

print(
    "TRY accepted-species final values confirmed : 33"
)

print(
    "Omo canonical TRY targets confirmed         : 87"
)

print(
    "Omo canonical taxa resolved by TRY          : 46"
)

print(
    "Omo canonical taxa unresolved after TRY     : 41"
)

print(
    "Standardized unit confirmed                 : mg"
)

print(
    "Final aggregation method confirmed          : "
    f"{FINAL_AGGREGATION_METHOD}"
)


# =============================================================================
# 22. EXPORT
# =============================================================================

final_try_species.to_csv(
    OUT_TRY_SPECIES_VALUES,
    index=False
)

canonical_try.to_csv(
    OUT_TRY_CANONICAL_VALUES,
    index=False
)

unresolved_canonical.to_csv(
    OUT_TRY_CANONICAL_GAPS,
    index=False
)

quality_summary.to_csv(
    OUT_TRY_SPECIES_QUALITY,
    index=False
)

canonical_status_summary.to_csv(
    OUT_TRY_CANONICAL_STATUS,
    index=False
)

final_summary.to_csv(
    OUT_TRY_FINAL_SUMMARY,
    index=False
)


# =============================================================================
# 23. REPORT EXPORTS
# =============================================================================

print("\nFinal TRY exports")
print("-" * 100)

print(
    f"TRY species values       : "
    f"{OUT_TRY_SPECIES_VALUES}"
)

print(
    f"Omo canonical TRY values : "
    f"{OUT_TRY_CANONICAL_VALUES}"
)

print(
    f"TRY canonical gaps       : "
    f"{OUT_TRY_CANONICAL_GAPS}"
)

print(
    f"TRY quality summary      : "
    f"{OUT_TRY_SPECIES_QUALITY}"
)

print(
    f"TRY canonical status     : "
    f"{OUT_TRY_CANONICAL_STATUS}"
)

print(
    f"TRY final summary        : "
    f"{OUT_TRY_FINAL_SUMMARY}"
)


# =============================================================================
# 24. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "TRY Request 51911 species-level Seed Mass acquisition "
    "completed successfully."
)

print(
    f"Final TRY accepted-species values : "
    f"{len(final_try_species)}"
)

print(
    f"Omo canonical taxa resolved       : "
    f"{len(resolved_canonical)} / 87"
)

print(
    f"Omo canonical taxa unresolved     : "
    f"{len(unresolved_canonical)} / 87"
)

print(
    "Final estimator: median of source-lineage-reconciled "
    "positive Seed Mass measurements."
)

print(
    "No BIEN/TRY multi-source consolidation has yet been performed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3L — BIEN × TRY CONSOLIDATION AND TRUE RESIDUAL-GAP DETERMINATION
# =============================================================================
#
# Purpose
# -------
# Reconcile the two independently completed species-level Seed Mass evidence
# streams:
#
#     BIEN
#     TRY Request 51911
#
# This section:
#
#   1. validates BIEN and TRY canonical outputs;
#   2. reconstructs the full 127-taxon canonical backbone;
#   3. classifies taxa into:
#
#          BIEN_ONLY
#          TRY_ONLY
#          BIEN_AND_TRY
#          UNRESOLVED_BY_BIEN_AND_TRY
#
#   4. compares BIEN and TRY values where both exist;
#   5. quantifies agreement without choosing a preferred source yet;
#   6. restores previously validated acquisition routing using:
#
#          species-level non-TRY rescue targets
#          genus-stage targets
#
#   7. identifies the true remaining Seed Mass acquisition gaps.
#
# IMPORTANT
# ---------
# No BIEN-vs-TRY source priority is imposed here.
# No overlapping values are averaged.
# No fallback estimate is created here.
# No genus/family estimate is created here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3L — BIEN × TRY CONSOLIDATION "
    "AND TRUE RESIDUAL-GAP DETERMINATION"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

# -------------------------------------------------------------------------
# Final BIEN species-level evidence
# -------------------------------------------------------------------------

BIEN_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_final_species_values.csv"
)

# -------------------------------------------------------------------------
# Final TRY canonical evidence
# -------------------------------------------------------------------------

TRY_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_omo_canonical_values.csv"
)

# -------------------------------------------------------------------------
# Full 127-taxon canonical Seed Mass backbone
# -------------------------------------------------------------------------

BACKBONE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_acquisition_manifest.csv"
)

# -------------------------------------------------------------------------
# Previously validated species-level non-TRY rescue targets
# -------------------------------------------------------------------------

NON_TRY_RESCUE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_species_non_try_rescue_targets.csv"
)

# -------------------------------------------------------------------------
# Previously validated genus-stage targets
# -------------------------------------------------------------------------

GENUS_STAGE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_genus_stage_targets.csv"
)


# =============================================================================
# 2. OUTPUT PATHS
# =============================================================================

OUT_CROSS_SOURCE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_cross_source_consolidation.csv"
)

OUT_OVERLAP = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_overlap_diagnostic.csv"
)

OUT_COVERAGE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_127_taxon_coverage.csv"
)

OUT_TRUE_GAPS = (
    OUTPUT_DIR /
    "omo_seed_mass_true_residual_gaps_after_bien_try.csv"
)

OUT_CATEGORY_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_coverage_category_summary.csv"
)

OUT_OVERLAP_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_overlap_summary.csv"
)

OUT_ROUTING_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_post_bien_try_routing_summary.csv"
)

OUT_GLOBAL_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_global_summary.csv"
)


# =============================================================================
# 3. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    BIEN_FILE,
    TRY_FILE,
    BACKBONE_FILE,
    NON_TRY_RESCUE_FILE,
    GENUS_STAGE_FILE,
]

for path in required_files:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


# =============================================================================
# 4. LOAD INPUT TABLES
# =============================================================================

bien = pd.read_csv(
    BIEN_FILE,
    low_memory=False
)

try_values = pd.read_csv(
    TRY_FILE,
    low_memory=False
)

backbone = pd.read_csv(
    BACKBONE_FILE,
    low_memory=False
)

non_try_rescue = pd.read_csv(
    NON_TRY_RESCUE_FILE,
    low_memory=False
)

genus_stage = pd.read_csv(
    GENUS_STAGE_FILE,
    low_memory=False
)


print("\nInput tables")
print("-" * 100)

print(
    f"BIEN final species values       : "
    f"{len(bien)} rows"
)

print(
    f"TRY canonical values            : "
    f"{len(try_values)} rows"
)

print(
    f"Canonical Seed Mass backbone    : "
    f"{len(backbone)} rows"
)

print(
    f"Species-level non-TRY rescue    : "
    f"{len(non_try_rescue)} rows"
)

print(
    f"Genus-stage targets             : "
    f"{len(genus_stage)} rows"
)


# =============================================================================
# 5. HELPER — FIND COLUMN
# =============================================================================

def find_column(df, candidates, required=True):

    lookup = {
        str(col).strip().lower(): col
        for col in df.columns
    }

    for candidate in candidates:

        key = candidate.strip().lower()

        if key in lookup:
            return lookup[key]

    if required:

        raise KeyError(
            "Required field not found.\n"
            f"Candidates tested: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


# =============================================================================
# 6. IDENTIFY CANONICAL NAME FIELDS
# =============================================================================

BIEN_CANON_COL = find_column(
    bien,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)

TRY_CANON_COL = find_column(
    try_values,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)

BACKBONE_CANON_COL = find_column(
    backbone,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)

NON_TRY_CANON_COL = find_column(
    non_try_rescue,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)

GENUS_CANON_COL = find_column(
    genus_stage,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)


# =============================================================================
# 7. VALIDATE 127-TAXON BACKBONE
# =============================================================================

assert len(backbone) == 127, (
    f"Expected 127 rows in Seed Mass backbone, "
    f"found {len(backbone)}."
)

assert (
    backbone[
        BACKBONE_CANON_COL
    ]
    .nunique()
    == 127
), (
    "Seed Mass backbone does not contain "
    "127 unique canonical taxa."
)


canonical_backbone = (
    backbone[
        [
            BACKBONE_CANON_COL
        ]
    ]
    .rename(
        columns={
            BACKBONE_CANON_COL:
                "Canonical_Species"
        }
    )
    .copy()
)


assert canonical_backbone[
    "Canonical_Species"
].notna().all()

assert canonical_backbone[
    "Canonical_Species"
].is_unique


print("\nCanonical backbone")
print("-" * 100)

print(
    f"Rows              : "
    f"{len(canonical_backbone)}"
)

print(
    f"Unique taxa       : "
    f"{canonical_backbone['Canonical_Species'].nunique()}"
)


# =============================================================================
# 8. PREPARE BIEN CANONICAL VALUES
# =============================================================================

BIEN_VALUE_COL = find_column(
    bien,
    [
        "Seed_Mass_mg",
        "Seed Mass mg",
        "Standardized_Trait_Value",
    ]
)

BIEN_SOURCE_COL = find_column(
    bien,
    [
        "Trait_Source",
    ],
    required=False
)

BIEN_QUALITY_COL = find_column(
    bien,
    [
        "Trait_Quality_Flag",
    ],
    required=False
)

BIEN_STRENGTH_COL = find_column(
    bien,
    [
        "Evidence_Strength",
    ],
    required=False
)


bien_prepared = pd.DataFrame({
    "Canonical_Species":
        bien[
            BIEN_CANON_COL
        ],

    "BIEN_Seed_Mass_mg":
        pd.to_numeric(
            bien[
                BIEN_VALUE_COL
            ],
            errors="coerce"
        ),
})


if BIEN_SOURCE_COL is not None:

    bien_prepared[
        "BIEN_Trait_Source"
    ] = bien[
        BIEN_SOURCE_COL
    ]

else:

    bien_prepared[
        "BIEN_Trait_Source"
    ] = "SPECIES_BIEN"


if BIEN_QUALITY_COL is not None:

    bien_prepared[
        "BIEN_Trait_Quality_Flag"
    ] = bien[
        BIEN_QUALITY_COL
    ]

else:

    bien_prepared[
        "BIEN_Trait_Quality_Flag"
    ] = pd.NA


if BIEN_STRENGTH_COL is not None:

    bien_prepared[
        "BIEN_Evidence_Strength"
    ] = bien[
        BIEN_STRENGTH_COL
    ]

else:

    bien_prepared[
        "BIEN_Evidence_Strength"
    ] = pd.NA


# -----------------------------------------------------------------------------
# BIEN validation
# -----------------------------------------------------------------------------

assert bien_prepared[
    "Canonical_Species"
].notna().all()

assert bien_prepared[
    "Canonical_Species"
].is_unique, (
    "BIEN final species table contains duplicate canonical taxa."
)

assert bien_prepared[
    "BIEN_Seed_Mass_mg"
].notna().all()

assert bien_prepared[
    "BIEN_Seed_Mass_mg"
].gt(0).all()


print("\nValidated BIEN canonical evidence")
print("-" * 100)

print(
    f"Canonical taxa with BIEN values : "
    f"{len(bien_prepared)}"
)


# =============================================================================
# 9. PREPARE TRY CANONICAL VALUES
# =============================================================================

try_resolved = try_values[
    try_values[
        "Seed_Mass_mg"
    ]
    .notna()
].copy()


try_prepared = pd.DataFrame({
    "Canonical_Species":
        try_resolved[
            TRY_CANON_COL
        ],

    "TRY_Seed_Mass_mg":
        pd.to_numeric(
            try_resolved[
                "Seed_Mass_mg"
            ],
            errors="coerce"
        ),
})


if "TRY_Accepted_Species" in try_resolved.columns:

    try_prepared[
        "TRY_Accepted_Species"
    ] = try_resolved[
        "TRY_Accepted_Species"
    ]

else:

    try_prepared[
        "TRY_Accepted_Species"
    ] = pd.NA


if "Trait_Source" in try_resolved.columns:

    try_prepared[
        "TRY_Trait_Source"
    ] = try_resolved[
        "Trait_Source"
    ]

else:

    try_prepared[
        "TRY_Trait_Source"
    ] = "SPECIES_TRY"


if "Trait_Quality_Flag" in try_resolved.columns:

    try_prepared[
        "TRY_Trait_Quality_Flag"
    ] = try_resolved[
        "Trait_Quality_Flag"
    ]

else:

    try_prepared[
        "TRY_Trait_Quality_Flag"
    ] = pd.NA


if "Evidence_Strength" in try_resolved.columns:

    try_prepared[
        "TRY_Evidence_Strength"
    ] = try_resolved[
        "Evidence_Strength"
    ]

else:

    try_prepared[
        "TRY_Evidence_Strength"
    ] = pd.NA


# -----------------------------------------------------------------------------
# TRY validation
# -----------------------------------------------------------------------------

assert len(try_prepared) == 46, (
    f"Expected 46 canonical TRY-resolved taxa, "
    f"found {len(try_prepared)}."
)

assert try_prepared[
    "Canonical_Species"
].notna().all()

assert try_prepared[
    "Canonical_Species"
].is_unique

assert try_prepared[
    "TRY_Seed_Mass_mg"
].notna().all()

assert try_prepared[
    "TRY_Seed_Mass_mg"
].gt(0).all()


print("\nValidated TRY canonical evidence")
print("-" * 100)

print(
    f"Canonical taxa with TRY values : "
    f"{len(try_prepared)}"
)


# =============================================================================
# 10. CONFIRM BIEN / TRY TAXA BELONG TO 127-TAXON BACKBONE
# =============================================================================

backbone_set = set(
    canonical_backbone[
        "Canonical_Species"
    ]
)

bien_set = set(
    bien_prepared[
        "Canonical_Species"
    ]
)

try_set = set(
    try_prepared[
        "Canonical_Species"
    ]
)


bien_outside_backbone = (
    bien_set
    - backbone_set
)

try_outside_backbone = (
    try_set
    - backbone_set
)


assert not bien_outside_backbone, (
    "BIEN contains canonical taxa outside the 127-taxon backbone:\n"
    f"{sorted(bien_outside_backbone)}"
)

assert not try_outside_backbone, (
    "TRY contains canonical taxa outside the 127-taxon backbone:\n"
    f"{sorted(try_outside_backbone)}"
)


# =============================================================================
# 11. MERGE BIEN AND TRY ONTO FULL 127-TAXON BACKBONE
# =============================================================================

coverage = (
    canonical_backbone
    .merge(
        bien_prepared,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
    .merge(
        try_prepared,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


assert len(coverage) == 127

assert coverage[
    "Canonical_Species"
].is_unique


# =============================================================================
# 12. CLASSIFY BIEN / TRY COVERAGE
# =============================================================================

has_bien = (
    coverage[
        "BIEN_Seed_Mass_mg"
    ]
    .notna()
)

has_try = (
    coverage[
        "TRY_Seed_Mass_mg"
    ]
    .notna()
)


coverage[
    "BIEN_TRY_Coverage_Class"
] = "UNRESOLVED_BY_BIEN_AND_TRY"


coverage.loc[
    has_bien & ~has_try,
    "BIEN_TRY_Coverage_Class"
] = "BIEN_ONLY"


coverage.loc[
    ~has_bien & has_try,
    "BIEN_TRY_Coverage_Class"
] = "TRY_ONLY"


coverage.loc[
    has_bien & has_try,
    "BIEN_TRY_Coverage_Class"
] = "BIEN_AND_TRY"


# =============================================================================
# 13. COVERAGE CATEGORY SUMMARY
# =============================================================================

category_summary = (
    coverage[
        "BIEN_TRY_Coverage_Class"
    ]
    .value_counts()
    .rename_axis(
        "BIEN_TRY_Coverage_Class"
    )
    .reset_index(
        name="N_Taxa"
    )
)


category_summary[
    "Percent_of_127"
] = (
    category_summary[
        "N_Taxa"
    ]
    / 127
    * 100
)


print("\nBIEN × TRY canonical coverage")
print("-" * 100)

display(
    category_summary
)


# =============================================================================
# 14. EXTRACT OVERLAP TAXA
# =============================================================================

overlap = coverage[
    coverage[
        "BIEN_TRY_Coverage_Class"
    ]
    .eq(
        "BIEN_AND_TRY"
    )
].copy()


print("\nTaxa with both BIEN and TRY species-level evidence")
print("-" * 100)

print(
    f"N taxa : "
    f"{len(overlap)}"
)


# =============================================================================
# 15. COMPUTE BIEN-vs-TRY AGREEMENT DIAGNOSTICS
# =============================================================================
#
# Seed Mass spans orders of magnitude.
#
# Therefore:
#
#     raw arithmetic difference
#
# is less informative than:
#
#     ratio
#     fold difference
#     log10 difference
#
# =============================================================================

if len(overlap) > 0:

    overlap[
        "BIEN_TRY_Ratio"
    ] = (
        overlap[
            "TRY_Seed_Mass_mg"
        ]
        /
        overlap[
            "BIEN_Seed_Mass_mg"
        ]
    )


    overlap[
        "BIEN_TRY_Fold_Difference"
    ] = np.maximum(
        overlap[
            "BIEN_TRY_Ratio"
        ],
        1
        /
        overlap[
            "BIEN_TRY_Ratio"
        ]
    )


    overlap[
        "Log10_TRY_minus_BIEN"
    ] = (
        np.log10(
            overlap[
                "TRY_Seed_Mass_mg"
            ]
        )
        -
        np.log10(
            overlap[
                "BIEN_Seed_Mass_mg"
            ]
        )
    )


    overlap[
        "Absolute_Log10_Difference"
    ] = (
        overlap[
            "Log10_TRY_minus_BIEN"
        ]
        .abs()
    )


    overlap[
        "Agreement_Class"
    ] = "LARGE_DIFFERENCE_GT_5_FOLD"


    overlap.loc[
        overlap[
            "BIEN_TRY_Fold_Difference"
        ]
        .le(5),
        "Agreement_Class"
    ] = "SUBSTANTIAL_DIFFERENCE_LE_5_FOLD"


    overlap.loc[
        overlap[
            "BIEN_TRY_Fold_Difference"
        ]
        .le(2),
        "Agreement_Class"
    ] = "MODERATE_AGREEMENT_LE_2_FOLD"


    overlap.loc[
        overlap[
            "BIEN_TRY_Fold_Difference"
        ]
        .le(1.5),
        "Agreement_Class"
    ] = "CLOSE_AGREEMENT_LE_1_5_FOLD"


else:

    overlap[
        "BIEN_TRY_Ratio"
    ] = pd.Series(
        dtype=float
    )

    overlap[
        "BIEN_TRY_Fold_Difference"
    ] = pd.Series(
        dtype=float
    )

    overlap[
        "Log10_TRY_minus_BIEN"
    ] = pd.Series(
        dtype=float
    )

    overlap[
        "Absolute_Log10_Difference"
    ] = pd.Series(
        dtype=float
    )

    overlap[
        "Agreement_Class"
    ] = pd.Series(
        dtype="string"
    )


# =============================================================================
# 16. DISPLAY OVERLAP DIAGNOSTIC
# =============================================================================

print("\nBIEN × TRY overlap diagnostic")
print("-" * 100)

if len(overlap) > 0:

    display(
        overlap[
            [
                col
                for col in [
                    "Canonical_Species",
                    "BIEN_Seed_Mass_mg",
                    "TRY_Seed_Mass_mg",
                    "BIEN_TRY_Ratio",
                    "BIEN_TRY_Fold_Difference",
                    "Log10_TRY_minus_BIEN",
                    "Agreement_Class",
                    "BIEN_Trait_Quality_Flag",
                    "BIEN_Evidence_Strength",
                    "TRY_Trait_Quality_Flag",
                    "TRY_Evidence_Strength",
                ]
                if col in overlap.columns
            ]
        ]
        .sort_values(
            "BIEN_TRY_Fold_Difference",
            ascending=False
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 17. OVERLAP AGREEMENT SUMMARY
# =============================================================================

if len(overlap) > 0:

    overlap_summary = (
        overlap[
            "Agreement_Class"
        ]
        .value_counts()
        .rename_axis(
            "Agreement_Class"
        )
        .reset_index(
            name="N_Taxa"
        )
    )

else:

    overlap_summary = pd.DataFrame(
        columns=[
            "Agreement_Class",
            "N_Taxa",
        ]
    )


print("\nBIEN × TRY agreement summary")
print("-" * 100)

display(
    overlap_summary
)


# =============================================================================
# 18. IDENTIFY CURRENT SPECIES-LEVEL COVERAGE
# =============================================================================

coverage[
    "Any_Species_Level_Seed_Mass"
] = (
    coverage[
        "BIEN_Seed_Mass_mg"
    ]
    .notna()
    |
    coverage[
        "TRY_Seed_Mass_mg"
    ]
    .notna()
)


species_resolved = coverage[
    coverage[
        "Any_Species_Level_Seed_Mass"
    ]
].copy()


species_unresolved = coverage[
    ~coverage[
        "Any_Species_Level_Seed_Mass"
    ]
].copy()


print("\nCurrent species-level Seed Mass coverage")
print("-" * 100)

print(
    f"Taxa resolved by BIEN and/or TRY : "
    f"{len(species_resolved)}"
)

print(
    f"Taxa unresolved by both          : "
    f"{len(species_unresolved)}"
)


# =============================================================================
# 19. RECONSTRUCT POST-BIEN/TRY ACQUISITION ROUTING
# =============================================================================

non_try_species_set = set(
    non_try_rescue[
        NON_TRY_CANON_COL
    ]
    .dropna()
    .astype(str)
)


genus_stage_set = set(
    genus_stage[
        GENUS_CANON_COL
    ]
    .dropna()
    .astype(str)
)


assert len(non_try_species_set) == 9, (
    f"Expected 9 predefined non-TRY rescue taxa, "
    f"found {len(non_try_species_set)}."
)

assert len(genus_stage_set) == 3, (
    f"Expected 3 predefined genus-stage taxa, "
    f"found {len(genus_stage_set)}."
)


# -----------------------------------------------------------------------------
# Confirm all routed taxa belong to canonical backbone
# -----------------------------------------------------------------------------

non_try_outside_backbone = (
    non_try_species_set
    - backbone_set
)

genus_outside_backbone = (
    genus_stage_set
    - backbone_set
)


assert not non_try_outside_backbone, (
    "Non-TRY rescue targets outside canonical backbone:\n"
    f"{sorted(non_try_outside_backbone)}"
)

assert not genus_outside_backbone, (
    "Genus-stage targets outside canonical backbone:\n"
    f"{sorted(genus_outside_backbone)}"
)


coverage[
    "Previously_Routed_Non_TRY_Species_Rescue"
] = (
    coverage[
        "Canonical_Species"
    ]
    .isin(
        non_try_species_set
    )
)


coverage[
    "Previously_Routed_Genus_Stage"
] = (
    coverage[
        "Canonical_Species"
    ]
    .isin(
        genus_stage_set
    )
)


# =============================================================================
# 20. ASSIGN POST-BIEN/TRY NEXT ACTION
# =============================================================================

coverage[
    "Post_BIEN_TRY_Next_Action"
] = (
    "NO_FURTHER_SPECIES_ACQUISITION_REQUIRED"
)


# -----------------------------------------------------------------------------
# Genus-stage targets remain genus-level only if still unresolved
# -----------------------------------------------------------------------------

coverage.loc[
    (
        ~coverage[
            "Any_Species_Level_Seed_Mass"
        ]
        &
        coverage[
            "Previously_Routed_Genus_Stage"
        ]
    ),
    "Post_BIEN_TRY_Next_Action"
] = "PROCEED_TO_GENUS_LEVEL_STAGE"


# -----------------------------------------------------------------------------
# All remaining unresolved species-level taxa proceed to targeted rescue
# -----------------------------------------------------------------------------

coverage.loc[
    (
        ~coverage[
            "Any_Species_Level_Seed_Mass"
        ]
        &
        ~coverage[
            "Previously_Routed_Genus_Stage"
        ]
    ),
    "Post_BIEN_TRY_Next_Action"
] = "TARGETED_SPECIES_NON_TRY_RESCUE"


# =============================================================================
# 21. TRUE RESIDUAL GAPS
# =============================================================================

true_gaps = coverage[
    ~coverage[
        "Any_Species_Level_Seed_Mass"
    ]
].copy()


print("\nTrue Seed Mass gaps after BIEN + TRY")
print("-" * 100)

print(
    f"N unresolved canonical taxa : "
    f"{len(true_gaps)}"
)


if len(true_gaps) > 0:

    display(
        true_gaps[
            [
                "Canonical_Species",
                "BIEN_TRY_Coverage_Class",
                "Previously_Routed_Non_TRY_Species_Rescue",
                "Previously_Routed_Genus_Stage",
                "Post_BIEN_TRY_Next_Action",
            ]
        ]
        .sort_values(
            [
                "Post_BIEN_TRY_Next_Action",
                "Canonical_Species",
            ]
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 22. POST-BIEN/TRY ROUTING SUMMARY
# =============================================================================

routing_summary = (
    coverage[
        "Post_BIEN_TRY_Next_Action"
    ]
    .value_counts()
    .rename_axis(
        "Post_BIEN_TRY_Next_Action"
    )
    .reset_index(
        name="N_Taxa"
    )
)


routing_summary[
    "Percent_of_127"
] = (
    routing_summary[
        "N_Taxa"
    ]
    / 127
    * 100
)


print("\nPost-BIEN/TRY Seed Mass routing summary")
print("-" * 100)

display(
    routing_summary
)


# =============================================================================
# 23. GLOBAL CONSOLIDATION COUNTS
# =============================================================================

n_bien = int(
    coverage[
        "BIEN_Seed_Mass_mg"
    ]
    .notna()
    .sum()
)


n_try = int(
    coverage[
        "TRY_Seed_Mass_mg"
    ]
    .notna()
    .sum()
)


n_overlap = len(
    overlap
)


n_species_resolved = int(
    coverage[
        "Any_Species_Level_Seed_Mass"
    ]
    .sum()
)


n_true_gaps = int(
    (
        ~coverage[
            "Any_Species_Level_Seed_Mass"
        ]
    )
    .sum()
)


n_non_try_rescue = int(
    coverage[
        "Post_BIEN_TRY_Next_Action"
    ]
    .eq(
        "TARGETED_SPECIES_NON_TRY_RESCUE"
    )
    .sum()
)


n_genus_stage = int(
    coverage[
        "Post_BIEN_TRY_Next_Action"
    ]
    .eq(
        "PROCEED_TO_GENUS_LEVEL_STAGE"
    )
    .sum()
)


# =============================================================================
# 24. GLOBAL CONSOLIDATION SUMMARY
# =============================================================================

global_summary = pd.DataFrame({
    "Metric": [
        "Canonical Omo taxa",
        "Taxa with BIEN species evidence",
        "Taxa with TRY species evidence",
        "Taxa with both BIEN and TRY evidence",
        "Taxa resolved by BIEN and/or TRY",
        "True residual taxa after BIEN and TRY",
        "Taxa routed to targeted species non-TRY rescue",
        "Taxa routed to genus-level stage",
    ],

    "Value": [
        127,
        n_bien,
        n_try,
        n_overlap,
        n_species_resolved,
        n_true_gaps,
        n_non_try_rescue,
        n_genus_stage,
    ],
})


print("\nBIEN × TRY consolidation summary")
print("-" * 100)

display(
    global_summary
)


# =============================================================================
# 25. STRONG VALIDATION
# =============================================================================

assert len(coverage) == 127

assert coverage[
    "Canonical_Species"
].is_unique

assert (
    category_summary[
        "N_Taxa"
    ]
    .sum()
    == 127
)

assert (
    n_species_resolved
    + n_true_gaps
    == 127
)

assert len(
    non_try_species_set
) == 9

assert len(
    genus_stage_set
) == 3

assert coverage[
    "Post_BIEN_TRY_Next_Action"
].notna().all()

assert (
    n_non_try_rescue
    + n_genus_stage
    == n_true_gaps
), (
    "Residual routing does not account for all unresolved taxa."
)


# =============================================================================
# 26. CROSS-SOURCE VALIDATION REPORT
# =============================================================================

print("\nCross-source validation")
print("-" * 100)

print(
    "Full canonical backbone confirmed           : "
    "127 taxa"
)

print(
    f"BIEN canonical species evidence              : "
    f"{n_bien}"
)

print(
    f"TRY canonical species evidence               : "
    f"{n_try}"
)

print(
    f"BIEN × TRY overlap                           : "
    f"{n_overlap}"
)

print(
    f"Species-level taxa currently resolved        : "
    f"{n_species_resolved}"
)

print(
    f"True residual taxa after BIEN + TRY          : "
    f"{n_true_gaps}"
)

print(
    f"Targeted species non-TRY rescue              : "
    f"{n_non_try_rescue}"
)

print(
    f"Genus-level stage                            : "
    f"{n_genus_stage}"
)

print(
    "No BIEN/TRY source-priority rule has yet been imposed."
)


# =============================================================================
# 27. EXPORT
# =============================================================================

coverage.to_csv(
    OUT_CROSS_SOURCE,
    index=False
)

overlap.to_csv(
    OUT_OVERLAP,
    index=False
)

coverage.to_csv(
    OUT_COVERAGE,
    index=False
)

true_gaps.to_csv(
    OUT_TRUE_GAPS,
    index=False
)

category_summary.to_csv(
    OUT_CATEGORY_SUMMARY,
    index=False
)

overlap_summary.to_csv(
    OUT_OVERLAP_SUMMARY,
    index=False
)

routing_summary.to_csv(
    OUT_ROUTING_SUMMARY,
    index=False
)

global_summary.to_csv(
    OUT_GLOBAL_SUMMARY,
    index=False
)


# =============================================================================
# 28. REPORT EXPORTS
# =============================================================================

print("\nSection 9.3L exports")
print("-" * 100)

print(
    f"Cross-source consolidation : "
    f"{OUT_CROSS_SOURCE}"
)

print(
    f"BIEN × TRY overlap         : "
    f"{OUT_OVERLAP}"
)

print(
    f"127-taxon coverage         : "
    f"{OUT_COVERAGE}"
)

print(
    f"True residual gaps         : "
    f"{OUT_TRUE_GAPS}"
)

print(
    f"Coverage summary           : "
    f"{OUT_CATEGORY_SUMMARY}"
)

print(
    f"Overlap summary            : "
    f"{OUT_OVERLAP_SUMMARY}"
)

print(
    f"Post-BIEN/TRY routing      : "
    f"{OUT_ROUTING_SUMMARY}"
)

print(
    f"Global summary             : "
    f"{OUT_GLOBAL_SUMMARY}"
)


# =============================================================================
# 29. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "BIEN × TRY Seed Mass consolidation diagnostic "
    "completed successfully."
)

print(
    f"Species-level taxa resolved by BIEN and/or TRY : "
    f"{n_species_resolved} / 127"
)

print(
    f"True residual taxa                           : "
    f"{n_true_gaps} / 127"
)

print(
    f"Targeted species non-TRY rescue              : "
    f"{n_non_try_rescue}"
)

print(
    f"Genus-level stage                            : "
    f"{n_genus_stage}"
)

print(
    "Overlapping BIEN and TRY evidence remains preserved separately."
)

print(
    "No final cross-source source-priority decision has yet been made."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3L-R1 — AUDIT POST-ALIAS-RESCUE BIEN SEED MASS OUTPUTS
# =============================================================================
#
# Purpose
# -------
# Section 9.3L accidentally used the original 15-species BIEN table:
#
#     omo_seed_mass_bien_final_species_values.csv
#
# However, the validated Seed Mass workflow subsequently added BIEN evidence
# through taxonomic-alias rescue and Spondias definition review.
#
# Earlier validated project state established:
#
#     BIEN species-level canonical coverage = 28 taxa
#
# This diagnostic searches existing Seed Mass BIEN outputs to determine:
#
#   1. which files contain species-level/canonical Seed Mass values;
#   2. how many taxa each contains;
#   3. whether a final 28-taxon consolidated table already exists;
#   4. which component tables would be needed if reconstruction is required.
#
# IMPORTANT
# ---------
# No evidence is modified or consolidated here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3L-R1 — AUDIT POST-ALIAS-RESCUE "
    "BIEN SEED MASS OUTPUTS"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# =============================================================================
# 2. DISCOVER BIEN SEED MASS CSV OUTPUTS
# =============================================================================

candidate_files = sorted(
    OUTPUT_DIR.glob(
        "omo_seed_mass_bien*.csv"
    )
)

# Include Spondias resolution because it arose from BIEN alias rescue
# but has a different filename prefix.
spondias_files = sorted(
    OUTPUT_DIR.glob(
        "omo_seed_mass_spondias*.csv"
    )
)

candidate_files = (
    candidate_files
    + spondias_files
)


print("\nCandidate BIEN / BIEN-derived Seed Mass files")
print("-" * 100)

print(
    f"N files : {len(candidate_files)}"
)

for path in candidate_files:
    print(f" - {path.name}")


# =============================================================================
# 3. HELPER FUNCTIONS
# =============================================================================

def find_optional_column(df, candidates):

    lookup = {
        str(col).strip().lower(): col
        for col in df.columns
    }

    for candidate in candidates:

        key = candidate.lower()

        if key in lookup:
            return lookup[key]

    return None


canonical_candidates = [
    "Canonical_Species",
    "Omo_Canonical_Species",
    "Exact_Omo_Match",
    "Omo_Canonical_Taxon",
]

value_candidates = [
    "Seed_Mass_mg",
    "Trait_Value_mg",
    "Trait_Value",
    "Standardized_Trait_Value",
    "Median_mg",
    "Median_Accepted_mg",
]

source_candidates = [
    "Trait_Source",
    "Source",
]

quality_candidates = [
    "Trait_Quality_Flag",
    "Evidence_Quality_Flag",
]


# =============================================================================
# 4. AUDIT EVERY CANDIDATE TABLE
# =============================================================================

audit_rows = []

for path in candidate_files:

    try:

        df = pd.read_csv(
            path,
            low_memory=False
        )

    except Exception as exc:

        audit_rows.append({
            "File": path.name,
            "N_Rows": np.nan,
            "N_Columns": np.nan,
            "Canonical_Field": pd.NA,
            "Value_Field": pd.NA,
            "N_Unique_Canonical_Taxa": np.nan,
            "N_Positive_Numeric_Values": np.nan,
            "Trait_Source_Field": pd.NA,
            "Quality_Field": pd.NA,
            "Audit_Status": f"READ_ERROR: {exc}",
        })

        continue


    canonical_col = find_optional_column(
        df,
        canonical_candidates
    )

    value_col = find_optional_column(
        df,
        value_candidates
    )

    source_col = find_optional_column(
        df,
        source_candidates
    )

    quality_col = find_optional_column(
        df,
        quality_candidates
    )


    if canonical_col is not None:

        n_unique_taxa = (
            df[
                canonical_col
            ]
            .dropna()
            .astype(str)
            .str.strip()
            .nunique()
        )

    else:

        n_unique_taxa = np.nan


    if value_col is not None:

        numeric_values = pd.to_numeric(
            df[
                value_col
            ],
            errors="coerce"
        )

        n_positive = int(
            numeric_values
            .gt(0)
            .fillna(False)
            .sum()
        )

    else:

        n_positive = np.nan


    if (
        canonical_col is not None
        and
        value_col is not None
    ):

        status = (
            "POTENTIAL_SPECIES_VALUE_TABLE"
        )

    elif canonical_col is not None:

        status = (
            "HAS_CANONICAL_TAXA_NO_VALUE_FIELD"
        )

    elif value_col is not None:

        status = (
            "HAS_VALUE_NO_CANONICAL_FIELD"
        )

    else:

        status = (
            "NOT_A_CANONICAL_VALUE_TABLE"
        )


    audit_rows.append({
        "File":
            path.name,

        "N_Rows":
            len(df),

        "N_Columns":
            len(df.columns),

        "Canonical_Field":
            canonical_col,

        "Value_Field":
            value_col,

        "N_Unique_Canonical_Taxa":
            n_unique_taxa,

        "N_Positive_Numeric_Values":
            n_positive,

        "Trait_Source_Field":
            source_col,

        "Quality_Field":
            quality_col,

        "Audit_Status":
            status,
    })


audit = pd.DataFrame(
    audit_rows
)


# =============================================================================
# 5. DISPLAY POTENTIAL VALUE TABLES
# =============================================================================

potential = audit[
    audit[
        "Audit_Status"
    ].eq(
        "POTENTIAL_SPECIES_VALUE_TABLE"
    )
].copy()


potential = (
    potential
    .sort_values(
        [
            "N_Unique_Canonical_Taxa",
            "N_Positive_Numeric_Values",
        ],
        ascending=False
    )
    .reset_index(drop=True)
)


print("\nPotential BIEN canonical Seed Mass value tables")
print("-" * 100)

if len(potential) > 0:
    display(potential)
else:
    print("NONE")


# =============================================================================
# 6. LOOK SPECIFICALLY FOR A 28-TAXON TABLE
# =============================================================================

twenty_eight_candidates = potential[
    potential[
        "N_Unique_Canonical_Taxa"
    ].eq(28)
].copy()


print("\nPotential complete 28-taxon BIEN consolidation")
print("-" * 100)

print(
    f"N matching files : "
    f"{len(twenty_eight_candidates)}"
)

if len(twenty_eight_candidates) > 0:

    display(
        twenty_eight_candidates
    )

else:

    print(
        "No single existing file was detected with exactly "
        "28 canonical taxa and a recognized Seed Mass value field."
    )


# =============================================================================
# 7. INSPECT KNOWN COMPONENT TABLES
# =============================================================================

known_components = [
    "omo_seed_mass_bien_final_species_values.csv",
    "omo_seed_mass_bien_alias_new_accepted_species_summary.csv",
    "omo_seed_mass_spondias_species_value.csv",
    "omo_seed_mass_bien_alias_prior_accepted_reused.csv",
]


component_rows = []

for filename in known_components:

    path = OUTPUT_DIR / filename

    if not path.exists():

        component_rows.append({
            "File": filename,
            "Exists": False,
            "Rows": np.nan,
            "Unique_Canonical_Taxa": np.nan,
            "Columns": pd.NA,
        })

        continue


    df = pd.read_csv(
        path,
        low_memory=False
    )

    canonical_col = find_optional_column(
        df,
        canonical_candidates
    )

    if canonical_col is not None:

        n_taxa = (
            df[
                canonical_col
            ]
            .dropna()
            .astype(str)
            .str.strip()
            .nunique()
        )

    else:

        n_taxa = np.nan


    component_rows.append({
        "File":
            filename,

        "Exists":
            True,

        "Rows":
            len(df),

        "Unique_Canonical_Taxa":
            n_taxa,

        "Columns":
            " | ".join(
                map(str, df.columns)
            ),
    })


component_summary = pd.DataFrame(
    component_rows
)


print("\nKnown BIEN rescue component tables")
print("-" * 100)

display(
    component_summary
)


# =============================================================================
# 8. EXPORT AUDIT
# =============================================================================

OUT_AUDIT = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_post_alias_output_audit.csv"
)

OUT_COMPONENTS = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_post_alias_component_audit.csv"
)


audit.to_csv(
    OUT_AUDIT,
    index=False
)

component_summary.to_csv(
    OUT_COMPONENTS,
    index=False
)


# =============================================================================
# 9. REPORT
# =============================================================================

print("\nAudit exports")
print("-" * 100)

print(
    f"All BIEN output audit : "
    f"{OUT_AUDIT}"
)

print(
    f"Component audit       : "
    f"{OUT_COMPONENTS}"
)


print("\n" + "=" * 100)

print(
    "Post-alias-rescue BIEN Seed Mass output audit completed."
)

print(
    "No BIEN evidence has been modified."
)

print(
    "Do not use the current 61/127 BIEN × TRY coverage "
    "as the final consolidation until the validated "
    "28-taxon BIEN evidence layer is restored."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3L-R2 — VALIDATE CONSOLIDATED 28-TAXON BIEN EVIDENCE LAYER
# =============================================================================
#
# Purpose
# -------
# Validate the existing consolidated BIEN Seed Mass table:
#
#     omo_seed_mass_bien_all_species_values_current.csv
#
# against its two principal components:
#
#     1. original BIEN species values              = 15 taxa
#     2. final BIEN taxonomic-rescue values        = 13 taxa
#
# Expected:
#
#     15 + 13 = 28 unique canonical taxa
#
# No evidence is modified here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3L-R2 — VALIDATE CONSOLIDATED "
    "28-TAXON BIEN EVIDENCE LAYER"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

CONSOLIDATED_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_all_species_values_current.csv"
)

BASE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_final_species_values.csv"
)

RESCUE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_final_taxonomic_rescue_values.csv"
)

BACKBONE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_acquisition_manifest.csv"
)

OUT_VALIDATION = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_28_taxon_consolidation_validation.csv"
)

OUT_SET_DIAGNOSTIC = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_28_taxon_set_diagnostic.csv"
)


# =============================================================================
# 2. VALIDATE FILE EXISTENCE
# =============================================================================

for path in [
    CONSOLIDATED_FILE,
    BASE_FILE,
    RESCUE_FILE,
    BACKBONE_FILE,
]:

    assert path.exists(), (
        f"Required file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD TABLES
# =============================================================================

combined = pd.read_csv(
    CONSOLIDATED_FILE,
    low_memory=False
)

base = pd.read_csv(
    BASE_FILE,
    low_memory=False
)

rescue = pd.read_csv(
    RESCUE_FILE,
    low_memory=False
)

backbone = pd.read_csv(
    BACKBONE_FILE,
    low_memory=False
)


print("\nInput tables")
print("-" * 100)

print(
    f"Consolidated BIEN table : {len(combined)} rows"
)

print(
    f"Original BIEN values    : {len(base)} rows"
)

print(
    f"Taxonomic rescue values : {len(rescue)} rows"
)

print(
    f"Canonical backbone      : {len(backbone)} rows"
)


# =============================================================================
# 4. REQUIRED CONSOLIDATED FIELDS
# =============================================================================

required_fields = {
    "Canonical_Species",
    "Trait_Code",
    "Seed_Mass_mg",
    "Standardized_Trait_Unit",
    "Trait_Source",
    "Trait_Quality_Flag",
}

missing_fields = (
    required_fields
    - set(combined.columns)
)

assert not missing_fields, (
    "Consolidated BIEN table is missing required fields:\n"
    f"{sorted(missing_fields)}"
)


# =============================================================================
# 5. BASIC CONSOLIDATED-TABLE VALIDATION
# =============================================================================

combined["Seed_Mass_mg"] = pd.to_numeric(
    combined["Seed_Mass_mg"],
    errors="coerce"
)

assert len(combined) == 28, (
    f"Expected 28 consolidated BIEN taxa, found {len(combined)}."
)

assert (
    combined[
        "Canonical_Species"
    ].nunique()
    == 28
), (
    "Consolidated BIEN table does not contain 28 unique canonical taxa."
)

assert combined[
    "Canonical_Species"
].notna().all()

assert combined[
    "Seed_Mass_mg"
].notna().all()

assert combined[
    "Seed_Mass_mg"
].gt(0).all()

assert (
    combined[
        "Standardized_Trait_Unit"
    ]
    .astype("string")
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
)

assert (
    combined[
        "Trait_Code"
    ]
    .astype("string")
    .str.strip()
    .eq("SEED_MASS")
    .all()
)


print("\nConsolidated BIEN integrity")
print("-" * 100)

print("Rows                 : 28")
print("Unique canonical taxa: 28")
print("Missing Seed Mass    : 0")
print("Non-positive values  : 0")
print("Standardized unit    : mg")
print("Trait code           : SEED_MASS")


# =============================================================================
# 6. VALIDATE AGAINST 127-TAXON BACKBONE
# =============================================================================

backbone_set = set(
    backbone[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)

combined_set = set(
    combined[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)

outside_backbone = (
    combined_set
    - backbone_set
)

assert not outside_backbone, (
    "Consolidated BIEN table contains taxa outside the "
    "127-taxon canonical backbone:\n"
    f"{sorted(outside_backbone)}"
)


print("\nBackbone membership")
print("-" * 100)

print(
    f"Consolidated BIEN taxa in backbone : "
    f"{len(combined_set)} / 28"
)

print(
    "Outside backbone                   : 0"
)


# =============================================================================
# 7. COMPARE AGAINST ORIGINAL + RESCUE COMPONENT SETS
# =============================================================================

base_set = set(
    base[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)

rescue_set = set(
    rescue[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)

component_union = (
    base_set
    | rescue_set
)

component_overlap = (
    base_set
    & rescue_set
)

combined_only = (
    combined_set
    - component_union
)

component_only = (
    component_union
    - combined_set
)


print("\nComponent-set reconciliation")
print("-" * 100)

print(
    f"Original BIEN taxa          : "
    f"{len(base_set)}"
)

print(
    f"Taxonomic rescue taxa       : "
    f"{len(rescue_set)}"
)

print(
    f"Base/rescue overlap         : "
    f"{len(component_overlap)}"
)

print(
    f"Union of components         : "
    f"{len(component_union)}"
)

print(
    f"Consolidated BIEN taxa      : "
    f"{len(combined_set)}"
)

print(
    f"Consolidated-only taxa      : "
    f"{len(combined_only)}"
)

print(
    f"Component-only taxa         : "
    f"{len(component_only)}"
)


# =============================================================================
# 8. SET DIAGNOSTIC TABLE
# =============================================================================

all_diagnostic_taxa = sorted(
    combined_set
    | component_union
)

set_diagnostic = pd.DataFrame(
    {
        "Canonical_Species":
            all_diagnostic_taxa
    }
)

set_diagnostic[
    "In_Original_BIEN"
] = set_diagnostic[
    "Canonical_Species"
].isin(
    base_set
)

set_diagnostic[
    "In_Taxonomic_Rescue"
] = set_diagnostic[
    "Canonical_Species"
].isin(
    rescue_set
)

set_diagnostic[
    "In_Consolidated_28"
] = set_diagnostic[
    "Canonical_Species"
].isin(
    combined_set
)


set_diagnostic[
    "Set_Status"
] = np.where(
    (
        set_diagnostic[
            "In_Consolidated_28"
        ]
        ==
        (
            set_diagnostic[
                "In_Original_BIEN"
            ]
            |
            set_diagnostic[
                "In_Taxonomic_Rescue"
            ]
        )
    ),
    "CONSISTENT",
    "REVIEW"
)


print("\nSet consistency diagnostic")
print("-" * 100)

display(
    set_diagnostic
)


# =============================================================================
# 9. PROVENANCE SUMMARY
# =============================================================================

provenance_summary = (
    combined[
        "Trait_Source"
    ]
    .value_counts()
    .rename_axis(
        "Trait_Source"
    )
    .reset_index(
        name="N_Taxa"
    )
)


print("\nConsolidated BIEN provenance")
print("-" * 100)

display(
    provenance_summary
)


# =============================================================================
# 10. QUALITY SUMMARY
# =============================================================================

quality_summary = (
    combined[
        "Trait_Quality_Flag"
    ]
    .value_counts()
    .rename_axis(
        "Trait_Quality_Flag"
    )
    .reset_index(
        name="N_Taxa"
    )
)


print("\nConsolidated BIEN quality summary")
print("-" * 100)

display(
    quality_summary
)


# =============================================================================
# 11. NUMERIC SUMMARY
# =============================================================================

print("\nConsolidated BIEN Seed Mass numeric summary")
print("-" * 100)

print(
    combined[
        "Seed_Mass_mg"
    ]
    .describe()
    .to_string()
)


# =============================================================================
# 12. FINAL VALIDATION
# =============================================================================

assert len(base_set) == 15, (
    f"Expected 15 original BIEN taxa, found {len(base_set)}."
)

assert len(rescue_set) == 13, (
    f"Expected 13 final taxonomic-rescue taxa, found {len(rescue_set)}."
)

assert len(component_overlap) == 0, (
    "Original BIEN and final rescue tables unexpectedly overlap."
)

assert len(component_union) == 28

assert component_union == combined_set, (
    "The consolidated 28-taxon BIEN table does not exactly equal "
    "the union of the validated original and rescue components."
)

assert (
    set_diagnostic[
        "Set_Status"
    ]
    .eq("CONSISTENT")
    .all()
)


# =============================================================================
# 13. VALIDATION SUMMARY
# =============================================================================

validation_summary = pd.DataFrame({
    "Metric": [
        "Original BIEN canonical taxa",
        "Final taxonomic-rescue taxa",
        "Overlap between components",
        "Component union",
        "Consolidated BIEN taxa",
        "Consolidated taxa outside backbone",
        "Missing/non-positive Seed Mass values",
    ],

    "Value": [
        len(base_set),
        len(rescue_set),
        len(component_overlap),
        len(component_union),
        len(combined_set),
        len(outside_backbone),
        int(
            (
                combined[
                    "Seed_Mass_mg"
                ]
                .isna()
                |
                combined[
                    "Seed_Mass_mg"
                ]
                .le(0)
            )
            .sum()
        ),
    ],
})


print("\n28-taxon BIEN consolidation validation summary")
print("-" * 100)

display(
    validation_summary
)


# =============================================================================
# 14. EXPORT
# =============================================================================

validation_summary.to_csv(
    OUT_VALIDATION,
    index=False
)

set_diagnostic.to_csv(
    OUT_SET_DIAGNOSTIC,
    index=False
)


print("\nValidation exports")
print("-" * 100)

print(
    f"Validation summary : "
    f"{OUT_VALIDATION}"
)

print(
    f"Set diagnostic     : "
    f"{OUT_SET_DIAGNOSTIC}"
)


# =============================================================================
# 15. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Consolidated 28-taxon BIEN Seed Mass evidence "
    "validated successfully."
)

print(
    "The table can now replace the stale 15-taxon BIEN input "
    "in Section 9.3L."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.3L-R3 — CORRECTED BIEN × TRY CONSOLIDATION
#                    AND TRUE RESIDUAL-GAP DETERMINATION
# =============================================================================
#
# Correction
# ----------
# The earlier 9.3L run used the stale 15-taxon BIEN table.
#
# This corrected run uses the validated current BIEN evidence layer:
#
#     omo_seed_mass_bien_all_species_values_current.csv
#
# Validated in Section 9.3L-R2:
#
#     Original BIEN taxa          : 15
#     Taxonomic rescue taxa       : 13
#     Consolidated BIEN taxa      : 28
#     Missing/non-positive values : 0
#
# No source-selection or fallback estimation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.3L-R3 — CORRECTED BIEN × TRY CONSOLIDATION "
    "AND TRUE RESIDUAL-GAP DETERMINATION"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BIEN_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_all_species_values_current.csv"
)

TRY_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_51911_final_omo_canonical_values.csv"
)

BACKBONE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_acquisition_manifest.csv"
)

NON_TRY_RESCUE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_species_non_try_rescue_targets.csv"
)

GENUS_STAGE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_genus_stage_targets.csv"
)


# Outputs — overwrite the earlier stale 9.3L diagnostic with corrected state
OUT_CROSS_SOURCE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_cross_source_consolidation.csv"
)

OUT_OVERLAP = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_overlap_diagnostic.csv"
)

OUT_COVERAGE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_127_taxon_coverage.csv"
)

OUT_TRUE_GAPS = (
    OUTPUT_DIR /
    "omo_seed_mass_true_residual_gaps_after_bien_try.csv"
)

OUT_CATEGORY_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_coverage_category_summary.csv"
)

OUT_OVERLAP_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_overlap_summary.csv"
)

OUT_ROUTING_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_post_bien_try_routing_summary.csv"
)

OUT_GLOBAL_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_try_global_summary.csv"
)


# =============================================================================
# 2. LOAD
# =============================================================================

for path in [
    BIEN_FILE,
    TRY_FILE,
    BACKBONE_FILE,
    NON_TRY_RESCUE_FILE,
    GENUS_STAGE_FILE,
]:
    assert path.exists(), f"Missing required file:\n{path}"


bien = pd.read_csv(
    BIEN_FILE,
    low_memory=False
)

try_values = pd.read_csv(
    TRY_FILE,
    low_memory=False
)

backbone = pd.read_csv(
    BACKBONE_FILE,
    low_memory=False
)

non_try_rescue = pd.read_csv(
    NON_TRY_RESCUE_FILE,
    low_memory=False
)

genus_stage = pd.read_csv(
    GENUS_STAGE_FILE,
    low_memory=False
)


print("\nInput tables")
print("-" * 100)

print(f"Validated BIEN canonical values : {len(bien)}")
print(f"TRY canonical request table     : {len(try_values)}")
print(f"Canonical backbone              : {len(backbone)}")
print(f"Original non-TRY rescue targets : {len(non_try_rescue)}")
print(f"Genus-stage targets             : {len(genus_stage)}")


# =============================================================================
# 3. BASIC VALIDATION
# =============================================================================

assert len(bien) == 28
assert bien["Canonical_Species"].nunique() == 28

assert len(backbone) == 127
assert backbone["Canonical_Species"].nunique() == 127

assert len(non_try_rescue) == 9
assert len(genus_stage) == 3


bien["Seed_Mass_mg"] = pd.to_numeric(
    bien["Seed_Mass_mg"],
    errors="coerce"
)

assert bien["Seed_Mass_mg"].notna().all()
assert bien["Seed_Mass_mg"].gt(0).all()


# =============================================================================
# 4. PREPARE 127-TAXON BACKBONE
# =============================================================================

canonical_backbone = (
    backbone[
        ["Canonical_Species"]
    ]
    .drop_duplicates()
    .copy()
)

assert len(canonical_backbone) == 127
assert canonical_backbone["Canonical_Species"].is_unique


# =============================================================================
# 5. PREPARE BIEN
# =============================================================================

bien_prepared = bien[
    [
        col for col in [
            "Canonical_Species",
            "Seed_Mass_mg",
            "Trait_Source",
            "Trait_Quality_Flag",
            "Evidence_Strength",
        ]
        if col in bien.columns
    ]
].copy()


bien_prepared = bien_prepared.rename(
    columns={
        "Seed_Mass_mg":
            "BIEN_Seed_Mass_mg",

        "Trait_Source":
            "BIEN_Trait_Source",

        "Trait_Quality_Flag":
            "BIEN_Trait_Quality_Flag",

        "Evidence_Strength":
            "BIEN_Evidence_Strength",
    }
)


assert len(bien_prepared) == 28
assert bien_prepared["Canonical_Species"].is_unique


# =============================================================================
# 6. PREPARE RESOLVED TRY
# =============================================================================

try_resolved = try_values[
    try_values["Seed_Mass_mg"].notna()
].copy()


try_resolved["Seed_Mass_mg"] = pd.to_numeric(
    try_resolved["Seed_Mass_mg"],
    errors="coerce"
)


assert len(try_resolved) == 46
assert try_resolved["Canonical_Species"].is_unique
assert try_resolved["Seed_Mass_mg"].gt(0).all()


try_prepared = try_resolved[
    [
        col for col in [
            "Canonical_Species",
            "TRY_Accepted_Species",
            "Seed_Mass_mg",
            "Trait_Source",
            "Trait_Quality_Flag",
            "Evidence_Strength",
        ]
        if col in try_resolved.columns
    ]
].copy()


try_prepared = try_prepared.rename(
    columns={
        "Seed_Mass_mg":
            "TRY_Seed_Mass_mg",

        "Trait_Source":
            "TRY_Trait_Source",

        "Trait_Quality_Flag":
            "TRY_Trait_Quality_Flag",

        "Evidence_Strength":
            "TRY_Evidence_Strength",
    }
)


# =============================================================================
# 7. BACKBONE MEMBERSHIP
# =============================================================================

backbone_set = set(
    canonical_backbone["Canonical_Species"]
)

bien_set = set(
    bien_prepared["Canonical_Species"]
)

try_set = set(
    try_prepared["Canonical_Species"]
)


assert not (
    bien_set - backbone_set
)

assert not (
    try_set - backbone_set
)


# =============================================================================
# 8. MERGE BIEN + TRY
# =============================================================================

coverage = (
    canonical_backbone
    .merge(
        bien_prepared,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
    .merge(
        try_prepared,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


assert len(coverage) == 127


# =============================================================================
# 9. COVERAGE CLASS
# =============================================================================

has_bien = coverage[
    "BIEN_Seed_Mass_mg"
].notna()

has_try = coverage[
    "TRY_Seed_Mass_mg"
].notna()


coverage["BIEN_TRY_Coverage_Class"] = (
    "UNRESOLVED_BY_BIEN_AND_TRY"
)


coverage.loc[
    has_bien & ~has_try,
    "BIEN_TRY_Coverage_Class"
] = "BIEN_ONLY"


coverage.loc[
    ~has_bien & has_try,
    "BIEN_TRY_Coverage_Class"
] = "TRY_ONLY"


coverage.loc[
    has_bien & has_try,
    "BIEN_TRY_Coverage_Class"
] = "BIEN_AND_TRY"


category_summary = (
    coverage[
        "BIEN_TRY_Coverage_Class"
    ]
    .value_counts()
    .rename_axis(
        "BIEN_TRY_Coverage_Class"
    )
    .reset_index(
        name="N_Taxa"
    )
)


category_summary["Percent_of_127"] = (
    category_summary["N_Taxa"]
    / 127
    * 100
)


print("\nCorrected BIEN × TRY canonical coverage")
print("-" * 100)

display(category_summary)


# =============================================================================
# 10. OVERLAP
# =============================================================================

overlap = coverage[
    coverage[
        "BIEN_TRY_Coverage_Class"
    ].eq(
        "BIEN_AND_TRY"
    )
].copy()


print("\nBIEN × TRY overlap")
print("-" * 100)

print(
    f"N overlapping canonical taxa : "
    f"{len(overlap)}"
)


if len(overlap) > 0:

    overlap["BIEN_TRY_Ratio"] = (
        overlap["TRY_Seed_Mass_mg"]
        /
        overlap["BIEN_Seed_Mass_mg"]
    )

    overlap["BIEN_TRY_Fold_Difference"] = np.maximum(
        overlap["BIEN_TRY_Ratio"],
        1 / overlap["BIEN_TRY_Ratio"]
    )

    overlap["Log10_TRY_minus_BIEN"] = (
        np.log10(
            overlap["TRY_Seed_Mass_mg"]
        )
        -
        np.log10(
            overlap["BIEN_Seed_Mass_mg"]
        )
    )

    overlap["Agreement_Class"] = (
        "LARGE_DIFFERENCE_GT_5_FOLD"
    )

    overlap.loc[
        overlap[
            "BIEN_TRY_Fold_Difference"
        ].le(5),
        "Agreement_Class"
    ] = "SUBSTANTIAL_DIFFERENCE_LE_5_FOLD"

    overlap.loc[
        overlap[
            "BIEN_TRY_Fold_Difference"
        ].le(2),
        "Agreement_Class"
    ] = "MODERATE_AGREEMENT_LE_2_FOLD"

    overlap.loc[
        overlap[
            "BIEN_TRY_Fold_Difference"
        ].le(1.5),
        "Agreement_Class"
    ] = "CLOSE_AGREEMENT_LE_1_5_FOLD"

    display(
        overlap[
            [
                "Canonical_Species",
                "BIEN_Seed_Mass_mg",
                "TRY_Seed_Mass_mg",
                "BIEN_TRY_Fold_Difference",
                "Agreement_Class",
            ]
        ]
        .sort_values(
            "BIEN_TRY_Fold_Difference",
            ascending=False
        )
        .reset_index(drop=True)
    )

else:

    print(
        "NONE — expected because TRY was requested from "
        "the post-BIEN residual target set."
    )


# =============================================================================
# 11. OVERLAP SUMMARY
# =============================================================================

if len(overlap) > 0:

    overlap_summary = (
        overlap[
            "Agreement_Class"
        ]
        .value_counts()
        .rename_axis(
            "Agreement_Class"
        )
        .reset_index(
            name="N_Taxa"
        )
    )

else:

    overlap_summary = pd.DataFrame(
        columns=[
            "Agreement_Class",
            "N_Taxa",
        ]
    )


# =============================================================================
# 12. CURRENT SPECIES-LEVEL COVERAGE
# =============================================================================

coverage[
    "Any_Species_Level_Seed_Mass"
] = (
    coverage[
        "BIEN_Seed_Mass_mg"
    ].notna()
    |
    coverage[
        "TRY_Seed_Mass_mg"
    ].notna()
)


n_species_resolved = int(
    coverage[
        "Any_Species_Level_Seed_Mass"
    ]
    .sum()
)

n_true_gaps = (
    127 - n_species_resolved
)


print("\nCorrected species-level Seed Mass coverage")
print("-" * 100)

print(
    f"Taxa resolved by BIEN and/or TRY : "
    f"{n_species_resolved}"
)

print(
    f"Taxa unresolved by both          : "
    f"{n_true_gaps}"
)


# =============================================================================
# 13. ROUTING SETS
# =============================================================================

non_try_set = set(
    non_try_rescue[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)

genus_set = set(
    genus_stage[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


assert len(non_try_set) == 9
assert len(genus_set) == 3


coverage[
    "Previously_Routed_Non_TRY_Species_Rescue"
] = (
    coverage[
        "Canonical_Species"
    ]
    .isin(
        non_try_set
    )
)


coverage[
    "Previously_Routed_Genus_Stage"
] = (
    coverage[
        "Canonical_Species"
    ]
    .isin(
        genus_set
    )
)


# =============================================================================
# 14. POST-BIEN/TRY NEXT ACTION
# =============================================================================

coverage[
    "Post_BIEN_TRY_Next_Action"
] = (
    "NO_FURTHER_SPECIES_ACQUISITION_REQUIRED"
)


unresolved = (
    ~coverage[
        "Any_Species_Level_Seed_Mass"
    ]
)


coverage.loc[
    (
        unresolved
        &
        coverage[
            "Previously_Routed_Genus_Stage"
        ]
    ),
    "Post_BIEN_TRY_Next_Action"
] = "PROCEED_TO_GENUS_LEVEL_STAGE"


coverage.loc[
    (
        unresolved
        &
        ~coverage[
            "Previously_Routed_Genus_Stage"
        ]
    ),
    "Post_BIEN_TRY_Next_Action"
] = "TARGETED_SPECIES_NON_TRY_RESCUE"


# =============================================================================
# 15. TRUE RESIDUAL GAPS
# =============================================================================

true_gaps = coverage[
    unresolved
].copy()


print("\nCorrected true Seed Mass gaps after BIEN + TRY")
print("-" * 100)

print(
    f"N unresolved canonical taxa : "
    f"{len(true_gaps)}"
)


display(
    true_gaps[
        [
            "Canonical_Species",
            "BIEN_TRY_Coverage_Class",
            "Previously_Routed_Non_TRY_Species_Rescue",
            "Previously_Routed_Genus_Stage",
            "Post_BIEN_TRY_Next_Action",
        ]
    ]
    .sort_values(
        [
            "Post_BIEN_TRY_Next_Action",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 16. ROUTING SUMMARY
# =============================================================================

routing_summary = (
    coverage[
        "Post_BIEN_TRY_Next_Action"
    ]
    .value_counts()
    .rename_axis(
        "Post_BIEN_TRY_Next_Action"
    )
    .reset_index(
        name="N_Taxa"
    )
)


routing_summary[
    "Percent_of_127"
] = (
    routing_summary["N_Taxa"]
    / 127
    * 100
)


print("\nCorrected post-BIEN/TRY routing summary")
print("-" * 100)

display(routing_summary)


# =============================================================================
# 17. GLOBAL SUMMARY
# =============================================================================

n_bien = int(
    coverage[
        "BIEN_Seed_Mass_mg"
    ]
    .notna()
    .sum()
)

n_try = int(
    coverage[
        "TRY_Seed_Mass_mg"
    ]
    .notna()
    .sum()
)

n_overlap = len(overlap)

n_non_try_rescue = int(
    coverage[
        "Post_BIEN_TRY_Next_Action"
    ]
    .eq(
        "TARGETED_SPECIES_NON_TRY_RESCUE"
    )
    .sum()
)

n_genus_stage = int(
    coverage[
        "Post_BIEN_TRY_Next_Action"
    ]
    .eq(
        "PROCEED_TO_GENUS_LEVEL_STAGE"
    )
    .sum()
)


global_summary = pd.DataFrame({
    "Metric": [
        "Canonical Omo taxa",
        "Taxa with validated BIEN species evidence",
        "Taxa with TRY species evidence",
        "Taxa with both BIEN and TRY evidence",
        "Taxa resolved by BIEN and/or TRY",
        "True residual taxa after BIEN and TRY",
        "Taxa routed to targeted species non-TRY rescue",
        "Taxa routed to genus-level stage",
    ],

    "Value": [
        127,
        n_bien,
        n_try,
        n_overlap,
        n_species_resolved,
        n_true_gaps,
        n_non_try_rescue,
        n_genus_stage,
    ],
})


print("\nCorrected BIEN × TRY consolidation summary")
print("-" * 100)

display(global_summary)


# =============================================================================
# 18. STRONG VALIDATION
# =============================================================================

assert n_bien == 28, (
    f"Expected 28 BIEN taxa, found {n_bien}."
)

assert n_try == 46, (
    f"Expected 46 TRY taxa, found {n_try}."
)

# TRY was constructed from the post-BIEN residual pool.
assert n_overlap == 0, (
    "Unexpected BIEN × TRY overlap detected."
)

assert n_species_resolved == 74, (
    f"Expected 74 taxa resolved by BIEN/TRY, "
    f"found {n_species_resolved}."
)

assert n_true_gaps == 53, (
    f"Expected 53 true residual taxa, "
    f"found {n_true_gaps}."
)

assert n_genus_stage == 3, (
    f"Expected 3 genus-stage taxa, found {n_genus_stage}."
)

assert n_non_try_rescue == 50, (
    f"Expected 50 species-level non-TRY rescue taxa, "
    f"found {n_non_try_rescue}."
)

assert (
    n_species_resolved
    + n_non_try_rescue
    + n_genus_stage
    == 127
)


print("\nCorrected cross-source validation")
print("-" * 100)

print("Validated BIEN evidence            : 28")
print("TRY species evidence               : 46")
print("BIEN × TRY overlap                 : 0")
print("Species-level resolved             : 74")
print("True residual taxa                 : 53")
print("Targeted species non-TRY rescue    : 50")
print("Genus-level stage                  : 3")


# =============================================================================
# 19. EXPORT
# =============================================================================

coverage.to_csv(
    OUT_CROSS_SOURCE,
    index=False
)

overlap.to_csv(
    OUT_OVERLAP,
    index=False
)

coverage.to_csv(
    OUT_COVERAGE,
    index=False
)

true_gaps.to_csv(
    OUT_TRUE_GAPS,
    index=False
)

category_summary.to_csv(
    OUT_CATEGORY_SUMMARY,
    index=False
)

overlap_summary.to_csv(
    OUT_OVERLAP_SUMMARY,
    index=False
)

routing_summary.to_csv(
    OUT_ROUTING_SUMMARY,
    index=False
)

global_summary.to_csv(
    OUT_GLOBAL_SUMMARY,
    index=False
)


# =============================================================================
# 20. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Corrected BIEN × TRY Seed Mass consolidation "
    "completed successfully."
)

print(
    f"Species-level taxa resolved by BIEN and/or TRY : "
    f"{n_species_resolved} / 127"
)

print(
    f"True residual taxa                           : "
    f"{n_true_gaps} / 127"
)

print(
    f"Targeted species non-TRY rescue              : "
    f"{n_non_try_rescue}"
)

print(
    f"Genus-level stage                            : "
    f"{n_genus_stage}"
)

print(
    "The stale 15-taxon BIEN consolidation outputs have now been "
    "replaced by the validated 28-taxon BIEN state."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4A — PREPARE AND TAXONOMICALLY AUDIT THE 50 RESIDUAL
#                SPECIES-LEVEL SEED MASS RESCUE TARGETS
# =============================================================================
#
# Purpose
# -------
# Prepare the 50 species-level residual Seed Mass targets for external rescue.
#
# This section:
#
#   1. loads the validated post-BIEN/TRY residual gap table;
#   2. isolates the 50 species-level non-TRY rescue taxa;
#   3. derives basic genus/species-name structure;
#   4. searches existing Notebook 07B taxonomic outputs for previously
#      validated alternative / accepted / query names;
#   5. builds a controlled canonical-to-search-target crosswalk;
#   6. identifies cases where multiple Omo canonical strings resolve to
#      the same biological search target;
#   7. exports a deduplicated external-search manifest.
#
# IMPORTANT
# ---------
# - No Seed Mass value is assigned here.
# - No new synonymy is invented.
# - Only previously validated project taxonomic evidence is used to change
#   a search target.
# - Unresolved names remain explicitly unresolved for later manual review.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print(
    "SECTION 9.4A — PREPARE AND TAXONOMICALLY AUDIT THE 50 "
    "RESIDUAL SPECIES-LEVEL SEED MASS RESCUE TARGETS"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

RESIDUAL_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_true_residual_gaps_after_bien_try.csv"
)

# Search across existing functional-trait CSV outputs for previously validated
# taxonomic evidence.
TAXONOMIC_SEARCH_DIR = OUTPUT_DIR


# Outputs
OUT_TARGET_CROSSWALK = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_crosswalk.csv"
)

OUT_UNIQUE_SEARCH_MANIFEST = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_unique_search_manifest.csv"
)

OUT_UNRESOLVED_TAXONOMY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_review_required.csv"
)

OUT_DUPLICATE_TARGET_GROUPS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_shared_search_targets.csv"
)

OUT_AUDIT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_audit_summary.csv"
)


# =============================================================================
# 2. LOAD RESIDUAL GAP TABLE
# =============================================================================

assert RESIDUAL_FILE.exists(), (
    f"Residual gap table not found:\n{RESIDUAL_FILE}"
)

residual = pd.read_csv(
    RESIDUAL_FILE,
    low_memory=False
)

print("\nResidual gap table")
print("-" * 100)

print(
    f"Rows : {len(residual)}"
)


required_residual_fields = {
    "Canonical_Species",
    "Post_BIEN_TRY_Next_Action",
}

missing = (
    required_residual_fields
    - set(residual.columns)
)

assert not missing, (
    "Residual table missing required fields:\n"
    f"{sorted(missing)}"
)


# =============================================================================
# 3. ISOLATE THE 50 SPECIES-LEVEL NON-TRY RESCUE TARGETS
# =============================================================================

targets = residual[
    residual[
        "Post_BIEN_TRY_Next_Action"
    ].eq(
        "TARGETED_SPECIES_NON_TRY_RESCUE"
    )
].copy()


targets = (
    targets[
        [
            "Canonical_Species"
        ]
    ]
    .drop_duplicates()
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


assert len(targets) == 50, (
    f"Expected 50 species-level rescue targets, found {len(targets)}."
)

assert targets[
    "Canonical_Species"
].is_unique


print("\nSpecies-level non-TRY rescue targets")
print("-" * 100)

print(
    f"N canonical taxa : "
    f"{len(targets)}"
)

display(targets)


# =============================================================================
# 4. BASIC NAME STRUCTURE
# =============================================================================

def normalize_spaces(value):

    if pd.isna(value):
        return pd.NA

    return re.sub(
        r"\s+",
        " ",
        str(value).strip()
    )


targets[
    "Canonical_Species"
] = targets[
    "Canonical_Species"
].apply(
    normalize_spaces
)


targets[
    "Canonical_Genus"
] = (
    targets[
        "Canonical_Species"
    ]
    .str.split()
    .str[0]
)


targets[
    "Canonical_Name_Token_Count"
] = (
    targets[
        "Canonical_Species"
    ]
    .str.split()
    .str.len()
)


targets[
    "Species_Level_Name_Structure"
] = np.where(
    targets[
        "Canonical_Name_Token_Count"
    ].ge(2),
    "BINOMIAL_OR_MULTITOKEN",
    "GENUS_ONLY"
)


print("\nTarget name structure")
print("-" * 100)

display(
    targets[
        "Species_Level_Name_Structure"
    ]
    .value_counts()
    .rename_axis(
        "Name_Structure"
    )
    .reset_index(
        name="N_Taxa"
    )
)


# =============================================================================
# 5. DISCOVER EXISTING TAXONOMIC EVIDENCE TABLES
# =============================================================================
#
# We search CSV outputs for fields that commonly store validated alternative,
# accepted, corrected, query, or normalized names.
# =============================================================================

csv_files = sorted(
    TAXONOMIC_SEARCH_DIR.glob("*.csv")
)


candidate_taxonomic_fields = {
    "Canonical_Species",
    "Omo_Canonical_Species",
    "Source_Species",
    "Query_Name",
    "Preferred_TRY_Query_Name",
    "Validated_BIEN_Query_Name",
    "Alternative_Query_Name",
    "Candidate_Validated_Name",
    "Resolved_TRY_Query_Name",
    "TRY_Accepted_Species",
    "Accepted_Species",
    "Accepted_Name",
    "Normalized_Name",
    "Validated_Taxonomic_Name",
    "Validated_Name",
    "Matched_Name",
    "SpeciesName",
    "AccSpeciesName",
}


evidence_rows = []


for path in csv_files:

    # Avoid recursively using the current new outputs if rerunning the cell.
    if path.name.startswith(
        "omo_seed_mass_non_try_rescue_"
    ):
        continue

    try:

        df = pd.read_csv(
            path,
            low_memory=False
        )

    except Exception:
        continue

    present_fields = [
        col
        for col in df.columns
        if col in candidate_taxonomic_fields
    ]

    if len(present_fields) < 2:
        continue

    canonical_fields = [
        col
        for col in [
            "Canonical_Species",
            "Omo_Canonical_Species",
        ]
        if col in df.columns
    ]

    if not canonical_fields:
        continue


    for canonical_col in canonical_fields:

        subset = df[
            df[
                canonical_col
            ]
            .astype(str)
            .isin(
                set(
                    targets[
                        "Canonical_Species"
                    ].astype(str)
                )
            )
        ].copy()

        if subset.empty:
            continue


        alternative_fields = [
            col
            for col in present_fields
            if col != canonical_col
        ]


        for _, row in subset.iterrows():

            canonical_name = (
                row[
                    canonical_col
                ]
            )

            if pd.isna(canonical_name):
                continue

            for alt_col in alternative_fields:

                alt_value = row[
                    alt_col
                ]

                if pd.isna(alt_value):
                    continue

                alt_value = normalize_spaces(
                    alt_value
                )

                canonical_name_norm = normalize_spaces(
                    canonical_name
                )

                if (
                    pd.isna(alt_value)
                    or
                    alt_value == ""
                    or
                    alt_value == canonical_name_norm
                ):
                    continue


                evidence_rows.append({
                    "Canonical_Species":
                        canonical_name_norm,

                    "Candidate_Search_Name":
                        alt_value,

                    "Evidence_Field":
                        alt_col,

                    "Evidence_Source_File":
                        path.name,
                })


taxonomic_evidence = pd.DataFrame(
    evidence_rows
)


if len(taxonomic_evidence) > 0:

    taxonomic_evidence = (
        taxonomic_evidence
        .drop_duplicates()
        .reset_index(drop=True)
    )

else:

    taxonomic_evidence = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Candidate_Search_Name",
            "Evidence_Field",
            "Evidence_Source_File",
        ]
    )


print("\nPreviously validated alternative-name evidence discovered")
print("-" * 100)

print(
    f"Evidence rows : "
    f"{len(taxonomic_evidence)}"
)

print(
    f"Canonical taxa with candidate alternatives : "
    f"{taxonomic_evidence['Canonical_Species'].nunique()}"
)


# =============================================================================
# 6. SCORE EVIDENCE FIELDS BY RELIABILITY
# =============================================================================
#
# Higher priority means stronger evidence that the field represents a
# deliberately validated search/accepted name.
# =============================================================================

field_priority = {
    "Validated_Taxonomic_Name": 100,
    "Validated_Name": 95,
    "Candidate_Validated_Name": 90,
    "Alternative_Query_Name": 85,
    "Validated_BIEN_Query_Name": 85,
    "Resolved_TRY_Query_Name": 85,
    "Preferred_TRY_Query_Name": 80,
    "TRY_Accepted_Species": 75,
    "AccSpeciesName": 70,
    "Accepted_Species": 70,
    "Accepted_Name": 70,
    "Normalized_Name": 65,
    "Matched_Name": 60,
    "Query_Name": 55,
    "SpeciesName": 40,
    "Source_Species": 30,
}


if len(taxonomic_evidence) > 0:

    taxonomic_evidence[
        "Evidence_Priority"
    ] = (
        taxonomic_evidence[
            "Evidence_Field"
        ]
        .map(
            field_priority
        )
        .fillna(10)
        .astype(int)
    )

else:

    taxonomic_evidence[
        "Evidence_Priority"
    ] = pd.Series(
        dtype=int
    )


# =============================================================================
# 7. SELECT BEST PREVIOUSLY VALIDATED SEARCH NAME
# =============================================================================

best_rows = []


for canonical_name in targets[
    "Canonical_Species"
]:

    subset = taxonomic_evidence[
        taxonomic_evidence[
            "Canonical_Species"
        ].eq(
            canonical_name
        )
    ].copy()

    if subset.empty:

        best_rows.append({
            "Canonical_Species":
                canonical_name,

            "Validated_Search_Name":
                canonical_name,

            "Taxonomic_Search_Status":
                "USE_CANONICAL_NAME",

            "Selected_Evidence_Field":
                pd.NA,

            "Selected_Evidence_Source":
                pd.NA,

            "Selected_Evidence_Priority":
                np.nan,

            "N_Alternative_Name_Candidates":
                0,
        })

        continue


    # Prefer highest evidence priority.
    # If tied, use the most frequently occurring candidate name.
    candidate_frequency = (
        subset[
            "Candidate_Search_Name"
        ]
        .value_counts()
        .rename(
            "Candidate_Frequency"
        )
    )


    subset = subset.merge(
        candidate_frequency,
        left_on="Candidate_Search_Name",
        right_index=True,
        how="left"
    )


    subset = subset.sort_values(
        [
            "Evidence_Priority",
            "Candidate_Frequency",
            "Candidate_Search_Name",
        ],
        ascending=[
            False,
            False,
            True,
        ]
    )


    best = subset.iloc[0]


    best_rows.append({
        "Canonical_Species":
            canonical_name,

        "Validated_Search_Name":
            best[
                "Candidate_Search_Name"
            ],

        "Taxonomic_Search_Status":
            "PREVIOUSLY_VALIDATED_ALTERNATIVE_NAME",

        "Selected_Evidence_Field":
            best[
                "Evidence_Field"
            ],

        "Selected_Evidence_Source":
            best[
                "Evidence_Source_File"
            ],

        "Selected_Evidence_Priority":
            best[
                "Evidence_Priority"
            ],

        "N_Alternative_Name_Candidates":
            subset[
                "Candidate_Search_Name"
            ]
            .nunique(),
    })


selected_names = pd.DataFrame(
    best_rows
)


# =============================================================================
# 8. BUILD CROSSWALK
# =============================================================================

crosswalk = targets.merge(
    selected_names,
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)


crosswalk[
    "Search_Genus"
] = (
    crosswalk[
        "Validated_Search_Name"
    ]
    .astype("string")
    .str.split()
    .str[0]
)


crosswalk[
    "Search_Name_Token_Count"
] = (
    crosswalk[
        "Validated_Search_Name"
    ]
    .astype("string")
    .str.split()
    .str.len()
)


# =============================================================================
# 9. FLAG CASES REQUIRING TAXONOMIC REVIEW
# =============================================================================
#
# Review is required when:
# - selected search name is genus-only;
# - search genus changes without a strong validated field;
# - multiple conflicting alternative candidates exist and the selected
#   evidence priority is weak.
# =============================================================================

crosswalk[
    "Taxonomic_Review_Flag"
] = "NO_REVIEW_REQUIRED"


crosswalk.loc[
    crosswalk[
        "Search_Name_Token_Count"
    ].lt(2),
    "Taxonomic_Review_Flag"
] = "REVIEW_GENUS_ONLY_SEARCH_NAME"


genus_changed = (
    crosswalk[
        "Canonical_Genus"
    ].astype(str)
    !=
    crosswalk[
        "Search_Genus"
    ].astype(str)
)


weak_priority = (
    pd.to_numeric(
        crosswalk[
            "Selected_Evidence_Priority"
        ],
        errors="coerce"
    )
    .fillna(0)
    .lt(55)
)


crosswalk.loc[
    genus_changed
    &
    weak_priority,
    "Taxonomic_Review_Flag"
] = "REVIEW_GENUS_CHANGE_WEAK_EVIDENCE"


conflicting_candidates = (
    crosswalk[
        "N_Alternative_Name_Candidates"
    ]
    .gt(1)
    &
    pd.to_numeric(
        crosswalk[
            "Selected_Evidence_Priority"
        ],
        errors="coerce"
    )
    .fillna(0)
    .lt(70)
)


crosswalk.loc[
    conflicting_candidates,
    "Taxonomic_Review_Flag"
] = "REVIEW_MULTIPLE_WEAK_ALTERNATIVES"


# =============================================================================
# 10. IDENTIFY SHARED BIOLOGICAL SEARCH TARGETS
# =============================================================================

target_counts = (
    crosswalk[
        "Validated_Search_Name"
    ]
    .value_counts()
    .rename_axis(
        "Validated_Search_Name"
    )
    .reset_index(
        name="N_Omo_Canonical_Taxa"
    )
)


crosswalk = crosswalk.merge(
    target_counts,
    on="Validated_Search_Name",
    how="left",
    validate="many_to_one"
)


crosswalk[
    "Shared_Search_Target"
] = (
    crosswalk[
        "N_Omo_Canonical_Taxa"
    ]
    .gt(1)
)


# =============================================================================
# 11. SHARED-TARGET GROUPS
# =============================================================================

shared_groups = crosswalk[
    crosswalk[
        "Shared_Search_Target"
    ]
].copy()


if len(shared_groups) > 0:

    shared_target_summary = (
        shared_groups
        .groupby(
            "Validated_Search_Name",
            dropna=False
        )
        .agg(
            N_Omo_Canonical_Taxa=(
                "Canonical_Species",
                "size"
            ),

            Omo_Canonical_Taxa=(
                "Canonical_Species",
                lambda x: " | ".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),
        )
        .reset_index()
    )

else:

    shared_target_summary = pd.DataFrame(
        columns=[
            "Validated_Search_Name",
            "N_Omo_Canonical_Taxa",
            "Omo_Canonical_Taxa",
        ]
    )


print("\nShared biological search targets")
print("-" * 100)

print(
    f"N shared search targets : "
    f"{len(shared_target_summary)}"
)

if len(shared_target_summary) > 0:
    display(shared_target_summary)
else:
    print("NONE")


# =============================================================================
# 12. BUILD UNIQUE EXTERNAL SEARCH MANIFEST
# =============================================================================
#
# A search name may represent multiple project canonical strings.
# We search each biological target once and map the evidence back later.
# =============================================================================

search_manifest = (
    crosswalk[
        [
            "Validated_Search_Name",
            "Search_Genus",
            "Taxonomic_Review_Flag",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Validated_Search_Name"
        ]
    )
    .copy()
)


search_manifest[
    "Search_Target_ID"
] = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(
        1,
        len(search_manifest) + 1
    )
]


search_manifest[
    "Requested_Trait"
] = "seed mass"

search_manifest[
    "Standardized_Unit"
] = "mg"

search_manifest[
    "Preferred_Measurement_Entity"
] = (
    "individual seed dry mass"
)

search_manifest[
    "Search_Status"
] = np.where(
    search_manifest[
        "Taxonomic_Review_Flag"
    ].eq(
        "NO_REVIEW_REQUIRED"
    ),
    "READY_FOR_EXTERNAL_SEARCH",
    "TAXONOMIC_REVIEW_BEFORE_SEARCH"
)


# Put ID first.
search_manifest = search_manifest[
    [
        "Search_Target_ID",
        "Validated_Search_Name",
        "Search_Genus",
        "Requested_Trait",
        "Standardized_Unit",
        "Preferred_Measurement_Entity",
        "Taxonomic_Review_Flag",
        "Search_Status",
    ]
]


# =============================================================================
# 13. ATTACH SEARCH TARGET IDS BACK TO CROSSWALK
# =============================================================================

crosswalk = crosswalk.merge(
    search_manifest[
        [
            "Search_Target_ID",
            "Validated_Search_Name",
            "Search_Status",
        ]
    ],
    on="Validated_Search_Name",
    how="left",
    validate="many_to_one"
)


# =============================================================================
# 14. TAXONOMIC REVIEW TABLE
# =============================================================================

taxonomic_review = crosswalk[
    ~crosswalk[
        "Taxonomic_Review_Flag"
    ].eq(
        "NO_REVIEW_REQUIRED"
    )
].copy()


print("\nTaxonomic review required before external search")
print("-" * 100)

print(
    f"N canonical taxa : "
    f"{len(taxonomic_review)}"
)

if len(taxonomic_review) > 0:

    display(
        taxonomic_review[
            [
                "Canonical_Species",
                "Validated_Search_Name",
                "Selected_Evidence_Field",
                "Selected_Evidence_Source",
                "N_Alternative_Name_Candidates",
                "Taxonomic_Review_Flag",
            ]
        ]
        .sort_values(
            "Canonical_Species"
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 15. DISPLAY FINAL CROSSWALK
# =============================================================================

print("\nCanonical-to-external-search crosswalk")
print("-" * 100)

display(
    crosswalk[
        [
            "Canonical_Species",
            "Canonical_Genus",
            "Validated_Search_Name",
            "Search_Genus",
            "Taxonomic_Search_Status",
            "Selected_Evidence_Field",
            "Selected_Evidence_Source",
            "N_Alternative_Name_Candidates",
            "N_Omo_Canonical_Taxa",
            "Shared_Search_Target",
            "Taxonomic_Review_Flag",
            "Search_Target_ID",
            "Search_Status",
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 16. DISPLAY UNIQUE SEARCH MANIFEST
# =============================================================================

print("\nUnique external Seed Mass search manifest")
print("-" * 100)

print(
    f"Canonical rescue taxa : "
    f"{len(crosswalk)}"
)

print(
    f"Unique biological search targets : "
    f"{len(search_manifest)}"
)

display(
    search_manifest
    .sort_values(
        "Validated_Search_Name"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 17. AUDIT SUMMARY
# =============================================================================

audit_summary = pd.DataFrame({
    "Metric": [
        "Canonical species-level rescue taxa",
        "Canonical taxa using canonical name for search",
        "Canonical taxa using previously validated alternative name",
        "Unique biological search targets",
        "Shared biological search targets",
        "Canonical taxa requiring taxonomic review",
        "Unique targets ready for external search",
        "Unique targets requiring review before search",
    ],

    "Value": [
        len(crosswalk),

        int(
            crosswalk[
                "Taxonomic_Search_Status"
            ]
            .eq(
                "USE_CANONICAL_NAME"
            )
            .sum()
        ),

        int(
            crosswalk[
                "Taxonomic_Search_Status"
            ]
            .eq(
                "PREVIOUSLY_VALIDATED_ALTERNATIVE_NAME"
            )
            .sum()
        ),

        len(search_manifest),

        len(shared_target_summary),

        len(taxonomic_review),

        int(
            search_manifest[
                "Search_Status"
            ]
            .eq(
                "READY_FOR_EXTERNAL_SEARCH"
            )
            .sum()
        ),

        int(
            search_manifest[
                "Search_Status"
            ]
            .eq(
                "TAXONOMIC_REVIEW_BEFORE_SEARCH"
            )
            .sum()
        ),
    ],
})


print("\nSection 9.4A audit summary")
print("-" * 100)

display(audit_summary)


# =============================================================================
# 18. STRONG VALIDATION
# =============================================================================

assert len(crosswalk) == 50

assert crosswalk[
    "Canonical_Species"
].is_unique

assert crosswalk[
    "Validated_Search_Name"
].notna().all()

assert crosswalk[
    "Search_Target_ID"
].notna().all()

assert search_manifest[
    "Validated_Search_Name"
].is_unique

assert (
    search_manifest[
        "Search_Target_ID"
    ]
    .is_unique
)

assert (
    crosswalk[
        "Search_Target_ID"
    ]
    .isin(
        set(
            search_manifest[
                "Search_Target_ID"
            ]
        )
    )
    .all()
)


# =============================================================================
# 19. EXPORT
# =============================================================================

crosswalk.to_csv(
    OUT_TARGET_CROSSWALK,
    index=False
)

search_manifest.to_csv(
    OUT_UNIQUE_SEARCH_MANIFEST,
    index=False
)

taxonomic_review.to_csv(
    OUT_UNRESOLVED_TAXONOMY,
    index=False
)

shared_target_summary.to_csv(
    OUT_DUPLICATE_TARGET_GROUPS,
    index=False
)

audit_summary.to_csv(
    OUT_AUDIT_SUMMARY,
    index=False
)


# =============================================================================
# 20. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4A exports")
print("-" * 100)

print(
    f"Taxonomic crosswalk       : "
    f"{OUT_TARGET_CROSSWALK}"
)

print(
    f"Unique search manifest    : "
    f"{OUT_UNIQUE_SEARCH_MANIFEST}"
)

print(
    f"Taxonomic review required : "
    f"{OUT_UNRESOLVED_TAXONOMY}"
)

print(
    f"Shared search targets     : "
    f"{OUT_DUPLICATE_TARGET_GROUPS}"
)

print(
    f"Audit summary             : "
    f"{OUT_AUDIT_SUMMARY}"
)


# =============================================================================
# 21. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Residual Seed Mass species-level rescue target "
    "preparation completed successfully."
)

print(
    f"Omo canonical rescue taxa       : "
    f"{len(crosswalk)}"
)

print(
    f"Unique biological search targets: "
    f"{len(search_manifest)}"
)

print(
    f"Taxonomic review cases          : "
    f"{len(taxonomic_review)}"
)

print(
    "No Seed Mass values have been assigned."
)

print(
    "No external trait search has yet been performed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4A-R — CORRECT SPECIES-LEVEL TAXONOMIC SEARCH TARGETS
# =============================================================================
#
# Purpose
# -------
# Repair the provisional Section 9.4A search-target manifest by enforcing
# strict species-level taxonomic rules.
#
# Rules
# -----
# 1. A valid replacement search name must be species-level.
# 2. Genus-only alternatives never replace a usable canonical binomial.
# 3. Literal placeholders GENUS / SPECIES are invalid.
# 4. Congeneric pseudo-alias tables are explicitly excluded.
# 5. Previously validated species-level aliases may replace the canonical name.
# 6. Otherwise the canonical Omo species string remains the search target.
#
# No Seed Mass evidence is acquired here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print(
    "SECTION 9.4A-R — CORRECT SPECIES-LEVEL "
    "TAXONOMIC SEARCH TARGETS"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

CROSSWALK_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_crosswalk.csv"
)

# Earlier alias evidence tables that contain genuinely reviewed species names.
VALIDATED_ALIAS_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_validated_taxonomic_aliases.csv"
)

BIEN_ALIAS_STATUS_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_bien_alias_requery_status.csv"
)

TRY_GAP_CANDIDATES_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_try_gap_validated_name_candidates.csv"
)


OUT_CORRECTED_CROSSWALK = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_crosswalk_corrected.csv"
)

OUT_CORRECTED_MANIFEST = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_unique_search_manifest_corrected.csv"
)

OUT_SHARED_TARGETS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_shared_search_targets_corrected.csv"
)

OUT_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_review_required_corrected.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_audit_summary_corrected.csv"
)


# =============================================================================
# 2. LOAD 50-TAXON CROSSWALK
# =============================================================================

assert CROSSWALK_FILE.exists()

base = pd.read_csv(
    CROSSWALK_FILE,
    low_memory=False
)

assert len(base) == 50
assert base["Canonical_Species"].is_unique


# =============================================================================
# 3. HELPERS
# =============================================================================

INVALID_TAXON_TOKENS = {
    "",
    "genus",
    "species",
    "sp",
    "sp.",
    "spp",
    "spp.",
    "na",
    "nan",
    "none",
    "unknown",
    "unresolved",
}


def clean_name(value):

    if pd.isna(value):
        return None

    value = re.sub(
        r"\s+",
        " ",
        str(value).strip()
    )

    if value.lower() in INVALID_TAXON_TOKENS:
        return None

    return value


def is_species_level(value):

    value = clean_name(value)

    if value is None:
        return False

    tokens = value.split()

    if len(tokens) < 2:
        return False

    if (
        tokens[0].lower() in INVALID_TAXON_TOKENS
        or
        tokens[1].lower() in INVALID_TAXON_TOKENS
    ):
        return False

    return True


# =============================================================================
# 4. START FROM CANONICAL BINOMIALS
# =============================================================================
#
# Default scientific rule:
#
#     SEARCH THE CANONICAL SPECIES ITSELF
#
# unless a stronger, previously validated species-level accepted/alias name
# exists.
# =============================================================================

corrected = base[
    [
        "Canonical_Species"
    ]
].copy()


corrected[
    "Corrected_Search_Name"
] = corrected[
    "Canonical_Species"
].apply(clean_name)


corrected[
    "Search_Name_Source"
] = "OMO_CANONICAL_SPECIES"


corrected[
    "Taxonomic_Resolution_Class"
] = "CANONICAL_SPECIES_RETAINED"


# =============================================================================
# 5. COLLECT ONLY LEGITIMATE SPECIES-LEVEL VALIDATED ALIASES
# =============================================================================

validated_rows = []


def add_alias_candidates(
    path,
    canonical_fields,
    alias_fields,
    source_class
):

    if not path.exists():
        return

    df = pd.read_csv(
        path,
        low_memory=False
    )

    canonical_col = next(
        (
            c for c in canonical_fields
            if c in df.columns
        ),
        None
    )

    if canonical_col is None:
        return

    for alias_col in alias_fields:

        if alias_col not in df.columns:
            continue

        subset = df[
            [
                canonical_col,
                alias_col
            ]
        ].copy()

        subset.columns = [
            "Canonical_Species",
            "Candidate_Name"
        ]

        subset["Canonical_Species"] = (
            subset["Canonical_Species"]
            .apply(clean_name)
        )

        subset["Candidate_Name"] = (
            subset["Candidate_Name"]
            .apply(clean_name)
        )

        subset = subset[
            subset[
                "Canonical_Species"
            ].isin(
                set(
                    corrected[
                        "Canonical_Species"
                    ]
                )
            )
        ]

        subset = subset[
            subset[
                "Candidate_Name"
            ].apply(
                is_species_level
            )
        ]

        for _, row in subset.iterrows():

            validated_rows.append({
                "Canonical_Species":
                    row["Canonical_Species"],

                "Candidate_Name":
                    row["Candidate_Name"],

                "Evidence_Source":
                    path.name,

                "Evidence_Field":
                    alias_col,

                "Evidence_Class":
                    source_class,
            })


# -------------------------------------------------------------------------
# Explicitly reviewed Seed Mass alias register
# -------------------------------------------------------------------------

add_alias_candidates(
    VALIDATED_ALIAS_FILE,
    [
        "Canonical_Species",
        "Omo_Canonical_Species",
    ],
    [
        "Validated_Species_Alias",
        "Validated_Name",
        "Accepted_Name",
        "Alternative_Query_Name",
    ],
    "VALIDATED_TAXONOMIC_ALIAS"
)


# -------------------------------------------------------------------------
# BIEN alias-requery status
# -------------------------------------------------------------------------

add_alias_candidates(
    BIEN_ALIAS_STATUS_FILE,
    [
        "Canonical_Species",
        "Omo_Canonical_Species",
    ],
    [
        "Validated_BIEN_Query_Name",
    ],
    "VALIDATED_BIEN_QUERY_NAME"
)


# -------------------------------------------------------------------------
# TRY controlled gap candidates
# -------------------------------------------------------------------------

add_alias_candidates(
    TRY_GAP_CANDIDATES_FILE,
    [
        "Canonical_Species",
    ],
    [
        "Candidate_Validated_Name",
    ],
    "VALIDATED_TRY_GAP_NAME"
)


validated_aliases = pd.DataFrame(
    validated_rows
)


if len(validated_aliases) == 0:

    validated_aliases = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Candidate_Name",
            "Evidence_Source",
            "Evidence_Field",
            "Evidence_Class",
        ]
    )

else:

    validated_aliases = (
        validated_aliases
        .drop_duplicates()
        .reset_index(drop=True)
    )


print("\nLegitimate species-level alias evidence")
print("-" * 100)

print(
    f"Evidence rows       : "
    f"{len(validated_aliases)}"
)

print(
    f"Canonical taxa      : "
    f"{validated_aliases['Canonical_Species'].nunique()}"
)


# =============================================================================
# 6. EVIDENCE PRIORITY
# =============================================================================

priority = {
    "VALIDATED_TAXONOMIC_ALIAS": 100,
    "VALIDATED_BIEN_QUERY_NAME": 90,
    "VALIDATED_TRY_GAP_NAME": 85,
}


if len(validated_aliases) > 0:

    validated_aliases[
        "Evidence_Priority"
    ] = (
        validated_aliases[
            "Evidence_Class"
        ]
        .map(priority)
        .fillna(0)
    )


# =============================================================================
# 7. SELECT BEST GENUINE SPECIES-LEVEL ALIAS
# =============================================================================

selected_alias_rows = []


for canonical in corrected[
    "Canonical_Species"
]:

    subset = validated_aliases[
        validated_aliases[
            "Canonical_Species"
        ].eq(
            canonical
        )
    ].copy()

    if subset.empty:
        continue

    subset = subset.sort_values(
        [
            "Evidence_Priority",
            "Candidate_Name",
        ],
        ascending=[
            False,
            True,
        ]
    )

    best = subset.iloc[0]

    selected_alias_rows.append({
        "Canonical_Species":
            canonical,

        "Selected_Alias":
            best[
                "Candidate_Name"
            ],

        "Selected_Alias_Source":
            best[
                "Evidence_Source"
            ],

        "Selected_Alias_Field":
            best[
                "Evidence_Field"
            ],

        "Selected_Alias_Class":
            best[
                "Evidence_Class"
            ],

        "N_Valid_Species_Aliases":
            subset[
                "Candidate_Name"
            ].nunique(),
    })


selected_aliases = pd.DataFrame(
    selected_alias_rows
)


if len(selected_aliases) > 0:

    corrected = corrected.merge(
        selected_aliases,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )

else:

    corrected[
        "Selected_Alias"
    ] = pd.NA

    corrected[
        "Selected_Alias_Source"
    ] = pd.NA

    corrected[
        "Selected_Alias_Field"
    ] = pd.NA

    corrected[
        "Selected_Alias_Class"
    ] = pd.NA

    corrected[
        "N_Valid_Species_Aliases"
    ] = 0


# =============================================================================
# 8. APPLY SPECIES-LEVEL ALIAS ONLY WHEN VALID
# =============================================================================

alias_available = (
    corrected[
        "Selected_Alias"
    ]
    .notna()
)


corrected.loc[
    alias_available,
    "Corrected_Search_Name"
] = corrected.loc[
    alias_available,
    "Selected_Alias"
]


corrected.loc[
    alias_available,
    "Search_Name_Source"
] = corrected.loc[
    alias_available,
    "Selected_Alias_Source"
]


corrected.loc[
    alias_available,
    "Taxonomic_Resolution_Class"
] = (
    "PREVIOUSLY_VALIDATED_SPECIES_LEVEL_ALIAS"
)


# =============================================================================
# 9. FINAL SPECIES-LEVEL STRUCTURAL CHECK
# =============================================================================

corrected[
    "Search_Name_Is_Species_Level"
] = corrected[
    "Corrected_Search_Name"
].apply(
    is_species_level
)


corrected[
    "Corrected_Search_Genus"
] = (
    corrected[
        "Corrected_Search_Name"
    ]
    .astype("string")
    .str.split()
    .str[0]
)


# Any canonical input that itself is malformed remains a review case.
corrected[
    "Taxonomic_Review_Flag"
] = np.where(
    corrected[
        "Search_Name_Is_Species_Level"
    ],
    "NO_REVIEW_REQUIRED",
    "SPECIES_LEVEL_NAME_REVIEW_REQUIRED"
)


# Multiple competing genuine species-level aliases deserve explicit review.
corrected.loc[
    pd.to_numeric(
        corrected[
            "N_Valid_Species_Aliases"
        ],
        errors="coerce"
    )
    .fillna(0)
    .gt(1),
    "Taxonomic_Review_Flag"
] = "MULTIPLE_VALIDATED_SPECIES_NAMES_REVIEW"


# =============================================================================
# 10. IDENTIFY SHARED SEARCH TARGETS
# =============================================================================

search_counts = (
    corrected[
        "Corrected_Search_Name"
    ]
    .value_counts()
    .rename_axis(
        "Corrected_Search_Name"
    )
    .reset_index(
        name="N_Omo_Canonical_Taxa"
    )
)


corrected = corrected.merge(
    search_counts,
    on="Corrected_Search_Name",
    how="left",
    validate="many_to_one"
)


corrected[
    "Shared_Search_Target"
] = (
    corrected[
        "N_Omo_Canonical_Taxa"
    ]
    .gt(1)
)


shared = corrected[
    corrected[
        "Shared_Search_Target"
    ]
].copy()


if len(shared) > 0:

    shared_summary = (
        shared
        .groupby(
            "Corrected_Search_Name"
        )
        .agg(
            N_Omo_Canonical_Taxa=(
                "Canonical_Species",
                "size"
            ),

            Omo_Canonical_Taxa=(
                "Canonical_Species",
                lambda x: " | ".join(
                    sorted(
                        x.astype(str)
                    )
                )
            ),
        )
        .reset_index()
    )

else:

    shared_summary = pd.DataFrame(
        columns=[
            "Corrected_Search_Name",
            "N_Omo_Canonical_Taxa",
            "Omo_Canonical_Taxa",
        ]
    )


print("\nCorrected shared biological search targets")
print("-" * 100)

print(
    f"N shared targets : "
    f"{len(shared_summary)}"
)

if len(shared_summary) > 0:
    display(shared_summary)
else:
    print("NONE")


# =============================================================================
# 11. BUILD CORRECTED UNIQUE SEARCH MANIFEST
# =============================================================================

manifest = (
    corrected[
        [
            "Corrected_Search_Name",
            "Corrected_Search_Genus",
            "Taxonomic_Review_Flag",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Corrected_Search_Name"
        ]
    )
    .sort_values(
        "Corrected_Search_Name"
    )
    .reset_index(drop=True)
)


manifest[
    "Search_Target_ID"
] = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(
        1,
        len(manifest) + 1
    )
]


manifest[
    "Requested_Trait"
] = "seed mass"

manifest[
    "Standardized_Unit"
] = "mg"

manifest[
    "Preferred_Measurement_Entity"
] = "individual seed dry mass"


manifest[
    "Search_Status"
] = np.where(
    manifest[
        "Taxonomic_Review_Flag"
    ].eq(
        "NO_REVIEW_REQUIRED"
    ),
    "READY_FOR_EXTERNAL_SEARCH",
    "TAXONOMIC_REVIEW_BEFORE_SEARCH"
)


# =============================================================================
# 12. ATTACH IDs TO CROSSWALK
# =============================================================================

corrected = corrected.merge(
    manifest[
        [
            "Corrected_Search_Name",
            "Search_Target_ID",
            "Search_Status",
        ]
    ],
    on="Corrected_Search_Name",
    how="left",
    validate="many_to_one"
)


# =============================================================================
# 13. REVIEW TABLE
# =============================================================================

review = corrected[
    ~corrected[
        "Taxonomic_Review_Flag"
    ].eq(
        "NO_REVIEW_REQUIRED"
    )
].copy()


# =============================================================================
# 14. DISPLAY CORRECTED CROSSWALK
# =============================================================================

print("\nCorrected species-level rescue crosswalk")
print("-" * 100)

display(
    corrected[
        [
            "Canonical_Species",
            "Corrected_Search_Name",
            "Taxonomic_Resolution_Class",
            "Selected_Alias_Source",
            "Selected_Alias_Field",
            "N_Valid_Species_Aliases",
            "N_Omo_Canonical_Taxa",
            "Shared_Search_Target",
            "Taxonomic_Review_Flag",
            "Search_Target_ID",
            "Search_Status",
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 15. DISPLAY CORRECTED SEARCH MANIFEST
# =============================================================================

print("\nCorrected unique Seed Mass search manifest")
print("-" * 100)

print(
    f"Canonical rescue taxa          : "
    f"{len(corrected)}"
)

print(
    f"Unique species search targets  : "
    f"{len(manifest)}"
)

print(
    f"Ready for external search      : "
    f"{manifest['Search_Status'].eq('READY_FOR_EXTERNAL_SEARCH').sum()}"
)

print(
    f"Require taxonomic review       : "
    f"{manifest['Search_Status'].eq('TAXONOMIC_REVIEW_BEFORE_SEARCH').sum()}"
)

display(manifest)


# =============================================================================
# 16. SUMMARY
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Canonical rescue taxa",
        "Canonical names retained for search",
        "Previously validated species aliases used",
        "Unique species-level search targets",
        "Shared biological search targets",
        "Canonical taxa requiring taxonomic review",
        "Unique search targets ready for external search",
        "Unique search targets requiring taxonomic review",
    ],

    "Value": [
        len(corrected),

        int(
            corrected[
                "Taxonomic_Resolution_Class"
            ].eq(
                "CANONICAL_SPECIES_RETAINED"
            ).sum()
        ),

        int(
            corrected[
                "Taxonomic_Resolution_Class"
            ].eq(
                "PREVIOUSLY_VALIDATED_SPECIES_LEVEL_ALIAS"
            ).sum()
        ),

        len(manifest),

        len(shared_summary),

        len(review),

        int(
            manifest[
                "Search_Status"
            ].eq(
                "READY_FOR_EXTERNAL_SEARCH"
            ).sum()
        ),

        int(
            manifest[
                "Search_Status"
            ].eq(
                "TAXONOMIC_REVIEW_BEFORE_SEARCH"
            ).sum()
        ),
    ],
})


print("\nCorrected Section 9.4A audit summary")
print("-" * 100)

display(summary)


# =============================================================================
# 17. STRONG VALIDATION
# =============================================================================

assert len(corrected) == 50

assert corrected[
    "Canonical_Species"
].is_unique

# No literal placeholders may survive.
assert not corrected[
    "Corrected_Search_Name"
].astype(str).str.lower().isin(
    INVALID_TAXON_TOKENS
).any()

# Every final search name must remain species level unless explicitly reviewed.
assert (
    corrected.loc[
        corrected[
            "Taxonomic_Review_Flag"
        ].eq(
            "NO_REVIEW_REQUIRED"
        ),
        "Search_Name_Is_Species_Level"
    ]
    .all()
)

# Forbidden provenance must never appear.
assert not (
    corrected[
        "Selected_Alias_Source"
    ]
    .fillna("")
    .str.contains(
        "congeneric_pseudo_aliases_excluded",
        case=False,
        regex=False
    )
    .any()
), (
    "Excluded congeneric pseudo-alias evidence entered the corrected crosswalk."
)

assert manifest[
    "Corrected_Search_Name"
].is_unique

assert manifest[
    "Search_Target_ID"
].is_unique


# =============================================================================
# 18. EXPORT
# =============================================================================

corrected.to_csv(
    OUT_CORRECTED_CROSSWALK,
    index=False
)

manifest.to_csv(
    OUT_CORRECTED_MANIFEST,
    index=False
)

shared_summary.to_csv(
    OUT_SHARED_TARGETS,
    index=False
)

review.to_csv(
    OUT_REVIEW,
    index=False
)

summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 19. REPORT EXPORTS
# =============================================================================

print("\nCorrected Section 9.4A exports")
print("-" * 100)

print(
    f"Corrected crosswalk : "
    f"{OUT_CORRECTED_CROSSWALK}"
)

print(
    f"Corrected manifest  : "
    f"{OUT_CORRECTED_MANIFEST}"
)

print(
    f"Shared targets      : "
    f"{OUT_SHARED_TARGETS}"
)

print(
    f"Review cases        : "
    f"{OUT_REVIEW}"
)

print(
    f"Audit summary       : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 20. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Corrected residual species-level taxonomic "
    "search-target preparation completed successfully."
)

print(
    "Genus-only strings, literal GENUS/SPECIES placeholders, "
    "and excluded congeneric pseudo-aliases are no longer "
    "permitted as species-level rescue names."
)

print(
    "No Seed Mass values have been assigned."
)

print(
    "No external search has yet been performed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4A-R2 — AUTHORITATIVE CORRECTION OF RESIDUAL
#                   TAXONOMIC SEARCH TARGETS
# =============================================================================
#
# Purpose
# -------
# Correct a small number of biologically inappropriate species collapses that
# survived the earlier project taxonomic-alias register.
#
# This section:
#
#   1. starts from the corrected 50-taxon Section 9.4A-R crosswalk;
#   2. preserves all sound orthographic / accepted-name corrections;
#   3. reverses false congeneric/species collapses;
#   4. applies a small explicit authoritative correction layer;
#   5. rebuilds the canonical-to-search-target crosswalk;
#   6. regenerates the unique species-level external-search manifest;
#   7. reports shared biological search targets;
#   8. identifies any remaining taxonomic review cases.
#
# IMPORTANT
# ---------
# - No Seed Mass value is assigned here.
# - No external trait evidence is acquired here.
# - This section changes search names only.
# - Omo canonical identifiers remain unchanged.
#
# Authoritative corrections applied
# ---------------------------------
# Diospyros dendo             -> Diospyros dendo
# Diospyrous dendo            -> Diospyros dendo
# Diospyros soubreana         -> Diospyros soubreana
# Sterculia oblonga           -> Sterculia oblonga
# Sterculia tragacantha       -> Sterculia tragacantha
# Musa sapiensis              -> Musa × paradisiaca
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re

print("=" * 100)
print(
    "SECTION 9.4A-R2 — AUTHORITATIVE CORRECTION OF "
    "RESIDUAL TAXONOMIC SEARCH TARGETS"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

INPUT_CROSSWALK = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_crosswalk_corrected.csv"
)


# Final outputs
OUT_FINAL_CROSSWALK = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_crosswalk_final.csv"
)

OUT_FINAL_MANIFEST = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_unique_search_manifest_final.csv"
)

OUT_SHARED_TARGETS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_shared_search_targets_final.csv"
)

OUT_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_review_required_final.csv"
)

OUT_CORRECTION_REGISTER = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_authoritative_taxonomic_corrections.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_finalization_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT
# =============================================================================

assert INPUT_CROSSWALK.exists(), (
    f"Corrected Section 9.4A-R crosswalk not found:\n"
    f"{INPUT_CROSSWALK}"
)


crosswalk = pd.read_csv(
    INPUT_CROSSWALK,
    low_memory=False
)


print("\nInput crosswalk")
print("-" * 100)

print(
    f"Rows             : "
    f"{len(crosswalk)}"
)

print(
    f"Canonical taxa   : "
    f"{crosswalk['Canonical_Species'].nunique()}"
)


assert len(crosswalk) == 50, (
    f"Expected 50 residual canonical taxa, found {len(crosswalk)}."
)

assert (
    crosswalk[
        "Canonical_Species"
    ].nunique()
    == 50
)

assert crosswalk[
    "Canonical_Species"
].is_unique


# =============================================================================
# 3. HELPER FUNCTIONS
# =============================================================================

INVALID_TAXON_TOKENS = {
    "",
    "genus",
    "species",
    "sp",
    "sp.",
    "spp",
    "spp.",
    "na",
    "nan",
    "none",
    "unknown",
    "unresolved",
}


def clean_name(value):

    if pd.isna(value):
        return None

    value = re.sub(
        r"\s+",
        " ",
        str(value).strip()
    )

    if value.lower() in INVALID_TAXON_TOKENS:
        return None

    return value


def species_level_name(value):

    value = clean_name(value)

    if value is None:
        return False

    tokens = value.split()

    # Musa × paradisiaca contains 3 tokens and is valid.
    if len(tokens) < 2:
        return False

    if tokens[0].lower() in INVALID_TAXON_TOKENS:
        return False

    return True


# =============================================================================
# 4. NORMALIZE CURRENT SEARCH-NAME FIELD
# =============================================================================

assert "Corrected_Search_Name" in crosswalk.columns, (
    "Expected Corrected_Search_Name from Section 9.4A-R."
)


crosswalk[
    "Pre_R2_Search_Name"
] = (
    crosswalk[
        "Corrected_Search_Name"
    ]
    .apply(clean_name)
)


crosswalk[
    "Final_Search_Name"
] = crosswalk[
    "Pre_R2_Search_Name"
]


crosswalk[
    "R2_Correction_Applied"
] = False


crosswalk[
    "R2_Correction_Class"
] = "NO_R2_CORRECTION"


crosswalk[
    "R2_Correction_Rationale"
] = (
    "Section 9.4A-R species-level search target retained."
)


# =============================================================================
# 5. AUTHORITATIVE CORRECTION REGISTER
# =============================================================================
#
# These corrections are deliberately explicit rather than inferred.
#
# They repair cases where earlier alias tables incorrectly collapsed accepted
# species into congeners, or where a misspelled project name should map to the
# correct accepted species.
# =============================================================================

authoritative_corrections = pd.DataFrame(
    [
        {
            "Canonical_Species":
                "Diospyros dendo",

            "Final_Search_Name":
                "Diospyros dendo",

            "Correction_Class":
                "ACCEPTED_SPECIES_RETAINED",

            "Correction_Rationale":
                (
                    "Diospyros dendo is retained as its own species-level "
                    "search target and must not be collapsed into "
                    "Diospyros mespiliformis."
                ),
        },

        {
            "Canonical_Species":
                "Diospyrous dendo",

            "Final_Search_Name":
                "Diospyros dendo",

            "Correction_Class":
                "ORTHOGRAPHIC_CORRECTION",

            "Correction_Rationale":
                (
                    "Project spelling Diospyrous dendo is normalized to "
                    "Diospyros dendo."
                ),
        },

        {
            "Canonical_Species":
                "Diospyros soubreana",

            "Final_Search_Name":
                "Diospyros soubreana",

            "Correction_Class":
                "ACCEPTED_SPECIES_RETAINED",

            "Correction_Rationale":
                (
                    "Diospyros soubreana is retained as its own "
                    "species-level search target and must not be collapsed "
                    "into Diospyros mespiliformis."
                ),
        },

        {
            "Canonical_Species":
                "Sterculia oblonga",

            "Final_Search_Name":
                "Sterculia oblonga",

            "Correction_Class":
                "ACCEPTED_SPECIES_RETAINED",

            "Correction_Rationale":
                (
                    "Sterculia oblonga is retained as its own species-level "
                    "search target and must not be collapsed into "
                    "Sterculia rhinopetala."
                ),
        },

        {
            "Canonical_Species":
                "Sterculia tragacantha",

            "Final_Search_Name":
                "Sterculia tragacantha",

            "Correction_Class":
                "ACCEPTED_SPECIES_RETAINED",

            "Correction_Rationale":
                (
                    "Sterculia tragacantha is retained as its own "
                    "species-level search target and must not be collapsed "
                    "into Sterculia rhinopetala."
                ),
        },

        {
            "Canonical_Species":
                "Musa sapiensis",

            "Final_Search_Name":
                "Musa × paradisiaca",

            "Correction_Class":
                "VALIDATED_ACCEPTED_NAME",

            "Correction_Rationale":
                (
                    "Musa sapiensis is routed to Musa × paradisiaca as the "
                    "preferred validated species-level search name."
                ),
        },
    ]
)


print("\nAuthoritative correction register")
print("-" * 100)

display(
    authoritative_corrections
)


# =============================================================================
# 6. CONFIRM CORRECTION TAXA EXIST IN THE 50-TAXON RESIDUAL SET
# =============================================================================

correction_taxa = set(
    authoritative_corrections[
        "Canonical_Species"
    ]
)

crosswalk_taxa = set(
    crosswalk[
        "Canonical_Species"
    ]
)


missing_correction_taxa = (
    correction_taxa
    -
    crosswalk_taxa
)


assert not missing_correction_taxa, (
    "One or more authoritative correction taxa are absent from "
    "the 50-taxon rescue crosswalk:\n"
    f"{sorted(missing_correction_taxa)}"
)


# =============================================================================
# 7. APPLY AUTHORITATIVE CORRECTIONS
# =============================================================================

correction_lookup = (
    authoritative_corrections
    .set_index(
        "Canonical_Species"
    )
)


for canonical_name in correction_lookup.index:

    mask = (
        crosswalk[
            "Canonical_Species"
        ]
        .eq(
            canonical_name
        )
    )

    new_name = correction_lookup.loc[
        canonical_name,
        "Final_Search_Name"
    ]

    correction_class = correction_lookup.loc[
        canonical_name,
        "Correction_Class"
    ]

    rationale = correction_lookup.loc[
        canonical_name,
        "Correction_Rationale"
    ]


    crosswalk.loc[
        mask,
        "Final_Search_Name"
    ] = new_name

    crosswalk.loc[
        mask,
        "R2_Correction_Applied"
    ] = True

    crosswalk.loc[
        mask,
        "R2_Correction_Class"
    ] = correction_class

    crosswalk.loc[
        mask,
        "R2_Correction_Rationale"
    ] = rationale


# =============================================================================
# 8. FINAL STRUCTURAL TAXONOMIC CHECK
# =============================================================================

crosswalk[
    "Final_Search_Name"
] = (
    crosswalk[
        "Final_Search_Name"
    ]
    .apply(clean_name)
)


crosswalk[
    "Final_Search_Name_Is_Species_Level"
] = (
    crosswalk[
        "Final_Search_Name"
    ]
    .apply(
        species_level_name
    )
)


crosswalk[
    "Final_Search_Genus"
] = (
    crosswalk[
        "Final_Search_Name"
    ]
    .astype("string")
    .str.split()
    .str[0]
)


# =============================================================================
# 9. REBUILD FINAL TAXONOMIC REVIEW FLAG
# =============================================================================

crosswalk[
    "Final_Taxonomic_Review_Flag"
] = "NO_REVIEW_REQUIRED"


crosswalk.loc[
    ~crosswalk[
        "Final_Search_Name_Is_Species_Level"
    ],
    "Final_Taxonomic_Review_Flag"
] = (
    "SPECIES_LEVEL_NAME_REVIEW_REQUIRED"
)


# Preserve earlier multiple-valid-name warning only where still relevant.
if "Taxonomic_Review_Flag" in crosswalk.columns:

    previous_multiple_name_review = (
        crosswalk[
            "Taxonomic_Review_Flag"
        ]
        .astype("string")
        .eq(
            "MULTIPLE_VALIDATED_SPECIES_NAMES_REVIEW"
        )
    )

    # Musa is now explicitly resolved, so do not retain that old warning.
    previous_multiple_name_review = (
        previous_multiple_name_review
        &
        ~crosswalk[
            "Canonical_Species"
        ].eq(
            "Musa sapiensis"
        )
    )

    crosswalk.loc[
        previous_multiple_name_review,
        "Final_Taxonomic_Review_Flag"
    ] = (
        "MULTIPLE_VALIDATED_SPECIES_NAMES_REVIEW"
    )


# =============================================================================
# 10. IDENTIFY FINAL SHARED BIOLOGICAL SEARCH TARGETS
# =============================================================================
#
# Section 9.4A-R already contained N_Omo_Canonical_Taxa based on the earlier
# provisional search targets.
#
# Section 9.4A-R2 changes several search names, so that old count must NOT be
# reused. Remove any inherited shared-target count/status fields and recompute
# them from Final_Search_Name.
# =============================================================================

preexisting_shared_target_fields = [
    "N_Omo_Canonical_Taxa",
    "Shared_Search_Target",
    "Final_Shared_Search_Target",
]

fields_to_drop = [
    col
    for col in preexisting_shared_target_fields
    if col in crosswalk.columns
]

if fields_to_drop:

    print("\nRemoving inherited shared-target fields before R2 recount")
    print("-" * 100)

    for col in fields_to_drop:
        print(f" - {col}")

    crosswalk = crosswalk.drop(
        columns=fields_to_drop
    )

else:

    print(
        "\nNo inherited shared-target fields required removal."
    )


# -----------------------------------------------------------------------------
# Recalculate the number of Omo canonical taxa represented by each FINAL
# biological search target.
# -----------------------------------------------------------------------------

target_counts = (
    crosswalk[
        "Final_Search_Name"
    ]
    .value_counts()
    .rename_axis(
        "Final_Search_Name"
    )
    .reset_index(
        name="N_Omo_Canonical_Taxa"
    )
)


crosswalk = crosswalk.merge(
    target_counts,
    on="Final_Search_Name",
    how="left",
    validate="many_to_one"
)


# -----------------------------------------------------------------------------
# Validate that the merge generated the intended field without suffixes.
# -----------------------------------------------------------------------------

assert "N_Omo_Canonical_Taxa" in crosswalk.columns, (
    "Final shared-target count field was not created."
)

unexpected_count_suffixes = [
    col
    for col in crosswalk.columns
    if col.startswith("N_Omo_Canonical_Taxa_")
]

assert not unexpected_count_suffixes, (
    "Unexpected shared-target merge suffixes remain:\n"
    f"{unexpected_count_suffixes}"
)


crosswalk[
    "Final_Shared_Search_Target"
] = (
    crosswalk[
        "N_Omo_Canonical_Taxa"
    ]
    .gt(1)
)


# -----------------------------------------------------------------------------
# Extract canonical taxa sharing the same final biological search target.
# -----------------------------------------------------------------------------

shared_rows = crosswalk[
    crosswalk[
        "Final_Shared_Search_Target"
    ]
].copy()


if len(shared_rows) > 0:

    shared_summary = (
        shared_rows
        .groupby(
            "Final_Search_Name",
            dropna=False
        )
        .agg(
            N_Omo_Canonical_Taxa=(
                "Canonical_Species",
                "size"
            ),

            Omo_Canonical_Taxa=(
                "Canonical_Species",
                lambda x: " | ".join(
                    sorted(
                        set(
                            x.astype(str)
                        )
                    )
                )
            ),
        )
        .reset_index()
    )

else:

    shared_summary = pd.DataFrame(
        columns=[
            "Final_Search_Name",
            "N_Omo_Canonical_Taxa",
            "Omo_Canonical_Taxa",
        ]
    )


print("\nFinal shared biological search targets")
print("-" * 100)

print(
    f"N shared targets : "
    f"{len(shared_summary)}"
)

if len(shared_summary) > 0:

    display(
        shared_summary
    )

else:

    print("NONE")

# =============================================================================
# 11. BUILD FINAL UNIQUE SPECIES SEARCH MANIFEST
# =============================================================================

manifest = (
    crosswalk[
        [
            "Final_Search_Name",
            "Final_Search_Genus",
            "Final_Taxonomic_Review_Flag",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Final_Search_Name"
        ]
    )
    .sort_values(
        "Final_Search_Name"
    )
    .reset_index(drop=True)
)


manifest[
    "Search_Target_ID"
] = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(
        1,
        len(manifest) + 1
    )
]


manifest[
    "Requested_Trait"
] = "seed mass"


manifest[
    "Standardized_Unit"
] = "mg"


manifest[
    "Preferred_Measurement_Entity"
] = "individual seed dry mass"


manifest[
    "Search_Status"
] = np.where(
    manifest[
        "Final_Taxonomic_Review_Flag"
    ]
    .eq(
        "NO_REVIEW_REQUIRED"
    ),
    "READY_FOR_EXTERNAL_SEARCH",
    "TAXONOMIC_REVIEW_BEFORE_SEARCH"
)


# =============================================================================
# 12. ATTACH FINAL SEARCH TARGET IDS BACK TO CANONICAL CROSSWALK
# =============================================================================
#
# Section 9.4A-R already supplied provisional Search_Target_ID and
# Search_Status fields.
#
# Because Section 9.4A-R2 changes the biological search targets, those
# inherited identifiers/statuses are obsolete and must be removed before
# attaching the newly generated FINAL search manifest.
# =============================================================================

inherited_search_fields = [
    "Search_Target_ID",
    "Search_Status",
]

fields_to_drop = [
    col
    for col in inherited_search_fields
    if col in crosswalk.columns
]

if fields_to_drop:

    print("\nRemoving inherited search-target fields before final R2 assignment")
    print("-" * 100)

    for col in fields_to_drop:
        print(f" - {col}")

    crosswalk = crosswalk.drop(
        columns=fields_to_drop
    )

else:

    print(
        "\nNo inherited search-target fields required removal."
    )


# -----------------------------------------------------------------------------
# Also guard against suffix columns left by a previous partial/rerun attempt.
# -----------------------------------------------------------------------------

stale_suffix_fields = [
    col
    for col in crosswalk.columns
    if (
        col.startswith("Search_Target_ID_")
        or
        col.startswith("Search_Status_")
    )
]

if stale_suffix_fields:

    print("\nRemoving stale search-target suffix fields")
    print("-" * 100)

    for col in stale_suffix_fields:
        print(f" - {col}")

    crosswalk = crosswalk.drop(
        columns=stale_suffix_fields
    )


# -----------------------------------------------------------------------------
# Attach the FINAL R2 search-target ID and status.
# -----------------------------------------------------------------------------

crosswalk = crosswalk.merge(
    manifest[
        [
            "Final_Search_Name",
            "Search_Target_ID",
            "Search_Status",
        ]
    ],
    on="Final_Search_Name",
    how="left",
    validate="many_to_one"
)


# -----------------------------------------------------------------------------
# Strong validation immediately after merge.
# -----------------------------------------------------------------------------

assert "Search_Target_ID" in crosswalk.columns, (
    "Final Search_Target_ID was not created after manifest merge."
)

assert "Search_Status" in crosswalk.columns, (
    "Final Search_Status was not created after manifest merge."
)


unexpected_search_suffixes = [
    col
    for col in crosswalk.columns
    if (
        col.startswith("Search_Target_ID_")
        or
        col.startswith("Search_Status_")
    )
]

assert not unexpected_search_suffixes, (
    "Unexpected search-target merge suffixes remain:\n"
    f"{unexpected_search_suffixes}"
)


assert crosswalk[
    "Search_Target_ID"
].notna().all(), (
    "One or more canonical taxa failed to receive a final Search_Target_ID."
)

assert crosswalk[
    "Search_Status"
].notna().all(), (
    "One or more canonical taxa failed to receive a final Search_Status."
)


print("\nFinal R2 search-target assignment")
print("-" * 100)

print(
    f"Canonical taxa assigned IDs : "
    f"{crosswalk['Search_Target_ID'].notna().sum()} / {len(crosswalk)}"
)

print(
    f"Unique final target IDs     : "
    f"{crosswalk['Search_Target_ID'].nunique()}"
)

# =============================================================================
# 13. FINAL REVIEW REGISTER
# =============================================================================

review = crosswalk[
    ~crosswalk[
        "Final_Taxonomic_Review_Flag"
    ]
    .eq(
        "NO_REVIEW_REQUIRED"
    )
].copy()


print("\nFinal taxonomic review cases")
print("-" * 100)

print(
    f"N canonical taxa : "
    f"{len(review)}"
)


if len(review) > 0:

    display(
        review[
            [
                col
                for col in [
                    "Canonical_Species",
                    "Pre_R2_Search_Name",
                    "Final_Search_Name",
                    "R2_Correction_Applied",
                    "R2_Correction_Class",
                    "Final_Taxonomic_Review_Flag",
                ]
                if col in review.columns
            ]
        ]
        .sort_values(
            "Canonical_Species"
        )
        .reset_index(drop=True)
    )

else:

    print("NONE")


# =============================================================================
# 14. DISPLAY AUTHORITATIVELY CORRECTED TAXA
# =============================================================================

corrected_rows = crosswalk[
    crosswalk[
        "R2_Correction_Applied"
    ]
].copy()


print("\nAuthoritatively corrected search targets")
print("-" * 100)

display(
    corrected_rows[
        [
            "Canonical_Species",
            "Pre_R2_Search_Name",
            "Final_Search_Name",
            "R2_Correction_Class",
            "R2_Correction_Rationale",
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 15. DISPLAY FINAL CANONICAL-TO-SEARCH CROSSWALK
# =============================================================================

print("\nFinal canonical-to-search-target crosswalk")
print("-" * 100)

display(
    crosswalk[
        [
            col
            for col in [
                "Canonical_Species",
                "Pre_R2_Search_Name",
                "Final_Search_Name",
                "Final_Search_Genus",
                "R2_Correction_Applied",
                "R2_Correction_Class",
                "N_Omo_Canonical_Taxa",
                "Final_Shared_Search_Target",
                "Final_Taxonomic_Review_Flag",
                "Search_Target_ID",
                "Search_Status",
            ]
            if col in crosswalk.columns
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 16. DISPLAY FINAL UNIQUE SEARCH MANIFEST
# =============================================================================

print("\nFinal unique species-level Seed Mass search manifest")
print("-" * 100)

print(
    f"Canonical rescue taxa       : "
    f"{len(crosswalk)}"
)

print(
    f"Unique biological targets   : "
    f"{len(manifest)}"
)

print(
    f"Ready for external search   : "
    f"{manifest['Search_Status'].eq('READY_FOR_EXTERNAL_SEARCH').sum()}"
)

print(
    f"Require taxonomic review    : "
    f"{manifest['Search_Status'].eq('TAXONOMIC_REVIEW_BEFORE_SEARCH').sum()}"
)


display(
    manifest
)


# =============================================================================
# 17. VALIDATE SPECIFIC AUTHORITATIVE CORRECTIONS
# =============================================================================

expected_corrections = {
    "Diospyros dendo":
        "Diospyros dendo",

    "Diospyrous dendo":
        "Diospyros dendo",

    "Diospyros soubreana":
        "Diospyros soubreana",

    "Sterculia oblonga":
        "Sterculia oblonga",

    "Sterculia tragacantha":
        "Sterculia tragacantha",

    "Musa sapiensis":
        "Musa × paradisiaca",
}


for canonical, expected_search in expected_corrections.items():

    observed = (
        crosswalk.loc[
            crosswalk[
                "Canonical_Species"
            ].eq(
                canonical
            ),
            "Final_Search_Name"
        ]
        .iloc[0]
    )

    assert observed == expected_search, (
        f"Unexpected final search target for {canonical}:\n"
        f"Expected : {expected_search}\n"
        f"Observed : {observed}"
    )


# =============================================================================
# 18. FINAL SUMMARY
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Canonical species-level rescue taxa",
        "Authoritative R2 corrections applied",
        "Unique final biological search targets",
        "Shared final biological search targets",
        "Canonical taxa requiring final taxonomic review",
        "Unique targets ready for external search",
        "Unique targets requiring taxonomic review",
    ],

    "Value": [
        len(crosswalk),

        int(
            crosswalk[
                "R2_Correction_Applied"
            ].sum()
        ),

        len(manifest),

        len(shared_summary),

        len(review),

        int(
            manifest[
                "Search_Status"
            ]
            .eq(
                "READY_FOR_EXTERNAL_SEARCH"
            )
            .sum()
        ),

        int(
            manifest[
                "Search_Status"
            ]
            .eq(
                "TAXONOMIC_REVIEW_BEFORE_SEARCH"
            )
            .sum()
        ),
    ],
})


print("\nSection 9.4A-R2 finalization summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 19. STRONG VALIDATION
# =============================================================================

assert len(crosswalk) == 50

assert crosswalk[
    "Canonical_Species"
].is_unique


assert crosswalk[
    "Final_Search_Name"
].notna().all()


assert not crosswalk[
    "Final_Search_Name"
].astype(str).str.lower().isin(
    INVALID_TAXON_TOKENS
).any()


# No unreviewed genus-only targets are allowed.
assert (
    crosswalk.loc[
        crosswalk[
            "Final_Taxonomic_Review_Flag"
        ].eq(
            "NO_REVIEW_REQUIRED"
        ),
        "Final_Search_Name_Is_Species_Level"
    ]
    .all()
)


# No false collapse to Diospyros mespiliformis for these accepted taxa.
for taxon in [
    "Diospyros dendo",
    "Diospyros soubreana",
]:

    observed = crosswalk.loc[
        crosswalk[
            "Canonical_Species"
        ].eq(
            taxon
        ),
        "Final_Search_Name"
    ].iloc[0]

    assert observed != "Diospyros mespiliformis"


# No false collapse to Sterculia rhinopetala.
for taxon in [
    "Sterculia oblonga",
    "Sterculia tragacantha",
]:

    observed = crosswalk.loc[
        crosswalk[
            "Canonical_Species"
        ].eq(
            taxon
        ),
        "Final_Search_Name"
    ].iloc[0]

    assert observed != "Sterculia rhinopetala"


assert manifest[
    "Final_Search_Name"
].is_unique


assert manifest[
    "Search_Target_ID"
].is_unique


assert crosswalk[
    "Search_Target_ID"
].notna().all()


# =============================================================================
# 20. EXPORT
# =============================================================================

crosswalk.to_csv(
    OUT_FINAL_CROSSWALK,
    index=False
)

manifest.to_csv(
    OUT_FINAL_MANIFEST,
    index=False
)

shared_summary.to_csv(
    OUT_SHARED_TARGETS,
    index=False
)

review.to_csv(
    OUT_REVIEW,
    index=False
)

authoritative_corrections.to_csv(
    OUT_CORRECTION_REGISTER,
    index=False
)

summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 21. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4A-R2 exports")
print("-" * 100)

print(
    f"Final taxonomic crosswalk : "
    f"{OUT_FINAL_CROSSWALK}"
)

print(
    f"Final search manifest      : "
    f"{OUT_FINAL_MANIFEST}"
)

print(
    f"Shared search targets      : "
    f"{OUT_SHARED_TARGETS}"
)

print(
    f"Remaining review cases     : "
    f"{OUT_REVIEW}"
)

print(
    f"Correction register        : "
    f"{OUT_CORRECTION_REGISTER}"
)

print(
    f"Finalization summary       : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 22. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Authoritative residual Seed Mass taxonomic "
    "search-target correction completed successfully."
)

print(
    f"Omo canonical rescue taxa        : "
    f"{len(crosswalk)}"
)

print(
    f"Unique final search targets       : "
    f"{len(manifest)}"
)

print(
    f"Remaining taxonomic review cases  : "
    f"{len(review)}"
)

print(
    "No Seed Mass values have been assigned."
)

print(
    "No external trait search has yet been performed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4B — TARGETED NON-TRY SEED MASS EVIDENCE ACQUISITION FRAMEWORK
# =============================================================================
#
# Purpose
# -------
# Prepare a controlled external-evidence acquisition framework for the
# 39 unique species-level Seed Mass rescue targets finalized in Section
# 9.4A-R2.
#
# This section:
#
#   1. loads the final 39-target taxonomic search manifest;
#   2. loads the 50-canonical-taxon crosswalk;
#   3. validates the search-target structure;
#   4. defines the external evidence-source hierarchy;
#   5. creates one acquisition register row per biological search target;
#   6. creates an empty measurement-level evidence ledger schema;
#   7. creates a source-search checklist;
#   8. preserves canonical-to-search-target multiplicity;
#   9. prepares standardized QC fields for future evidence entry;
#  10. assigns NO Seed Mass values.
#
# Evidence-source hierarchy
# -------------------------
# Preferred acquisition order:
#
#   1. Kew Seed Information Database / SID-derived authoritative records
#   2. Published primary literature with explicit seed-mass measurement
#   3. Reputable structured plant-trait database
#   4. Published secondary compilation with traceable source
#
# Measurement requirement
# -----------------------
# Preferred entity:
#
#     individual seed dry mass
#
# Measurements representing:
#
#     fruit mass
#     endocarp mass
#     stone mass
#     diaspore mass
#     propagule mass
#
# must NOT automatically be treated as standardized Seed Mass.
#
# IMPORTANT
# ---------
# This section performs NO web search.
# This section acquires NO measurements.
# This section assigns NO Seed Mass values.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4B — TARGETED NON-TRY SEED MASS "
    "EVIDENCE ACQUISITION FRAMEWORK"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


FINAL_MANIFEST_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_unique_search_manifest_final.csv"
)

FINAL_CROSSWALK_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_rescue_taxonomic_crosswalk_final.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_ACQUISITION_REGISTER = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)

OUT_SOURCE_CHECKLIST = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_source_search_checklist.csv"
)

OUT_EVIDENCE_LEDGER_TEMPLATE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_evidence_ledger_template.csv"
)

OUT_CANONICAL_TARGET_MAP = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)

OUT_SOURCE_HIERARCHY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_source_hierarchy.csv"
)

OUT_FRAMEWORK_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_framework_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

for path in [
    FINAL_MANIFEST_FILE,
    FINAL_CROSSWALK_FILE,
]:

    assert path.exists(), (
        f"Required Section 9.4A-R2 output not found:\n{path}"
    )


# =============================================================================
# 3. LOAD FINAL TAXONOMIC INPUTS
# =============================================================================

manifest = pd.read_csv(
    FINAL_MANIFEST_FILE,
    low_memory=False
)

crosswalk = pd.read_csv(
    FINAL_CROSSWALK_FILE,
    low_memory=False
)


print("\nInput tables")
print("-" * 100)

print(
    f"Final biological search targets : "
    f"{len(manifest)}"
)

print(
    f"Omo canonical rescue taxa       : "
    f"{len(crosswalk)}"
)


# =============================================================================
# 4. REQUIRED FIELD VALIDATION
# =============================================================================

required_manifest_fields = {
    "Final_Search_Name",
    "Final_Search_Genus",
    "Search_Target_ID",
    "Requested_Trait",
    "Standardized_Unit",
    "Preferred_Measurement_Entity",
    "Search_Status",
}

required_crosswalk_fields = {
    "Canonical_Species",
    "Final_Search_Name",
    "Search_Target_ID",
}


missing_manifest = (
    required_manifest_fields
    - set(manifest.columns)
)

missing_crosswalk = (
    required_crosswalk_fields
    - set(crosswalk.columns)
)


assert not missing_manifest, (
    "Final search manifest missing required fields:\n"
    f"{sorted(missing_manifest)}"
)

assert not missing_crosswalk, (
    "Final canonical crosswalk missing required fields:\n"
    f"{sorted(missing_crosswalk)}"
)


# =============================================================================
# 5. STRONG INPUT VALIDATION
# =============================================================================

assert len(manifest) == 39, (
    f"Expected 39 final biological targets, found {len(manifest)}."
)

assert manifest[
    "Final_Search_Name"
].is_unique

assert manifest[
    "Search_Target_ID"
].is_unique

assert len(crosswalk) == 50, (
    f"Expected 50 canonical rescue taxa, found {len(crosswalk)}."
)

assert crosswalk[
    "Canonical_Species"
].is_unique

assert crosswalk[
    "Search_Target_ID"
].notna().all()

assert (
    crosswalk[
        "Search_Target_ID"
    ]
    .isin(
        set(
            manifest[
                "Search_Target_ID"
            ]
        )
    )
    .all()
)

assert (
    manifest[
        "Search_Status"
    ]
    .eq(
        "READY_FOR_EXTERNAL_SEARCH"
    )
    .all()
), (
    "One or more final biological targets are not ready for external search."
)

assert (
    manifest[
        "Standardized_Unit"
    ]
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("mg")
    .all()
)


print("\nFinal search-manifest validation")
print("-" * 100)

print("Unique biological targets : 39")
print("Canonical rescue taxa     : 50")
print("Targets ready for search  : 39")
print("Standardized unit         : mg")


# =============================================================================
# 6. DEFINE EXTERNAL EVIDENCE SOURCE HIERARCHY
# =============================================================================

source_hierarchy = pd.DataFrame(
    [
        {
            "Source_Priority": 1,
            "Source_Class":
                "KEW_SID_OR_EQUIVALENT_AUTHORITATIVE_SEED_DATABASE",
            "Preferred": True,
            "Description":
                (
                    "Kew Seed Information Database or directly traceable "
                    "authoritative seed-mass records."
                ),
            "Acceptable_Evidence":
                (
                    "Species-specific seed mass or 1000-seed weight with "
                    "measurement entity clearly representing seed."
                ),
        },

        {
            "Source_Priority": 2,
            "Source_Class":
                "PRIMARY_PEER_REVIEWED_LITERATURE",
            "Preferred": True,
            "Description":
                (
                    "Peer-reviewed primary study reporting species-specific "
                    "seed mass."
                ),
            "Acceptable_Evidence":
                (
                    "Explicit measured seed mass or convertible 100/1000-seed "
                    "weight with clear units and species identity."
                ),
        },

        {
            "Source_Priority": 3,
            "Source_Class":
                "STRUCTURED_TRAIT_DATABASE",
            "Preferred": True,
            "Description":
                (
                    "Reputable structured plant-trait database not already "
                    "represented by the completed BIEN/TRY evidence streams."
                ),
            "Acceptable_Evidence":
                (
                    "Species-specific numeric Seed Mass with source provenance "
                    "and interpretable measurement entity."
                ),
        },

        {
            "Source_Priority": 4,
            "Source_Class":
                "TRACEABLE_SECONDARY_COMPILATION",
            "Preferred": False,
            "Description":
                (
                    "Published secondary compilation, flora, monograph or "
                    "technical source with traceable species-specific evidence."
                ),
            "Acceptable_Evidence":
                (
                    "Use only when higher-priority evidence is unavailable and "
                    "the measurement definition is sufficiently explicit."
                ),
        },
    ]
)


print("\nExternal Seed Mass source hierarchy")
print("-" * 100)

display(
    source_hierarchy
)


# =============================================================================
# 7. CREATE CANONICAL-TO-SEARCH-TARGET MAP
# =============================================================================

canonical_target_map = crosswalk[
    [
        col
        for col in [
            "Canonical_Species",
            "Final_Search_Name",
            "Final_Search_Genus",
            "Search_Target_ID",
            "N_Omo_Canonical_Taxa",
            "Final_Shared_Search_Target",
            "R2_Correction_Applied",
            "R2_Correction_Class",
            "Final_Taxonomic_Review_Flag",
        ]
        if col in crosswalk.columns
    ]
].copy()


canonical_target_map = (
    canonical_target_map
    .sort_values(
        [
            "Search_Target_ID",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


print("\nCanonical-to-search-target structure")
print("-" * 100)

print(
    f"Canonical taxa       : "
    f"{len(canonical_target_map)}"
)

print(
    f"Unique search targets: "
    f"{canonical_target_map['Search_Target_ID'].nunique()}"
)


# =============================================================================
# 8. COUNT CANONICAL TAXA PER SEARCH TARGET
# =============================================================================

canonical_multiplicity = (
    canonical_target_map
    .groupby(
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        dropna=False
    )
    .agg(
        N_Omo_Canonical_Taxa=(
            "Canonical_Species",
            "size"
        ),

        Omo_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    set(
                        x.astype(str)
                    )
                )
            )
        ),
    )
    .reset_index()
)


# =============================================================================
# 9. CREATE TARGET-LEVEL ACQUISITION REGISTER
# =============================================================================

acquisition_register = (
    manifest
    .merge(
        canonical_multiplicity,
        on=[
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        how="left",
        validate="one_to_one"
    )
    .copy()
)


acquisition_register[
    "Acquisition_Phase"
] = "SPECIES_LEVEL_NON_TRY_RESCUE"


acquisition_register[
    "Preferred_Source_Priority"
] = 1


acquisition_register[
    "Current_Acquisition_Status"
] = "NOT_YET_SEARCHED"


acquisition_register[
    "Evidence_Found"
] = False


acquisition_register[
    "N_Candidate_Evidence_Records"
] = 0


acquisition_register[
    "N_Accepted_Evidence_Records"
] = 0


acquisition_register[
    "N_Rejected_Evidence_Records"
] = 0


acquisition_register[
    "Best_Source_Class"
] = pd.NA


acquisition_register[
    "Final_Target_Resolution_Status"
] = "PENDING_EXTERNAL_EVIDENCE"


acquisition_register[
    "Notes"
] = pd.NA


acquisition_register = acquisition_register[
    [
        "Search_Target_ID",
        "Final_Search_Name",
        "Final_Search_Genus",
        "N_Omo_Canonical_Taxa",
        "Omo_Canonical_Taxa",
        "Requested_Trait",
        "Standardized_Unit",
        "Preferred_Measurement_Entity",
        "Acquisition_Phase",
        "Preferred_Source_Priority",
        "Current_Acquisition_Status",
        "Evidence_Found",
        "N_Candidate_Evidence_Records",
        "N_Accepted_Evidence_Records",
        "N_Rejected_Evidence_Records",
        "Best_Source_Class",
        "Final_Target_Resolution_Status",
        "Notes",
    ]
]


print("\nTarget-level external acquisition register")
print("-" * 100)

display(
    acquisition_register
)


# =============================================================================
# 10. CREATE SOURCE-BY-TARGET SEARCH CHECKLIST
# =============================================================================
#
# Each of the 39 biological targets receives one row per source class.
#
# This produces:
#
#     39 targets × 4 source classes = 156 search-checklist rows
#
# No search result is entered here.
# =============================================================================

source_checklist = (
    manifest[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Final_Search_Genus",
        ]
    ]
    .assign(
        _join_key=1
    )
    .merge(
        source_hierarchy.assign(
            _join_key=1
        ),
        on="_join_key",
        how="inner"
    )
    .drop(
        columns="_join_key"
    )
)


source_checklist[
    "Search_Attempt_Status"
] = "NOT_SEARCHED"


source_checklist[
    "Search_Date"
] = pd.NA


source_checklist[
    "Source_or_Database_Name"
] = pd.NA


source_checklist[
    "Search_Query_Used"
] = pd.NA


source_checklist[
    "Evidence_Located"
] = False


source_checklist[
    "N_Candidate_Records"
] = 0


source_checklist[
    "Search_Notes"
] = pd.NA


source_checklist = (
    source_checklist
    .sort_values(
        [
            "Search_Target_ID",
            "Source_Priority",
        ]
    )
    .reset_index(drop=True)
)


assert len(source_checklist) == (
    39 * len(source_hierarchy)
)


print("\nExternal source-search checklist")
print("-" * 100)

print(
    f"Search targets       : "
    f"{manifest['Search_Target_ID'].nunique()}"
)

print(
    f"Source classes       : "
    f"{len(source_hierarchy)}"
)

print(
    f"Checklist rows       : "
    f"{len(source_checklist)}"
)


# =============================================================================
# 11. CREATE EMPTY MEASUREMENT-LEVEL EVIDENCE LEDGER
# =============================================================================
#
# Each external measurement found later should occupy ONE ROW in this table.
#
# Critical distinction:
#
#     Search target = biological taxon being searched
#     Evidence row  = one reported numeric/qualitative source record
#
# Multiple measurements for one species are therefore preserved.
# =============================================================================

evidence_ledger_columns = [
    # -------------------------------------------------------------------------
    # Internal identification
    # -------------------------------------------------------------------------
    "Evidence_Record_ID",
    "Search_Target_ID",
    "Final_Search_Name",
    "Canonical_Species_Map",

    # -------------------------------------------------------------------------
    # Taxonomic provenance
    # -------------------------------------------------------------------------
    "Source_Reported_Taxon",
    "Taxonomic_Match_Status",
    "Taxonomic_Notes",

    # -------------------------------------------------------------------------
    # Trait identity
    # -------------------------------------------------------------------------
    "Trait_Code",
    "Reported_Trait_Name",
    "Measurement_Entity",
    "Measurement_Entity_Class",

    # -------------------------------------------------------------------------
    # Numeric evidence
    # -------------------------------------------------------------------------
    "Reported_Value",
    "Reported_Unit",
    "Reported_Seed_Count",
    "Converted_Individual_Seed_Mass_mg",
    "Conversion_Formula",
    "Conversion_Status",

    # -------------------------------------------------------------------------
    # Methodological evidence
    # -------------------------------------------------------------------------
    "Dry_Mass_Confirmed",
    "Measurement_Method",
    "Sample_Size",
    "Population_or_Location",
    "Study_Year",

    # -------------------------------------------------------------------------
    # Source provenance
    # -------------------------------------------------------------------------
    "Source_Priority",
    "Source_Class",
    "Source_Name",
    "Source_URL_or_DOI",
    "Citation",
    "Page_or_Table",
    "Access_Date",

    # -------------------------------------------------------------------------
    # QC
    # -------------------------------------------------------------------------
    "Evidence_QC_Status",
    "Evidence_QC_Flag",
    "QC_Rationale",

    # -------------------------------------------------------------------------
    # Final evidence disposition
    # -------------------------------------------------------------------------
    "Evidence_Action",
    "Eligible_For_Species_Aggregation",
]


evidence_ledger = pd.DataFrame(
    columns=evidence_ledger_columns
)


# =============================================================================
# 12. DEFINE CONTROLLED VOCABULARIES
# =============================================================================
#
# These are exported as documentation through the framework summary/structure.
# =============================================================================

measurement_entity_classes = [
    "TRUE_SEED",
    "SEED_COMPATIBLE",
    "1000_SEED_WEIGHT",
    "MULTI_SEED_WEIGHT",
    "DIASPORE_OR_DISPERSAL_UNIT",
    "ENDOCARP_OR_STONE",
    "FRUIT",
    "PROPAGULE",
    "AMBIGUOUS",
]


evidence_qc_statuses = [
    "PENDING_REVIEW",
    "ACCEPT",
    "HOLD",
    "REJECT",
]


evidence_actions = [
    "RETAIN_FOR_SPECIES_AGGREGATION",
    "RETAIN_IN_LEDGER_ONLY",
    "EXCLUDE_MEASUREMENT",
]


# =============================================================================
# 13. PREPARE EVIDENCE ENTRY EXAMPLE STRUCTURE
# =============================================================================
#
# No real evidence values are entered.
# This only documents how future records should be represented.
# =============================================================================

ledger_schema_description = pd.DataFrame(
    [
        {
            "Field":
                "Evidence_Record_ID",
            "Meaning":
                "Unique internal ID for each external evidence record.",
        },
        {
            "Field":
                "Search_Target_ID",
            "Meaning":
                "Final 9.4A-R2 biological search target identifier.",
        },
        {
            "Field":
                "Source_Reported_Taxon",
            "Meaning":
                "Taxon name exactly as reported by the external source.",
        },
        {
            "Field":
                "Measurement_Entity",
            "Meaning":
                "What was physically weighed: seed, 1000 seeds, fruit, stone, etc.",
        },
        {
            "Field":
                "Converted_Individual_Seed_Mass_mg",
            "Meaning":
                "Standardized individual seed mass after defensible conversion.",
        },
        {
            "Field":
                "Source_Class",
            "Meaning":
                "Controlled external evidence-source class.",
        },
        {
            "Field":
                "Evidence_QC_Status",
            "Meaning":
                "ACCEPT, HOLD, REJECT or PENDING_REVIEW.",
        },
        {
            "Field":
                "Eligible_For_Species_Aggregation",
            "Meaning":
                "Whether the measurement can contribute to a final species value.",
        },
    ]
)


# =============================================================================
# 14. FRAMEWORK SUMMARY
# =============================================================================

framework_summary = pd.DataFrame(
    {
        "Metric": [
            "Canonical species-level rescue taxa",
            "Unique biological search targets",
            "Targets ready for external search",
            "External source classes",
            "Source-search checklist rows",
            "Evidence ledger records currently entered",
            "Seed Mass values currently assigned",
        ],

        "Value": [
            len(crosswalk),
            len(manifest),

            int(
                manifest[
                    "Search_Status"
                ]
                .eq(
                    "READY_FOR_EXTERNAL_SEARCH"
                )
                .sum()
            ),

            len(source_hierarchy),

            len(source_checklist),

            len(evidence_ledger),

            0,
        ],
    }
)


print("\nSection 9.4B framework summary")
print("-" * 100)

display(
    framework_summary
)


# =============================================================================
# 15. STRONG VALIDATION
# =============================================================================

assert len(manifest) == 39

assert len(crosswalk) == 50

assert len(acquisition_register) == 39

assert acquisition_register[
    "Search_Target_ID"
].is_unique

assert (
    acquisition_register[
        "N_Omo_Canonical_Taxa"
    ]
    .sum()
    == 50
)

assert len(source_hierarchy) == 4

assert len(source_checklist) == 156

assert (
    source_checklist[
        "Search_Target_ID"
    ]
    .nunique()
    == 39
)

assert (
    source_checklist
    .groupby(
        "Search_Target_ID"
    )
    .size()
    .eq(4)
    .all()
)

assert evidence_ledger.empty

assert len(
    evidence_ledger.columns
) == len(
    evidence_ledger_columns
)

assert (
    acquisition_register[
        "Current_Acquisition_Status"
    ]
    .eq(
        "NOT_YET_SEARCHED"
    )
    .all()
)

assert (
    acquisition_register[
        "Evidence_Found"
    ]
    .eq(False)
    .all()
)

assert (
    acquisition_register[
        "Final_Target_Resolution_Status"
    ]
    .eq(
        "PENDING_EXTERNAL_EVIDENCE"
    )
    .all()
)


print("\nFramework validation")
print("-" * 100)

print("Canonical rescue taxa        : 50")
print("Unique biological targets    : 39")
print("Source hierarchy levels      : 4")
print("Search-checklist rows        : 156")
print("Evidence records entered     : 0")
print("Seed Mass values assigned    : 0")


# =============================================================================
# 16. EXPORT
# =============================================================================

acquisition_register.to_csv(
    OUT_ACQUISITION_REGISTER,
    index=False
)

source_checklist.to_csv(
    OUT_SOURCE_CHECKLIST,
    index=False
)

evidence_ledger.to_csv(
    OUT_EVIDENCE_LEDGER_TEMPLATE,
    index=False
)

canonical_target_map.to_csv(
    OUT_CANONICAL_TARGET_MAP,
    index=False
)

source_hierarchy.to_csv(
    OUT_SOURCE_HIERARCHY,
    index=False
)

framework_summary.to_csv(
    OUT_FRAMEWORK_SUMMARY,
    index=False
)


# =============================================================================
# 17. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4B exports")
print("-" * 100)

print(
    f"Acquisition register      : "
    f"{OUT_ACQUISITION_REGISTER}"
)

print(
    f"Source-search checklist   : "
    f"{OUT_SOURCE_CHECKLIST}"
)

print(
    f"Evidence ledger template  : "
    f"{OUT_EVIDENCE_LEDGER_TEMPLATE}"
)

print(
    f"Canonical-target map      : "
    f"{OUT_CANONICAL_TARGET_MAP}"
)

print(
    f"Source hierarchy          : "
    f"{OUT_SOURCE_HIERARCHY}"
)

print(
    f"Framework summary         : "
    f"{OUT_FRAMEWORK_SUMMARY}"
)


# =============================================================================
# 18. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Targeted non-TRY Seed Mass external-evidence "
    "acquisition framework prepared successfully."
)

print(
    f"Omo canonical rescue taxa        : "
    f"{len(crosswalk)}"
)

print(
    f"Unique biological search targets : "
    f"{len(manifest)}"
)

print(
    f"Source-search tasks prepared      : "
    f"{len(source_checklist)}"
)

print(
    "No external evidence has yet been entered."
)

print(
    "No Seed Mass values have been assigned."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4C — EXTERNAL SEED MASS EVIDENCE ACQUISITION: BATCH 1
# =============================================================================
#
# Targets
# -------
# OMO_SM_RESCUE_001  Albizia auxilary
# OMO_SM_RESCUE_002  Alstonia boonei
# OMO_SM_RESCUE_003  Amphimas tetracoides
# OMO_SM_RESCUE_004  Aningeria pierrei
# OMO_SM_RESCUE_005  Anthocleista vogelii
# OMO_SM_RESCUE_006  Anthonotha macrophylla
# OMO_SM_RESCUE_007  Bambusa vulgaris
# OMO_SM_RESCUE_008  Baphia nitida
# OMO_SM_RESCUE_009  Brachystegia eurycoma
# OMO_SM_RESCUE_010  Citrus sinensis
#
# Purpose
# -------
# Enter externally located Batch-1 evidence into the controlled evidence
# ledger without calculating final species-level Seed Mass estimates.
#
# ACCEPTABLE candidate evidence:
#   Alstonia boonei
#   Anthonotha macrophylla
#   Brachystegia eurycoma
#   Citrus sinensis
#
# HOLD evidence:
#   Amphimas tetracoides   — congeneric record only
#   Bambusa vulgaris      — species attribution requires confirmation
#
# No quantitative evidence yet:
#   Albizia auxilary
#   Aningeria pierrei
#   Anthocleista vogelii
#   Baphia nitida
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 9.4C — EXTERNAL SEED MASS EVIDENCE ACQUISITION: BATCH 1")
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

ACQUISITION_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)

SOURCE_CHECKLIST_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_source_search_checklist.csv"
)

CANONICAL_MAP_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)

OUT_BATCH1_EVIDENCE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_evidence.csv"
)

OUT_BATCH1_TARGET_STATUS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_target_status.csv"
)

OUT_BATCH1_NO_VALUE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_no_numeric_evidence.csv"
)

OUT_BATCH1_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_summary.csv"
)


# =============================================================================
# 2. LOAD CONTROL TABLES
# =============================================================================

for path in [
    ACQUISITION_REGISTER_FILE,
    SOURCE_CHECKLIST_FILE,
    CANONICAL_MAP_FILE,
]:
    assert path.exists(), f"Required file not found:\n{path}"


acquisition = pd.read_csv(
    ACQUISITION_REGISTER_FILE,
    low_memory=False
)

checklist = pd.read_csv(
    SOURCE_CHECKLIST_FILE,
    low_memory=False
)

canonical_map = pd.read_csv(
    CANONICAL_MAP_FILE,
    low_memory=False
)


# =============================================================================
# 3. DEFINE BATCH 1 TARGETS
# =============================================================================

batch_ids = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(1, 11)
]

batch_targets = acquisition[
    acquisition["Search_Target_ID"].isin(batch_ids)
].copy()


assert len(batch_targets) == 10


print("\nBatch 1 targets")
print("-" * 100)

display(
    batch_targets[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


# =============================================================================
# 4. BATCH 1 EXTERNAL EVIDENCE REGISTER
# =============================================================================
#
# One row = one distinct externally located evidence record.
#
# Source URLs are intentionally not required for computation; citation and
# source descriptors preserve provenance and URLs can be retained separately
# in the literature/source archive if desired.
# =============================================================================

evidence_rows = [

    # -------------------------------------------------------------------------
    # Alstonia boonei
    # PROTA: approximately 33,000 seeds per kg
    #
    # 1 kg = 1,000,000 mg
    # individual mass = 1,000,000 / 33,000
    #                 = 30.30303 mg
    # -------------------------------------------------------------------------
    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B01_001",

        "Search_Target_ID":
            "OMO_SM_RESCUE_002",

        "Final_Search_Name":
            "Alstonia boonei",

        "Source_Reported_Taxon":
            "Alstonia boonei",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "seeds per kilogram",

        "Measurement_Entity":
            "seed",

        "Measurement_Entity_Class":
            "TRUE_SEED",

        "Reported_Value":
            33000.0,

        "Reported_Unit":
            "seeds kg-1",

        "Reported_Seed_Count":
            33000,

        "Converted_Individual_Seed_Mass_mg":
            1_000_000 / 33000,

        "Conversion_Formula":
            "1000000 mg kg-1 / 33000 seeds kg-1",

        "Conversion_Status":
            "DIRECT_COUNT_TO_MASS_CONVERSION",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "seed-count-per-unit-mass compilation",

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "PROTA / Plant Resources of Tropical Africa",

        "Citation":
            "PROTA treatment of Alstonia boonei",

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "SPECIES_SPECIFIC_SEED_COUNT_CONVERSION",

        "QC_Rationale":
            (
                "Species-specific PROTA propagation record explicitly reports "
                "approximately 33,000 seeds per kg. Conversion to individual "
                "mass is mathematically direct."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,
    },


    # -------------------------------------------------------------------------
    # Amphimas tetracoides
    #
    # PROTA value is for Amphimas pterocarpoides, NOT demonstrated here to be
    # conspecific with the project target.
    # -------------------------------------------------------------------------
    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B01_002",

        "Search_Target_ID":
            "OMO_SM_RESCUE_003",

        "Final_Search_Name":
            "Amphimas tetracoides",

        "Source_Reported_Taxon":
            "Amphimas pterocarpoides",

        "Taxonomic_Match_Status":
            "CONGENERIC_NONMATCH",

        "Reported_Trait_Name":
            "1000-seed weight",

        "Measurement_Entity":
            "1000 seeds",

        "Measurement_Entity_Class":
            "1000_SEED_WEIGHT",

        "Reported_Value":
            800.0,

        "Reported_Unit":
            "g per 1000 seeds",

        "Reported_Seed_Count":
            1000,

        "Converted_Individual_Seed_Mass_mg":
            800.0,

        "Conversion_Formula":
            "800 g / 1000 seeds = 0.8 g seed-1 = 800 mg seed-1",

        "Conversion_Status":
            "NUMERIC_CONVERSION_VALID_TAXONOMIC_MATCH_UNRESOLVED",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "1000-seed weight",

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "PROTA / Plant Resources of Tropical Africa",

        "Citation":
            "PROTA treatment of Amphimas pterocarpoides",

        "Evidence_QC_Status":
            "HOLD",

        "Evidence_QC_Flag":
            "CONGENERIC_EVIDENCE_NOT_SPECIES_TRANSFERABLE",

        "QC_Rationale":
            (
                "The 800 mg individual equivalent is valid for Amphimas "
                "pterocarpoides, but no authoritative evidence has yet "
                "established that Amphimas tetracoides is the same taxon."
            ),

        "Evidence_Action":
            "RETAIN_IN_LEDGER_ONLY",

        "Eligible_For_Species_Aggregation":
            False,
    },


    # -------------------------------------------------------------------------
    # Anthonotha macrophylla
    # PROTA: approximately 60 seeds per kg
    # -------------------------------------------------------------------------
    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B01_003",

        "Search_Target_ID":
            "OMO_SM_RESCUE_006",

        "Final_Search_Name":
            "Anthonotha macrophylla",

        "Source_Reported_Taxon":
            "Anthonotha macrophylla",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "seeds per kilogram",

        "Measurement_Entity":
            "seed",

        "Measurement_Entity_Class":
            "TRUE_SEED",

        "Reported_Value":
            60.0,

        "Reported_Unit":
            "seeds kg-1",

        "Reported_Seed_Count":
            60,

        "Converted_Individual_Seed_Mass_mg":
            1_000_000 / 60,

        "Conversion_Formula":
            "1000000 mg kg-1 / 60 seeds kg-1",

        "Conversion_Status":
            "DIRECT_COUNT_TO_MASS_CONVERSION",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "seed-count-per-unit-mass compilation",

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "PROTA / Timbers 2",

        "Citation":
            "PROTA Timbers 2 treatment of Anthonotha macrophylla",

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "SPECIES_SPECIFIC_SEED_COUNT_CONVERSION",

        "QC_Rationale":
            (
                "Species-specific source reports approximately 60 seeds per "
                "kg. The implied individual seed mass is approximately "
                "16.667 g."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,
    },


    # -------------------------------------------------------------------------
    # Bambusa vulgaris
    #
    # Secondary source states approximately 25,000 bamboo seeds per kg in a
    # Bambusa vulgaris discussion, but the count is not sufficiently explicit
    # as a species-specific primary measurement.
    # -------------------------------------------------------------------------
    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B01_004",

        "Search_Target_ID":
            "OMO_SM_RESCUE_007",

        "Final_Search_Name":
            "Bambusa vulgaris",

        "Source_Reported_Taxon":
            "Bambusa vulgaris",

        "Taxonomic_Match_Status":
            "SPECIES_CONTEXT_ATTRIBUTION_REQUIRES_CONFIRMATION",

        "Reported_Trait_Name":
            "seed count per kilogram",

        "Measurement_Entity":
            "seed",

        "Measurement_Entity_Class":
            "SEED_COMPATIBLE",

        "Reported_Value":
            25000.0,

        "Reported_Unit":
            "seeds kg-1",

        "Reported_Seed_Count":
            25000,

        "Converted_Individual_Seed_Mass_mg":
            40.0,

        "Conversion_Formula":
            "1000000 mg kg-1 / 25000 seeds kg-1 = 40 mg seed-1",

        "Conversion_Status":
            "NUMERIC_CONVERSION_VALID_SOURCE_ATTRIBUTION_REVIEW",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "seed-count-per-unit-mass secondary report",

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "Bamboo cultivation literature",

        "Citation":
            "Published bamboo propagation/cultivation source citing seed count",

        "Evidence_QC_Status":
            "HOLD",

        "Evidence_QC_Flag":
            "SPECIES_ATTRIBUTION_REQUIRES_CONFIRMATION",

        "QC_Rationale":
            (
                "The numerical conversion is valid, but the reported "
                "25,000 seeds/kg is not yet sufficiently demonstrated as a "
                "direct Bambusa vulgaris species-specific measurement."
            ),

        "Evidence_Action":
            "RETAIN_IN_LEDGER_ONLY",

        "Eligible_For_Species_Aggregation":
            False,
    },


    # -------------------------------------------------------------------------
    # Brachystegia eurycoma
    #
    # Asoiro et al. 2020:
    # mature harvested dried seed
    # average 1000-unit mass = 8960 g
    # individual equivalent = 8960 mg
    # -------------------------------------------------------------------------
    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B01_005",

        "Search_Target_ID":
            "OMO_SM_RESCUE_009",

        "Final_Search_Name":
            "Brachystegia eurycoma",

        "Source_Reported_Taxon":
            "Brachystegia eurycoma",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "1000-unit seed mass",

        "Measurement_Entity":
            "1000 dried seeds",

        "Measurement_Entity_Class":
            "1000_SEED_WEIGHT",

        "Reported_Value":
            8960.0,

        "Reported_Unit":
            "g per 1000 seeds",

        "Reported_Seed_Count":
            1000,

        "Converted_Individual_Seed_Mass_mg":
            8960.0,

        "Conversion_Formula":
            "8960 g / 1000 seeds = 8.96 g seed-1 = 8960 mg seed-1",

        "Conversion_Status":
            "DIRECT_1000_SEED_WEIGHT_CONVERSION",

        "Dry_Mass_Confirmed":
            True,

        "Measurement_Method":
            (
                "Mature harvested dried seeds; physical-property "
                "measurement in laboratory study"
            ),

        "Source_Priority":
            2,

        "Source_Class":
            "PRIMARY_PEER_REVIEWED_LITERATURE",

        "Source_Name":
            "Heliyon",

        "Citation":
            (
                "Asoiro, Ezeoha, Anyanwu & Aneke (2020), "
                "Physical properties of Irvingia gabonensis, Detarium "
                "microcapum, Mucuna pruriens and Brachystegia eurycoma seeds, "
                "Heliyon 6:e04885."
            ),

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "PRIMARY_DRY_SEED_MASS_MEASUREMENT",

        "QC_Rationale":
            (
                "Primary peer-reviewed study explicitly examined mature "
                "harvested dried Brachystegia eurycoma seeds and reported "
                "1000-unit mass."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,
    },


    # -------------------------------------------------------------------------
    # Citrus sinensis
    # CIFOR-ICRAF: 1000-seed weight = 104.9 g
    # range = 30.6–285.7 g
    # -------------------------------------------------------------------------
    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B01_006",

        "Search_Target_ID":
            "OMO_SM_RESCUE_010",

        "Final_Search_Name":
            "Citrus sinensis",

        "Source_Reported_Taxon":
            "Citrus sinensis",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "weight 1000 seeds",

        "Measurement_Entity":
            "1000 seeds",

        "Measurement_Entity_Class":
            "1000_SEED_WEIGHT",

        "Reported_Value":
            104.9,

        "Reported_Unit":
            "g per 1000 seeds",

        "Reported_Seed_Count":
            1000,

        "Converted_Individual_Seed_Mass_mg":
            104.9,

        "Conversion_Formula":
            "104.9 g / 1000 seeds = 104.9 mg seed-1",

        "Conversion_Status":
            "DIRECT_1000_SEED_WEIGHT_CONVERSION",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "compiled 1000-seed weight",

        "Source_Priority":
            3,

        "Source_Class":
            "STRUCTURED_TRAIT_DATABASE",

        "Source_Name":
            "CIFOR-ICRAF species attribute compilation",

        "Citation":
            "CIFOR-ICRAF Citrus sinensis species attributes",

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "SPECIES_SPECIFIC_1000_SEED_WEIGHT",

        "QC_Rationale":
            (
                "Species-specific 1000-seed weight reported as 104.9 g "
                "with source range 30.6–285.7 g."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,
    },
]


batch_evidence = pd.DataFrame(
    evidence_rows
)


# =============================================================================
# 5. ATTACH CANONICAL SPECIES MAP
# =============================================================================

canonical_lookup = (
    canonical_map
    .groupby(
        "Search_Target_ID"
    )["Canonical_Species"]
    .apply(
        lambda x: " | ".join(
            sorted(
                set(
                    x.astype(str)
                )
            )
        )
    )
    .to_dict()
)


batch_evidence[
    "Canonical_Species_Map"
] = batch_evidence[
    "Search_Target_ID"
].map(
    canonical_lookup
)


batch_evidence[
    "Trait_Code"
] = "SEED_MASS"


# =============================================================================
# 6. NUMERIC INTEGRITY
# =============================================================================

batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
] = pd.to_numeric(
    batch_evidence[
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
)


assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].notna().all()

assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].gt(0).all()


# =============================================================================
# 7. BATCH 1 TARGET STATUS
# =============================================================================

target_status = batch_targets[
    [
        "Search_Target_ID",
        "Final_Search_Name",
        "Omo_Canonical_Taxa",
    ]
].copy()


evidence_counts = (
    batch_evidence
    .groupby(
        "Search_Target_ID"
    )
    .agg(
        N_Candidate_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        N_Accepted_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "ACCEPT").sum()
            )
        ),

        N_Held_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "HOLD").sum()
            )
        ),

        N_Rejected_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "REJECT").sum()
            )
        ),
    )
    .reset_index()
)


target_status = target_status.merge(
    evidence_counts,
    on="Search_Target_ID",
    how="left",
    validate="one_to_one"
)


for col in [
    "N_Candidate_Evidence_Records",
    "N_Accepted_Evidence_Records",
    "N_Held_Evidence_Records",
    "N_Rejected_Evidence_Records",
]:

    target_status[col] = (
        target_status[col]
        .fillna(0)
        .astype(int)
    )


target_status[
    "Batch1_Evidence_Status"
] = "NO_NUMERIC_EVIDENCE_LOCATED"


target_status.loc[
    target_status[
        "N_Held_Evidence_Records"
    ].gt(0),
    "Batch1_Evidence_Status"
] = "NUMERIC_EVIDENCE_HELD_FOR_REVIEW"


target_status.loc[
    target_status[
        "N_Accepted_Evidence_Records"
    ].gt(0),
    "Batch1_Evidence_Status"
] = "ACCEPTED_EXTERNAL_EVIDENCE_AVAILABLE"


# =============================================================================
# 8. NO-NUMERIC-EVIDENCE REGISTER
# =============================================================================

no_value = target_status[
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0)
].copy()


print("\nBatch 1 evidence disposition")
print("-" * 100)

display(
    target_status
)


print("\nTargets with no usable numeric evidence yet")
print("-" * 100)

display(
    no_value[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


# =============================================================================
# 9. EVIDENCE-QC SUMMARY
# =============================================================================

qc_summary = (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .value_counts()
    .rename_axis(
        "Evidence_QC_Status"
    )
    .reset_index(
        name="N_Evidence_Records"
    )
)


print("\nBatch 1 evidence QC summary")
print("-" * 100)

display(qc_summary)


# =============================================================================
# 10. ACCEPTED-EVIDENCE PREVIEW
# =============================================================================

accepted = batch_evidence[
    batch_evidence[
        "Evidence_QC_Status"
    ].eq(
        "ACCEPT"
    )
].copy()


print("\nAccepted Batch 1 external evidence")
print("-" * 100)

display(
    accepted[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Converted_Individual_Seed_Mass_mg",
            "Source_Class",
            "Source_Name",
            "Evidence_QC_Flag",
        ]
    ]
)


# =============================================================================
# 11. HELD-EVIDENCE PREVIEW
# =============================================================================

held = batch_evidence[
    batch_evidence[
        "Evidence_QC_Status"
    ].eq(
        "HOLD"
    )
].copy()


print("\nHeld Batch 1 external evidence")
print("-" * 100)

display(
    held[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Source_Reported_Taxon",
            "Converted_Individual_Seed_Mass_mg",
            "Evidence_QC_Flag",
            "QC_Rationale",
        ]
    ]
)


# =============================================================================
# 12. BATCH SUMMARY
# =============================================================================

batch_summary = pd.DataFrame(
    {
        "Metric": [
            "Batch biological search targets",
            "External candidate evidence records",
            "Accepted evidence records",
            "Held evidence records",
            "Rejected evidence records",
            "Targets with accepted external evidence",
            "Targets represented only by held evidence",
            "Targets with no numeric evidence located",
            "Final species Seed Mass values assigned",
        ],

        "Value": [
            len(batch_targets),
            len(batch_evidence),

            int(
                batch_evidence[
                    "Evidence_QC_Status"
                ].eq("ACCEPT").sum()
            ),

            int(
                batch_evidence[
                    "Evidence_QC_Status"
                ].eq("HOLD").sum()
            ),

            int(
                batch_evidence[
                    "Evidence_QC_Status"
                ].eq("REJECT").sum()
            ),

            int(
                target_status[
                    "N_Accepted_Evidence_Records"
                ].gt(0).sum()
            ),

            int(
                (
                    target_status[
                        "N_Accepted_Evidence_Records"
                    ].eq(0)
                    &
                    target_status[
                        "N_Held_Evidence_Records"
                    ].gt(0)
                ).sum()
            ),

            len(no_value),

            0,
        ],
    }
)


print("\nSection 9.4C Batch 1 summary")
print("-" * 100)

display(batch_summary)


# =============================================================================
# 13. STRONG VALIDATION
# =============================================================================

assert len(batch_targets) == 10

assert len(batch_evidence) == 6

assert (
    batch_evidence[
        "Evidence_Record_ID"
    ].is_unique
)

assert (
    batch_evidence[
        "Search_Target_ID"
    ].isin(
        set(batch_ids)
    ).all()
)

assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("ACCEPT").sum()
    == 4
)

assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("HOLD").sum()
    == 2
)

assert len(no_value) == 4

assert set(
    no_value[
        "Final_Search_Name"
    ]
) == {
    "Albizia auxilary",
    "Aningeria pierrei",
    "Anthocleista vogelii",
    "Baphia nitida",
}


# =============================================================================
# 14. EXPORT
# =============================================================================

batch_evidence.to_csv(
    OUT_BATCH1_EVIDENCE,
    index=False
)

target_status.to_csv(
    OUT_BATCH1_TARGET_STATUS,
    index=False
)

no_value.to_csv(
    OUT_BATCH1_NO_VALUE,
    index=False
)

batch_summary.to_csv(
    OUT_BATCH1_SUMMARY,
    index=False
)


# =============================================================================
# 15. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "External Seed Mass evidence acquisition "
    "Batch 1 completed."
)

print(
    f"Biological targets searched      : "
    f"{len(batch_targets)}"
)

print(
    f"Candidate evidence records       : "
    f"{len(batch_evidence)}"
)

print(
    f"Accepted evidence records        : "
    f"{len(accepted)}"
)

print(
    f"Held evidence records            : "
    f"{len(held)}"
)

print(
    f"Targets without numeric evidence : "
    f"{len(no_value)}"
)

print(
    "No final species-level Seed Mass estimates have been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4D — RESOLVE BATCH 1 HELD EXTERNAL SEED MASS EVIDENCE
# =============================================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print("SECTION 9.4D — RESOLVE BATCH 1 HELD EXTERNAL SEED MASS EVIDENCE")
print("=" * 100)

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BATCH1_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_evidence.csv"
)

OUT_RESOLVED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_evidence_resolved.csv"
)

OUT_ACCEPTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_accepted_evidence.csv"
)

OUT_REJECTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_rejected_evidence.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_hold_resolution_summary.csv"
)

assert BATCH1_FILE.exists()

evidence = pd.read_csv(
    BATCH1_FILE,
    low_memory=False
)

print("\nInput Batch 1 evidence")
print("-" * 100)

print(f"Rows : {len(evidence)}")


# =============================================================================
# 1. APPLY HELD-EVIDENCE DECISIONS
# =============================================================================

resolution = {
    "OMO_SM_EXT_B01_002": {
        "Evidence_QC_Status": "REJECT",
        "Evidence_QC_Flag":
            "CONGENERIC_NONTRANSFERABLE_SPECIES_EVIDENCE",
        "QC_Rationale":
            (
                "Amphimas pterocarpoides is treated as an accepted species "
                "separate from the unresolved project name Amphimas tetracoides. "
                "The 800 mg value therefore cannot be transferred across species."
            ),
        "Evidence_Action":
            "EXCLUDE_MEASUREMENT",
        "Eligible_For_Species_Aggregation":
            False,
    },

    "OMO_SM_EXT_B01_004": {
        "Evidence_QC_Status": "REJECT",
        "Evidence_QC_Flag":
            "BIOLOGICALLY_INCONSISTENT_SPECIES_ATTRIBUTION",
        "QC_Rationale":
            (
                "Authoritative bamboo references report that Bambusa vulgaris "
                "does not produce seed. The 25,000 seeds/kg attribution is "
                "therefore not accepted as species-specific Bambusa vulgaris "
                "Seed Mass evidence."
            ),
        "Evidence_Action":
            "EXCLUDE_MEASUREMENT",
        "Eligible_For_Species_Aggregation":
            False,
    },
}


for record_id, decision in resolution.items():

    mask = evidence[
        "Evidence_Record_ID"
    ].eq(
        record_id
    )

    assert mask.sum() == 1, (
        f"Expected one row for {record_id}, found {mask.sum()}."
    )

    for field, value in decision.items():

        evidence.loc[
            mask,
            field
        ] = value


# =============================================================================
# 2. VALIDATE NO HELD RECORDS REMAIN
# =============================================================================

assert (
    evidence[
        "Evidence_QC_Status"
    ]
    .eq("HOLD")
    .sum()
    == 0
)


accepted = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq("ACCEPT")
].copy()


rejected = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq("REJECT")
].copy()


print("\nResolved Batch 1 QC disposition")
print("-" * 100)

display(
    evidence[
        [
            "Evidence_Record_ID",
            "Search_Target_ID",
            "Final_Search_Name",
            "Source_Reported_Taxon",
            "Converted_Individual_Seed_Mass_mg",
            "Evidence_QC_Status",
            "Evidence_QC_Flag",
            "Evidence_Action",
        ]
    ]
)


# =============================================================================
# 3. FINAL COUNTS
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Batch 1 candidate evidence records",
        "Accepted evidence records",
        "Rejected evidence records",
        "Held evidence records remaining",
        "Targets with accepted evidence",
        "Rejected congeneric-transfer records",
        "Rejected biologically inconsistent attribution records",
    ],

    "Value": [
        len(evidence),

        len(accepted),

        len(rejected),

        int(
            evidence[
                "Evidence_QC_Status"
            ].eq("HOLD").sum()
        ),

        accepted[
            "Search_Target_ID"
        ].nunique(),

        int(
            rejected[
                "Evidence_QC_Flag"
            ].eq(
                "CONGENERIC_NONTRANSFERABLE_SPECIES_EVIDENCE"
            ).sum()
        ),

        int(
            rejected[
                "Evidence_QC_Flag"
            ].eq(
                "BIOLOGICALLY_INCONSISTENT_SPECIES_ATTRIBUTION"
            ).sum()
        ),
    ],
})


print("\nSection 9.4D hold-resolution summary")
print("-" * 100)

display(summary)


# =============================================================================
# 4. STRONG VALIDATION
# =============================================================================

assert len(evidence) == 6
assert len(accepted) == 4
assert len(rejected) == 2

assert set(
    rejected[
        "Final_Search_Name"
    ]
) == {
    "Amphimas tetracoides",
    "Bambusa vulgaris",
}

assert (
    accepted[
        "Eligible_For_Species_Aggregation"
    ]
    .astype(bool)
    .all()
)

assert (
    rejected[
        "Eligible_For_Species_Aggregation"
    ]
    .astype(bool)
    .sum()
    == 0
)


# =============================================================================
# 5. EXPORT
# =============================================================================

evidence.to_csv(
    OUT_RESOLVED,
    index=False
)

accepted.to_csv(
    OUT_ACCEPTED,
    index=False
)

rejected.to_csv(
    OUT_REJECTED,
    index=False
)

summary.to_csv(
    OUT_SUMMARY,
    index=False
)


print("\nResolution exports")
print("-" * 100)

print(f"Resolved evidence : {OUT_RESOLVED}")
print(f"Accepted evidence : {OUT_ACCEPTED}")
print(f"Rejected evidence : {OUT_REJECTED}")
print(f"Summary           : {OUT_SUMMARY}")


print("\n" + "=" * 100)

print(
    "Batch 1 held external Seed Mass evidence resolved successfully."
)

print("Accepted records : 4")
print("Rejected records : 2")
print("Held records     : 0")

print(
    "Amphimas tetracoides remains unresolved at species level."
)

print(
    "Bambusa vulgaris remains unresolved for Seed Mass because "
    "the held seed-count attribution was rejected."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4E — EXTERNAL SEED MASS EVIDENCE ACQUISITION: BATCH 2
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print("SECTION 9.4E — EXTERNAL SEED MASS EVIDENCE ACQUISITION: BATCH 2")
print("=" * 100)

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

ACQUISITION_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)

CANONICAL_MAP_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)

OUT_BATCH2_EVIDENCE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_evidence.csv"
)

OUT_BATCH2_TARGET_STATUS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_target_status.csv"
)

OUT_BATCH2_NO_VALUE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_no_numeric_evidence.csv"
)

OUT_BATCH2_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_summary.csv"
)

acquisition = pd.read_csv(
    ACQUISITION_REGISTER_FILE,
    low_memory=False
)

canonical_map = pd.read_csv(
    CANONICAL_MAP_FILE,
    low_memory=False
)

batch_ids = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(11, 21)
]

batch_targets = acquisition[
    acquisition["Search_Target_ID"].isin(batch_ids)
].copy()

assert len(batch_targets) == 10

print("\nBatch 2 targets")
print("-" * 100)

display(
    batch_targets[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


# =============================================================================
# 1. EXTERNAL EVIDENCE ROWS
# =============================================================================

evidence_rows = [

    # Cleistopholis patens
    {
        "Evidence_Record_ID": "OMO_SM_EXT_B02_001",
        "Search_Target_ID": "OMO_SM_RESCUE_011",
        "Final_Search_Name": "Cleistopholis patens",
        "Source_Reported_Taxon": "Cleistopholis patens",
        "Taxonomic_Match_Status": "EXACT_SPECIES_MATCH",
        "Reported_Trait_Name": "seeds per kilogram",
        "Measurement_Entity": "seed",
        "Measurement_Entity_Class": "TRUE_SEED",
        "Reported_Value": 650.0,
        "Reported_Unit": "approximate midpoint of 600-700 seeds kg-1",
        "Reported_Seed_Count": 650,
        "Converted_Individual_Seed_Mass_mg": 1_000_000 / 650,
        "Conversion_Formula": "1000000 mg kg-1 / 650 seeds kg-1",
        "Conversion_Status": "RANGE_MIDPOINT_DERIVED",
        "Dry_Mass_Confirmed": pd.NA,
        "Measurement_Method": "species-specific seed-count range",
        "Source_Priority": 4,
        "Source_Class": "TRACEABLE_SECONDARY_COMPILATION",
        "Source_Name": "PROTA / Plant Resources of Tropical Africa",
        "Citation": "PROTA treatment of Cleistopholis patens",
        "Evidence_QC_Status": "HOLD",
        "Evidence_QC_Flag": "RANGE_DERIVED_MIDPOINT_REQUIRES_REVIEW",
        "QC_Rationale": (
            "Source reports 600–700 seeds/kg rather than a single exact value. "
            "Midpoint-derived individual mass is retained only as provisional evidence."
        ),
        "Evidence_Action": "RETAIN_IN_LEDGER_ONLY",
        "Eligible_For_Species_Aggregation": False,
    },

    # Cola nitida — PROTA range midpoint
    {
        "Evidence_Record_ID": "OMO_SM_EXT_B02_002",
        "Search_Target_ID": "OMO_SM_RESCUE_012",
        "Final_Search_Name": "Cola nitida",
        "Source_Reported_Taxon": "Cola nitida",
        "Taxonomic_Match_Status": "EXACT_SPECIES_MATCH",
        "Reported_Trait_Name": "1000-seed weight",
        "Measurement_Entity": "1000 seeds",
        "Measurement_Entity_Class": "1000_SEED_WEIGHT",
        "Reported_Value": 2000.0,
        "Reported_Unit": "midpoint of 500-3500 g per 1000 seeds",
        "Reported_Seed_Count": 1000,
        "Converted_Individual_Seed_Mass_mg": 2000.0,
        "Conversion_Formula": (
            "midpoint(500,3500) g / 1000 seeds = 2000 mg seed-1"
        ),
        "Conversion_Status": "RANGE_MIDPOINT_DERIVED",
        "Dry_Mass_Confirmed": pd.NA,
        "Measurement_Method": "compiled 1000-seed weight range",
        "Source_Priority": 4,
        "Source_Class": "TRACEABLE_SECONDARY_COMPILATION",
        "Source_Name": "PROTA",
        "Citation": "PROTA treatment of Cola nitida",
        "Evidence_QC_Status": "HOLD",
        "Evidence_QC_Flag": "HIGH_RANGE_VARIABILITY_REQUIRES_REVIEW",
        "QC_Rationale": (
            "Reported 1000-seed weight spans 500–3500 g, indicating substantial "
            "within-species variation. Midpoint is provisional only."
        ),
        "Evidence_Action": "RETAIN_IN_LEDGER_ONLY",
        "Eligible_For_Species_Aggregation": False,
    },

    # Cola nitida — primary nursery study, 15 g selected seeds
    {
        "Evidence_Record_ID": "OMO_SM_EXT_B02_003",
        "Search_Target_ID": "OMO_SM_RESCUE_012",
        "Final_Search_Name": "Cola nitida",
        "Source_Reported_Taxon": "Cola nitida",
        "Taxonomic_Match_Status": "EXACT_SPECIES_MATCH",
        "Reported_Trait_Name": "individual seed/nut weight",
        "Measurement_Entity": "seed/nut",
        "Measurement_Entity_Class": "TRUE_SEED",
        "Reported_Value": 15.0,
        "Reported_Unit": "g per seed",
        "Reported_Seed_Count": 1,
        "Converted_Individual_Seed_Mass_mg": 15000.0,
        "Conversion_Formula": "15 g seed-1 × 1000 = 15000 mg seed-1",
        "Conversion_Status": "DIRECT_UNIT_CONVERSION",
        "Dry_Mass_Confirmed": False,
        "Measurement_Method": "selected fresh nursery seed material",
        "Source_Priority": 2,
        "Source_Class": "PRIMARY_PEER_REVIEWED_LITERATURE",
        "Source_Name": "International Journal of Horticultural Science",
        "Citation": (
            "Nursery study of Cola nitida using seeds/nuts of 15 g weight"
        ),
        "Evidence_QC_Status": "HOLD",
        "Evidence_QC_Flag": "SELECTED_FRESH_SEED_MASS_NOT_DRY_MASS",
        "QC_Rationale": (
            "Study intentionally selected 15 g fresh seeds/nuts. Value is species-specific "
            "but not representative dry seed mass."
        ),
        "Evidence_Action": "RETAIN_IN_LEDGER_ONLY",
        "Eligible_For_Species_Aggregation": False,
    },

    # Hexalobus crispiflorus
    {
        "Evidence_Record_ID": "OMO_SM_EXT_B02_004",
        "Search_Target_ID": "OMO_SM_RESCUE_018",
        "Final_Search_Name": "Hexalobus crispiflorus",
        "Source_Reported_Taxon": "Hexalobus crispiflorus",
        "Taxonomic_Match_Status": "EXACT_SPECIES_MATCH",
        "Reported_Trait_Name": "seeds per kilogram",
        "Measurement_Entity": "seed",
        "Measurement_Entity_Class": "TRUE_SEED",
        "Reported_Value": 600.0,
        "Reported_Unit": "seeds kg-1",
        "Reported_Seed_Count": 600,
        "Converted_Individual_Seed_Mass_mg": 1_000_000 / 600,
        "Conversion_Formula": "1000000 mg kg-1 / 600 seeds kg-1",
        "Conversion_Status": "DIRECT_COUNT_TO_MASS_CONVERSION",
        "Dry_Mass_Confirmed": pd.NA,
        "Measurement_Method": "species-specific seed-count-per-unit-mass compilation",
        "Source_Priority": 4,
        "Source_Class": "TRACEABLE_SECONDARY_COMPILATION",
        "Source_Name": "PROTA",
        "Citation": "PROTA treatment of Hexalobus crispiflorus",
        "Evidence_QC_Status": "ACCEPT",
        "Evidence_QC_Flag": "SPECIES_SPECIFIC_SEED_COUNT_CONVERSION",
        "QC_Rationale": (
            "Source explicitly reports about 600 seeds per kilogram for the target species."
        ),
        "Evidence_Action": "RETAIN_FOR_SPECIES_AGGREGATION",
        "Eligible_For_Species_Aggregation": True,
    },

    # Irvingia gabonensis — primary morphometric study
    {
        "Evidence_Record_ID": "OMO_SM_EXT_B02_005",
        "Search_Target_ID": "OMO_SM_RESCUE_020",
        "Final_Search_Name": "Irvingia gabonensis",
        "Source_Reported_Taxon": "Irvingia gabonensis",
        "Taxonomic_Match_Status": "EXACT_SPECIES_MATCH",
        "Reported_Trait_Name": "individual seed mass",
        "Measurement_Entity": "seed",
        "Measurement_Entity_Class": "TRUE_SEED",
        "Reported_Value": 12.72,
        "Reported_Unit": "g per seed",
        "Reported_Seed_Count": 1,
        "Converted_Individual_Seed_Mass_mg": 12720.0,
        "Conversion_Formula": "12.72 g × 1000 = 12720 mg",
        "Conversion_Status": "DIRECT_UNIT_CONVERSION",
        "Dry_Mass_Confirmed": pd.NA,
        "Measurement_Method": "morphometric measurement",
        "Source_Priority": 2,
        "Source_Class": "PRIMARY_PEER_REVIEWED_LITERATURE",
        "Source_Name": "Economic Botany / morphometric study",
        "Citation": (
            "Padonou et al. pilot assessment of Irvingia gabonensis morphotypes "
            "in southwestern Benin"
        ),
        "Evidence_QC_Status": "ACCEPT",
        "Evidence_QC_Flag": "PRIMARY_SPECIES_SPECIFIC_SEED_MASS",
        "QC_Rationale": (
            "Primary study reports mean seed mass directly for Irvingia gabonensis."
        ),
        "Evidence_Action": "RETAIN_FOR_SPECIES_AGGREGATION",
        "Eligible_For_Species_Aggregation": True,
    },

    # Irvingia gabonensis — primary 1000-unit mass
    {
        "Evidence_Record_ID": "OMO_SM_EXT_B02_006",
        "Search_Target_ID": "OMO_SM_RESCUE_020",
        "Final_Search_Name": "Irvingia gabonensis",
        "Source_Reported_Taxon": "Irvingia gabonensis",
        "Taxonomic_Match_Status": "EXACT_SPECIES_MATCH",
        "Reported_Trait_Name": "1000-unit seed mass",
        "Measurement_Entity": "1000 seeds",
        "Measurement_Entity_Class": "1000_SEED_WEIGHT",
        "Reported_Value": 11130.0,
        "Reported_Unit": "g per 1000 seeds",
        "Reported_Seed_Count": 1000,
        "Converted_Individual_Seed_Mass_mg": 11130.0,
        "Conversion_Formula": "11130 g / 1000 seeds = 11130 mg seed-1",
        "Conversion_Status": "DIRECT_1000_SEED_WEIGHT_CONVERSION",
        "Dry_Mass_Confirmed": True,
        "Measurement_Method": "laboratory physical-property measurement",
        "Source_Priority": 2,
        "Source_Class": "PRIMARY_PEER_REVIEWED_LITERATURE",
        "Source_Name": "Heliyon",
        "Citation": (
            "Asoiro et al. (2020), Physical properties of Irvingia gabonensis "
            "and other African culinary seeds"
        ),
        "Evidence_QC_Status": "ACCEPT",
        "Evidence_QC_Flag": "PRIMARY_DRY_SEED_MASS_MEASUREMENT",
        "QC_Rationale": (
            "Primary laboratory study reports mean 1000-unit mass of Irvingia "
            "gabonensis seeds."
        ),
        "Evidence_Action": "RETAIN_FOR_SPECIES_AGGREGATION",
        "Eligible_For_Species_Aggregation": True,
    },
]

batch_evidence = pd.DataFrame(
    evidence_rows
)


# =============================================================================
# 2. ATTACH CANONICAL MAP
# =============================================================================

canonical_lookup = (
    canonical_map
    .groupby(
        "Search_Target_ID"
    )["Canonical_Species"]
    .apply(
        lambda x: " | ".join(
            sorted(
                set(
                    x.astype(str)
                )
            )
        )
    )
    .to_dict()
)

batch_evidence[
    "Canonical_Species_Map"
] = batch_evidence[
    "Search_Target_ID"
].map(
    canonical_lookup
)

batch_evidence["Trait_Code"] = "SEED_MASS"


# =============================================================================
# 3. NUMERIC VALIDATION
# =============================================================================

batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
] = pd.to_numeric(
    batch_evidence[
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
)

assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].notna().all()

assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].gt(0).all()


# =============================================================================
# 4. TARGET STATUS
# =============================================================================

target_status = batch_targets[
    [
        "Search_Target_ID",
        "Final_Search_Name",
        "Omo_Canonical_Taxa",
    ]
].copy()

evidence_counts = (
    batch_evidence
    .groupby(
        "Search_Target_ID"
    )
    .agg(
        N_Candidate_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),
        N_Accepted_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "ACCEPT").sum()
            )
        ),
        N_Held_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "HOLD").sum()
            )
        ),
        N_Rejected_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "REJECT").sum()
            )
        ),
    )
    .reset_index()
)

target_status = target_status.merge(
    evidence_counts,
    on="Search_Target_ID",
    how="left",
    validate="one_to_one"
)

for col in [
    "N_Candidate_Evidence_Records",
    "N_Accepted_Evidence_Records",
    "N_Held_Evidence_Records",
    "N_Rejected_Evidence_Records",
]:
    target_status[col] = (
        target_status[col]
        .fillna(0)
        .astype(int)
    )

target_status[
    "Batch2_Evidence_Status"
] = "NO_NUMERIC_EVIDENCE_LOCATED"

target_status.loc[
    target_status[
        "N_Held_Evidence_Records"
    ].gt(0),
    "Batch2_Evidence_Status"
] = "NUMERIC_EVIDENCE_HELD_FOR_REVIEW"

target_status.loc[
    target_status[
        "N_Accepted_Evidence_Records"
    ].gt(0),
    "Batch2_Evidence_Status"
] = "ACCEPTED_EXTERNAL_EVIDENCE_AVAILABLE"


# =============================================================================
# 5. NO-VALUE TARGETS
# =============================================================================

no_value = target_status[
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0)
].copy()


print("\nBatch 2 evidence disposition")
print("-" * 100)

display(target_status)

print("\nTargets with no numeric evidence located")
print("-" * 100)

display(
    no_value[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


# =============================================================================
# 6. QC SUMMARY
# =============================================================================

qc_summary = (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .value_counts()
    .rename_axis(
        "Evidence_QC_Status"
    )
    .reset_index(
        name="N_Evidence_Records"
    )
)

print("\nBatch 2 evidence QC summary")
print("-" * 100)

display(qc_summary)


# =============================================================================
# 7. BATCH SUMMARY
# =============================================================================

batch_summary = pd.DataFrame({
    "Metric": [
        "Batch biological search targets",
        "External candidate evidence records",
        "Accepted evidence records",
        "Held evidence records",
        "Rejected evidence records",
        "Targets with accepted external evidence",
        "Targets represented only by held evidence",
        "Targets with no numeric evidence located",
        "Final species Seed Mass values assigned",
    ],

    "Value": [
        len(batch_targets),
        len(batch_evidence),
        int(
            batch_evidence[
                "Evidence_QC_Status"
            ].eq("ACCEPT").sum()
        ),
        int(
            batch_evidence[
                "Evidence_QC_Status"
            ].eq("HOLD").sum()
        ),
        int(
            batch_evidence[
                "Evidence_QC_Status"
            ].eq("REJECT").sum()
        ),
        int(
            target_status[
                "N_Accepted_Evidence_Records"
            ].gt(0).sum()
        ),
        int(
            (
                target_status[
                    "N_Accepted_Evidence_Records"
                ].eq(0)
                &
                target_status[
                    "N_Held_Evidence_Records"
                ].gt(0)
            ).sum()
        ),
        len(no_value),
        0,
    ],
})


print("\nSection 9.4E Batch 2 summary")
print("-" * 100)

display(batch_summary)


# =============================================================================
# 8. STRONG VALIDATION
# =============================================================================

assert len(batch_targets) == 10
assert len(batch_evidence) == 6

assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("ACCEPT").sum()
    == 3
)

assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("HOLD").sum()
    == 3
)

assert len(no_value) == 6


# =============================================================================
# 9. EXPORT
# =============================================================================

batch_evidence.to_csv(
    OUT_BATCH2_EVIDENCE,
    index=False
)

target_status.to_csv(
    OUT_BATCH2_TARGET_STATUS,
    index=False
)

no_value.to_csv(
    OUT_BATCH2_NO_VALUE,
    index=False
)

batch_summary.to_csv(
    OUT_BATCH2_SUMMARY,
    index=False
)


print("\n" + "=" * 100)

print(
    "External Seed Mass evidence acquisition "
    "Batch 2 completed."
)

print(
    f"Biological targets searched      : "
    f"{len(batch_targets)}"
)

print(
    f"Candidate evidence records       : "
    f"{len(batch_evidence)}"
)

print(
    f"Accepted evidence records        : "
    f"{batch_evidence['Evidence_QC_Status'].eq('ACCEPT').sum()}"
)

print(
    f"Held evidence records            : "
    f"{batch_evidence['Evidence_QC_Status'].eq('HOLD').sum()}"
)

print(
    f"Targets without numeric evidence : "
    f"{len(no_value)}"
)

print(
    "No final species-level Seed Mass estimates have been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4F — RESOLVE BATCH 2 HELD EXTERNAL SEED MASS EVIDENCE
# =============================================================================
#
# Purpose
# -------
# Resolve the three records placed on HOLD in Section 9.4E.
#
# Decisions
# ---------
#
# Cleistopholis patens
#     Source evidence : 600–700 seeds kg-1
#     Representative : midpoint count = 650 seeds kg-1
#     Converted mass : 1,000,000 / 650 = 1538.461538 mg seed-1
#     Decision       : ACCEPT
#
# Cola nitida — PROTA range
#     Source evidence : 500–3500 g per 1000 seeds
#     Representative : midpoint = 2000 g per 1000 seeds
#     Converted mass : 2000 mg seed-1
#     Decision       : ACCEPT
#
# Cola nitida — selected nursery material
#     Source evidence : 15 g selected fresh seed/nut
#     Converted mass : 15000 mg
#     Decision       : REJECT
#
# IMPORTANT
# ---------
# Range-derived representative estimates remain explicitly flagged.
# No final species-level aggregation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4F — RESOLVE BATCH 2 HELD "
    "EXTERNAL SEED MASS EVIDENCE"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BATCH2_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_evidence.csv"
)


OUT_RESOLVED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_evidence_resolved.csv"
)

OUT_ACCEPTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_accepted_evidence.csv"
)

OUT_REJECTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_rejected_evidence.csv"
)

OUT_RANGE_DERIVED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_range_derived_accepted.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_hold_resolution_summary.csv"
)


# =============================================================================
# 2. LOAD BATCH 2 EVIDENCE
# =============================================================================

assert BATCH2_FILE.exists(), (
    f"Batch 2 evidence file not found:\n{BATCH2_FILE}"
)

evidence = pd.read_csv(
    BATCH2_FILE,
    low_memory=False
)


print("\nInput Batch 2 evidence")
print("-" * 100)

print(
    f"Rows : {len(evidence)}"
)


assert len(evidence) == 6

assert evidence[
    "Evidence_Record_ID"
].is_unique


print(
    f"Held records entering review : "
    f"{evidence['Evidence_QC_Status'].eq('HOLD').sum()}"
)


assert (
    evidence[
        "Evidence_QC_Status"
    ]
    .eq("HOLD")
    .sum()
    == 3
)


# =============================================================================
# 3. PRESERVE ORIGINAL HOLD INFORMATION
# =============================================================================

evidence[
    "Pre_9_4F_QC_Status"
] = evidence[
    "Evidence_QC_Status"
]

evidence[
    "Pre_9_4F_QC_Flag"
] = evidence[
    "Evidence_QC_Flag"
]

evidence[
    "Hold_Review_Resolved"
] = False

evidence[
    "Hold_Review_Decision"
] = pd.NA


# =============================================================================
# 4. RESOLUTION DECISIONS
# =============================================================================

resolution = {

    # -------------------------------------------------------------------------
    # Cleistopholis patens
    #
    # PROTA explicitly reports 600–700 seeds/kg.
    #
    # Representative conversion uses midpoint count:
    #
    #     650 seeds/kg
    #
    #     1,000,000 mg / 650
    #         = 1538.461538 mg seed-1
    #
    # This is not treated as an exact measured individual mass.
    # -------------------------------------------------------------------------

    "OMO_SM_EXT_B02_001": {

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "SPECIES_SPECIFIC_RANGE_DERIVED_SEED_MASS",

        "QC_Rationale":
            (
                "Authoritative species-specific source explicitly reports "
                "600–700 seeds per kilogram. A representative individual "
                "Seed Mass of 1538.461538 mg is derived from the midpoint "
                "count of 650 seeds/kg. The estimate is accepted while "
                "retaining explicit range-derived provenance."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,

        "Hold_Review_Resolved":
            True,

        "Hold_Review_Decision":
            "ACCEPT_RANGE_DERIVED_SPECIES_EVIDENCE",
    },


    # -------------------------------------------------------------------------
    # Cola nitida — PROTA 1000-seed-weight range
    #
    # 500–3500 g / 1000 seeds
    #
    # Equivalent individual Seed Mass range:
    #
    #     500–3500 mg
    #
    # Representative midpoint:
    #
    #     2000 mg
    #
    # -------------------------------------------------------------------------

    "OMO_SM_EXT_B02_002": {

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "HIGH_VARIABILITY_RANGE_DERIVED_SEED_MASS",

        "QC_Rationale":
            (
                "Species-specific source reports a 1000-seed-weight range "
                "of 500–3500 g, equivalent to 500–3500 mg per seed. "
                "The 2000 mg midpoint is accepted as a representative "
                "range-derived Seed Mass while the very high reported "
                "within-species range remains explicitly flagged."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,

        "Hold_Review_Resolved":
            True,

        "Hold_Review_Decision":
            "ACCEPT_HIGH_VARIABILITY_RANGE_DERIVED",
    },


    # -------------------------------------------------------------------------
    # Cola nitida — selected 15 g nursery seeds
    #
    # Seeds below 15 g were deliberately excluded.
    # This is therefore selected experimental planting material rather than
    # an unbiased species-level Seed Mass estimate.
    # -------------------------------------------------------------------------

    "OMO_SM_EXT_B02_003": {

        "Evidence_QC_Status":
            "REJECT",

        "Evidence_QC_Flag":
            "SELECTED_NURSERY_SEED_MASS_NOT_SPECIES_REPRESENTATIVE",

        "QC_Rationale":
            (
                "The study deliberately selected 15 g fresh Cola nitida "
                "seeds/nuts and excluded smaller material. The value is "
                "therefore a selected nursery-treatment criterion rather "
                "than a representative species-level dry Seed Mass and is "
                "excluded from standardized trait aggregation."
            ),

        "Evidence_Action":
            "EXCLUDE_MEASUREMENT",

        "Eligible_For_Species_Aggregation":
            False,

        "Hold_Review_Resolved":
            True,

        "Hold_Review_Decision":
            "REJECT_SELECTED_EXPERIMENTAL_MATERIAL",
    },
}


# =============================================================================
# 5. APPLY RESOLUTION
# =============================================================================

for record_id, decision in resolution.items():

    mask = evidence[
        "Evidence_Record_ID"
    ].eq(
        record_id
    )

    assert mask.sum() == 1, (
        f"Expected exactly one row for {record_id}; "
        f"found {mask.sum()}."
    )

    assert evidence.loc[
        mask,
        "Pre_9_4F_QC_Status"
    ].iloc[0] == "HOLD", (
        f"{record_id} was not a held Batch 2 record."
    )

    for field, value in decision.items():

        evidence.loc[
            mask,
            field
        ] = value


# =============================================================================
# 6. RANGE-DERIVED EVIDENCE METADATA
# =============================================================================
#
# Add explicit reported-range metadata rather than leaving only the midpoint.
# =============================================================================

evidence[
    "Reported_Range_Min"
] = np.nan

evidence[
    "Reported_Range_Max"
] = np.nan

evidence[
    "Reported_Range_Unit"
] = pd.NA

evidence[
    "Converted_Seed_Mass_Range_Min_mg"
] = np.nan

evidence[
    "Converted_Seed_Mass_Range_Max_mg"
] = np.nan


# Cleistopholis patens
mask = evidence[
    "Evidence_Record_ID"
].eq(
    "OMO_SM_EXT_B02_001"
)

evidence.loc[
    mask,
    "Reported_Range_Min"
] = 600.0

evidence.loc[
    mask,
    "Reported_Range_Max"
] = 700.0

evidence.loc[
    mask,
    "Reported_Range_Unit"
] = "seeds kg-1"

# Inverse relation:
# 700 seeds/kg -> smaller individual mass
# 600 seeds/kg -> larger individual mass

evidence.loc[
    mask,
    "Converted_Seed_Mass_Range_Min_mg"
] = (
    1_000_000 / 700
)

evidence.loc[
    mask,
    "Converted_Seed_Mass_Range_Max_mg"
] = (
    1_000_000 / 600
)


# Cola nitida PROTA
mask = evidence[
    "Evidence_Record_ID"
].eq(
    "OMO_SM_EXT_B02_002"
)

evidence.loc[
    mask,
    "Reported_Range_Min"
] = 500.0

evidence.loc[
    mask,
    "Reported_Range_Max"
] = 3500.0

evidence.loc[
    mask,
    "Reported_Range_Unit"
] = "g per 1000 seeds"

evidence.loc[
    mask,
    "Converted_Seed_Mass_Range_Min_mg"
] = 500.0

evidence.loc[
    mask,
    "Converted_Seed_Mass_Range_Max_mg"
] = 3500.0


# =============================================================================
# 7. VALIDATE RANGE-DERIVED REPRESENTATIVE VALUES
# =============================================================================

cleistopholis = evidence[
    evidence[
        "Evidence_Record_ID"
    ].eq(
        "OMO_SM_EXT_B02_001"
    )
].iloc[0]


expected_cleistopholis = (
    1_000_000 / 650
)


assert np.isclose(
    cleistopholis[
        "Converted_Individual_Seed_Mass_mg"
    ],
    expected_cleistopholis,
    rtol=1e-9,
), (
    "Unexpected Cleistopholis representative Seed Mass."
)


cola_range = evidence[
    evidence[
        "Evidence_Record_ID"
    ].eq(
        "OMO_SM_EXT_B02_002"
    )
].iloc[0]


assert np.isclose(
    cola_range[
        "Converted_Individual_Seed_Mass_mg"
    ],
    2000.0,
), (
    "Unexpected Cola nitida range-derived midpoint."
)


# =============================================================================
# 8. FINAL QC DISPOSITION
# =============================================================================

accepted = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq(
        "ACCEPT"
    )
].copy()


rejected = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq(
        "REJECT"
    )
].copy()


held = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq(
        "HOLD"
    )
].copy()


range_derived = accepted[
    accepted[
        "Evidence_QC_Flag"
    ].isin(
        [
            "SPECIES_SPECIFIC_RANGE_DERIVED_SEED_MASS",
            "HIGH_VARIABILITY_RANGE_DERIVED_SEED_MASS",
        ]
    )
].copy()


print("\nResolved Batch 2 QC disposition")
print("-" * 100)

display(
    evidence[
        [
            "Evidence_Record_ID",
            "Search_Target_ID",
            "Final_Search_Name",
            "Converted_Individual_Seed_Mass_mg",
            "Evidence_QC_Status",
            "Evidence_QC_Flag",
            "Evidence_Action",
            "Hold_Review_Decision",
        ]
    ]
)


# =============================================================================
# 9. RANGE-DERIVED ACCEPTED RECORDS
# =============================================================================

print("\nAccepted range-derived evidence")
print("-" * 100)

display(
    range_derived[
        [
            "Final_Search_Name",
            "Reported_Range_Min",
            "Reported_Range_Max",
            "Reported_Range_Unit",
            "Converted_Seed_Mass_Range_Min_mg",
            "Converted_Individual_Seed_Mass_mg",
            "Converted_Seed_Mass_Range_Max_mg",
            "Evidence_QC_Flag",
        ]
    ]
)


# =============================================================================
# 10. ACCEPTED SPECIES EVIDENCE SUMMARY
# =============================================================================

accepted_species_summary = (
    accepted
    .groupby(
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        dropna=False
    )
    .agg(
        N_Accepted_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        Min_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "min"
        ),

        Median_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "median"
        ),

        Max_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "max"
        ),
    )
    .reset_index()
)


print("\nAccepted Batch 2 species evidence")
print("-" * 100)

display(
    accepted_species_summary
)


# =============================================================================
# 11. HOLD-RESOLUTION SUMMARY
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Batch 2 candidate evidence records",
        "Records initially accepted",
        "Records initially held",
        "Held records accepted after review",
        "Held records rejected after review",
        "Final accepted evidence records",
        "Final rejected evidence records",
        "Held evidence records remaining",
        "Targets with accepted evidence",
        "Range-derived accepted records",
    ],

    "Value": [
        len(evidence),

        int(
            evidence[
                "Pre_9_4F_QC_Status"
            ].eq(
                "ACCEPT"
            ).sum()
        ),

        int(
            evidence[
                "Pre_9_4F_QC_Status"
            ].eq(
                "HOLD"
            ).sum()
        ),

        int(
            (
                evidence[
                    "Pre_9_4F_QC_Status"
                ].eq("HOLD")
                &
                evidence[
                    "Evidence_QC_Status"
                ].eq("ACCEPT")
            ).sum()
        ),

        int(
            (
                evidence[
                    "Pre_9_4F_QC_Status"
                ].eq("HOLD")
                &
                evidence[
                    "Evidence_QC_Status"
                ].eq("REJECT")
            ).sum()
        ),

        len(accepted),

        len(rejected),

        len(held),

        accepted[
            "Search_Target_ID"
        ].nunique(),

        len(range_derived),
    ],
})


print("\nSection 9.4F hold-resolution summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 12. STRONG VALIDATION
# =============================================================================

assert len(evidence) == 6

assert len(accepted) == 5, (
    f"Expected 5 final accepted records, found {len(accepted)}."
)

assert len(rejected) == 1, (
    f"Expected 1 final rejected record, found {len(rejected)}."
)

assert len(held) == 0

assert (
    accepted[
        "Search_Target_ID"
    ].nunique()
    == 4
), (
    "Expected accepted evidence for four Batch 2 biological targets."
)

assert set(
    accepted[
        "Final_Search_Name"
    ]
) == {
    "Cleistopholis patens",
    "Cola nitida",
    "Hexalobus crispiflorus",
    "Irvingia gabonensis",
}


assert set(
    rejected[
        "Final_Search_Name"
    ]
) == {
    "Cola nitida"
}


assert (
    accepted[
        "Eligible_For_Species_Aggregation"
    ]
    .astype(bool)
    .all()
)


assert (
    rejected[
        "Eligible_For_Species_Aggregation"
    ]
    .astype(bool)
    .sum()
    == 0
)


# =============================================================================
# 13. EXPORT
# =============================================================================

evidence.to_csv(
    OUT_RESOLVED,
    index=False
)

accepted.to_csv(
    OUT_ACCEPTED,
    index=False
)

rejected.to_csv(
    OUT_REJECTED,
    index=False
)

range_derived.to_csv(
    OUT_RANGE_DERIVED,
    index=False
)

summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 14. REPORT EXPORTS
# =============================================================================

print("\nResolution exports")
print("-" * 100)

print(
    f"Resolved evidence     : "
    f"{OUT_RESOLVED}"
)

print(
    f"Accepted evidence     : "
    f"{OUT_ACCEPTED}"
)

print(
    f"Rejected evidence     : "
    f"{OUT_REJECTED}"
)

print(
    f"Range-derived records : "
    f"{OUT_RANGE_DERIVED}"
)

print(
    f"Summary               : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 15. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Batch 2 held external Seed Mass evidence "
    "resolved successfully."
)

print(
    f"Accepted records : "
    f"{len(accepted)}"
)

print(
    f"Rejected records : "
    f"{len(rejected)}"
)

print(
    f"Held records     : "
    f"{len(held)}"
)

print(
    "Cleistopholis patens range-derived evidence accepted "
    "with explicit provenance."
)

print(
    "Cola nitida PROTA range-derived evidence accepted with "
    "high-variability flag; selected 15 g nursery material excluded."
)

print(
    "No final species-level Seed Mass estimates have yet been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4G — EXTERNAL SEED MASS EVIDENCE ACQUISITION: BATCH 3
# =============================================================================
#
# Authoritative Batch 3 targets
# -----------------------------
#
# OMO_SM_RESCUE_021  Lonchocarpus cilliata
# OMO_SM_RESCUE_022  Macaranga barteri
# OMO_SM_RESCUE_023  Maesobotrya barteri
# OMO_SM_RESCUE_024  Millettia thonningii
# OMO_SM_RESCUE_025  Mitragyna ciliata
# OMO_SM_RESCUE_026  Musa × paradisiaca
# OMO_SM_RESCUE_027  Myrianthus arboreus
# OMO_SM_RESCUE_028  Nesogordonia papaverifera
# OMO_SM_RESCUE_029  Nothospondias staudtii
# OMO_SM_RESCUE_030  Picralima nitida
#
# Current defensible quantitative evidence
# -----------------------------------------
#
# Nesogordonia papaverifera
#     25,000 dry seeds kg-1
#     1,000,000 mg / 25,000
#     = 40 mg seed-1
#     Decision: ACCEPT
#
# Picralima nitida
#     mean individual seed mass = 2.26 g
#     moisture = 47.19% wet basis
#     Decision: HOLD for dry-mass/moisture-normalization review
#
# No defensible numeric species-level Seed Mass evidence yet
# ----------------------------------------------------------
# Lonchocarpus cilliata
# Macaranga barteri
# Maesobotrya barteri
# Millettia thonningii
# Mitragyna ciliata
# Musa × paradisiaca
# Myrianthus arboreus
# Nothospondias staudtii
#
# IMPORTANT
# ---------
# No final species-level Seed Mass aggregation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4G — EXTERNAL SEED MASS "
    "EVIDENCE ACQUISITION: BATCH 3"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

ACQUISITION_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)

CANONICAL_MAP_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)

OUT_BATCH3_EVIDENCE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_evidence.csv"
)

OUT_BATCH3_TARGET_STATUS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_target_status.csv"
)

OUT_BATCH3_NO_VALUE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_no_numeric_evidence.csv"
)

OUT_BATCH3_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_summary.csv"
)


# =============================================================================
# 2. LOAD CONTROL TABLES
# =============================================================================

for path in [
    ACQUISITION_REGISTER_FILE,
    CANONICAL_MAP_FILE,
]:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


acquisition = pd.read_csv(
    ACQUISITION_REGISTER_FILE,
    low_memory=False
)

canonical_map = pd.read_csv(
    CANONICAL_MAP_FILE,
    low_memory=False
)


# =============================================================================
# 3. EXTRACT AUTHORITATIVE BATCH 3 TARGETS
# =============================================================================

batch_ids = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(21, 31)
]


batch_targets = (
    acquisition.loc[
        acquisition[
            "Search_Target_ID"
        ].isin(batch_ids)
    ]
    .copy()
)


batch_targets["_Batch_Order"] = (
    batch_targets[
        "Search_Target_ID"
    ]
    .str.extract(
        r"(\d+)$",
        expand=False
    )
    .astype(int)
)


batch_targets = (
    batch_targets
    .sort_values(
        "_Batch_Order"
    )
    .drop(
        columns="_Batch_Order"
    )
    .reset_index(drop=True)
)


assert len(batch_targets) == 10

assert (
    batch_targets[
        "Search_Target_ID"
    ]
    .tolist()
    == batch_ids
)


print("\nAuthoritative Batch 3 targets")
print("-" * 100)

display(
    batch_targets[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


# =============================================================================
# 4. VERIFY AGAINST THE AUTHORITATIVE TARGET TABLE
# =============================================================================

expected_targets = {
    "OMO_SM_RESCUE_021":
        "Lonchocarpus cilliata",

    "OMO_SM_RESCUE_022":
        "Macaranga barteri",

    "OMO_SM_RESCUE_023":
        "Maesobotrya barteri",

    "OMO_SM_RESCUE_024":
        "Millettia thonningii",

    "OMO_SM_RESCUE_025":
        "Mitragyna ciliata",

    "OMO_SM_RESCUE_026":
        "Musa × paradisiaca",

    "OMO_SM_RESCUE_027":
        "Myrianthus arboreus",

    "OMO_SM_RESCUE_028":
        "Nesogordonia papaverifera",

    "OMO_SM_RESCUE_029":
        "Nothospondias staudtii",

    "OMO_SM_RESCUE_030":
        "Picralima nitida",
}


for target_id, expected_name in expected_targets.items():

    observed = batch_targets.loc[
        batch_targets[
            "Search_Target_ID"
        ].eq(
            target_id
        ),
        "Final_Search_Name"
    ]

    assert len(observed) == 1

    assert observed.iloc[0] == expected_name, (
        f"Unexpected Batch 3 target identity for {target_id}.\n"
        f"Expected : {expected_name}\n"
        f"Observed : {observed.iloc[0]}"
    )


# =============================================================================
# 5. EXTERNAL EVIDENCE ROWS
# =============================================================================

evidence_rows = [

    # -------------------------------------------------------------------------
    # Nesogordonia papaverifera
    #
    # CIRAD technical sheet reports:
    #
    #     25,000 dry seeds per kilogram
    #
    # 1 kg = 1,000,000 mg
    #
    # individual dry seed mass:
    #
    #     1,000,000 / 25,000
    #     = 40 mg
    #
    # -------------------------------------------------------------------------

    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B03_001",

        "Search_Target_ID":
            "OMO_SM_RESCUE_028",

        "Final_Search_Name":
            "Nesogordonia papaverifera",

        "Source_Reported_Taxon":
            "Nesogordonia papaverifera",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "dry seeds per kilogram",

        "Measurement_Entity":
            "dry seed",

        "Measurement_Entity_Class":
            "TRUE_SEED",

        "Reported_Value":
            25000.0,

        "Reported_Unit":
            "dry seeds kg-1",

        "Reported_Seed_Count":
            25000,

        "Converted_Individual_Seed_Mass_mg":
            40.0,

        "Conversion_Formula":
            "1000000 mg kg-1 / 25000 dry seeds kg-1",

        "Conversion_Status":
            "DIRECT_DRY_SEED_COUNT_TO_MASS_CONVERSION",

        "Dry_Mass_Confirmed":
            True,

        "Measurement_Method":
            (
                "Forestry propagation technical sheet; "
                "dry seed count per kilogram"
            ),

        "Sample_Size":
            pd.NA,

        "Population_or_Location":
            "West African forestry context",

        "Study_Year":
            2021,

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "CIRAD — Kotibé technical sheet",

        "Citation":
            (
                "CIRAD technical sheet for "
                "Nesogordonia papaverifera (Kotibé)."
            ),

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "SPECIES_SPECIFIC_DRY_SEED_COUNT_CONVERSION",

        "QC_Rationale":
            (
                "Species-specific forestry source explicitly reports "
                "25,000 dry seeds per kilogram. Conversion to an "
                "individual dry Seed Mass of 40 mg is direct."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,
    },


    # -------------------------------------------------------------------------
    # Picralima nitida
    #
    # Primary physical-properties study:
    #
    # mean individual seed mass = 2.26 g
    # min                       = 1.71 g
    # max                       = 2.67 g
    #
    # corresponding moisture:
    # mean = 47.19% wet basis
    #
    # 1000-seed mass:
    # mean = 1709.14 g
    #
    # This is excellent species-specific measurement evidence but not directly
    # dry-mass comparable.
    # -------------------------------------------------------------------------

    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B03_002",

        "Search_Target_ID":
            "OMO_SM_RESCUE_030",

        "Final_Search_Name":
            "Picralima nitida",

        "Source_Reported_Taxon":
            "Picralima nitida",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "individual seed mass",

        "Measurement_Entity":
            "seed",

        "Measurement_Entity_Class":
            "TRUE_SEED",

        "Reported_Value":
            2.26,

        "Reported_Unit":
            "g per seed",

        "Reported_Seed_Count":
            1,

        "Converted_Individual_Seed_Mass_mg":
            2260.0,

        "Conversion_Formula":
            "2.26 g seed-1 × 1000 mg g-1",

        "Conversion_Status":
            "DIRECT_UNIT_CONVERSION_MOISTURE_REVIEW_REQUIRED",

        "Dry_Mass_Confirmed":
            False,

        "Measurement_Method":
            (
                "Physical-property measurement at measured "
                "seed moisture content"
            ),

        "Sample_Size":
            pd.NA,

        "Population_or_Location":
            "Nigeria",

        "Study_Year":
            2019,

        "Source_Priority":
            2,

        "Source_Class":
            "PRIMARY_PEER_REVIEWED_LITERATURE",

        "Source_Name":
            (
                "Journal of The Institution of Engineers "
                "(India), Series A"
            ),

        "Citation":
            (
                "Influence of Moisture Content and Compression Axis "
                "on Mechanical, Physical, and Phytochemical Properties "
                "of Akuamma (Picralima nitida) Fruits and Seeds."
            ),

        "Evidence_QC_Status":
            "HOLD",

        "Evidence_QC_Flag":
            "DIRECT_SEED_MASS_HIGH_MOISTURE_CONTENT_REVIEW",

        "QC_Rationale":
            (
                "The species-specific individual Seed Mass is directly "
                "measured, but the mean of 2.26 g corresponds to "
                "47.19% wet-basis moisture. The value is held pending "
                "explicit dry-mass normalization review."
            ),

        "Evidence_Action":
            "RETAIN_IN_LEDGER_ONLY",

        "Eligible_For_Species_Aggregation":
            False,
    },
]


batch_evidence = pd.DataFrame(
    evidence_rows
)


# =============================================================================
# 6. ADD PICRALIMA MEASUREMENT CONTEXT
# =============================================================================

batch_evidence[
    "Reported_Moisture_Content_Percent"
] = np.nan

batch_evidence[
    "Reported_Moisture_Basis"
] = pd.NA

batch_evidence[
    "Reported_Value_Min"
] = np.nan

batch_evidence[
    "Reported_Value_Max"
] = np.nan

batch_evidence[
    "Associated_1000_Seed_Mass_g"
] = np.nan


picralima_mask = (
    batch_evidence[
        "Evidence_Record_ID"
    ]
    .eq(
        "OMO_SM_EXT_B03_002"
    )
)


batch_evidence.loc[
    picralima_mask,
    "Reported_Moisture_Content_Percent"
] = 47.19


batch_evidence.loc[
    picralima_mask,
    "Reported_Moisture_Basis"
] = "wet basis"


batch_evidence.loc[
    picralima_mask,
    "Reported_Value_Min"
] = 1.71


batch_evidence.loc[
    picralima_mask,
    "Reported_Value_Max"
] = 2.67


batch_evidence.loc[
    picralima_mask,
    "Associated_1000_Seed_Mass_g"
] = 1709.14


# =============================================================================
# 7. ATTACH CANONICAL TAXON MAP
# =============================================================================

canonical_lookup = (
    canonical_map
    .groupby(
        "Search_Target_ID"
    )[
        "Canonical_Species"
    ]
    .apply(
        lambda x: " | ".join(
            sorted(
                set(
                    x.astype(str)
                )
            )
        )
    )
    .to_dict()
)


batch_evidence[
    "Canonical_Species_Map"
] = (
    batch_evidence[
        "Search_Target_ID"
    ]
    .map(
        canonical_lookup
    )
)


batch_evidence[
    "Trait_Code"
] = "SEED_MASS"


# =============================================================================
# 8. NUMERIC VALIDATION
# =============================================================================

batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
] = pd.to_numeric(
    batch_evidence[
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
)


assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].notna().all()


assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].gt(0).all()


assert np.isclose(
    batch_evidence.loc[
        batch_evidence[
            "Search_Target_ID"
        ].eq(
            "OMO_SM_RESCUE_028"
        ),
        "Converted_Individual_Seed_Mass_mg"
    ].iloc[0],
    40.0
)


assert np.isclose(
    batch_evidence.loc[
        batch_evidence[
            "Search_Target_ID"
        ].eq(
            "OMO_SM_RESCUE_030"
        ),
        "Converted_Individual_Seed_Mass_mg"
    ].iloc[0],
    2260.0
)


# =============================================================================
# 9. BUILD TARGET-LEVEL EVIDENCE STATUS
# =============================================================================

target_status = batch_targets[
    [
        "Search_Target_ID",
        "Final_Search_Name",
        "Omo_Canonical_Taxa",
    ]
].copy()


evidence_counts = (
    batch_evidence
    .groupby(
        "Search_Target_ID"
    )
    .agg(
        N_Candidate_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        N_Accepted_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "ACCEPT").sum()
            )
        ),

        N_Held_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "HOLD").sum()
            )
        ),

        N_Rejected_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "REJECT").sum()
            )
        ),
    )
    .reset_index()
)


target_status = target_status.merge(
    evidence_counts,
    on="Search_Target_ID",
    how="left",
    validate="one_to_one"
)


for col in [
    "N_Candidate_Evidence_Records",
    "N_Accepted_Evidence_Records",
    "N_Held_Evidence_Records",
    "N_Rejected_Evidence_Records",
]:

    target_status[col] = (
        target_status[col]
        .fillna(0)
        .astype(int)
    )


target_status[
    "Batch3_Evidence_Status"
] = "NO_NUMERIC_EVIDENCE_LOCATED"


target_status.loc[
    target_status[
        "N_Held_Evidence_Records"
    ].gt(0),
    "Batch3_Evidence_Status"
] = "NUMERIC_EVIDENCE_HELD_FOR_REVIEW"


target_status.loc[
    target_status[
        "N_Accepted_Evidence_Records"
    ].gt(0),
    "Batch3_Evidence_Status"
] = "ACCEPTED_EXTERNAL_EVIDENCE_AVAILABLE"


# =============================================================================
# 10. NO-EVIDENCE INTERPRETIVE NOTES
# =============================================================================

target_status[
    "No_Evidence_Interpretive_Note"
] = pd.NA


# Musa × paradisiaca is biologically unusual because cultivated forms
# generally have severely reduced/absent viable seed production.
target_status.loc[
    target_status[
        "Final_Search_Name"
    ].eq(
        "Musa × paradisiaca"
    )
    &
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0),
    "No_Evidence_Interpretive_Note"
] = (
    "Cultivated Musa × paradisiaca is typically highly sterile or "
    "seedless; absence of usable species-level Seed Mass evidence may "
    "reflect reproductive biology rather than ordinary search failure."
)


# Maesobotrya paper found, but reported seedling rather than seed mass.
target_status.loc[
    target_status[
        "Final_Search_Name"
    ].eq(
        "Maesobotrya barteri"
    )
    &
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0),
    "No_Evidence_Interpretive_Note"
] = (
    "Published germination study located, but reported fresh/dry "
    "seedling biomass rather than individual Seed Mass; no compatible "
    "numeric Seed Mass retained."
)


ordinary_no_evidence = (
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0)
    &
    target_status[
        "No_Evidence_Interpretive_Note"
    ].isna()
)


target_status.loc[
    ordinary_no_evidence,
    "No_Evidence_Interpretive_Note"
] = (
    "No defensible species-specific numeric Seed Mass evidence "
    "located in Batch 3 search."
)


# =============================================================================
# 11. NO-NUMERIC-EVIDENCE TABLE
# =============================================================================

no_value = target_status[
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0)
].copy()


print("\nBatch 3 evidence disposition")
print("-" * 100)

display(
    target_status
)


print("\nTargets with no usable numeric evidence yet")
print("-" * 100)

display(
    no_value[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
            "No_Evidence_Interpretive_Note",
        ]
    ]
)


# =============================================================================
# 12. QC SUMMARY
# =============================================================================

qc_summary = (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .value_counts()
    .rename_axis(
        "Evidence_QC_Status"
    )
    .reset_index(
        name="N_Evidence_Records"
    )
)


print("\nBatch 3 evidence QC summary")
print("-" * 100)

display(
    qc_summary
)


# =============================================================================
# 13. ACCEPTED EVIDENCE
# =============================================================================

accepted = batch_evidence[
    batch_evidence[
        "Evidence_QC_Status"
    ].eq(
        "ACCEPT"
    )
].copy()


print("\nAccepted Batch 3 external evidence")
print("-" * 100)

display(
    accepted[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Converted_Individual_Seed_Mass_mg",
            "Dry_Mass_Confirmed",
            "Source_Class",
            "Source_Name",
            "Evidence_QC_Flag",
        ]
    ]
)


# =============================================================================
# 14. HELD EVIDENCE
# =============================================================================

held = batch_evidence[
    batch_evidence[
        "Evidence_QC_Status"
    ].eq(
        "HOLD"
    )
].copy()


print("\nHeld Batch 3 external evidence")
print("-" * 100)

display(
    held[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Converted_Individual_Seed_Mass_mg",
            "Reported_Moisture_Content_Percent",
            "Reported_Moisture_Basis",
            "Associated_1000_Seed_Mass_g",
            "Evidence_QC_Flag",
            "QC_Rationale",
        ]
    ]
)


# =============================================================================
# 15. BATCH SUMMARY
# =============================================================================

batch_summary = pd.DataFrame({
    "Metric": [
        "Batch biological search targets",
        "External candidate evidence records",
        "Accepted evidence records",
        "Held evidence records",
        "Rejected evidence records",
        "Targets with accepted external evidence",
        "Targets represented only by held evidence",
        "Targets with no numeric evidence located",
        "Final species Seed Mass values assigned",
    ],

    "Value": [
        len(batch_targets),

        len(batch_evidence),

        int(
            batch_evidence[
                "Evidence_QC_Status"
            ]
            .eq("ACCEPT")
            .sum()
        ),

        int(
            batch_evidence[
                "Evidence_QC_Status"
            ]
            .eq("HOLD")
            .sum()
        ),

        int(
            batch_evidence[
                "Evidence_QC_Status"
            ]
            .eq("REJECT")
            .sum()
        ),

        int(
            target_status[
                "N_Accepted_Evidence_Records"
            ]
            .gt(0)
            .sum()
        ),

        int(
            (
                target_status[
                    "N_Accepted_Evidence_Records"
                ].eq(0)
                &
                target_status[
                    "N_Held_Evidence_Records"
                ].gt(0)
            )
            .sum()
        ),

        len(no_value),

        0,
    ],
})


print("\nSection 9.4G Batch 3 summary")
print("-" * 100)

display(
    batch_summary
)


# =============================================================================
# 16. STRONG VALIDATION
# =============================================================================

assert len(batch_targets) == 10

assert len(batch_evidence) == 2


assert batch_evidence[
    "Evidence_Record_ID"
].is_unique


assert (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .eq("ACCEPT")
    .sum()
    == 1
)


assert (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .eq("HOLD")
    .sum()
    == 1
)


assert (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .eq("REJECT")
    .sum()
    == 0
)


assert len(no_value) == 8


assert set(
    accepted[
        "Final_Search_Name"
    ]
) == {
    "Nesogordonia papaverifera"
}


assert set(
    held[
        "Final_Search_Name"
    ]
) == {
    "Picralima nitida"
}


assert set(
    no_value[
        "Final_Search_Name"
    ]
) == {
    "Lonchocarpus cilliata",
    "Macaranga barteri",
    "Maesobotrya barteri",
    "Millettia thonningii",
    "Mitragyna ciliata",
    "Musa × paradisiaca",
    "Myrianthus arboreus",
    "Nothospondias staudtii",
}


# =============================================================================
# 17. EXPORT
# =============================================================================

batch_evidence.to_csv(
    OUT_BATCH3_EVIDENCE,
    index=False
)

target_status.to_csv(
    OUT_BATCH3_TARGET_STATUS,
    index=False
)

no_value.to_csv(
    OUT_BATCH3_NO_VALUE,
    index=False
)

batch_summary.to_csv(
    OUT_BATCH3_SUMMARY,
    index=False
)


# =============================================================================
# 18. REPORT EXPORTS
# =============================================================================

print("\nBatch 3 exports")
print("-" * 100)

print(
    f"Evidence register : "
    f"{OUT_BATCH3_EVIDENCE}"
)

print(
    f"Target status     : "
    f"{OUT_BATCH3_TARGET_STATUS}"
)

print(
    f"No-value targets  : "
    f"{OUT_BATCH3_NO_VALUE}"
)

print(
    f"Batch summary     : "
    f"{OUT_BATCH3_SUMMARY}"
)


# =============================================================================
# 19. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "External Seed Mass evidence acquisition "
    "Batch 3 completed successfully."
)

print(
    f"Biological targets searched      : "
    f"{len(batch_targets)}"
)

print(
    f"Candidate evidence records       : "
    f"{len(batch_evidence)}"
)

print(
    f"Accepted evidence records        : "
    f"{len(accepted)}"
)

print(
    f"Held evidence records            : "
    f"{len(held)}"
)

print(
    f"Targets without numeric evidence : "
    f"{len(no_value)}"
)

print(
    "No final species-level Seed Mass estimates have been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4H — RESOLVE PICRALIMA NITIDA MOISTURE-NORMALIZATION HOLD
# =============================================================================
#
# Purpose
# -------
# Resolve the single held Batch-3 external Seed Mass record:
#
#     Picralima nitida
#
# Original evidence
# -----------------
# Reported individual seed mass : 2.26 g = 2260 mg
# Reported moisture content      : 47.19% wet basis
#
# For moisture content expressed on a wet basis:
#
#     MC_wb = (Wet mass - Dry mass) / Wet mass
#
# therefore:
#
#     Dry mass = Wet mass × (1 - MC_wb)
#
#     Dry mass = 2260 × (1 - 0.4719)
#              = 1193.506 mg
#
# Decision
# --------
# ACCEPT_AFTER_MOISTURE_NORMALIZATION
#
# The original wet-mass measurement is preserved in the evidence ledger.
# Only the normalized dry-mass-equivalent value is eligible for subsequent
# species-level Seed Mass aggregation.
#
# IMPORTANT
# ---------
# No species-level aggregation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4H — RESOLVE PICRALIMA NITIDA "
    "MOISTURE-NORMALIZATION HOLD"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BATCH3_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_evidence.csv"
)


OUT_RESOLVED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_evidence_resolved.csv"
)

OUT_ACCEPTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_accepted_evidence.csv"
)

OUT_REJECTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_rejected_evidence.csv"
)

OUT_PICRALIMA_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_picralima_nitida_moisture_normalization_review.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_hold_resolution_summary.csv"
)


# =============================================================================
# 2. LOAD BATCH 3 EVIDENCE
# =============================================================================

assert BATCH3_FILE.exists(), (
    f"Batch 3 evidence file not found:\n{BATCH3_FILE}"
)

evidence = pd.read_csv(
    BATCH3_FILE,
    low_memory=False
)


print("\nInput Batch 3 evidence")
print("-" * 100)

print(
    f"Rows : {len(evidence)}"
)

print(
    f"Held records entering review : "
    f"{evidence['Evidence_QC_Status'].eq('HOLD').sum()}"
)


assert len(evidence) == 2, (
    f"Expected 2 Batch 3 evidence records, found {len(evidence)}."
)

assert evidence[
    "Evidence_Record_ID"
].is_unique

assert (
    evidence[
        "Evidence_QC_Status"
    ]
    .eq("HOLD")
    .sum()
    == 1
), (
    "Expected exactly one held Batch 3 evidence record."
)


# =============================================================================
# 3. IDENTIFY PICRALIMA RECORD
# =============================================================================

PICRALIMA_RECORD_ID = (
    "OMO_SM_EXT_B03_002"
)

picralima_mask = (
    evidence[
        "Evidence_Record_ID"
    ]
    .eq(
        PICRALIMA_RECORD_ID
    )
)


assert picralima_mask.sum() == 1, (
    "Expected exactly one Picralima nitida record."
)


picralima = evidence.loc[
    picralima_mask
].iloc[0]


assert picralima[
    "Final_Search_Name"
] == "Picralima nitida"


assert picralima[
    "Evidence_QC_Status"
] == "HOLD"


# =============================================================================
# 4. PRESERVE ORIGINAL MEASUREMENT AND QC STATE
# =============================================================================

evidence[
    "Pre_9_4H_QC_Status"
] = evidence[
    "Evidence_QC_Status"
]


evidence[
    "Pre_9_4H_QC_Flag"
] = evidence[
    "Evidence_QC_Flag"
]


evidence[
    "Original_Wet_Seed_Mass_mg"
] = np.nan


evidence[
    "Moisture_Normalized_Dry_Seed_Mass_mg"
] = np.nan


evidence[
    "Moisture_Normalization_Applied"
] = False


evidence[
    "Moisture_Normalization_Formula"
] = pd.NA


evidence[
    "Moisture_Normalization_Assumption"
] = pd.NA


evidence[
    "Hold_Review_Resolved"
] = False


evidence[
    "Hold_Review_Decision"
] = pd.NA


# =============================================================================
# 5. EXTRACT ORIGINAL MASS AND MOISTURE CONTENT
# =============================================================================

wet_mass_mg = pd.to_numeric(
    evidence.loc[
        picralima_mask,
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
).iloc[0]


moisture_percent = pd.to_numeric(
    evidence.loc[
        picralima_mask,
        "Reported_Moisture_Content_Percent"
    ],
    errors="coerce"
).iloc[0]


moisture_basis = (
    evidence.loc[
        picralima_mask,
        "Reported_Moisture_Basis"
    ]
    .iloc[0]
)


assert pd.notna(wet_mass_mg), (
    "Picralima wet-mass value is missing."
)

assert pd.notna(moisture_percent), (
    "Picralima moisture content is missing."
)

assert wet_mass_mg > 0

assert 0 < moisture_percent < 100

assert str(
    moisture_basis
).strip().lower() == "wet basis", (
    "Moisture conversion formula implemented here applies only "
    "to wet-basis moisture content."
)


# =============================================================================
# 6. CALCULATE DRY-MASS EQUIVALENT
# =============================================================================
#
# Wet-basis moisture fraction:
#
#     MC_wb = moisture_percent / 100
#
# Dry mass:
#
#     Dry = Wet × (1 - MC_wb)
# =============================================================================

moisture_fraction = (
    moisture_percent
    / 100.0
)


dry_mass_mg = (
    wet_mass_mg
    *
    (
        1.0
        -
        moisture_fraction
    )
)


print("\nPicralima nitida moisture-normalization calculation")
print("-" * 100)

print(
    f"Original wet seed mass       : "
    f"{wet_mass_mg:.6f} mg"
)

print(
    f"Reported moisture content    : "
    f"{moisture_percent:.4f}%"
)

print(
    f"Moisture basis               : "
    f"{moisture_basis}"
)

print(
    f"Dry-matter fraction          : "
    f"{1.0 - moisture_fraction:.6f}"
)

print(
    f"Normalized dry Seed Mass     : "
    f"{dry_mass_mg:.6f} mg"
)


# =============================================================================
# 7. NUMERIC VALIDATION
# =============================================================================

EXPECTED_DRY_MASS_MG = (
    1193.506
)


assert np.isclose(
    dry_mass_mg,
    EXPECTED_DRY_MASS_MG,
    rtol=1e-10,
    atol=1e-6
), (
    "Unexpected Picralima dry-mass calculation.\n"
    f"Expected : {EXPECTED_DRY_MASS_MG}\n"
    f"Observed : {dry_mass_mg}"
)


assert dry_mass_mg < wet_mass_mg, (
    "Normalized dry mass must be lower than wet mass."
)


# =============================================================================
# 8. PRESERVE ORIGINAL VALUE AND APPLY NORMALIZATION
# =============================================================================

evidence.loc[
    picralima_mask,
    "Original_Wet_Seed_Mass_mg"
] = wet_mass_mg


evidence.loc[
    picralima_mask,
    "Moisture_Normalized_Dry_Seed_Mass_mg"
] = dry_mass_mg


evidence.loc[
    picralima_mask,
    "Moisture_Normalization_Applied"
] = True


evidence.loc[
    picralima_mask,
    "Moisture_Normalization_Formula"
] = (
    "dry_mass = wet_mass * "
    "(1 - moisture_percent/100)"
)


evidence.loc[
    picralima_mask,
    "Moisture_Normalization_Assumption"
] = (
    "Reported 47.19% wet-basis moisture content is treated as "
    "applicable to the same seed-material condition represented "
    "by the reported mean individual mass of 2.26 g."
)


# =============================================================================
# 9. REPLACE AGGREGATION VALUE WITH NORMALIZED DRY MASS
# =============================================================================
#
# The existing Converted_Individual_Seed_Mass_mg field becomes the
# standardized aggregation-ready value.
#
# Original wet mass remains preserved separately.
# =============================================================================

evidence.loc[
    picralima_mask,
    "Converted_Individual_Seed_Mass_mg"
] = dry_mass_mg


evidence.loc[
    picralima_mask,
    "Conversion_Status"
] = (
    "MOISTURE_NORMALIZED_TO_DRY_MASS_EQUIVALENT"
)


# =============================================================================
# 10. RESOLVE QC HOLD
# =============================================================================

evidence.loc[
    picralima_mask,
    "Evidence_QC_Status"
] = "ACCEPT"


evidence.loc[
    picralima_mask,
    "Evidence_QC_Flag"
] = (
    "PRIMARY_SEED_MASS_ACCEPTED_AFTER_MOISTURE_NORMALIZATION"
)


evidence.loc[
    picralima_mask,
    "QC_Rationale"
] = (
    "Primary species-specific individual Seed Mass was reported "
    "as 2.26 g at 47.19% wet-basis moisture. The original "
    "2260 mg wet-mass value was converted to a dry-mass-equivalent "
    "value using dry = wet × (1 - MC_wb), yielding 1193.506 mg. "
    "The original wet value and moisture context remain preserved "
    "in the evidence ledger."
)


evidence.loc[
    picralima_mask,
    "Evidence_Action"
] = (
    "RETAIN_FOR_SPECIES_AGGREGATION"
)


evidence.loc[
    picralima_mask,
    "Eligible_For_Species_Aggregation"
] = True


evidence.loc[
    picralima_mask,
    "Hold_Review_Resolved"
] = True


evidence.loc[
    picralima_mask,
    "Hold_Review_Decision"
] = (
    "ACCEPT_AFTER_MOISTURE_NORMALIZATION"
)


# =============================================================================
# 11. CONFIRM NESOGORDONIA RECORD REMAINS UNCHANGED
# =============================================================================

nesogordonia_mask = (
    evidence[
        "Final_Search_Name"
    ]
    .eq(
        "Nesogordonia papaverifera"
    )
)


assert nesogordonia_mask.sum() == 1


nesogordonia = evidence.loc[
    nesogordonia_mask
].iloc[0]


assert nesogordonia[
    "Evidence_QC_Status"
] == "ACCEPT"


assert np.isclose(
    pd.to_numeric(
        nesogordonia[
            "Converted_Individual_Seed_Mass_mg"
        ],
        errors="coerce"
    ),
    40.0
), (
    "Previously accepted Nesogordonia value changed unexpectedly."
)


# =============================================================================
# 12. FINAL QC DISPOSITION
# =============================================================================

accepted = evidence[
    evidence[
        "Evidence_QC_Status"
    ]
    .eq(
        "ACCEPT"
    )
].copy()


held = evidence[
    evidence[
        "Evidence_QC_Status"
    ]
    .eq(
        "HOLD"
    )
].copy()


rejected = evidence[
    evidence[
        "Evidence_QC_Status"
    ]
    .eq(
        "REJECT"
    )
].copy()


print("\nResolved Batch 3 QC disposition")
print("-" * 100)

display(
    evidence[
        [
            "Evidence_Record_ID",
            "Search_Target_ID",
            "Final_Search_Name",
            "Original_Wet_Seed_Mass_mg",
            "Reported_Moisture_Content_Percent",
            "Converted_Individual_Seed_Mass_mg",
            "Moisture_Normalization_Applied",
            "Evidence_QC_Status",
            "Evidence_QC_Flag",
            "Evidence_Action",
            "Hold_Review_Decision",
        ]
    ]
)


# =============================================================================
# 13. PICRALIMA REVIEW REGISTER
# =============================================================================

picralima_review = evidence.loc[
    picralima_mask,
    [
        "Evidence_Record_ID",
        "Search_Target_ID",
        "Final_Search_Name",
        "Source_Reported_Taxon",
        "Original_Wet_Seed_Mass_mg",
        "Reported_Moisture_Content_Percent",
        "Reported_Moisture_Basis",
        "Moisture_Normalized_Dry_Seed_Mass_mg",
        "Converted_Individual_Seed_Mass_mg",
        "Moisture_Normalization_Formula",
        "Moisture_Normalization_Assumption",
        "Pre_9_4H_QC_Status",
        "Evidence_QC_Status",
        "Evidence_QC_Flag",
        "QC_Rationale",
        "Evidence_Action",
        "Eligible_For_Species_Aggregation",
        "Hold_Review_Decision",
    ]
].copy()


print("\nPicralima nitida normalization review")
print("-" * 100)

display(
    picralima_review
)


# =============================================================================
# 14. ACCEPTED BATCH 3 SPECIES EVIDENCE SUMMARY
# =============================================================================

accepted_species_summary = (
    accepted
    .groupby(
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        dropna=False
    )
    .agg(
        N_Accepted_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        Min_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "min"
        ),

        Median_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "median"
        ),

        Max_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "max"
        ),
    )
    .reset_index()
)


print("\nAccepted Batch 3 species evidence")
print("-" * 100)

display(
    accepted_species_summary
)


# =============================================================================
# 15. HOLD-RESOLUTION SUMMARY
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Batch 3 candidate evidence records",
        "Records initially accepted",
        "Records initially held",
        "Held records accepted after review",
        "Held records rejected after review",
        "Final accepted evidence records",
        "Final rejected evidence records",
        "Held evidence records remaining",
        "Targets with accepted external evidence",
        "Moisture-normalized accepted records",
    ],

    "Value": [
        len(evidence),

        int(
            evidence[
                "Pre_9_4H_QC_Status"
            ]
            .eq(
                "ACCEPT"
            )
            .sum()
        ),

        int(
            evidence[
                "Pre_9_4H_QC_Status"
            ]
            .eq(
                "HOLD"
            )
            .sum()
        ),

        int(
            (
                evidence[
                    "Pre_9_4H_QC_Status"
                ]
                .eq("HOLD")
                &
                evidence[
                    "Evidence_QC_Status"
                ]
                .eq("ACCEPT")
            )
            .sum()
        ),

        int(
            (
                evidence[
                    "Pre_9_4H_QC_Status"
                ]
                .eq("HOLD")
                &
                evidence[
                    "Evidence_QC_Status"
                ]
                .eq("REJECT")
            )
            .sum()
        ),

        len(accepted),

        len(rejected),

        len(held),

        accepted[
            "Search_Target_ID"
        ]
        .nunique(),

        int(
            evidence[
                "Moisture_Normalization_Applied"
            ]
            .fillna(False)
            .astype(bool)
            .sum()
        ),
    ],
})


print("\nSection 9.4H hold-resolution summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 16. STRONG VALIDATION
# =============================================================================

assert len(evidence) == 2


assert len(accepted) == 2, (
    f"Expected 2 final accepted Batch 3 records, "
    f"found {len(accepted)}."
)


assert len(held) == 0


assert len(rejected) == 0


assert set(
    accepted[
        "Final_Search_Name"
    ]
) == {
    "Nesogordonia papaverifera",
    "Picralima nitida",
}


final_picralima = evidence.loc[
    picralima_mask
].iloc[0]


assert np.isclose(
    pd.to_numeric(
        final_picralima[
            "Converted_Individual_Seed_Mass_mg"
        ],
        errors="coerce"
    ),
    1193.506
)


assert bool(
    final_picralima[
        "Eligible_For_Species_Aggregation"
    ]
)


assert final_picralima[
    "Hold_Review_Decision"
] == (
    "ACCEPT_AFTER_MOISTURE_NORMALIZATION"
)


# =============================================================================
# 17. EXPORT
# =============================================================================

evidence.to_csv(
    OUT_RESOLVED,
    index=False
)


accepted.to_csv(
    OUT_ACCEPTED,
    index=False
)


rejected.to_csv(
    OUT_REJECTED,
    index=False
)


picralima_review.to_csv(
    OUT_PICRALIMA_REVIEW,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 18. REPORT EXPORTS
# =============================================================================

print("\nResolution exports")
print("-" * 100)

print(
    f"Resolved Batch 3 evidence : "
    f"{OUT_RESOLVED}"
)

print(
    f"Accepted evidence         : "
    f"{OUT_ACCEPTED}"
)

print(
    f"Rejected evidence         : "
    f"{OUT_REJECTED}"
)

print(
    f"Picralima review          : "
    f"{OUT_PICRALIMA_REVIEW}"
)

print(
    f"Summary                   : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 19. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Picralima nitida moisture-normalization "
    "review completed successfully."
)

print(
    f"Original wet Seed Mass       : "
    f"{wet_mass_mg:.3f} mg"
)

print(
    f"Moisture content             : "
    f"{moisture_percent:.2f}% wet basis"
)

print(
    f"Normalized dry Seed Mass     : "
    f"{dry_mass_mg:.3f} mg"
)

print(
    "Picralima nitida is now eligible for "
    "species-level Seed Mass aggregation."
)

print(
    f"Final accepted Batch 3 records : "
    f"{len(accepted)}"
)

print(
    f"Held Batch 3 records remaining : "
    f"{len(held)}"
)

print(
    "No final species-level Seed Mass aggregation "
    "has yet been performed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4I — CONFIRM AUTHORITATIVE BATCH 4 TARGETS
# =============================================================================
#
# Purpose
# -------
# Extract the final remaining species-level non-TRY Seed Mass rescue targets:
#
#     OMO_SM_RESCUE_031
#     ...
#     OMO_SM_RESCUE_039
#
# Species identities are inherited directly from the validated acquisition
# register. Nothing is hard-coded.
#
# No external evidence is entered here.
# =============================================================================

from pathlib import Path
import pandas as pd

print("=" * 100)
print(
    "SECTION 9.4I — CONFIRM AUTHORITATIVE "
    "BATCH 4 SEED MASS TARGETS"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

ACQUISITION_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)


# =============================================================================
# 2. LOAD FINAL ACQUISITION REGISTER
# =============================================================================

assert ACQUISITION_REGISTER_FILE.exists(), (
    f"Acquisition register not found:\n"
    f"{ACQUISITION_REGISTER_FILE}"
)

acquisition = pd.read_csv(
    ACQUISITION_REGISTER_FILE,
    low_memory=False
)


# =============================================================================
# 3. DEFINE FINAL BATCH IDS
# =============================================================================

batch4_ids = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(31, 40)
]


batch4_targets = (
    acquisition.loc[
        acquisition[
            "Search_Target_ID"
        ].isin(
            batch4_ids
        )
    ]
    .copy()
)


# =============================================================================
# 4. SORT BY TARGET NUMBER
# =============================================================================

batch4_targets[
    "_Batch_Order"
] = (
    batch4_targets[
        "Search_Target_ID"
    ]
    .str.extract(
        r"(\d+)$",
        expand=False
    )
    .astype(int)
)


batch4_targets = (
    batch4_targets
    .sort_values(
        "_Batch_Order"
    )
    .drop(
        columns="_Batch_Order"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 5. STRUCTURAL VALIDATION
# =============================================================================

assert len(batch4_targets) == 9, (
    f"Expected 9 Batch 4 targets, "
    f"found {len(batch4_targets)}."
)


assert (
    batch4_targets[
        "Search_Target_ID"
    ].tolist()
    == batch4_ids
), (
    "Final Batch 4 target IDs do not correspond exactly "
    "to OMO_SM_RESCUE_031–039."
)


assert (
    batch4_targets[
        "Final_Search_Name"
    ]
    .notna()
    .all()
)


assert (
    batch4_targets[
        "Omo_Canonical_Taxa"
    ]
    .notna()
    .all()
)


# =============================================================================
# 6. DISPLAY AUTHORITATIVE FINAL BATCH
# =============================================================================

print("\nAuthoritative Batch 4 targets")
print("-" * 100)

display(
    batch4_targets[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


print("\nBatch 4 validation")
print("-" * 100)

print(
    f"Final biological targets : "
    f"{len(batch4_targets)}"
)

print(
    f"Unique target IDs        : "
    f"{batch4_targets['Search_Target_ID'].nunique()}"
)

print(
    f"Unique search names      : "
    f"{batch4_targets['Final_Search_Name'].nunique()}"
)


print("\n" + "=" * 100)

print(
    "Authoritative Batch 4 target extraction "
    "completed successfully."
)

print(
    "No external Seed Mass evidence has yet been assigned."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4J — EXTERNAL SEED MASS EVIDENCE ACQUISITION: BATCH 4
# =============================================================================
#
# Authoritative Batch 4 targets
# -----------------------------
#
# OMO_SM_RESCUE_031  Pterocarpus osun
# OMO_SM_RESCUE_032  Rinorea dentata
# OMO_SM_RESCUE_033  Rothmannia hispida
# OMO_SM_RESCUE_034  Scottellia coriacea
# OMO_SM_RESCUE_035  Sterculia oblonga
# OMO_SM_RESCUE_036  Sterculia tragacantha
# OMO_SM_RESCUE_037  Strombosia pustulata
# OMO_SM_RESCUE_038  Trichilia gilgiana
# OMO_SM_RESCUE_039  Uapaca heudelotii
#
# Current defensible quantitative evidence
# -----------------------------------------
#
# Sterculia oblonga
#     PROTA: about 1100 g per 1000 seeds
#     = 1100 mg seed-1
#     ACCEPT
#
# Strombosia pustulata
#     PROTA: about 800 fruit stones per kg
#     = 1250 mg per stone
#     HOLD — stone/endocarp vs true-seed comparability
#
# No defensible numeric Seed Mass evidence yet
# ---------------------------------------------
# Pterocarpus osun
# Rinorea dentata
# Rothmannia hispida
# Scottellia coriacea
# Sterculia tragacantha
# Trichilia gilgiana
# Uapaca heudelotii
#
# IMPORTANT
# ---------
# No final species-level aggregation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4J — EXTERNAL SEED MASS "
    "EVIDENCE ACQUISITION: BATCH 4"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

ACQUISITION_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)

CANONICAL_MAP_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)

OUT_BATCH4_EVIDENCE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_evidence.csv"
)

OUT_BATCH4_TARGET_STATUS = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_target_status.csv"
)

OUT_BATCH4_NO_VALUE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_no_numeric_evidence.csv"
)

OUT_BATCH4_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_summary.csv"
)


# =============================================================================
# 2. LOAD CONTROL TABLES
# =============================================================================

for path in [
    ACQUISITION_REGISTER_FILE,
    CANONICAL_MAP_FILE,
]:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


acquisition = pd.read_csv(
    ACQUISITION_REGISTER_FILE,
    low_memory=False
)

canonical_map = pd.read_csv(
    CANONICAL_MAP_FILE,
    low_memory=False
)


# =============================================================================
# 3. EXTRACT AUTHORITATIVE BATCH 4 TARGETS
# =============================================================================

batch_ids = [
    f"OMO_SM_RESCUE_{i:03d}"
    for i in range(31, 40)
]


batch_targets = (
    acquisition.loc[
        acquisition[
            "Search_Target_ID"
        ].isin(batch_ids)
    ]
    .copy()
)


batch_targets["_Batch_Order"] = (
    batch_targets[
        "Search_Target_ID"
    ]
    .str.extract(
        r"(\d+)$",
        expand=False
    )
    .astype(int)
)


batch_targets = (
    batch_targets
    .sort_values(
        "_Batch_Order"
    )
    .drop(
        columns="_Batch_Order"
    )
    .reset_index(drop=True)
)


assert len(batch_targets) == 9

assert (
    batch_targets[
        "Search_Target_ID"
    ].tolist()
    == batch_ids
)


print("\nAuthoritative Batch 4 targets")
print("-" * 100)

display(
    batch_targets[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
        ]
    ]
)


# =============================================================================
# 4. VERIFY AUTHORITATIVE TARGET IDENTITIES
# =============================================================================

expected_targets = {
    "OMO_SM_RESCUE_031":
        "Pterocarpus osun",

    "OMO_SM_RESCUE_032":
        "Rinorea dentata",

    "OMO_SM_RESCUE_033":
        "Rothmannia hispida",

    "OMO_SM_RESCUE_034":
        "Scottellia coriacea",

    "OMO_SM_RESCUE_035":
        "Sterculia oblonga",

    "OMO_SM_RESCUE_036":
        "Sterculia tragacantha",

    "OMO_SM_RESCUE_037":
        "Strombosia pustulata",

    "OMO_SM_RESCUE_038":
        "Trichilia gilgiana",

    "OMO_SM_RESCUE_039":
        "Uapaca heudelotii",
}


for target_id, expected_name in expected_targets.items():

    observed = batch_targets.loc[
        batch_targets[
            "Search_Target_ID"
        ].eq(
            target_id
        ),
        "Final_Search_Name"
    ]

    assert len(observed) == 1

    assert observed.iloc[0] == expected_name, (
        f"Unexpected Batch 4 target identity for {target_id}.\n"
        f"Expected : {expected_name}\n"
        f"Observed : {observed.iloc[0]}"
    )


# =============================================================================
# 5. EXTERNAL EVIDENCE ROWS
# =============================================================================

evidence_rows = [

    # -------------------------------------------------------------------------
    # Sterculia oblonga
    #
    # PROTA:
    #     1000-seed weight ≈ 1100 g
    #
    # Therefore:
    #     1100 g / 1000 seeds
    #     = 1.1 g seed-1
    #     = 1100 mg seed-1
    #
    # -------------------------------------------------------------------------

    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B04_001",

        "Search_Target_ID":
            "OMO_SM_RESCUE_035",

        "Final_Search_Name":
            "Sterculia oblonga",

        "Source_Reported_Taxon":
            "Sterculia oblonga",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "1000-seed weight",

        "Measurement_Entity":
            "1000 seeds",

        "Measurement_Entity_Class":
            "1000_SEED_WEIGHT",

        "Reported_Value":
            1100.0,

        "Reported_Unit":
            "g per 1000 seeds",

        "Reported_Seed_Count":
            1000,

        "Converted_Individual_Seed_Mass_mg":
            1100.0,

        "Conversion_Formula":
            "1100 g / 1000 seeds = 1100 mg seed-1",

        "Conversion_Status":
            "DIRECT_1000_SEED_WEIGHT_CONVERSION",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "species-specific propagation compilation",

        "Sample_Size":
            pd.NA,

        "Population_or_Location":
            pd.NA,

        "Study_Year":
            2008,

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "PROTA / Plant Resources of Tropical Africa",

        "Citation":
            "PROTA treatment of Sterculia oblonga",

        "Evidence_QC_Status":
            "ACCEPT",

        "Evidence_QC_Flag":
            "SPECIES_SPECIFIC_1000_SEED_WEIGHT",

        "QC_Rationale":
            (
                "PROTA explicitly reports an approximate 1000-seed "
                "weight of 1100 g for Sterculia oblonga. The conversion "
                "to an individual Seed Mass of 1100 mg is direct."
            ),

        "Evidence_Action":
            "RETAIN_FOR_SPECIES_AGGREGATION",

        "Eligible_For_Species_Aggregation":
            True,
    },


    # -------------------------------------------------------------------------
    # Strombosia pustulata
    #
    # PROTA:
    #     about 800 fruit stones per kg
    #
    # Therefore:
    #     1,000,000 mg / 800
    #     = 1250 mg per stone
    #
    # HOWEVER:
    # the reported unit is a fruit stone/endocarp enclosing one seed,
    # not an explicitly isolated seed.
    #
    # Therefore HOLD.
    # -------------------------------------------------------------------------

    {
        "Evidence_Record_ID":
            "OMO_SM_EXT_B04_002",

        "Search_Target_ID":
            "OMO_SM_RESCUE_037",

        "Final_Search_Name":
            "Strombosia pustulata",

        "Source_Reported_Taxon":
            "Strombosia pustulata",

        "Taxonomic_Match_Status":
            "EXACT_SPECIES_MATCH",

        "Reported_Trait_Name":
            "fruit stones per kilogram",

        "Measurement_Entity":
            "one-seeded fruit stone",

        "Measurement_Entity_Class":
            "ENDOCARP_OR_STONE",

        "Reported_Value":
            800.0,

        "Reported_Unit":
            "fruit stones kg-1",

        "Reported_Seed_Count":
            800,

        "Converted_Individual_Seed_Mass_mg":
            1250.0,

        "Conversion_Formula":
            "1000000 mg kg-1 / 800 fruit stones kg-1",

        "Conversion_Status":
            "DIRECT_DISPERSAL_UNIT_MASS_CONVERSION_DEFINITION_REVIEW",

        "Dry_Mass_Confirmed":
            pd.NA,

        "Measurement_Method":
            "species propagation compilation",

        "Sample_Size":
            pd.NA,

        "Population_or_Location":
            "African tropical forest",

        "Study_Year":
            2008,

        "Source_Priority":
            4,

        "Source_Class":
            "TRACEABLE_SECONDARY_COMPILATION",

        "Source_Name":
            "PROTA / Plant Resources of Tropical Africa",

        "Citation":
            "PROTA treatment of Strombosia pustulata",

        "Evidence_QC_Status":
            "HOLD",

        "Evidence_QC_Flag":
            "SEED_VS_ENDOCARP_DISPERSAL_UNIT_AMBIGUITY",

        "QC_Rationale":
            (
                "PROTA reports approximately 800 fruit stones per kg. "
                "The fruit is a drupe with a one-seeded stone, so the "
                "1250 mg conversion represents a stone/endocarp dispersal "
                "unit rather than unequivocally isolated Seed Mass. "
                "The record is held for definition review."
            ),

        "Evidence_Action":
            "RETAIN_IN_LEDGER_ONLY",

        "Eligible_For_Species_Aggregation":
            False,
    },
]


batch_evidence = pd.DataFrame(
    evidence_rows
)


# =============================================================================
# 6. ATTACH CANONICAL TAXON MAP
# =============================================================================

canonical_lookup = (
    canonical_map
    .groupby(
        "Search_Target_ID"
    )[
        "Canonical_Species"
    ]
    .apply(
        lambda x: " | ".join(
            sorted(
                set(
                    x.astype(str)
                )
            )
        )
    )
    .to_dict()
)


batch_evidence[
    "Canonical_Species_Map"
] = (
    batch_evidence[
        "Search_Target_ID"
    ]
    .map(
        canonical_lookup
    )
)


batch_evidence[
    "Trait_Code"
] = "SEED_MASS"


# =============================================================================
# 7. NUMERIC VALIDATION
# =============================================================================

batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
] = pd.to_numeric(
    batch_evidence[
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
)


assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].notna().all()


assert batch_evidence[
    "Converted_Individual_Seed_Mass_mg"
].gt(0).all()


assert np.isclose(
    batch_evidence.loc[
        batch_evidence[
            "Search_Target_ID"
        ].eq(
            "OMO_SM_RESCUE_035"
        ),
        "Converted_Individual_Seed_Mass_mg"
    ].iloc[0],
    1100.0
)


assert np.isclose(
    batch_evidence.loc[
        batch_evidence[
            "Search_Target_ID"
        ].eq(
            "OMO_SM_RESCUE_037"
        ),
        "Converted_Individual_Seed_Mass_mg"
    ].iloc[0],
    1250.0
)


# =============================================================================
# 8. BUILD TARGET-LEVEL STATUS
# =============================================================================

target_status = batch_targets[
    [
        "Search_Target_ID",
        "Final_Search_Name",
        "Omo_Canonical_Taxa",
    ]
].copy()


evidence_counts = (
    batch_evidence
    .groupby(
        "Search_Target_ID"
    )
    .agg(
        N_Candidate_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        N_Accepted_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "ACCEPT").sum()
            )
        ),

        N_Held_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "HOLD").sum()
            )
        ),

        N_Rejected_Evidence_Records=(
            "Evidence_QC_Status",
            lambda x: int(
                (x == "REJECT").sum()
            )
        ),
    )
    .reset_index()
)


target_status = target_status.merge(
    evidence_counts,
    on="Search_Target_ID",
    how="left",
    validate="one_to_one"
)


for col in [
    "N_Candidate_Evidence_Records",
    "N_Accepted_Evidence_Records",
    "N_Held_Evidence_Records",
    "N_Rejected_Evidence_Records",
]:

    target_status[col] = (
        target_status[col]
        .fillna(0)
        .astype(int)
    )


target_status[
    "Batch4_Evidence_Status"
] = "NO_NUMERIC_EVIDENCE_LOCATED"


target_status.loc[
    target_status[
        "N_Held_Evidence_Records"
    ].gt(0),
    "Batch4_Evidence_Status"
] = "NUMERIC_EVIDENCE_HELD_FOR_REVIEW"


target_status.loc[
    target_status[
        "N_Accepted_Evidence_Records"
    ].gt(0),
    "Batch4_Evidence_Status"
] = "ACCEPTED_EXTERNAL_EVIDENCE_AVAILABLE"


# =============================================================================
# 9. NO-EVIDENCE INTERPRETIVE NOTES
# =============================================================================

target_status[
    "No_Evidence_Interpretive_Note"
] = pd.NA


# Pterocarpus osun: seed-composition evidence exists, but not mass.
target_status.loc[
    target_status[
        "Final_Search_Name"
    ].eq(
        "Pterocarpus osun"
    )
    &
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0),
    "No_Evidence_Interpretive_Note"
] = (
    "Published Nigerian seed-composition evidence located, "
    "but no compatible individual or multi-seed mass value "
    "was reported."
)


# Trichilia gilgiana: seed dimensions available, not mass.
target_status.loc[
    target_status[
        "Final_Search_Name"
    ].eq(
        "Trichilia gilgiana"
    )
    &
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0),
    "No_Evidence_Interpretive_Note"
] = (
    "PROTA reports seed morphology and dimensions but no numeric "
    "Seed Mass value compatible with the trait definition."
)


ordinary_no_evidence = (
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0)
    &
    target_status[
        "No_Evidence_Interpretive_Note"
    ].isna()
)


target_status.loc[
    ordinary_no_evidence,
    "No_Evidence_Interpretive_Note"
] = (
    "No defensible species-specific numeric Seed Mass evidence "
    "located in Batch 4 search."
)


# =============================================================================
# 10. NO-NUMERIC-EVIDENCE REGISTER
# =============================================================================

no_value = target_status[
    target_status[
        "N_Candidate_Evidence_Records"
    ].eq(0)
].copy()


print("\nBatch 4 evidence disposition")
print("-" * 100)

display(
    target_status
)


print("\nTargets with no usable numeric evidence yet")
print("-" * 100)

display(
    no_value[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
            "No_Evidence_Interpretive_Note",
        ]
    ]
)


# =============================================================================
# 11. QC SUMMARY
# =============================================================================

qc_summary = (
    batch_evidence[
        "Evidence_QC_Status"
    ]
    .value_counts()
    .rename_axis(
        "Evidence_QC_Status"
    )
    .reset_index(
        name="N_Evidence_Records"
    )
)


print("\nBatch 4 evidence QC summary")
print("-" * 100)

display(
    qc_summary
)


# =============================================================================
# 12. ACCEPTED EVIDENCE
# =============================================================================

accepted = batch_evidence[
    batch_evidence[
        "Evidence_QC_Status"
    ].eq(
        "ACCEPT"
    )
].copy()


print("\nAccepted Batch 4 external evidence")
print("-" * 100)

display(
    accepted[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Converted_Individual_Seed_Mass_mg",
            "Source_Class",
            "Source_Name",
            "Evidence_QC_Flag",
        ]
    ]
)


# =============================================================================
# 13. HELD EVIDENCE
# =============================================================================

held = batch_evidence[
    batch_evidence[
        "Evidence_QC_Status"
    ].eq(
        "HOLD"
    )
].copy()


print("\nHeld Batch 4 external evidence")
print("-" * 100)

display(
    held[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Measurement_Entity",
            "Measurement_Entity_Class",
            "Converted_Individual_Seed_Mass_mg",
            "Evidence_QC_Flag",
            "QC_Rationale",
        ]
    ]
)


# =============================================================================
# 14. BATCH SUMMARY
# =============================================================================

batch_summary = pd.DataFrame({
    "Metric": [
        "Batch biological search targets",
        "External candidate evidence records",
        "Accepted evidence records",
        "Held evidence records",
        "Rejected evidence records",
        "Targets with accepted external evidence",
        "Targets represented only by held evidence",
        "Targets with no numeric evidence located",
        "Final species Seed Mass values assigned",
    ],

    "Value": [
        len(batch_targets),

        len(batch_evidence),

        int(
            batch_evidence[
                "Evidence_QC_Status"
            ].eq("ACCEPT").sum()
        ),

        int(
            batch_evidence[
                "Evidence_QC_Status"
            ].eq("HOLD").sum()
        ),

        int(
            batch_evidence[
                "Evidence_QC_Status"
            ].eq("REJECT").sum()
        ),

        int(
            target_status[
                "N_Accepted_Evidence_Records"
            ].gt(0).sum()
        ),

        int(
            (
                target_status[
                    "N_Accepted_Evidence_Records"
                ].eq(0)
                &
                target_status[
                    "N_Held_Evidence_Records"
                ].gt(0)
            ).sum()
        ),

        len(no_value),

        0,
    ],
})


print("\nSection 9.4J Batch 4 summary")
print("-" * 100)

display(
    batch_summary
)


# =============================================================================
# 15. STRONG VALIDATION
# =============================================================================

assert len(batch_targets) == 9

assert len(batch_evidence) == 2

assert batch_evidence[
    "Evidence_Record_ID"
].is_unique


assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("ACCEPT").sum()
    == 1
)


assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("HOLD").sum()
    == 1
)


assert (
    batch_evidence[
        "Evidence_QC_Status"
    ].eq("REJECT").sum()
    == 0
)


assert len(no_value) == 7


assert set(
    accepted[
        "Final_Search_Name"
    ]
) == {
    "Sterculia oblonga"
}


assert set(
    held[
        "Final_Search_Name"
    ]
) == {
    "Strombosia pustulata"
}


assert set(
    no_value[
        "Final_Search_Name"
    ]
) == {
    "Pterocarpus osun",
    "Rinorea dentata",
    "Rothmannia hispida",
    "Scottellia coriacea",
    "Sterculia tragacantha",
    "Trichilia gilgiana",
    "Uapaca heudelotii",
}


# =============================================================================
# 16. EXPORT
# =============================================================================

batch_evidence.to_csv(
    OUT_BATCH4_EVIDENCE,
    index=False
)

target_status.to_csv(
    OUT_BATCH4_TARGET_STATUS,
    index=False
)

no_value.to_csv(
    OUT_BATCH4_NO_VALUE,
    index=False
)

batch_summary.to_csv(
    OUT_BATCH4_SUMMARY,
    index=False
)


# =============================================================================
# 17. REPORT EXPORTS
# =============================================================================

print("\nBatch 4 exports")
print("-" * 100)

print(
    f"Evidence register : "
    f"{OUT_BATCH4_EVIDENCE}"
)

print(
    f"Target status     : "
    f"{OUT_BATCH4_TARGET_STATUS}"
)

print(
    f"No-value targets  : "
    f"{OUT_BATCH4_NO_VALUE}"
)

print(
    f"Batch summary     : "
    f"{OUT_BATCH4_SUMMARY}"
)


# =============================================================================
# 18. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "External Seed Mass evidence acquisition "
    "Batch 4 completed successfully."
)

print(
    f"Biological targets searched      : "
    f"{len(batch_targets)}"
)

print(
    f"Candidate evidence records       : "
    f"{len(batch_evidence)}"
)

print(
    f"Accepted evidence records        : "
    f"{len(accepted)}"
)

print(
    f"Held evidence records            : "
    f"{len(held)}"
)

print(
    f"Targets without numeric evidence : "
    f"{len(no_value)}"
)

print(
    "No final species-level Seed Mass estimates have been calculated."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4K — RESOLVE STROMBOSIA PUSTULATA SEED-vs-STONE DEFINITION HOLD
# =============================================================================
#
# Purpose
# -------
# Resolve the single held Batch-4 external Seed Mass record:
#
#     Strombosia pustulata
#
# Original evidence
# -----------------
# Reported entity       : one-seeded fruit stone
# Reported count        : approximately 800 stones per kg
# Converted equivalent : 1,000,000 mg / 800
#                      = 1250 mg per stone
#
# Trait-definition issue
# ----------------------
# The measured unit is a fruit stone/endocarp enclosing the seed.
# It is therefore not directly comparable with standardized individual
# true-seed mass.
#
# Decision
# --------
# EXCLUDE_FROM_STANDARDIZED_SEED_MASS
#
# The record remains preserved in the evidence ledger for provenance,
# but is not eligible for species-level Seed Mass aggregation.
#
# No final species-level aggregation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4K — RESOLVE STROMBOSIA PUSTULATA "
    "SEED-vs-STONE DEFINITION HOLD"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BATCH4_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_evidence.csv"
)

OUT_RESOLVED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_evidence_resolved.csv"
)

OUT_ACCEPTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_accepted_evidence.csv"
)

OUT_REJECTED = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_rejected_evidence.csv"
)

OUT_STROMBOSIA_REVIEW = (
    OUTPUT_DIR /
    "omo_seed_mass_strombosia_pustulata_definition_review.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_hold_resolution_summary.csv"
)


# =============================================================================
# 2. LOAD BATCH 4 EVIDENCE
# =============================================================================

assert BATCH4_FILE.exists(), (
    f"Batch 4 evidence file not found:\n{BATCH4_FILE}"
)

evidence = pd.read_csv(
    BATCH4_FILE,
    low_memory=False
)

print("\nInput Batch 4 evidence")
print("-" * 100)

print(
    f"Rows : {len(evidence)}"
)

print(
    f"Held records entering review : "
    f"{evidence['Evidence_QC_Status'].eq('HOLD').sum()}"
)

assert len(evidence) == 2, (
    f"Expected 2 Batch 4 evidence records, found {len(evidence)}."
)

assert evidence[
    "Evidence_Record_ID"
].is_unique

assert (
    evidence[
        "Evidence_QC_Status"
    ].eq("HOLD").sum()
    == 1
), (
    "Expected exactly one held Batch 4 record."
)


# =============================================================================
# 3. IDENTIFY STROMBOSIA RECORD
# =============================================================================

STROMBOSIA_RECORD_ID = (
    "OMO_SM_EXT_B04_002"
)

strombosia_mask = (
    evidence[
        "Evidence_Record_ID"
    ].eq(
        STROMBOSIA_RECORD_ID
    )
)

assert strombosia_mask.sum() == 1, (
    "Expected exactly one Strombosia pustulata record."
)

strombosia = evidence.loc[
    strombosia_mask
].iloc[0]

assert strombosia[
    "Final_Search_Name"
] == "Strombosia pustulata"

assert strombosia[
    "Evidence_QC_Status"
] == "HOLD"

assert strombosia[
    "Measurement_Entity_Class"
] == "ENDOCARP_OR_STONE"


# =============================================================================
# 4. PRESERVE PRE-REVIEW STATE
# =============================================================================

evidence[
    "Pre_9_4K_QC_Status"
] = evidence[
    "Evidence_QC_Status"
]

evidence[
    "Pre_9_4K_QC_Flag"
] = evidence[
    "Evidence_QC_Flag"
]

evidence[
    "Definition_Review_Resolved"
] = False

evidence[
    "Definition_Review_Decision"
] = pd.NA

evidence[
    "Measurement_Entity_Compatibility"
] = pd.NA


# =============================================================================
# 5. VALIDATE STONE-EQUIVALENT NUMERIC CONVERSION
# =============================================================================

stone_mass_mg = pd.to_numeric(
    evidence.loc[
        strombosia_mask,
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
).iloc[0]

assert pd.notna(stone_mass_mg)

assert np.isclose(
    stone_mass_mg,
    1250.0
), (
    f"Unexpected Strombosia stone-equivalent mass: {stone_mass_mg}"
)


# =============================================================================
# 6. APPLY DEFINITION-REVIEW DECISION
# =============================================================================

evidence.loc[
    strombosia_mask,
    "Evidence_QC_Status"
] = "REJECT"

evidence.loc[
    strombosia_mask,
    "Evidence_QC_Flag"
] = (
    "STONE_ENDOCARP_NOT_STANDARDIZED_TRUE_SEED_MASS"
)

evidence.loc[
    strombosia_mask,
    "QC_Rationale"
] = (
    "The reported approximately 800 units per kilogram refers to "
    "one-seeded fruit stones/endocarps rather than isolated true seeds. "
    "Although the numerical conversion to 1250 mg per stone is valid, "
    "the measurement entity is not directly comparable with standardized "
    "individual Seed Mass. The record is therefore preserved in the "
    "evidence ledger but excluded from trait aggregation."
)

evidence.loc[
    strombosia_mask,
    "Evidence_Action"
] = "RETAIN_IN_LEDGER_ONLY"

evidence.loc[
    strombosia_mask,
    "Eligible_For_Species_Aggregation"
] = False

evidence.loc[
    strombosia_mask,
    "Definition_Review_Resolved"
] = True

evidence.loc[
    strombosia_mask,
    "Definition_Review_Decision"
] = (
    "EXCLUDE_FROM_STANDARDIZED_SEED_MASS"
)

evidence.loc[
    strombosia_mask,
    "Measurement_Entity_Compatibility"
] = (
    "NONCOMPARABLE_ENDOCARP_OR_STONE"
)


# =============================================================================
# 7. CONFIRM STERCULIA RECORD REMAINS ACCEPTED
# =============================================================================

sterculia_mask = (
    evidence[
        "Final_Search_Name"
    ].eq(
        "Sterculia oblonga"
    )
)

assert sterculia_mask.sum() == 1

sterculia = evidence.loc[
    sterculia_mask
].iloc[0]

assert sterculia[
    "Evidence_QC_Status"
] == "ACCEPT"

assert np.isclose(
    pd.to_numeric(
        sterculia[
            "Converted_Individual_Seed_Mass_mg"
        ],
        errors="coerce"
    ),
    1100.0
)


# =============================================================================
# 8. FINAL QC DISPOSITION
# =============================================================================

accepted = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq(
        "ACCEPT"
    )
].copy()

rejected = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq(
        "REJECT"
    )
].copy()

held = evidence[
    evidence[
        "Evidence_QC_Status"
    ].eq(
        "HOLD"
    )
].copy()


print("\nResolved Batch 4 QC disposition")
print("-" * 100)

display(
    evidence[
        [
            "Evidence_Record_ID",
            "Search_Target_ID",
            "Final_Search_Name",
            "Measurement_Entity",
            "Measurement_Entity_Class",
            "Converted_Individual_Seed_Mass_mg",
            "Evidence_QC_Status",
            "Evidence_QC_Flag",
            "Evidence_Action",
            "Definition_Review_Decision",
        ]
    ]
)


# =============================================================================
# 9. STROMBOSIA DEFINITION REVIEW REGISTER
# =============================================================================

strombosia_review = evidence.loc[
    strombosia_mask,
    [
        "Evidence_Record_ID",
        "Search_Target_ID",
        "Final_Search_Name",
        "Source_Reported_Taxon",
        "Reported_Trait_Name",
        "Measurement_Entity",
        "Measurement_Entity_Class",
        "Reported_Value",
        "Reported_Unit",
        "Converted_Individual_Seed_Mass_mg",
        "Pre_9_4K_QC_Status",
        "Evidence_QC_Status",
        "Evidence_QC_Flag",
        "QC_Rationale",
        "Measurement_Entity_Compatibility",
        "Evidence_Action",
        "Eligible_For_Species_Aggregation",
        "Definition_Review_Decision",
    ]
].copy()


print("\nStrombosia pustulata definition review")
print("-" * 100)

display(
    strombosia_review
)


# =============================================================================
# 10. ACCEPTED BATCH 4 SPECIES EVIDENCE
# =============================================================================

accepted_species_summary = (
    accepted
    .groupby(
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        dropna=False
    )
    .agg(
        N_Accepted_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        Min_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "min"
        ),

        Median_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "median"
        ),

        Max_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "max"
        ),
    )
    .reset_index()
)


print("\nAccepted Batch 4 species evidence")
print("-" * 100)

display(
    accepted_species_summary
)


# =============================================================================
# 11. HOLD-RESOLUTION SUMMARY
# =============================================================================

summary = pd.DataFrame({
    "Metric": [
        "Batch 4 candidate evidence records",
        "Records initially accepted",
        "Records initially held",
        "Held records accepted after review",
        "Held records rejected after review",
        "Final accepted evidence records",
        "Final rejected evidence records",
        "Held evidence records remaining",
        "Targets with accepted external evidence",
        "Definition-incompatible records excluded",
    ],

    "Value": [
        len(evidence),

        int(
            evidence[
                "Pre_9_4K_QC_Status"
            ].eq(
                "ACCEPT"
            ).sum()
        ),

        int(
            evidence[
                "Pre_9_4K_QC_Status"
            ].eq(
                "HOLD"
            ).sum()
        ),

        int(
            (
                evidence[
                    "Pre_9_4K_QC_Status"
                ].eq("HOLD")
                &
                evidence[
                    "Evidence_QC_Status"
                ].eq("ACCEPT")
            ).sum()
        ),

        int(
            (
                evidence[
                    "Pre_9_4K_QC_Status"
                ].eq("HOLD")
                &
                evidence[
                    "Evidence_QC_Status"
                ].eq("REJECT")
            ).sum()
        ),

        len(accepted),

        len(rejected),

        len(held),

        accepted[
            "Search_Target_ID"
        ].nunique(),

        int(
            evidence[
                "Definition_Review_Decision"
            ]
            .eq(
                "EXCLUDE_FROM_STANDARDIZED_SEED_MASS"
            )
            .sum()
        ),
    ],
})


print("\nSection 9.4K hold-resolution summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 12. STRONG VALIDATION
# =============================================================================

assert len(evidence) == 2

assert len(accepted) == 1, (
    f"Expected 1 accepted Batch 4 record, found {len(accepted)}."
)

assert len(rejected) == 1, (
    f"Expected 1 rejected Batch 4 record, found {len(rejected)}."
)

assert len(held) == 0

assert set(
    accepted[
        "Final_Search_Name"
    ]
) == {
    "Sterculia oblonga"
}

assert set(
    rejected[
        "Final_Search_Name"
    ]
) == {
    "Strombosia pustulata"
}

assert not bool(
    rejected[
        "Eligible_For_Species_Aggregation"
    ]
    .astype(bool)
    .any()
)

assert (
    rejected[
        "Definition_Review_Decision"
    ]
    .eq(
        "EXCLUDE_FROM_STANDARDIZED_SEED_MASS"
    )
    .all()
)


# =============================================================================
# 13. EXPORT
# =============================================================================

evidence.to_csv(
    OUT_RESOLVED,
    index=False
)

accepted.to_csv(
    OUT_ACCEPTED,
    index=False
)

rejected.to_csv(
    OUT_REJECTED,
    index=False
)

strombosia_review.to_csv(
    OUT_STROMBOSIA_REVIEW,
    index=False
)

summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 14. REPORT EXPORTS
# =============================================================================

print("\nResolution exports")
print("-" * 100)

print(
    f"Resolved Batch 4 evidence : "
    f"{OUT_RESOLVED}"
)

print(
    f"Accepted evidence         : "
    f"{OUT_ACCEPTED}"
)

print(
    f"Rejected evidence         : "
    f"{OUT_REJECTED}"
)

print(
    f"Strombosia review         : "
    f"{OUT_STROMBOSIA_REVIEW}"
)

print(
    f"Summary                   : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 15. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Strombosia pustulata seed-vs-stone "
    "definition review completed successfully."
)

print(
    f"Stone-equivalent mass preserved : "
    f"{stone_mass_mg:.3f} mg"
)

print(
    "The stone/endocarp value is excluded from standardized "
    "true-seed mass aggregation."
)

print(
    f"Final accepted Batch 4 records : "
    f"{len(accepted)}"
)

print(
    f"Final rejected Batch 4 records : "
    f"{len(rejected)}"
)

print(
    f"Held Batch 4 records remaining : "
    f"{len(held)}"
)

print(
    "No final species-level Seed Mass aggregation "
    "has yet been performed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4L — CONSOLIDATE EXTERNAL SPECIES-LEVEL SEED MASS RESCUE
#                AND DETERMINE TRUE RESIDUAL GAP
# =============================================================================
#
# Purpose
# -------
# Consolidate all QC-resolved species-level external Seed Mass evidence from
# Sections 9.4C–9.4K.
#
# This section:
#
#   1. loads accepted evidence from external rescue Batches 1–4;
#   2. validates that no HOLD evidence enters consolidation;
#   3. combines all accepted measurement-level records;
#   4. aggregates accepted measurements by biological search target;
#   5. uses the MEDIAN where >1 accepted measurement exists;
#   6. preserves source and QC provenance;
#   7. maps target-level values back to Omo canonical taxa;
#   8. determines unresolved species-level rescue taxa;
#   9. restores the separately deferred genus-level targets;
#  10. reconstructs post-BIEN/TRY/external-rescue coverage;
#  11. prepares the true residual set for the next acquisition stage.
#
# Expected external-rescue structure
# ----------------------------------
# Biological search targets resolved : 11
#
# Important:
# Some biological search targets map to more than one Omo canonical identifier.
# Therefore the number of newly resolved CANONICAL taxa will be >11.
#
# No genus-level estimation is performed here.
# No family-level estimation is performed here.
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 100)
print(
    "SECTION 9.4L — CONSOLIDATE EXTERNAL SPECIES-LEVEL "
    "SEED MASS RESCUE AND DETERMINE TRUE RESIDUAL GAP"
)
print("=" * 100)


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# -----------------------------------------------------------------------------
# QC-resolved accepted evidence from four external batches
# -----------------------------------------------------------------------------

BATCH1_ACCEPTED_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch01_accepted_evidence.csv"
)

BATCH2_ACCEPTED_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch02_accepted_evidence.csv"
)

BATCH3_ACCEPTED_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch03_accepted_evidence.csv"
)

BATCH4_ACCEPTED_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_batch04_accepted_evidence.csv"
)


# -----------------------------------------------------------------------------
# Full species-rescue taxonomic structure from Section 9.4B
# -----------------------------------------------------------------------------

ACQUISITION_REGISTER_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_acquisition_register.csv"
)

CANONICAL_TARGET_MAP_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)


# -----------------------------------------------------------------------------
# Genus-level targets frozen earlier
# -----------------------------------------------------------------------------

GENUS_STAGE_FILE = (
    OUTPUT_DIR /
    "omo_seed_mass_genus_stage_targets.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_COMBINED_EVIDENCE = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_accepted_evidence_all_batches.csv"
)

OUT_TARGET_VALUES = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_species_target_values.csv"
)

OUT_CANONICAL_VALUES = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_canonical_values.csv"
)

OUT_UNRESOLVED_SPECIES = (
    OUTPUT_DIR /
    "omo_seed_mass_post_external_rescue_unresolved_species.csv"
)

OUT_TRUE_RESIDUAL = (
    OUTPUT_DIR /
    "omo_seed_mass_post_external_rescue_true_residual_targets.csv"
)

OUT_COVERAGE_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_post_external_rescue_coverage_summary.csv"
)

OUT_PROVENANCE_SUMMARY = (
    OUTPUT_DIR /
    "omo_seed_mass_non_try_external_provenance_summary.csv"
)


# =============================================================================
# 2. VALIDATE REQUIRED INPUTS
# =============================================================================

required_files = [
    BATCH1_ACCEPTED_FILE,
    BATCH2_ACCEPTED_FILE,
    BATCH3_ACCEPTED_FILE,
    BATCH4_ACCEPTED_FILE,
    ACQUISITION_REGISTER_FILE,
    CANONICAL_TARGET_MAP_FILE,
    GENUS_STAGE_FILE,
]

for path in required_files:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD ACCEPTED BATCH EVIDENCE
# =============================================================================

batch1 = pd.read_csv(
    BATCH1_ACCEPTED_FILE,
    low_memory=False
)

batch2 = pd.read_csv(
    BATCH2_ACCEPTED_FILE,
    low_memory=False
)

batch3 = pd.read_csv(
    BATCH3_ACCEPTED_FILE,
    low_memory=False
)

batch4 = pd.read_csv(
    BATCH4_ACCEPTED_FILE,
    low_memory=False
)


for df, batch_name in [
    (batch1, "BATCH_1"),
    (batch2, "BATCH_2"),
    (batch3, "BATCH_3"),
    (batch4, "BATCH_4"),
]:

    df["External_Rescue_Batch"] = batch_name


print("\nAccepted evidence inputs")
print("-" * 100)

print(
    f"Batch 1 accepted records : {len(batch1)}"
)

print(
    f"Batch 2 accepted records : {len(batch2)}"
)

print(
    f"Batch 3 accepted records : {len(batch3)}"
)

print(
    f"Batch 4 accepted records : {len(batch4)}"
)


# Expected from completed QC sections.
assert len(batch1) == 4
assert len(batch2) == 5
assert len(batch3) == 2
assert len(batch4) == 1


# =============================================================================
# 4. COMBINE ALL ACCEPTED EXTERNAL EVIDENCE
# =============================================================================

accepted_all = pd.concat(
    [
        batch1,
        batch2,
        batch3,
        batch4,
    ],
    ignore_index=True,
    sort=False
)


print("\nCombined accepted external evidence")
print("-" * 100)

print(
    f"Accepted measurement records : "
    f"{len(accepted_all)}"
)

print(
    f"Biological search targets    : "
    f"{accepted_all['Search_Target_ID'].nunique()}"
)


# =============================================================================
# 5. STRONG MEASUREMENT-LEVEL VALIDATION
# =============================================================================

required_evidence_fields = {
    "Evidence_Record_ID",
    "Search_Target_ID",
    "Final_Search_Name",
    "Converted_Individual_Seed_Mass_mg",
    "Evidence_QC_Status",
    "Evidence_QC_Flag",
    "Evidence_Action",
    "Eligible_For_Species_Aggregation",
}

missing_fields = (
    required_evidence_fields
    -
    set(accepted_all.columns)
)

assert not missing_fields, (
    "Combined external evidence is missing required fields:\n"
    f"{sorted(missing_fields)}"
)


assert accepted_all[
    "Evidence_Record_ID"
].is_unique


assert (
    accepted_all[
        "Evidence_QC_Status"
    ]
    .eq("ACCEPT")
    .all()
), (
    "Non-accepted evidence entered external-rescue consolidation."
)


# Convert aggregation field safely.
accepted_all[
    "Converted_Individual_Seed_Mass_mg"
] = pd.to_numeric(
    accepted_all[
        "Converted_Individual_Seed_Mass_mg"
    ],
    errors="coerce"
)


assert accepted_all[
    "Converted_Individual_Seed_Mass_mg"
].notna().all()


assert accepted_all[
    "Converted_Individual_Seed_Mass_mg"
].gt(0).all()


# Boolean validation that avoids string "False" evaluating as True.
eligible_normalized = (
    accepted_all[
        "Eligible_For_Species_Aggregation"
    ]
    .astype(str)
    .str.strip()
    .str.lower()
    .isin(
        [
            "true",
            "1",
            "yes",
        ]
    )
)

assert eligible_normalized.all(), (
    "An accepted record is not marked eligible for species aggregation."
)


assert (
    accepted_all[
        "Search_Target_ID"
    ].nunique()
    == 11
), (
    "Expected accepted external evidence for 11 biological targets."
)


# =============================================================================
# 6. LOAD SPECIES-RESCUE TAXONOMIC STRUCTURE
# =============================================================================

acquisition = pd.read_csv(
    ACQUISITION_REGISTER_FILE,
    low_memory=False
)

canonical_map = pd.read_csv(
    CANONICAL_TARGET_MAP_FILE,
    low_memory=False
)

genus_stage = pd.read_csv(
    GENUS_STAGE_FILE,
    low_memory=False
)


print("\nExternal-rescue taxonomic structure")
print("-" * 100)

print(
    f"Biological rescue targets : "
    f"{len(acquisition)}"
)

print(
    f"Canonical rescue taxa     : "
    f"{len(canonical_map)}"
)

print(
    f"Genus-stage rows          : "
    f"{len(genus_stage)}"
)


assert len(acquisition) == 39

assert acquisition[
    "Search_Target_ID"
].is_unique

assert len(canonical_map) == 50

assert canonical_map[
    "Canonical_Species"
].is_unique


# =============================================================================
# 7. VALIDATE ACCEPTED TARGETS AGAINST FINAL MANIFEST
# =============================================================================

assert (
    accepted_all[
        "Search_Target_ID"
    ]
    .isin(
        set(
            acquisition[
                "Search_Target_ID"
            ]
        )
    )
    .all()
)


target_name_check = (
    accepted_all[
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ]
    ]
    .drop_duplicates()
    .merge(
        acquisition[
            [
                "Search_Target_ID",
                "Final_Search_Name",
            ]
        ],
        on="Search_Target_ID",
        how="left",
        suffixes=(
            "_Evidence",
            "_Manifest",
        ),
        validate="one_to_one"
    )
)


assert (
    target_name_check[
        "Final_Search_Name_Evidence"
    ]
    .eq(
        target_name_check[
            "Final_Search_Name_Manifest"
        ]
    )
    .all()
), (
    "Evidence/manifest taxonomic-name mismatch detected."
)


# =============================================================================
# 8. AGGREGATE ACCEPTED EXTERNAL EVIDENCE BY BIOLOGICAL TARGET
# =============================================================================
#
# Aggregation rule:
#
#     one accepted record  -> that value
#     multiple records     -> median
#
# The median is consistent with earlier Seed Mass aggregation and robust to
# moderate source-to-source variability.
# =============================================================================

def join_unique(series):

    values = (
        series
        .dropna()
        .astype(str)
        .str.strip()
    )

    values = [
        value
        for value in values
        if value
    ]

    return " | ".join(
        sorted(
            set(values)
        )
    )


target_values = (
    accepted_all
    .groupby(
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        dropna=False
    )
    .agg(
        N_Accepted_Evidence_Records=(
            "Evidence_Record_ID",
            "size"
        ),

        Min_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "min"
        ),

        Median_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "median"
        ),

        Max_Accepted_mg=(
            "Converted_Individual_Seed_Mass_mg",
            "max"
        ),

        Source_Classes=(
            "Source_Class",
            join_unique
        ),

        Source_Names=(
            "Source_Name",
            join_unique
        ),

        Evidence_QC_Flags=(
            "Evidence_QC_Flag",
            join_unique
        ),

        External_Rescue_Batches=(
            "External_Rescue_Batch",
            join_unique
        ),
    )
    .reset_index()
)


target_values[
    "Seed_Mass_mg"
] = target_values[
    "Median_Accepted_mg"
]


target_values[
    "Trait_Code"
] = "SEED_MASS"


target_values[
    "Standardized_Trait_Unit"
] = "mg"


target_values[
    "Aggregation_Method"
] = np.where(
    target_values[
        "N_Accepted_Evidence_Records"
    ].eq(1),
    "SINGLE_ACCEPTED_EXTERNAL_MEASUREMENT",
    "MEDIAN_OF_ACCEPTED_EXTERNAL_MEASUREMENTS"
)


target_values[
    "Trait_Source"
] = (
    "SPECIES_EXTERNAL_NON_TRY_RESCUE"
)


target_values[
    "Trait_Quality_Flag"
] = np.where(
    target_values[
        "N_Accepted_Evidence_Records"
    ].eq(1),
    "EXTERNAL_SPECIES_EVIDENCE_SINGLE_ACCEPTED_RECORD",
    "EXTERNAL_SPECIES_EVIDENCE_MULTIPLE_ACCEPTED_RECORDS"
)


# =============================================================================
# 9. SPECIAL QUALITY FLAGS PRESERVED FROM QC
# =============================================================================

# Keep important methodological transformations visible at target level.
for idx, row in target_values.iterrows():

    flags = str(
        row[
            "Evidence_QC_Flags"
        ]
    )

    if (
        "MOISTURE_NORMALIZATION"
        in flags
    ):

        target_values.loc[
            idx,
            "Trait_Quality_Flag"
        ] = (
            "EXTERNAL_SPECIES_EVIDENCE_MOISTURE_NORMALIZED"
        )

    elif (
        "HIGH_VARIABILITY_RANGE_DERIVED"
        in flags
    ):

        target_values.loc[
            idx,
            "Trait_Quality_Flag"
        ] = (
            "EXTERNAL_SPECIES_EVIDENCE_RANGE_DERIVED_HIGH_VARIABILITY"
        )

    elif (
        "RANGE_DERIVED"
        in flags
    ):

        target_values.loc[
            idx,
            "Trait_Quality_Flag"
        ] = (
            "EXTERNAL_SPECIES_EVIDENCE_RANGE_DERIVED"
        )


# =============================================================================
# 10. DISPLAY TARGET-LEVEL EXTERNAL VALUES
# =============================================================================

print("\nConsolidated biological-target Seed Mass values")
print("-" * 100)

display(
    target_values[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Seed_Mass_mg",
            "N_Accepted_Evidence_Records",
            "Min_Accepted_mg",
            "Median_Accepted_mg",
            "Max_Accepted_mg",
            "Aggregation_Method",
            "Trait_Source",
            "Trait_Quality_Flag",
        ]
    ]
    .sort_values(
        "Search_Target_ID"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 11. TARGET-LEVEL VALIDATION
# =============================================================================

assert len(target_values) == 11

assert target_values[
    "Search_Target_ID"
].is_unique

assert target_values[
    "Seed_Mass_mg"
].notna().all()

assert target_values[
    "Seed_Mass_mg"
].gt(0).all()


# Irvingia should have two accepted records and median 11,925 mg.
irvingia = target_values[
    target_values[
        "Final_Search_Name"
    ].eq(
        "Irvingia gabonensis"
    )
]

assert len(irvingia) == 1

assert int(
    irvingia[
        "N_Accepted_Evidence_Records"
    ].iloc[0]
) == 2

assert np.isclose(
    irvingia[
        "Seed_Mass_mg"
    ].iloc[0],
    11925.0
)


# Picralima must retain moisture-normalized value.
picralima = target_values[
    target_values[
        "Final_Search_Name"
    ].eq(
        "Picralima nitida"
    )
]

assert len(picralima) == 1

assert np.isclose(
    picralima[
        "Seed_Mass_mg"
    ].iloc[0],
    1193.506
)


# =============================================================================
# 12. MAP TARGET VALUES BACK TO OMO CANONICAL TAXA
# =============================================================================

canonical_values = canonical_map.merge(
    target_values,
    on=[
        "Search_Target_ID",
        "Final_Search_Name",
    ],
    how="inner",
    validate="many_to_one"
)


print("\nCanonical taxa resolved by external species rescue")
print("-" * 100)

print(
    f"Resolved biological targets : "
    f"{target_values['Search_Target_ID'].nunique()}"
)

print(
    f"Resolved canonical taxa     : "
    f"{canonical_values['Canonical_Species'].nunique()}"
)


display(
    canonical_values[
        [
            "Canonical_Species",
            "Search_Target_ID",
            "Final_Search_Name",
            "Seed_Mass_mg",
            "N_Accepted_Evidence_Records",
            "Aggregation_Method",
            "Trait_Source",
            "Trait_Quality_Flag",
        ]
    ]
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 13. VALIDATE CANONICAL RESCUE EXPANSION
# =============================================================================

assert canonical_values[
    "Canonical_Species"
].is_unique


# From the validated 39-target/50-canonical taxonomic crosswalk,
# the 11 externally resolved biological targets should resolve 14 canonical
# Omo identifiers because several targets represent validated spelling variants.
assert (
    canonical_values[
        "Canonical_Species"
    ].nunique()
    == 14
), (
    "Expected the 11 biological rescue targets to map to "
    "14 Omo canonical taxa."
)


# =============================================================================
# 14. DETERMINE UNRESOLVED SPECIES-LEVEL RESCUE TAXA
# =============================================================================

resolved_canonical_set = set(
    canonical_values[
        "Canonical_Species"
    ]
)

all_species_rescue_set = set(
    canonical_map[
        "Canonical_Species"
    ]
)


unresolved_species_names = (
    all_species_rescue_set
    -
    resolved_canonical_set
)


unresolved_species = (
    canonical_map[
        canonical_map[
            "Canonical_Species"
        ].isin(
            unresolved_species_names
        )
    ]
    .copy()
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


unresolved_species[
    "Seed_Mass_Resolution_Status"
] = (
    "UNRESOLVED_AFTER_BIEN_TRY_AND_EXTERNAL_SPECIES_RESCUE"
)


unresolved_species[
    "Next_Acquisition_Action"
] = (
    "SPECIES_LEVEL_FALLBACK_OR_CONGENERIC_REVIEW"
)


print("\nSpecies-level taxa unresolved after external rescue")
print("-" * 100)

print(
    f"N unresolved canonical species taxa : "
    f"{len(unresolved_species)}"
)

display(
    unresolved_species[
        [
            "Canonical_Species",
            "Search_Target_ID",
            "Final_Search_Name",
            "Seed_Mass_Resolution_Status",
            "Next_Acquisition_Action",
        ]
    ]
)


assert len(unresolved_species) == 36, (
    f"Expected 36 unresolved species-level canonical taxa, "
    f"found {len(unresolved_species)}."
)


# =============================================================================
# 15. STANDARDIZE GENUS-STAGE TABLE
# =============================================================================

# Find canonical-name field robustly.
genus_canonical_candidates = [
    "Canonical_Species",
    "Canonical Species",
]

GENUS_CANONICAL_COL = next(
    (
        col
        for col in genus_canonical_candidates
        if col in genus_stage.columns
    ),
    None
)

assert GENUS_CANONICAL_COL is not None, (
    "Could not locate canonical-species field in genus-stage table.\n"
    f"Columns: {list(genus_stage.columns)}"
)


genus_residual = pd.DataFrame({
    "Canonical_Species":
        genus_stage[
            GENUS_CANONICAL_COL
        ].astype(str),

    "Seed_Mass_Resolution_Status":
        "GENUS_LEVEL_ACQUISITION_PENDING",

    "Next_Acquisition_Action":
        "PROCEED_TO_GENUS_LEVEL_STAGE",
})


genus_residual = (
    genus_residual
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ]
    )
    .reset_index(drop=True)
)


print("\nGenus-level residual targets")
print("-" * 100)

display(
    genus_residual
)


assert len(genus_residual) == 3, (
    f"Expected 3 genus-stage canonical targets, "
    f"found {len(genus_residual)}."
)


# =============================================================================
# 16. BUILD TRUE POST-EXTERNAL-RESCUE RESIDUAL SET
# =============================================================================

species_residual_for_union = unresolved_species[
    [
        "Canonical_Species",
        "Seed_Mass_Resolution_Status",
        "Next_Acquisition_Action",
    ]
].copy()


true_residual = pd.concat(
    [
        species_residual_for_union,
        genus_residual,
    ],
    ignore_index=True,
    sort=False
)


true_residual = (
    true_residual
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ]
    )
    .sort_values(
        [
            "Next_Acquisition_Action",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


print("\nTrue residual Seed Mass targets after external species rescue")
print("-" * 100)

print(
    f"Species-level residual taxa : "
    f"{len(unresolved_species)}"
)

print(
    f"Genus-stage residual taxa   : "
    f"{len(genus_residual)}"
)

print(
    f"Total true residual taxa    : "
    f"{len(true_residual)}"
)


display(
    true_residual
)


assert len(true_residual) == 39, (
    f"Expected 39 total residual canonical taxa, "
    f"found {len(true_residual)}."
)


# =============================================================================
# 17. RECONSTRUCT OVERALL 127-TAXON COVERAGE
# =============================================================================
#
# Section 9.3L-R3 established:
#
#     74 canonical taxa resolved by BIEN and/or TRY
#     53 residual taxa
#
# Those 53 comprised:
#
#     50 species-level external-rescue taxa
#      3 genus-stage taxa
#
# External species rescue has now resolved 14 of the 50 canonical
# species-level taxa.
#
# Therefore:
#
#     74 + 14 = 88 resolved canonical taxa
#     127 - 88 = 39 residual taxa
#
# These figures are validated against the reconstructed residual set above.
# =============================================================================

N_CANONICAL_TAXA = 127

N_BIEN_TRY_RESOLVED = 74

N_EXTERNAL_CANONICAL_RESOLVED = (
    canonical_values[
        "Canonical_Species"
    ].nunique()
)

N_TOTAL_RESOLVED = (
    N_BIEN_TRY_RESOLVED
    +
    N_EXTERNAL_CANONICAL_RESOLVED
)

N_TRUE_RESIDUAL = (
    len(true_residual)
)


assert N_EXTERNAL_CANONICAL_RESOLVED == 14

assert N_TOTAL_RESOLVED == 88

assert N_TRUE_RESIDUAL == 39

assert (
    N_TOTAL_RESOLVED
    +
    N_TRUE_RESIDUAL
    ==
    N_CANONICAL_TAXA
)


# =============================================================================
# 18. COVERAGE SUMMARY
# =============================================================================

coverage_summary = pd.DataFrame({
    "Coverage_Category": [
        "RESOLVED_BY_BIEN_AND_OR_TRY",
        "RESOLVED_BY_EXTERNAL_SPECIES_RESCUE",
        "UNRESOLVED_SPECIES_LEVEL_AFTER_EXTERNAL_RESCUE",
        "GENUS_LEVEL_ACQUISITION_PENDING",
    ],

    "N_Canonical_Taxa": [
        N_BIEN_TRY_RESOLVED,
        N_EXTERNAL_CANONICAL_RESOLVED,
        len(unresolved_species),
        len(genus_residual),
    ],
})


coverage_summary[
    "Percent_of_127"
] = (
    coverage_summary[
        "N_Canonical_Taxa"
    ]
    /
    N_CANONICAL_TAXA
    *
    100
)


print("\nPost-external-rescue Seed Mass coverage")
print("-" * 100)

display(
    coverage_summary
)


print("\nOverall Seed Mass coverage checkpoint")
print("-" * 100)

print(
    f"Canonical Omo taxa                        : "
    f"{N_CANONICAL_TAXA}"
)

print(
    f"Resolved by BIEN and/or TRY               : "
    f"{N_BIEN_TRY_RESOLVED}"
)

print(
    f"Newly resolved by external species rescue : "
    f"{N_EXTERNAL_CANONICAL_RESOLVED}"
)

print(
    f"Total species-level resolved              : "
    f"{N_TOTAL_RESOLVED}"
)

print(
    f"Remaining species-level rescue taxa       : "
    f"{len(unresolved_species)}"
)

print(
    f"Genus-level acquisition pending           : "
    f"{len(genus_residual)}"
)

print(
    f"True residual taxa                        : "
    f"{N_TRUE_RESIDUAL}"
)

print(
    f"Overall canonical coverage                : "
    f"{100 * N_TOTAL_RESOLVED / N_CANONICAL_TAXA:.2f}%"
)


# =============================================================================
# 19. EXTERNAL-RESCUE PROVENANCE SUMMARY
# =============================================================================

provenance_summary = (
    target_values
    .groupby(
        "Trait_Quality_Flag"
    )
    .agg(
        N_Biological_Targets=(
            "Search_Target_ID",
            "size"
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
    .reset_index()
)


print("\nExternal species-rescue provenance summary")
print("-" * 100)

display(
    provenance_summary
)


# =============================================================================
# 20. EXTERNAL TARGET NUMERIC SUMMARY
# =============================================================================

print("\nExternal rescue Seed Mass numeric summary")
print("-" * 100)

print(
    target_values[
        "Seed_Mass_mg"
    ].describe()
)


# =============================================================================
# 21. FINAL VALIDATION SUMMARY
# =============================================================================

validation_summary = pd.DataFrame({
    "Metric": [
        "Accepted external measurement records",
        "Resolved biological search targets",
        "Canonical taxa resolved by external rescue",
        "Canonical taxa previously resolved by BIEN/TRY",
        "Total canonical taxa now resolved",
        "Species-level taxa still unresolved",
        "Genus-stage taxa still unresolved",
        "True residual canonical taxa",
        "Canonical Omo taxa",
    ],

    "Value": [
        len(accepted_all),
        len(target_values),
        N_EXTERNAL_CANONICAL_RESOLVED,
        N_BIEN_TRY_RESOLVED,
        N_TOTAL_RESOLVED,
        len(unresolved_species),
        len(genus_residual),
        len(true_residual),
        N_CANONICAL_TAXA,
    ],
})


print("\nSection 9.4L consolidation summary")
print("-" * 100)

display(
    validation_summary
)


# =============================================================================
# 22. FINAL STRONG ASSERTIONS
# =============================================================================

assert len(accepted_all) == 12

assert len(target_values) == 11

assert len(canonical_values) == 14

assert len(unresolved_species) == 36

assert len(genus_residual) == 3

assert len(true_residual) == 39

assert N_TOTAL_RESOLVED == 88

assert (
    N_TOTAL_RESOLVED
    +
    len(true_residual)
    ==
    127
)


# No resolved canonical external taxon may remain in species residual.
assert not (
    set(
        canonical_values[
            "Canonical_Species"
        ]
    )
    &
    set(
        unresolved_species[
            "Canonical_Species"
        ]
    )
)


# Species and genus residual groups must not overlap.
assert not (
    set(
        unresolved_species[
            "Canonical_Species"
        ]
    )
    &
    set(
        genus_residual[
            "Canonical_Species"
        ]
    )
)


# =============================================================================
# 23. EXPORT
# =============================================================================

accepted_all.to_csv(
    OUT_COMBINED_EVIDENCE,
    index=False
)


target_values.to_csv(
    OUT_TARGET_VALUES,
    index=False
)


canonical_values.to_csv(
    OUT_CANONICAL_VALUES,
    index=False
)


unresolved_species.to_csv(
    OUT_UNRESOLVED_SPECIES,
    index=False
)


true_residual.to_csv(
    OUT_TRUE_RESIDUAL,
    index=False
)


coverage_summary.to_csv(
    OUT_COVERAGE_SUMMARY,
    index=False
)


provenance_summary.to_csv(
    OUT_PROVENANCE_SUMMARY,
    index=False
)


# =============================================================================
# 24. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4L exports")
print("-" * 100)

print(
    f"Combined accepted evidence : "
    f"{OUT_COMBINED_EVIDENCE}"
)

print(
    f"Biological target values   : "
    f"{OUT_TARGET_VALUES}"
)

print(
    f"Canonical external values  : "
    f"{OUT_CANONICAL_VALUES}"
)

print(
    f"Unresolved species targets : "
    f"{OUT_UNRESOLVED_SPECIES}"
)

print(
    f"True residual targets      : "
    f"{OUT_TRUE_RESIDUAL}"
)

print(
    f"Coverage summary           : "
    f"{OUT_COVERAGE_SUMMARY}"
)

print(
    f"Provenance summary         : "
    f"{OUT_PROVENANCE_SUMMARY}"
)


# =============================================================================
# 25. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "External species-level Seed Mass rescue "
    "consolidation completed successfully."
)

print(
    f"Accepted external measurement records : "
    f"{len(accepted_all)}"
)

print(
    f"Resolved biological search targets    : "
    f"{len(target_values)}"
)

print(
    f"Resolved Omo canonical taxa           : "
    f"{N_EXTERNAL_CANONICAL_RESOLVED}"
)

print(
    f"Total Omo taxa resolved after "
    f"BIEN + TRY + external rescue          : "
    f"{N_TOTAL_RESOLVED} / {N_CANONICAL_TAXA}"
)

print(
    f"Remaining species-level taxa          : "
    f"{len(unresolved_species)}"
)

print(
    f"Genus-level targets pending           : "
    f"{len(genus_residual)}"
)

print(
    f"True residual canonical taxa          : "
    f"{len(true_residual)}"
)

print(
    "No genus-level Seed Mass value has yet been assigned."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4M
# CLASSIFY POST-EXTERNAL-RESCUE RESIDUAL SEED MASS GAP
# =============================================================================
#
# Purpose
# -------
# 1. Start from the authoritative residual gap produced by Section 9.4L.
# 2. Separate unresolved canonical rows from unresolved biological targets.
# 3. Preserve taxonomic consolidation established in Section 9.4A-R2.
# 4. Classify each unresolved biological target into the next defensible
#    Seed Mass acquisition / fallback pathway.
# 5. Keep the explicit genus-stage taxa separate.
# 6. Export an auditable residual classification register.
#
# IMPORTANT
# ---------
# This section performs CLASSIFICATION ONLY.
#
# It does NOT:
#   - calculate genus means,
#   - calculate family means,
#   - impute Seed Mass,
#   - assign final Seed Mass values,
#   - alter the validated 88/127 coverage state from Section 9.4L.
#
# Expected starting state from Section 9.4L
# -----------------------------------------
# Canonical taxa                         : 127
# Resolved Seed Mass taxa               : 88
# Residual canonical taxa               : 39
# Residual species-level canonical taxa : 36
# Explicit genus-stage canonical taxa   : 3
#
# The 36 species-level canonical rows should collapse to 28 unresolved
# biological search targets.
# =============================================================================


from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"


# Authoritative canonical-to-biological-target map from Section 9.4B
CROSSWALK_FILE = (
    OUT_DIR /
    "omo_seed_mass_non_try_external_canonical_target_map.csv"
)


# Authoritative Section 9.4L residual-gap outputs
RESIDUAL_FILE_CANDIDATES = [

    OUT_DIR /
    "omo_seed_mass_post_external_rescue_true_residual_targets.csv",

    OUT_DIR /
    "omo_seed_mass_post_external_rescue_unresolved_species.csv",
]


# Outputs from this section
OUT_CANONICAL_CLASSIFICATION = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_canonical_classification.csv"
)

OUT_BIOLOGICAL_CLASSIFICATION = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_biological_target_classification.csv"
)

OUT_GENUS_STAGE = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_genus_stage_targets.csv"
)

OUT_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_classification_summary.csv"
)

# =============================================================================
# 2. HELPER FUNCTIONS
# =============================================================================

def clean_text_series(series):
    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "": pd.NA,
                "nan": pd.NA,
                "None": pd.NA,
                "<NA>": pd.NA,
            }
        )
    )


def find_column(df, candidates, required=True):
    """
    Find a column using exact or normalized matching.
    """

    normalized_lookup = {
        str(col).strip().lower().replace(" ", "_"): col
        for col in df.columns
    }

    for candidate in candidates:

        if candidate in df.columns:
            return candidate

        key = (
            str(candidate)
            .strip()
            .lower()
            .replace(" ", "_")
        )

        if key in normalized_lookup:
            return normalized_lookup[key]

    if required:
        raise KeyError(
            "Required field not found.\n"
            f"Candidates tested: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


def first_existing_file(paths):

    for path in paths:
        if path.exists():
            return path

    return None


def print_rule(char="-", width=100):
    print(char * width)


# =============================================================================
# 3. LOAD AUTHORITATIVE CROSSWALK
# =============================================================================

if not CROSSWALK_FILE.exists():
    raise FileNotFoundError(
        "Authoritative Section 9.4A-R2 crosswalk not found:\n"
        f"{CROSSWALK_FILE}"
    )


crosswalk = pd.read_csv(CROSSWALK_FILE)


CW_CANON_COL = find_column(
    crosswalk,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)

CW_SEARCH_COL = find_column(
    crosswalk,
    [
        "Final_Search_Name",
        "Final Search Name",
    ]
)

CW_TARGET_ID_COL = find_column(
    crosswalk,
    [
        "Search_Target_ID",
        "Search Target ID",
    ]
)


crosswalk[CW_CANON_COL] = clean_text_series(
    crosswalk[CW_CANON_COL]
)

crosswalk[CW_SEARCH_COL] = clean_text_series(
    crosswalk[CW_SEARCH_COL]
)

crosswalk[CW_TARGET_ID_COL] = clean_text_series(
    crosswalk[CW_TARGET_ID_COL]
)


# =============================================================================
# 4. LOCATE SECTION 9.4L RESIDUAL EXPORT
# =============================================================================

RESIDUAL_FILE = first_existing_file(
    RESIDUAL_FILE_CANDIDATES
)


if RESIDUAL_FILE is not None:

    residual = pd.read_csv(
        RESIDUAL_FILE
    )

    print(
        "Section 9.4L residual file located:"
    )
    print(
        RESIDUAL_FILE
    )

else:

    # -------------------------------------------------------------------------
    # Controlled in-memory fallback
    # -------------------------------------------------------------------------
    #
    # Section 9.4L may already have left the authoritative residual table
    # in memory under one of these likely variable names.
    # -------------------------------------------------------------------------

    residual = None

    possible_variables = [
        "residual_gap",
        "residual_taxa",
        "remaining_gap",
        "remaining_taxa",
        "post_external_residual",
        "final_residual",
    ]

    for var_name in possible_variables:

        if (
            var_name in globals()
            and isinstance(
                globals()[var_name],
                pd.DataFrame
            )
        ):

            candidate_df = (
                globals()[var_name]
                .copy()
            )

            try:

                test_col = find_column(
                    candidate_df,
                    [
                        "Canonical_Species",
                        "Canonical Species",
                    ],
                    required=False
                )

                if test_col is not None:

                    residual = candidate_df

                    print(
                        "Using Section 9.4L residual "
                        f"table from memory: {var_name}"
                    )

                    break

            except Exception:
                pass


if residual is None:

    raise FileNotFoundError(
        "\nCould not locate the authoritative "
        "Section 9.4L residual-gap table.\n\n"
        "Checked the following paths:\n"
        +
        "\n".join(
            str(x)
            for x in RESIDUAL_FILE_CANDIDATES
        )
        +
        "\n\nNo suitable Section 9.4L residual "
        "DataFrame was found in memory either."
    )


# =============================================================================
# 5. STANDARDIZE RESIDUAL TABLE
# =============================================================================

RES_CANON_COL = find_column(
    residual,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)


residual[RES_CANON_COL] = clean_text_series(
    residual[RES_CANON_COL]
)


residual = (
    residual[
        residual[
            RES_CANON_COL
        ].notna()
    ]
    .copy()
)


residual = (
    residual
    .drop_duplicates(
        subset=[
            RES_CANON_COL
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 6. VALIDATE SECTION 9.4L STARTING STATE
# =============================================================================

print()
print("=" * 100)
print(
    "SECTION 9.4M — CLASSIFY POST-EXTERNAL-RESCUE "
    "RESIDUAL SEED MASS GAP"
)
print("=" * 100)

print()
print("Input residual state")
print_rule()

print(
    f"Residual canonical taxa : "
    f"{len(residual)}"
)


assert len(residual) == 39, (
    "\nSection 9.4M expected exactly 39 "
    "residual canonical taxa from Section 9.4L.\n"
    f"Observed: {len(residual)}\n"
    "Do not proceed until the 9.4L residual "
    "state is reconciled."
)


# =============================================================================
# 7. IDENTIFY EXPLICIT GENUS-STAGE TAXA
# =============================================================================

EXPLICIT_GENUS_STAGE_TAXA = {
    "Cola species",
    "Drypetes",
    "Drypetes species",
}


residual[
    "Residual_Level"
] = np.where(
    residual[
        RES_CANON_COL
    ].isin(
        EXPLICIT_GENUS_STAGE_TAXA
    ),
    "EXPLICIT_GENUS_STAGE",
    "SPECIES_LEVEL_RESIDUAL"
)


level_summary = (
    residual[
        "Residual_Level"
    ]
    .value_counts()
    .rename_axis(
        "Residual_Level"
    )
    .reset_index(
        name="N_Canonical_Taxa"
    )
)


print()
print("Residual-level classification")
print_rule()
display(level_summary)


n_species_residual = int(
    (
        residual[
            "Residual_Level"
        ]
        ==
        "SPECIES_LEVEL_RESIDUAL"
    ).sum()
)

n_genus_stage = int(
    (
        residual[
            "Residual_Level"
        ]
        ==
        "EXPLICIT_GENUS_STAGE"
    ).sum()
)


assert n_species_residual == 36, (
    "Expected 36 unresolved species-level "
    f"canonical taxa; observed {n_species_residual}."
)

assert n_genus_stage == 3, (
    "Expected 3 explicit genus-stage canonical "
    f"taxa; observed {n_genus_stage}."
)


# =============================================================================
# 8. MAP SPECIES-LEVEL RESIDUAL ROWS BACK TO BIOLOGICAL SEARCH TARGETS
# =============================================================================

species_residual = (
    residual[
        residual[
            "Residual_Level"
        ]
        ==
        "SPECIES_LEVEL_RESIDUAL"
    ]
    .copy()
)


species_map = (
    species_residual[
        [
            RES_CANON_COL
        ]
    ]
    .rename(
        columns={
            RES_CANON_COL:
            "Canonical_Species"
        }
    )
    .merge(
        crosswalk[
            [
                CW_CANON_COL,
                CW_TARGET_ID_COL,
                CW_SEARCH_COL,
            ]
        ].rename(
            columns={
                CW_CANON_COL:
                "Canonical_Species",
                CW_TARGET_ID_COL:
                "Search_Target_ID",
                CW_SEARCH_COL:
                "Final_Search_Name",
            }
        ),
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


missing_target_map = (
    species_map[
        "Search_Target_ID"
    ].isna()
)


if missing_target_map.any():

    print()
    print(
        "Species-level residual taxa lacking "
        "9.4A-R2 search-target mapping"
    )
    print_rule()

    display(
        species_map.loc[
            missing_target_map,
            [
                "Canonical_Species"
            ]
        ]
    )

    raise AssertionError(
        "Some residual species-level taxa do not "
        "map to the authoritative 9.4A-R2 crosswalk."
    )


# =============================================================================
# 9. COLLAPSE CANONICAL ROWS TO BIOLOGICAL TARGETS
# =============================================================================

biological_targets = (
    species_map
    .groupby(
        [
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        as_index=False
    )
    .agg(
        N_Omo_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),
        Omo_Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    pd.unique(
                        x.dropna()
                    )
                )
            )
        )
    )
    .sort_values(
        "Search_Target_ID"
    )
    .reset_index(drop=True)
)


print()
print(
    "Residual species-level biological targets"
)
print_rule()

print(
    "Species-level residual canonical taxa : "
    f"{len(species_map)}"
)

print(
    "Residual biological search targets    : "
    f"{len(biological_targets)}"
)

print(
    "Canonical duplication removed         : "
    f"{len(species_map) - len(biological_targets)}"
)


assert len(species_map) == 36

assert len(biological_targets) == 28, (
    "\nExpected the 36 residual species-level "
    "canonical rows to collapse to 28 biological "
    f"targets; observed {len(biological_targets)}."
)


# =============================================================================
# 10. DEFINE CLASSIFICATION REGISTER
# =============================================================================
#
# IMPORTANT:
# These classifications specify the NEXT DEFENSIBLE SEARCH / FALLBACK
# pathway. They do not assign any numerical value.
#
# Classes
# -------
# TAXONOMIC_NAME_RESCUE_FIRST
#     A validated/alternative accepted search name exists and should be
#     exhausted before trait-level fallback.
#
# CONGENERIC_EVIDENCE_ASSESSMENT
#     Species-level external search has been exhausted sufficiently for
#     the current workflow. Next assess available congeners.
#
# SPECIAL_BIOLOGICAL_CASE
#     Ordinary seed-mass fallback may be biologically inappropriate and
#     requires explicit treatment.
#
# =============================================================================

TAXONOMIC_NAME_RESCUE_FIRST = {

    # Existing project corrections / accepted-name search routes
    "Aningeria pierrei",

    "Corynanthe pachyceras",

    "Homalium longistylum",

    "Mitragyna ciliata",

    "Musa × paradisiaca",
}


SPECIAL_BIOLOGICAL_CASE = {

    # Cultivated hybrid / often sterile or seedless.
    "Musa × paradisiaca",
}


def classify_species_target(search_name):

    if search_name in SPECIAL_BIOLOGICAL_CASE:

        return (
            "SPECIAL_BIOLOGICAL_CASE",
            "REVIEW_SEED_BEARING_STATUS_BEFORE_FALLBACK",
            (
                "Cultivated Musa × paradisiaca is commonly "
                "sterile or seedless; ordinary genus/family "
                "Seed Mass transfer should not be applied "
                "without explicit biological justification."
            )
        )

    if search_name in TAXONOMIC_NAME_RESCUE_FIRST:

        return (
            "TAXONOMIC_NAME_RESCUE_FIRST",
            "TARGETED_ACCEPTED_NAME_AND_SYNONYM_SEARCH",
            (
                "A taxonomically corrected or preferred "
                "search name exists in the project workflow. "
                "Exhaust species-specific evidence under "
                "accepted names and validated synonyms before "
                "using genus-level fallback."
            )
        )

    return (
        "CONGENERIC_EVIDENCE_ASSESSMENT",
        "ASSESS_GENUS_LEVEL_SEED_MASS_EVIDENCE",
        (
            "Current species-level BIEN, TRY and targeted "
            "external rescue did not yield an accepted Seed "
            "Mass value. Assess defensible congeneric evidence "
            "before considering family-level fallback."
        )
    )


classification_rows = biological_targets.apply(
    lambda row: classify_species_target(
        row["Final_Search_Name"]
    ),
    axis=1
)


biological_targets[
    "Residual_Classification"
] = [
    x[0]
    for x in classification_rows
]

biological_targets[
    "Next_Action"
] = [
    x[1]
    for x in classification_rows
]

biological_targets[
    "Classification_Rationale"
] = [
    x[2]
    for x in classification_rows
]


biological_targets[
    "Numerical_Seed_Mass_Assigned"
] = False


# =============================================================================
# 11. EXPLICIT GENUS-STAGE REGISTER
# =============================================================================

genus_stage = (
    residual.loc[
        residual[
            "Residual_Level"
        ].eq(
            "EXPLICIT_GENUS_STAGE"
        ),
        [
            RES_CANON_COL,
            "Residual_Level",
        ]
    ]
    .rename(
        columns={
            RES_CANON_COL:
            "Canonical_Species"
        }
    )
    .copy()
)


def genus_from_target(name):

    if name == "Cola species":
        return "Cola"

    if name in {
        "Drypetes",
        "Drypetes species",
    }:
        return "Drypetes"

    return str(name).split()[0]


genus_stage[
    "Genus_Search_Target"
] = (
    genus_stage[
        "Canonical_Species"
    ]
    .map(
        genus_from_target
    )
)


genus_stage[
    "Residual_Classification"
] = (
    "EXPLICIT_GENUS_LEVEL_TARGET"
)


genus_stage[
    "Next_Action"
] = (
    "ACQUIRE_OR_DERIVE_GENUS_LEVEL_SEED_MASS"
)


genus_stage[
    "Classification_Rationale"
] = (
    "The project taxon is already explicitly genus-level or "
    "unresolved below genus. Species-level imputation is therefore "
    "not appropriate; genus-level evidence should be evaluated directly."
)


genus_stage[
    "Numerical_Seed_Mass_Assigned"
] = False


# -----------------------------------------------------------------------------
# Validation
# -----------------------------------------------------------------------------

assert len(genus_stage) == 3, (
    f"Expected 3 explicit genus-stage canonical taxa, "
    f"found {len(genus_stage)}."
)

assert (
    genus_stage[
        "Residual_Level"
    ]
    .eq(
        "EXPLICIT_GENUS_STAGE"
    )
    .all()
)

assert set(
    genus_stage[
        "Genus_Search_Target"
    ]
) == {
    "Cola",
    "Drypetes",
}


# =============================================================================
# 12. BUILD CANONICAL-LEVEL CLASSIFICATION TABLE
# =============================================================================

canonical_species_classification = (
    species_map
    .merge(
        biological_targets[
            [
                "Search_Target_ID",
                "Final_Search_Name",
                "Residual_Classification",
                "Next_Action",
                "Classification_Rationale",
                "Numerical_Seed_Mass_Assigned",
            ]
        ],
        on=[
            "Search_Target_ID",
            "Final_Search_Name",
        ],
        how="left",
        validate="many_to_one"
    )
)


canonical_species_classification[
    "Residual_Level"
] = (
    "SPECIES_LEVEL_RESIDUAL"
)


canonical_genus_classification = (
    genus_stage
    .copy()
)


canonical_genus_classification[
    "Search_Target_ID"
] = pd.NA

canonical_genus_classification[
    "Final_Search_Name"
] = canonical_genus_classification[
    "Genus_Search_Target"
]


canonical_classification = pd.concat(
    [
        canonical_species_classification[
            [
                "Canonical_Species",
                "Search_Target_ID",
                "Final_Search_Name",
                "Residual_Level",
                "Residual_Classification",
                "Next_Action",
                "Classification_Rationale",
                "Numerical_Seed_Mass_Assigned",
            ]
        ],
        canonical_genus_classification[
            [
                "Canonical_Species",
                "Search_Target_ID",
                "Final_Search_Name",
                "Residual_Level",
                "Residual_Classification",
                "Next_Action",
                "Classification_Rationale",
                "Numerical_Seed_Mass_Assigned",
            ]
        ],
    ],
    ignore_index=True
)


canonical_classification = (
    canonical_classification
    .sort_values(
        [
            "Residual_Level",
            "Final_Search_Name",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 13. CLASSIFICATION SUMMARY
# =============================================================================

species_class_summary = (
    biological_targets[
        "Residual_Classification"
    ]
    .value_counts()
    .rename_axis(
        "Residual_Classification"
    )
    .reset_index(
        name="N_Biological_Targets"
    )
)


species_action_summary = (
    biological_targets[
        "Next_Action"
    ]
    .value_counts()
    .rename_axis(
        "Next_Action"
    )
    .reset_index(
        name="N_Biological_Targets"
    )
)


print()
print(
    "Species-level residual classification"
)
print_rule()

display(
    biological_targets[
        [
            "Search_Target_ID",
            "Final_Search_Name",
            "Omo_Canonical_Taxa",
            "N_Omo_Canonical_Taxa",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
)


print()
print("Classification summary")
print_rule()

display(
    species_class_summary
)


print()
print("Next-action summary")
print_rule()

display(
    species_action_summary
)


print()
print("Explicit genus-stage targets")
print_rule()

display(
    genus_stage[
        [
            "Canonical_Species",
            "Genus_Search_Target",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 14. VALIDATION
# =============================================================================

assert len(
    canonical_classification
) == 39


assert (
    canonical_classification[
        "Canonical_Species"
    ]
    .nunique()
    ==
    39
)


assert (
    canonical_classification[
        "Numerical_Seed_Mass_Assigned"
    ]
    .eq(False)
    .all()
)


assert (
    biological_targets[
        "Search_Target_ID"
    ]
    .nunique()
    ==
    28
)


assert (
    biological_targets[
        "Final_Search_Name"
    ]
    .nunique()
    ==
    28
)


assert (
    biological_targets[
        "Numerical_Seed_Mass_Assigned"
    ]
    .eq(False)
    .all()
)


assert (
    genus_stage[
        "Canonical_Species"
    ]
    .nunique()
    ==
    3
)


assert set(
    genus_stage[
        "Genus_Search_Target"
    ]
) == {
    "Cola",
    "Drypetes",
}


# Check that the canonical multiplicity reconciles exactly.
assert (
    biological_targets[
        "N_Omo_Canonical_Taxa"
    ].sum()
    ==
    36
)


# =============================================================================
# 15. BUILD SECTION SUMMARY
# =============================================================================

summary_rows = [

    {
        "Metric":
            "Residual canonical taxa entering 9.4M",
        "Value":
            39,
    },

    {
        "Metric":
            "Residual species-level canonical taxa",
        "Value":
            36,
    },

    {
        "Metric":
            "Residual species-level biological targets",
        "Value":
            28,
    },

    {
        "Metric":
            "Canonical duplication collapsed",
        "Value":
            8,
    },

    {
        "Metric":
            "Explicit genus-stage canonical taxa",
        "Value":
            3,
    },

    {
        "Metric":
            "Unique explicit genus search targets",
        "Value":
            genus_stage[
                "Genus_Search_Target"
            ].nunique(),
    },

    {
        "Metric":
            "Seed Mass values assigned in Section 9.4M",
        "Value":
            0,
    },

]


for _, row in species_class_summary.iterrows():

    summary_rows.append(
        {
            "Metric":
                "Species biological targets — "
                + row[
                    "Residual_Classification"
                ],
            "Value":
                int(
                    row[
                        "N_Biological_Targets"
                    ]
                ),
        }
    )


classification_summary = pd.DataFrame(
    summary_rows
)


print()
print("Section 9.4M summary")
print_rule()

display(
    classification_summary
)


# =============================================================================
# 16. EXPORT
# =============================================================================

canonical_classification.to_csv(
    OUT_CANONICAL_CLASSIFICATION,
    index=False
)

biological_targets.to_csv(
    OUT_BIOLOGICAL_CLASSIFICATION,
    index=False
)

genus_stage.to_csv(
    OUT_GENUS_STAGE,
    index=False
)

classification_summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 17. FINAL REPORT
# =============================================================================

print()
print("Classification exports")
print_rule()

print(
    "Canonical classification : "
    f"{OUT_CANONICAL_CLASSIFICATION}"
)

print(
    "Biological targets       : "
    f"{OUT_BIOLOGICAL_CLASSIFICATION}"
)

print(
    "Genus-stage targets      : "
    f"{OUT_GENUS_STAGE}"
)

print(
    "Summary                  : "
    f"{OUT_SUMMARY}"
)


print()
print("=" * 100)
print(
    "Post-external-rescue residual Seed Mass "
    "gap classified successfully."
)
print(
    f"Residual canonical taxa                : "
    f"{len(canonical_classification)}"
)
print(
    f"Species-level canonical residuals      : "
    f"{len(species_map)}"
)
print(
    f"Species-level biological targets       : "
    f"{len(biological_targets)}"
)
print(
    f"Explicit genus-stage canonical taxa    : "
    f"{len(genus_stage)}"
)
print(
    f"Unique explicit genus targets          : "
    f"{genus_stage['Genus_Search_Target'].nunique()}"
)
print(
    "Seed Mass values assigned in 9.4M     : 0"
)
print(
    "The residual gap is now classified for "
    "taxonomic rescue, congeneric evidence "
    "assessment, special-case review, and "
    "explicit genus-level acquisition."
)
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4N
# RESIDUAL GENUS-EVIDENCE AVAILABILITY AUDIT
# =============================================================================
#
# Purpose
# -------
# Audit how much validated species-level Seed Mass evidence is already
# available for the genera required by the post-external-rescue residual gap.
#
# Inputs
# ------
# Section 9.4M:
#   - 23 species biological targets classified for congeneric assessment
#   - 3 explicit genus-stage canonical taxa representing Cola and Drypetes
#
# Existing Seed Mass evidence:
#   - validated BIEN species-level values
#   - validated TRY species-level values
#   - validated external species-level rescue values
#
# This section:
#   1. reconstructs the genus-audit target list;
#   2. locates the validated BIEN and TRY species-value tables;
#   3. loads the validated external species-target values from Section 9.4L;
#   4. standardizes biological species names;
#   5. prevents aliases / spelling variants from being counted as
#      independent species;
#   6. pools accepted species-level Seed Mass evidence;
#   7. summarizes evidence depth by target genus;
#   8. classifies genus evidence as:
#
#        STRONG_GENUS_EVIDENCE       >= 3 independent species
#        MODERATE_GENUS_EVIDENCE      2 independent species
#        SINGLE_SPECIES_GENUS_EVIDENCE 1 independent species
#        NO_GENUS_EVIDENCE            0 species
#
# IMPORTANT
# ---------
# NO genus Seed Mass estimate is assigned in this section.
# NO family estimate is assigned.
# NO residual canonical taxon is resolved here.
# =============================================================================


from pathlib import Path
import pandas as pd
import numpy as np
import re


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# Section 9.4M outputs
BIOLOGICAL_CLASSIFICATION_FILE = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_biological_target_classification.csv"
)

GENUS_STAGE_FILE = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_genus_stage_targets.csv"
)


# Section 9.4L validated external rescue values
EXTERNAL_TARGET_VALUES_FILE = (
    OUT_DIR /
    "omo_seed_mass_non_try_external_species_target_values.csv"
)


# Outputs from 9.4N
OUT_GENUS_AUDIT_TARGETS = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_audit_targets.csv"
)

OUT_SPECIES_EVIDENCE_POOL = (
    OUT_DIR /
    "omo_seed_mass_genus_audit_species_evidence_pool.csv"
)

OUT_GENUS_EVIDENCE_DETAIL = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_detail.csv"
)

OUT_GENUS_EVIDENCE_AUDIT = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_audit.csv"
)

OUT_GENUS_EVIDENCE_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_summary.csv"
)


# =============================================================================
# 2. HELPERS
# =============================================================================

def clean_text_series(series):

    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "": pd.NA,
                "nan": pd.NA,
                "None": pd.NA,
                "<NA>": pd.NA,
            }
        )
    )


def find_column(
    df,
    candidates,
    required=True
):

    lookup = {
        str(col)
        .strip()
        .lower()
        .replace(" ", "_"):
            col
        for col in df.columns
    }

    for candidate in candidates:

        if candidate in df.columns:
            return candidate

        key = (
            str(candidate)
            .strip()
            .lower()
            .replace(" ", "_")
        )

        if key in lookup:
            return lookup[key]

    if required:

        raise KeyError(
            "Required field not found.\n"
            f"Candidates tested: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


def first_token_genus(name):

    if pd.isna(name):
        return pd.NA

    value = str(name).strip()

    if not value:
        return pd.NA

    return value.split()[0]


def join_unique(series):

    values = (
        series
        .dropna()
        .astype(str)
        .str.strip()
    )

    values = [
        x
        for x in values
        if x
    ]

    return " | ".join(
        sorted(
            set(values)
        )
    )


def candidate_seed_value_column(df):

    candidates = [
        "Seed_Mass_mg",
        "Standardized_Trait_Value",
        "Median_Seed_Mass_mg",
        "Median_Seed_Mass",
        "SeedMass_mg",
        "Seed_Mass",
        "Trait_Value",
        "Value_mg",
    ]

    return find_column(
        df,
        candidates,
        required=False
    )


def candidate_biological_name_column(df):

    # Priority is important.
    #
    # Prefer the accepted / source / resolved biological name rather than
    # the project canonical string whenever possible.

    candidates = [
        "Final_Search_Name",
        "Preferred_TRY_Query_Name",
        "Resolved_TRY_Query_Name",
        "TRY_Query_Name",
        "AccSpeciesName",
        "Accepted_Species_Name",
        "Resolved_Species_Name",
        "Trait_Source_Name",
        "Source_Species_Name",
        "BIEN_Species",
        "Species",
        "Canonical_Species",
    ]

    return find_column(
        df,
        candidates,
        required=False
    )


# =============================================================================
# 3. LOAD SECTION 9.4M CLASSIFICATION TABLES
# =============================================================================

for path in [
    BIOLOGICAL_CLASSIFICATION_FILE,
    GENUS_STAGE_FILE,
    EXTERNAL_TARGET_VALUES_FILE,
]:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


biological_targets = pd.read_csv(
    BIOLOGICAL_CLASSIFICATION_FILE,
    low_memory=False
)

genus_stage = pd.read_csv(
    GENUS_STAGE_FILE,
    low_memory=False
)

external_values = pd.read_csv(
    EXTERNAL_TARGET_VALUES_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4N — RESIDUAL GENUS-EVIDENCE "
    "AVAILABILITY AUDIT"
)
print("=" * 100)


# =============================================================================
# 4. RECONSTRUCT GENUS-AUDIT TARGET LIST
# =============================================================================

CONGENERIC_CLASS = (
    "CONGENERIC_EVIDENCE_ASSESSMENT"
)


congeneric_targets = (
    biological_targets[
        biological_targets[
            "Residual_Classification"
        ].eq(
            CONGENERIC_CLASS
        )
    ]
    .copy()
)


assert len(congeneric_targets) == 23, (
    "Expected 23 biological targets classified for "
    f"congeneric assessment; found {len(congeneric_targets)}."
)


congeneric_targets[
    "Genus_Search_Target"
] = (
    congeneric_targets[
        "Final_Search_Name"
    ]
    .map(
        first_token_genus
    )
)


species_genus_targets = (
    congeneric_targets[
        "Genus_Search_Target"
    ]
    .dropna()
    .drop_duplicates()
    .sort_values()
    .tolist()
)


assert len(species_genus_targets) == 21, (
    "Expected the 23 congeneric biological targets to represent "
    f"21 unique genera; found {len(species_genus_targets)}."
)


explicit_genus_targets = (
    genus_stage[
        "Genus_Search_Target"
    ]
    .dropna()
    .astype(str)
    .str.strip()
    .drop_duplicates()
    .sort_values()
    .tolist()
)


assert set(
    explicit_genus_targets
) == {
    "Cola",
    "Drypetes",
}


all_genus_targets = sorted(
    set(
        species_genus_targets
    )
    |
    set(
        explicit_genus_targets
    )
)


assert len(all_genus_targets) == 23, (
    "Expected 23 unique genera in the complete genus audit; "
    f"found {len(all_genus_targets)}."
)


# =============================================================================
# 5. BUILD AUDIT-TARGET REGISTER
# =============================================================================

genus_target_rows = []


for genus in all_genus_targets:

    species_target_rows = congeneric_targets[
        congeneric_targets[
            "Genus_Search_Target"
        ].eq(
            genus
        )
    ]

    explicit_rows = genus_stage[
        genus_stage[
            "Genus_Search_Target"
        ].eq(
            genus
        )
    ]

    genus_target_rows.append(
        {
            "Genus_Search_Target":
                genus,

            "N_Residual_Biological_Species_Targets":
                len(species_target_rows),

            "Residual_Biological_Targets":
                join_unique(
                    species_target_rows[
                        "Final_Search_Name"
                    ]
                ),

            "N_Explicit_Genus_Canonical_Taxa":
                len(explicit_rows),

            "Explicit_Genus_Canonical_Taxa":
                join_unique(
                    explicit_rows[
                        "Canonical_Species"
                    ]
                ),

            "Genus_Audit_Reason":
                (
                    "SPECIES_CONGENERIC_ASSESSMENT"
                    if len(species_target_rows) > 0
                    else
                    "EXPLICIT_GENUS_LEVEL_TARGET"
                ),
        }
    )


genus_audit_targets = pd.DataFrame(
    genus_target_rows
)


print("\nGenus audit targets")
print("-" * 100)

display(
    genus_audit_targets
)


# =============================================================================
# 6. LOCATE VALIDATED BIEN AND TRY SPECIES-VALUE TABLES
# =============================================================================
#
# Exact filenames changed during Sections 9.3L-R1/R2/R3 and 9.3G-K.
# Rather than hard-coding a stale filename, inspect CSV outputs and select
# tables by validated structure.
#
# BIEN expectation:
#     28 unique canonical taxa
#
# TRY expectation:
#     46 canonical taxa represented by accepted TRY species evidence
#
# Exclude summaries, reviews and raw measurement files.
# =============================================================================

all_csv_files = sorted(
    OUT_DIR.glob("*.csv")
)


def score_candidate_file(
    path,
    source_keyword,
    expected_rows
):

    name = path.name.lower()

    if source_keyword not in name:
        return None

    exclusion_terms = [
        "summary",
        "validation",
        "diagnostic",
        "review",
        "routing",
        "manifest",
        "checklist",
        "raw",
        "metadata",
        "provenance",
        "unresolved",
        "request",
    ]

    if any(
        term in name
        for term in exclusion_terms
    ):
        return None

    try:
        df = pd.read_csv(
            path,
            low_memory=False
        )
    except Exception:
        return None

    canonical_col = find_column(
        df,
        [
            "Canonical_Species",
            "Canonical Species",
        ],
        required=False
    )

    value_col = candidate_seed_value_column(
        df
    )

    if (
        canonical_col is None
        or value_col is None
    ):
        return None

    numeric = pd.to_numeric(
        df[value_col],
        errors="coerce"
    )

    valid_rows = (
        df[
            canonical_col
        ].notna()
        &
        numeric.notna()
        &
        numeric.gt(0)
    )

    n_valid = int(
        valid_rows.sum()
    )

    n_unique = int(
        df.loc[
            valid_rows,
            canonical_col
        ].astype(str)
        .nunique()
    )

    score = 0

    # Strong weight for expected validated structure.
    if n_unique == expected_rows:
        score += 100

    if n_valid == expected_rows:
        score += 40

    # Prefer filenames indicating final/consolidated/value tables.
    preference_terms = [
        "final",
        "consolidated",
        "canonical",
        "value",
        "values",
        "accepted",
    ]

    score += sum(
        5
        for term in preference_terms
        if term in name
    )

    return {
        "Path": path,
        "Score": score,
        "N_Valid_Rows": n_valid,
        "N_Unique_Canonical": n_unique,
        "Canonical_Column": canonical_col,
        "Value_Column": value_col,
    }


bien_candidates = []

try_candidates = []


for path in all_csv_files:

    result = score_candidate_file(
        path,
        source_keyword="bien",
        expected_rows=28
    )

    if result is not None:
        bien_candidates.append(
            result
        )


    result = score_candidate_file(
        path,
        source_keyword="try",
        expected_rows=46
    )

    if result is not None:
        try_candidates.append(
            result
        )


bien_candidates = sorted(
    bien_candidates,
    key=lambda x: x["Score"],
    reverse=True
)

try_candidates = sorted(
    try_candidates,
    key=lambda x: x["Score"],
    reverse=True
)


if len(bien_candidates) == 0:

    raise FileNotFoundError(
        "No suitable validated BIEN Seed Mass species-value "
        "table could be identified."
    )


if len(try_candidates) == 0:

    raise FileNotFoundError(
        "No suitable validated TRY Seed Mass species-value "
        "table could be identified."
    )


BIEN_FILE = bien_candidates[0]["Path"]

TRY_FILE = try_candidates[0]["Path"]


print("\nAutomatically selected validated species-value inputs")
print("-" * 100)

print(
    f"BIEN : {BIEN_FILE}"
)

print(
    f"TRY  : {TRY_FILE}"
)


# =============================================================================
# 7. LOAD SELECTED BIEN AND TRY TABLES
# =============================================================================

bien = pd.read_csv(
    BIEN_FILE,
    low_memory=False
)

try_values = pd.read_csv(
    TRY_FILE,
    low_memory=False
)


BIEN_CANON_COL = find_column(
    bien,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)

TRY_CANON_COL = find_column(
    try_values,
    [
        "Canonical_Species",
        "Canonical Species",
    ]
)


BIEN_VALUE_COL = candidate_seed_value_column(
    bien
)

TRY_VALUE_COL = candidate_seed_value_column(
    try_values
)


assert BIEN_VALUE_COL is not None

assert TRY_VALUE_COL is not None


BIEN_BIOLOGICAL_NAME_COL = (
    candidate_biological_name_column(
        bien
    )
)

TRY_BIOLOGICAL_NAME_COL = (
    candidate_biological_name_column(
        try_values
    )
)


assert BIEN_BIOLOGICAL_NAME_COL is not None

assert TRY_BIOLOGICAL_NAME_COL is not None


print("\nSelected source fields")
print("-" * 100)

print(
    f"BIEN canonical field       : {BIEN_CANON_COL}"
)

print(
    f"BIEN biological-name field : {BIEN_BIOLOGICAL_NAME_COL}"
)

print(
    f"BIEN value field           : {BIEN_VALUE_COL}"
)

print()

print(
    f"TRY canonical field        : {TRY_CANON_COL}"
)

print(
    f"TRY biological-name field  : {TRY_BIOLOGICAL_NAME_COL}"
)

print(
    f"TRY value field            : {TRY_VALUE_COL}"
)


# =============================================================================
# 8. STANDARDIZE BIEN EVIDENCE
# =============================================================================

bien_pool = pd.DataFrame({
    "Canonical_Species":
        clean_text_series(
            bien[
                BIEN_CANON_COL
            ]
        ),

    "Biological_Species_Name":
        clean_text_series(
            bien[
                BIEN_BIOLOGICAL_NAME_COL
            ]
        ),

    "Seed_Mass_mg":
        pd.to_numeric(
            bien[
                BIEN_VALUE_COL
            ],
            errors="coerce"
        ),

    "Evidence_Source":
        "BIEN",
})


# =============================================================================
# 9. STANDARDIZE TRY EVIDENCE
# =============================================================================

try_pool = pd.DataFrame({
    "Canonical_Species":
        clean_text_series(
            try_values[
                TRY_CANON_COL
            ]
        ),

    "Biological_Species_Name":
        clean_text_series(
            try_values[
                TRY_BIOLOGICAL_NAME_COL
            ]
        ),

    "Seed_Mass_mg":
        pd.to_numeric(
            try_values[
                TRY_VALUE_COL
            ],
            errors="coerce"
        ),

    "Evidence_Source":
        "TRY",
})


# =============================================================================
# 10. STANDARDIZE EXTERNAL EVIDENCE
# =============================================================================

EXT_NAME_COL = find_column(
    external_values,
    [
        "Final_Search_Name",
        "Canonical_Species",
    ]
)

EXT_VALUE_COL = find_column(
    external_values,
    [
        "Seed_Mass_mg",
        "Median_Accepted_mg",
    ]
)


external_pool = pd.DataFrame({
    "Canonical_Species":
        clean_text_series(
            external_values[
                EXT_NAME_COL
            ]
        ),

    "Biological_Species_Name":
        clean_text_series(
            external_values[
                EXT_NAME_COL
            ]
        ),

    "Seed_Mass_mg":
        pd.to_numeric(
            external_values[
                EXT_VALUE_COL
            ],
            errors="coerce"
        ),

    "Evidence_Source":
        "EXTERNAL_SPECIES_RESCUE",
})


# =============================================================================
# 11. COMBINE SPECIES EVIDENCE POOL
# =============================================================================

species_pool = pd.concat(
    [
        bien_pool,
        try_pool,
        external_pool,
    ],
    ignore_index=True,
    sort=False
)


species_pool = (
    species_pool[
        species_pool[
            "Biological_Species_Name"
        ].notna()
        &
        species_pool[
            "Seed_Mass_mg"
        ].notna()
        &
        species_pool[
            "Seed_Mass_mg"
        ].gt(0)
    ]
    .copy()
)


species_pool[
    "Biological_Genus"
] = (
    species_pool[
        "Biological_Species_Name"
    ]
    .map(
        first_token_genus
    )
)


# =============================================================================
# 12. COLLAPSE ALIASES WITHIN EACH SOURCE
# =============================================================================
#
# Several Omo canonical rows represent spelling variants of one biological
# species. They must not inflate genus evidence depth.
#
# Collapse to:
#
#     Evidence_Source × Biological_Species_Name
#
# before calculating genus evidence.
# =============================================================================

species_pool_collapsed = (
    species_pool
    .groupby(
        [
            "Evidence_Source",
            "Biological_Genus",
            "Biological_Species_Name",
        ],
        as_index=False
    )
    .agg(
        N_Canonical_Rows=(
            "Canonical_Species",
            "size"
        ),

        Canonical_Species_Represented=(
            "Canonical_Species",
            join_unique
        ),

        Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),
    )
)


# =============================================================================
# 13. COLLAPSE ANY SAME BIOLOGICAL SPECIES ACROSS SOURCES
# =============================================================================
#
# Normally BIEN/TRY/external species sets are non-overlapping because of the
# acquisition hierarchy, but this second collapse protects against accidental
# duplication.
# =============================================================================

biological_species_pool = (
    species_pool_collapsed
    .groupby(
        [
            "Biological_Genus",
            "Biological_Species_Name",
        ],
        as_index=False
    )
    .agg(
        N_Evidence_Sources=(
            "Evidence_Source",
            "nunique"
        ),

        Evidence_Sources=(
            "Evidence_Source",
            join_unique
        ),

        Canonical_Species_Represented=(
            "Canonical_Species_Represented",
            join_unique
        ),

        Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),
    )
)


print("\nValidated biological-species evidence pool")
print("-" * 100)

print(
    f"BIEN evidence rows             : {len(bien_pool)}"
)

print(
    f"TRY evidence rows              : {len(try_pool)}"
)

print(
    f"External biological targets    : {len(external_pool)}"
)

print(
    f"Combined source rows           : {len(species_pool)}"
)

print(
    f"Independent biological species : {len(biological_species_pool)}"
)


# =============================================================================
# 14. RESTRICT DETAIL TABLE TO TARGET GENERA
# =============================================================================

genus_evidence_detail = (
    biological_species_pool[
        biological_species_pool[
            "Biological_Genus"
        ].isin(
            all_genus_targets
        )
    ]
    .copy()
    .sort_values(
        [
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nSpecies evidence available within residual target genera")
print("-" * 100)

display(
    genus_evidence_detail
)


# =============================================================================
# 15. GENUS-LEVEL EVIDENCE SUMMARY
# =============================================================================

genus_summary_existing = (
    genus_evidence_detail
    .groupby(
        "Biological_Genus",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        Species_Represented=(
            "Biological_Species_Name",
            join_unique
        ),

        N_Evidence_Sources=(
            "Evidence_Sources",
            lambda x: len(
                set(
                    " | ".join(
                        x.astype(str)
                    ).split(
                        " | "
                    )
                )
            )
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


# =============================================================================
# 16. ENSURE ZERO-EVIDENCE GENERA ARE RETAINED
# =============================================================================

genus_audit = (
    genus_audit_targets
    .merge(
        genus_summary_existing,
        left_on="Genus_Search_Target",
        right_on="Biological_Genus",
        how="left",
        validate="one_to_one"
    )
)


genus_audit[
    "N_Independent_Species"
] = (
    genus_audit[
        "N_Independent_Species"
    ]
    .fillna(0)
    .astype(int)
)


genus_audit[
    "N_Evidence_Sources"
] = (
    genus_audit[
        "N_Evidence_Sources"
    ]
    .fillna(0)
    .astype(int)
)


for col in [
    "Species_Represented",
    "Evidence_Sources",
]:

    genus_audit[col] = (
        genus_audit[col]
        .fillna("")
    )


# =============================================================================
# 17. CLASSIFY GENUS EVIDENCE DEPTH
# =============================================================================

def genus_evidence_class(n_species):

    if n_species >= 3:
        return "STRONG_GENUS_EVIDENCE"

    if n_species == 2:
        return "MODERATE_GENUS_EVIDENCE"

    if n_species == 1:
        return "SINGLE_SPECIES_GENUS_EVIDENCE"

    return "NO_GENUS_EVIDENCE"


genus_audit[
    "Genus_Evidence_Class"
] = (
    genus_audit[
        "N_Independent_Species"
    ]
    .map(
        genus_evidence_class
    )
)


# =============================================================================
# 18. ASSIGN NEXT AUDIT ACTION — NO VALUE ASSIGNMENT
# =============================================================================

def genus_next_action(row):

    n = row[
        "N_Independent_Species"
    ]

    if n >= 3:
        return (
            "CANDIDATE_FOR_GENUS_ESTIMATE_RULE_DEVELOPMENT"
        )

    if n == 2:
        return (
            "REVIEW_TWO_SPECIES_GENUS_ESTIMATE_FEASIBILITY"
        )

    if n == 1:
        return (
            "REVIEW_SINGLE_CONGENER_TRANSFER_RISK"
        )

    return (
        "NO_CONGENERIC_EVIDENCE_PROCEED_TO_FAMILY_AUDIT"
    )


genus_audit[
    "Next_Audit_Action"
] = (
    genus_audit.apply(
        genus_next_action,
        axis=1
    )
)


genus_audit[
    "Numerical_Genus_Value_Assigned"
] = False


# =============================================================================
# 19. FINAL DISPLAY
# =============================================================================

display_columns = [
    "Genus_Search_Target",
    "N_Residual_Biological_Species_Targets",
    "Residual_Biological_Targets",
    "N_Explicit_Genus_Canonical_Taxa",
    "Explicit_Genus_Canonical_Taxa",
    "N_Independent_Species",
    "Species_Represented",
    "Evidence_Sources",
    "Min_Seed_Mass_mg",
    "Median_Seed_Mass_mg",
    "Max_Seed_Mass_mg",
    "Genus_Evidence_Class",
    "Next_Audit_Action",
]


print("\nResidual genus-evidence audit")
print("-" * 100)

display(
    genus_audit[
        display_columns
    ]
    .sort_values(
        [
            "Genus_Evidence_Class",
            "Genus_Search_Target",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 20. SUMMARY BY EVIDENCE CLASS
# =============================================================================

audit_summary = (
    genus_audit[
        "Genus_Evidence_Class"
    ]
    .value_counts()
    .rename_axis(
        "Genus_Evidence_Class"
    )
    .reset_index(
        name="N_Genera"
    )
)


audit_summary[
    "Percent_of_23_Genera"
] = (
    audit_summary[
        "N_Genera"
    ]
    /
    len(genus_audit)
    *
    100
)


print("\nGenus-evidence availability summary")
print("-" * 100)

display(
    audit_summary
)


# =============================================================================
# 21. GENERA WITH NO CONGENERIC EVIDENCE
# =============================================================================

no_genus_evidence = genus_audit[
    genus_audit[
        "Genus_Evidence_Class"
    ].eq(
        "NO_GENUS_EVIDENCE"
    )
].copy()


print("\nGenera with no accepted species-level Seed Mass evidence")
print("-" * 100)

print(
    f"N genera : {len(no_genus_evidence)}"
)


if len(no_genus_evidence) > 0:

    display(
        no_genus_evidence[
            [
                "Genus_Search_Target",
                "Residual_Biological_Targets",
                "Explicit_Genus_Canonical_Taxa",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 22. GENERA WITH ONE-SPECIES EVIDENCE
# =============================================================================

single_species_evidence = genus_audit[
    genus_audit[
        "Genus_Evidence_Class"
    ].eq(
        "SINGLE_SPECIES_GENUS_EVIDENCE"
    )
].copy()


print("\nGenera represented by only one accepted congeneric species")
print("-" * 100)

print(
    f"N genera : {len(single_species_evidence)}"
)


if len(single_species_evidence) > 0:

    display(
        single_species_evidence[
            [
                "Genus_Search_Target",
                "N_Independent_Species",
                "Species_Represented",
                "Median_Seed_Mass_mg",
                "Evidence_Sources",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 23. STRONG VALIDATION
# =============================================================================

assert len(genus_audit) == 23, (
    f"Expected 23 unique genus audit targets; found {len(genus_audit)}."
)


assert genus_audit[
    "Genus_Search_Target"
].is_unique


assert set(
    genus_audit[
        "Genus_Search_Target"
    ]
) == set(
    all_genus_targets
)


assert (
    genus_audit[
        "Numerical_Genus_Value_Assigned"
    ]
    .eq(False)
    .all()
)


assert (
    genus_audit[
        "N_Independent_Species"
    ]
    .ge(0)
    .all()
)


# No biological species should occur more than once in the collapsed pool.
assert not biological_species_pool[
    [
        "Biological_Genus",
        "Biological_Species_Name",
    ]
].duplicated().any()


# External rescue should contribute exactly 11 biological species targets.
assert external_pool[
    "Biological_Species_Name"
].nunique() == 11


# =============================================================================
# 24. SECTION SUMMARY
# =============================================================================

section_summary_rows = [
    {
        "Metric":
            "Species-level congeneric-assessment targets",
        "Value":
            len(congeneric_targets),
    },

    {
        "Metric":
            "Unique genera from species congeneric targets",
        "Value":
            len(species_genus_targets),
    },

    {
        "Metric":
            "Explicit genus-stage canonical taxa",
        "Value":
            len(genus_stage),
    },

    {
        "Metric":
            "Unique explicit genus targets",
        "Value":
            len(explicit_genus_targets),
    },

    {
        "Metric":
            "Total unique genera audited",
        "Value":
            len(genus_audit),
    },

    {
        "Metric":
            "Independent biological species in pooled evidence",
        "Value":
            len(biological_species_pool),
    },

    {
        "Metric":
            "Genus Seed Mass values assigned in 9.4N",
        "Value":
            0,
    },
]


for _, row in audit_summary.iterrows():

    section_summary_rows.append(
        {
            "Metric":
                "Genera — "
                +
                row[
                    "Genus_Evidence_Class"
                ],

            "Value":
                int(
                    row[
                        "N_Genera"
                    ]
                ),
        }
    )


section_summary = pd.DataFrame(
    section_summary_rows
)


print("\nSection 9.4N summary")
print("-" * 100)

display(
    section_summary
)


# =============================================================================
# 25. EXPORT
# =============================================================================

genus_audit_targets.to_csv(
    OUT_GENUS_AUDIT_TARGETS,
    index=False
)


biological_species_pool.to_csv(
    OUT_SPECIES_EVIDENCE_POOL,
    index=False
)


genus_evidence_detail.to_csv(
    OUT_GENUS_EVIDENCE_DETAIL,
    index=False
)


genus_audit.to_csv(
    OUT_GENUS_EVIDENCE_AUDIT,
    index=False
)


section_summary.to_csv(
    OUT_GENUS_EVIDENCE_SUMMARY,
    index=False
)


# =============================================================================
# 26. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4N exports")
print("-" * 100)

print(
    f"Genus audit targets     : "
    f"{OUT_GENUS_AUDIT_TARGETS}"
)

print(
    f"Species evidence pool   : "
    f"{OUT_SPECIES_EVIDENCE_POOL}"
)

print(
    f"Genus evidence detail   : "
    f"{OUT_GENUS_EVIDENCE_DETAIL}"
)

print(
    f"Genus evidence audit    : "
    f"{OUT_GENUS_EVIDENCE_AUDIT}"
)

print(
    f"Audit summary           : "
    f"{OUT_GENUS_EVIDENCE_SUMMARY}"
)


# =============================================================================
# 27. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Residual genus-level Seed Mass evidence "
    "availability audit completed successfully."
)

print(
    f"Species congeneric-assessment targets : "
    f"{len(congeneric_targets)}"
)

print(
    f"Unique species-target genera          : "
    f"{len(species_genus_targets)}"
)

print(
    f"Explicit genus-stage targets          : "
    f"{len(explicit_genus_targets)}"
)

print(
    f"Total unique genera audited           : "
    f"{len(genus_audit)}"
)

print(
    "Genus Seed Mass values assigned       : 0"
)

print(
    "The evidence depth of every required genus "
    "has now been audited before genus-level estimation."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4N-R
# AUTHORITATIVE GENUS NORMALIZATION AND CORRECTED
# RESIDUAL GENUS-EVIDENCE AUDIT
# =============================================================================
#
# Purpose
# -------
# Correct the diagnostic genus audit produced in Section 9.4N.
#
# Section 9.4N revealed that biological genus identity could be split by
# orthographic variants in the project / BIEN evidence layer, e.g.:
#
#     Diospyrous  -> Diospyros
#
# This section:
#
#   1. loads the validated biological-species evidence pool from 9.4N;
#   2. preserves every raw biological name;
#   3. applies ONLY controlled, explicit taxonomic/string normalizations;
#   4. collapses orthographic aliases of the same biological species;
#   5. recalculates evidence depth for all 23 required genera;
#   6. compares the corrected audit with the original 9.4N audit;
#   7. exports the corrected evidence pool and audit.
#
# IMPORTANT
# ---------
# This is still an AUDIT ONLY.
#
# NO genus Seed Mass estimate is assigned.
# NO family estimate is assigned.
# NO residual taxon is resolved.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# Section 9.4N outputs
GENUS_AUDIT_TARGET_FILE = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_audit_targets.csv"
)

SPECIES_POOL_FILE = (
    OUT_DIR /
    "omo_seed_mass_genus_audit_species_evidence_pool.csv"
)

ORIGINAL_GENUS_AUDIT_FILE = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_audit.csv"
)


# 9.4N-R outputs
OUT_NORMALIZATION_DIAGNOSTIC = (
    OUT_DIR /
    "omo_seed_mass_genus_evidence_taxonomic_normalization_diagnostic.csv"
)

OUT_CORRECTED_SPECIES_POOL = (
    OUT_DIR /
    "omo_seed_mass_genus_audit_species_evidence_pool_normalized.csv"
)

OUT_CORRECTED_DETAIL = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_detail_normalized.csv"
)

OUT_CORRECTED_AUDIT = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_audit_corrected.csv"
)

OUT_AUDIT_COMPARISON = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_audit_before_after_comparison.csv"
)

OUT_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_evidence_audit_corrected_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

for path in [
    GENUS_AUDIT_TARGET_FILE,
    SPECIES_POOL_FILE,
    ORIGINAL_GENUS_AUDIT_FILE,
]:

    assert path.exists(), (
        f"Required Section 9.4N file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD 9.4N TABLES
# =============================================================================

genus_targets = pd.read_csv(
    GENUS_AUDIT_TARGET_FILE,
    low_memory=False
)

species_pool_raw = pd.read_csv(
    SPECIES_POOL_FILE,
    low_memory=False
)

original_audit = pd.read_csv(
    ORIGINAL_GENUS_AUDIT_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4N-R — AUTHORITATIVE GENUS NORMALIZATION "
    "AND CORRECTED RESIDUAL GENUS-EVIDENCE AUDIT"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Genus audit targets                : "
    f"{len(genus_targets)}"
)

print(
    f"Biological-species evidence rows   : "
    f"{len(species_pool_raw)}"
)

print(
    f"Original genus-audit rows          : "
    f"{len(original_audit)}"
)


assert len(genus_targets) == 23

assert len(original_audit) == 23


# =============================================================================
# 4. REQUIRED FIELDS
# =============================================================================

required_pool_fields = {
    "Biological_Genus",
    "Biological_Species_Name",
    "Seed_Mass_mg",
    "Evidence_Sources",
    "Canonical_Species_Represented",
}

missing_pool_fields = (
    required_pool_fields
    -
    set(species_pool_raw.columns)
)

assert not missing_pool_fields, (
    "Species evidence pool is missing required columns:\n"
    f"{sorted(missing_pool_fields)}"
)


assert (
    "Genus_Search_Target"
    in genus_targets.columns
)

assert (
    "Genus_Search_Target"
    in original_audit.columns
)


# =============================================================================
# 5. HELPER FUNCTIONS
# =============================================================================

def clean_text(value):

    if pd.isna(value):
        return pd.NA

    value = str(value).strip()

    if value in {
        "",
        "nan",
        "None",
        "<NA>",
    }:
        return pd.NA

    return value


def join_unique(series):

    values = []

    for value in series.dropna():

        text = str(value).strip()

        if not text:
            continue

        # Existing fields may already contain pipe-delimited values.
        pieces = [
            x.strip()
            for x in text.split("|")
            if x.strip()
        ]

        values.extend(
            pieces
        )

    return " | ".join(
        sorted(
            set(values)
        )
    )


def genus_from_species_name(name):

    name = clean_text(name)

    if pd.isna(name):
        return pd.NA

    return str(name).split()[0]


# =============================================================================
# 6. CONTROLLED GENUS NORMALIZATION REGISTER
# =============================================================================
#
# ONLY explicit project orthographic variants are normalized.
#
# No fuzzy matching.
# No automatic synonym invention.
# =============================================================================

GENUS_NORMALIZATION = {

    "Diospyrous":
        "Diospyros",

    "Macranga":
        "Macaranga",

    "Rothmania":
        "Rothmannia",

    "Nothospondia":
        "Nothospondias",

    "Corynante":
        "Corynanthe",

    "Scotellia":
        "Scottellia",

    "Nesodogornia":
        "Nesogordonia",

    "Nuclea":
        "Nauclea",

    "Ricinidendron":
        "Ricinodendron",

    "Termialia":
        "Terminalia",

    "Irvinga":
        "Irvingia",

    "Spondia":
        "Spondias",
}


# =============================================================================
# 7. CONTROLLED BIOLOGICAL-SPECIES NORMALIZATION
# =============================================================================
#
# These are spelling variants already present in the validated BIEN /
# project evidence layer and represent the SAME biological species.
#
# Their collapse is necessary so they are not counted as multiple independent
# congeners.
#
# =============================================================================

SPECIES_NAME_NORMALIZATION = {

    # -------------------------------------------------------------------------
    # Diospyros
    #
    # Both project strings represent the same biological species.
    # -------------------------------------------------------------------------

    "Diospyrous melistoformis":
        "Diospyros mespiliformis",

    "Diospyrous mespiliformis":
        "Diospyros mespiliformis",


    # -------------------------------------------------------------------------
    # Elaeis guineensis
    # -------------------------------------------------------------------------

    "Elaeis guineense":
        "Elaeis guineensis",

    "Elaeis guinensis":
        "Elaeis guineensis",


    # -------------------------------------------------------------------------
    # Nauclea diderrichii
    # -------------------------------------------------------------------------

    "Nauclea diderrichi":
        "Nauclea diderrichii",

    "Nuclea diderrichi":
        "Nauclea diderrichii",


    # -------------------------------------------------------------------------
    # Ricinodendron heudelotii
    # -------------------------------------------------------------------------

    "Ricinidendron heudeloti":
        "Ricinodendron heudelotii",

    "Ricinodendron heudeloti":
        "Ricinodendron heudelotii",

    "Ricinodendron heudeotii":
        "Ricinodendron heudelotii",


    # -------------------------------------------------------------------------
    # Terminalia superba
    # -------------------------------------------------------------------------

    "Termialia superba":
        "Terminalia superba",


    # -------------------------------------------------------------------------
    # Spondias mombin
    # -------------------------------------------------------------------------

    "Spondia mombin":
        "Spondias mombin",
}


# =============================================================================
# 8. PRESERVE RAW NAME STATE
# =============================================================================

species_pool = (
    species_pool_raw
    .copy()
)


species_pool[
    "Raw_Biological_Species_Name"
] = (
    species_pool[
        "Biological_Species_Name"
    ]
    .map(
        clean_text
    )
)


species_pool[
    "Raw_Biological_Genus"
] = (
    species_pool[
        "Raw_Biological_Species_Name"
    ]
    .map(
        genus_from_species_name
    )
)


# =============================================================================
# 9. NORMALIZE BIOLOGICAL SPECIES NAMES
# =============================================================================

def normalize_species_name(name):

    name = clean_text(name)

    if pd.isna(name):
        return pd.NA


    # First apply exact validated species-name correction.
    if name in SPECIES_NAME_NORMALIZATION:

        return (
            SPECIES_NAME_NORMALIZATION[
                name
            ]
        )


    pieces = str(name).split()

    if len(pieces) == 0:
        return pd.NA


    raw_genus = pieces[0]

    normalized_genus = (
        GENUS_NORMALIZATION
        .get(
            raw_genus,
            raw_genus
        )
    )


    pieces[0] = normalized_genus

    return " ".join(
        pieces
    )


species_pool[
    "Normalized_Biological_Species_Name"
] = (
    species_pool[
        "Raw_Biological_Species_Name"
    ]
    .map(
        normalize_species_name
    )
)


species_pool[
    "Normalized_Biological_Genus"
] = (
    species_pool[
        "Normalized_Biological_Species_Name"
    ]
    .map(
        genus_from_species_name
    )
)


# =============================================================================
# 10. NORMALIZATION FLAGS
# =============================================================================

species_pool[
    "Species_Name_Normalization_Applied"
] = (
    species_pool[
        "Raw_Biological_Species_Name"
    ]
    .astype("string")
    .ne(
        species_pool[
            "Normalized_Biological_Species_Name"
        ]
        .astype("string")
    )
)


species_pool[
    "Genus_Normalization_Applied"
] = (
    species_pool[
        "Raw_Biological_Genus"
    ]
    .astype("string")
    .ne(
        species_pool[
            "Normalized_Biological_Genus"
        ]
        .astype("string")
    )
)


species_pool[
    "Any_Taxonomic_Normalization_Applied"
] = (
    species_pool[
        "Species_Name_Normalization_Applied"
    ]
    |
    species_pool[
        "Genus_Normalization_Applied"
    ]
)


# =============================================================================
# 11. NORMALIZATION DIAGNOSTIC
# =============================================================================

normalization_diagnostic = (
    species_pool[
        species_pool[
            "Any_Taxonomic_Normalization_Applied"
        ]
    ][
        [
            "Raw_Biological_Genus",
            "Raw_Biological_Species_Name",
            "Normalized_Biological_Genus",
            "Normalized_Biological_Species_Name",
            "Seed_Mass_mg",
            "Evidence_Sources",
            "Canonical_Species_Represented",
        ]
    ]
    .copy()
    .sort_values(
        [
            "Normalized_Biological_Genus",
            "Normalized_Biological_Species_Name",
            "Raw_Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nControlled taxonomic normalizations applied")
print("-" * 100)

print(
    f"Evidence rows affected : "
    f"{len(normalization_diagnostic)}"
)


if len(normalization_diagnostic) > 0:

    display(
        normalization_diagnostic
    )

else:

    print("NONE")


# =============================================================================
# 12. NUMERIC VALIDATION
# =============================================================================

species_pool[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_pool[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert species_pool[
    "Seed_Mass_mg"
].notna().all()


assert species_pool[
    "Seed_Mass_mg"
].gt(0).all()


assert species_pool[
    "Normalized_Biological_Species_Name"
].notna().all()


assert species_pool[
    "Normalized_Biological_Genus"
].notna().all()


# =============================================================================
# 13. COLLAPSE ORTHOGRAPHIC ALIASES TO TRUE BIOLOGICAL SPECIES
# =============================================================================
#
# If several raw spellings resolve to one biological species, they count as
# ONE independent species.
#
# Species-level value:
#     median of all accepted values representing the normalized species.
#
# =============================================================================

normalized_species_pool = (
    species_pool
    .groupby(
        [
            "Normalized_Biological_Genus",
            "Normalized_Biological_Species_Name",
        ],
        as_index=False
    )
    .agg(
        N_PreNormalization_Rows=(
            "Raw_Biological_Species_Name",
            "size"
        ),

        Raw_Species_Names_Represented=(
            "Raw_Biological_Species_Name",
            join_unique
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),

        Canonical_Species_Represented=(
            "Canonical_Species_Represented",
            join_unique
        ),

        Min_Source_Value_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Max_Source_Value_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


# Restore standard field names for downstream compatibility.
normalized_species_pool[
    "Biological_Genus"
] = normalized_species_pool[
    "Normalized_Biological_Genus"
]


normalized_species_pool[
    "Biological_Species_Name"
] = normalized_species_pool[
    "Normalized_Biological_Species_Name"
]


# =============================================================================
# 14. DUPLICATE-COLLAPSE DIAGNOSTIC
# =============================================================================

collapsed_aliases = normalized_species_pool[
    normalized_species_pool[
        "N_PreNormalization_Rows"
    ].gt(1)
].copy()


print("\nBiological species represented by multiple pre-normalization rows")
print("-" * 100)

print(
    f"N biological species : "
    f"{len(collapsed_aliases)}"
)


if len(collapsed_aliases) > 0:

    display(
        collapsed_aliases[
            [
                "Biological_Genus",
                "Biological_Species_Name",
                "N_PreNormalization_Rows",
                "Raw_Species_Names_Represented",
                "Seed_Mass_mg",
                "Evidence_Sources",
                "Canonical_Species_Represented",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 15. VALIDATE UNIQUE BIOLOGICAL SPECIES
# =============================================================================

assert not normalized_species_pool[
    [
        "Biological_Genus",
        "Biological_Species_Name",
    ]
].duplicated().any()


print("\nSpecies evidence-pool normalization")
print("-" * 100)

print(
    f"Pre-normalization biological rows : "
    f"{len(species_pool_raw)}"
)

print(
    f"Normalized independent species    : "
    f"{len(normalized_species_pool)}"
)

print(
    f"Alias/spelling rows collapsed     : "
    f"{len(species_pool_raw) - len(normalized_species_pool)}"
)


# =============================================================================
# 16. NORMALIZE AUDIT-TARGET GENERA
# =============================================================================

genus_targets[
    "Raw_Genus_Search_Target"
] = (
    genus_targets[
        "Genus_Search_Target"
    ]
    .astype(str)
    .str.strip()
)


genus_targets[
    "Genus_Search_Target"
] = (
    genus_targets[
        "Raw_Genus_Search_Target"
    ]
    .map(
        lambda x:
            GENUS_NORMALIZATION
            .get(
                x,
                x
            )
    )
)


# Should still remain 23 unique biological genera.
assert (
    genus_targets[
        "Genus_Search_Target"
    ].nunique()
    ==
    23
)


# =============================================================================
# 17. REBUILD TARGET-GENUS EVIDENCE DETAIL
# =============================================================================

target_genus_set = set(
    genus_targets[
        "Genus_Search_Target"
    ]
)


corrected_detail = (
    normalized_species_pool[
        normalized_species_pool[
            "Biological_Genus"
        ].isin(
            target_genus_set
        )
    ]
    .copy()
    .sort_values(
        [
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nCorrected species evidence within residual target genera")
print("-" * 100)

display(
    corrected_detail[
        [
            "Biological_Genus",
            "Biological_Species_Name",
            "N_PreNormalization_Rows",
            "Raw_Species_Names_Represented",
            "Evidence_Sources",
            "Canonical_Species_Represented",
            "Seed_Mass_mg",
        ]
    ]
)


# =============================================================================
# 18. REBUILD GENUS SUMMARY
# =============================================================================

genus_summary = (
    corrected_detail
    .groupby(
        "Biological_Genus",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        Species_Represented=(
            "Biological_Species_Name",
            join_unique
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


# =============================================================================
# 19. MERGE ZERO-EVIDENCE GENERA BACK INTO AUDIT
# =============================================================================

corrected_audit = (
    genus_targets
    .drop(
        columns=[
            "Raw_Genus_Search_Target"
        ]
    )
    .merge(
        genus_summary,
        left_on="Genus_Search_Target",
        right_on="Biological_Genus",
        how="left",
        validate="one_to_one"
    )
)


corrected_audit[
    "N_Independent_Species"
] = (
    corrected_audit[
        "N_Independent_Species"
    ]
    .fillna(0)
    .astype(int)
)


for col in [
    "Species_Represented",
    "Evidence_Sources",
]:

    corrected_audit[col] = (
        corrected_audit[col]
        .fillna("")
    )


# =============================================================================
# 20. EVIDENCE-DEPTH CLASSIFICATION
# =============================================================================

def classify_genus_evidence(n_species):

    if n_species >= 3:

        return (
            "STRONG_GENUS_EVIDENCE"
        )

    if n_species == 2:

        return (
            "MODERATE_GENUS_EVIDENCE"
        )

    if n_species == 1:

        return (
            "SINGLE_SPECIES_GENUS_EVIDENCE"
        )

    return (
        "NO_GENUS_EVIDENCE"
    )


corrected_audit[
    "Genus_Evidence_Class"
] = (
    corrected_audit[
        "N_Independent_Species"
    ]
    .map(
        classify_genus_evidence
    )
)


# =============================================================================
# 21. NEXT AUDIT ACTION
# =============================================================================

def next_audit_action(n_species):

    if n_species >= 3:

        return (
            "CANDIDATE_FOR_GENUS_ESTIMATE_RULE_DEVELOPMENT"
        )

    if n_species == 2:

        return (
            "REVIEW_TWO_SPECIES_GENUS_ESTIMATE_FEASIBILITY"
        )

    if n_species == 1:

        return (
            "REVIEW_SINGLE_CONGENER_TRANSFER_RISK"
        )

    return (
        "NO_CONGENERIC_EVIDENCE_PROCEED_TO_FAMILY_AUDIT"
    )


corrected_audit[
    "Next_Audit_Action"
] = (
    corrected_audit[
        "N_Independent_Species"
    ]
    .map(
        next_audit_action
    )
)


corrected_audit[
    "Numerical_Genus_Value_Assigned"
] = False


# =============================================================================
# 22. CORRECTED AUDIT DISPLAY
# =============================================================================

print("\nCorrected residual genus-evidence audit")
print("-" * 100)

display(
    corrected_audit[
        [
            "Genus_Search_Target",
            "N_Residual_Biological_Species_Targets",
            "Residual_Biological_Targets",
            "N_Explicit_Genus_Canonical_Taxa",
            "Explicit_Genus_Canonical_Taxa",
            "N_Independent_Species",
            "Species_Represented",
            "Evidence_Sources",
            "Min_Seed_Mass_mg",
            "Median_Seed_Mass_mg",
            "Max_Seed_Mass_mg",
            "Genus_Evidence_Class",
            "Next_Audit_Action",
        ]
    ]
    .sort_values(
        [
            "Genus_Evidence_Class",
            "Genus_Search_Target",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 23. SPECIAL DIAGNOSTIC — DIOSPYROS
# =============================================================================

diospyros_corrected = corrected_audit[
    corrected_audit[
        "Genus_Search_Target"
    ].eq(
        "Diospyros"
    )
].copy()


assert len(
    diospyros_corrected
) == 1


print("\nDiospyros correction diagnostic")
print("-" * 100)

display(
    diospyros_corrected[
        [
            "Genus_Search_Target",
            "N_Independent_Species",
            "Species_Represented",
            "Evidence_Sources",
            "Min_Seed_Mass_mg",
            "Median_Seed_Mass_mg",
            "Max_Seed_Mass_mg",
            "Genus_Evidence_Class",
            "Next_Audit_Action",
        ]
    ]
)


assert (
    diospyros_corrected[
        "N_Independent_Species"
    ].iloc[0]
    >= 1
), (
    "Diospyros still has zero evidence after normalization. "
    "Inspect the normalization diagnostic before proceeding."
)


# =============================================================================
# 24. ORIGINAL-vs-CORRECTED AUDIT COMPARISON
# =============================================================================

comparison = (
    original_audit[
        [
            "Genus_Search_Target",
            "N_Independent_Species",
            "Species_Represented",
            "Median_Seed_Mass_mg",
            "Genus_Evidence_Class",
        ]
    ]
    .rename(
        columns={
            "N_Independent_Species":
                "Original_N_Independent_Species",

            "Species_Represented":
                "Original_Species_Represented",

            "Median_Seed_Mass_mg":
                "Original_Median_Seed_Mass_mg",

            "Genus_Evidence_Class":
                "Original_Genus_Evidence_Class",
        }
    )
    .merge(
        corrected_audit[
            [
                "Genus_Search_Target",
                "N_Independent_Species",
                "Species_Represented",
                "Median_Seed_Mass_mg",
                "Genus_Evidence_Class",
            ]
        ]
        .rename(
            columns={
                "N_Independent_Species":
                    "Corrected_N_Independent_Species",

                "Species_Represented":
                    "Corrected_Species_Represented",

                "Median_Seed_Mass_mg":
                    "Corrected_Median_Seed_Mass_mg",

                "Genus_Evidence_Class":
                    "Corrected_Genus_Evidence_Class",
            }
        ),
        on="Genus_Search_Target",
        how="outer",
        validate="one_to_one"
    )
)


comparison[
    "N_Species_Change"
] = (
    comparison[
        "Corrected_N_Independent_Species"
    ]
    -
    comparison[
        "Original_N_Independent_Species"
    ]
)


comparison[
    "Evidence_Class_Changed"
] = (
    comparison[
        "Original_Genus_Evidence_Class"
    ]
    .astype(str)
    .ne(
        comparison[
            "Corrected_Genus_Evidence_Class"
        ]
        .astype(str)
    )
)


comparison[
    "Audit_Result_Changed"
] = (
    comparison[
        "N_Species_Change"
    ].ne(0)
    |
    comparison[
        "Evidence_Class_Changed"
    ]
)


changed_audit_rows = comparison[
    comparison[
        "Audit_Result_Changed"
    ]
].copy()


print("\nGenus audit results changed by normalization")
print("-" * 100)

print(
    f"N genera changed : "
    f"{len(changed_audit_rows)}"
)


if len(changed_audit_rows) > 0:

    display(
        changed_audit_rows
    )

else:

    print("NONE")


# =============================================================================
# 25. CORRECTED EVIDENCE-CLASS SUMMARY
# =============================================================================

corrected_summary = (
    corrected_audit[
        "Genus_Evidence_Class"
    ]
    .value_counts()
    .rename_axis(
        "Genus_Evidence_Class"
    )
    .reset_index(
        name="N_Genera"
    )
)


corrected_summary[
    "Percent_of_23_Genera"
] = (
    corrected_summary[
        "N_Genera"
    ]
    /
    23
    *
    100
)


print("\nCorrected genus-evidence availability summary")
print("-" * 100)

display(
    corrected_summary
)


# =============================================================================
# 26. NO-EVIDENCE GENERA AFTER CORRECTION
# =============================================================================

no_evidence = corrected_audit[
    corrected_audit[
        "Genus_Evidence_Class"
    ].eq(
        "NO_GENUS_EVIDENCE"
    )
].copy()


print("\nGenera with no congeneric evidence after normalization")
print("-" * 100)

print(
    f"N genera : "
    f"{len(no_evidence)}"
)


if len(no_evidence) > 0:

    display(
        no_evidence[
            [
                "Genus_Search_Target",
                "Residual_Biological_Targets",
                "Explicit_Genus_Canonical_Taxa",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 27. ONE-SPECIES EVIDENCE AFTER CORRECTION
# =============================================================================

single_evidence = corrected_audit[
    corrected_audit[
        "Genus_Evidence_Class"
    ].eq(
        "SINGLE_SPECIES_GENUS_EVIDENCE"
    )
].copy()


print("\nSingle-species genus evidence after normalization")
print("-" * 100)

print(
    f"N genera : "
    f"{len(single_evidence)}"
)


if len(single_evidence) > 0:

    display(
        single_evidence[
            [
                "Genus_Search_Target",
                "Species_Represented",
                "Median_Seed_Mass_mg",
                "Evidence_Sources",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 28. TWO-OR-MORE SPECIES EVIDENCE
# =============================================================================

multi_evidence = corrected_audit[
    corrected_audit[
        "N_Independent_Species"
    ].ge(2)
].copy()


print("\nGenera with at least two independent species")
print("-" * 100)

print(
    f"N genera : "
    f"{len(multi_evidence)}"
)


if len(multi_evidence) > 0:

    display(
        multi_evidence[
            [
                "Genus_Search_Target",
                "N_Independent_Species",
                "Species_Represented",
                "Min_Seed_Mass_mg",
                "Median_Seed_Mass_mg",
                "Max_Seed_Mass_mg",
                "Evidence_Sources",
                "Genus_Evidence_Class",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 29. STRONG VALIDATION
# =============================================================================

assert len(
    corrected_audit
) == 23


assert corrected_audit[
    "Genus_Search_Target"
].is_unique


assert (
    corrected_audit[
        "Numerical_Genus_Value_Assigned"
    ]
    .eq(False)
    .all()
)


assert (
    corrected_audit[
        "N_Independent_Species"
    ]
    .ge(0)
    .all()
)


assert (
    corrected_summary[
        "N_Genera"
    ].sum()
    ==
    23
)


# Critical correction:
# Diospyros must no longer be falsely classified as zero-evidence.
assert (
    diospyros_corrected[
        "Genus_Evidence_Class"
    ].iloc[0]
    !=
    "NO_GENUS_EVIDENCE"
)


# No duplicate normalized biological species may remain.
assert not normalized_species_pool[
    [
        "Biological_Genus",
        "Biological_Species_Name",
    ]
].duplicated().any()


# No genus-level Seed Mass values have been assigned.
assert (
    corrected_audit[
        "Numerical_Genus_Value_Assigned"
    ].sum()
    ==
    0
)


# =============================================================================
# 30. SECTION SUMMARY
# =============================================================================

section_summary_rows = [

    {
        "Metric":
            "Genus audit targets",
        "Value":
            len(corrected_audit),
    },

    {
        "Metric":
            "Pre-normalization biological evidence rows",
        "Value":
            len(species_pool_raw),
    },

    {
        "Metric":
            "Normalized independent biological species",
        "Value":
            len(normalized_species_pool),
    },

    {
        "Metric":
            "Alias/spelling rows collapsed",
        "Value":
            (
                len(species_pool_raw)
                -
                len(normalized_species_pool)
            ),
    },

    {
        "Metric":
            "Evidence rows with taxonomic normalization",
        "Value":
            len(normalization_diagnostic),
    },

    {
        "Metric":
            "Genus audit results changed",
        "Value":
            len(changed_audit_rows),
    },

    {
        "Metric":
            "Diospyros independent evidence species",
        "Value":
            int(
                diospyros_corrected[
                    "N_Independent_Species"
                ].iloc[0]
            ),
    },

    {
        "Metric":
            "Genus Seed Mass values assigned in 9.4N-R",
        "Value":
            0,
    },
]


for _, row in corrected_summary.iterrows():

    section_summary_rows.append(
        {
            "Metric":
                "Corrected genera — "
                +
                row[
                    "Genus_Evidence_Class"
                ],

            "Value":
                int(
                    row[
                        "N_Genera"
                    ]
                ),
        }
    )


section_summary = pd.DataFrame(
    section_summary_rows
)


print("\nSection 9.4N-R summary")
print("-" * 100)

display(
    section_summary
)


# =============================================================================
# 31. EXPORT
# =============================================================================

normalization_diagnostic.to_csv(
    OUT_NORMALIZATION_DIAGNOSTIC,
    index=False
)


normalized_species_pool.to_csv(
    OUT_CORRECTED_SPECIES_POOL,
    index=False
)


corrected_detail.to_csv(
    OUT_CORRECTED_DETAIL,
    index=False
)


corrected_audit.to_csv(
    OUT_CORRECTED_AUDIT,
    index=False
)


comparison.to_csv(
    OUT_AUDIT_COMPARISON,
    index=False
)


section_summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 32. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4N-R exports")
print("-" * 100)

print(
    f"Normalization diagnostic : "
    f"{OUT_NORMALIZATION_DIAGNOSTIC}"
)

print(
    f"Normalized species pool  : "
    f"{OUT_CORRECTED_SPECIES_POOL}"
)

print(
    f"Corrected evidence detail : "
    f"{OUT_CORRECTED_DETAIL}"
)

print(
    f"Corrected genus audit     : "
    f"{OUT_CORRECTED_AUDIT}"
)

print(
    f"Before/after comparison   : "
    f"{OUT_AUDIT_COMPARISON}"
)

print(
    f"Corrected audit summary   : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 33. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Authoritative genus normalization and corrected "
    "residual genus-evidence audit completed successfully."
)

print(
    f"Genus audit targets                     : "
    f"{len(corrected_audit)}"
)

print(
    f"Normalized independent biological taxa  : "
    f"{len(normalized_species_pool)}"
)

print(
    f"Taxonomic normalization rows            : "
    f"{len(normalization_diagnostic)}"
)

print(
    f"Genus audit results changed             : "
    f"{len(changed_audit_rows)}"
)

print(
    f"Diospyros evidence species after correction : "
    f"{int(diospyros_corrected['N_Independent_Species'].iloc[0])}"
)

print(
    "Genus Seed Mass values assigned              : 0"
)

print(
    "Corrected genus evidence depth is now ready "
    "for genus-transfer rule development."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4O
# GENUS-LEVEL SEED MASS TRANSFER RULE DEVELOPMENT AND CANDIDATE ASSIGNMENT
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. SECTION HEADER
# =============================================================================

print("=" * 100)
print(
    "SECTION 9.4O — GENUS-LEVEL SEED MASS TRANSFER RULE DEVELOPMENT "
    "AND CANDIDATE ASSIGNMENT"
)
print("=" * 100)


# =============================================================================
# 2. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

CORRECTED_GENUS_AUDIT_FILE = (
    OUTPUT_DIR
    / "omo_seed_mass_residual_genus_evidence_audit_corrected.csv"
)

NORMALIZED_SPECIES_POOL_FILE = (
    OUTPUT_DIR
    / "omo_seed_mass_genus_audit_species_evidence_pool_normalized.csv"
)

CANONICAL_CLASSIFICATION_FILE = (
    OUTPUT_DIR
    / "omo_seed_mass_post_external_rescue_canonical_classification.csv"
)

OUT_TRANSFER_RULES = (
    OUTPUT_DIR
    / "omo_seed_mass_genus_transfer_rules.csv"
)

OUT_GENUS_ASSIGNMENTS = (
    OUTPUT_DIR
    / "omo_seed_mass_genus_level_assignments.csv"
)

OUT_SINGLE_CONGENER_REVIEW = (
    OUTPUT_DIR
    / "omo_seed_mass_single_congener_transfer_review.csv"
)

OUT_FAMILY_STAGE_TARGETS = (
    OUTPUT_DIR
    / "omo_seed_mass_family_stage_targets.csv"
)

OUT_CANONICAL_GENUS_ASSIGNMENTS = (
    OUTPUT_DIR
    / "omo_seed_mass_canonical_genus_assignments.csv"
)

OUT_SUMMARY = (
    OUTPUT_DIR
    / "omo_seed_mass_genus_transfer_summary.csv"
)


# =============================================================================
# 3. VALIDATE REQUIRED INPUTS
# =============================================================================

required_files = [
    CORRECTED_GENUS_AUDIT_FILE,
    NORMALIZED_SPECIES_POOL_FILE,
    CANONICAL_CLASSIFICATION_FILE,
]

for file_path in required_files:
    if not file_path.exists():
        raise FileNotFoundError(
            "Required Section 9.4O input file not found:\n"
            f"{file_path}"
        )


# =============================================================================
# 4. LOAD AUTHORITATIVE INPUTS
# =============================================================================

genus_audit = pd.read_csv(
    CORRECTED_GENUS_AUDIT_FILE
)

species_pool = pd.read_csv(
    NORMALIZED_SPECIES_POOL_FILE
)

canonical_classification = pd.read_csv(
    CANONICAL_CLASSIFICATION_FILE
)


print("\nInput structure")
print("-" * 100)
print(
    f"Corrected genus audit rows       : {len(genus_audit)}"
)
print(
    f"Normalized species evidence rows : {len(species_pool)}"
)
print(
    f"Canonical residual rows          : {len(canonical_classification)}"
)


# =============================================================================
# 5. VALIDATE AUTHORITATIVE 9.4N-R STATE
# =============================================================================

required_audit_columns = [
    "Genus_Search_Target",
    "N_Independent_Species",
    "Species_Represented",
    "Evidence_Sources",
    "Min_Seed_Mass_mg",
    "Median_Seed_Mass_mg",
    "Max_Seed_Mass_mg",
    "Genus_Evidence_Class",
    "Next_Audit_Action",
]

missing_audit_columns = [
    col
    for col in required_audit_columns
    if col not in genus_audit.columns
]

if missing_audit_columns:
    raise KeyError(
        "Corrected genus audit is missing required columns:\n"
        f"{missing_audit_columns}"
    )


assert len(genus_audit) == 23, (
    "Section 9.4N-R should contain exactly 23 genus audit targets."
)

assert genus_audit[
    "Genus_Search_Target"
].is_unique, (
    "Genus_Search_Target must be unique in the corrected genus audit."
)


expected_class_counts = {
    "NO_GENUS_EVIDENCE": 15,
    "SINGLE_SPECIES_GENUS_EVIDENCE": 6,
    "MODERATE_GENUS_EVIDENCE": 2,
}

observed_class_counts = (
    genus_audit[
        "Genus_Evidence_Class"
    ]
    .value_counts()
    .to_dict()
)

for evidence_class, expected_n in expected_class_counts.items():

    observed_n = observed_class_counts.get(
        evidence_class,
        0
    )

    assert observed_n == expected_n, (
        f"Unexpected 9.4N-R count for {evidence_class}.\n"
        f"Expected : {expected_n}\n"
        f"Observed : {observed_n}"
    )


# =============================================================================
# 6. DEFINE CONSERVATIVE GENUS-TRANSFER RULES
# =============================================================================

transfer_rules = pd.DataFrame(
    [
        {
            "Rule_ID": "GTR_01",
            "Genus_Evidence_Class":
                "MODERATE_GENUS_EVIDENCE",
            "Minimum_Independent_Species": 2,
            "Automatic_Transfer_Permitted": True,
            "Estimator":
                "MEDIAN_OF_INDEPENDENT_CONGENERIC_SPECIES",
            "Assignment_Class":
                "GENUS_ESTIMATE_MODERATE_EVIDENCE",
            "Next_Action":
                "ASSIGN_GENUS_MEDIAN",
            "Rule_Rationale":
                (
                    "At least two independent congeneric species with "
                    "compatible Seed Mass evidence are available. "
                    "The genus median is permitted as a conservative "
                    "genus-level fallback estimate."
                ),
        },
        {
            "Rule_ID": "GTR_02",
            "Genus_Evidence_Class":
                "SINGLE_SPECIES_GENUS_EVIDENCE",
            "Minimum_Independent_Species": 1,
            "Automatic_Transfer_Permitted": False,
            "Estimator":
                "NONE_WITHOUT_EXPLICIT_REVIEW",
            "Assignment_Class":
                "SINGLE_CONGENER_REVIEW_REQUIRED",
            "Next_Action":
                "REVIEW_SINGLE_CONGENER_TRANSFER_RISK",
            "Rule_Rationale":
                (
                    "Only one independent congeneric species is available. "
                    "A direct transfer could reproduce species-specific "
                    "rather than genus-level biology and therefore requires "
                    "explicit review before any numerical assignment."
                ),
        },
        {
            "Rule_ID": "GTR_03",
            "Genus_Evidence_Class":
                "NO_GENUS_EVIDENCE",
            "Minimum_Independent_Species": 0,
            "Automatic_Transfer_Permitted": False,
            "Estimator":
                "NONE",
            "Assignment_Class":
                "NO_GENUS_ESTIMATE_AVAILABLE",
            "Next_Action":
                "PROCEED_TO_FAMILY_LEVEL_EVIDENCE_AUDIT",
            "Rule_Rationale":
                (
                    "No independent congeneric Seed Mass evidence is "
                    "available in the normalized evidence pool. "
                    "The residual target must therefore proceed to "
                    "family-level evidence assessment."
                ),
        },
    ]
)


print("\nGenus-transfer rule framework")
print("-" * 100)

display(
    transfer_rules[
        [
            "Rule_ID",
            "Genus_Evidence_Class",
            "Minimum_Independent_Species",
            "Automatic_Transfer_Permitted",
            "Estimator",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 7. APPLY TRANSFER RULES TO CORRECTED GENUS AUDIT
# =============================================================================

rule_lookup = transfer_rules[
    [
        "Rule_ID",
        "Genus_Evidence_Class",
        "Automatic_Transfer_Permitted",
        "Estimator",
        "Assignment_Class",
    ]
].copy()


genus_decisions = genus_audit.merge(
    rule_lookup,
    on="Genus_Evidence_Class",
    how="left",
    validate="many_to_one",
)


assert genus_decisions[
    "Rule_ID"
].notna().all(), (
    "At least one genus evidence class was not assigned a transfer rule."
)


# =============================================================================
# 8. IDENTIFY AUTOMATIC GENUS-LEVEL ASSIGNMENTS
# =============================================================================

genus_assignments = genus_decisions[
    genus_decisions[
        "Automatic_Transfer_Permitted"
    ].eq(True)
].copy()


assert (
    genus_assignments[
        "N_Independent_Species"
    ] >= 2
).all(), (
    "Automatic genus transfer encountered a genus represented by "
    "fewer than two independent species."
)


genus_assignments[
    "Assigned_Seed_Mass_mg"
] = genus_assignments[
    "Median_Seed_Mass_mg"
]


genus_assignments[
    "Seed_Mass_Assignment_Level"
] = "GENUS_LEVEL"


genus_assignments[
    "Seed_Mass_Provenance_Class"
] = "GENUS_ESTIMATE"


genus_assignments[
    "Seed_Mass_Quality_Flag"
] = "MODERATE_GENUS_EVIDENCE"


genus_assignments[
    "Assignment_Method"
] = "MEDIAN_OF_INDEPENDENT_CONGENERIC_SPECIES"


genus_assignments[
    "Eligible_For_Final_Seed_Mass_Table"
] = True


# =============================================================================
# 9. AUTHORITATIVE EXPECTED GENUS ASSIGNMENTS
# =============================================================================

expected_genus_assignments = {
    "Albizia": 83.5,
    "Sterculia": 831.5,
}


observed_assignment_genera = set(
    genus_assignments[
        "Genus_Search_Target"
    ]
)

assert observed_assignment_genera == set(
    expected_genus_assignments
), (
    "Unexpected automatic genus-assignment set.\n"
    f"Expected : {sorted(expected_genus_assignments)}\n"
    f"Observed : {sorted(observed_assignment_genera)}"
)


for genus_name, expected_value in expected_genus_assignments.items():

    observed_value = genus_assignments.loc[
        genus_assignments[
            "Genus_Search_Target"
        ].eq(genus_name),
        "Assigned_Seed_Mass_mg",
    ]

    assert len(observed_value) == 1

    assert np.isclose(
        observed_value.iloc[0],
        expected_value,
        rtol=0,
        atol=1e-9,
    ), (
        f"Unexpected genus estimate for {genus_name}.\n"
        f"Expected : {expected_value}\n"
        f"Observed : {observed_value.iloc[0]}"
    )


# =============================================================================
# 10. SINGLE-CONGENER REVIEW REGISTER
# =============================================================================

single_congener_review = genus_decisions[
    genus_decisions[
        "Genus_Evidence_Class"
    ].eq(
        "SINGLE_SPECIES_GENUS_EVIDENCE"
    )
].copy()


single_congener_review[
    "Candidate_Seed_Mass_mg"
] = single_congener_review[
    "Median_Seed_Mass_mg"
]


single_congener_review[
    "Numerical_Seed_Mass_Assigned"
] = False


single_congener_review[
    "Transfer_Decision"
] = "PENDING_EXPLICIT_REVIEW"


single_congener_review[
    "Review_Question"
] = (
    "Is transfer from the single represented congener biologically "
    "and taxonomically defensible for the residual Omo target?"
)


expected_single_congener_genera = {
    "Anthocleista",
    "Cola",
    "Diospyros",
    "Macaranga",
    "Pterocarpus",
    "Trichilia",
}


assert set(
    single_congener_review[
        "Genus_Search_Target"
    ]
) == expected_single_congener_genera, (
    "Unexpected single-congener review set."
)


# =============================================================================
# 11. NO-GENUS-EVIDENCE / FAMILY-STAGE REGISTER
# =============================================================================

family_stage_targets = genus_decisions[
    genus_decisions[
        "Genus_Evidence_Class"
    ].eq(
        "NO_GENUS_EVIDENCE"
    )
].copy()


family_stage_targets[
    "Numerical_Seed_Mass_Assigned"
] = False


family_stage_targets[
    "Fallback_Level"
] = "FAMILY"


family_stage_targets[
    "Family_Audit_Status"
] = "READY_FOR_FAMILY_EVIDENCE_AUDIT"


assert len(family_stage_targets) == 15, (
    "Expected 15 genera to proceed to family-level audit."
)


# =============================================================================
# 12. MAP APPROVED GENUS ESTIMATES BACK TO CANONICAL RESIDUAL TAXA
# =============================================================================

required_canonical_columns = [
    "Canonical_Species",
    "Final_Search_Name",
    "Residual_Level",
    "Residual_Classification",
]

missing_canonical_columns = [
    col
    for col in required_canonical_columns
    if col not in canonical_classification.columns
]

if missing_canonical_columns:
    raise KeyError(
        "Canonical residual classification is missing required columns:\n"
        f"{missing_canonical_columns}"
    )


canonical_for_assignment = canonical_classification.copy()


def infer_genus(row):
    """
    Infer the genus used for genus-level fallback assignment.

    Priority:
    1. Final_Search_Name
    2. Canonical_Species
    """

    final_name = row.get(
        "Final_Search_Name",
        pd.NA
    )

    canonical_name = row.get(
        "Canonical_Species",
        pd.NA
    )

    if pd.notna(final_name):
        final_name = str(final_name).strip()

        if final_name:
            return final_name.split()[0]

    if pd.notna(canonical_name):
        canonical_name = str(canonical_name).strip()

        if canonical_name:
            return canonical_name.split()[0]

    return pd.NA


canonical_for_assignment[
    "Genus_Search_Target"
] = canonical_for_assignment.apply(
    infer_genus,
    axis=1,
)


assignment_lookup = genus_assignments[
    [
        "Genus_Search_Target",
        "Assigned_Seed_Mass_mg",
        "N_Independent_Species",
        "Species_Represented",
        "Evidence_Sources",
        "Assignment_Method",
        "Seed_Mass_Assignment_Level",
        "Seed_Mass_Provenance_Class",
        "Seed_Mass_Quality_Flag",
    ]
].copy()


canonical_genus_assignments = (
    canonical_for_assignment
    .merge(
        assignment_lookup,
        on="Genus_Search_Target",
        how="inner",
        validate="many_to_one",
    )
)


canonical_genus_assignments[
    "Numerical_Seed_Mass_Assigned"
] = True


canonical_genus_assignments[
    "Genus_Transfer_Status"
] = "APPROVED_GENUS_MEDIAN_TRANSFER"


# =============================================================================
# 13. ENSURE ONLY CURRENTLY UNASSIGNED RESIDUAL TAXA RECEIVE VALUES
# =============================================================================

if "Numerical_Seed_Mass_Assigned" in canonical_classification.columns:

    prior_assignment_status = (
        canonical_classification[
            [
                "Canonical_Species",
                "Numerical_Seed_Mass_Assigned",
            ]
        ]
        .copy()
    )

    prior_assignment_status[
        "Numerical_Seed_Mass_Assigned"
    ] = (
        prior_assignment_status[
            "Numerical_Seed_Mass_Assigned"
        ]
        .astype("boolean")
    )

    already_assigned = set(
        prior_assignment_status.loc[
            prior_assignment_status[
                "Numerical_Seed_Mass_Assigned"
            ].fillna(False),
            "Canonical_Species",
        ]
    )

    overlap = set(
        canonical_genus_assignments[
            "Canonical_Species"
        ]
    ).intersection(
        already_assigned
    )

    assert len(overlap) == 0, (
        "Genus-level assignment attempted for taxa that already "
        "have Seed Mass values:\n"
        f"{sorted(overlap)}"
    )


# =============================================================================
# 14. DISPLAY GENUS ASSIGNMENTS
# =============================================================================

print("\nApproved genus-level Seed Mass assignments")
print("-" * 100)

display(
    genus_assignments[
        [
            "Genus_Search_Target",
            "N_Independent_Species",
            "Species_Represented",
            "Evidence_Sources",
            "Min_Seed_Mass_mg",
            "Assigned_Seed_Mass_mg",
            "Max_Seed_Mass_mg",
            "Assignment_Method",
            "Seed_Mass_Quality_Flag",
        ]
    ]
)


# =============================================================================
# 15. DISPLAY CANONICAL TAXA RECEIVING GENUS ESTIMATES
# =============================================================================

print("\nCanonical residual taxa receiving approved genus estimates")
print("-" * 100)

display(
    canonical_genus_assignments[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Genus_Search_Target",
            "Assigned_Seed_Mass_mg",
            "N_Independent_Species",
            "Species_Represented",
            "Evidence_Sources",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
        ]
    ]
)


# =============================================================================
# 16. DISPLAY SINGLE-CONGENER CASES
# =============================================================================

print("\nSingle-congener cases retained for explicit review")
print("-" * 100)

display(
    single_congener_review[
        [
            "Genus_Search_Target",
            "Residual_Biological_Targets",
            "Explicit_Genus_Canonical_Taxa",
            "Species_Represented",
            "Candidate_Seed_Mass_mg",
            "Evidence_Sources",
            "Transfer_Decision",
        ]
    ]
)


# =============================================================================
# 17. DISPLAY FAMILY-STAGE GENERA
# =============================================================================

print("\nGenera proceeding directly to family-level evidence audit")
print("-" * 100)

display(
    family_stage_targets[
        [
            "Genus_Search_Target",
            "Residual_Biological_Targets",
            "Explicit_Genus_Canonical_Taxa",
            "Genus_Evidence_Class",
            "Family_Audit_Status",
        ]
    ]
)


# =============================================================================
# 18. SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Genus audit targets entering 9.4O",
            "Moderate-evidence genera",
            "Single-species-evidence genera",
            "No-genus-evidence genera",
            "Genus estimates approved",
            "Canonical residual taxa assigned genus estimates",
            "Single-congener cases retained for review",
            "Genera proceeding to family audit",
            "Seed Mass values assigned in Section 9.4O",
        ],
        "Value": [
            len(genus_decisions),
            int(
                (
                    genus_decisions[
                        "Genus_Evidence_Class"
                    ]
                    == "MODERATE_GENUS_EVIDENCE"
                ).sum()
            ),
            int(
                (
                    genus_decisions[
                        "Genus_Evidence_Class"
                    ]
                    == "SINGLE_SPECIES_GENUS_EVIDENCE"
                ).sum()
            ),
            int(
                (
                    genus_decisions[
                        "Genus_Evidence_Class"
                    ]
                    == "NO_GENUS_EVIDENCE"
                ).sum()
            ),
            len(genus_assignments),
            len(canonical_genus_assignments),
            len(single_congener_review),
            len(family_stage_targets),
            len(canonical_genus_assignments),
        ],
    }
)


print("\nSection 9.4O summary")
print("-" * 100)

display(summary)


# =============================================================================
# 19. FINAL VALIDATION
# =============================================================================

assert len(genus_assignments) == 2, (
    "Expected exactly two approved genus estimates."
)

assert len(single_congener_review) == 6, (
    "Expected exactly six single-congener review cases."
)

assert len(family_stage_targets) == 15, (
    "Expected exactly fifteen genera for family-level audit."
)

assert genus_assignments[
    "Assigned_Seed_Mass_mg"
].notna().all()

assert (
    genus_assignments[
        "Assigned_Seed_Mass_mg"
    ] > 0
).all()

assert canonical_genus_assignments[
    "Assigned_Seed_Mass_mg"
].notna().all()


# =============================================================================
# 20. EXPORT
# =============================================================================

transfer_rules.to_csv(
    OUT_TRANSFER_RULES,
    index=False,
)

genus_assignments.to_csv(
    OUT_GENUS_ASSIGNMENTS,
    index=False,
)

single_congener_review.to_csv(
    OUT_SINGLE_CONGENER_REVIEW,
    index=False,
)

family_stage_targets.to_csv(
    OUT_FAMILY_STAGE_TARGETS,
    index=False,
)

canonical_genus_assignments.to_csv(
    OUT_CANONICAL_GENUS_ASSIGNMENTS,
    index=False,
)

summary.to_csv(
    OUT_SUMMARY,
    index=False,
)


# =============================================================================
# 21. EXPORT REPORT
# =============================================================================

print("\nSection 9.4O exports")
print("-" * 100)

print(
    f"Transfer rules          : {OUT_TRANSFER_RULES}"
)

print(
    f"Genus assignments       : {OUT_GENUS_ASSIGNMENTS}"
)

print(
    f"Single-congener review  : {OUT_SINGLE_CONGENER_REVIEW}"
)

print(
    f"Family-stage targets    : {OUT_FAMILY_STAGE_TARGETS}"
)

print(
    f"Canonical assignments   : {OUT_CANONICAL_GENUS_ASSIGNMENTS}"
)

print(
    f"Summary                 : {OUT_SUMMARY}"
)


# =============================================================================
# 22. COMPLETION MESSAGE
# =============================================================================

print()
print("=" * 100)
print(
    "Genus-level Seed Mass transfer-rule development completed successfully."
)
print(
    f"Genus audit targets                     : {len(genus_decisions)}"
)
print(
    f"Approved genus estimates                : {len(genus_assignments)}"
)
print(
    f"Canonical taxa receiving genus estimates: {len(canonical_genus_assignments)}"
)
print(
    f"Single-congener cases awaiting review   : {len(single_congener_review)}"
)
print(
    f"Genera proceeding to family audit       : {len(family_stage_targets)}"
)
print(
    "Albizia and Sterculia are the only genera automatically assigned "
    "Seed Mass estimates under the conservative >=2-species rule."
)
print(
    "No single-congener Seed Mass value has been transferred automatically."
)
print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4P
# POST-GENUS-ASSIGNMENT RESIDUAL CONSOLIDATION
# AND FAMILY-LEVEL TARGET PREPARATION
# =============================================================================
#
# Purpose
# -------
# 1. Incorporate the two approved genus-level Seed Mass assignments from 9.4O.
# 2. Confirm that they do not overwrite positive species-level evidence.
# 3. Recalculate the authoritative residual canonical set.
# 4. Route the 37 residual canonical taxa into:
#
#      A. TAXONOMIC_NAME_RESCUE_PENDING
#      B. SINGLE_CONGENER_EXPLICIT_REVIEW
#      C. DIRECT_FAMILY_LEVEL_AUDIT
#      D. SPECIAL_BIOLOGICAL_CASE_REVIEW
#
# 5. Prepare the direct-family audit register.
#
# IMPORTANT
# ---------
# This section does NOT:
#   - assign family-level Seed Mass values,
#   - resolve single-congener cases,
#   - resolve taxonomic-name rescue cases,
#   - assign any new numerical values beyond carrying forward the two
#     approved genus assignments from Section 9.4O.
#
# Expected transition
# -------------------
# Residual canonical taxa before genus assignment : 39
# Approved genus assignments                      : 2
# Residual canonical taxa after genus assignment  : 37
#
# Expected residual routing
# -------------------------
# Taxonomic-name rescue pending : 5 canonical taxa
# Single-congener review        : derived from authoritative mapping
# Direct family audit           : derived from authoritative mapping
# Special biological case       : 1 canonical taxon
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

CANONICAL_CLASSIFICATION_FILE = (
    OUT_DIR /
    "omo_seed_mass_post_external_rescue_canonical_classification.csv"
)

CANONICAL_GENUS_ASSIGNMENTS_FILE = (
    OUT_DIR /
    "omo_seed_mass_canonical_genus_assignments.csv"
)

SINGLE_CONGENER_REVIEW_FILE = (
    OUT_DIR /
    "omo_seed_mass_single_congener_transfer_review.csv"
)

FAMILY_STAGE_GENUS_FILE = (
    OUT_DIR /
    "omo_seed_mass_family_stage_targets.csv"
)

EXTERNAL_CANONICAL_VALUES_FILE = (
    OUT_DIR /
    "omo_seed_mass_non_try_external_canonical_values.csv"
)

BIEN_VALUES_FILE = (
    OUT_DIR /
    "omo_seed_mass_bien_all_species_values_current.csv"
)

TRY_VALUES_FILE = (
    OUT_DIR /
    "omo_seed_mass_try_51911_final_omo_canonical_values.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_POST_GENUS_STATE = (
    OUT_DIR /
    "omo_seed_mass_post_genus_assignment_state.csv"
)

OUT_POST_GENUS_RESIDUAL = (
    OUT_DIR /
    "omo_seed_mass_post_genus_assignment_residual_canonical_taxa.csv"
)

OUT_TAXONOMIC_RESCUE_CANONICAL = (
    OUT_DIR /
    "omo_seed_mass_taxonomic_name_rescue_pending_targets.csv"
)

OUT_SINGLE_CONGENER_CANONICAL = (
    OUT_DIR /
    "omo_seed_mass_single_congener_review_canonical_targets.csv"
)

OUT_DIRECT_FAMILY_CANONICAL = (
    OUT_DIR /
    "omo_seed_mass_direct_family_audit_canonical_targets.csv"
)

OUT_SPECIAL_CASE_CANONICAL = (
    OUT_DIR /
    "omo_seed_mass_special_biological_case_targets.csv"
)

OUT_FAMILY_AUDIT_TARGETS = (
    OUT_DIR /
    "omo_seed_mass_family_audit_targets_prepared.csv"
)

OUT_FAMILY_AUDIT_GENUS_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_family_audit_genus_summary.csv"
)

OUT_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_post_genus_assignment_residual_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    CANONICAL_CLASSIFICATION_FILE,
    CANONICAL_GENUS_ASSIGNMENTS_FILE,
    SINGLE_CONGENER_REVIEW_FILE,
    FAMILY_STAGE_GENUS_FILE,
    EXTERNAL_CANONICAL_VALUES_FILE,
    BIEN_VALUES_FILE,
    TRY_VALUES_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

canonical_classification = pd.read_csv(
    CANONICAL_CLASSIFICATION_FILE,
    low_memory=False
)

genus_assignments = pd.read_csv(
    CANONICAL_GENUS_ASSIGNMENTS_FILE,
    low_memory=False
)

single_congener_review = pd.read_csv(
    SINGLE_CONGENER_REVIEW_FILE,
    low_memory=False
)

family_stage_genera = pd.read_csv(
    FAMILY_STAGE_GENUS_FILE,
    low_memory=False
)

external_values = pd.read_csv(
    EXTERNAL_CANONICAL_VALUES_FILE,
    low_memory=False
)

bien_values = pd.read_csv(
    BIEN_VALUES_FILE,
    low_memory=False
)

try_values = pd.read_csv(
    TRY_VALUES_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4P — POST-GENUS-ASSIGNMENT RESIDUAL CONSOLIDATION "
    "AND FAMILY-LEVEL TARGET PREPARATION"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Residual canonical classification : "
    f"{len(canonical_classification)}"
)

print(
    f"Approved canonical genus values   : "
    f"{len(genus_assignments)}"
)

print(
    f"Single-congener genera            : "
    f"{len(single_congener_review)}"
)

print(
    f"Direct family-stage genera        : "
    f"{len(family_stage_genera)}"
)


# =============================================================================
# 4. VALIDATE STARTING STATE
# =============================================================================

assert len(
    canonical_classification
) == 39, (
    f"Expected 39 canonical residual taxa entering 9.4P; "
    f"found {len(canonical_classification)}."
)


assert len(
    genus_assignments
) == 2, (
    f"Expected 2 approved canonical genus assignments; "
    f"found {len(genus_assignments)}."
)


assert len(
    single_congener_review
) == 6, (
    f"Expected 6 single-congener review genera; "
    f"found {len(single_congener_review)}."
)


assert len(
    family_stage_genera
) == 15, (
    f"Expected 15 direct family-stage genera; "
    f"found {len(family_stage_genera)}."
)


# =============================================================================
# 5. STANDARDIZE TEXT FIELDS
# =============================================================================

def clean_text_series(series):

    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "": pd.NA,
                "nan": pd.NA,
                "None": pd.NA,
                "<NA>": pd.NA,
            }
        )
    )


for df in [
    canonical_classification,
    genus_assignments,
    external_values,
    bien_values,
    try_values,
]:

    if "Canonical_Species" in df.columns:

        df[
            "Canonical_Species"
        ] = clean_text_series(
            df[
                "Canonical_Species"
            ]
        )


# =============================================================================
# 6. VALIDATE GENUS ASSIGNMENTS ARE MEMBERS OF RESIDUAL SET
# =============================================================================

residual_set_before = set(
    canonical_classification[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


genus_assignment_set = set(
    genus_assignments[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


assert genus_assignment_set.issubset(
    residual_set_before
), (
    "One or more approved genus assignments are not members "
    "of the 39-taxon residual set."
)


# =============================================================================
# 7. VERIFY NO OVERWRITE OF VALID POSITIVE SPECIES-LEVEL VALUES
# =============================================================================
#
# IMPORTANT
# ---------
# A canonical taxon is considered species-level resolved ONLY if the
# associated table contains a positive numeric Seed Mass value.
#
# Merely appearing in a BIEN / TRY / external table does not imply that
# the taxon already has an accepted species-level Seed Mass value.
# =============================================================================

def find_first_existing_column(
    df,
    candidates,
    required=True
):

    for col in candidates:

        if col in df.columns:
            return col

    if required:

        raise KeyError(
            "Could not locate required column.\n"
            f"Candidates tested: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


def positive_species_resolved_set(
    df,
    source_name
):

    canon_col = find_first_existing_column(
        df,
        [
            "Canonical_Species",
            "Canonical Species",
        ]
    )


    value_col = find_first_existing_column(
        df,
        [
            "Seed_Mass_mg",
            "Standardized_Trait_Value",
            "Median_Seed_Mass_mg",
            "Median_Seed_Mass",
            "SeedMass_mg",
            "Seed_Mass",
            "Trait_Value",
            "Value_mg",
        ]
    )


    numeric_value = pd.to_numeric(
        df[
            value_col
        ],
        errors="coerce"
    )


    valid_mask = (
        df[
            canon_col
        ].notna()
        &
        numeric_value.notna()
        &
        numeric_value.gt(0)
    )


    resolved = set(
        df.loc[
            valid_mask,
            canon_col
        ]
        .astype(str)
        .str.strip()
    )


    print(
        f"{source_name:<30}: "
        f"{len(resolved)} positive-value canonical taxa "
        f"(value field = {value_col})"
    )


    return resolved


print("\nValidated positive-value species-level coverage")
print("-" * 100)


bien_resolved_set = positive_species_resolved_set(
    bien_values,
    "BIEN"
)


try_resolved_set = positive_species_resolved_set(
    try_values,
    "TRY"
)


external_resolved_set = positive_species_resolved_set(
    external_values,
    "External species rescue"
)


species_resolved_set = (
    bien_resolved_set
    |
    try_resolved_set
    |
    external_resolved_set
)


print(
    f"{'Combined species-level set':<30}: "
    f"{len(species_resolved_set)} canonical taxa"
)


overlap_with_species_values = (
    genus_assignment_set
    &
    species_resolved_set
)


if overlap_with_species_values:

    print(
        "\nUnexpected overlap between genus assignments "
        "and positive species-level values"
    )
    print("-" * 100)

    for taxon in sorted(
        overlap_with_species_values
    ):

        print(
            f" - {taxon}"
        )


assert len(
    overlap_with_species_values
) == 0, (
    "Approved genus assignments overlap with taxa carrying "
    "positive species-level Seed Mass evidence:\n"
    f"{sorted(overlap_with_species_values)}"
)


assert (
    "Albizia auxilary"
    not in species_resolved_set
)


assert (
    "Sterculia tragacantha"
    not in species_resolved_set
)


print(
    "\nApproved genus assignments confirmed not to overwrite "
    "positive species-level evidence."
)


# =============================================================================
# 8. BUILD POST-GENUS ASSIGNMENT STATE
# =============================================================================

post_genus_state = (
    canonical_classification
    .copy()
)


post_genus_state[
    "Seed_Mass_Assigned_After_9_4O"
] = False


post_genus_state[
    "Assigned_Seed_Mass_mg"
] = np.nan


post_genus_state[
    "Seed_Mass_Assignment_Level"
] = pd.NA


post_genus_state[
    "Seed_Mass_Provenance_Class"
] = pd.NA


post_genus_state[
    "Seed_Mass_Quality_Flag"
] = pd.NA


assignment_lookup = genus_assignments[
    [
        "Canonical_Species",
        "Assigned_Seed_Mass_mg",
        "Seed_Mass_Assignment_Level",
        "Seed_Mass_Provenance_Class",
        "Seed_Mass_Quality_Flag",
    ]
].copy()


post_genus_state = (
    post_genus_state
    .merge(
        assignment_lookup,
        on="Canonical_Species",
        how="left",
        suffixes=(
            "",
            "_Genus"
        ),
        validate="one_to_one"
    )
)


assigned_mask = (
    post_genus_state[
        "Assigned_Seed_Mass_mg_Genus"
    ]
    .notna()
)


post_genus_state.loc[
    assigned_mask,
    "Seed_Mass_Assigned_After_9_4O"
] = True


post_genus_state.loc[
    assigned_mask,
    "Assigned_Seed_Mass_mg"
] = (
    post_genus_state.loc[
        assigned_mask,
        "Assigned_Seed_Mass_mg_Genus"
    ]
)


for base_col in [
    "Seed_Mass_Assignment_Level",
    "Seed_Mass_Provenance_Class",
    "Seed_Mass_Quality_Flag",
]:

    genus_col = (
        f"{base_col}_Genus"
    )

    post_genus_state.loc[
        assigned_mask,
        base_col
    ] = (
        post_genus_state.loc[
            assigned_mask,
            genus_col
        ]
    )


drop_cols = [
    col
    for col in post_genus_state.columns
    if col.endswith(
        "_Genus"
    )
]


post_genus_state = (
    post_genus_state
    .drop(
        columns=drop_cols
    )
)


# =============================================================================
# 9. VALIDATE THE TWO GENUS-ASSIGNED CANONICAL TAXA
# =============================================================================

assigned_canonical = (
    post_genus_state[
        post_genus_state[
            "Seed_Mass_Assigned_After_9_4O"
        ]
    ]
    .copy()
)


print("\nCanonical taxa resolved by genus assignment")
print("-" * 100)


display(
    assigned_canonical[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Assigned_Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
        ]
    ]
)


assert len(
    assigned_canonical
) == 2


assert set(
    assigned_canonical[
        "Canonical_Species"
    ]
) == {
    "Albizia auxilary",
    "Sterculia tragacantha",
}


# =============================================================================
# 10. RECALCULATE TRUE RESIDUAL CANONICAL SET
# =============================================================================

post_genus_residual = (
    post_genus_state[
        ~post_genus_state[
            "Seed_Mass_Assigned_After_9_4O"
        ]
    ]
    .copy()
)


post_genus_residual = (
    post_genus_residual
    .sort_values(
        [
            "Residual_Level",
            "Final_Search_Name",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


print("\nPost-genus-assignment residual state")
print("-" * 100)

print(
    f"Residual canonical taxa before genus assignment : "
    f"{len(canonical_classification)}"
)

print(
    f"Canonical taxa resolved by genus assignment     : "
    f"{len(assigned_canonical)}"
)

print(
    f"Residual canonical taxa after genus assignment  : "
    f"{len(post_genus_residual)}"
)


assert len(
    post_genus_residual
) == 37, (
    f"Expected 37 residual canonical taxa after genus assignment; "
    f"found {len(post_genus_residual)}."
)


# =============================================================================
# 11. IDENTIFY SINGLE-CONGENER GENERA
# =============================================================================

single_congener_genera = set(
    single_congener_review[
        "Genus_Search_Target"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert single_congener_genera == {
    "Anthocleista",
    "Cola",
    "Diospyros",
    "Macaranga",
    "Pterocarpus",
    "Trichilia",
}


# =============================================================================
# 12. IDENTIFY DIRECT FAMILY-STAGE GENERA
# =============================================================================

direct_family_genera = set(
    family_stage_genera[
        "Genus_Search_Target"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert len(
    direct_family_genera
) == 15


# =============================================================================
# 13. INFER GENUS FOR ALL RESIDUAL CANONICAL TAXA
# =============================================================================

def infer_genus(row):

    final_name = row.get(
        "Final_Search_Name",
        pd.NA
    )

    canonical_name = row.get(
        "Canonical_Species",
        pd.NA
    )


    if pd.notna(
        final_name
    ):

        final_name = str(
            final_name
        ).strip()

        if final_name:

            return (
                final_name
                .split()[0]
            )


    if pd.notna(
        canonical_name
    ):

        canonical_name = str(
            canonical_name
        ).strip()

        if canonical_name:

            return (
                canonical_name
                .split()[0]
            )


    return pd.NA


post_genus_residual[
    "Residual_Genus"
] = (
    post_genus_residual
    .apply(
        infer_genus,
        axis=1
    )
)


# =============================================================================
# 14. DEFINE TAXONOMIC-RESCUE BIOLOGICAL TARGETS
# =============================================================================

taxonomic_rescue_targets = {
    "Aningeria pierrei",
    "Corynanthe pachyceras",
    "Homalium longistylum",
    "Mitragyna ciliata",
}


# =============================================================================
# 15. ROUTE RESIDUAL TAXA
# =============================================================================
#
# Four distinct post-genus pathways:
#
#   1. TAXONOMIC_NAME_RESCUE_PENDING
#   2. SINGLE_CONGENER_EXPLICIT_REVIEW
#   3. DIRECT_FAMILY_LEVEL_AUDIT
#   4. SPECIAL_BIOLOGICAL_CASE_REVIEW
#
# =============================================================================

post_genus_residual[
    "Post_9_4O_Route"
] = (
    "UNCLASSIFIED"
)


# -----------------------------------------------------------------------------
# A. Taxonomic-name rescue first
# -----------------------------------------------------------------------------

post_genus_residual.loc[
    post_genus_residual[
        "Residual_Classification"
    ].eq(
        "TAXONOMIC_NAME_RESCUE_FIRST"
    ),
    "Post_9_4O_Route"
] = (
    "TAXONOMIC_NAME_RESCUE_PENDING"
)


# -----------------------------------------------------------------------------
# B. Single-congener explicit review
# -----------------------------------------------------------------------------

post_genus_residual.loc[
    post_genus_residual[
        "Residual_Genus"
    ].isin(
        single_congener_genera
    ),
    "Post_9_4O_Route"
] = (
    "SINGLE_CONGENER_EXPLICIT_REVIEW"
)


# -----------------------------------------------------------------------------
# C. Direct family-level audit
# -----------------------------------------------------------------------------

post_genus_residual.loc[
    post_genus_residual[
        "Residual_Genus"
    ].isin(
        direct_family_genera
    ),
    "Post_9_4O_Route"
] = (
    "DIRECT_FAMILY_LEVEL_AUDIT"
)


# -----------------------------------------------------------------------------
# D. Special biological case
# -----------------------------------------------------------------------------

post_genus_residual.loc[
    post_genus_residual[
        "Residual_Classification"
    ].eq(
        "SPECIAL_BIOLOGICAL_CASE"
    ),
    "Post_9_4O_Route"
] = (
    "SPECIAL_BIOLOGICAL_CASE_REVIEW"
)


# -----------------------------------------------------------------------------
# E. Reassert taxonomic rescue priority
#
# This is intentional. If any corrected-name taxon happens to share a genus
# with another routing group, taxonomic rescue still has priority.
# -----------------------------------------------------------------------------

post_genus_residual.loc[
    post_genus_residual[
        "Residual_Classification"
    ].eq(
        "TAXONOMIC_NAME_RESCUE_FIRST"
    ),
    "Post_9_4O_Route"
] = (
    "TAXONOMIC_NAME_RESCUE_PENDING"
)


# -----------------------------------------------------------------------------
# F. Reassert special biological priority
# -----------------------------------------------------------------------------

post_genus_residual.loc[
    post_genus_residual[
        "Residual_Classification"
    ].eq(
        "SPECIAL_BIOLOGICAL_CASE"
    ),
    "Post_9_4O_Route"
] = (
    "SPECIAL_BIOLOGICAL_CASE_REVIEW"
)


# =============================================================================
# 16. CHECK FOR UNCLASSIFIED RESIDUAL TAXA
# =============================================================================

unclassified = (
    post_genus_residual[
        post_genus_residual[
            "Post_9_4O_Route"
        ].eq(
            "UNCLASSIFIED"
        )
    ]
    .copy()
)


if len(
    unclassified
) > 0:

    print("\nUnclassified post-genus residual taxa")
    print("-" * 100)

    display(
        unclassified[
            [
                "Canonical_Species",
                "Final_Search_Name",
                "Residual_Genus",
                "Residual_Classification",
                "Next_Action",
            ]
        ]
    )


assert len(
    unclassified
) == 0, (
    "Some residual canonical taxa were not routed after 9.4O."
)


# =============================================================================
# 17. CREATE ROUTE-SPECIFIC CANONICAL TABLES
# =============================================================================

taxonomic_rescue_canonical = (
    post_genus_residual[
        post_genus_residual[
            "Post_9_4O_Route"
        ].eq(
            "TAXONOMIC_NAME_RESCUE_PENDING"
        )
    ]
    .copy()
)


single_congener_canonical = (
    post_genus_residual[
        post_genus_residual[
            "Post_9_4O_Route"
        ].eq(
            "SINGLE_CONGENER_EXPLICIT_REVIEW"
        )
    ]
    .copy()
)


direct_family_canonical = (
    post_genus_residual[
        post_genus_residual[
            "Post_9_4O_Route"
        ].eq(
            "DIRECT_FAMILY_LEVEL_AUDIT"
        )
    ]
    .copy()
)


special_case_canonical = (
    post_genus_residual[
        post_genus_residual[
            "Post_9_4O_Route"
        ].eq(
            "SPECIAL_BIOLOGICAL_CASE_REVIEW"
        )
    ]
    .copy()
)


# =============================================================================
# 18. ROUTE SUMMARY
# =============================================================================

route_summary = (
    post_genus_residual[
        "Post_9_4O_Route"
    ]
    .value_counts()
    .rename_axis(
        "Post_9_4O_Route"
    )
    .reset_index(
        name="N_Canonical_Taxa"
    )
)


print("\nPost-9.4O residual routing")
print("-" * 100)

display(
    route_summary
)


# =============================================================================
# 19. DISPLAY TAXONOMIC-RESCUE TARGETS
# =============================================================================

print("\nCanonical taxa retained for taxonomic-name rescue")
print("-" * 100)

display(
    taxonomic_rescue_canonical[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Residual_Level",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 20. DISPLAY SINGLE-CONGENER TARGETS
# =============================================================================

print("\nCanonical taxa retained for single-congener review")
print("-" * 100)

display(
    single_congener_canonical[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Residual_Level",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 21. DISPLAY DIRECT FAMILY-AUDIT TARGETS
# =============================================================================

print("\nCanonical taxa proceeding directly to family-level audit")
print("-" * 100)

display(
    direct_family_canonical[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Residual_Level",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 22. DISPLAY SPECIAL BIOLOGICAL CASE
# =============================================================================

print("\nSpecial biological case retained outside ordinary fallback")
print("-" * 100)

display(
    special_case_canonical[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 23. PREPARE DIRECT FAMILY-AUDIT TARGET REGISTER
# =============================================================================
#
# IMPORTANT:
# Only direct family-stage cases are included here.
#
# Excluded for now:
#   - taxonomic-name rescue cases,
#   - single-congener review cases,
#   - Musa special biological case.
# =============================================================================

family_audit_targets = (
    direct_family_canonical[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Residual_Level",
            "Residual_Classification",
            "Next_Action",
        ]
    ]
    .copy()
)


family_audit_targets[
    "Family_Audit_Status"
] = (
    "READY_FOR_FAMILY_TAXONOMIC_MAPPING"
)


family_audit_targets[
    "Numerical_Family_Value_Assigned"
] = False


# =============================================================================
# 24. FAMILY-AUDIT GENUS STRUCTURE
# =============================================================================

family_genus_summary = (
    family_audit_targets
    .groupby(
        "Residual_Genus",
        as_index=False
    )
    .agg(
        N_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Biological_Targets=(
            "Final_Search_Name",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
)


print("\nPrepared direct family-level audit genera")
print("-" * 100)

display(
    family_genus_summary
)


# =============================================================================
# 25. VALIDATE TAXONOMIC-RESCUE GROUP
# =============================================================================

assert len(
    taxonomic_rescue_canonical
) == 5, (
    f"Expected 5 canonical taxa awaiting taxonomic-name rescue; "
    f"found {len(taxonomic_rescue_canonical)}."
)


assert set(
    taxonomic_rescue_canonical[
        "Final_Search_Name"
    ]
) == {
    "Aningeria pierrei",
    "Corynanthe pachyceras",
    "Homalium longistylum",
    "Mitragyna ciliata",
}


assert set(
    taxonomic_rescue_canonical[
        "Final_Search_Name"
    ]
).issubset(
    taxonomic_rescue_targets
)


# =============================================================================
# 26. VALIDATE SINGLE-CONGENER GROUP
# =============================================================================

assert set(
    single_congener_canonical[
        "Residual_Genus"
    ]
    .dropna()
) == single_congener_genera


# =============================================================================
# 27. VALIDATE DIRECT FAMILY GROUP
# =============================================================================

assert set(
    direct_family_canonical[
        "Residual_Genus"
    ]
    .dropna()
) == direct_family_genera


assert family_audit_targets[
    "Numerical_Family_Value_Assigned"
].eq(
    False
).all()


# =============================================================================
# 28. VALIDATE SPECIAL BIOLOGICAL CASE
# =============================================================================

assert len(
    special_case_canonical
) == 1


assert (
    special_case_canonical[
        "Final_Search_Name"
    ]
    .eq(
        "Musa × paradisiaca"
    )
    .all()
)


# =============================================================================
# 29. FULL RESIDUAL RECONCILIATION
# =============================================================================

reconciled_residual_count = (
    len(
        taxonomic_rescue_canonical
    )
    +
    len(
        single_congener_canonical
    )
    +
    len(
        direct_family_canonical
    )
    +
    len(
        special_case_canonical
    )
)


assert reconciled_residual_count == 37, (
    "Post-genus routing does not reconcile to "
    "the 37 residual canonical taxa.\n"
    f"Taxonomic rescue : {len(taxonomic_rescue_canonical)}\n"
    f"Single congener  : {len(single_congener_canonical)}\n"
    f"Direct family    : {len(direct_family_canonical)}\n"
    f"Special case     : {len(special_case_canonical)}\n"
    f"Total            : {reconciled_residual_count}"
)


# =============================================================================
# 30. CHECK ROUTE SETS ARE MUTUALLY EXCLUSIVE
# =============================================================================

route_sets = {
    "TAXONOMIC_RESCUE":
        set(
            taxonomic_rescue_canonical[
                "Canonical_Species"
            ]
        ),

    "SINGLE_CONGENER":
        set(
            single_congener_canonical[
                "Canonical_Species"
            ]
        ),

    "DIRECT_FAMILY":
        set(
            direct_family_canonical[
                "Canonical_Species"
            ]
        ),

    "SPECIAL_CASE":
        set(
            special_case_canonical[
                "Canonical_Species"
            ]
        ),
}


route_names = list(
    route_sets.keys()
)


for i in range(
    len(
        route_names
    )
):

    for j in range(
        i + 1,
        len(
            route_names
        )
    ):

        name_a = route_names[
            i
        ]

        name_b = route_names[
            j
        ]

        overlap = (
            route_sets[
                name_a
            ]
            &
            route_sets[
                name_b
            ]
        )

        assert len(
            overlap
        ) == 0, (
            f"Residual routing overlap detected between "
            f"{name_a} and {name_b}:\n"
            f"{sorted(overlap)}"
        )


# =============================================================================
# 31. OVERALL SEED MASS COVERAGE UPDATE
# =============================================================================
#
# Before 9.4O:
#     88 / 127 resolved
#
# Approved genus assignments:
#      2 additional canonical taxa
#
# After 9.4O:
#     90 / 127 resolved
#
# Remaining:
#     37 / 127
# =============================================================================

N_TOTAL_CANONICAL = 127

N_RESOLVED_BEFORE_GENUS = 88

N_GENUS_ASSIGNED = len(
    assigned_canonical
)

N_RESOLVED_AFTER_GENUS = (
    N_RESOLVED_BEFORE_GENUS
    +
    N_GENUS_ASSIGNED
)


assert N_GENUS_ASSIGNED == 2

assert N_RESOLVED_AFTER_GENUS == 90

assert (
    N_TOTAL_CANONICAL
    -
    N_RESOLVED_AFTER_GENUS
    ==
    len(
        post_genus_residual
    )
)


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_GENUS
    /
    N_TOTAL_CANONICAL
)


# =============================================================================
# 32. SUMMARY TABLE
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Canonical taxa entering 9.4P",
            "Approved genus assignments incorporated",
            "Canonical taxa resolved after genus stage",
            "Residual canonical taxa after genus stage",
            "Taxonomic-name rescue pending canonical taxa",
            "Single-congener review canonical taxa",
            "Direct family-audit canonical taxa",
            "Special biological case canonical taxa",
            "Direct family-audit genera",
            "Overall canonical Seed Mass coverage percent",
            "Family Seed Mass values assigned in 9.4P",
        ],

        "Value": [
            len(
                canonical_classification
            ),

            N_GENUS_ASSIGNED,

            N_RESOLVED_AFTER_GENUS,

            len(
                post_genus_residual
            ),

            len(
                taxonomic_rescue_canonical
            ),

            len(
                single_congener_canonical
            ),

            len(
                direct_family_canonical
            ),

            len(
                special_case_canonical
            ),

            family_audit_targets[
                "Residual_Genus"
            ]
            .nunique(),

            coverage_percent,

            0,
        ],
    }
)


print("\nSection 9.4P summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 33. COVERAGE CHECKPOINT
# =============================================================================

print("\nCoverage checkpoint")
print("-" * 100)

print(
    f"Canonical Omo taxa                       : "
    f"{N_TOTAL_CANONICAL}"
)

print(
    f"Resolved before genus assignment         : "
    f"{N_RESOLVED_BEFORE_GENUS}"
)

print(
    f"New genus-level assignments              : "
    f"{N_GENUS_ASSIGNED}"
)

print(
    f"Total resolved after genus assignment    : "
    f"{N_RESOLVED_AFTER_GENUS}"
)

print(
    f"Residual canonical taxa                  : "
    f"{len(post_genus_residual)}"
)

print(
    f"Overall Seed Mass coverage               : "
    f"{coverage_percent:.2f}%"
)


# =============================================================================
# 34. EXPORT
# =============================================================================

post_genus_state.to_csv(
    OUT_POST_GENUS_STATE,
    index=False
)


post_genus_residual.to_csv(
    OUT_POST_GENUS_RESIDUAL,
    index=False
)


taxonomic_rescue_canonical.to_csv(
    OUT_TAXONOMIC_RESCUE_CANONICAL,
    index=False
)


single_congener_canonical.to_csv(
    OUT_SINGLE_CONGENER_CANONICAL,
    index=False
)


direct_family_canonical.to_csv(
    OUT_DIRECT_FAMILY_CANONICAL,
    index=False
)


special_case_canonical.to_csv(
    OUT_SPECIAL_CASE_CANONICAL,
    index=False
)


family_audit_targets.to_csv(
    OUT_FAMILY_AUDIT_TARGETS,
    index=False
)


family_genus_summary.to_csv(
    OUT_FAMILY_AUDIT_GENUS_SUMMARY,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 35. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4P exports")
print("-" * 100)

print(
    f"Post-genus state          : "
    f"{OUT_POST_GENUS_STATE}"
)

print(
    f"Residual canonical taxa   : "
    f"{OUT_POST_GENUS_RESIDUAL}"
)

print(
    f"Taxonomic rescue targets  : "
    f"{OUT_TAXONOMIC_RESCUE_CANONICAL}"
)

print(
    f"Single-congener targets   : "
    f"{OUT_SINGLE_CONGENER_CANONICAL}"
)

print(
    f"Direct family targets     : "
    f"{OUT_DIRECT_FAMILY_CANONICAL}"
)

print(
    f"Special-case targets      : "
    f"{OUT_SPECIAL_CASE_CANONICAL}"
)

print(
    f"Family audit preparation  : "
    f"{OUT_FAMILY_AUDIT_TARGETS}"
)

print(
    f"Family genus summary      : "
    f"{OUT_FAMILY_AUDIT_GENUS_SUMMARY}"
)

print(
    f"Summary                   : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 36. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Post-genus-assignment residual consolidation "
    "completed successfully."
)

print(
    f"Canonical residual taxa before genus stage : "
    f"{len(canonical_classification)}"
)

print(
    f"Genus-level canonical assignments          : "
    f"{len(assigned_canonical)}"
)

print(
    f"Canonical taxa resolved after genus stage  : "
    f"{N_RESOLVED_AFTER_GENUS} / {N_TOTAL_CANONICAL}"
)

print(
    f"Residual canonical taxa                    : "
    f"{len(post_genus_residual)}"
)

print(
    f"Taxonomic-name rescue pending              : "
    f"{len(taxonomic_rescue_canonical)}"
)

print(
    f"Single-congener review taxa                : "
    f"{len(single_congener_canonical)}"
)

print(
    f"Direct family-audit taxa                   : "
    f"{len(direct_family_canonical)}"
)

print(
    f"Special biological case taxa               : "
    f"{len(special_case_canonical)}"
)

print(
    f"Direct family-audit genera                 : "
    f"{family_audit_targets['Residual_Genus'].nunique()}"
)

print(
    "No family-level Seed Mass value has been assigned."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4Q
# EXPLICIT REVIEW OF SINGLE-CONGENER SEED MASS TRANSFER RISK
# =============================================================================
#
# Purpose
# -------
# Review the six genera for which the residual Seed Mass workflow has only
# ONE independent congeneric species available:
#
#     Anthocleista
#     Cola
#     Diospyros
#     Macaranga
#     Pterocarpus
#     Trichilia
#
# Section 9.4O deliberately withheld automatic transfer for these genera.
#
# Conservative review rule
# ------------------------
# A Seed Mass value derived from only one congeneric species is NOT transferred
# automatically unless additional independent corroboration is available.
#
# At the present validated project state, no second independent congeneric
# species or equivalent corroborating genus-level evidence has been established
# for these six genera.
#
# Therefore:
#
#     Review decision = DEFER_TO_FAMILY_AUDIT
#
# The candidate values remain preserved for provenance and later comparison
# with the family-level distribution.
#
# IMPORTANT
# ---------
# No Seed Mass value is assigned in Section 9.4Q.
#
# Expected state
# --------------
# Single-congener genera                  : 6
# Canonical taxa represented              : 10
# Single-congener transfers accepted      : 0
# Cases deferred to family audit          : 6 genera / 10 canonical taxa
#
# Overall resolved coverage remains       : 90 / 127
# Overall residual canonical taxa remains : 37
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

SINGLE_CONGENER_REVIEW_FILE = (
    OUT_DIR /
    "omo_seed_mass_single_congener_transfer_review.csv"
)

SINGLE_CONGENER_CANONICAL_FILE = (
    OUT_DIR /
    "omo_seed_mass_single_congener_review_canonical_targets.csv"
)

DIRECT_FAMILY_CANONICAL_FILE = (
    OUT_DIR /
    "omo_seed_mass_direct_family_audit_canonical_targets.csv"
)

POST_GENUS_RESIDUAL_FILE = (
    OUT_DIR /
    "omo_seed_mass_post_genus_assignment_residual_canonical_taxa.csv"
)

TAXONOMIC_RESCUE_FILE = (
    OUT_DIR /
    "omo_seed_mass_taxonomic_name_rescue_pending_targets.csv"
)

SPECIAL_CASE_FILE = (
    OUT_DIR /
    "omo_seed_mass_special_biological_case_targets.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_REVIEW_DECISIONS = (
    OUT_DIR /
    "omo_seed_mass_single_congener_review_decisions.csv"
)

OUT_CANONICAL_DECISIONS = (
    OUT_DIR /
    "omo_seed_mass_single_congener_review_canonical_decisions.csv"
)

OUT_DEFERRED_FAMILY_TARGETS = (
    OUT_DIR /
    "omo_seed_mass_single_congener_deferred_to_family_audit.csv"
)

OUT_AUGMENTED_FAMILY_TARGETS = (
    OUT_DIR /
    "omo_seed_mass_family_audit_targets_after_single_congener_review.csv"
)

OUT_FAMILY_GENUS_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_family_audit_genus_summary_after_single_congener_review.csv"
)

OUT_POST_9_4Q_RESIDUAL = (
    OUT_DIR /
    "omo_seed_mass_post_single_congener_review_residual_state.csv"
)

OUT_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_single_congener_review_summary.csv"
)


# =============================================================================
# 2. VALIDATE REQUIRED INPUTS
# =============================================================================

required_files = [
    SINGLE_CONGENER_REVIEW_FILE,
    SINGLE_CONGENER_CANONICAL_FILE,
    DIRECT_FAMILY_CANONICAL_FILE,
    POST_GENUS_RESIDUAL_FILE,
    TAXONOMIC_RESCUE_FILE,
    SPECIAL_CASE_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4Q input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

single_review = pd.read_csv(
    SINGLE_CONGENER_REVIEW_FILE,
    low_memory=False
)

single_canonical = pd.read_csv(
    SINGLE_CONGENER_CANONICAL_FILE,
    low_memory=False
)

direct_family = pd.read_csv(
    DIRECT_FAMILY_CANONICAL_FILE,
    low_memory=False
)

post_genus_residual = pd.read_csv(
    POST_GENUS_RESIDUAL_FILE,
    low_memory=False
)

taxonomic_rescue = pd.read_csv(
    TAXONOMIC_RESCUE_FILE,
    low_memory=False
)

special_case = pd.read_csv(
    SPECIAL_CASE_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4Q — EXPLICIT REVIEW OF SINGLE-CONGENER "
    "SEED MASS TRANSFER RISK"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Single-congener genera               : "
    f"{len(single_review)}"
)

print(
    f"Single-congener canonical taxa       : "
    f"{len(single_canonical)}"
)

print(
    f"Direct family-audit canonical taxa   : "
    f"{len(direct_family)}"
)

print(
    f"Post-genus residual canonical taxa   : "
    f"{len(post_genus_residual)}"
)

print(
    f"Taxonomic-rescue canonical taxa      : "
    f"{len(taxonomic_rescue)}"
)

print(
    f"Special biological-case taxa         : "
    f"{len(special_case)}"
)


# =============================================================================
# 4. VALIDATE AUTHORITATIVE 9.4P STATE
# =============================================================================

assert len(
    single_review
) == 6, (
    f"Expected 6 single-congener genera; "
    f"found {len(single_review)}."
)


assert len(
    single_canonical
) == 10, (
    f"Expected 10 canonical taxa represented by the "
    f"single-congener cases; found {len(single_canonical)}."
)


assert len(
    direct_family
) == 21, (
    f"Expected 21 direct family-audit canonical taxa; "
    f"found {len(direct_family)}."
)


assert len(
    post_genus_residual
) == 37, (
    f"Expected 37 post-genus residual canonical taxa; "
    f"found {len(post_genus_residual)}."
)


assert len(
    taxonomic_rescue
) == 5


assert len(
    special_case
) == 1


# =============================================================================
# 5. VALIDATE THE SIX SINGLE-CONGENER GENERA
# =============================================================================

EXPECTED_SINGLE_GENERA = {
    "Anthocleista",
    "Cola",
    "Diospyros",
    "Macaranga",
    "Pterocarpus",
    "Trichilia",
}


observed_single_genera = set(
    single_review[
        "Genus_Search_Target"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert observed_single_genera == EXPECTED_SINGLE_GENERA, (
    "Unexpected single-congener genus set.\n"
    f"Expected : {sorted(EXPECTED_SINGLE_GENERA)}\n"
    f"Observed : {sorted(observed_single_genera)}"
)


# =============================================================================
# 6. VALIDATE CANDIDATE VALUES
# =============================================================================

single_review[
    "Candidate_Seed_Mass_mg"
] = pd.to_numeric(
    single_review[
        "Candidate_Seed_Mass_mg"
    ],
    errors="coerce"
)


assert single_review[
    "Candidate_Seed_Mass_mg"
].notna().all()


assert single_review[
    "Candidate_Seed_Mass_mg"
].gt(0).all()


expected_candidate_values = {
    "Anthocleista": 1.710,
    "Cola": 2000.000,
    "Diospyros": 432.500,
    "Macaranga": 22.795,
    "Pterocarpus": 1922.680,
    "Trichilia": 500.000,
}


for genus_name, expected_value in expected_candidate_values.items():

    observed = single_review.loc[
        single_review[
            "Genus_Search_Target"
        ].eq(
            genus_name
        ),
        "Candidate_Seed_Mass_mg"
    ]

    assert len(
        observed
    ) == 1

    assert np.isclose(
        observed.iloc[0],
        expected_value,
        rtol=0,
        atol=1e-6
    ), (
        f"Unexpected candidate value for {genus_name}.\n"
        f"Expected : {expected_value}\n"
        f"Observed : {observed.iloc[0]}"
    )


# =============================================================================
# 7. DEFINE EXPLICIT REVIEW POLICY
# =============================================================================

REVIEW_POLICY = (
    "A single congeneric species is insufficient for automatic Seed Mass "
    "transfer unless independent corroborating evidence is available. "
    "No second independent congeneric species or equivalent corroborating "
    "genus-level evidence was established in the validated 9.4N-R evidence "
    "audit. Candidate values are therefore preserved but not assigned."
)


# =============================================================================
# 8. APPLY CASE-BY-CASE CONSERVATIVE REVIEW DECISIONS
# =============================================================================
#
# All six cases currently receive the SAME procedural decision:
#
#     DEFER_TO_FAMILY_AUDIT
#
# This does NOT mean that the candidate is biologically wrong.
#
# It means only that evidence depth is insufficient to justify transferring
# that candidate into the final Seed Mass table at this stage.
#
# =============================================================================

review_decisions = (
    single_review
    .copy()
)


review_decisions[
    "Single_Congener_Review_Decision"
] = (
    "DEFER_TO_FAMILY_AUDIT"
)


review_decisions[
    "Single_Congener_Transfer_Accepted"
] = False


review_decisions[
    "Assigned_Seed_Mass_mg"
] = np.nan


review_decisions[
    "Numerical_Seed_Mass_Assigned"
] = False


review_decisions[
    "Review_Rationale"
] = REVIEW_POLICY


review_decisions[
    "Candidate_Value_Status"
] = (
    "PRESERVED_FOR_FAMILY_CONTEXT_COMPARISON"
)


review_decisions[
    "Next_Action"
] = (
    "INCLUDE_IN_FAMILY_LEVEL_EVIDENCE_AUDIT"
)


# =============================================================================
# 9. ADD CASE-SPECIFIC INTERPRETIVE NOTES
# =============================================================================

case_notes = {

    "Anthocleista":
        (
            "Candidate derives from Anthocleista djalonensis. "
            "Only one congeneric species is represented; retain the "
            "1.710 mg value for comparison with Loganiaceae-family evidence "
            "before deciding whether transfer is defensible."
        ),

    "Cola":
        (
            "Candidate derives from Cola nitida and concerns an explicit "
            "genus-level project target (Cola species). Although this is "
            "taxonomically closer than a species-to-species transfer, one "
            "species alone does not characterize genus-wide Seed Mass "
            "variation. Retain 2000 mg for family-context review."
        ),

    "Diospyros":
        (
            "Candidate derives from Diospyros mespiliformis and would be "
            "transferred to three unresolved biological Diospyros targets. "
            "Given potentially substantial interspecific Seed Mass variation "
            "within the genus, the 432.5 mg value is withheld pending "
            "family-context assessment."
        ),

    "Macaranga":
        (
            "Candidate derives from Macaranga alnifolia. One congeneric "
            "species cannot establish genus-level Seed Mass variation; "
            "retain the 22.795 mg candidate without assigning it."
        ),

    "Pterocarpus":
        (
            "Candidate derives from Pterocarpus santalinoides. The "
            "1922.68 mg value may be informative but represents only one "
            "congener and is withheld pending broader family evidence."
        ),

    "Trichilia":
        (
            "Candidate derives from Trichilia monadelpha. The 500 mg "
            "candidate is retained for comparison with family-level "
            "evidence but is not transferred automatically."
        ),
}


review_decisions[
    "Case_Specific_Review_Note"
] = (
    review_decisions[
        "Genus_Search_Target"
    ]
    .map(
        case_notes
    )
)


assert review_decisions[
    "Case_Specific_Review_Note"
].notna().all()


# =============================================================================
# 10. DISPLAY GENUS-LEVEL REVIEW DECISIONS
# =============================================================================

print("\nSingle-congener genus review decisions")
print("-" * 100)

display(
    review_decisions[
        [
            "Genus_Search_Target",
            "Residual_Biological_Targets",
            "Explicit_Genus_Canonical_Taxa",
            "Species_Represented",
            "Candidate_Seed_Mass_mg",
            "Evidence_Sources",
            "Single_Congener_Review_Decision",
            "Numerical_Seed_Mass_Assigned",
            "Case_Specific_Review_Note",
        ]
    ]
)

# =============================================================================
# 11. MAP REVIEW DECISIONS TO CANONICAL TAXA
# =============================================================================
#
# IMPORTANT
# ---------
# single_canonical inherits some workflow fields from Sections 9.4M–9.4P,
# including Next_Action and Numerical_Seed_Mass_Assigned.
#
# The 9.4Q review decisions are now authoritative for these fields.
# Therefore, remove any pre-existing versions before merging so pandas does
# not create _x / _y suffixes.
# =============================================================================

decision_lookup = (
    review_decisions[
        [
            "Genus_Search_Target",
            "Species_Represented",
            "Candidate_Seed_Mass_mg",
            "Evidence_Sources",
            "Single_Congener_Review_Decision",
            "Single_Congener_Transfer_Accepted",
            "Numerical_Seed_Mass_Assigned",
            "Review_Rationale",
            "Case_Specific_Review_Note",
            "Candidate_Value_Status",
            "Next_Action",
        ]
    ]
    .rename(
        columns={
            "Genus_Search_Target":
                "Residual_Genus"
        }
    )
    .copy()
)


# -----------------------------------------------------------------------------
# Remove fields from the canonical table that will be replaced authoritatively
# by Section 9.4Q decisions.
# -----------------------------------------------------------------------------

decision_fields = [
    col
    for col in decision_lookup.columns
    if col != "Residual_Genus"
]


overlapping_decision_fields = [
    col
    for col in decision_fields
    if col in single_canonical.columns
]


if overlapping_decision_fields:

    print(
        "\nPre-existing canonical workflow fields replaced "
        "by authoritative 9.4Q review decisions"
    )
    print("-" * 100)

    for col in overlapping_decision_fields:
        print(
            f" - {col}"
        )


single_canonical_for_merge = (
    single_canonical
    .drop(
        columns=overlapping_decision_fields,
        errors="ignore"
    )
    .copy()
)


# -----------------------------------------------------------------------------
# Merge one genus-level review decision onto all corresponding canonical taxa.
# -----------------------------------------------------------------------------

canonical_decisions = (
    single_canonical_for_merge
    .merge(
        decision_lookup,
        on="Residual_Genus",
        how="left",
        validate="many_to_one"
    )
)


# =============================================================================
# 11A. VALIDATE CANONICAL DECISION MAPPING
# =============================================================================

assert len(
    canonical_decisions
) == 10, (
    f"Expected 10 canonical taxa after single-congener decision mapping; "
    f"found {len(canonical_decisions)}."
)


assert canonical_decisions[
    "Canonical_Species"
].is_unique


assert canonical_decisions[
    "Single_Congener_Review_Decision"
].notna().all()


assert canonical_decisions[
    "Numerical_Seed_Mass_Assigned"
].notna().all()


assert canonical_decisions[
    "Next_Action"
].notna().all()


assert canonical_decisions[
    "Single_Congener_Review_Decision"
].eq(
    "DEFER_TO_FAMILY_AUDIT"
).all()


assert canonical_decisions[
    "Numerical_Seed_Mass_Assigned"
].eq(
    False
).all()


assert canonical_decisions[
    "Next_Action"
].eq(
    "INCLUDE_IN_FAMILY_LEVEL_EVIDENCE_AUDIT"
).all()


# Confirm the six genera still map to the expected 10 canonical taxa.
assert set(
    canonical_decisions[
        "Residual_Genus"
    ]
) == {
    "Anthocleista",
    "Cola",
    "Diospyros",
    "Macaranga",
    "Pterocarpus",
    "Trichilia",
}

# =============================================================================
# 12. DISPLAY CANONICAL SINGLE-CONGENER DECISIONS
# =============================================================================

print("\nCanonical taxa affected by single-congener review")
print("-" * 100)

display(
    canonical_decisions[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Species_Represented",
            "Candidate_Seed_Mass_mg",
            "Single_Congener_Review_Decision",
            "Numerical_Seed_Mass_Assigned",
            "Next_Action",
        ]
    ]
)


# =============================================================================
# 13. CONFIRM NO NUMERIC VALUE WAS ASSIGNED
# =============================================================================

assert review_decisions[
    "Numerical_Seed_Mass_Assigned"
].eq(
    False
).all()


assert canonical_decisions[
    "Numerical_Seed_Mass_Assigned"
].eq(
    False
).all()


assert review_decisions[
    "Assigned_Seed_Mass_mg"
].isna().all()


# =============================================================================
# 14. BUILD DEFERRED FAMILY-AUDIT TABLE
# =============================================================================

deferred_family_targets = (
    canonical_decisions
    .copy()
)


deferred_family_targets[
    "Family_Audit_Entry_Route"
] = (
    "DEFERRED_AFTER_SINGLE_CONGENER_REVIEW"
)


deferred_family_targets[
    "Family_Audit_Status"
] = (
    "READY_FOR_FAMILY_TAXONOMIC_MAPPING"
)


deferred_family_targets[
    "Numerical_Family_Value_Assigned"
] = False


# =============================================================================
# 15. AUGMENT DIRECT FAMILY-AUDIT REGISTER
# =============================================================================

direct_family_augmented = (
    direct_family
    .copy()
)


direct_family_augmented[
    "Family_Audit_Entry_Route"
] = (
    "DIRECT_NO_GENUS_EVIDENCE"
)


direct_family_augmented[
    "Family_Audit_Status"
] = (
    "READY_FOR_FAMILY_TAXONOMIC_MAPPING"
)


direct_family_augmented[
    "Numerical_Family_Value_Assigned"
] = False


# Keep a common core schema for the combined family audit.
common_family_columns = [
    "Canonical_Species",
    "Final_Search_Name",
    "Residual_Genus",
    "Residual_Level",
    "Residual_Classification",
    "Family_Audit_Entry_Route",
    "Family_Audit_Status",
    "Numerical_Family_Value_Assigned",
]


for col in common_family_columns:

    if col not in direct_family_augmented.columns:
        direct_family_augmented[
            col
        ] = pd.NA

    if col not in deferred_family_targets.columns:
        deferred_family_targets[
            col
        ] = pd.NA


augmented_family_targets = pd.concat(
    [
        direct_family_augmented[
            common_family_columns
        ],
        deferred_family_targets[
            common_family_columns
        ],
    ],
    ignore_index=True
)


augmented_family_targets = (
    augmented_family_targets
    .sort_values(
        [
            "Residual_Genus",
            "Final_Search_Name",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 16. VALIDATE FAMILY-AUDIT EXPANSION
# =============================================================================
#
# 21 direct-family canonical taxa
# +10 single-congener canonical taxa
# =31 canonical taxa now ready for family audit.
#
# Taxonomic rescue (5) and Musa special case (1) remain outside.
# =============================================================================

assert len(
    augmented_family_targets
) == 31, (
    f"Expected 31 canonical taxa entering family audit after "
    f"single-congener review; found {len(augmented_family_targets)}."
)


assert augmented_family_targets[
    "Canonical_Species"
].nunique() == 31


assert not (
    set(
        augmented_family_targets[
            "Canonical_Species"
        ]
    )
    &
    set(
        taxonomic_rescue[
            "Canonical_Species"
        ]
    )
)


assert not (
    set(
        augmented_family_targets[
            "Canonical_Species"
        ]
    )
    &
    set(
        special_case[
            "Canonical_Species"
        ]
    )
)


# =============================================================================
# 17. FAMILY-AUDIT GENUS SUMMARY
# =============================================================================

family_genus_summary = (
    augmented_family_targets
    .groupby(
        "Residual_Genus",
        as_index=False
    )
    .agg(
        N_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Biological_Targets=(
            "Final_Search_Name",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Family_Audit_Entry_Routes=(
            "Family_Audit_Entry_Route",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
)


print("\nFamily-audit target structure after single-congener review")
print("-" * 100)

display(
    family_genus_summary
)


# =============================================================================
# 18. VALIDATE NUMBER OF FAMILY-AUDIT GENERA
# =============================================================================

n_family_audit_genera = (
    augmented_family_targets[
        "Residual_Genus"
    ]
    .nunique()
)


print(
    f"\nUnique genera entering family-level audit : "
    f"{n_family_audit_genera}"
)


# 15 original direct-family genera + 6 reviewed single-congener genera.
assert n_family_audit_genera == 21, (
    f"Expected 21 unique genera entering the family audit; "
    f"found {n_family_audit_genera}."
)


# =============================================================================
# 19. POST-9.4Q RESIDUAL STATE
# =============================================================================
#
# No new numerical values were assigned.
#
# Therefore total resolved coverage remains:
#
#     90 / 127
#
# Residual remains:
#
#     37 canonical taxa
#
# They are now partitioned as:
#
#     31 family-audit taxa
#      5 taxonomic-name rescue taxa
#      1 special biological case
#
# =============================================================================

post_9_4Q_residual = (
    post_genus_residual
    .copy()
)


family_set = set(
    augmented_family_targets[
        "Canonical_Species"
    ]
)


taxonomic_set = set(
    taxonomic_rescue[
        "Canonical_Species"
    ]
)


special_set = set(
    special_case[
        "Canonical_Species"
    ]
)


def assign_post_q_route(
    canonical_name
):

    if canonical_name in family_set:

        return (
            "FAMILY_LEVEL_EVIDENCE_AUDIT"
        )

    if canonical_name in taxonomic_set:

        return (
            "TAXONOMIC_NAME_RESCUE_PENDING"
        )

    if canonical_name in special_set:

        return (
            "SPECIAL_BIOLOGICAL_CASE_REVIEW"
        )

    return (
        "UNCLASSIFIED"
    )


post_9_4Q_residual[
    "Post_9_4Q_Route"
] = (
    post_9_4Q_residual[
        "Canonical_Species"
    ]
    .map(
        assign_post_q_route
    )
)


assert not post_9_4Q_residual[
    "Post_9_4Q_Route"
].eq(
    "UNCLASSIFIED"
).any()


# =============================================================================
# 20. FINAL POST-9.4Q ROUTING
# =============================================================================

post_q_route_summary = (
    post_9_4Q_residual[
        "Post_9_4Q_Route"
    ]
    .value_counts()
    .rename_axis(
        "Post_9_4Q_Route"
    )
    .reset_index(
        name="N_Canonical_Taxa"
    )
)


print("\nPost-9.4Q residual routing")
print("-" * 100)

display(
    post_q_route_summary
)


# =============================================================================
# 21. STRONG RECONCILIATION
# =============================================================================

assert len(
    post_9_4Q_residual
) == 37


assert len(
    augmented_family_targets
) == 31


assert len(
    taxonomic_rescue
) == 5


assert len(
    special_case
) == 1


assert (
    len(
        augmented_family_targets
    )
    +
    len(
        taxonomic_rescue
    )
    +
    len(
        special_case
    )
    ==
    37
)


# =============================================================================
# 22. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_AFTER_GENUS = 90

N_NEW_ASSIGNMENTS_9_4Q = 0

N_RESOLVED_AFTER_9_4Q = (
    N_RESOLVED_AFTER_GENUS
    +
    N_NEW_ASSIGNMENTS_9_4Q
)


assert N_RESOLVED_AFTER_9_4Q == 90


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4Q
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 23. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Single-congener genera reviewed",
            "Canonical taxa represented by single-congener cases",
            "Single-congener transfers accepted",
            "Single-congener transfers deferred",
            "Seed Mass values assigned in Section 9.4Q",
            "Direct family-audit canonical taxa before 9.4Q",
            "Single-congener canonical taxa added to family audit",
            "Total canonical taxa entering family audit",
            "Unique genera entering family audit",
            "Taxonomic-name rescue canonical taxa remaining",
            "Special biological-case canonical taxa remaining",
            "Total resolved canonical taxa after 9.4Q",
            "Residual canonical taxa after 9.4Q",
            "Overall Seed Mass coverage percent",
        ],

        "Value": [
            len(
                review_decisions
            ),

            len(
                canonical_decisions
            ),

            int(
                review_decisions[
                    "Single_Congener_Transfer_Accepted"
                ]
                .sum()
            ),

            int(
                review_decisions[
                    "Single_Congener_Review_Decision"
                ]
                .eq(
                    "DEFER_TO_FAMILY_AUDIT"
                )
                .sum()
            ),

            0,

            len(
                direct_family
            ),

            len(
                deferred_family_targets
            ),

            len(
                augmented_family_targets
            ),

            n_family_audit_genera,

            len(
                taxonomic_rescue
            ),

            len(
                special_case
            ),

            N_RESOLVED_AFTER_9_4Q,

            len(
                post_9_4Q_residual
            ),

            coverage_percent,
        ],
    }
)


print("\nSection 9.4Q summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 24. EXPORT
# =============================================================================

review_decisions.to_csv(
    OUT_REVIEW_DECISIONS,
    index=False
)


canonical_decisions.to_csv(
    OUT_CANONICAL_DECISIONS,
    index=False
)


deferred_family_targets.to_csv(
    OUT_DEFERRED_FAMILY_TARGETS,
    index=False
)


augmented_family_targets.to_csv(
    OUT_AUGMENTED_FAMILY_TARGETS,
    index=False
)


family_genus_summary.to_csv(
    OUT_FAMILY_GENUS_SUMMARY,
    index=False
)


post_9_4Q_residual.to_csv(
    OUT_POST_9_4Q_RESIDUAL,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 25. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4Q exports")
print("-" * 100)

print(
    f"Genus review decisions     : "
    f"{OUT_REVIEW_DECISIONS}"
)

print(
    f"Canonical review decisions : "
    f"{OUT_CANONICAL_DECISIONS}"
)

print(
    f"Deferred family targets    : "
    f"{OUT_DEFERRED_FAMILY_TARGETS}"
)

print(
    f"Augmented family targets   : "
    f"{OUT_AUGMENTED_FAMILY_TARGETS}"
)

print(
    f"Family genus summary       : "
    f"{OUT_FAMILY_GENUS_SUMMARY}"
)

print(
    f"Post-9.4Q residual state   : "
    f"{OUT_POST_9_4Q_RESIDUAL}"
)

print(
    f"Summary                    : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 26. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Single-congener Seed Mass transfer-risk review "
    "completed successfully."
)

print(
    f"Single-congener genera reviewed           : "
    f"{len(review_decisions)}"
)

print(
    f"Canonical taxa represented                : "
    f"{len(canonical_decisions)}"
)

print(
    "Single-congener transfers accepted        : 0"
)

print(
    f"Single-congener genera deferred            : "
    f"{len(review_decisions)}"
)

print(
    f"Canonical taxa now entering family audit  : "
    f"{len(augmented_family_targets)}"
)

print(
    f"Unique genera entering family audit       : "
    f"{n_family_audit_genera}"
)

print(
    f"Total Seed Mass coverage remains          : "
    f"{N_RESOLVED_AFTER_9_4Q} / {N_CANONICAL_TAXA} "
    f"({coverage_percent:.2f}%)"
)

print(
    f"Residual canonical taxa remain            : "
    f"{len(post_9_4Q_residual)}"
)

print(
    "No single-congener Seed Mass value has been assigned."
)

print(
    "Candidate values remain preserved for comparison "
    "with family-level evidence."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R
# FAMILY TAXONOMIC MAPPING AND FAMILY-LEVEL SEED MASS EVIDENCE AUDIT
# =============================================================================
#
# Purpose
# -------
# 1. Start from the 31 canonical taxa routed to family-level audit in 9.4Q.
# 2. Identify their validated botanical families from the project taxonomy.
# 3. Map the normalized independent biological-species Seed Mass evidence
#    pool from 9.4N-R to botanical families.
# 4. Prevent project spelling variants / aliases from inflating family
#    evidence depth.
# 5. Quantify independent species support within every required family.
# 6. Classify family evidence depth:
#
#       STRONG_FAMILY_EVIDENCE    >= 5 independent biological species
#       MODERATE_FAMILY_EVIDENCE   3–4 independent biological species
#       LIMITED_FAMILY_EVIDENCE    1–2 independent biological species
#       NO_FAMILY_EVIDENCE         0 independent biological species
#
# IMPORTANT
# ---------
# This is an AUDIT ONLY.
#
# NO family-level Seed Mass value is assigned.
# NO taxon is resolved in Section 9.4R.
#
# Expected starting state
# -----------------------
# Family-audit canonical taxa : 31
# Residual genera             : 21
# Overall resolved taxa       : 90 / 127
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# -----------------------------------------------------------------------------
# Authoritative 9.4Q family-audit target set
# -----------------------------------------------------------------------------

FAMILY_TARGET_FILE = (
    OUT_DIR /
    "omo_seed_mass_family_audit_targets_after_single_congener_review.csv"
)


# -----------------------------------------------------------------------------
# Authoritative normalized biological-species evidence pool from 9.4N-R
# -----------------------------------------------------------------------------

NORMALIZED_SPECIES_POOL_FILE = (
    OUT_DIR /
    "omo_seed_mass_genus_audit_species_evidence_pool_normalized.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_DISCOVERED_TAXONOMY_SOURCE = (
    OUT_DIR /
    "omo_seed_mass_family_audit_taxonomy_source_diagnostic.csv"
)

OUT_CANONICAL_FAMILY_MAP = (
    OUT_DIR /
    "omo_seed_mass_family_audit_canonical_family_map.csv"
)

OUT_EVIDENCE_FAMILY_MAPPING = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_species_mapping.csv"
)

OUT_FAMILY_EVIDENCE_DETAIL = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_detail.csv"
)

OUT_FAMILY_EVIDENCE_AUDIT = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_audit.csv"
)

OUT_FAMILY_EVIDENCE_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_audit_summary.csv"
)

OUT_UNMAPPED_EVIDENCE = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_unmapped_species.csv"
)


# =============================================================================
# 2. VALIDATE PRIMARY INPUT FILES
# =============================================================================

for path in [
    FAMILY_TARGET_FILE,
    NORMALIZED_SPECIES_POOL_FILE,
]:

    assert path.exists(), (
        f"Required Section 9.4R input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD PRIMARY INPUTS
# =============================================================================

family_targets = pd.read_csv(
    FAMILY_TARGET_FILE,
    low_memory=False
)

species_pool = pd.read_csv(
    NORMALIZED_SPECIES_POOL_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R — FAMILY TAXONOMIC MAPPING AND "
    "FAMILY-LEVEL SEED MASS EVIDENCE AUDIT"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Family-audit canonical taxa       : "
    f"{len(family_targets)}"
)

print(
    f"Residual genera                   : "
    f"{family_targets['Residual_Genus'].nunique()}"
)

print(
    f"Normalized biological species     : "
    f"{len(species_pool)}"
)


assert len(
    family_targets
) == 31, (
    f"Expected 31 canonical family-audit taxa; "
    f"found {len(family_targets)}."
)


assert (
    family_targets[
        "Residual_Genus"
    ].nunique()
    ==
    21
), (
    "Expected 21 unique residual genera entering family audit."
)


# =============================================================================
# 4. HELPERS
# =============================================================================

def clean_text(value):

    if pd.isna(value):
        return pd.NA

    value = str(value).strip()

    if value in {
        "",
        "nan",
        "None",
        "<NA>",
    }:
        return pd.NA

    return value


def clean_series(series):

    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "": pd.NA,
                "nan": pd.NA,
                "None": pd.NA,
                "<NA>": pd.NA,
            }
        )
    )


def find_column(
    df,
    candidates,
    required=True
):

    lookup = {
        str(col)
        .strip()
        .lower()
        .replace(" ", "_"):
            col
        for col in df.columns
    }

    for candidate in candidates:

        if candidate in df.columns:
            return candidate

        normalized_candidate = (
            str(candidate)
            .strip()
            .lower()
            .replace(" ", "_")
        )

        if normalized_candidate in lookup:
            return lookup[
                normalized_candidate
            ]

    if required:

        raise KeyError(
            "Required field not found.\n"
            f"Candidates tested: {candidates}\n"
            f"Available columns: {list(df.columns)}"
        )

    return None


def join_unique(series):

    values = []

    for value in series.dropna():

        text = str(value).strip()

        if not text:
            continue

        pieces = [
            x.strip()
            for x in text.split("|")
            if x.strip()
        ]

        values.extend(
            pieces
        )

    return " | ".join(
        sorted(
            set(values)
        )
    )


def split_pipe_values(value):

    if pd.isna(value):
        return []

    text = str(value).strip()

    if not text:
        return []

    return [
        x.strip()
        for x in text.split("|")
        if x.strip()
    ]

# =============================================================================
# 5. DISCOVER AUTHORITATIVE PROJECT TAXONOMY TABLE
#    USING ACTUAL 31-TARGET COVERAGE
# =============================================================================
#
# IMPORTANT
# ---------
# The previous diagnostic selected taxonomy sources largely on overall
# canonical-row count.
#
# For Section 9.4R, the decisive criterion must instead be:
#
#     How many of the CURRENT 31 family-audit canonical taxa
#     receive non-missing family assignments?
#
# A table with 127 names is not useful if those names belong to an earlier
# spelling/taxonomic state.
# =============================================================================

CANONICAL_COLUMN_CANDIDATES = [
    "Canonical_Species",
    "Canonical Species",
    "CanonicalSpecies",
    "Accepted_Species",
    "Accepted Species",
]


FAMILY_COLUMN_CANDIDATES = [
    "Family",
    "Family_Name",
    "Family Name",
    "Taxonomic_Family",
    "Accepted_Family",
    "APG_Family",
]


target_canonical_set = set(
    family_targets[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert len(
    target_canonical_set
) == 31


taxonomy_candidates = []


print("\nDiscovering project taxonomy source using target coverage")
print("-" * 100)


for csv_path in PROJECT_ROOT.rglob("*.csv"):

    # Do not use newly generated family-audit products as inputs.
    if (
        csv_path.parent == OUT_DIR
        and (
            "family_audit" in csv_path.name.lower()
            or
            "family_evidence" in csv_path.name.lower()
        )
    ):
        continue


    try:

        header_df = pd.read_csv(
            csv_path,
            nrows=5,
            low_memory=False
        )

    except Exception:

        continue


    canonical_col = find_column(
        header_df,
        CANONICAL_COLUMN_CANDIDATES,
        required=False
    )


    family_col = find_column(
        header_df,
        FAMILY_COLUMN_CANDIDATES,
        required=False
    )


    if (
        canonical_col is None
        or
        family_col is None
    ):
        continue


    try:

        candidate_df = pd.read_csv(
            csv_path,
            usecols=[
                canonical_col,
                family_col,
            ],
            low_memory=False
        )

    except Exception:

        continue


    candidate_df = candidate_df.rename(
        columns={
            canonical_col:
                "Canonical_Species",

            family_col:
                "Family",
        }
    )


    candidate_df[
        "Canonical_Species"
    ] = clean_series(
        candidate_df[
            "Canonical_Species"
        ]
    )


    candidate_df[
        "Family"
    ] = clean_series(
        candidate_df[
            "Family"
        ]
    )


    candidate_df = (
        candidate_df[
            candidate_df[
                "Canonical_Species"
            ].notna()
        ]
        .copy()
    )


    # -------------------------------------------------------------------------
    # Coverage specifically of the 31 current family-audit canonical taxa
    # -------------------------------------------------------------------------

    target_matches = candidate_df[
        candidate_df[
            "Canonical_Species"
        ].isin(
            target_canonical_set
        )
    ].copy()


    matched_target_names = set(
        target_matches[
            "Canonical_Species"
        ]
        .dropna()
    )


    matched_with_family = set(
        target_matches.loc[
            target_matches[
                "Family"
            ].notna(),
            "Canonical_Species"
        ]
    )


    n_target_names_found = len(
        matched_target_names
    )


    n_targets_with_family = len(
        matched_with_family
    )


    n_unique_canonical = (
        candidate_df[
            "Canonical_Species"
        ]
        .nunique()
    )


    n_unique_families = (
        candidate_df[
            "Family"
        ]
        .dropna()
        .nunique()
    )


    # -------------------------------------------------------------------------
    # Score:
    #
    # Target-family coverage is overwhelmingly more important than
    # overall file size.
    # -------------------------------------------------------------------------

    score = (
        n_targets_with_family * 10000
        +
        n_target_names_found * 100
    )


    # Exact backbone size is useful only as a secondary preference.
    if n_unique_canonical == 127:
        score += 100


    name_lower = (
        csv_path.name.lower()
    )


    for preference_term in [
        "species_metadata",
        "taxonomy",
        "taxonomic",
        "canonical",
        "metadata",
    ]:

        if preference_term in name_lower:
            score += 10


    taxonomy_candidates.append(
        {
            "Path":
                str(csv_path),

            "N_Target_Names_Found":
                n_target_names_found,

            "N_Targets_With_Family":
                n_targets_with_family,

            "Percent_Target_Family_Coverage":
                (
                    100
                    *
                    n_targets_with_family
                    /
                    31
                ),

            "N_Unique_Canonical":
                n_unique_canonical,

            "N_Unique_Families":
                n_unique_families,

            "Score":
                score,
        }
    )


taxonomy_candidate_table = pd.DataFrame(
    taxonomy_candidates
)


if len(
    taxonomy_candidate_table
) == 0:

    raise FileNotFoundError(
        "No project CSV containing both canonical-species "
        "and family fields could be located."
    )


taxonomy_candidate_table = (
    taxonomy_candidate_table
    .sort_values(
        [
            "N_Targets_With_Family",
            "N_Target_Names_Found",
            "Score",
        ],
        ascending=[
            False,
            False,
            False,
        ]
    )
    .reset_index(drop=True)
)


print(
    f"Candidate taxonomy tables located : "
    f"{len(taxonomy_candidate_table)}"
)


print("\nBest taxonomy candidates by 31-target coverage")
print("-" * 100)

display(
    taxonomy_candidate_table.head(
        15
    )
)


# =============================================================================
# 6. SELECT BEST-COVERAGE TAXONOMY SOURCE
# =============================================================================

selected_taxonomy = (
    taxonomy_candidate_table
    .iloc[0]
)


TAXONOMY_FILE = Path(
    selected_taxonomy[
        "Path"
    ]
)


print("\nSelected project taxonomy source")
print("-" * 100)

print(
    f"File                       : "
    f"{TAXONOMY_FILE}"
)

print(
    f"Family-audit targets found : "
    f"{int(selected_taxonomy['N_Target_Names_Found'])} / 31"
)

print(
    f"Targets with family        : "
    f"{int(selected_taxonomy['N_Targets_With_Family'])} / 31"
)

print(
    f"Target-family coverage     : "
    f"{selected_taxonomy['Percent_Target_Family_Coverage']:.2f}%"
)


# =============================================================================
# 7. LOAD SELECTED TAXONOMY TABLE ROBUSTLY
# =============================================================================

taxonomy_raw = pd.read_csv(
    TAXONOMY_FILE,
    low_memory=False
)


TAXONOMY_CANON_COL = find_column(
    taxonomy_raw,
    CANONICAL_COLUMN_CANDIDATES
)


TAXONOMY_FAMILY_COL = find_column(
    taxonomy_raw,
    FAMILY_COLUMN_CANDIDATES
)


print(
    f"Canonical field            : "
    f"{TAXONOMY_CANON_COL}"
)

print(
    f"Family field               : "
    f"{TAXONOMY_FAMILY_COL}"
)


taxonomy = (
    taxonomy_raw[
        [
            TAXONOMY_CANON_COL,
            TAXONOMY_FAMILY_COL,
        ]
    ]
    .rename(
        columns={
            TAXONOMY_CANON_COL:
                "Canonical_Species",

            TAXONOMY_FAMILY_COL:
                "Family",
        }
    )
    .copy()
)


taxonomy[
    "Canonical_Species"
] = clean_series(
    taxonomy[
        "Canonical_Species"
    ]
)


taxonomy[
    "Family"
] = clean_series(
    taxonomy[
        "Family"
    ]
)


taxonomy = (
    taxonomy[
        taxonomy[
            "Canonical_Species"
        ].notna()
    ]
    .drop_duplicates()
    .reset_index(drop=True)
)


# =============================================================================
# 7A. FAMILY-CONFLICT VALIDATION
# =============================================================================

taxonomy_family_counts = (
    taxonomy[
        taxonomy[
            "Family"
        ].notna()
    ]
    .groupby(
        "Canonical_Species"
    )[
        "Family"
    ]
    .nunique()
)


family_conflict_taxa = (
    taxonomy_family_counts[
        taxonomy_family_counts.gt(
            1
        )
    ]
)


assert len(
    family_conflict_taxa
) == 0, (
    "Some canonical taxa map to multiple families in the "
    "selected project taxonomy:\n"
    f"{family_conflict_taxa.to_dict()}"
)


taxonomy = (
    taxonomy
    .sort_values(
        [
            "Canonical_Species",
            "Family",
        ]
    )
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ],
        keep="first"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 8. MAP 31 FAMILY-AUDIT CANONICAL TAXA
# =============================================================================

canonical_family_map = (
    family_targets
    .merge(
        taxonomy,
        on="Canonical_Species",
        how="left",
        validate="many_to_one"
    )
)


unmapped_family_targets = (
    canonical_family_map[
        canonical_family_map[
            "Family"
        ].isna()
    ]
    .copy()
)


print("\nCanonical family-mapping coverage")
print("-" * 100)

print(
    f"Family-audit taxa : "
    f"{len(canonical_family_map)}"
)

print(
    f"Mapped to family  : "
    f"{canonical_family_map['Family'].notna().sum()}"
)

print(
    f"Unmapped          : "
    f"{len(unmapped_family_targets)}"
)


if len(
    unmapped_family_targets
) > 0:

    print("\nStill-unmapped family-audit taxa")
    print("-" * 100)

    display(
        unmapped_family_targets[
            [
                "Canonical_Species",
                "Final_Search_Name",
                "Residual_Genus",
            ]
        ]
    )


assert len(
    unmapped_family_targets
) == 0, (
    "\nThe best available project taxonomy source still "
    "does not cover all 31 current family-audit taxa.\n\n"
    "Do NOT manually guess the missing families yet.\n"
    "The displayed unmapped register should be reconciled "
    "against the validated canonical taxonomy first."
)


print("\nCanonical family mapping")
print("-" * 100)

display(
    canonical_family_map[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Family",
            "Family_Audit_Entry_Route",
        ]
    ]
    .sort_values(
        [
            "Family",
            "Residual_Genus",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 9. VALIDATE GENUS -> FAMILY CONSISTENCY IN RESIDUAL TARGETS
# =============================================================================

genus_family_counts = (
    canonical_family_map
    .groupby(
        "Residual_Genus"
    )[
        "Family"
    ]
    .nunique()
)


genus_family_conflicts = (
    genus_family_counts[
        genus_family_counts.gt(
            1
        )
    ]
)


assert len(
    genus_family_conflicts
) == 0, (
    "A residual genus maps to more than one family:\n"
    f"{genus_family_conflicts.to_dict()}"
)


target_genus_family = (
    canonical_family_map[
        [
            "Residual_Genus",
            "Family",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "Family",
            "Residual_Genus",
        ]
    )
    .reset_index(drop=True)
)


print("\nResidual genus-to-family mapping")
print("-" * 100)

display(
    target_genus_family
)


N_TARGET_FAMILIES = (
    canonical_family_map[
        "Family"
    ]
    .nunique()
)


print(
    f"\nUnique families represented by "
    f"21 residual genera : {N_TARGET_FAMILIES}"
)


# =============================================================================
# 10. VALIDATE NORMALIZED SPECIES EVIDENCE POOL
# =============================================================================

required_species_pool_columns = {
    "Biological_Genus",
    "Biological_Species_Name",
    "Canonical_Species_Represented",
    "Evidence_Sources",
    "Seed_Mass_mg",
}


missing_species_pool_columns = (
    required_species_pool_columns
    -
    set(
        species_pool.columns
    )
)


assert not missing_species_pool_columns, (
    "Normalized species evidence pool lacks required fields:\n"
    f"{sorted(missing_species_pool_columns)}"
)


species_pool[
    "Biological_Species_Name"
] = clean_series(
    species_pool[
        "Biological_Species_Name"
    ]
)


species_pool[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_pool[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert species_pool[
    "Seed_Mass_mg"
].notna().all()


assert species_pool[
    "Seed_Mass_mg"
].gt(
    0
).all()


assert not species_pool[
    "Biological_Species_Name"
].duplicated().any(), (
    "Normalized biological-species evidence pool unexpectedly "
    "contains duplicate biological species."
)


# =============================================================================
# 11. EXPLODE CANONICAL REPRESENTATION OF EVIDENCE SPECIES
# =============================================================================
#
# Each independent biological species may represent one or several project
# canonical spelling variants.
#
# Expand those canonical names only for FAMILY MAPPING.
#
# The independent biological species itself remains counted once.
# =============================================================================

evidence_mapping_rows = []


for _, row in species_pool.iterrows():

    represented_canonical = split_pipe_values(
        row[
            "Canonical_Species_Represented"
        ]
    )


    if len(
        represented_canonical
    ) == 0:

        evidence_mapping_rows.append(
            {
                "Biological_Genus":
                    row[
                        "Biological_Genus"
                    ],

                "Biological_Species_Name":
                    row[
                        "Biological_Species_Name"
                    ],

                "Canonical_Species":
                    pd.NA,

                "Seed_Mass_mg":
                    row[
                        "Seed_Mass_mg"
                    ],

                "Evidence_Sources":
                    row[
                        "Evidence_Sources"
                    ],
            }
        )

        continue


    for canonical_name in represented_canonical:

        evidence_mapping_rows.append(
            {
                "Biological_Genus":
                    row[
                        "Biological_Genus"
                    ],

                "Biological_Species_Name":
                    row[
                        "Biological_Species_Name"
                    ],

                "Canonical_Species":
                    canonical_name,

                "Seed_Mass_mg":
                    row[
                        "Seed_Mass_mg"
                    ],

                "Evidence_Sources":
                    row[
                        "Evidence_Sources"
                    ],
            }
        )


evidence_expanded = pd.DataFrame(
    evidence_mapping_rows
)


evidence_expanded[
    "Canonical_Species"
] = clean_series(
    evidence_expanded[
        "Canonical_Species"
    ]
)


# =============================================================================
# 12. MAP EVIDENCE SPECIES TO FAMILY THROUGH PROJECT CANONICAL TAXONOMY
# =============================================================================

evidence_expanded = (
    evidence_expanded
    .merge(
        taxonomy,
        on="Canonical_Species",
        how="left",
        validate="many_to_one"
    )
)


# =============================================================================
# 13. RESOLVE FAMILY PER INDEPENDENT BIOLOGICAL SPECIES
# =============================================================================

species_family_diagnostic = (
    evidence_expanded
    .groupby(
        [
            "Biological_Genus",
            "Biological_Species_Name",
        ],
        dropna=False
    )
    .agg(
        N_Project_Canonical_Rows=(
            "Canonical_Species",
            "nunique"
        ),

        Canonical_Species_Represented=(
            "Canonical_Species",
            join_unique
        ),

        N_Mapped_Families=(
            "Family",
            lambda x:
                x.dropna()
                .nunique()
        ),

        Families_Mapped=(
            "Family",
            join_unique
        ),

        Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),
    )
    .reset_index()
)


family_conflict_evidence = (
    species_family_diagnostic[
        species_family_diagnostic[
            "N_Mapped_Families"
        ].gt(
            1
        )
    ]
    .copy()
)


if len(
    family_conflict_evidence
) > 0:

    print(
        "\nEvidence species mapping to multiple families"
    )
    print("-" * 100)

    display(
        family_conflict_evidence
    )


assert len(
    family_conflict_evidence
) == 0, (
    "At least one independent biological evidence species "
    "maps to multiple botanical families."
)


species_family_diagnostic[
    "Family"
] = (
    species_family_diagnostic[
        "Families_Mapped"
    ]
    .replace(
        {
            "":
                pd.NA
        }
    )
)


# =============================================================================
# 14. UNMAPPED EVIDENCE SPECIES DIAGNOSTIC
# =============================================================================

unmapped_evidence = (
    species_family_diagnostic[
        species_family_diagnostic[
            "Family"
        ].isna()
    ]
    .copy()
)


print("\nIndependent evidence species lacking family mapping")
print("-" * 100)

print(
    f"N species : {len(unmapped_evidence)}"
)


if len(
    unmapped_evidence
) > 0:

    display(
        unmapped_evidence[
            [
                "Biological_Genus",
                "Biological_Species_Name",
                "Canonical_Species_Represented",
                "Evidence_Sources",
                "Seed_Mass_mg",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 15. BUILD MAPPED INDEPENDENT SPECIES EVIDENCE POOL
# =============================================================================

mapped_species_family = (
    species_family_diagnostic[
        species_family_diagnostic[
            "Family"
        ].notna()
    ]
    .copy()
)


assert not mapped_species_family[
    "Biological_Species_Name"
].duplicated().any()


print("\nFamily-mapped independent biological evidence pool")
print("-" * 100)

print(
    f"Independent biological species : "
    f"{len(species_pool)}"
)

print(
    f"Mapped to project families      : "
    f"{len(mapped_species_family)}"
)

print(
    f"Unmapped biological species     : "
    f"{len(unmapped_evidence)}"
)


# =============================================================================
# 16. RESTRICT EVIDENCE TO REQUIRED RESIDUAL FAMILIES
# =============================================================================

target_family_set = set(
    canonical_family_map[
        "Family"
    ]
)


family_evidence_detail = (
    mapped_species_family[
        mapped_species_family[
            "Family"
        ].isin(
            target_family_set
        )
    ]
    .copy()
    .sort_values(
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nIndependent Seed Mass evidence within required residual families")
print("-" * 100)

display(
    family_evidence_detail[
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
            "Seed_Mass_mg",
            "Evidence_Sources",
            "Canonical_Species_Represented",
        ]
    ]
)


# =============================================================================
# 17. SUMMARIZE FAMILY-LEVEL EVIDENCE
# =============================================================================

family_evidence_summary_existing = (
    family_evidence_detail
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        N_Independent_Genera=(
            "Biological_Genus",
            "nunique"
        ),

        Species_Represented=(
            "Biological_Species_Name",
            join_unique
        ),

        Genera_Represented=(
            "Biological_Genus",
            join_unique
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


# =============================================================================
# 18. BUILD REQUIRED-FAMILY TARGET STRUCTURE
# =============================================================================

required_family_targets = (
    canonical_family_map
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Residual_Genera=(
            "Residual_Genus",
            "nunique"
        ),

        Residual_Genera=(
            "Residual_Genus",
            join_unique
        ),

        N_Residual_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Residual_Canonical_Taxa=(
            "Canonical_Species",
            join_unique
        ),

        Residual_Biological_Targets=(
            "Final_Search_Name",
            join_unique
        ),

        Family_Audit_Entry_Routes=(
            "Family_Audit_Entry_Route",
            join_unique
        ),
    )
)


# =============================================================================
# 19. MERGE ZERO-EVIDENCE FAMILIES BACK INTO AUDIT
# =============================================================================

family_audit = (
    required_family_targets
    .merge(
        family_evidence_summary_existing,
        on="Family",
        how="left",
        validate="one_to_one"
    )
)


family_audit[
    "N_Independent_Species"
] = (
    family_audit[
        "N_Independent_Species"
    ]
    .fillna(
        0
    )
    .astype(
        int
    )
)


family_audit[
    "N_Independent_Genera"
] = (
    family_audit[
        "N_Independent_Genera"
    ]
    .fillna(
        0
    )
    .astype(
        int
    )
)


for col in [
    "Species_Represented",
    "Genera_Represented",
    "Evidence_Sources",
]:

    family_audit[
        col
    ] = (
        family_audit[
            col
        ]
        .fillna(
            ""
        )
    )


# =============================================================================
# 20. CLASSIFY FAMILY EVIDENCE DEPTH
# =============================================================================

def classify_family_evidence(
    n_species
):

    if n_species >= 5:

        return (
            "STRONG_FAMILY_EVIDENCE"
        )

    if n_species >= 3:

        return (
            "MODERATE_FAMILY_EVIDENCE"
        )

    if n_species >= 1:

        return (
            "LIMITED_FAMILY_EVIDENCE"
        )

    return (
        "NO_FAMILY_EVIDENCE"
    )


family_audit[
    "Family_Evidence_Class"
] = (
    family_audit[
        "N_Independent_Species"
    ]
    .map(
        classify_family_evidence
    )
)


# =============================================================================
# 21. NEXT AUDIT ACTION
# =============================================================================

def family_next_action(
    row
):

    n_species = (
        row[
            "N_Independent_Species"
        ]
    )

    n_genera = (
        row[
            "N_Independent_Genera"
        ]
    )


    if (
        n_species >= 5
        and
        n_genera >= 2
    ):

        return (
            "CANDIDATE_FOR_FAMILY_ESTIMATE_RULE_DEVELOPMENT"
        )


    if n_species >= 3:

        return (
            "REVIEW_FAMILY_ESTIMATE_FEASIBILITY"
        )


    if n_species >= 1:

        return (
            "LIMITED_FAMILY_EVIDENCE_REVIEW_TRANSFER_RISK"
        )


    return (
        "NO_FAMILY_EVIDENCE_REQUIRE_BROADER_FALLBACK"
    )


family_audit[
    "Next_Audit_Action"
] = (
    family_audit
    .apply(
        family_next_action,
        axis=1
    )
)


family_audit[
    "Numerical_Family_Value_Assigned"
] = False


# =============================================================================
# 22. DISPLAY AUTHORITATIVE FAMILY AUDIT
# =============================================================================

print("\nResidual family-level Seed Mass evidence audit")
print("-" * 100)


display(
    family_audit[
        [
            "Family",
            "N_Residual_Genera",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Genera_Represented",
            "Evidence_Sources",
            "Min_Seed_Mass_mg",
            "Median_Seed_Mass_mg",
            "Max_Seed_Mass_mg",
            "Family_Evidence_Class",
            "Next_Audit_Action",
        ]
    ]
    .sort_values(
        [
            "Family_Evidence_Class",
            "Family",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 23. FAMILY EVIDENCE-CLASS SUMMARY
# =============================================================================

family_class_summary = (
    family_audit[
        "Family_Evidence_Class"
    ]
    .value_counts()
    .rename_axis(
        "Family_Evidence_Class"
    )
    .reset_index(
        name="N_Families"
    )
)


family_class_summary[
    "Percent_of_Target_Families"
] = (
    family_class_summary[
        "N_Families"
    ]
    /
    len(
        family_audit
    )
    *
    100
)


print("\nFamily evidence-depth summary")
print("-" * 100)

display(
    family_class_summary
)


# =============================================================================
# 24. STRONG-FAMILY CANDIDATES
# =============================================================================

strong_family = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].eq(
            "STRONG_FAMILY_EVIDENCE"
        )
    ]
    .copy()
)


print("\nFamilies with strong existing Seed Mass evidence")
print("-" * 100)

print(
    f"N families : "
    f"{len(strong_family)}"
)


if len(
    strong_family
) > 0:

    display(
        strong_family[
            [
                "Family",
                "N_Independent_Species",
                "N_Independent_Genera",
                "Genera_Represented",
                "Median_Seed_Mass_mg",
                "Residual_Genera",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 25. MODERATE-FAMILY CANDIDATES
# =============================================================================

moderate_family = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].eq(
            "MODERATE_FAMILY_EVIDENCE"
        )
    ]
    .copy()
)


print("\nFamilies with moderate existing Seed Mass evidence")
print("-" * 100)

print(
    f"N families : "
    f"{len(moderate_family)}"
)


if len(
    moderate_family
) > 0:

    display(
        moderate_family[
            [
                "Family",
                "N_Independent_Species",
                "N_Independent_Genera",
                "Genera_Represented",
                "Median_Seed_Mass_mg",
                "Residual_Genera",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 26. LIMITED / NO FAMILY EVIDENCE
# =============================================================================

weak_family = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].isin(
            [
                "LIMITED_FAMILY_EVIDENCE",
                "NO_FAMILY_EVIDENCE",
            ]
        )
    ]
    .copy()
)


print("\nFamilies requiring cautious or broader fallback treatment")
print("-" * 100)

display(
    weak_family[
        [
            "Family",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Residual_Genera",
            "Family_Evidence_Class",
            "Next_Audit_Action",
        ]
    ]
)


# =============================================================================
# 27. STRONG VALIDATION
# =============================================================================

assert len(
    canonical_family_map
) == 31


assert canonical_family_map[
    "Canonical_Species"
].nunique() == 31


assert canonical_family_map[
    "Family"
].notna().all()


assert len(
    target_genus_family
) == 21


assert family_audit[
    "Family"
].is_unique


assert (
    family_audit[
        "N_Residual_Canonical_Taxa"
    ].sum()
    ==
    31
)


assert (
    family_audit[
        "N_Residual_Genera"
    ].sum()
    >=
    21
)


assert (
    family_audit[
        "Numerical_Family_Value_Assigned"
    ]
    .eq(
        False
    )
    .all()
)


assert (
    family_class_summary[
        "N_Families"
    ].sum()
    ==
    len(
        family_audit
    )
)


# =============================================================================
# 28. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4R = 90

N_FAMILY_VALUES_ASSIGNED_9_4R = 0

N_RESOLVED_AFTER_9_4R = (
    N_RESOLVED_BEFORE_9_4R
    +
    N_FAMILY_VALUES_ASSIGNED_9_4R
)


assert N_RESOLVED_AFTER_9_4R == 90


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4R
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 29. SECTION SUMMARY
# =============================================================================

summary_rows = [

    {
        "Metric":
            "Family-audit canonical taxa",
        "Value":
            len(
                canonical_family_map
            ),
    },

    {
        "Metric":
            "Residual genera entering family audit",
        "Value":
            len(
                target_genus_family
            ),
    },

    {
        "Metric":
            "Unique target families",
        "Value":
            len(
                family_audit
            ),
    },

    {
        "Metric":
            "Normalized independent biological evidence species",
        "Value":
            len(
                species_pool
            ),
    },

    {
        "Metric":
            "Independent evidence species mapped to family",
        "Value":
            len(
                mapped_species_family
            ),
    },

    {
        "Metric":
            "Independent evidence species unmapped to family",
        "Value":
            len(
                unmapped_evidence
            ),
    },

    {
        "Metric":
            "Family Seed Mass values assigned in 9.4R",
        "Value":
            0,
    },

    {
        "Metric":
            "Resolved canonical taxa after 9.4R",
        "Value":
            N_RESOLVED_AFTER_9_4R,
    },

    {
        "Metric":
            "Overall Seed Mass coverage percent",
        "Value":
            coverage_percent,
    },
]


for _, row in family_class_summary.iterrows():

    summary_rows.append(
        {
            "Metric":
                "Families — "
                +
                row[
                    "Family_Evidence_Class"
                ],

            "Value":
                int(
                    row[
                        "N_Families"
                    ]
                ),
        }
    )


section_summary = pd.DataFrame(
    summary_rows
)


print("\nSection 9.4R summary")
print("-" * 100)

display(
    section_summary
)


# =============================================================================
# 30. EXPORT
# =============================================================================

taxonomy_candidate_table.to_csv(
    OUT_DISCOVERED_TAXONOMY_SOURCE,
    index=False
)


canonical_family_map.to_csv(
    OUT_CANONICAL_FAMILY_MAP,
    index=False
)


species_family_diagnostic.to_csv(
    OUT_EVIDENCE_FAMILY_MAPPING,
    index=False
)


family_evidence_detail.to_csv(
    OUT_FAMILY_EVIDENCE_DETAIL,
    index=False
)


family_audit.to_csv(
    OUT_FAMILY_EVIDENCE_AUDIT,
    index=False
)


section_summary.to_csv(
    OUT_FAMILY_EVIDENCE_SUMMARY,
    index=False
)


unmapped_evidence.to_csv(
    OUT_UNMAPPED_EVIDENCE,
    index=False
)


# =============================================================================
# 31. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R exports")
print("-" * 100)

print(
    f"Taxonomy source diagnostic : "
    f"{OUT_DISCOVERED_TAXONOMY_SOURCE}"
)

print(
    f"Canonical family map       : "
    f"{OUT_CANONICAL_FAMILY_MAP}"
)

print(
    f"Evidence family mapping    : "
    f"{OUT_EVIDENCE_FAMILY_MAPPING}"
)

print(
    f"Family evidence detail     : "
    f"{OUT_FAMILY_EVIDENCE_DETAIL}"
)

print(
    f"Family evidence audit      : "
    f"{OUT_FAMILY_EVIDENCE_AUDIT}"
)

print(
    f"Unmapped evidence species  : "
    f"{OUT_UNMAPPED_EVIDENCE}"
)

print(
    f"Audit summary              : "
    f"{OUT_FAMILY_EVIDENCE_SUMMARY}"
)


# =============================================================================
# 32. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Family taxonomic mapping and Seed Mass evidence "
    "audit completed successfully."
)

print(
    f"Family-audit canonical taxa        : "
    f"{len(canonical_family_map)}"
)

print(
    f"Residual genera audited            : "
    f"{len(target_genus_family)}"
)

print(
    f"Unique botanical families audited  : "
    f"{len(family_audit)}"
)

print(
    f"Independent evidence species       : "
    f"{len(species_pool)}"
)

print(
    f"Evidence species mapped to family  : "
    f"{len(mapped_species_family)}"
)

print(
    f"Evidence species unmapped          : "
    f"{len(unmapped_evidence)}"
)

print(
    "Family Seed Mass values assigned  : 0"
)

print(
    f"Overall Seed Mass coverage remains : "
    f"{N_RESOLVED_AFTER_9_4R} / {N_CANONICAL_TAXA} "
    f"({coverage_percent:.2f}%)"
)

print(
    "Family evidence depth is now ready for "
    "family-transfer rule development."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R1
# AUTHORITATIVE FAMILY CROSSWALK FOR RESIDUAL SEED MASS GENERA
# =============================================================================

from pathlib import Path
import pandas as pd


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

FAMILY_TARGET_FILE = (
    OUT_DIR /
    "omo_seed_mass_family_audit_targets_after_single_congener_review.csv"
)

OUT_GENUS_FAMILY_CROSSWALK = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_family_authoritative_crosswalk.csv"
)

OUT_CANONICAL_FAMILY_MAP = (
    OUT_DIR /
    "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)

OUT_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_family_crosswalk_summary.csv"
)


# =============================================================================
# 2. LOAD AUTHORITATIVE 9.4Q FAMILY TARGETS
# =============================================================================

assert FAMILY_TARGET_FILE.exists(), (
    f"Required family target file not found:\n{FAMILY_TARGET_FILE}"
)

family_targets = pd.read_csv(
    FAMILY_TARGET_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R-R1 — AUTHORITATIVE FAMILY CROSSWALK "
    "FOR RESIDUAL SEED MASS GENERA"
)
print("=" * 100)

print("\nInput structure")
print("-" * 100)

print(
    f"Family-audit canonical taxa : {len(family_targets)}"
)

print(
    f"Residual genera             : "
    f"{family_targets['Residual_Genus'].nunique()}"
)


assert len(family_targets) == 31

assert (
    family_targets[
        "Residual_Genus"
    ].nunique()
    ==
    21
)


# =============================================================================
# 3. AUTHORITATIVE GENUS -> FAMILY CROSSWALK
# =============================================================================
#
# Current accepted family placement follows the modern project taxonomy /
# POWO-Kew backbone rather than historical family concepts retained in older
# floristic sources.
#
# Examples:
#   Drypetes     -> Putranjivaceae, not historical Euphorbiaceae
#   Anthocleista -> Gentianaceae, not historical Loganiaceae
#   Scottellia   -> Achariaceae, not historical Flacourtiaceae
#
# =============================================================================

GENUS_FAMILY_MAP = {

    "Amphimas":
        "Fabaceae",

    "Anthocleista":
        "Gentianaceae",

    "Bambusa":
        "Poaceae",

    "Baphia":
        "Fabaceae",

    "Cola":
        "Malvaceae",

    "Diospyros":
        "Ebenaceae",

    "Drypetes":
        "Putranjivaceae",

    "Grewia":
        "Malvaceae",

    "Lonchocarpus":
        "Fabaceae",

    "Macaranga":
        "Euphorbiaceae",

    "Maesobotrya":
        "Phyllanthaceae",

    "Millettia":
        "Fabaceae",

    "Myrianthus":
        "Urticaceae",

    "Nothospondias":
        "Anacardiaceae",

    "Pterocarpus":
        "Fabaceae",

    "Rinorea":
        "Violaceae",

    "Rothmannia":
        "Rubiaceae",

    "Scottellia":
        "Achariaceae",

    "Strombosia":
        "Olacaceae",

    "Trichilia":
        "Meliaceae",

    "Uapaca":
        "Phyllanthaceae",
}


# =============================================================================
# 4. BUILD CROSSWALK TABLE
# =============================================================================

genus_family_crosswalk = pd.DataFrame(
    [
        {
            "Residual_Genus":
                genus,

            "Family":
                family,

            "Family_Mapping_Source":
                "POWO_KEW_MODERN_TAXONOMIC_BACKBONE",

            "Family_Mapping_Status":
                "AUTHORITATIVE_GENUS_LEVEL_MAPPING",
        }

        for genus, family
        in GENUS_FAMILY_MAP.items()
    ]
)


genus_family_crosswalk = (
    genus_family_crosswalk
    .sort_values(
        [
            "Family",
            "Residual_Genus",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 5. VALIDATE AGAINST 9.4Q TARGET GENERA
# =============================================================================

observed_target_genera = set(
    family_targets[
        "Residual_Genus"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


mapped_genera = set(
    genus_family_crosswalk[
        "Residual_Genus"
    ]
)


missing_from_crosswalk = (
    observed_target_genera
    -
    mapped_genera
)


extra_in_crosswalk = (
    mapped_genera
    -
    observed_target_genera
)


print("\nCrosswalk coverage")
print("-" * 100)

print(
    f"Observed residual genera : "
    f"{len(observed_target_genera)}"
)

print(
    f"Mapped genera            : "
    f"{len(mapped_genera)}"
)

print(
    f"Missing genera           : "
    f"{len(missing_from_crosswalk)}"
)

print(
    f"Unexpected extra genera  : "
    f"{len(extra_in_crosswalk)}"
)


if missing_from_crosswalk:

    print(
        "\nMissing genera:"
    )

    for genus in sorted(
        missing_from_crosswalk
    ):
        print(
            f" - {genus}"
        )


assert len(
    missing_from_crosswalk
) == 0, (
    "Some 9.4Q residual genera lack an authoritative family mapping."
)


assert len(
    extra_in_crosswalk
) == 0, (
    "The family crosswalk contains genera not present in the "
    "authoritative 9.4Q family target set."
)


assert len(
    genus_family_crosswalk
) == 21


assert genus_family_crosswalk[
    "Residual_Genus"
].is_unique


assert genus_family_crosswalk[
    "Family"
].notna().all()


# =============================================================================
# 6. DISPLAY AUTHORITATIVE GENUS-FAMILY CROSSWALK
# =============================================================================

print("\nAuthoritative residual genus-to-family mapping")
print("-" * 100)

display(
    genus_family_crosswalk
)


# =============================================================================
# 7. MAP ALL 31 CANONICAL FAMILY-AUDIT TAXA
# =============================================================================

canonical_family_map = (
    family_targets
    .merge(
        genus_family_crosswalk[
            [
                "Residual_Genus",
                "Family",
                "Family_Mapping_Source",
                "Family_Mapping_Status",
            ]
        ],
        on="Residual_Genus",
        how="left",
        validate="many_to_one"
    )
)


# =============================================================================
# 8. VALIDATE COMPLETE CANONICAL COVERAGE
# =============================================================================

assert len(
    canonical_family_map
) == 31


assert canonical_family_map[
    "Canonical_Species"
].nunique() == 31


assert canonical_family_map[
    "Family"
].notna().all(), (
    "At least one family-audit canonical taxon remains unmapped."
)


assert canonical_family_map[
    "Family_Mapping_Status"
].eq(
    "AUTHORITATIVE_GENUS_LEVEL_MAPPING"
).all()


print("\nCanonical family mapping")
print("-" * 100)

display(
    canonical_family_map[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Family",
            "Family_Audit_Entry_Route",
            "Family_Mapping_Status",
        ]
    ]
    .sort_values(
        [
            "Family",
            "Residual_Genus",
            "Canonical_Species",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 9. FAMILY STRUCTURE
# =============================================================================

family_structure = (
    canonical_family_map
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Residual_Genera=(
            "Residual_Genus",
            "nunique"
        ),

        Residual_Genera=(
            "Residual_Genus",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        N_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Canonical_Taxa=(
            "Canonical_Species",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
    .sort_values(
        "Family"
    )
    .reset_index(drop=True)
)


print("\nResidual family structure")
print("-" * 100)

display(
    family_structure
)


# =============================================================================
# 10. STRONG RECONCILIATION
# =============================================================================

assert (
    family_structure[
        "N_Residual_Genera"
    ].sum()
    ==
    21
)


assert (
    family_structure[
        "N_Canonical_Taxa"
    ].sum()
    ==
    31
)


N_FAMILIES = (
    family_structure[
        "Family"
    ].nunique()
)


# =============================================================================
# 11. SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Family-audit canonical taxa",
            "Residual genera",
            "Residual genera mapped to family",
            "Unmapped residual genera",
            "Unique botanical families",
            "Family values assigned",
        ],

        "Value": [
            31,
            21,
            21,
            0,
            N_FAMILIES,
            0,
        ],
    }
)


print("\nSection 9.4R-R1 summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 12. EXPORT
# =============================================================================

genus_family_crosswalk.to_csv(
    OUT_GENUS_FAMILY_CROSSWALK,
    index=False
)


canonical_family_map.to_csv(
    OUT_CANONICAL_FAMILY_MAP,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 13. REPORT
# =============================================================================

print("\nSection 9.4R-R1 exports")
print("-" * 100)

print(
    f"Genus-family crosswalk : "
    f"{OUT_GENUS_FAMILY_CROSSWALK}"
)

print(
    f"Canonical family map   : "
    f"{OUT_CANONICAL_FAMILY_MAP}"
)

print(
    f"Summary                : "
    f"{OUT_SUMMARY}"
)


print("\n" + "=" * 100)

print(
    "Authoritative family mapping completed successfully."
)

print(
    f"Residual genera mapped       : "
    f"{len(genus_family_crosswalk)} / 21"
)

print(
    f"Canonical taxa mapped        : "
    f"{len(canonical_family_map)} / 31"
)

print(
    f"Unique botanical families    : "
    f"{N_FAMILIES}"
)

print(
    "Family Seed Mass values assigned : 0"
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R2
# FAMILY-LEVEL SEED MASS EVIDENCE AUDIT
# USING AUTHORITATIVE FAMILY MAPPING
# =============================================================================
#
# Purpose
# -------
# 1. Load the authoritative 21-genus -> 15-family crosswalk from 9.4R-R1.
# 2. Load the normalized independent biological-species Seed Mass evidence
#    pool from 9.4N-R.
# 3. Map all evidence species to family using controlled genus-family mapping.
# 4. Restrict evidence to the 15 residual target families.
# 5. Quantify:
#       - independent biological species per family
#       - independent genera per family
#       - Seed Mass min / median / mean / max
#       - evidence sources
# 6. Classify family evidence depth:
#
#       STRONG_FAMILY_EVIDENCE    >= 5 independent species
#       MODERATE_FAMILY_EVIDENCE   3–4 independent species
#       LIMITED_FAMILY_EVIDENCE    1–2 independent species
#       NO_FAMILY_EVIDENCE         0 independent species
#
# IMPORTANT
# ---------
# This section is an AUDIT ONLY.
#
# NO family-level Seed Mass value is assigned.
# NO canonical taxon is resolved in 9.4R-R2.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


# -----------------------------------------------------------------------------
# Authoritative 9.4R-R1 inputs
# -----------------------------------------------------------------------------

GENUS_FAMILY_CROSSWALK_FILE = (
    OUT_DIR /
    "omo_seed_mass_residual_genus_family_authoritative_crosswalk.csv"
)

CANONICAL_FAMILY_MAP_FILE = (
    OUT_DIR /
    "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)


# -----------------------------------------------------------------------------
# Authoritative normalized evidence pool from 9.4N-R
# -----------------------------------------------------------------------------

NORMALIZED_SPECIES_POOL_FILE = (
    OUT_DIR /
    "omo_seed_mass_genus_audit_species_evidence_pool_normalized.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_EVIDENCE_GENUS_FAMILY_CROSSWALK = (
    OUT_DIR /
    "omo_seed_mass_evidence_genus_family_crosswalk.csv"
)

OUT_EVIDENCE_SPECIES_FAMILY_MAP = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_species_mapping_authoritative.csv"
)

OUT_FAMILY_EVIDENCE_DETAIL = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_detail_authoritative.csv"
)

OUT_FAMILY_EVIDENCE_AUDIT = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_audit_authoritative.csv"
)

OUT_FAMILY_EVIDENCE_SUMMARY = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_audit_authoritative_summary.csv"
)

OUT_UNMAPPED_EVIDENCE_GENERA = (
    OUT_DIR /
    "omo_seed_mass_family_evidence_unmapped_genera.csv"
)


# =============================================================================
# 2. VALIDATE REQUIRED FILES
# =============================================================================

required_files = [
    GENUS_FAMILY_CROSSWALK_FILE,
    CANONICAL_FAMILY_MAP_FILE,
    NORMALIZED_SPECIES_POOL_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4R-R2 input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

target_genus_family = pd.read_csv(
    GENUS_FAMILY_CROSSWALK_FILE,
    low_memory=False
)

canonical_family_map = pd.read_csv(
    CANONICAL_FAMILY_MAP_FILE,
    low_memory=False
)

species_pool = pd.read_csv(
    NORMALIZED_SPECIES_POOL_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R-R2 — FAMILY-LEVEL SEED MASS EVIDENCE AUDIT "
    "USING AUTHORITATIVE FAMILY MAPPING"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Authoritative residual genera : "
    f"{len(target_genus_family)}"
)

print(
    f"Canonical family-audit taxa   : "
    f"{len(canonical_family_map)}"
)

print(
    f"Independent evidence species  : "
    f"{len(species_pool)}"
)


# =============================================================================
# 4. VALIDATE AUTHORITATIVE 9.4R-R1 STATE
# =============================================================================

assert len(
    target_genus_family
) == 21


assert target_genus_family[
    "Residual_Genus"
].is_unique


assert target_genus_family[
    "Family"
].notna().all()


assert len(
    canonical_family_map
) == 31


assert canonical_family_map[
    "Canonical_Species"
].nunique() == 31


assert canonical_family_map[
    "Family"
].notna().all()


target_family_set = set(
    target_genus_family[
        "Family"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert len(
    target_family_set
) == 15, (
    f"Expected 15 authoritative target families; "
    f"found {len(target_family_set)}."
)


# =============================================================================
# 5. VALIDATE NORMALIZED EVIDENCE POOL
# =============================================================================

required_species_columns = {
    "Biological_Genus",
    "Biological_Species_Name",
    "Seed_Mass_mg",
    "Evidence_Sources",
    "Canonical_Species_Represented",
}


missing_species_columns = (
    required_species_columns
    -
    set(
        species_pool.columns
    )
)


assert not missing_species_columns, (
    "Normalized species evidence pool is missing required fields:\n"
    f"{sorted(missing_species_columns)}"
)


species_pool[
    "Biological_Genus"
] = (
    species_pool[
        "Biological_Genus"
    ]
    .astype("string")
    .str.strip()
)


species_pool[
    "Biological_Species_Name"
] = (
    species_pool[
        "Biological_Species_Name"
    ]
    .astype("string")
    .str.strip()
)


species_pool[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_pool[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert species_pool[
    "Seed_Mass_mg"
].notna().all()


assert species_pool[
    "Seed_Mass_mg"
].gt(
    0
).all()


assert not species_pool[
    "Biological_Species_Name"
].duplicated().any(), (
    "Normalized species evidence pool contains duplicate biological species."
)


# =============================================================================
# 6. AUTHORITATIVE GENUS -> FAMILY MAP FOR ALL EVIDENCE GENERA
# =============================================================================
#
# The 9.4R-R1 crosswalk covers only the 21 residual genera.
#
# The evidence pool contains many other genera whose Seed Mass values may
# contribute legitimate family-level evidence.
#
# Therefore we need a controlled genus-family map for all genera represented
# in the normalized 63-species evidence pool.
#
# The mapping below uses the current project/POWO family concepts already
# established during Notebook 07B harmonization.
#
# =============================================================================

EVIDENCE_GENUS_FAMILY_MAP = {

    # -------------------------------------------------------------------------
    # Achariaceae
    # -------------------------------------------------------------------------
    "Scottellia": "Achariaceae",

    # -------------------------------------------------------------------------
    # Anacardiaceae
    # -------------------------------------------------------------------------
    "Nothospondias": "Anacardiaceae",
    "Spondias": "Anacardiaceae",

    # -------------------------------------------------------------------------
    # Annonaceae
    # -------------------------------------------------------------------------
    "Cleistopholis": "Annonaceae",
    "Hexalobus": "Annonaceae",

    # -------------------------------------------------------------------------
    # Apocynaceae
    # -------------------------------------------------------------------------
    "Alstonia": "Apocynaceae",
    "Picralima": "Apocynaceae",

    # -------------------------------------------------------------------------
    # Arecaceae
    # -------------------------------------------------------------------------
    "Elaeis": "Arecaceae",

    # -------------------------------------------------------------------------
    # Caricaceae
    # -------------------------------------------------------------------------
    "Carica": "Caricaceae",

    # -------------------------------------------------------------------------
    # Ebenaceae
    # -------------------------------------------------------------------------
    "Diospyros": "Ebenaceae",

    # -------------------------------------------------------------------------
    # Euphorbiaceae
    # -------------------------------------------------------------------------
    "Macaranga": "Euphorbiaceae",
    "Ricinodendron": "Euphorbiaceae",

    # -------------------------------------------------------------------------
    # Fabaceae
    # -------------------------------------------------------------------------
    "Albizia": "Fabaceae",
    "Amphimas": "Fabaceae",
    "Baphia": "Fabaceae",
    "Brachystegia": "Fabaceae",
    "Cassia": "Fabaceae",
    "Lonchocarpus": "Fabaceae",
    "Millettia": "Fabaceae",
    "Pterocarpus": "Fabaceae",

    # -------------------------------------------------------------------------
    # Gentianaceae
    # -------------------------------------------------------------------------
    "Anthocleista": "Gentianaceae",

    # -------------------------------------------------------------------------
    # Hypericaceae
    # -------------------------------------------------------------------------
    "Harungana": "Hypericaceae",

    # -------------------------------------------------------------------------
    # Malvaceae
    # -------------------------------------------------------------------------
    "Ceiba": "Malvaceae",
    "Cola": "Malvaceae",
    "Grewia": "Malvaceae",
    "Nesogordonia": "Malvaceae",
    "Sterculia": "Malvaceae",
    "Theobroma": "Malvaceae",

    # -------------------------------------------------------------------------
    # Meliaceae
    # -------------------------------------------------------------------------
    "Trichilia": "Meliaceae",

    # -------------------------------------------------------------------------
    # Moraceae
    # -------------------------------------------------------------------------
    "Artocarpus": "Moraceae",

    # -------------------------------------------------------------------------
    # Musaceae
    # -------------------------------------------------------------------------
    "Musa": "Musaceae",

    # -------------------------------------------------------------------------
    # Olacaceae
    # -------------------------------------------------------------------------
    "Strombosia": "Olacaceae",

    # -------------------------------------------------------------------------
    # Phyllanthaceae
    # -------------------------------------------------------------------------
    "Maesobotrya": "Phyllanthaceae",
    "Uapaca": "Phyllanthaceae",

    # -------------------------------------------------------------------------
    # Poaceae
    # -------------------------------------------------------------------------
    "Bambusa": "Poaceae",

    # -------------------------------------------------------------------------
    # Putranjivaceae
    # -------------------------------------------------------------------------
    "Drypetes": "Putranjivaceae",

    # -------------------------------------------------------------------------
    # Rubiaceae
    # -------------------------------------------------------------------------
    "Nauclea": "Rubiaceae",
    "Rothmannia": "Rubiaceae",

    # -------------------------------------------------------------------------
    # Sapindaceae
    # -------------------------------------------------------------------------
    "Blighia": "Sapindaceae",

    # -------------------------------------------------------------------------
    # Ulmaceae / Cannabaceae modern placement
    # -------------------------------------------------------------------------
    "Celtis": "Cannabaceae",
    "Trema": "Cannabaceae",

    # -------------------------------------------------------------------------
    # Urticaceae
    # -------------------------------------------------------------------------
    "Myrianthus": "Urticaceae",

    # -------------------------------------------------------------------------
    # Violaceae
    # -------------------------------------------------------------------------
    "Rinorea": "Violaceae",

    # -------------------------------------------------------------------------
    # Lamiaceae
    # -------------------------------------------------------------------------
    "Gmelina": "Lamiaceae",

    # -------------------------------------------------------------------------
    # Dilleniaceae
    # -------------------------------------------------------------------------
    "Dillenia": "Dilleniaceae",

    # -------------------------------------------------------------------------
    # Irvingiaceae
    # -------------------------------------------------------------------------
    "Irvingia": "Irvingiaceae",

    # -------------------------------------------------------------------------
    # Rutaceae
    # -------------------------------------------------------------------------
    "Citrus": "Rutaceae",
}


# =============================================================================
# 7. BUILD EVIDENCE GENUS-FAMILY CROSSWALK
# =============================================================================

evidence_genera = sorted(
    species_pool[
        "Biological_Genus"
    ]
    .dropna()
    .astype(str)
    .unique()
)


evidence_genus_family_crosswalk = pd.DataFrame(
    {
        "Biological_Genus":
            evidence_genera
    }
)


evidence_genus_family_crosswalk[
    "Family"
] = (
    evidence_genus_family_crosswalk[
        "Biological_Genus"
    ]
    .map(
        EVIDENCE_GENUS_FAMILY_MAP
    )
)


evidence_genus_family_crosswalk[
    "Family_Mapping_Status"
] = np.where(
    evidence_genus_family_crosswalk[
        "Family"
    ].notna(),
    "CONTROLLED_GENUS_FAMILY_MAPPING",
    "UNMAPPED_GENUS"
)


# =============================================================================
# 8. DIAGNOSE UNMAPPED EVIDENCE GENERA
# =============================================================================

unmapped_evidence_genera = (
    evidence_genus_family_crosswalk[
        evidence_genus_family_crosswalk[
            "Family"
        ].isna()
    ]
    .copy()
)


print("\nEvidence genus-family mapping")
print("-" * 100)

print(
    f"Evidence genera represented : "
    f"{len(evidence_genus_family_crosswalk)}"
)

print(
    f"Evidence genera mapped      : "
    f"{evidence_genus_family_crosswalk['Family'].notna().sum()}"
)

print(
    f"Evidence genera unmapped    : "
    f"{len(unmapped_evidence_genera)}"
)


if len(
    unmapped_evidence_genera
) > 0:

    display(
        unmapped_evidence_genera
    )


# =============================================================================
# 9. MAP INDEPENDENT EVIDENCE SPECIES TO FAMILY
# =============================================================================

evidence_species_family = (
    species_pool
    .merge(
        evidence_genus_family_crosswalk[
            [
                "Biological_Genus",
                "Family",
                "Family_Mapping_Status",
            ]
        ],
        on="Biological_Genus",
        how="left",
        validate="many_to_one"
    )
)


# =============================================================================
# 10. EVIDENCE SPECIES FAMILY-MAPPING DIAGNOSTIC
# =============================================================================

unmapped_species = (
    evidence_species_family[
        evidence_species_family[
            "Family"
        ].isna()
    ]
    .copy()
)


print("\nIndependent evidence species lacking family mapping")
print("-" * 100)

print(
    f"N species : "
    f"{len(unmapped_species)}"
)


if len(
    unmapped_species
) > 0:

    display(
        unmapped_species[
            [
                "Biological_Genus",
                "Biological_Species_Name",
                "Seed_Mass_mg",
                "Evidence_Sources",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 11. RESTRICT TO 15 TARGET FAMILIES
# =============================================================================

family_evidence_detail = (
    evidence_species_family[
        evidence_species_family[
            "Family"
        ].isin(
            target_family_set
        )
    ]
    .copy()
    .sort_values(
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nIndependent Seed Mass evidence within the 15 target families")
print("-" * 100)

display(
    family_evidence_detail[
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
            "Seed_Mass_mg",
            "Evidence_Sources",
            "Canonical_Species_Represented",
        ]
    ]
)


# =============================================================================
# 12. TARGET FAMILY STRUCTURE
# =============================================================================

def join_unique(series):

    values = []

    for value in series.dropna():

        text = str(value).strip()

        if not text:
            continue

        parts = [
            x.strip()
            for x in text.split("|")
            if x.strip()
        ]

        values.extend(
            parts
        )

    return " | ".join(
        sorted(
            set(values)
        )
    )


required_family_targets = (
    canonical_family_map
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Residual_Genera=(
            "Residual_Genus",
            "nunique"
        ),

        Residual_Genera=(
            "Residual_Genus",
            join_unique
        ),

        N_Residual_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Residual_Canonical_Taxa=(
            "Canonical_Species",
            join_unique
        ),

        Residual_Biological_Targets=(
            "Final_Search_Name",
            join_unique
        ),

        Family_Audit_Entry_Routes=(
            "Family_Audit_Entry_Route",
            join_unique
        ),
    )
)


# =============================================================================
# 13. FAMILY EVIDENCE SUMMARY
# =============================================================================

family_evidence_existing = (
    family_evidence_detail
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        N_Independent_Genera=(
            "Biological_Genus",
            "nunique"
        ),

        Species_Represented=(
            "Biological_Species_Name",
            join_unique
        ),

        Genera_Represented=(
            "Biological_Genus",
            join_unique
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


# =============================================================================
# 14. MERGE ZERO-EVIDENCE FAMILIES BACK
# =============================================================================

family_audit = (
    required_family_targets
    .merge(
        family_evidence_existing,
        on="Family",
        how="left",
        validate="one_to_one"
    )
)


family_audit[
    "N_Independent_Species"
] = (
    family_audit[
        "N_Independent_Species"
    ]
    .fillna(0)
    .astype(int)
)


family_audit[
    "N_Independent_Genera"
] = (
    family_audit[
        "N_Independent_Genera"
    ]
    .fillna(0)
    .astype(int)
)


for col in [
    "Species_Represented",
    "Genera_Represented",
    "Evidence_Sources",
]:

    family_audit[
        col
    ] = (
        family_audit[
            col
        ]
        .fillna("")
    )


# =============================================================================
# 15. CLASSIFY FAMILY EVIDENCE DEPTH
# =============================================================================

def classify_family_evidence(
    n_species
):

    if n_species >= 5:

        return (
            "STRONG_FAMILY_EVIDENCE"
        )

    if n_species >= 3:

        return (
            "MODERATE_FAMILY_EVIDENCE"
        )

    if n_species >= 1:

        return (
            "LIMITED_FAMILY_EVIDENCE"
        )

    return (
        "NO_FAMILY_EVIDENCE"
    )


family_audit[
    "Family_Evidence_Class"
] = (
    family_audit[
        "N_Independent_Species"
    ]
    .map(
        classify_family_evidence
    )
)


# =============================================================================
# 16. NEXT AUDIT ACTION
# =============================================================================

def determine_family_next_action(
    row
):

    n_species = (
        row[
            "N_Independent_Species"
        ]
    )

    n_genera = (
        row[
            "N_Independent_Genera"
        ]
    )


    if (
        n_species >= 5
        and
        n_genera >= 2
    ):

        return (
            "CANDIDATE_FOR_FAMILY_ESTIMATE_RULE_DEVELOPMENT"
        )


    if n_species >= 3:

        return (
            "REVIEW_FAMILY_ESTIMATE_FEASIBILITY"
        )


    if n_species >= 1:

        return (
            "LIMITED_FAMILY_EVIDENCE_REVIEW_TRANSFER_RISK"
        )


    return (
        "NO_FAMILY_EVIDENCE_REQUIRE_BROADER_FALLBACK"
    )


family_audit[
    "Next_Audit_Action"
] = (
    family_audit
    .apply(
        determine_family_next_action,
        axis=1
    )
)


family_audit[
    "Numerical_Family_Value_Assigned"
] = False


# =============================================================================
# 17. DISPLAY AUTHORITATIVE FAMILY AUDIT
# =============================================================================

print("\nAuthoritative residual family-level Seed Mass evidence audit")
print("-" * 100)

display(
    family_audit[
        [
            "Family",
            "N_Residual_Genera",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Genera_Represented",
            "Evidence_Sources",
            "Min_Seed_Mass_mg",
            "Median_Seed_Mass_mg",
            "Max_Seed_Mass_mg",
            "Family_Evidence_Class",
            "Next_Audit_Action",
        ]
    ]
    .sort_values(
        [
            "Family_Evidence_Class",
            "Family",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 18. EVIDENCE-CLASS SUMMARY
# =============================================================================

family_class_summary = (
    family_audit[
        "Family_Evidence_Class"
    ]
    .value_counts()
    .rename_axis(
        "Family_Evidence_Class"
    )
    .reset_index(
        name="N_Families"
    )
)


family_class_summary[
    "Percent_of_15_Families"
] = (
    family_class_summary[
        "N_Families"
    ]
    /
    15
    *
    100
)


print("\nFamily evidence-depth summary")
print("-" * 100)

display(
    family_class_summary
)


# =============================================================================
# 19. STRONG FAMILY EVIDENCE
# =============================================================================

strong_family = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].eq(
            "STRONG_FAMILY_EVIDENCE"
        )
    ]
    .copy()
)


print("\nFamilies with strong Seed Mass evidence")
print("-" * 100)

print(
    f"N families : "
    f"{len(strong_family)}"
)


if len(
    strong_family
) > 0:

    display(
        strong_family[
            [
                "Family",
                "N_Independent_Species",
                "N_Independent_Genera",
                "Genera_Represented",
                "Median_Seed_Mass_mg",
                "Residual_Genera",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 20. MODERATE FAMILY EVIDENCE
# =============================================================================

moderate_family = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].eq(
            "MODERATE_FAMILY_EVIDENCE"
        )
    ]
    .copy()
)


print("\nFamilies with moderate Seed Mass evidence")
print("-" * 100)

print(
    f"N families : "
    f"{len(moderate_family)}"
)


if len(
    moderate_family
) > 0:

    display(
        moderate_family[
            [
                "Family",
                "N_Independent_Species",
                "N_Independent_Genera",
                "Genera_Represented",
                "Median_Seed_Mass_mg",
                "Residual_Genera",
                "Next_Audit_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 21. LIMITED / NO FAMILY EVIDENCE
# =============================================================================

weak_family = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].isin(
            [
                "LIMITED_FAMILY_EVIDENCE",
                "NO_FAMILY_EVIDENCE",
            ]
        )
    ]
    .copy()
)


print("\nFamilies requiring cautious or broader fallback treatment")
print("-" * 100)

display(
    weak_family[
        [
            "Family",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Genera_Represented",
            "Residual_Genera",
            "Family_Evidence_Class",
            "Next_Audit_Action",
        ]
    ]
)


# =============================================================================
# 22. VALIDATE FAMILY COUNTS
# =============================================================================

assert len(
    family_audit
) == 15


assert family_audit[
    "Family"
].is_unique


assert (
    family_audit[
        "N_Residual_Canonical_Taxa"
    ].sum()
    ==
    31
)


assert (
    canonical_family_map[
        "Residual_Genus"
    ].nunique()
    ==
    21
)


assert (
    family_class_summary[
        "N_Families"
    ].sum()
    ==
    15
)


assert (
    family_audit[
        "Numerical_Family_Value_Assigned"
    ]
    .eq(False)
    .all()
)


# =============================================================================
# 23. VALIDATE TARGET FAMILY COVERAGE
# =============================================================================

assert set(
    family_audit[
        "Family"
    ]
) == target_family_set


# =============================================================================
# 24. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4R_R2 = 90

N_FAMILY_VALUES_ASSIGNED = 0

N_RESOLVED_AFTER_9_4R_R2 = (
    N_RESOLVED_BEFORE_9_4R_R2
    +
    N_FAMILY_VALUES_ASSIGNED
)


assert N_RESOLVED_AFTER_9_4R_R2 == 90


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4R_R2
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 25. SECTION SUMMARY
# =============================================================================

summary_rows = [

    {
        "Metric":
            "Family-audit canonical taxa",
        "Value":
            31,
    },

    {
        "Metric":
            "Residual genera",
        "Value":
            21,
    },

    {
        "Metric":
            "Unique target families",
        "Value":
            15,
    },

    {
        "Metric":
            "Independent biological evidence species",
        "Value":
            len(
                species_pool
            ),
    },

    {
        "Metric":
            "Evidence genera represented",
        "Value":
            len(
                evidence_genus_family_crosswalk
            ),
    },

    {
        "Metric":
            "Evidence genera unmapped to family",
        "Value":
            len(
                unmapped_evidence_genera
            ),
    },

    {
        "Metric":
            "Evidence species within target families",
        "Value":
            len(
                family_evidence_detail
            ),
    },

    {
        "Metric":
            "Family Seed Mass values assigned in 9.4R-R2",
        "Value":
            0,
    },

    {
        "Metric":
            "Resolved canonical taxa after 9.4R-R2",
        "Value":
            N_RESOLVED_AFTER_9_4R_R2,
    },

    {
        "Metric":
            "Overall Seed Mass coverage percent",
        "Value":
            coverage_percent,
    },
]


for _, row in family_class_summary.iterrows():

    summary_rows.append(
        {
            "Metric":
                "Families — "
                +
                row[
                    "Family_Evidence_Class"
                ],

            "Value":
                int(
                    row[
                        "N_Families"
                    ]
                ),
        }
    )


section_summary = pd.DataFrame(
    summary_rows
)


print("\nSection 9.4R-R2 summary")
print("-" * 100)

display(
    section_summary
)


# =============================================================================
# 26. EXPORT
# =============================================================================

evidence_genus_family_crosswalk.to_csv(
    OUT_EVIDENCE_GENUS_FAMILY_CROSSWALK,
    index=False
)


evidence_species_family.to_csv(
    OUT_EVIDENCE_SPECIES_FAMILY_MAP,
    index=False
)


family_evidence_detail.to_csv(
    OUT_FAMILY_EVIDENCE_DETAIL,
    index=False
)


family_audit.to_csv(
    OUT_FAMILY_EVIDENCE_AUDIT,
    index=False
)


section_summary.to_csv(
    OUT_FAMILY_EVIDENCE_SUMMARY,
    index=False
)


unmapped_evidence_genera.to_csv(
    OUT_UNMAPPED_EVIDENCE_GENERA,
    index=False
)


# =============================================================================
# 27. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R-R2 exports")
print("-" * 100)

print(
    f"Evidence genus-family map : "
    f"{OUT_EVIDENCE_GENUS_FAMILY_CROSSWALK}"
)

print(
    f"Evidence species-family   : "
    f"{OUT_EVIDENCE_SPECIES_FAMILY_MAP}"
)

print(
    f"Family evidence detail    : "
    f"{OUT_FAMILY_EVIDENCE_DETAIL}"
)

print(
    f"Family evidence audit     : "
    f"{OUT_FAMILY_EVIDENCE_AUDIT}"
)

print(
    f"Unmapped evidence genera  : "
    f"{OUT_UNMAPPED_EVIDENCE_GENERA}"
)

print(
    f"Audit summary             : "
    f"{OUT_FAMILY_EVIDENCE_SUMMARY}"
)


# =============================================================================
# 28. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Authoritative family-level Seed Mass evidence "
    "audit completed successfully."
)

print(
    f"Family-audit canonical taxa        : "
    f"{len(canonical_family_map)}"
)

print(
    f"Residual genera                    : "
    f"{canonical_family_map['Residual_Genus'].nunique()}"
)

print(
    f"Target botanical families          : "
    f"{len(family_audit)}"
)

print(
    f"Independent evidence species       : "
    f"{len(species_pool)}"
)

print(
    f"Evidence species in target families: "
    f"{len(family_evidence_detail)}"
)

print(
    f"Evidence genera unmapped           : "
    f"{len(unmapped_evidence_genera)}"
)

print(
    "Family Seed Mass values assigned  : 0"
)

print(
    f"Overall Seed Mass coverage remains : "
    f"{N_RESOLVED_AFTER_9_4R_R2} / {N_CANONICAL_TAXA} "
    f"({coverage_percent:.2f}%)"
)

print(
    "Family evidence depth is now ready for "
    "family-transfer rule development."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R2A
# COMPLETE AUTHORITATIVE FAMILY MAPPING OF ALL SEED MASS EVIDENCE GENERA
# =============================================================================
#
# Purpose
# -------
# Section 9.4R-R2 showed:
#
#     Independent evidence species : 63
#     Evidence genera represented  : 58
#     Evidence genera mapped       : 31
#     Evidence genera unmapped     : 27
#
# The incomplete evidence-genus family map artificially reduced the amount
# of usable family-level evidence.
#
# This section:
#
#   1. loads the normalized 63-species Seed Mass evidence pool;
#   2. reconstructs the full set of 58 biological genera;
#   3. applies an authoritative modern genus -> family crosswalk;
#   4. explicitly resolves the 27 genera previously left unmapped;
#   5. validates 58 / 58 genus-family coverage;
#   6. maps all 63 independent biological species to family;
#   7. identifies which species belong to the 15 residual target families;
#   8. exports the completed family mapping for Section 9.4R-R2B.
#
# IMPORTANT
# ---------
# NO family Seed Mass estimates are calculated here.
# NO family Seed Mass values are assigned.
# NO canonical taxon is resolved in this section.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

NORMALIZED_SPECIES_POOL_FILE = (
    OUT_DIR
    / "omo_seed_mass_genus_audit_species_evidence_pool_normalized.csv"
)

RESIDUAL_GENUS_FAMILY_FILE = (
    OUT_DIR
    / "omo_seed_mass_residual_genus_family_authoritative_crosswalk.csv"
)

CANONICAL_FAMILY_MAP_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)

PREVIOUS_EVIDENCE_GENUS_MAP_FILE = (
    OUT_DIR
    / "omo_seed_mass_evidence_genus_family_crosswalk.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_COMPLETE_GENUS_FAMILY_MAP = (
    OUT_DIR
    / "omo_seed_mass_evidence_genus_family_crosswalk_complete.csv"
)

OUT_COMPLETE_SPECIES_FAMILY_MAP = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_family_mapping_complete.csv"
)

OUT_PREVIOUSLY_UNMAPPED_RESOLUTION = (
    OUT_DIR
    / "omo_seed_mass_previously_unmapped_evidence_genera_family_resolution.csv"
)

OUT_TARGET_FAMILY_EVIDENCE_SPECIES = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_in_target_families_complete.csv"
)

OUT_NON_TARGET_FAMILY_EVIDENCE_SPECIES = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_outside_target_families.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_complete_evidence_genus_family_mapping_summary.csv"
)


# =============================================================================
# 2. VALIDATE REQUIRED INPUTS
# =============================================================================

required_files = [
    NORMALIZED_SPECIES_POOL_FILE,
    RESIDUAL_GENUS_FAMILY_FILE,
    CANONICAL_FAMILY_MAP_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4R-R2A input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

species_pool = pd.read_csv(
    NORMALIZED_SPECIES_POOL_FILE,
    low_memory=False
)

residual_genus_family = pd.read_csv(
    RESIDUAL_GENUS_FAMILY_FILE,
    low_memory=False
)

canonical_family_map = pd.read_csv(
    CANONICAL_FAMILY_MAP_FILE,
    low_memory=False
)


if PREVIOUS_EVIDENCE_GENUS_MAP_FILE.exists():

    previous_genus_map = pd.read_csv(
        PREVIOUS_EVIDENCE_GENUS_MAP_FILE,
        low_memory=False
    )

else:

    previous_genus_map = pd.DataFrame()


print("=" * 100)
print(
    "SECTION 9.4R-R2A — COMPLETE AUTHORITATIVE FAMILY MAPPING "
    "OF ALL SEED MASS EVIDENCE GENERA"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Independent biological evidence species : "
    f"{len(species_pool)}"
)

print(
    f"Residual genus-family crosswalk rows     : "
    f"{len(residual_genus_family)}"
)

print(
    f"Canonical family-audit taxa              : "
    f"{len(canonical_family_map)}"
)


# =============================================================================
# 4. VALIDATE NORMALIZED EVIDENCE POOL
# =============================================================================

required_species_columns = {
    "Biological_Genus",
    "Biological_Species_Name",
    "Seed_Mass_mg",
    "Evidence_Sources",
    "Canonical_Species_Represented",
}


missing_species_columns = (
    required_species_columns
    -
    set(
        species_pool.columns
    )
)


assert not missing_species_columns, (
    "Normalized evidence pool is missing required fields:\n"
    f"{sorted(missing_species_columns)}"
)


species_pool[
    "Biological_Genus"
] = (
    species_pool[
        "Biological_Genus"
    ]
    .astype("string")
    .str.strip()
)


species_pool[
    "Biological_Species_Name"
] = (
    species_pool[
        "Biological_Species_Name"
    ]
    .astype("string")
    .str.strip()
)


species_pool[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_pool[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert len(
    species_pool
) == 63


assert species_pool[
    "Seed_Mass_mg"
].notna().all()


assert species_pool[
    "Seed_Mass_mg"
].gt(
    0
).all()


assert not species_pool[
    "Biological_Species_Name"
].duplicated().any()


# =============================================================================
# 5. AUTHORITATIVE COMPLETE GENUS -> FAMILY CROSSWALK
# =============================================================================
#
# Modern family concepts are used consistently with 9.4R-R1.
#
# This map covers ALL genera currently represented in the normalized
# 63-species Seed Mass evidence pool.
#
# =============================================================================

COMPLETE_EVIDENCE_GENUS_FAMILY_MAP = {

    # -------------------------------------------------------------------------
    # Achariaceae
    # -------------------------------------------------------------------------
    "Scottellia":
        "Achariaceae",

    # -------------------------------------------------------------------------
    # Anacardiaceae
    # -------------------------------------------------------------------------
    "Nothospondias":
        "Anacardiaceae",

    "Spondias":
        "Anacardiaceae",

    # -------------------------------------------------------------------------
    # Annonaceae
    # -------------------------------------------------------------------------
    "Anonidium":
        "Annonaceae",

    "Cleistopholis":
        "Annonaceae",

    "Hexalobus":
        "Annonaceae",

    "Monodora":
        "Annonaceae",

    # -------------------------------------------------------------------------
    # Apocynaceae
    # -------------------------------------------------------------------------
    "Alstonia":
        "Apocynaceae",

    "Funtumia":
        "Apocynaceae",

    "Holarrhena":
        "Apocynaceae",

    "Picralima":
        "Apocynaceae",

    "Rauvolfia":
        "Apocynaceae",

    # -------------------------------------------------------------------------
    # Arecaceae
    # -------------------------------------------------------------------------
    "Elaeis":
        "Arecaceae",

    # -------------------------------------------------------------------------
    # Asparagaceae
    # -------------------------------------------------------------------------
    "Dracaena":
        "Asparagaceae",

    # -------------------------------------------------------------------------
    # Cannabaceae
    # -------------------------------------------------------------------------
    "Celtis":
        "Cannabaceae",

    "Trema":
        "Cannabaceae",

    # -------------------------------------------------------------------------
    # Caricaceae
    # -------------------------------------------------------------------------
    "Carica":
        "Caricaceae",

    # -------------------------------------------------------------------------
    # Clusiaceae
    # -------------------------------------------------------------------------
    "Garcinia":
        "Clusiaceae",

    # -------------------------------------------------------------------------
    # Combretaceae
    # -------------------------------------------------------------------------
    "Terminalia":
        "Combretaceae",

    # -------------------------------------------------------------------------
    # Dilleniaceae
    # -------------------------------------------------------------------------
    "Dillenia":
        "Dilleniaceae",

    # -------------------------------------------------------------------------
    # Ebenaceae
    # -------------------------------------------------------------------------
    "Diospyros":
        "Ebenaceae",

    # -------------------------------------------------------------------------
    # Euphorbiaceae
    # -------------------------------------------------------------------------
    "Macaranga":
        "Euphorbiaceae",

    "Ricinodendron":
        "Euphorbiaceae",

    # -------------------------------------------------------------------------
    # Fabaceae
    # -------------------------------------------------------------------------
    "Afzelia":
        "Fabaceae",

    "Albizia":
        "Fabaceae",

    "Amphimas":
        "Fabaceae",

    "Anthonotha":
        "Fabaceae",

    "Baphia":
        "Fabaceae",

    "Brachystegia":
        "Fabaceae",

    "Cassia":
        "Fabaceae",

    "Dialium":
        "Fabaceae",

    "Lonchocarpus":
        "Fabaceae",

    "Millettia":
        "Fabaceae",

    "Philenoptera":
        "Fabaceae",

    "Piptadeniastrum":
        "Fabaceae",

    "Pterocarpus":
        "Fabaceae",

    "Tetrapleura":
        "Fabaceae",

    # -------------------------------------------------------------------------
    # Gentianaceae
    # -------------------------------------------------------------------------
    "Anthocleista":
        "Gentianaceae",

    # -------------------------------------------------------------------------
    # Hypericaceae
    # -------------------------------------------------------------------------
    "Harungana":
        "Hypericaceae",

    # -------------------------------------------------------------------------
    # Irvingiaceae
    # -------------------------------------------------------------------------
    "Irvingia":
        "Irvingiaceae",

    # -------------------------------------------------------------------------
    # Lamiaceae
    # -------------------------------------------------------------------------
    "Gmelina":
        "Lamiaceae",

    # -------------------------------------------------------------------------
    # Malvaceae
    # -------------------------------------------------------------------------
    "Bombax":
        "Malvaceae",

    "Ceiba":
        "Malvaceae",

    "Cola":
        "Malvaceae",

    "Grewia":
        "Malvaceae",

    "Hildegardia":
        "Malvaceae",

    "Nesogordonia":
        "Malvaceae",

    "Sterculia":
        "Malvaceae",

    "Theobroma":
        "Malvaceae",

    # -------------------------------------------------------------------------
    # Meliaceae
    # -------------------------------------------------------------------------
    "Trichilia":
        "Meliaceae",

    # -------------------------------------------------------------------------
    # Moraceae
    # -------------------------------------------------------------------------
    "Artocarpus":
        "Moraceae",

    "Ficus":
        "Moraceae",

    "Milicia":
        "Moraceae",

    "Trilepisium":
        "Moraceae",

    # -------------------------------------------------------------------------
    # Musaceae
    # -------------------------------------------------------------------------
    "Musa":
        "Musaceae",

    # -------------------------------------------------------------------------
    # Myristicaceae
    # -------------------------------------------------------------------------
    "Pycnanthus":
        "Myristicaceae",

    # -------------------------------------------------------------------------
    # Olacaceae
    # -------------------------------------------------------------------------
    "Strombosia":
        "Olacaceae",

    # -------------------------------------------------------------------------
    # Phyllanthaceae
    # -------------------------------------------------------------------------
    "Bridelia":
        "Phyllanthaceae",

    "Maesobotrya":
        "Phyllanthaceae",

    "Margaritaria":
        "Phyllanthaceae",

    "Uapaca":
        "Phyllanthaceae",

    # -------------------------------------------------------------------------
    # Poaceae
    # -------------------------------------------------------------------------
    "Bambusa":
        "Poaceae",

    # -------------------------------------------------------------------------
    # Putranjivaceae
    # -------------------------------------------------------------------------
    "Drypetes":
        "Putranjivaceae",

    # -------------------------------------------------------------------------
    # Rubiaceae
    # -------------------------------------------------------------------------
    "Nauclea":
        "Rubiaceae",

    "Pavetta":
        "Rubiaceae",

    "Rothmannia":
        "Rubiaceae",

    # -------------------------------------------------------------------------
    # Rutaceae
    # -------------------------------------------------------------------------
    "Citrus":
        "Rutaceae",

    "Zanthoxylum":
        "Rutaceae",

    # -------------------------------------------------------------------------
    # Sapindaceae
    # -------------------------------------------------------------------------
    "Blighia":
        "Sapindaceae",

    "Lecaniodiscus":
        "Sapindaceae",

    # -------------------------------------------------------------------------
    # Sapotaceae
    # -------------------------------------------------------------------------
    "Chrysophyllum":
        "Sapotaceae",

    # -------------------------------------------------------------------------
    # Urticaceae
    # -------------------------------------------------------------------------
    "Musanga":
        "Urticaceae",

    "Myrianthus":
        "Urticaceae",

    # -------------------------------------------------------------------------
    # Violaceae
    # -------------------------------------------------------------------------
    "Rinorea":
        "Violaceae",
}


# =============================================================================
# 6. BUILD COMPLETE GENUS REGISTER FROM ACTUAL EVIDENCE POOL
# =============================================================================

evidence_genera = sorted(
    species_pool[
        "Biological_Genus"
    ]
    .dropna()
    .astype(str)
    .unique()
)


assert len(
    evidence_genera
) == 58, (
    f"Expected 58 evidence genera; "
    f"found {len(evidence_genera)}."
)


complete_genus_family = pd.DataFrame(
    {
        "Biological_Genus":
            evidence_genera
    }
)


complete_genus_family[
    "Family"
] = (
    complete_genus_family[
        "Biological_Genus"
    ]
    .map(
        COMPLETE_EVIDENCE_GENUS_FAMILY_MAP
    )
)


# =============================================================================
# 7. IDENTIFY WHETHER GENUS WAS PREVIOUSLY MAPPED OR UNMAPPED
# =============================================================================

previously_mapped_genera = set()


if len(
    previous_genus_map
) > 0:

    if {
        "Biological_Genus",
        "Family",
    }.issubset(
        previous_genus_map.columns
    ):

        previously_mapped_genera = set(
            previous_genus_map.loc[
                previous_genus_map[
                    "Family"
                ].notna(),
                "Biological_Genus"
            ]
            .astype(str)
            .str.strip()
        )


complete_genus_family[
    "Previous_9_4R_R2_Status"
] = np.where(
    complete_genus_family[
        "Biological_Genus"
    ].isin(
        previously_mapped_genera
    ),
    "PREVIOUSLY_MAPPED",
    "PREVIOUSLY_UNMAPPED"
)


complete_genus_family[
    "Family_Mapping_Source"
] = (
    "AUTHORITATIVE_MODERN_GENUS_FAMILY_CROSSWALK"
)


complete_genus_family[
    "Family_Mapping_Status"
] = np.where(
    complete_genus_family[
        "Family"
    ].notna(),
    "AUTHORITATIVE_FAMILY_MAPPING",
    "UNMAPPED_AFTER_R2A"
)


# =============================================================================
# 8. DIAGNOSE REMAINING UNMAPPED GENERA
# =============================================================================

still_unmapped = (
    complete_genus_family[
        complete_genus_family[
            "Family"
        ].isna()
    ]
    .copy()
)


print("\nComplete evidence genus-family mapping")
print("-" * 100)

print(
    f"Evidence genera represented : "
    f"{len(complete_genus_family)}"
)

print(
    f"Evidence genera mapped      : "
    f"{complete_genus_family['Family'].notna().sum()}"
)

print(
    f"Evidence genera unmapped    : "
    f"{len(still_unmapped)}"
)


if len(
    still_unmapped
) > 0:

    display(
        still_unmapped
    )


assert len(
    still_unmapped
) == 0, (
    "One or more evidence genera remain unmapped after "
    "the authoritative 9.4R-R2A correction."
)


# =============================================================================
# 9. VALIDATE THE 27 PREVIOUSLY UNMAPPED GENERA
# =============================================================================

EXPECTED_PREVIOUSLY_UNMAPPED = {
    "Afzelia",
    "Anonidium",
    "Anthonotha",
    "Bombax",
    "Bridelia",
    "Chrysophyllum",
    "Dialium",
    "Dracaena",
    "Ficus",
    "Funtumia",
    "Garcinia",
    "Hildegardia",
    "Holarrhena",
    "Lecaniodiscus",
    "Margaritaria",
    "Milicia",
    "Monodora",
    "Musanga",
    "Pavetta",
    "Philenoptera",
    "Piptadeniastrum",
    "Pycnanthus",
    "Rauvolfia",
    "Terminalia",
    "Tetrapleura",
    "Trilepisium",
    "Zanthoxylum",
}


observed_previously_unmapped = set(
    complete_genus_family.loc[
        complete_genus_family[
            "Previous_9_4R_R2_Status"
        ].eq(
            "PREVIOUSLY_UNMAPPED"
        ),
        "Biological_Genus"
    ]
)


assert observed_previously_unmapped == EXPECTED_PREVIOUSLY_UNMAPPED, (
    "The previous-unmapped genus set does not match the "
    "9.4R-R2 diagnostic.\n"
    f"Expected : {sorted(EXPECTED_PREVIOUSLY_UNMAPPED)}\n"
    f"Observed : {sorted(observed_previously_unmapped)}"
)


previously_unmapped_resolution = (
    complete_genus_family[
        complete_genus_family[
            "Biological_Genus"
        ].isin(
            EXPECTED_PREVIOUSLY_UNMAPPED
        )
    ]
    .copy()
    .sort_values(
        [
            "Family",
            "Biological_Genus",
        ]
    )
    .reset_index(drop=True)
)


print("\nPreviously unmapped evidence genera now resolved")
print("-" * 100)

display(
    previously_unmapped_resolution[
        [
            "Biological_Genus",
            "Family",
            "Previous_9_4R_R2_Status",
            "Family_Mapping_Source",
            "Family_Mapping_Status",
        ]
    ]
)


# =============================================================================
# 10. MAP ALL 63 EVIDENCE SPECIES TO FAMILY
# =============================================================================

species_family_complete = (
    species_pool
    .merge(
        complete_genus_family[
            [
                "Biological_Genus",
                "Family",
                "Family_Mapping_Source",
                "Family_Mapping_Status",
                "Previous_9_4R_R2_Status",
            ]
        ],
        on="Biological_Genus",
        how="left",
        validate="many_to_one"
    )
)


assert len(
    species_family_complete
) == 63


assert species_family_complete[
    "Family"
].notna().all()


assert species_family_complete[
    "Family_Mapping_Status"
].eq(
    "AUTHORITATIVE_FAMILY_MAPPING"
).all()


# =============================================================================
# 11. VALIDATE THE 29 PREVIOUSLY UNMAPPED SPECIES ARE NOW MAPPED
# =============================================================================

previously_unmapped_species = (
    species_family_complete[
        species_family_complete[
            "Previous_9_4R_R2_Status"
        ].eq(
            "PREVIOUSLY_UNMAPPED"
        )
    ]
    .copy()
)


print("\nEvidence species recovered by the 9.4R-R2A mapping correction")
print("-" * 100)

print(
    f"N species recovered : "
    f"{len(previously_unmapped_species)}"
)


display(
    previously_unmapped_species[
        [
            "Biological_Genus",
            "Biological_Species_Name",
            "Family",
            "Seed_Mass_mg",
            "Evidence_Sources",
        ]
    ]
    .sort_values(
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


assert len(
    previously_unmapped_species
) == 29, (
    f"Expected 29 previously unmapped evidence species; "
    f"found {len(previously_unmapped_species)}."
)


# =============================================================================
# 12. VALIDATE RESIDUAL TARGET FAMILY SET
# =============================================================================

target_family_set = set(
    canonical_family_map[
        "Family"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert len(
    target_family_set
) == 15


print("\nAuthoritative target families")
print("-" * 100)

for family in sorted(
    target_family_set
):

    print(
        f" - {family}"
    )


# =============================================================================
# 13. SPLIT EVIDENCE INTO TARGET vs NON-TARGET FAMILIES
# =============================================================================

target_family_evidence_species = (
    species_family_complete[
        species_family_complete[
            "Family"
        ].isin(
            target_family_set
        )
    ]
    .copy()
    .sort_values(
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


non_target_family_evidence_species = (
    species_family_complete[
        ~species_family_complete[
            "Family"
        ].isin(
            target_family_set
        )
    ]
    .copy()
    .sort_values(
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nEvidence species within the 15 target families after completion")
print("-" * 100)

print(
    f"N species : "
    f"{len(target_family_evidence_species)}"
)


display(
    target_family_evidence_species[
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
            "Seed_Mass_mg",
            "Evidence_Sources",
            "Previous_9_4R_R2_Status",
        ]
    ]
)


print("\nEvidence species outside the 15 target families")
print("-" * 100)

print(
    f"N species : "
    f"{len(non_target_family_evidence_species)}"
)


# =============================================================================
# 14. FAMILY REPRESENTATION DIAGNOSTIC
# =============================================================================

family_representation = (
    species_family_complete
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        N_Independent_Genera=(
            "Biological_Genus",
            "nunique"
        ),
    )
    .sort_values(
        [
            "N_Independent_Species",
            "Family",
        ],
        ascending=[
            False,
            True,
        ]
    )
    .reset_index(drop=True)
)


print("\nComplete family representation across all 63 evidence species")
print("-" * 100)

display(
    family_representation
)


# =============================================================================
# 15. TARGET-FAMILY REPRESENTATION CHECKPOINT
# =============================================================================

target_family_representation = (
    family_representation[
        family_representation[
            "Family"
        ].isin(
            target_family_set
        )
    ]
    .copy()
    .sort_values(
        "Family"
    )
    .reset_index(drop=True)
)


print("\nEvidence representation within the 15 target families")
print("-" * 100)

display(
    target_family_representation
)


# =============================================================================
# 16. CONFIRM IMPORTANT RECOVERED TARGET-FAMILY EVIDENCE
# =============================================================================
#
# These checks ensure that 9.4R-R2A genuinely repaired the major omissions
# exposed in 9.4R-R2.
#
# =============================================================================

recovered_target_genera = set(
    previously_unmapped_resolution.loc[
        previously_unmapped_resolution[
            "Family"
        ].isin(
            target_family_set
        ),
        "Biological_Genus"
    ]
)


expected_major_recoveries = {
    "Afzelia",
    "Anthonotha",
    "Bombax",
    "Bridelia",
    "Dialium",
    "Hildegardia",
    "Margaritaria",
    "Musanga",
    "Pavetta",
    "Philenoptera",
    "Piptadeniastrum",
    "Tetrapleura",
}


missing_major_recoveries = (
    expected_major_recoveries
    -
    recovered_target_genera
)


assert len(
    missing_major_recoveries
) == 0, (
    "Expected target-family evidence genera were not recovered:\n"
    f"{sorted(missing_major_recoveries)}"
)


# =============================================================================
# 17. STRONG VALIDATION
# =============================================================================

assert len(
    complete_genus_family
) == 58


assert complete_genus_family[
    "Biological_Genus"
].is_unique


assert complete_genus_family[
    "Family"
].notna().all()


assert len(
    species_family_complete
) == 63


assert species_family_complete[
    "Biological_Species_Name"
].nunique() == 63


assert species_family_complete[
    "Family"
].notna().all()


assert (
    len(
        target_family_evidence_species
    )
    +
    len(
        non_target_family_evidence_species
    )
    ==
    63
)


assert len(
    previously_unmapped_resolution
) == 27


assert len(
    previously_unmapped_species
) == 29


# =============================================================================
# 18. NO NUMERICAL FAMILY ASSIGNMENT
# =============================================================================

FAMILY_VALUES_ASSIGNED_9_4R_R2A = 0


assert FAMILY_VALUES_ASSIGNED_9_4R_R2A == 0


# =============================================================================
# 19. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Independent biological evidence species",
            "Evidence genera represented",
            "Previously mapped evidence genera",
            "Previously unmapped evidence genera",
            "Previously unmapped evidence species",
            "Evidence genera mapped after 9.4R-R2A",
            "Evidence genera remaining unmapped",
            "Evidence species mapped after 9.4R-R2A",
            "Evidence species remaining unmapped",
            "Residual target families",
            "Evidence species within target families",
            "Evidence species outside target families",
            "Family Seed Mass values assigned in 9.4R-R2A",
        ],

        "Value": [
            len(
                species_pool
            ),

            len(
                complete_genus_family
            ),

            int(
                complete_genus_family[
                    "Previous_9_4R_R2_Status"
                ].eq(
                    "PREVIOUSLY_MAPPED"
                ).sum()
            ),

            len(
                previously_unmapped_resolution
            ),

            len(
                previously_unmapped_species
            ),

            int(
                complete_genus_family[
                    "Family"
                ].notna()
                .sum()
            ),

            len(
                still_unmapped
            ),

            int(
                species_family_complete[
                    "Family"
                ].notna()
                .sum()
            ),

            int(
                species_family_complete[
                    "Family"
                ].isna()
                .sum()
            ),

            len(
                target_family_set
            ),

            len(
                target_family_evidence_species
            ),

            len(
                non_target_family_evidence_species
            ),

            0,
        ],
    }
)


print("\nSection 9.4R-R2A summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 20. EXPORT
# =============================================================================

complete_genus_family.to_csv(
    OUT_COMPLETE_GENUS_FAMILY_MAP,
    index=False
)


species_family_complete.to_csv(
    OUT_COMPLETE_SPECIES_FAMILY_MAP,
    index=False
)


previously_unmapped_resolution.to_csv(
    OUT_PREVIOUSLY_UNMAPPED_RESOLUTION,
    index=False
)


target_family_evidence_species.to_csv(
    OUT_TARGET_FAMILY_EVIDENCE_SPECIES,
    index=False
)


non_target_family_evidence_species.to_csv(
    OUT_NON_TARGET_FAMILY_EVIDENCE_SPECIES,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 21. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R-R2A exports")
print("-" * 100)

print(
    f"Complete genus-family map    : "
    f"{OUT_COMPLETE_GENUS_FAMILY_MAP}"
)

print(
    f"Complete species-family map  : "
    f"{OUT_COMPLETE_SPECIES_FAMILY_MAP}"
)

print(
    f"Recovered genus mappings     : "
    f"{OUT_PREVIOUSLY_UNMAPPED_RESOLUTION}"
)

print(
    f"Target-family evidence       : "
    f"{OUT_TARGET_FAMILY_EVIDENCE_SPECIES}"
)

print(
    f"Non-target-family evidence   : "
    f"{OUT_NON_TARGET_FAMILY_EVIDENCE_SPECIES}"
)

print(
    f"Summary                      : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 22. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Complete authoritative family mapping of the "
    "Seed Mass evidence pool completed successfully."
)

print(
    f"Independent evidence species      : "
    f"{len(species_pool)}"
)

print(
    f"Evidence genera                   : "
    f"{len(complete_genus_family)}"
)

print(
    f"Previously unmapped genera        : "
    f"{len(previously_unmapped_resolution)}"
)

print(
    f"Previously unmapped species       : "
    f"{len(previously_unmapped_species)}"
)

print(
    f"Evidence genera now mapped        : "
    f"{complete_genus_family['Family'].notna().sum()} / "
    f"{len(complete_genus_family)}"
)

print(
    f"Evidence species now mapped       : "
    f"{species_family_complete['Family'].notna().sum()} / "
    f"{len(species_family_complete)}"
)

print(
    f"Evidence species in target families : "
    f"{len(target_family_evidence_species)}"
)

print(
    "Family Seed Mass values assigned   : 0"
)

print(
    "The completed evidence-family map is now ready "
    "for corrected family evidence auditing in 9.4R-R2B."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R2B
# CORRECTED FAMILY-LEVEL SEED MASS EVIDENCE AUDIT
# =============================================================================
#
# Purpose
# -------
# 1. Use the COMPLETE 63-species family mapping from 9.4R-R2A.
# 2. Restrict evidence to the 15 authoritative residual target families.
# 3. Recompute family-level evidence depth using independent biological species.
# 4. Quantify:
#       - independent species
#       - independent genera
#       - raw Seed Mass min / median / mean / max
#       - log10 Seed Mass min / median / mean / max
#       - raw fold range
#       - log10 range
# 5. Classify family evidence depth conservatively.
# 6. Prepare family-transfer feasibility diagnostics.
#
# IMPORTANT
# ---------
# This remains an AUDIT ONLY.
#
# NO family-level Seed Mass value is assigned.
# NO canonical taxon is resolved in this section.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Authoritative inputs
# -----------------------------------------------------------------------------

COMPLETE_SPECIES_FAMILY_FILE = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_family_mapping_complete.csv"
)

CANONICAL_FAMILY_MAP_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)

GENUS_FAMILY_CROSSWALK_FILE = (
    OUT_DIR
    / "omo_seed_mass_residual_genus_family_authoritative_crosswalk.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_TARGET_FAMILY_EVIDENCE_DETAIL = (
    OUT_DIR
    / "omo_seed_mass_corrected_family_evidence_detail.csv"
)

OUT_CORRECTED_FAMILY_AUDIT = (
    OUT_DIR
    / "omo_seed_mass_corrected_family_evidence_audit.csv"
)

OUT_FAMILY_CLASS_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_corrected_family_evidence_class_summary.csv"
)

OUT_TRANSFER_FEASIBILITY = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_feasibility_register.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_corrected_family_evidence_audit_summary.csv"
)


# =============================================================================
# 2. VALIDATE REQUIRED INPUTS
# =============================================================================

required_files = [
    COMPLETE_SPECIES_FAMILY_FILE,
    CANONICAL_FAMILY_MAP_FILE,
    GENUS_FAMILY_CROSSWALK_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4R-R2B input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

species_family = pd.read_csv(
    COMPLETE_SPECIES_FAMILY_FILE,
    low_memory=False
)

canonical_family_map = pd.read_csv(
    CANONICAL_FAMILY_MAP_FILE,
    low_memory=False
)

residual_genus_family = pd.read_csv(
    GENUS_FAMILY_CROSSWALK_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R-R2B — CORRECTED FAMILY-LEVEL "
    "SEED MASS EVIDENCE AUDIT"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Complete independent evidence species : "
    f"{len(species_family)}"
)

print(
    f"Family-audit canonical taxa           : "
    f"{len(canonical_family_map)}"
)

print(
    f"Residual genus-family rows             : "
    f"{len(residual_genus_family)}"
)


# =============================================================================
# 4. VALIDATE AUTHORITATIVE INPUT STATE
# =============================================================================

assert len(
    species_family
) == 63, (
    f"Expected 63 independent evidence species; "
    f"found {len(species_family)}."
)


assert species_family[
    "Biological_Species_Name"
].nunique() == 63


assert species_family[
    "Family"
].notna().all()


assert len(
    canonical_family_map
) == 31


assert canonical_family_map[
    "Canonical_Species"
].nunique() == 31


assert canonical_family_map[
    "Family"
].notna().all()


assert len(
    residual_genus_family
) == 21


# =============================================================================
# 5. TARGET FAMILY SET
# =============================================================================

target_family_set = set(
    canonical_family_map[
        "Family"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assert len(
    target_family_set
) == 15, (
    f"Expected 15 target families; "
    f"found {len(target_family_set)}."
)


# =============================================================================
# 6. STANDARDIZE NUMERIC SEED MASS
# =============================================================================

species_family[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_family[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert species_family[
    "Seed_Mass_mg"
].notna().all()


assert species_family[
    "Seed_Mass_mg"
].gt(
    0
).all()


# =============================================================================
# 7. ADD LOG10 SEED MASS
# =============================================================================

species_family[
    "Log10_Seed_Mass"
] = np.log10(
    species_family[
        "Seed_Mass_mg"
    ]
)


assert species_family[
    "Log10_Seed_Mass"
].notna().all()


# =============================================================================
# 8. RESTRICT TO TARGET FAMILIES
# =============================================================================

target_family_evidence = (
    species_family[
        species_family[
            "Family"
        ].isin(
            target_family_set
        )
    ]
    .copy()
    .sort_values(
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nCorrected evidence species within target families")
print("-" * 100)

print(
    f"N species : "
    f"{len(target_family_evidence)}"
)


display(
    target_family_evidence[
        [
            "Family",
            "Biological_Genus",
            "Biological_Species_Name",
            "Seed_Mass_mg",
            "Log10_Seed_Mass",
            "Evidence_Sources",
        ]
    ]
)


assert len(
    target_family_evidence
) == 30, (
    f"Expected 30 evidence species within target families; "
    f"found {len(target_family_evidence)}."
)


# =============================================================================
# 9. HELPER
# =============================================================================

def join_unique(series):

    values = []

    for value in series.dropna():

        text = str(value).strip()

        if not text:
            continue

        pieces = [
            x.strip()
            for x in text.split("|")
            if x.strip()
        ]

        values.extend(
            pieces
        )

    return " | ".join(
        sorted(
            set(values)
        )
    )


# =============================================================================
# 10. BUILD TARGET FAMILY STRUCTURE
# =============================================================================

family_target_structure = (
    canonical_family_map
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Residual_Genera=(
            "Residual_Genus",
            "nunique"
        ),

        Residual_Genera=(
            "Residual_Genus",
            join_unique
        ),

        N_Residual_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Residual_Canonical_Taxa=(
            "Canonical_Species",
            join_unique
        ),

        Residual_Biological_Targets=(
            "Final_Search_Name",
            join_unique
        ),

        Family_Audit_Entry_Routes=(
            "Family_Audit_Entry_Route",
            join_unique
        ),
    )
)


assert len(
    family_target_structure
) == 15


# =============================================================================
# 11. FAMILY EVIDENCE SUMMARY
# =============================================================================

family_evidence_summary = (
    target_family_evidence
    .groupby(
        "Family",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        N_Independent_Genera=(
            "Biological_Genus",
            "nunique"
        ),

        Species_Represented=(
            "Biological_Species_Name",
            join_unique
        ),

        Genera_Represented=(
            "Biological_Genus",
            join_unique
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            join_unique
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),

        Min_Log10_Seed_Mass=(
            "Log10_Seed_Mass",
            "min"
        ),

        Median_Log10_Seed_Mass=(
            "Log10_Seed_Mass",
            "median"
        ),

        Mean_Log10_Seed_Mass=(
            "Log10_Seed_Mass",
            "mean"
        ),

        Max_Log10_Seed_Mass=(
            "Log10_Seed_Mass",
            "max"
        ),
    )
)


# =============================================================================
# 12. MERGE ZERO-EVIDENCE TARGET FAMILIES
# =============================================================================

family_audit = (
    family_target_structure
    .merge(
        family_evidence_summary,
        on="Family",
        how="left",
        validate="one_to_one"
    )
)


for col in [
    "N_Independent_Species",
    "N_Independent_Genera",
]:

    family_audit[
        col
    ] = (
        family_audit[
            col
        ]
        .fillna(0)
        .astype(int)
    )


for col in [
    "Species_Represented",
    "Genera_Represented",
    "Evidence_Sources",
]:

    family_audit[
        col
    ] = (
        family_audit[
            col
        ]
        .fillna("")
    )


# =============================================================================
# 13. ADD DISPERSION DIAGNOSTICS
# =============================================================================

family_audit[
    "Raw_Fold_Range"
] = np.where(
    family_audit[
        "Min_Seed_Mass_mg"
    ].gt(0),
    (
        family_audit[
            "Max_Seed_Mass_mg"
        ]
        /
        family_audit[
            "Min_Seed_Mass_mg"
        ]
    ),
    np.nan
)


family_audit[
    "Log10_Range"
] = (
    family_audit[
        "Max_Log10_Seed_Mass"
    ]
    -
    family_audit[
        "Min_Log10_Seed_Mass"
    ]
)


family_audit[
    "Geometric_Mean_Seed_Mass_mg"
] = np.where(
    family_audit[
        "Mean_Log10_Seed_Mass"
    ].notna(),
    10 ** (
        family_audit[
            "Mean_Log10_Seed_Mass"
        ]
    ),
    np.nan
)


# =============================================================================
# 14. FAMILY EVIDENCE CLASS
# =============================================================================

def classify_family_evidence(
    n_species
):

    if n_species >= 5:

        return (
            "STRONG_FAMILY_EVIDENCE"
        )

    if n_species >= 3:

        return (
            "MODERATE_FAMILY_EVIDENCE"
        )

    if n_species >= 1:

        return (
            "LIMITED_FAMILY_EVIDENCE"
        )

    return (
        "NO_FAMILY_EVIDENCE"
    )


family_audit[
    "Family_Evidence_Class"
] = (
    family_audit[
        "N_Independent_Species"
    ]
    .map(
        classify_family_evidence
    )
)


# =============================================================================
# 15. TAXONOMIC BREADTH CLASS
# =============================================================================

def classify_taxonomic_breadth(
    n_genera
):

    if n_genera >= 5:

        return (
            "BROAD_MULTI_GENUS_SUPPORT"
        )

    if n_genera >= 3:

        return (
            "MODERATE_MULTI_GENUS_SUPPORT"
        )

    if n_genera >= 2:

        return (
            "LIMITED_MULTI_GENUS_SUPPORT"
        )

    if n_genera == 1:

        return (
            "SINGLE_GENUS_SUPPORT"
        )

    return (
        "NO_GENUS_SUPPORT"
    )


family_audit[
    "Family_Taxonomic_Breadth_Class"
] = (
    family_audit[
        "N_Independent_Genera"
    ]
    .map(
        classify_taxonomic_breadth
    )
)


# =============================================================================
# 16. DISPERSION RISK CLASS
# =============================================================================
#
# Seed Mass can vary by orders of magnitude.
#
# Log10 range:
#     < 1       = less than 10-fold spread
#     1 to < 2  = 10–100-fold spread
#     >= 2      = >=100-fold spread
#
# =============================================================================

def classify_dispersion(
    log_range
):

    if pd.isna(
        log_range
    ):

        return (
            "NO_DISPERSION_ESTIMATE"
        )

    if log_range < 1:

        return (
            "LOW_DISPERSION"
        )

    if log_range < 2:

        return (
            "MODERATE_DISPERSION"
        )

    return (
        "HIGH_DISPERSION"
    )


family_audit[
    "Family_Dispersion_Risk"
] = (
    family_audit[
        "Log10_Range"
    ]
    .map(
        classify_dispersion
    )
)


# =============================================================================
# 17. FAMILY TRANSFER FEASIBILITY
# =============================================================================
#
# This is NOT yet a transfer decision.
#
# It only identifies whether the family has enough breadth to proceed
# to formal rule development in 9.4R-R3.
#
# =============================================================================

def determine_transfer_feasibility(
    row
):

    n_species = row[
        "N_Independent_Species"
    ]

    n_genera = row[
        "N_Independent_Genera"
    ]

    dispersion = row[
        "Family_Dispersion_Risk"
    ]


    if (
        n_species >= 5
        and
        n_genera >= 5
    ):

        if dispersion == "HIGH_DISPERSION":

            return (
                "STRONG_BUT_HIGH_DISPERSION_REVIEW_REQUIRED"
            )

        return (
            "STRONG_CANDIDATE_FOR_FAMILY_TRANSFER_RULE"
        )


    if (
        n_species >= 3
        and
        n_genera >= 2
    ):

        return (
            "MODERATE_CANDIDATE_REQUIRES_CAUTION"
        )


    if (
        n_species >= 1
    ):

        return (
            "INSUFFICIENT_FOR_AUTOMATIC_FAMILY_TRANSFER"
        )


    return (
        "NO_FAMILY_EVIDENCE"
    )


family_audit[
    "Family_Transfer_Feasibility"
] = (
    family_audit
    .apply(
        determine_transfer_feasibility,
        axis=1
    )
)


family_audit[
    "Numerical_Family_Value_Assigned"
] = False


# =============================================================================
# 18. DISPLAY CORRECTED FAMILY AUDIT
# =============================================================================

print("\nCorrected authoritative family evidence audit")
print("-" * 100)

display(
    family_audit[
        [
            "Family",
            "N_Residual_Genera",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Genera_Represented",
            "Min_Seed_Mass_mg",
            "Median_Seed_Mass_mg",
            "Geometric_Mean_Seed_Mass_mg",
            "Max_Seed_Mass_mg",
            "Raw_Fold_Range",
            "Log10_Range",
            "Family_Evidence_Class",
            "Family_Taxonomic_Breadth_Class",
            "Family_Dispersion_Risk",
            "Family_Transfer_Feasibility",
        ]
    ]
    .sort_values(
        [
            "Family_Evidence_Class",
            "Family",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 19. FAMILY CLASS SUMMARY
# =============================================================================

family_class_summary = (
    family_audit[
        "Family_Evidence_Class"
    ]
    .value_counts()
    .rename_axis(
        "Family_Evidence_Class"
    )
    .reset_index(
        name="N_Families"
    )
)


family_class_summary[
    "Percent_of_15_Families"
] = (
    family_class_summary[
        "N_Families"
    ]
    /
    15
    *
    100
)


print("\nCorrected family evidence-depth summary")
print("-" * 100)

display(
    family_class_summary
)


# =============================================================================
# 20. STRONG FAMILY CANDIDATES
# =============================================================================

strong_candidates = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].eq(
            "STRONG_FAMILY_EVIDENCE"
        )
    ]
    .copy()
)


print("\nStrong family-evidence candidates")
print("-" * 100)

display(
    strong_candidates[
        [
            "Family",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Median_Seed_Mass_mg",
            "Geometric_Mean_Seed_Mass_mg",
            "Raw_Fold_Range",
            "Log10_Range",
            "Family_Dispersion_Risk",
            "Residual_Genera",
            "Family_Transfer_Feasibility",
        ]
    ]
)


# =============================================================================
# 21. LIMITED / NO FAMILY EVIDENCE
# =============================================================================

limited_or_none = (
    family_audit[
        family_audit[
            "Family_Evidence_Class"
        ].isin(
            [
                "LIMITED_FAMILY_EVIDENCE",
                "NO_FAMILY_EVIDENCE",
            ]
        )
    ]
    .copy()
)


print("\nLimited or absent family evidence")
print("-" * 100)

display(
    limited_or_none[
        [
            "Family",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Residual_Genera",
            "Family_Evidence_Class",
            "Family_Transfer_Feasibility",
        ]
    ]
)


# =============================================================================
# 22. TRANSFER FEASIBILITY REGISTER
# =============================================================================

transfer_feasibility = (
    family_audit[
        [
            "Family",
            "N_Residual_Genera",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Median_Seed_Mass_mg",
            "Geometric_Mean_Seed_Mass_mg",
            "Raw_Fold_Range",
            "Log10_Range",
            "Family_Evidence_Class",
            "Family_Taxonomic_Breadth_Class",
            "Family_Dispersion_Risk",
            "Family_Transfer_Feasibility",
            "Numerical_Family_Value_Assigned",
        ]
    ]
    .copy()
)


print("\nFamily transfer-feasibility register")
print("-" * 100)

display(
    transfer_feasibility
)


# =============================================================================
# 23. VALIDATION
# =============================================================================

assert len(
    family_audit
) == 15


assert family_audit[
    "Family"
].is_unique


assert (
    family_audit[
        "N_Residual_Canonical_Taxa"
    ].sum()
    ==
    31
)


assert (
    family_audit[
        "N_Residual_Genera"
    ].sum()
    ==
    21
)


assert (
    family_class_summary[
        "N_Families"
    ].sum()
    ==
    15
)


assert (
    family_audit[
        "Numerical_Family_Value_Assigned"
    ]
    .eq(False)
    .all()
)


# =============================================================================
# 24. EXPECTED STRONG FAMILY CHECK
# =============================================================================

strong_family_names = set(
    strong_candidates[
        "Family"
    ]
)


assert {
    "Fabaceae",
    "Malvaceae",
}.issubset(
    strong_family_names
), (
    "Fabaceae and Malvaceae should both qualify as "
    "strong family-evidence groups after the corrected mapping."
)


# =============================================================================
# 25. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4R_R2B = 90

N_FAMILY_ASSIGNMENTS_9_4R_R2B = 0

N_RESOLVED_AFTER_9_4R_R2B = (
    N_RESOLVED_BEFORE_9_4R_R2B
    +
    N_FAMILY_ASSIGNMENTS_9_4R_R2B
)


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4R_R2B
    /
    N_CANONICAL_TAXA
)


assert N_RESOLVED_AFTER_9_4R_R2B == 90


# =============================================================================
# 26. SECTION SUMMARY
# =============================================================================

summary_rows = [

    {
        "Metric":
            "Independent evidence species",
        "Value":
            len(
                species_family
            ),
    },

    {
        "Metric":
            "Evidence species within target families",
        "Value":
            len(
                target_family_evidence
            ),
    },

    {
        "Metric":
            "Target families audited",
        "Value":
            len(
                family_audit
            ),
    },

    {
        "Metric":
            "Residual genera represented",
        "Value":
            int(
                family_audit[
                    "N_Residual_Genera"
                ].sum()
            ),
    },

    {
        "Metric":
            "Residual canonical taxa represented",
        "Value":
            int(
                family_audit[
                    "N_Residual_Canonical_Taxa"
                ].sum()
            ),
    },

    {
        "Metric":
            "Family Seed Mass values assigned in 9.4R-R2B",
        "Value":
            0,
    },

    {
        "Metric":
            "Resolved canonical taxa after 9.4R-R2B",
        "Value":
            N_RESOLVED_AFTER_9_4R_R2B,
    },

    {
        "Metric":
            "Overall Seed Mass coverage percent",
        "Value":
            coverage_percent,
    },
]


for _, row in family_class_summary.iterrows():

    summary_rows.append(
        {
            "Metric":
                "Families — "
                +
                row[
                    "Family_Evidence_Class"
                ],

            "Value":
                int(
                    row[
                        "N_Families"
                    ]
                ),
        }
    )


section_summary = pd.DataFrame(
    summary_rows
)


print("\nSection 9.4R-R2B summary")
print("-" * 100)

display(
    section_summary
)


# =============================================================================
# 27. EXPORT
# =============================================================================

target_family_evidence.to_csv(
    OUT_TARGET_FAMILY_EVIDENCE_DETAIL,
    index=False
)


family_audit.to_csv(
    OUT_CORRECTED_FAMILY_AUDIT,
    index=False
)


family_class_summary.to_csv(
    OUT_FAMILY_CLASS_SUMMARY,
    index=False
)


transfer_feasibility.to_csv(
    OUT_TRANSFER_FEASIBILITY,
    index=False
)


section_summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 28. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R-R2B exports")
print("-" * 100)

print(
    f"Target-family evidence detail : "
    f"{OUT_TARGET_FAMILY_EVIDENCE_DETAIL}"
)

print(
    f"Corrected family audit        : "
    f"{OUT_CORRECTED_FAMILY_AUDIT}"
)

print(
    f"Family class summary          : "
    f"{OUT_FAMILY_CLASS_SUMMARY}"
)

print(
    f"Transfer feasibility register : "
    f"{OUT_TRANSFER_FEASIBILITY}"
)

print(
    f"Section summary               : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 29. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Corrected family-level Seed Mass evidence audit "
    "completed successfully."
)

print(
    f"Independent evidence species        : "
    f"{len(species_family)}"
)

print(
    f"Evidence species in target families : "
    f"{len(target_family_evidence)}"
)

print(
    f"Target families audited             : "
    f"{len(family_audit)}"
)

print(
    f"Strong family-evidence groups       : "
    f"{len(strong_candidates)}"
)

print(
    "Family Seed Mass values assigned   : 0"
)

print(
    f"Overall Seed Mass coverage remains  : "
    f"{N_RESOLVED_AFTER_9_4R_R2B} / "
    f"{N_CANONICAL_TAXA} "
    f"({coverage_percent:.2f}%)"
)

print(
    "Family evidence is now ready for formal transfer-rule "
    "development in Section 9.4R-R3."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R3
# FAMILY-LEVEL SEED MASS TRANSFER RULE DEVELOPMENT
# AND CANDIDATE ASSIGNMENT
# =============================================================================
#
# Purpose
# -------
# 1. Formalize conservative family-level Seed Mass transfer rules.
# 2. Use:
#       - independent species depth
#       - independent genus breadth
#       - log10 Seed Mass dispersion
# 3. Distinguish:
#       - APPROVED_FAMILY_TRANSFER_CANDIDATE
#       - HIGH_DISPERSION_EXPLICIT_REVIEW
#       - INSUFFICIENT_FAMILY_EVIDENCE
#       - NO_FAMILY_EVIDENCE
# 4. Calculate a candidate family Seed Mass only when transfer criteria pass.
#
# Family-transfer rule
# --------------------
# Automatic family candidate permitted only if:
#
#     N independent species >= 5
#     N independent genera  >= 5
#     Log10 range           < 2.0
#
# Interpretation:
#
#     Log10 range < 2
#         family spread is <100-fold.
#
#     Log10 range >= 2
#         family spread is >=100-fold and therefore too heterogeneous for
#         automatic transfer even when evidence depth is strong.
#
# Estimator
# ---------
# For approved family candidates:
#
#     FAMILY ESTIMATE =
#         10 ** median(log10 individual species Seed Mass)
#
# This is a robust log-scale central tendency appropriate for a trait spanning
# multiple orders of magnitude.
#
# IMPORTANT
# ---------
# Section 9.4R-R3 develops and evaluates candidate family estimates.
#
# It DOES NOT yet propagate family estimates to canonical residual taxa.
# Canonical assignment is deferred to Section 9.4R-R4.
#
# Expected current state
# ----------------------
# Target families                    : 15
# Strong evidence families           : 2
# Expected approved family candidate : Malvaceae
# Expected explicit high-dispersion  : Fabaceae
#
# Seed Mass coverage remains         : 90 / 127
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs from 9.4R-R2B
# -----------------------------------------------------------------------------

CORRECTED_FAMILY_AUDIT_FILE = (
    OUT_DIR
    / "omo_seed_mass_corrected_family_evidence_audit.csv"
)

TARGET_FAMILY_EVIDENCE_FILE = (
    OUT_DIR
    / "omo_seed_mass_corrected_family_evidence_detail.csv"
)

TRANSFER_FEASIBILITY_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_feasibility_register.csv"
)

CANONICAL_FAMILY_MAP_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_TRANSFER_RULES = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_rules.csv"
)

OUT_FAMILY_TRANSFER_DECISIONS = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_decisions.csv"
)

OUT_APPROVED_FAMILY_CANDIDATES = (
    OUT_DIR
    / "omo_seed_mass_approved_family_transfer_candidates.csv"
)

OUT_HIGH_DISPERSION_REVIEW = (
    OUT_DIR
    / "omo_seed_mass_family_high_dispersion_review.csv"
)

OUT_LIMITED_OR_NO_EVIDENCE = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_noneligible_families.csv"
)

OUT_CANONICAL_CANDIDATE_PREVIEW = (
    OUT_DIR
    / "omo_seed_mass_family_candidate_canonical_preview.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_rule_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    CORRECTED_FAMILY_AUDIT_FILE,
    TARGET_FAMILY_EVIDENCE_FILE,
    TRANSFER_FEASIBILITY_FILE,
    CANONICAL_FAMILY_MAP_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4R-R3 input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

family_audit = pd.read_csv(
    CORRECTED_FAMILY_AUDIT_FILE,
    low_memory=False
)

family_evidence = pd.read_csv(
    TARGET_FAMILY_EVIDENCE_FILE,
    low_memory=False
)

transfer_feasibility = pd.read_csv(
    TRANSFER_FEASIBILITY_FILE,
    low_memory=False
)

canonical_family_map = pd.read_csv(
    CANONICAL_FAMILY_MAP_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R-R3 — FAMILY-LEVEL SEED MASS TRANSFER RULE "
    "DEVELOPMENT AND CANDIDATE ASSIGNMENT"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Target families                  : "
    f"{len(family_audit)}"
)

print(
    f"Target-family evidence species   : "
    f"{len(family_evidence)}"
)

print(
    f"Transfer-feasibility rows        : "
    f"{len(transfer_feasibility)}"
)

print(
    f"Canonical family-audit taxa      : "
    f"{len(canonical_family_map)}"
)


# =============================================================================
# 4. VALIDATE INPUT STATE
# =============================================================================

assert len(
    family_audit
) == 15


assert family_audit[
    "Family"
].is_unique


assert len(
    family_evidence
) == 30


assert len(
    canonical_family_map
) == 31


assert canonical_family_map[
    "Canonical_Species"
].nunique() == 31


assert (
    family_audit[
        "Numerical_Family_Value_Assigned"
    ]
    .eq(False)
    .all()
)


# =============================================================================
# 5. ENSURE NUMERIC FIELDS
# =============================================================================

numeric_cols = [
    "N_Independent_Species",
    "N_Independent_Genera",
    "Median_Seed_Mass_mg",
    "Geometric_Mean_Seed_Mass_mg",
    "Raw_Fold_Range",
    "Log10_Range",
]


for col in numeric_cols:

    family_audit[
        col
    ] = pd.to_numeric(
        family_audit[
            col
        ],
        errors="coerce"
    )


family_evidence[
    "Seed_Mass_mg"
] = pd.to_numeric(
    family_evidence[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert family_evidence[
    "Seed_Mass_mg"
].notna().all()


assert family_evidence[
    "Seed_Mass_mg"
].gt(0).all()


# =============================================================================
# 6. DEFINE FORMAL FAMILY TRANSFER RULES
# =============================================================================

transfer_rules = pd.DataFrame(
    [
        {
            "Rule_ID":
                "FTR_01",

            "Rule_Name":
                "APPROVED_STRONG_MODERATE_DISPERSION",

            "Minimum_Independent_Species":
                5,

            "Minimum_Independent_Genera":
                5,

            "Maximum_Log10_Range_Exclusive":
                2.0,

            "Automatic_Candidate_Permitted":
                True,

            "Estimator":
                "BACK_TRANSFORMED_MEDIAN_LOG10_SPECIES_SEED_MASS",

            "Decision":
                "APPROVED_FAMILY_TRANSFER_CANDIDATE",

            "Rationale":
                (
                    "Broad multi-species and multi-genus family evidence "
                    "with <100-fold Seed Mass spread is considered sufficient "
                    "for conservative family-level candidate estimation."
                ),
        },

        {
            "Rule_ID":
                "FTR_02",

            "Rule_Name":
                "STRONG_BUT_HIGH_DISPERSION",

            "Minimum_Independent_Species":
                5,

            "Minimum_Independent_Genera":
                5,

            "Maximum_Log10_Range_Exclusive":
                np.nan,

            "Automatic_Candidate_Permitted":
                False,

            "Estimator":
                "NONE_WITHOUT_EXPLICIT_REVIEW",

            "Decision":
                "HIGH_DISPERSION_EXPLICIT_REVIEW",

            "Rationale":
                (
                    "Strong evidence depth does not justify automatic family "
                    "transfer when Seed Mass spans >=100-fold across species."
                ),
        },

        {
            "Rule_ID":
                "FTR_03",

            "Rule_Name":
                "LIMITED_OR_INADEQUATE_FAMILY_SUPPORT",

            "Minimum_Independent_Species":
                np.nan,

            "Minimum_Independent_Genera":
                np.nan,

            "Maximum_Log10_Range_Exclusive":
                np.nan,

            "Automatic_Candidate_Permitted":
                False,

            "Estimator":
                "NONE",

            "Decision":
                "INSUFFICIENT_FAMILY_EVIDENCE",

            "Rationale":
                (
                    "Families with fewer than 5 independent species or fewer "
                    "than 5 independent genera are not eligible for automatic "
                    "family-level transfer."
                ),
        },

        {
            "Rule_ID":
                "FTR_04",

            "Rule_Name":
                "NO_FAMILY_EVIDENCE",

            "Minimum_Independent_Species":
                0,

            "Minimum_Independent_Genera":
                0,

            "Maximum_Log10_Range_Exclusive":
                np.nan,

            "Automatic_Candidate_Permitted":
                False,

            "Estimator":
                "NONE",

            "Decision":
                "NO_FAMILY_EVIDENCE",

            "Rationale":
                (
                    "No independent Seed Mass evidence exists for the family "
                    "within the validated evidence pool."
                ),
        },
    ]
)


print("\nFamily-transfer rule framework")
print("-" * 100)

display(
    transfer_rules
)


# =============================================================================
# 7. CLASSIFY FAMILY TRANSFER DECISION
# =============================================================================

def classify_transfer_decision(
    row
):

    n_species = int(
        row[
            "N_Independent_Species"
        ]
    )

    n_genera = int(
        row[
            "N_Independent_Genera"
        ]
    )

    log_range = row[
        "Log10_Range"
    ]


    if n_species == 0:

        return (
            "FTR_04",
            "NO_FAMILY_EVIDENCE"
        )


    if (
        n_species >= 5
        and
        n_genera >= 5
    ):

        if (
            pd.notna(log_range)
            and
            log_range < 2.0
        ):

            return (
                "FTR_01",
                "APPROVED_FAMILY_TRANSFER_CANDIDATE"
            )

        return (
            "FTR_02",
            "HIGH_DISPERSION_EXPLICIT_REVIEW"
        )


    return (
        "FTR_03",
        "INSUFFICIENT_FAMILY_EVIDENCE"
    )


decision_pairs = (
    family_audit
    .apply(
        classify_transfer_decision,
        axis=1
    )
)


family_decisions = (
    family_audit
    .copy()
)


family_decisions[
    "Family_Transfer_Rule_ID"
] = [
    pair[0]
    for pair in decision_pairs
]


family_decisions[
    "Family_Transfer_Decision"
] = [
    pair[1]
    for pair in decision_pairs
]


# =============================================================================
# 8. CALCULATE LOG-SCALE FAMILY CANDIDATE ESTIMATOR
# =============================================================================
#
# Candidate estimate:
#
#     10 ** median(log10(seed mass))
#
# It is calculated for every family with evidence for diagnostic purposes,
# but it becomes an APPROVED candidate only when FTR_01 passes.
#
# =============================================================================

family_candidate_rows = []


for family_name in family_decisions[
    "Family"
]:

    fam = family_evidence[
        family_evidence[
            "Family"
        ].eq(
            family_name
        )
    ].copy()


    if len(
        fam
    ) == 0:

        candidate_log10 = np.nan
        candidate_mg = np.nan

    else:

        fam[
            "Log10_Seed_Mass"
        ] = np.log10(
            fam[
                "Seed_Mass_mg"
            ]
        )


        candidate_log10 = (
            fam[
                "Log10_Seed_Mass"
            ]
            .median()
        )


        candidate_mg = (
            10 ** candidate_log10
        )


    family_candidate_rows.append(
        {
            "Family":
                family_name,

            "Candidate_Median_Log10_Seed_Mass":
                candidate_log10,

            "Candidate_Family_Seed_Mass_mg":
                candidate_mg,
        }
    )


candidate_table = pd.DataFrame(
    family_candidate_rows
)


family_decisions = (
    family_decisions
    .merge(
        candidate_table,
        on="Family",
        how="left",
        validate="one_to_one"
    )
)


# =============================================================================
# 9. APPROVAL FLAG
# =============================================================================

family_decisions[
    "Family_Candidate_Approved"
] = (
    family_decisions[
        "Family_Transfer_Decision"
    ]
    .eq(
        "APPROVED_FAMILY_TRANSFER_CANDIDATE"
    )
)


family_decisions[
    "Family_Estimator"
] = np.where(
    family_decisions[
        "Family_Candidate_Approved"
    ],
    "BACK_TRANSFORMED_MEDIAN_LOG10_SPECIES_SEED_MASS",
    "NONE_OR_REVIEW_REQUIRED"
)


family_decisions[
    "Family_Provenance_Class"
] = np.where(
    family_decisions[
        "Family_Candidate_Approved"
    ],
    "FAMILY_ESTIMATE_CANDIDATE",
    pd.NA
)


family_decisions[
    "Family_Quality_Flag"
] = np.select(
    [
        family_decisions[
            "Family_Transfer_Decision"
        ].eq(
            "APPROVED_FAMILY_TRANSFER_CANDIDATE"
        ),

        family_decisions[
            "Family_Transfer_Decision"
        ].eq(
            "HIGH_DISPERSION_EXPLICIT_REVIEW"
        ),

        family_decisions[
            "Family_Transfer_Decision"
        ].eq(
            "INSUFFICIENT_FAMILY_EVIDENCE"
        ),

        family_decisions[
            "Family_Transfer_Decision"
        ].eq(
            "NO_FAMILY_EVIDENCE"
        ),
    ],

    [
        "STRONG_MULTI_GENUS_MODERATE_DISPERSION",
        "STRONG_MULTI_GENUS_HIGH_DISPERSION_REVIEW",
        "LIMITED_FAMILY_SUPPORT_NO_TRANSFER",
        "NO_FAMILY_EVIDENCE",
    ],

    default="UNCLASSIFIED"
)


# =============================================================================
# 10. EXPLICITLY REMOVE CANDIDATE VALUES FROM NON-APPROVED FAMILIES
# =============================================================================
#
# Diagnostic estimates can be calculated internally, but the exported
# candidate-assignment field must remain missing unless the family passes FTR_01.
#
# =============================================================================

family_decisions[
    "Approved_Family_Seed_Mass_Candidate_mg"
] = np.where(
    family_decisions[
        "Family_Candidate_Approved"
    ],
    family_decisions[
        "Candidate_Family_Seed_Mass_mg"
    ],
    np.nan
)


# =============================================================================
# 11. DISPLAY FAMILY TRANSFER DECISIONS
# =============================================================================

print("\nFamily transfer-rule decisions")
print("-" * 100)

display(
    family_decisions[
        [
            "Family",
            "N_Residual_Genera",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Median_Seed_Mass_mg",
            "Geometric_Mean_Seed_Mass_mg",
            "Raw_Fold_Range",
            "Log10_Range",
            "Family_Evidence_Class",
            "Family_Dispersion_Risk",
            "Family_Transfer_Rule_ID",
            "Family_Transfer_Decision",
            "Candidate_Family_Seed_Mass_mg",
            "Approved_Family_Seed_Mass_Candidate_mg",
            "Family_Quality_Flag",
        ]
    ]
    .sort_values(
        [
            "Family_Transfer_Decision",
            "Family",
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 12. APPROVED FAMILY CANDIDATES
# =============================================================================

approved_candidates = (
    family_decisions[
        family_decisions[
            "Family_Candidate_Approved"
        ]
    ]
    .copy()
)


print("\nApproved family-level Seed Mass candidates")
print("-" * 100)

if len(
    approved_candidates
) > 0:

    display(
        approved_candidates[
            [
                "Family",
                "N_Independent_Species",
                "N_Independent_Genera",
                "Species_Represented",
                "Genera_Represented",
                "Median_Seed_Mass_mg",
                "Candidate_Median_Log10_Seed_Mass",
                "Approved_Family_Seed_Mass_Candidate_mg",
                "Raw_Fold_Range",
                "Log10_Range",
                "Residual_Genera",
                "N_Residual_Canonical_Taxa",
                "Family_Quality_Flag",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 13. HIGH-DISPERSION EXPLICIT REVIEW
# =============================================================================

high_dispersion_review = (
    family_decisions[
        family_decisions[
            "Family_Transfer_Decision"
        ]
        .eq(
            "HIGH_DISPERSION_EXPLICIT_REVIEW"
        )
    ]
    .copy()
)


print("\nStrong families withheld because of high dispersion")
print("-" * 100)

if len(
    high_dispersion_review
) > 0:

    display(
        high_dispersion_review[
            [
                "Family",
                "N_Independent_Species",
                "N_Independent_Genera",
                "Species_Represented",
                "Median_Seed_Mass_mg",
                "Candidate_Family_Seed_Mass_mg",
                "Raw_Fold_Range",
                "Log10_Range",
                "Residual_Genera",
                "Family_Transfer_Decision",
                "Family_Quality_Flag",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 14. NON-ELIGIBLE FAMILIES
# =============================================================================

noneligible_families = (
    family_decisions[
        family_decisions[
            "Family_Transfer_Decision"
        ]
        .isin(
            [
                "INSUFFICIENT_FAMILY_EVIDENCE",
                "NO_FAMILY_EVIDENCE",
            ]
        )
    ]
    .copy()
)


print("\nFamilies not eligible for family-level transfer")
print("-" * 100)

display(
    noneligible_families[
        [
            "Family",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "Family_Evidence_Class",
            "Family_Transfer_Decision",
        ]
    ]
)


# =============================================================================
# 15. EXPECTED DECISION VALIDATION
# =============================================================================

approved_family_names = set(
    approved_candidates[
        "Family"
    ]
)


assert approved_family_names == {
    "Malvaceae"
}, (
    "Expected Malvaceae to be the only automatically approved "
    "family-transfer candidate under FTR_01.\n"
    f"Observed approved families: {sorted(approved_family_names)}"
)


high_dispersion_names = set(
    high_dispersion_review[
        "Family"
    ]
)


assert high_dispersion_names == {
    "Fabaceae"
}, (
    "Expected Fabaceae to be the only strong family withheld "
    "for high-dispersion review."
)


# =============================================================================
# 16. VALIDATE MALVACEAE CANDIDATE ESTIMATOR
# =============================================================================

malvaceae_evidence = (
    family_evidence[
        family_evidence[
            "Family"
        ].eq(
            "Malvaceae"
        )
    ]
    .copy()
)


assert len(
    malvaceae_evidence
) == 8


malvaceae_log_median = (
    np.log10(
        malvaceae_evidence[
            "Seed_Mass_mg"
        ]
    )
    .median()
)


malvaceae_expected_candidate = (
    10 ** malvaceae_log_median
)


malvaceae_observed_candidate = (
    approved_candidates.loc[
        approved_candidates[
            "Family"
        ].eq(
            "Malvaceae"
        ),
        "Approved_Family_Seed_Mass_Candidate_mg"
    ]
    .iloc[0]
)


assert np.isclose(
    malvaceae_observed_candidate,
    malvaceae_expected_candidate,
    rtol=0,
    atol=1e-9
)


print("\nMalvaceae candidate estimator")
print("-" * 100)

print(
    f"N independent species             : "
    f"{len(malvaceae_evidence)}"
)

print(
    f"Arithmetic raw median             : "
    f"{malvaceae_evidence['Seed_Mass_mg'].median():.6f} mg"
)

print(
    f"Median log10 Seed Mass            : "
    f"{malvaceae_log_median:.6f}"
)

print(
    f"Back-transformed log10 median     : "
    f"{malvaceae_expected_candidate:.6f} mg"
)


# =============================================================================
# 17. CANONICAL CANDIDATE PREVIEW
# =============================================================================
#
# This is a PREVIEW ONLY.
#
# It identifies which unresolved canonical taxa WOULD receive the approved
# Malvaceae family candidate if 9.4R-R4 authorizes propagation.
#
# =============================================================================

approved_family_lookup = (
    approved_candidates[
        [
            "Family",
            "Approved_Family_Seed_Mass_Candidate_mg",
            "Family_Transfer_Rule_ID",
            "Family_Quality_Flag",
        ]
    ]
    .copy()
)


canonical_candidate_preview = (
    canonical_family_map
    .merge(
        approved_family_lookup,
        on="Family",
        how="inner",
        validate="many_to_one"
    )
)


canonical_candidate_preview[
    "Candidate_Assignment_Status"
] = (
    "PREVIEW_ONLY_NOT_YET_ASSIGNED"
)


canonical_candidate_preview[
    "Numerical_Seed_Mass_Assigned"
] = False


print("\nCanonical taxa potentially eligible for approved family candidate")
print("-" * 100)

display(
    canonical_candidate_preview[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Family",
            "Approved_Family_Seed_Mass_Candidate_mg",
            "Family_Transfer_Rule_ID",
            "Family_Quality_Flag",
            "Candidate_Assignment_Status",
        ]
    ]
)


# =============================================================================
# 18. VALIDATE CANONICAL PREVIEW
# =============================================================================
#
# Malvaceae residual structure from 9.4R-R1:
#
#     Cola species
#     Grewia pubescens
#
# Therefore exactly two canonical taxa should appear in the preview.
#
# =============================================================================

assert len(
    canonical_candidate_preview
) == 2, (
    f"Expected 2 Malvaceae residual canonical taxa in family "
    f"candidate preview; found {len(canonical_candidate_preview)}."
)


assert set(
    canonical_candidate_preview[
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Grewia pubescens",
}


assert (
    canonical_candidate_preview[
        "Numerical_Seed_Mass_Assigned"
    ]
    .eq(False)
    .all()
)


# =============================================================================
# 19. DECISION SUMMARY
# =============================================================================

decision_summary = (
    family_decisions[
        "Family_Transfer_Decision"
    ]
    .value_counts()
    .rename_axis(
        "Family_Transfer_Decision"
    )
    .reset_index(
        name="N_Families"
    )
)


print("\nFamily-transfer decision summary")
print("-" * 100)

display(
    decision_summary
)


# =============================================================================
# 20. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4R_R3 = 90

N_VALUES_ASSIGNED_9_4R_R3 = 0

N_RESOLVED_AFTER_9_4R_R3 = (
    N_RESOLVED_BEFORE_9_4R_R3
    +
    N_VALUES_ASSIGNED_9_4R_R3
)


assert N_RESOLVED_AFTER_9_4R_R3 == 90


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4R_R3
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 21. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Target families evaluated",
            "Approved automatic family-transfer candidates",
            "Strong high-dispersion families withheld",
            "Families with insufficient family evidence",
            "Families with no family evidence",
            "Canonical taxa in approved family-candidate preview",
            "Family Seed Mass values assigned in 9.4R-R3",
            "Resolved canonical taxa after 9.4R-R3",
            "Overall Seed Mass coverage percent",
        ],

        "Value": [
            len(
                family_decisions
            ),

            len(
                approved_candidates
            ),

            len(
                high_dispersion_review
            ),

            int(
                family_decisions[
                    "Family_Transfer_Decision"
                ]
                .eq(
                    "INSUFFICIENT_FAMILY_EVIDENCE"
                )
                .sum()
            ),

            int(
                family_decisions[
                    "Family_Transfer_Decision"
                ]
                .eq(
                    "NO_FAMILY_EVIDENCE"
                )
                .sum()
            ),

            len(
                canonical_candidate_preview
            ),

            0,

            N_RESOLVED_AFTER_9_4R_R3,

            coverage_percent,
        ],
    }
)


print("\nSection 9.4R-R3 summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 22. EXPORT
# =============================================================================

transfer_rules.to_csv(
    OUT_TRANSFER_RULES,
    index=False
)


family_decisions.to_csv(
    OUT_FAMILY_TRANSFER_DECISIONS,
    index=False
)


approved_candidates.to_csv(
    OUT_APPROVED_FAMILY_CANDIDATES,
    index=False
)


high_dispersion_review.to_csv(
    OUT_HIGH_DISPERSION_REVIEW,
    index=False
)


noneligible_families.to_csv(
    OUT_LIMITED_OR_NO_EVIDENCE,
    index=False
)


canonical_candidate_preview.to_csv(
    OUT_CANONICAL_CANDIDATE_PREVIEW,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 23. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R-R3 exports")
print("-" * 100)

print(
    f"Transfer rules             : "
    f"{OUT_TRANSFER_RULES}"
)

print(
    f"Family transfer decisions  : "
    f"{OUT_FAMILY_TRANSFER_DECISIONS}"
)

print(
    f"Approved family candidates : "
    f"{OUT_APPROVED_FAMILY_CANDIDATES}"
)

print(
    f"High-dispersion review     : "
    f"{OUT_HIGH_DISPERSION_REVIEW}"
)

print(
    f"Noneligible families       : "
    f"{OUT_LIMITED_OR_NO_EVIDENCE}"
)

print(
    f"Canonical candidate preview: "
    f"{OUT_CANONICAL_CANDIDATE_PREVIEW}"
)

print(
    f"Summary                    : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 24. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Family-level Seed Mass transfer-rule development "
    "completed successfully."
)

print(
    f"Target families evaluated                 : "
    f"{len(family_decisions)}"
)

print(
    f"Approved family-transfer candidates       : "
    f"{len(approved_candidates)}"
)

print(
    f"High-dispersion strong families withheld  : "
    f"{len(high_dispersion_review)}"
)

print(
    f"Canonical taxa in approved-family preview : "
    f"{len(canonical_candidate_preview)}"
)

print(
    "Family Seed Mass values assigned          : 0"
)

print(
    f"Overall Seed Mass coverage remains        : "
    f"{N_RESOLVED_AFTER_9_4R_R3} / "
    f"{N_CANONICAL_TAXA} "
    f"({coverage_percent:.2f}%)"
)

print(
    "Malvaceae is the only family currently approved "
    "for family-level candidate propagation."
)

print(
    "Fabaceae remains withheld pending explicit "
    "high-dispersion review."
)

print(
    "Canonical propagation, if approved, is deferred "
    "to Section 9.4R-R4."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R4
# APPROVED FAMILY-LEVEL SEED MASS ASSIGNMENT
# =============================================================================
#
# Purpose
# -------
# 1. Load the approved family candidate(s) from Section 9.4R-R3.
# 2. Propagate ONLY approved family-level Seed Mass values to eligible
#    unresolved canonical taxa.
# 3. Preserve full provenance and quality information.
# 4. Confirm that no non-approved family is assigned a value.
# 5. Recalculate Seed Mass coverage after family assignment.
#
# Current approved family state from 9.4R-R3
# ------------------------------------------------------------
# Approved family             : Malvaceae
# Approved family candidate   : 420.240217 mg
# Eligible canonical taxa     :
#       Cola species
#       Grewia pubescens
#
# Fabaceae:
#       withheld because of high family-level dispersion.
#
# IMPORTANT
# ---------
# No other family receives a value in this section.
#
# Expected coverage transition
# ----------------------------
# Before family assignment : 90 / 127
# New family assignments   :  2
# After family assignment  : 92 / 127
# Coverage                 : 72.44%
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

APPROVED_FAMILY_CANDIDATES_FILE = (
    OUT_DIR
    / "omo_seed_mass_approved_family_transfer_candidates.csv"
)

CANONICAL_CANDIDATE_PREVIEW_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_candidate_canonical_preview.csv"
)

FAMILY_TRANSFER_DECISIONS_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_decisions.csv"
)

POST_9_4Q_RESIDUAL_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_single_congener_review_residual_state.csv"
)

CANONICAL_FAMILY_MAP_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)

HIGH_DISPERSION_REVIEW_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_high_dispersion_review.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_FAMILY_ASSIGNMENTS = (
    OUT_DIR
    / "omo_seed_mass_family_level_assignments.csv"
)

OUT_CANONICAL_FAMILY_ASSIGNMENTS = (
    OUT_DIR
    / "omo_seed_mass_canonical_family_assignments.csv"
)

OUT_POST_FAMILY_STATE = (
    OUT_DIR
    / "omo_seed_mass_post_family_assignment_state.csv"
)

OUT_POST_FAMILY_RESIDUAL = (
    OUT_DIR
    / "omo_seed_mass_post_family_assignment_residual_canonical_taxa.csv"
)

OUT_HIGH_DISPERSION_PENDING = (
    OUT_DIR
    / "omo_seed_mass_high_dispersion_family_pending_targets.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_family_assignment_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    APPROVED_FAMILY_CANDIDATES_FILE,
    CANONICAL_CANDIDATE_PREVIEW_FILE,
    FAMILY_TRANSFER_DECISIONS_FILE,
    POST_9_4Q_RESIDUAL_FILE,
    CANONICAL_FAMILY_MAP_FILE,
    HIGH_DISPERSION_REVIEW_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4R-R4 input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

approved_family_candidates = pd.read_csv(
    APPROVED_FAMILY_CANDIDATES_FILE,
    low_memory=False
)

canonical_preview = pd.read_csv(
    CANONICAL_CANDIDATE_PREVIEW_FILE,
    low_memory=False
)

family_transfer_decisions = pd.read_csv(
    FAMILY_TRANSFER_DECISIONS_FILE,
    low_memory=False
)

post_q_residual = pd.read_csv(
    POST_9_4Q_RESIDUAL_FILE,
    low_memory=False
)

canonical_family_map = pd.read_csv(
    CANONICAL_FAMILY_MAP_FILE,
    low_memory=False
)

high_dispersion_review = pd.read_csv(
    HIGH_DISPERSION_REVIEW_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R-R4 — APPROVED FAMILY-LEVEL "
    "SEED MASS ASSIGNMENT"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Approved family candidates          : "
    f"{len(approved_family_candidates)}"
)

print(
    f"Canonical family candidate preview  : "
    f"{len(canonical_preview)}"
)

print(
    f"Family transfer-decision rows       : "
    f"{len(family_transfer_decisions)}"
)

print(
    f"Post-9.4Q residual canonical taxa   : "
    f"{len(post_q_residual)}"
)

print(
    f"Canonical family-map rows           : "
    f"{len(canonical_family_map)}"
)

print(
    f"High-dispersion review families     : "
    f"{len(high_dispersion_review)}"
)


# =============================================================================
# 4. VALIDATE AUTHORITATIVE R3 STATE
# =============================================================================

assert len(
    approved_family_candidates
) == 1, (
    f"Expected exactly one approved family candidate; "
    f"found {len(approved_family_candidates)}."
)


assert set(
    approved_family_candidates[
        "Family"
    ]
) == {
    "Malvaceae"
}


assert len(
    canonical_preview
) == 2, (
    f"Expected exactly two canonical candidate-preview rows; "
    f"found {len(canonical_preview)}."
)


assert set(
    canonical_preview[
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Grewia pubescens",
}


assert len(
    post_q_residual
) == 37


assert len(
    canonical_family_map
) == 31


assert len(
    high_dispersion_review
) == 1


assert set(
    high_dispersion_review[
        "Family"
    ]
) == {
    "Fabaceae"
}


# =============================================================================
# 5. VALIDATE APPROVED MALVACEAE CANDIDATE
# =============================================================================

approved_family_candidates[
    "Approved_Family_Seed_Mass_Candidate_mg"
] = pd.to_numeric(
    approved_family_candidates[
        "Approved_Family_Seed_Mass_Candidate_mg"
    ],
    errors="coerce"
)


assert approved_family_candidates[
    "Approved_Family_Seed_Mass_Candidate_mg"
].notna().all()


assert approved_family_candidates[
    "Approved_Family_Seed_Mass_Candidate_mg"
].gt(
    0
).all()


malvaceae_candidate = (
    approved_family_candidates.loc[
        approved_family_candidates[
            "Family"
        ].eq(
            "Malvaceae"
        ),
        "Approved_Family_Seed_Mass_Candidate_mg"
    ]
    .iloc[0]
)


assert np.isclose(
    malvaceae_candidate,
    420.240217,
    rtol=0,
    atol=1e-6
), (
    f"Unexpected Malvaceae candidate value: "
    f"{malvaceae_candidate}"
)


print("\nApproved family candidate")
print("-" * 100)

print(
    f"Family                     : Malvaceae"
)

print(
    f"Approved Seed Mass         : "
    f"{malvaceae_candidate:.6f} mg"
)

print(
    "Estimator                  : "
    "back-transformed median log10 species Seed Mass"
)


# =============================================================================
# 6. BUILD FAMILY-LEVEL ASSIGNMENT TABLE
# =============================================================================
#
# One row per approved family.
#
# =============================================================================

family_assignments = (
    approved_family_candidates
    [
        [
            "Family",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Species_Represented",
            "Genera_Represented",
            "Median_Seed_Mass_mg",
            "Candidate_Median_Log10_Seed_Mass",
            "Approved_Family_Seed_Mass_Candidate_mg",
            "Raw_Fold_Range",
            "Log10_Range",
            "Residual_Genera",
            "N_Residual_Canonical_Taxa",
            "Family_Transfer_Rule_ID",
            "Family_Quality_Flag",
        ]
    ]
    .copy()
)


family_assignments[
    "Seed_Mass_Assignment_Level"
] = (
    "FAMILY_LEVEL"
)


family_assignments[
    "Seed_Mass_Provenance_Class"
] = (
    "FAMILY_ESTIMATE"
)


family_assignments[
    "Seed_Mass_Assignment_Status"
] = (
    "APPROVED_AND_PROPAGATED"
)


family_assignments[
    "Numerical_Seed_Mass_Assigned"
] = True


# =============================================================================
# 7. BUILD CANONICAL FAMILY ASSIGNMENTS
# =============================================================================

canonical_family_assignments = (
    canonical_preview
    .copy()
)


canonical_family_assignments[
    "Assigned_Seed_Mass_mg"
] = pd.to_numeric(
    canonical_family_assignments[
        "Approved_Family_Seed_Mass_Candidate_mg"
    ],
    errors="coerce"
)


canonical_family_assignments[
    "Seed_Mass_Assignment_Level"
] = (
    "FAMILY_LEVEL"
)


canonical_family_assignments[
    "Seed_Mass_Provenance_Class"
] = (
    "FAMILY_ESTIMATE"
)


canonical_family_assignments[
    "Seed_Mass_Quality_Flag"
] = (
    canonical_family_assignments[
        "Family_Quality_Flag"
    ]
)


canonical_family_assignments[
    "Seed_Mass_Assignment_Method"
] = (
    "BACK_TRANSFORMED_MEDIAN_LOG10_SPECIES_SEED_MASS"
)


canonical_family_assignments[
    "Seed_Mass_Assignment_Status"
] = (
    "ASSIGNED_IN_SECTION_9_4R_R4"
)


canonical_family_assignments[
    "Numerical_Seed_Mass_Assigned"
] = True


# =============================================================================
# 8. VALIDATE CANONICAL FAMILY ASSIGNMENTS
# =============================================================================

assert len(
    canonical_family_assignments
) == 2


assert canonical_family_assignments[
    "Canonical_Species"
].is_unique


assert canonical_family_assignments[
    "Assigned_Seed_Mass_mg"
].notna().all()


assert canonical_family_assignments[
    "Assigned_Seed_Mass_mg"
].gt(
    0
).all()


assert canonical_family_assignments[
    "Family"
].eq(
    "Malvaceae"
).all()


assert canonical_family_assignments[
    "Seed_Mass_Provenance_Class"
].eq(
    "FAMILY_ESTIMATE"
).all()


assert canonical_family_assignments[
    "Numerical_Seed_Mass_Assigned"
].eq(
    True
).all()


print("\nApproved canonical family assignments")
print("-" * 100)

display(
    canonical_family_assignments[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Family",
            "Assigned_Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Family_Transfer_Rule_ID",
            "Seed_Mass_Assignment_Method",
        ]
    ]
)


# =============================================================================
# 9. VERIFY NO NON-APPROVED FAMILY IS ASSIGNED
# =============================================================================

approved_family_set = set(
    family_assignments[
        "Family"
    ]
)


assert approved_family_set == {
    "Malvaceae"
}


nonapproved_family_set = (
    set(
        family_transfer_decisions[
            "Family"
        ]
    )
    -
    approved_family_set
)


assert "Fabaceae" in nonapproved_family_set


assert not (
    set(
        canonical_family_assignments[
            "Family"
        ]
    )
    &
    nonapproved_family_set
), (
    "A non-approved family received a numerical assignment."
)


# =============================================================================
# 10. VERIFY ASSIGNED CANONICAL TAXA WERE UNRESOLVED BEFORE R4
# =============================================================================

pre_r4_residual_set = set(
    post_q_residual[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assigned_canonical_set = set(
    canonical_family_assignments[
        "Canonical_Species"
    ]
)


assert assigned_canonical_set.issubset(
    pre_r4_residual_set
)


# =============================================================================
# 11. BUILD POST-FAMILY-ASSIGNMENT STATE
# =============================================================================

post_family_state = (
    post_q_residual
    .copy()
)


post_family_state[
    "Seed_Mass_Assigned_After_9_4R_R4"
] = False


post_family_state[
    "Assigned_Seed_Mass_mg"
] = np.nan


post_family_state[
    "Seed_Mass_Assignment_Level"
] = pd.NA


post_family_state[
    "Seed_Mass_Provenance_Class"
] = pd.NA


post_family_state[
    "Seed_Mass_Quality_Flag"
] = pd.NA


assignment_lookup = (
    canonical_family_assignments[
        [
            "Canonical_Species",
            "Assigned_Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Family",
            "Family_Transfer_Rule_ID",
            "Seed_Mass_Assignment_Method",
        ]
    ]
    .copy()
)


# -----------------------------------------------------------------------------
# Remove any inherited fields with the same names before authoritative merge.
# -----------------------------------------------------------------------------

authoritative_fields = [
    col
    for col in assignment_lookup.columns
    if col != "Canonical_Species"
]


preexisting_overlap = [
    col
    for col in authoritative_fields
    if col in post_family_state.columns
]


if preexisting_overlap:

    post_family_state = (
        post_family_state
        .drop(
            columns=preexisting_overlap,
            errors="ignore"
        )
    )


post_family_state = (
    post_family_state
    .merge(
        assignment_lookup,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


assigned_mask = (
    post_family_state[
        "Assigned_Seed_Mass_mg"
    ]
    .notna()
)


post_family_state.loc[
    assigned_mask,
    "Seed_Mass_Assigned_After_9_4R_R4"
] = True


# =============================================================================
# 12. VALIDATE POST-FAMILY ASSIGNED STATE
# =============================================================================

post_family_assigned = (
    post_family_state[
        post_family_state[
            "Seed_Mass_Assigned_After_9_4R_R4"
        ]
    ]
    .copy()
)


assert len(
    post_family_assigned
) == 2


assert set(
    post_family_assigned[
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Grewia pubescens",
}


# =============================================================================
# 13. BUILD TRUE POST-FAMILY RESIDUAL SET
# =============================================================================

post_family_residual = (
    post_family_state[
        ~post_family_state[
            "Seed_Mass_Assigned_After_9_4R_R4"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(
    post_family_residual
) == 35, (
    f"Expected 35 residual canonical taxa after family assignment; "
    f"found {len(post_family_residual)}."
)


print("\nPost-family-assignment residual state")
print("-" * 100)

print(
    f"Residual canonical taxa before R4 : "
    f"{len(post_q_residual)}"
)

print(
    f"New family-level assignments      : "
    f"{len(post_family_assigned)}"
)

print(
    f"Residual canonical taxa after R4  : "
    f"{len(post_family_residual)}"
)


# =============================================================================
# 14. IDENTIFY FABACEAE HIGH-DISPERSION PENDING TARGETS
# =============================================================================

fabaceae_targets = (
    canonical_family_map[
        canonical_family_map[
            "Family"
        ].eq(
            "Fabaceae"
        )
    ]
    .copy()
)


fabaceae_pending_set = set(
    fabaceae_targets[
        "Canonical_Species"
    ]
)


high_dispersion_pending = (
    post_family_residual[
        post_family_residual[
            "Canonical_Species"
        ].isin(
            fabaceae_pending_set
        )
    ]
    .copy()
)


high_dispersion_pending[
    "Family"
] = "Fabaceae"


high_dispersion_pending[
    "Family_Transfer_Status"
] = (
    "WITHHELD_HIGH_DISPERSION"
)


high_dispersion_pending[
    "Next_Action_After_R4"
] = (
    "EXPLICIT_FABACEAE_HIGH_DISPERSION_REVIEW"
)


assert len(
    high_dispersion_pending
) == 5, (
    f"Expected 5 Fabaceae canonical residual taxa; "
    f"found {len(high_dispersion_pending)}."
)


print("\nFabaceae targets retained for explicit high-dispersion review")
print("-" * 100)

display(
    high_dispersion_pending[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Family",
            "Family_Transfer_Status",
            "Next_Action_After_R4",
        ]
    ]
)


# =============================================================================
# 15. CHECK MALVACEAE TARGETS ARE NO LONGER RESIDUAL
# =============================================================================

remaining_residual_set = set(
    post_family_residual[
        "Canonical_Species"
    ]
)


assert "Cola species" not in remaining_residual_set

assert "Grewia pubescens" not in remaining_residual_set


# =============================================================================
# 16. COVERAGE UPDATE
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_R4 = 90

N_NEW_FAMILY_ASSIGNMENTS = len(
    post_family_assigned
)

N_RESOLVED_AFTER_R4 = (
    N_RESOLVED_BEFORE_R4
    +
    N_NEW_FAMILY_ASSIGNMENTS
)


assert N_NEW_FAMILY_ASSIGNMENTS == 2

assert N_RESOLVED_AFTER_R4 == 92


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_R4
    /
    N_CANONICAL_TAXA
)


residual_percent = (
    100
    *
    len(
        post_family_residual
    )
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 17. POST-R4 ROUTING SUMMARY
# =============================================================================
#
# Preserve the broad residual routing inherited from 9.4Q.
#
# Family-level assignment removed two taxa from the former family-audit pool.
#
# =============================================================================

if "Post_9_4Q_Route" in post_family_residual.columns:

    residual_route_summary = (
        post_family_residual[
            "Post_9_4Q_Route"
        ]
        .value_counts()
        .rename_axis(
            "Post_R4_Residual_Route"
        )
        .reset_index(
            name="N_Canonical_Taxa"
        )
    )

    print("\nPost-R4 residual routing")
    print("-" * 100)

    display(
        residual_route_summary
    )


# =============================================================================
# 18. FAMILY ASSIGNMENT SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Canonical taxa before family assignment",
            "Approved family-level assignments",
            "Approved family represented",
            "Family Seed Mass estimate mg",
            "Canonical taxa resolved after family assignment",
            "Residual canonical taxa after family assignment",
            "Fabaceae high-dispersion canonical taxa retained",
            "Overall Seed Mass coverage percent",
            "Residual canonical percent",
        ],

        "Value": [
            len(
                post_q_residual
            ),

            N_NEW_FAMILY_ASSIGNMENTS,

            "Malvaceae",

            malvaceae_candidate,

            N_RESOLVED_AFTER_R4,

            len(
                post_family_residual
            ),

            len(
                high_dispersion_pending
            ),

            coverage_percent,

            residual_percent,
        ],
    }
)


print("\nSection 9.4R-R4 summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 19. STRONG RECONCILIATION
# =============================================================================

assert (
    N_RESOLVED_AFTER_R4
    +
    len(
        post_family_residual
    )
    ==
    N_CANONICAL_TAXA
)


assert (
    92
    +
    35
    ==
    127
)


assert np.isclose(
    coverage_percent,
    72.44094488188976,
    rtol=0,
    atol=1e-9
)


# =============================================================================
# 20. EXPORT
# =============================================================================

family_assignments.to_csv(
    OUT_FAMILY_ASSIGNMENTS,
    index=False
)


canonical_family_assignments.to_csv(
    OUT_CANONICAL_FAMILY_ASSIGNMENTS,
    index=False
)


post_family_state.to_csv(
    OUT_POST_FAMILY_STATE,
    index=False
)


post_family_residual.to_csv(
    OUT_POST_FAMILY_RESIDUAL,
    index=False
)


high_dispersion_pending.to_csv(
    OUT_HIGH_DISPERSION_PENDING,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 21. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R-R4 exports")
print("-" * 100)

print(
    f"Family assignments          : "
    f"{OUT_FAMILY_ASSIGNMENTS}"
)

print(
    f"Canonical family assignments: "
    f"{OUT_CANONICAL_FAMILY_ASSIGNMENTS}"
)

print(
    f"Post-family state           : "
    f"{OUT_POST_FAMILY_STATE}"
)

print(
    f"Post-family residual        : "
    f"{OUT_POST_FAMILY_RESIDUAL}"
)

print(
    f"Fabaceae pending review     : "
    f"{OUT_HIGH_DISPERSION_PENDING}"
)

print(
    f"Summary                     : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 22. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Approved family-level Seed Mass assignment "
    "completed successfully."
)

print(
    f"Approved family                     : "
    f"Malvaceae"
)

print(
    f"Family Seed Mass estimate           : "
    f"{malvaceae_candidate:.6f} mg"
)

print(
    f"Canonical taxa assigned             : "
    f"{N_NEW_FAMILY_ASSIGNMENTS}"
)

print(
    "Assigned taxa                      : "
    "Cola species | Grewia pubescens"
)

print(
    f"Resolved canonical taxa             : "
    f"{N_RESOLVED_AFTER_R4} / {N_CANONICAL_TAXA}"
)

print(
    f"Overall Seed Mass coverage          : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Residual canonical taxa             : "
    f"{len(post_family_residual)}"
)

print(
    f"Fabaceae high-dispersion taxa       : "
    f"{len(high_dispersion_pending)}"
)

print(
    "No non-approved family received a Seed Mass value."
)

print(
    "Fabaceae remains unresolved pending explicit "
    "high-dispersion review."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4R-R5
# EXPLICIT FABACEAE HIGH-DISPERSION SEED MASS REVIEW
# =============================================================================
#
# Purpose
# -------
# Section 9.4R-R3 classified Fabaceae as:
#
#     STRONG FAMILY EVIDENCE
#     11 independent species
#     10 independent genera
#     Raw Seed Mass range ≈ 666.7-fold
#     Log10 range ≈ 2.824
#
# Therefore a whole-family estimate was withheld.
#
# Section 9.4R-R5 now asks:
#
#   1. Which Fabaceae species and genera constitute the evidence pool?
#   2. Do any of the 5 unresolved Fabaceae target genera have SAME-GENUS
#      accepted Seed Mass evidence?
#   3. If same-genus evidence exists, how many independent biological
#      species support that genus?
#   4. Is there enough same-genus support to justify a genus estimate?
#   5. If only one congener exists, preserve it as contextual evidence but
#      DO NOT transfer automatically.
#   6. If no same-genus evidence exists, retain the target unresolved because
#      the whole-family distribution is too heterogeneous.
#
# Conservative decision rule
# --------------------------
#
#   >= 2 independent same-genus species:
#       CANDIDATE_FOR_GENUS_LEVEL_ESTIMATE
#
#   1 independent same-genus species:
#       SINGLE_CONGENER_ONLY_NO_AUTOMATIC_TRANSFER
#
#   0 same-genus species:
#       NO_CLOSE_GENUS_EVIDENCE_FAMILY_TOO_DISPERSED
#
# IMPORTANT
# ---------
# No family-wide Fabaceae estimate is assigned.
#
# Unless >=2 independent same-genus species are unexpectedly found,
# NO new numerical Seed Mass value will be assigned in this section.
#
# Expected current state:
#
#     Fabaceae pending canonical taxa : 5
#     Overall resolved taxa           : 92 / 127
#     Overall coverage                : 72.44%
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

FABACEAE_PENDING_FILE = (
    OUT_DIR
    / "omo_seed_mass_high_dispersion_family_pending_targets.csv"
)

COMPLETE_SPECIES_FAMILY_FILE = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_family_mapping_complete.csv"
)

FAMILY_TRANSFER_DECISIONS_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_decisions.csv"
)

POST_FAMILY_RESIDUAL_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_family_assignment_residual_canonical_taxa.csv"
)

SINGLE_CONGENER_REVIEW_FILE = (
    OUT_DIR
    / "omo_seed_mass_single_congener_review_decisions.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_FABACEAE_EVIDENCE = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_evidence_detail.csv"
)

OUT_FABACEAE_GENUS_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_evidence_genus_summary.csv"
)

OUT_FABACEAE_TARGET_REVIEW = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_high_dispersion_target_review.csv"
)

OUT_FABACEAE_SINGLE_CONGENER_CONTEXT = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_single_congener_context.csv"
)

OUT_FABACEAE_UNRESOLVED = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_post_review_unresolved_targets.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_high_dispersion_review_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    FABACEAE_PENDING_FILE,
    COMPLETE_SPECIES_FAMILY_FILE,
    FAMILY_TRANSFER_DECISIONS_FILE,
    POST_FAMILY_RESIDUAL_FILE,
    SINGLE_CONGENER_REVIEW_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4R-R5 input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

fabaceae_pending = pd.read_csv(
    FABACEAE_PENDING_FILE,
    low_memory=False
)

species_family = pd.read_csv(
    COMPLETE_SPECIES_FAMILY_FILE,
    low_memory=False
)

family_decisions = pd.read_csv(
    FAMILY_TRANSFER_DECISIONS_FILE,
    low_memory=False
)

post_family_residual = pd.read_csv(
    POST_FAMILY_RESIDUAL_FILE,
    low_memory=False
)

single_congener_review = pd.read_csv(
    SINGLE_CONGENER_REVIEW_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4R-R5 — EXPLICIT FABACEAE "
    "HIGH-DISPERSION SEED MASS REVIEW"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Fabaceae pending canonical taxa : "
    f"{len(fabaceae_pending)}"
)

print(
    f"Complete evidence species       : "
    f"{len(species_family)}"
)

print(
    f"Family decision rows            : "
    f"{len(family_decisions)}"
)

print(
    f"Post-family residual taxa       : "
    f"{len(post_family_residual)}"
)


# =============================================================================
# 4. VALIDATE AUTHORITATIVE R4 STATE
# =============================================================================

assert len(
    fabaceae_pending
) == 5, (
    f"Expected 5 Fabaceae pending canonical taxa; "
    f"found {len(fabaceae_pending)}."
)


assert set(
    fabaceae_pending[
        "Residual_Genus"
    ]
) == {
    "Amphimas",
    "Baphia",
    "Lonchocarpus",
    "Millettia",
    "Pterocarpus",
}


assert len(
    post_family_residual
) == 35


fabaceae_family_row = (
    family_decisions[
        family_decisions[
            "Family"
        ].eq(
            "Fabaceae"
        )
    ]
    .copy()
)


assert len(
    fabaceae_family_row
) == 1


assert (
    fabaceae_family_row[
        "Family_Transfer_Decision"
    ]
    .eq(
        "HIGH_DISPERSION_EXPLICIT_REVIEW"
    )
    .all()
)


# =============================================================================
# 5. EXTRACT AUTHORITATIVE FABACEAE EVIDENCE
# =============================================================================

species_family[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_family[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


fabaceae_evidence = (
    species_family[
        species_family[
            "Family"
        ].eq(
            "Fabaceae"
        )
    ]
    .copy()
    .sort_values(
        [
            "Biological_Genus",
            "Biological_Species_Name",
        ]
    )
    .reset_index(drop=True)
)


assert len(
    fabaceae_evidence
) == 11, (
    f"Expected 11 independent Fabaceae evidence species; "
    f"found {len(fabaceae_evidence)}."
)


assert (
    fabaceae_evidence[
        "Biological_Genus"
    ].nunique()
    ==
    10
)


assert fabaceae_evidence[
    "Seed_Mass_mg"
].notna().all()


assert fabaceae_evidence[
    "Seed_Mass_mg"
].gt(0).all()


fabaceae_evidence[
    "Log10_Seed_Mass"
] = np.log10(
    fabaceae_evidence[
        "Seed_Mass_mg"
    ]
)


print("\nValidated Fabaceae evidence pool")
print("-" * 100)

display(
    fabaceae_evidence[
        [
            "Biological_Genus",
            "Biological_Species_Name",
            "Seed_Mass_mg",
            "Log10_Seed_Mass",
            "Evidence_Sources",
        ]
    ]
)


# =============================================================================
# 6. FAMILY-LEVEL DIAGNOSTICS
# =============================================================================

FAB_MIN = (
    fabaceae_evidence[
        "Seed_Mass_mg"
    ]
    .min()
)

FAB_MEDIAN = (
    fabaceae_evidence[
        "Seed_Mass_mg"
    ]
    .median()
)

FAB_MAX = (
    fabaceae_evidence[
        "Seed_Mass_mg"
    ]
    .max()
)

FAB_LOG_MEDIAN = (
    fabaceae_evidence[
        "Log10_Seed_Mass"
    ]
    .median()
)

FAB_LOG_MIN = (
    fabaceae_evidence[
        "Log10_Seed_Mass"
    ]
    .min()
)

FAB_LOG_MAX = (
    fabaceae_evidence[
        "Log10_Seed_Mass"
    ]
    .max()
)

FAB_FOLD_RANGE = (
    FAB_MAX
    /
    FAB_MIN
)

FAB_LOG_RANGE = (
    FAB_LOG_MAX
    -
    FAB_LOG_MIN
)


print("\nFabaceae dispersion checkpoint")
print("-" * 100)

print(
    f"Independent species : {len(fabaceae_evidence)}"
)

print(
    f"Independent genera  : "
    f"{fabaceae_evidence['Biological_Genus'].nunique()}"
)

print(
    f"Minimum Seed Mass   : {FAB_MIN:.6f} mg"
)

print(
    f"Median Seed Mass    : {FAB_MEDIAN:.6f} mg"
)

print(
    f"Maximum Seed Mass   : {FAB_MAX:.6f} mg"
)

print(
    f"Raw fold range      : {FAB_FOLD_RANGE:.6f}"
)

print(
    f"Log10 range         : {FAB_LOG_RANGE:.6f}"
)


assert FAB_FOLD_RANGE > 100

assert FAB_LOG_RANGE >= 2


# =============================================================================
# 7. SUMMARIZE EVIDENCE BY FABACEAE GENUS
# =============================================================================

fabaceae_genus_summary = (
    fabaceae_evidence
    .groupby(
        "Biological_Genus",
        as_index=False
    )
    .agg(
        N_Independent_Species=(
            "Biological_Species_Name",
            "nunique"
        ),

        Species_Represented=(
            "Biological_Species_Name",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),

        Evidence_Sources=(
            "Evidence_Sources",
            lambda x: " | ".join(
                sorted(
                    x.dropna()
                    .astype(str)
                    .unique()
                )
            )
        ),
    )
    .sort_values(
        "Biological_Genus"
    )
    .reset_index(drop=True)
)


print("\nFabaceae evidence by genus")
print("-" * 100)

display(
    fabaceae_genus_summary
)


# =============================================================================
# 8. BUILD TARGET-GENUS REVIEW REGISTER
# =============================================================================
#
# IMPORTANT
# ---------
# Use R5_Next_Action rather than the generic Next_Action field.
#
# Earlier residual tables may already contain Next_Action fields inherited
# from Sections 9.4M–9.4R. A stage-specific name avoids pandas _x / _y
# collisions and preserves workflow provenance.
# =============================================================================

target_genera = sorted(
    fabaceae_pending[
        "Residual_Genus"
    ]
    .dropna()
    .astype(str)
    .str.strip()
    .unique()
)


review_rows = []


for genus in target_genera:

    genus_evidence = (
        fabaceae_evidence[
            fabaceae_evidence[
                "Biological_Genus"
            ].eq(
                genus
            )
        ]
        .copy()
    )


    n_same_genus_species = (
        genus_evidence[
            "Biological_Species_Name"
        ]
        .nunique()
    )


    if n_same_genus_species > 0:

        same_genus_species = " | ".join(
            sorted(
                genus_evidence[
                    "Biological_Species_Name"
                ]
                .dropna()
                .astype(str)
                .unique()
            )
        )


        same_genus_candidate = (
            genus_evidence[
                "Seed_Mass_mg"
            ]
            .median()
        )


        same_genus_min = (
            genus_evidence[
                "Seed_Mass_mg"
            ]
            .min()
        )


        same_genus_max = (
            genus_evidence[
                "Seed_Mass_mg"
            ]
            .max()
        )


        same_genus_sources = " | ".join(
            sorted(
                genus_evidence[
                    "Evidence_Sources"
                ]
                .dropna()
                .astype(str)
                .unique()
            )
        )


    else:

        same_genus_species = ""

        same_genus_candidate = np.nan

        same_genus_min = np.nan

        same_genus_max = np.nan

        same_genus_sources = ""


    # -------------------------------------------------------------------------
    # Conservative R5 review rule
    # -------------------------------------------------------------------------

    if n_same_genus_species >= 2:

        review_decision = (
            "CANDIDATE_FOR_GENUS_LEVEL_ESTIMATE"
        )

        r5_next_action = (
            "FORMAL_GENUS_ESTIMATE_REVIEW_REQUIRED"
        )


    elif n_same_genus_species == 1:

        review_decision = (
            "SINGLE_CONGENER_ONLY_NO_AUTOMATIC_TRANSFER"
        )

        r5_next_action = (
            "RETAIN_SINGLE_CONGENER_CONTEXT_AND_DEFER"
        )


    else:

        review_decision = (
            "NO_CLOSE_GENUS_EVIDENCE_FAMILY_TOO_DISPERSED"
        )

        r5_next_action = (
            "RETAIN_UNRESOLVED_AFTER_FABACEAE_REVIEW"
        )


    review_rows.append(
        {
            "Residual_Genus":
                genus,

            "N_Same_Genus_Independent_Species":
                n_same_genus_species,

            "Same_Genus_Species_Represented":
                same_genus_species,

            "Same_Genus_Evidence_Sources":
                same_genus_sources,

            "Same_Genus_Min_Seed_Mass_mg":
                same_genus_min,

            "Same_Genus_Candidate_Seed_Mass_mg":
                same_genus_candidate,

            "Same_Genus_Max_Seed_Mass_mg":
                same_genus_max,

            "Fabaceae_Family_Median_mg":
                FAB_MEDIAN,

            "Fabaceae_Raw_Fold_Range":
                FAB_FOLD_RANGE,

            "Fabaceae_Log10_Range":
                FAB_LOG_RANGE,

            "Fabaceae_Target_Review_Decision":
                review_decision,

            "R5_Next_Action":
                r5_next_action,
        }
    )


target_genus_review = pd.DataFrame(
    review_rows
)


# =============================================================================
# 8A. VALIDATE TARGET-GENUS REVIEW REGISTER
# =============================================================================

assert len(
    target_genus_review
) == 5, (
    f"Expected 5 Fabaceae target genera; "
    f"found {len(target_genus_review)}."
)


assert target_genus_review[
    "Residual_Genus"
].is_unique


assert set(
    target_genus_review[
        "Residual_Genus"
    ]
) == {
    "Amphimas",
    "Baphia",
    "Lonchocarpus",
    "Millettia",
    "Pterocarpus",
}


assert target_genus_review[
    "Fabaceae_Target_Review_Decision"
].notna().all()


assert target_genus_review[
    "R5_Next_Action"
].notna().all()


print("\nFabaceae target-genus evidence review")
print("-" * 100)

display(
    target_genus_review
)


# =============================================================================
# 9. MAP GENUS REVIEW TO FIVE CANONICAL TARGETS
# =============================================================================
#
# Remove only R5-owned fields if the cell is rerun in an interactive kernel.
# This prevents accidental suffix creation while preserving earlier workflow
# fields such as Next_Action_After_R4.
# =============================================================================

r5_fields = [
    "N_Same_Genus_Independent_Species",
    "Same_Genus_Species_Represented",
    "Same_Genus_Evidence_Sources",
    "Same_Genus_Min_Seed_Mass_mg",
    "Same_Genus_Candidate_Seed_Mass_mg",
    "Same_Genus_Max_Seed_Mass_mg",
    "Fabaceae_Family_Median_mg",
    "Fabaceae_Raw_Fold_Range",
    "Fabaceae_Log10_Range",
    "Fabaceae_Target_Review_Decision",
    "R5_Next_Action",
]


fabaceae_pending_for_merge = (
    fabaceae_pending
    .drop(
        columns=[
            col
            for col in r5_fields
            if col in fabaceae_pending.columns
        ],
        errors="ignore"
    )
    .copy()
)


fabaceae_target_review = (
    fabaceae_pending_for_merge
    .merge(
        target_genus_review,
        on="Residual_Genus",
        how="left",
        validate="many_to_one"
    )
)


# =============================================================================
# 9A. VALIDATE CANONICAL REVIEW MAPPING
# =============================================================================

assert len(
    fabaceae_target_review
) == 5


assert fabaceae_target_review[
    "Canonical_Species"
].is_unique


assert fabaceae_target_review[
    "Fabaceae_Target_Review_Decision"
].notna().all()


assert fabaceae_target_review[
    "R5_Next_Action"
].notna().all()


assert fabaceae_target_review[
    "N_Same_Genus_Independent_Species"
].notna().all()


# =============================================================================
# 10. SAME-GENUS SUPPORT DIAGNOSTIC
# =============================================================================

same_genus_supported = (
    fabaceae_target_review[
        fabaceae_target_review[
            "N_Same_Genus_Independent_Species"
        ].gt(0)
    ]
    .copy()
)


print("\nFabaceae targets with same-genus evidence")
print("-" * 100)

if len(
    same_genus_supported
) > 0:

    display(
        same_genus_supported[
            [
                "Canonical_Species",
                "Final_Search_Name",
                "Residual_Genus",
                "N_Same_Genus_Independent_Species",
                "Same_Genus_Species_Represented",
                "Same_Genus_Candidate_Seed_Mass_mg",
                "Fabaceae_Target_Review_Decision",
                "Next_Action",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 11. RECONCILE WITH EARLIER SINGLE-CONGENER REVIEW
# =============================================================================
#
# Pterocarpus was already reviewed in 9.4Q using:
#
#     Pterocarpus santalinoides
#
# and was deferred because one congener was insufficient.
#
# We preserve that conclusion unless new independent Pterocarpus species
# have appeared in the completed evidence pool.
#
# =============================================================================

pterocarpus_prior = (
    single_congener_review[
        single_congener_review[
            "Genus_Search_Target"
        ].eq(
            "Pterocarpus"
        )
    ]
    .copy()
)


assert len(
    pterocarpus_prior
) == 1


pterocarpus_current = (
    target_genus_review[
        target_genus_review[
            "Residual_Genus"
        ].eq(
            "Pterocarpus"
        )
    ]
    .copy()
)


assert len(
    pterocarpus_current
) == 1


print("\nPterocarpus cross-stage review")
print("-" * 100)

print(
    f"Prior 9.4Q candidate species : "
    f"{pterocarpus_prior['Species_Represented'].iloc[0]}"
)

print(
    f"Prior 9.4Q candidate value   : "
    f"{pterocarpus_prior['Candidate_Seed_Mass_mg'].iloc[0]} mg"
)

print(
    f"Current same-genus species   : "
    f"{pterocarpus_current['N_Same_Genus_Independent_Species'].iloc[0]}"
)

print(
    f"Current review decision      : "
    f"{pterocarpus_current['Fabaceae_Target_Review_Decision'].iloc[0]}"
)


# =============================================================================
# 12. FAMILY DISTRIBUTION CONTEXT FOR SAME-GENUS CANDIDATES
# =============================================================================

def percentile_rank(
    values,
    candidate
):

    if pd.isna(candidate):

        return np.nan

    values = np.asarray(
        values,
        dtype=float
    )

    return (
        100.0
        *
        np.mean(
            values <= candidate
        )
    )


fabaceae_values = (
    fabaceae_evidence[
        "Seed_Mass_mg"
    ]
    .to_numpy()
)


fabaceae_target_review[
    "Same_Genus_Candidate_Family_Percentile"
] = (
    fabaceae_target_review[
        "Same_Genus_Candidate_Seed_Mass_mg"
    ]
    .apply(
        lambda x:
            percentile_rank(
                fabaceae_values,
                x
            )
    )
)


fabaceae_target_review[
    "Same_Genus_Log10_Distance_From_Family_Median"
] = np.where(
    fabaceae_target_review[
        "Same_Genus_Candidate_Seed_Mass_mg"
    ].gt(0),

    np.abs(
        np.log10(
            fabaceae_target_review[
                "Same_Genus_Candidate_Seed_Mass_mg"
            ]
        )
        -
        FAB_LOG_MEDIAN
    ),

    np.nan
)


# =============================================================================
# 13. FINAL REVIEW DECISION
# =============================================================================
#
# Only >=2 independent same-genus species can move forward as a possible
# genus estimate.
#
# One congener remains insufficient.
#
# Zero congeners remains unresolved because whole-family Fabaceae transfer
# is already prohibited by high dispersion.
#
# =============================================================================

fabaceae_target_review[
    "Numerical_Seed_Mass_Assigned_In_9_4R_R5"
] = False


fabaceae_target_review[
    "Assigned_Seed_Mass_mg"
] = np.nan


def final_status(
    row
):

    n = (
        row[
            "N_Same_Genus_Independent_Species"
        ]
    )

    if n >= 2:

        return (
            "PENDING_FORMAL_GENUS_ESTIMATE_DECISION"
        )

    if n == 1:

        return (
            "UNRESOLVED_SINGLE_CONGENER_INSUFFICIENT"
        )

    return (
        "UNRESOLVED_NO_CLOSE_GENUS_EVIDENCE"
    )


fabaceae_target_review[
    "Post_9_4R_R5_Status"
] = (
    fabaceae_target_review
    .apply(
        final_status,
        axis=1
    )
)


# =============================================================================
# 14. DISPLAY FINAL FIVE-TAXON REVIEW
# =============================================================================

print("\nFinal Fabaceae high-dispersion target review")
print("-" * 100)

display(
    fabaceae_target_review[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "N_Same_Genus_Independent_Species",
            "Same_Genus_Species_Represented",
            "Same_Genus_Candidate_Seed_Mass_mg",
            "Same_Genus_Candidate_Family_Percentile",
            "Fabaceae_Raw_Fold_Range",
            "Fabaceae_Log10_Range",
            "Fabaceae_Target_Review_Decision",
            "Post_9_4R_R5_Status",
            "Numerical_Seed_Mass_Assigned_In_9_4R_R5",
        ]
    ]
)


# =============================================================================
# 15. DETERMINE WHETHER ANY GENUS QUALIFIES FOR FURTHER ESTIMATION
# =============================================================================

genus_candidate_targets = (
    fabaceae_target_review[
        fabaceae_target_review[
            "N_Same_Genus_Independent_Species"
        ].ge(2)
    ]
    .copy()
)


single_congener_context = (
    fabaceae_target_review[
        fabaceae_target_review[
            "N_Same_Genus_Independent_Species"
        ].eq(1)
    ]
    .copy()
)


no_genus_evidence_targets = (
    fabaceae_target_review[
        fabaceae_target_review[
            "N_Same_Genus_Independent_Species"
        ].eq(0)
    ]
    .copy()
)


print("\nFabaceae review disposition")
print("-" * 100)

print(
    f"Targets with >=2 same-genus species : "
    f"{len(genus_candidate_targets)}"
)

print(
    f"Targets with one congener only      : "
    f"{len(single_congener_context)}"
)

print(
    f"Targets with no same-genus evidence : "
    f"{len(no_genus_evidence_targets)}"
)


# =============================================================================
# 16. EXPECTED CONSERVATIVE STATE
# =============================================================================
#
# Based on the evidence assembled through 9.4R-R2A, Pterocarpus is expected
# to be the only target genus represented in the family evidence pool, and
# only by one independent species.
#
# If the data have changed, the following diagnostic intentionally stops us
# from silently applying an outdated assumption.
#
# =============================================================================

if len(
    genus_candidate_targets
) > 0:

    print(
        "\nATTENTION: At least one Fabaceae target genus now has "
        ">=2 independent evidence species."
    )

    print(
        "Do not automatically assign it here. It requires a dedicated "
        "genus-estimate decision before numerical propagation."
    )


# =============================================================================
# 17. BUILD POST-R5 UNRESOLVED REGISTER
# =============================================================================

fabaceae_unresolved = (
    fabaceae_target_review[
        ~fabaceae_target_review[
            "Numerical_Seed_Mass_Assigned_In_9_4R_R5"
        ]
    ]
    .copy()
)


assert (
    len(
        fabaceae_unresolved
    )
    +
    int(
        fabaceae_target_review[
            "Numerical_Seed_Mass_Assigned_In_9_4R_R5"
        ].sum()
    )
    ==
    5
)


# =============================================================================
# 18. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_R5 = 92

N_NEW_ASSIGNMENTS_R5 = int(
    fabaceae_target_review[
        "Numerical_Seed_Mass_Assigned_In_9_4R_R5"
    ]
    .sum()
)


N_RESOLVED_AFTER_R5 = (
    N_RESOLVED_BEFORE_R5
    +
    N_NEW_ASSIGNMENTS_R5
)


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_R5
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 19. STRONG VALIDATION
# =============================================================================

assert len(
    fabaceae_target_review
) == 5


assert fabaceae_target_review[
    "Canonical_Species"
].is_unique


assert (
    fabaceae_target_review[
        "N_Same_Genus_Independent_Species"
    ]
    .ge(0)
    .all()
)


# No whole-family assignment is permitted.
assert not (
    fabaceae_target_review[
        "Assigned_Seed_Mass_mg"
    ]
    .notna()
    &
    fabaceae_target_review[
        "N_Same_Genus_Independent_Species"
    ]
    .lt(2)
).any()


# =============================================================================
# 20. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Fabaceae independent evidence species",
            "Fabaceae independent evidence genera",
            "Fabaceae raw fold range",
            "Fabaceae log10 range",
            "Fabaceae residual canonical taxa reviewed",
            "Targets with >=2 same-genus evidence species",
            "Targets with one same-genus species",
            "Targets with no same-genus evidence",
            "Seed Mass values assigned in 9.4R-R5",
            "Fabaceae residual taxa remaining after review",
            "Resolved canonical taxa after 9.4R-R5",
            "Overall Seed Mass coverage percent",
        ],

        "Value": [
            len(
                fabaceae_evidence
            ),

            fabaceae_evidence[
                "Biological_Genus"
            ].nunique(),

            FAB_FOLD_RANGE,

            FAB_LOG_RANGE,

            len(
                fabaceae_target_review
            ),

            len(
                genus_candidate_targets
            ),

            len(
                single_congener_context
            ),

            len(
                no_genus_evidence_targets
            ),

            N_NEW_ASSIGNMENTS_R5,

            len(
                fabaceae_unresolved
            ),

            N_RESOLVED_AFTER_R5,

            coverage_percent,
        ],
    }
)


print("\nSection 9.4R-R5 summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 21. EXPORT
# =============================================================================

fabaceae_evidence.to_csv(
    OUT_FABACEAE_EVIDENCE,
    index=False
)


fabaceae_genus_summary.to_csv(
    OUT_FABACEAE_GENUS_SUMMARY,
    index=False
)


fabaceae_target_review.to_csv(
    OUT_FABACEAE_TARGET_REVIEW,
    index=False
)


single_congener_context.to_csv(
    OUT_FABACEAE_SINGLE_CONGENER_CONTEXT,
    index=False
)


fabaceae_unresolved.to_csv(
    OUT_FABACEAE_UNRESOLVED,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 22. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4R-R5 exports")
print("-" * 100)

print(
    f"Fabaceae evidence detail     : "
    f"{OUT_FABACEAE_EVIDENCE}"
)

print(
    f"Fabaceae genus summary       : "
    f"{OUT_FABACEAE_GENUS_SUMMARY}"
)

print(
    f"Fabaceae target review       : "
    f"{OUT_FABACEAE_TARGET_REVIEW}"
)

print(
    f"Single-congener context      : "
    f"{OUT_FABACEAE_SINGLE_CONGENER_CONTEXT}"
)

print(
    f"Post-review unresolved       : "
    f"{OUT_FABACEAE_UNRESOLVED}"
)

print(
    f"Summary                      : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 23. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Explicit Fabaceae high-dispersion Seed Mass review "
    "completed successfully."
)

print(
    f"Fabaceae evidence species             : "
    f"{len(fabaceae_evidence)}"
)

print(
    f"Fabaceae evidence genera              : "
    f"{fabaceae_evidence['Biological_Genus'].nunique()}"
)

print(
    f"Fabaceae canonical targets reviewed   : "
    f"{len(fabaceae_target_review)}"
)

print(
    f"Targets with >=2 same-genus species   : "
    f"{len(genus_candidate_targets)}"
)

print(
    f"Targets with one congener only        : "
    f"{len(single_congener_context)}"
)

print(
    f"Targets with no same-genus evidence   : "
    f"{len(no_genus_evidence_targets)}"
)

print(
    f"Seed Mass values assigned in R5       : "
    f"{N_NEW_ASSIGNMENTS_R5}"
)

print(
    f"Fabaceae targets remaining unresolved : "
    f"{len(fabaceae_unresolved)}"
)

print(
    f"Overall resolved canonical taxa       : "
    f"{N_RESOLVED_AFTER_R5} / {N_CANONICAL_TAXA}"
)

print(
    f"Overall Seed Mass coverage            : "
    f"{coverage_percent:.2f}%"
)

print(
    "Whole-family Fabaceae transfer remains prohibited "
    "because of extreme within-family Seed Mass dispersion."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4S
# FINAL TARGETED TAXONOMIC-NAME RESCUE FOR RESIDUAL SEED MASS TAXA
# =============================================================================
#
# Purpose
# -------
# Resolve the remaining taxonomic-name-rescue pathway after completion of:
#
#     BIEN + TRY species acquisition
#     external species rescue
#     genus-level review
#     family-level review
#     Fabaceae high-dispersion review
#
# Five canonical project taxa remain in the taxonomic-name-rescue pathway,
# representing four biological targets:
#
#     Aningeria robusta
#         -> Aningeria pierrei
#
#     Corynante pachycercus
#         -> Corynanthe pachyceras
#
#     Homalium alymeri
#     Homalium aylmeri
#         -> Homalium longistylum
#
#     Mitragyna cilliata
#         -> Mitragyna ciliata
#
# Section 9.4S asks one narrow question:
#
#     Does the already validated 63-species Seed Mass evidence pool contain
#     accepted species-level evidence for any of these reconciled biological
#     names?
#
# Conservative rule
# -----------------
# ACCEPT:
#     Accepted biological name occurs directly in the normalized independent
#     species evidence pool with a positive Seed Mass value.
#
# DO NOT ACCEPT:
#     Congeneric value
#     Family value
#     Fuzzy spelling similarity alone
#     Morphologically similar species
#
# IMPORTANT
# ---------
# This section uses previously validated taxonomic reconciliations only.
# It does not create new synonymies.
#
# Coverage entering 9.4S:
#
#     Resolved : 92 / 127
#     Residual : 35
#
# Final assignment count is intentionally determined from the evidence.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

TAXONOMIC_RESCUE_PENDING_FILE = (
    OUT_DIR
    / "omo_seed_mass_taxonomic_name_rescue_pending_targets.csv"
)

COMPLETE_SPECIES_FAMILY_FILE = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_family_mapping_complete.csv"
)

POST_FAMILY_RESIDUAL_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_family_assignment_residual_canonical_taxa.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_TAXONOMIC_CROSSWALK = (
    OUT_DIR
    / "omo_seed_mass_final_taxonomic_rescue_crosswalk.csv"
)

OUT_ACCEPTED_NAME_EVIDENCE = (
    OUT_DIR
    / "omo_seed_mass_final_taxonomic_rescue_accepted_name_evidence.csv"
)

OUT_CANONICAL_ASSIGNMENTS = (
    OUT_DIR
    / "omo_seed_mass_final_taxonomic_rescue_assignments.csv"
)

OUT_UNRESOLVED_TARGETS = (
    OUT_DIR
    / "omo_seed_mass_final_taxonomic_rescue_unresolved.csv"
)

OUT_POST_RESCUE_STATE = (
    OUT_DIR
    / "omo_seed_mass_post_taxonomic_rescue_state.csv"
)

OUT_POST_RESCUE_RESIDUAL = (
    OUT_DIR
    / "omo_seed_mass_post_taxonomic_rescue_residual_canonical_taxa.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_final_taxonomic_rescue_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    TAXONOMIC_RESCUE_PENDING_FILE,
    COMPLETE_SPECIES_FAMILY_FILE,
    POST_FAMILY_RESIDUAL_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4S input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

pending = pd.read_csv(
    TAXONOMIC_RESCUE_PENDING_FILE,
    low_memory=False
)

species_evidence = pd.read_csv(
    COMPLETE_SPECIES_FAMILY_FILE,
    low_memory=False
)

post_family_residual = pd.read_csv(
    POST_FAMILY_RESIDUAL_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4S — FINAL TARGETED TAXONOMIC-NAME "
    "RESCUE FOR RESIDUAL SEED MASS TAXA"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Taxonomic-rescue canonical taxa : "
    f"{len(pending)}"
)

print(
    f"Validated evidence species      : "
    f"{len(species_evidence)}"
)

print(
    f"Post-family residual taxa       : "
    f"{len(post_family_residual)}"
)


# =============================================================================
# 4. VALIDATE EXPECTED INPUT STATE
# =============================================================================

assert len(
    pending
) == 5, (
    f"Expected 5 taxonomic-rescue canonical taxa; "
    f"found {len(pending)}."
)


assert len(
    species_evidence
) == 63


assert len(
    post_family_residual
) == 35


assert species_evidence[
    "Biological_Species_Name"
].nunique() == 63


# =============================================================================
# 5. CLEAN KEY FIELDS
# =============================================================================

def clean_text_series(
    series
):

    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "":
                    pd.NA,

                "nan":
                    pd.NA,

                "None":
                    pd.NA,

                "<NA>":
                    pd.NA,
            }
        )
    )


pending[
    "Canonical_Species"
] = clean_text_series(
    pending[
        "Canonical_Species"
    ]
)


species_evidence[
    "Biological_Species_Name"
] = clean_text_series(
    species_evidence[
        "Biological_Species_Name"
    ]
)


species_evidence[
    "Seed_Mass_mg"
] = pd.to_numeric(
    species_evidence[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert species_evidence[
    "Seed_Mass_mg"
].notna().all()


assert species_evidence[
    "Seed_Mass_mg"
].gt(
    0
).all()


# =============================================================================
# 6. DEFINE VALIDATED TAXONOMIC RESCUE CROSSWALK
# =============================================================================
#
# These relationships were established previously in the project workflow.
#
# No new taxonomic relationship is inferred here.
#
# =============================================================================

validated_taxonomic_crosswalk = pd.DataFrame(
    [
        {
            "Canonical_Species":
                "Aningeria robusta",

            "Validated_Search_Name":
                "Aningeria pierrei",

            "Taxonomic_Rescue_Class":
                "VALIDATED_ACCEPTED_NAME",

            "Taxonomic_Rescue_Rationale":
                (
                    "Project taxon Aningeria robusta was previously "
                    "validated for Seed Mass searching under "
                    "Aningeria pierrei."
                ),
        },

        {
            "Canonical_Species":
                "Corynante pachycercus",

            "Validated_Search_Name":
                "Corynanthe pachyceras",

            "Taxonomic_Rescue_Class":
                "VALIDATED_ORTHOGRAPHIC_AND_ACCEPTED_NAME_RECONCILIATION",

            "Taxonomic_Rescue_Rationale":
                (
                    "Project spelling Corynante pachycercus was previously "
                    "reconciled to Corynanthe pachyceras."
                ),
        },

        {
            "Canonical_Species":
                "Homalium alymeri",

            "Validated_Search_Name":
                "Homalium longistylum",

            "Taxonomic_Rescue_Class":
                "VALIDATED_ACCEPTED_NAME",

            "Taxonomic_Rescue_Rationale":
                (
                    "Project name Homalium alymeri was previously routed "
                    "to Homalium longistylum for targeted Seed Mass search."
                ),
        },

        {
            "Canonical_Species":
                "Homalium aylmeri",

            "Validated_Search_Name":
                "Homalium longistylum",

            "Taxonomic_Rescue_Class":
                "VALIDATED_ORTHOGRAPHIC_AND_ACCEPTED_NAME_RECONCILIATION",

            "Taxonomic_Rescue_Rationale":
                (
                    "Project spelling Homalium aylmeri represents the same "
                    "validated rescue target as Homalium alymeri and was "
                    "routed to Homalium longistylum."
                ),
        },

        {
            "Canonical_Species":
                "Mitragyna cilliata",

            "Validated_Search_Name":
                "Mitragyna ciliata",

            "Taxonomic_Rescue_Class":
                "VALIDATED_ORTHOGRAPHIC_CORRECTION",

            "Taxonomic_Rescue_Rationale":
                (
                    "Project spelling Mitragyna cilliata was previously "
                    "normalized to Mitragyna ciliata."
                ),
        },
    ]
)


# =============================================================================
# 7. VALIDATE CROSSWALK AGAINST PENDING SET
# =============================================================================

pending_set = set(
    pending[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


crosswalk_set = set(
    validated_taxonomic_crosswalk[
        "Canonical_Species"
    ]
)


missing_from_crosswalk = (
    pending_set
    -
    crosswalk_set
)


extra_in_crosswalk = (
    crosswalk_set
    -
    pending_set
)


print("\nValidated taxonomic crosswalk reconciliation")
print("-" * 100)

print(
    f"Pending canonical taxa : "
    f"{len(pending_set)}"
)

print(
    f"Crosswalk canonical taxa: "
    f"{len(crosswalk_set)}"
)

print(
    f"Missing from crosswalk : "
    f"{len(missing_from_crosswalk)}"
)

print(
    f"Unexpected extra taxa  : "
    f"{len(extra_in_crosswalk)}"
)


if missing_from_crosswalk:

    print(
        "\nMissing crosswalk taxa:"
    )

    for taxon in sorted(
        missing_from_crosswalk
    ):
        print(
            f" - {taxon}"
        )


assert len(
    missing_from_crosswalk
) == 0, (
    "Some pending taxonomic-rescue taxa lack a validated "
    "Section 9.4S crosswalk entry."
)


assert len(
    extra_in_crosswalk
) == 0, (
    "Section 9.4S crosswalk contains taxa not present in "
    "the authoritative pending register."
)


# =============================================================================
# 8. DISPLAY VALIDATED CROSSWALK
# =============================================================================

print("\nFinal targeted taxonomic-rescue crosswalk")
print("-" * 100)

display(
    validated_taxonomic_crosswalk
)


# =============================================================================
# 9. BUILD BIOLOGICAL TARGET REGISTER
# =============================================================================
#
# Five canonical taxa collapse to four biological search targets because:
#
#     Homalium alymeri
#     Homalium aylmeri
#
# both resolve to:
#
#     Homalium longistylum
#
# =============================================================================

biological_targets = (
    validated_taxonomic_crosswalk
    .groupby(
        "Validated_Search_Name",
        as_index=False
    )
    .agg(
        N_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        Canonical_Taxa=(
            "Canonical_Species",
            lambda x:
                " | ".join(
                    sorted(
                        x.dropna()
                        .astype(str)
                        .unique()
                    )
                )
        ),

        Taxonomic_Rescue_Classes=(
            "Taxonomic_Rescue_Class",
            lambda x:
                " | ".join(
                    sorted(
                        x.dropna()
                        .astype(str)
                        .unique()
                    )
                )
        ),
    )
)


assert len(
    biological_targets
) == 4


print("\nBiological rescue targets")
print("-" * 100)

display(
    biological_targets
)


# =============================================================================
# 10. SEARCH NORMALIZED EVIDENCE POOL FOR ACCEPTED-NAME MATCHES
# =============================================================================

evidence_search = (
    biological_targets
    .merge(
        species_evidence,
        left_on="Validated_Search_Name",
        right_on="Biological_Species_Name",
        how="left",
        validate="one_to_one"
    )
)


evidence_search[
    "Accepted_Name_Evidence_Found"
] = (
    evidence_search[
        "Seed_Mass_mg"
    ]
    .notna()
)


# =============================================================================
# 11. CLASSIFY BIOLOGICAL TARGET OUTCOMES
# =============================================================================

evidence_search[
    "Taxonomic_Rescue_Evidence_Status"
] = np.where(
    evidence_search[
        "Accepted_Name_Evidence_Found"
    ],
    "VALIDATED_ACCEPTED_NAME_SPECIES_EVIDENCE_FOUND",
    "NO_ACCEPTED_NAME_SPECIES_EVIDENCE_IN_CURRENT_POOL"
)


evidence_search[
    "Taxonomic_Rescue_Action"
] = np.where(
    evidence_search[
        "Accepted_Name_Evidence_Found"
    ],
    "ASSIGN_ACCEPTED_NAME_SPECIES_VALUE_TO_PROJECT_CANONICAL_TAXA",
    "RETAIN_UNRESOLVED_FOR_BROADER_FALLBACK"
)


print("\nAccepted-name evidence search")
print("-" * 100)

display_columns = [
    "Validated_Search_Name",
    "N_Canonical_Taxa",
    "Canonical_Taxa",
    "Accepted_Name_Evidence_Found",
    "Taxonomic_Rescue_Evidence_Status",
    "Taxonomic_Rescue_Action",
]


for optional_col in [
    "Biological_Species_Name",
    "Seed_Mass_mg",
    "Evidence_Sources",
    "Family",
]:

    if optional_col in evidence_search.columns:

        display_columns.append(
            optional_col
        )


display(
    evidence_search[
        display_columns
    ]
)


# =============================================================================
# 12. RETAIN ONLY ACCEPTED-NAME SPECIES EVIDENCE
# =============================================================================

accepted_name_evidence = (
    evidence_search[
        evidence_search[
            "Accepted_Name_Evidence_Found"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


unresolved_biological_targets = (
    evidence_search[
        ~evidence_search[
            "Accepted_Name_Evidence_Found"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


print("\nTaxonomic rescue evidence disposition")
print("-" * 100)

print(
    f"Biological targets searched   : "
    f"{len(evidence_search)}"
)

print(
    f"Accepted-name evidence found  : "
    f"{len(accepted_name_evidence)}"
)

print(
    f"Biological targets unresolved : "
    f"{len(unresolved_biological_targets)}"
)


# =============================================================================
# 13. MAP ACCEPTED-NAME EVIDENCE BACK TO CANONICAL TAXA
# =============================================================================

assignment_lookup_columns = [
    "Validated_Search_Name",
    "Biological_Species_Name",
    "Seed_Mass_mg",
    "Evidence_Sources",
]


for optional_col in [
    "Family",
    "Family_Mapping_Source",
    "Family_Mapping_Status",
]:

    if optional_col in accepted_name_evidence.columns:

        assignment_lookup_columns.append(
            optional_col
        )


accepted_lookup = (
    accepted_name_evidence[
        assignment_lookup_columns
    ]
    .copy()
)


canonical_rescue = (
    validated_taxonomic_crosswalk
    .merge(
        accepted_lookup,
        on="Validated_Search_Name",
        how="left",
        validate="many_to_one"
    )
)


canonical_rescue[
    "Taxonomic_Rescue_Assigned"
] = (
    canonical_rescue[
        "Seed_Mass_mg"
    ]
    .notna()
)


canonical_rescue[
    "Assigned_Seed_Mass_mg"
] = (
    canonical_rescue[
        "Seed_Mass_mg"
    ]
)


canonical_rescue[
    "Seed_Mass_Assignment_Level"
] = np.where(
    canonical_rescue[
        "Taxonomic_Rescue_Assigned"
    ],
    "SPECIES_LEVEL_VALIDATED_ALIAS",
    pd.NA
)


canonical_rescue[
    "Seed_Mass_Provenance_Class"
] = np.where(
    canonical_rescue[
        "Taxonomic_Rescue_Assigned"
    ],
    "SPECIES_ACCEPTED_NAME_RESCUE",
    pd.NA
)


canonical_rescue[
    "Seed_Mass_Quality_Flag"
] = np.where(
    canonical_rescue[
        "Taxonomic_Rescue_Assigned"
    ],
    "VALIDATED_TAXONOMIC_EQUIVALENCE_EXISTING_SPECIES_EVIDENCE",
    "NO_ACCEPTED_NAME_SPECIES_EVIDENCE"
)


canonical_rescue[
    "Seed_Mass_Assignment_Method"
] = np.where(
    canonical_rescue[
        "Taxonomic_Rescue_Assigned"
    ],
    "DIRECT_TRANSFER_FROM_VALIDATED_ACCEPTED_NAME_SPECIES_EVIDENCE",
    pd.NA
)


canonical_rescue[
    "Numerical_Seed_Mass_Assigned"
] = (
    canonical_rescue[
        "Taxonomic_Rescue_Assigned"
    ]
)


# =============================================================================
# 14. SPLIT ASSIGNED AND UNRESOLVED CANONICAL TAXA
# =============================================================================

canonical_assignments = (
    canonical_rescue[
        canonical_rescue[
            "Taxonomic_Rescue_Assigned"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


unresolved_taxonomic_targets = (
    canonical_rescue[
        ~canonical_rescue[
            "Taxonomic_Rescue_Assigned"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


print("\nCanonical taxonomic-rescue assignments")
print("-" * 100)

if len(
    canonical_assignments
) > 0:

    display(
        canonical_assignments[
            [
                "Canonical_Species",
                "Validated_Search_Name",
                "Biological_Species_Name",
                "Assigned_Seed_Mass_mg",
                "Seed_Mass_Assignment_Level",
                "Seed_Mass_Provenance_Class",
                "Seed_Mass_Quality_Flag",
                "Seed_Mass_Assignment_Method",
            ]
        ]
    )

else:

    print("NONE")


print("\nCanonical taxonomic-rescue taxa remaining unresolved")
print("-" * 100)

if len(
    unresolved_taxonomic_targets
) > 0:

    display(
        unresolved_taxonomic_targets[
            [
                "Canonical_Species",
                "Validated_Search_Name",
                "Taxonomic_Rescue_Class",
                "Seed_Mass_Quality_Flag",
            ]
        ]
    )

else:

    print("NONE")


# =============================================================================
# 15. VERIFY ASSIGNED TAXA WERE ACTUALLY RESIDUAL BEFORE 9.4S
# =============================================================================

pre_s_residual_set = set(
    post_family_residual[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
    .str.strip()
)


assignment_set = set(
    canonical_assignments[
        "Canonical_Species"
    ]
)


assert assignment_set.issubset(
    pre_s_residual_set
), (
    "Section 9.4S attempted to assign a taxon that was not "
    "part of the post-family residual set."
)


# =============================================================================
# 16. BUILD POST-9.4S STATE
# =============================================================================

post_rescue_state = (
    post_family_residual
    .copy()
)


# Remove any possible rerun-generated fields before merge.
s_owned_fields = [
    "Validated_Search_Name",
    "Assigned_Seed_Mass_mg",
    "Seed_Mass_Assignment_Level",
    "Seed_Mass_Provenance_Class",
    "Seed_Mass_Quality_Flag",
    "Seed_Mass_Assignment_Method",
    "Numerical_Seed_Mass_Assigned_In_9_4S",
]


post_rescue_state = (
    post_rescue_state
    .drop(
        columns=[
            col
            for col in s_owned_fields
            if col in post_rescue_state.columns
        ],
        errors="ignore"
    )
    .copy()
)


if len(
    canonical_assignments
) > 0:

    assignment_lookup = (
        canonical_assignments[
            [
                "Canonical_Species",
                "Validated_Search_Name",
                "Assigned_Seed_Mass_mg",
                "Seed_Mass_Assignment_Level",
                "Seed_Mass_Provenance_Class",
                "Seed_Mass_Quality_Flag",
                "Seed_Mass_Assignment_Method",
            ]
        ]
        .copy()
    )


    post_rescue_state = (
        post_rescue_state
        .merge(
            assignment_lookup,
            on="Canonical_Species",
            how="left",
            validate="one_to_one"
        )
    )


else:

    post_rescue_state[
        "Validated_Search_Name"
    ] = pd.NA

    post_rescue_state[
        "Assigned_Seed_Mass_mg"
    ] = np.nan

    post_rescue_state[
        "Seed_Mass_Assignment_Level"
    ] = pd.NA

    post_rescue_state[
        "Seed_Mass_Provenance_Class"
    ] = pd.NA

    post_rescue_state[
        "Seed_Mass_Quality_Flag"
    ] = pd.NA

    post_rescue_state[
        "Seed_Mass_Assignment_Method"
    ] = pd.NA


post_rescue_state[
    "Numerical_Seed_Mass_Assigned_In_9_4S"
] = (
    post_rescue_state[
        "Assigned_Seed_Mass_mg"
    ]
    .notna()
)


# =============================================================================
# 17. BUILD TRUE POST-9.4S RESIDUAL
# =============================================================================

post_rescue_residual = (
    post_rescue_state[
        ~post_rescue_state[
            "Numerical_Seed_Mass_Assigned_In_9_4S"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


N_NEW_ASSIGNMENTS_9_4S = int(
    post_rescue_state[
        "Numerical_Seed_Mass_Assigned_In_9_4S"
    ]
    .sum()
)


assert (
    len(
        post_rescue_residual
    )
    ==
    35
    -
    N_NEW_ASSIGNMENTS_9_4S
)


# =============================================================================
# 18. COVERAGE UPDATE
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4S = 92

N_RESOLVED_AFTER_9_4S = (
    N_RESOLVED_BEFORE_9_4S
    +
    N_NEW_ASSIGNMENTS_9_4S
)


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4S
    /
    N_CANONICAL_TAXA
)


residual_percent = (
    100
    *
    len(
        post_rescue_residual
    )
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 19. CROSS-CHECK COUNT CONSISTENCY
# =============================================================================

assert (
    len(
        canonical_assignments
    )
    ==
    N_NEW_ASSIGNMENTS_9_4S
)


assert (
    N_RESOLVED_AFTER_9_4S
    +
    len(
        post_rescue_residual
    )
    ==
    127
)


# =============================================================================
# 20. BUILD BIOLOGICAL-TARGET OUTCOME SUMMARY
# =============================================================================

biological_outcome_summary = (
    evidence_search[
        [
            "Validated_Search_Name",
            "N_Canonical_Taxa",
            "Canonical_Taxa",
            "Accepted_Name_Evidence_Found",
            "Taxonomic_Rescue_Evidence_Status",
            "Taxonomic_Rescue_Action",
        ]
    ]
    .copy()
)


print("\nFinal biological-target rescue outcome")
print("-" * 100)

display(
    biological_outcome_summary
)


# =============================================================================
# 21. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Canonical taxonomic-rescue taxa entering 9.4S",
            "Biological accepted-name targets searched",
            "Biological targets with accepted-name Seed Mass evidence",
            "Biological targets without accepted-name Seed Mass evidence",
            "Canonical Seed Mass values assigned in 9.4S",
            "Canonical taxonomic-rescue taxa remaining unresolved",
            "Resolved canonical taxa after 9.4S",
            "Residual canonical taxa after 9.4S",
            "Overall Seed Mass coverage percent",
            "Residual canonical percent",
        ],

        "Value": [
            len(
                pending
            ),

            len(
                evidence_search
            ),

            len(
                accepted_name_evidence
            ),

            len(
                unresolved_biological_targets
            ),

            N_NEW_ASSIGNMENTS_9_4S,

            len(
                unresolved_taxonomic_targets
            ),

            N_RESOLVED_AFTER_9_4S,

            len(
                post_rescue_residual
            ),

            coverage_percent,

            residual_percent,
        ],
    }
)


print("\nSection 9.4S summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 22. EXPORT
# =============================================================================

validated_taxonomic_crosswalk.to_csv(
    OUT_TAXONOMIC_CROSSWALK,
    index=False
)


accepted_name_evidence.to_csv(
    OUT_ACCEPTED_NAME_EVIDENCE,
    index=False
)


canonical_assignments.to_csv(
    OUT_CANONICAL_ASSIGNMENTS,
    index=False
)


unresolved_taxonomic_targets.to_csv(
    OUT_UNRESOLVED_TARGETS,
    index=False
)


post_rescue_state.to_csv(
    OUT_POST_RESCUE_STATE,
    index=False
)


post_rescue_residual.to_csv(
    OUT_POST_RESCUE_RESIDUAL,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 23. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4S exports")
print("-" * 100)

print(
    f"Taxonomic rescue crosswalk : "
    f"{OUT_TAXONOMIC_CROSSWALK}"
)

print(
    f"Accepted-name evidence     : "
    f"{OUT_ACCEPTED_NAME_EVIDENCE}"
)

print(
    f"Canonical assignments      : "
    f"{OUT_CANONICAL_ASSIGNMENTS}"
)

print(
    f"Unresolved rescue targets  : "
    f"{OUT_UNRESOLVED_TARGETS}"
)

print(
    f"Post-rescue state          : "
    f"{OUT_POST_RESCUE_STATE}"
)

print(
    f"Post-rescue residual       : "
    f"{OUT_POST_RESCUE_RESIDUAL}"
)

print(
    f"Summary                    : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 24. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Final targeted taxonomic-name Seed Mass rescue "
    "completed successfully."
)

print(
    f"Canonical rescue taxa reviewed        : "
    f"{len(pending)}"
)

print(
    f"Biological accepted-name targets      : "
    f"{len(evidence_search)}"
)

print(
    f"Accepted-name targets with evidence   : "
    f"{len(accepted_name_evidence)}"
)

print(
    f"Canonical values assigned in 9.4S     : "
    f"{N_NEW_ASSIGNMENTS_9_4S}"
)

print(
    f"Taxonomic-rescue taxa still unresolved: "
    f"{len(unresolved_taxonomic_targets)}"
)

print(
    f"Resolved canonical taxa               : "
    f"{N_RESOLVED_AFTER_9_4S} / {N_CANONICAL_TAXA}"
)

print(
    f"Overall Seed Mass coverage            : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Total residual canonical taxa         : "
    f"{len(post_rescue_residual)}"
)

print(
    "Only direct accepted-name species evidence was eligible "
    "for assignment."
)

print(
    "No congeneric or family-level substitute was introduced "
    "during taxonomic rescue."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4T
# SPECIAL BIOLOGICAL CASE REVIEW — MUSA × PARADISIACA
# =============================================================================
#
# Purpose
# -------
# Review the remaining special biological case:
#
#     Project canonical taxon : Musa sapiensis
#     Validated search name   : Musa × paradisiaca
#
# This taxon is treated separately because cultivated Musa × paradisiaca
# is commonly sterile or functionally seedless.
#
# The question is NOT:
#
#     "Can we find any Musa seed mass?"
#
# The question is:
#
#     "Is a standardized species-level Seed Mass value biologically
#      applicable and defensible for the project taxon Musa × paradisiaca?"
#
# Conservative rule
# -----------------
# A numerical Seed Mass value may be assigned ONLY if:
#
#   1. direct taxon-specific evidence exists for Musa × paradisiaca;
#   2. the reported structure is a mature seed, not fruit pulp or another
#      reproductive component;
#   3. the taxon is demonstrated to produce mature seeds under the relevant
#      biological form represented by the project record;
#   4. the measurement is comparable with the standardized Seed Mass trait.
#
# Otherwise:
#
#     NO numerical Seed Mass is assigned.
#
# Possible outcomes
# -----------------
#
#     BIOLOGICALLY_APPLICABLE_DIRECT_SEED_EVIDENCE
#     NOT_APPLICABLE_FUNCTIONALLY_SEEDLESS
#     BIOLOGICALLY_UNRESOLVED_NO_COMPARABLE_SEED_EVIDENCE
#
# IMPORTANT
# ---------
# No genus-level Musa substitute is permitted here.
# No family-level Musaceae substitute is permitted here.
#
# Coverage entering 9.4T:
#
#     Resolved canonical taxa : 92 / 127
#     Residual canonical taxa : 35
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

POST_RESCUE_RESIDUAL_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_taxonomic_rescue_residual_canonical_taxa.csv"
)

COMPLETE_SPECIES_FAMILY_FILE = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_family_mapping_complete.csv"
)

POST_9_4Q_RESIDUAL_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_single_congener_review_residual_state.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_SPECIAL_CASE_REVIEW = (
    OUT_DIR
    / "omo_seed_mass_musa_paradisiaca_special_case_review.csv"
)

OUT_POST_SPECIAL_CASE_STATE = (
    OUT_DIR
    / "omo_seed_mass_post_special_case_state.csv"
)

OUT_POST_SPECIAL_CASE_RESIDUAL = (
    OUT_DIR
    / "omo_seed_mass_post_special_case_residual_canonical_taxa.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_musa_paradisiaca_special_case_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    POST_RESCUE_RESIDUAL_FILE,
    COMPLETE_SPECIES_FAMILY_FILE,
    POST_9_4Q_RESIDUAL_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4T input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

post_rescue_residual = pd.read_csv(
    POST_RESCUE_RESIDUAL_FILE,
    low_memory=False
)

species_evidence = pd.read_csv(
    COMPLETE_SPECIES_FAMILY_FILE,
    low_memory=False
)

post_q_residual = pd.read_csv(
    POST_9_4Q_RESIDUAL_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4T — SPECIAL BIOLOGICAL CASE REVIEW: "
    "MUSA × PARADISIACA"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Post-taxonomic-rescue residual taxa : "
    f"{len(post_rescue_residual)}"
)

print(
    f"Validated evidence species          : "
    f"{len(species_evidence)}"
)

print(
    f"Post-9.4Q residual rows             : "
    f"{len(post_q_residual)}"
)


# =============================================================================
# 4. VALIDATE EXPECTED STATE
# =============================================================================

assert len(
    post_rescue_residual
) == 35


assert len(
    species_evidence
) == 63


# =============================================================================
# 5. IDENTIFY MUSA SPECIAL-CASE TAXON
# =============================================================================

musa_rows = (
    post_rescue_residual[
        post_rescue_residual[
            "Canonical_Species"
        ]
        .astype(str)
        .str.strip()
        .eq(
            "Musa sapiensis"
        )
    ]
    .copy()
)


assert len(
    musa_rows
) == 1, (
    f"Expected exactly one Musa sapiensis residual row; "
    f"found {len(musa_rows)}."
)


musa_row = (
    musa_rows
    .iloc[0]
)


print("\nSpecial biological-case target")
print("-" * 100)

display(
    musa_rows
)


# =============================================================================
# 6. CONFIRM SPECIAL-CASE ROUTING FROM EARLIER WORKFLOW
# =============================================================================

musa_prior = (
    post_q_residual[
        post_q_residual[
            "Canonical_Species"
        ]
        .astype(str)
        .str.strip()
        .eq(
            "Musa sapiensis"
        )
    ]
    .copy()
)


assert len(
    musa_prior
) == 1


if "Post_9_4Q_Route" in musa_prior.columns:

    assert (
        musa_prior[
            "Post_9_4Q_Route"
        ]
        .eq(
            "SPECIAL_BIOLOGICAL_CASE_REVIEW"
        )
        .all()
    )


# =============================================================================
# 7. VALIDATED TAXONOMIC INTERPRETATION
# =============================================================================

SPECIAL_CASE_CANONICAL = (
    "Musa sapiensis"
)

VALIDATED_SEARCH_NAME = (
    "Musa × paradisiaca"
)

VALIDATED_GENUS = (
    "Musa"
)

VALIDATED_FAMILY = (
    "Musaceae"
)


# =============================================================================
# 8. SEARCH EXISTING VALIDATED EVIDENCE POOL
# =============================================================================

species_evidence[
    "Biological_Species_Name"
] = (
    species_evidence[
        "Biological_Species_Name"
    ]
    .astype("string")
    .str.strip()
)


direct_taxon_evidence = (
    species_evidence[
        species_evidence[
            "Biological_Species_Name"
        ]
        .eq(
            VALIDATED_SEARCH_NAME
        )
    ]
    .copy()
)


genus_context = (
    species_evidence[
        species_evidence[
            "Biological_Genus"
        ]
        .astype("string")
        .str.strip()
        .eq(
            VALIDATED_GENUS
        )
    ]
    .copy()
)


print("\nExisting evidence-pool search")
print("-" * 100)

print(
    f"Direct Musa × paradisiaca evidence rows : "
    f"{len(direct_taxon_evidence)}"
)

print(
    f"Other Musa genus-context rows            : "
    f"{len(genus_context)}"
)


if len(
    direct_taxon_evidence
) > 0:

    print("\nDirect taxon-specific evidence")
    print("-" * 100)

    display(
        direct_taxon_evidence
    )


if len(
    genus_context
) > 0:

    print("\nMusa genus context")
    print("-" * 100)

    display(
        genus_context[
            [
                "Biological_Species_Name",
                "Seed_Mass_mg",
                "Evidence_Sources",
                "Family",
            ]
        ]
    )


# =============================================================================
# 9. BIOLOGICAL APPLICABILITY REVIEW
# =============================================================================
#
# The currently validated project state contains no defensible direct
# standardized Seed Mass measurement for Musa × paradisiaca.
#
# Cultivated Musa × paradisiaca is commonly sterile/seedless, so transferring
# a Seed Mass from another Musa species would not represent the biological
# state of this target.
#
# Therefore:
#
#     Numerical Seed Mass assignment = FALSE
#
# This is treated as a biological applicability limitation rather than a
# conventional missing-value gap.
#
# =============================================================================

direct_comparable_seed_evidence_found = (
    len(
        direct_taxon_evidence
    )
    >
    0
)


if direct_comparable_seed_evidence_found:

    biological_case_status = (
        "DIRECT_TAXON_EVIDENCE_REQUIRES_MANUAL_BIOLOGICAL_COMPATIBILITY_REVIEW"
    )

    biological_applicability = (
        "UNRESOLVED_PENDING_DIRECT_EVIDENCE_REVIEW"
    )

    special_case_action = (
        "DO_NOT_ASSIGN_UNTIL_DIRECT_SEED_EVIDENCE_IS_CONFIRMED_COMPARABLE"
    )

else:

    biological_case_status = (
        "NO_DIRECT_COMPARABLE_SEED_MASS_EVIDENCE"
    )

    biological_applicability = (
        "NOT_APPLICABLE_OR_BIOLOGICALLY_UNRESOLVED_FUNCTIONALLY_SEEDLESS"
    )

    special_case_action = (
        "RETAIN_WITHOUT_NUMERICAL_SEED_MASS_AND_FLAG_SPECIAL_BIOLOGICAL_CASE"
    )


# =============================================================================
# 10. BUILD SPECIAL-CASE REVIEW RECORD
# =============================================================================

special_case_review = pd.DataFrame(
    [
        {
            "Canonical_Species":
                SPECIAL_CASE_CANONICAL,

            "Validated_Search_Name":
                VALIDATED_SEARCH_NAME,

            "Validated_Genus":
                VALIDATED_GENUS,

            "Validated_Family":
                VALIDATED_FAMILY,

            "Direct_Taxon_Seed_Mass_Evidence_Found":
                direct_comparable_seed_evidence_found,

            "N_Direct_Taxon_Evidence_Rows":
                len(
                    direct_taxon_evidence
                ),

            "N_Musa_Genus_Context_Rows":
                len(
                    genus_context
                ),

            "Biological_Case_Status":
                biological_case_status,

            "Trait_Applicability_Status":
                biological_applicability,

            "Seed_Mass_Provenance_Class":
                "SPECIAL_BIOLOGICAL_CASE",

            "Seed_Mass_Quality_Flag":
                (
                    "FUNCTIONALLY_SEEDLESS_OR_NO_COMPARABLE_STANDARDIZED_SEED_MASS"
                ),

            "Numerical_Seed_Mass_Assigned":
                False,

            "Assigned_Seed_Mass_mg":
                np.nan,

            "Special_Case_Action":
                special_case_action,

            "Special_Case_Rationale":
                (
                    "The project taxon Musa sapiensis was previously "
                    "reconciled to Musa × paradisiaca. No defensible direct "
                    "standardized Seed Mass value is present in the validated "
                    "evidence pool. Cultivated Musa × paradisiaca is commonly "
                    "sterile or functionally seedless, so congeneric or "
                    "family-level Seed Mass transfer would be biologically "
                    "misleading."
                ),
        }
    ]
)


# =============================================================================
# 11. DISPLAY SPECIAL-CASE DECISION
# =============================================================================

print("\nMusa × paradisiaca special biological-case review")
print("-" * 100)

display(
    special_case_review
)


# =============================================================================
# 12. VALIDATE NO NUMERICAL VALUE ASSIGNED
# =============================================================================

assert special_case_review[
    "Numerical_Seed_Mass_Assigned"
].eq(
    False
).all()


assert special_case_review[
    "Assigned_Seed_Mass_mg"
].isna().all()


# =============================================================================
# 13. BUILD POST-9.4T STATE
# =============================================================================
#
# Musa remains in the residual register because it has no numerical value,
# but is now explicitly classified as a SPECIAL BIOLOGICAL CASE rather than
# an ordinary acquisition failure.
#
# =============================================================================

post_special_case_state = (
    post_rescue_residual
    .copy()
)


# Remove rerun-owned fields if present.
t_owned_fields = [
    "Special_Case_Reviewed_In_9_4T",
    "Trait_Applicability_Status",
    "Special_Case_Action",
    "Seed_Mass_Provenance_Class_9_4T",
    "Seed_Mass_Quality_Flag_9_4T",
]


post_special_case_state = (
    post_special_case_state
    .drop(
        columns=[
            col
            for col in t_owned_fields
            if col in post_special_case_state.columns
        ],
        errors="ignore"
    )
    .copy()
)


post_special_case_state[
    "Special_Case_Reviewed_In_9_4T"
] = False


post_special_case_state[
    "Trait_Applicability_Status"
] = pd.NA


post_special_case_state[
    "Special_Case_Action"
] = pd.NA


post_special_case_state[
    "Seed_Mass_Provenance_Class_9_4T"
] = pd.NA


post_special_case_state[
    "Seed_Mass_Quality_Flag_9_4T"
] = pd.NA


musa_mask = (
    post_special_case_state[
        "Canonical_Species"
    ]
    .astype(str)
    .str.strip()
    .eq(
        SPECIAL_CASE_CANONICAL
    )
)


assert musa_mask.sum() == 1


post_special_case_state.loc[
    musa_mask,
    "Special_Case_Reviewed_In_9_4T"
] = True


post_special_case_state.loc[
    musa_mask,
    "Trait_Applicability_Status"
] = (
    biological_applicability
)


post_special_case_state.loc[
    musa_mask,
    "Special_Case_Action"
] = (
    special_case_action
)


post_special_case_state.loc[
    musa_mask,
    "Seed_Mass_Provenance_Class_9_4T"
] = (
    "SPECIAL_BIOLOGICAL_CASE"
)


post_special_case_state.loc[
    musa_mask,
    "Seed_Mass_Quality_Flag_9_4T"
] = (
    "FUNCTIONALLY_SEEDLESS_OR_NO_COMPARABLE_STANDARDIZED_SEED_MASS"
)


# =============================================================================
# 14. BUILD POST-SPECIAL-CASE RESIDUAL
# =============================================================================
#
# No numerical value is added in 9.4T, so the residual count remains 35.
#
# =============================================================================

post_special_case_residual = (
    post_special_case_state
    .copy()
    .reset_index(drop=True)
)


assert len(
    post_special_case_residual
) == 35


# =============================================================================
# 15. CLASSIFY RESIDUAL STATUS AFTER 9.4T
# =============================================================================
#
# We retain existing route information, but explicitly separate Musa as a
# biological non-standard case.
#
# =============================================================================

post_special_case_residual[
    "Post_9_4T_Residual_Class"
] = (
    "BROADER_FALLBACK_REQUIRED"
)


post_special_case_residual.loc[
    musa_mask,
    "Post_9_4T_Residual_Class"
] = (
    "SPECIAL_BIOLOGICAL_CASE_NO_STANDARD_NUMERIC_VALUE"
)


# =============================================================================
# 16. RESIDUAL CLASSIFICATION SUMMARY
# =============================================================================

residual_class_summary = (
    post_special_case_residual[
        "Post_9_4T_Residual_Class"
    ]
    .value_counts()
    .rename_axis(
        "Post_9_4T_Residual_Class"
    )
    .reset_index(
        name="N_Canonical_Taxa"
    )
)


print("\nPost-9.4T residual classification")
print("-" * 100)

display(
    residual_class_summary
)


# =============================================================================
# 17. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4T = 92

N_NEW_ASSIGNMENTS_9_4T = 0

N_RESOLVED_AFTER_9_4T = (
    N_RESOLVED_BEFORE_9_4T
    +
    N_NEW_ASSIGNMENTS_9_4T
)


assert N_RESOLVED_AFTER_9_4T == 92


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4T
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 18. STRONG RECONCILIATION
# =============================================================================

assert (
    N_RESOLVED_AFTER_9_4T
    +
    len(
        post_special_case_residual
    )
    ==
    127
)


assert special_case_review[
    "Seed_Mass_Provenance_Class"
].eq(
    "SPECIAL_BIOLOGICAL_CASE"
).all()


# =============================================================================
# 19. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Special biological-case taxa reviewed",
            "Direct Musa × paradisiaca evidence rows",
            "Musa genus-context evidence rows",
            "Numerical Seed Mass values assigned in 9.4T",
            "Special biological-case taxa retained without numeric value",
            "Resolved canonical taxa after 9.4T",
            "Residual canonical taxa after 9.4T",
            "Overall Seed Mass coverage percent",
        ],

        "Value": [
            1,

            len(
                direct_taxon_evidence
            ),

            len(
                genus_context
            ),

            0,

            1,

            N_RESOLVED_AFTER_9_4T,

            len(
                post_special_case_residual
            ),

            coverage_percent,
        ],
    }
)


print("\nSection 9.4T summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 20. EXPORT
# =============================================================================

special_case_review.to_csv(
    OUT_SPECIAL_CASE_REVIEW,
    index=False
)


post_special_case_state.to_csv(
    OUT_POST_SPECIAL_CASE_STATE,
    index=False
)


post_special_case_residual.to_csv(
    OUT_POST_SPECIAL_CASE_RESIDUAL,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 21. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4T exports")
print("-" * 100)

print(
    f"Special-case review        : "
    f"{OUT_SPECIAL_CASE_REVIEW}"
)

print(
    f"Post-special-case state    : "
    f"{OUT_POST_SPECIAL_CASE_STATE}"
)

print(
    f"Post-special-case residual : "
    f"{OUT_POST_SPECIAL_CASE_RESIDUAL}"
)

print(
    f"Summary                    : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 22. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Musa × paradisiaca special biological-case "
    "Seed Mass review completed successfully."
)

print(
    f"Direct taxon-specific Seed Mass evidence : "
    f"{len(direct_taxon_evidence)}"
)

print(
    f"Musa genus-context evidence rows         : "
    f"{len(genus_context)}"
)

print(
    "Numerical Seed Mass assigned             : 0"
)

print(
    "Trait status                             : "
    "special biological case / no standard numerical value"
)

print(
    f"Resolved canonical taxa                  : "
    f"{N_RESOLVED_AFTER_9_4T} / {N_CANONICAL_TAXA}"
)

print(
    f"Overall Seed Mass coverage               : "
    f"{coverage_percent:.2f}%"
)

print(
    f"Residual canonical taxa                  : "
    f"{len(post_special_case_residual)}"
)

print(
    "Musa × paradisiaca is retained without a numerical "
    "Seed Mass rather than receiving an inappropriate "
    "congeneric or family fallback."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4U
# CONSOLIDATE TRUE POST-RESCUE RESIDUAL GAP
# AND DEFINE BROADER FALLBACK HIERARCHY
# =============================================================================
#
# Purpose
# -------
# Consolidate the final residual Seed Mass state after:
#
#   - species-level BIEN / TRY acquisition
#   - external species rescue
#   - genus-level evidence review
#   - family-level evidence review
#   - approved Malvaceae family assignment
#   - Fabaceae high-dispersion review
#   - final taxonomic-name rescue
#   - Musa × paradisiaca special biological-case review
#
# Current state entering 9.4U:
#
#     Resolved canonical taxa : 92 / 127
#     Residual canonical taxa : 35
#
# Of the 35 residual taxa:
#
#     34 = ordinary unresolved taxa requiring broader fallback assessment
#      1 = special biological case:
#              Musa sapiensis -> Musa × paradisiaca
#
# Section 9.4U performs CLASSIFICATION ONLY.
#
# No numerical Seed Mass value is assigned here.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

POST_SPECIAL_CASE_RESIDUAL_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_special_case_residual_canonical_taxa.csv"
)

FAMILY_TRANSFER_DECISIONS_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_transfer_decisions.csv"
)

FAMILY_CANONICAL_MAP_FILE = (
    OUT_DIR
    / "omo_seed_mass_family_audit_canonical_family_map_authoritative.csv"
)

FABACEAE_REVIEW_FILE = (
    OUT_DIR
    / "omo_seed_mass_fabaceae_high_dispersion_target_review.csv"
)

FINAL_TAXONOMIC_RESCUE_FILE = (
    OUT_DIR
    / "omo_seed_mass_final_taxonomic_rescue_unresolved.csv"
)

SPECIAL_CASE_FILE = (
    OUT_DIR
    / "omo_seed_mass_musa_paradisiaca_special_case_review.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_CANONICAL_FALLBACK_CLASSIFICATION = (
    OUT_DIR
    / "omo_seed_mass_post_rescue_fallback_classification.csv"
)

OUT_FALLBACK_ROUTE_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_post_rescue_fallback_route_summary.csv"
)

OUT_BROADER_FALLBACK_TARGETS = (
    OUT_DIR
    / "omo_seed_mass_broader_fallback_numeric_targets.csv"
)

OUT_SPECIAL_NONNUMERIC_CASES = (
    OUT_DIR
    / "omo_seed_mass_special_nonnumeric_cases.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_post_rescue_fallback_classification_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    POST_SPECIAL_CASE_RESIDUAL_FILE,
    FAMILY_TRANSFER_DECISIONS_FILE,
    FAMILY_CANONICAL_MAP_FILE,
    FABACEAE_REVIEW_FILE,
    FINAL_TAXONOMIC_RESCUE_FILE,
    SPECIAL_CASE_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4U input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUT TABLES
# =============================================================================

residual = pd.read_csv(
    POST_SPECIAL_CASE_RESIDUAL_FILE,
    low_memory=False
)

family_decisions = pd.read_csv(
    FAMILY_TRANSFER_DECISIONS_FILE,
    low_memory=False
)

family_map = pd.read_csv(
    FAMILY_CANONICAL_MAP_FILE,
    low_memory=False
)

fabaceae_review = pd.read_csv(
    FABACEAE_REVIEW_FILE,
    low_memory=False
)

taxonomic_rescue_unresolved = pd.read_csv(
    FINAL_TAXONOMIC_RESCUE_FILE,
    low_memory=False
)

special_case = pd.read_csv(
    SPECIAL_CASE_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4U — CONSOLIDATE TRUE POST-RESCUE RESIDUAL GAP "
    "AND DEFINE BROADER FALLBACK HIERARCHY"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Residual canonical taxa after 9.4T : "
    f"{len(residual)}"
)

print(
    f"Family decision rows               : "
    f"{len(family_decisions)}"
)

print(
    f"Canonical family-map rows          : "
    f"{len(family_map)}"
)

print(
    f"Fabaceae review rows               : "
    f"{len(fabaceae_review)}"
)

print(
    f"Taxonomic-rescue unresolved rows   : "
    f"{len(taxonomic_rescue_unresolved)}"
)

print(
    f"Special biological-case rows       : "
    f"{len(special_case)}"
)


# =============================================================================
# 4. VALIDATE EXPECTED STATE
# =============================================================================

assert len(
    residual
) == 35, (
    f"Expected 35 residual canonical taxa after 9.4T; "
    f"found {len(residual)}."
)


assert residual[
    "Canonical_Species"
].nunique() == 35


assert len(
    family_decisions
) == 15


assert len(
    special_case
) == 1


# =============================================================================
# 5. CLEAN KEY FIELDS
# =============================================================================

def clean_text(series):

    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "":
                    pd.NA,
                "nan":
                    pd.NA,
                "None":
                    pd.NA,
                "<NA>":
                    pd.NA,
            }
        )
    )


for df in [
    residual,
    family_map,
    fabaceae_review,
    taxonomic_rescue_unresolved,
    special_case,
]:

    if "Canonical_Species" in df.columns:

        df[
            "Canonical_Species"
        ] = clean_text(
            df[
                "Canonical_Species"
            ]
        )


# =============================================================================
# 6. BUILD AUTHORITATIVE RESIDUAL BASE
# =============================================================================

fallback_state = (
    residual
    .copy()
)


# Remove any rerun-owned fields.
u_owned_fields = [
    "Fallback_Family",
    "Fallback_Family_Evidence_Class",
    "Fallback_Family_Transfer_Decision",
    "Fallback_Route",
    "Fallback_Priority",
    "Fallback_Action",
    "Fallback_Rationale",
    "Numeric_Fallback_Eligible",
    "Numerical_Seed_Mass_Assigned_In_9_4U",
]


fallback_state = (
    fallback_state
    .drop(
        columns=[
            col
            for col in u_owned_fields
            if col in fallback_state.columns
        ],
        errors="ignore"
    )
    .copy()
)


# =============================================================================
# 7. ADD FAMILY INFORMATION WHERE AVAILABLE
# =============================================================================

family_lookup = (
    family_map[
        [
            "Canonical_Species",
            "Family",
        ]
    ]
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ]
    )
    .rename(
        columns={
            "Family":
                "Fallback_Family"
        }
    )
)


fallback_state = (
    fallback_state
    .merge(
        family_lookup,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


# =============================================================================
# 8. ADD FAMILY EVIDENCE / TRANSFER DECISION
# =============================================================================

family_decision_lookup = (
    family_decisions[
        [
            "Family",
            "Family_Evidence_Class",
            "Family_Transfer_Decision",
            "N_Independent_Species",
            "N_Independent_Genera",
            "Log10_Range",
        ]
    ]
    .rename(
        columns={
            "Family":
                "Fallback_Family",

            "Family_Evidence_Class":
                "Fallback_Family_Evidence_Class",

            "Family_Transfer_Decision":
                "Fallback_Family_Transfer_Decision",

            "N_Independent_Species":
                "Fallback_Family_N_Independent_Species",

            "N_Independent_Genera":
                "Fallback_Family_N_Independent_Genera",

            "Log10_Range":
                "Fallback_Family_Log10_Range",
        }
    )
)


fallback_state = (
    fallback_state
    .merge(
        family_decision_lookup,
        on="Fallback_Family",
        how="left",
        validate="many_to_one"
    )
)


# =============================================================================
# 9. IDENTIFY SPECIAL CASE SETS
# =============================================================================

special_case_set = set(
    special_case[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


fabaceae_set = set(
    fabaceae_review[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


taxonomic_rescue_set = set(
    taxonomic_rescue_unresolved[
        "Canonical_Species"
    ]
    .dropna()
    .astype(str)
)


assert special_case_set == {
    "Musa sapiensis"
}


assert len(
    fabaceae_set
) == 5


assert len(
    taxonomic_rescue_set
) == 5


# =============================================================================
# 10. CLASSIFY FALLBACK ROUTES
# =============================================================================
#
# Hierarchy
# ---------
#
# 1. SPECIAL_BIOLOGICAL_CASE_NO_NUMERIC_FALLBACK
#
# 2. TAXONOMIC_RESCUE_EXHAUSTED_BROADER_FALLBACK
#
# 3. FABACEAE_HIGH_DISPERSION_BROADER_FALLBACK
#
# 4. FAMILY_EVIDENCE_INSUFFICIENT_BROADER_FALLBACK
#
# 5. NO_FAMILY_EVIDENCE_BROADER_FALLBACK
#
# 6. OTHER_BROADER_FALLBACK_REVIEW
#
# No numerical value is assigned in 9.4U.
#
# =============================================================================

def classify_fallback(
    row
):

    canonical = str(
        row[
            "Canonical_Species"
        ]
    )

    family_decision = row.get(
        "Fallback_Family_Transfer_Decision",
        pd.NA
    )


    # -------------------------------------------------------------------------
    # 1. Special biological case
    # -------------------------------------------------------------------------

    if canonical in special_case_set:

        return pd.Series(
            {
                "Fallback_Route":
                    "SPECIAL_BIOLOGICAL_CASE_NO_NUMERIC_FALLBACK",

                "Fallback_Priority":
                    0,

                "Fallback_Action":
                    "RETAIN_WITHOUT_STANDARD_NUMERIC_SEED_MASS",

                "Fallback_Rationale":
                    (
                        "Musa × paradisiaca is treated as a functionally "
                        "seedless or biologically non-standard Seed Mass case."
                    ),

                "Numeric_Fallback_Eligible":
                    False,
            }
        )


    # -------------------------------------------------------------------------
    # 2. Taxonomic rescue exhausted
    # -------------------------------------------------------------------------

    if canonical in taxonomic_rescue_set:

        return pd.Series(
            {
                "Fallback_Route":
                    "TAXONOMIC_RESCUE_EXHAUSTED_BROADER_FALLBACK",

                "Fallback_Priority":
                    1,

                "Fallback_Action":
                    "ASSESS_BROADER_NUMERIC_FALLBACK",

                "Fallback_Rationale":
                    (
                        "Accepted-name and synonym-level rescue was exhausted "
                        "without direct species Seed Mass evidence."
                    ),

                "Numeric_Fallback_Eligible":
                    True,
            }
        )


    # -------------------------------------------------------------------------
    # 3. Fabaceae high-dispersion residual
    # -------------------------------------------------------------------------

    if canonical in fabaceae_set:

        return pd.Series(
            {
                "Fallback_Route":
                    "FABACEAE_HIGH_DISPERSION_BROADER_FALLBACK",

                "Fallback_Priority":
                    2,

                "Fallback_Action":
                    "ASSESS_BROADER_NUMERIC_FALLBACK",

                "Fallback_Rationale":
                    (
                        "Fabaceae has strong evidence depth but extreme "
                        "within-family Seed Mass dispersion; genus-level "
                        "evidence was also insufficient for these targets."
                    ),

                "Numeric_Fallback_Eligible":
                    True,
            }
        )


    # -------------------------------------------------------------------------
    # 4. Insufficient family evidence
    # -------------------------------------------------------------------------

    if (
        pd.notna(
            family_decision
        )
        and
        family_decision
        ==
        "INSUFFICIENT_FAMILY_EVIDENCE"
    ):

        return pd.Series(
            {
                "Fallback_Route":
                    "FAMILY_EVIDENCE_INSUFFICIENT_BROADER_FALLBACK",

                "Fallback_Priority":
                    3,

                "Fallback_Action":
                    "ASSESS_BROADER_NUMERIC_FALLBACK",

                "Fallback_Rationale":
                    (
                        "Available family evidence did not meet conservative "
                        "family-transfer criteria."
                    ),

                "Numeric_Fallback_Eligible":
                    True,
            }
        )


    # -------------------------------------------------------------------------
    # 5. No family evidence
    # -------------------------------------------------------------------------

    if (
        pd.notna(
            family_decision
        )
        and
        family_decision
        ==
        "NO_FAMILY_EVIDENCE"
    ):

        return pd.Series(
            {
                "Fallback_Route":
                    "NO_FAMILY_EVIDENCE_BROADER_FALLBACK",

                "Fallback_Priority":
                    4,

                "Fallback_Action":
                    "ASSESS_BROADER_NUMERIC_FALLBACK",

                "Fallback_Rationale":
                    (
                        "No validated family-level Seed Mass evidence was "
                        "available in the 63-species evidence pool."
                    ),

                "Numeric_Fallback_Eligible":
                    True,
            }
        )


    # -------------------------------------------------------------------------
    # 6. Other residual case
    # -------------------------------------------------------------------------

    return pd.Series(
        {
            "Fallback_Route":
                "OTHER_BROADER_FALLBACK_REVIEW",

            "Fallback_Priority":
                5,

            "Fallback_Action":
                "ASSESS_BROADER_NUMERIC_FALLBACK",

            "Fallback_Rationale":
                (
                    "Residual taxon remains unresolved after species, genus, "
                    "family and special-case review."
                ),

            "Numeric_Fallback_Eligible":
                True,
        }
    )


fallback_fields = (
    fallback_state
    .apply(
        classify_fallback,
        axis=1
    )
)


fallback_state = pd.concat(
    [
        fallback_state,
        fallback_fields,
    ],
    axis=1
)


# =============================================================================
# 11. EXPLICITLY CONFIRM NO VALUE ASSIGNED
# =============================================================================

fallback_state[
    "Numerical_Seed_Mass_Assigned_In_9_4U"
] = False


fallback_state[
    "Assigned_Seed_Mass_mg_9_4U"
] = np.nan


assert fallback_state[
    "Numerical_Seed_Mass_Assigned_In_9_4U"
].eq(
    False
).all()


# =============================================================================
# 12. SPLIT NUMERIC FALLBACK vs SPECIAL NON-NUMERIC CASE
# =============================================================================

broader_fallback_targets = (
    fallback_state[
        fallback_state[
            "Numeric_Fallback_Eligible"
        ]
    ]
    .copy()
    .sort_values(
        [
            "Fallback_Priority",
            "Fallback_Family",
            "Residual_Genus",
            "Canonical_Species",
        ],
        na_position="last"
    )
    .reset_index(drop=True)
)


special_nonnumeric_cases = (
    fallback_state[
        ~fallback_state[
            "Numeric_Fallback_Eligible"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


# =============================================================================
# 13. VALIDATE EXPECTED 34 + 1 SPLIT
# =============================================================================

assert len(
    broader_fallback_targets
) == 34, (
    f"Expected 34 ordinary numeric-fallback candidates; "
    f"found {len(broader_fallback_targets)}."
)


assert len(
    special_nonnumeric_cases
) == 1


assert set(
    special_nonnumeric_cases[
        "Canonical_Species"
    ]
) == {
    "Musa sapiensis"
}


assert (
    len(
        broader_fallback_targets
    )
    +
    len(
        special_nonnumeric_cases
    )
    ==
    35
)


# =============================================================================
# 14. DISPLAY BROADER FALLBACK CLASSIFICATION
# =============================================================================

print("\nTrue post-rescue broader fallback classification")
print("-" * 100)

display(
    fallback_state[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Residual_Genus",
            "Fallback_Family",
            "Fallback_Family_Evidence_Class",
            "Fallback_Family_Transfer_Decision",
            "Fallback_Route",
            "Fallback_Priority",
            "Numeric_Fallback_Eligible",
            "Fallback_Action",
        ]
    ]
    .sort_values(
        [
            "Fallback_Priority",
            "Fallback_Family",
            "Residual_Genus",
            "Canonical_Species",
        ],
        na_position="last"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 15. ROUTE SUMMARY
# =============================================================================

route_summary = (
    fallback_state[
        "Fallback_Route"
    ]
    .value_counts()
    .rename_axis(
        "Fallback_Route"
    )
    .reset_index(
        name="N_Canonical_Taxa"
    )
)


route_summary[
    "Percent_of_35_Residual"
] = (
    route_summary[
        "N_Canonical_Taxa"
    ]
    /
    35
    *
    100
)


print("\nFallback-route summary")
print("-" * 100)

display(
    route_summary
)


# =============================================================================
# 16. FAMILY STRUCTURE OF ORDINARY FALLBACK TARGETS
# =============================================================================

family_fallback_summary = (
    broader_fallback_targets
    .groupby(
        "Fallback_Family",
        dropna=False,
        as_index=False
    )
    .agg(
        N_Canonical_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        N_Residual_Genera=(
            "Residual_Genus",
            "nunique"
        ),

        Residual_Genera=(
            "Residual_Genus",
            lambda x:
                " | ".join(
                    sorted(
                        x.dropna()
                        .astype(str)
                        .unique()
                    )
                )
        ),

        Canonical_Taxa=(
            "Canonical_Species",
            lambda x:
                " | ".join(
                    sorted(
                        x.dropna()
                        .astype(str)
                        .unique()
                    )
                )
        ),

        Fallback_Routes=(
            "Fallback_Route",
            lambda x:
                " | ".join(
                    sorted(
                        x.dropna()
                        .astype(str)
                        .unique()
                    )
                )
        ),
    )
    .sort_values(
        [
            "N_Canonical_Taxa",
            "Fallback_Family",
        ],
        ascending=[
            False,
            True,
        ],
        na_position="last"
    )
    .reset_index(drop=True)
)


print("\nFamily structure of the 34 ordinary broader-fallback targets")
print("-" * 100)

display(
    family_fallback_summary
)


# =============================================================================
# 17. TAXONOMIC-RESCUE RESIDUAL CHECK
# =============================================================================

taxonomic_fallback_rows = (
    fallback_state[
        fallback_state[
            "Canonical_Species"
        ].isin(
            taxonomic_rescue_set
        )
    ]
    .copy()
)


assert len(
    taxonomic_fallback_rows
) == 5


assert taxonomic_fallback_rows[
    "Fallback_Route"
].eq(
    "TAXONOMIC_RESCUE_EXHAUSTED_BROADER_FALLBACK"
).all()


# =============================================================================
# 18. FABACEAE RESIDUAL CHECK
# =============================================================================

fabaceae_fallback_rows = (
    fallback_state[
        fallback_state[
            "Canonical_Species"
        ].isin(
            fabaceae_set
        )
    ]
    .copy()
)


assert len(
    fabaceae_fallback_rows
) == 5


assert fabaceae_fallback_rows[
    "Fallback_Route"
].eq(
    "FABACEAE_HIGH_DISPERSION_BROADER_FALLBACK"
).all()


# =============================================================================
# 19. SPECIAL-CASE CHECK
# =============================================================================

assert special_nonnumeric_cases[
    "Fallback_Route"
].eq(
    "SPECIAL_BIOLOGICAL_CASE_NO_NUMERIC_FALLBACK"
).all()


assert special_nonnumeric_cases[
    "Numeric_Fallback_Eligible"
].eq(
    False
).all()


# =============================================================================
# 20. COVERAGE CHECKPOINT
# =============================================================================

N_CANONICAL_TAXA = 127

N_RESOLVED_BEFORE_9_4U = 92

N_NEW_ASSIGNMENTS_9_4U = 0

N_RESOLVED_AFTER_9_4U = (
    N_RESOLVED_BEFORE_9_4U
    +
    N_NEW_ASSIGNMENTS_9_4U
)


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4U
    /
    N_CANONICAL_TAXA
)


assert N_RESOLVED_AFTER_9_4U == 92


assert (
    N_RESOLVED_AFTER_9_4U
    +
    len(
        fallback_state
    )
    ==
    127
)


# =============================================================================
# 21. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Residual canonical taxa entering 9.4U",
            "Ordinary broader numeric-fallback candidates",
            "Special biological nonnumeric cases",
            "Taxonomic-rescue exhausted fallback taxa",
            "Fabaceae high-dispersion fallback taxa",
            "Seed Mass values assigned in 9.4U",
            "Resolved canonical taxa after 9.4U",
            "Residual canonical taxa after 9.4U",
            "Overall Seed Mass coverage percent",
        ],

        "Value": [
            len(
                fallback_state
            ),

            len(
                broader_fallback_targets
            ),

            len(
                special_nonnumeric_cases
            ),

            len(
                taxonomic_fallback_rows
            ),

            len(
                fabaceae_fallback_rows
            ),

            0,

            N_RESOLVED_AFTER_9_4U,

            len(
                fallback_state
            ),

            coverage_percent,
        ],
    }
)


print("\nSection 9.4U summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 22. EXPORT
# =============================================================================

fallback_state.to_csv(
    OUT_CANONICAL_FALLBACK_CLASSIFICATION,
    index=False
)


route_summary.to_csv(
    OUT_FALLBACK_ROUTE_SUMMARY,
    index=False
)


broader_fallback_targets.to_csv(
    OUT_BROADER_FALLBACK_TARGETS,
    index=False
)


special_nonnumeric_cases.to_csv(
    OUT_SPECIAL_NONNUMERIC_CASES,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 23. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4U exports")
print("-" * 100)

print(
    f"Canonical fallback classification : "
    f"{OUT_CANONICAL_FALLBACK_CLASSIFICATION}"
)

print(
    f"Fallback-route summary            : "
    f"{OUT_FALLBACK_ROUTE_SUMMARY}"
)

print(
    f"Broader numeric-fallback targets  : "
    f"{OUT_BROADER_FALLBACK_TARGETS}"
)

print(
    f"Special nonnumeric cases          : "
    f"{OUT_SPECIAL_NONNUMERIC_CASES}"
)

print(
    f"Summary                           : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 24. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "True post-rescue residual Seed Mass gap consolidated "
    "successfully."
)

print(
    f"Residual canonical taxa             : "
    f"{len(fallback_state)}"
)

print(
    f"Ordinary numeric-fallback candidates : "
    f"{len(broader_fallback_targets)}"
)

print(
    f"Special nonnumeric biological cases  : "
    f"{len(special_nonnumeric_cases)}"
)

print(
    f"Taxonomic-rescue fallback taxa       : "
    f"{len(taxonomic_fallback_rows)}"
)

print(
    f"Fabaceae high-dispersion taxa        : "
    f"{len(fabaceae_fallback_rows)}"
)

print(
    "Seed Mass values assigned in 9.4U   : 0"
)

print(
    f"Resolved canonical taxa              : "
    f"{N_RESOLVED_AFTER_9_4U} / {N_CANONICAL_TAXA}"
)

print(
    f"Overall Seed Mass coverage           : "
    f"{coverage_percent:.2f}%"
)

print(
    "The remaining 34 ordinary taxa are now explicitly "
    "classified for broader numeric fallback assessment."
)

print(
    "Musa × paradisiaca remains outside the ordinary "
    "numeric fallback pathway."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4V
# BROADER FALLBACK EVIDENCE DESIGN AND ESTIMATOR SELECTION
# =============================================================================
#
# Purpose
# -------
# The 9.4U residual consolidation identified:
#
#     34 ordinary numeric-fallback candidates
#      1 special biological nonnumeric case
#
# Section 9.4V does NOT assign any new Seed Mass values.
#
# Instead, it evaluates robust global fallback estimators using the validated
# independent biological-species Seed Mass evidence pool.
#
# Why this is necessary
# ---------------------
# Seed Mass spans several orders of magnitude. Therefore:
#
#     arithmetic mean on raw mg scale
#
# can be strongly influenced by very large-seeded taxa.
#
# We therefore compare:
#
#     1. Raw arithmetic mean
#     2. Raw median
#     3. Geometric mean
#     4. Back-transformed mean log10 Seed Mass
#     5. Back-transformed median log10 Seed Mass
#     6. Trimmed raw mean
#     7. Trimmed log10 mean, back-transformed
#
# We also perform sensitivity analyses:
#
#     - full evidence pool
#     - central 90% of the evidence distribution
#     - central 80% of the evidence distribution
#
# Objective
# ---------
# Select a conservative global fallback estimator for possible use in 9.4W.
#
# IMPORTANT
# ---------
# Section 9.4V assigns ZERO canonical Seed Mass values.
#
# Musa sapiensis -> Musa × paradisiaca remains excluded from numerical fallback.
#
# Coverage remains:
#
#     92 / 127 = 72.44%
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

COMPLETE_SPECIES_EVIDENCE_FILE = (
    OUT_DIR
    / "omo_seed_mass_evidence_species_family_mapping_complete.csv"
)

BROADER_FALLBACK_TARGETS_FILE = (
    OUT_DIR
    / "omo_seed_mass_broader_fallback_numeric_targets.csv"
)

SPECIAL_NONNUMERIC_FILE = (
    OUT_DIR
    / "omo_seed_mass_special_nonnumeric_cases.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_EVIDENCE_DISTRIBUTION = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_evidence_distribution.csv"
)

OUT_ESTIMATOR_COMPARISON = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_estimator_comparison.csv"
)

OUT_SENSITIVITY_ANALYSIS = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_sensitivity_analysis.csv"
)

OUT_SELECTED_ESTIMATOR = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_selected_estimator.csv"
)

OUT_FALLBACK_DESIGN = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_design_register.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_estimator_selection_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    COMPLETE_SPECIES_EVIDENCE_FILE,
    BROADER_FALLBACK_TARGETS_FILE,
    SPECIAL_NONNUMERIC_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4V input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

evidence = pd.read_csv(
    COMPLETE_SPECIES_EVIDENCE_FILE,
    low_memory=False
)

fallback_targets = pd.read_csv(
    BROADER_FALLBACK_TARGETS_FILE,
    low_memory=False
)

special_nonnumeric = pd.read_csv(
    SPECIAL_NONNUMERIC_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4V — BROADER FALLBACK EVIDENCE DESIGN "
    "AND ESTIMATOR SELECTION"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Validated independent evidence species : "
    f"{len(evidence)}"
)

print(
    f"Ordinary numeric-fallback candidates   : "
    f"{len(fallback_targets)}"
)

print(
    f"Special nonnumeric biological cases    : "
    f"{len(special_nonnumeric)}"
)


# =============================================================================
# 4. VALIDATE EXPECTED STATE
# =============================================================================

assert len(
    evidence
) == 63, (
    f"Expected 63 independent biological evidence species; "
    f"found {len(evidence)}."
)


assert len(
    fallback_targets
) == 34, (
    f"Expected 34 ordinary numeric-fallback candidates; "
    f"found {len(fallback_targets)}."
)


assert len(
    special_nonnumeric
) == 1


assert set(
    special_nonnumeric[
        "Canonical_Species"
    ]
) == {
    "Musa sapiensis"
}


# =============================================================================
# 5. PREPARE NUMERIC EVIDENCE
# =============================================================================

evidence[
    "Seed_Mass_mg"
] = pd.to_numeric(
    evidence[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


assert evidence[
    "Seed_Mass_mg"
].notna().all()


assert evidence[
    "Seed_Mass_mg"
].gt(
    0
).all()


assert evidence[
    "Biological_Species_Name"
].nunique() == 63


evidence[
    "Log10_Seed_Mass"
] = np.log10(
    evidence[
        "Seed_Mass_mg"
    ]
)


# =============================================================================
# 6. GLOBAL DISTRIBUTION DIAGNOSTICS
# =============================================================================

raw_values = (
    evidence[
        "Seed_Mass_mg"
    ]
    .to_numpy(
        dtype=float
    )
)

log_values = (
    evidence[
        "Log10_Seed_Mass"
    ]
    .to_numpy(
        dtype=float
    )
)


distribution_summary = pd.DataFrame(
    {
        "Metric": [
            "N independent species",
            "Minimum Seed Mass mg",
            "Q05 Seed Mass mg",
            "Q10 Seed Mass mg",
            "Q25 Seed Mass mg",
            "Median Seed Mass mg",
            "Q75 Seed Mass mg",
            "Q90 Seed Mass mg",
            "Q95 Seed Mass mg",
            "Maximum Seed Mass mg",
            "Raw arithmetic mean mg",
            "Raw standard deviation mg",
            "Minimum log10 Seed Mass",
            "Median log10 Seed Mass",
            "Mean log10 Seed Mass",
            "Maximum log10 Seed Mass",
            "Raw fold range",
            "Log10 range",
        ],

        "Value": [
            len(
                raw_values
            ),

            np.min(
                raw_values
            ),

            np.quantile(
                raw_values,
                0.05
            ),

            np.quantile(
                raw_values,
                0.10
            ),

            np.quantile(
                raw_values,
                0.25
            ),

            np.median(
                raw_values
            ),

            np.quantile(
                raw_values,
                0.75
            ),

            np.quantile(
                raw_values,
                0.90
            ),

            np.quantile(
                raw_values,
                0.95
            ),

            np.max(
                raw_values
            ),

            np.mean(
                raw_values
            ),

            np.std(
                raw_values,
                ddof=1
            ),

            np.min(
                log_values
            ),

            np.median(
                log_values
            ),

            np.mean(
                log_values
            ),

            np.max(
                log_values
            ),

            np.max(
                raw_values
            )
            /
            np.min(
                raw_values
            ),

            np.max(
                log_values
            )
            -
            np.min(
                log_values
            ),
        ],
    }
)


print("\nValidated global Seed Mass distribution")
print("-" * 100)

display(
    distribution_summary
)


# =============================================================================
# 7. HELPER — TRIMMED MEAN
# =============================================================================

def trimmed_mean(
    values,
    trim_fraction
):

    values = np.asarray(
        values,
        dtype=float
    )

    values = np.sort(
        values
    )

    n = len(
        values
    )

    k = int(
        np.floor(
            n
            *
            trim_fraction
        )
    )


    if (
        k == 0
        or
        2 * k >= n
    ):

        return float(
            np.mean(
                values
            )
        )


    trimmed = values[
        k:
        n - k
    ]


    return float(
        np.mean(
            trimmed
        )
    )


# =============================================================================
# 8. CALCULATE GLOBAL CANDIDATE ESTIMATORS
# =============================================================================

raw_mean = float(
    np.mean(
        raw_values
    )
)

raw_median = float(
    np.median(
        raw_values
    )
)

mean_log10 = float(
    np.mean(
        log_values
    )
)

median_log10 = float(
    np.median(
        log_values
    )
)

geometric_mean = float(
    10 ** mean_log10
)

backtransformed_log_median = float(
    10 ** median_log10
)

trimmed_raw_mean_10 = trimmed_mean(
    raw_values,
    0.10
)

trimmed_log_mean_10 = trimmed_mean(
    log_values,
    0.10
)

backtransformed_trimmed_log_mean_10 = float(
    10 ** trimmed_log_mean_10
)


estimator_comparison = pd.DataFrame(
    [
        {
            "Estimator_ID":
                "GFB_01",

            "Estimator":
                "RAW_ARITHMETIC_MEAN",

            "Scale":
                "RAW_MG",

            "Candidate_Seed_Mass_mg":
                raw_mean,

            "Robustness_Class":
                "LOW",

            "Recommended_For_Global_Fallback":
                False,

            "Rationale":
                (
                    "Raw arithmetic mean is highly sensitive to very "
                    "large-seeded taxa in a strongly right-skewed trait."
                ),
        },

        {
            "Estimator_ID":
                "GFB_02",

            "Estimator":
                "RAW_MEDIAN",

            "Scale":
                "RAW_MG",

            "Candidate_Seed_Mass_mg":
                raw_median,

            "Robustness_Class":
                "HIGH",

            "Recommended_For_Global_Fallback":
                True,

            "Rationale":
                (
                    "Median is robust to extreme values and represents "
                    "the central observed species on the original mg scale."
                ),
        },

        {
            "Estimator_ID":
                "GFB_03",

            "Estimator":
                "GEOMETRIC_MEAN",

            "Scale":
                "LOG10_BACKTRANSFORMED",

            "Candidate_Seed_Mass_mg":
                geometric_mean,

            "Robustness_Class":
                "MODERATE",

            "Recommended_For_Global_Fallback":
                True,

            "Rationale":
                (
                    "Geometric mean reflects central tendency on the "
                    "multiplicative log scale but remains sensitive to "
                    "the full distribution."
                ),
        },

        {
            "Estimator_ID":
                "GFB_04",

            "Estimator":
                "BACKTRANSFORMED_MEDIAN_LOG10",

            "Scale":
                "LOG10_BACKTRANSFORMED",

            "Candidate_Seed_Mass_mg":
                backtransformed_log_median,

            "Robustness_Class":
                "HIGH",

            "Recommended_For_Global_Fallback":
                True,

            "Rationale":
                (
                    "Median on log10 scale is highly robust and appropriate "
                    "for a trait spanning multiple orders of magnitude."
                ),
        },

        {
            "Estimator_ID":
                "GFB_05",

            "Estimator":
                "TRIMMED_RAW_MEAN_10_PERCENT",

            "Scale":
                "RAW_MG",

            "Candidate_Seed_Mass_mg":
                trimmed_raw_mean_10,

            "Robustness_Class":
                "MODERATE",

            "Recommended_For_Global_Fallback":
                False,

            "Rationale":
                (
                    "10% trimming reduces influence of extremes but still "
                    "operates on the strongly right-skewed raw scale."
                ),
        },

        {
            "Estimator_ID":
                "GFB_06",

            "Estimator":
                "BACKTRANSFORMED_TRIMMED_LOG10_MEAN_10_PERCENT",

            "Scale":
                "LOG10_BACKTRANSFORMED",

            "Candidate_Seed_Mass_mg":
                backtransformed_trimmed_log_mean_10,

            "Robustness_Class":
                "HIGH",

            "Recommended_For_Global_Fallback":
                True,

            "Rationale":
                (
                    "Trimming on log10 scale combines multiplicative trait "
                    "representation with resistance to extreme tails."
                ),
        },
    ]
)


print("\nGlobal fallback estimator comparison")
print("-" * 100)

display(
    estimator_comparison
)


# =============================================================================
# 9. IMPORTANT MATHEMATICAL CHECK
# =============================================================================
#
# Median is invariant to monotonic transformation.
#
# Therefore:
#
#     median(raw Seed Mass)
#
# and
#
#     10 ** median(log10 Seed Mass)
#
# should be numerically equal apart from floating-point rounding.
#
# =============================================================================

assert np.isclose(
    raw_median,
    backtransformed_log_median,
    rtol=0,
    atol=1e-9
), (
    "Raw median and back-transformed log10 median should be equal."
)


# =============================================================================
# 10. CENTRAL-DISTRIBUTION SENSITIVITY ANALYSIS
# =============================================================================

def central_subset(
    dataframe,
    lower_quantile,
    upper_quantile
):

    lower = dataframe[
        "Seed_Mass_mg"
    ].quantile(
        lower_quantile
    )

    upper = dataframe[
        "Seed_Mass_mg"
    ].quantile(
        upper_quantile
    )


    subset = (
        dataframe[
            dataframe[
                "Seed_Mass_mg"
            ]
            .between(
                lower,
                upper,
                inclusive="both"
            )
        ]
        .copy()
    )


    return (
        subset,
        lower,
        upper
    )


sensitivity_rows = []


sensitivity_specs = [
    (
        "FULL_100_PERCENT",
        0.00,
        1.00
    ),
    (
        "CENTRAL_90_PERCENT",
        0.05,
        0.95
    ),
    (
        "CENTRAL_80_PERCENT",
        0.10,
        0.90
    ),
]


for (
    label,
    lower_q,
    upper_q
) in sensitivity_specs:

    if (
        lower_q == 0
        and
        upper_q == 1
    ):

        subset = (
            evidence
            .copy()
        )

        lower_bound = (
            subset[
                "Seed_Mass_mg"
            ]
            .min()
        )

        upper_bound = (
            subset[
                "Seed_Mass_mg"
            ]
            .max()
        )

    else:

        (
            subset,
            lower_bound,
            upper_bound
        ) = central_subset(
            evidence,
            lower_q,
            upper_q
        )


    subset_raw = (
        subset[
            "Seed_Mass_mg"
        ]
        .to_numpy(
            dtype=float
        )
    )

    subset_log = np.log10(
        subset_raw
    )


    sensitivity_rows.append(
        {
            "Sensitivity_Set":
                label,

            "Lower_Quantile":
                lower_q,

            "Upper_Quantile":
                upper_q,

            "N_Species":
                len(
                    subset
                ),

            "Lower_Seed_Mass_Bound_mg":
                lower_bound,

            "Upper_Seed_Mass_Bound_mg":
                upper_bound,

            "Raw_Mean_mg":
                np.mean(
                    subset_raw
                ),

            "Raw_Median_mg":
                np.median(
                    subset_raw
                ),

            "Geometric_Mean_mg":
                10 ** np.mean(
                    subset_log
                ),

            "Backtransformed_Log10_Median_mg":
                10 ** np.median(
                    subset_log
                ),

            "Raw_Fold_Range":
                (
                    np.max(
                        subset_raw
                    )
                    /
                    np.min(
                        subset_raw
                    )
                ),

            "Log10_Range":
                (
                    np.max(
                        subset_log
                    )
                    -
                    np.min(
                        subset_log
                    )
                ),
        }
    )


sensitivity_analysis = pd.DataFrame(
    sensitivity_rows
)


print("\nGlobal estimator sensitivity analysis")
print("-" * 100)

display(
    sensitivity_analysis
)


# =============================================================================
# 11. ESTIMATOR STABILITY DIAGNOSTIC
# =============================================================================
#
# Compare how much each estimator changes when extremes are removed.
#
# A robust fallback estimator should remain relatively stable.
#
# =============================================================================

full_row = (
    sensitivity_analysis[
        sensitivity_analysis[
            "Sensitivity_Set"
        ].eq(
            "FULL_100_PERCENT"
        )
    ]
    .iloc[0]
)


central90_row = (
    sensitivity_analysis[
        sensitivity_analysis[
            "Sensitivity_Set"
        ].eq(
            "CENTRAL_90_PERCENT"
        )
    ]
    .iloc[0]
)


central80_row = (
    sensitivity_analysis[
        sensitivity_analysis[
            "Sensitivity_Set"
        ].eq(
            "CENTRAL_80_PERCENT"
        )
    ]
    .iloc[0]
)


def percent_change(
    original,
    comparison
):

    return (
        100
        *
        abs(
            comparison
            -
            original
        )
        /
        original
    )


stability = pd.DataFrame(
    [
        {
            "Estimator":
                "RAW_MEAN",

            "Full_Value_mg":
                full_row[
                    "Raw_Mean_mg"
                ],

            "Central90_Value_mg":
                central90_row[
                    "Raw_Mean_mg"
                ],

            "Central80_Value_mg":
                central80_row[
                    "Raw_Mean_mg"
                ],
        },

        {
            "Estimator":
                "RAW_MEDIAN",

            "Full_Value_mg":
                full_row[
                    "Raw_Median_mg"
                ],

            "Central90_Value_mg":
                central90_row[
                    "Raw_Median_mg"
                ],

            "Central80_Value_mg":
                central80_row[
                    "Raw_Median_mg"
                ],
        },

        {
            "Estimator":
                "GEOMETRIC_MEAN",

            "Full_Value_mg":
                full_row[
                    "Geometric_Mean_mg"
                ],

            "Central90_Value_mg":
                central90_row[
                    "Geometric_Mean_mg"
                ],

            "Central80_Value_mg":
                central80_row[
                    "Geometric_Mean_mg"
                ],
        },

        {
            "Estimator":
                "BACKTRANSFORMED_LOG10_MEDIAN",

            "Full_Value_mg":
                full_row[
                    "Backtransformed_Log10_Median_mg"
                ],

            "Central90_Value_mg":
                central90_row[
                    "Backtransformed_Log10_Median_mg"
                ],

            "Central80_Value_mg":
                central80_row[
                    "Backtransformed_Log10_Median_mg"
                ],
        },
    ]
)


stability[
    "Percent_Change_Central90"
] = stability.apply(
    lambda row:
        percent_change(
            row[
                "Full_Value_mg"
            ],
            row[
                "Central90_Value_mg"
            ]
        ),
    axis=1
)


stability[
    "Percent_Change_Central80"
] = stability.apply(
    lambda row:
        percent_change(
            row[
                "Full_Value_mg"
            ],
            row[
                "Central80_Value_mg"
            ]
        ),
    axis=1
)


stability[
    "Mean_Sensitivity_Percent"
] = (
    stability[
        [
            "Percent_Change_Central90",
            "Percent_Change_Central80",
        ]
    ]
    .mean(
        axis=1
    )
)


print("\nEstimator stability under tail removal")
print("-" * 100)

display(
    stability
    .sort_values(
        "Mean_Sensitivity_Percent"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 12. SELECT GLOBAL FALLBACK ESTIMATOR
# =============================================================================
#
# Conservative choice:
#
#     RAW_MEDIAN / BACKTRANSFORMED_MEDIAN_LOG10
#
# Because the median is invariant to monotonic log transformation, both are
# numerically identical.
#
# We retain the LOG-SCALE interpretation in provenance because Seed Mass is
# multiplicative and spans several orders of magnitude.
#
# =============================================================================

SELECTED_ESTIMATOR_ID = (
    "GFB_04"
)

SELECTED_ESTIMATOR = (
    "BACKTRANSFORMED_MEDIAN_LOG10"
)

SELECTED_VALUE_MG = (
    backtransformed_log_median
)


selected_estimator = pd.DataFrame(
    [
        {
            "Selected_Estimator_ID":
                SELECTED_ESTIMATOR_ID,

            "Selected_Estimator":
                SELECTED_ESTIMATOR,

            "Selected_Global_Fallback_Seed_Mass_mg":
                SELECTED_VALUE_MG,

            "Equivalent_Raw_Estimator":
                "RAW_MEDIAN",

            "Evidence_N_Independent_Species":
                len(
                    evidence
                ),

            "Evidence_N_Independent_Genera":
                evidence[
                    "Biological_Genus"
                ].nunique(),

            "Estimator_Scale":
                "LOG10_BACKTRANSFORMED",

            "Estimator_Robustness":
                "HIGH",

            "Fallback_Provenance_Class":
                "GLOBAL_FALLBACK_ESTIMATE",

            "Fallback_Quality_Flag":
                "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK",

            "Automatic_Propagation_In_9_4V":
                False,

            "Estimator_Selection_Rationale":
                (
                    "Seed Mass spans multiple orders of magnitude. "
                    "The median is robust to extreme values, and the "
                    "back-transformed median log10 Seed Mass is numerically "
                    "equivalent to the raw median while retaining an explicit "
                    "multiplicative-trait interpretation."
                ),
        }
    ]
)


print("\nSelected broader fallback estimator")
print("-" * 100)

display(
    selected_estimator
)


# =============================================================================
# 13. FALLBACK DESIGN REGISTER
# =============================================================================
#
# This establishes the intended hierarchy for 9.4W.
#
# =============================================================================

fallback_design = pd.DataFrame(
    [
        {
            "Fallback_Level_Order":
                1,

            "Fallback_Level":
                "DIRECT_OR_VALIDATED_SPECIES_EVIDENCE",

            "Use_Status":
                "ALREADY_EXHAUSTED_FOR_CURRENT_RESIDUALS",

            "Estimator":
                "SPECIES_VALUE",

            "Provenance_Class":
                "SPECIES_LEVEL",

            "Quality_Class":
                "HIGH_OR_MODERATE",
        },

        {
            "Fallback_Level_Order":
                2,

            "Fallback_Level":
                "GENUS_LEVEL",

            "Use_Status":
                "ALREADY_EXHAUSTED_OR_REJECTED_FOR_CURRENT_RESIDUALS",

            "Estimator":
                "GENUS_MEDIAN_WHEN_RULES_PASS",

            "Provenance_Class":
                "GENUS_ESTIMATE",

            "Quality_Class":
                "MODERATE",
        },

        {
            "Fallback_Level_Order":
                3,

            "Fallback_Level":
                "FAMILY_LEVEL",

            "Use_Status":
                (
                    "USED_ONLY_WHERE_CONSERVATIVE_FAMILY_RULES_PASSED; "
                    "CURRENT_RESIDUALS FAILED_OR_LACKED_SUPPORT"
                ),

            "Estimator":
                "ROBUST_FAMILY_LOG_SCALE_CENTRAL_TENDENCY",

            "Provenance_Class":
                "FAMILY_ESTIMATE",

            "Quality_Class":
                "LOW_TO_MODERATE",
        },

        {
            "Fallback_Level_Order":
                4,

            "Fallback_Level":
                "GLOBAL_TRAIT_LEVEL",

            "Use_Status":
                "CANDIDATE_FOR_34_ORDINARY_RESIDUAL_TAXA",

            "Estimator":
                SELECTED_ESTIMATOR,

            "Provenance_Class":
                "GLOBAL_FALLBACK_ESTIMATE",

            "Quality_Class":
                "LOW",
        },

        {
            "Fallback_Level_Order":
                5,

            "Fallback_Level":
                "BIOLOGICAL_NONAPPLICABILITY",

            "Use_Status":
                "MUSA_ONLY",

            "Estimator":
                "NO_NUMERIC_VALUE",

            "Provenance_Class":
                "SPECIAL_BIOLOGICAL_CASE",

            "Quality_Class":
                "NOT_APPLICABLE",
        },
    ]
)


print("\nBroader fallback hierarchy")
print("-" * 100)

display(
    fallback_design
)


# =============================================================================
# 14. CONFIRM TARGET POPULATION FOR POSSIBLE 9.4W PROPAGATION
# =============================================================================

assert fallback_targets[
    "Numeric_Fallback_Eligible"
].eq(
    True
).all()


assert "Musa sapiensis" not in set(
    fallback_targets[
        "Canonical_Species"
    ]
)


print("\nPotential 9.4W numeric fallback target population")
print("-" * 100)

print(
    f"Ordinary canonical taxa eligible for assessment : "
    f"{len(fallback_targets)}"
)

print(
    "Musa sapiensis included                         : "
    "NO"
)


# =============================================================================
# 15. NO ASSIGNMENT IN 9.4V
# =============================================================================

N_NEW_ASSIGNMENTS_9_4V = 0

N_RESOLVED_BEFORE_9_4V = 92

N_CANONICAL_TAXA = 127

N_RESOLVED_AFTER_9_4V = (
    N_RESOLVED_BEFORE_9_4V
    +
    N_NEW_ASSIGNMENTS_9_4V
)


coverage_percent = (
    100
    *
    N_RESOLVED_AFTER_9_4V
    /
    N_CANONICAL_TAXA
)


assert N_RESOLVED_AFTER_9_4V == 92


# =============================================================================
# 16. SECTION SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Independent species in global fallback evidence pool",
            "Independent genera in global fallback evidence pool",
            "Ordinary numeric-fallback canonical targets",
            "Special nonnumeric biological cases",
            "Selected global fallback estimator",
            "Selected global fallback Seed Mass mg",
            "Seed Mass values assigned in 9.4V",
            "Resolved canonical taxa after 9.4V",
            "Overall Seed Mass coverage percent",
        ],

        "Value": [
            len(
                evidence
            ),

            evidence[
                "Biological_Genus"
            ].nunique(),

            len(
                fallback_targets
            ),

            len(
                special_nonnumeric
            ),

            SELECTED_ESTIMATOR,

            SELECTED_VALUE_MG,

            0,

            N_RESOLVED_AFTER_9_4V,

            coverage_percent,
        ],
    }
)


print("\nSection 9.4V summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 17. EXPORT
# =============================================================================

evidence[
    [
        "Biological_Genus",
        "Biological_Species_Name",
        "Seed_Mass_mg",
        "Log10_Seed_Mass",
        "Evidence_Sources",
        "Family",
    ]
].to_csv(
    OUT_EVIDENCE_DISTRIBUTION,
    index=False
)


estimator_comparison.to_csv(
    OUT_ESTIMATOR_COMPARISON,
    index=False
)


sensitivity_analysis.to_csv(
    OUT_SENSITIVITY_ANALYSIS,
    index=False
)


selected_estimator.to_csv(
    OUT_SELECTED_ESTIMATOR,
    index=False
)


fallback_design.to_csv(
    OUT_FALLBACK_DESIGN,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 18. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4V exports")
print("-" * 100)

print(
    f"Evidence distribution       : "
    f"{OUT_EVIDENCE_DISTRIBUTION}"
)

print(
    f"Estimator comparison        : "
    f"{OUT_ESTIMATOR_COMPARISON}"
)

print(
    f"Sensitivity analysis        : "
    f"{OUT_SENSITIVITY_ANALYSIS}"
)

print(
    f"Selected estimator          : "
    f"{OUT_SELECTED_ESTIMATOR}"
)

print(
    f"Fallback hierarchy          : "
    f"{OUT_FALLBACK_DESIGN}"
)

print(
    f"Summary                     : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 19. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Broader Seed Mass fallback evidence design and "
    "estimator selection completed successfully."
)

print(
    f"Global evidence species             : "
    f"{len(evidence)}"
)

print(
    f"Global evidence genera              : "
    f"{evidence['Biological_Genus'].nunique()}"
)

print(
    f"Ordinary fallback target taxa       : "
    f"{len(fallback_targets)}"
)

print(
    f"Selected estimator                  : "
    f"{SELECTED_ESTIMATOR}"
)

print(
    f"Selected fallback Seed Mass         : "
    f"{SELECTED_VALUE_MG:.6f} mg"
)

print(
    "Fallback provenance                 : "
    "GLOBAL_FALLBACK_ESTIMATE"
)

print(
    "Fallback quality                    : "
    "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK"
)

print(
    "Seed Mass values assigned in 9.4V   : 0"
)

print(
    f"Resolved canonical taxa             : "
    f"{N_RESOLVED_AFTER_9_4V} / {N_CANONICAL_TAXA}"
)

print(
    f"Overall Seed Mass coverage          : "
    f"{coverage_percent:.2f}%"
)

print(
    "Musa × paradisiaca remains excluded from numerical fallback."
)

print(
    "The selected estimator is now ready for explicit "
    "propagation decision in Section 9.4W."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4W
# EXPLICIT GLOBAL FALLBACK PROPAGATION AND FINAL NUMERIC COVERAGE
# =============================================================================
#
# Purpose
# -------
# Section 9.4V selected the following conservative global Seed Mass fallback:
#
#     Estimator:
#         BACKTRANSFORMED_MEDIAN_LOG10
#
#     Equivalent raw estimator:
#         RAW_MEDIAN
#
#     Selected value:
#         131.19 mg
#
#     Evidence base:
#         63 independent species
#         58 independent genera
#
#     Provenance:
#         GLOBAL_FALLBACK_ESTIMATE
#
#     Quality:
#         LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK
#
# Section 9.4W now propagates this fallback ONLY to the 34 ordinary
# numeric-fallback canonical taxa identified in Section 9.4U.
#
# Musa sapiensis -> Musa × paradisiaca remains excluded because it was
# explicitly classified as:
#
#     SPECIAL_BIOLOGICAL_CASE_NO_NUMERIC_FALLBACK
#
# Expected state after propagation
# --------------------------------
#
#     Existing numerically resolved taxa : 92
#     Global fallback assignments        : 34
#     Numerically resolved taxa          : 126 / 127
#     Numeric Seed Mass coverage         : 99.21%
#     Nonnumeric special cases           : 1
#
# IMPORTANT
# ---------
# Global fallback values MUST retain their original residual/fallback route.
#
# This allows downstream analyses to distinguish, for example:
#
#     TAXONOMIC_RESCUE_EXHAUSTED_BROADER_FALLBACK
#     FABACEAE_HIGH_DISPERSION_BROADER_FALLBACK
#     FAMILY_EVIDENCE_INSUFFICIENT_BROADER_FALLBACK
#     NO_FAMILY_EVIDENCE_BROADER_FALLBACK
#
# from genuine species-, genus-, or family-supported trait values.
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Inputs
# -----------------------------------------------------------------------------

SELECTED_ESTIMATOR_FILE = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_selected_estimator.csv"
)

BROADER_FALLBACK_TARGETS_FILE = (
    OUT_DIR
    / "omo_seed_mass_broader_fallback_numeric_targets.csv"
)

SPECIAL_NONNUMERIC_FILE = (
    OUT_DIR
    / "omo_seed_mass_special_nonnumeric_cases.csv"
)

POST_SPECIAL_CASE_STATE_FILE = (
    OUT_DIR
    / "omo_seed_mass_post_special_case_state.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_GLOBAL_FALLBACK_ASSIGNMENTS = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_assignments.csv"
)

OUT_POST_GLOBAL_STATE = (
    OUT_DIR
    / "omo_seed_mass_post_global_fallback_state.csv"
)

OUT_FINAL_NUMERIC_RESIDUAL = (
    OUT_DIR
    / "omo_seed_mass_post_global_fallback_nonnumeric_residual.csv"
)

OUT_ROUTE_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_assignment_route_summary.csv"
)

OUT_PROVENANCE_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_assignment_provenance_summary.csv"
)

OUT_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_propagation_summary.csv"
)


# =============================================================================
# 2. VALIDATE INPUT FILES
# =============================================================================

required_files = [
    SELECTED_ESTIMATOR_FILE,
    BROADER_FALLBACK_TARGETS_FILE,
    SPECIAL_NONNUMERIC_FILE,
    POST_SPECIAL_CASE_STATE_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4W input file not found:\n{path}"
    )


# =============================================================================
# 3. LOAD INPUTS
# =============================================================================

selected_estimator = pd.read_csv(
    SELECTED_ESTIMATOR_FILE,
    low_memory=False
)

fallback_targets = pd.read_csv(
    BROADER_FALLBACK_TARGETS_FILE,
    low_memory=False
)

special_nonnumeric = pd.read_csv(
    SPECIAL_NONNUMERIC_FILE,
    low_memory=False
)

post_special_state = pd.read_csv(
    POST_SPECIAL_CASE_STATE_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4W — EXPLICIT GLOBAL FALLBACK PROPAGATION "
    "AND FINAL NUMERIC COVERAGE"
)
print("=" * 100)


print("\nInput structure")
print("-" * 100)

print(
    f"Selected estimator rows              : "
    f"{len(selected_estimator)}"
)

print(
    f"Ordinary numeric-fallback targets    : "
    f"{len(fallback_targets)}"
)

print(
    f"Special nonnumeric biological cases  : "
    f"{len(special_nonnumeric)}"
)

print(
    f"Post-special-case residual/state rows : "
    f"{len(post_special_state)}"
)


# =============================================================================
# 4. VALIDATE EXPECTED INPUT STATE
# =============================================================================

assert len(
    selected_estimator
) == 1, (
    f"Expected exactly one selected global fallback estimator; "
    f"found {len(selected_estimator)}."
)


assert len(
    fallback_targets
) == 34, (
    f"Expected 34 ordinary numeric-fallback targets; "
    f"found {len(fallback_targets)}."
)


assert len(
    special_nonnumeric
) == 1


assert set(
    special_nonnumeric[
        "Canonical_Species"
    ]
) == {
    "Musa sapiensis"
}


assert len(
    post_special_state
) == 35, (
    f"Expected 35 post-special-case residual/state rows; "
    f"found {len(post_special_state)}."
)


assert fallback_targets[
    "Canonical_Species"
].nunique() == 34


assert "Musa sapiensis" not in set(
    fallback_targets[
        "Canonical_Species"
    ]
)


# =============================================================================
# 5. LOAD AND VALIDATE SELECTED GLOBAL FALLBACK ESTIMATOR
# =============================================================================

SELECTED_ESTIMATOR_ID = (
    selected_estimator[
        "Selected_Estimator_ID"
    ]
    .iloc[0]
)


SELECTED_ESTIMATOR = (
    selected_estimator[
        "Selected_Estimator"
    ]
    .iloc[0]
)


GLOBAL_FALLBACK_VALUE_MG = pd.to_numeric(
    selected_estimator[
        "Selected_Global_Fallback_Seed_Mass_mg"
    ],
    errors="coerce"
).iloc[0]


GLOBAL_PROVENANCE_CLASS = (
    selected_estimator[
        "Fallback_Provenance_Class"
    ]
    .iloc[0]
)


GLOBAL_QUALITY_FLAG = (
    selected_estimator[
        "Fallback_Quality_Flag"
    ]
    .iloc[0]
)


assert SELECTED_ESTIMATOR_ID == "GFB_04"


assert SELECTED_ESTIMATOR == (
    "BACKTRANSFORMED_MEDIAN_LOG10"
)


assert pd.notna(
    GLOBAL_FALLBACK_VALUE_MG
)


assert GLOBAL_FALLBACK_VALUE_MG > 0


assert np.isclose(
    GLOBAL_FALLBACK_VALUE_MG,
    131.19,
    rtol=0,
    atol=1e-9
)


assert GLOBAL_PROVENANCE_CLASS == (
    "GLOBAL_FALLBACK_ESTIMATE"
)


assert GLOBAL_QUALITY_FLAG == (
    "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK"
)


print("\nAuthoritative global fallback estimator")
print("-" * 100)

print(
    f"Estimator ID        : "
    f"{SELECTED_ESTIMATOR_ID}"
)

print(
    f"Estimator           : "
    f"{SELECTED_ESTIMATOR}"
)

print(
    f"Fallback Seed Mass  : "
    f"{GLOBAL_FALLBACK_VALUE_MG:.6f} mg"
)

print(
    f"Provenance          : "
    f"{GLOBAL_PROVENANCE_CLASS}"
)

print(
    f"Quality flag        : "
    f"{GLOBAL_QUALITY_FLAG}"
)


# =============================================================================
# 6. VALIDATE FALLBACK TARGET ELIGIBILITY
# =============================================================================

assert "Numeric_Fallback_Eligible" in fallback_targets.columns


assert fallback_targets[
    "Numeric_Fallback_Eligible"
].eq(
    True
).all()


assert "Fallback_Route" in fallback_targets.columns


assert fallback_targets[
    "Fallback_Route"
].notna().all()


assert fallback_targets[
    "Canonical_Species"
].notna().all()


# =============================================================================
# 7. BUILD GLOBAL FALLBACK ASSIGNMENT TABLE
# =============================================================================

global_assignments = (
    fallback_targets
    .copy()
)


global_assignments[
    "Assigned_Seed_Mass_mg"
] = (
    GLOBAL_FALLBACK_VALUE_MG
)


global_assignments[
    "Seed_Mass_Assignment_Level"
] = (
    "GLOBAL_TRAIT_LEVEL"
)


global_assignments[
    "Seed_Mass_Provenance_Class"
] = (
    GLOBAL_PROVENANCE_CLASS
)


global_assignments[
    "Seed_Mass_Quality_Flag"
] = (
    GLOBAL_QUALITY_FLAG
)


global_assignments[
    "Seed_Mass_Assignment_Method"
] = (
    SELECTED_ESTIMATOR
)


global_assignments[
    "Global_Fallback_Estimator_ID"
] = (
    SELECTED_ESTIMATOR_ID
)


global_assignments[
    "Seed_Mass_Assignment_Status"
] = (
    "ASSIGNED_IN_SECTION_9_4W"
)


global_assignments[
    "Numerical_Seed_Mass_Assigned"
] = True


# =============================================================================
# 8. STRONG VALIDATION OF ASSIGNMENT TABLE
# =============================================================================

assert len(
    global_assignments
) == 34


assert global_assignments[
    "Canonical_Species"
].is_unique


assert global_assignments[
    "Assigned_Seed_Mass_mg"
].notna().all()


assert global_assignments[
    "Assigned_Seed_Mass_mg"
].gt(
    0
).all()


assert global_assignments[
    "Assigned_Seed_Mass_mg"
].eq(
    GLOBAL_FALLBACK_VALUE_MG
).all()


assert global_assignments[
    "Seed_Mass_Provenance_Class"
].eq(
    "GLOBAL_FALLBACK_ESTIMATE"
).all()


assert global_assignments[
    "Seed_Mass_Quality_Flag"
].eq(
    "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK"
).all()


assert global_assignments[
    "Numerical_Seed_Mass_Assigned"
].eq(
    True
).all()


assert "Musa sapiensis" not in set(
    global_assignments[
        "Canonical_Species"
    ]
)


# =============================================================================
# 9. DISPLAY GLOBAL FALLBACK ASSIGNMENTS
# =============================================================================

display_cols = [
    "Canonical_Species",
    "Final_Search_Name",
    "Residual_Genus",
    "Fallback_Family",
    "Fallback_Route",
    "Assigned_Seed_Mass_mg",
    "Seed_Mass_Assignment_Level",
    "Seed_Mass_Provenance_Class",
    "Seed_Mass_Quality_Flag",
    "Seed_Mass_Assignment_Method",
]


display_cols = [
    col
    for col in display_cols
    if col in global_assignments.columns
]


print("\nApproved global fallback assignments")
print("-" * 100)

display(
    global_assignments[
        display_cols
    ]
    .sort_values(
        [
            col
            for col in [
                "Fallback_Route",
                "Fallback_Family",
                "Residual_Genus",
                "Canonical_Species",
            ]
            if col in global_assignments.columns
        ]
    )
    .reset_index(drop=True)
)


# =============================================================================
# 10. ASSIGNMENT ROUTE SUMMARY
# =============================================================================

route_summary = (
    global_assignments[
        "Fallback_Route"
    ]
    .value_counts()
    .rename_axis(
        "Fallback_Route"
    )
    .reset_index(
        name="N_Canonical_Taxa"
    )
)


route_summary[
    "Percent_of_34_Global_Assignments"
] = (
    route_summary[
        "N_Canonical_Taxa"
    ]
    /
    len(
        global_assignments
    )
    *
    100
)


print("\nGlobal fallback assignments by prior residual route")
print("-" * 100)

display(
    route_summary
)


# =============================================================================
# 11. VERIFY EXPECTED KEY ROUTES
# =============================================================================

route_counts = dict(
    zip(
        route_summary[
            "Fallback_Route"
        ],
        route_summary[
            "N_Canonical_Taxa"
        ],
    )
)


assert route_counts.get(
    "TAXONOMIC_RESCUE_EXHAUSTED_BROADER_FALLBACK",
    0
) == 5


assert route_counts.get(
    "FABACEAE_HIGH_DISPERSION_BROADER_FALLBACK",
    0
) == 5


# =============================================================================
# 12. PRESERVE SPECIAL NONNUMERIC CASE
# =============================================================================

special_case_final = (
    special_nonnumeric
    .copy()
)


special_case_final[
    "Assigned_Seed_Mass_mg"
] = np.nan


special_case_final[
    "Seed_Mass_Assignment_Level"
] = (
    "BIOLOGICAL_NONAPPLICABILITY"
)


special_case_final[
    "Seed_Mass_Provenance_Class"
] = (
    "SPECIAL_BIOLOGICAL_CASE"
)


special_case_final[
    "Seed_Mass_Quality_Flag"
] = (
    "FUNCTIONALLY_SEEDLESS_OR_NO_COMPARABLE_STANDARDIZED_SEED_MASS"
)


special_case_final[
    "Seed_Mass_Assignment_Method"
] = (
    "NO_NUMERIC_VALUE"
)


special_case_final[
    "Seed_Mass_Assignment_Status"
] = (
    "RETAINED_WITHOUT_NUMERIC_VALUE"
)


special_case_final[
    "Numerical_Seed_Mass_Assigned"
] = False


assert len(
    special_case_final
) == 1


assert special_case_final[
    "Canonical_Species"
].eq(
    "Musa sapiensis"
).all()


assert special_case_final[
    "Assigned_Seed_Mass_mg"
].isna().all()


# =============================================================================
# 13. BUILD POST-GLOBAL-FALLBACK STATE
# =============================================================================
#
# post_special_state contains all 35 taxa that were residual after 9.4T.
#
# We now attach authoritative global assignments to 34 of them.
#
# =============================================================================

post_global_state = (
    post_special_state
    .copy()
)


w_owned_fields = [
    "Assigned_Seed_Mass_mg_9_4W",
    "Seed_Mass_Assignment_Level_9_4W",
    "Seed_Mass_Provenance_Class_9_4W",
    "Seed_Mass_Quality_Flag_9_4W",
    "Seed_Mass_Assignment_Method_9_4W",
    "Global_Fallback_Estimator_ID",
    "Numerical_Seed_Mass_Assigned_In_9_4W",
]


post_global_state = (
    post_global_state
    .drop(
        columns=[
            col
            for col in w_owned_fields
            if col in post_global_state.columns
        ],
        errors="ignore"
    )
    .copy()
)


assignment_lookup = (
    global_assignments[
        [
            "Canonical_Species",
            "Assigned_Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Seed_Mass_Assignment_Method",
            "Global_Fallback_Estimator_ID",
        ]
    ]
    .rename(
        columns={
            "Assigned_Seed_Mass_mg":
                "Assigned_Seed_Mass_mg_9_4W",

            "Seed_Mass_Assignment_Level":
                "Seed_Mass_Assignment_Level_9_4W",

            "Seed_Mass_Provenance_Class":
                "Seed_Mass_Provenance_Class_9_4W",

            "Seed_Mass_Quality_Flag":
                "Seed_Mass_Quality_Flag_9_4W",

            "Seed_Mass_Assignment_Method":
                "Seed_Mass_Assignment_Method_9_4W",
        }
    )
)


post_global_state = (
    post_global_state
    .merge(
        assignment_lookup,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


post_global_state[
    "Numerical_Seed_Mass_Assigned_In_9_4W"
] = (
    post_global_state[
        "Assigned_Seed_Mass_mg_9_4W"
    ]
    .notna()
)


# =============================================================================
# 14. VALIDATE POST-GLOBAL STATE
# =============================================================================

assert len(
    post_global_state
) == 35


assert int(
    post_global_state[
        "Numerical_Seed_Mass_Assigned_In_9_4W"
    ]
    .sum()
) == 34


musa_post_global = (
    post_global_state[
        post_global_state[
            "Canonical_Species"
        ].eq(
            "Musa sapiensis"
        )
    ]
)


assert len(
    musa_post_global
) == 1


assert musa_post_global[
    "Numerical_Seed_Mass_Assigned_In_9_4W"
].eq(
    False
).all()


assert musa_post_global[
    "Assigned_Seed_Mass_mg_9_4W"
].isna().all()


# =============================================================================
# 15. BUILD FINAL NONNUMERIC RESIDUAL
# =============================================================================

final_nonnumeric_residual = (
    post_global_state[
        ~post_global_state[
            "Numerical_Seed_Mass_Assigned_In_9_4W"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(
    final_nonnumeric_residual
) == 1


assert set(
    final_nonnumeric_residual[
        "Canonical_Species"
    ]
) == {
    "Musa sapiensis"
}


print("\nFinal nonnumeric Seed Mass residual")
print("-" * 100)

display(
    final_nonnumeric_residual
)


# =============================================================================
# 16. NUMERIC COVERAGE UPDATE
# =============================================================================

N_CANONICAL_TAXA = 127

N_NUMERIC_BEFORE_9_4W = 92

N_GLOBAL_ASSIGNMENTS_9_4W = len(
    global_assignments
)

N_NUMERIC_AFTER_9_4W = (
    N_NUMERIC_BEFORE_9_4W
    +
    N_GLOBAL_ASSIGNMENTS_9_4W
)


N_NONNUMERIC_AFTER_9_4W = (
    N_CANONICAL_TAXA
    -
    N_NUMERIC_AFTER_9_4W
)


assert N_GLOBAL_ASSIGNMENTS_9_4W == 34


assert N_NUMERIC_AFTER_9_4W == 126


assert N_NONNUMERIC_AFTER_9_4W == 1


numeric_coverage_percent = (
    100
    *
    N_NUMERIC_AFTER_9_4W
    /
    N_CANONICAL_TAXA
)


nonnumeric_percent = (
    100
    *
    N_NONNUMERIC_AFTER_9_4W
    /
    N_CANONICAL_TAXA
)


# =============================================================================
# 17. GLOBAL FALLBACK PROVENANCE SUMMARY
# =============================================================================

provenance_summary = pd.DataFrame(
    [
        {
            "Seed_Mass_Provenance_Class":
                "GLOBAL_FALLBACK_ESTIMATE",

            "Seed_Mass_Quality_Flag":
                "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK",

            "N_Canonical_Taxa":
                len(
                    global_assignments
                ),

            "Assigned_Seed_Mass_mg":
                GLOBAL_FALLBACK_VALUE_MG,

            "Estimator":
                SELECTED_ESTIMATOR,

            "Evidence_N_Independent_Species":
                int(
                    selected_estimator[
                        "Evidence_N_Independent_Species"
                    ].iloc[0]
                ),

            "Evidence_N_Independent_Genera":
                int(
                    selected_estimator[
                        "Evidence_N_Independent_Genera"
                    ].iloc[0]
                ),
        },

        {
            "Seed_Mass_Provenance_Class":
                "SPECIAL_BIOLOGICAL_CASE",

            "Seed_Mass_Quality_Flag":
                (
                    "FUNCTIONALLY_SEEDLESS_OR_NO_COMPARABLE_"
                    "STANDARDIZED_SEED_MASS"
                ),

            "N_Canonical_Taxa":
                1,

            "Assigned_Seed_Mass_mg":
                np.nan,

            "Estimator":
                "NO_NUMERIC_VALUE",

            "Evidence_N_Independent_Species":
                np.nan,

            "Evidence_N_Independent_Genera":
                np.nan,
        },
    ]
)


print("\nSection 9.4W assignment provenance")
print("-" * 100)

display(
    provenance_summary
)


# =============================================================================
# 18. SANITY CHECK — ALL 34 GLOBAL ASSIGNMENTS ARE IDENTICAL
# =============================================================================

assert global_assignments[
    "Assigned_Seed_Mass_mg"
].nunique() == 1


assert np.isclose(
    global_assignments[
        "Assigned_Seed_Mass_mg"
    ].iloc[0],
    GLOBAL_FALLBACK_VALUE_MG,
    atol=1e-12,
    rtol=0
)


# =============================================================================
# 19. IMPORTANT QUALITY CHECK
# =============================================================================
#
# Global fallback values must never masquerade as species observations.
#
# =============================================================================

for forbidden_provenance in [
    "SPECIES_LEVEL",
    "SPECIES_BIEN",
    "SPECIES_TRY",
    "GENUS_ESTIMATE",
    "FAMILY_ESTIMATE",
]:

    assert not global_assignments[
        "Seed_Mass_Provenance_Class"
    ].eq(
        forbidden_provenance
    ).any()


# =============================================================================
# 20. NUMERIC COVERAGE SUMMARY
# =============================================================================

summary = pd.DataFrame(
    {
        "Metric": [
            "Canonical Omo taxa",
            "Numerically resolved taxa before 9.4W",
            "Ordinary global fallback targets",
            "Global fallback Seed Mass assigned mg",
            "Global fallback assignments made in 9.4W",
            "Special biological nonnumeric cases",
            "Numerically resolved taxa after 9.4W",
            "Nonnumeric taxa after 9.4W",
            "Numeric Seed Mass coverage percent",
            "Nonnumeric percent",
        ],

        "Value": [
            N_CANONICAL_TAXA,

            N_NUMERIC_BEFORE_9_4W,

            len(
                fallback_targets
            ),

            GLOBAL_FALLBACK_VALUE_MG,

            N_GLOBAL_ASSIGNMENTS_9_4W,

            len(
                special_nonnumeric
            ),

            N_NUMERIC_AFTER_9_4W,

            N_NONNUMERIC_AFTER_9_4W,

            numeric_coverage_percent,

            nonnumeric_percent,
        ],
    }
)


print("\nSection 9.4W summary")
print("-" * 100)

display(
    summary
)


# =============================================================================
# 21. FINAL RECONCILIATION
# =============================================================================

assert (
    N_NUMERIC_AFTER_9_4W
    +
    N_NONNUMERIC_AFTER_9_4W
    ==
    N_CANONICAL_TAXA
)


assert (
    126
    +
    1
    ==
    127
)


assert np.isclose(
    numeric_coverage_percent,
    99.21259842519684,
    rtol=0,
    atol=1e-9
)


# =============================================================================
# 22. EXPORT
# =============================================================================

global_assignments.to_csv(
    OUT_GLOBAL_FALLBACK_ASSIGNMENTS,
    index=False
)


post_global_state.to_csv(
    OUT_POST_GLOBAL_STATE,
    index=False
)


final_nonnumeric_residual.to_csv(
    OUT_FINAL_NUMERIC_RESIDUAL,
    index=False
)


route_summary.to_csv(
    OUT_ROUTE_SUMMARY,
    index=False
)


provenance_summary.to_csv(
    OUT_PROVENANCE_SUMMARY,
    index=False
)


summary.to_csv(
    OUT_SUMMARY,
    index=False
)


# =============================================================================
# 23. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4W exports")
print("-" * 100)

print(
    f"Global fallback assignments : "
    f"{OUT_GLOBAL_FALLBACK_ASSIGNMENTS}"
)

print(
    f"Post-global state           : "
    f"{OUT_POST_GLOBAL_STATE}"
)

print(
    f"Final nonnumeric residual   : "
    f"{OUT_FINAL_NUMERIC_RESIDUAL}"
)

print(
    f"Route summary               : "
    f"{OUT_ROUTE_SUMMARY}"
)

print(
    f"Provenance summary          : "
    f"{OUT_PROVENANCE_SUMMARY}"
)

print(
    f"Summary                     : "
    f"{OUT_SUMMARY}"
)


# =============================================================================
# 24. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Explicit global Seed Mass fallback propagation "
    "completed successfully."
)

print(
    f"Selected global fallback estimator : "
    f"{SELECTED_ESTIMATOR}"
)

print(
    f"Global fallback Seed Mass           : "
    f"{GLOBAL_FALLBACK_VALUE_MG:.6f} mg"
)

print(
    f"Ordinary taxa receiving fallback    : "
    f"{N_GLOBAL_ASSIGNMENTS_9_4W}"
)

print(
    "Fallback provenance                 : "
    "GLOBAL_FALLBACK_ESTIMATE"
)

print(
    "Fallback quality                    : "
    "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK"
)

print(
    f"Numerically resolved canonical taxa : "
    f"{N_NUMERIC_AFTER_9_4W} / {N_CANONICAL_TAXA}"
)

print(
    f"Numeric Seed Mass coverage          : "
    f"{numeric_coverage_percent:.2f}%"
)

print(
    f"Nonnumeric canonical taxa           : "
    f"{N_NONNUMERIC_AFTER_9_4W}"
)

print(
    "Nonnumeric taxon                    : "
    "Musa sapiensis -> Musa × paradisiaca"
)

print(
    "Musa remains explicitly classified as a special "
    "biological case and has not been imputed."
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4X
# FINAL 127-TAXON SEED MASS CONSOLIDATION AND PROVENANCE VALIDATION
# UPDATED AUTHORITATIVE VERSION
# =============================================================================
#
# Authoritative final structure
# -----------------------------
#
# Species-level evidence:
#
#     TRY final canonical values             46
#     BIEN current consolidated values       28
#     External non-TRY rescue values         14
#                                            --
#                                             88
#
# Additional fallback levels:
#
#     Genus estimates                         2
#     Family estimates                        2
#     Global fallback estimates              34
#     Special biological nonnumeric case      1
#                                            ---
#                                            127
#
# Expected numeric coverage:
#
#     126 / 127 = 99.21%
#
# Only nonnumeric taxon:
#
#     Musa sapiensis -> Musa × paradisiaca
#
# Final evidence precedence:
#
#     SPECIES > GENUS > FAMILY > GLOBAL FALLBACK >
#     SPECIAL BIOLOGICAL CASE
#
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)


# -----------------------------------------------------------------------------
# Authoritative species-level inputs
# -----------------------------------------------------------------------------

TRY_SPECIES_FILE = (
    OUT_DIR
    / "omo_seed_mass_try_51911_final_omo_canonical_values.csv"
)

BIEN_SPECIES_FILE = (
    OUT_DIR
    / "omo_seed_mass_bien_all_species_values_current.csv"
)

EXTERNAL_SPECIES_FILE = (
    OUT_DIR
    / "omo_seed_mass_non_try_external_canonical_values.csv"
)


# -----------------------------------------------------------------------------
# Authoritative estimate / fallback inputs
# -----------------------------------------------------------------------------

GENUS_ASSIGNMENTS_FILE = (
    OUT_DIR
    / "omo_seed_mass_canonical_genus_assignments.csv"
)

FAMILY_ASSIGNMENTS_FILE = (
    OUT_DIR
    / "omo_seed_mass_canonical_family_assignments.csv"
)

GLOBAL_ASSIGNMENTS_FILE = (
    OUT_DIR
    / "omo_seed_mass_global_fallback_assignments.csv"
)

SPECIAL_CASE_FILE = (
    OUT_DIR
    / "omo_seed_mass_special_nonnumeric_cases.csv"
)


# -----------------------------------------------------------------------------
# Outputs
# -----------------------------------------------------------------------------

OUT_SPECIES_CONSOLIDATED = (
    OUT_DIR
    / "omo_seed_mass_final_species_level_consolidated.csv"
)

OUT_FINAL_TABLE = (
    OUT_DIR
    / "omo_seed_mass_final.csv"
)

OUT_PROVENANCE_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_final_provenance_summary.csv"
)

OUT_ASSIGNMENT_LEVEL_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_final_assignment_level_summary.csv"
)

OUT_QUALITY_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_final_quality_summary.csv"
)

OUT_VALIDATION_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_final_validation_summary.csv"
)

OUT_OVERLAP_AUDIT = (
    OUT_DIR
    / "omo_seed_mass_final_component_overlap_audit.csv"
)


# =============================================================================
# 2. HELPERS
# =============================================================================

def clean_text(series):

    return (
        series
        .astype("string")
        .str.strip()
        .replace(
            {
                "":
                    pd.NA,

                "nan":
                    pd.NA,

                "None":
                    pd.NA,

                "<NA>":
                    pd.NA,
            }
        )
    )


def first_existing_column(
    dataframe,
    candidates
):

    for col in candidates:

        if col in dataframe.columns:

            return col

    return None


def safe_series(
    dataframe,
    candidates,
    default
):

    col = first_existing_column(
        dataframe,
        candidates
    )

    if col is not None:

        return dataframe[
            col
        ].copy()

    return pd.Series(
        [default] * len(dataframe),
        index=dataframe.index
    )


# =============================================================================
# 3. VALIDATE ALL REQUIRED FILES
# =============================================================================

required_files = [
    TRY_SPECIES_FILE,
    BIEN_SPECIES_FILE,
    EXTERNAL_SPECIES_FILE,
    GENUS_ASSIGNMENTS_FILE,
    FAMILY_ASSIGNMENTS_FILE,
    GLOBAL_ASSIGNMENTS_FILE,
    SPECIAL_CASE_FILE,
]


for path in required_files:

    assert path.exists(), (
        f"Required Section 9.4X input file not found:\n{path}"
    )


# =============================================================================
# 4. LOAD INPUTS
# =============================================================================

try_raw = pd.read_csv(
    TRY_SPECIES_FILE,
    low_memory=False
)

bien_raw = pd.read_csv(
    BIEN_SPECIES_FILE,
    low_memory=False
)

external_raw = pd.read_csv(
    EXTERNAL_SPECIES_FILE,
    low_memory=False
)

genus_raw = pd.read_csv(
    GENUS_ASSIGNMENTS_FILE,
    low_memory=False
)

family_raw = pd.read_csv(
    FAMILY_ASSIGNMENTS_FILE,
    low_memory=False
)

global_raw = pd.read_csv(
    GLOBAL_ASSIGNMENTS_FILE,
    low_memory=False
)

special_raw = pd.read_csv(
    SPECIAL_CASE_FILE,
    low_memory=False
)


print("=" * 100)

print(
    "SECTION 9.4X — FINAL 127-TAXON SEED MASS "
    "CONSOLIDATION AND PROVENANCE VALIDATION"
)

print("=" * 100)


print("\nAuthoritative component inputs")
print("-" * 100)

print(
    f"TRY species-level rows       : {len(try_raw)}"
)

print(
    f"BIEN species-level rows      : {len(bien_raw)}"
)

print(
    f"External rescue rows         : {len(external_raw)}"
)

print(
    f"Genus assignment rows        : {len(genus_raw)}"
)

print(
    f"Family assignment rows       : {len(family_raw)}"
)

print(
    f"Global fallback rows         : {len(global_raw)}"
)

print(
    f"Special biological-case rows : {len(special_raw)}"
)


# =============================================================================
# 5. VALIDATE RAW INPUT COUNTS
# =============================================================================

assert len(
    try_raw
) == 87


assert len(
    bien_raw
) == 28


assert len(
    external_raw
) == 14


assert len(
    genus_raw
) == 2


assert len(
    family_raw
) == 2


assert len(
    global_raw
) == 34


assert len(
    special_raw
) == 1


# =============================================================================
# 6. STANDARDIZE TRY SPECIES VALUES
# =============================================================================

assert "Canonical_Species" in try_raw.columns
assert "Seed_Mass_mg" in try_raw.columns


try_raw[
    "Canonical_Species"
] = clean_text(
    try_raw[
        "Canonical_Species"
    ]
)


try_raw[
    "Seed_Mass_mg"
] = pd.to_numeric(
    try_raw[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


try_species = (
    try_raw[
        try_raw[
            "Seed_Mass_mg"
        ].gt(0)
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(
    try_species
) == 46, (
    f"Expected 46 positive TRY canonical values; "
    f"found {len(try_species)}."
)


assert try_species[
    "Canonical_Species"
].nunique() == 46


try_final_search = safe_series(
    try_species,
    [
        "Final_Search_Name",
        "Trait_Source_Name",
        "Biological_Species_Name",
        "TRY_Taxon_Name",
    ],
    pd.NA
)


try_source = safe_series(
    try_species,
    [
        "Evidence_Sources",
        "Source_Name",
        "Trait_Source_Name",
        "Source",
    ],
    "TRY"
)


try_quality = safe_series(
    try_species,
    [
        "Seed_Mass_Quality_Flag",
        "Trait_Quality_Flag",
        "Evidence_QC_Flag",
        "Quality_Flag",
    ],
    "ACCEPTED_TRY_SPECIES_EVIDENCE"
)


try_method = safe_series(
    try_species,
    [
        "Seed_Mass_Assignment_Method",
        "Aggregation_Method",
        "Estimator",
    ],
    "SPECIES_LEVEL_TRY_AGGREGATION"
)


try_final = pd.DataFrame(
    {
        "Canonical_Species":
            try_species[
                "Canonical_Species"
            ],

        "Final_Search_Name":
            try_final_search,

        "Seed_Mass_mg":
            try_species[
                "Seed_Mass_mg"
            ],

        "Seed_Mass_Assignment_Level":
            "SPECIES_LEVEL",

        "Seed_Mass_Provenance_Class":
            "SPECIES_TRY",

        "Seed_Mass_Quality_Flag":
            try_quality,

        "Seed_Mass_Assignment_Method":
            try_method,

        "Seed_Mass_Evidence_Source":
            try_source,

        "Final_Precedence_Rank":
            1,
    }
)


# =============================================================================
# 7. STANDARDIZE BIEN SPECIES VALUES
# =============================================================================

assert "Canonical_Species" in bien_raw.columns
assert "Seed_Mass_mg" in bien_raw.columns


bien_raw[
    "Canonical_Species"
] = clean_text(
    bien_raw[
        "Canonical_Species"
    ]
)


bien_raw[
    "Seed_Mass_mg"
] = pd.to_numeric(
    bien_raw[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


bien_species = (
    bien_raw[
        bien_raw[
            "Seed_Mass_mg"
        ].gt(0)
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(
    bien_species
) == 28


assert bien_species[
    "Canonical_Species"
].nunique() == 28


bien_final_search = safe_series(
    bien_species,
    [
        "Final_Search_Name",
        "Validated_Search_Name",
        "Biological_Species_Name",
        "Trait_Source_Name",
    ],
    pd.NA
)


bien_source = safe_series(
    bien_species,
    [
        "Evidence_Sources",
        "Source_Name",
        "Source",
        "Trait_Source_Name",
    ],
    "BIEN"
)


bien_quality = safe_series(
    bien_species,
    [
        "Seed_Mass_Quality_Flag",
        "Trait_Quality_Flag",
        "Evidence_QC_Flag",
        "Quality_Flag",
    ],
    "ACCEPTED_BIEN_SPECIES_EVIDENCE"
)


bien_method = safe_series(
    bien_species,
    [
        "Seed_Mass_Assignment_Method",
        "Aggregation_Method",
        "Estimator",
    ],
    "SPECIES_LEVEL_BIEN_AGGREGATION"
)


bien_final = pd.DataFrame(
    {
        "Canonical_Species":
            bien_species[
                "Canonical_Species"
            ],

        "Final_Search_Name":
            bien_final_search,

        "Seed_Mass_mg":
            bien_species[
                "Seed_Mass_mg"
            ],

        "Seed_Mass_Assignment_Level":
            "SPECIES_LEVEL",

        "Seed_Mass_Provenance_Class":
            "SPECIES_BIEN",

        "Seed_Mass_Quality_Flag":
            bien_quality,

        "Seed_Mass_Assignment_Method":
            bien_method,

        "Seed_Mass_Evidence_Source":
            bien_source,

        "Final_Precedence_Rank":
            1,
    }
)


# =============================================================================
# 8. STANDARDIZE EXTERNAL SPECIES RESCUE VALUES
# =============================================================================

assert "Canonical_Species" in external_raw.columns
assert "Seed_Mass_mg" in external_raw.columns


external_raw[
    "Canonical_Species"
] = clean_text(
    external_raw[
        "Canonical_Species"
    ]
)


external_raw[
    "Seed_Mass_mg"
] = pd.to_numeric(
    external_raw[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


external_species = (
    external_raw[
        external_raw[
            "Seed_Mass_mg"
        ].gt(0)
    ]
    .copy()
    .reset_index(drop=True)
)


assert len(
    external_species
) == 14


assert external_species[
    "Canonical_Species"
].nunique() == 14


external_final_search = safe_series(
    external_species,
    [
        "Final_Search_Name",
        "Validated_Search_Name",
        "Biological_Species_Name",
        "Source_Reported_Taxon",
    ],
    pd.NA
)


external_source = safe_series(
    external_species,
    [
        "Evidence_Sources",
        "Source_Name",
        "Source",
        "Source_Class",
    ],
    "EXTERNAL_SPECIES_RESCUE"
)


external_quality = safe_series(
    external_species,
    [
        "Seed_Mass_Quality_Flag",
        "Evidence_QC_Flag",
        "Trait_Quality_Flag",
        "Quality_Flag",
    ],
    "ACCEPTED_EXTERNAL_SPECIES_RESCUE"
)


external_method = safe_series(
    external_species,
    [
        "Seed_Mass_Assignment_Method",
        "Aggregation_Method",
        "Estimator",
    ],
    "SPECIES_LEVEL_EXTERNAL_RESCUE"
)


external_final = pd.DataFrame(
    {
        "Canonical_Species":
            external_species[
                "Canonical_Species"
            ],

        "Final_Search_Name":
            external_final_search,

        "Seed_Mass_mg":
            external_species[
                "Seed_Mass_mg"
            ],

        "Seed_Mass_Assignment_Level":
            "SPECIES_LEVEL",

        "Seed_Mass_Provenance_Class":
            "SPECIES_EXTERNAL_RESCUE",

        "Seed_Mass_Quality_Flag":
            external_quality,

        "Seed_Mass_Assignment_Method":
            external_method,

        "Seed_Mass_Evidence_Source":
            external_source,

        "Final_Precedence_Rank":
            1,
    }
)


# =============================================================================
# 9. VALIDATE SPECIES-LEVEL SOURCE DISJOINTNESS
# =============================================================================

try_set = set(
    try_final[
        "Canonical_Species"
    ]
)


bien_set = set(
    bien_final[
        "Canonical_Species"
    ]
)


external_set = set(
    external_final[
        "Canonical_Species"
    ]
)


species_overlap_rows = []


for name_a, set_a, name_b, set_b in [
    (
        "TRY",
        try_set,
        "BIEN",
        bien_set,
    ),
    (
        "TRY",
        try_set,
        "EXTERNAL",
        external_set,
    ),
    (
        "BIEN",
        bien_set,
        "EXTERNAL",
        external_set,
    ),
]:

    overlap = sorted(
        set_a
        &
        set_b
    )


    species_overlap_rows.append(
        {
            "Source_A":
                name_a,

            "Source_B":
                name_b,

            "N_Overlap":
                len(
                    overlap
                ),

            "Overlapping_Canonical_Taxa":
                " | ".join(
                    overlap
                ),
        }
    )


species_overlap_audit = pd.DataFrame(
    species_overlap_rows
)


print("\nSpecies-source overlap audit")
print("-" * 100)

display(
    species_overlap_audit
)


assert species_overlap_audit[
    "N_Overlap"
].eq(
    0
).all(), (
    "Species-level source overlap detected. "
    "TRY, BIEN and external rescue were expected "
    "to be mutually exclusive."
)


# =============================================================================
# 10. CONSOLIDATE FINAL SPECIES-LEVEL BLOCK
# =============================================================================

species_final = pd.concat(
    [
        try_final,
        bien_final,
        external_final,
    ],
    ignore_index=True
)


species_final = (
    species_final
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


assert len(
    species_final
) == 88


assert species_final[
    "Canonical_Species"
].nunique() == 88


assert species_final[
    "Seed_Mass_mg"
].notna().all()


assert species_final[
    "Seed_Mass_mg"
].gt(
    0
).all()


print("\nAuthoritative consolidated species-level block")
print("-" * 100)

print(
    f"TRY species             : {len(try_final)}"
)

print(
    f"BIEN species            : {len(bien_final)}"
)

print(
    f"External rescue species : {len(external_final)}"
)

print(
    f"Total species-level     : {len(species_final)}"
)


species_final.to_csv(
    OUT_SPECIES_CONSOLIDATED,
    index=False
)


# =============================================================================
# 11. STANDARDIZE GENUS ASSIGNMENTS
# =============================================================================

genus_value_col = first_existing_column(
    genus_raw,
    [
        "Assigned_Seed_Mass_mg",
        "Seed_Mass_mg",
    ]
)


assert genus_value_col is not None


genus_final = pd.DataFrame(
    {
        "Canonical_Species":
            clean_text(
                genus_raw[
                    "Canonical_Species"
                ]
            ),

        "Final_Search_Name":
            safe_series(
                genus_raw,
                [
                    "Final_Search_Name",
                    "Genus_Search_Target",
                ],
                pd.NA
            ),

        "Seed_Mass_mg":
            pd.to_numeric(
                genus_raw[
                    genus_value_col
                ],
                errors="coerce"
            ),

        "Seed_Mass_Assignment_Level":
            "GENUS_LEVEL",

        "Seed_Mass_Provenance_Class":
            "GENUS_ESTIMATE",

        "Seed_Mass_Quality_Flag":
            safe_series(
                genus_raw,
                [
                    "Seed_Mass_Quality_Flag",
                ],
                "MODERATE_GENUS_EVIDENCE"
            ),

        "Seed_Mass_Assignment_Method":
            safe_series(
                genus_raw,
                [
                    "Assignment_Method",
                    "Seed_Mass_Assignment_Method",
                ],
                "MEDIAN_OF_INDEPENDENT_CONGENERIC_SPECIES"
            ),

        "Seed_Mass_Evidence_Source":
            safe_series(
                genus_raw,
                [
                    "Evidence_Sources",
                ],
                "GENUS_LEVEL_EVIDENCE"
            ),

        "Final_Precedence_Rank":
            2,
    }
)


assert len(
    genus_final
) == 2


assert genus_final[
    "Seed_Mass_mg"
].gt(
    0
).all()


# =============================================================================
# 12. STANDARDIZE FAMILY ASSIGNMENTS
# =============================================================================

family_value_col = first_existing_column(
    family_raw,
    [
        "Assigned_Seed_Mass_mg",
        "Approved_Family_Seed_Mass_Candidate_mg",
        "Seed_Mass_mg",
    ]
)


assert family_value_col is not None


family_final = pd.DataFrame(
    {
        "Canonical_Species":
            clean_text(
                family_raw[
                    "Canonical_Species"
                ]
            ),

        "Final_Search_Name":
            safe_series(
                family_raw,
                [
                    "Final_Search_Name",
                ],
                pd.NA
            ),

        "Seed_Mass_mg":
            pd.to_numeric(
                family_raw[
                    family_value_col
                ],
                errors="coerce"
            ),

        "Seed_Mass_Assignment_Level":
            "FAMILY_LEVEL",

        "Seed_Mass_Provenance_Class":
            "FAMILY_ESTIMATE",

        "Seed_Mass_Quality_Flag":
            safe_series(
                family_raw,
                [
                    "Seed_Mass_Quality_Flag",
                    "Family_Quality_Flag",
                ],
                "STRONG_MULTI_GENUS_MODERATE_DISPERSION"
            ),

        "Seed_Mass_Assignment_Method":
            safe_series(
                family_raw,
                [
                    "Seed_Mass_Assignment_Method",
                ],
                "BACKTRANSFORMED_MEDIAN_LOG10_SPECIES_SEED_MASS"
            ),

        "Seed_Mass_Evidence_Source":
            safe_series(
                family_raw,
                [
                    "Family",
                ],
                "FAMILY_LEVEL_EVIDENCE"
            ),

        "Final_Precedence_Rank":
            3,
    }
)


assert len(
    family_final
) == 2


assert family_final[
    "Seed_Mass_mg"
].gt(
    0
).all()


# =============================================================================
# 13. STANDARDIZE GLOBAL FALLBACK ASSIGNMENTS
# =============================================================================

global_final = pd.DataFrame(
    {
        "Canonical_Species":
            clean_text(
                global_raw[
                    "Canonical_Species"
                ]
            ),

        "Final_Search_Name":
            safe_series(
                global_raw,
                [
                    "Final_Search_Name",
                ],
                pd.NA
            ),

        "Seed_Mass_mg":
            pd.to_numeric(
                global_raw[
                    "Assigned_Seed_Mass_mg"
                ],
                errors="coerce"
            ),

        "Seed_Mass_Assignment_Level":
            "GLOBAL_TRAIT_LEVEL",

        "Seed_Mass_Provenance_Class":
            "GLOBAL_FALLBACK_ESTIMATE",

        "Seed_Mass_Quality_Flag":
            "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK",

        "Seed_Mass_Assignment_Method":
            safe_series(
                global_raw,
                [
                    "Seed_Mass_Assignment_Method",
                ],
                "BACKTRANSFORMED_MEDIAN_LOG10"
            ),

        "Seed_Mass_Evidence_Source":
            "63_SPECIES_58_GENERA_GLOBAL_EVIDENCE_POOL",

        "Final_Precedence_Rank":
            4,
    }
)


assert len(
    global_final
) == 34


# -----------------------------------------------------------------------------
# Floating-point-safe fallback check
# -----------------------------------------------------------------------------

print("\nGlobal fallback value check")
print("-" * 100)

print(
    "Unique loaded values:",
    sorted(
        global_final[
            "Seed_Mass_mg"
        ]
        .dropna()
        .unique()
    )
)

print(
    f"Minimum : "
    f"{global_final['Seed_Mass_mg'].min():.15f}"
)

print(
    f"Maximum : "
    f"{global_final['Seed_Mass_mg'].max():.15f}"
)


assert global_final[
    "Seed_Mass_mg"
].notna().all()


assert np.isclose(
    global_final[
        "Seed_Mass_mg"
    ].to_numpy(
        dtype=float
    ),
    131.19,
    rtol=0,
    atol=1e-9
).all(), (
    "One or more global fallback Seed Mass values "
    "differ materially from 131.19 mg."
)


# =============================================================================
# 14. STANDARDIZE SPECIAL BIOLOGICAL CASE
# =============================================================================

special_final = pd.DataFrame(
    {
        "Canonical_Species":
            clean_text(
                special_raw[
                    "Canonical_Species"
                ]
            ),

        "Final_Search_Name":
            pd.Series(
                [
                    "Musa × paradisiaca"
                ],
                dtype="string"
            ),

        "Seed_Mass_mg":
            np.nan,

        "Seed_Mass_Assignment_Level":
            "BIOLOGICAL_NONAPPLICABILITY",

        "Seed_Mass_Provenance_Class":
            "SPECIAL_BIOLOGICAL_CASE",

        "Seed_Mass_Quality_Flag":
            (
                "FUNCTIONALLY_SEEDLESS_OR_NO_COMPARABLE_"
                "STANDARDIZED_SEED_MASS"
            ),

        "Seed_Mass_Assignment_Method":
            "NO_NUMERIC_VALUE",

        "Seed_Mass_Evidence_Source":
            "SPECIAL_BIOLOGICAL_CASE_REVIEW",

        "Final_Precedence_Rank":
            5,
    }
)


assert len(
    special_final
) == 1


assert special_final[
    "Canonical_Species"
].eq(
    "Musa sapiensis"
).all()


# =============================================================================
# 15. COMPONENT COUNT VALIDATION
# =============================================================================

component_summary = pd.DataFrame(
    [
        {
            "Component":
                "SPECIES_TRY",
            "N_Taxa":
                len(
                    try_final
                ),
        },

        {
            "Component":
                "SPECIES_BIEN",
            "N_Taxa":
                len(
                    bien_final
                ),
        },

        {
            "Component":
                "SPECIES_EXTERNAL_RESCUE",
            "N_Taxa":
                len(
                    external_final
                ),
        },

        {
            "Component":
                "GENUS_ESTIMATE",
            "N_Taxa":
                len(
                    genus_final
                ),
        },

        {
            "Component":
                "FAMILY_ESTIMATE",
            "N_Taxa":
                len(
                    family_final
                ),
        },

        {
            "Component":
                "GLOBAL_FALLBACK_ESTIMATE",
            "N_Taxa":
                len(
                    global_final
                ),
        },

        {
            "Component":
                "SPECIAL_BIOLOGICAL_CASE",
            "N_Taxa":
                len(
                    special_final
                ),
        },
    ]
)


print("\nFinal component structure")
print("-" * 100)

display(
    component_summary
)


assert component_summary[
    "N_Taxa"
].sum() == 127


# =============================================================================
# 16. COMBINE ALL FINAL COMPONENTS
# =============================================================================

all_components = pd.concat(
    [
        species_final,
        genus_final,
        family_final,
        global_final,
        special_final,
    ],
    ignore_index=True
)


all_components[
    "Canonical_Species"
] = clean_text(
    all_components[
        "Canonical_Species"
    ]
)


all_components[
    "Final_Search_Name"
] = clean_text(
    all_components[
        "Final_Search_Name"
    ]
)


# =============================================================================
# 17. CROSS-LEVEL OVERLAP AUDIT
# =============================================================================

component_sets = {
    "SPECIES":
        set(
            species_final[
                "Canonical_Species"
            ]
        ),

    "GENUS":
        set(
            genus_final[
                "Canonical_Species"
            ]
        ),

    "FAMILY":
        set(
            family_final[
                "Canonical_Species"
            ]
        ),

    "GLOBAL":
        set(
            global_final[
                "Canonical_Species"
            ]
        ),

    "SPECIAL":
        set(
            special_final[
                "Canonical_Species"
            ]
        ),
}


overlap_rows = []


component_names = list(
    component_sets.keys()
)


for i in range(
    len(
        component_names
    )
):

    for j in range(
        i + 1,
        len(
            component_names
        )
    ):

        a = component_names[
            i
        ]

        b = component_names[
            j
        ]


        overlap = sorted(
            component_sets[
                a
            ]
            &
            component_sets[
                b
            ]
        )


        overlap_rows.append(
            {
                "Component_A":
                    a,

                "Component_B":
                    b,

                "N_Overlap":
                    len(
                        overlap
                    ),

                "Overlapping_Canonical_Taxa":
                    " | ".join(
                        overlap
                    ),
            }
        )


overlap_audit = pd.DataFrame(
    overlap_rows
)


print("\nFinal cross-level overlap audit")
print("-" * 100)

display(
    overlap_audit
)


overlap_audit.to_csv(
    OUT_OVERLAP_AUDIT,
    index=False
)


assert overlap_audit[
    "N_Overlap"
].eq(
    0
).all(), (
    "Final evidence-level overlap detected. "
    "A canonical taxon appears in more than one final component."
)


# =============================================================================
# 18. BUILD AUTHORITATIVE FINAL TABLE
# =============================================================================

final_table = (
    all_components
    .sort_values(
        [
            "Canonical_Species",
            "Final_Precedence_Rank",
        ]
    )
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ],
        keep="first"
    )
    .sort_values(
        "Canonical_Species"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 19. ADD STANDARD TRAIT FIELDS
# =============================================================================

final_table[
    "Trait_Code"
] = "SEED_MASS"


final_table[
    "Trait_Name"
] = "Seed Mass"


final_table[
    "Standardized_Trait_Value"
] = (
    final_table[
        "Seed_Mass_mg"
    ]
)


final_table[
    "Standardized_Trait_Unit"
] = "mg"


final_table[
    "Numerical_Seed_Mass_Assigned"
] = (
    final_table[
        "Seed_Mass_mg"
    ]
    .notna()
)


final_table[
    "Seed_Mass_Data_Status"
] = np.where(
    final_table[
        "Seed_Mass_mg"
    ]
    .notna(),

    "NUMERIC_VALUE_AVAILABLE",

    "SPECIAL_BIOLOGICAL_CASE_NO_NUMERIC_VALUE"
)


# =============================================================================
# 20. FINAL HARD COUNT VALIDATION
# =============================================================================

assert len(
    final_table
) == 127, (
    f"Expected 127 final taxa; found {len(final_table)}."
)


assert final_table[
    "Canonical_Species"
].nunique() == 127


assert final_table[
    "Canonical_Species"
].is_unique


N_NUMERIC = int(
    final_table[
        "Seed_Mass_mg"
    ]
    .notna()
    .sum()
)


N_NONNUMERIC = int(
    final_table[
        "Seed_Mass_mg"
    ]
    .isna()
    .sum()
)


assert N_NUMERIC == 126


assert N_NONNUMERIC == 1


# =============================================================================
# 21. ONLY NONNUMERIC TAXON VALIDATION
# =============================================================================

nonnumeric = (
    final_table[
        final_table[
            "Seed_Mass_mg"
        ]
        .isna()
    ]
    .copy()
)


assert len(
    nonnumeric
) == 1


assert set(
    nonnumeric[
        "Canonical_Species"
    ]
) == {
    "Musa sapiensis"
}


assert nonnumeric[
    "Seed_Mass_Provenance_Class"
].eq(
    "SPECIAL_BIOLOGICAL_CASE"
).all()


# =============================================================================
# 22. NUMERIC VALUE VALIDATION
# =============================================================================

numeric_rows = (
    final_table[
        final_table[
            "Seed_Mass_mg"
        ]
        .notna()
    ]
    .copy()
)


assert numeric_rows[
    "Seed_Mass_mg"
].gt(
    0
).all()


# =============================================================================
# 23. VALIDATE FINAL ASSIGNMENT LEVEL COUNTS
# =============================================================================

assignment_level_summary = (
    final_table[
        "Seed_Mass_Assignment_Level"
    ]
    .value_counts()
    .rename_axis(
        "Seed_Mass_Assignment_Level"
    )
    .reset_index(
        name="N_Taxa"
    )
)


assignment_lookup = dict(
    zip(
        assignment_level_summary[
            "Seed_Mass_Assignment_Level"
        ],
        assignment_level_summary[
            "N_Taxa"
        ],
    )
)


assert assignment_lookup.get(
    "SPECIES_LEVEL",
    0
) == 88


assert assignment_lookup.get(
    "GENUS_LEVEL",
    0
) == 2


assert assignment_lookup.get(
    "FAMILY_LEVEL",
    0
) == 2


assert assignment_lookup.get(
    "GLOBAL_TRAIT_LEVEL",
    0
) == 34


assert assignment_lookup.get(
    "BIOLOGICAL_NONAPPLICABILITY",
    0
) == 1


assignment_level_summary[
    "Percent_of_127_Taxa"
] = (
    assignment_level_summary[
        "N_Taxa"
    ]
    /
    127
    *
    100
)


print("\nFinal assignment-level summary")
print("-" * 100)

display(
    assignment_level_summary
)


# =============================================================================
# 24. VALIDATE GENUS ESTIMATES
# =============================================================================

genus_rows = (
    final_table[
        final_table[
            "Seed_Mass_Assignment_Level"
        ]
        .eq(
            "GENUS_LEVEL"
        )
    ]
)


assert set(
    genus_rows[
        "Canonical_Species"
    ]
) == {
    "Albizia auxilary",
    "Sterculia tragacantha",
}


# =============================================================================
# 25. VALIDATE FAMILY ESTIMATES
# =============================================================================

family_rows = (
    final_table[
        final_table[
            "Seed_Mass_Assignment_Level"
        ]
        .eq(
            "FAMILY_LEVEL"
        )
    ]
)


assert set(
    family_rows[
        "Canonical_Species"
    ]
) == {
    "Cola species",
    "Grewia pubescens",
}


assert np.allclose(
    family_rows[
        "Seed_Mass_mg"
    ].to_numpy(
        dtype=float
    ),
    420.240217,
    rtol=0,
    atol=1e-6
)


# =============================================================================
# 26. VALIDATE GLOBAL FALLBACK
# =============================================================================

global_rows = (
    final_table[
        final_table[
            "Seed_Mass_Assignment_Level"
        ]
        .eq(
            "GLOBAL_TRAIT_LEVEL"
        )
    ]
)


assert len(
    global_rows
) == 34


assert np.isclose(
    global_rows[
        "Seed_Mass_mg"
    ].to_numpy(
        dtype=float
    ),
    131.19,
    rtol=0,
    atol=1e-9
).all()


assert global_rows[
    "Seed_Mass_Provenance_Class"
].eq(
    "GLOBAL_FALLBACK_ESTIMATE"
).all()


assert global_rows[
    "Seed_Mass_Quality_Flag"
].eq(
    "LOW_CONFIDENCE_GLOBAL_TRAIT_FALLBACK"
).all()


# =============================================================================
# 27. FINAL NUMERIC DISTRIBUTION
# =============================================================================

numeric_distribution = (
    numeric_rows[
        "Seed_Mass_mg"
    ]
    .describe()
)


print("\nFinal numerical Seed Mass distribution")
print("-" * 100)

display(
    numeric_distribution
)


# =============================================================================
# 28. FINAL PROVENANCE SUMMARY
# =============================================================================

provenance_summary = (
    final_table
    .groupby(
        [
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
        ],
        dropna=False,
        as_index=False
    )
    .agg(
        N_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        N_Numeric=(
            "Seed_Mass_mg",
            lambda x:
                int(
                    x.notna().sum()
                )
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


provenance_summary[
    "Percent_of_127_Taxa"
] = (
    provenance_summary[
        "N_Taxa"
    ]
    /
    127
    *
    100
)


print("\nFinal Seed Mass provenance summary")
print("-" * 100)

display(
    provenance_summary
)


# =============================================================================
# 29. FINAL QUALITY SUMMARY
# =============================================================================

quality_summary = (
    final_table[
        "Seed_Mass_Quality_Flag"
    ]
    .fillna(
        "UNSPECIFIED"
    )
    .value_counts()
    .rename_axis(
        "Seed_Mass_Quality_Flag"
    )
    .reset_index(
        name="N_Taxa"
    )
)


quality_summary[
    "Percent_of_127_Taxa"
] = (
    quality_summary[
        "N_Taxa"
    ]
    /
    127
    *
    100
)


print("\nFinal Seed Mass quality summary")
print("-" * 100)

display(
    quality_summary
)


# =============================================================================
# 30. FINAL VALIDATION SUMMARY
# =============================================================================

numeric_coverage_percent = (
    100
    *
    N_NUMERIC
    /
    127
)


nonnumeric_percent = (
    100
    *
    N_NONNUMERIC
    /
    127
)


validation_summary = pd.DataFrame(
    {
        "Metric": [
            "Final canonical taxa",
            "Unique canonical taxa",

            "Species-level taxa total",
            "Species TRY taxa",
            "Species BIEN taxa",
            "Species external-rescue taxa",

            "Genus-estimate taxa",
            "Family-estimate taxa",
            "Global-fallback taxa",
            "Special biological-case taxa",

            "Numeric Seed Mass values",
            "Nonnumeric Seed Mass values",

            "Numeric Seed Mass coverage percent",
            "Nonnumeric percent",

            "Minimum numeric Seed Mass mg",
            "Median numeric Seed Mass mg",
            "Mean numeric Seed Mass mg",
            "Maximum numeric Seed Mass mg",
        ],

        "Value": [
            len(
                final_table
            ),

            final_table[
                "Canonical_Species"
            ].nunique(),

            len(
                species_final
            ),

            len(
                try_final
            ),

            len(
                bien_final
            ),

            len(
                external_final
            ),

            len(
                genus_final
            ),

            len(
                family_final
            ),

            len(
                global_final
            ),

            len(
                special_final
            ),

            N_NUMERIC,

            N_NONNUMERIC,

            numeric_coverage_percent,

            nonnumeric_percent,

            numeric_rows[
                "Seed_Mass_mg"
            ].min(),

            numeric_rows[
                "Seed_Mass_mg"
            ].median(),

            numeric_rows[
                "Seed_Mass_mg"
            ].mean(),

            numeric_rows[
                "Seed_Mass_mg"
            ].max(),
        ],
    }
)


print("\nFinal Seed Mass validation")
print("-" * 100)

display(
    validation_summary
)


# =============================================================================
# 31. FINAL TABLE DISPLAY
# =============================================================================

print("\nAuthoritative final 127-taxon Seed Mass table")
print("-" * 100)

display(
    final_table[
        [
            "Canonical_Species",
            "Final_Search_Name",
            "Trait_Code",
            "Standardized_Trait_Value",
            "Standardized_Trait_Unit",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Seed_Mass_Assignment_Method",
            "Seed_Mass_Evidence_Source",
            "Seed_Mass_Data_Status",
        ]
    ]
)


# =============================================================================
# 32. FINAL HARD ASSERTIONS
# =============================================================================

assert len(
    final_table
) == 127


assert final_table[
    "Canonical_Species"
].is_unique


assert len(
    species_final
) == 88


assert len(
    genus_final
) == 2


assert len(
    family_final
) == 2


assert len(
    global_final
) == 34


assert len(
    special_final
) == 1


assert N_NUMERIC == 126


assert N_NONNUMERIC == 1


assert (
    88
    +
    2
    +
    2
    +
    34
    +
    1
    ==
    127
)


assert np.isclose(
    numeric_coverage_percent,
    99.21259842519684,
    rtol=0,
    atol=1e-9
)


# =============================================================================
# 33. EXPORT
# =============================================================================

final_table.to_csv(
    OUT_FINAL_TABLE,
    index=False
)


provenance_summary.to_csv(
    OUT_PROVENANCE_SUMMARY,
    index=False
)


assignment_level_summary.to_csv(
    OUT_ASSIGNMENT_LEVEL_SUMMARY,
    index=False
)


quality_summary.to_csv(
    OUT_QUALITY_SUMMARY,
    index=False
)


validation_summary.to_csv(
    OUT_VALIDATION_SUMMARY,
    index=False
)


# =============================================================================
# 34. REPORT EXPORTS
# =============================================================================

print("\nSection 9.4X exports")
print("-" * 100)

print(
    f"Consolidated species block : "
    f"{OUT_SPECIES_CONSOLIDATED}"
)

print(
    f"Final Seed Mass table      : "
    f"{OUT_FINAL_TABLE}"
)

print(
    f"Provenance summary         : "
    f"{OUT_PROVENANCE_SUMMARY}"
)

print(
    f"Assignment-level summary   : "
    f"{OUT_ASSIGNMENT_LEVEL_SUMMARY}"
)

print(
    f"Quality summary            : "
    f"{OUT_QUALITY_SUMMARY}"
)

print(
    f"Validation summary         : "
    f"{OUT_VALIDATION_SUMMARY}"
)

print(
    f"Component overlap audit    : "
    f"{OUT_OVERLAP_AUDIT}"
)


# =============================================================================
# 35. COMPLETION
# =============================================================================

print("\n" + "=" * 100)

print(
    "Final 127-taxon Seed Mass consolidation "
    "completed successfully."
)

print(
    f"Final canonical taxa             : "
    f"{len(final_table)}"
)

print(
    f"Unique canonical taxa            : "
    f"{final_table['Canonical_Species'].nunique()}"
)

print(
    f"Species-level evidence taxa      : "
    f"{len(species_final)}"
)

print(
    f"  TRY                             : "
    f"{len(try_final)}"
)

print(
    f"  BIEN                            : "
    f"{len(bien_final)}"
)

print(
    f"  External species rescue         : "
    f"{len(external_final)}"
)

print(
    f"Genus-level estimate taxa        : "
    f"{len(genus_final)}"
)

print(
    f"Family-level estimate taxa       : "
    f"{len(family_final)}"
)

print(
    f"Global fallback estimate taxa    : "
    f"{len(global_final)}"
)

print(
    f"Special biological-case taxa     : "
    f"{len(special_final)}"
)

print(
    f"Numeric Seed Mass values         : "
    f"{N_NUMERIC}"
)

print(
    f"Nonnumeric Seed Mass values      : "
    f"{N_NONNUMERIC}"
)

print(
    f"Numeric Seed Mass coverage       : "
    f"{numeric_coverage_percent:.2f}%"
)

print(
    "Only nonnumeric taxon            : "
    "Musa sapiensis -> Musa × paradisiaca"
)

print(
    "Final precedence validated       : "
    "SPECIES > GENUS > FAMILY > GLOBAL > SPECIAL"
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 9.4Y
# FINAL SEED MASS CLOSURE, PROVENANCE SUMMARY AND VALIDATION
# =============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


# =============================================================================
# 1. PATHS
# =============================================================================

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUT_DIR = (
    PROJECT_ROOT
    / "outputs"
    / "functional_traits"
)

FINAL_SEED_MASS_FILE = (
    OUT_DIR
    / "omo_seed_mass_final.csv"
)

OUT_CLOSURE_PROVENANCE = (
    OUT_DIR
    / "omo_seed_mass_final_closure_provenance_summary.csv"
)

OUT_CLOSURE_SUMMARY = (
    OUT_DIR
    / "omo_seed_mass_final_closure_summary.csv"
)


# =============================================================================
# 2. LOAD AUTHORITATIVE FINAL TABLE
# =============================================================================

assert FINAL_SEED_MASS_FILE.exists(), (
    f"Final Seed Mass table not found:\n{FINAL_SEED_MASS_FILE}"
)

seed_mass = pd.read_csv(
    FINAL_SEED_MASS_FILE,
    low_memory=False
)


print("=" * 100)
print(
    "SECTION 9.4Y — FINAL SEED MASS CLOSURE, "
    "PROVENANCE SUMMARY AND VALIDATION"
)
print("=" * 100)


# =============================================================================
# 3. CORE VALIDATION
# =============================================================================

seed_mass[
    "Seed_Mass_mg"
] = pd.to_numeric(
    seed_mass[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


N_TOTAL = len(
    seed_mass
)

N_UNIQUE = seed_mass[
    "Canonical_Species"
].nunique()

N_NUMERIC = int(
    seed_mass[
        "Seed_Mass_mg"
    ]
    .notna()
    .sum()
)

N_NONNUMERIC = int(
    seed_mass[
        "Seed_Mass_mg"
    ]
    .isna()
    .sum()
)


assert N_TOTAL == 127
assert N_UNIQUE == 127
assert N_NUMERIC == 126
assert N_NONNUMERIC == 1


# =============================================================================
# 4. NONNUMERIC SPECIAL CASE
# =============================================================================

nonnumeric = (
    seed_mass[
        seed_mass[
            "Seed_Mass_mg"
        ].isna()
    ]
    .copy()
)


assert len(
    nonnumeric
) == 1


assert nonnumeric[
    "Canonical_Species"
].iloc[0] == "Musa sapiensis"


assert nonnumeric[
    "Seed_Mass_Provenance_Class"
].iloc[0] == "SPECIAL_BIOLOGICAL_CASE"


# =============================================================================
# 5. FINAL NUMERIC DISTRIBUTION
# =============================================================================

numeric = (
    seed_mass[
        "Seed_Mass_mg"
    ]
    .dropna()
)


numeric_summary = {
    "Minimum_Seed_Mass_mg":
        numeric.min(),

    "Median_Seed_Mass_mg":
        numeric.median(),

    "Mean_Seed_Mass_mg":
        numeric.mean(),

    "Maximum_Seed_Mass_mg":
        numeric.max(),
}


# =============================================================================
# 6. FINAL PROVENANCE SUMMARY
# =============================================================================

provenance_summary = (
    seed_mass
    .groupby(
        [
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
        ],
        dropna=False,
        as_index=False
    )
    .agg(
        N_Taxa=(
            "Canonical_Species",
            "nunique"
        ),

        N_Numeric=(
            "Seed_Mass_mg",
            lambda x:
                int(
                    x.notna().sum()
                )
        ),

        Mean_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "mean"
        ),

        Median_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "median"
        ),

        Min_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "min"
        ),

        Max_Seed_Mass_mg=(
            "Seed_Mass_mg",
            "max"
        ),
    )
)


provenance_summary[
    "Percent_of_127_Taxa"
] = (
    provenance_summary[
        "N_Taxa"
    ]
    /
    127
    *
    100
)


print("\nFinal Seed Mass provenance summary")
print("-" * 100)

display(
    provenance_summary
)


# =============================================================================
# 7. ASSIGNMENT-LEVEL CHECK
# =============================================================================

assignment_counts = (
    seed_mass[
        "Seed_Mass_Assignment_Level"
    ]
    .value_counts()
)


assert assignment_counts.get(
    "SPECIES_LEVEL",
    0
) == 88


assert assignment_counts.get(
    "GENUS_LEVEL",
    0
) == 2


assert assignment_counts.get(
    "FAMILY_LEVEL",
    0
) == 2


assert assignment_counts.get(
    "GLOBAL_TRAIT_LEVEL",
    0
) == 34


assert assignment_counts.get(
    "BIOLOGICAL_NONAPPLICABILITY",
    0
) == 1


# =============================================================================
# 8. FINAL COVERAGE
# =============================================================================

numeric_coverage = (
    100
    *
    N_NUMERIC
    /
    N_TOTAL
)


nonnumeric_percent = (
    100
    *
    N_NONNUMERIC
    /
    N_TOTAL
)


# =============================================================================
# 9. CLOSURE SUMMARY
# =============================================================================

closure_summary = pd.DataFrame(
    {
        "Metric": [
            "Final canonical taxa",
            "Unique canonical taxa",
            "Species-level evidence taxa",
            "Genus-level estimate taxa",
            "Family-level estimate taxa",
            "Global fallback estimate taxa",
            "Special biological-case taxa",
            "Numeric Seed Mass values",
            "Nonnumeric Seed Mass values",
            "Numeric Seed Mass coverage percent",
            "Nonnumeric percent",
            "Minimum Seed Mass mg",
            "Median Seed Mass mg",
            "Mean Seed Mass mg",
            "Maximum Seed Mass mg",
            "Only nonnumeric canonical taxon",
            "Validated search name of nonnumeric taxon",
            "Final Seed Mass status",
        ],

        "Value": [
            N_TOTAL,
            N_UNIQUE,
            88,
            2,
            2,
            34,
            1,
            N_NUMERIC,
            N_NONNUMERIC,
            numeric_coverage,
            nonnumeric_percent,
            numeric_summary[
                "Minimum_Seed_Mass_mg"
            ],
            numeric_summary[
                "Median_Seed_Mass_mg"
            ],
            numeric_summary[
                "Mean_Seed_Mass_mg"
            ],
            numeric_summary[
                "Maximum_Seed_Mass_mg"
            ],
            "Musa sapiensis",
            "Musa × paradisiaca",
            "COMPLETE_WITH_ONE_BIOLOGICALLY_NONAPPLICABLE_NUMERIC_CASE",
        ],
    }
)


print("\nSection 9.4Y closure summary")
print("-" * 100)

display(
    closure_summary
)


# =============================================================================
# 10. EXPORT
# =============================================================================

provenance_summary.to_csv(
    OUT_CLOSURE_PROVENANCE,
    index=False
)


closure_summary.to_csv(
    OUT_CLOSURE_SUMMARY,
    index=False
)


# =============================================================================
# 11. FINAL COMPLETION MESSAGE
# =============================================================================

print("\n" + "=" * 100)

print(
    "FINAL SEED MASS TRAIT CLOSURE COMPLETED SUCCESSFULLY."
)

print(
    f"Final canonical taxa              : "
    f"{N_TOTAL}"
)

print(
    f"Unique canonical taxa             : "
    f"{N_UNIQUE}"
)

print(
    f"Species-level evidence taxa       : "
    f"{assignment_counts.get('SPECIES_LEVEL', 0)}"
)

print(
    f"Genus-level estimate taxa         : "
    f"{assignment_counts.get('GENUS_LEVEL', 0)}"
)

print(
    f"Family-level estimate taxa        : "
    f"{assignment_counts.get('FAMILY_LEVEL', 0)}"
)

print(
    f"Global fallback estimate taxa     : "
    f"{assignment_counts.get('GLOBAL_TRAIT_LEVEL', 0)}"
)

print(
    f"Special biological-case taxa      : "
    f"{assignment_counts.get('BIOLOGICAL_NONAPPLICABILITY', 0)}"
)

print(
    f"Numeric Seed Mass values          : "
    f"{N_NUMERIC}"
)

print(
    f"Numeric Seed Mass coverage        : "
    f"{numeric_coverage:.2f}%"
)

print(
    f"Median numeric Seed Mass          : "
    f"{numeric.median():.6f} mg"
)

print(
    "Only nonnumeric taxon             : "
    "Musa sapiensis -> Musa × paradisiaca"
)

print(
    "Seed Mass acquisition status      : COMPLETE"
)

print("=" * 100)
```

```python
# =============================================================================
# SECTION 10.0 — MAXIMUM HEIGHT TRAIT ACQUISITION AND HARMONIZATION
# Canonical backbone validation and acquisition configuration
# =============================================================================

from pathlib import Path

import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.0 — MAXIMUM HEIGHT TRAIT ACQUISITION AND HARMONIZATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.0A. Project paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = PROJECT_ROOT / "outputs" / "functional_traits"

SEED_MASS_FINAL_PATH = (
    OUTPUT_DIR / "omo_seed_mass_final.csv"
)

WD_FINAL_PATH = (
    OUTPUT_DIR / "omo_wood_density_final.csv"
)

SLA_FINAL_PATH = (
    OUTPUT_DIR / "omo_sla_final.csv"
)


# -----------------------------------------------------------------------------
# 10.0B. Trait configuration
# -----------------------------------------------------------------------------

MAX_HEIGHT_CONFIG = {
    "trait_code": "MAX_HEIGHT",
    "trait_name": "Maximum Height",
    "target_unit": "m",

    # Preferred ecological interpretation:
    # maximum mature plant/tree height, not plot-observed individual height.
    "trait_definition":
        "Maximum reported mature height of the taxon, expressed in metres.",

    # Acquisition philosophy
    "preferred_assignment_level": "SPECIES",
    "allow_genus_estimate": True,
    "allow_family_estimate": True,
    "allow_global_fallback": True,

    # No estimation is allowed silently.
    "require_provenance_for_estimates": True,

    # Trait values must remain biologically plausible and auditable.
    "require_positive_numeric_value": True,

    # Canonical names remain authoritative.
    "preserve_canonical_taxonomy": True,
}


# -----------------------------------------------------------------------------
# 10.0C. Load completed trait outputs
# -----------------------------------------------------------------------------

required_files = {
    "Seed Mass": SEED_MASS_FINAL_PATH,
    "Wood Density": WD_FINAL_PATH,
    "SLA": SLA_FINAL_PATH,
}

missing_files = [
    f"{label}: {path}"
    for label, path in required_files.items()
    if not path.exists()
]

if missing_files:
    raise FileNotFoundError(
        "One or more completed functional-trait outputs were not found:\n"
        + "\n".join(missing_files)
    )


seed_mass_final = pd.read_csv(SEED_MASS_FINAL_PATH)
wd_final = pd.read_csv(WD_FINAL_PATH)
sla_final = pd.read_csv(SLA_FINAL_PATH)


print("\nCompleted trait files loaded")
print("-" * 100)
print(f"Seed Mass rows : {len(seed_mass_final)}")
print(f"WD rows        : {len(wd_final)}")
print(f"SLA rows       : {len(sla_final)}")


# -----------------------------------------------------------------------------
# 10.0D. Identify canonical-species column safely
# -----------------------------------------------------------------------------

def identify_canonical_column(df, label):
    """
    Identify the canonical taxon column without silently guessing
    if multiple plausible columns exist.
    """

    preferred_columns = [
        "Canonical_Species",
        "Canonical_Name",
        "Canonical_Taxon",
        "Species",
    ]

    matches = [
        col for col in preferred_columns
        if col in df.columns
    ]

    if not matches:
        raise KeyError(
            f"Could not identify canonical taxon column in {label}.\n"
            f"Available columns:\n{list(df.columns)}"
        )

    return matches[0]


seed_canonical_col = identify_canonical_column(
    seed_mass_final,
    "Seed Mass final table"
)

wd_canonical_col = identify_canonical_column(
    wd_final,
    "Wood Density final table"
)

sla_canonical_col = identify_canonical_column(
    sla_final,
    "SLA final table"
)


# -----------------------------------------------------------------------------
# 10.0E. Extract canonical backbones
# -----------------------------------------------------------------------------

seed_taxa = (
    seed_mass_final[seed_canonical_col]
    .astype(str)
    .str.strip()
)

wd_taxa = (
    wd_final[wd_canonical_col]
    .astype(str)
    .str.strip()
)

sla_taxa = (
    sla_final[sla_canonical_col]
    .astype(str)
    .str.strip()
)


# -----------------------------------------------------------------------------
# 10.0F. Validate completed-trait backbone integrity
# -----------------------------------------------------------------------------

def validate_backbone(series, label):
    n_rows = len(series)
    n_unique = series.nunique(dropna=False)
    n_missing = series.replace(
        {"": np.nan, "nan": np.nan, "None": np.nan}
    ).isna().sum()

    print(f"\n{label}")
    print("-" * 100)
    print(f"Rows              : {n_rows}")
    print(f"Unique taxa       : {n_unique}")
    print(f"Missing taxon IDs : {n_missing}")

    if n_rows != 127:
        raise AssertionError(
            f"{label} contains {n_rows} rows; expected 127."
        )

    if n_unique != 127:
        raise AssertionError(
            f"{label} contains {n_unique} unique taxa; expected 127."
        )

    if n_missing != 0:
        raise AssertionError(
            f"{label} contains missing canonical taxon identifiers."
        )


validate_backbone(seed_taxa, "Seed Mass canonical backbone")
validate_backbone(wd_taxa, "Wood Density canonical backbone")
validate_backbone(sla_taxa, "SLA canonical backbone")


# -----------------------------------------------------------------------------
# 10.0G. Cross-trait taxonomic agreement
# -----------------------------------------------------------------------------

seed_set = set(seed_taxa)
wd_set = set(wd_taxa)
sla_set = set(sla_taxa)

print("\nCross-trait canonical backbone agreement")
print("-" * 100)

print(
    "Seed Mass = WD  :",
    seed_set == wd_set
)

print(
    "Seed Mass = SLA :",
    seed_set == sla_set
)

print(
    "WD = SLA        :",
    wd_set == sla_set
)


if not (
    seed_set == wd_set == sla_set
):
    print("\nTaxonomic discrepancies detected")
    print("-" * 100)

    print(
        "Seed only:",
        sorted(seed_set - wd_set - sla_set)
    )

    print(
        "WD only:",
        sorted(wd_set - seed_set - sla_set)
    )

    print(
        "SLA only:",
        sorted(sla_set - seed_set - wd_set)
    )

    raise AssertionError(
        "Completed functional-trait tables do not share "
        "the same 127-taxon canonical backbone."
    )


# -----------------------------------------------------------------------------
# 10.0H. Create Maximum Height target manifest
# -----------------------------------------------------------------------------

max_height_manifest = pd.DataFrame({
    "Canonical_Species": seed_taxa
})


# Infer target resolution conservatively.
# These are the two previously documented non-standard canonical targets.
known_genus_or_unresolved = {
    "Drypetes species",
    "Cola species",
}

max_height_manifest[
    "Max_Height_Target_Level"
] = np.where(
    max_height_manifest[
        "Canonical_Species"
    ].isin(known_genus_or_unresolved),
    "GENUS_LEVEL_OR_UNRESOLVED",
    "SPECIES_LEVEL"
)


# Empty acquisition fields — intentionally not populated yet.
max_height_manifest["Max_Height_m"] = np.nan
max_height_manifest["Max_Height_Assignment_Level"] = pd.NA
max_height_manifest["Max_Height_Provenance_Class"] = pd.NA
max_height_manifest["Max_Height_Source"] = pd.NA
max_height_manifest["Max_Height_Source_Taxon"] = pd.NA
max_height_manifest["Max_Height_Match_Status"] = pd.NA
max_height_manifest["Max_Height_Quality_Flag"] = "NOT_ACQUIRED"
max_height_manifest["Max_Height_Notes"] = pd.NA


# -----------------------------------------------------------------------------
# 10.0I. Manifest validation
# -----------------------------------------------------------------------------

assert len(max_height_manifest) == 127
assert (
    max_height_manifest["Canonical_Species"].nunique()
    == 127
)
assert (
    max_height_manifest["Max_Height_m"].notna().sum()
    == 0
)


print("\nMaximum Height target manifest")
print("-" * 100)

print(
    max_height_manifest[
        "Max_Height_Target_Level"
    ].value_counts(dropna=False)
)


print("\nInitial acquisition status")
print("-" * 100)

print(
    max_height_manifest[
        "Max_Height_Quality_Flag"
    ].value_counts(dropna=False)
)


print("\nMaximum Height trait configuration")
print("-" * 100)

for key, value in MAX_HEIGHT_CONFIG.items():
    print(f"{key:<35}: {value}")


# -----------------------------------------------------------------------------
# 10.0J. Explicit methodological rules
# -----------------------------------------------------------------------------

print("\nMaximum Height acquisition hierarchy")
print("-" * 100)

print(
    """
1. DIRECT SPECIES-LEVEL EVIDENCE
   Preferred whenever an accepted Maximum Height value can be linked
   directly to the canonical taxon or to a taxonomically verified synonym.

2. TAXONOMICALLY RECONCILED SPECIES EVIDENCE
   Accepted only when synonymy, orthographic variation, or current-name
   reconciliation is documented.

3. GENUS-LEVEL ESTIMATE
   Used only where credible species-level evidence cannot be acquired.

4. FAMILY-LEVEL ESTIMATE
   Used only where both species- and genus-level evidence are unavailable.

5. GLOBAL FALLBACK
   Reserved for otherwise unresolved taxa and explicitly labelled as such.

6. NO SILENT IMPUTATION
   Every non-species assignment must retain its assignment level,
   provenance class, source and quality flag.

7. CANONICAL TAXONOMY
   The existing 127-taxon Omo canonical backbone remains authoritative.
   External source names must never overwrite the canonical taxon name.

8. TRAIT INTERPRETATION
   Maximum Height refers to reported mature taxon height, not the maximum
   observed height of individuals measured in the Omo structural inventory.
"""
)


print("=" * 100)
print("SECTION 10.0 SETUP COMPLETE")
print("=" * 100)
print(
    f"Maximum Height targets prepared : "
    f"{len(max_height_manifest)}"
)
print(
    "Canonical backbone validated    : YES"
)
print(
    "Maximum Height values acquired  : 0"
)
print(
    "Next step                       : "
    "Section 10.1 — Existing-source audit and primary acquisition strategy"
)
```

```python
# =============================================================================
# SECTION 10.1 — EXISTING-SOURCE AUDIT FOR MAXIMUM HEIGHT
# Search project data assets for potentially relevant height information
# =============================================================================

from pathlib import Path
import re

import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.1 — EXISTING-SOURCE AUDIT FOR MAXIMUM HEIGHT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.1A. Search configuration
# -----------------------------------------------------------------------------

# Search the project tree, but avoid environments, Git internals and caches.
EXCLUDED_DIR_NAMES = {
    ".git",
    ".ipynb_checkpoints",
    "__pycache__",
    "venv",
    ".venv",
    "env",
    ".env",
    "site-packages",
}

SUPPORTED_EXTENSIONS = {
    ".csv",
    ".xlsx",
    ".xls",
    ".parquet",
}


# Height-related column-name patterns.
HEIGHT_PATTERNS = [
    r"\bheight\b",
    r"\bmax[_\s-]*height\b",
    r"\bmaximum[_\s-]*height\b",
    r"\bmature[_\s-]*height\b",
    r"\bplant[_\s-]*height\b",
    r"\btree[_\s-]*height\b",
    r"\bheight[_\s-]*m\b",
    r"\bheight[_\s-]*meter",
    r"\bheight[_\s-]*metre",
    r"\bhmax\b",
]


# Possible taxon-identification columns.
TAXON_PATTERNS = [
    r"species",
    r"canonical",
    r"taxon",
    r"scientific",
    r"accepted",
    r"binomial",
    r"plant[_\s-]*name",
]


# -----------------------------------------------------------------------------
# 10.1B. Helper functions
# -----------------------------------------------------------------------------

def is_excluded_path(path):
    """
    Return True if a file is located inside a directory
    that should not be audited.
    """
    return any(
        part in EXCLUDED_DIR_NAMES
        for part in path.parts
    )


def match_patterns(column_name, patterns):
    """
    Return True when a column name matches at least one
    supplied regular-expression pattern.
    """
    normalized = (
        str(column_name)
        .strip()
        .lower()
        .replace("-", "_")
    )

    return any(
        re.search(pattern, normalized)
        for pattern in patterns
    )


def read_table_preview(path, nrows=500):
    """
    Read enough of a structured file to inspect its schema.

    The audit does not need the complete dataset at this stage.
    """
    suffix = path.suffix.lower()

    if suffix == ".csv":
        return pd.read_csv(
            path,
            nrows=nrows,
            low_memory=False
        )

    elif suffix in {".xlsx", ".xls"}:
        return pd.read_excel(
            path,
            nrows=nrows
        )

    elif suffix == ".parquet":
        df = pd.read_parquet(path)
        return df.head(nrows)

    else:
        raise ValueError(
            f"Unsupported file type: {suffix}"
        )


# -----------------------------------------------------------------------------
# 10.1C. Identify structured project files
# -----------------------------------------------------------------------------

structured_files = []

for path in PROJECT_ROOT.rglob("*"):

    if (
        path.is_file()
        and path.suffix.lower() in SUPPORTED_EXTENSIONS
        and not is_excluded_path(path)
    ):
        structured_files.append(path)


structured_files = sorted(structured_files)


print("\nStructured project-file inventory")
print("-" * 100)
print(
    f"Structured files found : {len(structured_files)}"
)


extension_counts = pd.Series(
    [p.suffix.lower() for p in structured_files]
).value_counts()

print("\nFiles by extension")
print("-" * 100)
print(extension_counts.to_string())


# -----------------------------------------------------------------------------
# 10.1D. Schema audit for height-bearing columns
# -----------------------------------------------------------------------------

audit_records = []
read_failures = []

for path in structured_files:

    try:
        df_preview = read_table_preview(path)

    except Exception as exc:
        read_failures.append({
            "File": str(path.relative_to(PROJECT_ROOT)),
            "Error": str(exc),
        })
        continue

    columns = list(df_preview.columns)

    height_cols = [
        col for col in columns
        if match_patterns(col, HEIGHT_PATTERNS)
    ]

    taxon_cols = [
        col for col in columns
        if match_patterns(col, TAXON_PATTERNS)
    ]

    if height_cols:

        audit_records.append({
            "File": str(path.relative_to(PROJECT_ROOT)),
            "N_Columns": len(columns),
            "Height_Columns": height_cols,
            "Taxon_Columns": taxon_cols,
            "Preview_Rows_Read": len(df_preview),
        })


height_source_audit = pd.DataFrame(audit_records)


print("\nCandidate height-bearing structured files")
print("-" * 100)

if height_source_audit.empty:

    print(
        "No structured files with recognizable "
        "height-related columns were detected."
    )

else:

    print(
        f"Candidate files detected : "
        f"{len(height_source_audit)}"
    )

    display(
        height_source_audit
    )


# -----------------------------------------------------------------------------
# 10.1E. Detailed audit of candidate height columns
# -----------------------------------------------------------------------------

detail_records = []

for record in audit_records:

    path = PROJECT_ROOT / record["File"]

    try:
        df_preview = read_table_preview(
            path,
            nrows=5000
        )

    except Exception:
        continue

    for height_col in record["Height_Columns"]:

        raw = df_preview[height_col]

        numeric = pd.to_numeric(
            raw,
            errors="coerce"
        )

        detail_records.append({
            "File": record["File"],
            "Height_Column": height_col,
            "Taxon_Columns":
                ", ".join(record["Taxon_Columns"])
                if record["Taxon_Columns"]
                else "",
            "Rows_Examined": len(raw),
            "Numeric_Count": int(
                numeric.notna().sum()
            ),
            "Missing_Count": int(
                numeric.isna().sum()
            ),
            "Min_Numeric":
                float(numeric.min())
                if numeric.notna().any()
                else np.nan,
            "Median_Numeric":
                float(numeric.median())
                if numeric.notna().any()
                else np.nan,
            "Max_Numeric":
                float(numeric.max())
                if numeric.notna().any()
                else np.nan,
        })


height_column_detail = pd.DataFrame(
    detail_records
)


print("\nCandidate height-column diagnostics")
print("-" * 100)

if height_column_detail.empty:

    print(
        "No numeric height columns were available "
        "for detailed inspection."
    )

else:

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 100
    ):
        display(
            height_column_detail
        )


# -----------------------------------------------------------------------------
# 10.1F. Classify candidate sources conservatively
# -----------------------------------------------------------------------------

def provisional_height_source_class(file_name, height_column):
    """
    Preliminary classification only.

    This does NOT authorize use of the values.
    """

    text = (
        f"{file_name} {height_column}"
        .lower()
    )

    structural_terms = [
        "structural",
        "inventory",
        "dbh",
        "plot",
        "individual",
        "tree_measure",
    ]

    trait_terms = [
        "trait",
        "try",
        "bien",
        "max_height",
        "maximum_height",
        "maximum height",
        "mature_height",
        "mature height",
    ]

    if any(
        term in text
        for term in trait_terms
    ):
        return "POTENTIAL_TRAIT_SOURCE"

    if any(
        term in text
        for term in structural_terms
    ):
        return "LIKELY_OBSERVED_INVENTORY_HEIGHT"

    return "REQUIRES_MANUAL_CLASSIFICATION"


if not height_column_detail.empty:

    height_column_detail[
        "Provisional_Source_Class"
    ] = height_column_detail.apply(
        lambda row:
        provisional_height_source_class(
            row["File"],
            row["Height_Column"]
        ),
        axis=1
    )

    print("\nProvisional source classification")
    print("-" * 100)

    print(
        height_column_detail[
            "Provisional_Source_Class"
        ].value_counts(dropna=False)
    )

    print("\nDetailed classified candidates")
    print("-" * 100)

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 100
    ):
        display(
            height_column_detail[
                [
                    "File",
                    "Height_Column",
                    "Taxon_Columns",
                    "Numeric_Count",
                    "Min_Numeric",
                    "Median_Numeric",
                    "Max_Numeric",
                    "Provisional_Source_Class",
                ]
            ]
        )


# -----------------------------------------------------------------------------
# 10.1G. Read-failure report
# -----------------------------------------------------------------------------

print("\nStructured files that could not be inspected")
print("-" * 100)

if read_failures:

    print(
        f"Read failures : {len(read_failures)}"
    )

    read_failures_df = pd.DataFrame(
        read_failures
    )

    display(
        read_failures_df.head(50)
    )

else:

    print("Read failures : 0")


# -----------------------------------------------------------------------------
# 10.1H. Explicit safeguard
# -----------------------------------------------------------------------------

print("\nInterpretation safeguard")
print("-" * 100)

print(
    """
No height value discovered by this audit has been assigned to the
Maximum Height trait.

In particular:

• Plot-level or individual-tree height measurements represent observed
  Omo inventory structure.

• Such measurements are NOT automatically equivalent to a species'
  biological maximum mature height.

• A candidate dataset will be accepted as a Maximum Height source only
  after its ecological meaning, taxonomic linkage, units and provenance
  have been verified.

• Existing project data may nevertheless provide useful taxonomic,
  validation or supplementary evidence.
"""
)


# -----------------------------------------------------------------------------
# 10.1I. Audit summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.1 EXISTING-SOURCE AUDIT COMPLETE")
print("=" * 100)

print(
    f"Structured files scanned       : "
    f"{len(structured_files)}"
)

print(
    f"Height-bearing candidate files : "
    f"{len(height_source_audit)}"
)

print(
    f"Height columns identified      : "
    f"{len(height_column_detail)}"
)

print(
    f"Read failures                  : "
    f"{len(read_failures)}"
)

print(
    "Maximum Height values assigned : 0"
)

print(
    "Next decision                  : "
    "Evaluate candidate sources and establish the "
    "primary Maximum Height acquisition source."
)
```

```python
# =============================================================================
# SECTION 10.2 — ROW-LEVEL TRAIT-FIELD AUDIT FOR MAXIMUM HEIGHT
# Search existing structured files where trait identity may be stored in rows
# =============================================================================

import re

import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.2 — ROW-LEVEL TRAIT-FIELD AUDIT FOR MAXIMUM HEIGHT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.2A. Trait-field identification rules
# -----------------------------------------------------------------------------

TRAIT_FIELD_PATTERNS = [
    r"^trait$",
    r"trait[_\s-]*name",
    r"trait[_\s-]*id",
    r"trait[_\s-]*code",
    r"trait[_\s-]*description",
    r"trait[_\s-]*label",
    r"character",
    r"measurement[_\s-]*type",
    r"variable[_\s-]*name",
]


MAX_HEIGHT_VALUE_PATTERNS = [
    r"\bmaximum\s*height\b",
    r"\bmax(?:imum)?[_\s-]*height\b",
    r"\bmature\s*height\b",
    r"\bplant\s*height\b",
    r"\btree\s*height\b",
    r"\bheight\s*maximum\b",
    r"\bhmax\b",
]


# -----------------------------------------------------------------------------
# 10.2B. Helper functions
# -----------------------------------------------------------------------------

def normalize_search_text(value):
    """
    Convert arbitrary values to normalized searchable text.
    """
    if pd.isna(value):
        return ""

    return (
        str(value)
        .strip()
        .lower()
        .replace("_", " ")
        .replace("-", " ")
    )


def is_trait_field(column_name):
    """
    Determine whether a column plausibly stores trait identity.
    """
    text = normalize_search_text(column_name)

    return any(
        re.search(pattern, text)
        for pattern in TRAIT_FIELD_PATTERNS
    )


def contains_height_trait(series):
    """
    Search unique textual values in a candidate trait field
    for Maximum Height-like trait labels.
    """

    values = (
        series
        .dropna()
        .astype(str)
        .str.strip()
        .drop_duplicates()
    )

    matches = []

    for value in values:

        normalized = normalize_search_text(value)

        if any(
            re.search(pattern, normalized)
            for pattern in MAX_HEIGHT_VALUE_PATTERNS
        ):
            matches.append(value)

    return matches


# -----------------------------------------------------------------------------
# 10.2C. Search structured files for trait-identity fields
# -----------------------------------------------------------------------------

trait_field_records = []
row_level_height_hits = []
trait_read_failures = []


for path in structured_files:

    try:
        df = read_table_preview(
            path,
            nrows=10000
        )

    except Exception as exc:

        # Empty files already identified in Section 10.1 are harmless.
        trait_read_failures.append({
            "File": str(path.relative_to(PROJECT_ROOT)),
            "Error": str(exc),
        })

        continue


    candidate_trait_columns = [
        col for col in df.columns
        if is_trait_field(col)
    ]


    if not candidate_trait_columns:
        continue


    trait_field_records.append({
        "File": str(path.relative_to(PROJECT_ROOT)),
        "Trait_Fields": candidate_trait_columns,
        "Rows_Examined": len(df),
    })


    for trait_col in candidate_trait_columns:

        matching_values = contains_height_trait(
            df[trait_col]
        )

        if matching_values:

            row_level_height_hits.append({
                "File":
                    str(path.relative_to(PROJECT_ROOT)),

                "Trait_Field":
                    trait_col,

                "Matching_Trait_Values":
                    matching_values,

                "N_Matching_Labels":
                    len(matching_values),

                "Rows_Examined":
                    len(df),
            })


trait_field_audit = pd.DataFrame(
    trait_field_records
)

row_level_height_audit = pd.DataFrame(
    row_level_height_hits
)


# -----------------------------------------------------------------------------
# 10.2D. Trait-bearing file summary
# -----------------------------------------------------------------------------

print("\nFiles containing candidate trait-identity fields")
print("-" * 100)

print(
    f"Trait-bearing files detected : "
    f"{len(trait_field_audit)}"
)


if not trait_field_audit.empty:

    with pd.option_context(
        "display.max_rows", 100,
        "display.max_columns", None,
        "display.max_colwidth", 120
    ):

        display(
            trait_field_audit.head(100)
        )


# -----------------------------------------------------------------------------
# 10.2E. Maximum Height row-level results
# -----------------------------------------------------------------------------

print("\nRow-level Maximum Height trait matches")
print("-" * 100)

if row_level_height_audit.empty:

    print(
        "No Maximum Height-like trait labels were detected "
        "within existing trait-identity fields."
    )

else:

    print(
        f"Files with possible Maximum Height evidence : "
        f"{len(row_level_height_audit)}"
    )

    with pd.option_context(
        "display.max_rows", 100,
        "display.max_columns", None,
        "display.max_colwidth", 150
    ):

        display(
            row_level_height_audit
        )


# -----------------------------------------------------------------------------
# 10.2F. Inspect matching rows if any were found
# -----------------------------------------------------------------------------

matching_row_records = []


if not row_level_height_audit.empty:

    for _, hit in row_level_height_audit.iterrows():

        path = PROJECT_ROOT / hit["File"]

        try:
            df = read_table_preview(
                path,
                nrows=50000
            )

        except Exception:
            continue

        trait_col = hit["Trait_Field"]

        normalized_trait_values = (
            df[trait_col]
            .map(normalize_search_text)
        )

        mask = pd.Series(
            False,
            index=df.index
        )

        for pattern in MAX_HEIGHT_VALUE_PATTERNS:

            mask = mask | normalized_trait_values.str.contains(
                pattern,
                regex=True,
                na=False
            )

        matched_rows = df.loc[mask].copy()

        if matched_rows.empty:
            continue

        for idx, row in matched_rows.iterrows():

            matching_row_records.append({
                "File": hit["File"],
                "Row_Index": idx,
                "Trait_Field": trait_col,
                "Trait_Label": row.get(
                    trait_col,
                    pd.NA
                ),
                "Available_Columns":
                    ", ".join(
                        map(str, df.columns)
                    )
            })


matching_rows_summary = pd.DataFrame(
    matching_row_records
)


if not matching_rows_summary.empty:

    print("\nRows associated with possible Maximum Height traits")
    print("-" * 100)

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 150
    ):

        display(
            matching_rows_summary
        )


# -----------------------------------------------------------------------------
# 10.2G. Persist the empty Maximum Height target manifest
# -----------------------------------------------------------------------------

MAX_HEIGHT_MANIFEST_PATH = (
    OUTPUT_DIR /
    "omo_max_height_target_manifest.csv"
)

max_height_manifest.to_csv(
    MAX_HEIGHT_MANIFEST_PATH,
    index=False
)


print("\nMaximum Height target manifest export")
print("-" * 100)

print(
    f"Manifest path : {MAX_HEIGHT_MANIFEST_PATH}"
)

print(
    f"Rows          : {len(max_height_manifest)}"
)

print(
    f"Unique taxa   : "
    f"{max_height_manifest['Canonical_Species'].nunique()}"
)


# -----------------------------------------------------------------------------
# 10.2H. Acquisition readiness assessment
# -----------------------------------------------------------------------------

existing_height_evidence_found = (
    not row_level_height_audit.empty
)


print("\nAcquisition readiness assessment")
print("-" * 100)

print(
    f"Existing row-level Maximum Height evidence found : "
    f"{existing_height_evidence_found}"
)


if existing_height_evidence_found:

    print(
        "\nACTION REQUIRED:\n"
        "Candidate existing Maximum Height records must be "
        "examined before external acquisition begins."
    )

else:

    print(
        "\nNo existing dedicated Maximum Height evidence has "
        "been detected in the structured project assets."
    )

    print(
        "The project is therefore ready to establish an "
        "external primary Maximum Height acquisition source."
    )


# -----------------------------------------------------------------------------
# 10.2I. Final safeguards
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
This audit searches TRAIT VALUES stored inside trait-identity columns,
complementing the column-name audit performed in Section 10.1.

A negative result therefore means:

1. No dedicated Maximum Height column was detected in Section 10.1; and
2. No recognizable Maximum Height trait label was detected inside the
   existing structured trait tables in Section 10.2.

This does not prove that no height information exists anywhere in the
project. It establishes that no existing structured dataset has yet been
identified as an authoritative Maximum Height trait source.

No Maximum Height value has been assigned to any Omo taxon.
"""
)


print("=" * 100)
print("SECTION 10.2 ROW-LEVEL TRAIT AUDIT COMPLETE")
print("=" * 100)

print(
    f"Structured files available        : "
    f"{len(structured_files)}"
)

print(
    f"Trait-bearing files detected       : "
    f"{len(trait_field_audit)}"
)

print(
    f"Maximum Height candidate files     : "
    f"{len(row_level_height_audit)}"
)

print(
    f"Maximum Height values assigned     : 0"
)

print(
    f"Target manifest exported           : YES"
)

print(
    "Next step                         : "
    "Section 10.3 — Primary Maximum Height database acquisition"
)
```

```python
# =============================================================================
# SECTION 10.8 — MAXIMUM HEIGHT SUPPLEMENTARY TAXONOMIC RECONCILIATION
# Recover previously validated source-name mappings from Notebook 07B
# =============================================================================

from pathlib import Path
import re

import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.8 — MAXIMUM HEIGHT SUPPLEMENTARY TAXONOMIC RECONCILIATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.8A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BIEN_QUERY_LOG_PATH = (
    PROJECT_ROOT /
    "data" /
    "external" /
    "BIEN" /
    "omo_max_height_bien_species_query_log.csv"
)

BIEN_VALUES_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_provisional_species_values.csv"
)

MAX_HEIGHT_MANIFEST_PATH = (
    OUTPUT_DIR /
    "omo_max_height_target_manifest.csv"
)


# -----------------------------------------------------------------------------
# 10.8B. Load current Maximum Height state
# -----------------------------------------------------------------------------

height_manifest = pd.read_csv(
    MAX_HEIGHT_MANIFEST_PATH
)

bien_query_log = pd.read_csv(
    BIEN_QUERY_LOG_PATH
)

bien_values = pd.read_csv(
    BIEN_VALUES_PATH
)


assert len(height_manifest) == 127
assert height_manifest["Canonical_Species"].nunique() == 127
assert len(bien_values) == 3


bien_resolved_taxa = set(
    bien_values["Omo_Canonical_Species"]
    .astype(str)
    .str.strip()
)


residual_manifest = height_manifest[
    ~height_manifest["Canonical_Species"].isin(
        bien_resolved_taxa
    )
].copy()


print("\nCurrent Maximum Height state")
print("-" * 100)

print(f"Canonical taxa                : {len(height_manifest)}")
print(f"BIEN QC-passed taxa           : {len(bien_resolved_taxa)}")
print(f"Residual taxa                 : {len(residual_manifest)}")

print(
    "\nResidual target levels\n"
    + "-" * 100
)

print(
    residual_manifest[
        "Max_Height_Target_Level"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.8C. Search existing functional-trait CSV files for taxonomic mappings
# -----------------------------------------------------------------------------

csv_files = sorted(
    OUTPUT_DIR.glob("*.csv")
)


# Exclude Maximum Height files themselves so that this is genuinely
# recovering taxonomic evidence from earlier completed trait work.
mapping_source_files = [
    path for path in csv_files
    if "max_height" not in path.name.lower()
]


print("\nExisting functional-trait files available for reconciliation audit")
print("-" * 100)

print(f"CSV files scanned : {len(mapping_source_files)}")


# Candidate canonical-name fields
canonical_patterns = [
    "canonical_species",
    "canonical_name",
    "canonical_taxon",
]

# Candidate external/resolved/source taxon fields
resolved_patterns = [
    "source_species",
    "source_taxon",
    "trait_source_name",
    "matched_name",
    "name_matched",
    "accepted_name",
    "accepted_species",
    "resolved_name",
    "resolved_species",
    "verified_name",
    "verified_species",
    "bien_species",
    "try_species",
    "gwdd_species",
    "taxon_match",
]


def norm_col(col):
    """Normalize column name for pattern matching."""
    return re.sub(
        r"[^a-z0-9]+",
        "_",
        str(col).strip().lower()
    ).strip("_")


def identify_columns(columns, patterns):
    """Return columns matching any normalized pattern."""
    results = []

    for col in columns:
        normalized = norm_col(col)

        if any(
            pattern in normalized
            for pattern in patterns
        ):
            results.append(col)

    return results


mapping_records = []
read_failures = []


for path in mapping_source_files:

    try:
        df = pd.read_csv(
            path,
            low_memory=False
        )

    except Exception as exc:

        read_failures.append(
            {
                "File": path.name,
                "Error": str(exc),
            }
        )
        continue


    if df.empty:
        continue


    canonical_cols = identify_columns(
        df.columns,
        canonical_patterns
    )

    resolved_cols = identify_columns(
        df.columns,
        resolved_patterns
    )


    if not canonical_cols or not resolved_cols:
        continue


    for canonical_col in canonical_cols:

        canonical_values = (
            df[canonical_col]
            .astype("string")
            .str.strip()
        )


        for resolved_col in resolved_cols:

            if resolved_col == canonical_col:
                continue


            resolved_values = (
                df[resolved_col]
                .astype("string")
                .str.strip()
            )


            temp = pd.DataFrame(
                {
                    "Canonical_Species":
                        canonical_values,

                    "Recovered_Search_Name":
                        resolved_values,
                }
            )


            temp = temp[
                temp["Canonical_Species"].notna()
                &
                temp["Recovered_Search_Name"].notna()
            ].copy()


            temp = temp[
                temp["Canonical_Species"].isin(
                    residual_manifest[
                        "Canonical_Species"
                    ]
                )
            ].copy()


            temp = temp[
                temp["Recovered_Search_Name"].str.len() > 0
            ].copy()


            if temp.empty:
                continue


            temp["Evidence_File"] = path.name
            temp["Canonical_Field"] = canonical_col
            temp["Recovered_Name_Field"] = resolved_col


            mapping_records.append(temp)


# -----------------------------------------------------------------------------
# 10.8D. Assemble recovered mapping evidence
# -----------------------------------------------------------------------------

if mapping_records:

    recovered_mapping_raw = pd.concat(
        mapping_records,
        ignore_index=True
    )

else:

    recovered_mapping_raw = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Recovered_Search_Name",
            "Evidence_File",
            "Canonical_Field",
            "Recovered_Name_Field",
        ]
    )


# Remove obvious empty/string-null values
if not recovered_mapping_raw.empty:

    recovered_mapping_raw[
        "Recovered_Search_Name"
    ] = (
        recovered_mapping_raw[
            "Recovered_Search_Name"
        ]
        .astype(str)
        .str.strip()
    )


    recovered_mapping_raw = recovered_mapping_raw[
        ~recovered_mapping_raw[
            "Recovered_Search_Name"
        ].str.lower().isin(
            {
                "",
                "nan",
                "none",
                "<na>",
            }
        )
    ].copy()


    recovered_mapping_raw = (
        recovered_mapping_raw
        .drop_duplicates()
        .reset_index(drop=True)
    )


print("\nRecovered cross-trait mapping evidence")
print("-" * 100)

print(
    f"Mapping evidence rows : "
    f"{len(recovered_mapping_raw)}"
)

print(
    f"Residual taxa with at least one recovered name : "
    f"{recovered_mapping_raw['Canonical_Species'].nunique()}"
)


# -----------------------------------------------------------------------------
# 10.8E. Distinguish identical names from alternative search names
# -----------------------------------------------------------------------------

if not recovered_mapping_raw.empty:

    recovered_mapping_raw[
        "Name_Relationship"
    ] = np.where(
        recovered_mapping_raw[
            "Canonical_Species"
        ].str.casefold()
        ==
        recovered_mapping_raw[
            "Recovered_Search_Name"
        ].str.casefold(),

        "IDENTICAL_TO_CANONICAL",

        "ALTERNATIVE_SEARCH_NAME"
    )


    print("\nRecovered-name relationship")
    print("-" * 100)

    print(
        recovered_mapping_raw[
            "Name_Relationship"
        ].value_counts(dropna=False)
    )


    alternative_names = recovered_mapping_raw[
        recovered_mapping_raw[
            "Name_Relationship"
        ] == "ALTERNATIVE_SEARCH_NAME"
    ].copy()


else:

    alternative_names = recovered_mapping_raw.copy()


# -----------------------------------------------------------------------------
# 10.8F. Summarize alternative search names by canonical taxon
# -----------------------------------------------------------------------------

if not alternative_names.empty:

    alternative_summary = (
        alternative_names
        .groupby(
            "Canonical_Species",
            as_index=False
        )
        .agg(
            Recovered_Alternative_Names=(
                "Recovered_Search_Name",
                lambda x:
                    " | ".join(
                        sorted(
                            set(
                                str(v).strip()
                                for v in x
                                if str(v).strip()
                            )
                        )
                    )
            ),
            N_Alternative_Names=(
                "Recovered_Search_Name",
                "nunique"
            ),
            Evidence_Files=(
                "Evidence_File",
                lambda x:
                    " | ".join(
                        sorted(set(x))
                    )
            ),
        )
    )

else:

    alternative_summary = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Recovered_Alternative_Names",
            "N_Alternative_Names",
            "Evidence_Files",
        ]
    )


print("\nResidual taxa with alternative search names")
print("-" * 100)

print(
    f"Taxa with alternatives : "
    f"{len(alternative_summary)}"
)


if not alternative_summary.empty:

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 120,
    ):
        display(
            alternative_summary
        )


# -----------------------------------------------------------------------------
# 10.8G. Build supplementary acquisition search register
# -----------------------------------------------------------------------------

supplementary_search_register = (
    residual_manifest[
        [
            "Canonical_Species",
            "Max_Height_Target_Level",
        ]
    ]
    .copy()
)


supplementary_search_register = (
    supplementary_search_register
    .merge(
        alternative_summary,
        on="Canonical_Species",
        how="left"
    )
)


supplementary_search_register[
    "Primary_Search_Name"
] = supplementary_search_register[
    "Canonical_Species"
]


supplementary_search_register[
    "Alternative_Name_Available"
] = supplementary_search_register[
    "Recovered_Alternative_Names"
].notna()


supplementary_search_register[
    "Supplementary_Search_Status"
] = np.where(
    supplementary_search_register[
        "Alternative_Name_Available"
    ],
    "READY_WITH_RECONCILIATION",
    "READY_CANONICAL_ONLY"
)


# Explicitly preserve the two non-species targets
supplementary_search_register.loc[
    supplementary_search_register[
        "Max_Height_Target_Level"
    ] == "GENUS_LEVEL_OR_UNRESOLVED",
    "Supplementary_Search_Status"
] = "GENUS_LEVEL_SEARCH_REQUIRED"


# -----------------------------------------------------------------------------
# 10.8H. Validation
# -----------------------------------------------------------------------------

assert len(supplementary_search_register) == 124

assert (
    supplementary_search_register[
        "Canonical_Species"
    ].nunique()
    == 124
)


assert not set(
    supplementary_search_register[
        "Canonical_Species"
    ]
).intersection(
    bien_resolved_taxa
)


print("\nSupplementary Maximum Height search register")
print("-" * 100)

print(
    supplementary_search_register[
        "Supplementary_Search_Status"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.8I. Export reconciliation products
# -----------------------------------------------------------------------------

RAW_MAPPING_PATH = (
    OUTPUT_DIR /
    "omo_max_height_recovered_taxonomic_mapping_evidence.csv"
)

SEARCH_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_supplementary_search_register.csv"
)


recovered_mapping_raw.to_csv(
    RAW_MAPPING_PATH,
    index=False
)

supplementary_search_register.to_csv(
    SEARCH_REGISTER_PATH,
    index=False
)


print("\nReconciliation exports")
print("-" * 100)

print(
    f"Recovered mapping evidence : "
    f"{RAW_MAPPING_PATH}"
)

print(
    f"Supplementary register     : "
    f"{SEARCH_REGISTER_PATH}"
)


# -----------------------------------------------------------------------------
# 10.8J. Methodological safeguard
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
Recovered names are SEARCH AIDS, not replacements for the Omo canonical
taxonomy.

The authoritative 127-taxon canonical backbone remains unchanged.

An alternative name recovered from WD, SLA or Seed Mass work may be used
to locate Maximum Height evidence only when the relationship has already
been documented in Notebook 07B or can subsequently be verified.

No Maximum Height value has been assigned in this section.
"""
)


# -----------------------------------------------------------------------------
# 10.8K. Summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.8 TAXONOMIC RECONCILIATION AUDIT COMPLETE")
print("=" * 100)

print(
    f"Canonical taxa                    : "
    f"{len(height_manifest)}"
)

print(
    f"BIEN QC-passed species            : "
    f"{len(bien_resolved_taxa)}"
)

print(
    f"Residual taxa                     : "
    f"{len(supplementary_search_register)}"
)

print(
    f"Residual taxa with alternatives   : "
    f"{supplementary_search_register['Alternative_Name_Available'].sum()}"
)

n_genus_unresolved = (
    supplementary_search_register[
        "Max_Height_Target_Level"
    ]
    .eq("GENUS_LEVEL_OR_UNRESOLVED")
    .sum()
)

print(
    f"Genus/unresolved targets          : "
    f"{n_genus_unresolved}"
)

print(
    "New Maximum Height values assigned: 0"
)

print(
    "Next step                        : "
    "Evaluate the recovered taxonomic search register "
    "and select the supplementary structured height source."
)
```

```python
# =============================================================================
# SECTION 10.9 — CLEANING AND VALIDATION OF RECOVERED TAXONOMIC SEARCH NAMES
# Remove non-taxonomic artifacts and retain plausible botanical names only
# =============================================================================

from pathlib import Path
import re

import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.9 — CLEAN AND VALIDATE RECOVERED TAXONOMIC SEARCH NAMES")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.9A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

RAW_MAPPING_PATH = (
    OUTPUT_DIR /
    "omo_max_height_recovered_taxonomic_mapping_evidence.csv"
)

SEARCH_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_supplementary_search_register.csv"
)


# -----------------------------------------------------------------------------
# 10.9B. Load mapping evidence
# -----------------------------------------------------------------------------

recovered_mapping_raw = pd.read_csv(
    RAW_MAPPING_PATH,
    low_memory=False
)

supplementary_search_register = pd.read_csv(
    SEARCH_REGISTER_PATH,
    low_memory=False
)


print("\nInput state")
print("-" * 100)

print(
    f"Recovered mapping rows : "
    f"{len(recovered_mapping_raw)}"
)

print(
    f"Residual taxa          : "
    f"{len(supplementary_search_register)}"
)


# -----------------------------------------------------------------------------
# 10.9C. Botanical-name plausibility rules
# -----------------------------------------------------------------------------

def normalize_candidate(value):
    """
    Normalize a recovered search-name candidate.
    """

    if pd.isna(value):
        return None

    value = str(value).strip()

    if not value:
        return None

    return value


def looks_like_boolean_or_null(value):
    """
    Reject obvious non-name logical/null artifacts.
    """

    if value is None:
        return True

    return value.casefold() in {
        "true",
        "false",
        "nan",
        "none",
        "null",
        "<na>",
        "na",
    }


def looks_numeric(value):
    """
    Reject pure integer/decimal identifiers.
    """

    if value is None:
        return True

    return bool(
        re.fullmatch(
            r"[+-]?\d+(?:\.\d+)?",
            value
        )
    )


def looks_like_author_only(value):
    """
    Reject strings that appear to be author citations rather
    than taxon names.
    """

    if value is None:
        return True

    # Parenthesized author string
    if value.startswith("("):
        return True

    # Typical botanical author abbreviations without a clear genus
    author_tokens = [
        " ex ",
        " & ",
        " et al",
    ]

    if any(token in value for token in author_tokens):

        words = value.split()

        # A valid species name normally starts with an alphabetic genus
        # followed by a lowercase epithet.
        if len(words) < 2:
            return True

    return False


def plausible_genus_name(value):
    """
    Accept a single-word botanical genus-like name.
    """

    if value is None:
        return False

    return bool(
        re.fullmatch(
            r"[A-Z][A-Za-z-]+",
            value
        )
    )


def plausible_species_name(value):
    """
    Accept plausible binomial/trinomial botanical names.

    Allows:
        Genus species
        Genus species subspecies-like third token

    Does not attempt formal nomenclatural validation.
    """

    if value is None:
        return False

    value = re.sub(
        r"\s+",
        " ",
        value.strip()
    )

    pattern = (
        r"^[A-Z][A-Za-z-]+\s+"
        r"[a-z][A-Za-z-]+"
        r"(?:\s+[a-z][A-Za-z-]+)?$"
    )

    return bool(
        re.fullmatch(
            pattern,
            value
        )
    )


def classify_candidate(value):
    """
    Conservative recovered-name classification.
    """

    value = normalize_candidate(value)

    if value is None:
        return "REJECT_EMPTY"

    if looks_like_boolean_or_null(value):
        return "REJECT_LOGICAL_OR_NULL"

    if looks_numeric(value):
        return "REJECT_NUMERIC"

    if looks_like_author_only(value):
        return "REJECT_AUTHOR_OR_METADATA"

    if plausible_species_name(value):
        return "PLAUSIBLE_SPECIES_NAME"

    if plausible_genus_name(value):
        return "PLAUSIBLE_GENUS_NAME"

    return "REJECT_OTHER_NON_TAXON"


# -----------------------------------------------------------------------------
# 10.9D. Classify all recovered search-name candidates
# -----------------------------------------------------------------------------

recovered_mapping_qc = recovered_mapping_raw.copy()


recovered_mapping_qc[
    "Recovered_Search_Name"
] = (
    recovered_mapping_qc[
        "Recovered_Search_Name"
    ]
    .astype("string")
    .str.strip()
)


recovered_mapping_qc[
    "Search_Name_QC_Class"
] = recovered_mapping_qc[
    "Recovered_Search_Name"
].map(
    classify_candidate
)


print("\nRecovered-name QC classification")
print("-" * 100)

print(
    recovered_mapping_qc[
        "Search_Name_QC_Class"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.9E. Retain only plausible botanical names
# -----------------------------------------------------------------------------

accepted_classes = {
    "PLAUSIBLE_SPECIES_NAME",
    "PLAUSIBLE_GENUS_NAME",
}


clean_mapping = recovered_mapping_qc[
    recovered_mapping_qc[
        "Search_Name_QC_Class"
    ].isin(
        accepted_classes
    )
].copy()


clean_mapping = (
    clean_mapping
    .drop_duplicates(
        subset=[
            "Canonical_Species",
            "Recovered_Search_Name",
        ]
    )
    .reset_index(drop=True)
)


print("\nClean botanical search-name evidence")
print("-" * 100)

print(
    f"Accepted mapping rows : "
    f"{len(clean_mapping)}"
)

print(
    f"Taxa represented      : "
    f"{clean_mapping['Canonical_Species'].nunique()}"
)


# -----------------------------------------------------------------------------
# 10.9F. Remove canonical-identical names from alternative-name pool
# -----------------------------------------------------------------------------

clean_mapping[
    "Is_Identical_To_Canonical"
] = (
    clean_mapping[
        "Canonical_Species"
    ].str.casefold()
    ==
    clean_mapping[
        "Recovered_Search_Name"
    ].str.casefold()
)


true_alternatives = clean_mapping[
    ~clean_mapping[
        "Is_Identical_To_Canonical"
    ]
].copy()


print("\nClean alternative-name evidence")
print("-" * 100)

print(
    f"Alternative mapping rows : "
    f"{len(true_alternatives)}"
)

print(
    f"Taxa with alternatives   : "
    f"{true_alternatives['Canonical_Species'].nunique()}"
)


# -----------------------------------------------------------------------------
# 10.9G. Summarize clean alternatives
# -----------------------------------------------------------------------------

if not true_alternatives.empty:

    clean_alternative_summary = (
        true_alternatives
        .groupby(
            "Canonical_Species",
            as_index=False
        )
        .agg(
            Clean_Alternative_Names=(
                "Recovered_Search_Name",
                lambda x:
                    " | ".join(
                        sorted(
                            set(
                                str(v).strip()
                                for v in x
                            )
                        )
                    )
            ),
            N_Clean_Alternative_Names=(
                "Recovered_Search_Name",
                "nunique"
            ),
            Evidence_Files=(
                "Evidence_File",
                lambda x:
                    " | ".join(
                        sorted(set(x))
                    )
            ),
        )
    )

else:

    clean_alternative_summary = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "Clean_Alternative_Names",
            "N_Clean_Alternative_Names",
            "Evidence_Files",
        ]
    )


print("\nClean alternative search-name summary")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 200,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        clean_alternative_summary
    )


# -----------------------------------------------------------------------------
# 10.9H. Build cleaned supplementary search register
# -----------------------------------------------------------------------------

clean_search_register = (
    supplementary_search_register[
        [
            "Canonical_Species",
            "Max_Height_Target_Level",
        ]
    ]
    .copy()
)


clean_search_register = clean_search_register.merge(
    clean_alternative_summary,
    on="Canonical_Species",
    how="left"
)


clean_search_register[
    "Primary_Search_Name"
] = clean_search_register[
    "Canonical_Species"
]


clean_search_register[
    "Clean_Alternative_Available"
] = clean_search_register[
    "Clean_Alternative_Names"
].notna()


clean_search_register[
    "Search_Strategy"
] = np.where(
    clean_search_register[
        "Clean_Alternative_Available"
    ],
    "CANONICAL_PLUS_VERIFIED_ALTERNATIVES",
    "CANONICAL_ONLY"
)


clean_search_register.loc[
    clean_search_register[
        "Max_Height_Target_Level"
    ] == "GENUS_LEVEL_OR_UNRESOLVED",
    "Search_Strategy"
] = "GENUS_LEVEL_TARGET"


# -----------------------------------------------------------------------------
# 10.9I. Sanity checks
# -----------------------------------------------------------------------------

assert len(clean_search_register) == 124

assert (
    clean_search_register[
        "Canonical_Species"
    ].nunique()
    == 124
)


# Ensure obvious artifacts have disappeared
for bad_value in [
    "False",
    "True",
    "0",
    "1",
    "2",
    "4",
    "13",
    "30",
    "42",
    "138",
    "140",
    "147",
    "203",
    "735",
]:

    assert not (
        clean_mapping[
            "Recovered_Search_Name"
        ]
        .astype(str)
        .eq(bad_value)
        .any()
    )


print("\nClean supplementary search strategy")
print("-" * 100)

print(
    clean_search_register[
        "Search_Strategy"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.9J. Rejected-artifact diagnostic
# -----------------------------------------------------------------------------

rejected_mapping = recovered_mapping_qc[
    ~recovered_mapping_qc[
        "Search_Name_QC_Class"
    ].isin(
        accepted_classes
    )
].copy()


print("\nRejected recovered-name artifacts")
print("-" * 100)

print(
    f"Rejected rows : "
    f"{len(rejected_mapping)}"
)


print(
    rejected_mapping[
        "Search_Name_QC_Class"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.9K. Export cleaned products
# -----------------------------------------------------------------------------

QC_PATH = (
    OUTPUT_DIR /
    "omo_max_height_recovered_taxonomic_mapping_qc.csv"
)

CLEAN_MAPPING_PATH = (
    OUTPUT_DIR /
    "omo_max_height_clean_taxonomic_mapping.csv"
)

CLEAN_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_clean_supplementary_search_register.csv"
)

REJECTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_rejected_mapping_artifacts.csv"
)


recovered_mapping_qc.to_csv(
    QC_PATH,
    index=False
)

clean_mapping.to_csv(
    CLEAN_MAPPING_PATH,
    index=False
)

clean_search_register.to_csv(
    CLEAN_REGISTER_PATH,
    index=False
)

rejected_mapping.to_csv(
    REJECTED_PATH,
    index=False
)


print("\nCleaning exports")
print("-" * 100)

print(
    f"QC register        : {QC_PATH}"
)

print(
    f"Clean mappings     : {CLEAN_MAPPING_PATH}"
)

print(
    f"Clean search file  : {CLEAN_REGISTER_PATH}"
)

print(
    f"Rejected artifacts : {REJECTED_PATH}"
)


# -----------------------------------------------------------------------------
# 10.9L. Methodological safeguard
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
This cleaning stage removes logical values, numeric identifiers,
author-only strings and other metadata artifacts that were accidentally
captured during the broad reconciliation scan.

The resulting names are PLAUSIBLE botanical search terms only.

They are not automatically accepted as taxonomic synonyms. They remain
search aids whose relationship to the Omo canonical taxon must be
supported by existing Notebook 07B taxonomic evidence or verified during
supplementary acquisition.

The canonical Omo taxonomy remains unchanged.
"""
)


# -----------------------------------------------------------------------------
# 10.9M. Summary
# -----------------------------------------------------------------------------

n_clean_alt_taxa = int(
    clean_search_register[
        "Clean_Alternative_Available"
    ].sum()
)

n_genus_targets = int(
    clean_search_register[
        "Max_Height_Target_Level"
    ]
    .eq("GENUS_LEVEL_OR_UNRESOLVED")
    .sum()
)


print("=" * 100)
print("SECTION 10.9 TAXONOMIC SEARCH-NAME CLEANING COMPLETE")
print("=" * 100)

print(
    f"Residual taxa                  : "
    f"{len(clean_search_register)}"
)

print(
    f"Taxa with clean alternatives   : "
    f"{n_clean_alt_taxa}"
)

print(
    f"Canonical-only taxa             : "
    f"{len(clean_search_register) - n_clean_alt_taxa - n_genus_targets}"
)

print(
    f"Genus-level targets             : "
    f"{n_genus_targets}"
)

print(
    f"Rejected artifact rows          : "
    f"{len(rejected_mapping)}"
)

print(
    "New Maximum Height values      : 0"
)

print(
    "Next step                      : "
    "Select and test a supplementary structured "
    "Maximum Height source using the cleaned search register."
)
```

```python
# =============================================================================
# SECTION 10.11 — CONTROLLED OMO-TO-GIFT MATCHING AND COVERAGE AUDIT
# Exact canonical matching + controlled Section 10.9 alternative matching
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.11 — CONTROLLED OMO-TO-GIFT MATCHING AND COVERAGE AUDIT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.11A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

GIFT_DIR = (
    PROJECT_ROOT /
    "data" /
    "external" /
    "GIFT"
)

GIFT_PATH = (
    GIFT_DIR /
    "omo_max_height_gift_species_global_raw.csv"
)

SEARCH_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_clean_supplementary_search_register.csv"
)

BIEN_VALUES_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_provisional_species_values.csv"
)

TARGET_MANIFEST_PATH = (
    OUTPUT_DIR /
    "omo_max_height_target_manifest.csv"
)


# -----------------------------------------------------------------------------
# 10.11B. Load inputs
# -----------------------------------------------------------------------------

gift = pd.read_csv(
    GIFT_PATH,
    low_memory=False
)

search_register = pd.read_csv(
    SEARCH_REGISTER_PATH,
    low_memory=False
)

bien_values = pd.read_csv(
    BIEN_VALUES_PATH,
    low_memory=False
)

target_manifest = pd.read_csv(
    TARGET_MANIFEST_PATH,
    low_memory=False
)


print("\nInput files")
print("-" * 100)

print(f"GIFT rows              : {len(gift)}")
print(f"Residual search taxa   : {len(search_register)}")
print(f"BIEN provisional taxa  : {len(bien_values)}")
print(f"Canonical target taxa  : {len(target_manifest)}")


# -----------------------------------------------------------------------------
# 10.11C. Input validation
# -----------------------------------------------------------------------------

required_gift_columns = [
    "work_ID",
    "work_species",
    "trait_value_1.6.2",
    "cv_1.6.2",
    "n_1.6.2",
    "references_1.6.2",
]

missing_gift_columns = [
    col for col in required_gift_columns
    if col not in gift.columns
]

if missing_gift_columns:
    raise ValueError(
        "Required GIFT columns missing: "
        + ", ".join(missing_gift_columns)
    )


required_search_columns = [
    "Canonical_Species",
    "Max_Height_Target_Level",
    "Clean_Alternative_Names",
    "Search_Strategy",
]

missing_search_columns = [
    col for col in required_search_columns
    if col not in search_register.columns
]

if missing_search_columns:
    raise ValueError(
        "Required search-register columns missing: "
        + ", ".join(missing_search_columns)
    )


assert len(target_manifest) == 127
assert target_manifest["Canonical_Species"].nunique() == 127

assert len(search_register) == 124
assert search_register["Canonical_Species"].nunique() == 124

assert len(bien_values) == 3


print("\nSchema and backbone validation")
print("-" * 100)
print("Required GIFT fields present      : YES")
print("Required search fields present    : YES")
print("Canonical 127-taxon backbone      : VALID")
print("Residual 124-taxon register       : VALID")


# -----------------------------------------------------------------------------
# 10.11D. Normalize GIFT species and height fields
# -----------------------------------------------------------------------------

gift = gift.copy()

gift["work_species"] = (
    gift["work_species"]
    .astype("string")
    .str.strip()
)

gift["Gift_Name_Key"] = (
    gift["work_species"]
    .str.casefold()
)

gift["Max_Height_m"] = pd.to_numeric(
    gift["trait_value_1.6.2"],
    errors="coerce"
)


gift["GIFT_Value_Valid"] = (
    gift["Max_Height_m"].notna()
    &
    (gift["Max_Height_m"] > 0)
)


gift_valid = gift[
    gift["GIFT_Value_Valid"]
].copy()


print("\nGIFT trait validation")
print("-" * 100)

print(
    f"Numeric positive heights : "
    f"{len(gift_valid)} / {len(gift)}"
)

print(
    f"Unique GIFT species names : "
    f"{gift_valid['Gift_Name_Key'].nunique()}"
)

print(
    f"Duplicate GIFT name keys  : "
    f"{gift_valid['Gift_Name_Key'].duplicated().sum()}"
)


# -----------------------------------------------------------------------------
# 10.11E. Check uniqueness of GIFT working species
# -----------------------------------------------------------------------------

gift_name_counts = (
    gift_valid
    .groupby(
        "Gift_Name_Key",
        dropna=False
    )
    .size()
    .reset_index(name="N_GIFT_Rows")
)


duplicate_gift_names = gift_name_counts[
    gift_name_counts["N_GIFT_Rows"] > 1
].copy()


print("\nGIFT species-name uniqueness")
print("-" * 100)

print(
    f"Species-name keys represented more than once : "
    f"{len(duplicate_gift_names)}"
)


# -----------------------------------------------------------------------------
# 10.11F. Prepare residual canonical targets
# -----------------------------------------------------------------------------

search_register = search_register.copy()

search_register["Canonical_Name_Key"] = (
    search_register["Canonical_Species"]
    .astype("string")
    .str.strip()
    .str.casefold()
)


# -----------------------------------------------------------------------------
# 10.11G. Exact canonical-name matching
# -----------------------------------------------------------------------------

canonical_matches = search_register.merge(
    gift_valid,
    left_on="Canonical_Name_Key",
    right_on="Gift_Name_Key",
    how="inner",
    suffixes=("_OMO", "_GIFT")
)


canonical_matches[
    "GIFT_Match_Type"
] = "EXACT_CANONICAL"


print("\nExact canonical-name coverage")
print("-" * 100)

print(
    f"Matched GIFT rows       : "
    f"{len(canonical_matches)}"
)

print(
    f"Matched Omo taxa        : "
    f"{canonical_matches['Canonical_Species'].nunique()}"
)

print(
    f"Residual coverage       : "
    f"{100 * canonical_matches['Canonical_Species'].nunique() / len(search_register):.2f}%"
)


# -----------------------------------------------------------------------------
# 10.11H. Identify taxa unresolved by exact canonical matching
# -----------------------------------------------------------------------------

canonical_matched_taxa = set(
    canonical_matches[
        "Canonical_Species"
    ].dropna()
)


unresolved_after_canonical = search_register[
    ~search_register[
        "Canonical_Species"
    ].isin(
        canonical_matched_taxa
    )
].copy()


print("\nAfter exact canonical matching")
print("-" * 100)

print(
    f"Resolved taxa   : "
    f"{len(canonical_matched_taxa)}"
)

print(
    f"Unresolved taxa : "
    f"{len(unresolved_after_canonical)}"
)


# -----------------------------------------------------------------------------
# 10.11I. Expand CLEAN alternative names for unresolved taxa only
# -----------------------------------------------------------------------------

alternative_rows = []


for _, row in unresolved_after_canonical.iterrows():

    canonical = row["Canonical_Species"]
    alternatives = row.get(
        "Clean_Alternative_Names",
        np.nan
    )

    if pd.isna(alternatives):
        continue

    for alternative in str(alternatives).split("|"):

        alternative = alternative.strip()

        if not alternative:
            continue

        alternative_rows.append(
            {
                "Canonical_Species": canonical,
                "Alternative_Search_Name": alternative,
                "Alternative_Name_Key":
                    alternative.casefold(),
            }
        )


alternative_long = pd.DataFrame(
    alternative_rows
)


if not alternative_long.empty:

    alternative_long = (
        alternative_long
        .drop_duplicates()
        .reset_index(drop=True)
    )


print("\nControlled alternative-name search pool")
print("-" * 100)

print(
    f"Alternative search rows : "
    f"{len(alternative_long)}"
)

if not alternative_long.empty:

    print(
        f"Taxa represented        : "
        f"{alternative_long['Canonical_Species'].nunique()}"
    )

else:

    print("Taxa represented        : 0")


# -----------------------------------------------------------------------------
# 10.11J. Match cleaned alternatives to GIFT
# -----------------------------------------------------------------------------

if not alternative_long.empty:

    alternative_matches = alternative_long.merge(
        gift_valid,
        left_on="Alternative_Name_Key",
        right_on="Gift_Name_Key",
        how="inner"
    )

    alternative_matches[
        "GIFT_Match_Type"
    ] = "CLEAN_ALTERNATIVE"

else:

    alternative_matches = pd.DataFrame()


print("\nControlled alternative-name coverage")
print("-" * 100)

print(
    f"Matched GIFT rows : "
    f"{len(alternative_matches)}"
)

if not alternative_matches.empty:

    print(
        f"Matched Omo taxa  : "
        f"{alternative_matches['Canonical_Species'].nunique()}"
    )

else:

    print("Matched Omo taxa  : 0")


# -----------------------------------------------------------------------------
# 10.11K. Detect alternative-name ambiguity
# -----------------------------------------------------------------------------

if not alternative_matches.empty:

    alternative_match_summary = (
        alternative_matches
        .groupby(
            "Canonical_Species",
            as_index=False
        )
        .agg(
            N_GIFT_Alternative_Matches=(
                "work_species",
                "nunique"
            ),
            GIFT_Alternative_Names=(
                "work_species",
                lambda x:
                    " | ".join(
                        sorted(
                            set(
                                str(v).strip()
                                for v in x
                            )
                        )
                    )
            ),
            GIFT_Alternative_Heights_m=(
                "Max_Height_m",
                lambda x:
                    " | ".join(
                        str(v)
                        for v in sorted(
                            set(
                                float(v)
                                for v in x
                            )
                        )
                    )
            ),
        )
    )

    ambiguous_alternative_taxa = (
        alternative_match_summary[
            alternative_match_summary[
                "N_GIFT_Alternative_Matches"
            ] > 1
        ].copy()
    )

else:

    alternative_match_summary = pd.DataFrame(
        columns=[
            "Canonical_Species",
            "N_GIFT_Alternative_Matches",
            "GIFT_Alternative_Names",
            "GIFT_Alternative_Heights_m",
        ]
    )

    ambiguous_alternative_taxa = (
        alternative_match_summary.copy()
    )


print("\nAlternative-name ambiguity")
print("-" * 100)

print(
    f"Taxa with >1 GIFT alternative match : "
    f"{len(ambiguous_alternative_taxa)}"
)


if not ambiguous_alternative_taxa.empty:

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 120,
    ):
        display(
            ambiguous_alternative_taxa
        )


# -----------------------------------------------------------------------------
# 10.11L. Build provisional GIFT match ledger
# -----------------------------------------------------------------------------

canonical_ledger_columns = [
    "Canonical_Species",
    "work_ID",
    "work_species",
    "Max_Height_m",
    "cv_1.6.2",
    "n_1.6.2",
    "references_1.6.2",
    "GIFT_Match_Type",
]


canonical_ledger = canonical_matches[
    canonical_ledger_columns
].copy()


if not alternative_matches.empty:

    alternative_ledger = alternative_matches[
        canonical_ledger_columns
    ].copy()

else:

    alternative_ledger = pd.DataFrame(
        columns=canonical_ledger_columns
    )


gift_match_ledger = pd.concat(
    [
        canonical_ledger,
        alternative_ledger,
    ],
    ignore_index=True
)


gift_match_ledger = (
    gift_match_ledger
    .drop_duplicates()
    .reset_index(drop=True)
)


# -----------------------------------------------------------------------------
# 10.11M. Per-taxon GIFT match count
# -----------------------------------------------------------------------------

gift_taxon_match_counts = (
    gift_match_ledger
    .groupby(
        "Canonical_Species",
        as_index=False
    )
    .agg(
        N_GIFT_Matches=(
            "work_species",
            "nunique"
        ),
        N_GIFT_Height_Values=(
            "Max_Height_m",
            "nunique"
        ),
    )
)


unambiguous_gift_taxa = gift_taxon_match_counts[
    (gift_taxon_match_counts["N_GIFT_Matches"] == 1)
    &
    (gift_taxon_match_counts["N_GIFT_Height_Values"] == 1)
].copy()


ambiguous_gift_taxa = gift_taxon_match_counts[
    (gift_taxon_match_counts["N_GIFT_Matches"] > 1)
    |
    (gift_taxon_match_counts["N_GIFT_Height_Values"] > 1)
].copy()


print("\nCombined GIFT match status")
print("-" * 100)

print(
    f"Taxa with any GIFT match       : "
    f"{gift_taxon_match_counts['Canonical_Species'].nunique()}"
)

print(
    f"Unambiguous matched taxa       : "
    f"{len(unambiguous_gift_taxa)}"
)

print(
    f"Ambiguous matched taxa         : "
    f"{len(ambiguous_gift_taxa)}"
)


# -----------------------------------------------------------------------------
# 10.11N. Identify still-unresolved residual taxa
# -----------------------------------------------------------------------------

gift_matched_taxa = set(
    gift_taxon_match_counts[
        "Canonical_Species"
    ]
)


still_unresolved = search_register[
    ~search_register[
        "Canonical_Species"
    ].isin(
        gift_matched_taxa
    )
].copy()


print("\nResidual taxa after GIFT matching")
print("-" * 100)

print(
    f"Still unresolved taxa : "
    f"{len(still_unresolved)}"
)


if not still_unresolved.empty:

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 120,
    ):
        display(
            still_unresolved[
                [
                    "Canonical_Species",
                    "Max_Height_Target_Level",
                    "Clean_Alternative_Names",
                    "Search_Strategy",
                ]
            ]
        )


# -----------------------------------------------------------------------------
# 10.11O. BIEN–GIFT overlap audit
# -----------------------------------------------------------------------------

bien_overlap_targets = (
    bien_values[
        [
            "Omo_Canonical_Species",
            "Max_Height_m",
        ]
    ]
    .rename(
        columns={
            "Omo_Canonical_Species":
                "Canonical_Species",
            "Max_Height_m":
                "BIEN_Max_Height_m",
        }
    )
)


bien_overlap_targets[
    "Canonical_Name_Key"
] = (
    bien_overlap_targets[
        "Canonical_Species"
    ]
    .astype("string")
    .str.strip()
    .str.casefold()
)


bien_gift_overlap = bien_overlap_targets.merge(
    gift_valid[
        [
            "work_ID",
            "work_species",
            "Gift_Name_Key",
            "Max_Height_m",
            "cv_1.6.2",
            "n_1.6.2",
            "references_1.6.2",
        ]
    ],
    left_on="Canonical_Name_Key",
    right_on="Gift_Name_Key",
    how="left"
)


bien_gift_overlap = bien_gift_overlap.rename(
    columns={
        "Max_Height_m":
            "GIFT_Max_Height_m"
    }
)


bien_gift_overlap[
    "Absolute_Difference_m"
] = (
    bien_gift_overlap[
        "GIFT_Max_Height_m"
    ]
    -
    bien_gift_overlap[
        "BIEN_Max_Height_m"
    ]
).abs()


bien_gift_overlap[
    "Percent_Difference_vs_BIEN"
] = (
    bien_gift_overlap[
        "Absolute_Difference_m"
    ]
    /
    bien_gift_overlap[
        "BIEN_Max_Height_m"
    ]
    * 100
)


print("\nBIEN–GIFT overlap audit")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 50,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        bien_gift_overlap[
            [
                "Canonical_Species",
                "BIEN_Max_Height_m",
                "work_species",
                "GIFT_Max_Height_m",
                "n_1.6.2",
                "Absolute_Difference_m",
                "Percent_Difference_vs_BIEN",
            ]
        ]
    )


# -----------------------------------------------------------------------------
# 10.11P. Broad plausibility of Omo-matched GIFT values
# -----------------------------------------------------------------------------

gift_match_ledger[
    "Broad_Plausibility_OK"
] = (
    gift_match_ledger[
        "Max_Height_m"
    ].between(
        1,
        100,
        inclusive="both"
    )
)


plausibility_review = gift_match_ledger[
    ~gift_match_ledger[
        "Broad_Plausibility_OK"
    ]
].copy()


print("\nOmo-matched GIFT biological plausibility screen")
print("-" * 100)

print(
    f"Matched records requiring review (<1 m or >100 m) : "
    f"{len(plausibility_review)}"
)


if not plausibility_review.empty:

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
    ):
        display(
            plausibility_review
        )


# -----------------------------------------------------------------------------
# 10.11Q. Export audit products
# -----------------------------------------------------------------------------

MATCH_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_match_ledger.csv"
)

MATCH_COUNTS_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_match_counts.csv"
)

AMBIGUOUS_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_ambiguous_matches.csv"
)

UNRESOLVED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_unresolved_taxa.csv"
)

OVERLAP_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_gift_overlap_audit.csv"
)

PLAUSIBILITY_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_plausibility_review.csv"
)


gift_match_ledger.to_csv(
    MATCH_LEDGER_PATH,
    index=False
)

gift_taxon_match_counts.to_csv(
    MATCH_COUNTS_PATH,
    index=False
)

ambiguous_gift_taxa.to_csv(
    AMBIGUOUS_PATH,
    index=False
)

still_unresolved.to_csv(
    UNRESOLVED_PATH,
    index=False
)

bien_gift_overlap.to_csv(
    OVERLAP_PATH,
    index=False
)

plausibility_review.to_csv(
    PLAUSIBILITY_PATH,
    index=False
)


print("\nAudit exports")
print("-" * 100)

print(f"GIFT match ledger     : {MATCH_LEDGER_PATH}")
print(f"Match counts          : {MATCH_COUNTS_PATH}")
print(f"Ambiguous matches     : {AMBIGUOUS_PATH}")
print(f"Unresolved taxa       : {UNRESOLVED_PATH}")
print(f"BIEN–GIFT overlap     : {OVERLAP_PATH}")
print(f"Plausibility review   : {PLAUSIBILITY_PATH}")


# -----------------------------------------------------------------------------
# 10.11R. Methodological safeguard
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
This section performs controlled coverage discovery only.

Exact canonical names are searched first. Cleaned alternative names from
Section 10.9 are searched only for taxa that remain unresolved after the
canonical-name pass.

No GIFT value is accepted merely because a name matches.

Before incorporation, matched records must pass:
    • taxonomic reconciliation review where required;
    • ambiguity checks;
    • biological plausibility checks;
    • BIEN/GIFT overlap assessment where available; and
    • provenance assignment.

The existing Omo canonical taxonomy remains authoritative.
"""
)


# -----------------------------------------------------------------------------
# 10.11S. Final summary
# -----------------------------------------------------------------------------

n_exact_taxa = int(
    canonical_matches[
        "Canonical_Species"
    ].nunique()
)

n_alt_taxa = int(
    alternative_matches[
        "Canonical_Species"
    ].nunique()
) if not alternative_matches.empty else 0

n_any_gift = int(
    gift_taxon_match_counts[
        "Canonical_Species"
    ].nunique()
)


print("=" * 100)
print("SECTION 10.11 OMO-TO-GIFT COVERAGE AUDIT COMPLETE")
print("=" * 100)

print(f"Residual Omo targets              : {len(search_register)}")
print(f"Exact canonical GIFT matches      : {n_exact_taxa}")
print(f"Additional alternative matches    : {n_alt_taxa}")
print(f"Total taxa with GIFT evidence     : {n_any_gift}")
print(f"Unambiguous GIFT taxa             : {len(unambiguous_gift_taxa)}")
print(f"Ambiguous GIFT taxa               : {len(ambiguous_gift_taxa)}")
print(f"Still unresolved after GIFT       : {len(still_unresolved)}")
print(f"Plausibility-review records       : {len(plausibility_review)}")
print("New Maximum Height values assigned: 0")
print(
    "Next step                        : "
    "Section 10.12 — review GIFT matches, BIEN overlap and "
    "taxonomic reconciliation before acceptance."
)
```

```python
# =============================================================================
# SECTION 10.12 — GIFT ACCEPTANCE, SOURCE PRECEDENCE AND RESIDUAL CONSOLIDATION
# Accept unambiguous GIFT evidence while preserving BIEN precedence
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.12 — GIFT ACCEPTANCE, SOURCE PRECEDENCE AND RESIDUAL CONSOLIDATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.12A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

TARGET_MANIFEST_PATH = (
    OUTPUT_DIR /
    "omo_max_height_target_manifest.csv"
)

BIEN_VALUES_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_provisional_species_values.csv"
)

GIFT_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_match_ledger.csv"
)

GIFT_COUNTS_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_match_counts.csv"
)

GIFT_UNRESOLVED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_unresolved_taxa.csv"
)

BIEN_GIFT_OVERLAP_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_gift_overlap_audit.csv"
)


# -----------------------------------------------------------------------------
# 10.12B. Load validated inputs
# -----------------------------------------------------------------------------

target_manifest = pd.read_csv(
    TARGET_MANIFEST_PATH,
    low_memory=False
)

bien_values = pd.read_csv(
    BIEN_VALUES_PATH,
    low_memory=False
)

gift_ledger = pd.read_csv(
    GIFT_LEDGER_PATH,
    low_memory=False
)

gift_counts = pd.read_csv(
    GIFT_COUNTS_PATH,
    low_memory=False
)

gift_unresolved = pd.read_csv(
    GIFT_UNRESOLVED_PATH,
    low_memory=False
)

bien_gift_overlap = pd.read_csv(
    BIEN_GIFT_OVERLAP_PATH,
    low_memory=False
)


print("\nLoaded acquisition state")
print("-" * 100)

print(f"Canonical targets           : {len(target_manifest)}")
print(f"Validated BIEN species      : {len(bien_values)}")
print(
    f"GIFT matched taxa           : "
    f"{gift_counts['Canonical_Species'].nunique()}"
)
print(f"GIFT unresolved taxa        : {len(gift_unresolved)}")


# -----------------------------------------------------------------------------
# 10.12C. Validate expected state
# -----------------------------------------------------------------------------

assert len(target_manifest) == 127
assert target_manifest["Canonical_Species"].nunique() == 127

assert len(bien_values) == 3
assert bien_values["Omo_Canonical_Species"].nunique() == 3

assert gift_counts["Canonical_Species"].nunique() == 99
assert len(gift_unresolved) == 25


unambiguous_taxa = gift_counts[
    (gift_counts["N_GIFT_Matches"] == 1)
    &
    (gift_counts["N_GIFT_Height_Values"] == 1)
]["Canonical_Species"].copy()


ambiguous_taxa = gift_counts[
    (gift_counts["N_GIFT_Matches"] > 1)
    |
    (gift_counts["N_GIFT_Height_Values"] > 1)
]["Canonical_Species"].copy()


print("\nGIFT acceptance-state validation")
print("-" * 100)

print(f"Unambiguous taxa : {len(unambiguous_taxa)}")
print(f"Ambiguous taxa   : {len(ambiguous_taxa)}")
print(f"Unresolved taxa  : {len(gift_unresolved)}")


assert len(unambiguous_taxa) == 97
assert len(ambiguous_taxa) == 2


# -----------------------------------------------------------------------------
# 10.12D. Extract one authoritative GIFT row per unambiguous taxon
# -----------------------------------------------------------------------------

gift_unambiguous_records = gift_ledger[
    gift_ledger[
        "Canonical_Species"
    ].isin(
        set(unambiguous_taxa)
    )
].copy()


gift_unambiguous_check = (
    gift_unambiguous_records
    .groupby(
        "Canonical_Species"
    )
    .agg(
        N_Rows=("work_species", "size"),
        N_Source_Names=("work_species", "nunique"),
        N_Height_Values=("Max_Height_m", "nunique"),
    )
    .reset_index()
)


problem_unambiguous = gift_unambiguous_check[
    (gift_unambiguous_check["N_Source_Names"] != 1)
    |
    (gift_unambiguous_check["N_Height_Values"] != 1)
].copy()


if not problem_unambiguous.empty:
    raise AssertionError(
        "One or more supposedly unambiguous GIFT taxa "
        "contain multiple source names or values."
    )


gift_accepted = (
    gift_unambiguous_records
    .sort_values(
        [
            "Canonical_Species",
            "GIFT_Match_Type",
        ]
    )
    .drop_duplicates(
        subset=[
            "Canonical_Species"
        ],
        keep="first"
    )
    .copy()
)


assert len(gift_accepted) == 97


# -----------------------------------------------------------------------------
# 10.12E. Construct accepted GIFT species-level values
# -----------------------------------------------------------------------------

gift_accepted_values = pd.DataFrame(
    {
        "Canonical_Species":
            gift_accepted[
                "Canonical_Species"
            ],

        "Max_Height_m":
            pd.to_numeric(
                gift_accepted[
                    "Max_Height_m"
                ],
                errors="coerce"
            ),

        "Max_Height_Assignment_Level":
            "SPECIES",

        "Max_Height_Provenance_Class":
            "SPECIES_GIFT",

        "Max_Height_Source":
            "GIFT",

        "Max_Height_Source_Taxon":
            gift_accepted[
                "work_species"
            ],

        "Max_Height_Match_Status":
            gift_accepted[
                "GIFT_Match_Type"
            ],

        "Max_Height_Unit":
            "m",

        "GIFT_Work_ID":
            gift_accepted[
                "work_ID"
            ],

        "GIFT_N_Source_Values":
            gift_accepted[
                "n_1.6.2"
            ],

        "GIFT_CV":
            gift_accepted[
                "cv_1.6.2"
            ],

        "GIFT_References":
            gift_accepted[
                "references_1.6.2"
            ],

        "Max_Height_Quality_Flag":
            "QC_PASS_ACCEPTED",
    }
)


# -----------------------------------------------------------------------------
# 10.12F. Validate accepted GIFT values
# -----------------------------------------------------------------------------

gift_accepted_values[
    "Max_Height_m"
] = pd.to_numeric(
    gift_accepted_values[
        "Max_Height_m"
    ],
    errors="coerce"
)


assert gift_accepted_values["Max_Height_m"].notna().all()

assert (
    gift_accepted_values[
        "Max_Height_m"
    ] > 0
).all()


assert (
    gift_accepted_values[
        "Max_Height_m"
    ].between(
        1,
        100,
        inclusive="both"
    )
).all()


assert (
    gift_accepted_values[
        "Canonical_Species"
    ].nunique()
    == 97
)


n_exact_gift = int(
    gift_accepted_values[
        "Max_Height_Match_Status"
    ]
    .eq("EXACT_CANONICAL")
    .sum()
)

n_alternative_gift = int(
    gift_accepted_values[
        "Max_Height_Match_Status"
    ]
    .eq("CLEAN_ALTERNATIVE")
    .sum()
)


print("\nAccepted GIFT Maximum Height evidence")
print("-" * 100)

print(
    f"Accepted taxa       : "
    f"{len(gift_accepted_values)}"
)

print(
    f"Exact canonical     : "
    f"{n_exact_gift}"
)

print(
    f"Alternative-name    : "
    f"{n_alternative_gift}"
)

print(
    f"Minimum height (m)  : "
    f"{gift_accepted_values['Max_Height_m'].min():.4f}"
)

print(
    f"Median height (m)   : "
    f"{gift_accepted_values['Max_Height_m'].median():.4f}"
)

print(
    f"Mean height (m)     : "
    f"{gift_accepted_values['Max_Height_m'].mean():.4f}"
)

print(
    f"Maximum height (m)  : "
    f"{gift_accepted_values['Max_Height_m'].max():.4f}"
)


# -----------------------------------------------------------------------------
# 10.12G. Convert validated BIEN values to accepted standard schema
# -----------------------------------------------------------------------------

bien_accepted_values = pd.DataFrame(
    {
        "Canonical_Species":
            bien_values[
                "Omo_Canonical_Species"
            ],

        "Max_Height_m":
            pd.to_numeric(
                bien_values[
                    "Max_Height_m"
                ],
                errors="coerce"
            ),

        "Max_Height_Assignment_Level":
            "SPECIES",

        "Max_Height_Provenance_Class":
            "SPECIES_BIEN",

        "Max_Height_Source":
            "BIEN",

        "Max_Height_Source_Taxon":
            bien_values[
                "Omo_Canonical_Species"
            ],

        "Max_Height_Match_Status":
            "EXACT_CANONICAL",

        "Max_Height_Unit":
            "m",

        "GIFT_Work_ID":
            pd.NA,

        "GIFT_N_Source_Values":
            pd.NA,

        "GIFT_CV":
            pd.NA,

        "GIFT_References":
            pd.NA,

        "Max_Height_Quality_Flag":
            "QC_PASS_ACCEPTED",
    }
)


assert len(bien_accepted_values) == 3


# -----------------------------------------------------------------------------
# 10.12H. Enforce source precedence
# -----------------------------------------------------------------------------

bien_taxa = set(
    bien_accepted_values[
        "Canonical_Species"
    ]
)

gift_taxa = set(
    gift_accepted_values[
        "Canonical_Species"
    ]
)

overlap_between_accepted_sources = (
    bien_taxa.intersection(
        gift_taxa
    )
)


print("\nAccepted-source overlap")
print("-" * 100)

print(
    f"BIEN/GIFT duplicate accepted taxa : "
    f"{len(overlap_between_accepted_sources)}"
)


if overlap_between_accepted_sources:
    raise AssertionError(
        "GIFT accepted table unexpectedly contains "
        "taxa already assigned by BIEN."
    )


# -----------------------------------------------------------------------------
# 10.12I. Consolidate currently accepted species-level evidence
# -----------------------------------------------------------------------------

accepted_max_height = pd.concat(
    [
        bien_accepted_values,
        gift_accepted_values,
    ],
    ignore_index=True
)


assert len(accepted_max_height) == 100

assert (
    accepted_max_height[
        "Canonical_Species"
    ].nunique()
    == 100
)


print("\nCurrent accepted Maximum Height evidence")
print("-" * 100)

print(
    accepted_max_height[
        "Max_Height_Provenance_Class"
    ].value_counts()
)


current_coverage = (
    100 *
    len(accepted_max_height)
    /
    127
)


print(
    f"\nCurrently resolved taxa : "
    f"{len(accepted_max_height)} / 127"
)

print(
    f"Current coverage (%)    : "
    f"{current_coverage:.2f}%"
)


# -----------------------------------------------------------------------------
# 10.12J. Create ambiguity hold register
# -----------------------------------------------------------------------------

gift_ambiguity_hold = gift_ledger[
    gift_ledger[
        "Canonical_Species"
    ].isin(
        set(ambiguous_taxa)
    )
].copy()


gift_ambiguity_hold[
    "Review_Status"
] = "HOLD_TAXONOMIC_AMBIGUITY"


gift_ambiguity_hold[
    "Review_Reason"
] = (
    "More than one plausible GIFT source taxon/value "
    "matched the Omo canonical taxon."
)


print("\nGIFT ambiguity hold register")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 100,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        gift_ambiguity_hold[
            [
                "Canonical_Species",
                "work_species",
                "Max_Height_m",
                "GIFT_Match_Type",
                "n_1.6.2",
                "references_1.6.2",
                "Review_Status",
            ]
        ]
    )


# -----------------------------------------------------------------------------
# 10.12K. BIEN–GIFT source-precedence review
# -----------------------------------------------------------------------------

bien_gift_overlap = bien_gift_overlap.copy()


bien_gift_overlap[
    "Selected_Source"
] = "BIEN"


bien_gift_overlap[
    "Selection_Rationale"
] = (
    "BIEN record independently acquired and QC-passed "
    "before GIFT supplementary acquisition; retain "
    "source-precedence consistency."
)


print("\nBIEN–GIFT source-precedence decisions")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 20,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        bien_gift_overlap[
            [
                "Canonical_Species",
                "BIEN_Max_Height_m",
                "GIFT_Max_Height_m",
                "Absolute_Difference_m",
                "Percent_Difference_vs_BIEN",
                "Selected_Source",
                "Selection_Rationale",
            ]
        ]
    )


# -----------------------------------------------------------------------------
# 10.12L. Build complete acquisition-status ledger for all 127 taxa
# -----------------------------------------------------------------------------

status_ledger = target_manifest[
    [
        "Canonical_Species",
        "Max_Height_Target_Level",
    ]
].copy()


accepted_lookup = accepted_max_height[
    [
        "Canonical_Species",
        "Max_Height_m",
        "Max_Height_Assignment_Level",
        "Max_Height_Provenance_Class",
        "Max_Height_Source",
        "Max_Height_Source_Taxon",
        "Max_Height_Match_Status",
        "Max_Height_Quality_Flag",
    ]
].copy()


status_ledger = status_ledger.merge(
    accepted_lookup,
    on="Canonical_Species",
    how="left"
)


ambiguous_set = set(
    ambiguous_taxa
)


status_ledger[
    "Acquisition_Status"
] = np.select(
    [
        status_ledger[
            "Max_Height_m"
        ].notna(),

        status_ledger[
            "Canonical_Species"
        ].isin(
            ambiguous_set
        ),

        status_ledger[
            "Max_Height_Target_Level"
        ].eq(
            "GENUS_LEVEL_OR_UNRESOLVED"
        ),
    ],
    [
        "SPECIES_LEVEL_ACCEPTED",
        "HOLD_TAXONOMIC_AMBIGUITY",
        "GENUS_LEVEL_ACQUISITION_REQUIRED",
    ],
    default="SUPPLEMENTARY_ACQUISITION_REQUIRED",
)


print("\nComplete Maximum Height acquisition status")
print("-" * 100)

print(
    status_ledger[
        "Acquisition_Status"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.12M. Validate residual accounting
# -----------------------------------------------------------------------------

n_accepted = int(
    status_ledger[
        "Acquisition_Status"
    ]
    .eq(
        "SPECIES_LEVEL_ACCEPTED"
    )
    .sum()
)

n_ambiguous = int(
    status_ledger[
        "Acquisition_Status"
    ]
    .eq(
        "HOLD_TAXONOMIC_AMBIGUITY"
    )
    .sum()
)

n_genus_required = int(
    status_ledger[
        "Acquisition_Status"
    ]
    .eq(
        "GENUS_LEVEL_ACQUISITION_REQUIRED"
    )
    .sum()
)

n_supplementary = int(
    status_ledger[
        "Acquisition_Status"
    ]
    .eq(
        "SUPPLEMENTARY_ACQUISITION_REQUIRED"
    )
    .sum()
)


assert n_accepted == 100
assert n_ambiguous == 2
assert n_genus_required == 2
assert n_supplementary == 23

assert (
    n_accepted
    + n_ambiguous
    + n_genus_required
    + n_supplementary
    == 127
)


print("\nResidual accounting")
print("-" * 100)

print(
    f"Species-level accepted             : "
    f"{n_accepted}"
)

print(
    f"Taxonomic ambiguity hold           : "
    f"{n_ambiguous}"
)

print(
    f"Species supplementary acquisition  : "
    f"{n_supplementary}"
)

print(
    f"Genus-level acquisition required   : "
    f"{n_genus_required}"
)

total_remaining = (
    n_ambiguous
    + n_supplementary
    + n_genus_required
)

print(
    f"Total remaining unresolved/held    : "
    f"{total_remaining}"
)


# -----------------------------------------------------------------------------
# 10.12N. Export authoritative interim products
# -----------------------------------------------------------------------------

GIFT_ACCEPTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_accepted_species_values.csv"
)

ACCEPTED_ALL_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

AMBIGUITY_HOLD_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_taxonomic_ambiguity_hold.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)

SOURCE_PRECEDENCE_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_gift_source_precedence.csv"
)


gift_accepted_values.to_csv(
    GIFT_ACCEPTED_PATH,
    index=False
)

accepted_max_height.to_csv(
    ACCEPTED_ALL_PATH,
    index=False
)

gift_ambiguity_hold.to_csv(
    AMBIGUITY_HOLD_PATH,
    index=False
)

status_ledger.to_csv(
    STATUS_LEDGER_PATH,
    index=False
)

bien_gift_overlap.to_csv(
    SOURCE_PRECEDENCE_PATH,
    index=False
)


print("\nInterim exports")
print("-" * 100)

print(
    f"Accepted GIFT values     : "
    f"{GIFT_ACCEPTED_PATH}"
)

print(
    f"All accepted evidence    : "
    f"{ACCEPTED_ALL_PATH}"
)

print(
    f"Ambiguity hold           : "
    f"{AMBIGUITY_HOLD_PATH}"
)

print(
    f"Acquisition status       : "
    f"{STATUS_LEDGER_PATH}"
)

print(
    f"Source precedence        : "
    f"{SOURCE_PRECEDENCE_PATH}"
)


# -----------------------------------------------------------------------------
# 10.12O. Methodological interpretation
# -----------------------------------------------------------------------------

print("\nMethodological interpretation")
print("-" * 100)

print(
    """
BIEN and GIFT are treated as complementary species-level evidence sources.

BIEN retains precedence for the three taxa acquired and fully QC-validated
before supplementary GIFT acquisition.

GIFT contributes accepted species-level evidence only where:

    • the Omo taxon was not already resolved by BIEN;
    • the GIFT match is taxonomically unambiguous;
    • exactly one GIFT height value is associated with the matched taxon;
    • the height is positive and biologically plausible; and
    • the match is either exact canonical or derived from the cleaned,
      documented reconciliation register.

The two ambiguous GIFT matches remain on HOLD and have not been assigned.

No genus, family or global estimate has yet been introduced.
"""
)


# -----------------------------------------------------------------------------
# 10.12P. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.12 GIFT ACCEPTANCE AND RESIDUAL CONSOLIDATION COMPLETE")
print("=" * 100)

print(
    f"Canonical taxa                     : "
    f"{len(target_manifest)}"
)

print(
    f"Accepted BIEN species              : "
    f"{len(bien_accepted_values)}"
)

print(
    f"Accepted GIFT species              : "
    f"{len(gift_accepted_values)}"
)

print(
    f"Total species-level accepted       : "
    f"{len(accepted_max_height)}"
)

print(
    f"Current species-level coverage (%) : "
    f"{current_coverage:.2f}"
)

print(
    f"GIFT ambiguity holds               : "
    f"{n_ambiguous}"
)

print(
    f"Species still requiring evidence   : "
    f"{n_supplementary}"
)

print(
    f"Genus-level targets                : "
    f"{n_genus_required}"
)

print(
    f"Total unresolved/held              : "
    f"{total_remaining}"
)

print(
    "Next step                         : "
    "Section 10.13 — resolve the two GIFT taxonomic ambiguities "
    "before supplementary acquisition for the remaining taxa."
)
```

```python
# =============================================================================
# SECTION 10.13 — RESOLUTION OF GIFT TAXONOMIC AMBIGUITY HOLDS
# Explicitly resolve Blighia unijigata and Diospyrous dendo
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.13 — RESOLUTION OF GIFT TAXONOMIC AMBIGUITY HOLDS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.13A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

ACCEPTED_CURRENT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

AMBIGUITY_HOLD_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_taxonomic_ambiguity_hold.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


# -----------------------------------------------------------------------------
# 10.13B. Load Section 10.12 products
# -----------------------------------------------------------------------------

accepted_current = pd.read_csv(
    ACCEPTED_CURRENT_PATH,
    low_memory=False
)

ambiguity_hold = pd.read_csv(
    AMBIGUITY_HOLD_PATH,
    low_memory=False
)

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)


print("\nInput state")
print("-" * 100)

print(
    f"Currently accepted taxa : "
    f"{accepted_current['Canonical_Species'].nunique()}"
)

print(
    f"Ambiguity-hold rows     : "
    f"{len(ambiguity_hold)}"
)

print(
    f"Ambiguity-hold taxa     : "
    f"{ambiguity_hold['Canonical_Species'].nunique()}"
)


assert (
    accepted_current[
        "Canonical_Species"
    ].nunique()
    == 100
)

assert (
    ambiguity_hold[
        "Canonical_Species"
    ].nunique()
    == 2
)


# -----------------------------------------------------------------------------
# 10.13C. Explicit taxonomic decisions
# -----------------------------------------------------------------------------

# These are deliberate, externally verified taxonomic resolutions.
#
# 1. Blighia unijigata
#    -> Blighia unijugata Baker
#    Accepted species in POWO.
#    "unijigata" is treated as an orthographic error.
#
# 2. Diospyrous dendo
#    -> Diospyros dendo Welw. ex Hiern
#    Accepted species in POWO.
#    "Diospyrous" is treated as a genus spelling error.
#
# The competing GIFT candidates:
#    Blighia sapida
#    Diospyros mespiliformis
# are separate accepted species and are therefore rejected for these
# two canonical Omo taxa.


resolution_decisions = pd.DataFrame(
    [
        {
            "Canonical_Species":
                "Blighia unijigata",

            "Selected_GIFT_Taxon":
                "Blighia unijugata",

            "Rejected_GIFT_Taxon":
                "Blighia sapida",

            "Taxonomic_Decision":
                "ORTHOGRAPHIC_CORRECTION",

            "Resolution_Basis":
                (
                    "POWO recognizes Blighia unijugata Baker as an "
                    "accepted species occurring in Nigeria. "
                    "Blighia sapida is a separate accepted species."
                ),

            "Verification_Source":
                "Plants of the World Online (Kew)",

            "Resolution_Status":
                "APPROVED"
        },

        {
            "Canonical_Species":
                "Diospyrous dendo",

            "Selected_GIFT_Taxon":
                "Diospyros dendo",

            "Rejected_GIFT_Taxon":
                "Diospyros mespiliformis",

            "Taxonomic_Decision":
                "ORTHOGRAPHIC_CORRECTION",

            "Resolution_Basis":
                (
                    "POWO recognizes Diospyros dendo Welw. ex Hiern "
                    "as an accepted species native to Nigeria and "
                    "treats Diospyros mespiliformis as a distinct "
                    "accepted species."
                ),

            "Verification_Source":
                "Plants of the World Online (Kew)",

            "Resolution_Status":
                "APPROVED"
        },
    ]
)


print("\nApproved ambiguity resolutions")
print("-" * 100)

display(
    resolution_decisions
)


# -----------------------------------------------------------------------------
# 10.13D. Retrieve the selected GIFT evidence rows
# -----------------------------------------------------------------------------

selected_evidence = ambiguity_hold.merge(
    resolution_decisions[
        [
            "Canonical_Species",
            "Selected_GIFT_Taxon",
            "Taxonomic_Decision",
            "Resolution_Basis",
            "Verification_Source",
            "Resolution_Status",
        ]
    ],
    on="Canonical_Species",
    how="inner"
)


selected_evidence = selected_evidence[
    selected_evidence[
        "work_species"
    ].eq(
        selected_evidence[
            "Selected_GIFT_Taxon"
        ]
    )
].copy()


print("\nSelected GIFT evidence")
print("-" * 100)

display(
    selected_evidence[
        [
            "Canonical_Species",
            "work_species",
            "Max_Height_m",
            "n_1.6.2",
            "references_1.6.2",
            "Taxonomic_Decision",
            "Resolution_Status",
        ]
    ]
)


assert len(selected_evidence) == 2

assert (
    selected_evidence[
        "Canonical_Species"
    ].nunique()
    == 2
)


# -----------------------------------------------------------------------------
# 10.13E. Validate selected values
# -----------------------------------------------------------------------------

selected_evidence[
    "Max_Height_m"
] = pd.to_numeric(
    selected_evidence[
        "Max_Height_m"
    ],
    errors="coerce"
)


assert selected_evidence[
    "Max_Height_m"
].notna().all()


assert (
    selected_evidence[
        "Max_Height_m"
    ] > 0
).all()


assert (
    selected_evidence[
        "Max_Height_m"
    ].between(
        1,
        100,
        inclusive="both"
    )
).all()


expected_values = {
    "Blighia unijigata": 25.0,
    "Diospyrous dendo": 15.24018532,
}


for canonical, expected_value in expected_values.items():

    observed = float(
        selected_evidence.loc[
            selected_evidence[
                "Canonical_Species"
            ].eq(canonical),
            "Max_Height_m"
        ].iloc[0]
    )

    assert np.isclose(
        observed,
        expected_value,
        rtol=1e-6,
        atol=1e-6
    )


print("\nSelected-value validation")
print("-" * 100)

print("Selected records        : 2")
print("Numeric values          : PASS")
print("Positive values         : PASS")
print("Broad plausibility      : PASS")
print("Expected GIFT values    : PASS")


# -----------------------------------------------------------------------------
# 10.13F. Convert resolutions to accepted Maximum Height schema
# -----------------------------------------------------------------------------

resolved_values = pd.DataFrame(
    {
        "Canonical_Species":
            selected_evidence[
                "Canonical_Species"
            ],

        "Max_Height_m":
            selected_evidence[
                "Max_Height_m"
            ],

        "Max_Height_Assignment_Level":
            "SPECIES",

        "Max_Height_Provenance_Class":
            "SPECIES_GIFT",

        "Max_Height_Source":
            "GIFT",

        "Max_Height_Source_Taxon":
            selected_evidence[
                "work_species"
            ],

        "Max_Height_Match_Status":
            "VERIFIED_ORTHOGRAPHIC_RECONCILIATION",

        "Max_Height_Unit":
            "m",

        "GIFT_Work_ID":
            selected_evidence[
                "work_ID"
            ],

        "GIFT_N_Source_Values":
            selected_evidence[
                "n_1.6.2"
            ],

        "GIFT_CV":
            selected_evidence[
                "cv_1.6.2"
            ],

        "GIFT_References":
            selected_evidence[
                "references_1.6.2"
            ],

        "Max_Height_Quality_Flag":
            "QC_PASS_ACCEPTED",
    }
)


print("\nResolved Maximum Height assignments")
print("-" * 100)

display(
    resolved_values[
        [
            "Canonical_Species",
            "Max_Height_Source_Taxon",
            "Max_Height_m",
            "Max_Height_Provenance_Class",
            "Max_Height_Match_Status",
            "Max_Height_Quality_Flag",
        ]
    ]
)


# -----------------------------------------------------------------------------
# 10.13G. Append to accepted evidence
# -----------------------------------------------------------------------------

existing_taxa = set(
    accepted_current[
        "Canonical_Species"
    ]
)


new_taxa = set(
    resolved_values[
        "Canonical_Species"
    ]
)


assert existing_taxa.isdisjoint(
    new_taxa
)


# Align columns explicitly before concatenation.
for col in accepted_current.columns:

    if col not in resolved_values.columns:
        resolved_values[col] = pd.NA


for col in resolved_values.columns:

    if col not in accepted_current.columns:
        accepted_current[col] = pd.NA


resolved_values = resolved_values[
    accepted_current.columns
]


accepted_updated = pd.concat(
    [
        accepted_current,
        resolved_values,
    ],
    ignore_index=True
)


assert len(accepted_updated) == 102

assert (
    accepted_updated[
        "Canonical_Species"
    ].nunique()
    == 102
)


print("\nUpdated accepted species-level evidence")
print("-" * 100)

print(
    accepted_updated[
        "Max_Height_Provenance_Class"
    ].value_counts()
)


updated_coverage = (
    100 *
    accepted_updated[
        "Canonical_Species"
    ].nunique()
    /
    127
)


print(
    f"\nAccepted taxa     : "
    f"{accepted_updated['Canonical_Species'].nunique()} / 127"
)

print(
    f"Coverage (%)      : "
    f"{updated_coverage:.2f}"
)


# -----------------------------------------------------------------------------
# 10.13H. Update acquisition-status ledger
# -----------------------------------------------------------------------------

status_updated = status_ledger.copy()


resolution_taxa = set(
    resolution_decisions[
        "Canonical_Species"
    ]
)


mask_resolved = status_updated[
    "Canonical_Species"
].isin(
    resolution_taxa
)


status_updated.loc[
    mask_resolved,
    "Acquisition_Status"
] = "SPECIES_LEVEL_ACCEPTED"


resolved_lookup = resolved_values.set_index(
    "Canonical_Species"
)


fields_to_update = [
    "Max_Height_m",
    "Max_Height_Assignment_Level",
    "Max_Height_Provenance_Class",
    "Max_Height_Source",
    "Max_Height_Source_Taxon",
    "Max_Height_Match_Status",
    "Max_Height_Quality_Flag",
]


for field in fields_to_update:

    for canonical in resolution_taxa:

        status_updated.loc[
            status_updated[
                "Canonical_Species"
            ].eq(canonical),
            field
        ] = resolved_lookup.loc[
            canonical,
            field
        ]


# -----------------------------------------------------------------------------
# 10.13I. Validate updated acquisition accounting
# -----------------------------------------------------------------------------

status_counts = status_updated[
    "Acquisition_Status"
].value_counts()


n_accepted_updated = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "SPECIES_LEVEL_ACCEPTED"
    )
    .sum()
)

n_ambiguity_updated = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "HOLD_TAXONOMIC_AMBIGUITY"
    )
    .sum()
)

n_supplementary_updated = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "SUPPLEMENTARY_ACQUISITION_REQUIRED"
    )
    .sum()
)

n_genus_updated = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "GENUS_LEVEL_ACQUISITION_REQUIRED"
    )
    .sum()
)


assert n_accepted_updated == 102
assert n_ambiguity_updated == 0
assert n_supplementary_updated == 23
assert n_genus_updated == 2


assert (
    n_accepted_updated
    + n_ambiguity_updated
    + n_supplementary_updated
    + n_genus_updated
    == 127
)


print("\nUpdated Maximum Height acquisition status")
print("-" * 100)

print(status_counts)


print("\nUpdated residual accounting")
print("-" * 100)

print(
    f"Species-level accepted            : "
    f"{n_accepted_updated}"
)

print(
    f"Taxonomic ambiguity holds         : "
    f"{n_ambiguity_updated}"
)

print(
    f"Species requiring more evidence   : "
    f"{n_supplementary_updated}"
)

print(
    f"Genus-level targets               : "
    f"{n_genus_updated}"
)

print(
    f"Total remaining unresolved        : "
    f"{n_supplementary_updated + n_genus_updated}"
)


# -----------------------------------------------------------------------------
# 10.13J. Create formal taxonomic-resolution register
# -----------------------------------------------------------------------------

resolution_register = resolution_decisions.merge(
    selected_evidence[
        [
            "Canonical_Species",
            "work_species",
            "Max_Height_m",
            "n_1.6.2",
            "references_1.6.2",
        ]
    ],
    on="Canonical_Species",
    how="left"
)


resolution_register = resolution_register.rename(
    columns={
        "work_species":
            "Selected_Source_Taxon",

        "n_1.6.2":
            "GIFT_N_Source_Values",

        "references_1.6.2":
            "GIFT_References",
    }
)


print("\nFormal ambiguity-resolution register")
print("-" * 100)

display(
    resolution_register
)


# -----------------------------------------------------------------------------
# 10.13K. Export updated authoritative interim products
# -----------------------------------------------------------------------------

RESOLUTION_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_ambiguity_resolution_register.csv"
)

ACCEPTED_UPDATED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

STATUS_UPDATED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


resolution_register.to_csv(
    RESOLUTION_PATH,
    index=False
)

accepted_updated.to_csv(
    ACCEPTED_UPDATED_PATH,
    index=False
)

status_updated.to_csv(
    STATUS_UPDATED_PATH,
    index=False
)


print("\nUpdated exports")
print("-" * 100)

print(
    f"Resolution register : "
    f"{RESOLUTION_PATH}"
)

print(
    f"Accepted evidence   : "
    f"{ACCEPTED_UPDATED_PATH}"
)

print(
    f"Acquisition ledger  : "
    f"{STATUS_UPDATED_PATH}"
)


# -----------------------------------------------------------------------------
# 10.13L. Methodological interpretation
# -----------------------------------------------------------------------------

print("\nMethodological interpretation")
print("-" * 100)

print(
    """
The two GIFT ambiguity holds have been resolved using explicit
taxonomic verification rather than numerical preference.

Blighia unijigata:
    Accepted search taxon = Blighia unijugata
    Rejected candidate    = Blighia sapida
    Basis                 = orthographic reconciliation

Diospyrous dendo:
    Accepted search taxon = Diospyros dendo
    Rejected candidate    = Diospyros mespiliformis
    Basis                 = orthographic reconciliation

The Omo canonical strings themselves are retained unchanged so that the
established 127-taxon backbone remains stable.

The reconciled external names are stored separately as source taxa and
the provenance remains SPECIES_GIFT.
"""
)


# -----------------------------------------------------------------------------
# 10.13M. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.13 GIFT TAXONOMIC AMBIGUITY RESOLUTION COMPLETE")
print("=" * 100)

print(
    f"Ambiguity taxa reviewed            : "
    f"{len(resolution_decisions)}"
)

print(
    f"Ambiguity taxa resolved            : "
    f"{len(resolved_values)}"
)

print(
    f"Ambiguity holds remaining          : "
    f"{n_ambiguity_updated}"
)

print(
    f"Total species-level accepted       : "
    f"{n_accepted_updated}"
)

print(
    f"Species-level coverage (%)         : "
    f"{updated_coverage:.2f}"
)

print(
    f"Species still requiring evidence   : "
    f"{n_supplementary_updated}"
)

print(
    f"Genus-level targets                : "
    f"{n_genus_updated}"
)

print(
    f"Total unresolved                   : "
    f"{n_supplementary_updated + n_genus_updated}"
)

print(
    "Next step                         : "
    "Section 10.14 — supplementary acquisition strategy "
    "for the remaining 23 species-level and 2 genus-level taxa."
)
```

```python
# =============================================================================
# SECTION 10.15 — QC OF BIEN WHOLE-PLANT-HEIGHT SUPPLEMENTARY MEASUREMENTS
# Validate raw measurements and derive provisional observed species maxima
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.15 — QC OF BIEN WHOLE-PLANT-HEIGHT SUPPLEMENTARY MEASUREMENTS")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.15A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

BIEN_WHOLE_HEIGHT_RAW_PATH = (
    PROJECT_ROOT /
    "data" /
    "external" /
    "BIEN" /
    "omo_max_height_bien_whole_height_raw.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


# -----------------------------------------------------------------------------
# 10.15B. Load inputs
# -----------------------------------------------------------------------------

whole_height_raw = pd.read_csv(
    BIEN_WHOLE_HEIGHT_RAW_PATH,
    low_memory=False
)

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)


print("\nInput state")
print("-" * 100)

print(
    f"Raw BIEN whole-height records : "
    f"{len(whole_height_raw)}"
)

print(
    f"Omo taxa represented          : "
    f"{whole_height_raw['Omo_Canonical_Species'].nunique()}"
)


assert len(whole_height_raw) == 23
assert whole_height_raw["Omo_Canonical_Species"].nunique() == 3


# -----------------------------------------------------------------------------
# 10.15C. Required schema validation
# -----------------------------------------------------------------------------

required_columns = [
    "Omo_Canonical_Species",
    "scrubbed_species_binomial",
    "trait_name",
    "trait_value",
    "unit",
    "method",
    "source_citation",
    "verbatim_scientific_name",
    "name_submitted",
    "name_matched",
    "matched_taxonomic_status",
    "scrubbed_taxonomic_status",
    "scrubbed_taxon_canonical",
    "access",
    "id",
]


missing_columns = [
    col for col in required_columns
    if col not in whole_height_raw.columns
]


if missing_columns:
    raise ValueError(
        "Required BIEN fields missing: "
        + ", ".join(missing_columns)
    )


print("\nSchema validation")
print("-" * 100)

print("Required fields present : YES")


# -----------------------------------------------------------------------------
# 10.15D. Normalize measurement and taxonomy fields
# -----------------------------------------------------------------------------

qc = whole_height_raw.copy()


qc["Trait_Value_Numeric"] = pd.to_numeric(
    qc["trait_value"],
    errors="coerce"
)


qc["Canonical_Match"] = (
    qc["Omo_Canonical_Species"]
    .astype(str)
    .str.strip()
    .str.casefold()
    ==
    qc["scrubbed_species_binomial"]
    .astype(str)
    .str.strip()
    .str.casefold()
)


qc["Trait_Name_OK"] = (
    qc["trait_name"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .eq("whole plant height")
)


qc["Unit_OK"] = (
    qc["unit"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .eq("m")
)


qc["Numeric_OK"] = (
    qc["Trait_Value_Numeric"]
    .notna()
)


qc["Positive_OK"] = (
    qc["Numeric_OK"]
    &
    (qc["Trait_Value_Numeric"] > 0)
)


# Broad screen only.
# We do not automatically reject low values because generic whole-plant
# measurements may include juveniles or smaller individuals.
qc["Broad_Plausibility_OK"] = (
    qc["Numeric_OK"]
    &
    qc["Trait_Value_Numeric"]
    .between(
        0.1,
        100,
        inclusive="both"
    )
)


qc["Matched_Status_OK"] = (
    qc["matched_taxonomic_status"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .eq("accepted")
)


qc["Scrubbed_Status_OK"] = (
    qc["scrubbed_taxonomic_status"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .eq("accepted")
)


qc["Public_Access_OK"] = (
    qc["access"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .eq("public")
)


qc["Citation_Present"] = (
    qc["source_citation"]
    .notna()
    &
    qc["source_citation"]
    .astype(str)
    .str.strip()
    .ne("")
)


qc["BIEN_ID_Present"] = (
    qc["id"]
    .notna()
)


# -----------------------------------------------------------------------------
# 10.15E. Duplicate diagnostics
# -----------------------------------------------------------------------------

qc["Duplicate_BIEN_ID"] = (
    qc["id"].duplicated(keep=False)
)


measurement_signature = (
    qc[
        [
            "Omo_Canonical_Species",
            "trait_value",
            "unit",
            "source_citation",
            "method",
        ]
    ]
    .astype("string")
    .fillna("")
    .agg(" || ".join, axis=1)
)


qc["Duplicate_Measurement_Signature"] = (
    measurement_signature
    .duplicated(keep=False)
)


# -----------------------------------------------------------------------------
# 10.15F. Measurement-level QC status
# -----------------------------------------------------------------------------

qc["Measurement_QC_Status"] = np.where(
    qc["Canonical_Match"]
    &
    qc["Trait_Name_OK"]
    &
    qc["Unit_OK"]
    &
    qc["Numeric_OK"]
    &
    qc["Positive_OK"]
    &
    qc["Broad_Plausibility_OK"]
    &
    qc["Matched_Status_OK"]
    &
    qc["Scrubbed_Status_OK"]
    &
    qc["Public_Access_OK"]
    &
    qc["Citation_Present"]
    &
    qc["BIEN_ID_Present"]
    &
    ~qc["Duplicate_BIEN_ID"]
    &
    ~qc["Duplicate_Measurement_Signature"],

    "PASS",

    "REVIEW"
)


# -----------------------------------------------------------------------------
# 10.15G. QC summary
# -----------------------------------------------------------------------------

print("\nMeasurement QC summary")
print("-" * 100)


qc_summary = pd.Series(
    {
        "Canonical_Match":
            int(qc["Canonical_Match"].sum()),

        "Trait_Name_OK":
            int(qc["Trait_Name_OK"].sum()),

        "Unit_OK":
            int(qc["Unit_OK"].sum()),

        "Numeric_OK":
            int(qc["Numeric_OK"].sum()),

        "Positive_OK":
            int(qc["Positive_OK"].sum()),

        "Broad_Plausibility_OK":
            int(qc["Broad_Plausibility_OK"].sum()),

        "Matched_Status_OK":
            int(qc["Matched_Status_OK"].sum()),

        "Scrubbed_Status_OK":
            int(qc["Scrubbed_Status_OK"].sum()),

        "Public_Access_OK":
            int(qc["Public_Access_OK"].sum()),

        "Citation_Present":
            int(qc["Citation_Present"].sum()),

        "BIEN_ID_Present":
            int(qc["BIEN_ID_Present"].sum()),

        "Duplicate_BIEN_ID":
            int(qc["Duplicate_BIEN_ID"].sum()),

        "Duplicate_Measurement_Signature":
            int(
                qc[
                    "Duplicate_Measurement_Signature"
                ].sum()
            ),
    }
)


print(qc_summary)


print("\nOverall measurement QC status")
print("-" * 100)

print(
    qc[
        "Measurement_QC_Status"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.15H. Detailed raw measurement register
# -----------------------------------------------------------------------------

detail_columns = [
    "Omo_Canonical_Species",
    "scrubbed_species_binomial",
    "trait_value",
    "unit",
    "Trait_Value_Numeric",
    "method",
    "source_citation",
    "verbatim_scientific_name",
    "name_submitted",
    "name_matched",
    "matched_taxonomic_status",
    "scrubbed_taxonomic_status",
    "access",
    "id",
    "Duplicate_BIEN_ID",
    "Duplicate_Measurement_Signature",
    "Measurement_QC_Status",
]


print("\nDetailed BIEN whole-plant-height QC register")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 100,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        qc[
            detail_columns
        ]
        .sort_values(
            [
                "Omo_Canonical_Species",
                "Trait_Value_Numeric",
            ]
        )
    )


# -----------------------------------------------------------------------------
# 10.15I. Records requiring review
# -----------------------------------------------------------------------------

review_records = qc[
    qc[
        "Measurement_QC_Status"
    ].eq("REVIEW")
].copy()


print("\nMeasurements requiring review")
print("-" * 100)

print(
    f"Review records : "
    f"{len(review_records)}"
)


if not review_records.empty:

    with pd.option_context(
        "display.max_rows", 100,
        "display.max_columns", None,
        "display.max_colwidth", 120,
    ):
        display(
            review_records[
                detail_columns
            ]
        )


# -----------------------------------------------------------------------------
# 10.15J. Species-level variability diagnostics
# -----------------------------------------------------------------------------

passed = qc[
    qc[
        "Measurement_QC_Status"
    ].eq("PASS")
].copy()


if not passed.empty:

    species_variability = (
        passed
        .groupby(
            "Omo_Canonical_Species",
            as_index=False
        )
        .agg(
            N_Passed_Measurements=(
                "Trait_Value_Numeric",
                "size"
            ),
            Min_Observed_Height_m=(
                "Trait_Value_Numeric",
                "min"
            ),
            Median_Observed_Height_m=(
                "Trait_Value_Numeric",
                "median"
            ),
            Mean_Observed_Height_m=(
                "Trait_Value_Numeric",
                "mean"
            ),
            Max_Observed_Height_m=(
                "Trait_Value_Numeric",
                "max"
            ),
            SD_Observed_Height_m=(
                "Trait_Value_Numeric",
                "std"
            ),
            N_Source_Citations=(
                "source_citation",
                "nunique"
            ),
            N_Methods=(
                "method",
                lambda x:
                    x.dropna()
                    .astype(str)
                    .str.strip()
                    .replace("", np.nan)
                    .dropna()
                    .nunique()
            ),
        )
    )

else:

    species_variability = pd.DataFrame()


print("\nSpecies-level variability diagnostic")
print("-" * 100)

display(
    species_variability
)


# -----------------------------------------------------------------------------
# 10.15K. Derive provisional observed maximum per species
# -----------------------------------------------------------------------------

if not species_variability.empty:

    provisional_derived_max = (
        species_variability[
            [
                "Omo_Canonical_Species",
                "N_Passed_Measurements",
                "Min_Observed_Height_m",
                "Median_Observed_Height_m",
                "Mean_Observed_Height_m",
                "Max_Observed_Height_m",
                "SD_Observed_Height_m",
                "N_Source_Citations",
                "N_Methods",
            ]
        ]
        .copy()
    )


    provisional_derived_max = (
        provisional_derived_max
        .rename(
            columns={
                "Omo_Canonical_Species":
                    "Canonical_Species",

                "Max_Observed_Height_m":
                    "Provisional_Max_Height_m",
            }
        )
    )


    provisional_derived_max[
        "Proposed_Provenance_Class"
    ] = "SPECIES_BIEN_DERIVED_MAX"


    provisional_derived_max[
        "Proposed_Source"
    ] = "BIEN whole plant height"


    provisional_derived_max[
        "Proposed_Quality_Flag"
    ] = "PROVISIONAL_REQUIRES_ACCEPTANCE"


else:

    provisional_derived_max = pd.DataFrame()


print("\nProvisional species-level observed maxima")
print("-" * 100)

display(
    provisional_derived_max
)


# -----------------------------------------------------------------------------
# 10.15L. Source-citation summary
# -----------------------------------------------------------------------------

citation_summary = (
    qc[
        [
            "Omo_Canonical_Species",
            "source_citation",
        ]
    ]
    .drop_duplicates()
    .sort_values(
        [
            "Omo_Canonical_Species",
            "source_citation",
        ]
    )
)


print("\nSource citations represented")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 100,
    "display.max_colwidth", 150,
):
    display(
        citation_summary
    )


# -----------------------------------------------------------------------------
# 10.15M. Export QC products
# -----------------------------------------------------------------------------

QC_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_qc_register.csv"
)

VARIABILITY_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_species_variability.csv"
)

PROVISIONAL_MAX_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_provisional_derived_max.csv"
)

REVIEW_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_review_records.csv"
)


qc.to_csv(
    QC_REGISTER_PATH,
    index=False
)

species_variability.to_csv(
    VARIABILITY_PATH,
    index=False
)

provisional_derived_max.to_csv(
    PROVISIONAL_MAX_PATH,
    index=False
)

review_records.to_csv(
    REVIEW_PATH,
    index=False
)


print("\nQC exports")
print("-" * 100)

print(
    f"Measurement QC register : "
    f"{QC_REGISTER_PATH}"
)

print(
    f"Species variability     : "
    f"{VARIABILITY_PATH}"
)

print(
    f"Provisional maxima      : "
    f"{PROVISIONAL_MAX_PATH}"
)

print(
    f"Review records          : "
    f"{REVIEW_PATH}"
)


# -----------------------------------------------------------------------------
# 10.15N. Methodological safeguard
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
The maxima calculated in this section are DERIVED OBSERVED MAXIMA from
BIEN's generic 'whole plant height' measurements.

They are not equivalent in provenance to:

    • BIEN 'maximum whole plant height'; or
    • GIFT 'Plant_height_max'.

Accordingly, no value is accepted automatically.

A derived maximum may enter the Omo Maximum Height matrix only after
review confirms that:

    • the underlying measurements passed QC;
    • the species identification is exact;
    • duplicates have been removed;
    • the maximum value is biologically credible; and
    • its lower provenance class is explicitly retained.

No accepted Maximum Height values are modified in this section.
"""
)


# -----------------------------------------------------------------------------
# 10.15O. Final summary
# -----------------------------------------------------------------------------

n_pass = int(
    qc[
        "Measurement_QC_Status"
    ].eq("PASS").sum()
)

n_review = int(
    qc[
        "Measurement_QC_Status"
    ].eq("REVIEW").sum()
)

n_candidate_species = int(
    provisional_derived_max[
        "Canonical_Species"
    ].nunique()
) if not provisional_derived_max.empty else 0


print("=" * 100)
print("SECTION 10.15 BIEN WHOLE-HEIGHT QC COMPLETE")
print("=" * 100)

print(
    f"Raw measurements                    : "
    f"{len(qc)}"
)

print(
    f"QC PASS measurements                : "
    f"{n_pass}"
)

print(
    f"QC REVIEW measurements              : "
    f"{n_review}"
)

print(
    f"Species with provisional maxima     : "
    f"{n_candidate_species}"
)

print(
    "New accepted Maximum Height values  : 0"
)

print(
    "Next step                           : "
    "Review provisional BIEN-derived maxima before acceptance."
)
```

```python
# =============================================================================
# SECTION 10.15A — CORRECTED DUPLICATE INTERPRETATION
# Distinguish repeated biological measurements from true duplicate BIEN records
# =============================================================================

import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.15A — CORRECTED BIEN DUPLICATE INTERPRETATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.15A.1 — Start from the Section 10.15 QC register
# -----------------------------------------------------------------------------

qc_corrected = qc.copy()


# -----------------------------------------------------------------------------
# 10.15A.2 — True duplicate definition
#
# Different BIEN IDs represent distinct database records.
# Equal species/value/source combinations are therefore NOT automatically
# duplicates.
# -----------------------------------------------------------------------------

qc_corrected["True_Duplicate_BIEN_Record"] = (
    qc_corrected["id"].duplicated(keep=False)
)


# Repeated measurements are retained diagnostically.
# They may represent different individuals with the same measured height.

repeat_signature_cols = [
    "Omo_Canonical_Species",
    "trait_value",
    "unit",
    "source_citation",
    "method",
]

repeat_signature = (
    qc_corrected[repeat_signature_cols]
    .astype("string")
    .fillna("")
    .agg(" || ".join, axis=1)
)

qc_corrected["Repeated_Measurement_Value"] = (
    repeat_signature.duplicated(keep=False)
)


# -----------------------------------------------------------------------------
# 10.15A.3 — Recalculate measurement QC
# -----------------------------------------------------------------------------

qc_corrected["Measurement_QC_Status_Corrected"] = np.where(
    qc_corrected["Canonical_Match"]
    &
    qc_corrected["Trait_Name_OK"]
    &
    qc_corrected["Unit_OK"]
    &
    qc_corrected["Numeric_OK"]
    &
    qc_corrected["Positive_OK"]
    &
    qc_corrected["Broad_Plausibility_OK"]
    &
    qc_corrected["Matched_Status_OK"]
    &
    qc_corrected["Scrubbed_Status_OK"]
    &
    qc_corrected["Public_Access_OK"]
    &
    qc_corrected["Citation_Present"]
    &
    qc_corrected["BIEN_ID_Present"]
    &
    ~qc_corrected["True_Duplicate_BIEN_Record"],
    "PASS",
    "REVIEW"
)


# -----------------------------------------------------------------------------
# 10.15A.4 — Corrected QC summary
# -----------------------------------------------------------------------------

print("\nCorrected duplicate diagnostics")
print("-" * 100)

print(
    "Unique BIEN IDs                :",
    qc_corrected["id"].nunique()
)

print(
    "True duplicate BIEN records    :",
    int(
        qc_corrected[
            "True_Duplicate_BIEN_Record"
        ].sum()
    )
)

print(
    "Repeated measurement values    :",
    int(
        qc_corrected[
            "Repeated_Measurement_Value"
        ].sum()
    )
)


print("\nCorrected measurement QC")
print("-" * 100)

print(
    qc_corrected[
        "Measurement_QC_Status_Corrected"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.15A.5 — Recalculate species summaries using all unique valid BIEN records
# -----------------------------------------------------------------------------

passed_corrected = qc_corrected[
    qc_corrected[
        "Measurement_QC_Status_Corrected"
    ].eq("PASS")
].copy()


species_variability_corrected = (
    passed_corrected
    .groupby(
        "Omo_Canonical_Species",
        as_index=False
    )
    .agg(
        N_Valid_Measurements=(
            "Trait_Value_Numeric",
            "size"
        ),
        N_Unique_Height_Values=(
            "Trait_Value_Numeric",
            "nunique"
        ),
        Min_Observed_Height_m=(
            "Trait_Value_Numeric",
            "min"
        ),
        Median_Observed_Height_m=(
            "Trait_Value_Numeric",
            "median"
        ),
        Mean_Observed_Height_m=(
            "Trait_Value_Numeric",
            "mean"
        ),
        Max_Observed_Height_m=(
            "Trait_Value_Numeric",
            "max"
        ),
        SD_Observed_Height_m=(
            "Trait_Value_Numeric",
            "std"
        ),
        N_Source_Citations=(
            "source_citation",
            "nunique"
        ),
        N_BIEN_Records=(
            "id",
            "nunique"
        ),
    )
)


print("\nCorrected species-level variability")
print("-" * 100)

display(
    species_variability_corrected
)


# -----------------------------------------------------------------------------
# 10.15A.6 — Candidate derived maxima
# -----------------------------------------------------------------------------

candidate_maxima_corrected = (
    species_variability_corrected[
        [
            "Omo_Canonical_Species",
            "N_Valid_Measurements",
            "N_Unique_Height_Values",
            "Min_Observed_Height_m",
            "Median_Observed_Height_m",
            "Mean_Observed_Height_m",
            "Max_Observed_Height_m",
            "SD_Observed_Height_m",
            "N_Source_Citations",
            "N_BIEN_Records",
        ]
    ]
    .copy()
    .rename(
        columns={
            "Omo_Canonical_Species":
                "Canonical_Species",

            "Max_Observed_Height_m":
                "Candidate_Derived_Max_Height_m",
        }
    )
)


candidate_maxima_corrected[
    "Candidate_Provenance_Class"
] = "SPECIES_BIEN_DERIVED_MAX"


# Evidence-strength diagnostic only.
# This does NOT constitute acceptance.

candidate_maxima_corrected[
    "Evidence_Strength"
] = np.select(
    [
        (
            candidate_maxima_corrected[
                "N_Valid_Measurements"
            ] >= 5
        ),
        (
            candidate_maxima_corrected[
                "N_Valid_Measurements"
            ].between(2, 4)
        ),
    ],
    [
        "MULTIPLE_MEASUREMENTS",
        "LIMITED_MULTIPLE_MEASUREMENTS",
    ],
    default="SINGLE_MEASUREMENT"
)


candidate_maxima_corrected[
    "Acceptance_Status"
] = "HOLD_FOR_BIOLOGICAL_REVIEW"


print("\nCorrected candidate derived maxima")
print("-" * 100)

display(
    candidate_maxima_corrected
)


# -----------------------------------------------------------------------------
# 10.15A.7 — Explicit single-measurement warning
# -----------------------------------------------------------------------------

single_measurement_taxa = candidate_maxima_corrected[
    candidate_maxima_corrected[
        "N_Valid_Measurements"
    ].eq(1)
].copy()


print("\nSingle-measurement Maximum Height warning")
print("-" * 100)

if single_measurement_taxa.empty:

    print(
        "No candidate taxon is represented by only one measurement."
    )

else:

    display(
        single_measurement_taxa
    )

    print(
        "\nA single observed plant height must NOT be interpreted "
        "automatically as the biological Maximum Height of a species."
    )


# -----------------------------------------------------------------------------
# 10.15A.8 — Export corrected diagnostics
# -----------------------------------------------------------------------------

CORRECTED_QC_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_qc_corrected.csv"
)

CORRECTED_VARIABILITY_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_species_variability_corrected.csv"
)

CORRECTED_CANDIDATES_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_candidate_derived_max_corrected.csv"
)


qc_corrected.to_csv(
    CORRECTED_QC_PATH,
    index=False
)

species_variability_corrected.to_csv(
    CORRECTED_VARIABILITY_PATH,
    index=False
)

candidate_maxima_corrected.to_csv(
    CORRECTED_CANDIDATES_PATH,
    index=False
)


print("\nCorrected diagnostic exports")
print("-" * 100)

print(
    "Corrected QC register :",
    CORRECTED_QC_PATH
)

print(
    "Corrected variability :",
    CORRECTED_VARIABILITY_PATH
)

print(
    "Candidate maxima      :",
    CORRECTED_CANDIDATES_PATH
)


# -----------------------------------------------------------------------------
# 10.15A.9 — Methodological interpretation
# -----------------------------------------------------------------------------

print("\nMethodological interpretation")
print("-" * 100)

print(
    """
Repeated numerical height values are not treated as duplicates when
their BIEN record IDs differ.

This distinction is important because several trees or observations can
legitimately have identical measured heights.

Only true duplicate BIEN records are excluded automatically.

Even after measurement-level QC, a species-level observed maximum from
the generic BIEN 'whole plant height' trait is NOT automatically accepted
as biological Maximum Height.

In particular, taxa represented by a single measurement require strong
supplementary evidence before that observation could be used as a
Maximum Height estimate.

No Maximum Height assignment is made in this corrective section.
"""
)


# -----------------------------------------------------------------------------
# 10.15A.10 — Final summary
# -----------------------------------------------------------------------------

n_corrected_pass = int(
    qc_corrected[
        "Measurement_QC_Status_Corrected"
    ].eq("PASS").sum()
)

n_corrected_review = int(
    qc_corrected[
        "Measurement_QC_Status_Corrected"
    ].eq("REVIEW").sum()
)


print("=" * 100)
print("SECTION 10.15A CORRECTED DUPLICATE INTERPRETATION COMPLETE")
print("=" * 100)

print(
    f"Raw BIEN measurements             : "
    f"{len(qc_corrected)}"
)

print(
    f"Unique BIEN record IDs            : "
    f"{qc_corrected['id'].nunique()}"
)

print(
    f"True duplicate BIEN records       : "
    f"{int(qc_corrected['True_Duplicate_BIEN_Record'].sum())}"
)

print(
    f"Corrected QC PASS measurements    : "
    f"{n_corrected_pass}"
)

print(
    f"Corrected QC REVIEW measurements  : "
    f"{n_corrected_review}"
)

print(
    f"Candidate species maxima          : "
    f"{len(candidate_maxima_corrected)}"
)

print(
    "New Maximum Height values accepted: 0"
)

print(
    "Next step                        : "
    "biological review of the corrected BIEN-derived candidate maxima."
)
```

```python
# =============================================================================
# SECTION 10.16 — BIOLOGICAL REVIEW AND SUPPLEMENTARY SPECIES-LEVEL ACCEPTANCE
# Reject BIEN observed maxima where independent species descriptions
# demonstrate greater attainable Maximum Height
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.16 — BIOLOGICAL REVIEW AND SUPPLEMENTARY SPECIES-LEVEL ACCEPTANCE")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.16A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

CANDIDATE_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_whole_height_candidate_derived_max_corrected.csv"
)

ACCEPTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


# -----------------------------------------------------------------------------
# 10.16B. Load current state
# -----------------------------------------------------------------------------

candidate_bien = pd.read_csv(
    CANDIDATE_PATH,
    low_memory=False
)

accepted_current = pd.read_csv(
    ACCEPTED_PATH,
    low_memory=False
)

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)


print("\nCurrent state")
print("-" * 100)

print(
    f"BIEN-derived candidates : "
    f"{len(candidate_bien)}"
)

print(
    f"Accepted taxa            : "
    f"{accepted_current['Canonical_Species'].nunique()}"
)

print(
    f"Ledger taxa              : "
    f"{len(status_ledger)}"
)


assert len(candidate_bien) == 3

assert (
    accepted_current[
        "Canonical_Species"
    ].nunique()
    == 102
)

assert len(status_ledger) == 127


# -----------------------------------------------------------------------------
# 10.16C. Independent biological-reference evidence
# -----------------------------------------------------------------------------

supplementary_reference_evidence = pd.DataFrame(
    [
        {
            "Canonical_Species":
                "Anthonotha macrophylla",

            "BIEN_Observed_Max_m":
                3.0,

            "Reference_Max_Height_m":
                20.0,

            "Reference_Source":
                "PROTA / Plant Resources of Tropical Africa",

            "Reference_Evidence":
                (
                    "Species described as a shrub or small to "
                    "medium-sized tree up to 20 m tall."
                ),

            "Reference_Type":
                "SPECIES_DESCRIPTION",

            "Reference_URL_Label":
                "Anthonotha macrophylla (PROTA)",

            "Biological_Review_Decision":
                "REJECT_BIEN_DERIVED_MAX_ACCEPT_REFERENCE_MAX",

            "Review_Rationale":
                (
                    "BIEN contains only one 3 m observation, "
                    "which is clearly below the documented "
                    "species-level attainable height of 20 m."
                ),
        },

        {
            "Canonical_Species":
                "Harungana madagascariensis",

            "BIEN_Observed_Max_m":
                20.5,

            "Reference_Max_Height_m":
                25.0,

            "Reference_Source":
                "World Agroforestry Agroforestree Database",

            "Reference_Evidence":
                (
                    "Species described as generally 4–7 m tall, "
                    "sometimes reaching 10–25 m."
                ),

            "Reference_Type":
                "SPECIES_DESCRIPTION",

            "Reference_URL_Label":
                "Harungana madagascariensis Agroforestree profile",

            "Biological_Review_Decision":
                "REJECT_BIEN_DERIVED_MAX_ACCEPT_REFERENCE_MAX",

            "Review_Rationale":
                (
                    "The BIEN maximum of 20.5 m is biologically "
                    "credible but does not reach the documented "
                    "species-level maximum of 25 m."
                ),
        },

        {
            "Canonical_Species":
                "Trema orientalis",

            "BIEN_Observed_Max_m":
                11.0,

            "Reference_Max_Height_m":
                36.0,

            "Reference_Source":
                "PROSEA / Plant Resources of South-East Asia",

            "Reference_Evidence":
                (
                    "Species described as a deciduous shrub to "
                    "large tree reaching up to 36 m tall."
                ),

            "Reference_Type":
                "SPECIES_DESCRIPTION",

            "Reference_URL_Label":
                "Trema orientalis PROSEA profile",

            "Biological_Review_Decision":
                "REJECT_BIEN_DERIVED_MAX_ACCEPT_REFERENCE_MAX",

            "Review_Rationale":
                (
                    "The BIEN observed maximum of 11 m is well "
                    "below published species-level attainable "
                    "height; direct reference evidence reports "
                    "up to 36 m."
                ),
        },
    ]
)


print("\nBiological-review evidence")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 20,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        supplementary_reference_evidence
    )


# -----------------------------------------------------------------------------
# 10.16D. Cross-check BIEN candidate values
# -----------------------------------------------------------------------------

candidate_check = candidate_bien[
    [
        "Canonical_Species",
        "Candidate_Derived_Max_Height_m",
        "N_Valid_Measurements",
        "Evidence_Strength",
    ]
].merge(
    supplementary_reference_evidence[
        [
            "Canonical_Species",
            "BIEN_Observed_Max_m",
            "Reference_Max_Height_m",
        ]
    ],
    on="Canonical_Species",
    how="inner"
)


candidate_check[
    "BIEN_Value_Matches_Review"
] = np.isclose(
    candidate_check[
        "Candidate_Derived_Max_Height_m"
    ],
    candidate_check[
        "BIEN_Observed_Max_m"
    ]
)


candidate_check[
    "Reference_Exceeds_BIEN"
] = (
    candidate_check[
        "Reference_Max_Height_m"
    ]
    >
    candidate_check[
        "Candidate_Derived_Max_Height_m"
    ]
)


print("\nBIEN candidate cross-check")
print("-" * 100)

display(candidate_check)


assert len(candidate_check) == 3

assert (
    candidate_check[
        "BIEN_Value_Matches_Review"
    ].all()
)

assert (
    candidate_check[
        "Reference_Exceeds_BIEN"
    ].all()
)


# -----------------------------------------------------------------------------
# 10.16E. Formal BIEN-derived candidate decisions
# -----------------------------------------------------------------------------

bien_candidate_review = candidate_check.merge(
    supplementary_reference_evidence[
        [
            "Canonical_Species",
            "Reference_Source",
            "Reference_Evidence",
            "Biological_Review_Decision",
            "Review_Rationale",
        ]
    ],
    on="Canonical_Species",
    how="left"
)


bien_candidate_review[
    "Candidate_Accepted"
] = False


bien_candidate_review[
    "Candidate_Final_Status"
] = "REJECTED_AS_MAXIMUM_HEIGHT"


print("\nBIEN-derived candidate decisions")
print("-" * 100)

display(
    bien_candidate_review[
        [
            "Canonical_Species",
            "Candidate_Derived_Max_Height_m",
            "N_Valid_Measurements",
            "Reference_Max_Height_m",
            "Reference_Source",
            "Candidate_Final_Status",
            "Review_Rationale",
        ]
    ]
)


# -----------------------------------------------------------------------------
# 10.16F. Construct accepted supplementary reference values
# -----------------------------------------------------------------------------

reference_accepted_values = pd.DataFrame(
    {
        "Canonical_Species":
            supplementary_reference_evidence[
                "Canonical_Species"
            ],

        "Max_Height_m":
            supplementary_reference_evidence[
                "Reference_Max_Height_m"
            ],

        "Max_Height_Assignment_Level":
            "SPECIES",

        "Max_Height_Provenance_Class":
            "SPECIES_REFERENCE",

        "Max_Height_Source":
            supplementary_reference_evidence[
                "Reference_Source"
            ],

        "Max_Height_Source_Taxon":
            supplementary_reference_evidence[
                "Canonical_Species"
            ],

        "Max_Height_Match_Status":
            "EXACT_CANONICAL_REFERENCE",

        "Max_Height_Unit":
            "m",

        "GIFT_Work_ID":
            pd.NA,

        "GIFT_N_Source_Values":
            pd.NA,

        "GIFT_CV":
            pd.NA,

        "GIFT_References":
            pd.NA,

        "Max_Height_Quality_Flag":
            "QC_PASS_ACCEPTED_REFERENCE",
    }
)


# -----------------------------------------------------------------------------
# 10.16G. Validate new reference assignments
# -----------------------------------------------------------------------------

assert len(reference_accepted_values) == 3

assert (
    reference_accepted_values[
        "Canonical_Species"
    ].nunique()
    == 3
)

assert (
    reference_accepted_values[
        "Max_Height_m"
    ].notna()
    .all()
)

assert (
    reference_accepted_values[
        "Max_Height_m"
    ] > 0
).all()

assert (
    reference_accepted_values[
        "Max_Height_m"
    ].between(
        1,
        100,
        inclusive="both"
    )
).all()


existing_taxa = set(
    accepted_current[
        "Canonical_Species"
    ]
)

new_taxa = set(
    reference_accepted_values[
        "Canonical_Species"
    ]
)


assert existing_taxa.isdisjoint(new_taxa)


print("\nAccepted supplementary reference values")
print("-" * 100)

display(
    reference_accepted_values[
        [
            "Canonical_Species",
            "Max_Height_m",
            "Max_Height_Provenance_Class",
            "Max_Height_Source",
            "Max_Height_Quality_Flag",
        ]
    ]
)


# -----------------------------------------------------------------------------
# 10.16H. Align schemas and append accepted evidence
# -----------------------------------------------------------------------------

for col in accepted_current.columns:

    if col not in reference_accepted_values.columns:
        reference_accepted_values[col] = pd.NA


for col in reference_accepted_values.columns:

    if col not in accepted_current.columns:
        accepted_current[col] = pd.NA


reference_accepted_values = reference_accepted_values[
    accepted_current.columns
]


accepted_updated = pd.concat(
    [
        accepted_current,
        reference_accepted_values,
    ],
    ignore_index=True
)


assert len(accepted_updated) == 105

assert (
    accepted_updated[
        "Canonical_Species"
    ].nunique()
    == 105
)


# -----------------------------------------------------------------------------
# 10.16I. Updated provenance summary
# -----------------------------------------------------------------------------

print("\nUpdated accepted Maximum Height evidence")
print("-" * 100)

print(
    accepted_updated[
        "Max_Height_Provenance_Class"
    ].value_counts(dropna=False)
)


updated_coverage = (
    100 *
    accepted_updated[
        "Canonical_Species"
    ].nunique()
    /
    127
)


print(
    f"\nAccepted taxa  : "
    f"{accepted_updated['Canonical_Species'].nunique()} / 127"
)

print(
    f"Coverage (%)   : "
    f"{updated_coverage:.2f}"
)


# -----------------------------------------------------------------------------
# 10.16J. Update acquisition-status ledger
# -----------------------------------------------------------------------------

status_updated = status_ledger.copy()


new_reference_lookup = (
    reference_accepted_values
    .set_index(
        "Canonical_Species"
    )
)


for canonical in new_taxa:

    mask = status_updated[
        "Canonical_Species"
    ].eq(canonical)

    status_updated.loc[
        mask,
        "Acquisition_Status"
    ] = "SPECIES_LEVEL_ACCEPTED"

    update_fields = [
        "Max_Height_m",
        "Max_Height_Assignment_Level",
        "Max_Height_Provenance_Class",
        "Max_Height_Source",
        "Max_Height_Source_Taxon",
        "Max_Height_Match_Status",
        "Max_Height_Quality_Flag",
    ]

    for field in update_fields:

        status_updated.loc[
            mask,
            field
        ] = new_reference_lookup.loc[
            canonical,
            field
        ]


# -----------------------------------------------------------------------------
# 10.16K. Validate updated residual accounting
# -----------------------------------------------------------------------------

n_accepted = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "SPECIES_LEVEL_ACCEPTED"
    )
    .sum()
)

n_supplementary = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "SUPPLEMENTARY_ACQUISITION_REQUIRED"
    )
    .sum()
)

n_genus = int(
    status_updated[
        "Acquisition_Status"
    ]
    .eq(
        "GENUS_LEVEL_ACQUISITION_REQUIRED"
    )
    .sum()
)


assert n_accepted == 105
assert n_supplementary == 20
assert n_genus == 2

assert (
    n_accepted
    + n_supplementary
    + n_genus
    == 127
)


print("\nUpdated Maximum Height acquisition status")
print("-" * 100)

print(
    status_updated[
        "Acquisition_Status"
    ].value_counts(dropna=False)
)


print("\nResidual accounting")
print("-" * 100)

print(
    f"Species-level accepted           : "
    f"{n_accepted}"
)

print(
    f"Species requiring more evidence  : "
    f"{n_supplementary}"
)

print(
    f"Genus-level targets              : "
    f"{n_genus}"
)

print(
    f"Total unresolved                 : "
    f"{n_supplementary + n_genus}"
)


# -----------------------------------------------------------------------------
# 10.16L. Export review and updated evidence
# -----------------------------------------------------------------------------

BIOLOGICAL_REVIEW_PATH = (
    OUTPUT_DIR /
    "omo_max_height_bien_derived_max_biological_review.csv"
)

REFERENCE_EVIDENCE_PATH = (
    OUTPUT_DIR /
    "omo_max_height_supplementary_reference_evidence.csv"
)

ACCEPTED_UPDATED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

STATUS_UPDATED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


bien_candidate_review.to_csv(
    BIOLOGICAL_REVIEW_PATH,
    index=False
)

supplementary_reference_evidence.to_csv(
    REFERENCE_EVIDENCE_PATH,
    index=False
)

accepted_updated.to_csv(
    ACCEPTED_UPDATED_PATH,
    index=False
)

status_updated.to_csv(
    STATUS_UPDATED_PATH,
    index=False
)


print("\nUpdated exports")
print("-" * 100)

print(
    f"BIEN biological review : "
    f"{BIOLOGICAL_REVIEW_PATH}"
)

print(
    f"Reference evidence      : "
    f"{REFERENCE_EVIDENCE_PATH}"
)

print(
    f"Accepted evidence       : "
    f"{ACCEPTED_UPDATED_PATH}"
)

print(
    f"Acquisition ledger      : "
    f"{STATUS_UPDATED_PATH}"
)


# -----------------------------------------------------------------------------
# 10.16M. Methodological interpretation
# -----------------------------------------------------------------------------

print("\nMethodological interpretation")
print("-" * 100)

print(
    """
The generic BIEN 'whole plant height' measurements are retained as
supporting biological observations but are NOT interpreted as species
Maximum Height when independent species-level descriptions demonstrate
greater attainable height.

Accordingly:

Anthonotha macrophylla
    BIEN observed maximum : 3.0 m
    Accepted reference    : 20.0 m

Harungana madagascariensis
    BIEN observed maximum : 20.5 m
    Accepted reference    : 25.0 m

Trema orientalis
    BIEN observed maximum : 11.0 m
    Accepted reference    : 36.0 m

These assignments retain a distinct SPECIES_REFERENCE provenance class.

No genus-, family-, or global-level estimate has yet been introduced.
"""
)


# -----------------------------------------------------------------------------
# 10.16N. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.16 SUPPLEMENTARY SPECIES-LEVEL REVIEW COMPLETE")
print("=" * 100)

print(
    f"BIEN-derived candidates reviewed    : "
    f"{len(bien_candidate_review)}"
)

print(
    f"BIEN-derived maxima accepted        : 0"
)

print(
    f"Direct reference values accepted    : "
    f"{len(reference_accepted_values)}"
)

print(
    f"Total species-level accepted        : "
    f"{n_accepted}"
)

print(
    f"Species-level coverage (%)          : "
    f"{updated_coverage:.2f}"
)

print(
    f"Species still requiring evidence    : "
    f"{n_supplementary}"
)

print(
    f"Genus-level targets                 : "
    f"{n_genus}"
)

print(
    f"Total unresolved                    : "
    f"{n_supplementary + n_genus}"
)

print(
    "Next step                          : "
    "Section 10.17 — targeted supplementary acquisition "
    "for the remaining 20 species-level taxa."
)
```

```python
# =============================================================================
# SECTION 10.17 — TARGETED SUPPLEMENTARY ACQUISITION REGISTER
# Prepare the remaining 20 species-level and 2 genus-level targets
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.17 — TARGETED SUPPLEMENTARY ACQUISITION REGISTER")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.17A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)

CLEAN_SEARCH_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_clean_supplementary_search_register.csv"
)

ACCEPTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)


# -----------------------------------------------------------------------------
# 10.17B. Load current state
# -----------------------------------------------------------------------------

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)

clean_search_register = pd.read_csv(
    CLEAN_SEARCH_REGISTER_PATH,
    low_memory=False
)

accepted_current = pd.read_csv(
    ACCEPTED_PATH,
    low_memory=False
)


print("\nCurrent Maximum Height state")
print("-" * 100)

print(
    f"Acquisition ledger rows : "
    f"{len(status_ledger)}"
)

print(
    f"Accepted taxa           : "
    f"{accepted_current['Canonical_Species'].nunique()}"
)


assert len(status_ledger) == 127

assert (
    status_ledger[
        "Canonical_Species"
    ].nunique()
    == 127
)

assert (
    accepted_current[
        "Canonical_Species"
    ].nunique()
    == 105
)


# -----------------------------------------------------------------------------
# 10.17C. Extract remaining targets
# -----------------------------------------------------------------------------

species_residual = status_ledger[
    status_ledger[
        "Acquisition_Status"
    ].eq(
        "SUPPLEMENTARY_ACQUISITION_REQUIRED"
    )
].copy()


genus_residual = status_ledger[
    status_ledger[
        "Acquisition_Status"
    ].eq(
        "GENUS_LEVEL_ACQUISITION_REQUIRED"
    )
].copy()


print("\nResidual-state validation")
print("-" * 100)

print(
    f"Species-level residual taxa : "
    f"{len(species_residual)}"
)

print(
    f"Genus-level residual taxa   : "
    f"{len(genus_residual)}"
)


assert len(species_residual) == 20
assert len(genus_residual) == 2


# -----------------------------------------------------------------------------
# 10.17D. Attach cleaned alternative search names
# -----------------------------------------------------------------------------

search_fields = [
    "Canonical_Species",
    "Clean_Alternative_Names",
    "N_Clean_Alternative_Names",
    "Evidence_Files",
    "Search_Strategy",
]


available_search_fields = [
    col for col in search_fields
    if col in clean_search_register.columns
]


residual_register = pd.concat(
    [
        species_residual[
            [
                "Canonical_Species",
                "Max_Height_Target_Level",
                "Acquisition_Status",
            ]
        ],

        genus_residual[
            [
                "Canonical_Species",
                "Max_Height_Target_Level",
                "Acquisition_Status",
            ]
        ],
    ],
    ignore_index=True
)


residual_register = residual_register.merge(
    clean_search_register[
        available_search_fields
    ],
    on="Canonical_Species",
    how="left"
)


# -----------------------------------------------------------------------------
# 10.17E. Construct explicit primary and alternative search terms
# -----------------------------------------------------------------------------

residual_register[
    "Primary_Search_Name"
] = residual_register[
    "Canonical_Species"
]


residual_register[
    "Alternative_Search_Available"
] = residual_register[
    "Clean_Alternative_Names"
].notna()


# -----------------------------------------------------------------------------
# 10.17F. Separate species-level alternatives from genus-only fallbacks
# -----------------------------------------------------------------------------

def classify_alternative_names(value):

    if pd.isna(value):
        return pd.Series(
            {
                "Species_Alternative_Names": pd.NA,
                "Genus_Only_Search_Names": pd.NA,
            }
        )

    species_names = []
    genus_names = []

    for candidate in str(value).split("|"):

        candidate = candidate.strip()

        if not candidate:
            continue

        n_words = len(candidate.split())

        if n_words >= 2:
            species_names.append(candidate)

        elif n_words == 1:
            genus_names.append(candidate)


    species_text = (
        " | ".join(sorted(set(species_names)))
        if species_names
        else pd.NA
    )

    genus_text = (
        " | ".join(sorted(set(genus_names)))
        if genus_names
        else pd.NA
    )

    return pd.Series(
        {
            "Species_Alternative_Names":
                species_text,

            "Genus_Only_Search_Names":
                genus_text,
        }
    )


classified_names = residual_register[
    "Clean_Alternative_Names"
].apply(
    classify_alternative_names
)


residual_register = pd.concat(
    [
        residual_register,
        classified_names,
    ],
    axis=1
)


# -----------------------------------------------------------------------------
# 10.17G. Define targeted search priority
# -----------------------------------------------------------------------------

residual_register[
    "Targeted_Search_Priority"
] = np.select(
    [
        residual_register[
            "Max_Height_Target_Level"
        ].eq(
            "GENUS_LEVEL_OR_UNRESOLVED"
        ),

        residual_register[
            "Species_Alternative_Names"
        ].notna(),

        residual_register[
            "Genus_Only_Search_Names"
        ].notna(),
    ],
    [
        "GENUS_LEVEL_TARGET",
        "SPECIES_CANONICAL_PLUS_ALTERNATIVE",
        "SPECIES_CANONICAL_THEN_GENUS_SUPPORT",
    ],
    default="SPECIES_CANONICAL_ONLY",
)


# -----------------------------------------------------------------------------
# 10.17H. Define supplementary evidence hierarchy
# -----------------------------------------------------------------------------

residual_register[
    "Supplementary_Evidence_Hierarchy"
] = np.where(
    residual_register[
        "Max_Height_Target_Level"
    ].eq(
        "GENUS_LEVEL_OR_UNRESOLVED"
    ),
    (
        "GENUS_REFERENCE > "
        "GENUS_DATABASE > "
        "FAMILY_ESTIMATE > "
        "GLOBAL_FALLBACK"
    ),
    (
        "SPECIES_REFERENCE > "
        "SPECIES_DATABASE > "
        "VERIFIED_SYNONYM_REFERENCE > "
        "GENUS_ESTIMATE > "
        "FAMILY_ESTIMATE > "
        "GLOBAL_FALLBACK"
    )
)


# -----------------------------------------------------------------------------
# 10.17I. Prepare acquisition fields
# -----------------------------------------------------------------------------

residual_register[
    "Candidate_Max_Height_m"
] = np.nan

residual_register[
    "Candidate_Source"
] = pd.NA

residual_register[
    "Candidate_Source_Taxon"
] = pd.NA

residual_register[
    "Candidate_Source_Type"
] = pd.NA

residual_register[
    "Candidate_Reference"
] = pd.NA

residual_register[
    "Candidate_Match_Status"
] = pd.NA

residual_register[
    "Candidate_Provenance_Class"
] = pd.NA

residual_register[
    "Candidate_Quality_Flag"
] = "NOT_ACQUIRED"

residual_register[
    "Candidate_Notes"
] = pd.NA


# -----------------------------------------------------------------------------
# 10.17J. Validation
# -----------------------------------------------------------------------------

assert len(residual_register) == 22

assert (
    residual_register[
        "Canonical_Species"
    ].nunique()
    == 22
)


n_species = int(
    residual_register[
        "Max_Height_Target_Level"
    ]
    .eq(
        "SPECIES_LEVEL"
    )
    .sum()
)

n_genus = int(
    residual_register[
        "Max_Height_Target_Level"
    ]
    .eq(
        "GENUS_LEVEL_OR_UNRESOLVED"
    )
    .sum()
)


assert n_species == 20
assert n_genus == 2


# -----------------------------------------------------------------------------
# 10.17K. Display targeted acquisition register
# -----------------------------------------------------------------------------

print("\nRemaining targeted acquisition register")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 100,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        residual_register[
            [
                "Canonical_Species",
                "Max_Height_Target_Level",
                "Species_Alternative_Names",
                "Genus_Only_Search_Names",
                "Targeted_Search_Priority",
                "Supplementary_Evidence_Hierarchy",
            ]
        ]
    )


# -----------------------------------------------------------------------------
# 10.17L. Search-priority summary
# -----------------------------------------------------------------------------

print("\nTargeted search-priority summary")
print("-" * 100)

print(
    residual_register[
        "Targeted_Search_Priority"
    ].value_counts(dropna=False)
)


print("\nAlternative-name availability")
print("-" * 100)

print(
    "Species with species-level alternatives :",
    int(
        residual_register[
            "Species_Alternative_Names"
        ].notna().sum()
    )
)

print(
    "Species/genus targets with genus-only aids:",
    int(
        residual_register[
            "Genus_Only_Search_Names"
        ].notna().sum()
    )
)


# -----------------------------------------------------------------------------
# 10.17M. Explicit species-level unresolved list
# -----------------------------------------------------------------------------

print("\n20 unresolved species-level taxa")
print("-" * 100)

for i, taxon in enumerate(
    residual_register.loc[
        residual_register[
            "Max_Height_Target_Level"
        ].eq("SPECIES_LEVEL"),
        "Canonical_Species"
    ],
    start=1,
):
    print(f"{i:02d}. {taxon}")


print("\n2 unresolved genus-level targets")
print("-" * 100)

for i, taxon in enumerate(
    residual_register.loc[
        residual_register[
            "Max_Height_Target_Level"
        ].eq(
            "GENUS_LEVEL_OR_UNRESOLVED"
        ),
        "Canonical_Species"
    ],
    start=1,
):
    print(f"{i:02d}. {taxon}")


# -----------------------------------------------------------------------------
# 10.17N. Export targeted acquisition register
# -----------------------------------------------------------------------------

TARGETED_REGISTER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_targeted_supplementary_acquisition_register.csv"
)


residual_register.to_csv(
    TARGETED_REGISTER_PATH,
    index=False
)


print("\nTargeted acquisition export")
print("-" * 100)

print(
    f"Register path : "
    f"{TARGETED_REGISTER_PATH}"
)


# -----------------------------------------------------------------------------
# 10.17O. Methodological safeguard
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
This section prepares the remaining Maximum Height search targets only.

No value is assigned.

Species-level evidence remains preferred over all taxonomic estimates.

For taxa with previously verified alternative names, those names may be
used only as controlled search terms. They do not replace the canonical
Omo taxon.

Genus-level information for a species-level target may support biological
interpretation but must not be labelled as species-level evidence.

Family and global fallbacks remain last-resort options.
"""
)


# -----------------------------------------------------------------------------
# 10.17P. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.17 TARGETED SUPPLEMENTARY REGISTER COMPLETE")
print("=" * 100)

print(
    f"Current accepted Maximum Height taxa : "
    f"{accepted_current['Canonical_Species'].nunique()}"
)

print(
    f"Species-level targets remaining      : "
    f"{n_species}"
)

print(
    f"Genus-level targets remaining        : "
    f"{n_genus}"
)

print(
    f"Total targets remaining              : "
    f"{len(residual_register)}"
)

print(
    "New Maximum Height values assigned   : 0"
)

print(
    "Next step                            : "
    "targeted species-level Maximum Height acquisition "
    "for the remaining 20 taxa."
)
```

```python
# =============================================================================
# SECTION 10.19 — GENUS-LEVEL MAXIMUM HEIGHT DISTRIBUTION AUDIT
# Examine GIFT species-level Maximum Height distributions for the seven
# genera requiring genus-level estimation
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.19 — GENUS-LEVEL MAXIMUM HEIGHT DISTRIBUTION AUDIT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.19A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

GIFT_PATH = (
    PROJECT_ROOT /
    "data" /
    "external" /
    "GIFT" /
    "omo_max_height_gift_species_global_raw.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)

RECLASSIFICATION_PATH = (
    OUTPUT_DIR /
    "omo_max_height_final_evidence_level_reclassification.csv"
)


# -----------------------------------------------------------------------------
# 10.19B. Load inputs
# -----------------------------------------------------------------------------

gift = pd.read_csv(
    GIFT_PATH,
    low_memory=False
)

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)

reclassification = pd.read_csv(
    RECLASSIFICATION_PATH,
    low_memory=False
)


print("\nInput state")
print("-" * 100)

print(
    f"GIFT species records       : "
    f"{len(gift)}"
)

print(
    f"Acquisition ledger rows    : "
    f"{len(status_ledger)}"
)

print(
    f"Reclassification rows      : "
    f"{len(reclassification)}"
)


assert len(gift) == 70367
assert len(status_ledger) == 127


# -----------------------------------------------------------------------------
# 10.19C. Identify all seven genus-level targets
# -----------------------------------------------------------------------------

genus_targets = status_ledger[
    status_ledger[
        "Acquisition_Status"
    ].eq(
        "GENUS_LEVEL_ACQUISITION_REQUIRED"
    )
][
    [
        "Canonical_Species",
        "Max_Height_Target_Level",
        "Acquisition_Status",
    ]
].copy()


assert len(genus_targets) == 7


# Explicit mapping from Omo canonical target to genus used for estimation.
genus_assignment_map = {
    "Albizia auxilary":
        "Albizia",

    "Amphimas tetracoides":
        "Amphimas",

    "Diospyros undebunda":
        "Diospyros",

    "Drypetes":
        "Drypetes",

    "Lonchocarpus cilliata":
        "Lonchocarpus",

    "Cola species":
        "Cola",

    "Drypetes species":
        "Drypetes",
}


genus_targets[
    "Estimation_Genus"
] = genus_targets[
    "Canonical_Species"
].map(
    genus_assignment_map
)


assert genus_targets[
    "Estimation_Genus"
].notna().all()


print("\nSeven genus-level targets")
print("-" * 100)

display(
    genus_targets
)


# -----------------------------------------------------------------------------
# 10.19D. Prepare GIFT Maximum Height table
# -----------------------------------------------------------------------------

required_gift_columns = [
    "work_ID",
    "work_species",
    "trait_value_1.6.2",
    "cv_1.6.2",
    "n_1.6.2",
    "references_1.6.2",
]


missing_columns = [
    col for col in required_gift_columns
    if col not in gift.columns
]


if missing_columns:
    raise ValueError(
        "Required GIFT fields missing: "
        + ", ".join(missing_columns)
    )


gift = gift.copy()


gift[
    "Max_Height_m"
] = pd.to_numeric(
    gift[
        "trait_value_1.6.2"
    ],
    errors="coerce"
)


gift[
    "GIFT_Genus"
] = (
    gift[
        "work_species"
    ]
    .astype("string")
    .str.strip()
    .str.split()
    .str[0]
)


gift_valid = gift[
    gift[
        "Max_Height_m"
    ].notna()
    &
    (
        gift[
            "Max_Height_m"
        ] > 0
    )
    &
    gift[
        "GIFT_Genus"
    ].notna()
].copy()


print("\nGIFT preparation")
print("-" * 100)

print(
    f"Valid Maximum Height rows : "
    f"{len(gift_valid)}"
)

print(
    f"Unique GIFT genera        : "
    f"{gift_valid['GIFT_Genus'].nunique()}"
)


# -----------------------------------------------------------------------------
# 10.19E. Extract target-genus species records
# -----------------------------------------------------------------------------

target_genera = sorted(
    set(
        genus_targets[
            "Estimation_Genus"
        ]
    )
)


target_genus_records = gift_valid[
    gift_valid[
        "GIFT_Genus"
    ].isin(
        target_genera
    )
].copy()


print("\nTarget genera represented in GIFT")
print("-" * 100)

for genus in target_genera:

    n_species = int(
        target_genus_records.loc[
            target_genus_records[
                "GIFT_Genus"
            ].eq(genus),
            "work_species"
        ].nunique()
    )

    print(
        f"{genus:<15} : "
        f"{n_species} species"
    )


# -----------------------------------------------------------------------------
# 10.19F. Compute genus-level distributions
# -----------------------------------------------------------------------------

genus_summary_records = []


for genus in target_genera:

    subset = target_genus_records[
        target_genus_records[
            "GIFT_Genus"
        ].eq(genus)
    ].copy()


    values = subset[
        "Max_Height_m"
    ].dropna()


    n_species = subset[
        "work_species"
    ].nunique()


    if len(values) == 0:

        genus_summary_records.append(
            {
                "Genus":
                    genus,

                "N_GIFT_Species":
                    0,

                "Min_Species_Max_m":
                    np.nan,

                "Q1_Species_Max_m":
                    np.nan,

                "Median_Species_Max_m":
                    np.nan,

                "Mean_Species_Max_m":
                    np.nan,

                "Q3_Species_Max_m":
                    np.nan,

                "Max_Species_Max_m":
                    np.nan,

                "SD_Species_Max_m":
                    np.nan,
            }
        )

        continue


    genus_summary_records.append(
        {
            "Genus":
                genus,

            "N_GIFT_Species":
                int(n_species),

            "Min_Species_Max_m":
                float(values.min()),

            "Q1_Species_Max_m":
                float(values.quantile(0.25)),

            "Median_Species_Max_m":
                float(values.median()),

            "Mean_Species_Max_m":
                float(values.mean()),

            "Q3_Species_Max_m":
                float(values.quantile(0.75)),

            "Max_Species_Max_m":
                float(values.max()),

            "SD_Species_Max_m":
                float(values.std())
                if len(values) > 1
                else np.nan,
        }
    )


genus_summary = pd.DataFrame(
    genus_summary_records
)


print("\nGenus-level Maximum Height distributions")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 50,
    "display.max_columns", None,
):
    display(
        genus_summary
    )


# -----------------------------------------------------------------------------
# 10.19G. Inspect all species contributing to each genus
# -----------------------------------------------------------------------------

print("\nSpecies contributing to genus-level distributions")
print("-" * 100)


for genus in target_genera:

    subset = (
        target_genus_records[
            target_genus_records[
                "GIFT_Genus"
            ].eq(genus)
        ]
        [
            [
                "work_species",
                "Max_Height_m",
                "n_1.6.2",
                "cv_1.6.2",
                "references_1.6.2",
            ]
        ]
        .sort_values(
            "Max_Height_m",
            ascending=False
        )
        .reset_index(drop=True)
    )


    print(
        f"\n{genus}"
    )

    print("-" * 100)

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 100,
    ):
        display(
            subset
        )


# -----------------------------------------------------------------------------
# 10.19H. Build target-specific genus estimation audit
# -----------------------------------------------------------------------------

genus_estimation_audit = genus_targets.merge(
    genus_summary,
    left_on="Estimation_Genus",
    right_on="Genus",
    how="left"
)


# Proposed estimator is recorded for audit only.
# No value is assigned yet.
genus_estimation_audit[
    "Proposed_Estimator"
] = "MEDIAN_OF_SPECIES_MAXIMA"


genus_estimation_audit[
    "Proposed_Genus_Estimate_m"
] = genus_estimation_audit[
    "Median_Species_Max_m"
]


# Evidence-strength classification
genus_estimation_audit[
    "Genus_Evidence_Strength"
] = np.select(
    [
        genus_estimation_audit[
            "N_GIFT_Species"
        ].ge(10),

        genus_estimation_audit[
            "N_GIFT_Species"
        ].between(
            5,
            9,
            inclusive="both"
        ),

        genus_estimation_audit[
            "N_GIFT_Species"
        ].between(
            2,
            4,
            inclusive="both"
        ),
    ],
    [
        "STRONG_GENUS_SAMPLE",
        "MODERATE_GENUS_SAMPLE",
        "LIMITED_GENUS_SAMPLE",
    ],
    default="VERY_LIMITED_GENUS_SAMPLE",
)


genus_estimation_audit[
    "Acceptance_Status"
] = "HOLD_PENDING_GENUS_REVIEW"


print("\nTarget-specific genus estimation audit")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 50,
    "display.max_columns", None,
    "display.max_colwidth", 100,
):
    display(
        genus_estimation_audit
    )


# -----------------------------------------------------------------------------
# 10.19I. Diagnostic for the duplicated Drypetes genus
# -----------------------------------------------------------------------------

drypetes_targets = genus_estimation_audit[
    genus_estimation_audit[
        "Estimation_Genus"
    ].eq(
        "Drypetes"
    )
].copy()


print("\nDrypetes duplicated-target diagnostic")
print("-" * 100)

print(
    f"Omo targets using Drypetes genus estimate : "
    f"{len(drypetes_targets)}"
)

display(
    drypetes_targets[
        [
            "Canonical_Species",
            "Estimation_Genus",
            "N_GIFT_Species",
            "Median_Species_Max_m",
            "Genus_Evidence_Strength",
        ]
    ]
)


assert len(drypetes_targets) == 2


# -----------------------------------------------------------------------------
# 10.19J. Genus-estimator methodological comparison
# -----------------------------------------------------------------------------

genus_estimation_audit[
    "Median_vs_Max_Ratio"
] = (
    genus_estimation_audit[
        "Median_Species_Max_m"
    ]
    /
    genus_estimation_audit[
        "Max_Species_Max_m"
    ]
)


genus_estimation_audit[
    "Mean_vs_Median_Difference_m"
] = (
    genus_estimation_audit[
        "Mean_Species_Max_m"
    ]
    -
    genus_estimation_audit[
        "Median_Species_Max_m"
    ]
)


print("\nEstimator comparison diagnostic")
print("-" * 100)

display(
    genus_estimation_audit[
        [
            "Canonical_Species",
            "Estimation_Genus",
            "N_GIFT_Species",
            "Median_Species_Max_m",
            "Mean_Species_Max_m",
            "Max_Species_Max_m",
            "Median_vs_Max_Ratio",
            "Mean_vs_Median_Difference_m",
        ]
    ]
)


# -----------------------------------------------------------------------------
# 10.19K. Export genus audit products
# -----------------------------------------------------------------------------

GENUS_RECORDS_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_target_genus_species_records.csv"
)

GENUS_SUMMARY_PATH = (
    OUTPUT_DIR /
    "omo_max_height_gift_target_genus_summary.csv"
)

GENUS_AUDIT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_genus_estimation_audit.csv"
)


target_genus_records.to_csv(
    GENUS_RECORDS_PATH,
    index=False
)

genus_summary.to_csv(
    GENUS_SUMMARY_PATH,
    index=False
)

genus_estimation_audit.to_csv(
    GENUS_AUDIT_PATH,
    index=False
)


print("\nGenus audit exports")
print("-" * 100)

print(
    f"Target genus records : "
    f"{GENUS_RECORDS_PATH}"
)

print(
    f"Genus summary        : "
    f"{GENUS_SUMMARY_PATH}"
)

print(
    f"Estimation audit     : "
    f"{GENUS_AUDIT_PATH}"
)


# -----------------------------------------------------------------------------
# 10.19L. Methodological safeguard
# -----------------------------------------------------------------------------

print("\nMethodological safeguard")
print("-" * 100)

print(
    """
No genus-level Maximum Height estimate is assigned in this section.

The GIFT species-level maximum-height distributions are examined first
because genus estimates must reflect the central tendency of species
maximum heights rather than simply borrowing the tallest species in a
genus.

The provisional estimator is the MEDIAN of species-level maximum
heights within the genus.

This is preferred over the genus maximum because assigning the tallest
known species' height to an unresolved taxon would systematically
inflate Maximum Height.

The final estimator will be accepted only after reviewing:
    • number of GIFT species represented;
    • distribution spread;
    • extreme values;
    • biological plausibility; and
    • whether the genus sample is sufficiently informative.

No Maximum Height values are assigned here.
"""
)


# -----------------------------------------------------------------------------
# 10.19M. Final summary
# -----------------------------------------------------------------------------

n_genera_with_data = int(
    genus_summary[
        "N_GIFT_Species"
    ].gt(0).sum()
)


print("=" * 100)
print("SECTION 10.19 GENUS-LEVEL DISTRIBUTION AUDIT COMPLETE")
print("=" * 100)

print(
    f"Genus-level Omo targets         : "
    f"{len(genus_targets)}"
)

print(
    f"Unique genera required          : "
    f"{len(target_genera)}"
)

print(
    f"Genera with GIFT evidence       : "
    f"{n_genera_with_data}"
)

print(
    f"Genera without GIFT evidence    : "
    f"{len(target_genera) - n_genera_with_data}"
)

print(
    "New Maximum Height values      : 0"
)

print(
    "Next step                      : "
    "Section 10.20 — review genus distributions "
    "and approve defensible genus estimates."
)
```

```python
# =============================================================================
# SECTION 10.20 — APPROVAL OF DEFENSIBLE GENUS-LEVEL MAXIMUM HEIGHT ESTIMATES
# Accept median species-level maxima for sufficiently represented GIFT genera
# while retaining Amphimas on hold because only one species is represented
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.20 — APPROVAL OF DEFENSIBLE GENUS-LEVEL MAXIMUM HEIGHT ESTIMATES")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.20A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

GENUS_AUDIT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_genus_estimation_audit.csv"
)

ACCEPTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


# -----------------------------------------------------------------------------
# 10.20B. Load Section 10.19 audit and current state
# -----------------------------------------------------------------------------

genus_audit = pd.read_csv(
    GENUS_AUDIT_PATH,
    low_memory=False
)

accepted_current = pd.read_csv(
    ACCEPTED_PATH,
    low_memory=False
)

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)


print("\nCurrent state")
print("-" * 100)

print(
    f"Genus-level audit rows : "
    f"{len(genus_audit)}"
)

print(
    f"Accepted evidence taxa : "
    f"{accepted_current['Canonical_Species'].nunique()}"
)

print(
    f"Ledger rows            : "
    f"{len(status_ledger)}"
)


assert len(genus_audit) == 7

assert (
    accepted_current[
        "Canonical_Species"
    ].nunique()
    == 119
)

assert len(status_ledger) == 127


# -----------------------------------------------------------------------------
# 10.20C. Define genus-estimate acceptance rule
# -----------------------------------------------------------------------------

# Conservative rule:
#
# ACCEPT:
#     genus represented by at least 5 species in GIFT.
#
# HOLD:
#     genus represented by fewer than 5 species.
#
# The genus-level estimator is the median of species-level Maximum Height
# values, not the genus maximum.

genus_audit = genus_audit.copy()


genus_audit[
    "Genus_Estimate_Decision"
] = np.where(
    genus_audit[
        "N_GIFT_Species"
    ].ge(5),
    "ACCEPT_GENUS_MEDIAN",
    "HOLD_INSUFFICIENT_GENUS_SAMPLE"
)


genus_audit[
    "Approved_Genus_Estimate_m"
] = np.where(
    genus_audit[
        "Genus_Estimate_Decision"
    ].eq(
        "ACCEPT_GENUS_MEDIAN"
    ),
    genus_audit[
        "Median_Species_Max_m"
    ],
    np.nan
)


genus_audit[
    "Estimator_Rationale"
] = np.where(
    genus_audit[
        "Genus_Estimate_Decision"
    ].eq(
        "ACCEPT_GENUS_MEDIAN"
    ),
    (
        "Median of GIFT species-level maximum heights; "
        "genus represented by at least five species."
    ),
    (
        "Genus represented by fewer than five GIFT species; "
        "insufficient evidence for a robust genus-level estimate."
    )
)


# -----------------------------------------------------------------------------
# 10.20D. Display decision audit
# -----------------------------------------------------------------------------

print("\nGenus-level estimate decisions")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 50,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        genus_audit[
            [
                "Canonical_Species",
                "Estimation_Genus",
                "N_GIFT_Species",
                "Min_Species_Max_m",
                "Q1_Species_Max_m",
                "Median_Species_Max_m",
                "Mean_Species_Max_m",
                "Q3_Species_Max_m",
                "Max_Species_Max_m",
                "Genus_Evidence_Strength",
                "Genus_Estimate_Decision",
                "Approved_Genus_Estimate_m",
                "Estimator_Rationale",
            ]
        ]
    )


# -----------------------------------------------------------------------------
# 10.20E. Separate accepted and held targets
# -----------------------------------------------------------------------------

accepted_genus_audit = genus_audit[
    genus_audit[
        "Genus_Estimate_Decision"
    ].eq(
        "ACCEPT_GENUS_MEDIAN"
    )
].copy()


held_genus_audit = genus_audit[
    genus_audit[
        "Genus_Estimate_Decision"
    ].eq(
        "HOLD_INSUFFICIENT_GENUS_SAMPLE"
    )
].copy()


print("\nDecision summary")
print("-" * 100)

print(
    f"Genus-level targets accepted : "
    f"{len(accepted_genus_audit)}"
)

print(
    f"Genus-level targets held     : "
    f"{len(held_genus_audit)}"
)


assert len(accepted_genus_audit) == 6
assert len(held_genus_audit) == 1


# -----------------------------------------------------------------------------
# 10.20F. Verify expected held target
# -----------------------------------------------------------------------------

assert (
    held_genus_audit[
        "Canonical_Species"
    ].tolist()
    ==
    ["Amphimas tetracoides"]
)


print("\nHeld genus-level target")
print("-" * 100)

display(
    held_genus_audit[
        [
            "Canonical_Species",
            "Estimation_Genus",
            "N_GIFT_Species",
            "Median_Species_Max_m",
            "Genus_Estimate_Decision",
            "Estimator_Rationale",
        ]
    ]
)


# -----------------------------------------------------------------------------
# 10.20G. Construct accepted genus-level values
# -----------------------------------------------------------------------------

genus_accepted_values = pd.DataFrame(
    {
        "Canonical_Species":
            accepted_genus_audit[
                "Canonical_Species"
            ],

        "Max_Height_m":
            accepted_genus_audit[
                "Approved_Genus_Estimate_m"
            ],

        "Max_Height_Assignment_Level":
            "GENUS",

        "Max_Height_Provenance_Class":
            "GENUS_GIFT_MEDIAN",

        "Max_Height_Source":
            "GIFT genus distribution",

        "Max_Height_Source_Taxon":
            accepted_genus_audit[
                "Estimation_Genus"
            ],

        "Max_Height_Match_Status":
            "GENUS_LEVEL_ESTIMATE",

        "Max_Height_Unit":
            "m",

        "GIFT_Work_ID":
            pd.NA,

        "GIFT_N_Source_Values":
            accepted_genus_audit[
                "N_GIFT_Species"
            ],

        "GIFT_CV":
            pd.NA,

        "GIFT_References":
            pd.NA,

        "Max_Height_Quality_Flag":
            "GENUS_ESTIMATE_ACCEPTED",
    }
)


# -----------------------------------------------------------------------------
# 10.20H. Validate accepted genus estimates
# -----------------------------------------------------------------------------

assert len(genus_accepted_values) == 6

assert (
    genus_accepted_values[
        "Canonical_Species"
    ].nunique()
    == 6
)

assert (
    genus_accepted_values[
        "Max_Height_m"
    ].notna()
    .all()
)

assert (
    genus_accepted_values[
        "Max_Height_m"
    ] > 0
).all()

assert (
    genus_accepted_values[
        "Max_Height_m"
    ].between(
        1,
        100,
        inclusive="both"
    )
).all()


print("\nAccepted genus-level Maximum Height estimates")
print("-" * 100)

display(
    genus_accepted_values[
        [
            "Canonical_Species",
            "Max_Height_Source_Taxon",
            "Max_Height_m",
            "GIFT_N_Source_Values",
            "Max_Height_Provenance_Class",
            "Max_Height_Quality_Flag",
        ]
    ]
)


# -----------------------------------------------------------------------------
# 10.20I. Expected-value safeguard
# -----------------------------------------------------------------------------

expected_genus_values = {
    "Albizia auxilary":
        18.000000,

    "Cola species":
        12.192148,

    "Diospyros undebunda":
        13.358083,

    "Drypetes":
        15.000000,

    "Drypetes species":
        15.000000,

    "Lonchocarpus cilliata":
        15.000000,
}


for canonical, expected_value in expected_genus_values.items():

    observed = float(
        genus_accepted_values.loc[
            genus_accepted_values[
                "Canonical_Species"
            ].eq(canonical),
            "Max_Height_m"
        ].iloc[0]
    )

    assert np.isclose(
        observed,
        expected_value,
        rtol=1e-5,
        atol=1e-5
    )


print("\nExpected-value safeguard")
print("-" * 100)

print(
    "All six approved genus estimates match "
    "Section 10.19 medians : PASS"
)


# -----------------------------------------------------------------------------
# 10.20J. Ensure no overlap with current accepted evidence
# -----------------------------------------------------------------------------

existing_taxa = set(
    accepted_current[
        "Canonical_Species"
    ]
)

new_genus_taxa = set(
    genus_accepted_values[
        "Canonical_Species"
    ]
)


assert existing_taxa.isdisjoint(
    new_genus_taxa
)


print("\nAccepted-evidence overlap")
print("-" * 100)

print(
    "New genus estimates overlap accepted taxa : NO"
)


# -----------------------------------------------------------------------------
# 10.20K. Align schemas and append
# -----------------------------------------------------------------------------

all_columns = list(
    dict.fromkeys(
        list(
            accepted_current.columns
        )
        +
        list(
            genus_accepted_values.columns
        )
    )
)


accepted_aligned = accepted_current.reindex(
    columns=all_columns
)

genus_aligned = genus_accepted_values.reindex(
    columns=all_columns
)


accepted_updated = pd.concat(
    [
        accepted_aligned,
        genus_aligned,
    ],
    ignore_index=True
)


n_total_values = int(
    accepted_updated[
        "Canonical_Species"
    ].nunique()
)


assert n_total_values == 125


# -----------------------------------------------------------------------------
# 10.20L. Update acquisition-status ledger
# -----------------------------------------------------------------------------

status_updated = status_ledger.copy()


genus_lookup = (
    genus_accepted_values
    .set_index(
        "Canonical_Species"
    )
)


fields_to_update = [
    "Max_Height_m",
    "Max_Height_Assignment_Level",
    "Max_Height_Provenance_Class",
    "Max_Height_Source",
    "Max_Height_Source_Taxon",
    "Max_Height_Match_Status",
    "Max_Height_Quality_Flag",
]


for canonical in new_genus_taxa:

    mask = status_updated[
        "Canonical_Species"
    ].eq(canonical)

    status_updated.loc[
        mask,
        "Acquisition_Status"
    ] = "GENUS_LEVEL_ACCEPTED"

    for field in fields_to_update:

        status_updated.loc[
            mask,
            field
        ] = genus_lookup.loc[
            canonical,
            field
        ]


# Explicitly retain Amphimas on hold.
status_updated.loc[
    status_updated[
        "Canonical_Species"
    ].eq(
        "Amphimas tetracoides"
    ),
    "Acquisition_Status"
] = "GENUS_LEVEL_REVIEW_REQUIRED"


# -----------------------------------------------------------------------------
# 10.20M. Validate new acquisition accounting
# -----------------------------------------------------------------------------

n_species_accepted = int(
    status_updated[
        "Acquisition_Status"
    ].eq(
        "SPECIES_LEVEL_ACCEPTED"
    ).sum()
)

n_genus_accepted = int(
    status_updated[
        "Acquisition_Status"
    ].eq(
        "GENUS_LEVEL_ACCEPTED"
    ).sum()
)

n_genus_review = int(
    status_updated[
        "Acquisition_Status"
    ].eq(
        "GENUS_LEVEL_REVIEW_REQUIRED"
    ).sum()
)

n_species_complex = int(
    status_updated[
        "Acquisition_Status"
    ].eq(
        "SPECIES_COMPLEX_ACQUISITION_REQUIRED"
    ).sum()
)


assert n_species_accepted == 119
assert n_genus_accepted == 6
assert n_genus_review == 1
assert n_species_complex == 1


assert (
    n_species_accepted
    + n_genus_accepted
    + n_genus_review
    + n_species_complex
    == 127
)


print("\nUpdated acquisition status")
print("-" * 100)

print(
    status_updated[
        "Acquisition_Status"
    ].value_counts(dropna=False)
)


numeric_coverage = (
    100 *
    (
        n_species_accepted
        + n_genus_accepted
    )
    /
    127
)


print("\nUpdated coverage")
print("-" * 100)

print(
    f"Species-level accepted : "
    f"{n_species_accepted}"
)

print(
    f"Genus-level accepted   : "
    f"{n_genus_accepted}"
)

print(
    f"Numeric taxa resolved  : "
    f"{n_species_accepted + n_genus_accepted} / 127"
)

print(
    f"Numeric coverage (%)   : "
    f"{numeric_coverage:.2f}"
)

print(
    f"Genus review remaining : "
    f"{n_genus_review}"
)

print(
    f"Species-complex remain : "
    f"{n_species_complex}"
)


# -----------------------------------------------------------------------------
# 10.20N. Provenance summary
# -----------------------------------------------------------------------------

print("\nCurrent Maximum Height provenance summary")
print("-" * 100)

print(
    accepted_updated[
        "Max_Height_Provenance_Class"
    ].value_counts(dropna=False)
)


# -----------------------------------------------------------------------------
# 10.20O. Export approved genus estimates
# -----------------------------------------------------------------------------

GENUS_DECISION_PATH = (
    OUTPUT_DIR /
    "omo_max_height_genus_estimate_decisions.csv"
)

GENUS_ACCEPTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_genus_estimates_accepted.csv"
)

ACCEPTED_UPDATED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

STATUS_UPDATED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)


genus_audit.to_csv(
    GENUS_DECISION_PATH,
    index=False
)

genus_accepted_values.to_csv(
    GENUS_ACCEPTED_PATH,
    index=False
)

accepted_updated.to_csv(
    ACCEPTED_UPDATED_PATH,
    index=False
)

status_updated.to_csv(
    STATUS_UPDATED_PATH,
    index=False
)


print("\nExports")
print("-" * 100)

print(
    f"Genus decisions  : "
    f"{GENUS_DECISION_PATH}"
)

print(
    f"Accepted estimates: "
    f"{GENUS_ACCEPTED_PATH}"
)

print(
    f"Accepted evidence : "
    f"{ACCEPTED_UPDATED_PATH}"
)

print(
    f"Acquisition ledger: "
    f"{STATUS_UPDATED_PATH}"
)


# -----------------------------------------------------------------------------
# 10.20P. Methodological interpretation
# -----------------------------------------------------------------------------

print("\nMethodological interpretation")
print("-" * 100)

print(
    """
Six genus-level Maximum Height estimates are accepted using the median
of species-level Maximum Height values represented in GIFT.

The median is preferred because it reflects the central tendency of
species maximum stature within the genus and is substantially less
sensitive than the genus maximum to exceptionally tall species.

Accepted genera have at least five species represented in GIFT.

Amphimas tetracoides remains unresolved because GIFT contains only one
Amphimas species. Using that single species as a genus estimate would
effectively substitute one species for another and would not constitute
a genuine genus-level estimate.

The remaining unresolved cases after this section are therefore:

    1. Amphimas tetracoides
       -> genus-level review required

    2. Musa sapiensis
       -> cultivated species-complex acquisition required

No family-level or global fallback has yet been introduced.
"""
)


# -----------------------------------------------------------------------------
# 10.20Q. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.20 GENUS-LEVEL ESTIMATE APPROVAL COMPLETE")
print("=" * 100)

print(
    f"Genus-level targets reviewed      : "
    f"{len(genus_audit)}"
)

print(
    f"Genus estimates accepted          : "
    f"{n_genus_accepted}"
)

print(
    f"Genus estimates held              : "
    f"{n_genus_review}"
)

print(
    f"Species-level accepted            : "
    f"{n_species_accepted}"
)

print(
    f"Total numeric taxa currently      : "
    f"{n_species_accepted + n_genus_accepted}"
)

print(
    f"Current numeric coverage (%)      : "
    f"{numeric_coverage:.2f}"
)

print(
    f"Unresolved targets                : "
    f"{n_genus_review + n_species_complex}"
)

print(
    "Next step                        : "
    "Section 10.21 — final resolution of Amphimas tetracoides "
    "and Musa sapiensis."
)
```

```python
# =============================================================================
# SECTION 10.22 — FINAL MAXIMUM HEIGHT CONSOLIDATION,
# PROVENANCE SUMMARY AND VALIDATION
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 10.22 — FINAL MAXIMUM HEIGHT CONSOLIDATION, PROVENANCE SUMMARY AND VALIDATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 10.22A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

CURRENT_ACCEPTED_PATH = (
    OUTPUT_DIR /
    "omo_max_height_species_evidence_accepted_current.csv"
)

STATUS_LEDGER_PATH = (
    OUTPUT_DIR /
    "omo_max_height_acquisition_status_ledger.csv"
)

TARGET_MANIFEST_PATH = (
    OUTPUT_DIR /
    "omo_max_height_target_manifest.csv"
)


FINAL_HEIGHT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_final.csv"
)

PROVENANCE_SUMMARY_PATH = (
    OUTPUT_DIR /
    "omo_max_height_provenance_summary.csv"
)

ASSIGNMENT_SUMMARY_PATH = (
    OUTPUT_DIR /
    "omo_max_height_assignment_level_summary.csv"
)

VALIDATION_REPORT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_final_validation_report.csv"
)


# -----------------------------------------------------------------------------
# 10.22B. Load completed Maximum Height products
# -----------------------------------------------------------------------------

height_current = pd.read_csv(
    CURRENT_ACCEPTED_PATH,
    low_memory=False
)

status_ledger = pd.read_csv(
    STATUS_LEDGER_PATH,
    low_memory=False
)

target_manifest = pd.read_csv(
    TARGET_MANIFEST_PATH,
    low_memory=False
)


print("\nInput state")
print("-" * 100)

print(
    f"Completed evidence rows : "
    f"{len(height_current)}"
)

print(
    f"Unique evidence taxa    : "
    f"{height_current['Canonical_Species'].nunique()}"
)

print(
    f"Acquisition ledger rows : "
    f"{len(status_ledger)}"
)

print(
    f"Target manifest rows    : "
    f"{len(target_manifest)}"
)


# -----------------------------------------------------------------------------
# 10.22C. Canonical backbone validation
# -----------------------------------------------------------------------------

assert len(height_current) == 127

assert (
    height_current[
        "Canonical_Species"
    ].nunique()
    == 127
)

assert len(status_ledger) == 127

assert (
    status_ledger[
        "Canonical_Species"
    ].nunique()
    == 127
)

assert len(target_manifest) == 127

assert (
    target_manifest[
        "Canonical_Species"
    ].nunique()
    == 127
)


height_taxa = set(
    height_current[
        "Canonical_Species"
    ]
)

ledger_taxa = set(
    status_ledger[
        "Canonical_Species"
    ]
)

manifest_taxa = set(
    target_manifest[
        "Canonical_Species"
    ]
)


assert height_taxa == ledger_taxa
assert height_taxa == manifest_taxa


print("\nCanonical backbone validation")
print("-" * 100)

print("Rows                         : 127")
print("Unique taxa                  : 127")
print("Evidence = acquisition ledger: YES")
print("Evidence = target manifest   : YES")


# -----------------------------------------------------------------------------
# 10.22D. Standardize final numeric field
# -----------------------------------------------------------------------------

height_final = height_current.copy()


height_final[
    "Max_Height_m"
] = pd.to_numeric(
    height_final[
        "Max_Height_m"
    ],
    errors="coerce"
)


n_numeric = int(
    height_final[
        "Max_Height_m"
    ].notna().sum()
)

n_missing = int(
    height_final[
        "Max_Height_m"
    ].isna().sum()
)

n_nonpositive = int(
    (
        height_final[
            "Max_Height_m"
        ] <= 0
    ).sum()
)


assert n_numeric == 127
assert n_missing == 0
assert n_nonpositive == 0


print("\nNumeric validation")
print("-" * 100)

print(
    f"Numeric values       : "
    f"{n_numeric}"
)

print(
    f"Missing values       : "
    f"{n_missing}"
)

print(
    f"Zero/negative values : "
    f"{n_nonpositive}"
)

print(
    f"Minimum height (m)   : "
    f"{height_final['Max_Height_m'].min():.4f}"
)

print(
    f"Median height (m)    : "
    f"{height_final['Max_Height_m'].median():.4f}"
)

print(
    f"Mean height (m)      : "
    f"{height_final['Max_Height_m'].mean():.4f}"
)

print(
    f"Maximum height (m)   : "
    f"{height_final['Max_Height_m'].max():.4f}"
)


# -----------------------------------------------------------------------------
# 10.22E. Required provenance-field validation
# -----------------------------------------------------------------------------

required_fields = [
    "Canonical_Species",
    "Max_Height_m",
    "Max_Height_Assignment_Level",
    "Max_Height_Provenance_Class",
    "Max_Height_Source",
    "Max_Height_Source_Taxon",
    "Max_Height_Match_Status",
    "Max_Height_Unit",
    "Max_Height_Quality_Flag",
]


missing_required_fields = [
    col for col in required_fields
    if col not in height_final.columns
]


if missing_required_fields:

    raise ValueError(
        "Required final Maximum Height fields missing: "
        + ", ".join(missing_required_fields)
    )


print("\nRequired provenance fields")
print("-" * 100)

print("Required fields present : YES")


# -----------------------------------------------------------------------------
# 10.22F. Missing provenance diagnostics
# -----------------------------------------------------------------------------

provenance_missing_summary = pd.Series(
    {
        field:
            int(
                height_final[
                    field
                ].isna().sum()
            )
        for field in required_fields
    }
)


print("\nMissing-value audit by required field")
print("-" * 100)

print(
    provenance_missing_summary
)


# Numeric and core provenance fields must be complete.
core_complete_fields = [
    "Canonical_Species",
    "Max_Height_m",
    "Max_Height_Assignment_Level",
    "Max_Height_Provenance_Class",
    "Max_Height_Source",
    "Max_Height_Source_Taxon",
    "Max_Height_Match_Status",
    "Max_Height_Unit",
    "Max_Height_Quality_Flag",
]


for field in core_complete_fields:

    assert height_final[
        field
    ].notna().all()


# -----------------------------------------------------------------------------
# 10.22G. Unit validation
# -----------------------------------------------------------------------------

unit_counts = height_final[
    "Max_Height_Unit"
].value_counts(
    dropna=False
)


print("\nUnit validation")
print("-" * 100)

print(
    unit_counts
)


assert set(
    height_final[
        "Max_Height_Unit"
    ].dropna().unique()
) == {"m"}


# -----------------------------------------------------------------------------
# 10.22H. Assignment-level validation
# -----------------------------------------------------------------------------

assignment_counts = (
    height_final[
        "Max_Height_Assignment_Level"
    ]
    .value_counts(
        dropna=False
    )
)


print("\nAssignment-level summary")
print("-" * 100)

print(
    assignment_counts
)


expected_assignment_counts = {
    "SPECIES": 119,
    "GENUS": 7,
    "SPECIES_COMPLEX": 1,
}


for level, expected_n in expected_assignment_counts.items():

    observed_n = int(
        assignment_counts.get(
            level,
            0
        )
    )

    assert observed_n == expected_n


assert int(
    assignment_counts.sum()
) == 127


# -----------------------------------------------------------------------------
# 10.22I. Provenance-class validation
# -----------------------------------------------------------------------------

provenance_counts = (
    height_final[
        "Max_Height_Provenance_Class"
    ]
    .value_counts(
        dropna=False
    )
)


print("\nProvenance-class summary")
print("-" * 100)

print(
    provenance_counts
)


expected_provenance_counts = {
    "SPECIES_GIFT":
        99,

    "SPECIES_REFERENCE":
        17,

    "SPECIES_BIEN":
        3,

    "GENUS_GIFT_MEDIAN":
        6,

    "GENUS_GIFT_SINGLE_SPECIES_PROXY":
        1,

    "SPECIES_COMPLEX_REFERENCE":
        1,
}


for provenance_class, expected_n in (
    expected_provenance_counts.items()
):

    observed_n = int(
        provenance_counts.get(
            provenance_class,
            0
        )
    )

    assert observed_n == expected_n


assert int(
    provenance_counts.sum()
) == 127


# -----------------------------------------------------------------------------
# 10.22J. Acquisition-status closure validation
# -----------------------------------------------------------------------------

status_counts = (
    status_ledger[
        "Acquisition_Status"
    ]
    .value_counts(
        dropna=False
    )
)


print("\nFinal acquisition-status summary")
print("-" * 100)

print(
    status_counts
)


expected_status_counts = {
    "SPECIES_LEVEL_ACCEPTED":
        119,

    "GENUS_LEVEL_ACCEPTED":
        7,

    "SPECIES_COMPLEX_ACCEPTED":
        1,
}


for status, expected_n in expected_status_counts.items():

    observed_n = int(
        status_counts.get(
            status,
            0
        )
    )

    assert observed_n == expected_n


assert int(
    status_counts.sum()
) == 127


unresolved_status_mask = (
    status_ledger[
        "Acquisition_Status"
    ]
    .astype(str)
    .str.contains(
        "REQUIRED|HOLD|UNRESOLVED|NOT_ACQUIRED|REVIEW",
        case=False,
        regex=True,
        na=False
    )
)


n_unresolved_status = int(
    unresolved_status_mask.sum()
)


assert n_unresolved_status == 0


print(
    f"\nUnresolved/hold/review statuses : "
    f"{n_unresolved_status}"
)


# -----------------------------------------------------------------------------
# 10.22K. Quality-flag summary
# -----------------------------------------------------------------------------

quality_counts = (
    height_final[
        "Max_Height_Quality_Flag"
    ]
    .value_counts(
        dropna=False
    )
)


print("\nQuality-flag summary")
print("-" * 100)

print(
    quality_counts
)


# -----------------------------------------------------------------------------
# 10.22L. Biological range diagnostics
# -----------------------------------------------------------------------------

# These are broad diagnostics only.
# They do not imply that every tree species should lie within a narrow range.

height_final[
    "Final_Broad_Plausibility"
] = np.select(
    [
        height_final[
            "Max_Height_m"
        ].between(
            1,
            100,
            inclusive="both"
        ),
    ],
    [
        "PLAUSIBLE"
    ],
    default="REVIEW"
)


plausibility_counts = (
    height_final[
        "Final_Broad_Plausibility"
    ]
    .value_counts()
)


print("\nFinal broad biological plausibility")
print("-" * 100)

print(
    plausibility_counts
)


n_plausibility_review = int(
    height_final[
        "Final_Broad_Plausibility"
    ]
    .eq(
        "REVIEW"
    )
    .sum()
)


assert n_plausibility_review == 0


# -----------------------------------------------------------------------------
# 10.22M. Assignment-level statistical summary
# -----------------------------------------------------------------------------

assignment_summary = (
    height_final
    .groupby(
        "Max_Height_Assignment_Level",
        dropna=False
    )
    .agg(
        N_Taxa=(
            "Canonical_Species",
            "count"
        ),

        Mean_Max_Height_m=(
            "Max_Height_m",
            "mean"
        ),

        SD_Max_Height_m=(
            "Max_Height_m",
            "std"
        ),

        Min_Max_Height_m=(
            "Max_Height_m",
            "min"
        ),

        Median_Max_Height_m=(
            "Max_Height_m",
            "median"
        ),

        Max_Max_Height_m=(
            "Max_Height_m",
            "max"
        ),
    )
    .reset_index()
)


assignment_summary[
    "Percent_of_127_Taxa"
] = (
    assignment_summary[
        "N_Taxa"
    ]
    /
    127
    * 100
)


print("\nFinal assignment-level statistical summary")
print("-" * 100)

display(
    assignment_summary
)


# -----------------------------------------------------------------------------
# 10.22N. Provenance-class statistical summary
# -----------------------------------------------------------------------------

provenance_summary = (
    height_final
    .groupby(
        "Max_Height_Provenance_Class",
        dropna=False
    )
    .agg(
        N_Taxa=(
            "Canonical_Species",
            "count"
        ),

        Mean_Max_Height_m=(
            "Max_Height_m",
            "mean"
        ),

        SD_Max_Height_m=(
            "Max_Height_m",
            "std"
        ),

        Min_Max_Height_m=(
            "Max_Height_m",
            "min"
        ),

        Median_Max_Height_m=(
            "Max_Height_m",
            "median"
        ),

        Max_Max_Height_m=(
            "Max_Height_m",
            "max"
        ),
    )
    .reset_index()
)


provenance_summary[
    "Percent_of_127_Taxa"
] = (
    provenance_summary[
        "N_Taxa"
    ]
    /
    127
    * 100
)


provenance_summary = (
    provenance_summary
    .sort_values(
        [
            "N_Taxa",
            "Max_Height_Provenance_Class",
        ],
        ascending=[
            False,
            True,
        ]
    )
    .reset_index(
        drop=True
    )
)


print("\nFinal Maximum Height provenance summary")
print("-" * 100)

display(
    provenance_summary
)


# -----------------------------------------------------------------------------
# 10.22O. Explicit exceptional-case audit
# -----------------------------------------------------------------------------

exceptional_taxa = [
    "Amphimas tetracoides",
    "Musa sapiensis",
]


exceptional_audit = height_final[
    height_final[
        "Canonical_Species"
    ].isin(
        exceptional_taxa
    )
][
    [
        "Canonical_Species",
        "Max_Height_m",
        "Max_Height_Assignment_Level",
        "Max_Height_Provenance_Class",
        "Max_Height_Source",
        "Max_Height_Source_Taxon",
        "Max_Height_Match_Status",
        "Max_Height_Quality_Flag",
    ]
].copy()


print("\nExceptional-case audit")
print("-" * 100)

display(
    exceptional_audit
)


assert len(exceptional_audit) == 2


# -----------------------------------------------------------------------------
# 10.22P. Construct final authoritative table
# -----------------------------------------------------------------------------

# Put core fields first while preserving supplementary provenance columns.

core_column_order = [
    "Canonical_Species",
    "Max_Height_m",
    "Max_Height_Unit",
    "Max_Height_Assignment_Level",
    "Max_Height_Provenance_Class",
    "Max_Height_Source",
    "Max_Height_Source_Taxon",
    "Max_Height_Match_Status",
    "Max_Height_Quality_Flag",
]


remaining_columns = [
    col
    for col in height_final.columns
    if col not in core_column_order
]


height_final = height_final[
    core_column_order
    +
    remaining_columns
].copy()


# Preserve canonical manifest order.
manifest_order = (
    target_manifest[
        [
            "Canonical_Species"
        ]
    ]
    .reset_index()
    .rename(
        columns={
            "index":
                "_Canonical_Order"
        }
    )
)


height_final = height_final.merge(
    manifest_order,
    on="Canonical_Species",
    how="left"
)


assert height_final[
    "_Canonical_Order"
].notna().all()


height_final = (
    height_final
    .sort_values(
        "_Canonical_Order"
    )
    .drop(
        columns=[
            "_Canonical_Order"
        ]
    )
    .reset_index(
        drop=True
    )
)


assert len(height_final) == 127

assert (
    height_final[
        "Canonical_Species"
    ].nunique()
    == 127
)


# -----------------------------------------------------------------------------
# 10.22Q. Final validation report
# -----------------------------------------------------------------------------

validation_report = pd.DataFrame(
    [
        {
            "Validation_Check":
                "Canonical rows",

            "Expected":
                127,

            "Observed":
                len(height_final),

            "Status":
                "PASS"
                if len(height_final) == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Unique canonical taxa",

            "Expected":
                127,

            "Observed":
                height_final[
                    "Canonical_Species"
                ].nunique(),

            "Status":
                "PASS"
                if height_final[
                    "Canonical_Species"
                ].nunique() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Numeric Maximum Height",

            "Expected":
                127,

            "Observed":
                int(
                    height_final[
                        "Max_Height_m"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if height_final[
                    "Max_Height_m"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Missing Maximum Height",

            "Expected":
                0,

            "Observed":
                int(
                    height_final[
                        "Max_Height_m"
                    ].isna().sum()
                ),

            "Status":
                "PASS"
                if height_final[
                    "Max_Height_m"
                ].isna().sum() == 0
                else "FAIL",
        },

        {
            "Validation_Check":
                "Positive values",

            "Expected":
                127,

            "Observed":
                int(
                    (
                        height_final[
                            "Max_Height_m"
                        ] > 0
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    height_final[
                        "Max_Height_m"
                    ] > 0
                ).sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Species-level assignments",

            "Expected":
                119,

            "Observed":
                int(
                    (
                        height_final[
                            "Max_Height_Assignment_Level"
                        ] == "SPECIES"
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    height_final[
                        "Max_Height_Assignment_Level"
                    ] == "SPECIES"
                ).sum() == 119
                else "FAIL",
        },

        {
            "Validation_Check":
                "Genus-level assignments",

            "Expected":
                7,

            "Observed":
                int(
                    (
                        height_final[
                            "Max_Height_Assignment_Level"
                        ] == "GENUS"
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    height_final[
                        "Max_Height_Assignment_Level"
                    ] == "GENUS"
                ).sum() == 7
                else "FAIL",
        },

        {
            "Validation_Check":
                "Species-complex assignments",

            "Expected":
                1,

            "Observed":
                int(
                    (
                        height_final[
                            "Max_Height_Assignment_Level"
                        ] == "SPECIES_COMPLEX"
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    height_final[
                        "Max_Height_Assignment_Level"
                    ] == "SPECIES_COMPLEX"
                ).sum() == 1
                else "FAIL",
        },

        {
            "Validation_Check":
                "Unresolved acquisition statuses",

            "Expected":
                0,

            "Observed":
                n_unresolved_status,

            "Status":
                "PASS"
                if n_unresolved_status == 0
                else "FAIL",
        },

        {
            "Validation_Check":
                "Broad plausibility review records",

            "Expected":
                0,

            "Observed":
                n_plausibility_review,

            "Status":
                "PASS"
                if n_plausibility_review == 0
                else "FAIL",
        },
    ]
)


print("\nFinal validation report")
print("-" * 100)

display(
    validation_report
)


assert (
    validation_report[
        "Status"
    ].eq(
        "PASS"
    )
    .all()
)


# -----------------------------------------------------------------------------
# 10.22R. Export authoritative Maximum Height outputs
# -----------------------------------------------------------------------------

height_final.to_csv(
    FINAL_HEIGHT_PATH,
    index=False
)

provenance_summary.to_csv(
    PROVENANCE_SUMMARY_PATH,
    index=False
)

assignment_summary.to_csv(
    ASSIGNMENT_SUMMARY_PATH,
    index=False
)

validation_report.to_csv(
    VALIDATION_REPORT_PATH,
    index=False
)


print("\nAuthoritative Maximum Height exports")
print("-" * 100)

print(
    f"Final Maximum Height table : "
    f"{FINAL_HEIGHT_PATH}"
)

print(
    f"Provenance summary         : "
    f"{PROVENANCE_SUMMARY_PATH}"
)

print(
    f"Assignment summary         : "
    f"{ASSIGNMENT_SUMMARY_PATH}"
)

print(
    f"Validation report          : "
    f"{VALIDATION_REPORT_PATH}"
)


# -----------------------------------------------------------------------------
# 10.22S. Final closure statement
# -----------------------------------------------------------------------------

print("\nMaximum Height closure statement")
print("-" * 100)

print(
    """
Maximum Height acquisition and harmonization is now complete.

All 127 canonical Omo taxa have a positive numeric Maximum Height value.

Evidence strength remains explicitly differentiated:

    • direct/reconciled species-level evidence;
    • genus-level estimates based on GIFT species distributions;
    • one limited single-species genus proxy; and
    • one cultivated species-complex reference.

No canonical Omo taxon name has been overwritten by an external source.

No unresolved, review-required or not-acquired Maximum Height record
remains.

The file:

    outputs/functional_traits/omo_max_height_final.csv

is now the authoritative Notebook 07B Maximum Height trait table.
"""
)


# -----------------------------------------------------------------------------
# 10.22T. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 10.22 FINAL MAXIMUM HEIGHT CLOSURE COMPLETE")
print("=" * 100)

print(
    f"Canonical taxa                  : "
    f"{len(height_final)}"
)

print(
    f"Unique taxa                     : "
    f"{height_final['Canonical_Species'].nunique()}"
)

print(
    f"Numeric Maximum Height values   : "
    f"{height_final['Max_Height_m'].notna().sum()}"
)

print(
    f"Missing Maximum Height values   : "
    f"{height_final['Max_Height_m'].isna().sum()}"
)

print(
    f"Minimum Maximum Height (m)      : "
    f"{height_final['Max_Height_m'].min():.4f}"
)

print(
    f"Median Maximum Height (m)       : "
    f"{height_final['Max_Height_m'].median():.4f}"
)

print(
    f"Mean Maximum Height (m)         : "
    f"{height_final['Max_Height_m'].mean():.4f}"
)

print(
    f"Maximum Maximum Height (m)      : "
    f"{height_final['Max_Height_m'].max():.4f}"
)

print(
    f"Species-level assignments       : "
    f"{(height_final['Max_Height_Assignment_Level'] == 'SPECIES').sum()}"
)

print(
    f"Genus-level assignments         : "
    f"{(height_final['Max_Height_Assignment_Level'] == 'GENUS').sum()}"
)

print(
    f"Species-complex assignments     : "
    f"{(height_final['Max_Height_Assignment_Level'] == 'SPECIES_COMPLEX').sum()}"
)

print(
    f"Failed final validation checks  : "
    f"{(validation_report['Status'] != 'PASS').sum()}"
)

print(
    "Maximum Height trait status     : FORMALLY CLOSED"
)

print(
    "Next step                      : "
    "Final integration of WD + SLA + Seed Mass + Maximum Height "
    "across all 127 canonical taxa."
)
```

```python
# =============================================================================
# SECTION 11.0 — FINAL FOUR-TRAIT INTEGRATION SETUP AND BACKBONE AUDIT
# WD + SLA + Seed Mass + Maximum Height
# =============================================================================

from pathlib import Path
import pandas as pd


print("=" * 100)
print("SECTION 11.0 — FINAL FOUR-TRAIT INTEGRATION SETUP AND BACKBONE AUDIT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 11.0A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)


WD_PATH = (
    OUTPUT_DIR /
    "omo_wood_density_final.csv"
)

SLA_PATH = (
    OUTPUT_DIR /
    "omo_sla_final.csv"
)

SEED_PATH = (
    OUTPUT_DIR /
    "omo_seed_mass_final.csv"
)

HEIGHT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_final.csv"
)


# -----------------------------------------------------------------------------
# 11.0B. Load authoritative trait files
# -----------------------------------------------------------------------------

wd = pd.read_csv(
    WD_PATH,
    low_memory=False
)

sla = pd.read_csv(
    SLA_PATH,
    low_memory=False
)

seed = pd.read_csv(
    SEED_PATH,
    low_memory=False
)

height = pd.read_csv(
    HEIGHT_PATH,
    low_memory=False
)


print("\nAuthoritative trait files loaded")
print("-" * 100)

print(
    f"WD rows          : "
    f"{len(wd)}"
)

print(
    f"SLA rows         : "
    f"{len(sla)}"
)

print(
    f"Seed Mass rows   : "
    f"{len(seed)}"
)

print(
    f"Maximum Height   : "
    f"{len(height)}"
)


# -----------------------------------------------------------------------------
# 11.0C. Canonical taxon-column validation
# -----------------------------------------------------------------------------

trait_tables = {
    "WD": wd,
    "SLA": sla,
    "SEED_MASS": seed,
    "MAX_HEIGHT": height,
}


for trait_name, df in trait_tables.items():

    if "Canonical_Species" not in df.columns:

        raise ValueError(
            f"{trait_name} does not contain "
            "'Canonical_Species'."
        )


    print(
        f"\n{trait_name} canonical backbone"
    )

    print("-" * 100)

    print(
        f"Rows        : "
        f"{len(df)}"
    )

    print(
        f"Unique taxa : "
        f"{df['Canonical_Species'].nunique()}"
    )

    print(
        f"Missing taxa: "
        f"{df['Canonical_Species'].isna().sum()}"
    )


    assert len(df) == 127

    assert (
        df[
            "Canonical_Species"
        ].nunique()
        == 127
    )

    assert (
        df[
            "Canonical_Species"
        ].isna()
        .sum()
        == 0
    )


# -----------------------------------------------------------------------------
# 11.0D. Cross-trait taxon-set agreement
# -----------------------------------------------------------------------------

wd_taxa = set(
    wd[
        "Canonical_Species"
    ]
)

sla_taxa = set(
    sla[
        "Canonical_Species"
    ]
)

seed_taxa = set(
    seed[
        "Canonical_Species"
    ]
)

height_taxa = set(
    height[
        "Canonical_Species"
    ]
)


print("\nCross-trait canonical taxon-set agreement")
print("-" * 100)

print(
    f"WD = SLA          : "
    f"{wd_taxa == sla_taxa}"
)

print(
    f"WD = Seed Mass    : "
    f"{wd_taxa == seed_taxa}"
)

print(
    f"WD = Max Height   : "
    f"{wd_taxa == height_taxa}"
)

print(
    f"SLA = Seed Mass   : "
    f"{sla_taxa == seed_taxa}"
)

print(
    f"SLA = Max Height  : "
    f"{sla_taxa == height_taxa}"
)

print(
    f"Seed = Max Height : "
    f"{seed_taxa == height_taxa}"
)


assert wd_taxa == sla_taxa
assert wd_taxa == seed_taxa
assert wd_taxa == height_taxa


# -----------------------------------------------------------------------------
# 11.0E. Canonical row-order agreement
# -----------------------------------------------------------------------------

wd_order = wd[
    "Canonical_Species"
].tolist()

sla_order = sla[
    "Canonical_Species"
].tolist()

seed_order = seed[
    "Canonical_Species"
].tolist()

height_order = height[
    "Canonical_Species"
].tolist()


print("\nCross-trait canonical row-order agreement")
print("-" * 100)

print(
    f"WD = SLA          : "
    f"{wd_order == sla_order}"
)

print(
    f"WD = Seed Mass    : "
    f"{wd_order == seed_order}"
)

print(
    f"WD = Max Height   : "
    f"{wd_order == height_order}"
)


# Row-order agreement is desirable but not required,
# because the integration will merge explicitly by canonical taxon.


# -----------------------------------------------------------------------------
# 11.0F. Column inventories
# -----------------------------------------------------------------------------

print("\nWD columns")
print("-" * 100)

print(
    list(wd.columns)
)


print("\nSLA columns")
print("-" * 100)

print(
    list(sla.columns)
)


print("\nSeed Mass columns")
print("-" * 100)

print(
    list(seed.columns)
)


print("\nMaximum Height columns")
print("-" * 100)

print(
    list(height.columns)
)


# -----------------------------------------------------------------------------
# 11.0G. Candidate numeric trait-field discovery
# -----------------------------------------------------------------------------

numeric_field_candidates = {
    "WD": [
        col
        for col in wd.columns
        if any(
            token in col.lower()
            for token in [
                "wood",
                "wd",
                "density",
            ]
        )
    ],

    "SLA": [
        col
        for col in sla.columns
        if "sla" in col.lower()
    ],

    "SEED_MASS": [
        col
        for col in seed.columns
        if (
            "seed" in col.lower()
            and
            "mass" in col.lower()
        )
    ],

    "MAX_HEIGHT": [
        col
        for col in height.columns
        if (
            "height" in col.lower()
            and
            (
                col.endswith("_m")
                or
                "max_height" in col.lower()
            )
        )
    ],
}


print("\nCandidate numeric trait fields")
print("-" * 100)

for trait_name, candidates in (
    numeric_field_candidates.items()
):

    print(
        f"{trait_name:<12}: "
        f"{candidates}"
    )


# -----------------------------------------------------------------------------
# 11.0H. Missing-value overview
# -----------------------------------------------------------------------------

print("\nInitial missing-value overview")
print("-" * 100)


for trait_name, df in trait_tables.items():

    numeric_cols = df.select_dtypes(
        include="number"
    ).columns.tolist()

    print(
        f"\n{trait_name}"
    )

    if numeric_cols:

        missing_numeric = (
            df[
                numeric_cols
            ]
            .isna()
            .sum()
        )

        print(
            missing_numeric[
                missing_numeric > 0
            ]
        )

    else:

        print(
            "No numeric fields detected."
        )


# -----------------------------------------------------------------------------
# 11.0I. Integration safeguard
# -----------------------------------------------------------------------------

print("\nIntegration safeguard")
print("-" * 100)

print(
    """
No trait tables are merged in this section.

This section validates only that:

    • the four authoritative trait files exist;
    • each contains exactly 127 canonical taxa;
    • the canonical taxon sets are identical; and
    • the available value/provenance fields are known before integration.

The final functional-trait matrix will be constructed by explicit merge
on Canonical_Species rather than by row position.

Seed Mass biological nonapplicability will remain explicit and must not
be silently converted into a numeric value during integration.
"""
)


# -----------------------------------------------------------------------------
# 11.0J. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 11.0 FOUR-TRAIT INTEGRATION SETUP COMPLETE")
print("=" * 100)

print(
    "Trait files validated       : 4"
)

print(
    "Canonical taxa per trait    : 127"
)

print(
    "Cross-trait taxon agreement : PASS"
)

print(
    "Integrated matrix created   : NO"
)

print(
    "Next step                   : "
    "Section 11.1 — identify and standardize the authoritative "
    "value/provenance fields for each trait before merging."
)
```

```python
# =============================================================================
# SECTION 11.1 — AUTHORITATIVE TRAIT FIELD SELECTION AND STANDARDIZATION
# WD + SLA + Seed Mass + Maximum Height
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 11.1 — AUTHORITATIVE TRAIT FIELD SELECTION AND STANDARDIZATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 11.1A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

STANDARDIZED_DIR = (
    OUTPUT_DIR /
    "integration"
)

STANDARDIZED_DIR.mkdir(
    parents=True,
    exist_ok=True
)


WD_PATH = (
    OUTPUT_DIR /
    "omo_wood_density_final.csv"
)

SLA_PATH = (
    OUTPUT_DIR /
    "omo_sla_final.csv"
)

SEED_PATH = (
    OUTPUT_DIR /
    "omo_seed_mass_final.csv"
)

HEIGHT_PATH = (
    OUTPUT_DIR /
    "omo_max_height_final.csv"
)


# -----------------------------------------------------------------------------
# 11.1B. Load authoritative files
# -----------------------------------------------------------------------------

wd = pd.read_csv(
    WD_PATH,
    low_memory=False
)

sla = pd.read_csv(
    SLA_PATH,
    low_memory=False
)

seed = pd.read_csv(
    SEED_PATH,
    low_memory=False
)

height = pd.read_csv(
    HEIGHT_PATH,
    low_memory=False
)


for df in [wd, sla, seed, height]:

    assert len(df) == 127

    assert (
        df[
            "Canonical_Species"
        ].nunique()
        == 127
    )


print("\nAuthoritative trait files reloaded")
print("-" * 100)

print(f"WD rows          : {len(wd)}")
print(f"SLA rows         : {len(sla)}")
print(f"Seed Mass rows   : {len(seed)}")
print(f"Maximum Height   : {len(height)}")


# -----------------------------------------------------------------------------
# 11.1C. Define authoritative field map
# -----------------------------------------------------------------------------

authoritative_field_map = {
    "WD": {
        "value":
            "Standardized_Trait_Value",

        "unit":
            "Standardized_Trait_Unit",

        "provenance":
            "WD_Provenance_Class",

        "quality":
            "Trait_Quality_Flag",

        "source":
            "Trait_Source",

        "match":
            "Match_Status",

        "assignment":
            "Taxonomic_Level",
    },

    "SLA": {
        "value":
            "Standardized_SLA_Value",

        "unit":
            "Standardized_SLA_Unit",

        "provenance":
            "SLA_Provenance_Class",

        "quality":
            "SLA_Quality_Flag",

        "source":
            "Trait_Source",

        "match":
            "SLA_Match_Status",

        "assignment":
            "Taxonomic_Resolution",
    },

    "SEED_MASS": {
        "value":
            "Seed_Mass_mg",

        "unit":
            "Standardized_Trait_Unit",

        "provenance":
            "Seed_Mass_Provenance_Class",

        "quality":
            "Seed_Mass_Quality_Flag",

        "source":
            "Seed_Mass_Evidence_Source",

        "match":
            "Seed_Mass_Assignment_Method",

        "assignment":
            "Seed_Mass_Assignment_Level",
    },

    "MAX_HEIGHT": {
        "value":
            "Max_Height_m",

        "unit":
            "Max_Height_Unit",

        "provenance":
            "Max_Height_Provenance_Class",

        "quality":
            "Max_Height_Quality_Flag",

        "source":
            "Max_Height_Source",

        "match":
            "Max_Height_Match_Status",

        "assignment":
            "Max_Height_Assignment_Level",
    },
}


trait_frames = {
    "WD": wd,
    "SLA": sla,
    "SEED_MASS": seed,
    "MAX_HEIGHT": height,
}


print("\nAuthoritative integration field map")
print("-" * 100)

for trait, mapping in authoritative_field_map.items():

    print(f"\n{trait}")

    for role, column in mapping.items():

        print(
            f"  {role:<12}: "
            f"{column}"
        )


# -----------------------------------------------------------------------------
# 11.1D. Validate required authoritative fields
# -----------------------------------------------------------------------------

print("\nAuthoritative-field existence validation")
print("-" * 100)


for trait, mapping in authoritative_field_map.items():

    df = trait_frames[trait]

    required_columns = [
        "Canonical_Species"
    ] + list(
        mapping.values()
    )

    missing_columns = [
        col
        for col in required_columns
        if col not in df.columns
    ]

    if missing_columns:

        raise ValueError(
            f"{trait} missing required authoritative fields: "
            + ", ".join(missing_columns)
        )

    print(
        f"{trait:<12}: PASS"
    )


# -----------------------------------------------------------------------------
# 11.1E. WD authoritative-value validation
# -----------------------------------------------------------------------------

wd_value = pd.to_numeric(
    wd[
        "Standardized_Trait_Value"
    ],
    errors="coerce"
)


print("\nWD authoritative-field validation")
print("-" * 100)

print(
    f"Numeric values        : "
    f"{wd_value.notna().sum()} / 127"
)

print(
    f"Missing values        : "
    f"{wd_value.isna().sum()}"
)

print(
    f"Minimum               : "
    f"{wd_value.min():.4f}"
)

print(
    f"Median                : "
    f"{wd_value.median():.4f}"
)

print(
    f"Maximum               : "
    f"{wd_value.max():.4f}"
)

print(
    f"Units                  : "
    f"{sorted(wd['Standardized_Trait_Unit'].dropna().unique().tolist())}"
)


assert wd_value.notna().sum() == 127

assert (
    wd_value > 0
).all()


# -----------------------------------------------------------------------------
# 11.1F. SLA authoritative-field and redundancy validation
# -----------------------------------------------------------------------------

sla_value = pd.to_numeric(
    sla[
        "Standardized_SLA_Value"
    ],
    errors="coerce"
)


print("\nSLA authoritative-field validation")
print("-" * 100)

print(
    f"Numeric values        : "
    f"{sla_value.notna().sum()} / 127"
)

print(
    f"Missing values        : "
    f"{sla_value.isna().sum()}"
)

print(
    f"Minimum               : "
    f"{sla_value.min():.4f}"
)

print(
    f"Median                : "
    f"{sla_value.median():.4f}"
)

print(
    f"Maximum               : "
    f"{sla_value.max():.4f}"
)

print(
    f"Units                  : "
    f"{sorted(sla['Standardized_SLA_Unit'].dropna().unique().tolist())}"
)


assert sla_value.notna().sum() == 127

assert (
    sla_value > 0
).all()


# Compare duplicated SLA provenance fields where available.

sla_provenance_equal = (
    sla[
        "SLA_Provenance_Class"
    ]
    .astype("string")
    .fillna("<NA>")
    ==
    sla[
        "Provenance_Class"
    ]
    .astype("string")
    .fillna("<NA>")
).all()


sla_quality_equal = (
    sla[
        "SLA_Quality_Flag"
    ]
    .astype("string")
    .fillna("<NA>")
    ==
    sla[
        "Quality_Flag"
    ]
    .astype("string")
    .fillna("<NA>")
).all()


print(
    f"SLA provenance duplicate agreement : "
    f"{sla_provenance_equal}"
)

print(
    f"SLA quality duplicate agreement    : "
    f"{sla_quality_equal}"
)


# These are diagnostic.
# The explicitly trait-prefixed SLA fields remain authoritative.


# -----------------------------------------------------------------------------
# 11.1G. Seed Mass authoritative-value validation
# -----------------------------------------------------------------------------

seed_value = pd.to_numeric(
    seed[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)


seed_standardized = pd.to_numeric(
    seed[
        "Standardized_Trait_Value"
    ],
    errors="coerce"
)


print("\nSeed Mass authoritative-field validation")
print("-" * 100)

print(
    f"Numeric Seed_Mass_mg values          : "
    f"{seed_value.notna().sum()} / 127"
)

print(
    f"Missing Seed_Mass_mg values          : "
    f"{seed_value.isna().sum()}"
)

print(
    f"Numeric Standardized_Trait_Value     : "
    f"{seed_standardized.notna().sum()} / 127"
)


seed_numeric_overlap = (
    seed_value.notna()
    &
    seed_standardized.notna()
)


seed_value_agreement = np.isclose(
    seed_value[
        seed_numeric_overlap
    ],
    seed_standardized[
        seed_numeric_overlap
    ],
    rtol=1e-10,
    atol=1e-10
).all()


print(
    f"Seed_Mass_mg = standardized values   : "
    f"{seed_value_agreement}"
)


assert seed_value.notna().sum() == 126
assert seed_value.isna().sum() == 1
assert seed_value_agreement


# Identify the intentional nonnumeric record.

seed_missing_record = seed[
    seed_value.isna()
].copy()


print("\nSeed Mass intentional nonnumeric record")
print("-" * 100)

display(
    seed_missing_record[
        [
            "Canonical_Species",
            "Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Seed_Mass_Assignment_Method",
            "Seed_Mass_Data_Status",
        ]
    ]
)


assert len(seed_missing_record) == 1

assert (
    seed_missing_record[
        "Canonical_Species"
    ].iloc[0]
    ==
    "Musa sapiensis"
)


# -----------------------------------------------------------------------------
# 11.1H. Maximum Height authoritative-field validation
# -----------------------------------------------------------------------------

height_value = pd.to_numeric(
    height[
        "Max_Height_m"
    ],
    errors="coerce"
)


print("\nMaximum Height authoritative-field validation")
print("-" * 100)

print(
    f"Numeric values        : "
    f"{height_value.notna().sum()} / 127"
)

print(
    f"Missing values        : "
    f"{height_value.isna().sum()}"
)

print(
    f"Minimum               : "
    f"{height_value.min():.4f}"
)

print(
    f"Median                : "
    f"{height_value.median():.4f}"
)

print(
    f"Maximum               : "
    f"{height_value.max():.4f}"
)

print(
    f"Units                  : "
    f"{sorted(height['Max_Height_Unit'].dropna().unique().tolist())}"
)


assert height_value.notna().sum() == 127

assert (
    height_value > 0
).all()


# -----------------------------------------------------------------------------
# 11.1I. Construct standardized WD extract
# -----------------------------------------------------------------------------

wd_standardized = pd.DataFrame(
    {
        "Canonical_Species":
            wd[
                "Canonical_Species"
            ],

        "WD_Value":
            wd_value,

        "WD_Unit":
            wd[
                "Standardized_Trait_Unit"
            ],

        "WD_Assignment_Level":
            wd[
                "Taxonomic_Level"
            ],

        "WD_Provenance_Class":
            wd[
                "WD_Provenance_Class"
            ],

        "WD_Source":
            wd[
                "Trait_Source"
            ],

        "WD_Source_Taxon":
            wd[
                "Trait_Source_Name"
            ],

        "WD_Match_Status":
            wd[
                "Match_Status"
            ],

        "WD_Quality_Flag":
            wd[
                "Trait_Quality_Flag"
            ],
    }
)


# -----------------------------------------------------------------------------
# 11.1J. Construct standardized SLA extract
# -----------------------------------------------------------------------------

sla_standardized = pd.DataFrame(
    {
        "Canonical_Species":
            sla[
                "Canonical_Species"
            ],

        "SLA_Value":
            sla_value,

        "SLA_Unit":
            sla[
                "Standardized_SLA_Unit"
            ],

        "SLA_Assignment_Level":
            sla[
                "Taxonomic_Resolution"
            ],

        "SLA_Provenance_Class":
            sla[
                "SLA_Provenance_Class"
            ],

        "SLA_Source":
            sla[
                "Trait_Source"
            ],

        "SLA_Source_Taxon":
            sla[
                "SLA_Query_Name"
            ],

        "SLA_Match_Status":
            sla[
                "SLA_Match_Status"
            ],

        "SLA_Quality_Flag":
            sla[
                "SLA_Quality_Flag"
            ],
    }
)


# -----------------------------------------------------------------------------
# 11.1K. Construct standardized Seed Mass extract
# -----------------------------------------------------------------------------

seed_standardized_extract = pd.DataFrame(
    {
        "Canonical_Species":
            seed[
                "Canonical_Species"
            ],

        "Seed_Mass_mg":
            seed_value,

        "Seed_Mass_Unit":
            seed[
                "Standardized_Trait_Unit"
            ],

        "Seed_Mass_Assignment_Level":
            seed[
                "Seed_Mass_Assignment_Level"
            ],

        "Seed_Mass_Provenance_Class":
            seed[
                "Seed_Mass_Provenance_Class"
            ],

        "Seed_Mass_Source":
            seed[
                "Seed_Mass_Evidence_Source"
            ],

        "Seed_Mass_Source_Taxon":
            seed[
                "Final_Search_Name"
            ],

        "Seed_Mass_Match_Status":
            seed[
                "Seed_Mass_Assignment_Method"
            ],

        "Seed_Mass_Quality_Flag":
            seed[
                "Seed_Mass_Quality_Flag"
            ],

        "Seed_Mass_Data_Status":
            seed[
                "Seed_Mass_Data_Status"
            ],

        "Seed_Mass_Numeric_Assigned":
            seed[
                "Numerical_Seed_Mass_Assigned"
            ],
    }
)


# -----------------------------------------------------------------------------
# 11.1L. Construct standardized Maximum Height extract
# -----------------------------------------------------------------------------

height_standardized = pd.DataFrame(
    {
        "Canonical_Species":
            height[
                "Canonical_Species"
            ],

        "Max_Height_m":
            height_value,

        "Max_Height_Unit":
            height[
                "Max_Height_Unit"
            ],

        "Max_Height_Assignment_Level":
            height[
                "Max_Height_Assignment_Level"
            ],

        "Max_Height_Provenance_Class":
            height[
                "Max_Height_Provenance_Class"
            ],

        "Max_Height_Source":
            height[
                "Max_Height_Source"
            ],

        "Max_Height_Source_Taxon":
            height[
                "Max_Height_Source_Taxon"
            ],

        "Max_Height_Match_Status":
            height[
                "Max_Height_Match_Status"
            ],

        "Max_Height_Quality_Flag":
            height[
                "Max_Height_Quality_Flag"
            ],
    }
)


# -----------------------------------------------------------------------------
# 11.1M. Validate standardized extracts
# -----------------------------------------------------------------------------

standardized_extracts = {
    "WD":
        wd_standardized,

    "SLA":
        sla_standardized,

    "SEED_MASS":
        seed_standardized_extract,

    "MAX_HEIGHT":
        height_standardized,
}


print("\nStandardized extract validation")
print("-" * 100)


for trait, df in standardized_extracts.items():

    n_rows = len(df)

    n_taxa = df[
        "Canonical_Species"
    ].nunique()

    n_missing_taxa = df[
        "Canonical_Species"
    ].isna().sum()

    print(
        f"{trait:<12}: "
        f"rows={n_rows}, "
        f"unique_taxa={n_taxa}, "
        f"missing_taxa={n_missing_taxa}"
    )

    assert n_rows == 127
    assert n_taxa == 127
    assert n_missing_taxa == 0


# -----------------------------------------------------------------------------
# 11.1N. Core numeric completeness summary
# -----------------------------------------------------------------------------

numeric_completeness = pd.DataFrame(
    [
        {
            "Trait":
                "WD",

            "Numeric_Field":
                "WD_Value",

            "Numeric_Values":
                int(
                    wd_standardized[
                        "WD_Value"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    wd_standardized[
                        "WD_Value"
                    ].isna().sum()
                ),
        },

        {
            "Trait":
                "SLA",

            "Numeric_Field":
                "SLA_Value",

            "Numeric_Values":
                int(
                    sla_standardized[
                        "SLA_Value"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    sla_standardized[
                        "SLA_Value"
                    ].isna().sum()
                ),
        },

        {
            "Trait":
                "SEED_MASS",

            "Numeric_Field":
                "Seed_Mass_mg",

            "Numeric_Values":
                int(
                    seed_standardized_extract[
                        "Seed_Mass_mg"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    seed_standardized_extract[
                        "Seed_Mass_mg"
                    ].isna().sum()
                ),
        },

        {
            "Trait":
                "MAX_HEIGHT",

            "Numeric_Field":
                "Max_Height_m",

            "Numeric_Values":
                int(
                    height_standardized[
                        "Max_Height_m"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    height_standardized[
                        "Max_Height_m"
                    ].isna().sum()
                ),
        },
    ]
)


numeric_completeness[
    "Numeric_Coverage_Percent"
] = (
    numeric_completeness[
        "Numeric_Values"
    ]
    /
    127
    * 100
)


print("\nCore numeric completeness")
print("-" * 100)

display(
    numeric_completeness
)


# -----------------------------------------------------------------------------
# 11.1O. Export standardized integration extracts
# -----------------------------------------------------------------------------

WD_STANDARDIZED_PATH = (
    STANDARDIZED_DIR /
    "omo_wd_integration_extract.csv"
)

SLA_STANDARDIZED_PATH = (
    STANDARDIZED_DIR /
    "omo_sla_integration_extract.csv"
)

SEED_STANDARDIZED_PATH = (
    STANDARDIZED_DIR /
    "omo_seed_mass_integration_extract.csv"
)

HEIGHT_STANDARDIZED_PATH = (
    STANDARDIZED_DIR /
    "omo_max_height_integration_extract.csv"
)

COMPLETENESS_PATH = (
    STANDARDIZED_DIR /
    "omo_trait_numeric_completeness_preintegration.csv"
)


wd_standardized.to_csv(
    WD_STANDARDIZED_PATH,
    index=False
)

sla_standardized.to_csv(
    SLA_STANDARDIZED_PATH,
    index=False
)

seed_standardized_extract.to_csv(
    SEED_STANDARDIZED_PATH,
    index=False
)

height_standardized.to_csv(
    HEIGHT_STANDARDIZED_PATH,
    index=False
)

numeric_completeness.to_csv(
    COMPLETENESS_PATH,
    index=False
)


print("\nStandardized integration exports")
print("-" * 100)

print(
    f"WD extract          : "
    f"{WD_STANDARDIZED_PATH}"
)

print(
    f"SLA extract         : "
    f"{SLA_STANDARDIZED_PATH}"
)

print(
    f"Seed Mass extract   : "
    f"{SEED_STANDARDIZED_PATH}"
)

print(
    f"Maximum Height      : "
    f"{HEIGHT_STANDARDIZED_PATH}"
)

print(
    f"Completeness audit  : "
    f"{COMPLETENESS_PATH}"
)


# -----------------------------------------------------------------------------
# 11.1P. Integration safeguard
# -----------------------------------------------------------------------------

print("\nIntegration safeguard")
print("-" * 100)

print(
    """
The four authoritative trait files have now been reduced to explicit,
trait-specific integration schemas.

Diagnostic database fields are not treated as functional-trait values.

Authoritative numeric fields are:

    WD          -> Standardized_Trait_Value
    SLA         -> Standardized_SLA_Value
    Seed Mass   -> Seed_Mass_mg
    Max Height  -> Max_Height_m

The one missing Seed Mass value for Musa sapiensis remains intentionally
missing because its biological nonapplicability was formally established
during Section 9.

No numeric imputation has been introduced.

The four standardized extracts have NOT yet been merged.
"""
)


# -----------------------------------------------------------------------------
# 11.1Q. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 11.1 AUTHORITATIVE FIELD STANDARDIZATION COMPLETE")
print("=" * 100)

print(
    f"WD numeric values             : "
    f"{wd_standardized['WD_Value'].notna().sum()} / 127"
)

print(
    f"SLA numeric values            : "
    f"{sla_standardized['SLA_Value'].notna().sum()} / 127"
)

print(
    f"Seed Mass numeric values      : "
    f"{seed_standardized_extract['Seed_Mass_mg'].notna().sum()} / 127"
)

print(
    f"Maximum Height numeric values : "
    f"{height_standardized['Max_Height_m'].notna().sum()} / 127"
)

print(
    "Standardized extracts created : 4"
)

print(
    "Integrated matrix created     : NO"
)

print(
    "Next step                     : "
    "Section 11.2 — explicit four-trait merge and integrated "
    "matrix structural validation."
)
```

```python
# =============================================================================
# SECTION 11.2 — EXPLICIT FOUR-TRAIT MERGE AND STRUCTURAL VALIDATION
# WD + SLA + Seed Mass + Maximum Height
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 11.2 — EXPLICIT FOUR-TRAIT MERGE AND STRUCTURAL VALIDATION")
print("=" * 100)


# -----------------------------------------------------------------------------
# 11.2A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

INTEGRATION_DIR = (
    OUTPUT_DIR /
    "integration"
)


WD_PATH = (
    INTEGRATION_DIR /
    "omo_wd_integration_extract.csv"
)

SLA_PATH = (
    INTEGRATION_DIR /
    "omo_sla_integration_extract.csv"
)

SEED_PATH = (
    INTEGRATION_DIR /
    "omo_seed_mass_integration_extract.csv"
)

HEIGHT_PATH = (
    INTEGRATION_DIR /
    "omo_max_height_integration_extract.csv"
)


MERGED_INTERIM_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_four_trait_merged_interim.csv"
)

STRUCTURE_REPORT_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_four_trait_structure_report.csv"
)


# -----------------------------------------------------------------------------
# 11.2B. Load standardized extracts
# -----------------------------------------------------------------------------

wd = pd.read_csv(
    WD_PATH,
    low_memory=False
)

sla = pd.read_csv(
    SLA_PATH,
    low_memory=False
)

seed = pd.read_csv(
    SEED_PATH,
    low_memory=False
)

height = pd.read_csv(
    HEIGHT_PATH,
    low_memory=False
)


print("\nStandardized extracts loaded")
print("-" * 100)

print(f"WD rows        : {len(wd)}")
print(f"SLA rows       : {len(sla)}")
print(f"Seed Mass rows : {len(seed)}")
print(f"Height rows    : {len(height)}")


# -----------------------------------------------------------------------------
# 11.2C. Pre-merge uniqueness validation
# -----------------------------------------------------------------------------

trait_frames = {
    "WD": wd,
    "SLA": sla,
    "SEED_MASS": seed,
    "MAX_HEIGHT": height,
}


for trait_name, df in trait_frames.items():

    assert len(df) == 127

    assert (
        df[
            "Canonical_Species"
        ].nunique()
        == 127
    )

    assert (
        df[
            "Canonical_Species"
        ].duplicated()
        .sum()
        == 0
    )


print("\nPre-merge uniqueness validation")
print("-" * 100)
print("All four extracts contain exactly 127 unique canonical taxa : PASS")


# -----------------------------------------------------------------------------
# 11.2D. Establish authoritative merge backbone
# -----------------------------------------------------------------------------

# WD is used only as the initial merge backbone because all four taxon sets
# have already been shown to agree. Row position is not assumed.

functional_matrix = wd.copy()


print("\nInitial merge backbone")
print("-" * 100)

print(
    f"Rows        : "
    f"{len(functional_matrix)}"
)

print(
    f"Unique taxa : "
    f"{functional_matrix['Canonical_Species'].nunique()}"
)


# -----------------------------------------------------------------------------
# 11.2E. Merge SLA
# -----------------------------------------------------------------------------

functional_matrix = functional_matrix.merge(
    sla,
    on="Canonical_Species",
    how="outer",
    validate="one_to_one",
    indicator="_Merge_WD_SLA"
)


print("\nWD + SLA merge")
print("-" * 100)

print(
    functional_matrix[
        "_Merge_WD_SLA"
    ].value_counts(dropna=False)
)


assert (
    functional_matrix[
        "_Merge_WD_SLA"
    ].eq("both").all()
)

assert len(functional_matrix) == 127


functional_matrix = functional_matrix.drop(
    columns=[
        "_Merge_WD_SLA"
    ]
)


# -----------------------------------------------------------------------------
# 11.2F. Merge Seed Mass
# -----------------------------------------------------------------------------

functional_matrix = functional_matrix.merge(
    seed,
    on="Canonical_Species",
    how="outer",
    validate="one_to_one",
    indicator="_Merge_Seed"
)


print("\nWD + SLA + Seed Mass merge")
print("-" * 100)

print(
    functional_matrix[
        "_Merge_Seed"
    ].value_counts(dropna=False)
)


assert (
    functional_matrix[
        "_Merge_Seed"
    ].eq("both").all()
)

assert len(functional_matrix) == 127


functional_matrix = functional_matrix.drop(
    columns=[
        "_Merge_Seed"
    ]
)


# -----------------------------------------------------------------------------
# 11.2G. Merge Maximum Height
# -----------------------------------------------------------------------------

functional_matrix = functional_matrix.merge(
    height,
    on="Canonical_Species",
    how="outer",
    validate="one_to_one",
    indicator="_Merge_Height"
)


print("\nFour-trait merge")
print("-" * 100)

print(
    functional_matrix[
        "_Merge_Height"
    ].value_counts(dropna=False)
)


assert (
    functional_matrix[
        "_Merge_Height"
    ].eq("both").all()
)

assert len(functional_matrix) == 127


functional_matrix = functional_matrix.drop(
    columns=[
        "_Merge_Height"
    ]
)


# -----------------------------------------------------------------------------
# 11.2H. Structural validation after merge
# -----------------------------------------------------------------------------

print("\nIntegrated matrix structural validation")
print("-" * 100)

print(
    f"Rows              : "
    f"{len(functional_matrix)}"
)

print(
    f"Unique taxa       : "
    f"{functional_matrix['Canonical_Species'].nunique()}"
)

print(
    f"Duplicate taxa    : "
    f"{functional_matrix['Canonical_Species'].duplicated().sum()}"
)

print(
    f"Missing taxa      : "
    f"{functional_matrix['Canonical_Species'].isna().sum()}"
)

print(
    f"Total columns     : "
    f"{functional_matrix.shape[1]}"
)


assert len(functional_matrix) == 127

assert (
    functional_matrix[
        "Canonical_Species"
    ].nunique()
    == 127
)

assert (
    functional_matrix[
        "Canonical_Species"
    ].duplicated()
    .sum()
    == 0
)

assert (
    functional_matrix[
        "Canonical_Species"
    ].isna()
    .sum()
    == 0
)


# -----------------------------------------------------------------------------
# 11.2I. Validate core numeric trait fields
# -----------------------------------------------------------------------------

core_numeric_fields = {
    "WD":
        "WD_Value",

    "SLA":
        "SLA_Value",

    "SEED_MASS":
        "Seed_Mass_mg",

    "MAX_HEIGHT":
        "Max_Height_m",
}


numeric_validation_rows = []


for trait_name, field in core_numeric_fields.items():

    values = pd.to_numeric(
        functional_matrix[
            field
        ],
        errors="coerce"
    )

    functional_matrix[
        field
    ] = values


    n_numeric = int(
        values.notna().sum()
    )

    n_missing = int(
        values.isna().sum()
    )

    n_nonpositive = int(
        (
            values.notna()
            &
            (
                values <= 0
            )
        ).sum()
    )


    numeric_validation_rows.append(
        {
            "Trait":
                trait_name,

            "Field":
                field,

            "Numeric_Values":
                n_numeric,

            "Missing_Values":
                n_missing,

            "Nonpositive_Values":
                n_nonpositive,

            "Numeric_Coverage_Percent":
                100 * n_numeric / 127,
        }
    )


numeric_validation = pd.DataFrame(
    numeric_validation_rows
)


print("\nIntegrated numeric-trait validation")
print("-" * 100)

display(
    numeric_validation
)


assert (
    numeric_validation.loc[
        numeric_validation[
            "Trait"
        ].eq("WD"),
        "Numeric_Values"
    ].iloc[0]
    == 127
)

assert (
    numeric_validation.loc[
        numeric_validation[
            "Trait"
        ].eq("SLA"),
        "Numeric_Values"
    ].iloc[0]
    == 127
)

assert (
    numeric_validation.loc[
        numeric_validation[
            "Trait"
        ].eq("SEED_MASS"),
        "Numeric_Values"
    ].iloc[0]
    == 126
)

assert (
    numeric_validation.loc[
        numeric_validation[
            "Trait"
        ].eq("MAX_HEIGHT"),
        "Numeric_Values"
    ].iloc[0]
    == 127
)

assert (
    numeric_validation[
        "Nonpositive_Values"
    ].sum()
    == 0
)


# -----------------------------------------------------------------------------
# 11.2J. Validate expected Seed Mass exception
# -----------------------------------------------------------------------------

seed_missing = functional_matrix[
    functional_matrix[
        "Seed_Mass_mg"
    ].isna()
].copy()


print("\nIntegrated Seed Mass missing-value audit")
print("-" * 100)

display(
    seed_missing[
        [
            "Canonical_Species",
            "Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Seed_Mass_Data_Status",
        ]
    ]
)


assert len(seed_missing) == 1

assert (
    seed_missing[
        "Canonical_Species"
    ].iloc[0]
    ==
    "Musa sapiensis"
)


# -----------------------------------------------------------------------------
# 11.2K. Per-taxon trait completeness
# -----------------------------------------------------------------------------

numeric_trait_columns = [
    "WD_Value",
    "SLA_Value",
    "Seed_Mass_mg",
    "Max_Height_m",
]


functional_matrix[
    "N_Numeric_Core_Traits"
] = (
    functional_matrix[
        numeric_trait_columns
    ]
    .notna()
    .sum(axis=1)
)


functional_matrix[
    "Core_Trait_Completeness_Class"
] = np.select(
    [
        functional_matrix[
            "N_Numeric_Core_Traits"
        ].eq(4),

        functional_matrix[
            "N_Numeric_Core_Traits"
        ].eq(3),
    ],
    [
        "COMPLETE_4_OF_4",
        "COMPLETE_3_OF_4_WITH_DOCUMENTED_EXCEPTION",
    ],
    default="INCOMPLETE_UNEXPECTED"
)


completeness_counts = (
    functional_matrix[
        "Core_Trait_Completeness_Class"
    ]
    .value_counts(dropna=False)
)


print("\nPer-taxon numeric core-trait completeness")
print("-" * 100)

print(
    completeness_counts
)


assert int(
    completeness_counts.get(
        "COMPLETE_4_OF_4",
        0
    )
) == 126

assert int(
    completeness_counts.get(
        "COMPLETE_3_OF_4_WITH_DOCUMENTED_EXCEPTION",
        0
    )
) == 1

assert int(
    completeness_counts.get(
        "INCOMPLETE_UNEXPECTED",
        0
    )
) == 0


# -----------------------------------------------------------------------------
# 11.2L. Identify the documented 3-of-4 taxon
# -----------------------------------------------------------------------------

documented_exception = functional_matrix[
    functional_matrix[
        "Core_Trait_Completeness_Class"
    ].eq(
        "COMPLETE_3_OF_4_WITH_DOCUMENTED_EXCEPTION"
    )
].copy()


print("\nDocumented integration exception")
print("-" * 100)

display(
    documented_exception[
        [
            "Canonical_Species",
            "WD_Value",
            "SLA_Value",
            "Seed_Mass_mg",
            "Max_Height_m",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Data_Status",
            "Core_Trait_Completeness_Class",
        ]
    ]
)


assert len(documented_exception) == 1

assert (
    documented_exception[
        "Canonical_Species"
    ].iloc[0]
    ==
    "Musa sapiensis"
)


# -----------------------------------------------------------------------------
# 11.2M. Unit consistency audit
# -----------------------------------------------------------------------------

unit_fields = {
    "WD":
        "WD_Unit",

    "SLA":
        "SLA_Unit",

    "SEED_MASS":
        "Seed_Mass_Unit",

    "MAX_HEIGHT":
        "Max_Height_Unit",
}


unit_audit_rows = []


for trait_name, field in unit_fields.items():

    observed_units = sorted(
        functional_matrix[
            field
        ]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )


    unit_audit_rows.append(
        {
            "Trait":
                trait_name,

            "Unit_Field":
                field,

            "Observed_Units":
                " | ".join(
                    observed_units
                ),

            "N_Unique_Units":
                len(
                    observed_units
                ),
        }
    )


unit_audit = pd.DataFrame(
    unit_audit_rows
)


print("\nIntegrated unit audit")
print("-" * 100)

display(
    unit_audit
)


assert (
    unit_audit[
        "N_Unique_Units"
    ].eq(1)
    .all()
)


# -----------------------------------------------------------------------------
# 11.2N. Core provenance completeness
# -----------------------------------------------------------------------------

provenance_fields = [
    "WD_Provenance_Class",
    "SLA_Provenance_Class",
    "Seed_Mass_Provenance_Class",
    "Max_Height_Provenance_Class",
]


provenance_missing_rows = []


for field in provenance_fields:

    provenance_missing_rows.append(
        {
            "Field":
                field,

            "Missing":
                int(
                    functional_matrix[
                        field
                    ].isna().sum()
                ),
        }
    )


provenance_missing = pd.DataFrame(
    provenance_missing_rows
)


print("\nIntegrated provenance completeness")
print("-" * 100)

display(
    provenance_missing
)


assert (
    provenance_missing[
        "Missing"
    ].eq(0)
    .all()
)


# -----------------------------------------------------------------------------
# 11.2O. Structural validation report
# -----------------------------------------------------------------------------

structure_report = pd.DataFrame(
    [
        {
            "Validation_Check":
                "Integrated rows",

            "Expected":
                127,

            "Observed":
                len(
                    functional_matrix
                ),

            "Status":
                "PASS"
                if len(
                    functional_matrix
                ) == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Unique canonical taxa",

            "Expected":
                127,

            "Observed":
                functional_matrix[
                    "Canonical_Species"
                ].nunique(),

            "Status":
                "PASS"
                if functional_matrix[
                    "Canonical_Species"
                ].nunique() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Duplicate canonical taxa",

            "Expected":
                0,

            "Observed":
                int(
                    functional_matrix[
                        "Canonical_Species"
                    ].duplicated().sum()
                ),

            "Status":
                "PASS"
                if functional_matrix[
                    "Canonical_Species"
                ].duplicated().sum() == 0
                else "FAIL",
        },

        {
            "Validation_Check":
                "WD numeric values",

            "Expected":
                127,

            "Observed":
                int(
                    functional_matrix[
                        "WD_Value"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if functional_matrix[
                    "WD_Value"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "SLA numeric values",

            "Expected":
                127,

            "Observed":
                int(
                    functional_matrix[
                        "SLA_Value"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if functional_matrix[
                    "SLA_Value"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Seed Mass numeric values",

            "Expected":
                126,

            "Observed":
                int(
                    functional_matrix[
                        "Seed_Mass_mg"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if functional_matrix[
                    "Seed_Mass_mg"
                ].notna().sum() == 126
                else "FAIL",
        },

        {
            "Validation_Check":
                "Maximum Height numeric values",

            "Expected":
                127,

            "Observed":
                int(
                    functional_matrix[
                        "Max_Height_m"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if functional_matrix[
                    "Max_Height_m"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Complete 4-of-4 taxa",

            "Expected":
                126,

            "Observed":
                int(
                    (
                        functional_matrix[
                            "N_Numeric_Core_Traits"
                        ] == 4
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    functional_matrix[
                        "N_Numeric_Core_Traits"
                    ] == 4
                ).sum() == 126
                else "FAIL",
        },

        {
            "Validation_Check":
                "Documented 3-of-4 exception taxa",

            "Expected":
                1,

            "Observed":
                int(
                    (
                        functional_matrix[
                            "N_Numeric_Core_Traits"
                        ] == 3
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    functional_matrix[
                        "N_Numeric_Core_Traits"
                    ] == 3
                ).sum() == 1
                else "FAIL",
        },

        {
            "Validation_Check":
                "Unexpected incomplete taxa",

            "Expected":
                0,

            "Observed":
                int(
                    (
                        functional_matrix[
                            "N_Numeric_Core_Traits"
                        ] < 3
                    ).sum()
                ),

            "Status":
                "PASS"
                if (
                    functional_matrix[
                        "N_Numeric_Core_Traits"
                    ] < 3
                ).sum() == 0
                else "FAIL",
        },
    ]
)


print("\nFour-trait structural validation report")
print("-" * 100)

display(
    structure_report
)


assert (
    structure_report[
        "Status"
    ].eq("PASS")
    .all()
)


# -----------------------------------------------------------------------------
# 11.2P. Export interim merged matrix and structure report
# -----------------------------------------------------------------------------

functional_matrix.to_csv(
    MERGED_INTERIM_PATH,
    index=False
)

structure_report.to_csv(
    STRUCTURE_REPORT_PATH,
    index=False
)


print("\nInterim integration exports")
print("-" * 100)

print(
    f"Merged interim matrix : "
    f"{MERGED_INTERIM_PATH}"
)

print(
    f"Structure report      : "
    f"{STRUCTURE_REPORT_PATH}"
)


# -----------------------------------------------------------------------------
# 11.2Q. Integration interpretation
# -----------------------------------------------------------------------------

print("\nIntegration interpretation")
print("-" * 100)

print(
    """
The four authoritative trait extracts have now been merged explicitly
by Canonical_Species using one-to-one joins.

The integrated matrix contains exactly 127 canonical taxa.

Numeric core-trait coverage is:

    WD             127/127
    SLA            127/127
    Seed Mass      126/127
    Maximum Height 127/127

Exactly 126 taxa have all four numeric core traits.

Musa sapiensis has three numeric traits and one formally documented
Seed Mass biological nonapplicability. This is an expected ecological
exception, not an unresolved missing-data error.

No imputation has been introduced during integration.
"""
)


# -----------------------------------------------------------------------------
# 11.2R. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 11.2 FOUR-TRAIT MERGE AND STRUCTURAL VALIDATION COMPLETE")
print("=" * 100)

print(
    f"Integrated taxa                     : "
    f"{len(functional_matrix)}"
)

print(
    f"Unique canonical taxa               : "
    f"{functional_matrix['Canonical_Species'].nunique()}"
)

print(
    f"Complete four-numeric-trait taxa    : "
    f"{(functional_matrix['N_Numeric_Core_Traits'] == 4).sum()}"
)

print(
    f"Documented three-trait exception    : "
    f"{(functional_matrix['N_Numeric_Core_Traits'] == 3).sum()}"
)

print(
    f"Unexpected incomplete taxa          : "
    f"{(functional_matrix['N_Numeric_Core_Traits'] < 3).sum()}"
)

print(
    f"Failed structural validation checks : "
    f"{(structure_report['Status'] != 'PASS').sum()}"
)

print(
    "Integrated matrix status            : STRUCTURALLY VALID"
)

print(
    "Next step                           : "
    "Section 11.3 — integrated provenance, quality and trait-distribution audit."
)
```

```python
# =============================================================================
# SECTION 11.3 — INTEGRATED PROVENANCE, QUALITY AND TRAIT-DISTRIBUTION AUDIT
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 11.3 — INTEGRATED PROVENANCE, QUALITY AND TRAIT-DISTRIBUTION AUDIT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 11.3A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

INTEGRATION_DIR = (
    OUTPUT_DIR /
    "integration"
)

MERGED_INTERIM_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_four_trait_merged_interim.csv"
)

PROVENANCE_AUDIT_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_integrated_provenance_audit.csv"
)

QUALITY_AUDIT_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_integrated_quality_audit.csv"
)

DISTRIBUTION_AUDIT_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_trait_distribution_audit.csv"
)

EVIDENCE_PROFILE_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_taxon_evidence_profile.csv"
)


# -----------------------------------------------------------------------------
# 11.3B. Load structurally validated integrated matrix
# -----------------------------------------------------------------------------

functional_matrix = pd.read_csv(
    MERGED_INTERIM_PATH,
    low_memory=False
)


print("\nIntegrated matrix loaded")
print("-" * 100)

print(
    f"Rows        : "
    f"{len(functional_matrix)}"
)

print(
    f"Unique taxa : "
    f"{functional_matrix['Canonical_Species'].nunique()}"
)

print(
    f"Columns     : "
    f"{functional_matrix.shape[1]}"
)


assert len(functional_matrix) == 127

assert (
    functional_matrix[
        "Canonical_Species"
    ].nunique()
    == 127
)


# -----------------------------------------------------------------------------
# 11.3C. Core trait configuration
# -----------------------------------------------------------------------------

trait_config = {
    "WD": {
        "value":
            "WD_Value",

        "unit":
            "WD_Unit",

        "assignment":
            "WD_Assignment_Level",

        "provenance":
            "WD_Provenance_Class",

        "quality":
            "WD_Quality_Flag",

        "source":
            "WD_Source",
    },

    "SLA": {
        "value":
            "SLA_Value",

        "unit":
            "SLA_Unit",

        "assignment":
            "SLA_Assignment_Level",

        "provenance":
            "SLA_Provenance_Class",

        "quality":
            "SLA_Quality_Flag",

        "source":
            "SLA_Source",
    },

    "SEED_MASS": {
        "value":
            "Seed_Mass_mg",

        "unit":
            "Seed_Mass_Unit",

        "assignment":
            "Seed_Mass_Assignment_Level",

        "provenance":
            "Seed_Mass_Provenance_Class",

        "quality":
            "Seed_Mass_Quality_Flag",

        "source":
            "Seed_Mass_Source",
    },

    "MAX_HEIGHT": {
        "value":
            "Max_Height_m",

        "unit":
            "Max_Height_Unit",

        "assignment":
            "Max_Height_Assignment_Level",

        "provenance":
            "Max_Height_Provenance_Class",

        "quality":
            "Max_Height_Quality_Flag",

        "source":
            "Max_Height_Source",
    },
}


# -----------------------------------------------------------------------------
# 11.3D. Provenance-class audit
# -----------------------------------------------------------------------------

provenance_rows = []


for trait_name, cfg in trait_config.items():

    counts = (
        functional_matrix[
            cfg["provenance"]
        ]
        .value_counts(
            dropna=False
        )
    )


    for provenance_class, n_taxa in counts.items():

        provenance_rows.append(
            {
                "Trait":
                    trait_name,

                "Provenance_Class":
                    provenance_class,

                "N_Taxa":
                    int(n_taxa),

                "Percent_of_127":
                    100 * int(n_taxa) / 127,
            }
        )


provenance_audit = pd.DataFrame(
    provenance_rows
)


provenance_audit = (
    provenance_audit
    .sort_values(
        [
            "Trait",
            "N_Taxa",
            "Provenance_Class",
        ],
        ascending=[
            True,
            False,
            True,
        ]
    )
    .reset_index(
        drop=True
    )
)


print("\nIntegrated provenance-class audit")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 200,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        provenance_audit
    )


# Validate all 127 taxa are represented for each trait.

provenance_totals = (
    provenance_audit
    .groupby(
        "Trait"
    )[
        "N_Taxa"
    ]
    .sum()
)


print("\nProvenance totals by trait")
print("-" * 100)

print(
    provenance_totals
)


assert (
    provenance_totals
    .eq(127)
    .all()
)


# -----------------------------------------------------------------------------
# 11.3E. Assignment-level audit
# -----------------------------------------------------------------------------

assignment_rows = []


for trait_name, cfg in trait_config.items():

    counts = (
        functional_matrix[
            cfg["assignment"]
        ]
        .value_counts(
            dropna=False
        )
    )


    for assignment_level, n_taxa in counts.items():

        assignment_rows.append(
            {
                "Trait":
                    trait_name,

                "Assignment_Level":
                    assignment_level,

                "N_Taxa":
                    int(n_taxa),

                "Percent_of_127":
                    100 * int(n_taxa) / 127,
            }
        )


assignment_audit = pd.DataFrame(
    assignment_rows
)


assignment_audit = (
    assignment_audit
    .sort_values(
        [
            "Trait",
            "N_Taxa",
            "Assignment_Level",
        ],
        ascending=[
            True,
            False,
            True,
        ]
    )
    .reset_index(
        drop=True
    )
)


print("\nIntegrated assignment-level audit")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 200,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        assignment_audit
    )


assignment_totals = (
    assignment_audit
    .groupby(
        "Trait"
    )[
        "N_Taxa"
    ]
    .sum()
)


assert (
    assignment_totals
    .eq(127)
    .all()
)


# -----------------------------------------------------------------------------
# 11.3F. Quality-flag audit
# -----------------------------------------------------------------------------

quality_rows = []


for trait_name, cfg in trait_config.items():

    counts = (
        functional_matrix[
            cfg["quality"]
        ]
        .value_counts(
            dropna=False
        )
    )


    for quality_flag, n_taxa in counts.items():

        quality_rows.append(
            {
                "Trait":
                    trait_name,

                "Quality_Flag":
                    quality_flag,

                "N_Taxa":
                    int(n_taxa),

                "Percent_of_127":
                    100 * int(n_taxa) / 127,
            }
        )


quality_audit = pd.DataFrame(
    quality_rows
)


quality_audit = (
    quality_audit
    .sort_values(
        [
            "Trait",
            "N_Taxa",
            "Quality_Flag",
        ],
        ascending=[
            True,
            False,
            True,
        ]
    )
    .reset_index(
        drop=True
    )
)


print("\nIntegrated quality-flag audit")
print("-" * 100)

with pd.option_context(
    "display.max_rows", 200,
    "display.max_columns", None,
    "display.max_colwidth", 120,
):
    display(
        quality_audit
    )


quality_totals = (
    quality_audit
    .groupby(
        "Trait"
    )[
        "N_Taxa"
    ]
    .sum()
)


assert (
    quality_totals
    .eq(127)
    .all()
)


# -----------------------------------------------------------------------------
# 11.3G. Source diversity audit
# -----------------------------------------------------------------------------

source_rows = []


for trait_name, cfg in trait_config.items():

    source_series = (
        functional_matrix[
            cfg["source"]
        ]
        .astype("string")
    )


    n_nonmissing = int(
        source_series.notna().sum()
    )

    n_unique = int(
        source_series.dropna().nunique()
    )


    source_rows.append(
        {
            "Trait":
                trait_name,

            "Source_Field":
                cfg["source"],

            "N_Nonmissing_Sources":
                n_nonmissing,

            "N_Unique_Source_Labels":
                n_unique,
        }
    )


source_audit = pd.DataFrame(
    source_rows
)


print("\nIntegrated source diversity audit")
print("-" * 100)

display(
    source_audit
)


# -----------------------------------------------------------------------------
# 11.3H. Trait-distribution audit
# -----------------------------------------------------------------------------

distribution_rows = []


for trait_name, cfg in trait_config.items():

    values = pd.to_numeric(
        functional_matrix[
            cfg["value"]
        ],
        errors="coerce"
    )


    numeric_values = values.dropna()


    distribution_rows.append(
        {
            "Trait":
                trait_name,

            "Value_Field":
                cfg["value"],

            "Unit":
                " | ".join(
                    sorted(
                        functional_matrix[
                            cfg["unit"]
                        ]
                        .dropna()
                        .astype(str)
                        .unique()
                        .tolist()
                    )
                ),

            "N_Numeric":
                int(
                    numeric_values.size
                ),

            "N_Missing":
                int(
                    values.isna().sum()
                ),

            "Minimum":
                float(
                    numeric_values.min()
                ),

            "Q1":
                float(
                    numeric_values.quantile(0.25)
                ),

            "Median":
                float(
                    numeric_values.median()
                ),

            "Mean":
                float(
                    numeric_values.mean()
                ),

            "Q3":
                float(
                    numeric_values.quantile(0.75)
                ),

            "Maximum":
                float(
                    numeric_values.max()
                ),

            "SD":
                float(
                    numeric_values.std()
                ),

            "IQR":
                float(
                    numeric_values.quantile(0.75)
                    -
                    numeric_values.quantile(0.25)
                ),
        }
    )


distribution_audit = pd.DataFrame(
    distribution_rows
)


print("\nIntegrated trait-distribution audit")
print("-" * 100)

display(
    distribution_audit
)


# -----------------------------------------------------------------------------
# 11.3I. Broad outlier diagnostics using IQR
# -----------------------------------------------------------------------------

outlier_rows = []


for trait_name, cfg in trait_config.items():

    values = pd.to_numeric(
        functional_matrix[
            cfg["value"]
        ],
        errors="coerce"
    )


    numeric_values = values.dropna()


    q1 = numeric_values.quantile(
        0.25
    )

    q3 = numeric_values.quantile(
        0.75
    )

    iqr = q3 - q1


    lower_bound = (
        q1
        -
        1.5 * iqr
    )

    upper_bound = (
        q3
        +
        1.5 * iqr
    )


    outlier_mask = (
        values.notna()
        &
        (
            (values < lower_bound)
            |
            (values > upper_bound)
        )
    )


    outlier_subset = functional_matrix.loc[
        outlier_mask,
        [
            "Canonical_Species",
            cfg["value"],
            cfg["assignment"],
            cfg["provenance"],
            cfg["quality"],
        ]
    ].copy()


    if not outlier_subset.empty:

        for _, row in outlier_subset.iterrows():

            value = float(
                row[
                    cfg["value"]
                ]
            )


            outlier_rows.append(
                {
                    "Trait":
                        trait_name,

                    "Canonical_Species":
                        row[
                            "Canonical_Species"
                        ],

                    "Observed_Value":
                        value,

                    "Lower_IQR_Bound":
                        float(
                            lower_bound
                        ),

                    "Upper_IQR_Bound":
                        float(
                            upper_bound
                        ),

                    "Direction":
                        (
                            "LOW"
                            if value < lower_bound
                            else "HIGH"
                        ),

                    "Assignment_Level":
                        row[
                            cfg["assignment"]
                        ],

                    "Provenance_Class":
                        row[
                            cfg["provenance"]
                        ],

                    "Quality_Flag":
                        row[
                            cfg["quality"]
                        ],
                }
            )


outlier_audit = pd.DataFrame(
    outlier_rows
)


print("\nIQR-based broad outlier diagnostic")
print("-" * 100)

if outlier_audit.empty:

    print(
        "No IQR outliers detected."
    )

else:

    print(
        f"Outlier records detected : "
        f"{len(outlier_audit)}"
    )

    with pd.option_context(
        "display.max_rows", 300,
        "display.max_columns", None,
        "display.max_colwidth", 100,
    ):
        display(
            outlier_audit
        )


print(
    "\nImportant: IQR outliers are diagnostic only and are NOT "
    "automatically treated as erroneous trait values."
)


# -----------------------------------------------------------------------------
# 11.3J. Define broad evidence-strength classes
# -----------------------------------------------------------------------------

# This is an integration-level summary only.
# It does not alter original provenance classes.

def classify_evidence_strength(
    trait_name,
    provenance_class,
    assignment_level
):

    provenance_text = str(
        provenance_class
    ).upper()

    assignment_text = str(
        assignment_level
    ).upper()


    # Explicit biological special case.
    if (
        "SPECIAL_BIOLOGICAL_CASE"
        in provenance_text
        or
        "BIOLOGICAL_NONAPPLICABILITY"
        in assignment_text
    ):
        return "SPECIAL_BIOLOGICAL_CASE"


    # Lower-confidence estimate/proxy classes.
    if (
        "GLOBAL" in provenance_text
        or
        "FAMILY" in provenance_text
        or
        "GENUS" in provenance_text
        or
        "FALLBACK" in provenance_text
        or
        "PROXY" in provenance_text
        or
        "SPECIES_COMPLEX" in provenance_text
        or
        assignment_text in [
            "GENUS",
            "FAMILY",
            "SPECIES_COMPLEX",
        ]
    ):
        return "ESTIMATED_OR_SPECIALIZED_EVIDENCE"


    # Otherwise retain as direct/reconciled species evidence.
    return "DIRECT_OR_RECONCILED_SPECIES_EVIDENCE"


# -----------------------------------------------------------------------------
# 11.3K. Build per-taxon integrated evidence profile
# -----------------------------------------------------------------------------

evidence_profile = functional_matrix[
    [
        "Canonical_Species",
        "N_Numeric_Core_Traits",
        "Core_Trait_Completeness_Class",
    ]
].copy()


for trait_name, cfg in trait_config.items():

    evidence_strength_field = (
        f"{trait_name}_Evidence_Strength"
    )


    evidence_profile[
        evidence_strength_field
    ] = [
        classify_evidence_strength(
            trait_name,
            provenance,
            assignment
        )
        for provenance, assignment in zip(
            functional_matrix[
                cfg["provenance"]
            ],
            functional_matrix[
                cfg["assignment"]
            ],
        )
    ]


# Count estimated/specialized evidence across the four traits.

evidence_strength_fields = [
    "WD_Evidence_Strength",
    "SLA_Evidence_Strength",
    "SEED_MASS_Evidence_Strength",
    "MAX_HEIGHT_Evidence_Strength",
]


evidence_profile[
    "N_Direct_or_Reconciled_Traits"
] = (
    evidence_profile[
        evidence_strength_fields
    ]
    .eq(
        "DIRECT_OR_RECONCILED_SPECIES_EVIDENCE"
    )
    .sum(
        axis=1
    )
)


evidence_profile[
    "N_Estimated_or_Specialized_Traits"
] = (
    evidence_profile[
        evidence_strength_fields
    ]
    .eq(
        "ESTIMATED_OR_SPECIALIZED_EVIDENCE"
    )
    .sum(
        axis=1
    )
)


evidence_profile[
    "N_Special_Biological_Cases"
] = (
    evidence_profile[
        evidence_strength_fields
    ]
    .eq(
        "SPECIAL_BIOLOGICAL_CASE"
    )
    .sum(
        axis=1
    )
)


# -----------------------------------------------------------------------------
# 11.3L. Integrated evidence-profile classes
# -----------------------------------------------------------------------------

evidence_profile[
    "Integrated_Evidence_Profile"
] = np.select(
    [
        (
            evidence_profile[
                "N_Direct_or_Reconciled_Traits"
            ] == 4
        ),

        (
            evidence_profile[
                "N_Direct_or_Reconciled_Traits"
            ] == 3
        )
        &
        (
            evidence_profile[
                "N_Special_Biological_Cases"
            ] == 1
        ),

        (
            evidence_profile[
                "N_Estimated_or_Specialized_Traits"
            ] == 1
        ),

        (
            evidence_profile[
                "N_Estimated_or_Specialized_Traits"
            ] >= 2
        ),
    ],
    [
        "ALL_FOUR_DIRECT_OR_RECONCILED",

        "THREE_DIRECT_PLUS_SPECIAL_BIOLOGICAL_CASE",

        "ONE_TRAIT_ESTIMATED_OR_SPECIALIZED",

        "MULTIPLE_TRAITS_ESTIMATED_OR_SPECIALIZED",
    ],
    default="MIXED_OTHER",
)


profile_counts = (
    evidence_profile[
        "Integrated_Evidence_Profile"
    ]
    .value_counts(
        dropna=False
    )
)


print("\nIntegrated taxon evidence-profile summary")
print("-" * 100)

print(
    profile_counts
)


assert int(
    profile_counts.sum()
) == 127


# -----------------------------------------------------------------------------
# 11.3M. Taxa depending on multiple lower-confidence evidence streams
# -----------------------------------------------------------------------------

multi_estimate_taxa = evidence_profile[
    evidence_profile[
        "N_Estimated_or_Specialized_Traits"
    ] >= 2
].copy()


print("\nTaxa with two or more estimated/specialized trait assignments")
print("-" * 100)

print(
    f"N taxa : "
    f"{len(multi_estimate_taxa)}"
)


if not multi_estimate_taxa.empty:

    with pd.option_context(
        "display.max_rows", 200,
        "display.max_columns", None,
        "display.max_colwidth", 100,
    ):
        display(
            multi_estimate_taxa[
                [
                    "Canonical_Species",
                    "WD_Evidence_Strength",
                    "SLA_Evidence_Strength",
                    "SEED_MASS_Evidence_Strength",
                    "MAX_HEIGHT_Evidence_Strength",
                    "N_Direct_or_Reconciled_Traits",
                    "N_Estimated_or_Specialized_Traits",
                    "N_Special_Biological_Cases",
                    "Integrated_Evidence_Profile",
                ]
            ]
        )


# -----------------------------------------------------------------------------
# 11.3N. Musa sapiensis integration audit
# -----------------------------------------------------------------------------

musa_audit = functional_matrix[
    functional_matrix[
        "Canonical_Species"
    ].eq(
        "Musa sapiensis"
    )
].copy()


print("\nMusa sapiensis integrated audit")
print("-" * 100)

display(
    musa_audit[
        [
            "Canonical_Species",
            "WD_Value",
            "WD_Provenance_Class",
            "SLA_Value",
            "SLA_Provenance_Class",
            "Seed_Mass_mg",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Data_Status",
            "Max_Height_m",
            "Max_Height_Provenance_Class",
            "N_Numeric_Core_Traits",
            "Core_Trait_Completeness_Class",
        ]
    ]
)


assert len(musa_audit) == 1


# -----------------------------------------------------------------------------
# 11.3O. Integrated audit validation
# -----------------------------------------------------------------------------

audit_validation = pd.DataFrame(
    [
        {
            "Validation_Check":
                "Provenance totals equal 127 per trait",

            "Status":
                "PASS"
                if provenance_totals.eq(127).all()
                else "FAIL",
        },

        {
            "Validation_Check":
                "Assignment totals equal 127 per trait",

            "Status":
                "PASS"
                if assignment_totals.eq(127).all()
                else "FAIL",
        },

        {
            "Validation_Check":
                "Quality totals equal 127 per trait",

            "Status":
                "PASS"
                if quality_totals.eq(127).all()
                else "FAIL",
        },

        {
            "Validation_Check":
                "Evidence profile covers 127 taxa",

            "Status":
                "PASS"
                if len(evidence_profile) == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Musa documented exception retained",

            "Status":
                "PASS"
                if (
                    len(musa_audit) == 1
                    and
                    musa_audit[
                        "Seed_Mass_mg"
                    ].isna().all()
                )
                else "FAIL",
        },
    ]
)


print("\nIntegrated audit validation")
print("-" * 100)

display(
    audit_validation
)


assert (
    audit_validation[
        "Status"
    ].eq(
        "PASS"
    )
    .all()
)


# -----------------------------------------------------------------------------
# 11.3P. Export integrated audit products
# -----------------------------------------------------------------------------

provenance_audit.to_csv(
    PROVENANCE_AUDIT_PATH,
    index=False
)

quality_audit.to_csv(
    QUALITY_AUDIT_PATH,
    index=False
)

distribution_audit.to_csv(
    DISTRIBUTION_AUDIT_PATH,
    index=False
)

evidence_profile.to_csv(
    EVIDENCE_PROFILE_PATH,
    index=False
)


if not outlier_audit.empty:

    OUTLIER_AUDIT_PATH = (
        INTEGRATION_DIR /
        "omo_functional_traits_iqr_outlier_diagnostic.csv"
    )

    outlier_audit.to_csv(
        OUTLIER_AUDIT_PATH,
        index=False
    )


print("\nIntegrated audit exports")
print("-" * 100)

print(
    f"Provenance audit    : "
    f"{PROVENANCE_AUDIT_PATH}"
)

print(
    f"Quality audit       : "
    f"{QUALITY_AUDIT_PATH}"
)

print(
    f"Distribution audit  : "
    f"{DISTRIBUTION_AUDIT_PATH}"
)

print(
    f"Evidence profile    : "
    f"{EVIDENCE_PROFILE_PATH}"
)

if not outlier_audit.empty:

    print(
        f"Outlier diagnostic  : "
        f"{OUTLIER_AUDIT_PATH}"
    )


# -----------------------------------------------------------------------------
# 11.3Q. Interpretation safeguard
# -----------------------------------------------------------------------------

print("\nInterpretation safeguard")
print("-" * 100)

print(
    """
This section audits provenance, assignment level, quality flags,
distributional structure and evidence dependence across the four
integrated traits.

No trait value has been altered.

IQR outliers are treated only as distributional diagnostics. Extreme
trait values may be biologically valid and must not be removed solely
because they lie outside statistical IQR fences.

The integrated evidence-strength classes are summary labels only.
Original trait-specific provenance classes remain authoritative.

The Seed Mass biological nonapplicability for Musa sapiensis remains
explicit and unchanged.
"""
)


# -----------------------------------------------------------------------------
# 11.3R. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 11.3 INTEGRATED PROVENANCE, QUALITY AND DISTRIBUTION AUDIT COMPLETE")
print("=" * 100)

print(
    f"Integrated taxa audited              : "
    f"{len(functional_matrix)}"
)

print(
    f"Traits audited                       : "
    f"{len(trait_config)}"
)

print(
    f"Provenance audit rows                : "
    f"{len(provenance_audit)}"
)

print(
    f"Quality audit rows                   : "
    f"{len(quality_audit)}"
)

print(
    f"Trait distribution rows              : "
    f"{len(distribution_audit)}"
)

print(
    f"IQR diagnostic records               : "
    f"{len(outlier_audit)}"
)

print(
    f"Taxa with >=2 estimated/specialized traits : "
    f"{len(multi_estimate_taxa)}"
)

print(
    f"Failed integrated audit checks       : "
    f"{(audit_validation['Status'] != 'PASS').sum()}"
)

print(
    "Integrated trait values modified     : NO"
)

print(
    "Next step                            : "
    "Section 11.4 — final authoritative functional-trait matrix, "
    "closure validation and Notebook 07B export."
)
```

```python
# =============================================================================
# SECTION 11.4 — FINAL AUTHORITATIVE FUNCTIONAL-TRAIT MATRIX,
# CLOSURE VALIDATION AND NOTEBOOK 07B EXPORT
# =============================================================================

from pathlib import Path
import numpy as np
import pandas as pd


print("=" * 100)
print("SECTION 11.4 — FINAL AUTHORITATIVE FUNCTIONAL-TRAIT MATRIX, CLOSURE VALIDATION AND NOTEBOOK 07B EXPORT")
print("=" * 100)


# -----------------------------------------------------------------------------
# 11.4A. Paths
# -----------------------------------------------------------------------------

PROJECT_ROOT = Path(
    r"C:\Peter\SDM_data\omo_ecodb\omo-ecological-informatics"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "functional_traits"
)

INTEGRATION_DIR = (
    OUTPUT_DIR /
    "integration"
)

INTEGRATION_DIR.mkdir(
    parents=True,
    exist_ok=True
)


MERGED_INTERIM_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_four_trait_merged_interim.csv"
)

EVIDENCE_PROFILE_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_taxon_evidence_profile.csv"
)

PROVENANCE_AUDIT_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_integrated_provenance_audit.csv"
)

DISTRIBUTION_AUDIT_PATH = (
    INTEGRATION_DIR /
    "omo_functional_traits_trait_distribution_audit.csv"
)

WD_FINAL_PATH = (
    OUTPUT_DIR /
    "omo_wood_density_final.csv"
)

SLA_FINAL_PATH = (
    OUTPUT_DIR /
    "omo_sla_final.csv"
)

SEED_FINAL_PATH = (
    OUTPUT_DIR /
    "omo_seed_mass_final.csv"
)

HEIGHT_FINAL_PATH = (
    OUTPUT_DIR /
    "omo_max_height_final.csv"
)


FINAL_MATRIX_PATH = (
    OUTPUT_DIR /
    "omo_functional_traits_final.csv"
)

CORE_MATRIX_PATH = (
    OUTPUT_DIR /
    "omo_functional_traits_core_matrix.csv"
)

FINAL_VALIDATION_PATH = (
    OUTPUT_DIR /
    "omo_functional_traits_final_validation_report.csv"
)

FINAL_COMPLETENESS_PATH = (
    OUTPUT_DIR /
    "omo_functional_traits_final_completeness_summary.csv"
)

FINAL_PROVENANCE_PATH = (
    OUTPUT_DIR /
    "omo_functional_traits_final_provenance_summary.csv"
)

NOTEBOOK_CLOSURE_PATH = (
    OUTPUT_DIR /
    "omo_notebook_07B_closure_summary.csv"
)


# -----------------------------------------------------------------------------
# 11.4B. Load validated integrated products
# -----------------------------------------------------------------------------

integrated = pd.read_csv(
    MERGED_INTERIM_PATH,
    low_memory=False
)

evidence_profile = pd.read_csv(
    EVIDENCE_PROFILE_PATH,
    low_memory=False
)

provenance_audit = pd.read_csv(
    PROVENANCE_AUDIT_PATH,
    low_memory=False
)

distribution_audit = pd.read_csv(
    DISTRIBUTION_AUDIT_PATH,
    low_memory=False
)


print("\nValidated integration products loaded")
print("-" * 100)

print(
    f"Integrated matrix rows : "
    f"{len(integrated)}"
)

print(
    f"Evidence-profile rows  : "
    f"{len(evidence_profile)}"
)

print(
    f"Provenance audit rows  : "
    f"{len(provenance_audit)}"
)

print(
    f"Distribution rows      : "
    f"{len(distribution_audit)}"
)


assert len(integrated) == 127
assert integrated["Canonical_Species"].nunique() == 127
assert len(evidence_profile) == 127


# -----------------------------------------------------------------------------
# 11.4C. Load four authoritative trait source tables
# -----------------------------------------------------------------------------

wd_final = pd.read_csv(
    WD_FINAL_PATH,
    low_memory=False
)

sla_final = pd.read_csv(
    SLA_FINAL_PATH,
    low_memory=False
)

seed_final = pd.read_csv(
    SEED_FINAL_PATH,
    low_memory=False
)

height_final = pd.read_csv(
    HEIGHT_FINAL_PATH,
    low_memory=False
)


for df in [
    wd_final,
    sla_final,
    seed_final,
    height_final,
]:

    assert len(df) == 127
    assert df["Canonical_Species"].nunique() == 127


print("\nAuthoritative trait sources loaded")
print("-" * 100)

print("WD rows          : 127")
print("SLA rows         : 127")
print("Seed Mass rows   : 127")
print("Maximum Height   : 127")


# -----------------------------------------------------------------------------
# 11.4D. Attach final integrated evidence profile
# -----------------------------------------------------------------------------

profile_fields = [
    "Canonical_Species",
    "WD_Evidence_Strength",
    "SLA_Evidence_Strength",
    "SEED_MASS_Evidence_Strength",
    "MAX_HEIGHT_Evidence_Strength",
    "N_Direct_or_Reconciled_Traits",
    "N_Estimated_or_Specialized_Traits",
    "N_Special_Biological_Cases",
    "Integrated_Evidence_Profile",
]


final_matrix = integrated.merge(
    evidence_profile[
        profile_fields
    ],
    on="Canonical_Species",
    how="left",
    validate="one_to_one"
)


assert len(final_matrix) == 127
assert final_matrix["Integrated_Evidence_Profile"].notna().all()


# -----------------------------------------------------------------------------
# 11.4E. Standardize final core numeric fields
# -----------------------------------------------------------------------------

numeric_fields = [
    "WD_Value",
    "SLA_Value",
    "Seed_Mass_mg",
    "Max_Height_m",
]


for field in numeric_fields:

    final_matrix[field] = pd.to_numeric(
        final_matrix[field],
        errors="coerce"
    )


# -----------------------------------------------------------------------------
# 11.4F. Final trait-completeness fields
# -----------------------------------------------------------------------------

final_matrix[
    "N_Numeric_Core_Traits"
] = (
    final_matrix[
        numeric_fields
    ]
    .notna()
    .sum(axis=1)
)


final_matrix[
    "Final_Core_Trait_Status"
] = np.select(
    [
        final_matrix[
            "N_Numeric_Core_Traits"
        ].eq(4),

        (
            final_matrix[
                "Canonical_Species"
            ].eq(
                "Musa sapiensis"
            )
            &
            final_matrix[
                "N_Numeric_Core_Traits"
            ].eq(3)
            &
            final_matrix[
                "Seed_Mass_Provenance_Class"
            ].eq(
                "SPECIAL_BIOLOGICAL_CASE"
            )
        ),
    ],
    [
        "COMPLETE_4_OF_4",
        "COMPLETE_WITH_DOCUMENTED_BIOLOGICAL_EXCEPTION",
    ],
    default="UNEXPECTED_INCOMPLETE"
)


final_status_counts = (
    final_matrix[
        "Final_Core_Trait_Status"
    ]
    .value_counts(
        dropna=False
    )
)


print("\nFinal core-trait completeness")
print("-" * 100)

print(
    final_status_counts
)


assert int(
    final_status_counts.get(
        "COMPLETE_4_OF_4",
        0
    )
) == 126

assert int(
    final_status_counts.get(
        "COMPLETE_WITH_DOCUMENTED_BIOLOGICAL_EXCEPTION",
        0
    )
) == 1

assert int(
    final_status_counts.get(
        "UNEXPECTED_INCOMPLETE",
        0
    )
) == 0


# -----------------------------------------------------------------------------
# 11.4G. Final unit validation
# -----------------------------------------------------------------------------

expected_units = {
    "WD_Unit":
        "g cm-3",

    "SLA_Unit":
        "mm2 mg-1",

    "Seed_Mass_Unit":
        "mg",

    "Max_Height_Unit":
        "m",
}


print("\nFinal unit validation")
print("-" * 100)


for field, expected_unit in expected_units.items():

    observed_units = (
        final_matrix[
            field
        ]
        .dropna()
        .astype(str)
        .unique()
        .tolist()
    )

    print(
        f"{field:<22}: "
        f"{observed_units}"
    )

    assert set(observed_units) == {
        expected_unit
    }


# -----------------------------------------------------------------------------
# 11.4H. Final positive-value validation
# -----------------------------------------------------------------------------

positive_checks = {
    "WD_Value":
        127,

    "SLA_Value":
        127,

    "Seed_Mass_mg":
        126,

    "Max_Height_m":
        127,
}


print("\nFinal positive-value validation")
print("-" * 100)


for field, expected_positive_n in positive_checks.items():

    n_positive = int(
        (
            final_matrix[
                field
            ].notna()
            &
            (
                final_matrix[
                    field
                ] > 0
            )
        ).sum()
    )

    print(
        f"{field:<18}: "
        f"{n_positive}"
    )

    assert n_positive == expected_positive_n


# -----------------------------------------------------------------------------
# 11.4I. Validate Seed Mass biological exception
# -----------------------------------------------------------------------------

seed_exception = final_matrix[
    final_matrix[
        "Seed_Mass_mg"
    ].isna()
].copy()


print("\nFinal Seed Mass biological exception")
print("-" * 100)

display(
    seed_exception[
        [
            "Canonical_Species",
            "Seed_Mass_mg",
            "Seed_Mass_Assignment_Level",
            "Seed_Mass_Provenance_Class",
            "Seed_Mass_Quality_Flag",
            "Seed_Mass_Data_Status",
            "Final_Core_Trait_Status",
        ]
    ]
)


assert len(seed_exception) == 1

assert (
    seed_exception[
        "Canonical_Species"
    ].iloc[0]
    ==
    "Musa sapiensis"
)

assert (
    seed_exception[
        "Seed_Mass_Provenance_Class"
    ].iloc[0]
    ==
    "SPECIAL_BIOLOGICAL_CASE"
)


# -----------------------------------------------------------------------------
# 11.4J. Cross-check integrated numeric values against authoritative sources
# -----------------------------------------------------------------------------

# WD
wd_source_check = (
    wd_final[
        [
            "Canonical_Species",
            "Standardized_Trait_Value",
        ]
    ]
    .rename(
        columns={
            "Standardized_Trait_Value":
                "Authoritative_WD"
        }
    )
)


# SLA
sla_source_check = (
    sla_final[
        [
            "Canonical_Species",
            "Standardized_SLA_Value",
        ]
    ]
    .rename(
        columns={
            "Standardized_SLA_Value":
                "Authoritative_SLA"
        }
    )
)


# Seed Mass
seed_source_check = (
    seed_final[
        [
            "Canonical_Species",
            "Seed_Mass_mg",
        ]
    ]
    .rename(
        columns={
            "Seed_Mass_mg":
                "Authoritative_Seed_Mass"
        }
    )
)


# Maximum Height
height_source_check = (
    height_final[
        [
            "Canonical_Species",
            "Max_Height_m",
        ]
    ]
    .rename(
        columns={
            "Max_Height_m":
                "Authoritative_Max_Height"
        }
    )
)


source_check = (
    final_matrix[
        [
            "Canonical_Species",
            "WD_Value",
            "SLA_Value",
            "Seed_Mass_mg",
            "Max_Height_m",
        ]
    ]
    .merge(
        wd_source_check,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
    .merge(
        sla_source_check,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
    .merge(
        seed_source_check,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
    .merge(
        height_source_check,
        on="Canonical_Species",
        how="left",
        validate="one_to_one"
    )
)


source_check[
    "WD_Matches_Authoritative"
] = np.isclose(
    pd.to_numeric(
        source_check[
            "WD_Value"
        ],
        errors="coerce"
    ),
    pd.to_numeric(
        source_check[
            "Authoritative_WD"
        ],
        errors="coerce"
    ),
    rtol=1e-12,
    atol=1e-12
)


source_check[
    "SLA_Matches_Authoritative"
] = np.isclose(
    pd.to_numeric(
        source_check[
            "SLA_Value"
        ],
        errors="coerce"
    ),
    pd.to_numeric(
        source_check[
            "Authoritative_SLA"
        ],
        errors="coerce"
    ),
    rtol=1e-12,
    atol=1e-12
)


seed_integrated_numeric = pd.to_numeric(
    source_check[
        "Seed_Mass_mg"
    ],
    errors="coerce"
)

seed_authoritative_numeric = pd.to_numeric(
    source_check[
        "Authoritative_Seed_Mass"
    ],
    errors="coerce"
)


source_check[
    "Seed_Mass_Matches_Authoritative"
] = (
    (
        seed_integrated_numeric.isna()
        &
        seed_authoritative_numeric.isna()
    )
    |
    np.isclose(
        seed_integrated_numeric.fillna(0),
        seed_authoritative_numeric.fillna(0),
        rtol=1e-12,
        atol=1e-12
    )
)


source_check[
    "Max_Height_Matches_Authoritative"
] = np.isclose(
    pd.to_numeric(
        source_check[
            "Max_Height_m"
        ],
        errors="coerce"
    ),
    pd.to_numeric(
        source_check[
            "Authoritative_Max_Height"
        ],
        errors="coerce"
    ),
    rtol=1e-12,
    atol=1e-12
)


print("\nAuthoritative-source value agreement")
print("-" * 100)

print(
    "WD agreement         :",
    source_check[
        "WD_Matches_Authoritative"
    ].all()
)

print(
    "SLA agreement        :",
    source_check[
        "SLA_Matches_Authoritative"
    ].all()
)

print(
    "Seed Mass agreement  :",
    source_check[
        "Seed_Mass_Matches_Authoritative"
    ].all()
)

print(
    "Max Height agreement :",
    source_check[
        "Max_Height_Matches_Authoritative"
    ].all()
)


assert source_check[
    "WD_Matches_Authoritative"
].all()

assert source_check[
    "SLA_Matches_Authoritative"
].all()

assert source_check[
    "Seed_Mass_Matches_Authoritative"
].all()

assert source_check[
    "Max_Height_Matches_Authoritative"
].all()


# -----------------------------------------------------------------------------
# 11.4K. Preserve a compact authoritative core matrix
# -----------------------------------------------------------------------------

core_matrix = final_matrix[
    [
        "Canonical_Species",

        "WD_Value",
        "WD_Unit",

        "SLA_Value",
        "SLA_Unit",

        "Seed_Mass_mg",
        "Seed_Mass_Unit",

        "Max_Height_m",
        "Max_Height_Unit",

        "N_Numeric_Core_Traits",
        "Final_Core_Trait_Status",

        "Integrated_Evidence_Profile",
        "N_Direct_or_Reconciled_Traits",
        "N_Estimated_or_Specialized_Traits",
        "N_Special_Biological_Cases",
    ]
].copy()


assert len(core_matrix) == 127
assert core_matrix["Canonical_Species"].nunique() == 127


# -----------------------------------------------------------------------------
# 11.4L. Final provenance completeness
# -----------------------------------------------------------------------------

final_provenance_fields = [
    "WD_Provenance_Class",
    "SLA_Provenance_Class",
    "Seed_Mass_Provenance_Class",
    "Max_Height_Provenance_Class",
]


provenance_missing = {
    field:
        int(
            final_matrix[
                field
            ].isna().sum()
        )
    for field in final_provenance_fields
}


print("\nFinal provenance completeness")
print("-" * 100)

for field, n_missing in provenance_missing.items():

    print(
        f"{field:<30}: "
        f"{n_missing} missing"
    )

    assert n_missing == 0


# -----------------------------------------------------------------------------
# 11.4M. Final integrated evidence-profile summary
# -----------------------------------------------------------------------------

final_evidence_summary = (
    final_matrix[
        "Integrated_Evidence_Profile"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Integrated_Evidence_Profile"
    )
    .reset_index(
        name="N_Taxa"
    )
)


final_evidence_summary[
    "Percent_of_127"
] = (
    final_evidence_summary[
        "N_Taxa"
    ]
    /
    127
    * 100
)


print("\nFinal integrated evidence-profile summary")
print("-" * 100)

display(
    final_evidence_summary
)


assert final_evidence_summary[
    "N_Taxa"
].sum() == 127


# -----------------------------------------------------------------------------
# 11.4N. Final trait completeness summary
# -----------------------------------------------------------------------------

final_completeness_summary = pd.DataFrame(
    [
        {
            "Trait":
                "WOOD_DENSITY",

            "Value_Field":
                "WD_Value",

            "Numeric_Values":
                int(
                    final_matrix[
                        "WD_Value"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    final_matrix[
                        "WD_Value"
                    ].isna().sum()
                ),

            "Coverage_Percent":
                100
                * final_matrix[
                    "WD_Value"
                ].notna().sum()
                / 127,
        },

        {
            "Trait":
                "SPECIFIC_LEAF_AREA",

            "Value_Field":
                "SLA_Value",

            "Numeric_Values":
                int(
                    final_matrix[
                        "SLA_Value"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    final_matrix[
                        "SLA_Value"
                    ].isna().sum()
                ),

            "Coverage_Percent":
                100
                * final_matrix[
                    "SLA_Value"
                ].notna().sum()
                / 127,
        },

        {
            "Trait":
                "SEED_MASS",

            "Value_Field":
                "Seed_Mass_mg",

            "Numeric_Values":
                int(
                    final_matrix[
                        "Seed_Mass_mg"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    final_matrix[
                        "Seed_Mass_mg"
                    ].isna().sum()
                ),

            "Coverage_Percent":
                100
                * final_matrix[
                    "Seed_Mass_mg"
                ].notna().sum()
                / 127,
        },

        {
            "Trait":
                "MAXIMUM_HEIGHT",

            "Value_Field":
                "Max_Height_m",

            "Numeric_Values":
                int(
                    final_matrix[
                        "Max_Height_m"
                    ].notna().sum()
                ),

            "Missing_Values":
                int(
                    final_matrix[
                        "Max_Height_m"
                    ].isna().sum()
                ),

            "Coverage_Percent":
                100
                * final_matrix[
                    "Max_Height_m"
                ].notna().sum()
                / 127,
        },
    ]
)


print("\nFinal trait completeness summary")
print("-" * 100)

display(
    final_completeness_summary
)


# -----------------------------------------------------------------------------
# 11.4O. Final provenance summary
# -----------------------------------------------------------------------------

final_provenance_summary = provenance_audit.copy()


assert (
    final_provenance_summary
    .groupby("Trait")[
        "N_Taxa"
    ]
    .sum()
    .eq(127)
    .all()
)


# -----------------------------------------------------------------------------
# 11.4P. Final closure validation report
# -----------------------------------------------------------------------------

closure_validation = pd.DataFrame(
    [
        {
            "Validation_Check":
                "Final matrix rows",

            "Expected":
                127,

            "Observed":
                len(final_matrix),

            "Status":
                "PASS"
                if len(final_matrix) == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Unique canonical taxa",

            "Expected":
                127,

            "Observed":
                final_matrix[
                    "Canonical_Species"
                ].nunique(),

            "Status":
                "PASS"
                if final_matrix[
                    "Canonical_Species"
                ].nunique() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "WD numeric values",

            "Expected":
                127,

            "Observed":
                int(
                    final_matrix[
                        "WD_Value"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "WD_Value"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "SLA numeric values",

            "Expected":
                127,

            "Observed":
                int(
                    final_matrix[
                        "SLA_Value"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "SLA_Value"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Seed Mass numeric values",

            "Expected":
                126,

            "Observed":
                int(
                    final_matrix[
                        "Seed_Mass_mg"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "Seed_Mass_mg"
                ].notna().sum() == 126
                else "FAIL",
        },

        {
            "Validation_Check":
                "Maximum Height numeric values",

            "Expected":
                127,

            "Observed":
                int(
                    final_matrix[
                        "Max_Height_m"
                    ].notna().sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "Max_Height_m"
                ].notna().sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Complete 4-of-4 taxa",

            "Expected":
                126,

            "Observed":
                int(
                    final_matrix[
                        "Final_Core_Trait_Status"
                    ]
                    .eq(
                        "COMPLETE_4_OF_4"
                    )
                    .sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "Final_Core_Trait_Status"
                ].eq(
                    "COMPLETE_4_OF_4"
                ).sum() == 126
                else "FAIL",
        },

        {
            "Validation_Check":
                "Documented biological exception",

            "Expected":
                1,

            "Observed":
                int(
                    final_matrix[
                        "Final_Core_Trait_Status"
                    ]
                    .eq(
                        "COMPLETE_WITH_DOCUMENTED_BIOLOGICAL_EXCEPTION"
                    )
                    .sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "Final_Core_Trait_Status"
                ].eq(
                    "COMPLETE_WITH_DOCUMENTED_BIOLOGICAL_EXCEPTION"
                ).sum() == 1
                else "FAIL",
        },

        {
            "Validation_Check":
                "Unexpected incomplete taxa",

            "Expected":
                0,

            "Observed":
                int(
                    final_matrix[
                        "Final_Core_Trait_Status"
                    ]
                    .eq(
                        "UNEXPECTED_INCOMPLETE"
                    )
                    .sum()
                ),

            "Status":
                "PASS"
                if final_matrix[
                    "Final_Core_Trait_Status"
                ].eq(
                    "UNEXPECTED_INCOMPLETE"
                ).sum() == 0
                else "FAIL",
        },

        {
            "Validation_Check":
                "WD authoritative agreement",

            "Expected":
                127,

            "Observed":
                int(
                    source_check[
                        "WD_Matches_Authoritative"
                    ].sum()
                ),

            "Status":
                "PASS"
                if source_check[
                    "WD_Matches_Authoritative"
                ].sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "SLA authoritative agreement",

            "Expected":
                127,

            "Observed":
                int(
                    source_check[
                        "SLA_Matches_Authoritative"
                    ].sum()
                ),

            "Status":
                "PASS"
                if source_check[
                    "SLA_Matches_Authoritative"
                ].sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Seed Mass authoritative agreement",

            "Expected":
                127,

            "Observed":
                int(
                    source_check[
                        "Seed_Mass_Matches_Authoritative"
                    ].sum()
                ),

            "Status":
                "PASS"
                if source_check[
                    "Seed_Mass_Matches_Authoritative"
                ].sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Maximum Height authoritative agreement",

            "Expected":
                127,

            "Observed":
                int(
                    source_check[
                        "Max_Height_Matches_Authoritative"
                    ].sum()
                ),

            "Status":
                "PASS"
                if source_check[
                    "Max_Height_Matches_Authoritative"
                ].sum() == 127
                else "FAIL",
        },

        {
            "Validation_Check":
                "Missing provenance fields",

            "Expected":
                0,

            "Observed":
                int(
                    sum(
                        provenance_missing.values()
                    )
                ),

            "Status":
                "PASS"
                if sum(
                    provenance_missing.values()
                ) == 0
                else "FAIL",
        },
    ]
)


print("\nNotebook 07B closure validation report")
print("-" * 100)

display(
    closure_validation
)


assert (
    closure_validation[
        "Status"
    ].eq(
        "PASS"
    )
    .all()
)


# -----------------------------------------------------------------------------
# 11.4Q. Final canonical ordering
# -----------------------------------------------------------------------------

# Preserve WD canonical order as the final stable order,
# since all taxon sets are identical and merges were explicit.

canonical_order = (
    wd_final[
        [
            "Canonical_Species"
        ]
    ]
    .reset_index()
    .rename(
        columns={
            "index":
                "_Canonical_Order"
        }
    )
)


final_matrix = final_matrix.merge(
    canonical_order,
    on="Canonical_Species",
    how="left"
)


core_matrix = core_matrix.merge(
    canonical_order,
    on="Canonical_Species",
    how="left"
)


final_matrix = (
    final_matrix
    .sort_values(
        "_Canonical_Order"
    )
    .drop(
        columns=[
            "_Canonical_Order"
        ]
    )
    .reset_index(
        drop=True
    )
)


core_matrix = (
    core_matrix
    .sort_values(
        "_Canonical_Order"
    )
    .drop(
        columns=[
            "_Canonical_Order"
        ]
    )
    .reset_index(
        drop=True
    )
)


# -----------------------------------------------------------------------------
# 11.4R. Notebook closure summary
# -----------------------------------------------------------------------------

closure_summary = pd.DataFrame(
    [
        {
            "Notebook":
                "07B",

            "Component":
                "Wood Density",

            "Status":
                "CLOSED",

            "Canonical_Taxa":
                127,

            "Numeric_Values":
                127,

            "Missing_Numeric":
                0,
        },

        {
            "Notebook":
                "07B",

            "Component":
                "Specific Leaf Area",

            "Status":
                "CLOSED",

            "Canonical_Taxa":
                127,

            "Numeric_Values":
                127,

            "Missing_Numeric":
                0,
        },

        {
            "Notebook":
                "07B",

            "Component":
                "Seed Mass",

            "Status":
                "CLOSED",

            "Canonical_Taxa":
                127,

            "Numeric_Values":
                126,

            "Missing_Numeric":
                1,
        },

        {
            "Notebook":
                "07B",

            "Component":
                "Maximum Height",

            "Status":
                "CLOSED",

            "Canonical_Taxa":
                127,

            "Numeric_Values":
                127,

            "Missing_Numeric":
                0,
        },

        {
            "Notebook":
                "07B",

            "Component":
                "Four-Trait Integration",

            "Status":
                "CLOSED",

            "Canonical_Taxa":
                127,

            "Numeric_Values":
                126,

            "Missing_Numeric":
                1,
        },
    ]
)


# -----------------------------------------------------------------------------
# 11.4S. Export authoritative Notebook 07B products
# -----------------------------------------------------------------------------

final_matrix.to_csv(
    FINAL_MATRIX_PATH,
    index=False
)

core_matrix.to_csv(
    CORE_MATRIX_PATH,
    index=False
)

closure_validation.to_csv(
    FINAL_VALIDATION_PATH,
    index=False
)

final_completeness_summary.to_csv(
    FINAL_COMPLETENESS_PATH,
    index=False
)

final_provenance_summary.to_csv(
    FINAL_PROVENANCE_PATH,
    index=False
)

closure_summary.to_csv(
    NOTEBOOK_CLOSURE_PATH,
    index=False
)


print("\nAuthoritative Notebook 07B exports")
print("-" * 100)

print(
    f"Final full trait matrix       : "
    f"{FINAL_MATRIX_PATH}"
)

print(
    f"Final core trait matrix       : "
    f"{CORE_MATRIX_PATH}"
)

print(
    f"Final validation report       : "
    f"{FINAL_VALIDATION_PATH}"
)

print(
    f"Final completeness summary    : "
    f"{FINAL_COMPLETENESS_PATH}"
)

print(
    f"Final provenance summary      : "
    f"{FINAL_PROVENANCE_PATH}"
)

print(
    f"Notebook closure summary      : "
    f"{NOTEBOOK_CLOSURE_PATH}"
)


# -----------------------------------------------------------------------------
# 11.4T. Final methodological closure statement
# -----------------------------------------------------------------------------

print("\nNotebook 07B closure statement")
print("-" * 100)

print(
    """
Notebook 07B — Functional Trait Acquisition, Harmonization and Integration
is now complete.

The authoritative Omo functional-trait matrix contains exactly 127
canonical taxa and four core traits:

    • Wood Density
    • Specific Leaf Area
    • Seed Mass
    • Maximum Height

Numeric completeness:

    WD             : 127 / 127
    SLA            : 127 / 127
    Seed Mass      : 126 / 127
    Maximum Height : 127 / 127

Exactly 126 taxa have four numeric trait values.

Musa sapiensis retains a formally documented Seed Mass biological
nonapplicability and therefore has three numeric traits rather than four.

This is not treated as unresolved missing data and no artificial Seed
Mass value has been imputed.

All final integrated trait values agree exactly with their authoritative
trait-specific source tables.

Trait-specific provenance, assignment level and quality metadata are
preserved in the full final matrix.

Notebook 07B introduces no silent taxonomic substitution and no
undocumented imputation.
"""
)


# -----------------------------------------------------------------------------
# 11.4U. Final summary
# -----------------------------------------------------------------------------

print("=" * 100)
print("SECTION 11.4 NOTEBOOK 07B FINAL CLOSURE COMPLETE")
print("=" * 100)

print(
    f"Canonical taxa                        : "
    f"{len(final_matrix)}"
)

print(
    f"Unique canonical taxa                 : "
    f"{final_matrix['Canonical_Species'].nunique()}"
)

print(
    f"WD numeric values                     : "
    f"{final_matrix['WD_Value'].notna().sum()}"
)

print(
    f"SLA numeric values                    : "
    f"{final_matrix['SLA_Value'].notna().sum()}"
)

print(
    f"Seed Mass numeric values              : "
    f"{final_matrix['Seed_Mass_mg'].notna().sum()}"
)

print(
    f"Maximum Height numeric values         : "
    f"{final_matrix['Max_Height_m'].notna().sum()}"
)

print(
    f"Complete 4-of-4 taxa                  : "
    f"{(final_matrix['Final_Core_Trait_Status'] == 'COMPLETE_4_OF_4').sum()}"
)

print(
    f"Documented biological exceptions      : "
    f"{(final_matrix['Final_Core_Trait_Status'] == 'COMPLETE_WITH_DOCUMENTED_BIOLOGICAL_EXCEPTION').sum()}"
)

print(
    f"Unexpected incomplete taxa            : "
    f"{(final_matrix['Final_Core_Trait_Status'] == 'UNEXPECTED_INCOMPLETE').sum()}"
)

print(
    f"Failed closure validation checks      : "
    f"{(closure_validation['Status'] != 'PASS').sum()}"
)

print(
    "Authoritative final matrix            : "
    "outputs/functional_traits/omo_functional_traits_final.csv"
)

print(
    "Notebook 07B status                   : FORMALLY CLOSED"
)

print(
    "Next project step                     : "
    "begin the next planned functional-ecology stage using the "
    "validated four-trait matrix."
)
```
